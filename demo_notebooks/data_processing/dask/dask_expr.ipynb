## Example 1

Create a dask dataframe with dask-expr

```python
import dask_expr as dx

df = dx.datasets.timeseries()
df.head()

df.groupby("name").x.mean().compute()
df.x.mean().pprint()
df.x.mean().optimize().pprint()
```
