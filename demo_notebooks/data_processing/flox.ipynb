## Example 1

GroupBy single variable with numpy and xarray

```python
import numpy as np
import xarray as xr

from flox.xarray import xarray_reduce

labels = xr.DataArray(
    [1, 2, 3, 1, 2, 3, 0, 0, 0],
    dims="x",
    name="label",
)

da = xr.DataArray(
    np.ones((9,)), dims="x", name="array"
)

xarray_reduce(da, labels, func="sum")
```

## Example 2

GroupBy with Dask and Xarray

```python
import xarray as xr
from flox.xarray import xarray_reduce

da = xr.DataArray(
    np.ones((9,)), dims="x", name="array"
)
da_chunked = da.chunk(x=2)

labels = xr.DataArray(
    [1, 2, 3, 1, 2, 3, 0, 0, 0],
    dims="x",
    name="label",
)
labels_chunked = labels.chunk(x=3)

xarray_reduce(da_chunked, labels, func="sum")
```
