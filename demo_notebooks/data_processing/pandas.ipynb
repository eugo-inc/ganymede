## Example 2

Download a remote csv asynchronously and create a dataframe from it

```python
import pandas as pd
from fsspec import filesystem
from pathlib import Path
import asyncio

tmp = "/tmp/demo_notebooks"
Path(tmp).mkdir(exist_ok=True)

url = "https://raw.githubusercontent.com/pandas-dev/pandas/main/doc/data/titanic.csv"

async def download_csv():
    fs = filesystem("http", asynchronous=True)
    session = await fs.set_session()
    out = await fs._get(url, f"{tmp}/titanic.csv")
    await session.close()

asyncio.run(download_csv())

titanic = pd.read_csv(f"{tmp}/titanic.csv")
titanic.head()
```
