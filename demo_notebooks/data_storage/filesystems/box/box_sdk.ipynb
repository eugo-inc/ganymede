## Environment variables

This notebook reads its settings from environment variables. Set them before starting Jupyter:

- `BOX_CLIENT_ID`: client ID of your Box app
- `BOX_CLIENT_SECRET`: client secret of your Box app
- `BOX_ACCESS_TOKEN`: a Box developer or OAuth access token
- `BOX_FOLDER_ID`: optional, the folder to open; defaults to `0` (the root folder)

```python
import os

import fsspec
from boxsdk import Client, OAuth2

oauth = OAuth2(
    client_id=os.environ["BOX_CLIENT_ID"],
    client_secret=os.environ["BOX_CLIENT_SECRET"],
    access_token=os.environ["BOX_ACCESS_TOKEN"],
)

client = Client(oauth)
client.folder(os.environ.get("BOX_FOLDER_ID", "0"))
```
