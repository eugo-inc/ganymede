## Example 1

Open and write a file

```python
import fsspec
from pathlib import Path

Path("/tmp/ganymede_demo").mkdir(parents=True, exist_ok=True)

with fsspec.open("/tmp/ganymede_demo/test_file.txt", "w") as f:
    f.write("This file was written with fsspec")
```

## Example 2

Open and read the file

```python
import fsspec

with fsspec.open("/tmp/ganymede_demo/test_file.txt", "r") as f:
    print(f.read())
```

## Example 3

Use a filesystem object to list a directory and print file contents

```python
import fsspec

fs = fsspec.filesystem("file")
print(fs.ls("/tmp/ganymede_demo", detail=False))
fs.cat("/tmp/ganymede_demo/test_file.txt")
```
