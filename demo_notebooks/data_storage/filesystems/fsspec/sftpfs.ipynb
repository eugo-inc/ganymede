## Environment variables

This notebook reads its settings from environment variables. Set them before starting Jupyter:

- `SFTP_HOST`: hostname of your SFTP server
- `SFTP_PORT`: optional, defaults to `22`
- `SFTP_USER`: SFTP user name
- `SFTP_PRIVATE_KEY_PATH`: path to that user's private key file

## Example 1

List contents of the filesystem via SSH

```python
import os

from fsspec import filesystem

sftp_user = os.environ["SFTP_USER"]

# Extra keyword arguments are passed to paramiko's SSHClient.connect()
fs = filesystem("sftp",
                host=os.environ["SFTP_HOST"],
                port=int(os.environ.get("SFTP_PORT", "22")),
                username=sftp_user,
                key_filename=os.environ["SFTP_PRIVATE_KEY_PATH"])

fs.ls(f"/home/{sftp_user}")
```

## Example 2

List contents of the filesystem via SFTP

```python
import os

from fsspec.implementations.sftp import SFTPFileSystem

sftp_user = os.environ["SFTP_USER"]

sftp = SFTPFileSystem(host=os.environ["SFTP_HOST"],
                port=int(os.environ.get("SFTP_PORT", "22")),
                username=sftp_user,
                key_filename=os.environ["SFTP_PRIVATE_KEY_PATH"])

sftp.info(f"/home/{sftp_user}")
```
