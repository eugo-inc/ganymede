# OrderedMultidict

Description:
- An ordered multivalue dictionary that retains
method parity with Python's dict and helps power furl.

Resources:
- [GitHub] https://github.com/gruns/orderedmultidict

## Example 1

Create an ordered dict

```python
from orderedmultidict import omdict

omd = omdict()
omd[1] = 1
omd[1]

omd.add(1, 11)
omd.getlist(1)

omd.addlist(1, [111, 1111])
omd.getlist(1)

omd.allitems()
```
