## Example 1

Create a colormap

```python
from branca.colormap import linear

colormap_choice = linear.YlOrRd_04
vmin = 3.2
vmax = 10.3
colormap = colormap_choice.scale(vmin, vmax)
colormap.caption = 'Unemployment rate'
colormap
```
