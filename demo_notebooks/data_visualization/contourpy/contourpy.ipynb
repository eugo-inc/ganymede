# ContourPy

Description:
- Library for calculating contours of 2D quadrilateral grids

Resources:
- [GitHub] https://github.com/contourpy/contourpy
- [Docs] https://contourpy.readthedocs.io/en/v1.2.0/
- [Source Code] https://github.com/contourpy/contourpy/tree/main/lib/contourpy

## Example 1

Calculate filled contours and contour lines on a small grid

```python
from contourpy import contour_generator

z = [[0.0, 0.1], [0.2, 0.3]]
cont_gen = contour_generator(z=z)
cont_gen
```

```python
filled = cont_gen.filled(0.15, 0.25)
filled
```

```python
lines = cont_gen.lines(0.25)
lines
```

## Example 2

Draw filled contours and contour lines with Matplotlib, using contourpy's Matplotlib renderer

```python
from contourpy import contour_generator
from contourpy.util.mpl_renderer import MplRenderer as Renderer

z = [[0.0, 0.1], [0.2, 0.3]]
cont_gen = contour_generator(z=z)
lines = cont_gen.lines(0.25)
filled = cont_gen.filled(0.15, 0.25)

renderer = Renderer(figsize=(4, 2.5))
renderer.filled(filled, cont_gen.fill_type, color="gold")
renderer.lines(lines, cont_gen.line_type, color="red", linewidth=2)
# renderer.show()
```

## Example 3

Draw the same contours with Bokeh, using contourpy's Bokeh renderer

```python
from bokeh.io import output_notebook
from contourpy import contour_generator
from contourpy.util.bokeh_renderer import BokehRenderer

# Show Bokeh plots inside the notebook instead of opening a new browser tab
output_notebook()

z = [[0.0, 0.1], [0.2, 0.3]]
cont_gen = contour_generator(z=z)

renderer = BokehRenderer(figsize=(4, 2.5))
renderer.filled(cont_gen.filled(0.15, 0.25), cont_gen.fill_type, color="gold")
renderer.lines(cont_gen.lines(0.25), cont_gen.line_type, color="red", linewidth=2)
renderer.show()
```
