# Geoplot

Description:
- A high-level Python geospatial plotting library. It's an extension to cartopy and matplotlib.

Resources:
- [GitHub] https://github.com/ResidentMario/geoplot
- [Docs] https://residentmario.github.io/geoplot/index.html
- [Examples] https://github.com/ResidentMario/geoplot/tree/master/examples

## Example 1

Get a geodataframe of usa cities

```python
import pandas as pd
import geopandas as gpd
import geoplot as gplt

pd.set_option('max_columns', 6)

usa_cities = gpd.read_file(gplt.datasets.get_path('usa_cities'))
usa_cities.head()
```

## Example 2

Plot polygons of contiguous USA

```python
import geopandas as gpd
import geoplot as gplt

contiguous_usa = gpd.read_file(gplt.datasets.get_path('contiguous_usa'))
gplt.polyplot(contiguous_usa)
```
