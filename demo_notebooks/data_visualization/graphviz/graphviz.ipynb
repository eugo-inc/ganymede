# Graphviz

Description:
- This package facilitates the creation and rendering of graph descriptions in the DOT language of the Graphviz graph drawing software (upstream repo) from Python.

Resources:
- [GitHub] https://github.com/xflr6/graphviz
- [Docs] https://graphviz.readthedocs.io/en/stable/

## Example 1

Create a graph object with nodes and edges

```python
import graphviz
from IPython.display import Image

dot = graphviz.Digraph(comment='The Round Table')
dot.node('A', 'King Arthur')
dot.node('B', 'Sir Bedevere the Wise')
dot.node('L', 'Sir Lancelot the Brave')

dot.edges(['AB', 'AL'])
dot.edge('B', 'L', constraint='false')

# Lay the graph out with Graphviz's dot program and show the result as a PNG image.
# Ending the cell with just `dot` shows it as SVG in Jupyter; a PNG also shows on GitHub.
# dot.render('round-table.gv', view=True) would instead save the DOT source and a PDF and open it.
Image(dot.pipe(format='png'))
```

## Example 2

Create a graph object as PNG

```python
import graphviz
from IPython.display import Image

g = graphviz.Graph(format='png')
g.edge('Hello', 'World')

Image(g.pipe())  # pipe() renders in the graph's format, PNG here, and returns the bytes
```
