# PyGraphviz

Description:

## Example 1

Create a simple example to create a graphviz dot file and draw a graph.

```python
import pygraphviz as pgv
from pathlib import Path


tmp = "/tmp/demo_notebooks"
Path(tmp).mkdir(exist_ok=True)

A = pgv.AGraph()

A.add_edge(1, 2)
A.add_edge(2, 3)
A.add_edge(1, 3)

print(A.string())  # print to screen
A.write(f"{tmp}/simple.dot")  # write to simple.dot

B = pgv.AGraph(f"{tmp}/simple.dot")  # create a new graph from file
B.layout()  # layout with default (neato)
B.draw(f"{tmp}/simple.dot")  # draw png
```
