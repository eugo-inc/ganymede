# Railroad Diagrams

Description:
- Draw railroad (syntax) diagrams as SVG from Python.

Requirements:
- `pip install railroad-diagrams resvg-py` (resvg-py turns the SVG into a PNG image for display).

Resources:
- [GitHub] https://github.com/tabatkins/railroad-diagrams/tree/gh-pages
- [Python ReadMe] https://github.com/tabatkins/railroad-diagrams/blob/gh-pages/README-py.md
- [Online Playground] https://tabatkins.github.io/railroad-diagrams/generator.html

## Example 1

Construct a simple "choice" diagram

```python
import io

import resvg_py
from IPython.display import Image
from railroad import Diagram, Choice

d = Diagram("foo", Choice(0, "bar", "baz"))

# writeStandalone() writes the diagram as SVG together with railroad's default CSS
# (writeSvg() leaves the CSS out). resvg turns the SVG into a PNG, which shows in any
# notebook viewer, GitHub included.
svg = io.StringIO()
d.writeStandalone(svg.write)
Image(resvg_py.svg_to_bytes(svg_string=svg.getvalue(), zoom=2))
```
