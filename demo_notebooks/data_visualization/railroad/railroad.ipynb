## Example 1

Construct a simple "choice" diagram

```python
import sys

from railroad import Diagram, Choice

d = Diagram("foo", Choice(0, "bar", "baz"))
d.writeSvg(sys.stdout.write)
```

## Example 2

Create a simple "sequence" diagram

```python
import sys

from railroad import Diagram, Sequence

d = Diagram("bar", Sequence("1", "2", "3"))
d.writeSvg(sys.stdout.write)
```
