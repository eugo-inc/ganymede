# `Cftime`

Description:
- Time-handling functionality from netcdf4-python.

Resources:
- [GitHub] https://github.com/Unidata/cftime
- [Docs] https://unidata.github.io/cftime/

## Example 1

Calculate the number of days since a date in the past

```python
import cftime
from datetime import datetime

date = datetime(1582, 10, 4, 23)

num = cftime.date2num(date, 'days since 2000-01-01')
num
```
