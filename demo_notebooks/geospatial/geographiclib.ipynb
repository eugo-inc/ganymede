# Geographiclib

Description:
- A python implementation of the geodesic routines in GeographicLib. It allows the solution geodesic problems on an ellipsoid model of the earth.

Resources:
- [PyPi] https://pypi.org/project/geographiclib/#files
- [Docs] https://geographiclib.sourceforge.io/Python/2.0/code.html#geographiclib-api
- [Examples] https://geographiclib.sourceforge.io/Python/2.0/examples.html

## Example 1

The distance from Wellington, NZ (41.32S, 174.81E) to Salamanca, Spain (40.96N, 5.50W) using Inverse()

```python
from geographiclib.geodesic import Geodesic
import math

geod = Geodesic.WGS84

g = geod.Inverse(-41.32, 174.81, 40.96, -5.50)
print("The distance is {:.3f} m.".format(g['s12']))
```
