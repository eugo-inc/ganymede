## Environment variables

This notebook reads these environment variables. Set them before you start Jupyter:

- `AZURE_STORAGE_ACCOUNT_NAME`: your Azure Storage account name.
- `AZURE_STORAGE_CONTAINER`: a blob container in that storage account.
- `AZURE_STORAGE_ACCOUNT_KEY`: an access key for that storage account.

## Example 1: Synchronous - Private Data

Synchronously list the blobs in a private container

```python
import os
from azure.storage.blob import BlobServiceClient, BlobClient, ContainerClient

account_name = os.environ["AZURE_STORAGE_ACCOUNT_NAME"]
container_name = os.environ["AZURE_STORAGE_CONTAINER"]
account_key = os.environ["AZURE_STORAGE_ACCOUNT_KEY"]
account_url = f"https://{account_name}.blob.core.windows.net/{container_name}"

storage_options = {'account_name': account_name, 'account_key': account_key, 'account_url': account_url, 'container_name': container_name}
container_client = ContainerClient(**storage_options)
container_client.list_blobs()
```
