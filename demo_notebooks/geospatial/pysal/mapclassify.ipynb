# MapClassify

Description:
- An open-source python library for Choropleth map classification. It is part of PySAL the Python Spatial Analysis Library.

Resources:
- [GitHub] https://github.com/pysal/mapclassify
- [Docs] https://github.com/pysal/mapclassify/blob/main/docs/index.rst
- [Tutorials] https://github.com/pysal/mapclassify/tree/main/notebooks

## Example 1

Load built-in example data reporting employment density in 58 California counties:

```python
import mapclassify
y = mapclassify.load_example()

print(y.mean())
print(y.min(), y.max())
```

## Example 2

Create a boxplot

```python
import mapclassify
y = mapclassify.load_example()

mapclassify.BoxPlot(y)
```
