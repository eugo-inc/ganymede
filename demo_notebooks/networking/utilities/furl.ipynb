# Furl

Description:
- Make parsing and manipulating URLs easy

Resources:
- [GitHub] https://github.com/gruns/furl/blob/master/README.md

## Example 1

Get paths and query arguments

```python
from furl import furl
f = furl('http://www.google.com/?one=1&two=2')
f /= 'path'

del f.args['one']

f.args['three'] = '3'

f.url
```

## Example 2

Modify paths inline

```python
from furl import furl

furl('http://www.google.com/?one=1').add({'two':'2'}).url
furl('http://www.google.com/?one=1&two=2').set({'three':'3'}).url
furl('http://www.google.com/?one=1&two=2').remove(['one']).url
```
