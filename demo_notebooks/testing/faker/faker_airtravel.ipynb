Resources:

## Example 1

Generate a fake airport, airport name, airline, etc

```python
from faker import Faker
from faker_airtravel import AirTravelProvider

faker = Faker()
faker.add_provider(AirTravelProvider)

faker.airport_object()
faker.airport_name()
faker.airport_iata()
faker.airport_icao()
faker.airline()
faker.flight()
```
