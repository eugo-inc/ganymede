## Example 1

Generate a credit score, etc

```python
from faker import Faker
from faker_credit_score import CreditScore

faker = Faker()
faker.add_provider(CreditScore)

faker.credit_score_name()
faker.credit_score_provider()
faker.credit_score()
```
