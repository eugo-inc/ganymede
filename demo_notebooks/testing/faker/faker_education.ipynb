## Example 1

Generate a school name, district, etc

```python
from faker import Faker
from faker_education import SchoolProvider

faker = Faker()
faker.add_provider(SchoolProvider)

faker.school_object()
faker.school_name()
faker.school_nces_id()
faker.school_district()
faker.school_level()
faker.school_type()
faker.school_state()
```
