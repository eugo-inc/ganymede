# Faker Marketdata

Description:
- Work with fake market data

Resources:
- https://github.com/rubenafo/faker_marketdata

## Example 1:

Generate fake ticker data, etc

```python
from faker import Faker
from faker_marketdata import MarketDataProvider

Faker.seed(123)

faker = Faker()
faker.add_provider(MarketDataProvider)

faker.isin()
faker.sedol()
faker.cusip()
faker.ticker()
```
