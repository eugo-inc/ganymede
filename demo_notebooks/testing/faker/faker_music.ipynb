## Example 1:

Generate fake music genres, instruments, etc

```python
from faker import Faker
from faker_music import MusicProvider

faker = Faker()
faker.add_provider(MusicProvider)
faker.music_genre()
faker.music_subgenre()
faker.music_instrument()
faker.music_instrument_category()
faker.music_genre_object()
faker.music_instrument_object()
```
