## Example 1:

Generate fake pyspark dataframe

```python
from faker_pyspark import PySparkProvider
from faker import Faker

faker = Faker()
faker.add_provider(PySparkProvider)

df           = faker.pyspark_dataframe()
schema       = faker.pyspark_schema()
df_updated   = faker.pyspark_update_dataframe(df)
column_names = faker.pyspark_column_names()
data         = faker.pyspark_data_dict_using_schema(schema)
data         = faker.pyspark_data_dict()

print(df)
```
