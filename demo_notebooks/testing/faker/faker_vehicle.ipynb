# Faker Vehicle

Description:
- Work with fake vehicle data

Resources:
- https://github.com/kennethwsmith/faker_vehicle

## Example 1:

Generate fake vehicle categories, models, etc

```python
import json
from faker import Faker
from faker_vehicle import VehicleProvider

faker = Faker()
faker.add_provider(VehicleProvider)

faker.vehicle_year_make_model()
faker.vehicle_year_make_model_cat()
faker.vehicle_make_model()
faker.vehicle_make()
faker.vehicle_model()
faker.vehicle_year()
faker.vehicle_category()
json.dumps(faker.vehicle_object())

faker.machine_year_make_model()
faker.machine_year_make_model_cat()
faker.machine_make_model()
faker.machine_make()
faker.machine_model()
faker.machine_year()
faker.machine_category()
json.dumps(faker.machine_object())
```
