## Example 1: Time a block of code (context manager)

```python
from about_time import about_time

with about_time() as t:
    total = sum(i * i for i in range(200_000))

assert total == sum(i * i for i in range(200_000))
assert t.duration > 0, t.duration
print("block took", t.duration_human)
```

## Example 2: Time a call and keep its result

```python
from about_time import about_time

t = about_time(sorted, [5, 3, 9, 1])
assert t.result == [1, 3, 5, 9], t.result
assert t.duration >= 0
print("sorted() took", t.duration_human, "->", t.result)
```

## Example 3: Count items and throughput while iterating

```python
from about_time import about_time

t = about_time(x * 2 for x in range(10_000))
doubled = [x for x in t]
assert len(doubled) == 10_000 and t.count == 10_000, t.count
assert t.throughput > 0
print(t.count_human, "items at", t.throughput_human)
```

## Example 4: Human-readable durations

```python
from about_time import HumanDuration

samples = {0.0000012: "µs", 0.0012: "ms", 1.5: "s", 75.0: ":"}
for seconds, marker in samples.items():
    text = str(HumanDuration(seconds))
    assert marker in text, (seconds, text)
    print(f"{seconds!r:>10} s -> {text}")
```

## Example 5: Human-readable counts and rates with units

```python
from about_time import HumanCount, HumanThroughput

count = HumanCount(1_500_000, "B")
rate = HumanThroughput(2_500.0, "req")
assert str(count).endswith("MB"), str(count)
assert "req/s" in str(rate), str(rate)
print(count, "|", rate, "|", count.as_human(3))
```
