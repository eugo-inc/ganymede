```python
# Example 2: High contrast syntax highlighting with accessible themes
from pygments import highlight
from pygments.lexers import get_lexer_by_name
from pygments.formatters import HtmlFormatter
from pygments.styles import get_all_styles

print("Example 2: High contrast syntax highlighting with accessible themes")

# JavaScript code sample for demonstration
js_code = '''
// High contrast accessible highlighting demo
class AccessibilityHelper {
    constructor(config) {
        this.contrast = config.highContrast;
        this.theme = config.theme || 'a11y-dark';
    }
    
    applyTheme() {
        const elements = document.querySelectorAll('code');
        elements.forEach(el => {
            el.classList.add(this.theme);
        });
        return elements.length;
    }
    
    // Check accessibility compliance
    checkWCAG() {
        return this.contrast >= 4.5 ? 'AA' : 'A';
    }
}

const helper = new AccessibilityHelper({
    highContrast: 7.0,
    theme: 'accessible-dark'
});
'''

# Get all available styles (accessible-pygments styles are automatically registered)
all_styles = list(get_all_styles())
print(f"Available styles: {len(all_styles)} total")

# Look for accessible/a11y styles
accessible_styles = [style for style in all_styles if 'a11y' in style.lower()]
high_contrast_styles = [style for style in all_styles if 'high-contrast' in style.lower()]

print(f"A11y styles detected: {accessible_styles}")
print(f"High contrast styles detected: {high_contrast_styles}")

# Use JavaScript lexer for syntax highlighting
lexer = get_lexer_by_name('javascript')

# Test HTML output with accessible styles
assert accessible_styles, "accessible-pygments registered no a11y styles with Pygments"
test_styles = accessible_styles[:2]

for style_name in test_styles:
    try:
        print(f"\n--- Testing style: {style_name} ---")
        formatter = HtmlFormatter(style=style_name, cssclass=f'highlight-{style_name}')
        
        # Generate HTML with accessible styling
        highlighted_html = highlight(js_code, lexer, formatter)
        
        # Basic validation - check if HTML was generated
        html_length = len(highlighted_html)
        print(f"✓ Generated {html_length} characters of accessible HTML")
        
        # Check for accessibility features in the CSS
        css = formatter.get_style_defs()
        css_length = len(css)
        print(f"✓ Generated {css_length} characters of accessible CSS")
        
        # Show a snippet of the CSS for validation
        css_preview = css[:200] + "..." if len(css) > 200 else css
        print(f"✓ CSS preview: {css_preview}")
        
    except Exception as e:
        print(f"✗ Error with style {style_name}: {e}")
        raise

print("\n✓ High contrast accessible highlighting demonstrated")
print("Example 2 completed successfully!")
```

```python
# Example 3: Multi-language syntax highlighting with WCAG compliance
from pygments import highlight
from pygments.lexers import get_lexer_by_name, PythonLexer, JavaLexer, CssLexer
from pygments.formatters import HtmlFormatter
from pygments.styles import get_all_styles, get_style_by_name

print("Example 3: Multi-language syntax highlighting with WCAG compliance")

# Code samples in different languages
code_samples = {
    'python': '''
# Python accessibility helper
def calculate_contrast_ratio(color1, color2):
    """Calculate WCAG contrast ratio between two colors"""
    def luminance(color):
        return 0.2126 * color[0] + 0.7152 * color[1] + 0.0722 * color[2]
    
    l1 = luminance(color1)
    l2 = luminance(color2)
    
    if l1 > l2:
        return (l1 + 0.05) / (l2 + 0.05)
    else:
        return (l2 + 0.05) / (l1 + 0.05)

# Test WCAG AA compliance (4.5:1 ratio)
ratio = calculate_contrast_ratio([0.2, 0.2, 0.2], [0.9, 0.9, 0.9])
compliant = ratio >= 4.5
print(f"Contrast ratio: {ratio:.2f} ({'Compliant' if compliant else 'Not compliant'})")
''',
    
    'java': '''
// Java accessibility utilities
public class AccessibilityValidator {
    private static final double WCAG_AA_RATIO = 4.5;
    private static final double WCAG_AAA_RATIO = 7.0;
    
    public static boolean isWCAGCompliant(Color foreground, Color background) {
        double ratio = calculateContrastRatio(foreground, background);
        return ratio >= WCAG_AA_RATIO;
    }
    
    private static double calculateContrastRatio(Color c1, Color c2) {
        double l1 = getLuminance(c1);
        double l2 = getLuminance(c2);
        return Math.max(l1, l2) / Math.min(l1, l2);
    }
}
''',
    
    'css': '''
/* Accessible CSS for high contrast themes */
.a11y-high-contrast {
    /* WCAG AAA compliant colors */
    background-color: #000000;  /* Pure black */
    color: #ffffff;             /* Pure white */
    
    /* Ensure sufficient contrast for interactive elements */
    --focus-color: #ffff00;     /* High contrast yellow */
    --link-color: #00ffff;      /* High contrast cyan */
    --visited-color: #ff00ff;   /* High contrast magenta */
}

.code-highlight {
    border: 2px solid var(--focus-color);
    padding: 1rem;
    font-family: 'Courier New', monospace;
    font-size: 1.1rem;  /* Larger for better readability */
}
'''
}

# Get available styles (accessible-pygments already registered)
all_styles = list(get_all_styles())

# Test each language with accessible highlighting
lexers = {
    'python': PythonLexer(),
    'java': get_lexer_by_name('java'),
    'css': get_lexer_by_name('css')
}

# Use a11y styles specifically
a11y_styles = [s for s in all_styles if 'a11y' in s.lower()]
assert a11y_styles, "accessible-pygments registered no a11y styles with Pygments"
test_styles = a11y_styles[:1]

print(f"Testing with a11y style(s): {test_styles}")

for lang, code in code_samples.items():
    print(f"\n--- {lang.upper()} Code Highlighting ---")
    lexer = lexers[lang]
    
    for style_name in test_styles:
        try:
            # Create HTML formatter with accessible styling
            formatter = HtmlFormatter(
                style=style_name,
                cssclass=f'highlight-{lang}',
                linenos=False,  # Skip line numbers for cleaner output
                noclasses=True  # Inline styles for immediate visibility
            )
            
            # Highlight the code
            highlighted = highlight(code, lexer, formatter)
            
            # Validate output
            print(f"✓ {lang.capitalize()} highlighted with {style_name}")
            print(f"  Generated {len(highlighted)} characters of accessible HTML")
            
            # Check for basic accessibility features
            if 'style=' in highlighted:
                print(f"  ✓ Inline styles applied for immediate accessibility")
            
        except Exception as e:
            print(f"✗ Error highlighting {lang} with {style_name}: {e}")
            raise

print(f"\n✓ Multi-language accessible highlighting completed")
print("WCAG compliance features demonstrated across multiple languages")
print("Example 3 completed successfully!")
```

```python
# Example 4: Style comparison and accessibility metrics  
from pygments.styles import get_all_styles, get_style_by_name
from pygments.formatters import HtmlFormatter
from pygments import highlight
from pygments.lexers import PythonLexer
import re

print("Example 4: Style comparison and accessibility metrics")

# Sample code for consistency in comparison
sample_code = '''
def accessibility_demo():
    """Demonstrate accessible syntax highlighting"""
    # Keywords, strings, and comments
    message = "Testing WCAG compliance"
    threshold = 4.5  # WCAG AA contrast ratio
    
    if check_contrast_ratio() >= threshold:
        return {"status": "compliant", "level": "AA"}
    else:
        return {"status": "non-compliant", "warning": True}
'''

lexer = PythonLexer()
all_styles = list(get_all_styles())

print(f"Total styles available: {len(all_styles)}")

# Categorize styles
standard_styles = ['default', 'emacs', 'friendly', 'colorful', 'autumn']
a11y_styles = [s for s in all_styles if 'a11y' in s.lower()]

print(f"Standard styles (sample): {standard_styles}")
print(f"A11y accessible styles found: {a11y_styles}")

# Compare styles by analyzing their CSS output
def analyze_style_accessibility(style_name):
    """Basic analysis of style accessibility features"""
    try:
        formatter = HtmlFormatter(style=style_name, noclasses=True)
        highlighted = highlight(sample_code, lexer, formatter)
        
        # Extract color information from inline styles
        color_pattern = r'color:\s*#?([a-fA-F0-9]{3,6})'
        bg_pattern = r'background(?:-color)?:\s*#?([a-fA-F0-9]{3,6})'
        
        colors = re.findall(color_pattern, highlighted)
        backgrounds = re.findall(bg_pattern, highlighted)
        
        return {
            'style': style_name,
            'html_length': len(highlighted),
            'unique_colors': len(set(colors)),
            'unique_backgrounds': len(set(backgrounds)),
            'has_inline_styles': 'style=' in highlighted,
            'is_a11y': 'a11y' in style_name.lower()
        }
    except Exception as e:
        return {'style': style_name, 'error': str(e)}

# Test a mix of a11y and standard styles
assert a11y_styles, "accessible-pygments registered no a11y styles with Pygments"
test_styles = a11y_styles[:2] + standard_styles[:3]

print(f"\n--- Style Analysis ---")
results = []

for style_name in test_styles:
    analysis = analyze_style_accessibility(style_name)
    results.append(analysis)
    
    if 'error' in analysis:
        print(f"✗ {style_name}: {analysis['error']}")
    else:
        a11y_indicator = " (A11Y)" if analysis['is_a11y'] else ""
        print(f"✓ {style_name}{a11y_indicator}:")
        print(f"  HTML output: {analysis['html_length']} chars")
        print(f"  Unique colors: {analysis['unique_colors']}")
        print(f"  Unique backgrounds: {analysis['unique_backgrounds']}")
        print(f"  Inline styles: {analysis['has_inline_styles']}")

# Summary comparison
print(f"\n--- Accessibility Summary ---")
a11y_results = [r for r in results if r.get('is_a11y', False)]
standard_results = [r for r in results if r['style'] in standard_styles and 'error' not in r]

print(f"A11y styles tested: {len(a11y_results)}")
print(f"Standard styles tested: {len(standard_results)}")
failed = [r for r in results if 'error' in r]
assert not failed, f"styles failed to render: {failed}"
assert len(a11y_results) == len(a11y_styles[:2]), a11y_results

print("✓ Accessible-pygments provides WCAG-compliant alternatives:")
for style in a11y_styles:
    print(f"  - {style}")

print("✓ Style comparison and accessibility analysis completed")
print("Example 4 completed successfully!")
```
