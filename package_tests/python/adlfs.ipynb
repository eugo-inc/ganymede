## Example 5: adlfs functionality test 5

```python
# Example 5: fsspec maps the Azure protocols to adlfs and parses their URLs.
import fsspec
from adlfs import AzureBlobFileSystem

for protocol in ("abfs", "abfss", "az"):
    resolved = fsspec.get_filesystem_class(protocol)
    assert resolved is AzureBlobFileSystem, (protocol, resolved)

# The URL shapes Examples 2 and 4 print as prose, actually parsed.
assert AzureBlobFileSystem._strip_protocol("abfs://cont@acct.dfs.core.windows.net/dir/f.txt") == "cont/dir/f.txt"
assert AzureBlobFileSystem._strip_protocol("abfss://cont@acct.dfs.core.windows.net/dir/") == "cont/dir/"
assert AzureBlobFileSystem._strip_protocol("az://cont/dir/f.txt") == "cont/dir/f.txt"

# The entry point pandas and dask use, in one call: protocol lookup -> instantiation -> path stripping.
fs, path = fsspec.core.url_to_fs(
    "abfs://cont@acct.dfs.core.windows.net/dir/f.txt", account_name="acct", anon=True, skip_instance_cache=True
)
assert isinstance(fs, AzureBlobFileSystem), type(fs)
assert path == "cont/dir/f.txt", path
print("abfs/abfss/az -> AzureBlobFileSystem | url_to_fs ->", type(fs).__name__, path)
```

## Example 6: adlfs functionality test 6

```python
# Example 6: construct the filesystem and the Azure SDK client under it, with no network.
#
# Constructing AzureBlobFileSystem builds a real azure.storage.blob.aio.BlobServiceClient and runs
# split_path; an import-only test would not catch a break in either.
#
# `anon=True` takes the same construction path with no credential in the notebook, and keeps
# DefaultAzureCredential from probing the instance metadata service. `skip_instance_cache=True`:
# fsspec caches instances, and a cached one built by an earlier cell would let this pass for the
# wrong reason.
#
# This cell sends no network request.
from adlfs import AzureBlobFileSystem

fs = AzureBlobFileSystem(account_name="acct", anon=True, skip_instance_cache=True)
assert fs.service_client.url == "https://acct.blob.core.windows.net/", fs.service_client.url

blob_url = fs.service_client.get_container_client("cont").get_blob_client("dir/f.txt").url
assert blob_url == "https://acct.blob.core.windows.net/cont/dir/f.txt", blob_url

assert fs.split_path("abfs://cont@acct.dfs.core.windows.net/dir/f.txt") == ("cont", "dir/f.txt", None)
assert fs.split_path("az://cont/dir/f.txt") == ("cont", "dir/f.txt", None)
assert fs.split_path("cont/dir/f.txt") == ("cont", "dir/f.txt", None)
assert fs.split_path("cont") == ("cont", "", None)  # a bare container is a container, never a key
print("BlobServiceClient ->", fs.service_client.url, "| blob ->", blob_url)
```
