## Example 1: aenum functionality test 1

```python
import aenum

# Example 1: Basic aenum Enum functionality
print("Example 1: Testing basic aenum Enum")

try:
    # Create a simple enum using aenum
    class Color(aenum.Enum):
        RED = 1
        GREEN = 2
        BLUE = 3
    
    print("Created Color enum successfully")
    print(f"Color members: {list(Color)}")
    print(f"Color.RED: {Color.RED}")
    print(f"Color.RED.name: {Color.RED.name}")
    print(f"Color.RED.value: {Color.RED.value}")
    assert [c.name for c in Color] == ['RED', 'GREEN', 'BLUE'] and Color.RED.value == 1
    
    print("aenum example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aenum example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: aenum functionality test 2

```python
import aenum

# Example 2: Advanced aenum functionality with auto values
print("Example 2: Testing aenum with auto values")

try:
    # Create enum with automatic values
    class Planet(aenum.Enum):
        MERCURY = aenum.auto()
        VENUS = aenum.auto()
        EARTH = aenum.auto()
        MARS = aenum.auto()
    
    print("Created Planet enum with auto values")
    print(f"Planet members: {list(Planet)}")
    print(f"Planet.EARTH: {Planet.EARTH}")
    print(f"Planet.EARTH.value: {Planet.EARTH.value}")
    
    # Test enum iteration
    for planet in Planet:
        print(f"  {planet.name}: {planet.value}")
    assert [p.value for p in Planet] == [1, 2, 3, 4], list(Planet)
    
    print("aenum example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aenum example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: aenum functionality test 3

```python
import aenum

# Example 3: aenum Flag functionality
print("Example 3: Testing aenum Flag for bit operations")

try:
    # Create a Flag enum for bit operations
    class Permission(aenum.Flag):
        READ = aenum.auto()
        WRITE = aenum.auto()
        EXECUTE = aenum.auto()
    
    print("Created Permission flag enum")
    print(f"READ: {Permission.READ}")
    print(f"WRITE: {Permission.WRITE}")
    print(f"EXECUTE: {Permission.EXECUTE}")
    
    # Combine flags
    read_write = Permission.READ | Permission.WRITE
    print(f"READ | WRITE: {read_write}")
    print(f"Has READ permission: {Permission.READ in read_write}")
    assert Permission.READ in read_write and Permission.EXECUTE not in read_write
    
    print("aenum example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aenum example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: aenum functionality test 4

```python
import aenum

# Example 4: aenum IntEnum functionality
print("Example 4: Testing aenum IntEnum")

try:
    # IntEnum members are also integers
    class Status(aenum.IntEnum):
        PENDING = 1
        RUNNING = 2
        COMPLETE = 3
        ERROR = 4
    
    print("Created Status IntEnum")
    print(f"Status.RUNNING: {Status.RUNNING}")
    print(f"Status.RUNNING == 2: {Status.RUNNING == 2}")
    print(f"Status.COMPLETE > Status.PENDING: {Status.COMPLETE > Status.PENDING}")
    
    # Can use in arithmetic operations
    next_status = Status.PENDING + 1
    print(f"PENDING + 1 = {next_status} (equals RUNNING: {next_status == Status.RUNNING})")
    assert next_status == Status.RUNNING and Status.COMPLETE > Status.PENDING
    
    print("aenum example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aenum example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```
