## Example 1: affine functionality test 1

```python
from affine import Affine

# Example 1: Basic affine transformation matrix
print("Example 1: Testing basic affine transformation")

try:
    # affine provides 2D affine transformation matrices
    print("affine module imported successfully")
    
    # Create an identity transformation
    identity = Affine.identity()
    print(f"Identity matrix: {identity}")
    assert identity == Affine(1, 0, 0, 0, 1, 0), identity
    
    print("affine example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"affine example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: affine functionality test 2

```python
from affine import Affine

# Example 2: Affine translation transformation
print("Example 2: Testing affine translation")

try:
    # Create a translation transformation
    translate = Affine.translation(10, 20)
    print(f"Translation matrix (10, 20): {translate}")
    
    # Apply transformation to a point
    point = (1, 1)
    transformed_point = translate * point
    print(f"Point {point} translated to: {transformed_point}")
    assert transformed_point == (11, 21), transformed_point
    
    print("affine example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"affine example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: affine functionality test 3

```python
from affine import Affine

# Example 3: Affine scaling transformation
print("Example 3: Testing affine scaling")

try:
    # Create a scaling transformation
    scale = Affine.scale(2, 3)
    print(f"Scale matrix (2, 3): {scale}")
    
    # Apply scaling to a point
    point = (2, 4)
    scaled_point = scale * point
    print(f"Point {point} scaled to: {scaled_point}")
    assert scaled_point == (4, 12), scaled_point
    
    print("affine example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"affine example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: affine functionality test 4

```python
from affine import Affine
import math

# Example 4: Affine rotation transformation
print("Example 4: Testing affine rotation")

try:
    # Create a rotation transformation (45 degrees; Affine.rotation takes degrees)
    rotation = Affine.rotation(45)
    print(f"Rotation matrix (45°): {rotation}")
    
    # Apply rotation to a point
    point = (1, 0)
    rotated_point = rotation * point
    print(f"Point {point} rotated 45°: {rotated_point}")
    expected = (math.cos(math.radians(45)), math.sin(math.radians(45)))
    assert all(math.isclose(a, b, abs_tol=1e-12) for a, b in zip(rotated_point, expected)), rotated_point
    
    print("affine example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"affine example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: affine functionality test 5

```python
from affine import Affine

# Example 5: Affine transformation composition
print("Example 5: Testing affine transformation composition")

try:
    # Compose multiple transformations
    translate = Affine.translation(5, 5)
    scale = Affine.scale(2, 2)
    
    # Combine transformations (scale then translate)
    combined = translate * scale
    print(f"Combined transformation: {combined}")
    
    # Apply to a point
    point = (1, 1)
    result = combined * point
    print(f"Point {point} after scale+translate: {result}")
    
    # Check matrix properties
    print(f"Determinant: {combined.determinant}")
    assert result == (7, 7) and combined.determinant == 4, (result, combined.determinant)
    
    print("affine example 5 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"affine example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
