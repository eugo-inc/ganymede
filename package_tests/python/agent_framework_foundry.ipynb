# agent_framework_foundry

## Example 1: a Foundry chat client for a project endpoint

```python
# Example 1: FoundryChatClient keeps the model (deployment) it was given
from azure.core.credentials import AccessToken
from agent_framework_foundry import FoundryChatClient

class StaticToken:
    """A credential that hands out a fixed token: enough to build a client, never used to call Azure here."""
    def get_token(self, *scopes, **kwargs):
        return AccessToken("test-token", 4102444800)
    async def close(self):
        pass

client = FoundryChatClient(project_endpoint="https://example.services.ai.azure.com/api/projects/demo", model="gpt-4o-mini", credential=StaticToken())
assert client.model == "gpt-4o-mini", client.model
print("Example 1:", type(client).__name__, client.model)
```

## Example 2: a Foundry-hosted agent reference

```python
# Example 2: FoundryAgent points at an agent by name and version; nothing is fetched until it runs
from agent_framework_foundry import FoundryAgent

agent = FoundryAgent(project_endpoint="https://example.services.ai.azure.com/api/projects/demo", agent_name="support-bot",
                     agent_version="3", credential=StaticToken())
print("Example 2:", type(agent).__name__, agent.name if getattr(agent, "name", None) else "support-bot")
```

## Example 3: the embedding client

```python
# Example 3: FoundryEmbeddingClient for an embedding deployment
from agent_framework_foundry import FoundryEmbeddingClient

embedder = FoundryEmbeddingClient(project_endpoint="https://example.services.ai.azure.com/api/projects/demo", model="text-embedding-3-small",
                                  credential=StaticToken())
assert embedder.model == "text-embedding-3-small", embedder.model
print("Example 3:", type(embedder).__name__, embedder.model)
```

## Example 4: it builds on the OpenAI connector

```python
# Example 4: the Foundry chat client is an OpenAI-protocol client underneath (agent-framework-openai is its requirement)
import agent_framework_openai
from agent_framework_foundry import RawFoundryChatClient

bases = [klass.__module__.split(".")[0] for klass in RawFoundryChatClient.__mro__]
assert "agent_framework_openai" in bases, bases
print("Example 4: RawFoundryChatClient derives from", [b for b in dict.fromkeys(bases) if b.startswith("agent_framework")])
```

## Example 5: `agent_framework.foundry` re-exports this package

```python
# Example 5: the umbrella namespace hands back the same classes
import importlib.metadata
import agent_framework.foundry as via_core
import agent_framework_foundry as direct

assert via_core.FoundryChatClient is direct.FoundryChatClient and via_core.FoundryAgent is direct.FoundryAgent
print("Example 5:", importlib.metadata.version("agent-framework-foundry"), "re-exported through agent_framework.foundry")
```
