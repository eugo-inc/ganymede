## Example 1: aiodns functionality test 1

```python
import aiodns
import tracemalloc

tracemalloc.start()

# Example 1: Basic aiodns DNS resolver
print("Example 1: Testing basic aiodns DNS resolver")

def test_resolver():
    try:
        
        # Create DNS resolver
        resolver = aiodns.DNSResolver()
        print("DNS resolver created successfully")
        print(f"Resolver type: {type(resolver).__name__}")
        
        # Explicit nameservers are stored as given (no query is sent)
        configured = aiodns.DNSResolver(nameservers=['127.0.0.1'])
        assert configured.nameservers == ['127.0.0.1'], configured.nameservers
        
        print("Async DNS resolution capabilities available")
        print("aiodns example 1 completed successfully")
        
    except ImportError as e:
        print(f"Import error: {e}")
        raise
    except Exception as e:
        print(f"Runtime error: {e}")
        raise AssertionError(
            f"aiodns example (cell 1) failed: {type(e).__name__}: {e}"
        ) from e

# Run the async test
test_resolver()
tracemalloc.stop()
```

## Example 4: aiodns functionality test 4

```python
import aiodns


# Example 4: Error handling and DNS exceptions
print("Example 4: Error handling and DNS exceptions")

async def test_error_handling():
    try:
        print("Setting up DNS error handling examples...")
        
        # Common DNS errors and exceptions
        dns_errors = [
            'DNSError - Base DNS exception class',
            'ARES_ENOTFOUND - Domain name not found',
            'ARES_ETIMEOUT - DNS query timeout',
            'ARES_ECONNREFUSED - Connection refused by DNS server',
            'ARES_ENOMEM - Out of memory error',
            'ARES_EBADRESP - Invalid DNS response',
            'ARES_ENODATA - No data returned for query type'
        ]
        
        print(f"Common DNS errors:")
        for error in dns_errors:
            print(f"  • {error}")
        
        # Error handling patterns
        error_patterns = [
            "try:",
            "    resolver = aiodns.DNSResolver()",
            "    result = await resolver.query('example.com', 'A')",
            "    print(f'IP: {result[0].host}')",
            "except aiodns.error.DNSError as e:",
            "    error_code = e.args[0]",
            "    if error_code == aiodns.error.ARES_ENOTFOUND:",
            "        print('Domain not found')",
            "    elif error_code == aiodns.error.ARES_ETIMEOUT:",
            "        print('DNS query timeout')",
            "    else:",
            "        print(f'DNS error: {e}')",
            "except Exception as e:",
            "    print(f'Unexpected error: {e}')"
        ]
        
        print(f"\nError handling pattern:")
        for pattern in error_patterns:
            print(f"  {pattern}")
        
        # Retry and resilience strategies
        resilience_strategies = [
            "Implement exponential backoff for retries",
            "Use multiple DNS servers for redundancy",
            "Cache successful DNS responses locally",
            "Implement circuit breaker for failed servers",
            "Fall back to system DNS resolver if needed"
        ]
        
        print(f"\nResilience strategies:")
        for strategy in resilience_strategies:
            print(f"  • {strategy}")
        
        # DNS troubleshooting tips
        troubleshooting = [
            "Check DNS server connectivity",
            "Verify domain name spelling",
            "Test with different DNS servers",
            "Check network connectivity",
            "Validate query type for domain"
        ]
        
        print(f"\nTroubleshooting tips:")
        for tip in troubleshooting:
            print(f"  • {tip}")
        
        print("Example 4 completed successfully")
        
    except ImportError as e:
        print(f"Import error: {e}")
        raise
    except Exception as e:
        print(f"Runtime error: {e}")
        raise AssertionError(
            f"aiodns example (cell 7) failed: {type(e).__name__}: {e}"
        ) from e

# Run the async test
await test_error_handling()
print("Example 4 finished")
```
