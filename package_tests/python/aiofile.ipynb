## Example 1: aiofile functionality test 1

```python
import aiofile
import os
import tempfile

# Example 1: Basic aiofile async file operations
print("Example 1: Testing basic aiofile functionality")

async def test_basic_aiofile():
    try:
        print("aiofile module imported successfully")
        print("aiofile provides high-performance async file I/O")
        
        # Create temporary file
        temp_path = tempfile.mktemp()
        
        # Test async file writing (must use async with for aiofile)
        async with aiofile.async_open(temp_path, 'w') as f:
            await f.write("Hello, aiofile!")
        
        assert os.path.getsize(temp_path) == len("Hello, aiofile!"), os.path.getsize(temp_path)
        print(f"File written successfully: {temp_path}")
        print("Async file operations working")
        
        # Clean up
        if os.path.exists(temp_path):
            os.unlink(temp_path)
            
        print("aiofile example 1 completed successfully")
        
    except ImportError as e:
        print(f"Import error: {e}")
        raise
    except Exception as e:
        print(f"Runtime error: {e}")
        raise AssertionError(
            f"aiofile example (cell 1) failed: {type(e).__name__}: {e}"
        ) from e

# Run the async test
await test_basic_aiofile()
print("Example 1 finished")
```

## Example 2: aiofile functionality test 2

```python
import aiofile
import tempfile
import os

# Example 2: aiofile reading operations
print("Example 2: Testing aiofile reading operations")

async def test_file_reading():
    try:
        temp_path = tempfile.mktemp()
        
        # Write test content (must use async with and await)
        async with aiofile.async_open(temp_path, 'w') as f:
            await f.write("Line 1\nLine 2\nLine 3")
        
        # Read back content (must use async with and await)
        async with aiofile.async_open(temp_path, 'r') as f:
            content = await f.read()
            
        print(f"File content: {repr(content)}")
        assert content == 'Line 1\nLine 2\nLine 3', repr(content)
        print("Async file reading successful")
        
        # Clean up
        if os.path.exists(temp_path):
            os.unlink(temp_path)
            
        print("aiofile example 2 completed successfully")
        
    except ImportError as e:
        print(f"Import error: {e}")
        raise
    except Exception as e:
        print(f"Runtime error: {e}")
        raise AssertionError(
            f"aiofile example (cell 3) failed: {type(e).__name__}: {e}"
        ) from e

# Run the async test
await test_file_reading()
print("Example 2 finished")
```

## Example 3: aiofile functionality test 3

```python
import aiofile
import tempfile
import os

# Example 3: Advanced aiofile operations and line reading
print("Example 3: Advanced aiofile operations and line reading")

async def test_advanced_operations():
    try:
        print("Setting up advanced aiofile operations...")
        
        # Create test file with multiple lines
        temp_path = tempfile.mktemp()
        test_content = "First line\nSecond line\nThird line\nFourth line\n"
        
        # Write test content
        async with aiofile.async_open(temp_path, 'w') as f:
            await f.write(test_content)
        
        print(f"Test file created: {len(test_content)} characters")
        
        # Read line by line
        async with aiofile.async_open(temp_path, 'r') as f:
            lines = []
            async for line in f:
                lines.append(line.strip())
        
        print(f"Lines read: {len(lines)}")
        assert lines == ['First line', 'Second line', 'Third line', 'Fourth line'], lines
        for i, line in enumerate(lines, 1):
            print(f"  Line {i}: {line}")
        
        # File operations supported
        operations = [
            "async_open() - Open files asynchronously",
            "read() - Read entire file content",
            "write() - Write data to file", 
            "readline() - Read single line",
            "readlines() - Read all lines",
            "seek() - Change file position",
            "tell() - Get current position"
        ]
        
        print(f"\nSupported file operations:")
        for operation in operations:
            print(f"  • {operation}")
        
        # File modes
        file_modes = {
            'r': 'Read mode (text)',
            'w': 'Write mode (text)',
            'a': 'Append mode (text)',
            'rb': 'Read mode (binary)',
            'wb': 'Write mode (binary)',
            'ab': 'Append mode (binary)'
        }
        
        print(f"\nFile mode options:")
        for mode, description in file_modes.items():
            print(f"  '{mode}': {description}")
        
        # Clean up
        if os.path.exists(temp_path):
            os.unlink(temp_path)
        
        print("Example 3 completed successfully")
        
    except ImportError as e:
        print(f"Import error: {e}")
        raise
    except Exception as e:
        print(f"Runtime error: {e}")
        raise AssertionError(
            f"aiofile example (cell 5) failed: {type(e).__name__}: {e}"
        ) from e

# Run the async test
await test_advanced_operations()
print("Example 3 finished")
```

## Example 4: aiofile functionality test 4

```python
import aiofile
import tempfile
import os

# Example 4: Binary file operations and performance features
print("Example 4: Binary file operations and performance features")

async def test_binary_operations():
    try:
        print("Setting up binary file operations...")
        
        # Binary data example
        binary_data = b"Binary content: \x00\x01\x02\x03\xFF"
        temp_path = tempfile.mktemp()
        
        # Write binary data
        async with aiofile.async_open(temp_path, 'wb') as f:
            await f.write(binary_data)
        
        print(f"Binary data written: {len(binary_data)} bytes")
        
        # Read binary data back
        async with aiofile.async_open(temp_path, 'rb') as f:
            read_data = await f.read()
        
        print(f"Binary data read: {len(read_data)} bytes")
        print(f"Data matches: {binary_data == read_data}")
        assert read_data == binary_data, read_data
        
        # Performance advantages
        performance_benefits = [
            "Uses Linux aio_* system calls for true async I/O",
            "No thread pool overhead like aiofiles",
            "Direct kernel-level async operations",
            "Lower CPU usage for large file operations",
            "Better scalability for concurrent file access",
            "Reduced memory footprint"
        ]
        
        print(f"\nPerformance benefits:")
        for benefit in performance_benefits:
            print(f"  • {benefit}")
        
        # Use cases
        use_cases = [
            "High-throughput file processing",
            "Large file uploads/downloads",
            "Log file processing and analysis",
            "Media file handling",
            "Database file operations",
            "Backup and archival systems"
        ]
        
        print(f"\nOptimal use cases:")
        for use_case in use_cases:
            print(f"  • {use_case}")
        
        # Comparison with alternatives
        comparisons = {
            'aiofile': 'True async I/O via system calls',
            'aiofiles': 'Thread pool based (blocking I/O)',
            'asyncio.open': 'Not available in standard library',
            'standard open()': 'Synchronous blocking I/O'
        }
        
        print(f"\nComparison with alternatives:")
        for lib, description in comparisons.items():
            print(f"  {lib}: {description}")
        
        # Clean up
        if os.path.exists(temp_path):
            os.unlink(temp_path)
        
        print("Example 4 completed successfully")
        
    except ImportError as e:
        print(f"Import error: {e}")
        raise
    except Exception as e:
        print(f"Runtime error: {e}")
        raise AssertionError(
            f"aiofile example (cell 7) failed: {type(e).__name__}: {e}"
        ) from e

# Run the async test
await test_binary_operations()
print("Example 4 finished")
```
