## Example 1: aiofiles functionality test 1

```python
import aiofiles
import tempfile
import os

# Example 1: Basic aiofiles async file operations
print("Example 1: Testing basic aiofiles functionality")

try:
    # aiofiles provides async file I/O operations
    print("aiofiles module imported successfully")
    
    # Create a temporary file for testing
    with tempfile.NamedTemporaryFile(mode='w', delete=False) as tmp:
        tmp.write("Hello, aiofiles!")
        temp_path = tmp.name
    
    # Test async file reading
    async with aiofiles.open(temp_path, mode='r') as f:
        contents = await f.read()
        print(f"Read from file: {contents}")
    assert contents == "Hello, aiofiles!", contents
    
    # Clean up
    os.unlink(temp_path)
    
    print("aiofiles example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aiofiles example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished")
```

## Example 2: aiofiles functionality test 2

```python
import aiofiles
import tempfile
import os

# Example 2: aiofiles async file writing
print("Example 2: Testing aiofiles async file writing")

try:
    # Test async file writing
    temp_path = tempfile.mktemp()
    
    # Write to file asynchronously
    async with aiofiles.open(temp_path, mode='w') as f:
        await f.write("Async write operation")
        await f.write("\nSecond line")
    
    print(f"Successfully wrote to file: {temp_path}")
    
    # Verify by reading back
    async with aiofiles.open(temp_path, mode='r') as f:
        content = await f.read()
        print(f"File content: {repr(content)}")
    assert content == "Async write operation\nSecond line", content
    
    # Clean up
    os.unlink(temp_path)
    
    print("aiofiles example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aiofiles example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished")
```

## Example 3: aiofiles functionality test 3

```python
import aiofiles
import tempfile
import os

# Example 3: aiofiles line-by-line reading
print("Example 3: Testing aiofiles line-by-line reading")

try:
    # Create test file with multiple lines
    temp_path = tempfile.mktemp()
    
    # Write test data
    async with aiofiles.open(temp_path, mode='w') as f:
        await f.write("Line 1\nLine 2\nLine 3\n")
    
    print("Test file created with 3 lines")
    
    # Read line by line
    async with aiofiles.open(temp_path, mode='r') as f:
        line_count = 0
        async for line in f:
            line_count += 1
            print(f"Read line {line_count}: {line.strip()}")
    
    print(f"Total lines read: {line_count}")
    assert line_count == 3, line_count
    
    # Clean up
    os.unlink(temp_path)
    
    print("aiofiles example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aiofiles example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished")
```

## Example 4: aiofiles functionality test 4

```python
import aiofiles
import tempfile
import os

# Example 4: aiofiles binary file operations
print("Example 4: Testing aiofiles binary file operations")

try:
    # Test binary file operations
    temp_path = tempfile.mktemp()
    test_bytes = b"Binary data: \x00\x01\x02\x03"
    
    # Write binary data
    async with aiofiles.open(temp_path, mode='wb') as f:
        await f.write(test_bytes)
    
    print(f"Wrote {len(test_bytes)} bytes to file")
    
    # Read binary data back
    async with aiofiles.open(temp_path, mode='rb') as f:
        read_bytes = await f.read()
    
    print(f"Read {len(read_bytes)} bytes from file")
    print(f"Data match: {test_bytes == read_bytes}")
    assert read_bytes == test_bytes, read_bytes
    
    # Clean up
    os.unlink(temp_path)
    
    print("aiofiles example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aiofiles example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished")
```
