## Example 1: aioitertools functionality test 1

```python
import aioitertools

# Example 1: Basic aioitertools enumeration
print("Example 1: Testing aioitertools enumeration")

try:
    print("aioitertools module imported successfully")
    print("aioitertools provides async versions of itertools functions")
    
    # Test enumerate with async iteration
    items = [10, 20, 30]
    seen = []
    async for index, value in aioitertools.enumerate(items):
        print(f"Index {index}: {value}")
        seen.append((index, value))
        if index >= 1:  # Limit output for testing
            break
    
    assert seen == [(0, 10), (1, 20)], seen
    print("aioitertools example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aioitertools example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished")
```

## Example 2: aioitertools functionality test 2

```python
import aioitertools

# Example 2: aioitertools map
print("Example 2: Testing aioitertools map")

try:
    # function to double values
    def double(x):
        return x * 2
    
    # Test map with async iteration
    numbers = [1, 2, 3, 4]
    result = []
    async for doubled in aioitertools.map(double, numbers):
        result.append(doubled)
    
    print(f"Original: {numbers}")
    print(f"Doubled: {result}")
    assert result == [2, 4, 6, 8], result
    
    print("aioitertools example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aioitertools example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished")
```

## Example 3: aioitertools functionality test 3

```python
import aioitertools

# Example 3: aioitertools filterfalse
print("Example 3: Testing aioitertools filterfalse")

try:
    # predicate function
    def is_odd(x):
        return x % 2 == 1
    
    # Test filterfalse with async iteration (filters out odd numbers, keeping even)
    numbers = [1, 2, 3, 4, 5, 6, 7, 8]
    evens = []
    async for even in aioitertools.filterfalse(is_odd, numbers):
        evens.append(even)
    
    print(f"Original: {numbers}")
    print(f"Even numbers (filtered out odds): {evens}")
    assert evens == [2, 4, 6, 8], evens
    
    print("aioitertools example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aioitertools example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished")
```

## Example 4: aioitertools functionality test 4

```python
import aioitertools

# Example 4: aioitertools chain
print("Example 4: Testing aioitertools chain")

try:
    # Test chaining multiple iterables
    list1 = [1, 2, 3]
    list2 = [4, 5, 6]
    list3 = [7, 8, 9]
    
    chained = []
    async for item in aioitertools.chain(list1, list2, list3):
        chained.append(item)
    
    print(f"List 1: {list1}")
    print(f"List 2: {list2}")
    print(f"List 3: {list3}")
    print(f"Chained: {chained}")
    assert chained == list(range(1, 10)), chained
    
    print("aioitertools example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aioitertools example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished")
```
