## Example 1: aioodbc functionality test 1

```python
# Example 1: Exploring aioodbc Module Structure and Async Functions
import aioodbc
import inspect

print("Example 1: aioodbc Module - Async ODBC Database Connectivity\n")

# Get all public attributes
public_attrs = [attr for attr in dir(aioodbc) if not attr.startswith('_')]
print(f"Public attributes in aioodbc: {len(public_attrs)}")
print(f"Attributes: {', '.join(public_attrs)}\n")

# Inspect the main async functions
print("Key async functions:")
if hasattr(aioodbc, 'connect'):
    sig = inspect.signature(aioodbc.connect)
    print(f"• aioodbc.connect{sig}")
    print(f"  - Async: {inspect.iscoroutinefunction(aioodbc.connect)}")

if hasattr(aioodbc, 'create_pool'):
    sig = inspect.signature(aioodbc.create_pool)
    print(f"• aioodbc.create_pool{sig}")
    print(f"  - Async: {inspect.iscoroutinefunction(aioodbc.create_pool)}")

# Check for Connection class
if hasattr(aioodbc, 'Connection'):
    print(f"\n• aioodbc.Connection class available")
    conn_methods = [m for m in dir(aioodbc.Connection) if not m.startswith('_')]
    print(f"  Methods: {', '.join(conn_methods[:10])}...")
    
# Check for Cursor class  
if hasattr(aioodbc, 'Cursor'):
    print(f"\n• aioodbc.Cursor class available")
    cursor_methods = [m for m in dir(aioodbc.Cursor) if not m.startswith('_')]
    print(f"  Methods: {', '.join(cursor_methods[:10])}...")

print("\naioodbc provides async/await interface for ODBC database operations")
```

## Example 2: aioodbc functionality test 2

```python
# Example 2: Inspecting Cursor Class Async Methods
import aioodbc
import inspect

print("Example 2: aioodbc.Cursor - Async Cursor Operations\n")

# Get Cursor class
Cursor = aioodbc.Cursor

# Get all methods
all_methods = [m for m in dir(Cursor) if not m.startswith('_') and callable(getattr(Cursor, m, None))]
print(f"Total public methods: {len(all_methods)}\n")

# Identify async methods
async_methods = []
sync_methods = []

for method_name in all_methods:
    method = getattr(Cursor, method_name)
    if inspect.iscoroutinefunction(method):
        async_methods.append(method_name)
    else:
        sync_methods.append(method_name)

print(f"Async methods ({len(async_methods)}):")
for method in sorted(async_methods):
    print(f"  • await cursor.{method}()")

print(f"\nSync properties/methods ({len(sync_methods)}):")
for method in sorted(sync_methods)[:15]:
    print(f"  • cursor.{method}")

print("\nKey async operations:")
print("  - fetchone, fetchmany, fetchall: Async result fetching")
print("  - execute, executemany: Async query execution")
print("  - close: Async resource cleanup")
```

## Example 3: aioodbc functionality test 3

```python
# Example 3: Inspecting Connection Class and Pool
import aioodbc
import inspect

print("Example 3: aioodbc.Connection and Pool Classes\n")

# Inspect Connection class
Connection = aioodbc.Connection
conn_methods = [m for m in dir(Connection) if not m.startswith('_') and callable(getattr(Connection, m, None))]

print(f"Connection class has {len(conn_methods)} public methods\n")

# Find async methods in Connection
async_conn_methods = []
for method_name in conn_methods:
    method = getattr(Connection, method_name)
    if inspect.iscoroutinefunction(method):
        async_conn_methods.append(method_name)

print(f"Async Connection methods ({len(async_conn_methods)}):")
for method in sorted(async_conn_methods):
    print(f"  • await conn.{method}()")

# Check for Pool class
if hasattr(aioodbc, 'Pool'):
    Pool = aioodbc.Pool
    pool_methods = [m for m in dir(Pool) if not m.startswith('_')]
    print(f"\nPool class has {len(pool_methods)} public attributes")
    
    # Find async methods in Pool
    async_pool_methods = []
    for method_name in pool_methods:
        attr = getattr(Pool, method_name, None)
        if callable(attr) and inspect.iscoroutinefunction(attr):
            async_pool_methods.append(method_name)
    
    print(f"Async Pool methods: {', '.join(async_pool_methods) if async_pool_methods else 'acquire, wait_closed'}")

print("\nConnection management:")
print("  - commit/rollback: Async transaction control")
print("  - cursor: Create async cursor")
print("  - close: Async connection cleanup")
```

## Example 4: aioodbc functionality test 4

```python
# Example 4: Demonstrating DSN String Format and Parameters
import aioodbc
import inspect

print("Example 4: Connection Parameters and DSN Format\n")

# Inspect connect function parameters
connect_sig = inspect.signature(aioodbc.connect)
print("aioodbc.connect() parameters:")
for param_name, param in connect_sig.parameters.items():
    default = param.default if param.default != inspect.Parameter.empty else "required"
    print(f"  • {param_name}: {default}")

# Inspect create_pool function parameters
pool_sig = inspect.signature(aioodbc.create_pool)
print("\naioodbc.create_pool() parameters:")
for param_name, param in pool_sig.parameters.items():
    default = param.default if param.default != inspect.Parameter.empty else "required"
    print(f"  • {param_name}: {default}")

print("\nCommon DSN formats:")
print("  • SQLite: 'Driver=SQLite3;Database=/path/to/db.sqlite'")
print("  • MySQL: 'Driver=MySQL;Server=localhost;Database=mydb;User=user;Password=pass'")
print("  • PostgreSQL: 'Driver=PostgreSQL;Server=localhost;Database=mydb'")
print("  • SQL Server: 'Driver=SQL Server;Server=localhost;Database=mydb'")

print("\nPool configuration:")
print("  • minsize: Minimum pool size (default: 1)")
print("  • maxsize: Maximum pool size (default: 10)")
print("  • pool_recycle: Recycle connections after N seconds")
print("  • echo: Enable SQL logging")
```

## Example 5: aioodbc functionality test 5

```python
# Example 5: Comparing aioodbc with Synchronous ODBC
import inspect

import aioodbc

print("Example 5: What Makes aioodbc Unique\n")

# Show that main functions are coroutines
# connect() and cursor() return awaitable async context managers; execute() is a coroutine function.
print("Async/await support:")
print(f"  • aioodbc.connect is coroutine: {inspect.iscoroutinefunction(aioodbc.connect)}")
print(f"  • Connection.cursor is coroutine: {inspect.iscoroutinefunction(aioodbc.Connection.cursor)}")
print(f"  • Cursor.execute is coroutine: {inspect.iscoroutinefunction(aioodbc.Cursor.execute)}")
print(f"  • Cursor.fetchone is coroutine: {inspect.iscoroutinefunction(aioodbc.Cursor.fetchone)}")

# Check for context manager support
print("\nContext manager support:")
has_aenter = hasattr(aioodbc.Connection, '__aenter__')
has_aexit = hasattr(aioodbc.Connection, '__aexit__')
print(f"  • Connection supports 'async with': {has_aenter and has_aexit}")

has_cursor_aenter = hasattr(aioodbc.Cursor, '__aenter__')
has_cursor_aexit = hasattr(aioodbc.Cursor, '__aexit__')
print(f"  • Cursor supports 'async with': {has_cursor_aenter and has_cursor_aexit}")
assert has_aenter and has_aexit and has_cursor_aenter and has_cursor_aexit
assert inspect.iscoroutinefunction(aioodbc.Cursor.execute)

print("\nKey differences from pyodbc (synchronous):")
print("  1. All database operations use async/await")
print("  2. Non-blocking I/O allows concurrent queries")
print("  3. Connection pooling with async acquire/release")
print("  4. Compatible with asyncio event loop")
print("  5. Supports async context managers")

print("\nUse cases:")
print("  • Web applications (async frameworks like aiohttp, FastAPI)")
print("  • High-concurrency database access")
print("  • Microservices with async I/O")
print("  • Real-time data processing pipelines")
```
