## Example 3: aiorwlock functionality test 3

```python
import asyncio

import aiorwlock

# Example 3: aiorwlock multiple readers
print("Example 3: Testing aiorwlock multiple readers")

try:
    rwlock = aiorwlock.RWLock()
    print(f"Reader lock type: {type(rwlock.reader).__name__}")
    
    # Two tasks take the reader lock and each waits inside it until both are in.
    # If readers excluded each other, the second could not enter and the wait would time out.
    holders = peak = 0
    both_in = asyncio.Event()
    
    async def reader():
        global holders, peak
        async with rwlock.reader:
            holders += 1
            peak = max(peak, holders)
            if holders == 2:
                both_in.set()
            await asyncio.wait_for(both_in.wait(), timeout=5)
            holders -= 1
    
    await asyncio.gather(reader(), reader())
    print(f"Readers holding the lock at the same time: {peak}")
    assert peak == 2, peak
    
    print("aiorwlock example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aiorwlock example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished")
```

## Example 4: aiorwlock functionality test 4

```python
import asyncio
import time

import aiorwlock

# Example 4: aiorwlock reader-writer interaction
print("Example 4: Testing aiorwlock reader-writer interaction")

try:
    rwlock = aiorwlock.RWLock()
    print(f"Reader lock: {type(rwlock.reader).__name__}")
    print(f"Writer lock: {type(rwlock.writer).__name__}")
    
    # One writer and two readers contend for the lock; each records when it held it.
    spans = []
    
    async def hold(kind, lock):
        async with lock:
            start = time.monotonic()
            await asyncio.sleep(0.05)
            spans.append((kind, start, time.monotonic()))
    
    await asyncio.gather(hold("reader", rwlock.reader), hold("writer", rwlock.writer), hold("reader", rwlock.reader))
    writer_spans = [(s, e) for kind, s, e in spans if kind == "writer"]
    reader_spans = [(s, e) for kind, s, e in spans if kind == "reader"]
    assert len(writer_spans) == 1 and len(reader_spans) == 2, spans
    for ws, we in writer_spans:
        for rs, re_ in reader_spans:
            assert we <= rs or re_ <= ws, ("the writer overlapped a reader", spans)
    print("No reader held the lock while the writer did")
    
    print("aiorwlock example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aiorwlock example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished")
```
