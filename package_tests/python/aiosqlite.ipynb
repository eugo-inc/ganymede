## Example 1: aiosqlite functionality test 1

```python
import aiosqlite
import tempfile

# Example 1: Basic aiosqlite database connection
print("Example 1: Testing basic aiosqlite database connection")

try:
    print("aiosqlite module imported successfully")
    print("aiosqlite provides async SQLite database interface")
    
    # Create temporary database
    db_path = tempfile.mktemp(suffix='.db')
    
    # Test connection
    async with aiosqlite.connect(db_path) as db:
        print(f"Connected to SQLite database: {db_path}")
        async with db.execute('SELECT 1') as cursor:
            assert await cursor.fetchone() == (1,)
    
    # Clean up
    import os
    if os.path.exists(db_path):
        os.unlink(db_path)
        
    print("aiosqlite example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aiosqlite example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished")
```

## Example 2: aiosqlite functionality test 2

```python
import aiosqlite
import tempfile
import os

# Example 2: aiosqlite table creation
print("Example 2: Testing aiosqlite table creation")

try:
    # Create temporary database
    db_path = tempfile.mktemp(suffix='.db')
    
    async with aiosqlite.connect(db_path) as db:
        # Create a table
        await db.execute('''
            CREATE TABLE users (
                id INTEGER PRIMARY KEY,
                name TEXT NOT NULL,
                email TEXT
            )
        ''')
        await db.commit()
        
        async with db.execute("SELECT name FROM sqlite_master WHERE type = 'table'") as cursor:
            tables = [row[0] for row in await cursor.fetchall()]
        assert tables == ['users'], tables
        print("Table 'users' created")
    
    # Clean up
    if os.path.exists(db_path):
        os.unlink(db_path)
        
    print("aiosqlite example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aiosqlite example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished")
```

## Example 3: aiosqlite functionality test 3

```python
import aiosqlite
import tempfile
import os

# Example 3: aiosqlite data insertion
print("Example 3: Testing aiosqlite data insertion")

try:
    # Create temporary database
    db_path = tempfile.mktemp(suffix='.db')
    
    async with aiosqlite.connect(db_path) as db:
        # Create table
        await db.execute('CREATE TABLE products (id INTEGER PRIMARY KEY, name TEXT, price REAL)')
        
        # Insert data
        await db.execute('INSERT INTO products (name, price) VALUES (?, ?)', ('Laptop', 999.99))
        await db.execute('INSERT INTO products (name, price) VALUES (?, ?)', ('Mouse', 29.99))
        await db.commit()
        
        async with db.execute('SELECT name, price FROM products ORDER BY id') as cursor:
            rows = await cursor.fetchall()
        print(f"Products: {rows}")
        assert rows == [('Laptop', 999.99), ('Mouse', 29.99)], rows
    
    # Clean up
    if os.path.exists(db_path):
        os.unlink(db_path)
        
    print("aiosqlite example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aiosqlite example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished")
```

## Example 4: aiosqlite functionality test 4

```python
import aiosqlite
import tempfile
import os

# Example 4: aiosqlite data querying
print("Example 4: Testing aiosqlite data querying")

try:
    # Create temporary database with data
    db_path = tempfile.mktemp(suffix='.db')
    
    async with aiosqlite.connect(db_path) as db:
        # Setup data
        await db.execute('CREATE TABLE employees (id INTEGER PRIMARY KEY, name TEXT, department TEXT)')
        await db.execute('INSERT INTO employees (name, department) VALUES (?, ?)', ('Alice', 'Engineering'))
        await db.execute('INSERT INTO employees (name, department) VALUES (?, ?)', ('Bob', 'Marketing'))
        await db.execute('INSERT INTO employees (name, department) VALUES (?, ?)', ('Charlie', 'Engineering'))
        await db.commit()
        
        # Query data
        async with db.execute('SELECT name, department FROM employees WHERE department = ?', ('Engineering',)) as cursor:
            engineering_employees = await cursor.fetchall()
            
        print(f"Found {len(engineering_employees)} engineering employees:")
        for name, dept in engineering_employees:
            print(f"  {name} - {dept}")
        
        assert sorted(name for name, _ in engineering_employees) == ['Alice', 'Charlie'], engineering_employees
    
    # Clean up
    if os.path.exists(db_path):
        os.unlink(db_path)
        
    print("aiosqlite example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aiosqlite example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished")
```
