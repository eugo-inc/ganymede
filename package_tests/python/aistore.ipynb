## Example 1: aistore functionality test 1

```python
# Example 1: Exploring aistore Module Structure and Client
# aistore >= 1.5 removed the aistore.client shim module; Client is exported
# top-level (and lives at aistore.sdk.client).
import aistore
from aistore import Client
import inspect

print("Example 1: aistore Module - Distributed Object Storage\n")

# Get all public modules
public_attrs = [attr for attr in dir(aistore) if not attr.startswith('_')]
print(f"Public attributes in aistore: {len(public_attrs)}")
print(f"Attributes: {', '.join(public_attrs)}\n")

# Inspect Client class
print("Client class methods:")
client_methods = [m for m in dir(Client) if not m.startswith('_') and callable(getattr(Client, m))]
for method in sorted(client_methods)[:15]:
    print(f"  • {method}()")

# Inspect Client.__init__ signature
sig = inspect.signature(Client.__init__)
print(f"\nClient.__init__ signature:")
print(f"  {sig}")

# Check for main client methods
main_methods = ['bucket', 'cluster', 'job', 'etl']
print(f"\nMain client methods:")
for method in main_methods:
    if hasattr(Client, method):
        method_sig = inspect.signature(getattr(Client, method))
        print(f"  • {method}{method_sig}")

print("\naistore provides distributed object storage with Python SDK")
```

## Example 2: aistore functionality test 2

```python
# Example 2: Exploring Bucket Class and Operations
from aistore.sdk import Bucket
import inspect

print("Example 2: Bucket Class - Object Storage Container\n")

# Inspect Bucket class
bucket_methods = [m for m in dir(Bucket) if not m.startswith('_') and callable(getattr(Bucket, m))]
print(f"Bucket class has {len(bucket_methods)} methods\n")

# Group methods by category
object_methods = [m for m in bucket_methods if any(x in m.lower() for x in ['object', 'put', 'get', 'delete'])]
list_methods = [m for m in bucket_methods if 'list' in m.lower()]
bucket_ops = [m for m in bucket_methods if any(x in m.lower() for x in ['create', 'destroy', 'head', 'copy'])]

print(f"Object operations ({len(object_methods)}):")
for method in sorted(object_methods)[:12]:
    print(f"  • {method}()")

print(f"\nList operations ({len(list_methods)}):")
for method in sorted(list_methods):
    print(f"  • {method}()")

print(f"\nBucket operations ({len(bucket_ops)}):")
for method in sorted(bucket_ops):
    print(f"  • {method}()")

# Inspect key method signatures
key_methods = ['object', 'list_objects', 'create']
print(f"\nKey method signatures:")
for method_name in key_methods:
    if hasattr(Bucket, method_name):
        sig = inspect.signature(getattr(Bucket, method_name))
        print(f"  • {method_name}{sig}")

print("\nBucket provides container abstraction for objects")
```

## Example 3: aistore functionality test 3

```python
# Example 3: Exploring Object and Job Classes
from aistore.sdk import Object, Job
import inspect

print("Example 3: Object and Job Classes\n")

# Inspect Object class
print("Object class - Individual storage items:")
object_methods = [m for m in dir(Object) if not m.startswith('_') and callable(getattr(Object, m))]
print(f"Total methods: {len(object_methods)}\n")

# Categorize object methods
read_methods = [m for m in object_methods if any(x in m.lower() for x in ['get', 'read', 'head'])]
write_methods = [m for m in object_methods if any(x in m.lower() for x in ['put', 'write', 'append'])]
manage_methods = [m for m in object_methods if any(x in m.lower() for x in ['delete', 'rename', 'set'])]

print(f"Read operations ({len(read_methods)}):")
for method in sorted(read_methods):
    print(f"  • {method}()")

print(f"\nWrite operations ({len(write_methods)}):")
for method in sorted(write_methods):
    print(f"  • {method}()")

print(f"\nManagement operations ({len(manage_methods)}):")
for method in sorted(manage_methods)[:8]:
    print(f"  • {method}()")

# Inspect Job class for batch operations
print(f"\n\nJob class - Batch processing operations:")
job_methods = [m for m in dir(Job) if not m.startswith('_') and callable(getattr(Job, m))]
for method in sorted(job_methods)[:10]:
    print(f"  • {method}()")

# Check Job status and wait methods
status_methods = ['status', 'wait', 'wait_for_idle']
print(f"\nJob status methods:")
for method in status_methods:
    if hasattr(Job, method):
        print(f"  • {method}() - Available")

print("\nObject and Job classes enable data operations and batch processing")
```

## Example 4: aistore functionality test 4

```python
# Example 4: Exploring Cluster and ETL Classes
import importlib.util
import inspect

from aistore.sdk import Cluster
from aistore.sdk.etl.etl import Etl

print("Example 4: Cluster and ETL Classes\n")

print("Cluster class - Cluster management:")
cluster_methods = [m for m in dir(Cluster) if not m.startswith('_') and callable(getattr(Cluster, m))]
for method in sorted(cluster_methods)[:15]:
    print(f"  • {method}()")

# Inspect key cluster methods
key_methods = ['get_info', 'list_buckets']
print(f"\nKey cluster methods:")
for method in key_methods:
    sig = inspect.signature(getattr(Cluster, method))
    print(f"  • {method}{sig}")
assert set(key_methods) <= set(cluster_methods), cluster_methods

# The ETL class is `Etl` in aistore.sdk.etl.etl (also exported as aistore.sdk.Etl)
print(f"\n\nETL class - Data transformation:")
etl_methods = [m for m in dir(Etl) if not m.startswith('_') and callable(getattr(Etl, m))]
for method in sorted(etl_methods)[:12]:
    print(f"  • {method}()")

# ETL lifecycle operations
etl_ops = ['init', 'init_class', 'start', 'stop', 'view', 'delete']
print(f"\nETL operations: {etl_ops}")
assert set(etl_ops) <= set(etl_methods), etl_methods

# The PyTorch integration (aistore.pytorch) needs optional packages such as torch and
# alive_progress. Skip it only when one of those is not installed; any other error fails.
OPTIONAL_DEPS = {"torch", "alive_progress", "webdataset"}
try:
    from aistore.pytorch import AISMapDataset, AISIterDataset
except ModuleNotFoundError as e:
    missing = (e.name or "").split(".")[0]
    if missing not in OPTIONAL_DEPS or importlib.util.find_spec(missing) is not None:
        raise
    print(f"\nSKIP: optional dependency `{missing}` is not installed -> `aistore.pytorch` not exercised ({e})")
else:
    import torch.utils.data
    assert issubclass(AISMapDataset, torch.utils.data.Dataset), AISMapDataset.__mro__
    assert issubclass(AISIterDataset, torch.utils.data.IterableDataset), AISIterDataset.__mro__
    for dataset_cls in (AISMapDataset, AISIterDataset):
        dataset_methods = sorted(m for m in dir(dataset_cls) if not m.startswith('_') and callable(getattr(dataset_cls, m)))
        print(f"\n\nPyTorch integration: {dataset_cls.__name__}{inspect.signature(dataset_cls.__init__)}")
        print(f"  public methods: {dataset_methods}")
    assert callable(getattr(AISMapDataset, "get_obj_list", None)), "AISMapDataset has no get_obj_list()"
    assert callable(getattr(AISIterDataset, "__iter__", None)), "AISIterDataset is not iterable"

print("\nCluster and ETL enable distributed operations and transformations")
```

## Example 5: aistore functionality test 5

```python
# Example 5: Exploring SDK Submodules and Data Types
import aistore
from aistore import sdk
from aistore.sdk.namespace import Namespace
from aistore.sdk.provider import Provider
from aistore.sdk.types import BucketModel
import inspect

print("Example 5: SDK Submodules and Data Types\n")

# Explore sdk submodule
sdk_attrs = [attr for attr in dir(sdk) if not attr.startswith('_')]
print(f"SDK module attributes ({len(sdk_attrs)}):")
classes = [attr for attr in sdk_attrs if inspect.isclass(getattr(sdk, attr))]
print(f"Classes: {', '.join(sorted(classes)[:15])}")

# Provider types - storage backends (Provider lives in aistore.sdk.provider)
print(f"\n\nProvider types - Storage backends:")
for name, value in Provider.__members__.items():
    print(f"  • {name}: {value}")
assert {'AIS', 'AMAZON', 'GOOGLE'} <= set(Provider.__members__), list(Provider.__members__)

# A bucket is described by aistore.sdk.types.BucketModel
print(f"\n\nBucketModel - Bucket identity:")
for field in BucketModel.model_fields:
    print(f"  • {field}")
assert {'name', 'provider'} <= set(BucketModel.model_fields), list(BucketModel.model_fields)

# Namespace types
print(f"\n\nNamespace - Multi-cloud addressing:")
ns_attrs = [attr for attr in dir(Namespace) if not attr.startswith('_')]
for attr in sorted(ns_attrs)[:10]:
    print(f"  • {attr}")
assert Namespace(uuid='u', name='n').get_path() == '@u#n'

# Summary of main SDK components
print(f"\n\nMain SDK components:")
main_components = ['Client', 'Bucket', 'Object', 'Job', 'Cluster']
for component in main_components:
    has_component = hasattr(sdk, component)
    print(f"  • {component}: {'Available' if has_component else 'Not found'}")
assert all(hasattr(sdk, c) for c in main_components), [c for c in main_components if not hasattr(sdk, c)]

print("\naistore SDK provides comprehensive distributed storage API")
```
