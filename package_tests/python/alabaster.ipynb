## Example 1: alabaster functionality test 1

```python
# Example 1: Using alabaster.get_path() - Get theme directory path
import alabaster
import os

print("Example 1: alabaster.get_path() - Theme Directory Path\n")

# get_path() returns the path to the alabaster theme directory
# Note: It returns site-packages, need to get actual alabaster path
theme_path_base = alabaster.get_path()
print(f"Base path from get_path(): {theme_path_base}")

# Get actual alabaster package path
alabaster_path = os.path.dirname(alabaster.__file__)
print(f"Alabaster package path: {alabaster_path}")
print(f"Path exists: {os.path.exists(alabaster_path)}")

# List contents of the alabaster directory
if os.path.exists(alabaster_path):
    contents = os.listdir(alabaster_path)
    print(f"\nAlabaster directory contains {len(contents)} items:")
    
    # Separate by file type
    html_files = [f for f in contents if f.endswith('.html')]
    css_files = [f for f in contents if f.endswith('.css')]
    config_files = [f for f in contents if f.endswith('.conf')]
    py_files = [f for f in contents if f.endswith('.py')]
    
    print(f"  HTML templates: {len(html_files)} - {html_files}")
    print(f"  CSS files: {len(css_files)} - {css_files}")
    print(f"  Config files: {len(config_files)} - {config_files}")
    print(f"  Python files: {len(py_files)} - {py_files}")

print("\nget_path() provides base path for Sphinx to locate the theme")
```

## Example 2: alabaster functionality test 2

```python
# Example 2: Reading alabaster theme.conf configuration
import alabaster
import os

print("Example 2: Reading Alabaster Theme Configuration\n")

# Get actual alabaster package path
alabaster_path = os.path.dirname(alabaster.__file__)
theme_conf_path = os.path.join(alabaster_path, 'theme.conf')

print(f"Theme config path: {theme_conf_path}")
print(f"Config exists: {os.path.exists(theme_conf_path)}")

# Read and display theme.conf contents
if os.path.exists(theme_conf_path):
    print("\nTheme configuration contents:")
    with open(theme_conf_path, 'r') as f:
        config_content = f.read()
        print(config_content)
    
    # Parse configuration sections
    print("\nConfiguration analysis:")
    lines = config_content.split('\n')
    sections = [line.strip() for line in lines if line.strip().startswith('[')]
    options = [line.split('=')[0].strip() for line in lines if '=' in line and not line.strip().startswith('[')]
    
    print(f"  Sections found: {sections}")
    assert '[theme]' in sections, sections
    print(f"  Options defined: {len(options)}")
    if options:
        print(f"  Sample options: {options[:5]}")
else:
    print("\ntheme.conf not found at expected location")
    print("Checking for alabaster subdirectory...")
    alabaster_subdir = os.path.join(alabaster_path, 'alabaster')
    if os.path.exists(alabaster_subdir):
        theme_conf_alt = os.path.join(alabaster_subdir, 'theme.conf')
        if os.path.exists(theme_conf_alt):
            print(f"Found at: {theme_conf_alt}")

print("\ntheme.conf defines alabaster's inheritance and defaults")
```

## Example 3: alabaster functionality test 3

```python
# Example 3: Using alabaster.setup() - Sphinx extension setup
import alabaster
import inspect

print("Example 3: alabaster.setup() - Sphinx Extension Registration\n")

# Inspect the setup function
print("setup() function signature:")
sig = inspect.signature(alabaster.setup)
print(f"  {sig}")

# Get the setup function's docstring
if alabaster.setup.__doc__:
    print(f"\nDocstring: {alabaster.setup.__doc__[:200]}")

# Analyze setup function's code
print("\nsetup() function details:")
print(f"  Function name: {alabaster.setup.__name__}")
print(f"  Module: {alabaster.setup.__module__}")
print(f"  Callable: {callable(alabaster.setup)}")

# Get function source code
try:
    source_lines = inspect.getsource(alabaster.setup)
    print(f"\nFunction source code ({len(source_lines)} chars):")
    # Show first few lines
    lines = source_lines.split('\n')
    for line in lines[:10]:
        print(f"    {line}")
    if len(lines) > 10:
        print(f"    ... ({len(lines) - 10} more lines)")
except:
    print("\nSource code not available")
    raise

print("\nsetup() is called by Sphinx to register the alabaster theme")
```

```python
# Example 4: Using alabaster.update_context() - HTML context updates
import alabaster
import inspect

print("Example 4: alabaster.update_context() - HTML Context Updates\n")

# Inspect the update_context function
print("update_context() function signature:")
sig = inspect.signature(alabaster.update_context)
print(f"  {sig}")

# Get parameters
params = list(sig.parameters.keys())
print(f"\nFunction parameters: {params}")

# Get the function's docstring
if alabaster.update_context.__doc__:
    print(f"\nDocstring:")
    print(f"  {alabaster.update_context.__doc__}")

# Analyze function details
print("\nFunction details:")
print(f"  Function name: {alabaster.update_context.__name__}")
print(f"  Module: {alabaster.update_context.__module__}")
print(f"  Callable: {callable(alabaster.update_context)}")

# Get function source code
try:
    source_lines = inspect.getsource(alabaster.update_context)
    print(f"\nFunction source code:")
    lines = source_lines.split('\n')
    for i, line in enumerate(lines[:15]):
        print(f"    {line}")
    if len(lines) > 15:
        print(f"    ... ({len(lines) - 15} more lines)")
except Exception as e:
    print(f"\nSource code not available: {e}")
    raise

print("\nupdate_context() customizes HTML template rendering context")
```

## Example 5: alabaster functionality test 5

```python
# Example 5: Exploring alabaster theme templates
import alabaster
import os

print("Example 5: Alabaster HTML Templates and Static Files\n")

# Get alabaster package path
alabaster_path = os.path.dirname(alabaster.__file__)

# List all HTML templates
if os.path.exists(alabaster_path):
    all_files = os.listdir(alabaster_path)
    html_templates = [f for f in all_files if f.endswith('.html')]
    print(f"HTML templates ({len(html_templates)}):")
    for template in sorted(html_templates):
        template_path = os.path.join(alabaster_path, template)
        size = os.path.getsize(template_path)
        print(f"  • {template} ({size} bytes)")

# Check for static directory
static_dir = os.path.join(alabaster_path, 'static')
if os.path.exists(static_dir):
    print(f"\nStatic files directory: {static_dir}")
    static_files = os.listdir(static_dir)
    print(f"Static files ({len(static_files)}):")
    for sfile in sorted(static_files)[:10]:
        sfile_path = os.path.join(static_dir, sfile)
        if os.path.isfile(sfile_path):
            size = os.path.getsize(sfile_path)
            print(f"  • {sfile} ({size} bytes)")
        else:
            print(f"  • {sfile} (directory)")

# Read a sample template if available
if html_templates:
    sample_template = html_templates[0]
    template_path = os.path.join(alabaster_path, sample_template)
    print(f"\n{sample_template} preview (first 300 chars):")
    with open(template_path, 'r') as f:
        content = f.read(300)
        print(f"  {content}...")

print("\nAlabaster templates define the documentation page structure")
```
