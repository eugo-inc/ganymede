## Example 2: alembic functionality test 2

```python
from alembic.migration import MigrationContext
from sqlalchemy import create_engine, MetaData

# Example 2: Testing alembic migration context
print("Example 2: Testing alembic migration context")

try:
    # Create an in-memory SQLite database for testing
    engine = create_engine("sqlite:///:memory:")
    
    # Create migration context
    with engine.connect() as conn:
        context = MigrationContext.configure(conn)
        print("Migration context created successfully")
        print(f"Context type: {type(context).__name__}")
        # A fresh database has no alembic_version row yet
        assert context.get_current_revision() is None
        print("Current revision: None (nothing stamped yet)")
    
    print("alembic example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"alembic example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: alembic functionality test 3

```python
from alembic.operations import Operations
from alembic.migration import MigrationContext
import sqlalchemy as sa
from sqlalchemy import create_engine

# Example 3: Testing alembic operations
print("Example 3: Testing alembic database operations")

try:
    # Create engine and migration context
    engine = create_engine("sqlite:///:memory:")
    
    with engine.connect() as conn:
        context = MigrationContext.configure(conn)
        ops = Operations(context)
        
        print("Alembic operations object created")
        print(f"Operations type: {type(ops).__name__}")
        
        # Check available operation methods
        op_methods = [attr for attr in dir(ops) if not attr.startswith('_') and callable(getattr(ops, attr))]
        print(f"Available operations: {len(op_methods)}")
        
        # Run two schema operations and inspect the result
        ops.create_table('items', sa.Column('id', sa.Integer, primary_key=True))
        ops.add_column('items', sa.Column('name', sa.String(50)))
        columns = [col['name'] for col in sa.inspect(conn).get_columns('items')]
        print(f"Columns after create_table + add_column: {columns}")
        assert columns == ['id', 'name'], columns
    
    print("alembic example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"alembic example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```
