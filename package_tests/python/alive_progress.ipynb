## Example 1: A progress bar over a known total (receipt captured)

```python
import io
from alive_progress import alive_bar

out = io.StringIO()
with alive_bar(50, force_tty=False, file=out) as bar:
    for _ in range(50):
        bar()
receipt = out.getvalue()
assert "50/50" in receipt, receipt
print(receipt.strip())
```

## Example 2: Wrap an iterable with alive_it

```python
import io
from alive_progress import alive_it

out = io.StringIO()
squares = [x * x for x in alive_it(range(20), force_tty=False, file=out)]
assert squares[-1] == 361 and len(squares) == 20
assert "20/20" in out.getvalue(), out.getvalue()
print(out.getvalue().strip())
```

## Example 3: Manual mode: set the fraction done

```python
import io
from alive_progress import alive_bar

out = io.StringIO()
with alive_bar(manual=True, force_tty=False, file=out) as bar:
    for fraction in (0.25, 0.5, 1.0):
        bar(fraction)
assert "100%" in out.getvalue(), out.getvalue()
print(out.getvalue().strip())
```

## Example 4: Unknown total, title and text

```python
import io
from alive_progress import alive_bar

out = io.StringIO()
with alive_bar(title="items", force_tty=False, file=out) as bar:
    for i in range(7):
        bar.text(f"item {i}")
        bar()
assert "items" in out.getvalue() and "7" in out.getvalue(), out.getvalue()
print(out.getvalue().strip())
```

## Example 5: Global configuration

```python
import io
from alive_progress import alive_bar, config_handler

config_handler.set_global(length=10, force_tty=False)
try:
    out = io.StringIO()
    with alive_bar(3, file=out) as bar:
        for _ in range(3):
            bar()
    assert "3/3" in out.getvalue(), out.getvalue()
    print(out.getvalue().strip())
finally:
    config_handler.reset()
```
