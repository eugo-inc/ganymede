```python
from alpaca.trading.client import TradingClient
from alpaca.data.historical import StockHistoricalDataClient
from alpaca.common.enums import BaseURL
import alpaca

# Example 1: Basic Alpaca trading client setup
print("Example 1: Testing basic Alpaca trading client")

try:
    print("alpaca module imported successfully")
    print("Alpaca provides trading and market data APIs")
    
    # Without credentials both clients refuse to construct. alpaca-py takes keys only as arguments
    # (it reads no environment variables), so this holds on every host.
    for client_class in (TradingClient, StockHistoricalDataClient):
        try:
            client_class()
        except ValueError as e:
            assert "You must supply a method of authentication" in str(e), e
            print(f"{client_class.__name__} class available (requires API keys to connect)")
        else:
            raise AssertionError(f"{client_class.__name__}() constructed without any credentials")
    
    # With (placeholder) keys both construct offline: no request is sent until an API call is made.
    client = TradingClient(api_key="PKDEMOKEY", secret_key="demo-secret", paper=True)
    assert client._base_url == BaseURL.TRADING_PAPER, client._base_url
    print("Trading client structure created")
    
    data_client = StockHistoricalDataClient(api_key="PKDEMOKEY", secret_key="demo-secret")
    assert data_client._base_url == BaseURL.DATA, data_client._base_url
    print("Stock data client structure created")
        
    print("Alpaca client classes accessible")
    print("alpaca example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"alpaca_py example (cell 0) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

```python
from alpaca.data.historical import StockHistoricalDataClient
from alpaca.data.requests import StockBarsRequest
from alpaca.data.timeframe import TimeFrame
from datetime import datetime, timedelta

# Example 2: Market Data Client and Historical Data Requests
print("Example 2: Market Data Operations")

try:
    # Initialize market data client (no auth needed for some endpoints)
    data_client = StockHistoricalDataClient(
        api_key="demo_key",
        secret_key="demo_secret"
    )
    
    print("✓ Market data client initialized")
    
    # Create a sample historical data request
    request_params = StockBarsRequest(
        symbol_or_symbols=["AAPL", "MSFT", "GOOGL"],
        timeframe=TimeFrame.Day,
        start=datetime.now() - timedelta(days=7),
        end=datetime.now()
    )
    
    print("✓ Historical data request configured")
    print(f"Symbols: {request_params.symbol_or_symbols}")
    print(f"Timeframe: {request_params.timeframe}")
    print(f"Date range: {request_params.start.date()} to {request_params.end.date()}")
    
    # Show available timeframes
    available_timeframes = [
        TimeFrame.Minute, TimeFrame.Hour, TimeFrame.Day, TimeFrame.Week, TimeFrame.Month
    ]
    print(f"Available timeframes: {[tf.value for tf in available_timeframes]}")
    
except Exception as e:
    print(f"Note: API call would require valid credentials: {e}")
    print("✓ Market data request structure demonstrates proper usage")
    raise AssertionError(
        f"alpaca_py example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed\n")
```

```python
from alpaca.trading.requests import MarketOrderRequest, LimitOrderRequest
from alpaca.trading.enums import OrderSide, TimeInForce
from decimal import Decimal

# Example 3: Order Creation and Validation
print("Example 3: Trading Order Management")

try:
    # Create a market order request
    market_order = MarketOrderRequest(
        symbol="AAPL",
        qty=10,
        side=OrderSide.BUY,
        time_in_force=TimeInForce.DAY
    )
    
    print("✓ Market order request created")
    print(f"Symbol: {market_order.symbol}")
    print(f"Quantity: {market_order.qty}")
    print(f"Side: {market_order.side.value}")
    print(f"Time in force: {market_order.time_in_force.value}")
    
    # Create a limit order request
    limit_order = LimitOrderRequest(
        symbol="MSFT",
        qty=5,
        side=OrderSide.SELL,
        time_in_force=TimeInForce.GTC,
        limit_price=Decimal("300.50")
    )
    
    print("\n✓ Limit order request created")
    print(f"Symbol: {limit_order.symbol}")
    print(f"Quantity: {limit_order.qty}")
    print(f"Side: {limit_order.side.value}")
    print(f"Limit price: ${limit_order.limit_price}")
    print(f"Time in force: {limit_order.time_in_force.value}")
    
    # Validate order parameters
    print("\n✓ Order validation successful")
    print("Both orders ready for submission to Alpaca API")
    
except Exception as e:
    print(f"Error in order creation: {e}")
    raise AssertionError(
        f"alpaca_py example (cell 2) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 completed\n")
```

```python
from decimal import Decimal

# Example 4: Formatting account and position figures
# The figures below are sample values, not data from the Alpaca API; this cell only
# shows how such values are formatted and summed (no alpaca call is made).
print("Example 4: Formatting account and position figures (sample data)")

try:
    # Sample account figures
    mock_account_data = {
        'equity': Decimal('100000.00'),
        'cash': Decimal('25000.00'),
        'buying_power': Decimal('50000.00'),
        'portfolio_value': Decimal('100000.00'),
        'day_trade_count': 0
    }
    
    print("Sample account:")
    print(f"Portfolio Value: ${mock_account_data['portfolio_value']:,}")
    print(f"Available Cash: ${mock_account_data['cash']:,}")
    print(f"Buying Power: ${mock_account_data['buying_power']:,}")
    print(f"Day Trade Count: {mock_account_data['day_trade_count']}")
    
    # Sample positions
    mock_positions = [
        {
            'symbol': 'AAPL',
            'qty': Decimal('100'),
            'market_value': Decimal('15000.00'),
            'cost_basis': Decimal('14500.00'),
            'unrealized_pl': Decimal('500.00')
        },
        {
            'symbol': 'MSFT',
            'qty': Decimal('50'),
            'market_value': Decimal('16000.00'),
            'cost_basis': Decimal('15800.00'),
            'unrealized_pl': Decimal('200.00')
        }
    ]
    
    print("\nSample positions:")
    total_unrealized_pl = Decimal('0')
    for pos in mock_positions:
        print(f"{pos['symbol']}: {pos['qty']} shares, P&L: ${pos['unrealized_pl']}")
        total_unrealized_pl += pos['unrealized_pl']
    
    print(f"\nTotal Unrealized P&L: ${total_unrealized_pl}")
    assert total_unrealized_pl == Decimal('700.00'), total_unrealized_pl
    
except Exception as e:
    print(f"Error in portfolio management: {e}")
    raise AssertionError(
        f"alpaca_py example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed\n")
```

```python
# Example 5: Real-time Data Streaming Configuration
print("Example 5: Real-time Data Streaming")

try:
    # Import streaming classes
    from alpaca.data.live import StockDataStream
    
    # Configure data stream client (demonstration without actual connection)
    print("✓ StockDataStream class available")
    print("Supports real-time quotes, trades, and bars")
    print("Feeds available: IEX, SIP, OTC")
    
    # Define sample handler functions (demonstration)
    async def quote_handler(data):
        print(f"Quote: {data.symbol} - Bid: ${data.bid_price}, Ask: ${data.ask_price}")
    
    async def trade_handler(data):
        print(f"Trade: {data.symbol} - Price: ${data.price}, Size: {data.size}")
    
    async def bar_handler(data):
        print(f"Bar: {data.symbol} - OHLC: ${data.open}/${data.high}/${data.low}/${data.close}")
    
    print("\n✓ Event handlers defined")
    print("Handlers configured for: quotes, trades, minute bars")
    
    # Show subscription setup (would normally use stream_client.subscribe_*)
    symbols = ["AAPL", "MSFT", "GOOGL"]
    print(f"\n✓ Ready to subscribe to symbols: {symbols}")
    # Constructing the client and registering handlers is offline: nothing connects until stream.run().
    stream = StockDataStream("demo-key", "demo-secret")
    stream.subscribe_quotes(quote_handler, *symbols)
    stream.subscribe_trades(trade_handler, *symbols)
    stream.subscribe_bars(bar_handler, *symbols)
    print("Stream client ready for real-time data processing")
    
    # Streaming workflow
    print("\nStreaming workflow:")
    print("1. Initialize StockDataStream with API credentials")
    print("2. Subscribe to symbols for quotes/trades/bars")
    print("3. Register event handlers")
    print("4. Run stream in async context")
    
    print("✓ Streaming configuration complete")
    
except Exception as e:
    print(f"Configuration note: {e}")
    print("✓ Real-time streaming requires valid API credentials")
    raise AssertionError(
        f"alpaca_py example (cell 4) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 completed")
```
