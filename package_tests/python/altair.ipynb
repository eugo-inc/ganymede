## Example 1: altair functionality test 1

```python
import altair as alt
import pandas as pd

# Example 1: Basic Altair chart creation
print("Example 1: Testing basic Altair chart creation")

try:
    print("altair module imported successfully")
    print("Altair provides statistical visualization grammar")
    
    # Create simple dataset
    data = pd.DataFrame({
        'x': [1, 2, 3, 4, 5],
        'y': [2, 4, 6, 8, 10]
    })
    
    # Create basic chart
    chart = alt.Chart(data).mark_circle().encode(
        x='x',
        y='y'
    )
    
    print("Chart created successfully")
    print(f"Chart type: {type(chart).__name__}")
    spec = chart.to_dict()
    assert spec['mark']['type'] == 'circle', spec['mark']
    assert spec['encoding']['x']['field'] == 'x' and spec['encoding']['y']['field'] == 'y', spec['encoding']
    
    print("altair example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"altair example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: altair functionality test 2

```python
import altair as alt
import pandas as pd

# Example 2: Advanced chart types and marks
print("Example 2: Testing advanced Altair chart types")

try:
    # Create more complex dataset
    data = pd.DataFrame({
        'category': ['A', 'B', 'C', 'D', 'E'],
        'value': [23, 45, 56, 78, 32],
        'group': ['X', 'Y', 'X', 'Y', 'X']
    })
    
    # Bar chart
    bar_chart = alt.Chart(data).mark_bar().encode(
        x='category',
        y='value',
        color='group'
    )
    print("Bar chart created with color encoding")
    
    # Line chart with points
    line_data = pd.DataFrame({
        'x': range(10),
        'y': [i**2 for i in range(10)]
    })
    
    line_chart = alt.Chart(line_data).mark_line(point=True).encode(
        x='x',
        y='y'
    )
    print("Line chart with points created")
    
    # Area chart
    area_chart = alt.Chart(line_data).mark_area(opacity=0.7).encode(
        x='x',
        y='y'
    )
    print("Area chart created")
    
    marks = [c.to_dict()['mark']['type'] for c in (bar_chart, line_chart, area_chart)]
    assert marks == ['bar', 'line', 'area'], marks
    print("Multiple chart types demonstrated")
    print("altair example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"altair example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: altair functionality test 3

```python
import altair as alt
import pandas as pd
import numpy as np

# Example 3: Statistical transformations and aggregations
print("Example 3: Testing Altair statistical transformations")

try:
    # Create dataset for statistical analysis
    np.random.seed(42)
    data = pd.DataFrame({
        'x': np.random.normal(0, 1, 100),
        'y': np.random.normal(0, 1, 100),
        'category': np.random.choice(['A', 'B', 'C'], 100)
    })
    
    # Scatter plot with regression line
    scatter_plot = alt.Chart(data).mark_circle().encode(
        x='x',
        y='y',
        color='category'
    )
    print("Scatter plot created with categorical coloring")
    
    # Histogram with statistical binning
    histogram = alt.Chart(data).mark_bar().encode(
        alt.X('x:Q', bin=alt.Bin(maxbins=20)),
        y='count()',
        color='category'
    )
    print("Histogram created with automatic binning")
    
    # Box plot for statistical distribution
    box_plot = alt.Chart(data).mark_boxplot().encode(
        x='category',
        y='x'
    )
    print("Box plot created for distribution analysis")
    
    # Statistical summary
    mean_chart = alt.Chart(data).mark_rule(color='red').encode(
        y='mean(x):Q'
    )
    print("Statistical rule (mean line) created")
    
    assert histogram.to_dict()['encoding']['x']['bin'] == {'maxbins': 20}
    assert box_plot.to_dict()['mark']['type'] == 'boxplot'
    print("altair example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"altair example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: altair functionality test 4

```python
import altair as alt
import pandas as pd

# Example 4: Interactive visualizations and selections
print("Example 4: Testing Altair interactive features")

try:
    # Create time series data
    dates = pd.date_range('2023-01-01', periods=50, freq='D')
    data = pd.DataFrame({
        'date': dates,
        'value': range(50),
        'category': ['A'] * 25 + ['B'] * 25
    })
    
    # Interactive selection
    brush = alt.selection_interval()
    
    # Base chart with selection
    base = alt.Chart(data).mark_circle(size=100).encode(
        x='date:T',
        y='value:Q',
        color=alt.condition(brush, 'category:N', alt.value('lightgray'))
    ).add_params(brush)
    
    print("Interactive chart with brush selection created")
    
    # Chart with tooltip
    tooltip_chart = alt.Chart(data).mark_point().encode(
        x='date:T',
        y='value:Q',
        color='category:N',
        tooltip=['date:T', 'value:Q', 'category:N']
    )
    print("Chart with interactive tooltips created")
    
    # Multi-view chart
    chart1 = alt.Chart(data).mark_bar().encode(
        x='category:N',
        y='mean(value):Q'
    )
    
    chart2 = alt.Chart(data).mark_line().encode(
        x='date:T',
        y='value:Q',
        color='category:N'
    )
    
    combined = alt.hconcat(chart1, chart2)
    print("Multi-view concatenated chart created")
    
    params = base.to_dict()['params']
    assert [p['select']['type'] for p in params] == ['interval'], params
    assert len(tooltip_chart.to_dict()['encoding']['tooltip']) == 3
    assert len(combined.to_dict()['hconcat']) == 2
    print("altair example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"altair example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```
