```python
# Example 1: Construct an Amplitude client with a dummy API key.
# The key is stored on client.configuration — Amplitude exposes no client.api_key.
from amplitude import Amplitude

client = Amplitude("dummy-api-key-for-testing")
print(f"client type: {type(client).__name__}")
print(f"configuration.api_key = {client.configuration.api_key}")
print(f"configuration type: {type(client.configuration).__name__}")

assert isinstance(client, Amplitude)
assert client.configuration.api_key == "dummy-api-key-for-testing"
```

```python
# Example 2: Build a BaseEvent and inspect its serialized form.
from amplitude import BaseEvent

event = BaseEvent(
    event_type="test_event",
    user_id="user-123",
    event_properties={"plan": "pro", "count": 7},
)

body = event.get_event_body()
print(f"event_type = {body['event_type']}")
print(f"user_id = {body['user_id']}")
print(f"event_properties = {body['event_properties']}")

assert body["event_type"] == "test_event"
assert body["user_id"] == "user-123"
assert body["event_properties"]["plan"] == "pro"
```

```python
# Example 3: Identify event for user properties.
from amplitude import Identify, IdentifyEvent

identify = Identify()
identify.set("plan", "enterprise")
identify.set_once("signup_date", "2024-01-01")
identify.add("login_count", 1)

user_props = identify.user_properties
print(f"user_properties keys: {list(user_props.keys())}")
print(f"user_properties = {user_props}")

event = IdentifyEvent(user_id="user-456", user_properties=user_props)
print(f"identify event type: {event.event_type}")

assert len(user_props) >= 1
```

```python
# Example 4: Revenue event construction.
from amplitude import Revenue, RevenueEvent

revenue = Revenue(
    price=9.99,
    quantity=2,
    product_id="sku-001",
    revenue_type="purchase",
)

print(f"price = {revenue.price}")
print(f"quantity = {revenue.quantity}")
print(f"product_id = {revenue.product_id}")
print(f"is_valid = {revenue.is_valid()}")

rev_event = RevenueEvent(user_id="user-789", revenue_obj=revenue)
print(f"revenue event type: {rev_event.event_type}")

assert revenue.is_valid()
assert revenue.price == 9.99
```

```python
# Example 5: Configuration object inspection.
from amplitude import Amplitude
from amplitude.config import Config

client = Amplitude("dummy-key")
cfg = client.configuration

print(f"server_url = {cfg.server_url}")
print(f"flush_queue_size = {cfg.flush_queue_size}")
print(f"flush_interval_millis = {cfg.flush_interval_millis}")
print(f"flush_max_retries = {cfg.flush_max_retries}")
print(f"opt_out = {cfg.opt_out}")

assert isinstance(cfg, Config)
assert isinstance(cfg.flush_queue_size, int)
assert cfg.flush_queue_size > 0
assert cfg.opt_out is False
```
