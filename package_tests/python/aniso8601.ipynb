## Example 1: aniso8601 functionality test 1

```python
import datetime

import aniso8601

# Example 1: Basic aniso8601 datetime parsing
print("Example 1: Testing aniso8601 datetime parsing")

try:
    # aniso8601 parses ISO 8601 date/time strings
    print("aniso8601 module imported successfully")
    
    # Test parsing ISO 8601 datetime
    iso_datetime = "2023-06-15T10:30:45Z"
    parsed_dt = aniso8601.parse_datetime(iso_datetime)
    
    print(f"Parsed datetime: {iso_datetime} -> {parsed_dt}")
    print(f"Result type: {type(parsed_dt).__name__}")
    assert parsed_dt.replace(tzinfo=None) == datetime.datetime(2023, 6, 15, 10, 30, 45), parsed_dt
    assert parsed_dt.utcoffset() == datetime.timedelta(0), parsed_dt.utcoffset()
    
    print("aniso8601 example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aniso8601 example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: aniso8601 functionality test 2

```python
import datetime

import aniso8601

# Example 2: aniso8601 timezone parsing
print("Example 2: Testing aniso8601 timezone parsing")

try:
    # Test parsing datetime with timezone
    datetime_with_tz = "2023-06-15T10:30:45+05:00"
    parsed_tz = aniso8601.parse_datetime(datetime_with_tz)
    
    print(f"Timezone datetime: {datetime_with_tz}")
    print(f"Parsed result: {parsed_tz}")
    print(f"Timezone info: {parsed_tz.tzinfo}")
    assert parsed_tz.utcoffset() == datetime.timedelta(hours=5), parsed_tz.utcoffset()
    
    print("aniso8601 example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aniso8601 example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: aniso8601 functionality test 3

```python
import datetime

import aniso8601

# Example 3: aniso8601 duration parsing
print("Example 3: Testing aniso8601 duration parsing")

try:
    # Test parsing ISO 8601 duration
    duration_str = "P1Y2M3DT4H5M6S"
    parsed_duration = aniso8601.parse_duration(duration_str)
    
    print(f"Duration string: {duration_str}")
    print(f"Parsed duration: {parsed_duration}")
    print(f"Total seconds: {parsed_duration.total_seconds()}")
    # As a timedelta, a year counts as 365 days and a month as 30 days.
    assert parsed_duration == datetime.timedelta(days=365 + 60 + 3, hours=4, minutes=5, seconds=6), parsed_duration
    
    print("aniso8601 example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aniso8601 example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: aniso8601 functionality test 4

```python
import aniso8601

# Example 4: aniso8601 time parsing
print("Example 4: Testing aniso8601 time parsing")

try:
    # Test parsing ISO 8601 time only
    # Note: `parse_time` takes a bare time, not the `T` designator;
    # a leading `T` raises ISOFormatError. `parse_datetime` is the one that wants `<date>T<time>`.
    time_str = "14:30:45"
    parsed_time = aniso8601.parse_time(time_str)
    
    print(f"Time string: {time_str}")
    print(f"Parsed time: {parsed_time}")
    print(f"Hour: {parsed_time.hour}, Minute: {parsed_time.minute}, Second: {parsed_time.second}")
    assert (parsed_time.hour, parsed_time.minute, parsed_time.second) == (14, 30, 45), parsed_time
    
    print("aniso8601 example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"aniso8601 example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```
