# annotated_types: 5 Basic Usage Examples

```python
# Example 1: Using annotated types with min/max length constraints
import typing
from typing import Annotated
from annotated_types import MinLen, MaxLen

def process(data: Annotated[str, MinLen(3), MaxLen(10)]):
    print(f'Processing: {data}')
process('hello')

# The constraints travel with the type hint as plain, comparable metadata objects
hint = typing.get_type_hints(process, include_extras=True)['data']
assert typing.get_origin(hint) is Annotated
assert typing.get_args(hint) == (str, MinLen(3), MaxLen(10)), typing.get_args(hint)
assert typing.get_args(hint)[1].min_length == 3 and typing.get_args(hint)[2].max_length == 10
```

```python
# Example 2: Using annotated types with ge/le constraints
import typing
from typing import Annotated
from annotated_types import Ge, Le, Interval

def bounded(x: Annotated[int, Ge(0), Le(10)]):
    print(f'Bounded value: {x}')
bounded(5)

meta = typing.get_args(typing.get_type_hints(bounded, include_extras=True)['x'])[1:]
assert meta == (Ge(0), Le(10)), meta
# Interval is grouped metadata: it unpacks into the same individual bounds
assert list(Interval(ge=0, le=10)) == [Ge(0), Le(10)]
```

```python
# Example 3: Custom metadata
import typing
from typing import Annotated
class MyMeta: pass
def func(x: Annotated[int, MyMeta()]):
    print(f'With custom meta: {x}')
func(42)

args = typing.get_args(typing.get_type_hints(func, include_extras=True)['x'])
assert args[0] is int and isinstance(args[1], MyMeta), args
```

```python
# Example 4: Gt/Lt bounds on a float
import typing
from typing import Annotated
from annotated_types import Gt, Lt

def with_bounds(x: Annotated[float, Gt(0.0), Lt(10.0)]):
    print(f'Value with bounds: {x}')
with_bounds(3.14)

gt, lt = typing.get_args(typing.get_type_hints(with_bounds, include_extras=True)['x'])[1:]
assert (gt, lt) == (Gt(0.0), Lt(10.0)), (gt, lt)
assert gt.gt < 3.14 < lt.lt  # a validator reads the bounds straight off the metadata
```

```python
# Example 5: Integrating with type checkers
import typing
from typing import Annotated
from annotated_types import Len, MinLen, MaxLen

def check(data: Annotated[str, MinLen(2)]):
    print(f'Checked: {data}')
check('ok')

assert typing.get_args(typing.get_type_hints(check, include_extras=True)['data'])[1] == MinLen(2)
# Len(min, max) is grouped metadata that unpacks into MinLen and MaxLen
assert list(Len(2, 5)) == [MinLen(2), MaxLen(5)]
```
