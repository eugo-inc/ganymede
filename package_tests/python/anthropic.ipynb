## Example 2: anthropic functionality test 2

```python
import anthropic
from anthropic import Anthropic
import os

# Example 2: Anthropic API types and error handling
print("Example 2: Anthropic API types and error handling")

try:
    # Test different Anthropic client configurations
    print("Testing Anthropic client types and configurations...")
    
    # Show available classes and exceptions
    anthropic_classes = []
    anthropic_exceptions = []
    
    for attr_name in dir(anthropic):
        if not attr_name.startswith('_'):
            attr = getattr(anthropic, attr_name)
            if isinstance(attr, type):
                if 'Error' in attr_name or 'Exception' in attr_name:
                    anthropic_exceptions.append(attr_name)
                else:
                    anthropic_classes.append(attr_name)
    
    print(f"Anthropic classes: {anthropic_classes[:5]}")
    print(f"Exception types: {anthropic_exceptions[:5]}")
    
    # Test client initialization patterns
    print("\nTesting client initialization patterns:")
    
    # Option 1: Environment variable API key
    try:
        client = Anthropic()  # Uses ANTHROPIC_API_KEY env var
        print("  ✓ Environment variable configuration available")
    except Exception as e:
        print(f"  Environment configuration: {type(e).__name__}")
        raise
    
    # Option 2: Explicit API key
    try:
        client = Anthropic(api_key="test-key")
        print("  ✓ Explicit API key configuration available")
    except Exception as e:
        print(f"  Explicit API key: {type(e).__name__}")
        raise
    
    # Show client methods (without making API calls)
    if hasattr(anthropic, 'Anthropic'):
        client_methods = [method for method in dir(Anthropic) if not method.startswith('_')]
        print(f"\nClient methods available: {len(client_methods)}")
        
        # Look for key methods
        key_methods = ['messages', 'completions', 'chat']
        available_methods = [method for method in key_methods if method in client_methods]
        print(f"Key API methods: {available_methods}")

except Exception as e:
    print(f"Error testing Anthropic types: {e}")
    
    # Fallback to basic module inspection
    attrs = [attr for attr in dir(anthropic) if not attr.startswith('_')]
    print(f"Anthropic module attributes: {len(attrs)}")
    if attrs:
        print(f"Sample attributes: {attrs[:8]}")
    raise

print("Example 2 completed")
```

## Example 5: anthropic functionality test 5

```python
import anthropic
import sys

# Example 5: Anthropic package comprehensive analysis
print("Example 5: Anthropic package comprehensive analysis")

def analyze_anthropic_package():
    """Complete analysis of the Anthropic Python package"""
    
    print("Anthropic package analysis:")
    print(f"  Package: {anthropic.__name__}")
    print(f"  Type: {type(anthropic).__name__}")
    
    # Version information
    if hasattr(anthropic, '__version__'):
        print(f"  Version: {anthropic.__version__}")
    elif hasattr(anthropic, 'version'):
        print(f"  Version: {anthropic.version}")
    else:
        print("  Version: Not directly accessible")
    
    # Package structure analysis
    attrs = dir(anthropic)
    public_attrs = [attr for attr in attrs if not attr.startswith('_')]
    
    print(f"\nPackage structure:")
    print(f"  Total attributes: {len(attrs)}")
    print(f"  Public attributes: {len(public_attrs)}")
    
    # Categorize components
    classes = []
    functions = []
    modules = []
    exceptions = []
    
    for attr_name in public_attrs:
        attr = getattr(anthropic, attr_name)
        if isinstance(attr, type):
            if 'Error' in attr_name or 'Exception' in attr_name:
                exceptions.append(attr_name)
            else:
                classes.append(attr_name)
        elif callable(attr):
            functions.append(attr_name)
        elif hasattr(attr, '__file__'):
            modules.append(attr_name)
    
    print(f"\nComponent breakdown:")
    print(f"  Classes: {classes[:8]}")
    print(f"  Functions: {functions[:5]}")
    print(f"  Exceptions: {exceptions[:5]}")
    print(f"  Modules: {modules[:5]}")
    assert {'Anthropic', 'AsyncAnthropic'} <= set(classes), classes
    assert 'APIError' in exceptions and issubclass(anthropic.APIError, Exception), exceptions

def demonstrate_api_patterns():
    """Show common API usage patterns"""
    
    print("\nCommon Anthropic API patterns:")
    
    patterns = {
        'Simple Chat': '''
client = anthropic.Anthropic(api_key="your-key")
response = client.messages.create(
    model="claude-3-sonnet-20240229",
    max_tokens=1000,
    messages=[{"role": "user", "content": "Hello!"}]
)''',
        
        'Conversation': '''
messages = [
    {"role": "user", "content": "What is Python?"},
    {"role": "assistant", "content": "Python is a programming language..."},
    {"role": "user", "content": "Show me an example"}
]
response = client.messages.create(model="claude-3", messages=messages)''',
        
        'Streaming': '''
stream = client.messages.create(
    model="claude-3-sonnet-20240229",
    messages=[{"role": "user", "content": "Write a story"}],
    max_tokens=1000,
    stream=True
)
for chunk in stream:
    print(chunk.delta.text, end="")''',
        
        'Async Usage': '''
async with anthropic.AsyncAnthropic(api_key="key") as client:
    response = await client.messages.create(
        model="claude-3",
        messages=[{"role": "user", "content": "Hello"}]
    )'''
    }
    
    for pattern_name, code in patterns.items():
        print(f"\n{pattern_name}:")
        # Show just the first line for brevity
        first_line = code.strip().split('\n')[0]
        print(f"  {first_line}...")

def show_package_info():
    """Display key package information"""
    
    print(f"\nAnthropic package key features:")
    features = [
        "🤖 Access to Claude AI models (Opus, Sonnet, Haiku)",
        "💬 Conversational message-based API",
        "⚡ Async support with AsyncAnthropic",
        "🔄 Streaming responses for real-time output",
        "🎛️ Fine-grained control over generation parameters",
        "🔒 Built-in error handling and rate limiting",
        "📝 Type hints for better development experience",
        "🌐 HTTP client with automatic retries"
    ]
    
    for feature in features:
        print(f"  {feature}")
    
    print(f"\nSupported Claude models:")
    models = [
        "claude-3-opus-20240229 (Most capable, best for complex tasks)",
        "claude-3-sonnet-20240229 (Balanced performance and speed)",
        "claude-3-haiku-20240307 (Fastest, good for simple tasks)",
        "claude-2.1 (Previous generation, still capable)",
        "claude-2.0 (Legacy model)"
    ]
    
    for model in models:
        print(f"  {model}")

# Run all analysis functions
analyze_anthropic_package()
demonstrate_api_patterns()
show_package_info()

print("\nExample 5 completed - Anthropic package analysis")
print("Note: Anthropic provides Python SDK for Claude AI models")
print("Requires ANTHROPIC_API_KEY environment variable or explicit API key")
```
