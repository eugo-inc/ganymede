## Example 1: Basic Token Stream and Lexer Usage

```python
from antlr4 import *
from antlr4.Token import CommonToken
from antlr4.Token import Token

# Example 1: Basic Token Stream and Lexer Usage
print("Example 1: ANTLR Token Stream and Lexer")

try:
    # Create a simple input stream from string
    input_text = "Hello ANTLR world 123 + 456"
    input_stream = InputStream(input_text)
    
    print(f"✓ Input stream created from: '{input_text}'")
    print(f"Input stream size: {input_stream.size}")
    
    # Test basic input stream operations
    print(f"First character: '{input_stream.LA(1)}'")
    print(f"Character at position 6: '{chr(input_stream.data[6])}'")
    assert input_stream.size == len(input_text) and input_stream.LA(1) == ord('H')
    
    # Create a token (manual demonstration)
    token = CommonToken()
    token.type = Token.EOF
    token.text = "EOF"
    token.line = 1
    token.column = len(input_text)
    
    print(f"✓ Token created: type={token.type}, text='{token.text}', line={token.line}, col={token.column}")
    
    # Test common token stream
    token_list = []
    
    # Simulate tokenizing (normally done by generated lexer)
    words = input_text.split()
    for i, word in enumerate(words):
        token = CommonToken()
        token.type = i + 1  # Simple type assignment
        token.text = word
        token.line = 1
        token.column = input_text.find(word)
        token_list.append(token)
    
    # Add EOF token
    eof_token = CommonToken()
    eof_token.type = Token.EOF
    eof_token.text = "<EOF>"
    token_list.append(eof_token)
    
    print(f"✓ Created {len(token_list)} tokens")
    assert [t.text for t in token_list[:-1]] == words and token_list[-1].type == Token.EOF
    for token in token_list:
        print(f"  Token: '{token.text}' (type={token.type}, col={token.column})")
    
except Exception as e:
    print(f"Error in token stream example: {e}")
    raise

print("Example 1 completed\n")
```

## Example 3: Token Stream and Buffered Operations

```python
from antlr4 import *
from antlr4.Token import CommonToken

# Example 3: Token Stream and Buffered Operations
print("Example 3: Token Stream Operations")

try:
    # Create tokens for token stream
    tokens = []
    
    # Simulate a series of tokens (normally from lexer)
    token_data = [
        ("if", 1), ("(", 2), ("x", 3), (">", 4), ("10", 5), (")", 6),
        ("then", 7), ("print", 8), ("(", 2), ("'hello'", 9), (")", 6)
    ]
    
    for i, (text, token_type) in enumerate(token_data):
        token = CommonToken()
        token.type = token_type
        token.text = text
        token.tokenIndex = i
        token.line = 1
        token.column = i * 5  # Simple column calculation
        tokens.append(token)
    
    # Add EOF token
    eof_token = CommonToken()
    eof_token.type = Token.EOF
    eof_token.text = "<EOF>"
    eof_token.tokenIndex = len(tokens)
    tokens.append(eof_token)
    
    # Create buffered token stream
    from antlr4.BufferedTokenStream import BufferedTokenStream
    
    # Create a simple token source
    class SimpleTokenSource:
        def __init__(self, tokens):
            self.tokens = tokens
            self.index = 0
        
        def nextToken(self):
            if self.index < len(self.tokens):
                token = self.tokens[self.index]
                self.index += 1
                return token
            return self.tokens[-1]  # EOF token
    
    token_source = SimpleTokenSource(tokens)
    
    print(f"✓ Created token stream with {len(tokens)} tokens")
    
    # Test token operations
    sample_tokens = tokens[:5]  # First 5 tokens
    for token in sample_tokens:
        print(f"  Token[{token.tokenIndex}]: '{token.text}' (type={token.type})")
    
    # Test lookahead operations (LA = Look Ahead)
    print(f"\nTesting lookahead operations:")
    print(f"First token: '{tokens[0].text}'")
    print(f"Second token: '{tokens[1].text}'")
    print(f"EOF token: '{tokens[-1].text}'")
    
    # Test token filtering (skip whitespace, comments, etc.)
    filtered_tokens = [t for t in tokens if t.type != Token.EOF]
    print(f"\n✓ Filtered tokens (excluding EOF): {len(filtered_tokens)}")
    
except Exception as e:
    print(f"Error in token stream example: {e}")
    raise

print("Example 3 completed\n")
```

## Example 4: Parse Tree Walking and Visitors

```python
from antlr4 import *
from antlr4.tree.Tree import ParseTreeWalker

# Example 4: Parse Tree Walking and Visitors
print("Example 4: Parse Tree Walking and Visitors")

try:
    # Create a parse tree visitor base class
    from antlr4.tree.Tree import ParseTreeVisitor
    
    class SimpleVisitor(ParseTreeVisitor):
        def __init__(self):
            self.visited_nodes = []
        
        def visit(self, tree):
            """Visit a parse tree node"""
            self.visited_nodes.append(str(tree))
            print(f"Visiting node: {type(tree).__name__}")
            return super().visit(tree)
        
        def visitChildren(self, node):
            """Visit all children of a node"""
            result = []
            for i in range(node.getChildCount() if hasattr(node, 'getChildCount') else 0):
                child = node.getChild(i) if hasattr(node, 'getChild') else None
                if child:
                    result.append(self.visit(child))
            return result
    
    visitor = SimpleVisitor()
    print("✓ Parse tree visitor created")
    
    # Create a listener for tree walking
    from antlr4.tree.Tree import ParseTreeListener
    
    class SimpleListener(ParseTreeListener):
        def __init__(self):
            self.enter_count = 0
            self.exit_count = 0
            
        def enterEveryRule(self, ctx):
            """Called when entering any rule"""
            self.enter_count += 1
            print(f"Entering rule {self.enter_count}: {type(ctx).__name__}")
        
        def exitEveryRule(self, ctx):
            """Called when exiting any rule"""
            self.exit_count += 1
            print(f"Exiting rule {self.exit_count}: {type(ctx).__name__}")
    
    listener = SimpleListener()
    print("✓ Parse tree listener created")
    
    # Simulate tree walking with mock parse tree
    from antlr4.ParserRuleContext import ParserRuleContext
    
    # Create a simple mock tree structure
    root = ParserRuleContext()
    child1 = ParserRuleContext()
    child2 = ParserRuleContext()
    
    # Simulate walking the tree
    listener.enterEveryRule(root)
    listener.enterEveryRule(child1)
    listener.exitEveryRule(child1)
    listener.enterEveryRule(child2)
    listener.exitEveryRule(child2)
    listener.exitEveryRule(root)
    
    print(f"\n✓ Tree walking completed")
    print(f"Enter events: {listener.enter_count}")
    print(f"Exit events: {listener.exit_count}")
    
    # Test parse tree walker
    walker = ParseTreeWalker()
    print(f"✓ Parse tree walker created: {type(walker).__name__}")
    
except Exception as e:
    print(f"Error in parse tree walking example: {e}")
    raise AssertionError(
        f"antlr4_python3_runtime example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed\n")
```

## Example 5: Error Handling and Recovery

```python
from antlr4 import *
from antlr4.Token import CommonToken
from antlr4.error.ErrorListener import ErrorListener

# Example 5: Error Handling and Recovery
print("Example 5: ANTLR Error Handling and Recovery")

try:
    # Create custom error listener with detailed error reporting
    class DetailedErrorListener(ErrorListener):
        def __init__(self):
            super().__init__()
            self.errors = []
            self.warnings = []
        
        def syntaxError(self, recognizer, offendingSymbol, line, column, msg, e):
            """Handle syntax errors during parsing"""
            error = {
                'type': 'SyntaxError',
                'line': line,
                'column': column,
                'message': msg,
                'symbol': offendingSymbol.text if offendingSymbol else None
            }
            self.errors.append(error)
            print(f"Syntax Error at {line}:{column} - {msg}")
        
        def reportAmbiguity(self, recognizer, dfa, startIndex, stopIndex, exact, ambigAlts, configs):
            """Handle ambiguity in grammar"""
            warning = {
                'type': 'Ambiguity',
                'start': startIndex,
                'stop': stopIndex,
                'alternatives': str(ambigAlts)
            }
            self.warnings.append(warning)
            print(f"Ambiguity detected: indices {startIndex}-{stopIndex}")
        
        def reportAttemptingFullContext(self, recognizer, dfa, startIndex, stopIndex, conflictingAlts, configs):
            """Handle full context attempts"""
            print(f"Attempting full context: {startIndex}-{stopIndex}")
        
        def reportContextSensitivity(self, recognizer, dfa, startIndex, stopIndex, prediction, configs):
            """Handle context sensitivity"""
            print(f"Context sensitivity: prediction {prediction}")
    
    error_listener = DetailedErrorListener()
    print("✓ Detailed error listener created")
    
    # Test error recovery strategies
    from antlr4.error.ErrorStrategy import DefaultErrorStrategy
    
    class CustomErrorStrategy(DefaultErrorStrategy):
        def __init__(self):
            super().__init__()
            self.recovery_attempts = 0
        
        def recover(self, recognizer, e):
            """Custom error recovery"""
            self.recovery_attempts += 1
            print(f"Recovery attempt #{self.recovery_attempts}: {type(e).__name__}")
            super().recover(recognizer, e)
        
        def recoverInline(self, recognizer):
            """Inline error recovery"""
            print("Attempting inline recovery")
            return super().recoverInline(recognizer)
    
    error_strategy = CustomErrorStrategy()
    print("✓ Custom error strategy created")
    
    # Simulate error conditions and recovery
    mock_errors = [
        {'line': 1, 'col': 10, 'msg': 'Missing semicolon'},
        {'line': 2, 'col': 5, 'msg': 'Unexpected token'},
        {'line': 3, 'col': 15, 'msg': 'Mismatched parentheses'}
    ]
    
    print(f"\nSimulating error conditions:")
    for i, error in enumerate(mock_errors, 1):
        error_listener.syntaxError(
            recognizer=None,
            offendingSymbol=CommonToken(),
            line=error['line'],
            column=error['col'],
            msg=error['msg'],
            e=None
        )
    
    # Test error reporting
    print(f"\n✓ Error collection completed")
    print(f"Total syntax errors: {len(error_listener.errors)}")
    print(f"Total warnings: {len(error_listener.warnings)}")
    print(f"Recovery attempts: {error_strategy.recovery_attempts}")
    
    # Summary of ANTLR runtime capabilities
    print(f"\nANTLR4 Python Runtime Capabilities:")
    print("• Lexical analysis with token stream management")
    print("• Parser contexts and parse tree construction")  
    print("• Tree walking with visitors and listeners")
    print("• Comprehensive error handling and recovery")
    print("• Support for left-recursive grammar rules")
    print("• Adaptive LL(*) parsing algorithm")
    
except Exception as e:
    print(f"Error in error handling example: {e}")
    raise

print("Example 5 completed")
```
