## Example 2: anyio functionality test 2

```python
import anyio

# Example 2: anyio CancelScope functionality
print("Example 2: Testing anyio CancelScope")

async def test_cancel_scope():
    print("Testing anyio CancelScope functionality")
    
    # anyio 4.x: CancelScope takes an absolute `deadline` (anyio.current_time() based), not `timeout=`
    # (move_on_after(seconds) is the relative shorthand). A generous deadline must NOT cancel a short sleep.
    with anyio.CancelScope(deadline=anyio.current_time() + 5) as scope:
        print("Inside cancel scope")
        await anyio.sleep(0.05)  # This should complete
        print("Short sleep completed")
    assert not scope.cancel_called and not scope.cancelled_caught, "the 5 s deadline cancelled a 0.05 s sleep"
    
    # A deadline that passes first MUST cancel the body; the scope absorbs the cancellation.
    with anyio.CancelScope(deadline=anyio.current_time() + 0.05) as scope:
        await anyio.sleep(10)
        raise AssertionError("the 10 s sleep was not cancelled at the 0.05 s deadline")
    assert scope.cancelled_caught, "the scope did not record the deadline cancellation"
    print("Deadline cancellation caught by the scope")
    
    print("CancelScope test completed successfully")
    print("anyio example 2 completed successfully")

try:
    await test_cancel_scope()
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"anyio example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished")
```

## Example 3: anyio functionality test 3

```python
import anyio

# Example 3: anyio task groups
print("Example 3: Testing anyio task groups")

async def worker_task(task_id):
    await anyio.sleep(0.01)
    return f"Task {task_id} completed"

async def test_task_group():
    try:
        print("Testing anyio task group functionality")
        
        # Create and run multiple tasks concurrently
        async with anyio.create_task_group() as tg:
            tg.start_soon(worker_task, 1)
            tg.start_soon(worker_task, 2)
            tg.start_soon(worker_task, 3)
        
        print("All tasks in group completed")
        print("Task group functionality working")
        print("anyio example 3 completed successfully")
        
    except ImportError as e:
        print(f"Import error: {e}")
        raise
    except Exception as e:
        print(f"Runtime error: {e}")
        raise

try:
    await test_task_group()
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"anyio example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished")
```

## Example 4: anyio functionality test 4

```python
import anyio

# Example 4: anyio CapacityLimiter
print("Example 4: Testing anyio CapacityLimiter")

async def limited_task(task_id, limiter):
    async with limiter:
        print(f"Task {task_id} acquired limiter")
        await anyio.sleep(0.01)
        print(f"Task {task_id} releasing limiter")

async def test_capacity_limiter():
    try:
        print("Testing anyio CapacityLimiter functionality")
        
        # Create a limiter that allows max 2 concurrent operations
        limiter = anyio.CapacityLimiter(2)
        
        # Start multiple tasks that will be limited
        async with anyio.create_task_group() as tg:
            for i in range(3):
                tg.start_soon(limited_task, i + 1, limiter)
        
        print("CapacityLimiter test completed")
        print("anyio example 4 completed successfully")
        
    except ImportError as e:
        print(f"Import error: {e}")
        raise
    except Exception as e:
        print(f"Runtime error: {e}")
        raise

try:
    await test_capacity_limiter()
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"anyio example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished")
```
