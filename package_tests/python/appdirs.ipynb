## Example 4: appdirs functionality test 4

```python
import os

import appdirs

# Example 4: appdirs log directory
print("Example 4: Testing appdirs log directory")

try:
    # Test user log directory functionality
    app_name = "LoggingApp"
    app_author = "DevCompany"
    
    log_dir = appdirs.user_log_dir(app_name, app_author)
    print(f"User log directory: {log_dir}")
    
    # Test with version information
    versioned_dir = appdirs.user_data_dir(app_name, app_author, version="1.0")
    print(f"Versioned data dir: {versioned_dir}")
    assert os.path.basename(versioned_dir) == "1.0", versioned_dir  # the version is the last path component
    
    print("appdirs example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"appdirs example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```
