# archspec Package Testing

This notebook tests the archspec package - CPU architecture specification library.

archspec provides a standard way to describe CPU architectures and their capabilities for scientific computing.

## Example 1: CPU Architecture Detection


```python
import archspec.cpu

# Example 1: CPU Architecture Detection and Information
print("Example 1: CPU Architecture Detection")

try:
    # Detect current CPU architecture
    current_cpu = archspec.cpu.host()
    
    print(f"✓ Current CPU detected: {current_cpu}")
    print(f"CPU name: {current_cpu.name}")
    print(f"CPU family: {current_cpu.family}")
    print(f"CPU vendor: {current_cpu.vendor}")
    
    # Get CPU features
    features = current_cpu.features
    print(f"✓ CPU features count: {len(features)}")
    print(f"Sample features: {sorted(list(features))[:10]}")
    
    # Check specific features
    common_features = ['sse', 'sse2', 'avx', 'avx2', 'fma']
    available_features = []
    for feature in common_features:
        if feature in features:
            available_features.append(feature)
    
    print(f"✓ Common vector features available: {available_features}")
    
    # Get architecture info
    print(f"CPU generation: {current_cpu.generation if hasattr(current_cpu, 'generation') else 'N/A'}")
    print(f"CPU optimization target: {current_cpu}")
    
    # Test microarchitecture hierarchy
    parents = current_cpu.ancestors
    print(f"✓ Architecture hierarchy: {len(parents)} ancestor levels")
    for i, parent in enumerate(parents):
        print(f"  Level {i}: {parent}")
    assert current_cpu.name in archspec.cpu.TARGETS, current_cpu.name
    assert current_cpu.family == current_cpu or current_cpu.family in parents, (current_cpu.family, parents)
    
except Exception as e:
    print(f"Error in CPU detection: {e}")
    raise AssertionError(
        f"archspec example (cell 2) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 completed\n")
```

## Example 2: CPU Compatibility and Optimization

```python
import archspec.cpu

# Example 2: CPU Compatibility and Optimization Targets
print("Example 2: CPU Compatibility and Optimization")

try:
    # Get current CPU
    host_cpu = archspec.cpu.host()
    print(f"Host CPU: {host_cpu}")
    
    # Test CPU compatibility with different targets
    test_targets = ['x86_64', 'haswell', 'skylake', 'generic']
    
    print(f"\nTesting compatibility with common targets:")
    for target in test_targets:
        try:
            target_cpu = archspec.cpu.TARGETS.get(target)
            if target_cpu:
                # Check if host is compatible with target
                is_compatible = host_cpu >= target_cpu
                print(f"  {target}: {'✓ Compatible' if is_compatible else '✗ Not compatible'}")
            else:
                print(f"  {target}: Target not found")
        except Exception as e:
            print(f"  {target}: Error - {type(e).__name__}")
            raise
    
    # Find optimal target for current CPU
    print(f"\nFinding optimal optimization targets:")
    
    # Get all available targets
    available_targets = list(archspec.cpu.TARGETS.keys())
    print(f"Total available targets: {len(available_targets)}")
    
    # Find compatible targets
    compatible_targets = []
    for target_name in available_targets[:20]:  # Test first 20
        target = archspec.cpu.TARGETS[target_name]
        if host_cpu >= target:
            compatible_targets.append(target_name)
    
    print(f"Compatible targets (sample): {compatible_targets[:10]}")
    assert host_cpu >= host_cpu.family, (host_cpu, host_cpu.family)  # a CPU runs code built for its family
    
    # Test CPU feature sets
    print(f"\nCPU feature analysis:")
    features = host_cpu.features
    
    # Categorize features
    vector_features = [f for f in features if any(x in f for x in ['sse', 'avx', 'fma', 'simd'])]
    crypto_features = [f for f in features if any(x in f for x in ['aes', 'sha', 'crypto'])]
    
    print(f"Vector processing features: {vector_features}")
    print(f"Cryptography features: {crypto_features}")
    
    # Check for specific optimization opportunities
    if 'avx2' in features:
        print("✓ AVX2 available - can use vectorized operations")
    if 'fma' in features:
        print("✓ FMA available - can use fused multiply-add")
    if 'aes' in features:
        print("✓ AES-NI available - hardware crypto acceleration")
    
except Exception as e:
    print(f"Error in compatibility testing: {e}")
    raise AssertionError(
        f"archspec example (cell 4) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed\n")
```

## Example 3: Compiler and Build System Integration

```python
import archspec.cpu

# Example 3: Compiler and Build System Integration
print("Example 3: Compiler Flags and Build Integration")

try:
    # Get current CPU for optimization
    host_cpu = archspec.cpu.host()
    print(f"Optimizing for CPU: {host_cpu}")
    
    # Generate compiler flags for different compilers
    print(f"\nCompiler optimization flags:")
    
    # GCC flags
    if hasattr(host_cpu, 'to_dict'):
        cpu_info = host_cpu.to_dict()
        print(f"CPU info available: {list(cpu_info.keys())}")
    
    # Simulate compiler flag generation
    features = host_cpu.features
    
    # GCC/Clang style flags
    gcc_flags = []
    if 'sse2' in features:
        gcc_flags.append('-msse2')
    if 'avx' in features:
        gcc_flags.append('-mavx')
    if 'avx2' in features:
        gcc_flags.append('-mavx2')
    if 'fma' in features:
        gcc_flags.append('-mfma')
    
    print(f"GCC/Clang flags: {' '.join(gcc_flags)}")
    
    # Architecture-specific optimization
    print(f"\nArchitecture-specific optimizations:")
    
    # Check CPU generation for optimization level
    if hasattr(host_cpu, 'family'):
        family = host_cpu.family
        print(f"CPU family: {family}")
        
        # Generate -march flag equivalent
        if 'intel' in str(host_cpu.vendor).lower():
            print(f"Intel optimization: -march={host_cpu.name}")
        elif 'amd' in str(host_cpu.vendor).lower():
            print(f"AMD optimization: -march={host_cpu.name}")
    
    # Build system integration examples
    print(f"\nBuild system integration examples:")
    
    cmake_config = f'''
# CMake configuration
set(CMAKE_C_FLAGS "${{CMAKE_C_FLAGS}} {' '.join(gcc_flags[:3])}")
set(CMAKE_CXX_FLAGS "${{CMAKE_CXX_FLAGS}} {' '.join(gcc_flags[:3])}")
'''
    
    print("CMake integration:")
    for line in cmake_config.strip().split('\n'):
        print(f"  {line}")
    
    # Python setup.py integration
    setup_py_config = f'''
# setup.py configuration
extra_compile_args = {gcc_flags[:3]}
Extension('mymodule', sources=['src.c'], extra_compile_args=extra_compile_args)
'''
    
    print(f"\nsetup.py integration:")
    for line in setup_py_config.strip().split('\n'):
        print(f"  {line}")
    
    # Spack/Conda integration
    print(f"\nPackage manager integration:")
    spack_spec = f"spack install mypackage target={host_cpu.name}"
    print(f"Spack: {spack_spec}")
    
    # Environment variable setup
    print(f"\nEnvironment variables for optimization:")
    env_vars = {
        'CFLAGS': ' '.join(gcc_flags),
        'CXXFLAGS': ' '.join(gcc_flags),
        'TARGET_CPU': str(host_cpu.name)
    }
    
    for var, value in env_vars.items():
        print(f"export {var}=\"{value}\"")
    
except Exception as e:
    print(f"Error in compiler integration: {e}")
    
    # Show general concepts
    print("Compiler integration concepts:")
    concepts = [
        "Generate architecture-specific compiler flags",
        "Optimize for target CPU capabilities", 
        "Integrate with build systems (CMake, Make, etc.)",
        "Support cross-compilation scenarios",
        "Enable vectorization and CPU-specific optimizations"
    ]
    
    for concept in concepts:
        print(f"  • {concept}")
    raise  # re-raise: a broken package must fail this example, not print past it

print("Example 3 completed\n")
```

## Example 4: Cross-Platform Architecture Analysis

```python
import archspec.cpu

# Example 4: Cross-Platform Architecture Analysis
print("Example 4: Cross-Platform Architecture Support")

try:
    # Analyze current platform
    host_cpu = archspec.cpu.host()
    print(f"Current platform: {host_cpu}")
    
    # Explore CPU architecture families
    print(f"\nArchitecture family analysis:")
    
    # Get all available CPU targets
    all_targets = archspec.cpu.TARGETS
    print(f"Total CPU targets available: {len(all_targets)}")
    
    # Categorize by vendor/family
    intel_targets = []
    amd_targets = []
    arm_targets = []
    other_targets = []
    
    for target_name, target_cpu in list(all_targets.items())[:50]:  # Sample first 50
        vendor = str(target_cpu.vendor).lower()
        if 'intel' in vendor or 'genuine' in vendor:
            intel_targets.append(target_name)
        elif 'amd' in vendor or 'authentic' in vendor:
            amd_targets.append(target_name)
        elif 'arm' in vendor or target_name.startswith('arm'):
            arm_targets.append(target_name)
        else:
            other_targets.append(target_name)
    
    print(f"Intel/x86 targets: {intel_targets[:8]}")
    print(f"AMD targets: {amd_targets[:8]}")  
    print(f"ARM targets: {arm_targets[:8]}")
    print(f"Other targets: {other_targets[:8]}")
    
    # Test architecture generations
    print(f"\nArchitecture generation analysis:")
    
    # Look for common architecture generations
    generations = {}
    for target_name in list(all_targets.keys())[:30]:
        target = all_targets[target_name]
        if hasattr(target, 'generation'):
            generations.setdefault(target.generation, []).append(target_name)
    
    for gen, targets in list(generations.items())[:5]:
        print(f"Generation {gen}: {targets[:3]}")
    
    # Cross-compilation analysis
    print(f"\nCross-compilation compatibility:")
    
    # Test specific cross-platform scenarios
    cross_platform_tests = [
        ('x86_64', 'skylake'),      # x86_64 baseline to modern Intel
        ('x86_64', 'zen2'),         # x86_64 baseline to AMD
        ('haswell', 'skylake'),     # Intel generation upgrade
        ('zen', 'zen2')             # AMD generation upgrade
    ]
    
    for base_arch, target_arch in cross_platform_tests:
        try:
            base = all_targets.get(base_arch)
            target = all_targets.get(target_arch)
            
            if base and target:
                compatible = target >= base  # Can target run base?
                optimized = base < target    # Is target more optimized?
                
                print(f"{base_arch} → {target_arch}: "
                      f"Compatible={compatible}, Optimized={optimized}")
            else:
                print(f"{base_arch} → {target_arch}: Target not found")
                
        except Exception as e:
            print(f"{base_arch} → {target_arch}: Error - {type(e).__name__}")
            raise
    
    # The target database orders generations within a family
    assert all_targets['skylake'] > all_targets['haswell'] >= all_targets['x86_64']
    assert all_targets['zen2'] > all_targets['zen']
    assert not (all_targets['zen2'] >= all_targets['skylake'])  # different vendors' lines are not ordered
    
    # Feature set comparison
    print(f"\nFeature set comparison:")
    
    # Compare feature sets across different architectures
    if len(intel_targets) > 0 and len(amd_targets) > 0:
        try:
            intel_cpu = all_targets[intel_targets[0]]
            amd_cpu = all_targets[amd_targets[0]]
            
            intel_features = intel_cpu.features
            amd_features = amd_cpu.features
            
            common_features = intel_features & amd_features
            intel_only = intel_features - amd_features
            amd_only = amd_features - intel_features
            
            print(f"Common features: {len(common_features)}")
            print(f"Intel-specific: {len(intel_only)}")
            print(f"AMD-specific: {len(amd_only)}")
            
            if common_features:
                print(f"Sample common: {list(common_features)[:5]}")
                
        except Exception as e:
            print(f"Feature comparison error: {type(e).__name__}")
            raise
    
    # Platform detection summary
    print(f"\nPlatform capabilities:")
    print(f"✓ CPU architecture detection and classification")
    print(f"✓ Cross-platform compatibility analysis")
    print(f"✓ Feature set comparison across vendors")
    print(f"✓ Optimization target selection")
    print(f"✓ Build system integration support")

except Exception as e:
    print(f"Error in cross-platform analysis: {e}")
    
    # Show supported platforms
    print("Supported platform categories:")
    platforms = [
        "x86_64: Intel and AMD 64-bit processors",
        "ARM: ARM-based processors (mobile, embedded)",
        "PowerPC: IBM POWER architecture",
        "RISC-V: Open-source RISC architecture",
        "Various microarchitectures within each family"
    ]
    
    for platform in platforms:
        print(f"  • {platform}")
    raise  # re-raise: a broken package must fail this example, not print past it

print("Example 4 completed\n")
```

## Example 5: Scientific Computing Optimization

```python
import archspec.cpu
import platform

# Example 5: Scientific Computing Optimization and Package Analysis
print("Example 5: Scientific Computing Architecture Optimization")

def analyze_cpu_for_science():
    """Analyze CPU capabilities for scientific computing workloads"""
    
    host_cpu = archspec.cpu.host()
    print(f"Scientific computing analysis for: {host_cpu}")
    
    # Get CPU features relevant to scientific computing
    features = host_cpu.features
    
    # Categorize features by scientific computing relevance
    vector_features = [f for f in features if any(x in f for x in 
                      ['sse', 'avx', 'fma', 'vector', 'simd'])]
    math_features = [f for f in features if any(x in f for x in 
                    ['fma', 'f16c', 'bmi', 'popcnt'])]
    memory_features = [f for f in features if any(x in f for x in 
                      ['prefetch', 'clflush', 'movbe'])]
    
    print(f"\nScientific computing relevant features:")
    print(f"  Vector processing: {len(vector_features)} features")
    print(f"  Math operations: {len(math_features)} features") 
    print(f"  Memory optimization: {len(memory_features)} features")
    
    if vector_features:
        print(f"  Vector features: {vector_features}")
    if math_features:
        print(f"  Math features: {math_features}")
        
    # Analyze vectorization capabilities
    print(f"\nVectorization capability analysis:")
    vector_widths = {}
    
    if 'sse' in features:
        vector_widths['SSE'] = '128-bit (4x float, 2x double)'
    if 'avx' in features:
        vector_widths['AVX'] = '256-bit (8x float, 4x double)'
    if 'avx2' in features:
        vector_widths['AVX2'] = '256-bit + integer operations'
    if 'avx512f' in features:
        vector_widths['AVX-512'] = '512-bit (16x float, 8x double)'
        
    for instruction_set, width in vector_widths.items():
        print(f"  {instruction_set}: {width}")
    
    # Floating-point capabilities
    print(f"\nFloating-point optimization:")
    fp_optimizations = []
    
    if 'fma' in features:
        fp_optimizations.append("FMA: Fused multiply-add operations")
    if 'f16c' in features:
        fp_optimizations.append("F16C: Half-precision float conversion")
    if 'sse4_1' in features:
        fp_optimizations.append("SSE4.1: Enhanced floating-point operations")
        
    for opt in fp_optimizations:
        print(f"  ✓ {opt}")
        
    if not fp_optimizations:
        print(f"  Standard IEEE 754 floating-point")
    
    return host_cpu
    

def recommend_optimizations(cpu):
    """Recommend optimizations for scientific libraries"""
    
    print(f"\nRecommendations for scientific computing:")
    
    features = cpu.features
    
    # NumPy/SciPy recommendations
    print(f"\nNumPy/SciPy optimizations:")
    if 'avx2' in features and 'fma' in features:
        print(f"  ✓ Use Intel MKL or OpenBLAS with AVX2+FMA")
        print(f"  ✓ Enable BLAS/LAPACK hardware acceleration")
    elif 'avx' in features:
        print(f"  ✓ Use BLAS libraries with AVX support")
    else:
        print(f"  ⚠ Consider upgrading to newer CPU for better performance")
        
    # Machine learning recommendations
    print(f"\nMachine Learning optimizations:")
    if 'avx512f' in features:
        print(f"  ✓ Excellent for TensorFlow/PyTorch with AVX-512")
        print(f"  ✓ Deep learning training highly optimized")
    elif 'avx2' in features:
        print(f"  ✓ Good performance for ML inference")
        print(f"  ✓ Use optimized ML libraries (Intel Extension)")
    else:
        print(f"  ⚠ Consider GPU acceleration for ML workloads")
        
    # Compiler recommendations
    print(f"\nCompiler optimization recommendations:")
    compiler_flags = []
    
    if 'avx2' in features:
        compiler_flags.extend(['-mavx2', '-mfma'])
    elif 'avx' in features:
        compiler_flags.append('-mavx')
    if 'sse4_2' in features:
        compiler_flags.append('-msse4.2')
        
    if compiler_flags:
        print(f"  GCC/Clang: {' '.join(compiler_flags)}")
        print(f"  Intel: -xHost or -march=native")
    
    # Package-specific recommendations  
    print(f"\nPackage-specific optimizations:")
    optimizations = [
        "Conda: Install numpy/scipy with MKL backend",
        "pip: Use pip install numpy[mkl] if available", 
        "Docker: Use intel/python or continuumio/anaconda",
        "HPC: Module load intel compiler + MKL libraries",
        "Build: Set CFLAGS/CXXFLAGS with arch-specific flags"
    ]
    
    for opt in optimizations:
        print(f"  • {opt}")
        

def show_archspec_integration():
    """Show how archspec integrates with scientific tools"""
    
    print(f"\narchspec integration with scientific tools:")
    
    integrations = {
        'Spack': 'Package manager uses archspec for CPU targeting',
        'Conda': 'Can specify CPU optimization in package builds', 
        'CMake': 'CPU detection for optimized scientific libraries',
        'Meson': 'Build system integration for performance',
        'setuptools': 'Python extension compilation optimization',
        'Cython': 'Generate optimized code for target CPU',
        'Numba': 'Just-in-time compilation with CPU features',
        'JAX': 'XLA compilation targeting specific architectures'
    }
    
    for tool, description in integrations.items():
        print(f"  {tool}: {description}")

def show_package_summary():
    """Show archspec package capabilities summary"""
    
    print(f"\narchspec package capabilities:")
    
    capabilities = [
        "🔍 CPU architecture detection and identification",
        "🎯 Optimization target selection for compilers",
        "🏗️  Build system integration (CMake, Spack, etc.)",
        "🚀 Scientific computing performance optimization",
        "🔧 Cross-platform compatibility analysis", 
        "📊 Feature set comparison across architectures",
        "🎨 Compiler flag generation for optimal performance",
        "📦 Package manager integration for HPC/science"
    ]
    
    for capability in capabilities:
        print(f"  {capability}")
    
    print(f"\nPrimary use cases:")
    use_cases = [
        "High-performance computing (HPC) optimization",
        "Scientific Python package compilation",
        "Cross-platform software distribution", 
        "Automated build system configuration",
        "Performance benchmarking and analysis",
        "Container image optimization for computing"
    ]
    
    for use_case in use_cases:
        print(f"  • {use_case}")

# Run all analysis functions
cpu = analyze_cpu_for_science()
recommend_optimizations(cpu)
show_archspec_integration()
show_package_summary()

print(f"\nExample 5 completed - archspec scientific computing analysis")
print(f"Platform: {platform.platform()}")
print(f"archspec enables CPU-aware scientific computing optimization")
```
