## Example 2: argon2_cffi functionality test 2

```python
from argon2 import PasswordHasher
from argon2.exceptions import VerifyMismatchError

# Example 2: Argon2 password verification
print("Example 2: Testing Argon2 password verification")

try:
    # Create hasher
    ph = PasswordHasher()
    
    # Hash and verify password
    pw = "secure_password_456"
    hash_result = ph.hash(pw)
    
    # Test correct verification: verify() returns True, or raises VerifyMismatchError
    assert ph.verify(hash_result, pw) is True, "PasswordHasher.verify did not return True for the right password"
    print("Correct password verified successfully")
    
    # Test incorrect verification
    try:
        ph.verify(hash_result, "wrong_password")
    except VerifyMismatchError as e:
        assert "does not match" in str(e), e
        print("Wrong password correctly rejected")
    else:
        raise AssertionError("PasswordHasher.verify accepted 'wrong_password' for the hash of a different password")
    
    print("Password verification working correctly")
    print("argon2_cffi example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"argon2_cffi example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 4: argon2_cffi functionality test 4

```python
from argon2 import PasswordHasher
from argon2.low_level import hash_secret_raw, Type
import os

# Example 4: Low-level Argon2 API and hash types
print("Example 4: Testing low-level Argon2 API")

try:
    # Use low-level API for more control
    password = b"low_level_test_password"
    salt = os.urandom(16)  # Generate random salt
    
    print("Low-level Argon2 API testing")
    print(f"Password length: {len(password)} bytes")
    print(f"Salt length: {len(salt)} bytes")
    
    # Hash using low-level API with Argon2id (recommended variant)
    raw_hash = hash_secret_raw(
        secret=password,
        salt=salt,
        time_cost=2,
        memory_cost=65536,
        parallelism=1,
        hash_len=32,
        type=Type.ID  # Argon2id variant
    )
    
    print(f"Raw hash generated: {len(raw_hash)} bytes")
    print(f"Hash (first 16 bytes): {raw_hash[:16].hex()}")
    
    # Compare different Argon2 variants
    variants = {
        'Argon2d': Type.D,   # Data-dependent (faster, vulnerable to side-channel)
        'Argon2i': Type.I,   # Data-independent (slower, side-channel resistant)  
        'Argon2id': Type.ID  # Hybrid (recommended for password hashing)
    }
    
    print("\nComparing Argon2 variants:")
    for variant_name, variant_type in variants.items():
        try:
            variant_hash = hash_secret_raw(
                secret=password,
                salt=salt,
                time_cost=1,  # Reduced for demo
                memory_cost=1024,
                parallelism=1,
                hash_len=16,
                type=variant_type
            )
            print(f"{variant_name}: {variant_hash[:8].hex()}...")
        except Exception as e:
            print(f"{variant_name}: Error - {type(e).__name__}")
            raise
    
    print("Low-level API functionality verified")
    print("argon2_cffi example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"argon2_cffi example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: argon2_cffi functionality test 5

```python
from argon2 import PasswordHasher
import importlib.metadata
import time

# Example 5: Argon2 package capabilities and security analysis
print("Example 5: Argon2 security features and best practices")

try:
    print("argon2_cffi provides secure password hashing with Argon2")
    
    # Package version from the distribution metadata
    print(f"argon2-cffi version: {importlib.metadata.version('argon2-cffi')}")
    
    # Demonstrate security features
    ph = PasswordHasher()
    
    # Test timing resistance (hashes should take consistent time)
    passwords = ["short", "medium_length_password", "very_long_password_with_many_characters"]
    
    print("\nTesting timing consistency:")
    for password in passwords:
        start_time = time.perf_counter()
        hash_result = ph.hash(password)
        end_time = time.perf_counter()
        
        hash_time = (end_time - start_time) * 1000  # Convert to milliseconds
        print(f"Password '{password[:8]}...' - Hash time: {hash_time:.2f}ms")
    
    # Show hash format and components
    sample_hash = ph.hash("demonstration_password")
    print(f"\nArgon2 hash format analysis:")
    print(f"Full hash: {sample_hash}")
    
    # Parse hash components
    if sample_hash.startswith('$argon2'):
        parts = sample_hash.split('$')
        if len(parts) >= 4:
            print(f"Algorithm: {parts[1]}")
            print(f"Parameters: {parts[2]}")
            print(f"Salt+Hash: {parts[3][:20]}...")
    
    # Security recommendations
    print(f"\nSecurity best practices:")
    recommendations = [
        "Use Argon2id variant (default) for password hashing",
        "Adjust parameters based on available memory and acceptable delay",
        "Store only hashes, never plain passwords",
        "Use verify() method for password checking", 
        "Consider upgrading hashes if parameters become outdated",
        "Generate unique salts automatically (default behavior)"
    ]
    
    for rec in recommendations:
        print(f"• {rec}")
    
    # Parameter guidance
    print(f"\nParameter guidance:")
    param_info = [
        "time_cost: Number of iterations (higher = slower, more secure)",
        "memory_cost: Memory usage in KiB (higher = more memory, more secure)",
        "parallelism: Number of threads (balance with system capabilities)",
        "hash_len: Output hash length (32 bytes recommended)",
        "salt_len: Salt length (16 bytes minimum recommended)"
    ]
    
    for info in param_info:
        print(f"• {info}")
    
    # Use cases
    print(f"\nCommon use cases:")
    use_cases = [
        "Web application user authentication",
        "API key derivation from master passwords",
        "Secure storage of sensitive credentials",
        "Password-based encryption key derivation",
        "Multi-factor authentication token storage"
    ]
    
    for use_case in use_cases:
        print(f"• {use_case}")
    
    print("\nargon2_cffi provides industry-standard password security")
    print("argon2_cffi example 5 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
    raise
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"argon2_cffi example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
