## Example 2: argon2_cffi_bindings functionality test 2

```python
import _argon2_cffi_bindings as argon2_cffi_bindings
from _argon2_cffi_bindings import ffi, lib

# Example 2: Argon2 constants and error codes
print("Example 2: Testing Argon2 constants and error codes")

try:
    # Check Argon2 return codes
    print("Argon2 return codes and constants:")
    
    # Common Argon2 constants
    constants = {}
    constant_names = [
        'ARGON2_OK', 'ARGON2_OUTPUT_PTR_NULL', 'ARGON2_OUTPUT_TOO_SHORT',
        'ARGON2_OUTPUT_TOO_LONG', 'ARGON2_PWD_TOO_SHORT', 'ARGON2_PWD_PTR_MISMATCH',
        'ARGON2_SALT_TOO_SHORT', 'ARGON2_SALT_PTR_MISMATCH', 'ARGON2_AD_TOO_SHORT',
        'ARGON2_AD_PTR_MISMATCH', 'ARGON2_SECRET_TOO_SHORT', 'ARGON2_SECRET_PTR_MISMATCH'
    ]
    
    for const_name in constant_names:
        if hasattr(lib, const_name):
            constants[const_name] = getattr(lib, const_name)
    
    print(f"Available constants: {len(constants)}")
    assert len(constants) == len(constant_names), sorted(set(constant_names) - set(constants))
    assert constants['ARGON2_OK'] == 0
    assert all(code < 0 for name, code in constants.items() if name != 'ARGON2_OK'), constants  # errors are negative
    for name, value in list(constants.items())[:6]:
        print(f"  {name}: {value}")
    
    # Check Argon2 variant types
    variant_types = {}
    type_names = ['Argon2_d', 'Argon2_i', 'Argon2_id']
    
    for type_name in type_names:
        if hasattr(lib, type_name):
            variant_types[type_name] = getattr(lib, type_name)
    
    print(f"\nArgon2 variant types:")
    for name, value in variant_types.items():
        print(f"  {name}: {value}")
    assert variant_types == {'Argon2_d': 0, 'Argon2_i': 1, 'Argon2_id': 2}, variant_types
    
    # Every error code has a readable message
    message = ffi.string(lib.argon2_error_message(lib.ARGON2_SALT_TOO_SHORT)).decode()
    print(f"\nMessage for ARGON2_SALT_TOO_SHORT: {message}")
    assert message == "Salt is too short", message
    assert lib.ARGON2_VERSION_NUMBER == lib.ARGON2_VERSION_13 == 0x13  # current Argon2 version 1.3
    
    # Check version information
    version_functions = [name for name in dir(lib) if 'version' in name.lower()]
    if version_functions:
        print(f"\nVersion functions available: {version_functions}")
    
    # Memory allocation constants
    memory_constants = [name for name in dir(lib) if 'memory' in name.lower() or 'mem' in name.lower()]
    if memory_constants:
        print(f"Memory-related constants: {len(memory_constants)}")
    
    print("Argon2 constants and error handling available")
    print("argon2_cffi_bindings example 2 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"argon2_cffi_bindings example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: argon2_cffi_bindings functionality test 3

```python
import _argon2_cffi_bindings as argon2_cffi_bindings
from _argon2_cffi_bindings import ffi, lib

# Example 3: CFFI memory management and data structures
print("Example 3: Testing CFFI memory management")

try:
    print("Testing CFFI memory operations for Argon2")
    
    # Test CFFI memory allocation
    test_data = b"test_password_data"
    
    # Allocate memory using CFFI
    buffer_size = len(test_data)
    c_buffer = ffi.new("uint8_t[]", buffer_size)
    
    print(f"Allocated C buffer: {buffer_size} bytes")
    print(f"Buffer type: {type(c_buffer)}")
    
    # Copy Python data to C buffer
    ffi.memmove(c_buffer, test_data, buffer_size)
    print("Data copied to C buffer successfully")
    
    # Read back from C buffer
    read_back = ffi.buffer(c_buffer, buffer_size)[:]
    print(f"Data read back matches: {read_back == test_data}")
    assert read_back == test_data
    
    # Test creating different data types
    data_types = [
        ("uint8_t[]", 32, "8-bit unsigned array"),
        ("uint32_t[]", 8, "32-bit unsigned array"),
        ("char[]", 64, "character array")
    ]
    
    print(f"\nTesting CFFI data type allocations:")
    for ctype, size, description in data_types:
        try:
            test_buffer = ffi.new(ctype, size)
            assert len(test_buffer) == size and ffi.sizeof(test_buffer) == size * ffi.sizeof(ctype[:-2])
            print(f"  ✓ {description}: allocated {size} elements")
        except Exception as e:
            print(f"  ✗ {description}: {type(e).__name__}")
            raise
    
    # Test string conversion
    test_string = "argon2_test_string"
    c_string = ffi.new("char[]", test_string.encode('utf-8'))
    print(f"String to C conversion: {len(test_string)} chars")
    
    # Convert back
    back_to_python = ffi.string(c_string).decode('utf-8')
    print(f"C to Python conversion matches: {back_to_python == test_string}")
    assert back_to_python == test_string
    
    # Check FFI capabilities
    ffi_methods = [attr for attr in dir(ffi) if not attr.startswith('_')]
    print(f"\nFFI methods available: {len(ffi_methods)}")
    print(f"Key methods: {ffi_methods[:8]}")
    
    print("CFFI memory management working correctly")
    print("argon2_cffi_bindings example 3 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"argon2_cffi_bindings example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: argon2_cffi_bindings functionality test 4

```python
from _argon2_cffi_bindings import ffi, lib

# Example 4: Hash and verify a password through the Argon2 C functions
print("Example 4: Hashing and verifying with the Argon2 C API")

try:
    # Parameters of the reference Argon2id test vector: 2 passes, 65536 KiB of memory, parallelism 1, 32-byte tag
    password, salt = b"password", b"somesalt"
    t_cost, m_cost, parallelism, hash_len = 2, 1 << 16, 1, 32
    
    # Raw hash
    raw = ffi.new("uint8_t[]", hash_len)
    rc = lib.argon2_hash(t_cost, m_cost, parallelism, password, len(password), salt, len(salt),
                         raw, hash_len, ffi.NULL, 0, lib.Argon2_id, lib.ARGON2_VERSION_13)
    assert rc == lib.ARGON2_OK, ffi.string(lib.argon2_error_message(rc))
    raw_hex = bytes(ffi.buffer(raw, hash_len)).hex()
    print(f"Raw Argon2id hash: {raw_hex}")
    # Known answer from the reference implementation's test suite (Argon2id, version 0x13)
    assert raw_hex == "09316115d5cf24ed5a15a31a3ba326e5cf32edc24702987c02b6566f61913cf7", raw_hex
    print("✓ Raw hash matches the reference test vector")
    
    # Encoded hash: the PHC string format that stores the parameters and the salt with the hash
    encoded_len = lib.argon2_encodedlen(t_cost, m_cost, parallelism, len(salt), hash_len, lib.Argon2_id)
    encoded = ffi.new("char[]", encoded_len)
    rc = lib.argon2_hash(t_cost, m_cost, parallelism, password, len(password), salt, len(salt),
                         ffi.NULL, hash_len, encoded, encoded_len, lib.Argon2_id, lib.ARGON2_VERSION_13)
    assert rc == lib.ARGON2_OK, ffi.string(lib.argon2_error_message(rc))
    encoded_str = ffi.string(encoded).decode()
    print(f"Encoded hash: {encoded_str}")
    assert encoded_str == "$argon2id$v=19$m=65536,t=2,p=1$c29tZXNhbHQ$CTFhFdXPJO1aFaMaO6Mm5c8y7cJHAph8ArZWb2GRPPc", encoded_str
    
    # Verification: the right password matches, a wrong one does not
    assert lib.argon2_verify(encoded, password, len(password), lib.Argon2_id) == lib.ARGON2_OK
    wrong = b"passw0rd"
    rc = lib.argon2_verify(encoded, wrong, len(wrong), lib.Argon2_id)
    assert rc == lib.ARGON2_VERIFY_MISMATCH, rc
    print(f"✓ Correct password verifies; wrong password gives: {ffi.string(lib.argon2_error_message(rc)).decode()}")
    
    # Parameter validation: a salt shorter than 8 bytes is rejected before any hashing
    rc = lib.argon2_hash(t_cost, 8, 1, password, len(password), b"short", 5,
                         raw, hash_len, ffi.NULL, 0, lib.Argon2_id, lib.ARGON2_VERSION_13)
    assert rc == lib.ARGON2_SALT_TOO_SHORT, rc
    print(f"✓ 5-byte salt rejected: {ffi.string(lib.argon2_error_message(rc)).decode()}")
    
    print("argon2_cffi_bindings example 4 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"argon2_cffi_bindings example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```
