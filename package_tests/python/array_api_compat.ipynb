```python
# Example 1: is_*_array detection helpers from the top-level API.
import numpy as np
import array_api_compat

arr = np.arange(6).reshape(2, 3)

print("is_numpy_array:", array_api_compat.is_numpy_array(arr))
print("is_array_api_obj:", array_api_compat.is_array_api_obj(arr))
print("is_torch_array:", array_api_compat.is_torch_array(arr))
print("is_dask_array:", array_api_compat.is_dask_array(arr))

# A plain Python list should not be detected as any array.
print("list is_array_api_obj:", array_api_compat.is_array_api_obj([1, 2, 3]))

assert array_api_compat.is_numpy_array(arr) is True
assert array_api_compat.is_torch_array(arr) is False
```

```python
# Example 2: array_namespace() returns the compat-wrapped numpy namespace.
import numpy as np
from array_api_compat import array_namespace

x = np.asarray([1.0, 2.0, 3.0, 4.0])
y = np.asarray([10.0, 20.0, 30.0, 40.0])

xp = array_namespace(x, y)
print("namespace module:", xp.__name__)

# Use the namespace via the array API spec — should resolve to numpy operations.
result = xp.add(x, y)
print("add result:", result)

assert xp.__name__.endswith("numpy") or "numpy" in xp.__name__
assert np.allclose(result, np.array([11.0, 22.0, 33.0, 44.0]))
```

```python
# Example 3: numpy compat shim exposes array API spec functions like permute_dims.
import numpy as np
import array_api_compat.numpy as xnp

a = np.arange(12).reshape(3, 4)

# permute_dims is the array API spec name for transpose with explicit axes.
permuted = xnp.permute_dims(a, (1, 0))
print("original shape:", a.shape)
print("permuted shape:", permuted.shape)

assert permuted.shape == (4, 3)
assert np.array_equal(permuted, a.T)
```

```python
# Example 4: device + to_device helpers for cpu numpy arrays.
import numpy as np
from array_api_compat import device, to_device

a = np.asarray([1, 2, 3, 4, 5])

dev = device(a)
print("device:", dev)

# Round-trip to the same device — numpy only has "cpu".
moved = to_device(a, dev)
print("moved dtype/shape:", moved.dtype, moved.shape)

assert str(dev) == "cpu"
assert np.array_equal(moved, a)
```

```python
# Example 5: size() helper and is_*_namespace introspection.
import numpy as np
from array_api_compat import size, is_numpy_namespace, array_namespace

a = np.zeros((2, 3, 4))
n = size(a)
print("size:", n)

xp = array_namespace(a)
print("is_numpy_namespace:", is_numpy_namespace(xp))

assert n == 24
assert is_numpy_namespace(xp) is True
```
