```python
# Example 1: atleast_nd pads leading axes and keeps the data.
import numpy as np
import array_api_extra as xpx

x = np.asarray([1.0, 2.0, 3.0])
y = xpx.atleast_nd(x, ndim=3)
print("shape:", y.shape)

assert y.shape == (1, 1, 3)
assert np.array_equal(y[0, 0], x)
```

```python
# Example 2: at() does indexed updates, the array-API way to write x[i] = v.
# copy=True asks for a new array; the default (copy=None) may update a writeable NumPy array in place.
import numpy as np
import array_api_extra as xpx

x = np.zeros(5)
y = xpx.at(x)[1:3].set(7.0, copy=True)
z = xpx.at(y, 0).add(2.0, copy=True)
print("x:", x, "y:", y, "z:", z)

assert np.array_equal(x, np.zeros(5)), "copy=True must leave the input unchanged"
assert np.array_equal(y, np.asarray([0.0, 7.0, 7.0, 0.0, 0.0]))
assert np.array_equal(z, np.asarray([2.0, 7.0, 7.0, 0.0, 0.0]))
```

```python
# Example 3: apply_where evaluates each function only where its condition holds.
import numpy as np
import array_api_extra as xpx

x = np.asarray([-2.0, -1.0, 0.0, 4.0])
y = xpx.apply_where(x > 0, (x,), np.sqrt, fill_value=-1.0)
print("result:", y)

assert np.array_equal(y, np.asarray([-1.0, -1.0, -1.0, 2.0]))
```

```python
# Example 4: one_hot and nunique on integer labels.
import numpy as np
import array_api_extra as xpx

labels = np.asarray([0, 2, 1, 2])
encoded = xpx.one_hot(labels, 3)
print(encoded)

assert encoded.shape == (4, 3)
assert np.array_equal(encoded.argmax(axis=1), labels)
assert int(xpx.nunique(labels)) == 3
```

```python
# Example 5: cov, kron and isclose against NumPy's own results.
import numpy as np
import array_api_extra as xpx

rng = np.random.default_rng(0)
m = rng.random((3, 20))
c = xpx.cov(m)
k = xpx.kron(np.eye(2), np.ones((2, 2)))
print("cov shape:", c.shape, "kron shape:", k.shape)

assert np.allclose(c, np.cov(m))
assert np.array_equal(k, np.kron(np.eye(2), np.ones((2, 2))))
assert bool(np.all(xpx.isclose(np.asarray([1.0, 2.0]), np.asarray([1.0, 2.0 + 1e-12]))))
```
