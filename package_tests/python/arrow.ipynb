# arrow Package Testing

This notebook tests the arrow package - Better dates and times for Python.

## Example 1: Basic Arrow date/time creation and formatting


```python
import arrow
from datetime import datetime, timedelta, timezone

# Example 1: Basic Arrow date/time creation and formatting
print("Example 1: Basic Arrow functionality")

try:
    # Create Arrow objects in different ways
    now = arrow.now()
    utc_now = arrow.utcnow()
    custom_date = arrow.get('2023-12-25 15:30:00', 'YYYY-MM-DD HH:mm:ss')
    
    print(f"Current local time: {now}")
    print(f"Current UTC time: {utc_now}")
    print(f"Custom date: {custom_date}")
    
    # Format dates
    formatted = now.format('YYYY-MM-DD HH:mm:ss')
    humanized = now.humanize()
    
    print(f"Formatted: {formatted}")
    print(f"Humanized: {humanized}")
    
    assert custom_date.datetime == datetime(2023, 12, 25, 15, 30, tzinfo=timezone.utc), custom_date
    assert utc_now.utcoffset() == timedelta(0), utc_now
    assert abs((now - utc_now).total_seconds()) < 60  # the same instant, in two time zones
    assert arrow.get(formatted, 'YYYY-MM-DD HH:mm:ss').naive == now.floor('second').naive
    assert humanized == "just now", humanized
    
    print("Arrow basic functionality test completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"arrow example (cell 2) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished")
```

## Example 2: Timezone handling and conversion


```python
import arrow

# Example 2: Timezone handling and conversion
print("Example 2: Timezone operations")

try:
    # Create dates in different timezones
    utc_time = arrow.utcnow()
    ny_time = utc_time.to('America/New_York')
    tokyo_time = utc_time.to('Asia/Tokyo')
    london_time = utc_time.to('Europe/London')
    
    print(f"UTC time: {utc_time}")
    print(f"New York time: {ny_time}")
    print(f"Tokyo time: {tokyo_time}")
    print(f"London time: {london_time}")
    
    # Convert between timezones
    local_to_utc = arrow.now().to('UTC')
    print(f"Local to UTC: {local_to_utc}")
    assert ny_time == utc_time == tokyo_time == london_time  # one instant, different wall clocks
    
    # Fixed instants make the offsets exact, including daylight saving time
    summer = arrow.get('2023-07-01T12:00:00+00:00')
    winter = arrow.get('2023-01-15T12:00:00+00:00')
    assert summer.to('America/New_York').isoformat() == '2023-07-01T08:00:00-04:00'
    assert summer.to('Asia/Tokyo').isoformat() == '2023-07-01T21:00:00+09:00'
    assert summer.to('Europe/London').isoformat() == '2023-07-01T13:00:00+01:00'
    assert winter.to('America/New_York').isoformat() == '2023-01-15T07:00:00-05:00'
    assert winter.to('Europe/London').isoformat() == '2023-01-15T12:00:00+00:00'
    print("✓ New York, Tokyo and London offsets are correct in summer and winter")
    
    print("Timezone operations completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"arrow example (cell 4) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished")
```

## Example 3: Date arithmetic and manipulation


```python
import arrow

# Example 3: Date arithmetic and manipulation
print("Example 3: Date arithmetic")

try:
    # Start with a base date
    base_date = arrow.get('2023-06-15 12:00:00')
    print(f"Base date: {base_date}")
    
    # Add/subtract time units
    future_date = base_date.shift(days=7, hours=3, minutes=30)
    past_date = base_date.shift(months=-2, weeks=-1)
    
    print(f"Future date (+7 days, +3 hours, +30 min): {future_date}")
    print(f"Past date (-2 months, -1 week): {past_date}")
    
    # Get start/end of periods
    start_of_day = base_date.floor('day')
    end_of_month = base_date.ceil('month')
    
    print(f"Start of day: {start_of_day}")
    print(f"End of month: {end_of_month}")
    
    assert future_date == arrow.get('2023-06-22T15:30:00+00:00'), future_date
    assert past_date == arrow.get('2023-04-08T12:00:00+00:00'), past_date
    assert start_of_day == arrow.get('2023-06-15T00:00:00+00:00'), start_of_day
    assert end_of_month == arrow.get('2023-06-30T23:59:59.999999+00:00'), end_of_month
    
    print("Date arithmetic completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"arrow example (cell 6) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished")
```

## Example 4: Parsing various date formats


```python
import arrow

# Example 4: Parsing various date formats
print("Example 4: Date parsing")

try:
    # Parse different date string formats
    formats = [
        ('2023-12-25', 'YYYY-MM-DD'),
        ('25/12/2023 15:30', 'DD/MM/YYYY HH:mm'),
        ('Dec 25, 2023', 'MMM DD, YYYY'),
        ('2023-12-25T15:30:00Z', None)  # ISO format, no format needed
    ]
    
    expected = [
        '2023-12-25T00:00:00+00:00',
        '2023-12-25T15:30:00+00:00',
        '2023-12-25T00:00:00+00:00',
        '2023-12-25T15:30:00+00:00',
    ]
    for (date_str, fmt), want in zip(formats, expected):
        if fmt:
            parsed = arrow.get(date_str, fmt)
        else:
            parsed = arrow.get(date_str)
        
        print(f"'{date_str}' -> {parsed}")
        assert parsed.isoformat() == want, (date_str, parsed)
    
    # Parse timestamp
    timestamp = arrow.get(1703520600)  # Unix timestamp
    print(f"Unix timestamp 1703520600 -> {timestamp}")
    assert timestamp.isoformat() == '2023-12-25T16:10:00+00:00', timestamp
    assert timestamp.int_timestamp == 1703520600
    
    print("Date parsing completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"arrow example (cell 8) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished")
```

## Example 5: Humanized dates and ranges


```python
import arrow

# Example 5: Humanized dates and ranges
print("Example 5: Humanized dates and ranges")

try:
    # A fixed reference instant keeps the wording the same whatever day the notebook runs
    now = arrow.get('2023-08-15T12:00:00+00:00')
    
    # Create various time differences
    times = {
        'hour_ago': now.shift(hours=-1),
        'day_ago': now.shift(days=-1),
        'week_ago': now.shift(weeks=-1),
        'month_ago': now.shift(months=-1),
        'in_2_hours': now.shift(hours=2),
        'tomorrow': now.shift(days=1)
    }
    
    # Show humanized versions, relative to the reference instant
    expected = {
        'hour_ago': 'an hour ago',
        'day_ago': 'a day ago',
        'week_ago': 'a week ago',
        'month_ago': 'a month ago',
        'in_2_hours': 'in 2 hours',
        'tomorrow': 'in a day'
    }
    for label, time_obj in times.items():
        humanized = time_obj.humanize(now)
        print(f"{label}: {humanized}")
        assert humanized == expected[label], (label, humanized)
    
    # Generate date ranges
    start = arrow.get('2023-12-01')
    end = arrow.get('2023-12-07')
    
    print("\nDaily range:")
    days = [day.format('YYYY-MM-DD dddd') for day in arrow.Arrow.range('day', start, end)]
    for day in days:
        print(f"  {day}")
    assert len(days) == 7, days  # both ends included
    assert days[0] == '2023-12-01 Friday' and days[-1] == '2023-12-07 Thursday', days
    
    print("Humanized dates and ranges completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"arrow example (cell 10) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished")
```
