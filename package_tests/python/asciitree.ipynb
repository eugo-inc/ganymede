## Example 3: asciitree functionality test 3

```python
from asciitree import LeftAligned, draw_tree
import asciitree

# Example 3: Organization chart tree
print("Example 3: Testing organization chart visualization")

try:
    # Create an organizational structure
    organization = {
        'CEO': {
            'CTO': {
                'Dev Manager': {
                    'Senior Developer': {},
                    'Junior Developer': {},
                    'Intern': {}
                },
                'QA Manager': {
                    'QA Engineer 1': {},
                    'QA Engineer 2': {}
                }
            },
            'CFO': {
                'Accounting': {
                    'Accountant': {},
                    'Bookkeeper': {}
                },
                'Finance': {
                    'Financial Analyst': {}
                }
            },
            'CMO': {
                'Marketing': {
                    'Marketing Specialist': {},
                    'Content Writer': {}
                },
                'Sales': {
                    'Sales Rep 1': {},
                    'Sales Rep 2': {}
                }
            }
        }
    }
    
    # Create renderer with default settings
    tr = LeftAligned()
    org_chart = tr(organization)
    
    print("Organization chart:")
    print(org_chart)
    
    # Alternative drawing method: the legacy asciitree-0.2 `draw_tree(node, child_iter, text_str)` walks NODE
    # objects, not a dict, so the dict tree is handed over as (name, children) pairs
    print("\nAlternative drawing method:")
    alt_output = draw_tree(
        next(iter(organization.items())),
        child_iter=lambda node: list(node[1].items()),
        text_str=lambda node: node[0],
    )
    print(alt_output[:200] + "..." if len(str(alt_output)) > 200 else alt_output)
    # same tree, same nodes: one line per node in both renderings, rooted at CEO
    assert alt_output.splitlines()[0] == 'CEO' and len(alt_output.splitlines()) == len(org_chart.splitlines()), alt_output
    
    print("Organization chart visualization working")
    print("asciitree example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"asciitree example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: asciitree functionality test 4

```python
from asciitree import LeftAligned
import asciitree

# Example 4: Custom tree formatting and styling
print("Example 4: Testing custom tree formatting")

try:
    # Create a technical architecture tree
    architecture = {
        'Web Application': {
            'Frontend': {
                'React Components': {
                    'Header': {},
                    'Navigation': {},
                    'Content': {
                        'Dashboard': {},
                        'Settings': {}
                    }
                },
                'State Management': {
                    'Redux Store': {},
                    'Actions': {},
                    'Reducers': {}
                }
            },
            'Backend': {
                'API Layer': {
                    'REST Endpoints': {},
                    'Authentication': {},
                    'Middleware': {}
                },
                'Database': {
                    'PostgreSQL': {
                        'Users Table': {},
                        'Products Table': {},
                        'Orders Table': {}
                    }
                }
            },
            'Infrastructure': {
                'Docker Containers': {},
                'Load Balancer': {},
                'CDN': {}
            }
        }
    }
    
    # Test different rendering approaches
    tr = LeftAligned()
    
    # Basic rendering
    arch_tree = tr(architecture)
    print("Application architecture:")
    print(arch_tree)
    
    # Test if custom formatting is available
    try:
        # Try custom indentation if supported
        tr_custom = LeftAligned(indent=4, childstr='├── ', laststr='└── ')
        custom_tree = tr_custom(architecture)
        print("\nCustom formatted tree (if supported):")
        print(custom_tree[:300] + "..." if len(custom_tree) > 300 else custom_tree)
    except Exception as e:
        print(f"Custom formatting not available: {type(e).__name__}")
        raise
    
    print("Custom formatting demonstration completed")
    print("asciitree example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"asciitree example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```
