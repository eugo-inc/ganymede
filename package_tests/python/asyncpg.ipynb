## Example 6: asyncpg protocol round trip against an in-process fake server

```python
import asyncio, struct, asyncpg

def msg(kind, body=b""):
    return kind + struct.pack("!i", len(body) + 4) + body

async def serve(reader, writer):
    while True:  # SSLRequest / GSSENCRequest -> 'N' (no encryption); the next packet is the StartupMessage
        packet = await reader.readexactly(struct.unpack("!i", await reader.readexactly(4))[0] - 4)
        if struct.unpack("!i", packet[:4])[0] not in (80877103, 80877104):
            break
        writer.write(b"N")
    status = (b"server_version\x0016.4", b"client_encoding\x00UTF8", b"integer_datetimes\x00on")
    writer.write(msg(b"R", struct.pack("!i", 0)) + b"".join(msg(b"S", s + b"\0") for s in status)
                 + msg(b"K", struct.pack("!ii", 1, 2)) + msg(b"Z", b"I"))  # AuthOK, params, key, Ready
    int4_column = struct.pack("!h", 1) + b"x\0" + struct.pack("!ihihih", 0, 0, 23, 4, -1, 0)  # oid 23 = int4
    replies = {b"P": msg(b"1"), b"B": msg(b"2"), b"S": msg(b"Z", b"I"),  # Parse / Bind / Sync
               b"D": msg(b"t", struct.pack("!h", 0)) + msg(b"T", int4_column),  # Describe: 0 params, 1 column
               b"E": msg(b"D", struct.pack("!hii", 1, 4, 42)) + msg(b"C", b"SELECT 1\0")}  # Execute: one row
    while (kind := await reader.readexactly(1)) != b"X":  # X = Terminate; a message with no reply gets none
        await reader.readexactly(struct.unpack("!i", await reader.readexactly(4))[0] - 4)
        writer.write(replies.get(kind, b""))
    writer.close()

server = await asyncio.start_server(serve, "127.0.0.1", 0)
conn = await asyncpg.connect(host="127.0.0.1", port=server.sockets[0].getsockname()[1],
                             user="u", password="p", database="d", timeout=10, command_timeout=10)
try:
    assert (value := await conn.fetchval("SELECT 42::int4")) == 42, f"fetchval returned {value!r}, not 42"
    record = await conn.fetchrow("SELECT 42::int4")
    assert record["x"] == 42 and record.get("x") == 42 and dict(record.items()) == {"x": 42}, record
    # upstream #1333: Record.get() with no key must raise TypeError. asyncpg 0.31.x lacks that check
    # (fixed in 0.32.0): the call returns a bogus value or can crash the interpreter, so it is skipped there.
    if asyncpg.__version__.startswith("0.31."):
        print(f"skipped the key-less Record.get() probe: asyncpg {asyncpg.__version__} predates the upstream fix")
    else:
        try:
            record.get()
        except TypeError:
            pass
        else:
            raise AssertionError("Record.get() with no key did not raise TypeError")
finally:
    await conn.close()
    server.close()
print(f"✓ asyncpg {asyncpg.__version__} v3 round trip: fetchval -> 42, fetchrow -> {record!r}")
```

## Example 7: asyncpg maps a rejected password (SQLSTATE 28P01) to InvalidPasswordError

```python
# Example 7: the server rejects the password during startup (ErrorResponse, SQLSTATE 28P01). asyncpg must
# decode the ErrorResponse and raise its typed exception, not a generic error and not a hang.
import asyncio, struct, asyncpg

attempts = []

async def reject(reader, writer):
    attempts.append(1)
    while True:  # SSLRequest / GSSENCRequest -> 'N'; the StartupMessage is answered with a FATAL 28P01
        packet = await reader.readexactly(struct.unpack("!i", await reader.readexactly(4))[0] - 4)
        if struct.unpack("!i", packet[:4])[0] not in (80877103, 80877104):
            break
        writer.write(b"N")
    fields = b'SFATAL\0VFATAL\0C28P01\0Mpassword authentication failed for user "u"\0\0'
    writer.write(b"E" + struct.pack("!i", len(fields) + 4) + fields)
    writer.close()

server = await asyncio.start_server(reject, "127.0.0.1", 0)
try:
    await asyncpg.connect(host="127.0.0.1", port=server.sockets[0].getsockname()[1],
                          user="u", password="p", database="d", timeout=10)
except asyncpg.InvalidPasswordError as error:
    assert error.sqlstate == "28P01", error.sqlstate
    print(f"✓ 28P01 -> {type(error).__name__}: {error} ({len(attempts)} connection attempt(s))")
else:
    raise AssertionError("asyncpg.connect() succeeded against a server that rejected the password")
finally:
    server.close()
```
