# attrs: 5 Basic Usage Examples

```python
# Example 1: Basic class with attrs
import attr
@attr.s
class Point:
    x = attr.ib()
    y = attr.ib()
p = Point(1, 2)
print(p)
assert (p.x, p.y) == (1, 2) and repr(p) == "Point(x=1, y=2)"
assert p == Point(1, 2) and p != Point(2, 1)  # generated __eq__ compares the fields
assert attr.asdict(p) == {"x": 1, "y": 2}
```

```python
# Example 2: Default values
import attr
@attr.s
class User:
    name = attr.ib(default='anonymous')
    age = attr.ib(default=0)
u = User()
print(u)
assert (u.name, u.age) == ("anonymous", 0)
assert User("ada", 36).name == "ada" and User(age=5).age == 5
```

```python
# Example 3: Validators
import attr
@attr.s
class Product:
    price = attr.ib(validator=attr.validators.instance_of(float))
p = Product(9.99)
print(p)
assert p.price == 9.99
try:
    Product("9.99")  # a string is not a float
except TypeError as error:
    print(f"Validator rejected a str: {error.args[0]}")
else:
    raise AssertionError("the instance_of(float) validator accepted a str")
```

```python
# Example 4: Converters
import attr
@attr.s
class Item:
    count = attr.ib(converter=int)
i = Item('5')
print(i)
assert i.count == 5 and type(i.count) is int  # the converter ran on the way in
```

```python
# Example 5: Frozen classes (immutable)
import attr
@attr.s(frozen=True)
class Immutable:
    value = attr.ib()
obj = Immutable(10)
print(obj)
try:
    obj.value = 20
except attr.exceptions.FrozenInstanceError:
    print("Assignment to a frozen instance refused")
else:
    raise AssertionError("a frozen instance accepted an assignment")
assert obj.value == 10
changed = attr.evolve(obj, value=20)  # frozen instances are copied with changes instead
assert changed.value == 20 and obj.value == 10
assert hash(obj) == hash(Immutable(10))  # frozen and comparable, so hashable
```
