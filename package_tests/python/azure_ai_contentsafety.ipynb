```python
# Example 1: Top-level public API surface.
from azure.ai.contentsafety import ContentSafetyClient, BlocklistClient
from azure.core.credentials import AzureKeyCredential

# Construct clients with a dummy credential (no network call until a method is invoked).
endpoint = "https://example.cognitiveservices.azure.com/"
credential = AzureKeyCredential("dummy-key-not-used")

cs_client = ContentSafetyClient(endpoint=endpoint, credential=credential)
bl_client = BlocklistClient(endpoint=endpoint, credential=credential)

print(f"ContentSafetyClient: {type(cs_client).__name__}")
print(f"BlocklistClient: {type(bl_client).__name__}")
assert cs_client is not None
assert bl_client is not None
```

```python
# Example 2: Request models — AnalyzeTextOptions construction & serialization.
from azure.ai.contentsafety.models import AnalyzeTextOptions, TextCategory

options = AnalyzeTextOptions(
    text="This is a benign sentence used purely for shape testing.",
    categories=[TextCategory.HATE, TextCategory.VIOLENCE],
)

print(f"text attr: {options.text[:40]}...")
print(f"categories: {options.categories}")

# Models support dict-like serialization for the REST envelope.
serialized = options.as_dict() if hasattr(options, 'as_dict') else dict(options)
print(f"serialized keys: {sorted(serialized.keys())}")
assert 'text' in serialized
assert 'categories' in serialized
```

```python
# Example 3: Enums — TextCategory and ImageCategory exposed as string-valued enums.
from azure.ai.contentsafety.models import TextCategory, ImageCategory

text_cats = {c.name: c.value for c in TextCategory}
image_cats = {c.name: c.value for c in ImageCategory}

print("TextCategory values:")
for name, val in text_cats.items():
    print(f"  {name} = {val!r}")

print("ImageCategory values:")
for name, val in image_cats.items():
    print(f"  {name} = {val!r}")

# Spot-check the canonical four moderation categories.
assert 'HATE' in text_cats
assert 'VIOLENCE' in text_cats
assert 'SELF_HARM' in text_cats
assert 'SEXUAL' in text_cats
```

```python
# Example 5: Blocklist models — TextBlocklist + TextBlocklistItem construction.
from azure.ai.contentsafety.models import TextBlocklist, TextBlocklistItem

blocklist = TextBlocklist(
    blocklist_name="test-blocklist",
    description="A blocklist used for shape-testing the SDK models.",
)

item = TextBlocklistItem(
    text="forbidden-phrase",
    description="sample blocklist entry",
)

print(f"blocklist name: {blocklist.blocklist_name}")
print(f"blocklist description: {blocklist.description}")
print(f"item text: {item.text}")
print(f"item description: {item.description}")

assert blocklist.blocklist_name == "test-blocklist"
assert item.text == "forbidden-phrase"
print("\nazure.ai.contentsafety smoke test PASSED")
```
