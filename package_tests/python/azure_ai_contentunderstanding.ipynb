# azure_ai_contentunderstanding smoke test

Validates that `azure.ai.contentunderstanding` imports cleanly and exposes its primary client API.

```python
# Example 1: top-level public API surface
import azure.ai.contentunderstanding as acu

public_names = [n for n in dir(acu) if not n.startswith('_')]
print(f"public names: {len(public_names)}")
for name in public_names[:25]:
    obj = getattr(acu, name)
    print(f"  {name}: {type(obj).__name__}")

assert len(public_names) > 0, "package exposes no public names"
```

```python
# Example 2: ContentUnderstandingClient class import + signature inspection
import inspect

from azure.ai.contentunderstanding import ContentUnderstandingClient

assert inspect.isclass(ContentUnderstandingClient)

sig = inspect.signature(ContentUnderstandingClient.__init__)
params = list(sig.parameters)
print(f"ContentUnderstandingClient.__init__ params: {params}")

# The client takes the service endpoint and a key or token credential
assert params[:3] == ['self', 'endpoint', 'credential'], params
```

```python
# Example 3: models subpackage exposes enums + dataclasses used in requests/responses
from azure.ai.contentunderstanding import models

model_names = [n for n in dir(models) if not n.startswith('_')]
print(f"models.* exposes {len(model_names)} names")
for name in model_names[:15]:
    print(f"  {name}")

assert len(model_names) > 0, "models subpackage looks empty"
```

```python
# Example 4: aio (async) client variant is importable
import importlib

aio_mod = importlib.import_module('azure.ai.contentunderstanding.aio')
AsyncClient = getattr(aio_mod, 'ContentUnderstandingClient', None)

print(f"aio module: {aio_mod}")
print(f"aio.ContentUnderstandingClient: {AsyncClient}")

assert AsyncClient is not None, "async client missing from .aio"
```

```python
# Example 5: client instantiation with a fake credential (no network call)
from azure.core.credentials import AzureKeyCredential

from azure.ai.contentunderstanding import ContentUnderstandingClient

endpoint = "https://fake.cognitiveservices.azure.com/"
credential = AzureKeyCredential("fake-key-for-smoke-test")

client = ContentUnderstandingClient(endpoint=endpoint, credential=credential)

print(f"client = {client!r}")
print(f"type = {type(client).__name__}")

assert client is not None
assert type(client).__name__ == "ContentUnderstandingClient"
```
