```python
# Example 1: import the main client class + verify it's a class with expected signature.
import inspect

from azure.ai.documentintelligence import DocumentIntelligenceClient

assert inspect.isclass(DocumentIntelligenceClient)
sig = inspect.signature(DocumentIntelligenceClient.__init__)
params = list(sig.parameters)
print(f"DocumentIntelligenceClient.__init__ params: {params}")
assert 'endpoint' in params
assert 'credential' in params
print(f"DocumentIntelligenceClient MRO: {[c.__name__ for c in DocumentIntelligenceClient.__mro__]}")
```

```python
# Example 2: instantiate the client with a fake key credential — no network call yet.
from azure.ai.documentintelligence import DocumentIntelligenceClient
from azure.core.credentials import AzureKeyCredential

endpoint = "https://fake-resource.cognitiveservices.azure.com/"
credential = AzureKeyCredential("fake-key-not-real")

client = DocumentIntelligenceClient(endpoint=endpoint, credential=credential)
print(f"client type: {type(client).__name__}")
print(f"client endpoint: {client._config.endpoint}")

# Verify common methods exist on the client (signatures only).
for method_name in ('begin_analyze_document', 'begin_classify_document'):
    assert hasattr(client, method_name), f"missing method: {method_name}"
    print(f"  has method: {method_name}")

client.close()
```

```python
# Example 3: the administration client variant (model build / management).
import inspect

from azure.ai.documentintelligence import DocumentIntelligenceAdministrationClient
from azure.core.credentials import AzureKeyCredential

assert inspect.isclass(DocumentIntelligenceAdministrationClient)
admin = DocumentIntelligenceAdministrationClient(
    endpoint="https://fake-resource.cognitiveservices.azure.com/",
    credential=AzureKeyCredential("fake-key"),
)
print(f"admin client type: {type(admin).__name__}")

# GA (1.0.x) method names: list_document_models was renamed to list_models.
for method_name in ('begin_build_document_model', 'list_models', 'get_model', 'get_resource_details'):
    assert hasattr(admin, method_name), f"missing method: {method_name}"
    print(f"  has method: {method_name}")

admin.close()
```

```python
# Example 4: explore the models module — request/response dataclasses generated from the spec.
from azure.ai.documentintelligence import models as adi_models

expected_models = (
    'AnalyzeDocumentRequest',
    'AnalyzeResult',
    'DocumentAnalysisFeature',
    'DocumentField',
    'AnalyzeOutputOption',
)
for name in expected_models:
    assert hasattr(adi_models, name), f"missing model: {name}"
    obj = getattr(adi_models, name)
    print(f"  {name}: {type(obj).__name__}")

# Instantiate a simple request body to confirm the model layer round-trips.
req = adi_models.AnalyzeDocumentRequest(url_source="https://example.com/sample.pdf")
as_dict = req.as_dict()
print(f"AnalyzeDocumentRequest.as_dict(): {as_dict}")
assert as_dict.get("urlSource") == "https://example.com/sample.pdf"
```

```python
# Example 5: async client signature and package version.
import inspect

from azure.ai.documentintelligence.aio import DocumentIntelligenceClient as AsyncClient
from azure.ai.documentintelligence import _version

assert inspect.isclass(AsyncClient)
async_sig = inspect.signature(AsyncClient.__init__)
assert 'endpoint' in async_sig.parameters
assert 'credential' in async_sig.parameters
print(f"AsyncClient signature: {async_sig}")

version = _version.VERSION
print(f"package VERSION: {version}")
assert version.startswith("1."), f"unexpected major version: {version}"
print("PASS: azure_ai_documentintelligence smoke test complete")
```
