```python
# Example 1: Verify primary client classes are importable from top-level.
from azure.ai.inference import (
    ChatCompletionsClient,
    EmbeddingsClient,
    ImageEmbeddingsClient,
)

for cls in (ChatCompletionsClient, EmbeddingsClient, ImageEmbeddingsClient):
    print(f"{cls.__name__}: {cls.__module__}")
    assert callable(cls)

assert ChatCompletionsClient.__name__ == "ChatCompletionsClient"
assert EmbeddingsClient.__name__ == "EmbeddingsClient"
assert ImageEmbeddingsClient.__name__ == "ImageEmbeddingsClient"
```

```python
# Example 2: Construct message model objects (no network calls).
from azure.ai.inference.models import (
    SystemMessage,
    UserMessage,
    AssistantMessage,
)

sys_msg = SystemMessage(content="You are a helpful assistant.")
user_msg = UserMessage(content="Hello, world!")
asst_msg = AssistantMessage(content="Hi there!")

print(f"system: role={sys_msg.role!r}, content={sys_msg.content!r}")
print(f"user: role={user_msg.role!r}, content={user_msg.content!r}")
print(f"assistant: role={asst_msg.role!r}, content={asst_msg.content!r}")

assert sys_msg.role == "system"
assert user_msg.role == "user"
assert asst_msg.role == "assistant"
```

```python
# Example 3: Construct a ChatCompletionsClient (no network, just validate init).
from azure.ai.inference import ChatCompletionsClient
from azure.core.credentials import AzureKeyCredential

client = ChatCompletionsClient(
    endpoint="https://example.invalid/openai/deployments/test",
    credential=AzureKeyCredential("fake-key-not-used"),
)

print(f"client type: {type(client).__name__}")
print(f"endpoint: {client._config.endpoint}")
assert isinstance(client, ChatCompletionsClient)
assert hasattr(client, "complete")
assert hasattr(client, "close")
client.close()
```

```python
# Example 4: Inspect enums and the JSON-schema response-format model.
# Structured output is described with a JsonSchemaFormat.
from azure.ai.inference.models import ChatRole, CompletionsFinishReason, JsonSchemaFormat

print(f"ChatRole.SYSTEM.value = {ChatRole.SYSTEM.value}")
print(f"ChatRole.USER.value = {ChatRole.USER.value}")
print(f"ChatRole.ASSISTANT.value = {ChatRole.ASSISTANT.value}")
assert ChatRole.USER.value == "user"
assert ChatRole.ASSISTANT.value == "assistant"
assert ChatRole("user") is ChatRole.USER  # lookup by the value sent on the wire

fmt = JsonSchemaFormat(
    name="weather",
    schema={"type": "object", "properties": {"temp": {"type": "number"}}},
    description="Weather report schema",
    strict=True,
)
print(f"JsonSchemaFormat name={fmt.name!r} strict={fmt.strict}")
assert fmt.name == "weather"
assert fmt["schema"]["type"] == "object"

finish_reasons = [r.value for r in CompletionsFinishReason]
print(f"finish reasons: {finish_reasons}")
assert "stop" in finish_reasons
```

```python
# Example 5: Construct EmbeddingsClient + verify async clients exist.
from azure.ai.inference import EmbeddingsClient
from azure.ai.inference.aio import ChatCompletionsClient as AsyncChatClient
from azure.ai.inference.aio import EmbeddingsClient as AsyncEmbeddingsClient
from azure.core.credentials import AzureKeyCredential

emb_client = EmbeddingsClient(
    endpoint="https://example.invalid/",
    credential=AzureKeyCredential("fake-key"),
)
print(f"sync EmbeddingsClient: {type(emb_client).__name__}")
assert hasattr(emb_client, "embed")
emb_client.close()

print(f"async ChatCompletionsClient: {AsyncChatClient.__module__}")
print(f"async EmbeddingsClient: {AsyncEmbeddingsClient.__module__}")
assert AsyncChatClient.__name__ == "ChatCompletionsClient"
assert AsyncEmbeddingsClient.__name__ == "EmbeddingsClient"
print("all azure.ai.inference smoke checks passed")
```
