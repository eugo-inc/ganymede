```python
# Example 1: import the primary client class and inspect its surface.
from azure.ai.projects import AIProjectClient

assert AIProjectClient is not None
assert hasattr(AIProjectClient, '__init__')

# Every generated Azure client has send_request() for raw HTTP calls and close();
# the operation groups are instance attributes, checked in Example 2.
expected_methods = ['close', 'send_request']
for name in expected_methods:
    assert hasattr(AIProjectClient, name), f"AIProjectClient missing {name}"

print(f"AIProjectClient: {AIProjectClient}")
print(f"Module: {AIProjectClient.__module__}")
print(f"MRO: {[c.__name__ for c in AIProjectClient.__mro__]}")
```

```python
# Example 2: instantiate AIProjectClient with a fake credential + endpoint (no network call).
from azure.ai.projects import AIProjectClient
from azure.core.credentials import AzureKeyCredential

# Construction should succeed without contacting the service.
client = AIProjectClient(
    endpoint="https://fake-resource.services.ai.azure.com/api/projects/fake-project",
    credential=AzureKeyCredential("fake-key-for-construction-only"),
)

assert client is not None
# Operation groups are attributes created together with the client.
for op_group in ['connections', 'datasets', 'deployments', 'indexes']:
    assert hasattr(client, op_group), f"AIProjectClient has no {op_group} operations"
    print(f"  has operation group: {op_group} -> {type(getattr(client, op_group)).__name__}")

client.close()
print("client constructed and closed cleanly")
```

```python
# Example 3: enumerate the models module — the bulk of the public surface.
from azure.ai.projects import models

public_names = [n for n in dir(models) if not n.startswith('_')]
print(f"azure.ai.projects.models exposes {len(public_names)} public names")
assert len(public_names) > 0

# Sample a few well-known names that should exist across versions.
sample = public_names[:10]
print("sample:", sample)

# Each public name should resolve to something (class, enum, or alias).
for name in sample:
    obj = getattr(models, name)
    assert obj is not None
```

```python
# Example 4: aio (async) client mirrors the sync client.
from azure.ai.projects.aio import AIProjectClient as AsyncAIProjectClient
from azure.core.credentials import AzureKeyCredential

assert AsyncAIProjectClient is not None
assert AsyncAIProjectClient.__name__ == 'AIProjectClient'

# Construct (no I/O).
aclient = AsyncAIProjectClient(
    endpoint="https://fake-resource.services.ai.azure.com/api/projects/fake-project",
    credential=AzureKeyCredential("fake-key"),
)
assert aclient is not None

# Async close is awaitable; in Jupyter we can await it directly.
await aclient.close()
print("async client constructed and closed cleanly")
```

```python
# Example 5: confirm key transitive deps required by the SDK are importable.
import azure.core
import azure.identity
import azure.storage.blob
import isodate
import openai
import typing_extensions

deps = {
    'azure.core': azure.core,
    'azure.identity': azure.identity,
    'azure.storage.blob': azure.storage.blob,
    'isodate': isodate,
    'openai': openai,
    'typing_extensions': typing_extensions,
}
for name, mod in deps.items():
    ver = getattr(mod, '__version__', 'n/a')
    print(f"  {name}: {ver}")
    assert mod is not None

assert all(d is not None for d in deps.values()), "missing transitive dep"
print("\nazure-ai-projects runtime deps OK")
```
