# azure_identity: 5 Basic Usage Examples

```python
# Example 1: Authenticate with DefaultAzureCredential
from azure.identity import DefaultAzureCredential
credential = DefaultAzureCredential()
print('DefaultAzureCredential created:', type(credential))
```

```python
# Example 2: Use EnvironmentCredential
import os
from azure.identity import EnvironmentCredential
os.environ['AZURE_CLIENT_ID'] = 'your-client-id'
os.environ['AZURE_TENANT_ID'] = 'your-tenant-id'
os.environ['AZURE_CLIENT_SECRET'] = 'your-client-secret'
credential = EnvironmentCredential()
print('EnvironmentCredential created:', type(credential))
```

```python
# Example 3: Use AzureCliCredential (requires Azure CLI login)
from azure.identity import AzureCliCredential
credential = AzureCliCredential()
print('AzureCliCredential created:', type(credential))
```

```python
# Example 4: DefaultAzureCredential returns a token, or reports every credential it tried
from azure.core.exceptions import ClientAuthenticationError
from azure.identity import DefaultAzureCredential
credential = DefaultAzureCredential(exclude_environment_credential=True, exclude_managed_identity_credential=True)
try:
    token = credential.get_token('https://management.azure.com/.default')
except ClientAuthenticationError as e:
    print('No Azure identity available; the chain reported each credential it tried')
    assert 'DefaultAzureCredential failed to retrieve a token' in str(e), str(e)
    assert 'AzureCliCredential' in str(e) and 'SharedTokenCacheCredential' in str(e), str(e)
else:
    # A developer sign-in on this machine (Azure CLI, azd, Azure PowerShell, VS Code or the shared
    # token cache) supplied the token.
    assert token.token and token.expires_on > 0, token.expires_on
    print('A developer sign-in on this machine supplied a token')
```
