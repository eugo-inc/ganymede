## Example 1: Basic HTML parsing with BeautifulSoup

```python
from bs4 import BeautifulSoup

# Example 1: Basic HTML parsing with BeautifulSoup
html_content = """
<html>
<head><title>Test Page</title></head>
<body>
    <h1>Welcome to Beautiful Soup</h1>
    <p class="intro">This is a paragraph with class 'intro'.</p>
    <p id="main">This is the main paragraph.</p>
    <ul>
        <li>First item</li>
        <li>Second item</li>
        <li>Third item</li>
    </ul>
</body>
</html>
"""

soup = BeautifulSoup(html_content, 'html.parser')
print("Example 1: Basic HTML parsing")
print(f"Title: {soup.title.string}")
print(f"H1 text: {soup.h1.string}")
print(f"First paragraph: {soup.p.string}")
print("Beautiful Soup basic parsing completed")
```

## Example 2: CSS selectors and find methods

```python
from bs4 import BeautifulSoup

# Example 2: CSS selector and find methods
html_content = """
<html>
<body>
    <div class="container">
        <p class="highlight">Important text</p>
        <p class="normal">Regular text</p>
        <a href="https://example.com" class="link">External link</a>
        <a href="/internal" class="link">Internal link</a>
    </div>
</body>
</html>
"""

soup = BeautifulSoup(html_content, 'html.parser')
print("Example 2: CSS selectors and find methods")

# Using CSS selectors
highlight = soup.select('.highlight')[0]
print(f"Highlight text: {highlight.string}")

# Find by class
links = soup.find_all('a', class_='link')
print(f"Found {len(links)} links:")
for link in links:
    print(f"  {link.string}: {link.get('href')}")

# Find by tag
container = soup.find('div', class_='container')
print(f"Container has {len(container.find_all('p'))} paragraphs")
print("CSS selector example completed")
```

## Example 3: Modifying HTML content

```python
from bs4 import BeautifulSoup

# Example 3: Modifying HTML content
html_content = """
<html>
<body>
    <h1>Original Title</h1>
    <p>Original paragraph content.</p>
    <ul id="list">
        <li>Item 1</li>
        <li>Item 2</li>
    </ul>
</body>
</html>
"""

soup = BeautifulSoup(html_content, 'html.parser')
print("Example 3: Modifying HTML content")
print("Before modification:")
print(f"Title: {soup.h1.string}")
print(f"Paragraph: {soup.p.string}")
print(f"List items: {len(soup.find_all('li'))}")

# Modify title
soup.h1.string = "Modified Title"

# Modify paragraph text
soup.p.string = "This paragraph has been updated!"

# Add new list item
new_item = soup.new_tag("li")
new_item.string = "Item 3"
soup.find('ul').append(new_item)

print("\nAfter modification:")
print(f"Title: {soup.h1.string}")
print(f"Paragraph: {soup.p.string}")
print(f"List items: {len(soup.find_all('li'))}")
print("HTML modification completed")
```

## Example 4: Extracting data from tables

```python
from bs4 import BeautifulSoup

# Example 4: Extracting data from tables
html_content = """
<html>
<body>
    <table id="data-table">
        <thead>
            <tr><th>Name</th><th>Age</th><th>City</th></tr>
        </thead>
        <tbody>
            <tr><td>Alice</td><td>25</td><td>New York</td></tr>
            <tr><td>Bob</td><td>30</td><td>Chicago</td></tr>
            <tr><td>Charlie</td><td>35</td><td>San Francisco</td></tr>
        </tbody>
    </table>
</body>
</html>
"""

soup = BeautifulSoup(html_content, 'html.parser')
print("Example 4: Table data extraction")

# Find the table
table = soup.find('table', id='data-table')
print(f"Found table with ID: {table.get('id')}")

# Extract headers
headers = [th.string for th in table.find('thead').find_all('th')]
print(f"Headers: {headers}")

# Extract data rows
rows = []
for tr in table.find('tbody').find_all('tr'):
    row = [td.string for td in tr.find_all('td')]
    rows.append(row)

print("Table data:")
for i, row in enumerate(rows):
    print(f"  Row {i+1}: {dict(zip(headers, row))}")

print(f"Extracted {len(rows)} data rows from table")
print("Table extraction completed")
```

## Example 5: Advanced Beautiful Soup features

```python
from bs4 import BeautifulSoup
import re

# Example 5: Advanced Beautiful Soup features
html_content = """
<html>
<body>
    <div class="content">
        <p data-info="important">This is <strong>very</strong> important text.</p>
        <p>This is <em>regular</em> text with <a href="#link">a link</a>.</p>
        <span class="date">2024-01-15</span>
        <span class="phone">123-456-7890</span>
    </div>
</body>
</html>
"""

soup = BeautifulSoup(html_content, 'html.parser')
print("Example 5: Advanced Beautiful Soup features")

# Extract text content, preserving structure
print("Text extraction:")
content_div = soup.find('div', class_='content')
print(f"All text: {content_div.get_text()}")
print(f"Separated text: {content_div.get_text(separator=' | ')}")

# Find elements by attribute
important_p = soup.find('p', {'data-info': 'important'})
print(f"\nImportant paragraph: {important_p.get_text()}")

# Extract specific patterns using regex
phone_pattern = re.compile(r'\d{3}-\d{3}-\d{4}')
phone_span = soup.find('span', string=phone_pattern)
if phone_span:
    print(f"Found phone: {phone_span.string}")

# Get nested tags
nested_tags = soup.find_all(['strong', 'em', 'a'])
print(f"\nNested formatting tags found: {len(nested_tags)}")
for tag in nested_tags:
    print(f"  {tag.name}: {tag.string}")

print("Beautiful Soup advanced features completed")
```
