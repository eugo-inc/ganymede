```python
import bigquery_magics
from google.auth.credentials import AnonymousCredentials
from google.cloud.bigquery import QueryJobConfig

# Example 3: bigquery_magics.context holds the project, credentials and default job settings that %%bigquery uses.
print("Example 3: configure the shared context")

context = bigquery_magics.context

# Setting the project and credentials up front means the magic never looks for application default credentials.
context.project = "example-project"
context.credentials = AnonymousCredentials()
assert context.project == "example-project", context.project
assert isinstance(context.credentials, AnonymousCredentials), type(context.credentials)

# The default job config applies to every query; magic arguments such as --maximum_bytes_billed override it per cell.
context.default_query_job_config = QueryJobConfig(
    maximum_bytes_billed=100_000_000,
    use_query_cache=False,
    labels={"team": "analytics"},
)
job_config = context.default_query_job_config
assert job_config.maximum_bytes_billed == 100_000_000, job_config.maximum_bytes_billed
assert job_config.use_query_cache is False, job_config.use_query_cache
api_repr = job_config.to_api_repr()
assert api_repr["query"]["maximumBytesBilled"] == "100000000", api_repr
assert api_repr["labels"] == {"team": "analytics"}, api_repr

# progress_bar_type=None turns the progress bar off; default_variable names the DataFrame when a cell gives none.
context.progress_bar_type = None
context.default_variable = "results"
assert context.progress_bar_type is None
assert context.default_variable == "results", context.default_variable
context.default_variable = None

print(f"project={context.project}, maximumBytesBilled={api_repr['query']['maximumBytesBilled']}, labels={api_repr['labels']}")
print("Example 3 finished\n")
```

```python
import contextlib
import io

import bigquery_magics
from IPython import get_ipython
from IPython.core.error import UsageError

# Example 4: %%bigquery checks its arguments and the query before it opens a connection.
print("Example 4: argument and query checks in the %%bigquery magic")

ipython = get_ipython()
assert ipython is not None, "this example needs an IPython kernel"
if not bigquery_magics.is_registered:
    ipython.run_line_magic("load_ext", "bigquery_magics")


def rejected(line, query, expected):
    """Run the cell magic and return the exception it raises."""
    try:
        ipython.run_cell_magic("bigquery", line, query)
    except expected as error:
        return error
    raise AssertionError(f"%%bigquery {line!r} was accepted")


error = rejected('--params {"n": 5', "SELECT @n", SyntaxError)
assert "--params" in str(error), error

error = rejected('--params {"a": 1} --params {"b": 2}', "SELECT 1", ValueError)
assert "Duplicate --params" in str(error), error

error = rejected("--engine spark", "SELECT 1", ValueError)
assert "Invalid engine: spark" in str(error), error

error = rejected("--no_such_option", "SELECT 1", UsageError)
print(f"unknown option -> {type(error).__name__}")

# A query can come from a variable with $name; the variable must exist and hold text.
ipython.user_ns.pop("missing_query", None)
error = rejected("", "$missing_query", NameError)
assert "missing_query does not exist" in str(error), error

ipython.user_ns["numeric_query"] = 42
error = rejected("", "$numeric_query", TypeError)
assert "must be a string" in str(error), error
del ipython.user_ns["numeric_query"]

# An empty cell is reported on stderr and returns nothing.
stderr = io.StringIO()
with contextlib.redirect_stderr(stderr):
    result = ipython.run_cell_magic("bigquery", "", "   \n")
assert result is None, result
assert "Query is missing." in stderr.getvalue(), stderr.getvalue()

print("bad --params JSON, a repeated --params, an unknown engine, an unknown option and bad $query variables were all rejected")
print("Example 4 finished\n")
```

```python
from bigquery_magics import pyformat
from google.cloud.bigquery import ArrayQueryParameter, QueryJobConfig, ScalarQueryParameter

# Example 5: --pyformat text substitution versus real query parameters.
print("Example 5: --pyformat substitution and query parameters")

# --pyformat fills {name} fields from notebook variables. Strings are not escaped, which is what lets it
# fill in table names, where query parameters are not allowed.
template = "SELECT order_id FROM `{table}` WHERE amount >= {minimum} LIMIT {limit}"
sql = pyformat.pyformat(template, {"table": "example-project.sales.orders", "minimum": 100.5, "limit": 10})
assert sql == "SELECT order_id FROM `example-project.sales.orders` WHERE amount >= 100.5 LIMIT 10", sql
print(f"formatted: {sql}")

try:
    pyformat.pyformat("SELECT {ids}", {"ids": [1, 2, 3]})
except TypeError as error:
    assert "unsupported type" in str(error), error
else:
    raise AssertionError("a list was substituted into the SQL text")

try:
    pyformat.pyformat("SELECT {undefined_name}", {})
except KeyError as error:
    assert "undefined_name" in str(error), error
else:
    raise AssertionError("an undefined variable was substituted")

# Values belong in query parameters, which BigQuery sends separately from the SQL text.
job_config = QueryJobConfig(
    query_parameters=[
        ScalarQueryParameter("min_amount", "FLOAT64", 100.5),
        ArrayQueryParameter("states", "STRING", ["WA", "OR"]),
    ]
)
query = job_config.to_api_repr()["query"]
assert query["parameterMode"] == "NAMED", query
by_name = {p["name"]: p for p in query["queryParameters"]}
assert by_name["min_amount"]["parameterType"]["type"] == "FLOAT64", by_name
assert by_name["min_amount"]["parameterValue"]["value"] == 100.5, by_name
assert by_name["states"]["parameterType"] == {"type": "ARRAY", "arrayType": {"type": "STRING"}}, by_name
assert [v["value"] for v in by_name["states"]["parameterValue"]["arrayValues"]] == ["WA", "OR"], by_name
print(f"parameters: {sorted(by_name)} in {query['parameterMode']} mode")
print("Example 5 finished\n")
```
