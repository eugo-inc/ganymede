## Example 1: black functionality test 1

```python
import black
from black import format_str, Mode

# Example 1: Basic code formatting with Black
print("Example 1: Basic code formatting with Black")

try:
    # Sample unformatted Python code
    unformatted_code = """def hello(name,age=None):
    if age is None:age=25
    print(f"Hello {name}, you are {age} years old!")
    return{"name":name,"age":age}"""
    
    print("✓ Black code formatter imported")
    print("Original unformatted code:")
    print(unformatted_code)
    print()
    
    # Format the code using Black
    mode = Mode()
    formatted_code = format_str(unformatted_code, mode=mode)
    
    print("Black formatted code:")
    print(formatted_code)
    
    # Compare lengths and characteristics
    print(f"Original length: {len(unformatted_code)} characters")
    print(f"Formatted length: {len(formatted_code)} characters")
    original_lines = unformatted_code.count('\n') + 1
    formatted_lines = formatted_code.count('\n') + 1
    print(f"Original lines: {original_lines}")
    print(f"Formatted lines: {formatted_lines}")
    
    print("✓ Code formatting completed successfully")
    print("black example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"black example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: black functionality test 2

```python
import black
from black import Mode, TargetVersion

# Example 2: Black formatting modes and options
print("Example 2: Black formatting modes and options")

try:
    # Complex unformatted code sample
    complex_code = """import os,sys,json
def process_data(items,config={'timeout':30,'retries':3}):
    results=[]
    for item in items:
        if item['active']:
            result={'id':item['id'],'processed':True,'timestamp':item.get('timestamp',None)}
            results.append(result)
    return results

class DataProcessor:
    def __init__(self,config=None):
        self.config=config if config else {}
    def run(self):pass"""
    
    print("Testing different Black formatting modes:")
    print("\nOriginal code:")
    print(complex_code)
    print("\n" + "="*50)
    
    # Standard mode
    standard_mode = Mode()
    formatted_standard = black.format_str(complex_code, mode=standard_mode)
    print("\n1. Standard mode formatting:")
    print(formatted_standard)
    
    # Different line length
    short_lines_mode = Mode(line_length=60)
    formatted_short = black.format_str(complex_code, mode=short_lines_mode)
    print("2. Short line length (60 chars):")
    print(formatted_short)
    
    # Different Python target version
    try:
        py38_mode = Mode(target_versions={TargetVersion.PY38})
        formatted_py38 = black.format_str(complex_code, mode=py38_mode)
        print("3. Python 3.8+ target:")
        print(formatted_py38[:200] + "..." if len(formatted_py38) > 200 else formatted_py38)
    except Exception as e:
        print(f"3. Python 3.8+ target: {type(e).__name__}")
        raise
    
    # String normalization options
    no_string_normalize_mode = Mode(string_normalization=False)
    code_with_quotes = '''def test(): return "hello" + 'world' '''
    formatted_quotes = black.format_str(code_with_quotes, mode=no_string_normalize_mode)
    print("4. String normalization disabled:")
    print(f"Original: {code_with_quotes}")
    print(f"Formatted: {formatted_quotes}")
    
    print("✓ Multiple formatting modes demonstrated")
    print("black example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"black example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: black functionality test 3

```python
import black
from black import InvalidInput, format_str, Mode

# Example 3: Black error handling and code validation
print("Example 3: Black error handling and code validation")

try:
    print("Testing Black's error handling capabilities:")
    
    # Valid code that can be formatted
    valid_code = """
def calculate(a,b):
    if a>b:return a*b
    else:return a+b
"""
    
    try:
        formatted_valid = format_str(valid_code, mode=Mode())
        print("✓ Valid code formatted successfully:")
        print(formatted_valid)
    except Exception as e:
        print(f"Valid code error: {type(e).__name__}: {e}")
        raise
    
    # Invalid syntax that Black cannot format
    invalid_codes = [
        "def broken_function(:\n    pass",  # Syntax error
        "if True\n    print('missing colon')",  # Missing colon
        "def func():\n  return\n    pass",  # Mixed indentation
    ]
    
    print("Testing invalid code handling:")
    for i, invalid_code in enumerate(invalid_codes, 1):
        # Black refuses source it cannot parse with InvalidInput ("Cannot parse: <line>:<col> ...", lower-case in
        # newer releases); formatting it anyway is a bug.
        try:
            formatted_invalid = format_str(invalid_code, mode=Mode())
        except InvalidInput as e:
            assert str(e).lower().startswith("cannot parse"), f"unexpected InvalidInput message: {e!r}"
            print(f"  {i}. ✓ InvalidInput caught: {type(e).__name__}")
        else:
            raise AssertionError(f"black formatted invalid code {invalid_code!r} into {formatted_invalid!r}")
    
    # Test formatting validation
    print("\nTesting format validation:")
    
    # Code that's already well formatted
    well_formatted_code = '''def example_function(param1: str, param2: int = 42) -> str:
    """Example function with proper formatting."""
    if param2 > 10:
        return f"Hello {param1}, number is {param2}"
    return f"Hi {param1}"
'''
    
    try:
        result = format_str(well_formatted_code, mode=Mode())
        if result.strip() == well_formatted_code.strip():
            print("✓ Well-formatted code unchanged")
        else:
            print("✓ Well-formatted code had minor changes")
            print(f"Changes: {len(result)} vs {len(well_formatted_code)} chars")
    except Exception as e:
        print(f"Well-formatted code error: {type(e).__name__}")
        raise
    
    # Code with common formatting issues
    problematic_code = """
def messy_function(a,b,c,d,e,f,g,h,i,j,k,l,m,n,o,p):
    if a and b and c and d and e and f and g and h and i and j and k and l and m and n and o and p:
        return a+b+c+d+e+f+g+h+i+j+k+l+m+n+o+p
    else:
        return None
"""
    
    try:
        formatted_problematic = format_str(problematic_code, mode=Mode())
        print("✓ Problematic code reformatted:")
        problematic_lines = problematic_code.count('\n')
        reformatted_lines = formatted_problematic.count('\n')
        print(f"Original lines: {problematic_lines}")
        print(f"Formatted lines: {reformatted_lines}")
        print("First few lines of result:")
        lines = formatted_problematic.split('\n')[:5]
        for line in lines:
            if line.strip():
                print(f"  {line}")
    except Exception as e:
        print(f"Problematic code error: {type(e).__name__}")
        raise
    
    print("✓ Error handling and validation completed")
    print("black example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"black example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: black functionality test 4

```python
import black
from black import format_str, Mode, FileMode
import tempfile
import os

# Example 4: Black file processing and batch formatting
print("Example 4: Black file processing and batch formatting")

try:
    # Create temporary Python files with different formatting issues
    temp_dir = tempfile.mkdtemp()
    
    # Sample files with different code styles
    sample_files = {
        "messy_imports.py": """import os,sys
import json,re
from datetime import datetime,timedelta
from collections import defaultdict,Counter""",
        
        "poor_spacing.py": """def calculate(a,b,c):
    result=a+b*c
    if result>100:return result
    elif result<0:return 0
    else:return result//2""",
        
        "long_lines.py": """def very_long_function_name_with_many_parameters(parameter_one, parameter_two, parameter_three, parameter_four, parameter_five, parameter_six):
    return parameter_one + parameter_two + parameter_three + parameter_four + parameter_five + parameter_six"""
    }
    
    print(f"✓ Creating {len(sample_files)} temporary Python files")
    
    formatted_results = {}
    
    for filename, content in sample_files.items():
        file_path = os.path.join(temp_dir, filename)
        
        # Write original file
        with open(file_path, 'w') as f:
            f.write(content)
        
        print(f"\nProcessing {filename}:")
        print(f"Original content:")
        print(content)
        
        # Format the content
        try:
            mode = Mode()
            formatted_content = format_str(content, mode=mode)
            formatted_results[filename] = formatted_content
            
            print(f"Formatted content:")
            print(formatted_content)
            
            # Write formatted file
            formatted_path = os.path.join(temp_dir, f"formatted_{filename}")
            with open(formatted_path, 'w') as f:
                f.write(formatted_content)
            
            print(f"✓ Saved formatted version")
            
        except Exception as e:
            print(f"Error formatting {filename}: {type(e).__name__}: {e}")
            raise
    
    # Compare formatting results
    print(f"\nFormatting comparison summary:")
    for filename, formatted_content in formatted_results.items():
        original_content = sample_files[filename]
        
        original_lines = original_content.count('\n') + 1
        formatted_lines = formatted_content.count('\n') + 1
        
        print(f"  {filename}:")
        print(f"    Lines: {original_lines} -> {formatted_lines}")
        print(f"    Characters: {len(original_content)} -> {len(formatted_content)}")
        print(f"    Changed: {'Yes' if original_content != formatted_content else 'No'}")
    
    # Test FileMode for different configurations
    print(f"\nTesting different file modes:")
    sample_code = "def test(a,b,c,d,e,f): return a+b+c+d+e+f"
    
    modes = [
        ("Standard", Mode()),
        ("Short lines", Mode(line_length=50)),
        ("Long lines", Mode(line_length=120)),
    ]
    
    for mode_name, mode in modes:
        try:
            result = format_str(sample_code, mode=mode)
            lines = result.count('\n') + 1
            print(f"  {mode_name}: {lines} lines, max length ~{mode.line_length}")
        except Exception as e:
            print(f"  {mode_name}: Error - {type(e).__name__}")
            raise
    
    # Cleanup
    try:
        for filename in sample_files:
            os.unlink(os.path.join(temp_dir, filename))
            formatted_file = os.path.join(temp_dir, f"formatted_{filename}")
            if os.path.exists(formatted_file):
                os.unlink(formatted_file)
        os.rmdir(temp_dir)
    except:
        pass
        raise
    
    print("✓ File processing and batch formatting completed")
    print("black example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"black example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```
