## Example 1: bleach functionality test 1

```python
import bleach

# Example 1: Basic HTML sanitization with Bleach
print("Example 1: Basic HTML sanitization with Bleach")

try:
    # Sample HTML content with potentially dangerous elements
    unsafe_html = """
    <p>This is a <strong>safe</strong> paragraph.</p>
    <script>alert('This is dangerous!');</script>
    <p>Here's a <a href="https://example.com">safe link</a></p>
    <iframe src="https://evil.com"></iframe>
    <p style="color: red;">Styled paragraph</p>
    <img src="https://example.com/image.jpg" alt="Image">
    """
    
    print("✓ Bleach HTML sanitizer imported")
    print("Original unsafe HTML:")
    print(unsafe_html)
    
    # Basic sanitization - only allow safe tags
    safe_tags = ['p', 'strong', 'a', 'img']
    safe_attributes = {
        'a': ['href'],
        'img': ['src', 'alt']
    }
    
    cleaned_html = bleach.clean(
        unsafe_html,
        tags=safe_tags,
        attributes=safe_attributes,
        strip=True
    )
    
    print("\nAfter Bleach sanitization:")
    print(cleaned_html)
    
    # Count removed elements
    original_lines = len(unsafe_html.split('\n'))
    cleaned_lines = len(cleaned_html.split('\n'))
    
    print(f"\nSanitization results:")
    print(f"Original content lines: {original_lines}")
    print(f"Cleaned content lines: {cleaned_lines}")
    print(f"Dangerous scripts removed: {'script' not in cleaned_html}")
    print(f"Dangerous iframes removed: {'iframe' not in cleaned_html}")
    print(f"Inline styles removed: {'style=' not in cleaned_html}")
    
    print("✓ Basic HTML sanitization completed successfully")
    print("bleach example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"bleach example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: bleach functionality test 2

```python
import bleach
from bleach import ALLOWED_TAGS, ALLOWED_ATTRIBUTES

# Example 2: URL and link sanitization
print("Example 2: URL and link sanitization")

try:
    # HTML with various types of URLs and links
    html_with_links = """
    <p>Visit our <a href="https://example.com">website</a></p>
    <p>Email us: <a href="mailto:test@example.com">test@example.com</a></p>
    <p>Dangerous: <a href="javascript:alert('XSS')">Click here</a></p>
    <p>File link: <a href="file:///etc/passwd">System file</a></p>
    <p>FTP: <a href="ftp://files.example.com">FTP Server</a></p>
    <p>Relative: <a href="/internal/page">Internal page</a></p>
    <p>Data URI: <a href="data:text/html,<script>alert('XSS')</script>">Data</a></p>
    """
    
    print("Testing URL and link sanitization:")
    print("Original HTML with various links:")
    print(html_with_links)
    
    # Default allowed protocols
    print(f"\nDefault allowed protocols: {bleach.ALLOWED_PROTOCOLS}")
    
    # Safe URL sanitization
    safe_links = bleach.clean(
        html_with_links,
        tags=['p', 'a'],
        attributes={'a': ['href']},
        protocols=['http', 'https', 'mailto'],
        strip=True
    )
    
    print("\nAfter safe URL sanitization:")
    print(safe_links)
    
    # More restrictive - only HTTPS
    https_only = bleach.clean(
        html_with_links,
        tags=['p', 'a'],
        attributes={'a': ['href']},
        protocols=['https'],
        strip=True
    )
    
    print("\nHTTPS-only sanitization:")
    print(https_only)
    
    # Link text sanitization (remove URLs but keep text)
    def test_linkify():
        plain_text = "Visit https://example.com or email test@example.com"
        linkified = bleach.linkify(plain_text)
        print(f"\nLinkify demo:")
        print(f"Original: {plain_text}")
        print(f"Linkified: {linkified}")
    
    try:
        test_linkify()
    except Exception as e:
        print(f"Linkify test: {type(e).__name__}")
        raise
    
    # URL validation results
    dangerous_protocols = ['javascript:', 'data:', 'file://']
    safe_protocols = ['https://', 'http://', 'mailto:']
    
    print(f"\nURL security analysis:")
    for protocol in dangerous_protocols:
        if protocol not in safe_links:
            print(f"  ✓ Blocked dangerous protocol: {protocol}")
        else:
            print(f"  ✗ Dangerous protocol present: {protocol}")
    
    for protocol in safe_protocols:
        if protocol in safe_links:
            print(f"  ✓ Safe protocol preserved: {protocol}")
    
    print("✓ URL and link sanitization completed successfully")
    print("bleach example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"bleach example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: bleach functionality test 3

```python
import bleach

# Example 3: Advanced sanitization with custom filters
print("Example 3: Advanced sanitization with custom filters")

try:
    # Complex HTML with various security threats
    complex_html = """
    <div class="content" id="main">
        <h1 style="color: red; font-size: 24px;">Title</h1>
        <p class="intro">Introduction with <em>emphasis</em> and <strong>bold</strong> text.</p>
        <ul>
            <li>List item 1</li>
            <li onclick="alert('XSS')">Dangerous list item</li>
            <li>List item 3</li>
        </ul>
        <img src="https://example.com/image.jpg" alt="Image" onload="alert('Image XSS')">
        <a href="https://example.com" target="_blank" rel="noopener">External link</a>
        <form action="/submit" method="post">
            <input type="text" name="username" value="default">
            <button type="submit">Submit</button>
        </form>
        <table border="1">
            <tr><th>Header</th></tr>
            <tr><td>Data</td></tr>
        </table>
    </div>
    """
    
    print("Testing advanced sanitization:")
    print("Original complex HTML:")
    print(complex_html[:300] + "..." if len(complex_html) > 300 else complex_html)
    
    # Rich text sanitization - allow formatting but remove dangerous elements
    rich_text_tags = ['h1', 'h2', 'h3', 'p', 'div', 'span', 'strong', 'em', 'u', 'ul', 'ol', 'li', 'a', 'img']
    rich_text_attributes = {
        'div': ['class', 'id'],
        'p': ['class'],
        'h1': ['class'],
        'h2': ['class'],  
        'h3': ['class'],
        'a': ['href', 'target', 'rel'],
        'img': ['src', 'alt', 'width', 'height'],
        'span': ['class'],
    }
    
    rich_cleaned = bleach.clean(
        complex_html,
        tags=rich_text_tags,
        attributes=rich_text_attributes,
        protocols=['http', 'https'],
        strip=True
    )
    
    print("\nRich text sanitization (allows formatting):")
    print(rich_cleaned)
    
    # Minimal sanitization - only basic text formatting
    minimal_tags = ['p', 'strong', 'em', 'br']
    minimal_cleaned = bleach.clean(
        complex_html,
        tags=minimal_tags,
        attributes={},
        strip=True
    )
    
    print("\nMinimal sanitization (text only):")
    print(minimal_cleaned)
    
    # Custom attribute filtering
    def custom_attribute_filter(tag, name, value):
        # Allow specific CSS styles only
        if name == 'style':
            allowed_styles = ['color', 'font-size', 'font-weight']
            # Simple style parsing (in reality, you'd want more robust parsing)
            if any(style in value for style in allowed_styles):
                return True
        return False
    
    # Sanitization with custom filters
    try:
        custom_cleaned = bleach.clean(
            complex_html,
            tags=['h1', 'p', 'strong', 'em'],
            attributes={
                '*': custom_attribute_filter
            },
            strip=True
        )
        print("\nCustom attribute filtering:")
        print(custom_cleaned[:200] + "..." if len(custom_cleaned) > 200 else custom_cleaned)
    except Exception as e:
        print(f"Custom filtering: {type(e).__name__}")
        raise
    
    # Security analysis
    security_threats = [
        'onclick=', 'onload=', 'onerror=', 'javascript:', 'vbscript:',
        '<script', '<iframe', '<object', '<embed', '<form'
    ]
    
    print(f"\nSecurity threat analysis:")
    for threat in security_threats:
        original_present = threat.lower() in complex_html.lower()
        cleaned_present = threat.lower() in rich_cleaned.lower()
        
        if original_present and not cleaned_present:
            print(f"  ✓ Removed threat: {threat}")
        elif original_present and cleaned_present:
            print(f"  ✗ Threat still present: {threat}")
        elif not original_present:
            print(f"  - Not present: {threat}")
    
    print("✓ Advanced sanitization with custom filters completed")
    print("bleach example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"bleach example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: bleach functionality test 4

```python
import bleach

# Example 4: User-generated content sanitization scenarios
print("Example 4: User-generated content sanitization scenarios")

try:
    # Simulate various user-generated content scenarios
    user_contents = {
        "blog_comment": """
        <p>Great article! I especially liked the part about <strong>security</strong>.</p>
        <p>Here's a useful link: <a href="https://security.example.com">Security Guide</a></p>
        <script>document.cookie = "stolen=true";</script>
        <p>Thanks for sharing! 👍</p>
        """,
        
        "forum_post": """
        <h2>My Experience with Python</h2>
        <p>I've been using Python for <em>5 years</em> and here are my thoughts:</p>
        <ul>
            <li>Great for beginners</li>
            <li onclick="window.location='http://malicious.com'">Easy to learn</li>
            <li>Powerful libraries</li>
        </ul>
        <img src="x" onerror="alert('XSS in forum post')">
        """,
        
        "product_review": """
        <div class="review">
            <h3>5-star product!</h3>
            <p>This product is <strong>amazing</strong>. I highly recommend it.</p>
            <p style="background: url('javascript:alert(1)')">Styled text</p>
            <iframe src="https://competitor.com/tracking"></iframe>
        </div>
        """,
        
        "chat_message": """
        Hey everyone! Check out this cool site: https://example.com
        <script>fetch('/api/steal-data')</script>
        Anyone want to collaborate on a project?
        """
    }
    
    print("Testing user-generated content sanitization scenarios:")
    
    # Define sanitization profiles for different contexts
    sanitization_profiles = {
        "strict": {
            "tags": ['p', 'br'],
            "attributes": {},
            "description": "Text-only with line breaks"
        },
        "basic": {
            "tags": ['p', 'br', 'strong', 'em', 'ul', 'ol', 'li'],
            "attributes": {},
            "description": "Basic text formatting"
        },
        "rich": {
            "tags": ['p', 'br', 'strong', 'em', 'ul', 'ol', 'li', 'h1', 'h2', 'h3', 'a', 'blockquote'],
            "attributes": {'a': ['href']},
            "description": "Rich text with safe links"
        },
        "permissive": {
            "tags": ['p', 'br', 'strong', 'em', 'ul', 'ol', 'li', 'h1', 'h2', 'h3', 'a', 'img', 'div'],
            "attributes": {
                'a': ['href', 'target'],
                'img': ['src', 'alt'],
                'div': ['class']
            },
            "description": "Permissive with images and classes"
        }
    }
    
    # Test each content type with different profiles
    for content_name, content in user_contents.items():
        print(f"\n{'='*50}")
        print(f"Content Type: {content_name}")
        print(f"Original length: {len(content)} characters")
        
        # Show original content (truncated)
        print(f"Original: {content[:100]}...")
        
        for profile_name, profile in sanitization_profiles.items():
            try:
                sanitized = bleach.clean(
                    content,
                    tags=profile["tags"],
                    attributes=profile["attributes"],
                    protocols=['http', 'https', 'mailto'],
                    strip=True
                )
                
                # Analyze results
                original_threats = sum([
                    content.lower().count('<script'),
                    content.lower().count('onclick='),
                    content.lower().count('onerror='),
                    content.lower().count('javascript:'),
                    content.lower().count('<iframe')
                ])
                
                remaining_threats = sum([
                    sanitized.lower().count('<script'),
                    sanitized.lower().count('onclick='),
                    sanitized.lower().count('onerror='),
                    sanitized.lower().count('javascript:'),
                    sanitized.lower().count('<iframe')
                ])
                
                print(f"\n  {profile_name.upper()} profile ({profile['description']}):")
                print(f"    Length: {len(content)} -> {len(sanitized)} chars")
                print(f"    Threats: {original_threats} -> {remaining_threats}")
                print(f"    Result: {sanitized[:80]}...")
                
            except Exception as e:
                print(f"    Error with {profile_name}: {type(e).__name__}")
                raise
    
    # Demonstrate common XSS attack vectors and their mitigation
    print(f"\n{'='*50}")
    print("Common XSS attack vectors test:")
    
    xss_vectors = [
        "<script>alert('XSS')</script>",
        "<img src=x onerror=alert('XSS')>",
        "<iframe src=javascript:alert('XSS')></iframe>", 
        "<a href='javascript:alert(\"XSS\")'>Click</a>",
        "<div onclick='alert(\"XSS\")'>Click me</div>",
        "<style>body{background:url('javascript:alert(1)')}</style>"
    ]
    
    for vector in xss_vectors:
        cleaned = bleach.clean(vector, tags=['p', 'a'], attributes={'a': ['href']}, protocols=['http', 'https'])
        is_safe = cleaned == "" or "javascript:" not in cleaned.lower()
        status = "✓ BLOCKED" if is_safe else "✗ THREAT REMAINS"
        print(f"  {status}: {vector[:40]}...")
    
    print("✓ User-generated content sanitization scenarios completed")
    print("bleach example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"bleach example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```
