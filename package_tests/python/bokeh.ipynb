## Example 1: bokeh functionality test 1

```python
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
import numpy as np

# Example 1: Basic line plot with Bokeh
print("=== Bokeh Basic Line Plot Example ===")

try:
    # Enable notebook output
    output_notebook()
    
    # Generate sample data
    x = np.linspace(0, 4*np.pi, 100)
    y = np.sin(x)
    y2 = np.cos(x)
    
    # Create a new plot with tools
    p = figure(
        title="Trigonometric Functions",
        x_axis_label='x',
        y_axis_label='y',
        width=400,
        height=300
    )
    
    # Add line renderers
    p.line(x, y, legend_label="sin(x)", line_width=2, color='blue')
    p.line(x, y2, legend_label="cos(x)", line_width=2, color='red')
    
    print(f"Created plot with {len(x)} data points")
    print("Plot features:")
    print("  - Interactive pan and zoom")
    print("  - Hover tooltips")
    print("  - Legend with toggle functionality")
    print("  - Responsive design")
    
    # Display the plot
    show(p)
    print("Basic line plot example completed successfully!")
    
except Exception as e:
    print(f"Error in basic line plot: {e}")
    raise AssertionError(
        f"bokeh example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 2: bokeh functionality test 2

```python
from bokeh.plotting import figure, show
from bokeh.models import HoverTool, ColumnDataSource
from bokeh.io import output_notebook
import numpy as np

# Example 2: Interactive scatter plot with hover tooltips
print("=== Bokeh Interactive Scatter Plot with Hover Example ===")

try:
    # Enable notebook output
    output_notebook()
    
    # Generate random data
    n = 500
    x = np.random.random(n) * 100
    y = np.random.random(n) * 100
    colors = np.random.choice(['red', 'green', 'blue', 'orange', 'purple'], n)
    sizes = np.random.randint(10, 25, n)
    
    # Create figure with custom tools
    p = figure(
        title="Interactive Scatter Plot",
        x_axis_label='X Values',
        y_axis_label='Y Values',
        width=500,
        height=400,
        tools="pan,wheel_zoom,box_zoom,reset,save"
    )
    
    # Add hover tool with custom tooltips
    hover = HoverTool(tooltips=[
        ("Index", "$index"),
        ("(X,Y)", "($x, $y)"),
        ("Size", "@size"),
        ("Color", "@color")
    ])
    p.add_tools(hover)
    
    # Create ColumnDataSource
    source = ColumnDataSource(data=dict(x=x, y=y, size=sizes, color=colors))

    # Add scatter renderer
    scatter = p.scatter(
        'x', 'y',
        size='size',
        color='color',
        alpha=0.6,
        source=source
    )
    
    print(f"Created scatter plot with {n} points")
    print("Interactive features:")
    print("  - Hover tooltips showing data details")
    print("  - Pan and zoom functionality")
    print("  - Box zoom selection")
    print("  - Reset and save tools")
    print(f"  - Color variety: {len(set(colors))} different colors")
    
    # Display the plot
    show(p)
    print("Interactive scatter plot example completed successfully!")
    
except Exception as e:
    print(f"Error in interactive scatter plot: {e}")
    raise AssertionError(
        f"bokeh example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: bokeh functionality test 3

```python
from bokeh.plotting import figure, show
from bokeh.layouts import column, row
from bokeh.io import output_notebook
from bokeh.transform import factor_cmap
from bokeh.palettes import Category20c
import numpy as np

# Example 3: Multi-plot dashboard with different chart types
print("=== Bokeh Multi-Plot Dashboard Example ===")

try:
    # Enable notebook output
    output_notebook()
    
    # Data for bar chart
    categories = ['Category A', 'Category B', 'Category C', 'Category D', 'Category E']
    values = [23, 45, 56, 78, 32]
    
    # Create bar chart
    bar_plot = figure(
        x_range=categories,
        title="Bar Chart",
        width=300,
        height=250
    )
    bar_plot.vbar(
        x=categories, 
        top=values, 
        width=0.5,
        color=factor_cmap('x', palette=Category20c[len(categories)], factors=categories)
    )
    
    # Data for line chart with multiple series
    x = np.linspace(0, 2*np.pi, 100)
    line_plot = figure(
        title="Multiple Line Series",
        width=300,
        height=250
    )
    line_plot.line(x, np.sin(x), line_width=2, color='blue', legend_label='sin(x)')
    line_plot.line(x, np.cos(x), line_width=2, color='red', legend_label='cos(x)')
    line_plot.line(x, np.tan(x/2), line_width=2, color='green', legend_label='tan(x/2)')
    
    # Data for area chart
    x2 = np.linspace(0, 10, 50)
    y1 = np.exp(-x2/5) * np.sin(x2)
    y2 = np.exp(-x2/3) * np.cos(x2)
    
    area_plot = figure(
        title="Stacked Area Chart",
        width=300,
        height=250
    )
    area_plot.varea(x=x2, y1=0, y2=y1, alpha=0.7, color='lightblue')
    area_plot.varea(x=x2, y1=y1, y2=y1+y2, alpha=0.7, color='lightcoral')
    
    # Create dashboard layout
    dashboard = column(
        row(bar_plot, line_plot),
        area_plot
    )
    
    print("Dashboard created with:")
    print(f"  - Bar chart with {len(categories)} categories")
    print(f"  - Line chart with 3 mathematical functions")
    print(f"  - Stacked area chart with {len(x2)} data points")
    print("  - Responsive grid layout")
    print("  - Individual plot interactions")
    
    # Display the dashboard
    show(dashboard)
    print("Multi-plot dashboard example completed successfully!")
    
except Exception as e:
    print(f"Error in multi-plot dashboard: {e}")
    raise AssertionError(
        f"bokeh example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: bokeh functionality test 4

```python
from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource, Select, CustomJS
from bokeh.layouts import column
from bokeh.io import output_notebook
import numpy as np
import pandas as pd

# Example 4: Interactive data visualization with widgets
print("=== Bokeh Interactive Widgets Example ===")

try:
    # Enable notebook output
    output_notebook()
    
    # Create sample dataset
    np.random.seed(42)
    n = 100
    df = pd.DataFrame({
        'x': np.random.randn(n),
        'y': np.random.randn(n),
        'category': np.random.choice(['A', 'B', 'C'], n),
        'size': np.random.randint(10, 30, n),
        'alpha': np.random.rand(n)
    })
    
    # Create ColumnDataSource
    source = ColumnDataSource(df)
    
    # Create the main plot
    p = figure(
        title="Interactive Data Visualization",
        width=500,
        height=400,
        tools="pan,wheel_zoom,box_zoom,reset,save"
    )
    
    # Add scatter plot
    scatter = p.scatter(
        'x', 'y',
        size='size',
        alpha='alpha',
        color='navy',
        source=source
    )
    
    # Create a dropdown widget for color mapping
    select = Select(
        title="Color by:",
        value="category",
        options=[("category", "Category"), ("size", "Size"), ("alpha", "Alpha")]
    )
    
    # JavaScript callback for interactivity
    callback = CustomJS(args=dict(source=source, scatter=scatter), code="""
        const data = source.data;
        const selected = cb_obj.value;
        
        if (selected === 'category') {
            scatter.glyph.fill_color = 'navy';
        } else if (selected === 'size') {
            scatter.glyph.fill_color = 'red';
        } else {
            scatter.glyph.fill_color = 'green';
        }
        
        source.change.emit();
    """)
    
    select.js_on_change('value', callback)
    
    # Create layout
    layout = column(select, p)
    
    print("Interactive visualization created with:")
    print(f"  - Dataset with {len(df)} data points")
    print(f"  - {len(df.columns)} data dimensions")
    print("  - Interactive dropdown widget for color mapping")
    print("  - JavaScript callbacks for real-time updates")
    print("  - Pan, zoom, and selection tools")
    print("  - Responsive layout design")
    
    # Display the interactive plot
    show(layout)
    print("Interactive widgets example completed successfully!")
    
except Exception as e:
    print(f"Error in interactive widgets: {e}")
    raise AssertionError(
        f"bokeh example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 5: bokeh functionality test 5

```python
import bokeh
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
from bokeh.models import HoverTool, ColumnDataSource
from bokeh.transform import linear_cmap
from bokeh.palettes import Viridis256
import numpy as np

# Example 5: Comprehensive Bokeh capabilities and features overview
print("=== Bokeh Comprehensive Capabilities Overview ===")

try:
    print(f"Bokeh version: {bokeh.__version__}")
    
    # Enable notebook output
    output_notebook()
    
    # Create comprehensive demonstration plot
    n = 1000
    x = np.random.random(n) * 100
    y = np.random.random(n) * 100
    colors = np.random.randint(0, 256, n)
    sizes = np.random.randint(8, 25, n)
    
    # Create figure with all available tools
    p = figure(
        title="Comprehensive Bokeh Features Demo",
        width=600,
        height=500,
        tools="pan,wheel_zoom,box_zoom,poly_select,lasso_select,reset,save,crosshair"
    )
    
    # Add advanced hover tooltip
    hover = HoverTool(tooltips=[
        ("Index", "$index"),
        ("Position", "($x{0.00}, $y{0.00})"),
        ("Size", "@size"),
        ("Color Value", "@color_val")
    ])
    p.add_tools(hover)
    
    # Create ColumnDataSource
    source = ColumnDataSource(data=dict(x=x, y=y, size=sizes, color_val=colors))
    
    # Create scatter with color mapping
    scatter = p.scatter(
        'x', 'y',
        size='size',
        color=linear_cmap('color_val', Viridis256, 0, 255),
        alpha=0.7,
        source=source
    )
    
    # Customize plot appearance
    p.title.text_font_size = "16pt"
    p.title.align = "center"
    p.xaxis.axis_label = "X Coordinate"
    p.yaxis.axis_label = "Y Coordinate"
    p.grid.grid_line_alpha = 0.3
    
    # Display the comprehensive plot
    show(p)
    
    print("\n=== Bokeh Key Features and Capabilities ===")
    print("✓ Interactive Web-Based Visualizations")
    print("  - HTML/JavaScript output for web browsers")
    print("  - Real-time interactivity without server requirements")
    print("  - Responsive design for different screen sizes")
    
    print("\n✓ Rich Plot Types and Glyphs")
    print("  - Lines, scatter plots, bars, areas, patches")
    print("  - Heatmaps, contour plots, network graphs")
    print("  - Custom and compound glyphs")
    
    print("\n✓ Advanced Interaction Tools")
    print("  - Pan, zoom, select, crosshair, hover")
    print("  - Box select, lasso select, polygon select")
    print("  - Reset, save, wheel zoom capabilities")
    
    print("\n✓ Widgets and Applications")
    print("  - Sliders, dropdowns, buttons, text inputs")
    print("  - JavaScript and Python callbacks")
    print("  - Server applications with Bokeh Server")
    
    print("\n✓ Data Integration")
    print("  - NumPy and Pandas integration")
    print("  - ColumnDataSource for efficient data handling")
    print("  - Streaming and real-time data updates")
    
    print("\n✓ Layouts and Dashboards")
    print("  - Grid layouts, tabs, panels")
    print("  - Multiple plot coordination")
    print("  - Responsive and adaptive designs")
    
    print("\n✓ Export and Deployment Options")
    print("  - Static HTML files, PNG/SVG export")
    print("  - Jupyter notebook integration")
    print("  - Standalone web applications")
    
    print(f"\nDemo plot created with {n} interactive data points")
    print("Bokeh comprehensive overview completed successfully!")
    
except Exception as e:
    print(f"Error in comprehensive overview: {e}")
    raise AssertionError(
        f"bokeh example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
```
