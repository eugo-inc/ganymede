## Example 1: boto3_stubs functionality test 1

```python
from typing import TYPE_CHECKING
import sys
import importlib.metadata
import importlib.util

# Example 1: Basic boto3-stubs package overview and type checking setup
print("=== Boto3-Stubs Type Hints Package Overview ===")

try:
    # boto3-stubs is a stub-only package (PEP 561): its .pyi files live in a `boto3-stubs/` directory,
    # which is not an importable module, so check the installed distribution's files instead.
    stub_files = [str(f) for f in (importlib.metadata.files('boto3-stubs') or [])
                  if str(f).startswith('boto3-stubs/') and str(f).endswith('.pyi')]
    assert stub_files, "boto3-stubs is installed but ships no boto3-stubs/*.pyi stub files"
    print(f"boto3-stubs {importlib.metadata.version('boto3-stubs')}: {len(stub_files)} .pyi stub files")
    print(f"Sample stub files: {sorted(stub_files)[:3]}")
    
    # Demonstrate type checking context
    print(f"\nTYPE_CHECKING constant: {TYPE_CHECKING}")
    print("TYPE_CHECKING is True when type checkers like mypy run")
    print("TYPE_CHECKING is False during runtime")
    
    # Show how type stubs work conceptually
    if TYPE_CHECKING:
        # This block only executes during type checking, not at runtime
        print("This would import type stubs during type checking")
    else:
        print("Runtime: Type stubs are not imported during execution")
    
    # Check for related packages
    related_packages = ['botocore', 'boto3']
    available_packages = []
    
    for package in related_packages:
        try:
            spec = importlib.util.find_spec(package)
            if spec is not None:
                available_packages.append(package)
        except:
            pass
            raise
    
    print(f"\nRelated AWS packages available: {available_packages}")
    
    print("\n=== Purpose of boto3-stubs ===")
    print("• Provides type annotations for boto3 and botocore")
    print("• Enables static type checking with mypy, PyCharm, VS Code")
    print("• Improves IDE autocomplete and error detection")
    print("• Does not affect runtime behavior")
    print("• Companion package to boto3 for development")
    print("• Contains .pyi files (Python Interface files)")
    
    print("Basic boto3-stubs overview completed successfully!")
    
except Exception as e:
    print(f"Error in basic boto3-stubs demonstration: {e}")
    raise AssertionError(
        f"boto3_stubs example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: boto3_stubs functionality test 3

```python
from typing import TYPE_CHECKING, Dict, List, Any
import boto3

# Example 3: Practical type annotation usage patterns with boto3-stubs
print("=== Boto3-Stubs Type Annotation Usage Patterns ===")

try:
    # Example of how type annotations would look with boto3-stubs
    # Note: These are demonstration patterns, actual imports happen during type checking
    
    print("=== Common Type Annotation Patterns ===")
    
    # Pattern 1: Client type annotations
    def create_s3_client_typed() -> Any:  # Would be S3Client with proper stubs
        """Create an S3 client with type hints."""
        client = boto3.client('s3', region_name='us-east-1')
        return client
    
    print("✓ Pattern 1: Client creation with return type annotation")
    s3_client = create_s3_client_typed()
    print(f"  Created client: {type(s3_client).__name__}")
    
    # Pattern 2: Response type annotations  
    def list_buckets_typed(client: Any) -> Dict[str, Any]:  # Would use proper response types
        """List S3 buckets with typed response."""
        try:
            # This would fail without credentials, but shows the pattern
            response = {'Buckets': [], 'Owner': {}}  # Mock response structure
            return response
        except Exception:
            return {'Buckets': [], 'Owner': {}}
    
    print("✓ Pattern 2: API response type annotations")
    response = list_buckets_typed(s3_client)
    print(f"  Response structure: {list(response.keys())}")
    
    # Pattern 3: Parameter type annotations
    def put_object_typed(
        client: Any,  # Would be S3Client
        bucket: str,
        key: str,
        body: bytes
    ) -> Dict[str, Any]:  # Would be PutObjectResponse
        """Put object with fully typed parameters."""
        print(f"  Would upload {len(body)} bytes to {bucket}/{key}")
        return {'ETag': '"mock-etag"', 'VersionId': 'mock-version'}
    
    print("✓ Pattern 3: Method parameter type annotations")
    mock_result = put_object_typed(s3_client, "test-bucket", "test-key", b"test data")
    print(f"  Mock result: {mock_result}")
    
    # Pattern 4: Resource type annotations
    def create_dynamodb_resource_typed() -> Any:  # Would be DynamoDBServiceResource
        """Create DynamoDB resource with type hints."""
        resource = boto3.resource('dynamodb', region_name='us-east-1')
        return resource
    
    print("✓ Pattern 4: Resource creation with type annotations")
    dynamodb_resource = create_dynamodb_resource_typed()
    print(f"  Created resource: {type(dynamodb_resource).__name__}")
    
    print("\n=== Benefits of Type Annotations ===")
    print("• IDE autocomplete for methods and parameters")
    print("• Static type checking catches errors before runtime")
    print("• Better code documentation and readability")
    print("• IntelliSense support in editors")
    print("• Reduced debugging time for AWS API calls")
    
    print("Type annotation usage patterns completed successfully!")
    
except Exception as e:
    print(f"Error in type annotation patterns: {e}")
    raise AssertionError(
        f"boto3_stubs example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```
