## Example 3: botocore_stubs functionality test 3

```python
from typing import TYPE_CHECKING, Dict, List, Any, Union, Optional
import botocore.session
import botocore.config

# Example 3: Type annotation patterns for botocore low-level operations
print("=== Botocore-Stubs Type Annotation Patterns ===")

try:
    # Demonstrate type-safe botocore usage patterns
    # Note: These show patterns that would be type-checked with stubs
    
    print("=== Session Type Annotation Patterns ===")
    
    # Pattern 1: Session creation with type hints
    def create_typed_session(
        profile_name: Optional[str] = None,
        region_name: Optional[str] = None
    ) -> Any:  # Would be botocore.session.Session with proper stubs
        """Create a botocore session with type annotations."""
        session = botocore.session.Session(profile=profile_name)
        if region_name:
            session.set_config_variable('region', region_name)
        return session
    
    print("✓ Pattern 1: Session creation with optional parameters")
    session = create_typed_session(region_name='us-east-1')
    print(f"  Created session: {type(session).__name__}")
    
    # Pattern 2: Client creation with configuration
    def create_typed_client(
        session: Any,  # Would be Session type
        service_name: str,
        region_name: Optional[str] = None,
        config: Optional[Any] = None  # Would be Config type
    ) -> Any:  # Would be BaseClient type
        """Create a service client with full type annotations."""
        client = session.create_client(
            service_name, 
            region_name=region_name,
            config=config
        )
        return client
    
    print("✓ Pattern 2: Client creation with typed parameters")
    s3_client = create_typed_client(session, 's3', 'us-east-1')
    print(f"  Created client: {type(s3_client).__name__}")
    
    # Pattern 3: Operation introspection with typed parameters
    def inspect_service_operations(
        client: Any,  # Would be BaseClient
        service_name: str
    ) -> Dict[str, Any]:  # Would be specific response type
        """Inspect service operations with type safety."""
        try:
            # Get service model information
            service_model = client._service_model
            operations = list(service_model.operation_names)
            
            return {
                'service_name': service_model.service_name,
                'protocol': service_model.protocol,
                'operations_count': len(operations),
                'sample_operations': operations[:5]
            }
        except Exception as e:
            return {'Error': {'Code': 'InspectionError', 'Message': str(e)}}
    
    print("✓ Pattern 3: Service operation introspection")
    service_info = inspect_service_operations(s3_client, 's3')
    print(f"  Service info: {service_info.get('service_name', 'Unknown')}")
    print(f"  Protocol: {service_info.get('protocol', 'Unknown')}")
    print(f"  Operations count: {service_info.get('operations_count', 0)}")
    
    # Pattern 4: Configuration with typed parameters
    def create_typed_config(
        region_name: Optional[str] = None,
        max_attempts: Optional[int] = None,
        connect_timeout: Optional[float] = None
    ) -> Any:  # Would be Config type
        """Create configuration with typed parameters."""
        config_dict = {}
        if region_name:
            config_dict['region_name'] = region_name
        if max_attempts:
            config_dict['retries'] = {'max_attempts': max_attempts}
        if connect_timeout:
            config_dict['connect_timeout'] = connect_timeout
        
        return botocore.config.Config(**config_dict)
    
    print("✓ Pattern 4: Configuration creation with types")
    config = create_typed_config('us-west-2', 3, 10.0)
    print(f"  Created config: {type(config).__name__}")
    print(f"  Config region: {config.region_name}")
    
    print("\n=== Advanced Typing Patterns ===")
    advanced_patterns = [
        "Generic client types for service-specific operations",
        "Union types for multiple response formats",
        "Protocol types for credential provider interfaces",
        "Literal types for enumerated values (regions, protocols)",
        "TypedDict for structured response payloads",
        "Callable types for event handler functions"
    ]
    
    for pattern in advanced_patterns:
        print(f"• {pattern}")
    
    print("Type annotation patterns demonstration completed successfully!")
    
except Exception as e:
    print(f"Error in type annotation patterns: {e}")
    raise AssertionError(
        f"botocore_stubs example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```
