## Example 1: bottleneck functionality test 1

```python
import bottleneck as bn
import numpy as np
import time

# Example 1: Basic bottleneck operations and performance comparison
print("=== Bottleneck Basic Operations and Performance ===")

try:
    # Show bottleneck version
    print(f"Bottleneck version: {bn.__version__}")
    
    # Create sample data
    data = np.random.rand(100000)
    print(f"Created array with {len(data)} elements")
    
    # Compare bottleneck vs NumPy performance
    print("\n=== Performance Comparison: bottleneck vs NumPy ===")
    
    # Test nanmean performance
    data_with_nans = data.copy()
    data_with_nans[::1000] = np.nan  # Add some NaN values
    
    # NumPy nanmean
    start_time = time.time()
    numpy_result = np.nanmean(data_with_nans)
    numpy_time = time.time() - start_time
    
    # Bottleneck nanmean
    start_time = time.time()
    bottleneck_result = bn.nanmean(data_with_nans)
    bottleneck_time = time.time() - start_time
    
    print(f"Array size: {len(data_with_nans):,} elements")
    print(f"NaN values: {np.isnan(data_with_nans).sum()} ({np.isnan(data_with_nans).sum()/len(data_with_nans)*100:.1f}%)")
    print(f"NumPy nanmean: {numpy_result:.6f} (time: {numpy_time:.6f}s)")
    print(f"Bottleneck nanmean: {bottleneck_result:.6f} (time: {bottleneck_time:.6f}s)")
    print(f"Speedup: {numpy_time/bottleneck_time:.1f}x faster")
    print(f"Results match: {np.isclose(numpy_result, bottleneck_result)}")
    
    # Show available functions
    bn_functions = [attr for attr in dir(bn) if not attr.startswith('_') and callable(getattr(bn, attr))]
    print(f"\nBottleneck provides {len(bn_functions)} optimized functions")
    print(f"Sample functions: {bn_functions[:8]}")
    
    print("Basic bottleneck operations completed successfully!")
    
except Exception as e:
    print(f"Error in basic bottleneck operations: {e}")
    raise AssertionError(
        f"bottleneck example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 2: bottleneck functionality test 2

```python
import bottleneck as bn
import numpy as np

# Example 2: NaN-aware statistical functions
print("=== Bottleneck NaN-Aware Statistical Functions ===")

try:
    # Create test data with missing values
    np.random.seed(42)
    data = np.random.randn(1000)
    # Introduce some NaN values
    nan_indices = np.random.choice(len(data), size=50, replace=False)
    data[nan_indices] = np.nan
    
    print(f"Dataset: {len(data)} values, {np.isnan(data).sum()} NaN values")
    
    # Test various NaN-aware functions
    print("\n=== NaN-Aware Statistics ===")
    
    functions_to_test = [
        ('nanmean', 'Mean ignoring NaN values'),
        ('nanstd', 'Standard deviation ignoring NaN'),
        ('nanvar', 'Variance ignoring NaN'),
        ('nanmin', 'Minimum value ignoring NaN'),
        ('nanmax', 'Maximum value ignoring NaN'),
        ('nansum', 'Sum ignoring NaN values'),
        ('nanmedian', 'Median ignoring NaN')
    ]
    
    for func_name, description in functions_to_test:
        if hasattr(bn, func_name):
            bn_func = getattr(bn, func_name)
            result = bn_func(data)
            print(f"{func_name:>10}: {result:.6f} - {description}")
    
    # Test with 2D arrays
    print("\n=== 2D Array Operations ===")
    data_2d = np.random.randn(50, 100)
    data_2d[np.random.rand(50, 100) < 0.1] = np.nan  # 10% NaN values
    
    print(f"2D array shape: {data_2d.shape}")
    print(f"Total NaN values: {np.isnan(data_2d).sum()}")
    
    # Axis-specific operations
    mean_axis0 = bn.nanmean(data_2d, axis=0)  # Mean along columns
    mean_axis1 = bn.nanmean(data_2d, axis=1)  # Mean along rows
    
    print(f"Column means shape: {mean_axis0.shape}")
    print(f"Row means shape: {mean_axis1.shape}")
    print(f"First 5 column means: {mean_axis0[:5]}")
    
    # Count non-NaN values
    count_nonnan = bn.nansum(~np.isnan(data_2d), axis=0)
    print(f"Non-NaN counts per column (first 5): {count_nonnan[:5].astype(int)}")
    
    print("NaN-aware statistical functions completed successfully!")
    
except Exception as e:
    print(f"Error in NaN-aware statistics: {e}")
    raise AssertionError(
        f"bottleneck example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: bottleneck functionality test 3

```python
import bottleneck as bn
import numpy as np

# Example 3: Moving window functions and ranking operations
print("=== Bottleneck Moving Window Functions ===")

try:
    # Create time series data
    np.random.seed(42)
    time_series = np.cumsum(np.random.randn(1000) * 0.1) + 100  # Random walk
    print(f"Time series length: {len(time_series)}")
    print(f"Value range: {time_series.min():.3f} to {time_series.max():.3f}")
    
    # Moving window statistics
    print("\n=== Moving Window Statistics ===")
    
    window_size = 20
    
    # Moving average (mean)
    moving_mean = bn.move_mean(time_series, window=window_size)
    print(f"Moving mean (window={window_size}): computed {len(moving_mean)} values")
    print(f"First 5 moving means: {moving_mean[window_size-1:window_size+4]}")
    
    # Moving standard deviation
    moving_std = bn.move_std(time_series, window=window_size)
    print(f"Moving std (window={window_size}): computed {len(moving_std)} values")
    print(f"First 5 moving stds: {moving_std[window_size-1:window_size+4]}")
    
    # Moving min and max
    moving_min = bn.move_min(time_series, window=window_size)
    moving_max = bn.move_max(time_series, window=window_size)
    print(f"Moving range at position {window_size}: [{moving_min[window_size]:.3f}, {moving_max[window_size]:.3f}]")
    
    # Moving sum
    moving_sum = bn.move_sum(time_series, window=window_size)
    print(f"Moving sum at position {window_size}: {moving_sum[window_size]:.3f}")
    
    # Ranking operations
    print("\n=== Ranking Operations ===")
    
    sample_data = np.array([3.2, 1.1, 4.4, 2.2, 5.5, 1.1, 3.3])
    print(f"Sample data: {sample_data}")
    
    # Rank data (1-based ranking)
    ranks = bn.rankdata(sample_data)
    print(f"Ranks: {ranks}")
    
    # API correction: `method=` is SciPy's rankdata signature, not bottleneck's. Bottleneck is
    # `rankdata(a, axis=None)` and always averages ties — passing `method` raises
    # `TypeError: wrong number of keyword arguments`. What bottleneck does offer is per-axis ranking.
    ranks_axis0 = bn.rankdata(sample_data.reshape(-1, 1), axis=0).ravel()
    
    print(f"Rank (ties averaged - the only method bottleneck implements): {ranks}")
    print(f"Rank along axis 0: {ranks_axis0}")
    
    # Nanrankdata for arrays with NaN values
    data_with_nans = sample_data.copy()
    data_with_nans[1] = np.nan
    data_with_nans[4] = np.nan
    
    print(f"Data with NaNs: {data_with_nans}")
    nan_ranks = bn.nanrankdata(data_with_nans)
    print(f"NaN-aware ranks: {nan_ranks}")
    
    # Moving window with NaN handling
    print("\n=== Moving Windows with NaN Values ===")
    ts_with_nans = time_series.copy()
    ts_with_nans[100:110] = np.nan  # Insert NaN block
    
    # API drift: the `move_nan*` family is gone. NaN handling is now a parameter of the ordinary
    # moving function: `min_count` is how many non-NaN observations a window needs to produce a
    # value, so `min_count=1` is the old nan-mean behaviour.
    moving_nanmean = bn.move_mean(ts_with_nans, window=window_size, min_count=1)
    print(f"Moving mean with NaN handling around NaN region:")
    print(f"Positions 95-115: {moving_nanmean[95:116]}")
    
    print("Moving window functions completed successfully!")
    
except Exception as e:
    print(f"Error in moving window functions: {e}")
    raise AssertionError(
        f"bottleneck example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: bottleneck functionality test 4

```python
import bottleneck as bn
import numpy as np
import time

# Example 4: Performance benchmarking and real-world applications
print("=== Bottleneck Performance Benchmarking ===")

try:
    # Large dataset performance comparison
    print("=== Large Dataset Performance Test ===")
    
    # Create large datasets
    sizes = [10000, 100000, 1000000]
    
    for size in sizes:
        print(f"\nArray size: {size:,} elements")
        
        # Create data with missing values
        data = np.random.randn(size)
        data[np.random.rand(size) < 0.05] = np.nan  # 5% missing values
        
        # Test multiple functions
        functions_to_test = [
            ('nanmean', np.nanmean, bn.nanmean),
            ('nanstd', np.nanstd, bn.nanstd),
            ('nansum', np.nansum, bn.nansum)
        ]
        
        for func_name, np_func, bn_func in functions_to_test:
            # NumPy timing
            start = time.time()
            np_result = np_func(data)
            np_time = time.time() - start
            
            # Bottleneck timing
            start = time.time()
            bn_result = bn_func(data)
            bn_time = time.time() - start
            
            speedup = np_time / bn_time if bn_time > 0 else float('inf')
            
            print(f"  {func_name}: {speedup:.1f}x speedup (NumPy: {np_time:.4f}s, BN: {bn_time:.4f}s)")
    
    # Real-world application: Financial data analysis
    print("\n=== Real-World Application: Financial Time Series ===")
    
    # Simulate stock price data
    np.random.seed(42)
    n_days = 1000
    n_stocks = 100
    
    # Generate correlated stock returns
    base_returns = np.random.randn(n_days) * 0.02
    stock_returns = np.random.randn(n_days, n_stocks) * 0.01 + base_returns.reshape(-1, 1)
    
    # Convert to price levels
    initial_prices = 100 * np.random.rand(n_stocks) + 50
    stock_prices = np.cumprod(1 + stock_returns, axis=0) * initial_prices
    
    # Introduce some missing data (market holidays, etc.)
    missing_mask = np.random.rand(n_days, n_stocks) < 0.02
    stock_prices[missing_mask] = np.nan
    
    print(f"Stock data: {n_days} days × {n_stocks} stocks")
    print(f"Missing data points: {np.isnan(stock_prices).sum()} ({np.isnan(stock_prices).mean()*100:.2f}%)")
    
    # Calculate financial metrics
    start_time = time.time()
    
    # Daily returns (ignoring missing data)
    daily_returns = np.diff(stock_prices, axis=0) / stock_prices[:-1]
    
    # Portfolio statistics
    portfolio_mean = bn.nanmean(stock_prices, axis=1)  # Equal-weighted portfolio
    portfolio_volatility = bn.move_std(portfolio_mean, window=30)  # 30-day rolling volatility
    
    # Stock-level statistics
    stock_means = bn.nanmean(stock_prices, axis=0)
    stock_stds = bn.nanstd(stock_prices, axis=0)
    
    # Moving averages for technical analysis
    short_ma = bn.move_mean(portfolio_mean, window=10)
    long_ma = bn.move_mean(portfolio_mean, window=50)
    
    analysis_time = time.time() - start_time
    
    print(f"Analysis completed in {analysis_time:.4f} seconds")
    print(f"Portfolio mean price: {bn.nanmean(portfolio_mean):.2f}")
    print(f"Portfolio volatility (30-day): {bn.nanmean(portfolio_volatility[30:]):.4f}")
    print(f"Stock price range: {bn.nanmin(stock_means):.2f} - {bn.nanmax(stock_means):.2f}")
    
    # Trading signal example
    signal_strength = (short_ma - long_ma) / stock_stds.mean()
    recent_signals = signal_strength[-10:]
    print(f"Recent trading signals: {recent_signals[~np.isnan(recent_signals)]}")
    
    print("Performance benchmarking completed successfully!")
    
except Exception as e:
    print(f"Error in performance benchmarking: {e}")
    raise AssertionError(
        f"bottleneck example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```
