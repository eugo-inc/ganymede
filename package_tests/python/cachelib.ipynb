## Example 1: Simple Memory Cache Operations

```python
from cachelib import SimpleCache
import time

print("Example 1: Simple Memory Cache Operations")

try:
    # Create a simple in-memory cache
    cache = SimpleCache()
    
    print("✓ Simple cache created")
    print(f"Default timeout: {cache.default_timeout} seconds")
    
    # Store various types of values
    cache.set('user:123', {'name': 'John Doe', 'email': 'john@example.com'})
    cache.set('session:abc', 'active_session_data', timeout=300)
    cache.set('counter', 42)
    cache.set('config:app', {'debug': True, 'version': '1.0.0'})
    
    print("✓ Stored 4 cache entries with different data types")
    
    # Retrieve values
    user_data = cache.get('user:123')
    session_data = cache.get('session:abc')
    counter = cache.get('counter')
    config = cache.get('config:app')
    
    print(f"✓ Retrieved user data: {user_data}")
    print(f"✓ Retrieved session: {session_data}")
    print(f"✓ Retrieved counter: {counter}")
    print(f"✓ Retrieved config: {config}")
    
    # Test cache miss
    missing = cache.get('nonexistent:key')
    print(f"✓ Cache miss returns: {missing}")
    
    # Handle missing keys with manual default
    value = cache.get('missing:key')
    default_value = value if value is not None else 'default_value'
    print(f"✓ Manual default handling: {default_value}")
    
    # Check if key exists
    exists = cache.has('user:123')
    not_exists = cache.has('fake:key')
    print(f"✓ Key 'user:123' exists: {exists}")
    print(f"✓ Key 'fake:key' exists: {not_exists}")
    
    # Delete a key
    deleted = cache.delete('counter')
    print(f"✓ Deleted 'counter': {deleted}")
    
    # Verify deletion
    counter_after = cache.get('counter')
    print(f"✓ Counter after deletion: {counter_after}")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"cachelib example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 2: Multiple Cache Backends Comparison

```python
from cachelib import SimpleCache, NullCache, BaseCache
import tempfile

print("Example 2: Multiple Cache Backends Comparison")

try:
    # Test data
    test_key = 'comparison:test'
    test_value = {'message': 'Hello cachelib!', 'count': 42}
    
    # Create different cache instances
    memory_cache = SimpleCache()
    null_cache = NullCache()
    
    print("✓ Created multiple cache backends")
    
    # Test all cache backends with same operations
    caches = {
        'SimpleCache': memory_cache,
        'NullCache': null_cache,
    }
    
    print("\nTesting set/get operations:")
    for name, cache in caches.items():
        # Set value
        set_result = cache.set(test_key, test_value)
        
        # Get value
        get_result = cache.get(test_key)
        
        # Check existence
        has_result = cache.has(test_key)
        
        print(f"✓ {name}:")
        print(f"  - Set: {set_result}")
        print(f"  - Get: {get_result}")
        print(f"  - Has: {has_result}")
        print(f"  - Type: {type(cache).__name__}")
    
    # Show available cache types
    available_caches = [
        'SimpleCache', 'FileSystemCache', 'NullCache', 
        'RedisCache', 'MemcachedCache', 'DynamoDbCache',
        'MongoDbCache', 'UWSGICache'
    ]
    
    print(f"\n✓ Available cache backends in cachelib:")
    for cache_type in available_caches:
        print(f"  - {cache_type}")
    
    # Show BaseCache inheritance
    print(f"\n✓ All caches inherit from: {BaseCache.__name__}")
    print(f"✓ Unified API across all backends")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"cachelib example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: Cache with Timeout and Expiration

```python
from cachelib import SimpleCache
import time

print("Example 3: Cache with Timeout and Expiration")

try:
    # Create cache with custom default timeout
    cache = SimpleCache(default_timeout=2)  # 2 second default
    
    print(f"✓ Cache with {cache.default_timeout}s default timeout")
    
    # Store values with different timeout strategies
    cache.set('short:lived', 'expires soon', timeout=1)  # 1 second
    cache.set('medium:lived', 'expires later', timeout=3)  # 3 seconds  
    cache.set('default:timeout', 'uses default')  # Uses default (2s)
    cache.set('no:timeout', 'never expires', timeout=0)  # Never expires
    
    print("✓ Stored values with different timeouts")
    
    # Check initial values
    print(f"✓ Short lived: {cache.get('short:lived')}")
    print(f"✓ Medium lived: {cache.get('medium:lived')}")
    print(f"✓ Default timeout: {cache.get('default:timeout')}")
    print(f"✓ No timeout: {cache.get('no:timeout')}")
    
    # Wait for short timeout to expire
    print("Waiting 1.5 seconds...")
    time.sleep(1.5)
    
    print("After 1.5 seconds:")
    print(f"✓ Short lived: {cache.get('short:lived')} (should be None)")
    print(f"✓ Medium lived: {cache.get('medium:lived')} (should exist)")
    print(f"✓ Default timeout: {cache.get('default:timeout')} (should exist)")
    print(f"✓ No timeout: {cache.get('no:timeout')} (should exist)")
    
    # Wait for more timeouts
    print("Waiting another 2 seconds...")
    time.sleep(2)
    
    print("After 3.5 seconds total:")
    print(f"✓ Medium lived: {cache.get('medium:lived')} (should be None)")
    print(f"✓ Default timeout: {cache.get('default:timeout')} (should be None)")
    print(f"✓ No timeout: {cache.get('no:timeout')} (should still exist)")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"cachelib example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: NullCache for Development

```python
from cachelib import NullCache

print("Example 4: NullCache for Development")

try:
    # Create a null cache (stores nothing)
    null_cache = NullCache()
    
    print("✓ NullCache created")
    print(f"Default timeout: {null_cache.default_timeout}")
    
    # Attempt to store values (will not be stored)
    result1 = null_cache.set('test:key1', 'value1')
    result2 = null_cache.set('test:key2', {'data': 'value2'})
    result3 = null_cache.set('test:key3', [1, 2, 3])
    
    print(f"✓ Set operations return: {result1}, {result2}, {result3}")
    
    # Try to retrieve values (will always return None)
    value1 = null_cache.get('test:key1')
    value2 = null_cache.get('test:key2')
    value3 = null_cache.get('test:key3')
    
    print(f"✓ Get key1: {value1}")
    print(f"✓ Get key2: {value2}")
    print(f"✓ Get key3: {value3}")
    
    # Test has() method (will always return False)
    has1 = null_cache.has('test:key1')
    has2 = null_cache.has('nonexistent')
    
    print(f"✓ Has key1: {has1}")
    print(f"✓ Has nonexistent: {has2}")
    
    # Test delete() method
    deleted = null_cache.delete('test:key1')
    print(f"✓ Delete operation: {deleted}")
    
    print("✓ NullCache is useful for disabling caching during development")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"cachelib example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 5: FileSystem Cache

```python
from cachelib import FileSystemCache
import tempfile
import shutil
import os

print("Example 5: FileSystem Cache")

try:
    # Create a temporary directory for cache
    cache_dir = tempfile.mkdtemp()
    print(f"✓ Created cache directory: {cache_dir}")
    
    # Create filesystem cache
    fs_cache = FileSystemCache(cache_dir)
    
    print("✓ FileSystem cache created")
    print(f"Cache directory: {fs_cache._path}")
    
    # Store data persistently
    fs_cache.set('persistent:user', {'id': 456, 'name': 'Jane Smith'})
    fs_cache.set('persistent:settings', {'theme': 'dark', 'lang': 'en'})
    fs_cache.set('large:data', list(range(100)))
    
    print("✓ Stored persistent data in filesystem")
    
    # List files created
    cache_files = os.listdir(cache_dir)
    print(f"✓ Cache files created: {len(cache_files)} files")
    
    # Retrieve data
    user = fs_cache.get('persistent:user')
    settings = fs_cache.get('persistent:settings')
    large_data = fs_cache.get('large:data')
    
    print(f"✓ Retrieved user: {user}")
    print(f"✓ Retrieved settings: {settings}")
    print(f"✓ Retrieved large data length: {len(large_data) if large_data else 0}")
    
    # Test cache operations
    exists = fs_cache.has('persistent:user')
    print(f"✓ User exists: {exists}")
    
    deleted = fs_cache.delete('large:data')
    print(f"✓ Deleted large data: {deleted}")
    
    # Clean up
    shutil.rmtree(cache_dir, ignore_errors=True)
    print("✓ Cache directory cleaned up")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"cachelib example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
```
