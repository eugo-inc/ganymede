# cachetools: 5 Basic Usage Examples

```python
# Example 1: Using LRUCache
from cachetools import LRUCache
cache = LRUCache(maxsize=2)
cache['a'] = 1
cache['b'] = 2
cache['c'] = 3
print(list(cache.items()))  # Only last 2 items kept
```

```python
# Example 2: Using TTLCache
from cachetools import TTLCache
import time
cache = TTLCache(maxsize=2, ttl=1)
cache['a'] = 1
time.sleep(1.1)
print('a' in cache)  # Should be False
```

```python
# Example 3: Using cached decorator
from cachetools import cached, LRUCache
cache = LRUCache(maxsize=2)
@cached(cache)
def square(x):
    print(f'Calculating {x}^2')
    return x * x
print(square(2))
print(square(2))  # Cached
```

```python
# Example 4: Using keys module
from cachetools import keys
print(keys.hashkey('a', 1, b=2))
```

```python
# Example 5: Using LFUCache
from cachetools import LFUCache
cache = LFUCache(maxsize=2)
cache['a'] = 1
cache['b'] = 2
cache['a']
cache['c'] = 3
print(list(cache.items()))  # Least frequently used evicted
```
