## Example 1: Basic Cache with Memory Management

```python
from cachey import Cache

print("Example 1: Basic Cache with Memory Management")

try:
    # Create a cache with memory limit (100KB)
    cache = Cache(available_bytes=100*1024)
    
    print(f"✓ Cache created with {cache.available_bytes:,} bytes limit")
    print(f"Initial total bytes: {cache.total_bytes:,}")
    print(f"Cache scorer type: {type(cache.scorer).__name__}")
    
    # Store items with put() method (requires cost parameter)
    small_data = [1, 2, 3, 4, 5]
    medium_data = list(range(100))
    large_data = list(range(1000))
    
    # Put items with different costs
    cache.put('small', small_data, cost=1)    # Low cost
    cache.put('medium', medium_data, cost=5)  # Medium cost  
    cache.put('large', large_data, cost=20)   # High cost
    
    print(f"✓ Stored 3 items with different costs")
    print(f"Total bytes used: {cache.total_bytes:,}")
    
    # Retrieve items with get() method
    small = cache.get('small')
    medium = cache.get('medium')
    large = cache.get('large')
    
    print(f"✓ Retrieved small data: {len(small) if small else 0} items")
    print(f"✓ Retrieved medium data: {len(medium) if medium else 0} items")
    print(f"✓ Retrieved large data: {len(large) if large else 0} items")
    
    # Check cache statistics
    print(f"✓ Items in cache: {len(cache.data)}")
    print(f"✓ Cache memory limit: {cache.limit:,} bytes")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"cachey example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 2: Custom Cost-Based Caching

```python
from cachey import Cache, Scorer

print("Example 2: Custom Cost-Based Caching")

try:
    # Create a cache with scorer for cost-based eviction
    cache = Cache(available_bytes=512*1024)
    
    print(f"✓ Cache with default scorer created")
    print(f"Available bytes: {cache.available_bytes:,}")
    print(f"Scorer type: {type(cache.scorer).__name__}")
    
    # Store items with different costs using put() method
    cache.put('cheap_data', list(range(100)), cost=1)    # Low cost
    cache.put('medium_data', list(range(200)), cost=10)  # Medium cost  
    cache.put('expensive_data', list(range(300)), cost=100)  # High cost
    
    print("✓ Stored items with different computational costs")
    print(f"Total bytes used: {cache.total_bytes:,}")
    
    # Access items to affect their scores
    cheap = cache.get('cheap_data')  # Low cost, frequent access
    expensive = cache.get('expensive_data')  # High cost, preserve in cache
    
    print("✓ Accessed items to update cache scores")
    print(f"Retrieved cheap data: {len(cheap) if cheap else 0} items")
    print(f"Retrieved expensive data: {len(expensive) if expensive else 0} items")
    
    # Show cache state
    print(f"Items in cache: {len(cache.data)}")
    cache_keys = list(cache.data.keys())
    for key in cache_keys:
        print(f"  - {key}: stored in cache")
    
    # Test with more data to trigger eviction based on cost
    for i in range(5):
        key = f'temp_item_{i}'
        cache.put(key, list(range(500)), cost=5)  # Medium cost items
    
    print(f"✓ Added temporary items, final cache has {len(cache.data)} items")
    print("✓ Cost-based caching manages items by access patterns and cost")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"cachey example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: Cache Eviction and Memory Pressure

```python
from cachey import Cache
import time

print("Example 3: Cache Eviction and Memory Pressure")

try:
    # Create a small cache to demonstrate eviction
    cache = Cache(available_bytes=1024)  # Very small: 1KB
    
    print(f"✓ Small cache created: {cache.available_bytes:,} bytes")
    
    # Fill cache beyond capacity using put() method
    data_items = {}
    for i in range(6):  # Reduced to 6 for cleaner output
        key = f'item_{i}'
        data = [1] * 50  # Each item takes some space
        data_items[key] = data
        cache.put(key, data, cost=1)  # Use put() method with cost
        print(f"Added {key}, total bytes: {cache.total_bytes}")
        
        # Check if items were evicted
        remaining_keys = list(cache.data.keys())
        print(f"  Keys in cache: {len(remaining_keys)} - {remaining_keys}")
    
    print("✓ Cache automatically evicted items when memory limit exceeded")
    
    # Try to access evicted items using get() method
    for key in ['item_0', 'item_1', 'item_2']:
        value = cache.get(key)
        if value is not None:
            print(f"✓ {key} still in cache: {len(value)} items")
        else:
            print(f"✗ {key} was evicted from cache")
    
    # Show final cache state
    print(f"Final cache state:")
    print(f"  Total bytes: {cache.total_bytes}")
    print(f"  Items remaining: {len(cache.data)}")
    print(f"  Keys: {list(cache.data.keys())}")
    
    # Demonstrate cache pressure with larger items
    print(f"\nTesting with larger items:")
    large_item = list(range(200))  # Larger item
    cache.put('large_item', large_item, cost=10)
    print(f"Added large item, total bytes: {cache.total_bytes}")
    print(f"Remaining keys: {list(cache.data.keys())}")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"cachey example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: NBBytes Function for Size Calculation

```python
from cachey import nbytes
import numpy as np

print("Example 4: NBBytes Function for Size Calculation")

try:
    # Test nbytes function with different data types
    data_samples = {
        'small_list': [1, 2, 3, 4, 5],
        'large_list': list(range(1000)),
        'string_data': 'Hello, this is a test string!',
        'dict_data': {'a': 1, 'b': 2, 'c': list(range(100))},
        'nested_structure': {'users': [{'id': i, 'name': f'user_{i}'} for i in range(50)]}
    }
    
    print("✓ Testing nbytes function with various data types:")
    total_bytes = 0
    
    for name, data in data_samples.items():
        size = nbytes(data)
        total_bytes += size
        print(f"  {name}: {size:,} bytes")
    
    print(f"✓ Total size calculated: {total_bytes:,} bytes")
    
    # Test with numpy arrays if available
    try:
        arr_1d = np.array(range(1000))
        arr_2d = np.random.random((100, 100))
        
        size_1d = nbytes(arr_1d)
        size_2d = nbytes(arr_2d)
        
        print(f"✓ NumPy 1D array (1000 elements): {size_1d:,} bytes")
        print(f"✓ NumPy 2D array (100x100): {size_2d:,} bytes")
        
    except ImportError:
        print("✓ NumPy not available, skipping array size tests")
    
    # Compare with sys.getsizeof
    import sys
    test_data = list(range(100))
    cachey_size = nbytes(test_data)
    sys_size = sys.getsizeof(test_data)
    
    print(f"✓ Comparison for list(range(100)):")
    print(f"  cachey.nbytes: {cachey_size} bytes")
    print(f"  sys.getsizeof: {sys_size} bytes")
    
    print("✓ nbytes provides accurate memory size calculation for cache management")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"cachey example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 5: Cache Decorator Usage

```python
from cachey import Cache
import time

print("Example 5: Cache Memoization Usage")

try:
    # Create a cache instance
    cache_instance = Cache(available_bytes=1024*1024)  # 1MB cache
    
    print("✓ Created cache instance for memoization")
    
    # Define functions to be memoized
    def expensive_computation(n):
        """Simulate an expensive computation"""
        print(f"Computing expensive result for n={n}...")
        time.sleep(0.1)  # Simulate work
        return sum(range(n * 100))
    
    def fibonacci(n):
        """Fibonacci calculation"""
        if n <= 1:
            return n
        return fibonacci(n-1) + fibonacci(n-2)
    
    # Memoize the functions using cache
    memo_expensive = cache_instance.memoize(expensive_computation)
    memo_fibonacci = cache_instance.memoize(fibonacci)
    
    print("✓ Created memoized functions")
    
    # Test expensive computation caching
    print("\nTesting memoized expensive_computation:")
    result1 = memo_expensive(10)
    print(f"✓ First call result: {result1}")
    
    result2 = memo_expensive(10)  # Should be cached
    print(f"✓ Second call result (cached): {result2}")
    
    result3 = memo_expensive(20)  # New computation
    print(f"✓ Different input result: {result3}")
    
    # Test fibonacci with memoization
    print("\nTesting memoized fibonacci:")
    fib_start = time.time()
    fib_result = memo_fibonacci(15)
    fib_time = time.time() - fib_start
    
    print(f"✓ fibonacci(15) = {fib_result}")
    print(f"✓ Computation time: {fib_time:.4f} seconds")
    
    # Second call should be much faster  
    fib_start2 = time.time()
    fib_result2 = memo_fibonacci(15)
    fib_time2 = time.time() - fib_start2
    
    print(f"✓ Second fibonacci(15) = {fib_result2}")
    print(f"✓ Cached computation time: {fib_time2:.6f} seconds")
    
    # Show cache statistics
    print(f"\n✓ Cache contains {len(cache_instance.data)} items")
    print(f"✓ Total bytes used: {cache_instance.total_bytes:,}")
    
    print("✓ Cache memoization provides automatic caching with memory management")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"cachey example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
```
