## Example 1: Basic Version Management and Schema Evolution

```python
import cadwyn
from cadwyn import Cadwyn, VersionChange, Version, VersionBundle
from pydantic import BaseModel
from datetime import date

# Example 1: Basic Version Management and Schema Evolution
print("Example 1: Basic Version Management")

try:
    # Define base model
    class UserSchema(BaseModel):
        id: int
        name: str
        email: str
        created_at: date
    
    # Define version changes
    class VersionChange_2024_01_15(VersionChange):
        description = "Added user roles"
        instructions_to_migrate_to_previous_version = [
            # This would contain migration instructions
        ]
    
    # Create version bundle with versions in DESCENDING order (newest to oldest)
    versions = VersionBundle(
        Version(date(2024, 3, 1)),  # Newest first
        Version(date(2024, 2, 1)),  # Middle
        Version(date(2024, 1, 1))   # Oldest last
    )
    
    # Create Cadwyn instance with required versions argument
    cadwyn_app = Cadwyn(versions=versions)
    
    print("✓ Cadwyn instance created with version bundle")
    print(f"Cadwyn version: {cadwyn.__version__ if hasattr(cadwyn, '__version__') else 'N/A'}")
    print(f"Version bundle type: {type(versions).__name__}")
    
    # Test version registration
    print("✓ Version management features:")
    print("  - Automatic schema versioning")
    print("  - Migration support")
    print("  - FastAPI integration")
    print("  - Pydantic model evolution")
    
    # Create sample schema instances
    user_v1 = UserSchema(
        id=1,
        name="Alice Smith",
        email="alice@example.com",
        created_at=date(2024, 1, 1)
    )
    
    print(f"✓ User schema v1: {user_v1}")
    
    # Test schema validation
    user_data = {
        "id": 2,
        "name": "Bob Johnson", 
        "email": "bob@example.com",
        "created_at": "2024-01-15"
    }
    
    validated_user = UserSchema(**user_data)
    print(f"✓ Validated user: {validated_user}")
    
    # Demonstrate version change tracking
    print("✓ Version change information:")
    print(f"  Description: {VersionChange_2024_01_15.description}")
    print(f"  Migration instructions available: {len(VersionChange_2024_01_15.instructions_to_migrate_to_previous_version)}")
    
    print("✓ Cadwyn provides:")
    print("  - Automatic API versioning")
    print("  - Schema migration tracking")
    print("  - Backward compatibility")
    print("  - Descending version order (newest to oldest)")
    
except Exception as e:
    print(f"Error in version management: {e}")
    raise AssertionError(
        f"cadwyn example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 completed\n")
```

## Example 2: Schema Generation and Migration

```python
from cadwyn import generate_versioned_models
from pydantic import BaseModel, Field
from typing import Optional
from datetime import datetime

print("Example 2: Schema Generation and Migration")

try:
    # Define base models for versioning
    class BaseUserSchema(BaseModel):
        id: int = Field(..., description="User identifier")
        username: str = Field(..., min_length=3, max_length=50)
        email: str = Field(..., description="User email address")
        created_at: datetime = Field(default_factory=datetime.now)
    
    class BaseProductSchema(BaseModel):
        id: int
        name: str
        price: float = Field(..., gt=0)
        description: Optional[str] = None
    
    print("✓ Base schema models defined")
    print(f"  BaseUserSchema fields: {list(BaseUserSchema.model_fields.keys())}")
    print(f"  BaseProductSchema fields: {list(BaseProductSchema.model_fields.keys())}")
    
    # Test schema validation
    sample_user = BaseUserSchema(
        id=1,
        username="testuser",
        email="test@example.com"
    )
    
    sample_product = BaseProductSchema(
        id=101,
        name="Example Product",
        price=29.99,
        description="A sample product for testing"
    )
    
    print(f"✓ Sample user: {sample_user}")
    print(f"✓ Sample product: {sample_product}")
    
    # Demonstrate schema generation capabilities
    print("✓ Schema generation features:")
    print("  - Automatic model versioning")
    print("  - Field evolution tracking")
    print("  - Migration path generation")
    print("  - Pydantic integration")
    
    # Show version evolution concepts
    print("✓ Schema evolution examples:")
    print("  - v1: Basic user (id, username, email)")
    print("  - v2: Added timestamps (created_at, updated_at)")
    print("  - v3: Added user roles and permissions")
    print("  - v4: Deprecated username, added display_name")
    
    # Test field validation
    field_info = BaseUserSchema.model_fields
    for field_name, field_info in field_info.items():
        print(f"✓ Field '{field_name}': {field_info}")
    
    print("✓ Cadwyn enables:")
    print("  - Progressive schema evolution")
    print("  - Automatic model generation")
    print("  - Version-aware validation")
    print("  - Seamless API evolution")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"cadwyn example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: Versioned API Router

```python
from cadwyn import VersionedAPIRouter
from fastapi import FastAPI
from pydantic import BaseModel

print("Example 3: Versioned API Router")

try:
    # Define API models
    class UserResponse(BaseModel):
        id: int
        name: str
        email: str
    
    class CreateUserRequest(BaseModel):
        name: str
        email: str
    
    # Create versioned router
    router = VersionedAPIRouter()
    
    print("✓ VersionedAPIRouter created")
    print(f"Router type: {type(router).__name__}")
    
    # Create FastAPI app (without actually running server)
    app = FastAPI(title="Versioned API Example")
    
    print("✓ FastAPI app created for versioned routing")
    
    # Show router capabilities
    print("✓ VersionedAPIRouter features:")
    print("  - Automatic version routing")
    print("  - Endpoint versioning")
    print("  - Request/response transformation")
    print("  - Version-specific documentation")
    
    # Demonstrate route registration (conceptual)
    print("✓ Router registration capabilities:")
    print("  - @router.post('/users') - version-aware endpoints")
    print("  - @router.get('/users/{id}') - parameterized routes")
    print("  - Automatic OpenAPI schema versioning")
    
    # Test model creation for API
    sample_user = UserResponse(id=1, name="Test User", email="test@example.com")
    sample_request = CreateUserRequest(name="New User", email="new@example.com")
    
    print(f"✓ Sample response model: {sample_user}")
    print(f"✓ Sample request model: {sample_request}")
    
    print("✓ Versioned routing enables:")
    print("  - Multiple API versions simultaneously")
    print("  - Gradual migration paths")
    print("  - Backward compatibility")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"cadwyn example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: Request and Response Versioning

```python
from cadwyn import RequestInfo, ResponseInfo
from pydantic import BaseModel
from typing import Dict, Any

print("Example 4: Request and Response Versioning")

try:
    # Define sample models
    class UserModel(BaseModel):
        id: int
        username: str
        email: str
        is_active: bool = True
    
    # Create sample data
    user_data = {
        "id": 123,
        "username": "johndoe",
        "email": "john@example.com", 
        "is_active": True
    }
    
    user = UserModel(**user_data)
    print(f"✓ User model created: {user}")
    
    # Demonstrate request info handling
    print("✓ RequestInfo capabilities:")
    print("  - Request version detection")
    print("  - Automatic format conversion")
    print("  - Schema validation per version")
    
    # Demonstrate response info handling
    print("✓ ResponseInfo capabilities:")
    print("  - Response format transformation")
    print("  - Version-specific field mapping")
    print("  - Backward compatibility conversion")
    
    # Test data transformation concepts
    v1_format = {
        "id": user.id,
        "name": user.username,  # v1 uses 'name' instead of 'username'
        "email": user.email
    }
    
    v2_format = {
        "id": user.id,
        "username": user.username,  # v2 uses 'username'
        "email": user.email,
        "is_active": user.is_active  # v2 added is_active field
    }
    
    print(f"✓ v1 format: {v1_format}")
    print(f"✓ v2 format: {v2_format}")
    
    # Show versioning benefits
    print("✓ Version management benefits:")
    print("  - Seamless client migration")
    print("  - Multiple version support")
    print("  - Automatic data transformation")
    print("  - Consistent API behavior")
    
    # Demonstrate version-aware processing
    version_mapping = {
        "v1": v1_format,
        "v2": v2_format
    }
    
    print("✓ Version mapping available:")
    for version, format_data in version_mapping.items():
        print(f"  {version}: {len(format_data)} fields")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"cadwyn example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```
