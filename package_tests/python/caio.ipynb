## Example 1: Basic Context Creation and Information

```python
import caio
import os

print("Example 1: Basic Context Creation and Information")

try:
    print(f"✓ CAIO module info: {caio}")
    
    # Show available variants
    print("✓ Available AIO variants:")
    for i, variant in enumerate(caio.variants, 1):
        print(f"  {i}. {variant.__name__}")
    
    print("✓ Available AsyncIO variants:")
    for i, variant in enumerate(caio.variants_asyncio, 1):
        print(f"  {i}. {variant.__name__}")
    
    # Show preferred implementations
    print(f"✓ Preferred AIO backend: {caio.preferred.__name__}")
    print(f"✓ Preferred AsyncIO backend: {caio.preferred_asyncio.__name__}")
    
    # Create a basic context
    ctx = caio.Context()
    print(f"✓ Context created: {ctx}")
    print(f"  Max requests: {ctx.max_requests}")
    print(f"  Context type: {type(ctx).__name__}")
    
    # Test context properties
    print("✓ Context capabilities:")
    print("  - Asynchronous file operations")
    print("  - Configurable request pool")
    print("  - Multiple backend support")
    print("  - Thread-safe operations")
    
    # Show context methods
    ctx_methods = [method for method in dir(ctx) 
                  if not method.startswith('_') and callable(getattr(ctx, method))]
    print(f"✓ Context methods available: {len(ctx_methods)}")
    print(f"  Methods: {ctx_methods}")
    
    # Context is automatically cleaned up
    print("✓ Context will be automatically cleaned up")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"caio example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 2: Synchronous File Operations

```python
import caio
import tempfile
import os

print("Example 2: File Operations and Context Management")

try:
    # Create a temporary file for testing
    with tempfile.NamedTemporaryFile(mode='w+', delete=False) as tf:
        test_file = tf.name
        test_data = b"Hello CAIO! This is test data for asynchronous I/O operations."
        
    # Write initial data
    with open(test_file, 'wb') as f:
        f.write(test_data)
    
    print(f"✓ Test file created: {os.path.basename(test_file)}")
    print(f"Test data length: {len(test_data)} bytes")
    
    # Create CAIO context
    ctx = caio.Context()
    print(f"✓ CAIO context created")
    print(f"  Max requests: {ctx.max_requests}")
    
    # Demonstrate context capabilities
    print("✓ Context features:")
    print("  - High-performance async I/O operations")
    print("  - File descriptor based operations")
    print("  - Multiple concurrent requests support")
    print("  - Thread-safe operation submission")
    
    # Show operation types available
    print("✓ Available operation types:")
    print("  - Read operations: caio.Operation.read()")
    print("  - Write operations: caio.Operation.write()")
    print("  - Sync operations: caio.Operation.fsync()")
    print("  - Data sync operations: caio.Operation.fdsync()")
    
    # File operations demonstration (conceptual)
    with open(test_file, 'rb') as f:
        fd = f.fileno()
        print(f"✓ File descriptor: {fd}")
        print(f"File size: {os.path.getsize(test_file)} bytes")
        
        # Read file content normally for verification
        content = f.read()
        print(f"✓ File content verified: {len(content)} bytes")
        print(f"Content preview: {content.decode()[:50]}...")
    
    # Context management
    print("✓ Context management:")
    print(f"  Context active: {ctx is not None}")
    print("  - Automatic resource management")
    print("  - Efficient memory usage")
    print("  - Platform-optimized backends")
    
    # Clean up
    os.unlink(test_file)
    print("✓ Test file cleaned up")
    
except Exception as e:
    print(f"❌ Error: {e}")
    # Clean up on error
    try:
        if 'test_file' in locals() and os.path.exists(test_file):
            os.unlink(test_file)
    except:
        pass
        raise
    raise
```

## Example 3: AsyncIO Integration

```python
import caio
import asyncio

print("Example 3: AsyncIO Integration")

try:
    print("✓ Testing AsyncIO integration:")
    
    # Show AsyncIO backend info
    asyncio_backend = caio.preferred_asyncio
    print(f"✓ Preferred AsyncIO backend: {asyncio_backend.__name__}")
    
    # Python 3.12 deprecated, and 3.14 removed, implicit loop creation in asyncio.get_event_loop(); pass a loop explicitly.
    _asyncio_loop = asyncio.new_event_loop()
    async_ctx = caio.AsyncioContext(loop=_asyncio_loop)
    print(f"✓ AsyncIO context created: {type(async_ctx).__name__}")
    
    # Show AsyncIO context capabilities
    print("✓ AsyncIO context features:")
    print("  - Coroutine-based I/O operations")
    print("  - Integration with asyncio event loop")
    print("  - Non-blocking file operations")
    print("  - Concurrent operation support")
    
    # Show available AsyncIO variants
    print("✓ Available AsyncIO backends:")
    for i, variant in enumerate(caio.variants_asyncio, 1):
        print(f"  {i}. {variant.__name__}")
    
    # Demonstrate async operation concepts
    print("✓ Async operation patterns:")
    print("  - await ctx.read(fd, size, offset)")
    print("  - await ctx.write(fd, data, offset)")
    print("  - Seamless asyncio integration")
    print("  - Event loop compatibility")
    
    # Show context comparison
    regular_ctx = caio.Context()
    print(f"✓ Context comparison:")
    print(f"  Regular context: {type(regular_ctx).__name__}")
    print(f"  AsyncIO context: {type(async_ctx).__name__}")
    
    # Show async benefits
    print("✓ AsyncIO integration benefits:")
    print("  - Non-blocking I/O in async applications")
    print("  - Better resource utilization")
    print("  - Scalable concurrent operations")
    print("  - Python asyncio ecosystem compatibility")
    
    print("✓ AsyncIO integration demonstrated successfully")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"caio example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
finally:
    if '_asyncio_loop' in dir():
        _asyncio_loop.close()
```

## Example 4: Context Configuration and Performance

```python
import caio
import time

print("Example 4: Context Configuration and Performance")

try:
    # Test different context configurations
    configs = [
        {"max_requests": 64, "name": "Small"},
        {"max_requests": 256, "name": "Medium"},
        {"max_requests": 512, "name": "Large"}
    ]
    
    print("✓ Testing different context configurations:")
    
    contexts = []
    for config in configs:
        # max_requests is the portable capacity setting across caio backends.
        ctx = caio.Context(
            max_requests=config["max_requests"]
        )
        contexts.append((ctx, config["name"]))
        
        print(f"  {config['name']}: {ctx.max_requests} max_requests")
    
    # Show configuration benefits
    print("✓ Context configuration benefits:")
    print("  - Tunable request pool size")
    print("  - Adjustable max concurrent requests")
    print("  - Performance optimization")
    print("  - Resource management")
    print("  - Memory usage control")
    
    # Performance characteristics
    print("✓ Performance characteristics:")
    for ctx, name in contexts:
        print(f"  {name} context:")
        print(f"    Max requests: {ctx.max_requests}")
        print(f"    Suitable for: {'High throughput' if ctx.max_requests > 256 else 'Light usage'}")
    
    # Show context methods
    sample_ctx = contexts[0][0]
    ctx_methods = [method for method in dir(sample_ctx) 
                  if not method.startswith('_') and callable(getattr(sample_ctx, method))]
    print(f"✓ Context methods available: {ctx_methods}")
    
    # Configuration recommendations
    print("✓ Configuration recommendations:")
    print("  - Small: Light applications with few concurrent operations")
    print("  - Medium: Moderate load with balanced resource usage")
    print("  - Large: High-throughput applications with many operations")
    
    print("✓ Context configuration testing completed")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"caio example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```
