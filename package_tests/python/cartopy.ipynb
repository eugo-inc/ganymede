## Example 1: Basic Coordinate Reference Systems (CRS)

```python
import cartopy.crs as ccrs
import matplotlib.pyplot as plt

print("Example 1: Basic Coordinate Reference Systems (CRS)")

try:
    # Create different coordinate reference systems
    plate_carree = ccrs.PlateCarree()
    mercator = ccrs.Mercator()
    robinson = ccrs.Robinson()
    mollweide = ccrs.Mollweide()
    
    print(f"✓ PlateCarree projection: {plate_carree}")
    print(f"✓ Mercator projection: {mercator}")
    print(f"✓ Robinson projection: {robinson}")
    print(f"✓ Mollweide projection: {mollweide}")
    
    # Show CRS properties
    print(f"✓ PlateCarree globe: {plate_carree.globe}")
    print(f"✓ Available CRS classes: {len([attr for attr in dir(ccrs) if not attr.startswith('_')])}")
    
    # Test CRS transformation capabilities
    print("✓ CRS transformation features:")
    print("  - Geographic coordinate transformations")
    print("  - Map projection conversions")
    print("  - Globe and datum specifications")
    
    print("✓ Basic CRS operations completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cartopy example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Natural Earth Features

```python
import cartopy.feature as cfeature
import cartopy.crs as ccrs

print("Example 2: Natural Earth Features")

try:
    # Access Natural Earth features
    coastlines = cfeature.COASTLINE
    land = cfeature.LAND
    ocean = cfeature.OCEAN
    lakes = cfeature.LAKES
    borders = cfeature.BORDERS
    
    print(f"✓ Coastlines feature: {coastlines}")
    print(f"✓ Land feature: {land}")
    print(f"✓ Ocean feature: {ocean}")
    print(f"✓ Lakes feature: {lakes}")
    print(f"✓ Borders feature: {borders}")
    
    # Show feature properties
    print(f"✓ Coastlines category: {coastlines.category}")
    print(f"✓ Coastlines name: {coastlines.name}")
    print(f"✓ Coastlines scale: {coastlines.scale}")
    
    # Test additional features if available
    try:
        rivers = cfeature.RIVERS_LAKE_CENTERLINES
        print(f"✓ Rivers feature: {rivers}")
    except AttributeError:
        print("✓ Rivers feature: Not available in this cartopy version")
    
    # Create custom features
    states_provinces = cfeature.NaturalEarthFeature(
        category='cultural',
        name='admin_1_states_provinces_lines',
        scale='50m',
        facecolor='none'
    )
    print(f"✓ Custom states feature: {states_provinces}")
    
    # Create custom physical features
    custom_lakes = cfeature.NaturalEarthFeature(
        category='physical',
        name='lakes',
        scale='50m',
        facecolor='lightblue',
        edgecolor='blue'
    )
    print(f"✓ Custom lakes feature: {custom_lakes}")
    
    # Show available feature types
    print("✓ Available Natural Earth features:")
    print("  - Physical: coastlines, land, ocean, lakes")
    print("  - Cultural: borders, populated places, roads")
    print("  - Multiple scales: 10m, 50m, 110m resolution")
    print("  - Custom styling: colors, line widths, transparency")
    
    print("✓ Natural Earth features demonstration completed")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cartopy example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: Creating Basic Maps

```python
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import matplotlib.pyplot as plt

print("Example 3: Creating Basic Maps")

try:
    # Test map creation components
    fig = plt.figure(figsize=(10, 6))
    
    # Create map axis with PlateCarree projection
    projection = ccrs.PlateCarree()
    ax = plt.axes(projection=projection)
    
    print(f"✓ Map figure created: {fig}")
    print(f"✓ Map axes created with projection: {projection}")
    
    # Test adding features (without actually rendering)
    print("✓ Map features available for adding:")
    print(f"  - Coastlines: {cfeature.COASTLINE}")
    print(f"  - Land masses: {cfeature.LAND}")
    print(f"  - Ocean areas: {cfeature.OCEAN}")
    print(f"  - Political boundaries: {cfeature.BORDERS}")
    
    # Test extent setting capabilities
    print("✓ Map extent options:")
    print("  - set_global(): Global coverage")
    print("  - set_extent([lon_min, lon_max, lat_min, lat_max]): Custom bounds")
    
    # Test gridlines and labels
    print("✓ Map annotation options:")
    print("  - Gridlines with coordinate labels")
    print("  - Custom tick marks and spacing")
    print("  - Coordinate system transformations")
    
    # Close figure to free memory
    plt.close(fig)
    print("✓ Map components tested successfully")
    
    # Test different map projections
    projections = {
        'PlateCarree': ccrs.PlateCarree(),
        'Mercator': ccrs.Mercator(),
        'Robinson': ccrs.Robinson()
    }
    
    print("✓ Available map projections:")
    for name, proj in projections.items():
        print(f"  - {name}: {proj}")
    
    print("✓ Basic map creation demonstrated successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cartopy example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: Coordinate Transformations

```python
import cartopy.crs as ccrs
import numpy as np

print("Example 4: Coordinate Transformations")

try:
    # Create different coordinate reference systems
    plate_carree = ccrs.PlateCarree()
    mercator = ccrs.Mercator()
    lambert = ccrs.LambertConformal()
    
    print(f"✓ PlateCarree CRS: {plate_carree}")
    print(f"✓ Mercator CRS: {mercator}")
    print(f"✓ Lambert Conformal CRS: {lambert}")
    
    # Test coordinate transformation concepts
    print("✓ Coordinate transformation capabilities:")
    print("  - Geographic to projected coordinate conversion")
    print("  - Datum transformations between reference systems")
    print("  - Map projection parameter handling")
    
    # Sample coordinates for demonstration
    sample_points = [
        (-74.0, 40.7),  # New York
        (-0.1, 51.5),   # London
        (2.3, 48.9),    # Paris
        (139.7, 35.7)   # Tokyo
    ]
    
    print("✓ Sample geographic coordinates:")
    for i, (lon, lat) in enumerate(sample_points):
        print(f"  Point {i+1}: Longitude {lon}°, Latitude {lat}°")
    
    # Test CRS properties and methods
    print("✓ CRS properties:")
    print(f"  PlateCarree globe: {plate_carree.globe}")
    print(f"  Mercator boundaries: {hasattr(mercator, 'x_limits')}")
    print(f"  Lambert parameters: {hasattr(lambert, 'proj4_params')}")
    
    # Demonstrate transformation concepts
    print("✓ Transformation types available:")
    print("  - transform_point(): Single coordinate transformation")
    print("  - transform_points(): Bulk coordinate transformation") 
    print("  - as_geodetic(): Convert to geodetic coordinates")
    
    print("✓ Map projection features:")
    print("  - Conformal projections (preserving angles)")
    print("  - Equal-area projections (preserving areas)")
    print("  - Equidistant projections (preserving distances)")
    
    print("✓ Coordinate transformation concepts demonstrated")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cartopy example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: Advanced Cartopy Features and Integration

```python
import cartopy
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import cartopy.io.img_tiles as cimgt

print("Example 5: Advanced Cartopy Features and Integration")

try:
    # Show cartopy version and modules
    print(f"✓ Cartopy package: {cartopy}")
    print(f"✓ Cartopy version available: {hasattr(cartopy, '__version__')}")
    
    # Advanced coordinate reference systems
    polar_stereo = ccrs.NorthPolarStereo()
    ortho = ccrs.Orthographic(central_longitude=0, central_latitude=0)
    
    print(f"✓ Polar Stereographic: {polar_stereo}")
    print(f"✓ Orthographic: {ortho}")
    
    # Test specialized map projections (check availability first)
    advanced_projections = {}
    
    # Add projections that are available
    if hasattr(ccrs, 'Geostationary'):
        advanced_projections['Geostationary'] = ccrs.Geostationary
    if hasattr(ccrs, 'InterruptedGoodeHomolosine'):
        advanced_projections['InterruptedGoodeHomolosine'] = ccrs.InterruptedGoodeHomolosine
    if hasattr(ccrs, 'RotatedPole'):
        advanced_projections['RotatedPole'] = ccrs.RotatedPole
    if hasattr(ccrs, 'LambertAzimuthalEqualArea'):
        advanced_projections['LambertAzimuthalEqualArea'] = ccrs.LambertAzimuthalEqualArea
    if hasattr(ccrs, 'Stereographic'):
        advanced_projections['Stereographic'] = ccrs.Stereographic
    
    print("✓ Advanced projection classes:")
    for name, proj_class in advanced_projections.items():
        print(f"  - {name}: {proj_class}")
    
    # Test some commonly available projections
    try:
        sinusoidal = ccrs.Sinusoidal()
        print(f"✓ Sinusoidal projection: {sinusoidal}")
    except AttributeError:
        print("✓ Sinusoidal projection: Not available")
    
    try:
        miller = ccrs.Miller()
        print(f"✓ Miller projection: {miller}")
    except AttributeError:
        print("✓ Miller projection: Not available")
    
    # Test image tiles integration
    try:
        stamen_terrain = cimgt.Stamen('terrain-background')
        print(f"✓ Stamen terrain tiles: {stamen_terrain}")
        print("✓ Web tile integration available")
    except Exception:
        print("✓ Tile services require network access or updated API")
        raise
    
    # Show cartopy io modules
    print("✓ Cartopy I/O capabilities:")
    print("  - cartopy.io.img_tiles: Web map tile services")
    print("  - cartopy.io.shapereader: Shapefile reading")
    import cartopy.io.srtm
    print("  - cartopy.io.srtm: SRTM elevation data")
    
    # Feature customization
    custom_feature = cfeature.NaturalEarthFeature(
        'physical', 'coastline', '110m',
        edgecolor='blue', facecolor='none', linewidth=2
    )
    print(f"✓ Custom feature styling: {custom_feature}")
    
    # Show cartopy utilities
    print("✓ Additional cartopy modules:")
    print("  - cartopy.geodesic: Geodesic calculations")
    print("  - cartopy.mpl: Matplotlib integration")
    print("  - cartopy.util: Utility functions")
    
    print("✓ Cartopy integration capabilities:")
    print("  - Matplotlib plotting backend")
    print("  - NumPy array handling")
    print("  - Shapely geometry support")
    print("  - PyProj coordinate transformations")
    
    # Test core projections that should be available
    core_projections = [
        ('PlateCarree', ccrs.PlateCarree()),
        ('Mercator', ccrs.Mercator()),
        ('Robinson', ccrs.Robinson()),
        ('Mollweide', ccrs.Mollweide())
    ]
    
    print("✓ Core projections verified:")
    for name, proj in core_projections:
        print(f"  - {name}: Available")
    
    print("✓ Advanced cartopy features demonstrated successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cartopy example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
