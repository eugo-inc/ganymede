## Example 1: Basic Exchange Information

```python
import ccxt

print("Example 1: Basic Exchange Information")

try:
    # Show CCXT version and basic info
    print(f"✓ CCXT version: {ccxt.__version__}")
    
    # List all supported exchanges
    exchanges = ccxt.exchanges
    print(f"✓ Total supported exchanges: {len(exchanges)}")
    print(f"  Sample exchanges: {exchanges[:10]}")
    
    # Create exchange instances (without API keys for testing)
    binance = ccxt.binance()
    coinbase = ccxt.coinbase()
    kraken = ccxt.kraken()
    
    print(f"✓ Binance exchange: {binance.id}")
    print(f"  Name: {binance.name}")
    print(f"  Countries: {binance.countries}")
    
    print(f"✓ Coinbase exchange: {coinbase.id}")
    print(f"  Name: {coinbase.name}")
    print(f"  Rate limit: {coinbase.rateLimit}ms")
    
    print(f"✓ Kraken exchange: {kraken.id}")
    print(f"  Name: {kraken.name}")
    if hasattr(kraken, 'cors'):
        print(f"  Has CORS: {kraken.cors}")
    else:
        print("  CORS: Not available in this version")
    
    # Show exchange capabilities
    print("✓ Exchange capabilities example (Binance):")
    capabilities = binance.has
    key_capabilities = ['fetchTickers', 'fetchOrderBook', 'fetchTrades', 'fetchBalance']
    for cap in key_capabilities:
        status = capabilities.get(cap, False)
        print(f"  {cap}: {status}")
    
    assert "binance" in exchanges and binance.id == "binance", (binance.id, len(exchanges))
    print("✓ Basic exchange information completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    raise AssertionError(f"ccxt example (cell 1) ImportError: {e}") from e
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"ccxt example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Market Data and Symbols

```python
import ccxt

print("Example 2: Market Data and Symbols")

try:
    # Create exchange instance
    exchange = ccxt.binance()
    print(f"✓ Created {exchange.name} exchange instance")
    
    # Test market structure without API calls
    print("✓ Market data structure:")
    print(f"  Exchange ID: {exchange.id}")
    print(f"  Exchange name: {exchange.name}")
    print(f"  User agent: {exchange.userAgent}")
    
    # Show expected market data types
    print("✓ Expected market data types:")
    print("  - Ticker: Real-time price information")
    print("  - Order book: Bid/ask prices and volumes")  
    print("  - Trades: Recent trade history")
    print("  - OHLCV: Candlestick/bar data")
    
    # Test symbol formatting
    print("✓ Symbol format examples:")
    common_symbols = ['BTC/USDT', 'ETH/USDT', 'BNB/USDT', 'ADA/USDT', 'DOT/USDT']
    for symbol in common_symbols:
        print(f"  {symbol}")
    
    # Show exchange properties
    print("✓ Exchange configuration:")
    print(f"  Timeout: {exchange.timeout}ms")
    print(f"  Rate limit: {exchange.rateLimit}ms")
    print(f"  Enable rate limit: {exchange.enableRateLimit}")
    
    # Test market capabilities
    market_functions = [
        'fetchTicker', 'fetchTickers', 'fetchOrderBook', 
        'fetchTrades', 'fetchOHLCV', 'fetchBalance'
    ]
    
    print("✓ Market data capabilities:")
    for func in market_functions:
        has_capability = exchange.has.get(func, False)
        print(f"  {func}: {'✓' if has_capability else '✗'}")
    
    # Test timeframe support
    timeframes = exchange.timeframes if hasattr(exchange, 'timeframes') else {}
    if timeframes:
        print(f"✓ Supported timeframes: {list(timeframes.keys())[:10]}")
    else:
        print("✓ Timeframes: Available when loaded")
    
    assert exchange.has.get("fetchOrderBook") is True and "1h" in exchange.timeframes, (exchange.has.get("fetchOrderBook"), list(exchange.timeframes))
    print("✓ Market data structure demonstration completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    raise AssertionError(f"ccxt example (cell 3) ImportError: {e}") from e
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"ccxt example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: Exchange Configuration and Methods

```python
import ccxt

print("Example 3: Exchange Configuration and Methods")

try:
    # Test multiple exchange configurations
    exchanges_config = {
        'binance': {'sandbox': False, 'timeout': 10000},
        'coinbase': {'timeout': 15000, 'verbose': False},
        'kraken': {'timeout': 20000, 'enableRateLimit': True}
    }
    
    exchanges = {}
    for name, config in exchanges_config.items():
        exchange_class = getattr(ccxt, name)
        exchange = exchange_class(config)
        exchanges[name] = exchange
        
        print(f"✓ {name.title()} configured:")
        print(f"  Timeout: {exchange.timeout}ms")
        print(f"  Rate limit: {exchange.rateLimit}ms")
        print(f"  Enable rate limit: {exchange.enableRateLimit}")
    
    # Test API method categories
    binance = exchanges['binance']
    print("✓ Exchange API method categories:")
    
    # Public methods (no authentication required)
    public_methods = [
        'fetchTicker', 'fetchTickers', 'fetchOrderBook', 'fetchTrades', 
        'fetchOHLCV', 'fetchMarketsAndCurrencies', 'fetchStatus'
    ]
    
    print("  Public methods (no auth required):")
    for method in public_methods:
        has_method = hasattr(binance, method)
        print(f"    {method}: {'✓' if has_method else '✗'}")
    
    # Private methods (authentication required)
    private_methods = [
        'fetchBalance', 'fetchOrders', 'fetchOpenOrders', 'createOrder',
        'cancelOrder', 'fetchMyTrades', 'fetchDeposits', 'fetchWithdrawals'
    ]
    
    print("  Private methods (auth required):")
    for method in private_methods:
        has_method = hasattr(binance, method)
        print(f"    {method}: {'✓' if has_method else '✗'}")
    
    # Test error handling structure
    print("✓ CCXT error types:")
    error_types = [
        'BaseError', 'ExchangeError', 'NetworkError', 'RequestTimeout',
        'AuthenticationError', 'PermissionDenied', 'InsufficientFunds'
    ]
    
    for error_type in error_types:
        error_class = getattr(ccxt, error_type, None)
        if error_class:
            print(f"  {error_type}: {error_class}")
    
    # Test exchange sandbox mode
    print("✓ Sandbox mode configuration:")
    for name, exchange in exchanges.items():
        sandbox_available = hasattr(exchange, 'sandbox')
        print(f"  {name}: Sandbox {'available' if sandbox_available else 'not available'}")
    
    assert (exchanges["binance"].timeout, exchanges["kraken"].timeout) == (10000, 20000), (exchanges["binance"].timeout, exchanges["kraken"].timeout)
    assert issubclass(ccxt.RequestTimeout, ccxt.NetworkError) and issubclass(ccxt.NetworkError, ccxt.BaseError)
    print("✓ Exchange configuration and methods completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    raise AssertionError(f"ccxt example (cell 5) ImportError: {e}") from e
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"ccxt example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: Trading Pairs and Currency Information

```python
import ccxt

print("Example 4: Trading Pairs and Currency Information")

try:
    # Create exchange instance
    exchange = ccxt.binance()
    print(f"✓ Created {exchange.name} exchange instance")
    
    # Test symbol parsing and formatting
    print("✓ Symbol parsing and formatting:")
    
    # Test common trading pairs
    symbols_to_test = [
        'BTC/USDT', 'ETH/USDT', 'BNB/USDT', 'ADA/USDT', 
        'DOT/USDT', 'LINK/USDT', 'XRP/USDT', 'SOL/USDT'
    ]
    
    for symbol in symbols_to_test:
        # Parse symbol components
        parts = symbol.split('/')
        base = parts[0]
        quote = parts[1]
        print(f"  {symbol}: Base={base}, Quote={quote}")
    
    # Test market structure
    print("✓ Market data structure:")
    print("  Market object contains:")
    print("    - symbol: Trading pair (e.g., 'BTC/USDT')")
    print("    - base: Base currency (e.g., 'BTC')")
    print("    - quote: Quote currency (e.g., 'USDT')")
    print("    - active: Whether market is active")
    print("    - precision: Decimal precision info")
    print("    - limits: Min/max amounts")
    
    # Test precision and limits structure
    print("✓ Precision structure:")
    print("  - amount: Decimal places for order amounts")
    print("  - price: Decimal places for prices") 
    print("  - cost: Decimal places for total cost")
    
    print("✓ Limits structure:")
    print("  - amount: {min: number, max: number}")
    print("  - price: {min: number, max: number}")
    print("  - cost: {min: number, max: number}")
    
    # Test fee structure
    print("✓ Fee information structure:")
    print("  - maker: Maker fee percentage")
    print("  - taker: Taker fee percentage")
    print("  - currency: Fee currency")
    
    # Show common currency codes
    major_currencies = [
        'BTC', 'ETH', 'USDT', 'USDC', 'BNB', 'ADA', 'XRP', 
        'DOT', 'LINK', 'LTC', 'BCH', 'SOL', 'MATIC'
    ]
    
    print(f"✓ Major cryptocurrencies ({len(major_currencies)}):")
    for i in range(0, len(major_currencies), 4):
        group = major_currencies[i:i+4]
        print(f"  {', '.join(group)}")
    
    # Test order types
    order_types = ['market', 'limit', 'stop', 'stop_limit']
    print(f"✓ Common order types:")
    for order_type in order_types:
        print(f"  {order_type}")
    
    assert exchange.safe_currency_code("XBT") == "BTC", exchange.safe_currency_code("XBT")
    print("✓ Trading pairs and currency information completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    raise AssertionError(f"ccxt example (cell 7) ImportError: {e}") from e
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"ccxt example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: Advanced CCXT Features and Utilities

```python
import ccxt
from datetime import datetime

print("Example 5: Advanced CCXT Features and Utilities")

try:
    # Test multiple exchange comparisons
    exchanges_to_compare = ['binance', 'coinbase', 'kraken', 'okx', 'bybit']
    
    print("✓ Exchange comparison:")
    print(f"  Comparing {len(exchanges_to_compare)} major exchanges")
    
    exchange_info = {}
    for exchange_name in exchanges_to_compare:
        try:
            exchange_class = getattr(ccxt, exchange_name)
            exchange = exchange_class()
            
            exchange_info[exchange_name] = {
                'name': exchange.name,
                'countries': exchange.countries,
                'rateLimit': exchange.rateLimit,
                'has_fetchTicker': exchange.has.get('fetchTicker', False),
                'has_fetchOHLCV': exchange.has.get('fetchOHLCV', False),
                'has_fetchBalance': exchange.has.get('fetchBalance', False)
            }
            
            print(f"  ✓ {exchange_name}: {exchange.name}")
            
        except Exception as e:
            print(f"  ❌ {exchange_name}: Error loading - {e}")
            raise
    
    # Show feature comparison
    print("✓ Feature comparison:")
    features = ['has_fetchTicker', 'has_fetchOHLCV', 'has_fetchBalance']
    for feature in features:
        print(f"  {feature}:")
        for name, info in exchange_info.items():
            status = '✓' if info[feature] else '✗'
            print(f"    {name}: {status}")
    
    # Test timestamp utilities
    print("✓ Timestamp utilities:")
    current_time = datetime.now()
    print(f"  Current time: {current_time}")
    
    # Test common timeframe values
    timeframes = ['1m', '5m', '15m', '1h', '4h', '1d', '1w', '1M']
    print(f"  Common timeframes: {timeframes}")
    
    # Show CCXT constants and utilities
    print("✓ CCXT utilities and constants:")
    
    # Test decimal precision helpers
    print("  Precision handling:")
    print("    - Decimal precision for prices and amounts")
    print("    - Rounding methods for order placement")
    print("    - Fee calculation utilities")
    
    # Show error handling
    print("✓ Error handling categories:")
    error_categories = {
        'Network errors': ['NetworkError', 'RequestTimeout', 'ExchangeNotAvailable'],
        'Authentication': ['AuthenticationError', 'PermissionDenied', 'AccountSuspended'],
        'Trading errors': ['InsufficientFunds', 'InvalidOrder', 'OrderNotFound'],
        'Rate limiting': ['RateLimitExceeded', 'DDoSProtection']
    }
    
    for category, errors in error_categories.items():
        print(f"  {category}:")
        for error in errors:
            print(f"    - {error}")
    
    # Show async support
    print("✓ Async support:")
    print("  - ccxt.async_support for asynchronous operations")
    print("  - Concurrent API calls for multiple exchanges")
    print("  - Non-blocking market data fetching")
    
    # Show pro version features
    print("✓ CCXT Pro features (separate package):")
    print("  - WebSocket connections for real-time data")
    print("  - Streaming tickers and order books")
    print("  - Lower latency trading")
    
    assert set(exchange_info) == set(exchanges_to_compare), sorted(exchange_info)
    assert exchange.parse8601(exchange.iso8601(1704067200000)) == 1704067200000, exchange.iso8601(1704067200000)
    print("✓ Advanced CCXT features and utilities completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    raise AssertionError(f"ccxt example (cell 9) ImportError: {e}") from e
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"ccxt example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
