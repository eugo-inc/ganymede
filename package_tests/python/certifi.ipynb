# certifi: 5 Basic Usage Examples

```python
# Example 1: Get the path to the certifi CA bundle
import certifi
print(certifi.where())
```

```python
# Example 2: Use certifi with requests
import requests
import certifi
response = requests.get('https://www.google.com', verify=certifi.where())
print(response.status_code)
```

```python
# Example 3: Use certifi with urllib3
import urllib3
import certifi
http = urllib3.PoolManager(cert_reqs='CERT_REQUIRED', ca_certs=certifi.where())
r = http.request('GET', 'https://www.google.com')
print(r.status)
```

```python
# Example 4: Check certifi version
import certifi
print(certifi.__version__)
```

```python
# Example 5: Read the CA bundle file
import certifi
with open(certifi.where(), 'r') as f:
    print(f.readline())
```
