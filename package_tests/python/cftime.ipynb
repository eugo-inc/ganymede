## Example 1: Basic DateTime Objects and Calendar Systems

```python
import cftime

print("Example 1: Basic DateTime Objects and Calendar Systems")

try:
    # Show cftime version and basic info
    print(f"✓ cftime version: {cftime.__version__}")
    
    # Create basic datetime objects
    dt_standard = cftime.datetime(2024, 1, 15, 12, 30, 45)
    print(f"✓ Standard datetime created: {dt_standard}")
    
    # Show datetime components
    print("✓ DateTime components:")
    print(f"  Year: {dt_standard.year}")
    print(f"  Month: {dt_standard.month}")
    print(f"  Day: {dt_standard.day}")
    print(f"  Hour: {dt_standard.hour}")
    print(f"  Minute: {dt_standard.minute}")
    print(f"  Second: {dt_standard.second}")
    
    # Show supported calendar systems
    calendar_systems = [
        'standard',     # Proleptic Gregorian calendar
        'gregorian',    # Mixed Julian/Gregorian calendar  
        'proleptic_gregorian',  # Proleptic Gregorian
        'julian',       # Julian calendar
        '365_day',      # No leap year calendar (365 days/year)
        '360_day',      # All months 30 days (360 days/year)
        'noleap',       # Alias for 365_day
        'all_leap'      # All years are leap years (366 days/year)
    ]
    
    print("✓ Supported calendar systems:")
    for calendar in calendar_systems:
        print(f"  - {calendar}")
    
    # Create datetime with specific calendar
    dt_365day = cftime.datetime(2024, 1, 15, calendar='365_day')
    dt_360day = cftime.datetime(2024, 1, 15, calendar='360_day')
    dt_julian = cftime.datetime(2024, 1, 15, calendar='julian')
    
    print("✓ DateTime objects with different calendars:")
    print(f"  365-day calendar: {dt_365day}")
    print(f"  360-day calendar: {dt_360day}")
    print(f"  Julian calendar: {dt_julian}")
    
    # Show calendar property
    print("✓ Calendar properties:")
    print(f"  Standard calendar: {dt_standard.calendar}")
    print(f"  365-day calendar: {dt_365day.calendar}")
    print(f"  360-day calendar: {dt_360day.calendar}")
    
    # Test datetime arithmetic
    print("✓ DateTime arithmetic capabilities:")
    print("  - Addition and subtraction of time deltas")
    print("  - Calendar-aware date calculations")
    print("  - Leap year handling per calendar system")
    
    print("✓ Basic datetime objects and calendar systems completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("Note: Install with 'pip install cftime'")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cftime example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Time Delta Operations and Calendar Arithmetic

```python
import cftime
from datetime import timedelta

print("Example 2: Time Delta Operations and Calendar Arithmetic")

try:
    # Create base datetime objects with different calendars
    base_standard = cftime.datetime(2024, 1, 1, 0, 0, 0, calendar='standard')
    base_365day = cftime.datetime(2024, 1, 1, 0, 0, 0, calendar='365_day') 
    base_360day = cftime.datetime(2024, 1, 1, 0, 0, 0, calendar='360_day')
    
    print("✓ Base datetime objects created:")
    print(f"  Standard calendar: {base_standard}")
    print(f"  365-day calendar: {base_365day}")
    print(f"  360-day calendar: {base_360day}")
    
    # Test timedelta arithmetic
    print("✓ Timedelta arithmetic:")
    
    # Add days
    delta_30_days = timedelta(days=30)
    
    new_standard = base_standard + delta_30_days
    new_365day = base_365day + delta_30_days
    new_360day = base_360day + delta_30_days
    
    print(f"  Standard + 30 days: {new_standard}")
    print(f"  365-day + 30 days: {new_365day}")
    print(f"  360-day + 30 days: {new_360day}")
    
    # Add hours and minutes
    delta_time = timedelta(hours=12, minutes=30)
    
    new_with_time = base_standard + delta_time
    print(f"  Standard + 12.5 hours: {new_with_time}")
    
    # Test leap year handling
    print("✓ Leap year handling across calendars:")
    
    # February 28th in different calendars
    feb28_standard = cftime.datetime(2024, 2, 28, calendar='standard')
    feb28_365day = cftime.datetime(2024, 2, 28, calendar='365_day')
    
    # Add one day
    one_day = timedelta(days=1)
    
    mar1_standard = feb28_standard + one_day
    mar1_365day = feb28_365day + one_day
    
    print(f"  2024-02-28 + 1 day (standard): {mar1_standard}")
    print(f"  2024-02-28 + 1 day (365_day): {mar1_365day}")
    
    # Test subtraction
    print("✓ DateTime subtraction:")
    
    end_date = cftime.datetime(2024, 12, 31, 23, 59, 59)
    start_date = cftime.datetime(2024, 1, 1, 0, 0, 0)
    
    difference = end_date - start_date
    print(f"  2024-12-31 - 2024-01-01: {difference}")
    print(f"  Total days: {difference.days}")
    print(f"  Total seconds: {difference.total_seconds()}")
    
    # Calendar-specific arithmetic
    print("✓ Calendar-specific date arithmetic:")
    
    # Test year boundaries in 360-day calendar
    dec30_360 = cftime.datetime(2023, 12, 30, calendar='360_day')  # Last day of 360-day year
    jan1_360 = dec30_360 + one_day  # Should be Jan 1 next year
    
    print(f"  2023-12-30 + 1 day (360_day): {jan1_360}")
    
    # Test month arithmetic
    print("✓ Month and year operations:")
    
    # Different ways to add months (conceptual)
    jan15 = cftime.datetime(2024, 1, 15, calendar='standard')
    print(f"  Base date: {jan15}")
    
    # Add 31 days (approximately 1 month)
    month_later = jan15 + timedelta(days=31)
    print(f"  + 31 days: {month_later}")
    
    # Add a year (365 or 366 days depending on calendar)
    year_later_365 = cftime.datetime(2024, 1, 15, calendar='365_day') + timedelta(days=365)
    year_later_360 = cftime.datetime(2024, 1, 15, calendar='360_day') + timedelta(days=360)
    
    print(f"  + 1 year (365_day): {year_later_365}")
    print(f"  + 1 year (360_day): {year_later_360}")
    
    print("✓ Time delta operations and calendar arithmetic completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cftime example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: String Formatting and Parsing

```python
import cftime

print("Example 3: String Formatting and Parsing")

try:
    # Create datetime objects for formatting tests
    dt = cftime.datetime(2024, 7, 15, 14, 30, 45, calendar='standard')
    dt_360 = cftime.datetime(2024, 7, 15, 14, 30, 45, calendar='360_day')
    
    print("✓ DateTime objects for formatting:")
    print(f"  Standard: {dt}")
    print(f"  360-day: {dt_360}")
    
    # Test string representation
    print("✓ String representations:")
    print(f"  str(dt): {str(dt)}")
    print(f"  repr(dt): {repr(dt)}")
    
    # Test strftime formatting
    print("✓ strftime formatting:")
    format_patterns = [
        '%Y-%m-%d',           # ISO date
        '%Y-%m-%d %H:%M:%S',  # ISO datetime
        '%A, %B %d, %Y',      # Long format
        '%Y%j',               # Year + day of year
        '%U',                 # Week number
    ]
    
    for pattern in format_patterns:
        try:
            formatted = dt.strftime(pattern)
            print(f"  {pattern}: {formatted}")
        except Exception as e:
            print(f"  {pattern}: Error - {e}")
            raise
    
    # Test calendar-specific formatting
    print("✓ Calendar-specific information:")
    print(f"  Calendar: {dt.calendar}")
    print(f"  Day of year: {dt.dayofyr}")
    print(f"  Day of week: {dt.dayofwk}")
    
    # Compare day of year across calendars
    print("✓ Day of year comparison:")
    same_date_std = cftime.datetime(2024, 7, 15, calendar='standard')
    same_date_365 = cftime.datetime(2024, 7, 15, calendar='365_day')
    same_date_360 = cftime.datetime(2024, 7, 15, calendar='360_day')
    
    print(f"  July 15, 2024 (standard): Day {same_date_std.dayofyr}")
    print(f"  July 15, 2024 (365_day): Day {same_date_365.dayofyr}")
    print(f"  July 15, 2024 (360_day): Day {same_date_360.dayofyr}")
    
    # Test date parsing (conceptual - actual parsing varies by implementation)
    print("✓ Date parsing concepts:")
    print("  cftime supports parsing from various string formats")
    print("  Common patterns: ISO 8601, CF conventions")
    print("  Calendar-aware parsing for netCDF time units")
    
    # Show CF time unit examples
    print("✓ CF time unit examples:")
    cf_units = [
        "days since 1850-01-01",
        "hours since 2000-01-01 00:00:00",
        "seconds since 1970-01-01T00:00:00Z",
        "days since 0001-01-01 00:00:00",
        "months since 2000-01-01"
    ]
    
    for unit in cf_units:
        print(f"  {unit}")
    
    # Test isoformat method
    print("✓ ISO format:")
    print(f"  isoformat(): {dt.isoformat()}")
    print(f"  isoformat() with T separator: {dt.isoformat('T')}")
    
    # Test replace method
    print("✓ Replace method:")
    new_year = dt.replace(year=2025)
    new_month = dt.replace(month=12)
    new_time = dt.replace(hour=0, minute=0, second=0)
    
    print(f"  Replace year: {new_year}")
    print(f"  Replace month: {new_month}")
    print(f"  Replace time: {new_time}")
    
    print("✓ String formatting and parsing completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cftime example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 5: Advanced Calendar Operations and Comparisons

```python
import cftime

print("Example 5: Advanced Calendar Operations and Comparisons")

try:
    # Create datetime objects for comprehensive testing
    dt_std = cftime.datetime(2024, 6, 15, 12, 0, 0, calendar='standard')
    dt_365 = cftime.datetime(2024, 6, 15, 12, 0, 0, calendar='365_day')
    dt_360 = cftime.datetime(2024, 6, 15, 12, 0, 0, calendar='360_day')
    dt_julian = cftime.datetime(2024, 6, 15, 12, 0, 0, calendar='julian')
    
    print("✓ DateTime objects for comparison:")
    print(f"  Standard: {dt_std}")
    print(f"  365-day: {dt_365}")
    print(f"  360-day: {dt_360}")
    print(f"  Julian: {dt_julian}")
    
    # Test comparison operations
    print("✓ DateTime comparison operations:")
    same_calendar_dt1 = cftime.datetime(2024, 6, 14, calendar='standard')
    same_calendar_dt2 = cftime.datetime(2024, 6, 16, calendar='standard')
    
    print(f"  2024-06-14 < 2024-06-15: {same_calendar_dt1 < dt_std}")
    print(f"  2024-06-16 > 2024-06-15: {same_calendar_dt2 > dt_std}")
    print(f"  Equal comparison: {dt_std == cftime.datetime(2024, 6, 15, 12, 0, 0, calendar='standard')}")
    
    # Calendar-specific year length
    print("✓ Calendar-specific year characteristics:")
    year_lengths = {}
    
    # Calculate days in year for each calendar
    for cal_name, dt_obj in [('standard', dt_std), ('365_day', dt_365), ('360_day', dt_360)]:
        jan1 = cftime.datetime(2024, 1, 1, calendar=cal_name)
        dec31_or_last = None
        
        if cal_name == '360_day':
            dec31_or_last = cftime.datetime(2024, 12, 30, calendar=cal_name)  # 360-day year ends on 12/30
        else:
            dec31_or_last = cftime.datetime(2024, 12, 31, calendar=cal_name)
        
        year_length = (dec31_or_last - jan1).days + 1
        year_lengths[cal_name] = year_length
        print(f"  {cal_name} calendar 2024: {year_length} days")
    
    # Test leap year detection
    print("✓ Leap year behavior:")
    leap_year = 2024  # Known leap year
    non_leap_year = 2023  # Known non-leap year
    
    # February length comparison
    feb_lengths = {}
    for cal_name in ['standard', '365_day', '360_day']:
        if cal_name == 'standard':
            # In standard calendar, 2024 is a leap year
            feb_28 = cftime.datetime(2024, 2, 28, calendar=cal_name)
            try:
                feb_29 = cftime.datetime(2024, 2, 29, calendar=cal_name)
                feb_lengths[cal_name] = 29  # Has Feb 29
            except:
                feb_lengths[cal_name] = 28  # No Feb 29
        elif cal_name == '365_day':
            feb_lengths[cal_name] = 28  # Always 28 days
        else:  # 360_day
            feb_lengths[cal_name] = 30  # Always 30 days
            
        print(f"  February 2024 in {cal_name}: {feb_lengths[cal_name]} days")
    
    # Calendar conversion concepts
    print("✓ Calendar conversion concepts:")
    print("  - Direct conversion between calendars not always possible")
    print("  - Day-of-year numbers differ between calendar systems")
    print("  - Time series alignment requires careful consideration")
    print("  - Statistical methods needed for calendar harmonization")
    
    # Show utility functions and methods
    print("✓ Available datetime methods and properties:")
    dt_methods = [
        'year', 'month', 'day', 'hour', 'minute', 'second',
        'microsecond', 'calendar', 'dayofwk', 'dayofyr',
        'strftime', 'isoformat', 'replace', 'timetuple'
    ]
    
    print("  Properties and methods:")
    for method in dt_methods:
        if hasattr(dt_std, method):
            print(f"    {method}")
    
    # Integration with scientific Python ecosystem
    print("✓ Scientific Python integration:")
    integration_info = [
        "xarray: Native support for cftime datetime objects",
        "pandas: Limited support, may require conversion",
        "matplotlib: Can plot cftime datetime axes",
        "numpy: Works with datetime64 for some operations",
        "netCDF4: Primary library for CF-compliant time handling"
    ]
    
    for info in integration_info:
        print(f"  - {info}")
    
    # Performance and best practices
    print("✓ Performance and best practices:")
    best_practices = [
        "Use consistent calendar systems within datasets",
        "Cache datetime objects for repeated operations",
        "Consider timezone handling for global data",
        "Validate calendar compatibility before operations",
        "Use vectorized operations when possible"
    ]
    
    for practice in best_practices:
        print(f"  - {practice}")
    
    print("✓ Advanced calendar operations and comparisons completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cftime example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
