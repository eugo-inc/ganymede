## Example 1: Basic Encoding Detection

```python
import chardet

print("Example 1: Basic Encoding Detection")

try:
    # Show chardet version and info
    print(f"✓ chardet version: {chardet.__version__}")
    
    # Test with various text samples
    test_samples = [
        b'Hello World!',  # ASCII
        b'Hello World! \xe4\xb8\xad\xe6\x96\x87',  # UTF-8 with Chinese
        b'Hello World! \xc4\x85\xc4\x99\xc5\x82',  # UTF-8 with Polish
        b'\xff\xfeH\x00e\x00l\x00l\x00o\x00',  # UTF-16LE
        b'Hello World! \xe7\xe4\xe9\xf2\xe8',  # Windows-1252
    ]
    
    print("✓ Testing encoding detection on various samples:")
    
    for i, sample in enumerate(test_samples, 1):
        result = chardet.detect(sample)
        
        print(f"  Sample {i}:")
        print(f"    Bytes: {sample[:20]}{'...' if len(sample) > 20 else ''}")
        print(f"    Detected encoding: {result['encoding']}")
        print(f"    Confidence: {result['confidence']:.3f}")
        print(f"    Language: {result.get('language', 'N/A')}")
        print()
    
    # Test with longer text for better accuracy
    longer_text = """
    This is a longer sample of English text that should be detected
    as ASCII or UTF-8 with high confidence. The more text we provide,
    the better chardet can determine the encoding.
    """.encode('utf-8')
    
    result_long = chardet.detect(longer_text)
    print("✓ Detection with longer text sample:")
    print(f"  Encoding: {result_long['encoding']}")
    print(f"  Confidence: {result_long['confidence']:.3f}")
    
    # Test detect function return structure
    print("✓ Detection result structure:")
    result_keys = result_long.keys()
    for key in result_keys:
        print(f"  {key}: {result_long[key]}")
    
    print("✓ Basic encoding detection completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("Note: chardet is usually pre-installed with Python")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"chardet example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Multi-Language Text Detection

```python
import chardet

print("Example 2: Multi-Language Text Detection")

try:
    # Test with different languages and scripts
    multilingual_samples = {
        'English': "The quick brown fox jumps over the lazy dog.",
        'Chinese': "你好世界，这是中文测试。",
        'Japanese': "こんにちは世界、これは日本語のテストです。",
        'Russian': "Привет мир, это русский тест.",
        'Arabic': "مرحبا بالعالم، هذا اختبار عربي.",
        'German': "Hallo Welt, das ist ein deutscher Test mit Umlauten: äöü.",
        'French': "Bonjour le monde, c'est un test français avec accents: éàç."
    }
    
    print("✓ Testing multi-language text detection:")
    
    for language, text in multilingual_samples.items():
        # Encode text in UTF-8
        byte_text = text.encode('utf-8')
        result = chardet.detect(byte_text)
        
        print(f"  {language}:")
        print(f"    Text: {text[:50]}{'...' if len(text) > 50 else ''}")
        print(f"    Detected encoding: {result['encoding']}")
        print(f"    Confidence: {result['confidence']:.3f}")
        print(f"    Detected language: {result.get('language', 'N/A')}")
        print()
    
    # Test mixed content
    mixed_content = """
    English: Hello World!
    中文: 你好世界!
    Русский: Привет мир!
    Deutsch: Hallo Welt!
    """.encode('utf-8')
    
    mixed_result = chardet.detect(mixed_content)
    print("✓ Mixed language content detection:")
    print(f"  Encoding: {mixed_result['encoding']}")
    print(f"  Confidence: {mixed_result['confidence']:.3f}")
    print(f"  Language: {mixed_result.get('language', 'N/A')}")
    
    # Test different encodings of the same text
    test_text = "Hello World! Testing different encodings: café, naïve, résumé"
    
    encodings_to_test = ['utf-8', 'latin-1', 'cp1252', 'ascii']
    print("✓ Same text in different encodings:")
    
    for encoding in encodings_to_test:
        try:
            encoded_text = test_text.encode(encoding)
            result = chardet.detect(encoded_text)
            
            print(f"  {encoding} encoding:")
            print(f"    Detected as: {result['encoding']}")
            print(f"    Confidence: {result['confidence']:.3f}")
            
        except UnicodeEncodeError:
            print(f"  {encoding} encoding: Cannot encode this text")
        print()
    
    # Show confidence interpretation
    print("✓ Confidence score interpretation:")
    confidence_levels = [
        (0.0, 0.3, "Low confidence - uncertain detection"),
        (0.3, 0.7, "Medium confidence - reasonable detection"),
        (0.7, 0.95, "High confidence - reliable detection"),
        (0.95, 1.0, "Very high confidence - certain detection")
    ]
    
    for min_conf, max_conf, description in confidence_levels:
        print(f"  {min_conf:.1f} - {max_conf:.1f}: {description}")
    
    print("✓ Multi-language text detection completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"chardet example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: chardet functionality test 3

```python
import chardet
import io

print("Example 3: Incremental Character Detection")

try:
    from chardet import UniversalDetector
    
    print("✓ Testing incremental detection with UniversalDetector:")
    
    # Create detector instance
    detector = UniversalDetector()
    
    # Sample text in different chunks
    text_chunks = [
        b'This is the beginning of a longer text sample ',
        b'that will be processed incrementally. ',
        b'The UniversalDetector can analyze text ',
        b'piece by piece and provide detection results ',
        b'when enough data has been processed.'
    ]
    
    print("✓ Processing text chunks incrementally:")
    
    for i, chunk in enumerate(text_chunks, 1):
        detector.feed(chunk)
        print(f"  Chunk {i}: {chunk.decode('utf-8', errors='ignore')[:50]}...")
        print(f"    Done: {detector.done}")
        
        if detector.done:
            break
    
    # Get final result
    detector.close()
    result = detector.result
    
    print(f"✓ Final incremental detection result:")
    print(f"  Encoding: {result['encoding']}")
    print(f"  Confidence: {result['confidence']:.3f}")
    print(f"  Language: {result.get('language', 'N/A')}")
    assert result["encoding"] == "ascii", result  # language is a guess that varies; not asserted
    
    # Test with mixed encoding chunks
    print("✓ Testing with mixed content:")
    
    detector2 = UniversalDetector()
    mixed_chunks = [
        "Hello World! ".encode('utf-8'),
        "Café résumé ".encode('utf-8'), 
        "naïve façade".encode('utf-8')
    ]
    
    for chunk in mixed_chunks:
        detector2.feed(chunk)
        if detector2.done:
            break
    
    detector2.close()
    mixed_result = detector2.result
    
    print(f"  Mixed content encoding: {mixed_result['encoding']}")
    print(f"  Mixed content confidence: {mixed_result['confidence']:.3f}")
    assert mixed_result["encoding"] == "utf-8", mixed_result
    
    # Test reset functionality
    print("✓ Testing detector reset:")
    detector.reset()
    print(f"  Detector reset: {detector.done}")
    
    # Feed new data after reset
    detector.feed(b'New text after reset')
    detector.close()
    reset_result = detector.result
    print(f"  Post-reset encoding: {reset_result['encoding']}")
    assert reset_result["encoding"] == "ascii", reset_result
    
    print("✓ Incremental character detection completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"chardet example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: chardet functionality test 4

```python
import chardet

print("Example 4: File and Stream Encoding Detection")

try:
    import tempfile
    import os
    
    print("✓ Testing file-based encoding detection:")
    
    # Create temporary files with different encodings
    test_files = []
    
    # UTF-8 file
    with tempfile.NamedTemporaryFile(mode='wb', delete=False) as f:
        content = "Hello World! This is UTF-8 content with special chars: café, naïve, résumé"
        f.write(content.encode('utf-8'))
        test_files.append(('utf-8', f.name))
    
    # Latin-1 file  
    with tempfile.NamedTemporaryFile(mode='wb', delete=False) as f:
        content = "Hello World! This is Latin-1 content: café"
        f.write(content.encode('latin-1'))
        test_files.append(('latin-1', f.name))
    
    # ASCII file
    with tempfile.NamedTemporaryFile(mode='wb', delete=False) as f:
        content = "Hello World! This is plain ASCII content."
        f.write(content.encode('ascii'))
        test_files.append(('ascii', f.name))
    
    print("✓ Detecting encoding of temporary files:")
    
    for expected_encoding, filepath in test_files:
        with open(filepath, 'rb') as f:
            raw_data = f.read()
            result = chardet.detect(raw_data)
            
            print(f"  File with {expected_encoding} content:")
            print(f"    Detected: {result['encoding']}")
            print(f"    Confidence: {result['confidence']:.3f}")
            print(f"    File size: {len(raw_data)} bytes")
            print()
    
    # Test partial file reading
    print("✓ Testing partial file reading:")
    
    with open(test_files[0][1], 'rb') as f:
        # Read only first 20 bytes
        partial_data = f.read(20)
        partial_result = chardet.detect(partial_data)
        
        print(f"  Partial read (20 bytes):")
        print(f"    Detected: {partial_result['encoding']}")
        print(f"    Confidence: {partial_result['confidence']:.3f}")
    
    # Test with binary stream
    print("✓ Testing with binary stream:")
    
    from io import BytesIO
    
    stream_content = "Stream content with UTF-8: 你好世界".encode('utf-8')
    stream = BytesIO(stream_content)
    
    stream_data = stream.read()
    stream_result = chardet.detect(stream_data)
    
    print(f"  Stream detection:")
    print(f"    Encoding: {stream_result['encoding']}")
    print(f"    Confidence: {stream_result['confidence']:.3f}")
    
    # Cleanup temporary files
    for _, filepath in test_files:
        try:
            os.unlink(filepath)
        except:
            pass
            raise
    
    # Test empty and minimal data
    print("✓ Testing edge cases:")
    
    empty_result = chardet.detect(b'')
    print(f"  Empty data: {empty_result}")
    
    minimal_result = chardet.detect(b'A')
    print(f"  Single byte: {minimal_result['encoding']}, confidence: {minimal_result['confidence']:.3f}")
    
    print("✓ File and stream encoding detection completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"chardet example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: chardet functionality test 5

```python
import chardet

print("Example 5: Advanced Encoding Scenarios and Performance")

try:
    import time
    
    print("✓ Testing performance with different data sizes:")
    
    # Generate test data of different sizes
    base_text = "The quick brown fox jumps over the lazy dog. " * 100
    test_sizes = [100, 1000, 10000, 50000]
    
    for size in test_sizes:
        # Create text of specific size
        text_data = (base_text * (size // len(base_text) + 1))[:size]
        byte_data = text_data.encode('utf-8')
        
        # Time the detection
        start_time = time.time()
        result = chardet.detect(byte_data)
        detection_time = time.time() - start_time
        
        print(f"  Size: {size:5d} bytes")
        print(f"    Detection time: {detection_time:.4f} seconds")
        print(f"    Detected: {result['encoding']}")
        print(f"    Confidence: {result['confidence']:.3f}")
        print()
    
    # Test with challenging encoding scenarios
    print("✓ Testing challenging encoding scenarios:")
    
    challenging_cases = {
        'Mixed ASCII/UTF-8': b'ASCII text \xe2\x9c\x93 with UTF-8 checkmark',
        'Windows-1252': b'Windows-1252: \x93smart quotes\x94 and \x96em dash\x97',
        'ISO-8859-1': b'ISO-8859-1: caf\xe9, na\xefve, r\xe9sum\xe9',
        'UTF-16 BOM': b'\xff\xfeU\x00T\x00F\x00-\x001\x006\x00 \x00t\x00e\x00x\x00t\x00',
        'UTF-8 BOM': b'\xef\xbb\xbfUTF-8 with BOM'
    }
    
    for case_name, data in challenging_cases.items():
        result = chardet.detect(data)
        print(f"  {case_name}:")
        print(f"    Detected: {result['encoding']}")
        print(f"    Confidence: {result['confidence']:.3f}")
        print(f"    Data preview: {data[:30]}...")
        print()
    
    # Test confidence thresholds
    print("✓ Testing confidence-based decisions:")
    
    confidence_samples = [
        b'A',  # Very short - low confidence
        b'Hello',  # Short ASCII - medium confidence
        b'Hello World! This is a longer English text sample.',  # Long ASCII - high confidence
        b'\xe4\xb8\xad\xe6\x96\x87\xe6\xb5\x8b\xe8\xaf\x95'  # Chinese UTF-8 - high confidence
    ]
    
    confidence_thresholds = [0.5, 0.7, 0.9]
    
    for i, sample in enumerate(confidence_samples, 1):
        result = chardet.detect(sample)
        confidence = result['confidence']
        encoding = result['encoding']
        
        print(f"  Sample {i}: confidence = {confidence:.3f}")
        
        # Determine reliability based on thresholds
        if confidence >= 0.9:
            reliability = "Very reliable"
        elif confidence >= 0.7:
            reliability = "Reliable"
        elif confidence >= 0.5:
            reliability = "Moderately reliable"
        else:
            reliability = "Low reliability"
        
        print(f"    Encoding: {encoding} ({reliability})")
        print(f"    Sample: {sample}")
        print()
    
    # Show chardet version and supported encodings info
    print("✓ Chardet package information:")
    print(f"  Version: {chardet.__version__}")
    
    # Show available modules
    chardet_modules = [attr for attr in dir(chardet) if not attr.startswith('_')]
    print(f"  Available modules: {len(chardet_modules)}")
    print(f"  Key modules: {chardet_modules[:5]}")
    
    print("✓ Advanced encoding scenarios and performance testing completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"chardet example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
