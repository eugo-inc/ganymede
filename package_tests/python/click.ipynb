# click: 5 Basic Usage Examples

```python
# Example 1: Basic command
import click
@click.command()
def hello():
    click.echo('Hello, world!')
# hello()  # Uncomment to run as script
```

```python
# Example 2: Command with arguments
import click
@click.command()
@click.argument('name')
def greet(name):
    click.echo(f'Hello, {name}!')
# greet()  # Uncomment to run as script
```

```python
# Example 3: Command with options
import click
@click.command()
@click.option('--count', default=1, help='Number of greetings.')
def repeat(count):
    for _ in range(count):
        click.echo('Hi!')
# repeat()  # Uncomment to run as script
```

```python
# Example 4: Group commands
import click
@click.group()
def cli(): pass
@cli.command()
def foo():
    click.echo('foo')
@cli.command()
def bar():
    click.echo('bar')
# cli()  # Uncomment to run as script
```

```python
# Example 5: Prompting for user input
import click
@click.command()
def ask():
    name = click.prompt('Your name')
    click.echo(f'Hello, {name}!')
# ask()  # Uncomment to run as script
```
