## Example 4: Plugin Discovery and Loading Mechanics

```python
import click
import click_plugins
import os
import sys
import tempfile
from importlib.metadata import EntryPoint, entry_points
from click.testing import CliRunner
from click_plugins import with_plugins
from click_plugins.core import BrokenCommand

print("Example 4: Plugin Discovery and Entry Point Management")

try:
    print("✓ Plugin discovery mechanisms:")
    
    print("✓ Entry point discovery with importlib.metadata:")
    print("  - with_plugins() registers one command per entry point it is given")
    print("  - Entry points are defined in package metadata")
    print("  - Installed packages are scanned for matching entry points")
    
    # Example of finding entry points manually
    def find_entry_points(group_name):
        """Find entry points in a given group."""
        return list(entry_points(group=group_name))
    
    # Test with common entry point group names
    test_groups = [
        'console_scripts',
        'click.commands',
        'pytest.plugins', 
        'setuptools.commands'
    ]
    
    print("✓ Scanning common entry point groups:")
    for group in test_groups:
        found = find_entry_points(group)
        print(f"  {group}: {len(found)} entry points found")
        if found and len(found) <= 3:  # Show examples for small groups
            for ep in found:
                print(f"    - {ep.name}")
    assert len(find_entry_points('console_scripts')) > 0
    
    # Load plugins through with_plugins(): one entry point resolves to a real click command; the other names a module that
    # does not exist and must become click_plugins' BrokenCommand instead of taking the CLI down (measured)
    plugin_dir = tempfile.mkdtemp()
    with open(os.path.join(plugin_dir, "nb_demo_backup_plugin.py"), "w") as plugin_file:
        plugin_file.write(
            "import click\n\n@click.command()\n@click.option('--destination', required=True)\n"
            "def backup(destination):\n    \"\"\"Backup application data.\"\"\"\n    click.echo(f'Creating backup to {destination}')\n"
        )
    sys.path.insert(0, plugin_dir)
    plugins = [
        EntryPoint(name="backup", value="nb_demo_backup_plugin:backup", group="myapp.plugins"),
        EntryPoint(name="deploy", value="nb_demo_missing_module:deploy", group="myapp.plugins"),
    ]
    
    @with_plugins(plugins)
    @click.group()
    def cli():
        """MyApp - Extensible CLI application."""
    
    runner = CliRunner()
    result = runner.invoke(cli, ["backup", "--destination", "/tmp/backups"])
    assert (result.exit_code, result.output) == (0, "Creating backup to /tmp/backups\n"), (result.exit_code, result.output)
    assert isinstance(cli.commands["deploy"], BrokenCommand), cli.commands
    broken = runner.invoke(cli, ["deploy"])
    assert broken.exit_code == 1 and "could not be loaded" in broken.output, (broken.exit_code, broken.output)
    print(f"✓ with_plugins registered {sorted(cli.commands)}; 'deploy' is a BrokenCommand, as click_plugins intends")
    
    # Show the plugin loading workflow
    print("✓ Plugin loading workflow:")
    workflow_steps = [
        "1. @with_plugins decorator applied to click.Group",
        "2. Entry point groups specified in decorator", 
        "3. importlib.metadata yields the matching entry points",
        "4. Entry points are loaded and validated",
        "5. Plugin commands registered with CLI group",
        "6. Commands become available for execution"
    ]
    
    for step in workflow_steps:
        print(f"  {step}")
    
    # Show entry point definition examples
    print("✓ Entry point definition examples:")
    
    # Setup.py style
    setup_entry_points = '''
    entry_points={
        "myapp.plugins": [
            "backup = myapp_backup.commands:backup",
            "deploy = myapp_deploy.cli:deploy_group", 
            "analyze = myapp_tools.analysis:analyze_cmd"
        ]
    }
    '''
    
    print("  setup.py format:")
    for line in setup_entry_points.strip().split('\n'):
        print(f"    {line}")
    
    # pyproject.toml style
    pyproject_entry_points = '''
    [project.entry-points."myapp.plugins"]
    backup = "myapp_backup.commands:backup"
    deploy = "myapp_deploy.cli:deploy_group"
    analyze = "myapp_tools.analysis:analyze_cmd"
    '''
    
    print("  pyproject.toml format:")
    for line in pyproject_entry_points.strip().split('\n'):
        print(f"    {line}")
    
    # Show plugin validation and error handling
    print("✓ Plugin validation and error handling:")
    print("  - Entry points must resolve to valid click commands")
    print("  - Loading errors are typically silently ignored")
    print("  - Invalid plugins don't break the main application")
    print("  - Use logging to debug plugin loading issues")
    
    # Show debugging plugin discovery
    print("✓ Debugging plugin discovery:")
    debug_tips = [
        "Check entry point syntax in setup.py/pyproject.toml",
        "Verify plugin package is properly installed",
        "Use importlib.metadata.entry_points(group=...) to test manually",
        "Check import paths and function names",
        "Enable verbose logging during development"
    ]
    
    for tip in debug_tips:
        print(f"  - {tip}")
    
    # Show advanced entry point patterns
    print("✓ Advanced entry point patterns:")
    print("  - Group plugins by functionality (commands, extensions, hooks)")
    print("  - Use version-specific entry points for compatibility")
    print("  - Implement plugin priority and ordering systems")
    print("  - Support conditional plugin loading")
    
    print("✓ Plugin discovery and entry point management completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"click_plugins example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: Real-World Plugin Architecture Example

```python
import click
import click_plugins
from click_plugins import with_plugins

print("Example 5: Real-World Plugin Architecture and Best Practices")

try:
    
    print("✓ Real-world plugin architecture:")
    
    # Show complete plugin ecosystem example
    print("✓ Complete plugin ecosystem structure:")
    ecosystem_structure = '''
    my-cli-tool/                    # Main application
    ├── setup.py                   # Main app setup
    ├── mycli/
    │   ├── __init__.py
    │   ├── main.py               # CLI with @with_plugins
    │   └── core_commands.py      # Built-in commands
    
    my-cli-backup-plugin/          # Backup plugin package
    ├── setup.py                  # Plugin setup with entry points
    ├── mycli_backup/
    │   ├── __init__.py
    │   └── commands.py           # Plugin commands
    
    my-cli-deploy-plugin/          # Deploy plugin package  
    ├── setup.py                  # Plugin setup with entry points
    ├── mycli_deploy/
    │   ├── __init__.py
    │   └── commands.py           # Plugin commands
    '''
    
    for line in ecosystem_structure.strip().split('\n')[:12]:
        print(f"  {line}")
    
    # Show main application setup
    print("✓ Main application CLI setup:")
    main_cli_example = '''
    # mycli/main.py
    import click
    from click_plugins import with_plugins
    
    @with_plugins(['mycli.plugins'])
    @click.group()
    @click.option('--verbose', is_flag=True, help='Verbose output')
    @click.pass_context
    def cli(ctx, verbose):
        """MyCLI - Extensible command line tool."""
        ctx.ensure_object(dict)
        ctx.obj['verbose'] = verbose
    
    # Built-in commands
    @cli.command()
    @click.pass_context
    def status(ctx):
        """Show application status."""
        if ctx.obj['verbose']:
            click.echo("Application status: Running (verbose)")
        else:
            click.echo("Status: OK")
    
    if __name__ == '__main__':
        cli()
    '''
    
    for line in main_cli_example.strip().split('\n')[:16]:
        print(f"  {line}")
    
    # Show plugin package setup
    print("✓ Plugin package setup example:")
    plugin_setup = '''
    # my-cli-backup-plugin/setup.py
    from setuptools import setup, find_packages
    
    setup(
        name="mycli-backup-plugin",
        version="1.0.0",
        packages=find_packages(),
        install_requires=[
            "click>=7.0",
            "my-cli-tool>=1.0.0"  # Main app dependency
        ],
        entry_points={
            "mycli.plugins": [
                "backup = mycli_backup.commands:backup",
                "restore = mycli_backup.commands:restore"
            ]
        },
        author="Plugin Developer",
        description="Backup functionality for MyCLI"
    )
    '''
    
    for line in plugin_setup.strip().split('\n'):
        print(f"  {line}")
    
    # Show plugin command implementation
    print("✓ Plugin command implementation:")
    plugin_commands = '''
    # mycli_backup/commands.py
    import click
    
    @click.command()
    @click.option('--destination', required=True, help='Backup destination')
    @click.option('--compress', is_flag=True, help='Compress backup')
    @click.pass_context
    def backup(ctx, destination, compress):
        """Create application backup."""
        verbose = ctx.obj.get('verbose', False)
        
        if verbose:
            click.echo(f"Creating backup to {destination}")
            click.echo(f"Compression: {'enabled' if compress else 'disabled'}")
        
        # Backup logic here
        click.echo("Backup completed successfully")
    
    @click.command()
    @click.argument('backup_file')
    def restore(backup_file):
        """Restore from backup file."""
        click.echo(f"Restoring from {backup_file}")
        # Restore logic here
        click.echo("Restore completed successfully")
    '''
    
    for line in plugin_commands.strip().split('\n')[:16]:
        print(f"  {line}")
    
    # Show installation and usage workflow
    print("✓ Installation and usage workflow:")
    workflow = [
        "1. Install main application: pip install my-cli-tool",
        "2. Install plugins: pip install mycli-backup-plugin",
        "3. Plugins are automatically discovered and loaded",
        "4. New commands appear in CLI help automatically",
        "5. Use commands: mycli backup --destination /tmp/backup"
    ]
    
    for step in workflow:
        print(f"  {step}")
    
    # Show best practices for plugin development
    print("✓ Plugin development best practices:")
    best_practices = [
        "Use consistent naming conventions (app-plugin-name)",
        "Define clear dependencies on main application",
        "Follow semantic versioning for compatibility",
        "Include comprehensive documentation and help text",
        "Handle errors gracefully with informative messages",
        "Use click's context system for shared state",
        "Test plugins with the main application",
        "Provide plugin configuration options"
    ]
    
    for practice in best_practices:
        print(f"  - {practice}")
    
    # Show real-world examples
    print("✓ Real-world plugin examples:")
    real_world = [
        "pip: Extensible package manager with plugin support",
        "flask: Web framework with blueprint plugin system",
        "pytest: Testing framework with extensive plugin ecosystem", 
        "cookiecutter: Template engine with hook plugins",
        "pre-commit: Git hook framework with plugin architecture"
    ]
    
    for example in real_world:
        print(f"  - {example}")
    
    print("✓ Real-world plugin architecture and best practices completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"click_plugins example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
