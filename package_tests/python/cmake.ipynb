# cmake: 5 Basic Usage Examples

This notebook demonstrates basic usage of the [cmake](https://pypi.org/project/cmake/) Python package, which provides access to the CMake build system and its executable.



```python
# Example 1: Check CMake Version
import cmake
print("CMake version:", cmake.__version__)
```

```python
# Example 2: Find CMake Executable Path
print("CMake executable directory:", cmake.CMAKE_BIN_DIR)
```

```python
# Example 3: Run a Simple CMake Command (show help)
import subprocess
import sys
help_output = subprocess.check_output([f'{cmake.CMAKE_BIN_DIR}/cmake', '--help'])
print(help_output.decode('utf-8').splitlines()[0])
```

```python
# Example 4: Create and Configure a Minimal CMake Project
import os, tempfile
with tempfile.TemporaryDirectory() as tmpdir:
    cmakelists = os.path.join(tmpdir, 'CMakeLists.txt')
    with open(cmakelists, 'w') as f:
        f.write("""
cmake_minimum_required(VERSION 3.10)
project(HelloCMake)
add_custom_target(hello ALL COMMAND ${CMAKE_COMMAND} -E echo Hello, CMake!)
""")
    subprocess.check_call([f'{cmake.CMAKE_BIN_DIR}/cmake', '-S', tmpdir, '-B', tmpdir])
    subprocess.check_call([f'{cmake.CMAKE_BIN_DIR}/cmake', '--build', tmpdir])
```

```python
# Example 5: Query CMake Capabilities (JSON)
import json
cap = subprocess.check_output([f'{cmake.CMAKE_BIN_DIR}/cmake', '-E', 'capabilities'])
print(json.loads(cap))
```
