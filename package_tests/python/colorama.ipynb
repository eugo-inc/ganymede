# colorama: 5 Basic Usage Examples

```python
# Example 1: Basic colored text output
from colorama import init, Fore, Back, Style

# Initialize colorama for cross-platform compatibility
init()

# Basic colored text examples
print(Fore.RED + 'This text is red!')
print(Fore.GREEN + 'This text is green!')
print(Fore.BLUE + 'This text is blue!')
print(Style.RESET_ALL + 'Back to normal text')

# Verify colorama is working
print(f"Colorama initialized: {hasattr(Fore, 'RED')}")
```

```python
# Example 2: Background colors and text styles
from colorama import Fore, Back, Style

# Background colors
print(Back.YELLOW + Fore.BLACK + 'Black on yellow')
print(Back.MAGENTA + Fore.WHITE + 'White on magenta')

# Text styles
print(Style.BRIGHT + Fore.CYAN + 'Bright cyan text')
print(Style.DIM + Fore.RED + 'Dim red text')

# Combined styles
print(Style.BRIGHT + Back.BLUE + Fore.YELLOW + 'Bright yellow on blue')
print(Style.RESET_ALL + 'Reset to normal')
```

```python
# Example 3: Cursor positioning
from colorama import init, Cursor, Fore

# Initialize with autoreset
init(autoreset=True)

# Cursor positioning examples
print("Original text")
print(Cursor.UP(1) + Cursor.FORWARD(15) + Fore.GREEN + "[MOVED!]")

# Test cursor functions
print(f"\nCursor UP: '{Cursor.UP(1)}'")
print(f"Cursor FORWARD: '{Cursor.FORWARD(5)}'")

# Verify cursor module works
print(f"Cursor module available: {hasattr(Cursor, 'UP')}")
```

```python
# Example 4: Platform detection and ANSI codes
from colorama import init, Fore, Back, Style
import sys, platform

# Check platform
print(f"Platform: {platform.system()}")

# Initialize with options
init(autoreset=False, convert=True)

# Test ANSI codes
print(f"ANSI RED: {repr(Fore.RED)}")
print(f"ANSI RESET: {repr(Style.RESET_ALL)}")

# Colored output
colored = Fore.BLUE + "Blue text" + Style.RESET_ALL
print(colored)
```
