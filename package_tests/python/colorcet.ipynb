## Example 1: Basic Colormap Access and Information

```python
import colorcet as cc

print("Example 1: Basic Colormap Access and Information")

try:
    print("✓ colorcet package information:")
    
    # Show total number of colormaps available
    all_attrs = [attr for attr in dir(cc) if not attr.startswith('_')]
    colormap_attrs = [attr for attr in all_attrs if isinstance(getattr(cc, attr), list)]
    
    print(f"  Total attributes: {len(all_attrs)}")
    print(f"  Available colormaps: {len(colormap_attrs)}")
    
    # Show some popular colormaps and their properties
    popular_maps = ['fire', 'rainbow', 'viridis', 'coolwarm', 'gray']
    
    print("✓ Popular colormaps:")
    for cmap_name in popular_maps:
        if hasattr(cc, cmap_name):
            cmap = getattr(cc, cmap_name)
            print(f"  {cmap_name}: {len(cmap)} colors")
            # Show first and last colors as hex
            if len(cmap) > 0:
                first_color = cmap[0] if isinstance(cmap[0], str) else f"#{cmap[0]:06x}"
                last_color = cmap[-1] if isinstance(cmap[-1], str) else f"#{cmap[-1]:06x}"
                print(f"    Range: {first_color} → {last_color}")
    
    # Show colormap categories
    print("✓ Colormap categories:")
    categories = {
        'Linear': [attr for attr in colormap_attrs if 'linear' in attr.lower()],
        'Diverging': [attr for attr in colormap_attrs if 'diverging' in attr.lower()], 
        'Cyclic': [attr for attr in colormap_attrs if 'cyclic' in attr.lower()],
        'Rainbow': [attr for attr in colormap_attrs if 'rainbow' in attr.lower()],
        'Glasbey': [attr for attr in colormap_attrs if 'glasbey' in attr.lower()]
    }
    
    for category, maps in categories.items():
        print(f"  {category}: {len(maps)} colormaps")
        if maps:
            print(f"    Examples: {maps[:3]}")
    
    # Test colormap access patterns
    print("✓ Colormap access patterns:")
    
    if hasattr(cc, 'fire'):
        fire_map = cc.fire
        print(f"  Direct access: cc.fire → {len(fire_map)} colors")
    
    # Show matplotlib integration
    if hasattr(cc, 'cm'):
        print(f"  Matplotlib colormaps: cc.cm available")
        print(f"  Usage: plt.imshow(data, cmap=cc.cm.fire)")
    
    # Show palette access
    if hasattr(cc, 'palette'):
        print(f"  Palette access: cc.palette available")
        print(f"  Usage: cc.palette['fire'] or cc.palette.fire")
    
    # Show aliases
    if hasattr(cc, 'aliases'):
        aliases = cc.aliases
        print(f"  Colormap aliases: {len(aliases)} aliases available")
        sample_aliases = list(aliases.keys())[:5]
        print(f"    Sample aliases: {sample_aliases}")
    
    print("✓ Basic colormap access and information completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"colorcet example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Matplotlib Integration and Colormap Usage

```python
import colorcet as cc
import numpy as np

print("Example 2: Matplotlib Integration and Colormap Usage")

try:
    # Show matplotlib integration capabilities
    print("✓ Matplotlib colormap integration:")
    
    # Check if cm module is available for matplotlib
    if hasattr(cc, 'cm'):
        print("  cc.cm module available for matplotlib")
        
        # List some common colormaps available through cc.cm
        cm_attrs = [attr for attr in dir(cc.cm) if not attr.startswith('_')]
        print(f"  Available through cc.cm: {len(cm_attrs)} colormaps")
        
        # Show examples of how to use with matplotlib
        print("  Usage patterns:")
        print("    plt.imshow(data, cmap=cc.cm.fire)")
        print("    plt.contourf(X, Y, Z, cmap=cc.cm.rainbow)")
        print("    plt.scatter(x, y, c=values, cmap=cc.cm.coolwarm)")
    
    # Demonstrate color value extraction and manipulation
    print("✓ Colormap data extraction:")
    
    if hasattr(cc, 'fire'):
        fire_cmap = cc.fire
        print(f"  Fire colormap: {len(fire_cmap)} discrete colors")
        
        # Show color sampling
        if len(fire_cmap) > 10:
            sample_indices = np.linspace(0, len(fire_cmap)-1, 5, dtype=int)
            print("  Sample colors from fire colormap:")
            for i, idx in enumerate(sample_indices):
                color = fire_cmap[idx]
                print(f"    Position {i/4:.1f}: {color}")
    
    # Show colormap comparison
    print("✓ Colormap comparison:")
    cmaps_to_compare = ['fire', 'rainbow', 'coolwarm', 'gray', 'blues']
    
    for cmap_name in cmaps_to_compare:
        if hasattr(cc, cmap_name):
            cmap = getattr(cc, cmap_name)
            if isinstance(cmap, list) and len(cmap) > 0:
                print(f"  {cmap_name}: {len(cmap)} colors, range {cmap[0]} → {cmap[-1]}")
    
    print("✓ Matplotlib integration and colormap usage completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"colorcet example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: Perceptually Uniform Colormap Features

```python
import colorcet as cc

print("Example 3: Perceptually Uniform Colormap Features")

try:
    # Demonstrate perceptually uniform properties
    print("✓ Perceptually uniform colormap benefits:")
    print("  - Avoids rainbow colormap pitfalls")
    print("  - Equal perceptual steps between colors")
    print("  - Better for scientific visualization")
    print("  - Accessible for color vision deficiencies")
    
    # Show different colormap categories and their properties
    print("✓ Colormap categories and properties:")
    
    category_examples = {
        'Sequential': ['fire', 'viridis', 'plasma', 'inferno', 'blues', 'grays'],
        'Diverging': ['coolwarm', 'bwr', 'RdBu', 'RdYlBu'],
        'Cyclic': ['twilight', 'hsv'],
        'Qualitative': ['glasbey', 'glasbey_light', 'glasbey_dark']
    }
    
    for category, cmap_names in category_examples.items():
        print(f"  {category} colormaps:")
        available_in_category = []
        for cmap_name in cmap_names:
            if hasattr(cc, cmap_name):
                cmap = getattr(cc, cmap_name)
                if isinstance(cmap, list):
                    available_in_category.append(f"{cmap_name}({len(cmap)})")
        
        if available_in_category:
            print(f"    Available: {', '.join(available_in_category[:3])}")
    
    # Demonstrate colormap properties
    print("✓ Colormap analysis:")
    
    if hasattr(cc, 'fire'):
        fire_map = cc.fire
        print(f"  Fire colormap analysis:")
        print(f"    Length: {len(fire_map)} colors")
        print(f"    Type: {type(fire_map[0]).__name__ if fire_map else 'unknown'}")
        
        # Show color format (hex strings vs tuples)
        if fire_map:
            first_color = fire_map[0]
            if isinstance(first_color, str):
                print(f"    Format: Hex strings (e.g., {first_color})")
            else:
                print(f"    Format: Color tuples (e.g., {first_color})")
    
    # Show colormap discovery
    print("✓ Colormap discovery:")
    all_cmaps = [attr for attr in dir(cc) if not attr.startswith('_') and isinstance(getattr(cc, attr), list)]
    
    if all_cmaps:
        print(f"  Total available: {len(all_cmaps)}")
        print(f"  Longest names: {sorted(all_cmaps, key=len, reverse=True)[:3]}")
        print(f"  Shortest names: {sorted(all_cmaps, key=len)[:3]}")
    
    print("✓ Perceptually uniform colormap features completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"colorcet example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: Practical Data Visualization Applications

```python
import colorcet as cc
import numpy as np

print("Example 4: Practical Data Visualization Applications")

try:
    # Create sample data for visualization examples
    print("✓ Creating sample data for visualization:")
    
    # 2D data for heatmap
    x = np.linspace(-2, 2, 20)
    y = np.linspace(-2, 2, 20)
    X, Y = np.meshgrid(x, y)
    Z = np.exp(-(X**2 + Y**2))  # 2D Gaussian
    
    print(f"  Sample data shape: {Z.shape}")
    print(f"  Data range: {Z.min():.3f} to {Z.max():.3f}")
    
    # 1D data for line plots with color coding
    time_series = np.sin(np.linspace(0, 4*np.pi, 100)) + 0.1*np.random.randn(100)
    categories = np.random.choice(['A', 'B', 'C', 'D'], 50)
    
    print(f"  Time series length: {len(time_series)}")
    print(f"  Categories: {np.unique(categories)}")
    
    # Demonstrate colormap selection for different data types
    print("✓ Colormap recommendations by data type:")
    
    data_type_recommendations = {
        'Sequential (single-hue progression)': ['fire', 'viridis', 'plasma', 'blues'],
        'Diverging (center point important)': ['coolwarm', 'bwr'],
        'Qualitative (categorical data)': ['glasbey', 'glasbey_light'],
        'Cyclic (circular/periodic data)': ['twilight', 'hsv']
    }
    
    for data_type, recommended_cmaps in data_type_recommendations.items():
        print(f"  {data_type}:")
        available_recommended = []
        for cmap_name in recommended_cmaps:
            if hasattr(cc, cmap_name):
                available_recommended.append(cmap_name)
        
        if available_recommended:
            print(f"    Recommended: {', '.join(available_recommended[:2])}")
    
    # Show color extraction for discrete use
    print("✓ Color extraction for discrete visualization:")
    
    if hasattr(cc, 'fire'):
        fire_cmap = cc.fire
        
        # Extract specific number of colors
        n_colors_needed = 5
        if len(fire_cmap) >= n_colors_needed:
            indices = np.linspace(0, len(fire_cmap)-1, n_colors_needed, dtype=int)
            selected_colors = [fire_cmap[i] for i in indices]
            
            print(f"  Extracted {n_colors_needed} colors from fire colormap:")
            for i, color in enumerate(selected_colors):
                print(f"    Color {i+1}: {color}")
    
    # Show colormap combinations and alternatives
    print("✓ Colormap usage patterns:")
    
    usage_examples = [
        "Heatmaps: cc.fire (sequential data)",
        "Scatter plots: cc.viridis (continuous values)", 
        "Correlation matrices: cc.coolwarm (diverging from center)",
        "Multi-category plots: cc.glasbey (distinct categories)",
        "Periodic data: cc.twilight (cyclic patterns)"
    ]
    
    print("  Common applications:")
    for example in usage_examples:
        print(f"    {example}")
    
    print("✓ Practical data visualization applications completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"colorcet example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```
