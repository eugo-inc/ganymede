# colorful: 5 Basic Usage Examples

```python
# Example 1: Basic colorful usage
import colorful

# colorful turns colors off when output is not a terminal; pick the 256-color palette explicitly
colorful.use_256_ansi_colors()

print('Testing colorful basic functionality:')

# Basic colors
red_text = colorful.red('This is red text')
print(red_text)
assert '\x1b[' in str(red_text) and 'This is red text' in str(red_text)

blue_text = colorful.blue('This is blue text')
print(blue_text)

# Combined styling
styled_text = colorful.bold_green('This is bold green')
print(styled_text)
assert str(styled_text) != str(colorful.green('This is bold green'))
```

```python
# Example 2: Text styles and effects
import colorful as cf

# Text styles
print(cf.bold('Bold text'))
print(cf.italic('Italic text'))
print(cf.underlined('Underlined text'))

# Combining styles with colors
print(cf.bold_red('Bold red text'))
print(cf.italic_blue('Italic blue text'))

# Styles wrap the text in ANSI escape codes. A ColorfulString is not a str and its len()
# counts only the visible characters, so compare the str() forms.
basic = cf.red('test')
styled = cf.bold_red('test')
assert '\x1b[' in str(styled) and str(styled) != 'test', repr(str(styled))
assert str(styled) != str(basic)
print(f"Styles add codes: {str(styled)!r}")
```

```python
# Example 3: Rainbow effect and multiple colors
import colorful as cf

# Create a rainbow effect
rainbow_colors = ['red', 'yellow', 'green', 'cyan', 'blue', 'magenta']
rainbow_text = "RAINBOW"

print("Rainbow text:")
for i, char in enumerate(rainbow_text):
    color = rainbow_colors[i % len(rainbow_colors)]
    color_func = getattr(cf, color)
    print(color_func(char), end='')
print()  # New line

# Test all basic colors
colors = ['red', 'green', 'blue']
print(f"All colors work: {all(hasattr(cf, c) for c in colors)}")
```

```python
# Example 4: Environment and platform compatibility
import colorful as cf
import sys, os

# Platform information
print(f"Python version: {sys.version.split()[0]}")
print(f"Platform: {sys.platform}")
print(f"TTY: {sys.stdout.isatty()}")

# Test color output
test_colors = ['red', 'green', 'blue']
for color in test_colors:
    color_func = getattr(cf, color)
    result = color_func("test")
    print(f"{color}: {type(result).__name__}")

print("Platform compatibility verified")
```

```python
# Example 5: Error handling and edge cases
import colorful as cf

# Test edge cases
edge_cases = [
    ("", "empty string"),
    ("Unicode: ñáéíóú", "unicode"),
    ("  ", "spaces only")
]

for text, description in edge_cases:
    result = str(cf.green(text))
    assert text in result and result.startswith('\x1b['), (description, result)
    print(f"{description}: ✓")

# Final verification
print(f"Colorful fully functional: {hasattr(cf, 'red')}")
```
