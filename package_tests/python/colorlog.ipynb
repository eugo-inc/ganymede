# Colorlog Package Examples

This notebook demonstrates 5 basic, versatile, and comprehensive examples of the `colorlog` package.

Colorlog provides colored logging for Python applications.


## Example 1: Basic colored logging setup

```python
import logging
import colorlog

# Create a colored logger
logger = colorlog.getLogger('example_logger')
logger.setLevel(logging.DEBUG)

# Create colored handler
handler = colorlog.StreamHandler()
handler.setFormatter(colorlog.ColoredFormatter(
    '%(log_color)s%(levelname)s:%(name)s:%(message)s'
))

logger.addHandler(handler)

# Test different log levels
logger.debug('This is a debug message')
logger.info('This is an info message')
logger.warning('This is a warning message')
logger.error('This is an error message')
logger.critical('This is a critical message')

# Verify colorlog functionality
print(f"\nColorlog module imported: {colorlog.__name__}")
print(f"Logger created: {logger.name}")
print(f"Handler type: {type(handler).__name__}")
print(f"Formatter type: {type(handler.formatter).__name__}")
```

## Example 2: Custom color schemes and formats

```python
import logging
import colorlog

# Create logger with custom colors
custom_logger = colorlog.getLogger('custom_logger')
custom_logger.setLevel(logging.DEBUG)

# Custom color scheme
custom_handler = colorlog.StreamHandler()
custom_formatter = colorlog.ColoredFormatter(
    '%(log_color)s[%(asctime)s] %(levelname)-8s %(name)s: %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S',
    log_colors={
        'DEBUG': 'cyan',
        'INFO': 'green',
        'WARNING': 'yellow',
        'ERROR': 'red',
        'CRITICAL': 'red,bg_white',
    }
)

custom_handler.setFormatter(custom_formatter)
custom_logger.addHandler(custom_handler)

print("Custom colored logging:")
custom_logger.debug('Debug with timestamp')
custom_logger.info('Info with timestamp')
custom_logger.warning('Warning with timestamp')
custom_logger.error('Error with timestamp')
custom_logger.critical('Critical with timestamp')

# Verify customization works
print(f"\nCustom formatter configured: {hasattr(custom_formatter, 'log_colors')}")
print(f"Colorlog supports customization: {hasattr(colorlog, 'ColoredFormatter')}")
```

## Example 3: Integration with standard logging

```python
import logging
import colorlog
import sys

# Configure root logger with colorlog
root_logger = logging.getLogger()
root_logger.setLevel(logging.DEBUG)

# Clear existing handlers
root_logger.handlers.clear()

# Add colorlog handler
console_handler = colorlog.StreamHandler(sys.stdout)
console_formatter = colorlog.ColoredFormatter(
    '%(log_color)s%(levelname)-8s%(reset)s %(blue)s%(name)s%(reset)s %(message)s'
)
console_handler.setFormatter(console_formatter)
root_logger.addHandler(console_handler)

# Test with different logger names
loggers = [
    logging.getLogger('app.main'),
    logging.getLogger('app.database'),
    logging.getLogger('app.auth')
]

messages = [
    ('Application started', 'info'),
    ('Database connection established', 'info'),
    ('User authentication failed', 'warning')
]

print("Multi-logger colored output:")
for i, (message, level) in enumerate(messages):
    logger = loggers[i % len(loggers)]
    getattr(logger, level)(message)

# Test logging with exceptions
exception_logger = logging.getLogger('exception_test')
print("\nException logging:")
try:
    result = 10 / 0
except ZeroDivisionError:
    exception_logger.exception('Division by zero occurred')

# Verify integration
print(f"\nIntegration verification:")
print(f"Root logger handlers: {len(root_logger.handlers)}")
formatter_is_colorlog = any(isinstance(h.formatter, colorlog.ColoredFormatter) for h in root_logger.handlers)
print(f"Colorlog formatter in use: {formatter_is_colorlog}")
assert len(root_logger.handlers) == 1 and formatter_is_colorlog, root_logger.handlers
```

## Example 4: Environment-based configuration

```python
import logging
import colorlog
import os
import sys

# Check environment variables
print("Environment variables affecting colorlog:")
env_vars = ['NO_COLOR', 'FORCE_COLOR', 'TERM', 'COLORTERM']
for var in env_vars:
    value = os.environ.get(var, 'Not set')
    print(f"  {var}: {value}")

# Create conditional colored logging
def setup_logger(name, use_color=None):
    """Setup logger with optional color support"""
    logger = logging.getLogger(name)
    logger.setLevel(logging.DEBUG)
    
    # Clear existing handlers
    logger.handlers.clear()
    
    handler = logging.StreamHandler()
    
    # Determine if colors should be used
    if use_color is None:
        # Auto-detect based on environment
        use_color = (
            hasattr(handler.stream, 'isatty') and 
            handler.stream.isatty() and 
            os.environ.get('NO_COLOR') is None
        )
    
    if use_color:
        formatter = colorlog.ColoredFormatter(
            '%(log_color)s%(levelname)-8s%(reset)s %(message)s'
        )
    else:
        formatter = logging.Formatter(
            '%(levelname)-8s %(message)s'
        )
    
    handler.setFormatter(formatter)
    logger.addHandler(handler)
    
    return logger, use_color

# Test different configurations
print("\nAuto-detected color support:")
auto_logger, auto_color = setup_logger('auto')
print(f"Using colors: {auto_color}")
auto_logger.info('Auto-configured logging')
auto_logger.warning('This may or may not be colored')

print("\nForced color mode:")
color_logger, _ = setup_logger('forced_color', use_color=True)
color_logger.info('Forced color logging')
color_logger.error('This should be colored')

print("\nNo color mode:")
plain_logger, _ = setup_logger('no_color', use_color=False)
plain_logger.info('Plain logging')
plain_logger.error('This should not be colored')

# Test TTY detection
print(f"\nTTY detection:")
print(f"stdout.isatty(): {sys.stdout.isatty()}")
print(f"stderr.isatty(): {sys.stderr.isatty()}")

# Verify colorlog environment handling
print(f"\nColorlog handles environment properly: {hasattr(colorlog, 'ColoredFormatter')}")
```

## Example 5: Advanced features and performance

```python
import logging
import colorlog
import os
import time
import io

# Performance comparison
def performance_test():
    """Compare performance of colored vs plain logging"""
    # Setup loggers
    buffer1 = io.StringIO()
    buffer2 = io.StringIO()
    
    # Colored logger
    colored_logger = colorlog.getLogger('colored_perf')
    colored_logger.setLevel(logging.INFO)
    colored_handler = colorlog.StreamHandler(buffer1)
    colored_handler.setFormatter(colorlog.ColoredFormatter(
        '%(log_color)s%(levelname)s:%(name)s:%(message)s'
    ))
    colored_logger.addHandler(colored_handler)
    
    # Plain logger
    plain_logger = logging.getLogger('plain_perf')
    plain_logger.setLevel(logging.INFO)
    plain_handler = logging.StreamHandler(buffer2)
    plain_handler.setFormatter(logging.Formatter(
        '%(levelname)s:%(name)s:%(message)s'
    ))
    plain_logger.addHandler(plain_handler)
    
    # Test messages
    messages = [f"Test message {i}" for i in range(100)]
    
    # Colored logging performance
    start_time = time.time()
    for msg in messages:
        colored_logger.info(msg)
    colored_time = time.time() - start_time
    
    # Plain logging performance
    start_time = time.time()
    for msg in messages:
        plain_logger.info(msg)
    plain_time = time.time() - start_time
    
    return colored_time, plain_time

print("Performance comparison:")
colored_time, plain_time = performance_test()
print(f"Colored logging: {colored_time:.4f} seconds")
print(f"Plain logging: {plain_time:.4f} seconds")
if plain_time > 0:
    print(f"Overhead: {((colored_time - plain_time) / plain_time * 100):.1f}%")

# Test complex formatting
complex_logger = colorlog.getLogger('complex')
complex_logger.setLevel(logging.DEBUG)

complex_handler = colorlog.StreamHandler()
complex_formatter = colorlog.ColoredFormatter(
    '%(log_color)s%(asctime)s [%(levelname)-8s] %(name)s - %(message)s%(reset)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
complex_handler.setFormatter(complex_formatter)
complex_logger.addHandler(complex_handler)

print("\nComplex formatting:")
complex_logger.debug('Debug message with full context')
complex_logger.info('Info message with full context')
complex_logger.warning('Warning message with full context')

# Test large message handling
print("\nLarge message handling:")
large_message = "Large message: " + "A" * 100
complex_logger.info(large_message)

# Test color escape sequence handling
print("\nColor escape sequences:")
test_formatter = colorlog.ColoredFormatter(
    '%(log_color)s%(levelname)s%(reset)s - %(message)s'
)

# Get the format string with color codes
sample_record = logging.LogRecord(
    name='test', level=logging.INFO, pathname='', lineno=0,
    msg='Sample message', args=(), exc_info=None
)

formatted = test_formatter.format(sample_record)
print(f"Formatted message length: {len(formatted)}")
print(f"Contains escape codes: {len(formatted) > len('INFO - Sample message')}")

forced = colorlog.ColoredFormatter(
    '%(log_color)s%(levelname)s%(reset)s - %(message)s', force_color=True
).format(sample_record)
saved_force_color = os.environ.pop('FORCE_COLOR', None)
try:
    uncoloured = colorlog.ColoredFormatter(
        '%(log_color)s%(levelname)s%(reset)s - %(message)s', no_color=True
    ).format(sample_record)
finally:
    if saved_force_color is not None:
        os.environ['FORCE_COLOR'] = saved_force_color
print(f"Forced colour: {forced!r}")
print(f"No colour:     {uncoloured!r}")
assert forced.startswith('\x1b[32mINFO\x1b[0m'), f"force_color=True did not colour INFO green: {forced!r}"
assert uncoloured == 'INFO - Sample message', f"no_color=True left escape codes in: {uncoloured!r}"
assert forced.replace('\x1b[32m', '').replace('\x1b[0m', '') == uncoloured, forced

# Final verification
print(f"\nColorlog verification:")
print(f"ColoredFormatter available: {hasattr(colorlog, 'ColoredFormatter')}")
print(f"StreamHandler available: {hasattr(colorlog, 'StreamHandler')}")
print(f"getLogger function available: {hasattr(colorlog, 'getLogger')}")

all_functions_work = all([
    hasattr(colorlog, 'ColoredFormatter'),
    hasattr(colorlog, 'StreamHandler'),
    hasattr(colorlog, 'getLogger')
])
print(f"\nColorlog is fully functional: {all_functions_work}")
print("\nColorlog testing completed!")
```
