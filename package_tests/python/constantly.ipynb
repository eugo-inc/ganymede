```python
# Example 1: NamedConstant — symbolic, value-less constants on a container class.
from constantly import NamedConstant, Names

class Status(Names):
    OK = NamedConstant()
    WARN = NamedConstant()
    ERROR = NamedConstant()

print(f"Status.OK = {Status.OK!r}")
print(f"Status.OK.name = {Status.OK.name}")
print(f"Status.iterconstants() => {list(Status.iterconstants())}")

assert Status.OK is Status.OK
assert Status.OK is not Status.WARN
assert Status.OK.name == 'OK'
assert Status.lookupByName('ERROR') is Status.ERROR
```

```python
# Example 2: ValueConstant — constants that wrap an arbitrary user-supplied value.
from constantly import ValueConstant, Values

class HTTPStatus(Values):
    OK = ValueConstant(200)
    NOT_FOUND = ValueConstant(404)
    SERVER_ERROR = ValueConstant(500)

print(f"HTTPStatus.OK.value = {HTTPStatus.OK.value}")
print(f"HTTPStatus.NOT_FOUND.value = {HTTPStatus.NOT_FOUND.value}")

found = HTTPStatus.lookupByValue(404)
print(f"lookupByValue(404) => {found!r}")

assert HTTPStatus.OK.value == 200
assert found is HTTPStatus.NOT_FOUND
assert HTTPStatus.lookupByName('SERVER_ERROR') is HTTPStatus.SERVER_ERROR
```

```python
# Example 3: FlagConstant — bitwise-combinable flags.
from constantly import FlagConstant, Flags

class Permission(Flags):
    READ = FlagConstant()
    WRITE = FlagConstant()
    EXECUTE = FlagConstant()

rw = Permission.READ | Permission.WRITE
print(f"rw = {rw!r}")
print(f"READ in rw: {Permission.READ in rw}")
print(f"EXECUTE in rw: {Permission.EXECUTE in rw}")

assert Permission.READ in rw
assert Permission.WRITE in rw
assert Permission.EXECUTE not in rw
```

```python
# Example 4: Lookup, iteration, and identity invariants.
from constantly import NamedConstant, Names

class Color(Names):
    RED = NamedConstant()
    GREEN = NamedConstant()
    BLUE = NamedConstant()

names = [c.name for c in Color.iterconstants()]
print(f"Names: {names}")

# lookupByName retrieves the same singleton each call.
a = Color.lookupByName('RED')
b = Color.lookupByName('RED')
assert a is b is Color.RED

# Unknown name raises ValueError.
try:
    Color.lookupByName('PURPLE')
    raised = False
except ValueError:
    raised = True
assert raised, "expected ValueError for unknown name"
print("unknown-name lookup raised ValueError as expected")
```

```python
# Example 5: FlagConstant bitwise operators — AND, XOR, NOT.
from constantly import FlagConstant, Flags

class Feature(Flags):
    A = FlagConstant()
    B = FlagConstant()
    C = FlagConstant()

ab = Feature.A | Feature.B
bc = Feature.B | Feature.C

intersection = ab & bc
xor = ab ^ bc

print(f"ab & bc = {intersection!r}")
print(f"ab ^ bc = {xor!r}")

assert Feature.B in intersection
assert Feature.A not in intersection
assert Feature.A in xor and Feature.C in xor and Feature.B not in xor
print("All FlagConstant bitwise invariants hold")
```
