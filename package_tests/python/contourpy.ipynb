## Example 1: Basic Contour Generation and Algorithms

```python
import contourpy
import numpy as np

print("Example 1: Basic Contour Generation and Algorithms")

try:
    print("✓ contourpy package information:")
    print(f"  Module type: {type(contourpy)}")
    
    # Check version
    if hasattr(contourpy, '__version__'):
        print(f"  Version: {contourpy.__version__}")
    else:
        print("  Version: Available via package metadata")
    
    # Show main functions and capabilities
    attrs = [attr for attr in dir(contourpy) if not attr.startswith('_')]
    print(f"  Public attributes: {len(attrs)}")
    
    # Show key functions
    key_functions = ['contour_generator', 'LineType', 'FillType']
    available_functions = [f for f in key_functions if hasattr(contourpy, f)]
    print(f"  Key components: {available_functions}")
    
    # Create sample 2D data for contour generation
    print("✓ Sample data generation:")
    x = np.linspace(-3, 3, 20)
    y = np.linspace(-3, 3, 20) 
    X, Y = np.meshgrid(x, y)
    Z = np.exp(-(X**2 + Y**2))  # 2D Gaussian function
    
    print(f"  Grid dimensions: {Z.shape}")
    print(f"  Value range: {Z.min():.3f} to {Z.max():.3f}")
    
    # Demonstrate contour generator creation
    if hasattr(contourpy, 'contour_generator'):
        print("✓ Contour generator creation:")
        
        # Create basic contour generator
        cont_gen = contourpy.contour_generator(X, Y, Z)
        print(f"  Generator type: {type(cont_gen)}")
        
        # Test contour line generation
        test_level = 0.5
        lines = cont_gen.lines(test_level)
        print(f"  Contour lines at level {test_level}: {len(lines)} lines")
        
        if lines:
            first_line = lines[0]
            print(f"  First line points: {len(first_line)} coordinates")
            print(f"  Line shape: {first_line.shape}")
            
    # Show available algorithms
    print("✓ Available contour algorithms:")
    algorithms = ['mpl2005', 'mpl2014', 'serial', 'threaded']
    algorithm_info = [
        "mpl2005: Original matplotlib algorithm",
        "mpl2014: Updated matplotlib algorithm", 
        "serial: Single-threaded implementation",
        "threaded: Multi-threaded for performance"
    ]
    
    for info in algorithm_info:
        print(f"  • {info}")
    
    print("✓ Basic contour generation and algorithms completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"contourpy example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Contour Lines and Level Sets

```python
import contourpy
import numpy as np

print("Example 2: Contour Lines and Level Sets")

try:
    # Create more complex 2D data
    print("✓ Complex 2D function for contour analysis:")
    
    x = np.linspace(-2, 2, 25)
    y = np.linspace(-2, 2, 25)
    X, Y = np.meshgrid(x, y)
    
    # Create a more interesting function (combination of Gaussians)
    Z1 = np.exp(-((X-0.5)**2 + (Y-0.5)**2))
    Z2 = 0.7 * np.exp(-((X+0.5)**2 + (Y+0.5)**2))  
    Z = Z1 + Z2
    
    print(f"  Combined function range: {Z.min():.3f} to {Z.max():.3f}")
    print(f"  Grid resolution: {Z.shape}")
    
    # Demonstrate contour line generation at multiple levels
    if hasattr(contourpy, 'contour_generator'):
        print("✓ Multi-level contour line generation:")
        
        cont_gen = contourpy.contour_generator(X, Y, Z)
        
        # Generate contours at different levels
        levels = [0.1, 0.3, 0.5, 0.7, 0.9]
        
        for level in levels:
            lines = cont_gen.lines(level)
            print(f"  Level {level}: {len(lines)} contour lines")
            
            # Analyze line segments
            total_points = sum(len(line) for line in lines)
            if lines:
                avg_points = total_points / len(lines)
                print(f"    Total points: {total_points}, Avg per line: {avg_points:.1f}")
                
        # Demonstrate automatic level selection
        print("✓ Automatic level generation:")
        n_levels = 8
        auto_levels = np.linspace(Z.min(), Z.max(), n_levels)[1:-1]  # Exclude extremes
        
        print(f"  Generated {len(auto_levels)} automatic levels:")
        for i, level in enumerate(auto_levels):
            print(f"    Level {i+1}: {level:.3f}")
            
        # Show line properties and analysis
        print("✓ Contour line analysis:")
        
        # Analyze a specific level
        analysis_level = 0.4
        lines = cont_gen.lines(analysis_level)
        
        if lines:
            print(f"  Analysis for level {analysis_level}:")
            print(f"    Number of separate contour lines: {len(lines)}")
            
            for i, line in enumerate(lines[:3]):  # Show first 3 lines
                x_coords, y_coords = line[:, 0], line[:, 1]
                x_range = x_coords.max() - x_coords.min()
                y_range = y_coords.max() - y_coords.min()
                
                print(f"    Line {i+1}: {len(line)} points")
                print(f"      X range: {x_range:.3f}, Y range: {y_range:.3f}")
                
                # Check if line is closed (first and last points are close)
                if len(line) > 2:
                    distance = np.sqrt((line[0,0] - line[-1,0])**2 + 
                                     (line[0,1] - line[-1,1])**2)
                    is_closed = distance < 0.01  # Small threshold
                    print(f"      Closed contour: {is_closed}")
    
    print("✓ Contour lines and level sets completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"contourpy example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: Filled Contours and Polygons

```python
import contourpy
import numpy as np

print("Example 3: Filled Contours and Polygons")

try:
    # Create sample data for filled contour demonstration
    print("✓ Data preparation for filled contours:")
    
    x = np.linspace(-2, 2, 20)
    y = np.linspace(-2, 2, 20)
    X, Y = np.meshgrid(x, y)
    
    # Create a function with interesting topography
    Z = np.sin(X) * np.cos(Y) + 0.3 * np.sin(3*X) * np.sin(3*Y)
    
    print(f"  Function range: {Z.min():.3f} to {Z.max():.3f}")
    print(f"  Grid shape: {Z.shape}")
    
    # Demonstrate filled contour generation
    if hasattr(contourpy, 'contour_generator'):
        print("✓ Filled contour generation:")
        
        cont_gen = contourpy.contour_generator(X, Y, Z)
        
        # Generate filled contours between level pairs
        level_pairs = [
            (-1.0, -0.5),
            (-0.5, 0.0), 
            (0.0, 0.5),
            (0.5, 1.0)
        ]
        
        for i, (lower, upper) in enumerate(level_pairs):
            try:
                filled_polys = cont_gen.filled(lower, upper)
                print(f"  Region {i+1} ({lower} to {upper}): {len(filled_polys)} polygons")
                
                # Analyze polygon complexity
                if filled_polys:
                    total_vertices = sum(len(poly) for poly in filled_polys)
                    avg_vertices = total_vertices / len(filled_polys)
                    print(f"    Total vertices: {total_vertices}, Avg per polygon: {avg_vertices:.1f}")
                    
            except Exception as e:
                print(f"    Error generating filled contours for {lower}-{upper}: {e}")
                raise
                
        # Show single-level filled contours  
        print("✓ Single-level filled contour analysis:")
        
        single_levels = [-0.3, 0.0, 0.3]
        
        for level in single_levels:
            try:
                # Get filled regions above this level
                filled_above = cont_gen.filled(level, Z.max())
                # Get filled regions below this level  
                filled_below = cont_gen.filled(Z.min(), level)
                
                print(f"  Level {level}:")
                print(f"    Regions above: {len(filled_above)} polygons")
                print(f"    Regions below: {len(filled_below)} polygons")
                
            except Exception as e:
                print(f"    Error for level {level}: {e}")
                raise
                
    # Show fill type options
    print("✓ Fill type options:")
    
    if hasattr(contourpy, 'FillType'):
        print("  Available fill types for different output formats:")
        fill_types = [
            "OuterCode: Outer boundary with hole indicators",
            "OuterOffset: Outer boundary with offset arrays",
            "ChunkCombinedCode: Combined chunks with codes",
            "ChunkCombinedOffset: Combined chunks with offsets"
        ]
        
        for fill_type in fill_types:
            print(f"    • {fill_type}")
    else:
        print("  Fill types: Various polygon formats available")
        
    # Show polygon properties
    print("✓ Polygon properties:")
    properties = [
        "Vertices: Corner points defining polygon boundary",
        "Holes: Interior regions excluded from fill",
        "Orientation: Clockwise vs counter-clockwise winding",
        "Nesting: Polygons can contain other polygons",
        "Codes: Path codes for complex polygon rendering"
    ]
    
    for prop in properties:
        print(f"  • {prop}")
    
    print("✓ Filled contours and polygons completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"contourpy example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```
