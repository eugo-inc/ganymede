## Example 2: Custom Name Patterns and Word Categories

```python
import coolname

print("Example 2: Custom Name Patterns and Word Categories")

try:
    # Explore different name generation patterns
    print("✓ Name generation patterns:")
    
    if hasattr(coolname, 'generate'):
        # Generate names with different approaches
        print("  Various generated names:")
        
        # Generate multiple names to show variety
        names = [coolname.generate() for _ in range(8)]
        
        for i, name in enumerate(names, 1):
            name_str = ' '.join(name) if isinstance(name, list) else str(name)
            print(f"    {i}. {name_str}")
            
        # Analyze word patterns
        print("✓ Word pattern analysis:")
        
        sample_names = [coolname.generate() for _ in range(3)]
        for i, name in enumerate(sample_names, 1):
            print(f"  Name {i}: {name}")
            if len(name) >= 3:
                print(f"    First word (adj): '{name[0]}'")
                print(f"    Second word (adj): '{name[1]}'") 
                print(f"    Third word (noun): '{name[2]}'")
                if len(name) > 3:
                    print(f"    Remaining: {name[3:]}")
                    
    # Show slug variations and formats
    print("✓ Slug format variations:")
    
    if hasattr(coolname, 'generate_slug'):
        # Generate multiple slugs to show format consistency
        slugs = [coolname.generate_slug() for _ in range(6)]
        
        print("  Generated slugs:")
        for i, slug in enumerate(slugs, 1):
            word_count = len(slug.split('-'))
            char_count = len(slug)
            print(f"    {i}. {slug} ({word_count} words, {char_count} chars)")
            
        # Analyze slug characteristics
        print("✓ Slug characteristics:")
        characteristics = [
            "Lowercase format for URL compatibility",
            "Hyphen-separated words for readability",
            "No special characters or spaces",
            "Pronounceable and memorable combinations",
            "Consistent word order and structure"
        ]
        
        for char in characteristics:
            print(f"    • {char}")
            
    # Show word category examples
    print("✓ Word category examples:")
    
    # Extract individual words from generated names to show categories
    all_words = []
    sample_names = [coolname.generate() for _ in range(5)]
    
    for name in sample_names:
        all_words.extend(name)
        
    # Categorize words by position (rough categorization)
    adjectives = [name[0] for name in sample_names if len(name) > 0]
    second_words = [name[1] for name in sample_names if len(name) > 1]
    nouns = [name[2] for name in sample_names if len(name) > 2]
    
    print(f"  First position (adjectives): {adjectives}")
    print(f"  Second position (adjectives): {second_words}")
    print(f"  Third position (nouns): {nouns}")
    
    # Show unique word count
    unique_words = set(all_words)
    print(f"  Unique words in sample: {len(unique_words)}")
    print(f"  Total words generated: {len(all_words)}")
    
    print("✓ Custom name patterns and word categories completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"coolname example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: Randomization and Reproducibility

```python
import coolname
import random

print("Example 3: Randomization and Reproducibility")

try:
    # Demonstrate random name generation properties
    print("✓ Random generation properties:")
    
    # Generate names to check for randomness
    if hasattr(coolname, 'generate'):
        # Generate a batch of names
        batch_size = 10
        name_batch = [coolname.generate() for _ in range(batch_size)]
        
        print(f"  Generated {batch_size} names:")
        for i, name in enumerate(name_batch, 1):
            name_str = ' '.join(name) if isinstance(name, list) else str(name)
            print(f"    {i:2}. {name_str}")
            
        # Check for uniqueness
        name_strings = [' '.join(name) for name in name_batch]
        unique_names = set(name_strings)
        print(f"  Uniqueness: {len(unique_names)}/{len(name_strings)} unique names")
        
        # Check for duplicates
        if len(unique_names) < len(name_strings):
            print("  Found duplicates (rare but possible)")
        else:
            print("  All names unique (expected with large word space)")
            
    # Show slug randomization
    print("✓ Slug randomization:")
    
    if hasattr(coolname, 'generate_slug'):
        slug_batch = [coolname.generate_slug() for _ in range(8)]
        
        print("  Generated slugs:")
        for i, slug in enumerate(slug_batch, 1):
            print(f"    {i}. {slug}")
            
        # Analyze slug patterns
        word_counts = [len(slug.split('-')) for slug in slug_batch]
        char_counts = [len(slug) for slug in slug_batch]
        
        print(f"  Word count range: {min(word_counts)}-{max(word_counts)}")
        print(f"  Character count range: {min(char_counts)}-{max(char_counts)}")
        
    # Demonstrate reproducibility considerations
    print("✓ Reproducibility and seeding:")
    
    # Check if coolname has seeding capabilities
    if hasattr(coolname, 'replace_random'):
        print("  Seeding capabilities available via replace_random()")
        
        # Try to demonstrate seeded generation
        try:
            # Set seed for Python's random module
            random.seed(42)
            
            # Generate names with seeded randomness
            seeded_names = [coolname.generate() for _ in range(3)]
            print("  Seeded generation attempt:")
            for i, name in enumerate(seeded_names, 1):
                name_str = ' '.join(name) if isinstance(name, list) else str(name)
                print(f"    {i}. {name_str}")
                
        except Exception as e:
            print(f"  Seeding note: {e}")
            raise
            
    # Show combination space analysis
    print("✓ Combination space analysis:")
    
    if hasattr(coolname, 'get_combinations_count'):
        try:
            combo_count = coolname.get_combinations_count()
            print(f"  Total combinations available: {combo_count:,}")
            print(f"  Collision probability very low with {combo_count} possibilities")
            
        except Exception as e:
            print(f"  Combination analysis: Function available but may need parameters")
            raise
    else:
        print("  Large combination space ensures low collision probability")
        
    # Show randomness distribution
    print("✓ Randomness distribution:")
    
    # Sample first words to check distribution
    if hasattr(coolname, 'generate'):
        first_words = [coolname.generate()[0] for _ in range(20)]
        word_freq = {}
        
        for word in first_words:
            word_freq[word] = word_freq.get(word, 0) + 1
            
        print("  First word frequency in 20 samples:")
        for word, freq in sorted(word_freq.items()):
            print(f"    '{word}': {freq} times")
            
        print(f"  Unique first words: {len(word_freq)}/20")
    
    print("✓ Randomization and reproducibility completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"coolname example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: Practical Use Cases and Applications

```python
import coolname
import uuid

print("Example 4: Practical Use Cases and Applications")

try:
    # Demonstrate practical applications
    print("✓ Practical application scenarios:")
    
    # Use case 1: User-friendly identifiers
    print("  Use case 1: User-friendly resource identifiers")
    
    if hasattr(coolname, 'generate_slug'):
        # Generate identifiers for different resources
        resource_types = ['server', 'database', 'project', 'container', 'backup']
        
        for resource in resource_types:
            identifier = coolname.generate_slug()
            print(f"    {resource}-{identifier}")
            
    # Use case 2: Temporary names and handles
    print("  Use case 2: Temporary names and session handles")
    
    if hasattr(coolname, 'generate'):
        # Generate readable session identifiers
        session_names = []
        for i in range(4):
            name = coolname.generate()
            session_handle = '-'.join(name[:3])  # Use first 3 words
            session_names.append(session_handle)
            print(f"    Session {i+1}: {session_handle}")
            
    # Use case 3: Default names for user content
    print("  Use case 3: Default names for user-generated content")
    
    content_types = ['document', 'playlist', 'folder', 'bookmark-collection']
    
    for content_type in content_types:
        if hasattr(coolname, 'generate'):
            name_words = coolname.generate()
            readable_name = ' '.join(name_words[:3]).title()
            print(f"    {content_type}: \"{readable_name}\"")
            
    # Use case 4: Test data generation
    print("  Use case 4: Test data and mock object names")
    
    if hasattr(coolname, 'generate_slug'):
        # Generate test user names
        test_users = []
        for i in range(5):
            username = coolname.generate_slug()
            # Limit length for practical usernames
            if len(username) > 30:
                words = username.split('-')
                username = '-'.join(words[:3])  # Take first 3 words
            test_users.append(username)
            print(f"    test_user_{i+1}: {username}")
            
    # Show comparison with other identifier methods
    print("✓ Comparison with other identifier methods:")
    
    # Compare coolname with UUID
    comparison_data = []
    
    for i in range(3):
        # Generate different types of identifiers
        if hasattr(coolname, 'generate_slug'):
            cool_slug = coolname.generate_slug()
        else:
            cool_slug = "unavailable"
            
        uuid_hex = str(uuid.uuid4())
        uuid_short = str(uuid.uuid4()).split('-')[0]
        
        comparison_data.append({
            'coolname': cool_slug,
            'uuid_full': uuid_hex,
            'uuid_short': uuid_short
        })
        
    print("  Identifier comparison:")
    for i, data in enumerate(comparison_data, 1):
        print(f"    Set {i}:")
        print(f"      Coolname:   {data['coolname']} (human-readable)")
        print(f"      UUID full:  {data['uuid_full']} (machine-optimized)")
        print(f"      UUID short: {data['uuid_short']} (compact)")
        
    # Show advantages of coolname approach
    print("✓ Advantages of coolname identifiers:")
    
    advantages = [
        "Human-readable and memorable",
        "Pronounceable for verbal communication", 
        "Culturally neutral word combinations",
        "Lower chance of offensive combinations",
        "Aesthetically pleasing for user interfaces",
        "Natural sorting behavior",
        "Good balance of uniqueness and readability"
    ]
    
    for advantage in advantages:
        print(f"    • {advantage}")
        
    # Show integration patterns
    print("✓ Integration patterns:")
    
    integration_examples = [
        "Web URLs: /project/{coolname-slug}/dashboard",
        "Database keys: user_{coolname}_profile", 
        "File names: backup_{coolname}_{timestamp}.sql",
        "Container names: app-{coolname}-worker",
        "API endpoints: /api/v1/resources/{coolname}"
    ]
    
    for example in integration_examples:
        print(f"    • {example}")
    
    print("✓ Practical use cases and applications completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"coolname example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```
