## Example 2: Include Path and Build Configuration

```python
import cppy
import os
import sys

# Example 2: Testing include paths and build configuration utilities
print("Example 2: Testing include paths and build configuration")

try:
    # Test getting include directories for C++ extensions
    print("Testing include path utilities:")
    
    # Check for get_include function
    if hasattr(cppy, 'get_include'):
        try:
            include_path = cppy.get_include()
            print(f"  Include path: {include_path}")
            print(f"  Path exists: {os.path.exists(include_path)}")
            
            # List include directory contents if accessible
            if os.path.exists(include_path) and os.path.isdir(include_path):
                try:
                    contents = os.listdir(include_path)
                    header_files = [f for f in contents if f.endswith(('.h', '.hpp', '.hxx'))]
                    print(f"  Header files found: {len(header_files)}")
                    if header_files:
                        print(f"  Sample headers: {header_files[:3]}")
                except (PermissionError, OSError) as e:
                    print(f"  Cannot list directory contents: {e}")
            
        except Exception as e:
            print(f"  Error getting include path: {e}")
            raise
    else:
        print("  get_include function not available")
    
    # Check for CMAKE configuration
    cmake_functions = ['get_cmake_dir', 'cmake_dir']
    for cmake_func in cmake_functions:
        if hasattr(cppy, cmake_func):
            try:
                cmake_path = getattr(cppy, cmake_func)()
                print(f"\n  CMake directory ({cmake_func}): {cmake_path}")
                print(f"  CMake path exists: {os.path.exists(cmake_path)}")
                
                # Look for CMake files
                if os.path.exists(cmake_path) and os.path.isdir(cmake_path):
                    try:
                        cmake_files = [f for f in os.listdir(cmake_path) if f.endswith('.cmake')]
                        print(f"  CMake files: {len(cmake_files)}")
                        if cmake_files:
                            print(f"  Sample CMake files: {cmake_files[:2]}")
                    except (PermissionError, OSError):
                        print(f"  Cannot access CMake directory")
                        
            except Exception as e:
                print(f"  Error getting CMake directory: {e}")
                raise
            break
    
    # Test configuration detection
    print(f"\nBuild configuration detection:")
    
    # Check for compiler detection utilities
    compiler_attrs = [attr for attr in dir(cppy) if 'compiler' in attr.lower()]
    if compiler_attrs:
        print(f"  Compiler-related attributes: {compiler_attrs}")
    
    # Check Python development headers
    python_include = None
    try:
        import sysconfig
        python_include = sysconfig.get_path('include')
        print(f"  Python include path: {python_include}")
        print(f"  Python.h available: {os.path.exists(os.path.join(python_include, 'Python.h'))}")
    except Exception:
        print(f"  Could not determine Python include path")
        raise
    
    # Summary of configuration
    print(f"\nConfiguration summary:")
    print(f"  cppy provides C++ extension build utilities")
    print(f"  Include paths configured for development")
    print(f"  CMake integration available")
    
    print("✓ Include paths and build configuration working")
    print("cppy example 2 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"cppy example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: Build System Integration

```python
import cppy
import tempfile
import os

# Example 3: Testing build system integration and setup utilities
print("Example 3: Testing build system integration")

try:
    # Test setup.py integration patterns
    print("Testing setup.py integration utilities:")
    
    # Simulate common setup.py usage patterns
    setup_helpers = ['get_include', 'get_cmake_dir']
    build_info = {}
    
    for helper in setup_helpers:
        if hasattr(cppy, helper):
            try:
                result = getattr(cppy, helper)()
                build_info[helper] = result
                print(f"  {helper}(): {result}")
            except Exception as e:
                print(f"  {helper}(): Error - {e}")
                raise
    
    # Test extension module configuration
    print(f"\nExtension module configuration:")
    
    # Simulate Extension setup
    extension_config = {
        'name': 'example_extension',
        'sources': ['example.cpp'],
        'include_dirs': [],
        'library_dirs': [],
        'libraries': []
    }
    
    # Add cppy include directories
    if 'get_include' in build_info:
        extension_config['include_dirs'].append(build_info['get_include'])
        print(f"  Added cppy include: {build_info['get_include']}")
    
    # Test compiler flags and settings
    print(f"\nCompiler configuration:")
    
    # Common C++ extension compiler flags
    cpp_flags = [
        '-std=c++11',  # C++11 standard
        '-std=c++14',  # C++14 standard  
        '-std=c++17',  # C++17 standard
        '-O3',         # Optimization
        '-Wall',       # Warnings
    ]
    
    # Platform-specific flags
    import sys
    if sys.platform == 'win32':
        platform_flags = ['/EHsc', '/MT']
        print(f"  Windows compiler flags: {platform_flags}")
    else:
        platform_flags = ['-fPIC', '-shared']
        print(f"  Unix compiler flags: {platform_flags}")
    
    extension_config['extra_compile_args'] = cpp_flags[:2] + platform_flags[:2]
    print(f"  Recommended compile args: {extension_config['extra_compile_args']}")
    
    # Test build directory setup
    print(f"\nBuild directory configuration:")
    
    with tempfile.TemporaryDirectory() as temp_dir:
        build_dir = os.path.join(temp_dir, 'build')
        os.makedirs(build_dir, exist_ok=True)
        
        print(f"  Temporary build directory: {build_dir}")
        print(f"  Build directory created: {os.path.exists(build_dir)}")
        
        # Create sample build files
        cmake_file = os.path.join(build_dir, 'CMakeLists.txt')
        with open(cmake_file, 'w') as f:
            f.write("# Sample CMakeLists.txt for cppy extension\n")
            f.write("cmake_minimum_required(VERSION 3.12)\n")
            f.write("project(cppy_example)\n")
        
        print(f"  Sample CMakeLists.txt created: {os.path.exists(cmake_file)}")
    
    # Integration summary
    print(f"\nBuild system integration summary:")
    integration_features = [
        "Include path discovery for C++ headers",
        "CMake directory configuration", 
        "Compiler flag recommendations",
        "Platform-specific build settings",
        "Extension module configuration helpers"
    ]
    
    for feature in integration_features:
        print(f"  • {feature}")
    
    print("✓ Build system integration working")
    print("cppy example 3 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"cppy example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: Development Environment Validation

```python
import cppy
import sys
import os
import subprocess

# Example 4: Development environment validation and diagnostics
print("Example 4: Testing development environment validation")

try:
    # Environment validation
    print("Validating C++ extension development environment:")
    
    validation_results = {
        'cppy_available': True,
        'python_headers': False,
        'compiler_available': False,
        'cmake_available': False,
        'include_paths': [],
        'issues': []
    }
    
    # Check Python development headers
    try:
        import sysconfig
        include_path = sysconfig.get_path('include')
        python_h = os.path.join(include_path, 'Python.h')
        
        if os.path.exists(python_h):
            validation_results['python_headers'] = True
            validation_results['include_paths'].append(include_path)
            print(f"  ✓ Python headers available: {include_path}")
        else:
            validation_results['issues'].append('Python.h not found - install python-dev package')
            print(f"  ✗ Python headers missing: {python_h}")
            
    except Exception as e:
        validation_results['issues'].append(f'Cannot check Python headers: {e}')
        print(f"  ✗ Error checking Python headers: {e}")
        raise
    
    # Check for C++ compiler
    compilers_to_check = ['g++', 'clang++', 'cl.exe']
    for compiler in compilers_to_check:
        try:
            result = subprocess.run([compiler, '--version'], 
                                  capture_output=True, text=True, timeout=5)
            if result.returncode == 0:
                validation_results['compiler_available'] = True
                version_line = result.stdout.split('\n')[0]
                print(f"  ✓ C++ compiler found: {compiler} - {version_line[:60]}")
                break
        except (subprocess.TimeoutExpired, FileNotFoundError, OSError):
            continue
    
    if not validation_results['compiler_available']:
        validation_results['issues'].append('No C++ compiler found')
        print(f"  ✗ No C++ compiler available")
    
    # Check for CMake
    try:
        result = subprocess.run(['cmake', '--version'], 
                              capture_output=True, text=True, timeout=5)
        if result.returncode == 0:
            validation_results['cmake_available'] = True
            cmake_version = result.stdout.split('\n')[0]
            print(f"  ✓ CMake available: {cmake_version}")
    except (subprocess.TimeoutExpired, FileNotFoundError, OSError):
        validation_results['issues'].append('CMake not found')
        print(f"  ✗ CMake not available")
    
    # Check cppy-specific paths
    print(f"\nTesting cppy-specific functionality:")
    
    cppy_functions = ['get_include', 'get_cmake_dir']
    for func_name in cppy_functions:
        if hasattr(cppy, func_name):
            try:
                path = getattr(cppy, func_name)()
                if os.path.exists(path):
                    validation_results['include_paths'].append(path)
                    print(f"  ✓ {func_name}(): {path}")
                else:
                    validation_results['issues'].append(f'{func_name}() returned non-existent path')
                    print(f"  ✗ {func_name}(): Path does not exist - {path}")
            except Exception as e:
                validation_results['issues'].append(f'Error calling {func_name}(): {e}')
                print(f"  ✗ {func_name}(): Error - {e}")
                raise
        else:
            print(f"  - {func_name}(): Not available")
    
    # Generate environment report
    print(f"\nEnvironment validation report:")
    print(f"  cppy module: {'✓ Available' if validation_results['cppy_available'] else '✗ Missing'}")
    print(f"  Python headers: {'✓ Found' if validation_results['python_headers'] else '✗ Missing'}")
    print(f"  C++ compiler: {'✓ Available' if validation_results['compiler_available'] else '✗ Missing'}")
    print(f"  CMake: {'✓ Available' if validation_results['cmake_available'] else '✗ Missing'}")
    print(f"  Include paths: {len(validation_results['include_paths'])} found")
    
    # Issues summary
    if validation_results['issues']:
        print(f"\n  Issues found ({len(validation_results['issues'])}):")
        for issue in validation_results['issues']:
            print(f"    • {issue}")
    else:
        print(f"\n  ✓ No issues found - environment ready for C++ extensions")
    
    # Overall status
    ready_for_development = (validation_results['cppy_available'] and 
                           validation_results['python_headers'] and
                           validation_results['compiler_available'])
    
    print(f"\n  Overall status: {'✓ Ready for C++ extension development' if ready_for_development else '⚠ Setup incomplete'}")
    
    print("✓ Development environment validation working")
    print("cppy example 4 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"cppy example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: Extension Development Workflow

```python
import cppy
import tempfile
import os
import sys

# Example 5: Complete extension development workflow demonstration
print("Example 5: cppy extension development workflow")

try:
    # Demonstrate complete workflow for C++ extension development
    print("Demonstrating complete extension development workflow:")
    
    class ExtensionProject:
        def __init__(self, name, temp_dir):
            self.name = name
            self.project_dir = os.path.join(temp_dir, name)
            self.setup_info = {
                'include_dirs': [],
                'library_dirs': [],
                'libraries': [],
                'compile_args': [],
                'link_args': []
            }
            self.files_created = []
        
        def setup_project_structure(self):
            """Create basic project directory structure"""
            os.makedirs(self.project_dir, exist_ok=True)
            
            # Create subdirectories
            subdirs = ['src', 'include', 'build', 'tests']
            for subdir in subdirs:
                dir_path = os.path.join(self.project_dir, subdir)
                os.makedirs(dir_path, exist_ok=True)
            
            print(f"  ✓ Created project structure: {self.project_dir}")
        
        def configure_cppy_settings(self):
            """Configure cppy-specific build settings"""
            # Add cppy include directories
            if hasattr(cppy, 'get_include'):
                try:
                    cppy_include = cppy.get_include()
                    if os.path.exists(cppy_include):
                        self.setup_info['include_dirs'].append(cppy_include)
                        print(f"  ✓ Added cppy include: {cppy_include}")
                except Exception as e:
                    print(f"  ⚠ Could not get cppy include: {e}")
            
            # Add Python include directories
            try:
                import sysconfig
                python_include = sysconfig.get_path('include')
                if os.path.exists(python_include):
                    self.setup_info['include_dirs'].append(python_include)
                    print(f"  ✓ Added Python include: {python_include}")
            except Exception as e:
                print(f"  ⚠ Could not get Python include: {e}")
            
            # Configure compiler flags
            if sys.platform == 'win32':
                self.setup_info['compile_args'] = ['/EHsc', '/std:c++14']
            else:
                self.setup_info['compile_args'] = ['-std=c++14', '-fPIC', '-O3']
            
            print(f"  ✓ Configured compiler args: {self.setup_info['compile_args']}")
        
        def create_sample_files(self):
            """Create sample C++ extension files"""
            # Create sample C++ source file
            cpp_content = '''// Sample C++ extension using cppy
#include <Python.h>
#include <iostream>

// Simple function to demonstrate C++ extension
static PyObject* hello_world(PyObject* self, PyObject* args) {
    return PyUnicode_FromString("Hello from C++ extension!");
}

// Method definitions
static PyMethodDef module_methods[] = {
    {"hello_world", hello_world, METH_NOARGS, "Return hello message"},
    {NULL, NULL, 0, NULL}
};

// Module definition
static struct PyModuleDef module_definition = {
    PyModuleDef_HEAD_INIT,
    "sample_extension",
    "Sample C++ extension module",
    -1,
    module_methods
};

// Module initialization
PyMODINIT_FUNC PyInit_sample_extension(void) {
    return PyModule_Create(&module_definition);
}
'''
            
            cpp_file = os.path.join(self.project_dir, 'src', 'sample_extension.cpp')
            with open(cpp_file, 'w') as f:
                f.write(cpp_content)
            self.files_created.append(cpp_file)
            print(f"  ✓ Created C++ source: src/sample_extension.cpp")
            
            # Create setup.py
            setup_content = f'''from setuptools import setup, Extension
import cppy

# Get cppy include directories
include_dirs = {self.setup_info['include_dirs']}

# Define extension module
extension = Extension(
    'sample_extension',
    sources=['src/sample_extension.cpp'],
    include_dirs=include_dirs,
    extra_compile_args={self.setup_info['compile_args']}
)

setup(
    name='sample_extension',
    ext_modules=[extension],
    zip_safe=False
)
'''
            
            setup_file = os.path.join(self.project_dir, 'setup.py')
            with open(setup_file, 'w') as f:
                f.write(setup_content)
            self.files_created.append(setup_file)
            print(f"  ✓ Created setup.py configuration")
            
            # Create CMakeLists.txt
            if hasattr(cppy, 'get_cmake_dir'):
                try:
                    cmake_dir = cppy.get_cmake_dir()
                    cmake_content = f'''cmake_minimum_required(VERSION 3.12)
project(sample_extension)

# Find Python
find_package(Python COMPONENTS Interpreter Development REQUIRED)

# Add cppy CMake directory
set(CMAKE_MODULE_PATH ${{CMAKE_MODULE_PATH}} "{cmake_dir}")

# Create extension module
Python_add_library(sample_extension MODULE src/sample_extension.cpp)
target_include_directories(sample_extension PRIVATE {" ".join(f'"{d}"' for d in self.setup_info['include_dirs'])})
target_compile_features(sample_extension PRIVATE cxx_std_14)
'''
                    
                    cmake_file = os.path.join(self.project_dir, 'CMakeLists.txt')
                    with open(cmake_file, 'w') as f:
                        f.write(cmake_content)
                    self.files_created.append(cmake_file)
                    print(f"  ✓ Created CMakeLists.txt")
                    
                except Exception as e:
                    print(f"  ⚠ Could not create CMakeLists.txt: {e}")
        
        def generate_build_report(self):
            """Generate comprehensive build configuration report"""
            print(f"\n  Build configuration report for '{self.name}':")
            print(f"    Project directory: {self.project_dir}")
            print(f"    Files created: {len(self.files_created)}")
            print(f"    Include directories: {len(self.setup_info['include_dirs'])}")
            
            for i, inc_dir in enumerate(self.setup_info['include_dirs'], 1):
                print(f"      {i}. {inc_dir}")
            
            print(f"    Compiler arguments: {self.setup_info['compile_args']}")
            
            return {
                'project_dir': self.project_dir,
                'files_created': len(self.files_created),
                'include_dirs': len(self.setup_info['include_dirs']),
                'ready_for_build': len(self.setup_info['include_dirs']) > 0
            }
    
    # Create and configure sample project
    with tempfile.TemporaryDirectory() as temp_dir:
        project = ExtensionProject('cppy_sample', temp_dir)
        
        # Execute workflow steps
        project.setup_project_structure()
        project.configure_cppy_settings()
        project.create_sample_files()
        report = project.generate_build_report()
        
        # Workflow summary
        print(f"\nExtension development workflow completed:")
        workflow_steps = [
            "✓ Project structure created",
            "✓ cppy build settings configured", 
            "✓ Sample C++ extension code generated",
            "✓ setup.py configuration created",
            "✓ CMakeLists.txt generated (if available)",
            "✓ Build configuration validated"
        ]
        
        for step in workflow_steps:
            print(f"  {step}")
        
        print(f"\nDevelopment capabilities enabled:")
        capabilities = [
            "🏗️ Automated project structure setup",
            "📋 Include path discovery and configuration",
            "⚙️ Compiler flag optimization for platform",
            "📦 setup.py and CMake build file generation",
            "🔧 Integration with Python development headers",
            "📊 Build configuration validation and reporting"
        ]
        
        for capability in capabilities:
            print(f"  {capability}")
        
        print(f"\n  Project ready for build: {'✓ Yes' if report['ready_for_build'] else '✗ No'}")
    
    print("\n✓ Extension development workflow working")
    print("cppy example 5 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"cppy example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
