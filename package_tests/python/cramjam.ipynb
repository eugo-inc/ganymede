## Example 1: Basic Snappy compression and decompression

```python
# Example 1: Basic Snappy compression and decompression
import cramjam

print("Example 1: Basic Snappy Compression/Decompression\n")

# Original data
original_data = b"Hello, this is a test message for cramjam Snappy compression! " * 10
print(f"Original data size: {len(original_data)} bytes")
print(f"Original data preview: {original_data[:50]}...\n")

# Compress using Snappy
compressed = cramjam.snappy.compress(original_data)
print(f"Compressed size: {len(compressed)} bytes")
print(f"Compression ratio: {len(original_data) / len(compressed):.2f}x")
print(f"Space saved: {(1 - len(compressed) / len(original_data)) * 100:.1f}%\n")

# Decompress
decompressed = cramjam.snappy.decompress(compressed)
print(f"Decompressed size: {len(decompressed)} bytes")
print(f"Data matches original: {bytes(decompressed) == original_data}")
print(f"Decompressed preview: {bytes(decompressed)[:50]}...")

print("\nSnappy provides fast compression with moderate compression ratios")
```

## Example 2: Comparing multiple compression algorithms

```python
# Example 2: Comparing multiple compression algorithms
import cramjam

print("Example 2: Comparing Compression Algorithms\n")

# Test data with repetitive patterns (compresses well)
test_data = b"The quick brown fox jumps over the lazy dog. " * 50
print(f"Original data: {len(test_data)} bytes\n")

# Test different algorithms
algorithms = [
    ('Snappy', cramjam.snappy),
    ('Gzip', cramjam.gzip),
    ('Brotli', cramjam.brotli),
    ('LZ4', cramjam.lz4),
    ('Zstd', cramjam.zstd),
]

print("Compression comparison:")
print(f"{'Algorithm':<12} {'Size (bytes)':<15} {'Ratio':<10} {'Space Saved'}")
print("-" * 55)

results = []
for name, algo in algorithms:
    compressed = algo.compress(test_data)
    ratio = len(test_data) / len(compressed)
    saved = (1 - len(compressed) / len(test_data)) * 100
    results.append((name, len(compressed), ratio, saved))
    print(f"{name:<12} {len(compressed):<15} {ratio:<10.2f} {saved:.1f}%")

# Find best compression
best = min(results, key=lambda x: x[1])
print(f"\nBest compression: {best[0]} with {best[3]:.1f}% space saved")

print("\nDifferent algorithms offer tradeoffs between speed and compression")
```

## Example 3: Brotli compression levels

```python
# Example 3: Using Brotli with compression levels
import cramjam

print("Example 3: Brotli Compression Levels\n")

# Test data
data = b"Python compression library using Rust for fast performance. " * 20
print(f"Original size: {len(data)} bytes\n")

# Brotli supports compression levels 0-11
print("Brotli compression at different levels:")
print(f"{'Level':<8} {'Size (bytes)':<15} {'Ratio':<10} {'Space Saved'}")
print("-" * 50)

for level in [1, 4, 7, 11]:
    compressed = cramjam.brotli.compress(data, level=level)
    ratio = len(data) / len(compressed)
    saved = (1 - len(compressed) / len(data)) * 100
    print(f"{level:<8} {len(compressed):<15} {ratio:<10.2f} {saved:.1f}%")
    
    # Verify decompression works
    decompressed = cramjam.brotli.decompress(compressed)
    assert bytes(decompressed) == data, f"Decompression failed at level {level}"

print("\nHigher levels provide better compression but take more time")
print("Level 1: Fast, moderate compression")
print("Level 11: Slow, maximum compression")
```

## Example 4: Compressing binary data

```python
# Example 4: Compressing binary data (simulated file content)
import cramjam
import random

print("Example 4: Compressing Binary Data\n")

# Create different types of binary data
# 1. Highly repetitive data (compresses well)
repetitive_data = bytes([0xFF, 0x00] * 500)

# 2. Random data (compresses poorly)
random.seed(42)
random_data = bytes([random.randint(0, 255) for _ in range(1000)])

# 3. Text-like data (compresses moderately)
text_data = b"ABCDEFGHIJKLMNOPQRSTUVWXYZ" * 40

datasets = [
    ("Repetitive", repetitive_data),
    ("Random", random_data),
    ("Text-like", text_data)
]

print("Zstd compression on different data types:")
print(f"{'Type':<15} {'Original':<12} {'Compressed':<12} {'Ratio'}")
print("-" * 50)

for name, data in datasets:
    compressed = cramjam.zstd.compress(data)
    ratio = len(data) / len(compressed)
    print(f"{name:<15} {len(data):<12} {len(compressed):<12} {ratio:.2f}x")
    
    # Verify round-trip
    decompressed = cramjam.zstd.decompress(compressed)
    assert bytes(decompressed) == data

print("\nCompression effectiveness depends on data patterns")
print("Repetitive data: Excellent compression")
print("Random data: Poor compression (may even expand)")
print("Structured data: Good to moderate compression")
```

## Example 5: LZ4 fast compression

```python
# Example 5: Using LZ4 for fast compression (ideal for real-time data)
import cramjam
import time

print("Example 5: LZ4 Fast Compression Performance\n")

# Create a larger dataset to see performance difference
large_data = b"Performance testing with cramjam LZ4 algorithm. " * 1000
print(f"Data size: {len(large_data):,} bytes ({len(large_data) / 1024:.1f} KB)\n")

# Test LZ4 compression
print("LZ4 compression (optimized for speed):")
start = time.perf_counter()
compressed = cramjam.lz4.compress(large_data)
compress_time = time.perf_counter() - start
print(f"  Compression time: {compress_time * 1000:.2f} ms")
print(f"  Compressed size: {len(compressed):,} bytes")
print(f"  Compression ratio: {len(large_data) / len(compressed):.2f}x")

# Test decompression
start = time.perf_counter()
decompressed = cramjam.lz4.decompress(compressed)
decompress_time = time.perf_counter() - start
print(f"  Decompression time: {decompress_time * 1000:.2f} ms")
print(f"  Data integrity: {bytes(decompressed) == large_data}")

# Calculate throughput
compress_throughput = (len(large_data) / 1024 / 1024) / compress_time
decompress_throughput = (len(large_data) / 1024 / 1024) / decompress_time
print(f"\nThroughput:")
print(f"  Compression: {compress_throughput:.1f} MB/s")
print(f"  Decompression: {decompress_throughput:.1f} MB/s")

print("\nLZ4 is ideal for:")
print("  • Real-time data compression")
print("  • Network data transmission")
print("  • In-memory data compression")
print("  • Applications requiring fast decompression")
```
