## Example 1: Basic CRC32C Checksums

```python
import crc32c

print("Example 1: Basic CRC32C Checksums")

try:
    # Test basic checksum calculation
    data1 = b"Hello, World!"
    checksum1 = crc32c.crc32c(data1)
    
    print(f"✓ Data: {data1}")
    print(f"✓ CRC32C checksum: 0x{checksum1:08x}")
    
    # Test with different data
    data2 = b"The quick brown fox jumps over the lazy dog"
    checksum2 = crc32c.crc32c(data2)
    
    print(f"✓ Data: {data2}")
    print(f"✓ CRC32C checksum: 0x{checksum2:08x}")
    
    # Test empty data
    empty_data = b""
    empty_checksum = crc32c.crc32c(empty_data)
    print(f"✓ Empty data checksum: 0x{empty_checksum:08x}")
    
    # Test incremental checksum calculation
    part1 = b"Hello, "
    part2 = b"World!"
    
    # Calculate in parts
    partial_checksum = crc32c.crc32c(part1)
    final_checksum = crc32c.crc32c(part2, partial_checksum)
    
    # Calculate all at once
    direct_checksum = crc32c.crc32c(part1 + part2)
    
    print(f"✓ Incremental checksum: 0x{final_checksum:08x}")
    print(f"✓ Direct checksum: 0x{direct_checksum:08x}")
    print(f"✓ Checksums match: {final_checksum == direct_checksum}")
    
    # Test with numeric data
    numeric_data = b"1234567890"
    numeric_checksum = crc32c.crc32c(numeric_data)
    print(f"✓ Numeric data checksum: 0x{numeric_checksum:08x}")
    
    # Verify reproducibility
    repeat_checksum = crc32c.crc32c(data1)
    print(f"✓ Reproducible: {checksum1 == repeat_checksum}")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"crc32c example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 completed")
```

## Example 2: Incremental Checksum Calculation

```python
import crc32c

print("Example 2: Incremental Checksum Calculation")

try:
    # Demonstrate incremental CRC32C calculation
    print("✓ Incremental checksum processing:")
    
    # Process data in chunks
    chunks = [b"This is ", b"a test ", b"of incremental ", b"CRC32C ", b"calculation."]
    
    # Calculate incrementally
    incremental_checksum = 0  # Start with 0 (no initial value)
    for i, chunk in enumerate(chunks):
        incremental_checksum = crc32c.crc32c(chunk, incremental_checksum)
        print(f"  After chunk {i+1}: 0x{incremental_checksum:08x}")
    
    # Calculate all at once for comparison
    full_data = b"".join(chunks)
    direct_checksum = crc32c.crc32c(full_data)
    
    print(f"  Incremental result: 0x{incremental_checksum:08x}")
    print(f"  Direct calculation: 0x{direct_checksum:08x}")
    print(f"  Results match: {incremental_checksum == direct_checksum}")
    
    # Practical example: File-like processing
    print("✓ File-like stream processing:")
    
    def simulate_file_stream(data, chunk_size=8):
        """Simulate reading a file in chunks"""
        for i in range(0, len(data), chunk_size):
            yield data[i:i + chunk_size]
    
    # Large data example
    large_data = b"The quick brown fox jumps over the lazy dog. " * 10
    
    # Process as stream
    stream_checksum = 0
    chunk_count = 0
    for chunk in simulate_file_stream(large_data, 16):
        stream_checksum = crc32c.crc32c(chunk, stream_checksum)
        chunk_count += 1
    
    # Direct calculation
    direct_large_checksum = crc32c.crc32c(large_data)
    
    print(f"  Processed {chunk_count} chunks")
    print(f"  Stream checksum: 0x{stream_checksum:08x}")
    print(f"  Direct checksum: 0x{direct_large_checksum:08x}")
    print(f"  Stream processing works: {stream_checksum == direct_large_checksum}")
    
    # Demonstrate with different initial values
    print("✓ Custom initial values:")
    
    test_data = b"Testing with initial value"
    custom_initial = 0x12345678
    
    # With custom initial value
    custom_checksum = crc32c.crc32c(test_data, custom_initial)
    
    # Equivalent: initial value as separate computation
    zero_checksum = crc32c.crc32c(test_data)
    
    print(f"  Test data: {test_data}")
    print(f"  With initial 0x{custom_initial:08x}: 0x{custom_checksum:08x}")
    print(f"  With initial 0x00000000: 0x{zero_checksum:08x}")
    print(f"  Different results: {custom_checksum != zero_checksum}")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"crc32c example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed")
```

## Example 3: Performance and Hardware Acceleration

```python
import crc32c
import time
import hashlib

print("Example 3: Performance and Hardware Acceleration")

try:
    # Test hardware acceleration detection
    print("✓ CRC32C implementation details:")
    
    # Check if hardware acceleration is available
    print("  crc32c module loaded successfully")
    
    # Check module attributes
    crc32c_attrs = [attr for attr in dir(crc32c) if not attr.startswith('_')]
    print(f"  Available functions: {crc32c_attrs}")
    
    # Performance comparison with different data sizes
    print("✓ Performance benchmarking:")
    
    test_sizes = [1024, 10240, 102400]  # 1KB, 10KB, 100KB
    
    for size in test_sizes:
        # Generate test data
        test_data = b'A' * size
        
        # Time CRC32C calculation
        start_time = time.perf_counter()
        iterations = 1000 if size <= 10240 else 100
        
        for _ in range(iterations):
            checksum = crc32c.crc32c(test_data)
        
        end_time = time.perf_counter()
        avg_time = (end_time - start_time) / iterations
        
        throughput = (size / max(avg_time, 1e-9)) / (1024 * 1024)  # MB/s
        
        print(f"  {size:6d} bytes: {avg_time*1000:6.2f}ms, {throughput:6.1f} MB/s")
    
    # Compare with Python's standard CRC32
    print("✓ CRC32C vs standard CRC32 comparison:")
    
    test_data = b"Performance comparison test data" * 100
    
    # CRC32C (hardware accelerated)
    start_time = time.perf_counter()
    crc32c_result = crc32c.crc32c(test_data)
    crc32c_time = time.perf_counter() - start_time
    
    # Standard CRC32 (from zlib)
    import zlib
    start_time = time.perf_counter()
    crc32_result = zlib.crc32(test_data) & 0xffffffff
    crc32_time = time.perf_counter() - start_time
    
    print(f"  CRC32C result: 0x{crc32c_result:08x} ({crc32c_time*1000:.2f}ms)")
    print(f"  CRC32 result:  0x{crc32_result:08x} ({crc32_time*1000:.2f}ms)")
    print(f"  CRC32C speedup: {crc32_time / max(crc32c_time, 1e-9):.1f}x faster")
    
    # Memory efficiency test
    print("✓ Memory efficiency:")
    
    # Large data processing in chunks (memory efficient)
    large_size = 1024 * 1024  # 1MB
    chunk_size = 8192  # 8KB chunks
    
    # Simulate processing large data without loading all into memory
    def generate_chunks(total_size, chunk_size):
        """Generate data chunks without storing all in memory"""
        remaining = total_size
        chunk_data = b'M' * min(chunk_size, remaining)
        while remaining > 0:
            current_chunk = min(chunk_size, remaining)
            yield chunk_data[:current_chunk]
            remaining -= current_chunk
    
    # Process in chunks
    start_time = time.perf_counter()
    chunk_checksum = 0
    chunk_count = 0
    
    for chunk in generate_chunks(large_size, chunk_size):
        chunk_checksum = crc32c.crc32c(chunk, chunk_checksum)
        chunk_count += 1
    
    chunk_time = time.perf_counter() - start_time
    
    print(f"  Processed {chunk_count} chunks ({large_size} bytes)")
    print(f"  Chunked checksum: 0x{chunk_checksum:08x}")
    print(f"  Processing time: {chunk_time:.3f}s")
    print(f"  Throughput: {(large_size / max(chunk_time, 1e-9)) / (1024*1024):.1f} MB/s")
    
    # Validation: small data consistency
    print("✓ Consistency validation:")
    
    validation_data = b"Consistency test"
    
    # Multiple calculations should be identical
    checksums = [crc32c.crc32c(validation_data) for _ in range(5)]
    all_same = all(c == checksums[0] for c in checksums)
    
    print(f"  Validation data: {validation_data}")
    print(f"  All checksums identical: {all_same}")
    print(f"  Checksum value: 0x{checksums[0]:08x}")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"crc32c example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 completed")
```

## Example 4: Data Integrity and Validation Use Cases

```python
import crc32c
import json
import struct

print("Example 4: Data Integrity and Validation Use Cases")

try:
    # File integrity verification simulation
    print("✓ File integrity verification:")
    
    # Simulate file data
    file_data = b"This is important file content that needs integrity verification." * 20
    
    # Calculate checksum for storage
    stored_checksum = crc32c.crc32c(file_data)
    print(f"  Original checksum: 0x{stored_checksum:08x}")
    
    # Simulate file corruption (single bit flip)
    corrupted_data = bytearray(file_data)
    corrupted_data[100] ^= 1  # Flip one bit
    corrupted_data = bytes(corrupted_data)
    
    # Verify integrity
    current_checksum = crc32c.crc32c(corrupted_data)
    integrity_ok = (stored_checksum == current_checksum)
    
    print(f"  Current checksum:  0x{current_checksum:08x}")
    print(f"  Integrity check: {'✓ PASS' if integrity_ok else '❌ CORRUPTED'}")
    
    # Network data validation simulation
    print("✓ Network packet validation:")
    
    # Simulate network packet with header and payload
    def create_packet(payload_data):
        """Create a packet with CRC32C checksum in header"""
        payload_checksum = crc32c.crc32c(payload_data)
        # Header: magic(4) + length(4) + checksum(4) = 12 bytes
        header = struct.pack('<III', 0xABCDEF00, len(payload_data), payload_checksum)
        return header + payload_data
    
    def verify_packet(packet_data):
        """Verify packet integrity using embedded checksum"""
        if len(packet_data) < 12:
            return False, "Packet too short"
        
        # Extract header
        magic, length, expected_checksum = struct.unpack('<III', packet_data[:12])
        
        if magic != 0xABCDEF00:
            return False, "Invalid magic number"
        
        if len(packet_data) != 12 + length:
            return False, "Length mismatch"
        
        # Extract and verify payload
        payload = packet_data[12:]
        actual_checksum = crc32c.crc32c(payload)
        
        return (expected_checksum == actual_checksum), f"Checksum {'match' if expected_checksum == actual_checksum else 'mismatch'}"
    
    # Test with valid packet
    original_payload = b"Network payload data for transmission"
    valid_packet = create_packet(original_payload)
    
    is_valid, message = verify_packet(valid_packet)
    print(f"  Valid packet: {'✓ PASS' if is_valid else '❌ FAIL'} - {message}")
    
    # Test with corrupted packet
    corrupted_packet = bytearray(valid_packet)
    corrupted_packet[-5] ^= 0xFF  # Corrupt payload
    corrupted_packet = bytes(corrupted_packet)
    
    is_valid, message = verify_packet(corrupted_packet)
    print(f"  Corrupted packet: {'✓ PASS' if is_valid else '❌ FAIL'} - {message}")
    
    # Database record integrity
    print("✓ Database record integrity:")
    
    # Simulate database records with checksums
    class IntegrityRecord:
        def __init__(self, data):
            self.data = data
            self.checksum = crc32c.crc32c(json.dumps(data, sort_keys=True).encode())
        
        def verify(self):
            current_checksum = crc32c.crc32c(json.dumps(self.data, sort_keys=True).encode())
            return current_checksum == self.checksum
    
    # Create records
    record1 = IntegrityRecord({"id": 1, "name": "Alice", "balance": 1000.50})
    record2 = IntegrityRecord({"id": 2, "name": "Bob", "balance": 750.25})
    
    print(f"  Record 1 integrity: {'✓ VALID' if record1.verify() else '❌ INVALID'}")
    print(f"  Record 2 integrity: {'✓ VALID' if record2.verify() else '❌ INVALID'}")
    
    # Simulate data tampering
    record1.data["balance"] = 9999.99  # Unauthorized change
    print(f"  Modified record 1: {'✓ VALID' if record1.verify() else '❌ INVALID'}")
    
    # Backup verification
    print("✓ Backup data verification:")
    
    # Simulate backup chunks with individual checksums
    backup_chunks = [
        b"Chunk 1: Critical system configuration data",
        b"Chunk 2: User account information and preferences",
        b"Chunk 3: Application data and settings",
        b"Chunk 4: Database indexes and metadata"
    ]
    
    # Create backup manifest with checksums
    backup_manifest = {}
    for i, chunk in enumerate(backup_chunks):
        chunk_id = f"chunk_{i+1}"
        backup_manifest[chunk_id] = {
            "size": len(chunk),
            "checksum": crc32c.crc32c(chunk)
        }
    
    print(f"  Created manifest for {len(backup_chunks)} chunks")
    
    # Verify backup integrity
    all_chunks_valid = True
    for i, chunk in enumerate(backup_chunks):
        chunk_id = f"chunk_{i+1}"
        expected_checksum = backup_manifest[chunk_id]["checksum"]
        actual_checksum = crc32c.crc32c(chunk)
        
        chunk_valid = (expected_checksum == actual_checksum)
        all_chunks_valid = all_chunks_valid and chunk_valid
        
        print(f"    {chunk_id}: {'✓ VALID' if chunk_valid else '❌ INVALID'}")
    
    print(f"  Overall backup integrity: {'✓ VALID' if all_chunks_valid else '❌ COMPROMISED'}")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"crc32c example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed")
```
