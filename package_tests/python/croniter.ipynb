## Example 1: Basic Cron Expression Parsing and Iteration

```python
from croniter import croniter
from datetime import datetime

print("Example 1: Basic Cron Expression Parsing and Iteration")

try:
    print("✓ croniter package information:")
    print(f"  Croniter class: {croniter}")
    print(f"  Module functionality: Cron expression parsing and scheduling")
    
    # Create sample datetime for calculations
    base_time = datetime(2024, 1, 1, 9, 30, 0)  # Monday morning
    print(f"  Base time: {base_time}")
    
    # Demonstrate basic cron expressions
    print("✓ Basic cron expression examples:")
    
    cron_examples = {
        "0 9 * * *": "Daily at 9:00 AM",
        "0 12 * * 1-5": "Weekdays at noon",
        "*/15 * * * *": "Every 15 minutes",
        "0 0 1 * *": "First day of each month at midnight",
        "0 18 * * 5": "Every Friday at 6:00 PM"
    }
    
    for cron_expr, description in cron_examples.items():
        print(f"  '{cron_expr}': {description}")
        
        # Create croniter instance
        cron = croniter(cron_expr, base_time)
        
        # Get next execution time
        next_time = cron.get_next(datetime)
        print(f"    Next execution: {next_time}")
        
    # Demonstrate forward iteration
    print("✓ Forward iteration example:")
    daily_cron = croniter("0 14 * * *", base_time)  # Daily at 2 PM
    
    print("  Next 5 daily executions at 2 PM:")
    for i in range(5):
        next_exec = daily_cron.get_next(datetime)
        day_name = next_exec.strftime("%A")
        print(f"    {i+1}. {next_exec} ({day_name})")
        
    # Demonstrate backward iteration  
    print("✓ Backward iteration example:")
    weekly_cron = croniter("0 10 * * 0", base_time)  # Sundays at 10 AM
    
    print("  Previous 3 Sunday executions at 10 AM:")
    for i in range(3):
        prev_exec = weekly_cron.get_prev(datetime)
        print(f"    {i+1}. {prev_exec}")
        
    print("✓ Basic cron expression parsing and iteration completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"croniter example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Advanced Cron Patterns and Matching

```python
from croniter import croniter
from datetime import datetime

print("Example 2: Advanced Cron Patterns and Matching")

try:
    # Demonstrate advanced cron patterns
    print("✓ Advanced cron expression patterns:")
    
    base_time = datetime(2024, 3, 15, 14, 30, 0)  # Friday afternoon
    
    advanced_patterns = {
        "0 9-17/2 * * 1-5": "Business hours, every 2 hours (9,11,13,15,17)",
        "*/10 8-18 * * 1-5": "Every 10 minutes during work hours on weekdays",
        "0 0 */3 * *": "Every 3 days at midnight",
        "15,45 * * * *": "At 15 and 45 minutes past each hour",
        "0 12 1-7 * 1": "First Monday of each month at noon",
        "0 2 L * *": "Last day of each month at 2 AM",
        "0 6 * * 1#2": "Second Monday of each month at 6 AM"
    }
    
    for pattern, description in advanced_patterns.items():
        print(f"  '{pattern}': {description}")
        
        try:
            cron = croniter(pattern, base_time)
            next_time = cron.get_next(datetime)
            print(f"    Next: {next_time}")
        except Exception as e:
            print(f"    Note: Pattern may require extended croniter support ({e})")
            raise
            
    # Demonstrate croniter matching functionality
    print("✓ Cron expression matching:")
    
    # Test times for matching
    test_times = [
        datetime(2024, 3, 18, 12, 0, 0),  # Monday noon
        datetime(2024, 3, 18, 12, 30, 0), # Monday 12:30 PM
        datetime(2024, 3, 19, 12, 0, 0),  # Tuesday noon  
        datetime(2024, 3, 23, 12, 0, 0),  # Saturday noon
    ]
    
    # Test pattern: weekdays at noon
    weekday_noon = "0 12 * * 1-5"
    
    print(f"  Testing pattern '{weekday_noon}' (weekdays at noon):")
    for test_time in test_times:
        matches = croniter.match(weekday_noon, test_time)
        day_name = test_time.strftime("%A")
        print(f"    {test_time} ({day_name}): {'✓ Matches' if matches else '✗ No match'}")
        
    # Demonstrate range calculations
    print("✓ Range calculations:")
    
    # Calculate all executions in a time range
    start_time = datetime(2024, 3, 18, 0, 0, 0)  # Monday
    end_time = datetime(2024, 3, 22, 23, 59, 59)  # Friday
    
    hourly_pattern = "0 */3 * * 1-5"  # Every 3 hours on weekdays
    cron = croniter(hourly_pattern, start_time)
    
    executions = []
    current_time = cron.get_next(datetime)
    
    while current_time <= end_time:
        executions.append(current_time)
        current_time = cron.get_next(datetime)
        
    print(f"  Pattern '{hourly_pattern}' executions from {start_time.date()} to {end_time.date()}:")
    print(f"  Total executions: {len(executions)}")
    
    for i, exec_time in enumerate(executions[:5]):  # Show first 5
        day_name = exec_time.strftime("%A")
        print(f"    {i+1}. {exec_time} ({day_name})")
        
    if len(executions) > 5:
        print(f"    ... and {len(executions) - 5} more")
    
    print("✓ Advanced cron patterns and matching completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"croniter example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: Practical Scheduling Applications

```python
from croniter import croniter
from datetime import datetime, timedelta
import time

print("Example 3: Practical Scheduling Applications")

try:
    # Demonstrate practical scheduling scenarios
    print("✓ Common scheduling scenarios:")
    
    current_time = datetime.now()
    print(f"  Current time: {current_time}")
    
    # Common business scheduling patterns
    business_schedules = {
        "Daily backups": "0 2 * * *",
        "Weekly reports": "0 9 * * 1",
        "Monthly billing": "0 0 1 * *",
        "Quarterly reviews": "0 10 1 */3 *",
        "Business hour maintenance": "0 */2 9-17 * * 1-5",
        "Weekend data processing": "0 6 * * 6,0"
    }
    
    print("  Business scheduling patterns:")
    for task, pattern in business_schedules.items():
        cron = croniter(pattern, current_time)
        next_run = cron.get_next(datetime)
        time_until = next_run - current_time
        
        print(f"    {task}:")
        print(f"      Pattern: {pattern}")
        print(f"      Next run: {next_run}")
        print(f"      Time until: {time_until}")
        
    # Demonstrate scheduling calculator
    print("✓ Scheduling calculator:")
    
    def calculate_schedule_window(cron_expr, start_date, days_ahead=7):
        """Calculate all executions within a time window"""
        end_date = start_date + timedelta(days=days_ahead)
        cron = croniter(cron_expr, start_date)
        
        executions = []
        current = cron.get_next(datetime)
        
        while current <= end_date:
            executions.append(current)
            current = cron.get_next(datetime)
            
        return executions
    
    # Calculate maintenance windows
    maintenance_pattern = "0 3 * * 1,3,5"  # Mon, Wed, Fri at 3 AM
    start_date = datetime(2024, 6, 17, 0, 0, 0)  # Monday
    
    maintenance_times = calculate_schedule_window(maintenance_pattern, start_date, 14)
    
    print(f"  Maintenance schedule (pattern: {maintenance_pattern}):")
    print(f"  From {start_date.date()} for 14 days:")
    
    for i, maint_time in enumerate(maintenance_times, 1):
        day_name = maint_time.strftime("%A")
        print(f"    {i}. {maint_time} ({day_name})")
        
    # Demonstrate scheduling conflict detection
    print("✓ Schedule conflict detection:")
    
    def check_schedule_conflicts(patterns, check_time):
        """Check if multiple schedules conflict at a specific time"""
        conflicts = []
        
        for name, pattern in patterns.items():
            if croniter.match(pattern, check_time):
                conflicts.append(name)
                
        return conflicts
    
    # Define multiple schedules
    schedules = {
        "Database backup": "0 2 * * *",
        "Log rotation": "0 2 * * 0",
        "System updates": "0 2 * * 1"
    }
    
    # Check for conflicts on specific times
    test_times = [
        datetime(2024, 6, 17, 2, 0, 0),  # Monday 2 AM
        datetime(2024, 6, 23, 2, 0, 0),  # Sunday 2 AM
        datetime(2024, 6, 18, 2, 0, 0),  # Tuesday 2 AM
    ]
    
    for check_time in test_times:
        conflicts = check_schedule_conflicts(schedules, check_time)
        day_name = check_time.strftime("%A")
        
        print(f"  {check_time} ({day_name}):")
        if conflicts:
            print(f"    ⚠️ Conflicts: {', '.join(conflicts)}")
        else:
            print(f"    ✓ No conflicts")
            
    # Show scheduling optimization
    print("✓ Schedule optimization tips:")
    
    optimization_tips = [
        "Spread resource-intensive tasks across different times",
        "Use off-peak hours for maintenance and backups", 
        "Consider timezone differences for global operations",
        "Implement retry logic for failed scheduled tasks",
        "Monitor execution times to optimize scheduling",
        "Use ranges and steps to distribute load evenly"
    ]
    
    for tip in optimization_tips:
        print(f"  • {tip}")
        
    # Demonstrate dynamic scheduling
    print("✓ Dynamic scheduling example:")
    
    # Calculate next few business day mornings
    business_morning = "0 9 * * 1-5"  # 9 AM weekdays
    cron = croniter(business_morning, current_time)
    
    print("  Next 5 business day mornings:")
    for i in range(5):
        next_morning = cron.get_next(datetime)
        day_name = next_morning.strftime("%A")
        days_away = (next_morning.date() - current_time.date()).days
        
        print(f"    {i+1}. {next_morning} ({day_name}) - {days_away} days away")
    
    print("✓ Practical scheduling applications completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"croniter example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```
