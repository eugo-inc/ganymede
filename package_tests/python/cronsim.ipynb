```python
# Example 1: Basic CronSim iteration — every minute.
from datetime import datetime
from cronsim import CronSim

start = datetime(2026, 1, 1, 0, 0)
it = CronSim("* * * * *", start)

first_five = [next(it) for _ in range(5)]
for ts in first_five:
    print(ts)

assert first_five[0] == datetime(2026, 1, 1, 0, 1)
assert first_five[4] == datetime(2026, 1, 1, 0, 5)
assert len(first_five) == 5
```

```python
# Example 2: Hourly schedule — at minute 0 of every hour.
from datetime import datetime
from cronsim import CronSim

start = datetime(2026, 3, 15, 10, 30)
it = CronSim("0 * * * *", start)

next_three = [next(it) for _ in range(3)]
for ts in next_three:
    print(ts)

assert next_three[0] == datetime(2026, 3, 15, 11, 0)
assert next_three[1] == datetime(2026, 3, 15, 12, 0)
assert next_three[2] == datetime(2026, 3, 15, 13, 0)
```

```python
# Example 3: Day-of-week schedule — 9:00 AM every Monday.
from datetime import datetime
from cronsim import CronSim

start = datetime(2026, 1, 1, 0, 0)
it = CronSim("0 9 * * MON", start)

first = next(it)
second = next(it)
print(f"first Monday 9am: {first}")
print(f"second Monday 9am: {second}")

assert first == datetime(2026, 1, 5, 9, 0)
assert second == datetime(2026, 1, 12, 9, 0)
assert first.weekday() == 0  # Monday
```

```python
# Example 4: Step values and ranges.
from datetime import datetime
from cronsim import CronSim

# Every 15 minutes, between 9 and 17, weekdays only.
start = datetime(2026, 6, 1, 8, 0)
it = CronSim("*/15 9-17 * * MON-FRI", start)

first_four = [next(it) for _ in range(4)]
for ts in first_four:
    print(ts)

assert first_four[0] == datetime(2026, 6, 1, 9, 0)
assert first_four[1] == datetime(2026, 6, 1, 9, 15)
assert first_four[2] == datetime(2026, 6, 1, 9, 30)
assert first_four[3] == datetime(2026, 6, 1, 9, 45)
```

```python
# Example 5: Invalid expressions raise CronSimError.
from datetime import datetime
from cronsim import CronSim, CronSimError

start = datetime(2026, 1, 1, 0, 0)

bad_expressions = [
    "not a cron expression",
    "60 * * * *",          # minute out of range
    "* 25 * * *",          # hour out of range
    "* * 32 * *",          # day out of range
]

errors_caught = 0
for expr in bad_expressions:
    try:
        CronSim(expr, start)
    except CronSimError as e:
        errors_caught += 1
        print(f"caught for {expr!r}: {e}")

assert errors_caught == len(bad_expressions), f"expected {len(bad_expressions)} errors, got {errors_caught}"
```
