## Example 1: Basic CSS Selector Parsing and Translation

```python
import cssselect

print("Example 1: Basic CSS Selector Parsing and Translation")

try:
    print("✓ cssselect package information:")
    print(f"  Module type: {type(cssselect)}")
    
    # Check version
    if hasattr(cssselect, '__version__'):
        print(f"  Version: {cssselect.__version__}")
    else:
        print("  Version: Available via package metadata")
    
    # Show main classes and functionality
    attrs = [attr for attr in dir(cssselect) if not attr.startswith('_')]
    print(f"  Public attributes: {len(attrs)}")
    print(f"  Key components: {attrs}")
    
    # Demonstrate CSS to XPath translation
    print("✓ CSS to XPath translation:")
    
    if hasattr(cssselect, 'GenericTranslator'):
        translator = cssselect.GenericTranslator()
        
        # Basic CSS selectors and their XPath equivalents
        css_selectors = [
            "div",                    # Element selector
            ".content",               # Class selector  
            "#header",                # ID selector
            "div.content",            # Element with class
            "div p",                  # Descendant selector
            "div > p",                # Child selector
            "p.highlight",            # Element with class
        ]
        
        print("  Basic CSS selectors to XPath:")
        for css_sel in css_selectors:
            xpath = translator.css_to_xpath(css_sel)
            print(f"    '{css_sel}' → {xpath}")
            
        # Show complex selectors
        print("  Complex CSS selectors:")
        complex_selectors = [
            "div.container p.text",           # Multiple classes
            "ul li:first-child",              # Pseudo-selector
            "input[type='text']",             # Attribute selector
            "div.header, div.footer",         # Multiple selectors
        ]
        
        for css_sel in complex_selectors:
            try:
                xpath = translator.css_to_xpath(css_sel)
                print(f"    '{css_sel}' → {xpath[:80]}...")
            except Exception as e:
                print(f"    '{css_sel}' → Error: {e}")
                raise
                
    # Demonstrate parsing functionality
    print("✓ CSS selector parsing:")
    
    if hasattr(cssselect, 'parse'):
        # Parse CSS selectors into components
        sample_selector = "div.content p:nth-child(2)"
        
        try:
            parsed = cssselect.parse(sample_selector)
            print(f"  Parsed selector '{sample_selector}':")
            print(f"    Type: {type(parsed)}")
            print(f"    Components: {len(parsed)} selector(s)")
            
            # Show individual components
            for i, selector in enumerate(parsed):
                print(f"    Selector {i+1}: {selector}")
                
        except Exception as e:
            print(f"    Parsing error: {e}")
            raise
            
    print("✓ Basic CSS selector parsing and translation completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cssselect example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Advanced Selectors and Pseudo-classes

```python
import cssselect

print("Example 2: Advanced Selectors and Pseudo-classes")

try:
    # Demonstrate advanced CSS selector types
    print("✓ Advanced CSS selector types:")
    
    translator = cssselect.GenericTranslator()
    
    # Attribute selectors
    print("  Attribute selectors:")
    attribute_selectors = [
        "input[type]",                    # Has attribute
        "input[type='text']",             # Exact value
        "input[type^='text']",            # Starts with
        "input[type$='word']",            # Ends with
        "input[type*='xt']",              # Contains
        "a[href~='example']",             # Word in space-separated list
        "div[class|='nav']",              # Starts with value or value-
    ]
    
    for selector in attribute_selectors:
        try:
            xpath = translator.css_to_xpath(selector)
            print(f"    '{selector}' → {xpath[:60]}...")
        except Exception as e:
            print(f"    '{selector}' → Error: {str(e)[:40]}...")
            raise
            
    # Pseudo-class selectors
    print("  Pseudo-class selectors:")
    pseudo_selectors = [
        "p:first-child",                  # First child
        "p:last-child",                   # Last child
        "li:nth-child(2)",                # Nth child
        "li:nth-child(odd)",              # Odd children
        "li:nth-child(even)",             # Even children
        "tr:nth-child(2n+1)",             # Formula-based
        "input:checked",                  # Form states
        "a:visited",                      # Link states
    ]
    
    for selector in pseudo_selectors:
        try:
            xpath = translator.css_to_xpath(selector)
            print(f"    '{selector}' → {xpath[:60]}...")
        except Exception as e:
            print(f"    '{selector}' → Unsupported or error: {str(e)[:30]}...")
            raise
            
    # Combinator selectors
    print("  Combinator selectors:")
    combinator_selectors = [
        "div p",                          # Descendant (space)
        "div > p",                        # Child (>)
        "h1 + p",                         # Adjacent sibling (+)
        "h1 ~ p",                         # General sibling (~)
        "div.container > ul li",          # Mixed combinators
    ]
    
    for selector in combinator_selectors:
        try:
            xpath = translator.css_to_xpath(selector)
            print(f"    '{selector}' → {xpath[:60]}...")
        except Exception as e:
            print(f"    '{selector}' → Error: {str(e)[:30]}...")
            raise
            
    # Show CSS selector validation
    print("✓ CSS selector validation:")
    
    def validate_css_selector(selector_str):
        """Validate CSS selector by attempting to parse it"""
        try:
            parsed = cssselect.parse(selector_str)
            xpath = translator.css_to_xpath(selector_str)
            return True, f"Valid - parsed into {len(parsed)} component(s)"
        except Exception as e:
            return False, f"Invalid - {str(e)[:50]}"
    
    # Test various selectors for validity
    test_selectors = [
        "div.content",                    # Valid basic
        "div > p:first-child",            # Valid complex
        "input[type='text']:focus",       # Valid with pseudo
        "div..content",                   # Invalid (double dot)
        "div > > p",                      # Invalid (double combinator)
        "div[class='test'",              # Invalid (missing bracket)
    ]
    
    print("  Selector validation results:")
    for selector in test_selectors:
        is_valid, message = validate_css_selector(selector)
        status = "✓" if is_valid else "✗"
        print(f"    {status} '{selector}': {message}")
        
    # Show grouping and multiple selectors
    print("✓ Grouped selectors:")
    
    group_selectors = [
        "h1, h2, h3",                     # Multiple elements
        ".warning, .error",               # Multiple classes
        "div.header, div.footer",         # Complex groups
        "p:first-child, p:last-child",    # Multiple pseudo-classes
    ]
    
    for selector in group_selectors:
        parsed = cssselect.parse(selector)
        print(f"  '{selector}' → {len(parsed)} individual selectors")
        assert len(parsed) == selector.count(",") + 1
        
        # Show each individual selector (a parsed Selector is translated with selector_to_xpath;
        # its str() is a debug repr, not CSS text)
        for i, individual in enumerate(parsed):
            xpath = translator.selector_to_xpath(individual)
            print(f"    {i+1}. {xpath[:50]}... ← {individual.canonical()}")
            assert xpath == translator.css_to_xpath(individual.canonical()), (xpath, individual.canonical())
    
    print("✓ Advanced selectors and pseudo-classes completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cssselect example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: HTML and XML Document Processing

```python
import cssselect

print("Example 3: HTML and XML Document Processing")

try:
    # Demonstrate HTML-specific translator
    print("✓ HTML-specific CSS translation:")
    
    if hasattr(cssselect, 'HTMLTranslator'):
        html_translator = cssselect.HTMLTranslator()
        print("  HTMLTranslator available for HTML-specific optimizations")
        
        # HTML-specific selectors
        html_selectors = [
            "input[type='text']",         # HTML input types
            "a:link",                     # Link pseudo-classes
            "input:checked",              # Form states
            "option:checked",             # Form element states (':selected' is a jQuery-ism cssselect does not know)
            "tr:nth-child(even)",         # Table row patterns
            "td:first-child",             # Table cell selection
        ]
        
        print("  HTML-optimized XPath translations:")
        for selector in html_selectors:
            xpath = html_translator.css_to_xpath(selector)
            print(f"    '{selector}' → {xpath[:70]}...")
            assert xpath.startswith("descendant-or-self::"), xpath
    
    # Compare HTML vs Generic translators
    print("✓ HTML vs Generic translator comparison:")
    
    if hasattr(cssselect, 'HTMLTranslator') and hasattr(cssselect, 'GenericTranslator'):
        html_translator = cssselect.HTMLTranslator()
        generic_translator = cssselect.GenericTranslator()
        
        comparison_selectors = [
            "input[type='text']",
            "a:visited",
            "tr:nth-child(odd)",
        ]
        
        for selector in comparison_selectors:
            print(f"  Selector: '{selector}'")
            try:
                html_xpath = html_translator.css_to_xpath(selector)
                print(f"    HTML: {html_xpath[:60]}...")
            except Exception as e:
                print(f"    HTML: Error - {e}")
                raise
                
            try:
                generic_xpath = generic_translator.css_to_xpath(selector)
                print(f"    Generic: {generic_xpath[:60]}...")
            except Exception as e:
                print(f"    Generic: Error - {e}")
                raise
                
    # Demonstrate XML namespace handling
    print("✓ XML namespace considerations:")
    
    # XML selectors with namespace implications
    xml_considerations = [
        "Element names are case-sensitive in XML",
        "Namespace prefixes need proper handling",
        "Attribute matching may differ from HTML",
        "Pseudo-classes may not apply in XML context"
    ]
    
    for consideration in xml_considerations:
        print(f"  • {consideration}")
        
    # Show document type-specific optimizations
    print("✓ Document type optimizations:")
    
    # Create sample HTML-like and XML-like selectors
    document_specific = {
        'HTML Forms': [
            "form input[type='submit']",
            "select option:checked",
            "textarea:focus",
            "fieldset legend"
        ],
        'HTML Tables': [
            "table tbody tr:nth-child(even)",
            "th:first-child",
            "td.numeric",
            "table caption"
        ],
        'HTML Structure': [
            "header nav ul li",
            "main article section",
            "footer address",
            "aside .widget"
        ],
        'XML Data': [
            "record item[type='primary']",
            "document metadata",
            "data-set entry",
            "config setting[enabled='true']"
        ]
    }
    
    translator = cssselect.GenericTranslator()
    
    for doc_type, selectors in document_specific.items():
        print(f"  {doc_type} selectors:")
        for selector in selectors:
            try:
                xpath = translator.css_to_xpath(selector)
                print(f"    '{selector}' → Valid XPath generated")
            except Exception as e:
                print(f"    '{selector}' → Error: {str(e)[:30]}...")
                raise
                
    # Show practical integration patterns
    print("✓ Integration patterns:")
    
    integration_examples = [
        "lxml: element.cssselect() method integration",
        "BeautifulSoup: CSS selector support via cssselect",
        "Scrapy: CSS selector extraction rules",
        "PyQuery: jQuery-like CSS selectors for Python",
        "XML parsing: Element selection in XML documents"
    ]
    
    for example in integration_examples:
        print(f"  • {example}")
        
    # Demonstrate selector optimization
    print("✓ Selector optimization tips:")
    
    optimization_tips = [
        "Use specific selectors to improve XPath efficiency",
        "Avoid overly complex nested selectors when possible",
        "Consider browser compatibility for pseudo-classes",
        "Test XPath output for performance in target XML library",
        "Use appropriate translator (HTML vs Generic) for context"
    ]
    
    for tip in optimization_tips:
        print(f"  • {tip}")
    
    print("✓ HTML and XML document processing completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cssselect example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: Error Handling and Edge Cases

```python
import cssselect

print("Example 4: Error Handling and Edge Cases")

try:
    # Demonstrate error types and handling
    print("✓ CSS selector error types:")
    
    # Check available exception classes
    error_classes = ['SelectorError', 'SelectorSyntaxError', 'ExpressionError']
    
    for error_class in error_classes:
        if hasattr(cssselect, error_class):
            error_type = getattr(cssselect, error_class)
            print(f"  {error_class}: {error_type}")
        else:
            print(f"  {error_class}: Not available")
            
    # Test various invalid selectors
    print("✓ Invalid selector handling:")
    
    translator = cssselect.GenericTranslator()
    
    # selector -> the error cssselect must raise: a syntax error at parse time, or an
    # ExpressionError when it parses but has no XPath translation
    invalid_selectors = {
        "": cssselect.SelectorSyntaxError,                      # Empty selector
        "   ": cssselect.SelectorSyntaxError,                   # Whitespace only
        "div..class": cssselect.SelectorSyntaxError,            # Double dots
        "div > > p": cssselect.SelectorSyntaxError,             # Double combinators
        "div[class='test'": cssselect.SelectorSyntaxError,     # Unclosed bracket
        "div]class='test']": cssselect.SelectorSyntaxError,    # Wrong bracket order
        "div:unknown-pseudo": cssselect.ExpressionError,       # Unknown pseudo-class
        "div::unknown-element": cssselect.ExpressionError,     # Unknown pseudo-element
        "@media screen": cssselect.SelectorSyntaxError,        # CSS at-rule (not selector)
        "div { color: red; }": cssselect.SelectorSyntaxError,  # CSS rule instead of selector
    }
    
    for invalid_selector, expected_error in invalid_selectors.items():
        print(f"  Testing: '{invalid_selector}'")
        try:
            parsed = cssselect.parse(invalid_selector)
            xpath = translator.css_to_xpath(invalid_selector)
        except cssselect.SelectorError as e:
            assert type(e) is expected_error, (invalid_selector, type(e), expected_error)
            print(f"    Rejected as expected ({type(e).__name__}) - {str(e)[:60]}")
        else:
            raise AssertionError(f"expected {expected_error.__name__} for {invalid_selector!r}, got XPath {xpath!r}")
            
    # Demonstrate edge cases that work
    print("✓ Edge cases that are valid:")
    
    edge_case_selectors = [
        "*",                             # Universal selector
        "div *",                         # Universal descendant
        "[class]",                       # Any element with class attribute
        ":root",                         # Root pseudo-class
        "div:not(.excluded)",            # Negation pseudo-class
        "div:nth-child(n+2)",            # Complex nth-child formula
        "div[class~='item']",            # Word match in attribute
        "div[data-*='value']",           # Attribute prefix matching
    ]
    
    for selector in edge_case_selectors:
        try:
            parsed = cssselect.parse(selector)
            xpath = translator.css_to_xpath(selector)
            print(f"  '{selector}' → Valid (XPath: {xpath[:50]}...)")
        except Exception as e:
            print(f"  '{selector}' → Error: {str(e)[:40]}")
            raise
            
    # Show error recovery and validation
    print("✓ Error recovery patterns:")
    
    def safe_css_to_xpath(selector_str, fallback_xpath="*"):
        """Safely convert CSS to XPath with fallback"""
        try:
            return translator.css_to_xpath(selector_str), None
        except Exception as e:
            return fallback_xpath, str(e)
    
    # Test error recovery: (selector, is_valid) — the fallback must be used for exactly the invalid ones
    test_selectors = [
        ("div.content", True),
        ("div..invalid", False),
        ("p:nth-child(2)", True),
        ("p[missing-quote='test", False),
    ]
    
    print("  Error recovery results:")
    for selector, is_valid in test_selectors:
        xpath, error = safe_css_to_xpath(selector)
        assert (error is None) == is_valid, (selector, is_valid, error)
        if error:
            assert xpath == "*", (selector, xpath)
            print(f"    '{selector}' → rejected, fallback XPath {xpath!r} used ({error[:30]})")
        else:
            print(f"    '{selector}' → Success: {xpath[:50]}...")
            
    # Demonstrate CSS selector complexity limits
    print("✓ Complexity and performance considerations:")
    
    # Test increasingly complex selectors
    complexity_tests = [
        "div",                                                           # Simple
        "div.class#id[attr='value']",                                   # Multiple conditions
        "div > ul > li:nth-child(2) > a[href^='http']",                # Deeply nested
        "div.a > ul.b > li.c:nth-child(odd) > a.d[href*='example']",   # Very specific
    ]
    
    print("  Complexity handling:")
    for i, selector in enumerate(complexity_tests, 1):
        try:
            xpath = translator.css_to_xpath(selector)
            print(f"    Level {i}: '{selector}' → Success ({len(xpath)} chars)")
        except Exception as e:
            print(f"    Level {i}: '{selector}' → Error: {str(e)[:40]}")
            raise
            
    # Show debugging and introspection
    print("✓ Debugging and introspection:")
    
    debug_info = [
        "Use cssselect.parse() to inspect parsed structure",
        "Check translator.css_to_xpath() output before use",
        "Validate selectors in development environment",
        "Handle exceptions gracefully in production code",
        "Test XPath output with target XML library"
    ]
    
    for info in debug_info:
        print(f"  • {info}")
    
    print("✓ Error handling and edge cases completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cssselect example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: Performance and Integration Best Practices

```python
import cssselect
import time

print("Example 5: Performance and Integration Best Practices")

try:
    # Demonstrate performance characteristics
    print("✓ Performance benchmarking:")
    
    translator = cssselect.GenericTranslator()
    
    # Test selector parsing performance
    test_selectors = [
        "div",                          # Simple
        "div.content p",                # Moderate  
        "div > ul li:nth-child(odd)",   # Complex
        "form input[type='text']:focus" # Very complex
    ]
    
    # Benchmark parsing time
    parse_times = []
    translation_times = []
    
    for selector in test_selectors:
        # Time parsing
        start_time = time.time()
        for _ in range(1000):
            parsed = cssselect.parse(selector)
        parse_time = (time.time() - start_time) * 1000
        parse_times.append(parse_time)
        
        # Time translation
        start_time = time.time()
        for _ in range(1000):
            xpath = translator.css_to_xpath(selector)
        translation_time = (time.time() - start_time) * 1000
        translation_times.append(translation_time)
        
        print(f"  '{selector}':")
        print(f"    Parse 1000x: {parse_time:.1f}ms ({parse_time/1000:.3f}ms each)")
        print(f"    Translate 1000x: {translation_time:.1f}ms ({translation_time/1000:.3f}ms each)")
        
    # Performance optimization tips
    print("✓ Performance optimization:")
    
    optimization_tips = [
        "Cache translator instances for repeated use",
        "Pre-parse and validate selectors at startup",
        "Use simpler selectors when possible for better performance",
        "Consider XPath complexity when designing CSS selectors",
        "Profile XPath execution in target XML processing library"
    ]
    
    for tip in optimization_tips:
        print(f"  • {tip}")
        
    # Demonstrate caching pattern
    print("✓ Caching pattern example:")
    
    class CachedCSSTranslator:
        """CSS translator with caching for better performance"""
        
        def __init__(self, use_html_translator=False):
            if use_html_translator and hasattr(cssselect, 'HTMLTranslator'):
                self.translator = cssselect.HTMLTranslator()
            else:
                self.translator = cssselect.GenericTranslator()
            self.cache = {}
            
        def css_to_xpath(self, css_selector):
            """Convert CSS to XPath with caching"""
            if css_selector not in self.cache:
                self.cache[css_selector] = self.translator.css_to_xpath(css_selector)
            return self.cache[css_selector]
            
        def clear_cache(self):
            """Clear the translation cache"""
            self.cache.clear()
            
        def cache_stats(self):
            """Get cache statistics"""
            return {
                'cache_size': len(self.cache),
                'cached_selectors': list(self.cache.keys())
            }
    
    # Test caching performance
    cached_translator = CachedCSSTranslator()
    test_selector = "div.content p.highlight"
    
    # First call (cache miss)
    start_time = time.time()
    xpath1 = cached_translator.css_to_xpath(test_selector)
    first_call_time = (time.time() - start_time) * 1000
    
    # Second call (cache hit) 
    start_time = time.time()
    xpath2 = cached_translator.css_to_xpath(test_selector)
    second_call_time = (time.time() - start_time) * 1000
    
    print(f"  Cached translator performance:")
    print(f"    First call (cache miss): {first_call_time:.3f}ms")
    print(f"    Second call (cache hit): {second_call_time:.3f}ms")
    print(f"    Cache stats: {cached_translator.cache_stats()}")
    
    # Show integration patterns
    print("✓ Integration patterns:")
    
    integration_patterns = {
        'lxml': 'element.cssselect() uses cssselect internally',
        'PyQuery': 'jQuery-like interface with CSS selector support',
        'Scrapy': 'CSS selector support in item extraction',
        'BeautifulSoup': 'CSS selector methods via SoupSieve (not cssselect)',
        'Custom XML libs': 'Direct XPath integration with cssselect translation'
    }
    
    for library, description in integration_patterns.items():
        print(f"  {library}: {description}")
        
    # Show practical usage patterns
    print("✓ Practical usage patterns:")
    
    usage_examples = [
        "Web scraping: Convert familiar CSS selectors to XPath",
        "XML processing: Use CSS syntax for easier element selection", 
        "Testing: CSS selectors for web element location",
        "Data extraction: Structured document parsing",
        "Template processing: Dynamic content selection"
    ]
    
    for example in usage_examples:
        print(f"  • {example}")
        
    # Best practices summary
    print("✓ Best practices summary:")
    
    best_practices = [
        "Validate CSS selectors early in development",
        "Use appropriate translator type (HTML vs Generic)",
        "Cache translations for frequently used selectors",
        "Test XPath output with your target XML library",
        "Handle parsing errors gracefully in production",
        "Profile performance for complex selector workloads",
        "Document CSS selector patterns for team consistency"
    ]
    
    for practice in best_practices:
        print(f"  • {practice}")
        
    # Show library ecosystem
    print("✓ Related libraries and ecosystem:")
    
    ecosystem_libs = {
        'lxml': 'High-performance XML/HTML processing with CSS selector support',
        'pyquery': 'jQuery-like manipulation with CSS selectors',  
        'scrapy': 'Web scraping framework with CSS selector extraction',
        'beautifulsoup4': 'HTML parsing (uses SoupSieve for CSS selectors)',
        'selenium': 'Web automation with CSS selector element location',
        'requests-html': 'HTML parsing with CSS selector support'
    }
    
    for lib, description in ecosystem_libs.items():
        print(f"  {lib}: {description}")
        
    # Final summary
    print("✓ Package summary:")
    print("  cssselect provides reliable CSS-to-XPath translation with")
    print("  excellent performance and broad compatibility, making it")
    print("  essential for XML/HTML processing applications that prefer")
    print("  CSS selector syntax over XPath expressions.")
    
    print("✓ Performance and integration best practices completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cssselect example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
