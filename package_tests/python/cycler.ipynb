## Example 1: Basic Cycler Creation and Iteration

```python
from cycler import cycler

print("Example 1: Basic Cycler Creation and Iteration")

try:
    print("✓ cycler package information:")
    print(f"  Cycler function: {cycler}")
    print(f"  Used for matplotlib style cycling")
    
    # Create basic cyclers
    print("✓ Basic cycler creation:")
    
    # Color cycler
    color_cycle = cycler('color', ['red', 'green', 'blue', 'orange'])
    print(f"  Color cycler: {color_cycle}")
    print(f"  Cycler type: {type(color_cycle)}")
    
    # Line style cycler
    linestyle_cycle = cycler('linestyle', ['-', '--', '-.', ':'])
    print(f"  Linestyle cycler: {linestyle_cycle}")
    
    # Marker cycler
    marker_cycle = cycler('marker', ['o', 's', '^', 'D'])
    print(f"  Marker cycler: {marker_cycle}")
    
    # Demonstrate cycler iteration
    print("✓ Cycler iteration:")
    
    print("  Color cycle iteration:")
    for i, style_dict in enumerate(color_cycle):
        print(f"    Step {i+1}: {style_dict}")
        if i >= 7:  # Show cycling behavior (more than 4 colors)
            break
            
    print("  Linestyle cycle iteration:")
    for i, style_dict in enumerate(linestyle_cycle):
        print(f"    Step {i+1}: {style_dict}")
        if i >= 5:  # Show cycling behavior
            break
            
    # Show cycler properties
    print("✓ Cycler properties:")
    
    print(f"  Color cycler length: {len(color_cycle)}")
    print(f"  Color cycler keys: {color_cycle.keys}")
    
    # Show cycling behavior with by_key()
    print("✓ Access by key:")
    try:
        if hasattr(color_cycle, 'by_key'):
            by_key_dict = color_cycle.by_key()
            print(f"  By key structure: {by_key_dict}")
        else:
            print("  by_key() method: Available in cycler objects")
    except Exception as e:
        print(f"  by_key() note: {e}")
        raise
        
    # Demonstrate manual cycling
    print("✓ Manual cycling through values:")
    
    colors = ['red', 'green', 'blue']
    for i in range(8):
        current_color = colors[i % len(colors)]
        print(f"  Plot {i+1}: color='{current_color}'")
        
    print("✓ Basic cycler creation and iteration completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: cycler is part of the matplotlib ecosystem")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cycler example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Cycler Composition and Combination

```python
from cycler import cycler

print("Example 2: Cycler Composition and Combination")

try:
    # Demonstrate cycler arithmetic operations
    print("✓ Cycler composition operations:")
    
    # Create base cyclers
    color_cycle = cycler('color', ['red', 'green', 'blue'])
    linestyle_cycle = cycler('linestyle', ['-', '--', '-.'])
    marker_cycle = cycler('marker', ['o', 's', '^'])
    
    print(f"  Base color cycler: {len(color_cycle)} items")
    print(f"  Base linestyle cycler: {len(linestyle_cycle)} items")
    print(f"  Base marker cycler: {len(marker_cycle)} items")
    
    # Addition (concatenation) - requires NON-OVERLAPPING properties
    print("✓ Cycler addition (non-overlapping properties):")
    
    # Create cycler with different property for addition
    combined_cycle = color_cycle + linestyle_cycle + marker_cycle
    print(f"  Combined cycle (color + linestyle + marker): {len(combined_cycle)} items")
    
    print("  Combined cycle properties:")
    for i, style in enumerate(combined_cycle):
        print(f"    {i+1}. {style}")
        
    # Multiplication (Cartesian product)
    print("✓ Cycler multiplication (Cartesian product):")
    
    color_line_combo = color_cycle * linestyle_cycle
    print(f"  Color × Linestyle combinations: {len(color_line_combo)} items")
    
    print("  First 6 combinations:")
    for i, style in enumerate(color_line_combo):
        print(f"    {i+1}. {style}")
        if i >= 5:
            break
            
    # Triple combination
    print("✓ Triple cycler combination:")
    
    full_combo = color_cycle * linestyle_cycle * marker_cycle
    print(f"  Color × Linestyle × Marker: {len(full_combo)} combinations")
    
    print("  First 5 complete styles:")
    for i, style in enumerate(full_combo):
        print(f"    {i+1}. {style}")
        if i >= 4:
            break
            
    # Different length cyclers (using multiplication, not addition)
    print("✓ Different length cycler combinations:")
    
    short_colors = cycler('color', ['red', 'blue'])
    long_lines = cycler('linestyle', ['-', '--', '-.', ':', '-'])
    
    mixed_combo = short_colors * long_lines
    print(f"  Short colors ({len(short_colors)}) × Long lines ({len(long_lines)}): {len(mixed_combo)} total")
    
    print("  Mixed length combinations:")
    for i, style in enumerate(mixed_combo):
        print(f"    {i+1}. {style}")
        if i >= 6:
            break
            
    # Demonstrate cycler slicing
    print("✓ Cycler slicing:")
    
    large_combo = color_cycle * linestyle_cycle
    
    # Get subset of combinations
    subset = large_combo[:4]  # First 4 combinations
    print(f"  First 4 combinations from {len(large_combo)} total:")
    for i, style in enumerate(subset):
        print(f"    {i+1}. {style}")
        
    # Show cycler key information
    print("✓ Combined cycler keys:")
    
    print(f"  Color cycler keys: {list(color_cycle.keys)}")
    print(f"  Combined cycler keys: {list(full_combo.keys)}")
    
    # Demonstrate cycler equality and comparison
    print("✓ Cycler comparison:")
    
    same_colors = cycler('color', ['red', 'green', 'blue'])
    different_colors = cycler('color', ['red', 'green', 'yellow'])
    
    print(f"  Original == Same colors: {color_cycle == same_colors}")
    print(f"  Original == Different colors: {color_cycle == different_colors}")
    
    # Show proper addition usage (different properties only)
    print("✓ Proper cycler addition examples:")
    
    color_prop = cycler('color', ['red', 'green'])
    width_prop = cycler('linewidth', [1, 2])  # Different property name
    alpha_prop = cycler('alpha', [0.8, 0.6])  # Another different property
    
    combined_props = color_prop + width_prop + alpha_prop
    print(f"  Combined different properties: {len(combined_props)} items")
    for i, style in enumerate(combined_props):
        print(f"    {i+1}. {style}")
    
    # Show concatenation of same property using list extension
    print("✓ Extending values within same property:")
    
    # Instead of adding overlapping cyclers, create one with extended values
    extended_colors = cycler('color', ['red', 'green', 'blue', 'orange', 'purple'])
    print(f"  Extended color list: {len(extended_colors)} colors")
    for i, style in enumerate(extended_colors):
        print(f"    {i+1}. {style}")
    
    print("✓ Cycler composition and combination completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: cycler enables complex matplotlib styling patterns")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cycler example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: Matplotlib Integration and Styling

```python
from cycler import cycler
import matplotlib.pyplot as plt
import numpy as np

print("Example 3: Matplotlib Integration and Styling")

try:
    # Demonstrate matplotlib rcParams integration
    print("✓ Matplotlib rcParams integration:")
    
    # Create custom property cycle
    custom_cycle = (cycler('color', ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4']) +
                   cycler('linestyle', ['-', '--', '-.', ':']))
    
    print(f"  Custom cycle: {len(custom_cycle)} style combinations")
    print(f"  Cycle keys: {list(custom_cycle.keys)}")
    
    # Show how cycler integrates with matplotlib
    print("✓ Matplotlib property cycling:")
    
    # Generate sample data
    x = np.linspace(0, 2*np.pi, 100)
    
    print("  Sample plot styling with cycler:")
    print("  plt.rc('axes', prop_cycle=custom_cycle)")
    
    # Show what each plot would get
    for i, style in enumerate(custom_cycle):
        y = np.sin(x + i*np.pi/4)
        print(f"    Plot {i+1}: {style}")
        # In actual matplotlib: plt.plot(x, y) would automatically use these styles
        
    # Demonstrate style property types
    print("✓ Common matplotlib style properties:")
    
    style_properties = {
        'color': ['red', 'green', 'blue', '#FF6B6B'],
        'linestyle': ['-', '--', '-.', ':'],
        'marker': ['o', 's', '^', 'D', 'v', '<', '>', 'p'],
        'linewidth': [1, 2, 1.5, 3],
        'markersize': [6, 8, 10, 12],
        'alpha': [1.0, 0.8, 0.6, 0.4]
    }
    
    for prop, values in style_properties.items():
        sample_cycler = cycler(prop, values[:3])  # Use first 3 values
        print(f"  {prop}: {sample_cycler}")
        
    # Show advanced styling combinations
    print("✓ Advanced styling combinations:")
    
    # Scientific plot styling
    scientific_cycle = (cycler('color', ['black', 'red', 'blue', 'green']) *
                       cycler('linestyle', ['-', '--']) *
                       cycler('marker', ['', 'o']))
    
    print(f"  Scientific style combinations: {len(scientific_cycle)}")
    print("  First 4 scientific styles:")
    for i, style in enumerate(scientific_cycle[:4]):
        print(f"    {i+1}. {style}")
        
    # Presentation-friendly styling
    presentation_cycle = (cycler('color', ['#E74C3C', '#3498DB', '#2ECC71', '#F39C12']) +
                         cycler('linewidth', [2, 2.5, 2, 3]))
    
    print(f"  Presentation style combinations: {len(presentation_cycle)}")
    print("  Presentation styles:")
    for i, style in enumerate(presentation_cycle):
        print(f"    {i+1}. {style}")
        
    # Demonstrate cycler with subplot styling
    print("✓ Subplot styling patterns:")
    
    subplot_styles = [
        "Use same cycler across all subplots for consistency",
        "Create different cyclers for different subplot types",
        "Reset cycler for each subplot: ax.set_prop_cycle()",
        "Manual style assignment for specific control"
    ]
    
    for pattern in subplot_styles:
        print(f"  • {pattern}")
        
    # Show cycler state management
    print("✓ Cycler state and reset:")
    
    demo_cycle = cycler('color', ['red', 'green', 'blue'])
    
    print("  Cycling through states:")
    cycler_iter = iter(demo_cycle)
    
    for i in range(5):  # More than the cycle length
        try:
            style = next(cycler_iter)
            print(f"    Iteration {i+1}: {style}")
        except StopIteration:
            # Cycler exhausted, create new iterator
            cycler_iter = iter(demo_cycle)
            style = next(cycler_iter)
            print(f"    Iteration {i+1} (reset): {style}")
            
    print("  Note: matplotlib handles cycler reset automatically")
    
    print("✓ Matplotlib integration and styling completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: cycler requires matplotlib for full functionality")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cycler example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: Advanced Cycler Patterns and Custom Properties

```python
from cycler import cycler
import numpy as np

print("Example 4: Advanced Cycler Patterns and Custom Properties")

try:
    # Demonstrate complex cycler patterns
    print("✓ Complex cycler patterns:")
    
    # Gradient-based color cycling
    def generate_color_gradient(start_color, end_color, steps):
        """Generate gradient colors between two hex colors"""
        # Simple gradient simulation
        colors = []
        for i in range(steps):
            # This would normally interpolate between colors
            intensity = i / (steps - 1)
            colors.append(f"C{i % 10}")  # Use matplotlib default colors for demo
        return colors
    
    gradient_colors = generate_color_gradient('#FF0000', '#0000FF', 5)
    gradient_cycle = cycler('color', gradient_colors)
    
    print(f"  Gradient color cycle: {len(gradient_colors)} colors")
    for i, style in enumerate(gradient_cycle):
        print(f"    {i+1}. {style}")
        
    # Mathematical property cycling
    print("✓ Mathematical property patterns:")
    
    # Linewidth that varies mathematically
    linewidths = [1 + 2 * np.sin(i * np.pi / 4) for i in range(6)]
    linewidths = [max(0.5, lw) for lw in linewidths]  # Ensure positive
    math_cycle = cycler('linewidth', linewidths)
    
    print(f"  Mathematical linewidth cycle:")
    for i, style in enumerate(math_cycle):
        print(f"    {i+1}. {style}")
        
    # Demonstrate custom property cycling
    print("✓ Custom property cycling:")
    
    # Custom properties (could be used with custom matplotlib artists)
    custom_properties = {
        'transparency': [0.9, 0.7, 0.5, 0.3],
        'hatch_pattern': ['///', '\\\\\\', '|||', '---'],
        'edge_color': ['black', 'gray', 'darkgray', 'lightgray'],
        'fill_style': ['full', 'left', 'right', 'bottom']
    }
    
    # Create cycler by multiplication (all same length)
    custom_cycle = (cycler('transparency', custom_properties['transparency']) *
                   cycler('hatch_pattern', custom_properties['hatch_pattern']) *
                   cycler('edge_color', custom_properties['edge_color']) * 
                   cycler('fill_style', custom_properties['fill_style']))
    
    print(f"  Custom property cycle: {len(custom_cycle)} combinations")
    print("  First 4 custom properties:")
    for i, style in enumerate(custom_cycle):
        print(f"    {i+1}. {style}")
        if i >= 3:
            break
        
    # Conditional cycler creation
    print("✓ Conditional cycler patterns:")
    
    def create_theme_cycler(theme='default'):
        """Create cycler based on theme"""
        themes = {
            'default': {
                'color': ['blue', 'orange', 'green', 'red'],
                'linestyle': ['-', '-', '-', '-']
            },
            'scientific': {
                'color': ['black', 'red', 'blue', 'green'],
                'linestyle': ['-', '--', '-.', ':'],
                'marker': ['', 'o', 's', '^']
            },
            'colorful': {
                'color': ['#FF6B6B', '#4ECDC4', '#45B7D1', '#FFA07A'],
                'linewidth': [2, 2.5, 3, 2.5]
            }
        }
        
        if theme not in themes:
            theme = 'default'
            
        theme_props = themes[theme]
        result_cycler = None
        
        for prop, values in theme_props.items():
            prop_cycler = cycler(prop, values)
            if result_cycler is None:
                result_cycler = prop_cycler
            else:
                # Use multiplication since all theme properties have equal length
                result_cycler *= prop_cycler
                    
        return result_cycler
    
    # Test theme cyclers
    themes = ['default', 'scientific', 'colorful']
    for theme in themes:
        theme_cycler = create_theme_cycler(theme)
        print(f"  {theme.capitalize()} theme: {len(theme_cycler)} styles")
        print(f"    Keys: {list(theme_cycler.keys)}")
        
    # Show cycler introspection
    print("✓ Cycler introspection and analysis:")
    
    analysis_cycler = cycler('color', ['red', 'green']) * cycler('marker', ['o', 's', '^'])
    
    print(f"  Analysis cycler properties:")
    print(f"    Length: {len(analysis_cycler)}")
    print(f"    Keys: {list(analysis_cycler.keys)}")
    
    # Analyze cycler structure
    if hasattr(analysis_cycler, 'by_key'):
        try:
            by_key_data = analysis_cycler.by_key()
            print(f"    By-key structure: {by_key_data}")
        except Exception as e:
            print(f"    By-key analysis: {e}")
            raise
            
    # Show cycler modification patterns
    print("✓ Cycler modification patterns:")
    
    base_cycler = cycler('color', ['red', 'green', 'blue'])
    
    # Instead of trying to add overlapping cyclers, show other modification patterns
    print("  Extending color values directly:")
    extended_colors = ['red', 'green', 'blue', 'orange', 'purple', 'brown']
    extended_cycler = cycler('color', extended_colors)
    print(f"  Extended color cycler: {len(extended_cycler)} colors")
    
    # Subset cycler
    subset = base_cycler[:2]
    print(f"  Subset length: {len(subset)}")
    
    # Combine with different properties (multiplication)
    enhanced = base_cycler * cycler('alpha', [1.0, 0.7, 0.4])
    print(f"  Enhanced with alpha: {len(enhanced)} combinations")
    
    # Show addition with non-overlapping properties
    color_and_width = base_cycler + cycler('linewidth', [1, 2, 3])
    print(f"  Color + linewidth (non-overlapping): {len(color_and_width)} styles")
    
    # Show practical cycler building
    print("✓ Practical cycler building:")
    
    # Build a comprehensive publication-ready cycler
    pub_colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']
    pub_markers = ['o', 's', '^', 'D']
    pub_lines = ['-', '--', '-.', ':']
    
    publication_cycler = (cycler('color', pub_colors) * 
                         cycler('marker', pub_markers) * 
                         cycler('linestyle', pub_lines))
    
    print(f"  Publication cycler: {len(publication_cycler)} combinations")
    print("  First 3 publication styles:")
    for i, style in enumerate(publication_cycler):
        print(f"    {i+1}. {style}")
        if i >= 2:
            break
    
    print("✓ Advanced cycler patterns and custom properties completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: cycler supports complex styling patterns")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cycler example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: Performance and Best Practices

```python
from cycler import cycler
import time

print("Example 5: Performance and Best Practices")

try:
    # Best practice: Pre-define cyclers for reuse
    print("✓ Performance and best practices:")
    
    base_colors = cycler('color', ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728'])
    base_styles = cycler('linestyle', ['-', '--', '-.', ':'])
    
    # Performance tip: Use multiplication for combining equal-length cyclers
    efficient_cycler = base_colors * base_styles
    print(f"  Efficient cycler length: {len(efficient_cycler)}")
    
    # Memory efficient: Iterate without storing all combinations
    print("✓ Memory efficient iteration:")
    for i, style_dict in enumerate(efficient_cycler):
        if i < 3:  # Show first 3 only
            print(f"    Style {i}: {style_dict}")
    
    # Best practice: Validate cycler before use
    def validate_cycler(cyc):
        """Validate a cycler has expected properties"""
        if hasattr(cyc, 'keys'):
            return len(cyc.keys) > 0
        return False
    
    is_valid = validate_cycler(efficient_cycler)
    print(f"  Cycler validation passed: {is_valid}")
    
    # Performance comparison: Non-overlapping addition vs multiplication
    print("✓ Performance comparison:")

    start_time = time.time()
    # Addition with non-overlapping properties (different property names)
    addition_cycler = base_colors + cycler('linewidth', [1, 2, 3, 4])
    addition_time = time.time() - start_time

    start_time = time.time()
    multiplication_cycler = base_colors * base_styles
    multiplication_time = time.time() - start_time

    print(f"  Addition operation time: {addition_time:.6f}s")
    print(f"  Multiplication operation time: {multiplication_time:.6f}s")
    print(f"  Addition cycler length: {len(addition_cycler)}")
    print(f"  Multiplication cycler length: {len(multiplication_cycler)}")
    
    # Best practices for large cyclers
    print("✓ Large cycler best practices:")
    
    # Create large cycler efficiently
    large_colors = [f'C{i}' for i in range(10)]
    large_styles = ['-', '--', '-.', ':', '-', '--', '-.', ':', '-', '--']  # Same length
    
    # Use multiplication for comprehensive combinations
    large_cycler = cycler('color', large_colors) * cycler('linestyle', large_styles)
    print(f"  Large cycler combinations: {len(large_cycler)}")
    
    # Best practice: Use slicing for subsets
    subset_cycler = large_cycler[:10]
    print(f"  Subset cycler: {len(subset_cycler)} styles")
    
    # Memory usage tip: Use generators for very large cyclers
    print("✓ Memory optimization tips:")
    
    def cycler_generator(cycler_obj, max_items=None):
        """Generator for memory-efficient cycler iteration"""
        count = 0
        for item in cycler_obj:
            if max_items and count >= max_items:
                break
            yield item
            count += 1
    
    # Demonstrate generator usage
    gen_count = 0
    for style in cycler_generator(large_cycler, 5):
        print(f"    Generated style {gen_count + 1}: {len(str(style))} chars")
        gen_count += 1
        
    # Best practice: Cycler reuse and caching
    print("✓ Cycler reuse and caching:")
    
    # Cache commonly used cyclers (fix the unhashable list issue)
    _cycler_cache = {}
    
    def get_cached_cycler(name, **kwargs):
        """Get or create cached cycler"""
        # Convert lists to tuples for hashing
        hashable_kwargs = {}
        for key, value in kwargs.items():
            if isinstance(value, list):
                hashable_kwargs[key] = tuple(value)
            else:
                hashable_kwargs[key] = value
                
        cache_key = (name, tuple(sorted(hashable_kwargs.items())))
        
        if cache_key not in _cycler_cache:
            if name == 'basic':
                colors = kwargs.get('colors', ['red', 'blue'])
                _cycler_cache[cache_key] = cycler('color', colors)
            elif name == 'scientific':
                _cycler_cache[cache_key] = cycler('color', ['black']) * cycler('marker', ['o', 's'])
        return _cycler_cache[cache_key]
    
    # Test cached cyclers
    cached1 = get_cached_cycler('basic', colors=['red', 'green', 'blue'])
    cached2 = get_cached_cycler('scientific')
    
    print(f"  Cached basic cycler: {len(cached1)} items")
    print(f"  Cached scientific cycler: {len(cached2)} items")
    print(f"  Cache size: {len(_cycler_cache)} cyclers")
    
    # Performance tip: Avoid recreating cyclers in loops
    print("✓ Loop optimization:")
    
    # Good: Create cycler once
    plot_cycler = cycler('color', ['red', 'green', 'blue'])
    
    print("  Best practice: Create cycler outside loop")
    print(f"  Plot cycler ready: {len(plot_cycler)} styles")
    
    # Show cycler iteration patterns
    cycler_iter = iter(plot_cycler)
    for i in range(5):
        try:
            style = next(cycler_iter)
            print(f"    Plot {i+1}: {style}")
        except StopIteration:
            # Reset iterator when exhausted
            cycler_iter = iter(plot_cycler)
            style = next(cycler_iter)
            print(f"    Plot {i+1} (reset): {style}")
    
    # Show best practices summary
    print("✓ Best practices summary:")
    best_practices = [
        "Use multiplication (*) for Cartesian products",
        "Use addition (+) only for non-overlapping properties", 
        "Cache frequently used cyclers to avoid recreation",
        "Use slicing for cycler subsets",
        "Create cyclers outside loops for performance",
        "Use generators for memory-efficient large cycler iteration"
    ]
    
    for i, practice in enumerate(best_practices, 1):
        print(f"  {i}. {practice}")
    
    print("✓ Performance and best practices completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: cycler is essential for matplotlib styling")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"cycler example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
