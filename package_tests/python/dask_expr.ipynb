## Example 1: dask_expr functionality test 1

```python
# Example 1: Expression trees in dask.dataframe (dask-expr backend)
import dask.dataframe as dd
import pandas as pd
import numpy as np

print("Example 1: Dask-Expr Backend - Expression Trees\n")

# Create sample data
print("Creating sample DataFrame:")
pdf = pd.DataFrame({
    'x': np.random.random(1000),
    'y': np.random.random(1000),
    'z': np.random.random(1000),
    'name': np.random.choice(['Alice', 'Bob', 'Charlie'], 1000)
})
df = dd.from_pandas(pdf, npartitions=4)
print(f"  DataFrame type: {type(df)}")
print(f"  Columns: {list(df.columns)}")
print(f"  Partitions: {df.npartitions}\n")

# Show expression tree for mean operation
print("Expression tree for df.x.mean():")
mean_expr = df.x.mean()
print(mean_expr.pprint())

# Compute result
print("\nComputed result:")
result = mean_expr.compute()
print(f"  Mean of x: {result:.6f}")

print("\nDask-expr backend represents operations as expression trees")
```

## Example 2: dask_expr functionality test 2

```python
# Example 2: Query optimization with dask-expr backend
import dask.dataframe as dd
import pandas as pd
import numpy as np

print("Example 2: Query Optimization - Unoptimized vs Optimized\n")

# Create dataset
pdf = pd.DataFrame({
    'a': np.random.random(5000),
    'b': np.random.random(5000),
    'c': np.random.random(5000),
    'category': np.random.choice(['X', 'Y', 'Z'], 5000)
})
df = dd.from_pandas(pdf, npartitions=10)
print(f"Created DataFrame with {len(pdf)} rows, {df.npartitions} partitions\n")

# Create expression
expr = df.a.sum()

# Show unoptimized expression
print("Unoptimized expression:")
print(expr.pprint())

# Show optimized expression
print("\nOptimized expression:")
optimized = expr.optimize()
print(optimized.pprint())

# Compute results
print("\nComputed sum of column 'a':")
result = expr.compute()
print(f"  Sum: {result:.2f}")

print("\nOptimization transforms expression trees for better performance")
```

## Example 3: dask_expr functionality test 3

```python
# Example 3: GroupBy expression optimization
import dask.dataframe as dd
import pandas as pd
import numpy as np

print("Example 3: GroupBy Expression Optimization\n")

# Create dataset with groups
pdf = pd.DataFrame({
    'user_id': np.random.randint(100, 200, 10000),
    'amount': np.random.uniform(10, 500, 10000),
    'quantity': np.random.randint(1, 10, 10000),
    'region': np.random.choice(['North', 'South', 'East', 'West'], 10000)
})
df = dd.from_pandas(pdf, npartitions=8)
print(f"Dataset: {len(pdf)} rows, {df.npartitions} partitions")
print(f"Columns: {list(df.columns)}\n")

# GroupBy operation
print("GroupBy expression: df.groupby('region').amount.mean()")
grouped_expr = df.groupby('region').amount.mean()

# Show expression tree
print("\nExpression tree:")
print(grouped_expr.pprint())

# Compute result
print("\nComputed results by region:")
result = grouped_expr.compute()
print(result)

print("\nDask-expr optimizes GroupBy for parallel execution")
```

## Example 4: dask_expr functionality test 4

```python
# Example 4: Column projection and filtering optimization
import dask.dataframe as dd
import pandas as pd
import numpy as np

print("Example 4: Projection and Filter Optimization\n")

# Create dataset with many columns
pdf = pd.DataFrame({
    'col_a': np.random.random(8000),
    'col_b': np.random.random(8000),
    'col_c': np.random.random(8000),
    'col_d': np.random.random(8000),
    'col_e': np.random.random(8000),
    'category': np.random.choice(['Type1', 'Type2', 'Type3'], 8000)
})
df = dd.from_pandas(pdf, npartitions=8)
print(f"Full dataset: {list(df.columns)}")
print(f"Partitions: {df.npartitions}\n")

# Select columns and filter
print("Operation: df[['col_a', 'col_b']][df.col_a > 0.5].col_a.sum()")
filtered_expr = df[['col_a', 'col_b']][df.col_a > 0.5].col_a.sum()

# Show expression tree
print("\nExpression tree:")
print(filtered_expr.pprint())

# Show optimized version
print("\nOptimized expression:")
print(filtered_expr.optimize().pprint())

# Compute
print("\nComputed result:")
result = filtered_expr.compute()
print(f"  Sum of col_a (where col_a > 0.5): {result:.2f}")

print("\nProjection optimization reads only necessary columns")
```

## Example 5: dask_expr functionality test 5

```python
# Example 5: Complex multi-operation expression trees
import dask.dataframe as dd
import pandas as pd
import numpy as np

print("Example 5: Complex Expression Trees\n")

# Create dataset
pdf = pd.DataFrame({
    'value': np.random.random(6000),
    'weight': np.random.uniform(0.5, 2.0, 6000),
    'group': np.random.choice(['A', 'B', 'C', 'D'], 6000),
    'flag': np.random.choice([True, False], 6000)
})
df = dd.from_pandas(pdf, npartitions=6)
print(f"Dataset: {len(pdf)} rows, {df.npartitions} partitions\n")

# Create complex expression
print("Complex operation: weighted average by group")
df['weighted_value'] = df.value * df.weight
result_expr = df.groupby('group').weighted_value.sum() / df.groupby('group').weight.sum()

# Show expression tree
print("\nExpression tree:")
print(result_expr.pprint())

# Compute result
print("\nWeighted average by group:")
result = result_expr.compute()
print(result)

# Additional complex aggregation
print("\n\nMultiple aggregations:")
agg_expr = df.groupby('group').agg({'value': ['mean', 'sum'], 'weight': 'mean'})
print("\nExpression tree for multiple aggregations:")
print(agg_expr.pprint())

print("\nComputed aggregations:")
agg_result = agg_expr.compute()
print(agg_result)

print("\nDask-expr optimizes complex multi-operation expressions")
```
