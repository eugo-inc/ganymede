## Example 3: Job Management and Scheduling

Demonstrate job creation, scheduling, and workflow management.

```python
# Example 3: Job Workflows and Task Orchestration

from databricks.sdk.service import jobs, compute
from databricks.sdk.service.jobs import (
    JobSettings, Task, NotebookTask, SparkPythonTask, PythonWheelTask,
    TaskDependency, CronSchedule, JobCluster, JobEmailNotifications
)
from databricks.sdk.service.compute import Library, PythonPyPiLibrary, ClusterSpec

print("Example 3: Job workflows and multi-task orchestration")
print("\n=== Creating Multi-Task Job Workflow ===")

# Create comprehensive ETL job with multiple task types
etl_job = JobSettings(
    name="customer-360-etl-pipeline",
    tasks=[
        # Task 1: Data extraction notebook
        Task(
            task_key="extract_raw_data",
            notebook_task=NotebookTask(
                notebook_path="/Workspace/ETL/Extraction/extract_customer_data",
                base_parameters={
                    "env": "production",
                    "source_database": "raw_db",
                    "target_path": "/mnt/bronze/customers"
                }
            ),
            existing_cluster_id="cluster-001",
            timeout_seconds=3600,
            max_retries=2,
            retry_on_timeout=True
        ),
        
        # Task 2: Data validation Python script
        Task(
            task_key="validate_data_quality",
            depends_on=[TaskDependency(task_key="extract_raw_data")],
            spark_python_task=SparkPythonTask(
                python_file="dbfs:/FileStore/scripts/data_quality_checks.py",
                parameters=[
                    "--input-path", "/mnt/bronze/customers",
                    "--validation-rules", "dbfs:/config/validation_rules.json",
                    "--output-report", "/mnt/quality_reports/customers"
                ]
            ),
            existing_cluster_id="cluster-001",
            timeout_seconds=1800
        ),
        
        # Task 3: Data transformation
        Task(
            task_key="transform_to_silver",
            depends_on=[TaskDependency(task_key="validate_data_quality")],
            notebook_task=NotebookTask(
                notebook_path="/Workspace/ETL/Transformation/bronze_to_silver",
                base_parameters={
                    "input_path": "/mnt/bronze/customers",
                    "output_path": "/mnt/silver/customers",
                    "apply_pii_masking": "true"
                }
            ),
            existing_cluster_id="cluster-001"
        ),
        
        # Task 4: Aggregate to gold layer
        Task(
            task_key="aggregate_to_gold",
            depends_on=[TaskDependency(task_key="transform_to_silver")],
            notebook_task=NotebookTask(
                notebook_path="/Workspace/ETL/Aggregation/silver_to_gold",
                base_parameters={
                    "input_path": "/mnt/silver/customers",
                    "output_path": "/mnt/gold/customer_360"
                }
            ),
            existing_cluster_id="cluster-001"
        )
    ],
    schedule=CronSchedule(
        quartz_cron_expression="0 0 2 * * ?",  # Daily at 2 AM
        timezone_id="America/Los_Angeles",
        pause_status=jobs.PauseStatus.UNPAUSED
    ),
    max_concurrent_runs=1,
    email_notifications=JobEmailNotifications(
        on_start=["data-eng-team@company.com"],
        on_success=["data-eng-team@company.com"],
        on_failure=["data-eng-team@company.com", "oncall@company.com"],
        no_alert_for_skipped_runs=True
    ),
    timeout_seconds=14400,  # 4 hours
    tags={
        "pipeline": "customer-360",
        "layer": "all",
        "sla": "critical"
    }
)

print(f"✅ ETL Pipeline Job Configuration:")
print(f"   Job name: {etl_job.name}")
print(f"   Total tasks: {len(etl_job.tasks)}")
print(f"   Schedule: {etl_job.schedule.quartz_cron_expression}")
print(f"   Timezone: {etl_job.schedule.timezone_id}")
print(f"   Max concurrent runs: {etl_job.max_concurrent_runs}")
print(f"   Tags: {etl_job.tags}")

# Show task dependency graph
print(f"\n=== Task Dependency Graph ===")
for task in etl_job.tasks:
    deps = task.depends_on if task.depends_on else []
    dep_str = ", ".join([d.task_key for d in deps]) if deps else "None"
    print(f"   {task.task_key}")
    print(f"      └─ Depends on: {dep_str}")

# Create ML training job with Python wheel and job cluster
ml_training_job = JobSettings(
    name="customer-churn-model-training",
    tasks=[
        Task(
            task_key="train_model",
            python_wheel_task=PythonWheelTask(
                package_name="churn_model_training",
                entry_point="train",
                parameters=[
                    "--data-path", "/mnt/gold/customer_360",
                    "--model-output", "/mnt/models/churn",
                    "--experiment-name", "/Experiments/ChurnPrediction"
                ]
            ),
            job_cluster_key="ml_cluster",
            libraries=[
                Library(pypi=PythonPyPiLibrary(package="scikit-learn==1.3.0")),
                Library(pypi=PythonPyPiLibrary(package="xgboost==2.0.0")),
                Library(pypi=PythonPyPiLibrary(package="mlflow==2.8.0"))
            ]
        ),
        Task(
            task_key="evaluate_model",
            depends_on=[TaskDependency(task_key="train_model")],
            notebook_task=NotebookTask(
                notebook_path="/Workspace/ML/Evaluation/model_evaluation",
                base_parameters={
                    "model_path": "/mnt/models/churn",
                    "test_data": "/mnt/gold/customer_360_test"
                }
            ),
            job_cluster_key="ml_cluster"
        )
    ],
    job_clusters=[
        JobCluster(
            job_cluster_key="ml_cluster",
            new_cluster=ClusterSpec(
                spark_version="13.3.x-cpu-ml-scala2.12",
                node_type_id="i3.xlarge",
                num_workers=4,
                spark_conf={
                    "spark.databricks.delta.preview.enabled": "true"
                }
            )
        )
    ],
    schedule=CronSchedule(
        quartz_cron_expression="0 0 3 * * ?",  # Daily at 3 AM
        timezone_id="UTC"
    ),
    tags={
        "model": "churn-prediction",
        "ml-stage": "training"
    }
)

print(f"\n✅ ML Training Job Configuration:")
print(f"   Job name: {ml_training_job.name}")
print(f"   Tasks: {len(ml_training_job.tasks)}")
print(f"   Job clusters: {len(ml_training_job.job_clusters)}")
print(f"   Cluster: {ml_training_job.job_clusters[0].new_cluster.spark_version}")
print(f"   Libraries per task: {len(ml_training_job.tasks[0].libraries)} ML libraries")
for lib in ml_training_job.tasks[0].libraries:
    print(f"      • {lib.pypi.package}")

# Demonstrate task types
print(f"\n=== Available Task Types ===")
task_types = [
    "NotebookTask - Execute Databricks notebooks",
    "SparkPythonTask - Run Python scripts",
    "PythonWheelTask - Execute packaged Python applications",
    "SparkJarTask - Run JVM-based applications",
    "SparkSubmitTask - Submit Spark jobs",
    "DbtTask - Execute dbt transformations",
    "SqlTask - Run SQL queries",
    "DashboardTask - Refresh dashboards",
    "ConditionTask - Conditional branching"
]

for task_type in task_types:
    print(f"   • {task_type}")

# Demonstrate scheduling capabilities
print(f"\n=== Scheduling Capabilities ===")
scheduling_features = [
    "Cron expressions with timezone support",
    "Pause/unpause scheduled jobs",
    "Max concurrent runs limit",
    "Timeout configurations per task",
    "Retry policies with exponential backoff",
    "Email notifications on job events",
    "Webhook integrations",
    "Job dependencies and triggers"
]

for feature in scheduling_features:
    print(f"   ✓ {feature}")

print(f"\n✅ Job workflows created successfully")
print("Example 3 completed")
```

## Example 4: SQL Warehouse and Query Management

```python
# Example 4: SQL Warehouse and Query Management

from databricks.sdk.service import sql
from databricks.sdk.service.sql import (
    EndpointInfo, CreateWarehouseRequestWarehouseType,
    SpotInstancePolicy, Channel, QueryParameter, Query
)

print("Example 4: SQL warehouse configuration and query management")
print("\n=== SQL Warehouse Configurations ===")

# Create serverless SQL warehouse specification
serverless_warehouse = EndpointInfo(
    name="analytics-serverless-pro",
    cluster_size="2X-Small",
    min_num_clusters=1,
    max_num_clusters=10,
    auto_stop_mins=20,
    enable_photon=True,
    enable_serverless_compute=True,
    warehouse_type=CreateWarehouseRequestWarehouseType.PRO,
    channel=Channel(name=sql.ChannelName.CHANNEL_NAME_CURRENT),
    tags=sql.EndpointTags(
        custom_tags=[
            sql.EndpointTagPair(key="team", value="analytics"),
            sql.EndpointTagPair(key="environment", value="production"),
            sql.EndpointTagPair(key="billing", value="shared")
        ]
    )
)

print(f"✅ Serverless SQL Warehouse:")
print(f"   Name: {serverless_warehouse.name}")
print(f"   Size: {serverless_warehouse.cluster_size}")
print(f"   Serverless: {serverless_warehouse.enable_serverless_compute}")
print(f"   Photon: {serverless_warehouse.enable_photon}")
print(f"   Type: {serverless_warehouse.warehouse_type}")
print(f"   Auto-scaling: {serverless_warehouse.min_num_clusters}-{serverless_warehouse.max_num_clusters}")

# Create classic SQL warehouse with spot instances
classic_warehouse = EndpointInfo(
    name="dev-spot-warehouse",
    cluster_size="Small",
    min_num_clusters=1,
    max_num_clusters=5,
    auto_stop_mins=60,
    enable_photon=True,
    enable_serverless_compute=False,
    warehouse_type=CreateWarehouseRequestWarehouseType.CLASSIC,
    spot_instance_policy=SpotInstancePolicy.COST_OPTIMIZED,
    tags=sql.EndpointTags(
        custom_tags=[
            sql.EndpointTagPair(key="team", value="engineering"),
            sql.EndpointTagPair(key="environment", value="development"),
            sql.EndpointTagPair(key="cost-optimization", value="spot")
        ]
    )
)

print(f"\n✅ Classic SQL Warehouse:")
print(f"   Name: {classic_warehouse.name}")
print(f"   Size: {classic_warehouse.cluster_size}")
print(f"   Spot policy: {classic_warehouse.spot_instance_policy}")
print(f"   Auto-stop: {classic_warehouse.auto_stop_mins} minutes")

# Create parameterized analytical query
revenue_query = Query(
    query_text="""
    SELECT 
        DATE_TRUNC('month', order_date) as period,
        region,
        product_category,
        COUNT(DISTINCT customer_id) as unique_customers,
        COUNT(*) as order_count,
        SUM(order_value) as total_revenue,
        AVG(order_value) as avg_order_value
    FROM sales_orders
    WHERE order_date >= :start_date
      AND order_date < :end_date
      AND region IN (:regions)
    GROUP BY 1, 2, 3
    ORDER BY period DESC, total_revenue DESC
    """,
    display_name="revenue-analysis-by-region",
    description="Flexible revenue analysis with date range and regional filtering",
    parameters=[
        QueryParameter(
            name="start_date",
            title="Start Date",
            date_value=sql.DateValue(date_value="2024-01-01")
        ),
        QueryParameter(
            name="end_date",
            title="End Date",
            date_value=sql.DateValue(date_value="2024-12-31")
        ),
        QueryParameter(
            name="regions",
            title="Regions",
            text_value=sql.TextValue(value="North America,Europe,Asia")
        )
    ]
)

print(f"\n✅ Parameterized Query:")
print(f"   Name: {revenue_query.display_name}")
print(f"   Parameters: {len(revenue_query.parameters)}")
for param in revenue_query.parameters:
    print(f"      • {param.title} ({param.name})")

# Create customer segmentation query
segmentation_query = Query(
    query_text="""
    WITH customer_metrics AS (
        SELECT
            customer_id,
            COUNT(DISTINCT order_id) as order_count,
            SUM(order_value) as lifetime_value,
            MAX(order_date) as last_order_date,
            MIN(order_date) as first_order_date,
            DATEDIFF(CURRENT_DATE(), MAX(order_date)) as days_since_last_order
        FROM orders
        WHERE order_date >= DATE_SUB(CURRENT_DATE(), 365)
        GROUP BY customer_id
    ),
    rfm_segments AS (
        SELECT
            *,
            CASE
                WHEN days_since_last_order <= 30 AND order_count >= 10 THEN 'Champion'
                WHEN days_since_last_order <= 60 AND order_count >= 5 THEN 'Loyal'
                WHEN lifetime_value >= 1000 THEN 'High Value'
                WHEN days_since_last_order > 180 THEN 'At Risk'
                ELSE 'Regular'
            END as segment
        FROM customer_metrics
    )
    SELECT
        segment,
        COUNT(*) as customer_count,
        AVG(lifetime_value) as avg_ltv,
        AVG(order_count) as avg_orders,
        AVG(days_since_last_order) as avg_recency
    FROM rfm_segments
    GROUP BY segment
    ORDER BY avg_ltv DESC
    """,
    display_name="customer-rfm-segmentation",
    description="RFM analysis for customer segmentation and targeting"
)

print(f"\n✅ Customer Segmentation Query:")
print(f"   Name: {segmentation_query.display_name}")
print(f"   Query type: RFM Analysis")
print(f"   Description: {segmentation_query.description}")

# Demonstrate query parameter types
print(f"\n=== Query Parameter Types ===")
param_types = [
    "text_value - String parameters (TextValue)",
    "date_value - Date parameters (DateValue)",
    "numeric_value - Numeric parameters (NumericValue)",
    "enum_value - Enumerated choices (EnumValue)",
    "date_range_value - Date range parameters (DateRangeValue)",
    "query_backed_value - Dynamic values from queries (QueryBackedValue)"
]

for ptype in param_types:
    print(f"   • {ptype}")

# Demonstrate warehouse sizes
print(f"\n=== Available Warehouse Sizes ===")
warehouse_sizes = [
    "2X-Small - Development and testing",
    "X-Small - Light workloads",
    "Small - Standard analytics",
    "Medium - Regular BI workloads",
    "Large - Heavy analytical workloads",
    "X-Large - Very large datasets",
    "2X-Large - Enterprise-scale analytics",
    "3X-Large - Maximum performance",
    "4X-Large - Extreme workloads"
]

for size in warehouse_sizes:
    print(f"   • {size}")

# Demonstrate warehouse types
print(f"\n=== Warehouse Types ===")
warehouse_types = [
    f"{CreateWarehouseRequestWarehouseType.PRO} - Enhanced performance and features",
    f"{CreateWarehouseRequestWarehouseType.CLASSIC} - Standard SQL warehouse",
    "SERVERLESS - Instant startup, auto-scaling compute"
]

for wtype in warehouse_types:
    print(f"   • {wtype}")

# SQL Analytics capabilities
print(f"\n=== SQL Analytics Capabilities ===")
capabilities = [
    "Serverless and classic warehouse types",
    "Auto-scaling based on query load",
    "Photon engine for accelerated queries",
    "Spot instance optimization for cost savings",
    "Query result caching",
    "Parameterized queries for reusability",
    "Query history and performance monitoring",
    "Dashboard and visualization support",
    "Multiple concurrent users",
    "Row and column-level security"
]

for capability in capabilities:
    print(f"   ✓ {capability}")

print(f"\n✅ SQL warehouse and query configurations created successfully")
print("Example 4 completed")
```

```python
# Example 5: Unity Catalog Data Governance

from databricks.sdk.service import catalog
from databricks.sdk.service.catalog import (
    CatalogInfo, SchemaInfo, TableInfo,
    TableType, ColumnInfo, ColumnMask,
    VolumeInfo, VolumeType,
    Privilege, SecurableType
)

print("Example 5: Unity Catalog data governance and security")
print("\n=== Unity Catalog Hierarchy ===")

# Create production catalog
prod_catalog = CatalogInfo(
    name="production",
    comment="Production data catalog with full governance",
    properties={
        "department": "data_platform",
        "owner": "data-governance-team",
        "data_classification": "company_confidential",
        "backup_enabled": "true",
        "retention_days": "2555"  # 7 years
    },
    storage_root="s3://company-prod-catalog-root/"
)

print(f"✅ Production Catalog:")
print(f"   Name: {prod_catalog.name}")
print(f"   Storage: {prod_catalog.storage_root}")
print(f"   Properties: {len(prod_catalog.properties)} governance properties")
for key, value in prod_catalog.properties.items():
    print(f"      • {key}: {value}")

# Create schema for customer data with PII
customer_schema = SchemaInfo(
    name="customer_data",
    catalog_name="production",
    full_name="production.customer_data",
    comment="Customer data schema with PII protection and access controls",
    properties={
        "contains_pii": "true",
        "data_retention_policy": "7_years",
        "encryption_at_rest": "true",
        "access_level": "restricted",
        "compliance": "GDPR,CCPA"
    }
)

print(f"\n✅ Customer Data Schema:")
print(f"   Full name: {customer_schema.full_name}")
print(f"   Contains PII: {customer_schema.properties['contains_pii']}")
print(f"   Compliance: {customer_schema.properties['compliance']}")

# Create governed table with column-level security
customers_table = TableInfo(
    name="customers",
    catalog_name="production",
    schema_name="customer_data",
    full_name="production.customer_data.customers",
    table_type=TableType.MANAGED,
    data_source_format=catalog.DataSourceFormat.DELTA,
    comment="Customer master table with row and column-level security",
    columns=[
        ColumnInfo(
            name="customer_id",
            type_text="BIGINT",
            type_name=catalog.ColumnTypeName.LONG,
            position=0,
            nullable=False,
            comment="Unique customer identifier (primary key)"
        ),
        ColumnInfo(
            name="email",
            type_text="STRING",
            type_name=catalog.ColumnTypeName.STRING,
            position=1,
            nullable=False,
            comment="Customer email address - PII",
            mask=ColumnMask(function_name="mask_email")
        ),
        ColumnInfo(
            name="phone",
            type_text="STRING",
            type_name=catalog.ColumnTypeName.STRING,
            position=2,
            nullable=True,
            comment="Customer phone number - PII",
            mask=ColumnMask(function_name="mask_phone")
        ),
        ColumnInfo(
            name="full_name",
            type_text="STRING",
            type_name=catalog.ColumnTypeName.STRING,
            position=3,
            nullable=False,
            comment="Customer full name - PII",
            mask=ColumnMask(function_name="mask_name")
        ),
        ColumnInfo(
            name="date_of_birth",
            type_text="DATE",
            type_name=catalog.ColumnTypeName.DATE,
            position=4,
            nullable=True,
            comment="Customer date of birth - PII",
            mask=ColumnMask(function_name="mask_dob")
        ),
        ColumnInfo(
            name="address",
            type_text="STRING",
            type_name=catalog.ColumnTypeName.STRING,
            position=5,
            nullable=True,
            comment="Customer address - PII",
            mask=ColumnMask(function_name="mask_address")
        ),
        ColumnInfo(
            name="account_status",
            type_text="STRING",
            type_name=catalog.ColumnTypeName.STRING,
            position=6,
            nullable=False,
            comment="Account status (active, suspended, closed)"
        ),
        ColumnInfo(
            name="created_timestamp",
            type_text="TIMESTAMP",
            type_name=catalog.ColumnTypeName.TIMESTAMP,
            position=7,
            nullable=False,
            comment="Account creation timestamp"
        ),
        ColumnInfo(
            name="updated_timestamp",
            type_text="TIMESTAMP",
            type_name=catalog.ColumnTypeName.TIMESTAMP,
            position=8,
            nullable=False,
            comment="Last update timestamp"
        )
    ],
    properties={
        "delta.enableChangeDataFeed": "true",
        "delta.autoOptimize.optimizeWrite": "true",
        "delta.autoOptimize.autoCompact": "true",
        "delta.deletedFileRetentionDuration": "interval 30 days",
        "data_classification": "PII"
    }
)

print(f"\n✅ Governed Table:")
print(f"   Table: {customers_table.full_name}")
print(f"   Type: {customers_table.table_type}")
print(f"   Format: {customers_table.data_source_format}")
print(f"   Columns: {len(customers_table.columns)}")

# Show masked columns
masked_columns = [col for col in customers_table.columns if col.mask]
print(f"   Masked columns ({len(masked_columns)}):")
for col in masked_columns:
    print(f"      • {col.name}: {col.mask.function_name}")

# Show table properties
print(f"   Delta properties:")
for key, value in customers_table.properties.items():
    print(f"      • {key}: {value}")

# Create analytics schema without PII
analytics_schema = SchemaInfo(
    name="analytics",
    catalog_name="production",
    full_name="production.analytics",
    comment="Analytics aggregated data - no PII",
    properties={
        "contains_pii": "false",
        "data_retention_policy": "2_years",
        "access_level": "general"
    }
)

print(f"\n✅ Analytics Schema:")
print(f"   Full name: {analytics_schema.full_name}")
print(f"   Contains PII: {analytics_schema.properties['contains_pii']}")
print(f"   Access level: {analytics_schema.properties['access_level']}")

# Create volume for unstructured data
ml_volume = VolumeInfo(
    catalog_name="production",
    schema_name="ml_models",
    name="model_artifacts",
    full_name="production.ml_models.model_artifacts",
    volume_type=VolumeType.MANAGED,
    comment="ML model artifacts and checkpoints storage"
)

print(f"\n✅ Unity Catalog Volume:")
print(f"   Volume: {ml_volume.full_name}")
print(f"   Type: {ml_volume.volume_type}")
print(f"   Purpose: {ml_volume.comment}")

# Demonstrate governance features
print(f"\n=== Unity Catalog Governance Features ===")
governance_features = [
    "Catalog-level access control",
    "Schema-level permissions",
    "Table and column-level security",
    "Dynamic data masking for PII",
    "Row-level security filters",
    "Audit logging and lineage tracking",
    "Data classification tags",
    "Retention policies and compliance",
    "Cross-cloud data sharing",
    "Volume management for unstructured data"
]

for feature in governance_features:
    print(f"   ✓ {feature}")

# Demonstrate privilege types
print(f"\n=== Available Privileges ===")
privilege_types = [
    Privilege.SELECT,
    Privilege.MODIFY,
    Privilege.CREATE,
    Privilege.USAGE,
    Privilege.READ_VOLUME,
    Privilege.WRITE_VOLUME,
    Privilege.EXECUTE
]

print(f"Securable privileges ({len(privilege_types)}):")
for priv in privilege_types:
    print(f"   • {priv}")

# Demonstrate securable types
print(f"\n=== Securable Types ===")
securable_types = [
    SecurableType.CATALOG,
    SecurableType.SCHEMA,
    SecurableType.TABLE,
    SecurableType.VOLUME,
    SecurableType.FUNCTION
]

print(f"Unity Catalog securables ({len(securable_types)}):")
for sec_type in securable_types:
    print(f"   • {sec_type}")

# Data governance best practices
print(f"\n=== Data Governance Best Practices ===")
best_practices = [
    "Implement least privilege access model",
    "Use dynamic masking for sensitive data",
    "Enable audit logging for compliance",
    "Tag all data assets with classifications",
    "Maintain data lineage for impact analysis",
    "Regular access reviews and certifications",
    "Automated retention policy enforcement",
    "Cross-functional data governance team"
]

for practice in best_practices:
    print(f"   • {practice}")

print(f"\n✅ Unity Catalog governance configuration created successfully")
print("Example 5 completed")
```
