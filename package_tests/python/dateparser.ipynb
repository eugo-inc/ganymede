```python
import dateparser
from datetime import datetime
import re

# Example 4: Custom Date Formats and Validation
print("Example 4: Custom Date Formats and Validation")

try:
    # Test custom date format parsing
    print("✓ Testing custom date formats:")
    
    # Various date formats from different systems/regions
    custom_formats = [
        ("20240115", "YYYYMMDD format"),
        ("15-01-2024", "DD-MM-YYYY format"),
        ("01/15/24", "MM/DD/YY format"),
        ("2024.01.15", "YYYY.MM.DD format"),
        ("15th Jan 2024", "Ordinal day format"),
        ("Jan 15th, 2024", "Month ordinal format"),
        ("Q1 2024", "Quarter format"),
        ("Week 3, 2024", "Week format"),
        ("2024-W03", "ISO week format"),
        ("Day 15 of January 2024", "Verbose format")
    ]
    
    for date_str, description in custom_formats:
        try:
            parsed = dateparser.parse(date_str)
            if parsed:
                print(f"  {description}: '{date_str}' → {parsed.strftime('%Y-%m-%d')}")
            else:
                print(f"  {description}: '{date_str}' → Could not parse")
        except Exception as e:
            print(f"  {description}: '{date_str}' → Error: {e}")
            raise
    
    # Test time format variations
    print("\n✓ Testing time format variations:")
    
    time_formats = [
        ("2024-01-15 14:30", "24-hour format"),
        ("2024-01-15 2:30 PM", "12-hour with AM/PM"),
        ("2024-01-15 2:30:45", "With seconds"),
        ("2024-01-15 14:30:45.123", "With milliseconds"),
        ("2024-01-15T14:30:45", "ISO format"),
        ("Jan 15, 2024 at 2:30pm", "Natural format"),
        ("15/01/2024 14h30", "European format"),
        ("2024-01-15 14:30:45 UTC", "With timezone")
    ]
    
    for time_str, description in time_formats:
        try:
            parsed = dateparser.parse(time_str)
            if parsed:
                print(f"  {description}: '{time_str}' → {parsed}")
            else:
                print(f"  {description}: '{time_str}' → Could not parse")
        except Exception as e:
            print(f"  {description}: '{time_str}' → Error: {e}")
            raise
    
    # Test date validation and edge cases
    print("\n✓ Testing date validation:")
    
    # Valid dates
    valid_dates = [
        "2024-02-29",  # Leap year
        "2024-12-31",  # End of year
        "2024-01-01",  # Start of year
        "2024-06-30",  # End of month
    ]
    
    for valid_date in valid_dates:
        parsed = dateparser.parse(valid_date)
        if parsed:
            print(f"  Valid: '{valid_date}' → {parsed.strftime('%Y-%m-%d %A')}")
    
    # Invalid dates: dateparser returns None for each
    invalid_dates = [
        "2023-02-29",  # Not a leap year
        "2024-02-30",  # Invalid day for February
        "2024-04-31",  # Invalid day for April
        "2024-00-15",  # Invalid month (zero)
        "2024-01-00",  # Invalid day (zero)
    ]
    
    print("\n  Testing invalid dates:")
    for invalid_date in invalid_dates:
        parsed = dateparser.parse(invalid_date)
        assert parsed is None, f"dateparser accepted invalid date {invalid_date!r} as {parsed}"
        print(f"    Correctly rejected: '{invalid_date}'")
    
    lenient = dateparser.parse("2024-13-01")
    assert lenient == datetime(2024, 1, 13), lenient
    print(f"    Lenient field order: '2024-13-01' → {lenient.date()}")
    strict = dateparser.parse("2024-13-01", date_formats=["%Y-%m-%d"], settings={"PARSERS": ["custom-formats"]})
    assert strict is None, f"the '%Y-%m-%d' custom format accepted '2024-13-01' as {strict}"
    print("    Strict '%Y-%m-%d' format only: '2024-13-01' correctly rejected")
    
    # Test ambiguous date handling
    print("\n✓ Testing ambiguous date handling:")
    
    ambiguous_dates = [
        "01/02/03",    # Could be any combination of day/month/year
        "1/2/2024",    # Day vs month ambiguity
        "12/11/2024",  # Month vs day
        "2024/01/02",  # Year first format
    ]
    
    for amb_date in ambiguous_dates:
        # Test with different date order settings
        default_parse = dateparser.parse(amb_date)
        dmy_parse = dateparser.parse(amb_date, settings={'DATE_ORDER': 'DMY'})
        mdy_parse = dateparser.parse(amb_date, settings={'DATE_ORDER': 'MDY'})
        
        print(f"  '{amb_date}':")
        print(f"    Default: {default_parse}")
        print(f"    DMY: {dmy_parse}")
        print(f"    MDY: {mdy_parse}")
    
    # Test partial date parsing
    print("\n✓ Testing partial date parsing:")
    
    partial_dates = [
        "2024",         # Year only
        "January 2024", # Month and year
        "2024-01",      # Year and month
        "15th",         # Day only (relative to current month/year)
        "January",      # Month only (relative to current year)
        "Q2",           # Quarter only (relative to current year)
    ]
    
    for partial_date in partial_dates:
        try:
            parsed = dateparser.parse(partial_date)
            if parsed:
                print(f"  '{partial_date}' → {parsed.strftime('%Y-%m-%d')}")
            else:
                print(f"  '{partial_date}' → Could not parse")
        except Exception as e:
            print(f"  '{partial_date}' → Error: {e}")
            raise
    
    # Test fuzzy parsing with validation
    print("\n✓ Testing fuzzy parsing with validation:")
    
    def validate_parsed_date(date_str, expected_year=None, expected_month=None):
        """Validate that parsed date meets expectations"""
        parsed = dateparser.parse(date_str)
        if not parsed:
            return False, "Could not parse"
        
        issues = []
        if expected_year and parsed.year != expected_year:
            issues.append(f"Expected year {expected_year}, got {parsed.year}")
        if expected_month and parsed.month != expected_month:
            issues.append(f"Expected month {expected_month}, got {parsed.month}")
        
        if issues:
            return False, "; ".join(issues)
        return True, "Valid"
    
    # Test validation
    validation_tests = [
        ("Meeting on January 15, 2024", 2024, 1),
        ("Deadline: March 30th", None, 3),
        ("Born in 1995", 1995, None),
        ("Christmas Day 2024", 2024, 12),
    ]
    
    for date_text, exp_year, exp_month in validation_tests:
        is_valid, message = validate_parsed_date(date_text, exp_year, exp_month)
        status = "✓" if is_valid else "✗"
        print(f"  {status} '{date_text}': {message}")
    
    # Test date range validation
    print("\n✓ Testing date range validation:")
    
    def is_date_in_range(date_str, start_year=2020, end_year=2030):
        """Check if parsed date falls within acceptable range"""
        parsed = dateparser.parse(date_str)
        if not parsed:
            return False, "Could not parse"
        
        if parsed.year < start_year:
            return False, f"Date {parsed.year} is before {start_year}"
        if parsed.year > end_year:
            return False, f"Date {parsed.year} is after {end_year}"
        
        return True, f"Date {parsed.year} is within range {start_year}-{end_year}"
    
    range_tests = [
        "2025-01-15",  # Valid range
        "2019-01-15",  # Before range
        "2031-01-15",  # After range
        "tomorrow",    # Relative (should be valid)
        "next year",   # Relative (should be valid)
    ]
    
    for date_test in range_tests:
        in_range, message = is_date_in_range(date_test)
        status = "✓" if in_range else "✗"
        print(f"  {status} '{date_test}': {message}")
    
    print("✓ Custom date formats and validation completed successfully")
    
except Exception as e:
    print(f"Error in custom formats example: {e}")
    raise AssertionError(
        f"dateparser example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed\n")
```

```python
import dateparser
from datetime import datetime, timedelta
import json
import re

# Example 5: Real-world Applications and Edge Cases
print("Example 5: Real-world Applications and Edge Cases")

try:
    # Log file parsing simulation
    print("✓ Log file parsing simulation:")
    
    # Simulate various log formats
    log_entries = [
        "2024-01-15 14:30:25 [INFO] User login successful",
        "Jan 15 14:31:05 server01 nginx: GET /api/users",
        "15/01/2024 2:32 PM - Database connection established",
        "2024-01-15T14:33:12.456Z - API request processed",
        "Mon Jan 15 14:34:00 2024 - Cache invalidated",
        "1642253645 - Timestamp in epoch format",  # Unix timestamp
        "2024-W03-1 14:35 - ISO week date format"
    ]
    
    def extract_dates_from_logs(log_entries):
        """Extract and parse dates from various log formats"""
        results = []
        for entry in log_entries:
            # Try to extract date portion (various strategies)
            date_patterns = [
                r'^\d{4}-\d{2}-\d{2}[\sT]\d{2}:\d{2}:\d{2}',  # ISO format
                r'^[A-Za-z]{3}\s+\d{1,2}\s+\d{2}:\d{2}:\d{2}',  # Syslog format
                r'^\d{1,2}/\d{1,2}/\d{4}\s+\d{1,2}:\d{2}\s+[AP]M',  # US format
                r'^\d{10}',  # Unix timestamp
                r'^[A-Za-z]{3}\s+[A-Za-z]{3}\s+\d{1,2}\s+\d{2}:\d{2}:\d{2}\s+\d{4}',  # Full format
            ]
            
            extracted_date = None
            for pattern in date_patterns:
                match = re.search(pattern, entry)
                if match:
                    extracted_date = match.group()
                    break
            
            if not extracted_date:
                # Fallback: try parsing the entire entry
                extracted_date = entry.split(' -')[0].split(' [')[0]
            
            try:
                # Handle Unix timestamp specially
                if extracted_date.isdigit() and len(extracted_date) == 10:
                    parsed_date = datetime.fromtimestamp(int(extracted_date))
                else:
                    parsed_date = dateparser.parse(extracted_date)
                
                results.append({
                    'original': entry,
                    'extracted': extracted_date,
                    'parsed': parsed_date,
                    'success': parsed_date is not None
                })
            except Exception as e:
                results.append({
                    'original': entry,
                    'extracted': extracted_date,
                    'parsed': None,
                    'success': False,
                    'error': str(e)
                })
        
        return results
    
    log_results = extract_dates_from_logs(log_entries)
    
    for result in log_results:
        assert result['success'], f"log timestamp did not parse: {result['extracted']!r} {result.get('error', '(parse returned None)')}"
        print(f"  ✓ Extracted: {result['parsed'].strftime('%Y-%m-%d %H:%M:%S')}")
    
    # Data processing pipeline
    print("\n✓ Data processing pipeline:")
    
    # Simulate processing dates from various data sources
    raw_data = [
        {"event": "user_signup", "timestamp": "2024-01-15 14:30:00", "source": "web"},
        {"event": "purchase", "timestamp": "Jan 15, 2024 at 2:35 PM", "source": "mobile"},
        {"event": "login", "timestamp": "15/01/2024 14:40", "source": "api"},
        {"event": "logout", "timestamp": "2024-01-15T14:45:00Z", "source": "web"},
        {"event": "view_product", "timestamp": "today at 2:50 PM", "source": "mobile"},
        {"event": "invalid_data", "timestamp": "not a date", "source": "unknown"}
    ]
    
    def process_event_data(events):
        """Process events and normalize timestamps"""
        processed_events = []
        failed_events = []
        
        for event in events:
            try:
                # Parse the timestamp
                parsed_time = dateparser.parse(event['timestamp'])
                
                if parsed_time:
                    # Normalize timezone information
                    if parsed_time.tzinfo is None:
                        # If naive, assume local time
                        parsed_time = parsed_time.replace(tzinfo=None)
                    else:
                        # Convert to naive UTC for consistency
                        parsed_time = parsed_time.utctimetuple()
                        parsed_time = datetime(*parsed_time[:6])
                    
                    # Normalize to UTC and add additional metadata
                    processed_event = {
                        'event_type': event['event'],
                        'source': event['source'],
                        'original_timestamp': event['timestamp'],
                        'normalized_timestamp': parsed_time.strftime('%Y-%m-%d %H:%M:%S'),
                        'unix_timestamp': int(parsed_time.timestamp()),
                        'day_of_week': parsed_time.strftime('%A'),
                        'hour_of_day': parsed_time.hour,
                        'is_weekend': parsed_time.weekday() >= 5
                    }
                    processed_events.append(processed_event)
                else:
                    failed_events.append(event)
                    
            except Exception as e:
                event['error'] = str(e)
                failed_events.append(event)
        
        return processed_events, failed_events
    
    processed, failed = process_event_data(raw_data)
    
    # every real timestamp parses; only the deliberately invalid record ('not a date') is refused
    assert [e['event'] for e in failed] == ['invalid_data'], failed
    assert len(processed) == 5, [e['event_type'] for e in processed]
    print(f"  Successfully processed: {len(processed)} events")
    print(f"  Refused (deliberately invalid): {len(failed)} event")
    
    for event in processed[:3]:  # Show first 3
        print(f"    {event['event_type']}: {event['normalized_timestamp']} ({event['day_of_week']})")
    
    print("  Deliberately invalid events:")
    for event in failed:
        print(f"    {event['event']}: '{event['timestamp']}'")
    
    # Date range queries and filtering
    print("\n✓ Date range queries and filtering:")
    
    def filter_events_by_date_range(events, start_date_str, end_date_str):
        """Filter events within a date range"""
        start_date = dateparser.parse(start_date_str)
        end_date = dateparser.parse(end_date_str)
        
        if not start_date or not end_date:
            return [], "Invalid date range"
        
        # Ensure both dates are naive for comparison
        if start_date.tzinfo is not None:
            start_date = start_date.replace(tzinfo=None)
        if end_date.tzinfo is not None:
            end_date = end_date.replace(tzinfo=None)
        
        filtered_events = []
        for event in events:
            event_time = dateparser.parse(event['original_timestamp'])
            if event_time:
                # Make event_time naive for comparison
                if event_time.tzinfo is not None:
                    event_time = event_time.replace(tzinfo=None)
                
                if start_date <= event_time <= end_date:
                    filtered_events.append(event)
        
        return filtered_events, None
    
    # Test date range filtering
    date_ranges = [
        ("2024-01-15 14:00", "2024-01-15 15:00"),  # 1 hour window
        ("2024-01-15", "2024-01-16"),              # Full day range
        ("2024-01-01", "2024-12-31"),              # Year range
    ]
    
    for start, end in date_ranges:
        filtered, error = filter_events_by_date_range(processed, start, end)
        assert error is None, f"filtering {start} to {end}: {error}"
        assert sorted(e['event_type'] for e in filtered) == ['login', 'logout', 'purchase', 'user_signup'], filtered
        print(f"  {start} to {end}: {len(filtered)} events")
    
    # Natural language scheduling
    print("\n✓ Natural language scheduling:")
    
    def parse_schedule_expressions(expressions):
        """Parse natural language scheduling expressions"""
        results = []
        
        for expr in expressions:
            try:
                # Parse the expression
                parsed_time = dateparser.parse(expr)
                
                if parsed_time:
                    # Calculate time until the event
                    now = datetime.now()
                    
                    # Make both times naive for comparison
                    if parsed_time.tzinfo is not None:
                        parsed_time = parsed_time.replace(tzinfo=None)
                    if now.tzinfo is not None:
                        now = now.replace(tzinfo=None)
                    
                    time_diff = parsed_time - now
                    
                    # Determine if it's in the past or future
                    if time_diff.total_seconds() > 0:
                        days = time_diff.days
                        hours = time_diff.seconds // 3600
                        time_desc = f"in {days} days, {hours} hours"
                    else:
                        days = abs(time_diff.days)
                        hours = abs(time_diff.seconds) // 3600
                        time_desc = f"{days} days, {hours} hours ago"
                    
                    results.append({
                        'expression': expr,
                        'parsed_time': parsed_time,
                        'formatted': parsed_time.strftime('%Y-%m-%d %H:%M:%S %A'),
                        'relative': time_desc,
                        'success': True
                    })
                else:
                    results.append({
                        'expression': expr,
                        'success': False,
                        'error': 'Could not parse'
                    })
                    
            except Exception as e:
                results.append({
                    'expression': expr,
                    'success': False,
                    'error': str(e)
                })
        
        return results
    
    schedule_expressions = [
        "tomorrow at 9 AM",
        "next Monday at 2 PM",
        "in 3 hours",
        "next Friday at 5:30 PM",
        "December 25th at midnight",
        "2024-12-31 23:59:59"
    ]
    
    schedule_results = parse_schedule_expressions(schedule_expressions)
    
    for result in schedule_results:
        if result['success']:
            print(f"  ✓ '{result['expression']}' → {result['formatted']}")
            print(f"    ({result['relative']})")
        else:
            print(f"  ✗ '{result['expression']}' → {result['error']}")
    
    # Edge cases and error handling
    print("\n✓ Edge cases and error handling:")
    
    edge_cases = [
        "",                           # Empty string
        "   ",                        # Whitespace only
        "not a date at all",         # Non-date text
        "32/13/2024",                # Invalid date
        "2024-02-30",                # Invalid date (February 30th)
        "25:99:99",                  # Invalid time
        "Monday the 45th",           # Impossible date
        "next yesterday",            # Contradictory
        "2024-01-01T25:00:00",       # Invalid hour
        "null",                      # Common programming null value
    ]
    
    def safely_parse_date(date_str):
        """Safely parse dates with comprehensive error handling"""
        if not date_str or not isinstance(date_str, str) or not date_str.strip():
            return None, "Empty or invalid input"
        
        try:
            # Clean the input
            cleaned = date_str.strip()
            
            # Try parsing
            result = dateparser.parse(cleaned)
            
            if result is None:
                return None, "Could not parse date"
            
            # Basic validation
            if result.year < 1900 or result.year > 2100:
                return None, f"Year {result.year} seems unrealistic"
            
            return result, "Success"
            
        except Exception as e:
            return None, f"Exception: {type(e).__name__}: {str(e)}"
    
    print("  Testing edge cases:")
    for edge_case in edge_cases:
        result, message = safely_parse_date(edge_case)
        status = "✓" if result else "✗"
        print(f"    {status} '{edge_case}' → {message}")
    
    # Performance considerations
    print("\n✓ Performance considerations:")
    
    import time
    
    # Test parsing performance with many dates
    test_dates = [
        "2024-01-15 14:30:00",
        "Jan 15, 2024 2:30 PM",
        "15/01/2024 14:30",
        "2024-01-15T14:30:00",
        "January 15, 2024",
    ] * 20  # 100 dates total
    
    start_time = time.time()
    
    success_count = 0
    for date_str in test_dates:
        try:
            result = dateparser.parse(date_str)
            if result:
                success_count += 1
        except:
            pass
            raise
    
    end_time = time.time()
    processing_time = end_time - start_time
    
    print(f"  Processed {len(test_dates)} dates in {processing_time:.3f} seconds")
    print(f"  Average: {(processing_time / len(test_dates) * 1000):.2f} ms per date")
    print(f"  Success rate: {(success_count / len(test_dates) * 100):.1f}%")
    
    print("✓ Real-world applications and edge cases completed successfully")
    
except Exception as e:
    print(f"Error in real-world applications example: {e}")
    raise AssertionError(
        f"dateparser example (cell 4) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 completed")
```
