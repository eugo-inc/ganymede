## Example 1: Basic Dictionary Merging

```python
import deepmerge

print("Example 1: Basic Dictionary Merging")

try:
    print("✓ deepmerge package information:")
    print(f"  Module type: {type(deepmerge).__name__}")
    
    # Show available functions
    attrs = [attr for attr in dir(deepmerge) if not attr.startswith('_')]
    print(f"  Public attributes: {len(attrs)}")
    key_attrs = [attr for attr in attrs if attr in ['merge', 'always_merger', 'Merger']]
    print(f"  Key components: {key_attrs}")
    
    # Basic dictionary merging
    print("✓ Basic dictionary merging:")
    
    dict1 = {
        'name': 'John',
        'age': 25,
        'skills': ['Python', 'JavaScript'],
        'contact': {
            'email': 'john@example.com',
            'phone': '123-456-7890'
        }
    }
    
    dict2 = {
        'age': 26,  # This will overwrite
        'location': 'New York',
        'skills': ['SQL', 'Docker'],  # This will be merged/replaced
        'contact': {
            'phone': '098-765-4321',  # This will overwrite
            'address': '123 Main St'  # This will be added
        }
    }
    
    print(f"  Dictionary 1: {dict1}")
    print(f"  Dictionary 2: {dict2}")
    
    # Simple merge using deepmerge.merge
    if hasattr(deepmerge, 'merge'):
        merged = deepmerge.merge(dict1, dict2)
        print(f"  Merged result: {merged}")
        print(f"  Original dict1 unchanged: {dict1}")
    
    # Alternative: use always_merger
    if hasattr(deepmerge, 'always_merger'):
        always_merged = deepmerge.always_merger.merge(dict1, dict2)
        print(f"  Always merger result: {always_merged}")
    
    # Test nested structure merging
    print("✓ Nested structure merging:")
    
    config1 = {
        'database': {
            'host': 'localhost',
            'port': 5432,
            'credentials': {
                'username': 'admin'
            }
        },
        'features': {
            'logging': True,
            'cache': False
        }
    }
    
    config2 = {
        'database': {
            'port': 5433,  # Override port
            'credentials': {
                'password': 'secret123'  # Add password
            },
            'pool_size': 10  # Add new field
        },
        'features': {
            'cache': True,  # Override cache
            'monitoring': True  # Add monitoring
        },
        'new_section': {
            'enabled': True
        }
    }
    
    print(f"  Config 1 database port: {config1['database']['port']}")
    print(f"  Config 2 database port: {config2['database']['port']}")
    
    # Merge configurations
    if hasattr(deepmerge, 'always_merger'):
        merged_config = deepmerge.always_merger.merge(config1, config2)
        print(f"  Merged database port: {merged_config['database']['port']}")
        print(f"  Merged database has password: {'password' in merged_config['database']['credentials']}")
        print(f"  New section added: {'new_section' in merged_config}")
        print(f"  Cache feature: {merged_config['features']['cache']}")
    
    # List merging behavior
    print("✓ List merging:")
    
    data1 = {
        'tags': ['python', 'web'],
        'numbers': [1, 2, 3]
    }
    
    data2 = {
        'tags': ['api', 'backend'],
        'numbers': [4, 5, 6]
    }
    
    print(f"  Data 1 tags: {data1['tags']}")
    print(f"  Data 2 tags: {data2['tags']}")
    
    if hasattr(deepmerge, 'always_merger'):
        merged_data = deepmerge.always_merger.merge(data1, data2)
        print(f"  Merged tags: {merged_data['tags']}")
        print(f"  Merged numbers: {merged_data['numbers']}")
    
    print("✓ Basic dictionary merging completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: deepmerge provides deep merging capabilities")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"deepmerge example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Custom Merge Strategies

```python
import deepmerge

print("Example 2: Custom Merge Strategies")

try:
    # Custom merger configuration
    print("✓ Custom merger strategies:")
    
    # Check if Merger class is available
    if hasattr(deepmerge, 'Merger'):
        # Create custom mergers with different strategies
        print("  Creating custom mergers:")
        
        # List append strategy
        list_append_merger = deepmerge.Merger(
            [(list, ["append"]), (dict, ["merge"]), (set, ["union"])],
            ["override"],
            ["override"]
        )
        
        # List replace strategy  
        list_replace_merger = deepmerge.Merger(
            [(list, ["override"]), (dict, ["merge"])],
            ["override"],
            ["override"]
        )
        
        print("  Custom mergers created successfully")
        
    else:
        print("  Merger class not available, using default mergers")
    
    # Test different list merging strategies
    print("✓ List merging strategies:")
    
    data1 = {
        'permissions': ['read', 'write'],
        'tags': ['python', 'web'],
        'config': {
            'enabled_features': ['auth', 'logging']
        }
    }
    
    data2 = {
        'permissions': ['execute', 'admin'],
        'tags': ['api', 'backend'],
        'config': {
            'enabled_features': ['monitoring', 'cache']
        }
    }
    
    print(f"  Data 1 permissions: {data1['permissions']}")
    print(f"  Data 2 permissions: {data2['permissions']}")
    
    # Default merger behavior (usually replaces lists)
    if hasattr(deepmerge, 'always_merger'):
        default_merge = deepmerge.always_merger.merge(data1, data2)
        print(f"  Default merge permissions: {default_merge['permissions']}")
    
    # Custom append strategy (if available)
    if hasattr(deepmerge, 'Merger'):
        try:
            append_merge = list_append_merger.merge(data1, data2)
            print(f"  Append merge permissions: {append_merge['permissions']}")
            print(f"  Append merge features: {append_merge['config']['enabled_features']}")
        except Exception as e:
            print(f"  Append merge not available: {e}")
            raise
    
    # Custom merge functions
    print("✓ Custom merge functions:")
    
    def merge_lists_unique(list1, list2):
        """Custom function to merge lists with unique values"""
        if not isinstance(list1, list) or not isinstance(list2, list):
            return list2  # Override if not both lists
        
        combined = list1 + list2
        return list(dict.fromkeys(combined))  # Preserve order, remove duplicates
    
    def merge_numbers_sum(num1, num2):
        """Custom function to sum numeric values"""
        if isinstance(num1, (int, float)) and isinstance(num2, (int, float)):
            return num1 + num2
        return num2  # Override if not both numbers
    
    # Apply custom merge logic manually
    data_with_numbers = {
        'scores': [85, 92, 78],
        'total_points': 100,
        'categories': ['math', 'science']
    }
    
    data_with_more = {
        'scores': [88, 95],
        'total_points': 150,
        'categories': ['english', 'science', 'math']  # Has duplicates with first
    }
    
    print(f"  Data 1 scores: {data_with_numbers['scores']}")
    print(f"  Data 2 scores: {data_with_more['scores']}")
    print(f"  Data 1 points: {data_with_numbers['total_points']}")
    print(f"  Data 2 points: {data_with_more['total_points']}")
    
    # Manual custom merging
    custom_merged = {}
    for key in set(data_with_numbers.keys()) | set(data_with_more.keys()):
        val1 = data_with_numbers.get(key)
        val2 = data_with_more.get(key)
        
        if key == 'scores':
            custom_merged[key] = merge_lists_unique(val1 or [], val2 or [])
        elif key == 'total_points':
            custom_merged[key] = merge_numbers_sum(val1 or 0, val2 or 0)
        elif key == 'categories':
            custom_merged[key] = merge_lists_unique(val1 or [], val2 or [])
        else:
            custom_merged[key] = val2 if val2 is not None else val1
    
    print(f"  Custom merged scores: {custom_merged['scores']}")
    print(f"  Custom merged points: {custom_merged['total_points']}")
    print(f"  Custom merged categories: {custom_merged['categories']}")
    
    # Conditional merging
    print("✓ Conditional merging strategies:")
    
    def conditional_merge(dict1, dict2, condition_fn):
        """Merge dictionaries based on conditions"""
        result = dict1.copy()
        
        for key, value in dict2.items():
            if key in result:
                if condition_fn(key, result[key], value):
                    # Apply merge logic based on condition
                    if isinstance(result[key], dict) and isinstance(value, dict):
                        result[key] = conditional_merge(result[key], value, condition_fn)
                    elif isinstance(result[key], list) and isinstance(value, list):
                        result[key] = merge_lists_unique(result[key], value)
                    else:
                        result[key] = value
                # else keep original value
            else:
                result[key] = value
                
        return result
    
    # Define condition: only merge if new value is "better" 
    def merge_if_better(key, old_val, new_val):
        """Merge condition: numbers if higher, strings if longer, lists always"""
        if isinstance(old_val, (int, float)) and isinstance(new_val, (int, float)):
            return new_val > old_val
        elif isinstance(old_val, str) and isinstance(new_val, str):
            return len(new_val) > len(old_val)
        elif isinstance(old_val, list):
            return True  # Always merge lists
        return True  # Default: merge
    
    old_data = {
        'score': 85,
        'name': 'John',
        'skills': ['Python'],
        'config': {'version': 1}
    }
    
    new_data = {
        'score': 92,  # Higher, should merge
        'name': 'Jo',  # Shorter, should not merge
        'skills': ['JavaScript', 'SQL'],  # Should merge
        'config': {'version': 2, 'debug': True}  # Should merge
    }
    
    conditional_result = conditional_merge(old_data, new_data, merge_if_better)
    
    print(f"  Conditional merge score: {conditional_result['score']} (was {old_data['score']})")
    print(f"  Conditional merge name: '{conditional_result['name']}' (was '{old_data['name']}')")
    print(f"  Conditional merge skills: {conditional_result['skills']}")
    
    print("✓ Custom merge strategies completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: Custom strategies may require specific deepmerge features")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"deepmerge example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: Configuration Management and Inheritance

```python
import deepmerge
import json

print("Example 3: Configuration Management and Inheritance")

try:
    # Configuration inheritance patterns
    print("✓ Configuration inheritance:")
    
    # Base configuration
    base_config = {
        'app': {
            'name': 'MyApp',
            'version': '1.0.0',
            'debug': False,
            'features': {
                'auth': True,
                'logging': True,
                'cache': False
            }
        },
        'database': {
            'type': 'postgresql',
            'host': 'localhost',
            'port': 5432,
            'pool': {
                'min_size': 1,
                'max_size': 10,
                'timeout': 30
            }
        }
    }
    
    # Development environment overrides
    dev_config = {
        'app': {
            'debug': True,
            'features': {
                'cache': True,
                'profiling': True
            }
        },
        'database': {
            'host': 'dev-db.local',
            'pool': {
                'max_size': 5
            }
        }
    }
    
    # Production environment overrides
    prod_config = {
        'app': {
            'features': {
                'profiling': False,
                'monitoring': True
            }
        },
        'database': {
            'host': 'prod-db.company.com',
            'port': 5433,
            'pool': {
                'min_size': 5,
                'max_size': 50,
                'timeout': 60
            }
        }
    }
    
    print(f"  Base config debug: {base_config['app']['debug']}")
    print(f"  Base config cache: {base_config['app']['features']['cache']}")
    print(f"  Base config pool max: {base_config['database']['pool']['max_size']}")
    
    # Merge configurations for different environments
    if hasattr(deepmerge, 'always_merger'):
        dev_final = deepmerge.always_merger.merge(base_config, dev_config)
        prod_final = deepmerge.always_merger.merge(base_config, prod_config)
        
        print(f"  Dev config debug: {dev_final['app']['debug']}")
        print(f"  Dev config cache: {dev_final['app']['features']['cache']}")
        print(f"  Dev config profiling: {dev_final['app']['features'].get('profiling', 'Not set')}")
        
        print(f"  Prod config debug: {prod_final['app']['debug']}")
        print(f"  Prod config monitoring: {prod_final['app']['features'].get('monitoring', 'Not set')}")
        print(f"  Prod config pool max: {prod_final['database']['pool']['max_size']}")
    
    # Multi-level configuration inheritance
    print("✓ Multi-level inheritance:")
    
    # Global defaults
    global_defaults = {
        'timeouts': {
            'connection': 5,
            'read': 30,
            'write': 10
        },
        'retry': {
            'max_attempts': 3,
            'backoff': 1.0
        }
    }
    
    # Service-specific defaults
    service_defaults = {
        'timeouts': {
            'read': 60  # Override for this service
        },
        'retry': {
            'max_attempts': 5  # Override for this service
        },
        'service_specific': {
            'batch_size': 100
        }
    }
    
    # Environment overrides
    env_overrides = {
        'timeouts': {
            'connection': 10  # Increase for this environment
        },
        'service_specific': {
            'batch_size': 50,  # Reduce for this environment
            'workers': 4
        }
    }
    
    # Merge in order: global -> service -> environment
    if hasattr(deepmerge, 'always_merger'):
        step1 = deepmerge.always_merger.merge(global_defaults, service_defaults)
        final_config = deepmerge.always_merger.merge(step1, env_overrides)
        
        print(f"  Final connection timeout: {final_config['timeouts']['connection']}")
        print(f"  Final read timeout: {final_config['timeouts']['read']}")
        print(f"  Final max attempts: {final_config['retry']['max_attempts']}")
        print(f"  Final batch size: {final_config['service_specific']['batch_size']}")
    
    # Configuration validation after merging
    print("✓ Post-merge validation:")
    
    def validate_config(config):
        """Validate merged configuration"""
        errors = []
        
        # Validate database config
        if 'database' in config:
            db = config['database']
            if db.get('port', 0) < 1024 and db.get('port', 0) > 0:
                errors.append("Database port should be >= 1024")
            
            if 'pool' in db:
                pool = db['pool']
                if pool.get('min_size', 0) > pool.get('max_size', 0):
                    errors.append("Pool min_size cannot be greater than max_size")
        
        # Validate app features
        if 'app' in config and 'features' in config['app']:
            features = config['app']['features']
            if features.get('debug') and features.get('monitoring'):
                errors.append("Debug and monitoring should not both be enabled")
        
        return errors
    
    # Validate our configurations
    if hasattr(deepmerge, 'always_merger'):
        dev_errors = validate_config(dev_final)
        prod_errors = validate_config(prod_final)
        
        print(f"  Dev config validation: {len(dev_errors)} errors")
        for error in dev_errors:
            print(f"    - {error}")
        
        print(f"  Prod config validation: {len(prod_errors)} errors")  
        for error in prod_errors:
            print(f"    - {error}")
    
    print("✓ Configuration management and inheritance completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: deepmerge is essential for configuration management")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"deepmerge example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: Complex Data Structure Merging

```python
import deepmerge

print("Example 4: Complex Data Structure Merging")

try:
    # API response merging
    print("✓ API response merging:")
    
    # First API response
    response1 = {
        'user': {
            'id': 123,
            'name': 'Alice Johnson',
            'email': 'alice@example.com',
            'profile': {
                'bio': 'Software Engineer',
                'skills': ['Python', 'JavaScript'],
                'experience': {
                    'years': 5,
                    'companies': ['TechCorp', 'StartupXYZ']
                }
            },
            'settings': {
                'notifications': True,
                'privacy': 'public'
            }
        },
        'metadata': {
            'source': 'api_v1',
            'timestamp': '2024-01-15T10:00:00Z'
        }
    }
    
    # Second API response (partial update)
    response2 = {
        'user': {
            'profile': {
                'skills': ['SQL', 'Docker'],  # Additional skills
                'experience': {
                    'years': 6,  # Updated
                    'companies': ['NewCompany']  # Additional company
                },
                'certifications': ['AWS', 'Docker']  # New field
            },
            'settings': {
                'theme': 'dark',  # New setting
                'privacy': 'private'  # Updated setting
            }
        },
        'metadata': {
            'source': 'api_v2',
            'timestamp': '2024-01-20T14:30:00Z'
        }
    }
    
    print(f"  Response 1 skills: {response1['user']['profile']['skills']}")
    print(f"  Response 2 skills: {response2['user']['profile']['skills']}")
    print(f"  Response 1 years: {response1['user']['profile']['experience']['years']}")
    print(f"  Response 2 years: {response2['user']['profile']['experience']['years']}")
    
    # Merge API responses
    if hasattr(deepmerge, 'always_merger'):
        merged_response = deepmerge.always_merger.merge(response1, response2)
        
        print(f"  Merged skills: {merged_response['user']['profile']['skills']}")
        print(f"  Merged years: {merged_response['user']['profile']['experience']['years']}")
        print(f"  Merged companies: {merged_response['user']['profile']['experience']['companies']}")
        print(f"  New certifications: {merged_response['user']['profile'].get('certifications', 'Not found')}")
        print(f"  Updated privacy: {merged_response['user']['settings']['privacy']}")
    
    # Schema merging for data validation
    print("✓ Schema merging:")
    
    base_schema = {
        'type': 'object',
        'properties': {
            'name': {'type': 'string', 'minLength': 1},
            'age': {'type': 'integer', 'minimum': 0, 'maximum': 150},
            'address': {
                'type': 'object',
                'properties': {
                    'street': {'type': 'string'},
                    'city': {'type': 'string'},
                    'zipcode': {'type': 'string', 'pattern': r'^\d{5}$'}
                },
                'required': ['street', 'city']
            }
        },
        'required': ['name', 'age']
    }
    
    extension_schema = {
        'properties': {
            'age': {'maximum': 120},  # More restrictive
            'email': {'type': 'string', 'format': 'email'},  # New field
            'address': {
                'properties': {
                    'country': {'type': 'string', 'default': 'USA'},  # New field
                    'zipcode': {'pattern': r'^\d{5}(-\d{4})?$'}  # More flexible
                }
            },
            'preferences': {  # Entirely new section
                'type': 'object',
                'properties': {
                    'newsletter': {'type': 'boolean', 'default': False}
                }
            }
        },
        'required': ['email']  # Additional requirement
    }
    
    print(f"  Base schema required: {base_schema['required']}")
    print(f"  Extension schema required: {extension_schema['required']}")
    print(f"  Base age maximum: {base_schema['properties']['age']['maximum']}")
    print(f"  Extension age maximum: {extension_schema['properties']['age']['maximum']}")
    
    # Merge schemas
    if hasattr(deepmerge, 'always_merger'):
        merged_schema = deepmerge.always_merger.merge(base_schema, extension_schema)
        
        print(f"  Merged required: {merged_schema['required']}")
        print(f"  Merged age maximum: {merged_schema['properties']['age']['maximum']}")
        print(f"  Has email field: {'email' in merged_schema['properties']}")
        print(f"  Address has country: {'country' in merged_schema['properties']['address']['properties']}")
    
    # Multi-level list merging
    print("✓ Multi-level list merging:")
    
    data_structure1 = {
        'departments': [
            {
                'name': 'Engineering',
                'teams': ['Backend', 'Frontend'],
                'members': [
                    {'name': 'Alice', 'role': 'Senior Dev'},
                    {'name': 'Bob', 'role': 'Junior Dev'}
                ]
            },
            {
                'name': 'Marketing',
                'teams': ['Digital', 'Content'],
                'members': [
                    {'name': 'Carol', 'role': 'Manager'}
                ]
            }
        ]
    }
    
    data_structure2 = {
        'departments': [
            {
                'name': 'Engineering',  # Same department
                'teams': ['DevOps', 'QA'],  # Additional teams
                'members': [
                    {'name': 'David', 'role': 'DevOps Engineer'}  # Additional member
                ]
            },
            {
                'name': 'Sales',  # New department
                'teams': ['Enterprise', 'SMB'],
                'members': [
                    {'name': 'Eve', 'role': 'Sales Rep'}
                ]
            }
        ]
    }
    
    print(f"  Structure 1 departments: {len(data_structure1['departments'])}")
    print(f"  Structure 2 departments: {len(data_structure2['departments'])}")
    print(f"  Engineering teams 1: {data_structure1['departments'][0]['teams']}")
    print(f"  Engineering teams 2: {data_structure2['departments'][0]['teams']}")
    
    # Note: Default deepmerge behavior with lists
    if hasattr(deepmerge, 'always_merger'):
        merged_structure = deepmerge.always_merger.merge(data_structure1, data_structure2)
        
        print(f"  Merged departments: {len(merged_structure['departments'])}")
        # Note: This will typically replace the departments list, not merge it
        print(f"  First department name: {merged_structure['departments'][0]['name']}")
        if len(merged_structure['departments']) > 0:
            print(f"  First dept teams: {merged_structure['departments'][0]['teams']}")
    
    # Custom deep merge for complex structures
    print("✓ Custom complex structure handling:")
    
    def smart_merge_by_key(list1, list2, key_field):
        """Merge lists of dictionaries by matching key field"""
        result = list1.copy()
        lookup = {item.get(key_field): i for i, item in enumerate(result)}
        
        for item2 in list2:
            key_value = item2.get(key_field)
            if key_value in lookup:
                # Merge existing item
                index = lookup[key_value]
                if hasattr(deepmerge, 'always_merger'):
                    result[index] = deepmerge.always_merger.merge(result[index], item2)
                else:
                    # Fallback: simple update
                    result[index].update(item2)
            else:
                # Add new item
                result.append(item2)
        
        return result
    
    # Apply smart merging to departments
    smart_merged = data_structure1.copy()
    smart_merged['departments'] = smart_merge_by_key(
        data_structure1['departments'],
        data_structure2['departments'], 
        'name'
    )
    
    print(f"  Smart merged departments: {len(smart_merged['departments'])}")
    eng_dept = next(d for d in smart_merged['departments'] if d['name'] == 'Engineering')
    print(f"  Engineering teams (smart): {eng_dept['teams']}")
    print(f"  Engineering members (smart): {len(eng_dept['members'])}")
    
    print("✓ Complex data structure merging completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: deepmerge handles complex nested structures")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"deepmerge example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: Real-world Applications and Best Practices

```python
import deepmerge
import copy

print("Example 5: Real-world Applications and Best Practices")

try:
    # Settings and preferences merging
    print("✓ Settings and preferences merging:")
    
    # Default application settings
    default_settings = {
        'ui': {
            'theme': 'light',
            'language': 'en',
            'sidebar': {
                'width': 250,
                'collapsed': False,
                'items': ['dashboard', 'reports', 'settings']
            }
        },
        'performance': {
            'cache_size': 100,
            'batch_size': 50,
            'timeout': 30
        },
        'features': {
            'notifications': True,
            'auto_save': True,
            'analytics': False
        }
    }
    
    # User customizations
    user_settings = {
        'ui': {
            'theme': 'dark',
            'sidebar': {
                'width': 300,
                'items': ['dashboard', 'analytics', 'reports']  # Reordered + added
            }
        },
        'performance': {
            'cache_size': 200  # Increased
        },
        'features': {
            'analytics': True,  # Enabled
            'experimental': True  # New feature
        }
    }
    
    print(f"  Default theme: {default_settings['ui']['theme']}")
    print(f"  User theme: {user_settings['ui']['theme']}")
    print(f"  Default sidebar width: {default_settings['ui']['sidebar']['width']}")
    print(f"  User sidebar width: {user_settings['ui']['sidebar']['width']}")
    
    # Merge user settings with defaults
    if hasattr(deepmerge, 'always_merger'):
        final_settings = deepmerge.always_merger.merge(default_settings, user_settings)
        
        print(f"  Final theme: {final_settings['ui']['theme']}")
        print(f"  Final sidebar width: {final_settings['ui']['sidebar']['width']}")
        print(f"  Final analytics: {final_settings['features']['analytics']}")
        print(f"  Final experimental: {final_settings['features'].get('experimental', False)}")
        print(f"  Final sidebar items: {final_settings['ui']['sidebar']['items']}")
    
    # Plugin system configuration merging
    print("✓ Plugin system configuration:")
    
    # Core plugin configuration
    core_plugins = {
        'auth': {
            'enabled': True,
            'config': {
                'session_timeout': 3600,
                'require_2fa': False
            }
        },
        'logging': {
            'enabled': True,
            'config': {
                'level': 'INFO',
                'format': 'json',
                'destinations': ['file', 'console']
            }
        }
    }
    
    # Environment-specific plugin overrides
    env_plugins = {
        'auth': {
            'config': {
                'session_timeout': 7200,  # Longer in this environment
                'require_2fa': True       # More secure
            }
        },
        'logging': {
            'config': {
                'level': 'DEBUG',         # More verbose
                'destinations': ['file', 'elasticsearch']  # Different output
            }
        },
        'monitoring': {  # Additional plugin
            'enabled': True,
            'config': {
                'metrics_interval': 60,
                'health_check': True
            }
        }
    }
    
    print(f"  Core auth timeout: {core_plugins['auth']['config']['session_timeout']}")
    print(f"  Env auth timeout: {env_plugins['auth']['config']['session_timeout']}")
    print(f"  Core logging level: {core_plugins['logging']['config']['level']}")
    print(f"  Env logging level: {env_plugins['logging']['config']['level']}")
    
    # Merge plugin configurations
    if hasattr(deepmerge, 'always_merger'):
        final_plugins = deepmerge.always_merger.merge(core_plugins, env_plugins)
        
        print(f"  Final auth timeout: {final_plugins['auth']['config']['session_timeout']}")
        print(f"  Final require 2FA: {final_plugins['auth']['config']['require_2fa']}")
        print(f"  Final logging level: {final_plugins['logging']['config']['level']}")
        print(f"  Final log destinations: {final_plugins['logging']['config']['destinations']}")
        print(f"  Monitoring plugin added: {'monitoring' in final_plugins}")
    
    # Data aggregation and reporting
    print("✓ Data aggregation merging:")
    
    # Report data from multiple sources
    sales_data = {
        'period': '2024-Q1',
        'metrics': {
            'revenue': 150000,
            'customers': 1200,
            'orders': 800
        },
        'regions': {
            'north': {'revenue': 60000, 'customers': 480},
            'south': {'revenue': 90000, 'customers': 720}
        }
    }
    
    marketing_data = {
        'period': '2024-Q1',
        'metrics': {
            'leads': 2500,
            'conversions': 320,
            'campaigns': 12
        },
        'regions': {
            'north': {'leads': 1000, 'conversions': 128},
            'south': {'leads': 1500, 'conversions': 192}
        }
    }
    
    print(f"  Sales revenue: ${sales_data['metrics']['revenue']:,}")
    print(f"  Marketing leads: {marketing_data['metrics']['leads']:,}")
    print(f"  North sales customers: {sales_data['regions']['north']['customers']}")
    print(f"  North marketing leads: {marketing_data['regions']['north']['leads']}")
    
    # Merge data for combined reporting
    if hasattr(deepmerge, 'always_merger'):
        combined_data = deepmerge.always_merger.merge(sales_data, marketing_data)
        
        print(f"  Combined metrics keys: {list(combined_data['metrics'].keys())}")
        print(f"  Combined north region keys: {list(combined_data['regions']['north'].keys())}")
        print(f"  Combined revenue: ${combined_data['metrics']['revenue']:,}")
        print(f"  Combined campaigns: {combined_data['metrics']['campaigns']}")
    
    # Best practices demonstration
    print("✓ Best practices:")
    
    best_practices = [
        "Always use deep copies when preserving original data",
        "Validate merged results for consistency and correctness",
        "Consider custom merge strategies for business-specific logic",
        "Document merge behavior for complex data structures",
        "Test merge operations with edge cases and empty values",
        "Use appropriate merge strategies for lists vs. dictionaries",
        "Consider performance implications for large data structures"
    ]
    
    for practice in best_practices:
        print(f"  • {practice}")
    
    # Performance and safety tips
    print("✓ Performance and safety tips:")
    
    # Demonstrate safe merging with copies
    original_data = {'important': 'data', 'config': {'value': 42}}
    update_data = {'config': {'value': 100, 'new_field': 'added'}}
    
    print(f"  Original before merge: {original_data}")
    
    # Safe merge (doesn't modify originals)
    if hasattr(deepmerge, 'always_merger'):
        safe_merged = deepmerge.always_merger.merge(
            copy.deepcopy(original_data), 
            copy.deepcopy(update_data)
        )
        
        print(f"  Original after merge: {original_data}")  # Unchanged
        print(f"  Safely merged: {safe_merged}")
    
    # Common use cases
    print("✓ Common use cases:")
    
    use_cases = [
        "Configuration management: env-specific overrides",
        "API response aggregation: combining partial data",
        "User preferences: defaults + customizations", 
        "Plugin systems: core + extension configurations",
        "Theme systems: base + variant + user customizations",
        "Feature flags: global + environment + user-specific",
        "Data pipeline: merging results from multiple sources"
    ]
    
    for use_case in use_cases:
        print(f"  • {use_case}")
    
    # Integration examples
    print("✓ Integration examples:")
    
    integrations = [
        "Django settings: BASE_SETTINGS + environment overrides",
        "Flask config: default + instance + environment configs",
        "FastAPI: OpenAPI schema merging for modular APIs",
        "Kubernetes: ConfigMap and Secret merging",
        "Docker Compose: base + override configurations",
        "Ansible: variable precedence and merging"
    ]
    
    for integration in integrations:
        print(f"  • {integration}")
    
    print("✓ Real-world applications and best practices completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: deepmerge is essential for configuration management")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"deepmerge example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
