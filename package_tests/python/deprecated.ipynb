## Example 1: Basic Function Deprecation

```python
import deprecated
import warnings

print("Example 1: Basic Function Deprecation")

try:
    print("✓ Basic function deprecation:")
    
    # 1. Simple deprecated function
    @deprecated.deprecated()
    def old_function():
        """This is a deprecated function."""
        return "Old functionality"
    
    @deprecated.deprecated(reason="Use new_function() instead")
    def legacy_calculator(a, b):
        """Deprecated calculator function."""
        return a + b
    
    # New replacement function
    def new_function():
        """This is the new improved function."""
        return "New improved functionality"
    
    def modern_calculator(a, b):
        """Modern calculator with validation."""
        if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
            raise TypeError("Arguments must be numbers")
        return a + b
    
    print("  Functions defined with @deprecated.deprecated() decorator")
    
    # 2. Capture deprecation warnings
    print("  Testing deprecated function calls:")
    
    # Enable all warnings to capture deprecation warnings
    warnings.filterwarnings("always", category=DeprecationWarning)
    
    # Capture warnings when calling deprecated functions
    with warnings.catch_warnings(record=True) as caught_warnings:
        warnings.simplefilter("always")
        
        # Call deprecated functions
        result1 = old_function()
        result2 = legacy_calculator(5, 3)
        
        print(f"    old_function() returned: '{result1}'")
        print(f"    legacy_calculator(5, 3) returned: {result2}")
        
        # Display captured warnings
        print(f"    Captured {len(caught_warnings)} deprecation warnings:")
        for warning in caught_warnings:
            print(f"      • {warning.category.__name__}: {warning.message}")
    
    # 3. Compare with modern alternatives
    print("  Modern alternatives:")
    new_result = new_function()
    modern_result = modern_calculator(5, 3)
    
    print(f"    new_function() returned: '{new_result}'")
    print(f"    modern_calculator(5, 3) returned: {modern_result}")
    
    # 4. Show decorator attributes
    print("  Decorator information:")
    
    if hasattr(old_function, '__deprecated__'):
        print(f"    old_function.__deprecated__: {old_function.__deprecated__}")
    
    print(f"    Function names preserved: {old_function.__name__}, {legacy_calculator.__name__}")
    print(f"    Docstrings preserved: {bool(old_function.__doc__)}, {bool(legacy_calculator.__doc__)}")
    
    print("✓ Basic function deprecation completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: deprecated package provides decorators for marking deprecated code")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"deprecated example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Class and Method Deprecation with Versioning

```python
import deprecated
import warnings

print("Example 2: Class and Method Deprecation with Versioning")

try:
    print("✓ Class and method deprecation with version information:")
    
    # 1. Deprecated class with version information
    @deprecated.deprecated(version='2.0.0', reason="Use ModernDataProcessor instead")
    class LegacyDataProcessor:
        """Deprecated data processor class."""
        
        def __init__(self, name):
            self.name = name
        
        @deprecated.deprecated(version='1.5.0', reason="Use process_data() instead")
        def old_process(self, data):
            """Deprecated processing method."""
            return f"Legacy processing: {data}"
        
        def process_data(self, data):
            """Current processing method."""
            return f"Current processing: {data}"
    
    # 2. Modern replacement class
    class ModernDataProcessor:
        """Modern data processor with improved functionality."""
        
        def __init__(self, name, config=None):
            self.name = name
            self.config = config or {}
        
        def process_data(self, data, validate=True):
            """Advanced data processing with validation."""
            if validate and not data:
                raise ValueError("Data cannot be empty")
            return f"Modern processing of '{data}' by {self.name}"
        
        def batch_process(self, data_list):
            """Batch processing capability."""
            return [self.process_data(item, validate=False) for item in data_list]
    
    # 3. Test deprecated class and methods
    print("  Testing deprecated class instantiation:")
    
    with warnings.catch_warnings(record=True) as caught_warnings:
        warnings.simplefilter("always")
        
        # Create deprecated class instance
        legacy_processor = LegacyDataProcessor("Legacy Processor")
        print(f"    Created: {legacy_processor.name}")
        
        # Call deprecated method
        result1 = legacy_processor.old_process("sample data")
        print(f"    old_process() result: {result1}")
        
        # Call current method
        result2 = legacy_processor.process_data("sample data")
        print(f"    process_data() result: {result2}")
        
        # Display warnings
        print(f"    Captured {len(caught_warnings)} warnings:")
        for warning in caught_warnings:
            print(f"      • {warning.message}")
    
    # 4. Compare with modern class
    print("  Modern class comparison:")
    
    modern_processor = ModernDataProcessor("Modern Processor", {"validation": True})
    modern_result = modern_processor.process_data("sample data")
    print(f"    Modern result: {modern_result}")
    
    # Batch processing example
    batch_data = ["item1", "item2", "item3"]
    batch_results = modern_processor.batch_process(batch_data)
    print(f"    Batch processing: {len(batch_results)} items processed")
    
    # 5. Inspect deprecation metadata
    print("  Deprecation metadata:")
    
    # Check if class has deprecation info
    if hasattr(LegacyDataProcessor, '__deprecated__'):
        print(f"    Class deprecated: {LegacyDataProcessor.__deprecated__}")
    
    # Check method deprecation
    if hasattr(LegacyDataProcessor.old_process, '__deprecated__'):
        print(f"    Method deprecated: {LegacyDataProcessor.old_process.__deprecated__}")
    
    # Show class comparison
    print(f"    Legacy class methods: {len([m for m in dir(LegacyDataProcessor) if not m.startswith('_')])}")
    print(f"    Modern class methods: {len([m for m in dir(ModernDataProcessor) if not m.startswith('_')])}")
    
    print("✓ Class and method deprecation with versioning completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: deprecated package supports class and method deprecation")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"deprecated example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: Custom Warning Categories and Advanced Configuration

```python
import deprecated
import warnings

print("Example 3: Custom Warning Categories and Advanced Configuration")

try:
    print("✓ Custom warning categories and configuration:")
    
    # 1. Custom warning category
    class APIDeprecationWarning(DeprecationWarning):
        """Custom warning category for API deprecations."""
        pass
    
    # 2. Functions with different deprecation configurations
    @deprecated.deprecated(
        version='1.0.0',
        reason="This API endpoint is being retired",
        category=APIDeprecationWarning
    )
    def deprecated_api_call(endpoint):
        """Deprecated API call function."""
        return f"API call to {endpoint} (deprecated)"
    
    @deprecated.deprecated(
        version='2.1.0',
        reason="Use async_process() for better performance",
        category=FutureWarning
    )
    def sync_process(data):
        """Synchronous processing (will be removed)."""
        return f"Sync processing: {data}"
    
    @deprecated.deprecated(
        reason="Internal function, use public API instead"
    )
    def _internal_helper(value):
        """Internal helper function."""
        return value * 2
    
    # 3. Test different warning categories
    print("  Testing custom warning categories:")
    
    # Configure warning filters for different categories
    warnings.filterwarnings("always", category=APIDeprecationWarning)
    warnings.filterwarnings("always", category=FutureWarning)
    warnings.filterwarnings("always", category=DeprecationWarning)
    
    with warnings.catch_warnings(record=True) as caught_warnings:
        warnings.simplefilter("always")
        
        # Call functions with different warning categories
        result1 = deprecated_api_call("/api/v1/users")
        result2 = sync_process("test data")
        result3 = _internal_helper(5)
        
        print(f"    deprecated_api_call() result: {result1}")
        print(f"    sync_process() result: {result2}")
        print(f"    _internal_helper() result: {result3}")
        
        # Analyze warnings by category
        warning_categories = {}
        for warning in caught_warnings:
            category = warning.category.__name__
            if category not in warning_categories:
                warning_categories[category] = []
            warning_categories[category].append(str(warning.message))
        
        print(f"    Warning analysis ({len(caught_warnings)} total):")
        for category, messages in warning_categories.items():
            print(f"      {category}: {len(messages)} warnings")
            for msg in messages:
                print(f"        • {msg[:60]}...")
    
    # 4. Deprecation with action recommendations
    @deprecated.deprecated(
        version='3.0.0',
        reason="Use NewConfigManager with better error handling and validation"
    )
    def create_config(settings):
        """Create configuration (deprecated)."""
        return {"config": settings, "version": "legacy"}
    
    def create_modern_config(settings, validate=True):
        """Create modern configuration with validation."""
        if validate:
            if not isinstance(settings, dict):
                raise TypeError("Settings must be a dictionary")
            if not settings:
                raise ValueError("Settings cannot be empty")
        
        return {
            "config": settings,
            "version": "modern",
            "validated": validate,
            "created_at": "2024-01-15"
        }
    
    # 5. Migration example
    print("  Migration example:")
    
    with warnings.catch_warnings(record=True) as migration_warnings:
        warnings.simplefilter("always")
        
        # Old way (deprecated)
        legacy_config = create_config({"debug": True, "timeout": 30})
        print(f"    Legacy config: {legacy_config}")
        
        # New way (recommended)
        modern_config = create_modern_config({"debug": True, "timeout": 30})
        print(f"    Modern config: {modern_config}")
        
        # Show migration guidance
        if migration_warnings:
            print("    Migration guidance from warnings:")
            for warning in migration_warnings:
                print(f"      • {warning.message}")
    
    # 6. Decorator inspection
    print("  Decorator inspection:")
    
    deprecated_funcs = [deprecated_api_call, sync_process, create_config]
    for func in deprecated_funcs:
        print(f"    Function: {func.__name__}")
        if hasattr(func, '__deprecated__'):
            print(f"      Deprecated: {func.__deprecated__}")
        if hasattr(func, '__wrapped__'):
            print(f"      Original function preserved: {func.__wrapped__.__name__}")
    
    print("✓ Custom warning categories and configuration completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: deprecated package supports custom warning categories")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"deprecated example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: Property and Attribute Deprecation

```python
import deprecated
import warnings

print("Example 4: Property and Attribute Deprecation")

try:
    print("✓ Property and attribute deprecation:")
    
    # 1. Class with deprecated properties
    class ConfigurationManager:
        """Configuration manager with deprecated and modern properties."""
        
        def __init__(self):
            self._timeout = 30
            self._max_retries = 3
            self._debug_enabled = False
            self._connection_string = "localhost:5432"
        
        # Deprecated property
        @property
        @deprecated.deprecated(version='2.0.0', reason="Use timeout_seconds instead")
        def timeout(self):
            """Deprecated timeout property (in seconds)."""
            return self._timeout
        
        @timeout.setter
        @deprecated.deprecated(version='2.0.0', reason="Use timeout_seconds instead") 
        def timeout(self, value):
            """Deprecated timeout setter."""
            self._timeout = value
        
        # Modern replacement property
        @property
        def timeout_seconds(self):
            """Timeout in seconds (current API)."""
            return self._timeout
        
        @timeout_seconds.setter
        def timeout_seconds(self, value):
            """Set timeout in seconds with validation."""
            if not isinstance(value, (int, float)) or value <= 0:
                raise ValueError("Timeout must be a positive number")
            self._timeout = value
        
        # Deprecated property with different naming
        @property
        @deprecated.deprecated(reason="Use retry_limit instead for clarity")
        def max_retries(self):
            """Maximum retry attempts (deprecated naming)."""
            return self._max_retries
        
        @property
        def retry_limit(self):
            """Maximum retry attempts (clear naming)."""
            return self._max_retries
        
        @retry_limit.setter
        def retry_limit(self, value):
            """Set retry limit with validation."""
            if not isinstance(value, int) or value < 0:
                raise ValueError("Retry limit must be a non-negative integer")
            self._max_retries = value
        
        # Method that uses deprecated internal property
        @deprecated.deprecated(reason="Use get_connection_info() instead")
        def get_connection_string(self):
            """Get database connection string (deprecated)."""
            return self._connection_string
        
        def get_connection_info(self):
            """Get connection information (modern API)."""
            return {
                "host": "localhost",
                "port": 5432,
                "connection_string": self._connection_string,
                "timeout": self._timeout
            }
    
    # 2. Test deprecated properties
    print("  Testing deprecated properties:")
    
    config = ConfigurationManager()
    
    with warnings.catch_warnings(record=True) as caught_warnings:
        warnings.simplefilter("always")
        
        # Access deprecated properties (getters)
        old_timeout = config.timeout
        old_retries = config.max_retries
        
        print(f"    Deprecated timeout: {old_timeout} seconds")
        print(f"    Deprecated max_retries: {old_retries}")
        
        # Set deprecated properties (setters)
        config.timeout = 45
        
        print(f"    Set deprecated timeout to: {config.timeout}")
        
        # Display property warnings
        property_warnings = [w for w in caught_warnings if 'property' in str(w.message).lower() or 'timeout' in str(w.message).lower()]
        print(f"    Property deprecation warnings: {len(property_warnings)}")
        for warning in property_warnings:
            print(f"      • {warning.message}")
    
    # 3. Compare with modern properties
    print("  Modern property alternatives:")
    
    # Use modern properties
    modern_timeout = config.timeout_seconds
    modern_retries = config.retry_limit
    
    print(f"    Modern timeout_seconds: {modern_timeout}")
    print(f"    Modern retry_limit: {modern_retries}")
    
    # Set modern properties with validation
    config.timeout_seconds = 60
    config.retry_limit = 5
    
    print(f"    Updated timeout_seconds: {config.timeout_seconds}")
    print(f"    Updated retry_limit: {config.retry_limit}")
    
    # 4. Test deprecated methods
    print("  Deprecated method comparison:")
    
    with warnings.catch_warnings(record=True) as method_warnings:
        warnings.simplefilter("always")
        
        # Call deprecated method
        old_connection = config.get_connection_string()
        print(f"    Deprecated connection string: {old_connection}")
        
        # Call modern method
        modern_info = config.get_connection_info()
        print(f"    Modern connection info: {modern_info}")
        
        # Show method warnings
        if method_warnings:
            print(f"    Method warnings: {len(method_warnings)}")
            for warning in method_warnings:
                print(f"      • {warning.message}")
    
    # 5. Property introspection
    print("  Property introspection:")
    
    # List all properties
    properties = [name for name in dir(ConfigurationManager) 
                 if isinstance(getattr(ConfigurationManager, name, None), property)]
    print(f"    Total properties: {len(properties)}")
    print(f"    Property names: {properties}")
    
    # Check which properties are deprecated
    deprecated_properties = []
    for prop_name in properties:
        prop = getattr(ConfigurationManager, prop_name)
        if hasattr(prop.fget, '__deprecated__'):
            deprecated_properties.append(prop_name)
    
    print(f"    Deprecated properties: {deprecated_properties}")
    print(f"    Modern properties: {[p for p in properties if p not in deprecated_properties]}")
    
    print("✓ Property and attribute deprecation completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: deprecated package works with properties and attributes")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"deprecated example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: deprecated Package Capabilities and Best Practices

```python
import deprecated
import warnings
import inspect

print("Example 5: deprecated Package Capabilities and Best Practices")

try:
    print("✓ deprecated package capabilities and best practices:")
    
    # 1. Package overview and capabilities
    print("  1. Package capabilities:")
    print(f"    Package name: {deprecated.__name__}")
    
    # List available decorators and utilities
    public_attrs = [attr for attr in dir(deprecated) if not attr.startswith('_')]
    print(f"    Public attributes: {public_attrs}")
    
    # Show main decorator
    if hasattr(deprecated, 'deprecated'):
        print(f"    Main decorator: {deprecated.deprecated}")
        print(f"    Decorator type: {type(deprecated.deprecated)}")
    
    # 2. Complete deprecation workflow example
    print("  2. Complete deprecation workflow:")
    
    # Legacy API being deprecated
    @deprecated.deprecated(
        version='1.0.0',
        reason="This function will be removed in v2.0. Use calculate_metrics() instead."
    )
    def old_calculate(data, method='avg'):
        """Legacy calculation function."""
        if method == 'avg':
            return sum(data) / len(data) if data else 0
        elif method == 'sum':
            return sum(data)
        else:
            return data[0] if data else 0
    
    # Modern replacement API
    def calculate_metrics(data, operation='mean'):
        """Modern metrics calculation with validation and extended operations."""
        if not isinstance(data, (list, tuple)):
            raise TypeError("Data must be a list or tuple")
        if not data:
            return 0
        
        if operation == 'mean':
            return sum(data) / len(data)
        elif operation == 'sum':
            return sum(data)
        elif operation == 'min':
            return min(data)
        elif operation == 'max':
            return max(data)
        elif operation == 'count':
            return len(data)
        else:
            raise ValueError(f"Unsupported operation: {operation}")
    
    # Test the migration workflow
    test_data = [1, 2, 3, 4, 5]
    
    with warnings.catch_warnings(record=True) as workflow_warnings:
        warnings.simplefilter("always")
        
        # Old way (deprecated)
        old_result = old_calculate(test_data, 'avg')
        print(f"    Legacy result: {old_result}")
        
        # New way (recommended)
        new_result = calculate_metrics(test_data, 'mean')
        extended_results = {
            'mean': calculate_metrics(test_data, 'mean'),
            'sum': calculate_metrics(test_data, 'sum'),
            'min': calculate_metrics(test_data, 'min'),
            'max': calculate_metrics(test_data, 'max')
        }
        
        print(f"    Modern result: {new_result}")
        print(f"    Extended capabilities: {extended_results}")
        
        # Show migration warning
        if workflow_warnings:
            print("    Migration advice:")
            for warning in workflow_warnings:
                print(f"      • {warning.message}")
    
    # 3. Best practices demonstration
    print("  3. Deprecation best practices:")
    
    best_practices = [
        "Always provide a clear reason for deprecation",
        "Specify version information when possible", 
        "Suggest specific alternatives or migration paths",
        "Use appropriate warning categories",
        "Maintain backward compatibility during deprecation period",
        "Document the deprecation in code and external docs",
        "Provide transition tools or utilities when needed",
        "Plan removal timeline and communicate it clearly"
    ]
    
    for i, practice in enumerate(best_practices, 1):
        print(f"    {i}. {practice}")
    
    # 4. Warning management and filtering
    print("  4. Warning management:")
    
    # Show how to handle warnings in production
    production_examples = {
        "Ignore specific deprecations": "warnings.filterwarnings('ignore', category=DeprecationWarning, module='myapp.legacy')",
        "Log warnings instead of printing": "logging.captureWarnings(True)",
        "Convert warnings to exceptions": "warnings.filterwarnings('error', category=DeprecationWarning)",
        "Show only once per location": "warnings.filterwarnings('once', category=DeprecationWarning)"
    }
    
    for use_case, code in production_examples.items():
        print(f"    {use_case}: {code}")
    
    # 5. Decorator introspection and metadata
    print("  5. Deprecation metadata inspection:")
    
    # Inspect deprecated function metadata
    deprecated_func = old_calculate
    
    print(f"    Function: {deprecated_func.__name__}")
    print(f"    Original name preserved: {deprecated_func.__name__}")
    print(f"    Documentation preserved: {bool(deprecated_func.__doc__)}")
    
    # Check for deprecation attributes
    if hasattr(deprecated_func, '__deprecated__'):
        print(f"    Deprecation marker: {deprecated_func.__deprecated__}")
    
    if hasattr(deprecated_func, '__wrapped__'):
        wrapped = deprecated_func.__wrapped__
        print(f"    Wrapped function: {wrapped.__name__}")
        print(f"    Signature preserved: {inspect.signature(wrapped)}")
    
    # Get function signature
    sig = inspect.signature(deprecated_func)
    print(f"    Function signature: {sig}")
    
    # 6. Usage statistics and recommendations
    print("  6. Package usage recommendations:")
    
    recommendations = [
        "Use deprecated package in libraries to maintain API compatibility",
        "Combine with semantic versioning for clear deprecation timeline", 
        "Integrate with CI/CD to catch deprecated API usage",
        "Use custom warning categories for different types of deprecations",
        "Document migration paths in deprecation messages",
        "Consider gradual deprecation for complex APIs",
        "Test deprecated code paths until removal",
        "Monitor deprecated API usage in production logs"
    ]
    
    for rec in recommendations:
        print(f"      • {rec}")
    
    # 7. Final verification
    print("  7. Package verification:")
    
    # Verify the package is working correctly
    verification_results = {
        "decorator_available": hasattr(deprecated, 'deprecated'),
        "warnings_captured": len(workflow_warnings) > 0,
        "metadata_preserved": hasattr(old_calculate, '__name__'),
        "functionality_intact": old_calculate([1, 2, 3]) == 2.0
    }
    
    print(f"    Verification results: {verification_results}")
    all_passed = all(verification_results.values())
    print(f"    All tests passed: {all_passed}")
    
    print("✓ deprecated package capabilities and best practices completed successfully")
    print("  → Ready for production use in API deprecation management")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: Install deprecated package: pip install Deprecated")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"deprecated example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
