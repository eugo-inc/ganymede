## Example 1: Basic Deprecation Warnings

```python
import deprecation
import warnings

print("Example 1: Basic Deprecation Warnings")

try:
    print("✓ Basic deprecation warnings with PEP 387 compliance:")
    
    # 1. Simple deprecation decorator
    @deprecation.deprecated(
        deprecated_in="1.0",
        removed_in="2.0", 
        current_version="1.5",
        details="Use new_api_function() instead for better performance"
    )
    def legacy_api_function(data):
        """Legacy API function that will be removed."""
        return f"Processing {data} with legacy method"
    
    # 2. Modern replacement function
    def new_api_function(data, options=None):
        """Modern API function with enhanced capabilities."""
        options = options or {}
        processing_type = options.get('type', 'standard')
        return f"Processing {data} with {processing_type} method (modern)"
    
    # 3. Test deprecation warnings
    print("  Testing deprecation warnings:")
    
    # Enable deprecation warnings
    warnings.filterwarnings("always", category=DeprecationWarning)
    
    with warnings.catch_warnings(record=True) as caught_warnings:
        warnings.simplefilter("always")
        
        # Call deprecated function
        result1 = legacy_api_function("sample data")
        print(f"    Legacy result: {result1}")
        
        # Call modern function
        result2 = new_api_function("sample data", {"type": "enhanced"})
        print(f"    Modern result: {result2}")
        
        # Display captured warnings
        print(f"    Captured {len(caught_warnings)} warnings:")
        for warning in caught_warnings:
            print(f"      • Category: {warning.category.__name__}")
            print(f"        Message: {warning.message}")
            print(f"        File: {warning.filename}:{warning.lineno}")
    
    # 4. Package information
    print("  Package information:")
    print(f"    Module name: {deprecation.__name__}")
    
    # List available functions
    public_funcs = [attr for attr in dir(deprecation) 
                   if not attr.startswith('_') and callable(getattr(deprecation, attr))]
    print(f"    Available functions: {public_funcs}")
    
    # 5. Deprecation decorator inspection
    print("  Decorator inspection:")
    
    if hasattr(legacy_api_function, '__deprecated__'):
        print(f"    Function marked as deprecated: {legacy_api_function.__deprecated__}")
    
    print(f"    Function name preserved: {legacy_api_function.__name__}")
    print(f"    Docstring preserved: {bool(legacy_api_function.__doc__)}")
    
    # Check for version information
    if hasattr(legacy_api_function, '_deprecated_details'):
        details = legacy_api_function._deprecated_details
        print(f"    Deprecation details available: {details}")
    
    print("✓ Basic deprecation warnings completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: deprecation package provides PEP 387 compliant deprecation warnings")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"deprecation example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Version-Based Deprecation Management

```python
import deprecation
import warnings

print("Example 2: Version-Based Deprecation Management")

try:
    print("✓ Version-based deprecation with timeline management:")
    
    # Define current package version
    CURRENT_VERSION = "2.1.0"
    
    # 1. Functions at different deprecation stages
    @deprecation.deprecated(
        deprecated_in="1.5.0",
        removed_in="2.0.0",
        current_version=CURRENT_VERSION,
        details="Function was removed - upgrade to use replacement_function()"
    )
    def old_removed_function():
        """Function that should have been removed."""
        return "This should not be called"
    
    @deprecation.deprecated(
        deprecated_in="2.0.0", 
        removed_in="3.0.0",
        current_version=CURRENT_VERSION,
        details="Use enhanced_processor() for better performance and features"
    )
    def basic_processor(data):
        """Basic processor that will be removed in v3.0."""
        return f"Basic processing: {data}"
    
    @deprecation.deprecated(
        deprecated_in="2.1.0",
        removed_in="3.5.0", 
        current_version=CURRENT_VERSION,
        details="Migrate to async_handler() for non-blocking operations"
    )
    def sync_handler(request):
        """Synchronous handler (recently deprecated)."""
        return f"Handling request: {request} (sync)"
    
    # 2. Modern replacement functions
    def enhanced_processor(data, config=None):
        """Enhanced processor with configuration options."""
        config = config or {"optimization": True}
        optimization = "optimized" if config.get("optimization") else "standard"
        return f"Enhanced processing: {data} ({optimization})"
    
    async def async_handler(request):
        """Asynchronous handler for non-blocking operations."""
        return f"Handling request: {request} (async)"
    
    # 3. Test different deprecation stages
    print("  Testing different deprecation stages:")
    
    with warnings.catch_warnings(record=True) as stage_warnings:
        warnings.simplefilter("always")
        
        # Call functions at different stages
        try:
            result1 = basic_processor("test data")
            print(f"    Recently deprecated (v2.0 → v3.0): {result1}")
        except Exception as e:
            print(f"    Recently deprecated failed: {e}")
            raise
        
        try:
            result2 = sync_handler("test request") 
            print(f"    Newly deprecated (v2.1 → v3.5): {result2}")
        except Exception as e:
            print(f"    Newly deprecated failed: {e}")
            raise
        
        # Display warnings with version context
        print(f"    Version-based warnings ({len(stage_warnings)} total):")
        for warning in stage_warnings:
            message = str(warning.message)
            print(f"      • {message}")
    
    # 4. Version comparison utilities
    print("  Version comparison:")
    
    def parse_version(version_str):
        """Simple version parser."""
        return tuple(map(int, version_str.split('.')))
    
    def should_warn(deprecated_in, removed_in, current):
        """Check if deprecation warning should be shown."""
        dep_version = parse_version(deprecated_in)
        rem_version = parse_version(removed_in)
        cur_version = parse_version(current)
        
        return dep_version <= cur_version < rem_version
    
    # Test version logic
    test_cases = [
        ("1.5.0", "2.0.0", CURRENT_VERSION),  # Should not warn (past removal)
        ("2.0.0", "3.0.0", CURRENT_VERSION),  # Should warn (active deprecation)
        ("2.1.0", "3.5.0", CURRENT_VERSION),  # Should warn (newly deprecated)
        ("2.2.0", "3.0.0", CURRENT_VERSION),  # Should not warn (not deprecated yet)
    ]
    
    for dep_in, rem_in, current in test_cases:
        should_show = should_warn(dep_in, rem_in, current)
        status = "WARN" if should_show else "SILENT"
        print(f"    {dep_in} → {rem_in} (current: {current}): {status}")
    
    # 5. Migration timeline
    print("  Migration timeline:")
    
    migration_plan = {
        "v1.5.0": "old_function deprecated",
        "v2.0.0": "old_function removed, basic_processor deprecated", 
        "v2.1.0": "sync_handler deprecated",
        "v3.0.0": "basic_processor will be removed",
        "v3.5.0": "sync_handler will be removed"
    }
    
    for version, action in migration_plan.items():
        current_marker = " ← CURRENT" if version == f"v{CURRENT_VERSION}" else ""
        print(f"    {version}: {action}{current_marker}")
    
    # 6. Best practices demonstration
    print("  Best practices for version-based deprecation:")
    
    best_practices = [
        "Always specify deprecated_in, removed_in, and current_version",
        "Follow semantic versioning for deprecation timeline",
        "Provide clear migration instructions in details",
        "Test both deprecated and replacement code paths",
        "Document version changes in CHANGELOG", 
        "Give users adequate time between deprecation and removal",
        "Use major version bumps for breaking changes",
        "Monitor usage metrics before removing deprecated features"
    ]
    
    for i, practice in enumerate(best_practices, 1):
        print(f"      {i}. {practice}")
    
    print("✓ Version-based deprecation management completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: deprecation package supports version-based deprecation management")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"deprecation example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: Class and Method Deprecation with PEP 387

```python
import deprecation
import warnings

print("Example 3: Class and Method Deprecation with PEP 387")

try:
    print("✓ Class and method deprecation following PEP 387:")
    
    PACKAGE_VERSION = "3.2.0"
    
    # 1. Deprecated class with version information
    @deprecation.deprecated(
        deprecated_in="3.0.0",
        removed_in="4.0.0", 
        current_version=PACKAGE_VERSION,
        details="LegacyDataStore is replaced by ModernDataStore with better performance and features"
    )
    class LegacyDataStore:
        """Deprecated data storage class."""
        
        def __init__(self, name):
            self.name = name
            self.data = {}
        
        @deprecation.deprecated(
            deprecated_in="2.5.0",
            removed_in="3.5.0",
            current_version=PACKAGE_VERSION,
            details="Use store_item() method instead"
        )
        def add_item(self, key, value):
            """Deprecated method for adding items."""
            self.data[key] = value
            return f"Added {key} to {self.name}"
        
        def store_item(self, key, value, metadata=None):
            """Current method for storing items with optional metadata."""
            self.data[key] = {"value": value, "metadata": metadata or {}}
            return f"Stored {key} in {self.name} with metadata"
        
        def get_item(self, key):
            """Get stored item."""
            return self.data.get(key)
    
    # 2. Modern replacement class
    class ModernDataStore:
        """Modern data storage with enhanced capabilities."""
        
        def __init__(self, name, config=None):
            self.name = name
            self.config = config or {}
            self.data = {}
            self.indexes = {}
        
        def store_item(self, key, value, metadata=None, index=True):
            """Store item with optional indexing and metadata."""
            item_data = {
                "value": value,
                "metadata": metadata or {},
                "timestamp": "2024-01-15T10:30:00Z"
            }
            self.data[key] = item_data
            
            if index and "searchable" in (metadata or {}):
                self.indexes[key] = value
            
            return f"Stored {key} in {self.name} (indexed: {index})"
        
        def get_item(self, key, include_metadata=False):
            """Get item with optional metadata."""
            item = self.data.get(key)
            if item and include_metadata:
                return item
            elif item:
                return item["value"]
            return None
        
        def search_items(self, query):
            """Search through indexed items."""
            results = []
            for key, value in self.indexes.items():
                if query.lower() in str(value).lower():
                    results.append((key, value))
            return results
    
    # 3. Test deprecated class and methods
    print("  Testing deprecated class and methods:")
    
    with warnings.catch_warnings(record=True) as class_warnings:
        warnings.simplefilter("always")
        
        # Create deprecated class instance (should warn)
        legacy_store = LegacyDataStore("Legacy Store")
        print(f"    Created legacy store: {legacy_store.name}")
        
        # Call deprecated method (should warn)
        result1 = legacy_store.add_item("item1", "value1")
        print(f"    Deprecated add_item(): {result1}")
        
        # Call current method (no warning)
        result2 = legacy_store.store_item("item2", "value2", {"type": "test"})
        print(f"    Current store_item(): {result2}")
        
        # Display class-level warnings
        print(f"    Class/method warnings ({len(class_warnings)} total):")
        for warning in class_warnings:
            message = str(warning.message)
            if "LegacyDataStore" in message:
                print(f"      CLASS: {message[:80]}...")
            else:
                print(f"      METHOD: {message[:80]}...")
    
    # 4. Compare with modern class
    print("  Modern class comparison:")
    
    modern_store = ModernDataStore("Modern Store", {"indexing": True})
    modern_result1 = modern_store.store_item("item1", "modern_value1", {"searchable": True})
    modern_result2 = modern_store.store_item("item2", "modern_value2", {"type": "example"})
    
    print(f"    Modern store_item() 1: {modern_result1}")
    print(f"    Modern store_item() 2: {modern_result2}")
    
    # Show enhanced capabilities
    search_results = modern_store.search_items("modern")
    print(f"    Search capability: {len(search_results)} results found")
    
    item_with_metadata = modern_store.get_item("item1", include_metadata=True)
    print(f"    Enhanced retrieval: {type(item_with_metadata)} with timestamp")
    
    # 5. Inheritance and deprecation
    print("  Inheritance patterns:")
    
    @deprecation.deprecated(
        deprecated_in="3.1.0",
        removed_in="4.5.0",
        current_version=PACKAGE_VERSION,
        details="BaseProcessor is being replaced by AbstractProcessor interface"
    )
    class BaseProcessor:
        """Deprecated base class."""
        
        def process(self, data):
            return f"Base processing: {data}"
    
    class AbstractProcessor:
        """Modern abstract interface."""
        
        def process(self, data):
            raise NotImplementedError("Subclasses must implement process()")
        
        def validate(self, data):
            """Validation method available in modern interface."""
            return isinstance(data, str) and len(data) > 0
    
    class ConcreteProcessor(AbstractProcessor):
        """Concrete implementation of modern interface."""
        
        def process(self, data):
            if self.validate(data):
                return f"Modern processing: {data} (validated)"
            else:
                raise ValueError("Invalid data provided")
    
    # Test inheritance pattern
    with warnings.catch_warnings(record=True) as inheritance_warnings:
        warnings.simplefilter("always")
        
        # This should warn about deprecated base class
        try:
            legacy_processor = BaseProcessor()
            legacy_result = legacy_processor.process("test")
            print(f"    Legacy processor: {legacy_result}")
        except Exception as e:
            print(f"    Legacy processor failed: {e}")
            raise
        
        # Modern approach
        modern_processor = ConcreteProcessor()
        modern_result = modern_processor.process("test")
        print(f"    Modern processor: {modern_result}")
        
        # Show inheritance warnings
        if inheritance_warnings:
            print(f"    Inheritance warnings: {len(inheritance_warnings)}")
            for warning in inheritance_warnings:
                print(f"      • {warning.message}")
    
    print("✓ Class and method deprecation with PEP 387 completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: deprecation package supports class and method deprecation")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"deprecation example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: Warning Management and Production Usage

```python
import deprecation
import warnings
import logging

print("Example 4: Warning Management and Production Usage")

try:
    print("✓ Warning management for production environments:")
    
    # 1. Configure logging for deprecation warnings
    print("  1. Logging configuration:")
    
    # Set up logging
    logging.basicConfig(level=logging.WARNING, format='%(levelname)s - %(message)s')
    logger = logging.getLogger(__name__)
    
    # Capture warnings in logging system
    logging.captureWarnings(True)
    warnings_logger = logging.getLogger('py.warnings')
    
    print("    Logging configured to capture warnings")
    print("    Deprecation warnings will be logged instead of printed")
    
    # 2. Different warning filter strategies
    print("  2. Warning filter strategies:")
    
    @deprecation.deprecated(
        deprecated_in="1.0.0",
        removed_in="2.0.0",
        current_version="1.5.0",
        details="Use production_ready_function() instead"
    )
    def test_function_alpha():
        """Test function for demonstration."""
        return "alpha result"
    
    @deprecation.deprecated(
        deprecated_in="1.2.0", 
        removed_in="2.5.0",
        current_version="1.5.0",
        details="Use production_ready_function() instead"
    )
    def test_function_beta():
        """Another test function."""
        return "beta result"
    
    def production_ready_function():
        """Production-ready replacement function."""
        return "production result"
    
    # Test different filter strategies
    filter_strategies = {
        "ignore_all": lambda: warnings.filterwarnings("ignore", category=DeprecationWarning),
        "show_once": lambda: warnings.filterwarnings("once", category=DeprecationWarning), 
        "show_always": lambda: warnings.filterwarnings("always", category=DeprecationWarning),
        "error_on_deprecation": lambda: warnings.filterwarnings("error", category=DeprecationWarning)
    }
    
    for strategy_name, configure_filter in filter_strategies.items():
        print(f"    Testing strategy: {strategy_name}")
        
        # Reset and apply filter
        warnings.resetwarnings()
        configure_filter()
        
        try:
            with warnings.catch_warnings(record=True) as strategy_warnings:
                if strategy_name != "error_on_deprecation":
                    # Call deprecated functions
                    result1 = test_function_alpha()
                    result2 = test_function_beta()
                    result3 = test_function_alpha()  # Second call to same function
                    
                    print(f"      Results: {result1}, {result2}, {result3}")
                    print(f"      Warnings captured: {len(strategy_warnings)}")
                else:
                    # The "error" filter turns the deprecation warning into a raised deprecation.DeprecatedWarning
                    try:
                        result = test_function_alpha()
                    except deprecation.DeprecatedWarning as e:
                        print(f"      Raised as expected: {type(e).__name__}")
                    else:
                        raise AssertionError(
                            f"the 'error' filter did not raise on a deprecated call; it returned {result!r}"
                        )
        except Exception as e:
            print(f"      Exception during strategy test: {e}")
            raise  # the assertion above, or any unexpected failure, must fail the example rather than print past it
    
    # 3. Custom warning handling
    print("  3. Custom warning handling:")
    
    class DeprecationTracker:
        """Track deprecation warnings for analysis."""
        
        def __init__(self):
            self.warnings_count = {}
            self.first_occurrence = {}
        
        def handle_warning(self, message, category, filename, lineno, file=None, line=None):
            """Custom warning handler."""
            key = str(message)
            self.warnings_count[key] = self.warnings_count.get(key, 0) + 1
            
            if key not in self.first_occurrence:
                self.first_occurrence[key] = {
                    "filename": filename,
                    "lineno": lineno,
                    "category": category.__name__
                }
            
            # Log the warning
            logger.warning(f"DEPRECATION [{self.warnings_count[key]}]: {message}")
        
        def get_report(self):
            """Generate deprecation usage report."""
            return {
                "total_warnings": sum(self.warnings_count.values()),
                "unique_warnings": len(self.warnings_count),
                "warning_counts": self.warnings_count.copy(),
                "first_occurrences": self.first_occurrence.copy()
            }
    
    # Use custom handler
    tracker = DeprecationTracker()
    
    # Install custom warning handler
    original_showwarning = warnings.showwarning
    warnings.showwarning = tracker.handle_warning
    warnings.filterwarnings("always", category=DeprecationWarning)
    
    try:
        # Generate some deprecation warnings
        for i in range(3):
            result_alpha = test_function_alpha()
            result_beta = test_function_beta()
        
        # Generate report
        report = tracker.get_report()
        print(f"    Deprecation report:")
        print(f"      Total warnings: {report['total_warnings']}")
        print(f"      Unique warnings: {report['unique_warnings']}")
        
        for warning_msg, count in report['warning_counts'].items():
            print(f"      '{warning_msg[:50]}...': {count} occurrences")
    
    finally:
        # Restore original warning handler
        warnings.showwarning = original_showwarning
    
    # 4. Production deployment considerations
    print("  4. Production deployment patterns:")
    
    production_patterns = [
        "Environment-based warning filters (suppress in prod, show in dev)",
        "Structured logging for deprecation warnings analysis",
        "Metrics collection for deprecated API usage tracking",
        "Gradual migration alerts and dashboards",
        "Automated testing to catch new deprecation warnings",
        "Version-aware warning suppression during upgrades",
        "User notification systems for breaking changes",
        "Documentation integration for migration guides"
    ]
    
    for i, pattern in enumerate(production_patterns, 1):
        print(f"      {i}. {pattern}")
    
    # 5. Migration helper utilities
    print("  5. Migration helper example:")
    
    class MigrationHelper:
        """Helper class for managing API migrations."""
        
        def __init__(self, current_version):
            self.current_version = current_version
            self.migration_map = {}
        
        def register_migration(self, old_func, new_func, deprecated_in, removed_in):
            """Register a function migration."""
            self.migration_map[old_func.__name__] = {
                "new_func": new_func,
                "deprecated_in": deprecated_in,
                "removed_in": removed_in,
                "old_func": old_func
            }
        
        def get_migrations_needed(self):
            """Get list of migrations needed for current version."""
            needed = []
            for func_name, info in self.migration_map.items():
                if self._should_migrate(info["deprecated_in"], info["removed_in"]):
                    needed.append({
                        "old_function": func_name,
                        "new_function": info["new_func"].__name__,
                        "deadline": info["removed_in"]
                    })
            return needed
        
        def _should_migrate(self, deprecated_in, removed_in):
            """Check if migration is needed."""
            return True  # Simplified logic
    
    # Test migration helper
    helper = MigrationHelper("1.5.0")
    helper.register_migration(test_function_alpha, production_ready_function, "1.0.0", "2.0.0")
    helper.register_migration(test_function_beta, production_ready_function, "1.2.0", "2.5.0")
    
    migrations_needed = helper.get_migrations_needed()
    print(f"    Migrations needed: {len(migrations_needed)}")
    for migration in migrations_needed:
        print(f"      {migration['old_function']} → {migration['new_function']} (deadline: {migration['deadline']})")
    
    print("✓ Warning management and production usage completed successfully")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: deprecation package supports production warning management")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"deprecation example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: deprecation Package Capabilities and PEP 387 Compliance

```python
import deprecation
import warnings
import inspect

print("Example 5: deprecation Package Capabilities and PEP 387 Compliance")

try:
    print("✓ deprecation package capabilities and PEP 387 compliance:")
    
    # 1. Package overview and PEP 387 features
    print("  1. Package and PEP 387 overview:")
    print(f"    Package name: {deprecation.__name__}")
    
    # List package capabilities
    public_attrs = [attr for attr in dir(deprecation) if not attr.startswith('_')]
    print(f"    Public attributes: {public_attrs}")
    
    # Show main decorator
    if hasattr(deprecation, 'deprecated'):
        print(f"    Main decorator: {deprecation.deprecated}")
        print(f"    Decorator type: {type(deprecation.deprecated)}")
    
    # PEP 387 compliance features
    pep387_features = [
        "Standardized deprecation warning format",
        "Version-based deprecation timeline",
        "Clear removal scheduling",
        "Migration guidance in warnings",
        "Consistent warning categories", 
        "Documentation integration support",
        "Backward compatibility preservation",
        "Gradual migration pathway"
    ]
    
    print("    PEP 387 compliance features:")
    for feature in pep387_features:
        print(f"      • {feature}")
    
    # 2. Complete PEP 387 compliant example
    print("  2. Complete PEP 387 workflow:")
    
    # Current package version
    PACKAGE_VERSION = "4.1.0"
    
    # Function following complete PEP 387 lifecycle
    @deprecation.deprecated(
        deprecated_in="3.0.0",
        removed_in="5.0.0",
        current_version=PACKAGE_VERSION,
        details="calculate_mean() is deprecated and will be removed in v5.0. "
                "Use statistics.mean() from the standard library or "
                "numpy.mean() for array operations. "
                "Migration guide: https://example.com/migration-guide"
    )
    def calculate_mean(numbers):
        """Calculate arithmetic mean (deprecated - use statistics.mean)."""
        return sum(numbers) / len(numbers) if numbers else 0
    
    # Modern alternatives
    import statistics
    
    def modern_calculate_stats(numbers):
        """Modern statistics calculation with multiple metrics."""
        if not numbers:
            return {"mean": 0, "median": 0, "mode": None}
        
        return {
            "mean": statistics.mean(numbers),
            "median": statistics.median(numbers), 
            "mode": statistics.mode(numbers) if len(set(numbers)) < len(numbers) else None,
            "count": len(numbers)
        }
    
    # 3. Test complete PEP 387 workflow
    print("  Testing PEP 387 compliant deprecation:")
    
    test_data = [1, 2, 3, 4, 5, 3, 2]
    
    with warnings.catch_warnings(record=True) as pep387_warnings:
        warnings.simplefilter("always")
        
        # Call deprecated function (PEP 387 compliant)
        deprecated_result = calculate_mean(test_data)
        print(f"    Deprecated result: {deprecated_result}")
        
        # Call modern alternatives
        modern_result = modern_calculate_stats(test_data)
        stdlib_result = statistics.mean(test_data)
        
        print(f"    Modern stats: {modern_result}")
        print(f"    Standard library: {stdlib_result}")
        
        # Analyze PEP 387 compliant warnings
        print(f"    PEP 387 warnings ({len(pep387_warnings)} total):")
        for warning in pep387_warnings:
            message = str(warning.message)
            print(f"      Message structure analysis:")
            # deprecation 2.1.0 words the version as "deprecated as of <version>" (measured), not "deprecated in"
            print(f"        • Contains version info: {'deprecated as of' in message.lower()}")
            print(f"        • Contains removal timeline: {'removed in' in message.lower()}")
            print(f"        • Contains migration guidance: {'migration' in message.lower()}")
            print(f"        • Full message: {message}")
    
    # 4. Decorator introspection and metadata
    print("  4. Deprecation metadata inspection:")
    
    # Inspect deprecated function
    deprecated_func = calculate_mean
    
    print(f"    Function: {deprecated_func.__name__}")
    print(f"    Signature: {inspect.signature(deprecated_func)}")
    print(f"    Docstring preserved: {bool(deprecated_func.__doc__)}")
    
    # Check deprecation metadata
    if hasattr(deprecated_func, '__deprecated__'):
        print(f"    Deprecation marker: {deprecated_func.__deprecated__}")
    
    if hasattr(deprecated_func, '__wrapped__'):
        wrapped = deprecated_func.__wrapped__
        print(f"    Original function: {wrapped.__name__}")
        print(f"    Original signature: {inspect.signature(wrapped)}")
    
    # Extract version information if available
    func_dict = getattr(deprecated_func, '__dict__', {})
    version_info = {k: v for k, v in func_dict.items() if 'version' in k.lower()}
    if version_info:
        print(f"    Version metadata: {version_info}")
    
    # 5. Best practices and recommendations
    print("  5. PEP 387 best practices:")
    
    best_practices = [
        "Always specify deprecated_in, removed_in, and current_version",
        "Provide at least two major versions between deprecation and removal",
        "Include specific migration instructions with alternatives",
        "Use clear, actionable language in deprecation messages",
        "Document deprecations in CHANGELOG and release notes",
        "Provide migration tools or scripts when possible",
        "Test deprecated code paths until actual removal",
        "Follow semantic versioning for deprecation lifecycle",
        "Consider user impact and adoption timeline",
        "Maintain backward compatibility during deprecation period"
    ]
    
    for i, practice in enumerate(best_practices, 1):
        print(f"      {i:2d}. {practice}")
    
    # 6. Integration patterns
    print("  6. Integration with development workflow:")
    
    integration_examples = {
        "CI/CD Testing": "pytest -W error::DeprecationWarning  # Fail on deprecation warnings",
        "Code Quality": "flake8 --select=W503,W504  # Catch deprecated patterns",
        "Documentation": "sphinx-build with deprecation directives",
        "Version Management": "semantic-release with deprecation tracking",
        "Monitoring": "log aggregation for deprecation warning analysis",
        "Migration Tools": "automated codemod scripts for bulk migration"
    }
    
    for tool, example in integration_examples.items():
        print(f"    {tool}: {example}")
    
    # 7. Final compliance verification
    print("  7. PEP 387 compliance verification:")
    
    compliance_checks = {
        "Warning format standardized": bool(pep387_warnings),
        "Version information provided": any("deprecated as of" in str(w.message).lower() for w in pep387_warnings),
        "Removal timeline specified": any("removed in" in str(w.message).lower() for w in pep387_warnings),
        "Migration guidance included": any("migration" in str(w.message).lower() for w in pep387_warnings),
        "Function signature preserved": inspect.signature(calculate_mean) is not None,
        "Docstring maintained": bool(calculate_mean.__doc__),
        "Backward compatibility": deprecated_result == statistics.mean(test_data)
    }
    
    print(f"    PEP 387 compliance results:")
    for check, passed in compliance_checks.items():
        status = "✓ PASS" if passed else "✗ FAIL"
        print(f"      {status}: {check}")
    
    all_passed = all(compliance_checks.values())
    print(f"    Overall compliance: {'✓ COMPLIANT' if all_passed else '✗ NON-COMPLIANT'}")
    assert all_passed, f"PEP 387 checks not met: {[check for check, passed in compliance_checks.items() if not passed]}"
    
    print("✓ deprecation package capabilities and PEP 387 compliance completed successfully")
    print("  → Ready for production use following Python deprecation standards")
    
except ImportError as e:
    print(f"❌ Import error: {e}")
    print("  Note: Install deprecation package: pip install deprecation")
except Exception as e:
    print(f"❌ Runtime error: {e}")
    raise AssertionError(
        f"deprecation example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
