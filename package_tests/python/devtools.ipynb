```python
print("Example 3: Debug Configuration and Environment Control")

try:
    from devtools import debug
    print("✓ Using devtools debug function")
    
    print("✓ Debug configuration and environment control:")
    
    # 1. Environment-based debug control
    print("  1. Environment-based debug control:")
    
    import os
    
    # Store original environment
    original_debug_env = os.environ.get('DEVTOOLS_DEBUG', '')
    
    def test_debug_with_env(env_value, test_name):
        """Test debug behavior with different environment values."""
        print(f"    Testing {test_name} (DEVTOOLS_DEBUG='{env_value}'):")
        
        # Set environment
        os.environ['DEVTOOLS_DEBUG'] = env_value
        
        # Test debug output
        test_data = {"environment": env_value, "enabled": env_value.lower() in ('1', 'true', 'yes')}
        debug("Environment test", test_data)
        
        return test_data
    
    # Test different environment settings
    env_tests = [
        ('1', 'enabled'),
        ('0', 'disabled'),
        ('true', 'enabled_true'),
        ('false', 'disabled_false')
    ]
    
    for env_val, test_name in env_tests[:2]:  # Test only first 2 to avoid spam
        test_debug_with_env(env_val, test_name)
    
    # Restore original environment
    if original_debug_env:
        os.environ['DEVTOOLS_DEBUG'] = original_debug_env
    else:
        os.environ.pop('DEVTOOLS_DEBUG', None)
    
    # 2. Conditional debugging patterns
    print("  2. Conditional debugging patterns:")
    
    class DebugConfig:
        """Debug configuration management."""
        
        def __init__(self, enabled=True, level='info', max_items=5):
            self.enabled = enabled
            self.level = level
            self.max_items = max_items
            self.debug_count = 0
        
        def debug_if(self, condition, *args):
            """Debug only if condition is met."""
            if self.enabled and condition and self.debug_count < self.max_items:
                debug(f"[{self.level.upper()}]", *args)
                self.debug_count += 1
            elif self.debug_count == self.max_items:
                debug("Debug limit reached - suppressing further output")
                self.debug_count += 1
        
        def reset_count(self):
            """Reset debug counter."""
            self.debug_count = 0
    
    # Test conditional debugging
    config = DebugConfig(enabled=True, level='debug', max_items=3)
    
    test_conditions = [
        (True, "condition_true", "This should be debugged"),
        (False, "condition_false", "This should NOT be debugged"),
        (True, "condition_true_2", "This should be debugged"),
        (True, "condition_true_3", "This should be debugged"),
        (True, "condition_true_4", "This should be limited"),
        (True, "condition_true_5", "This should be suppressed")
    ]
    
    for condition, name, message in test_conditions:
        config.debug_if(condition, name, message)
    
    # 3. Debug context management
    print("  3. Debug context management:")
    
    class DebugContext:
        """Context manager for debug sessions."""
        
        def __init__(self, session_name, enabled=True):
            self.session_name = session_name
            self.enabled = enabled
            self.start_time = None
        
        def __enter__(self):
            if self.enabled:
                import time
                self.start_time = time.time()
                debug("Session started", self.session_name)
            return self
        
        def __exit__(self, exc_type, exc_val, exc_tb):
            if self.enabled and self.start_time:
                import time
                duration = time.time() - self.start_time
                if exc_type:
                    debug("Session failed", self.session_name, f"duration: {duration:.4f}s", exc_type.__name__)
                else:
                    debug("Session completed", self.session_name, f"duration: {duration:.4f}s")
        
        def debug_in_context(self, *args):
            """Debug within this context."""
            if self.enabled:
                debug(f"[{self.session_name}]", *args)
    
    # Test context management
    with DebugContext("data_processing", enabled=True) as ctx:
        ctx.debug_in_context("Processing started")
        
        # Simulate some work
        data = [1, 2, 3, 4, 5]
        processed = [x * 2 for x in data]
        
        ctx.debug_in_context("Data processed", f"{len(data)} -> {len(processed)} items")
    
    # Test context with error
    try:
        with DebugContext("error_test", enabled=True) as ctx:
            ctx.debug_in_context("About to cause error")
            raise ValueError("Test error for context")
    except ValueError:
        pass  # Expected error
    
    # 4. Debug filtering and categorization
    print("  4. Debug filtering and categorization:")
    
    class CategorizedDebug:
        """Debug with category filtering."""
        
        def __init__(self, enabled_categories=None):
            self.enabled_categories = enabled_categories or ['info', 'warning', 'error']
        
        def debug_category(self, category, *args):
            """Debug with category filtering."""
            if category in self.enabled_categories:
                debug(f"[{category.upper()}]", *args)
        
        def info(self, *args):
            self.debug_category('info', *args)
        
        def warning(self, *args):
            self.debug_category('warning', *args)
        
        def error(self, *args):
            self.debug_category('error', *args)
        
        def trace(self, *args):
            self.debug_category('trace', *args)
    
    # Test categorized debugging
    cat_debug = CategorizedDebug(enabled_categories=['info', 'warning', 'error'])  # trace disabled
    
    cat_debug.info("This is an info message")
    cat_debug.warning("This is a warning message")
    cat_debug.error("This is an error message")
    cat_debug.trace("This trace message should be filtered out")
    
    # 5. Production-safe debugging
    print("  5. Production-safe debugging:")
    
    class ProductionDebug:
        """Production-safe debug wrapper."""
        
        def __init__(self, production_mode=False):
            self.production_mode = production_mode
            self.debug_buffer = []
            self.max_buffer_size = 10
        
        def safe_debug(self, *args, **kwargs):
            """Safe debug that buffers in production."""
            if not self.production_mode:
                debug(*args, **kwargs)
            else:
                # In production, buffer debug messages
                import time
                message = {
                    'timestamp': time.time(),
                    'args': args,
                    'kwargs': kwargs
                }
                
                self.debug_buffer.append(message)
                
                # Keep buffer size manageable
                if len(self.debug_buffer) > self.max_buffer_size:
                    self.debug_buffer.pop(0)
        
        def get_debug_buffer(self):
            """Get buffered debug messages."""
            return self.debug_buffer.copy()
        
        def clear_buffer(self):
            """Clear debug buffer."""
            self.debug_buffer.clear()
    
    # Test production debugging
    print("    Development mode:")
    dev_debug = ProductionDebug(production_mode=False)
    dev_debug.safe_debug("Development message 1")
    dev_debug.safe_debug("Development message 2")
    
    print("    Production mode:")
    prod_debug = ProductionDebug(production_mode=True)
    prod_debug.safe_debug("Production message 1")
    prod_debug.safe_debug("Production message 2")
    
    buffer = prod_debug.get_debug_buffer()
    print(f"    Buffered {len(buffer)} debug messages in production mode")
    
    # 6. Configuration summary
    print("  6. Debug configuration best practices:")
    
    best_practices = [
        "Use environment variables for debug control",
        "Implement conditional debugging based on context",
        "Use context managers for debug sessions",
        "Categorize debug messages by severity/type",
        "Buffer debug messages in production environments",
        "Limit debug output volume to prevent performance issues",
        "Use debug configuration classes for complex scenarios",
        "Reset debug state between test runs"
    ]
    
    for i, practice in enumerate(best_practices, 1):
        print(f"      {i}. {practice}")
    
    print("✓ Debug configuration and environment control completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    print("  Note: Error in debug configuration demonstration")
    raise AssertionError(
        f"devtools example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: Production Debugging and Best Practices

```python
print("Example 4: Production Debugging and Integration Patterns")

try:
    from devtools import debug
    print("✓ Using devtools for production debugging")
    
    print("✓ Production debugging and integration patterns:")
    
    # 1. Integration with logging systems
    print("  1. Integration with logging systems:")
    
    import logging
    import sys
    
    # Setup logging
    logging.basicConfig(
        level=logging.DEBUG,
        format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
        handlers=[logging.StreamHandler(sys.stdout)]
    )
    
    logger = logging.getLogger(__name__)
    
    class LoggingDebugIntegration:
        """Integrate devtools debug with standard logging."""
        
        def __init__(self, logger, debug_enabled=True):
            self.logger = logger
            self.debug_enabled = debug_enabled
        
        def debug_with_logging(self, level, *args, **kwargs):
            """Debug with both devtools and logging."""
            # Format message for logging
            message_parts = [str(arg) for arg in args]
            log_message = " | ".join(message_parts)
            
            # Log using standard logging
            if level == 'info':
                self.logger.info(log_message)
            elif level == 'warning':
                self.logger.warning(log_message)
            elif level == 'error':
                self.logger.error(log_message)
            else:
                self.logger.debug(log_message)
            
            # Also use devtools debug for detailed inspection
            if self.debug_enabled:
                debug(f"[{level.upper()}]", *args)
        
        def info(self, *args):
            self.debug_with_logging('info', *args)
        
        def warning(self, *args):
            self.debug_with_logging('warning', *args)
        
        def error(self, *args):
            self.debug_with_logging('error', *args)
    
    # Test logging integration
    log_debug = LoggingDebugIntegration(logger, debug_enabled=True)
    
    log_debug.info("Application startup", {"version": "1.0", "mode": "production"})
    log_debug.warning("Resource usage high", {"cpu": "85%", "memory": "78%"})
    log_debug.error("Database connection failed", {"host": "localhost", "port": 5432})
    
    # 2. Performance monitoring with debugging
    print("  2. Performance monitoring with debugging:")
    
    import time
    from functools import wraps
    
    class PerformanceDebugger:
        """Debug performance issues with detailed metrics."""
        
        def __init__(self, enabled=True, threshold=0.1):
            self.enabled = enabled
            self.threshold = threshold  # Log operations slower than this
            self.metrics = {}
        
        def monitor_performance(self, func_name=None):
            """Decorator to monitor function performance."""
            def decorator(func):
                name = func_name or func.__name__
                
                @wraps(func)
                def wrapper(*args, **kwargs):
                    if not self.enabled:
                        return func(*args, **kwargs)
                    
                    debug("Function entry", name, f"args_count: {len(args)}", f"kwargs_count: {len(kwargs)}")
                    
                    start_time = time.time()
                    start_memory = self._get_memory_usage()
                    
                    try:
                        result = func(*args, **kwargs)
                        
                        end_time = time.time()
                        end_memory = self._get_memory_usage()
                        duration = end_time - start_time
                        memory_delta = end_memory - start_memory
                        
                        # Store metrics
                        self.metrics[name] = {
                            'duration': duration,
                            'memory_delta': memory_delta,
                            'timestamp': end_time
                        }
                        
                        # Debug if performance threshold exceeded
                        if duration > self.threshold:
                            debug("Performance warning", name, f"duration: {duration:.4f}s", f"memory_delta: {memory_delta}KB")
                        else:
                            debug("Function exit", name, f"duration: {duration:.4f}s", "OK")
                        
                        return result
                        
                    except Exception as e:
                        end_time = time.time()
                        duration = end_time - start_time
                        debug("Function error", name, f"duration: {duration:.4f}s", type(e).__name__, str(e))
                        raise
                
                return wrapper
            return decorator
        
        def _get_memory_usage(self):
            """Get current memory usage (simplified)."""
            try:
                import psutil
                process = psutil.Process()
                return process.memory_info().rss / 1024  # KB
            except ImportError:
                return 0  # psutil not available
        
        def get_performance_summary(self):
            """Get performance metrics summary."""
            if not self.metrics:
                return "No performance data collected"
            
            summary = {}
            for func_name, metrics in self.metrics.items():
                summary[func_name] = {
                    'avg_duration': metrics['duration'],
                    'memory_impact': metrics['memory_delta']
                }
            
            debug("Performance summary", summary)
            return summary
    
    # Test performance monitoring
    perf_debugger = PerformanceDebugger(enabled=True, threshold=0.05)
    
    @perf_debugger.monitor_performance("fast_operation")
    def fast_operation(data):
        """Fast operation for testing."""
        return [x * 2 for x in data]
    
    @perf_debugger.monitor_performance("slow_operation")
    def slow_operation(data):
        """Slow operation for testing."""
        time.sleep(0.1)  # Simulate slow operation
        return sum(data)
    
    # Test operations
    test_data = list(range(100))
    fast_result = fast_operation(test_data)
    slow_result = slow_operation(test_data)
    
    perf_summary = perf_debugger.get_performance_summary()
    print(f"    Performance monitoring completed: {len(perf_summary)} functions monitored")
    
    # 3. Error tracking and debugging
    print("  3. Error tracking and debugging:")
    
    class ErrorTracker:
        """Track and debug errors with context."""
        
        def __init__(self):
            self.errors = []
            self.max_errors = 10
        
        def track_error(self, error, context=None):
            """Track error with debugging context."""
            import traceback
            import time
            
            error_info = {
                'timestamp': time.time(),
                'type': type(error).__name__,
                'message': str(error),
                'context': context or {},
                'traceback': traceback.format_exc()
            }
            
            self.errors.append(error_info)
            
            # Keep only recent errors
            if len(self.errors) > self.max_errors:
                self.errors.pop(0)
            
            # Debug the error
            debug("Error tracked", error_info['type'], error_info['message'], error_info['context'])
        
        def safe_execute(self, func, *args, **kwargs):
            """Execute function with error tracking."""
            try:
                debug("Safe execution", func.__name__, f"args: {len(args)}", f"kwargs: {len(kwargs)}")
                result = func(*args, **kwargs)
                debug("Safe execution success", func.__name__)
                return result
            except Exception as e:
                context = {
                    'function': func.__name__,
                    'args_count': len(args),
                    'kwargs_count': len(kwargs)
                }
                self.track_error(e, context)
                return None
        
        def get_error_summary(self):
            """Get error summary for debugging."""
            if not self.errors:
                return "No errors tracked"
            
            summary = {}
            for error in self.errors:
                error_type = error['type']
                if error_type not in summary:
                    summary[error_type] = 0
                summary[error_type] += 1
            
            debug("Error summary", summary)
            return summary
    
    # Test error tracking
    error_tracker = ErrorTracker()
    
    def error_prone_function(value):
        """Function that might raise errors."""
        if value == 0:
            raise ZeroDivisionError("Cannot divide by zero")
        elif value < 0:
            raise ValueError("Negative values not allowed")
        else:
            return 10 / value
    
    # Test with different inputs
    test_values = [5, 2, 0, -1, 3]
    
    for value in test_values:
        result = error_tracker.safe_execute(error_prone_function, value)
        print(f"    safe_execute({value}) = {result}")
    
    error_summary = error_tracker.get_error_summary()
    print(f"    Error tracking completed: {error_summary}")
    
    # 4. Development vs Production debugging
    print("  4. Development vs Production debugging modes:")
    
    class ModeAwareDebugger:
        """Debug differently based on environment mode."""
        
        def __init__(self, mode='development'):
            self.mode = mode
            self.debug_buffer = []
        
        def debug_mode_aware(self, *args, **kwargs):
            """Debug with mode awareness."""
            if self.mode == 'development':
                # Full debugging in development
                debug("[DEV]", *args)
            elif self.mode == 'staging':
                # Limited debugging in staging
                if len(args) > 0 and any(keyword in str(args[0]).lower() 
                                      for keyword in ['error', 'warning', 'critical']):
                    debug("[STAGING]", *args)
            elif self.mode == 'production':
                # Minimal debugging in production, buffer everything
                self.debug_buffer.append({
                    'timestamp': time.time(),
                    'args': args,
                    'kwargs': kwargs
                })
                
                # Only show critical issues
                if len(args) > 0 and 'error' in str(args[0]).lower():
                    debug("[PROD-ERROR]", *args)
        
        def set_mode(self, mode):
            """Change debugging mode."""
            debug("Debug mode changed", f"from {self.mode}", f"to {mode}")
            self.mode = mode
        
        def get_buffer(self):
            """Get production debug buffer."""
            return self.debug_buffer.copy()
    
    # Test mode-aware debugging
    mode_debugger = ModeAwareDebugger('development')
    
    test_messages = [
        ("info", "Application started"),
        ("warning", "High memory usage detected"),
        ("error", "Database connection failed"),
        ("trace", "Processing user request")
    ]
    
    for mode in ['development', 'staging', 'production']:
        print(f"    Testing {mode} mode:")
        mode_debugger.set_mode(mode)
        
        for level, message in test_messages[:2]:  # Test first 2 messages
            mode_debugger.debug_mode_aware(level, message)
    
    buffer_size = len(mode_debugger.get_buffer())
    print(f"    Production buffer contains {buffer_size} messages")
    
    # 5. Best practices summary
    print("  5. Production debugging best practices:")
    
    best_practices = [
        "Integrate devtools with standard logging frameworks",
        "Use performance monitoring to identify bottlenecks",
        "Track errors with contextual information",
        "Adapt debugging verbosity based on environment mode",
        "Buffer debug output in production for later analysis",
        "Set performance thresholds for automatic alerting",
        "Include function entry/exit debugging for flow analysis",
        "Combine multiple debugging approaches for comprehensive coverage",
        "Use decorators for transparent debugging integration",
        "Maintain debug configuration separate from application config"
    ]
    
    for i, practice in enumerate(best_practices, 1):
        print(f"      {i:2d}. {practice}")
    
    print("✓ Production debugging and integration patterns completed successfully")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    import traceback
    print(f"  Traceback: {traceback.format_exc()}")
    raise
    
print("Example 4 finished")
```

```python
print("Example 5: devtools Package Capabilities and Development Workflow")

try:
    import devtools
    from devtools import debug, pformat, pprint
    debug_function = debug
    pformat_function = pformat
    pprint_function = pprint
    print("✓ devtools package successfully imported")
    print(f"  Package version: {getattr(devtools, 'VERSION', 'unknown')}")
    
    print("✓ devtools package capabilities and development workflow:")
    
    # 1. Package overview and capabilities
    print("  1. Package capabilities overview:")
    print(f"    Package version: {getattr(devtools, 'VERSION', 'unknown')}")
    
    # List available attributes
    if hasattr(devtools, '__dict__'):
        public_attrs = [attr for attr in dir(devtools) if not attr.startswith('_')]
        print(f"    Public attributes: {public_attrs[:10]}...")  # Show first 10
    
    # Test basic functionality
    print("  2. Basic functionality verification:")
    
    # Test debug function
    debug_function("Package verification", "testing basic debug functionality")
    
    # Test with different data types
    test_data = {
        "string": "Hello devtools",
        "number": 42,
        "list": [1, 2, 3, 4, 5],
        "dict": {"nested": {"key": "value"}},
        "boolean": True
    }
    
    debug_function("Data type testing", test_data)
    
    # Test pformat
    # devtools' `pformat` is a `PrettyFormat` INSTANCE: its `__call__` takes no `width` (that is a constructor
    # argument), so a 60-column layout needs its own formatter.
    from devtools.prettier import PrettyFormat
    formatted = PrettyFormat(width=60)(test_data)
    print(f"    Formatted data length: {len(formatted)} characters")
    
    # 3. Complete development workflow integration
    print("  3. Complete development workflow:")
    
    class DevWorkflowManager:
        """Comprehensive development workflow with debugging."""
        
        def __init__(self, project_name, debug_enabled=True):
            self.project_name = project_name
            self.debug_enabled = debug_enabled
            self.workflow_state = {"phase": "initialization", "tasks": []}
            self.metrics = {"functions_called": 0, "errors_encountered": 0}
            
            debug_function("Workflow initialized", self.project_name, self.workflow_state)
        
        def start_development_phase(self, phase_name):
            """Start a new development phase."""
            debug_function("Phase transition", f"from {self.workflow_state['phase']}", f"to {phase_name}")
            
            self.workflow_state["phase"] = phase_name
            self.workflow_state["tasks"] = []
            
            debug_function("Development phase started", phase_name, self.workflow_state)
        
        def add_task(self, task_name, task_data=None):
            """Add a task to current phase."""
            task = {
                "name": task_name,
                "data": task_data or {},
                "status": "pending",
                "timestamp": time.time()
            }
            
            self.workflow_state["tasks"].append(task)
            debug_function("Task added", task_name, task)
        
        def execute_task(self, task_name):
            """Execute a specific task."""
            debug_function("Task execution started", task_name)
            self.metrics["functions_called"] += 1
            
            # Find and execute task
            for task in self.workflow_state["tasks"]:
                if task["name"] == task_name:
                    try:
                        task["status"] = "executing"
                        debug_function("Executing task", task)
                        
                        # Simulate task execution
                        import time
                        time.sleep(0.01)  # Small delay for realism
                        
                        task["status"] = "completed"
                        task["completion_time"] = time.time()
                        
                        debug_function("Task completed", task_name, "success")
                        return True
                        
                    except Exception as e:
                        task["status"] = "failed"
                        task["error"] = str(e)
                        self.metrics["errors_encountered"] += 1
                        
                        debug_function("Task failed", task_name, type(e).__name__, str(e))
                        return False
            
            debug_function("Task not found", task_name)
            return False
        
        def get_workflow_summary(self):
            """Get comprehensive workflow summary."""
            summary = {
                "project": self.project_name,
                "current_phase": self.workflow_state["phase"],
                "tasks_total": len(self.workflow_state["tasks"]),
                "tasks_completed": len([t for t in self.workflow_state["tasks"] if t["status"] == "completed"]),
                "tasks_failed": len([t for t in self.workflow_state["tasks"] if t["status"] == "failed"]),
                "metrics": self.metrics
            }
            
            debug_function("Workflow summary", summary)
            return summary
    
    # Test complete workflow
    import time
    
    workflow = DevWorkflowManager("TestProject", debug_enabled=True)
    
    # Planning phase
    workflow.start_development_phase("planning")
    workflow.add_task("requirements_analysis", {"priority": "high", "estimated_hours": 4})
    workflow.add_task("architecture_design", {"priority": "high", "estimated_hours": 6})
    workflow.execute_task("requirements_analysis")
    workflow.execute_task("architecture_design")
    
    # Development phase
    workflow.start_development_phase("development")
    workflow.add_task("core_implementation", {"priority": "critical", "estimated_hours": 20})
    workflow.add_task("unit_tests", {"priority": "high", "estimated_hours": 8})
    workflow.execute_task("core_implementation")
    workflow.execute_task("unit_tests")
    
    # Get final summary
    final_summary = workflow.get_workflow_summary()
    print(f"    Workflow completed: {final_summary['tasks_completed']}/{final_summary['tasks_total']} tasks")
    
    # 4. Advanced debugging patterns
    print("  4. Advanced debugging patterns:")
    
    class AdvancedDebugPatterns:
        """Demonstrate advanced debugging patterns."""
        
        def __init__(self):
            self.debug_contexts = []
            self.performance_data = {}
        
        def debug_with_context(self, context_name, data):
            """Debug with contextual information."""
            import inspect
            
            # Get caller information
            frame = inspect.currentframe().f_back
            caller_info = {
                "function": frame.f_code.co_name,
                "line": frame.f_lineno,
                "locals_count": len(frame.f_locals)
            }
            
            debug_context = {
                "context": context_name,
                "data": data,
                "caller": caller_info,
                "timestamp": time.time()
            }
            
            self.debug_contexts.append(debug_context)
            debug_function("Contextual debug", context_name, debug_context)
        
        def debug_performance_comparison(self, operation_name, *functions):
            """Debug performance comparison of multiple functions."""
            debug_function("Performance comparison started", operation_name, f"{len(functions)} functions")
            
            results = {}
            test_data = list(range(1000))
            
            for func in functions:
                start_time = time.time()
                
                try:
                    result = func(test_data)
                    end_time = time.time()
                    duration = end_time - start_time
                    
                    results[func.__name__] = {
                        "duration": duration,
                        "result_type": type(result).__name__,
                        "result_size": len(result) if hasattr(result, '__len__') else 1,
                        "success": True
                    }
                    
                    debug_function("Function performance", func.__name__, f"{duration:.6f}s", "success")
                    
                except Exception as e:
                    end_time = time.time()
                    duration = end_time - start_time
                    
                    results[func.__name__] = {
                        "duration": duration,
                        "error": str(e),
                        "success": False
                    }
                    
                    debug_function("Function performance", func.__name__, f"{duration:.6f}s", "failed", type(e).__name__)
            
            self.performance_data[operation_name] = results
            debug_function("Performance comparison completed", operation_name, results)
            
            return results
        
        def debug_data_flow(self, data_pipeline):
            """Debug data flow through a pipeline."""
            debug_function("Data flow debugging started", f"{len(data_pipeline)} steps")
            
            current_data = {"initial": "data"}
            flow_log = []
            
            for step_name, transform_func in data_pipeline:
                debug_function("Pipeline step", step_name, f"input: {type(current_data)}")
                
                try:
                    step_start = time.time()
                    transformed_data = transform_func(current_data)
                    step_duration = time.time() - step_start
                    
                    flow_entry = {
                        "step": step_name,
                        "input_type": type(current_data).__name__,
                        "output_type": type(transformed_data).__name__,
                        "duration": step_duration,
                        "success": True
                    }
                    
                    flow_log.append(flow_entry)
                    current_data = transformed_data
                    
                    debug_function("Pipeline step completed", step_name, flow_entry)
                    
                except Exception as e:
                    step_duration = time.time() - step_start
                    
                    flow_entry = {
                        "step": step_name,
                        "input_type": type(current_data).__name__,
                        "error": str(e),
                        "duration": step_duration,
                        "success": False
                    }
                    
                    flow_log.append(flow_entry)
                    debug_function("Pipeline step failed", step_name, flow_entry)
                    break
            
            debug_function("Data flow completed", f"{len(flow_log)} steps", flow_log)
            return current_data, flow_log
    
    # Test advanced patterns
    advanced = AdvancedDebugPatterns()
    
    # Test contextual debugging
    advanced.debug_with_context("user_authentication", {"user_id": 123, "method": "oauth"})
    advanced.debug_with_context("data_processing", {"records": 1500, "filters": ["active", "verified"]})
    
    # Test performance comparison
    def list_comp_method(data):
        return [x * 2 for x in data[:100]]  # Process subset for speed
    
    def map_method(data):
        return list(map(lambda x: x * 2, data[:100]))
    
    def loop_method(data):
        result = []
        for x in data[:100]:
            result.append(x * 2)
        return result
    
    perf_results = advanced.debug_performance_comparison("multiplication_methods", 
                                                        list_comp_method, map_method, loop_method)
    fastest_method = min(perf_results.keys(), key=lambda k: perf_results[k].get("duration", float('inf')))
    print(f"    Performance winner: {fastest_method}")
    
    # Test data flow debugging
    def add_metadata(data):
        return {**data, "metadata": {"processed": True, "version": "1.0"}}
    
    def validate_data(data):
        if not isinstance(data, dict):
            raise ValueError("Data must be a dictionary")
        return {**data, "validated": True}
    
    def format_output(data):
        return {"formatted": True, "data": data, "timestamp": time.time()}
    
    pipeline = [
        ("add_metadata", add_metadata),
        ("validate_data", validate_data),
        ("format_output", format_output)
    ]
    
    final_data, flow_log = advanced.debug_data_flow(pipeline)
    print(f"    Data pipeline: {len(flow_log)} steps, final type: {type(final_data).__name__}")
    
    # 5. Final verification and summary
    print("  5. Package verification and summary:")
    
    # debug() and pprint() write to stdout; capture what each printed for the checks below
    import contextlib
    import io
    debug_text = str(debug_function.format(test_data))
    pprint_capture = io.StringIO()
    with contextlib.redirect_stdout(pprint_capture):
        pprint_function(test_data)

    # Each check reads what the steps above produced: both development tasks complete, all three
    # multiplication methods and all three pipeline steps succeed
    verification_results = {
        "debug_function_works": "test_data: {" in debug_text and "'number': 42" in debug_text and "(dict) len=5" in debug_text,
        "pformat_function_works": "'string': 'Hello devtools'" in formatted,
        "pprint_function_works": "'string': 'Hello devtools'" in pprint_capture.getvalue(),
        "workflow_integration": final_summary["tasks_completed"] == final_summary["tasks_total"] == 2 and final_summary["tasks_failed"] == 0,
        "advanced_patterns": all(r["success"] for r in perf_results.values()) and [e["step"] for e in flow_log if e["success"]] == [step for step, _ in pipeline],
    }
    
    debug_function("Final verification", verification_results)
    
    # Summary statistics
    summary_stats = {
        "examples_completed": 5,
        "debug_calls_made": "50+",
        "patterns_demonstrated": ["basic", "advanced", "workflow", "performance", "error_handling"],
        "all_tests_passed": all(verification_results.values())
    }
    
    assert summary_stats["all_tests_passed"], verification_results
    print("    Verification status: ✓ ALL PASSED")
    print(f"    Examples completed: {summary_stats['examples_completed']}")
    
    debug_function("Package capabilities summary", summary_stats)
    
    print("✓ devtools package capabilities and development workflow completed successfully")
    print("  → Comprehensive debugging system ready for development use")
    
except Exception as e:
    print(f"❌ Runtime error: {e}")
    import traceback
    print(f"  Full traceback:")
    traceback.print_exc()
    raise
    
print("Example 5 finished")
```
