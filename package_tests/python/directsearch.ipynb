## Example 1: Basic Direct Search Optimization

Minimize a two-variable quadratic with `solve_directsearch` and check the solution.

```python
# Example 1: Basic Direct Search Optimization

import numpy as np
from directsearch import solve_directsearch

# Define a simple quadratic objective function
def quadratic_function(x):
    """Simple quadratic function: f(x) = (x[0]-2)^2 + (x[1]-3)^2"""
    return (x[0] - 2.0)**2 + (x[1] - 3.0)**2

print("=== Basic Direct Search Optimization ===")
print(f"Objective function: f(x) = (x[0]-2)^2 + (x[1]-3)^2")
print(f"Expected minimum: (2, 3) with value 0")

# Starting point
x0 = np.array([0.0, 0.0])
print(f"Starting point: {x0}")
print(f"Initial value: f({x0}) = {quadratic_function(x0):.6f}")

# Run optimization using direct search
result = solve_directsearch(
    f=quadratic_function,
    x0=x0,
    alpha0=1.0,          # Initial step size
    alpha_min=1e-6,      # Minimum step size
    maxevals=1000,       # Maximum function evaluations
    verbose=True,        # Show progress
    print_freq=50        # Print every 50 evaluations
)

print(f"\n=== Optimization Results ===")
print(f"Final point: {result.x}")
print(f"Final value: {result.f:.8f}")
print(f"Function evaluations: {result.nf}")
print(f"Exit flag: {result.flag}")

# Verify solution quality
error = np.linalg.norm(result.x - np.array([2.0, 3.0]))
print(f"Distance to true optimum: {error:.6f}")

print("\n✓ Basic direct search optimization complete!")
```

## Example 2: Polling strategies and step-size adaptation

Compare polling strategies (2n, n+1 and two random directions) and step-size expansion and contraction settings of `solve_directsearch` on the Rosenbrock function.

```python
# Example 2: Different polling strategies and step size adaptation

import numpy as np
from directsearch import solve_directsearch

# Define Rosenbrock function - challenging for direct search
def rosenbrock(x):
    """Rosenbrock function: f(x) = 100*(x[1]-x[0]^2)^2 + (1-x[0])^2"""
    return 100.0 * (x[1] - x[0]**2)**2 + (1.0 - x[0])**2

print("=== Comparing Different Polling Strategies ===")
print("Test function: Rosenbrock function")
print("Expected minimum: (1, 1) with value 0")

# Test with standard 2n polling (coordinate directions)
x0 = np.array([-1.2, 1.0])  # Standard Rosenbrock starting point
print(f"\nStarting point: {x0}")
print(f"Initial value: {rosenbrock(x0):.6f}")

# Test 1: Standard 2n polling (coordinate directions)
print("\n--- Test 1: Standard 2n Polling ---")
result_2n = solve_directsearch(
    f=rosenbrock,
    x0=x0,
    poll_type='2n',     # Coordinate directions: ±e_1, ±e_2, ...
    alpha0=1.0,
    alpha_min=1e-6,
    gamma_inc=2.0,      # Step size expansion factor
    gamma_dec=0.5,      # Step size contraction factor
    maxevals=2000,
    verbose=False
)

print(f"Final point: {result_2n.x}")
print(f"Final value: {result_2n.f:.8f}")
print(f"Function evaluations: {result_2n.nf}")
print(f"Distance to optimum: {np.linalg.norm(result_2n.x - np.array([1.0, 1.0])):.6f}")

# Test 2: n+1 polling (n+1 random directions)
print("\n--- Test 2: n+1 Polling (Random Simplex Directions) ---")
result_nplus1 = solve_directsearch(
    f=rosenbrock,
    x0=x0,
    poll_type='np1',    # n+1 random directions forming simplex
    alpha0=1.0,
    alpha_min=1e-6,
    gamma_inc=2.0,
    gamma_dec=0.5,
    maxevals=2000,
    verbose=False
)

print(f"Final point: {result_nplus1.x}")
print(f"Final value: {result_nplus1.f:.8f}")
print(f"Function evaluations: {result_nplus1.nf}")
print(f"Distance to optimum: {np.linalg.norm(result_nplus1.x - np.array([1.0, 1.0])):.6f}")

# Test 3: Random 2 directions
print("\n--- Test 3: Random 2 Directions ---")
result_random2 = solve_directsearch(
    f=rosenbrock,
    x0=x0,
    poll_type='random2',  # 2 random directions per iteration
    alpha0=1.0,
    alpha_min=1e-6,
    gamma_inc=2.0,
    gamma_dec=0.5,
    maxevals=2000,
    verbose=False
)

print(f"Final point: {result_random2.x}")
print(f"Final value: {result_random2.f:.8f}")
print(f"Function evaluations: {result_random2.nf}")
print(f"Distance to optimum: {np.linalg.norm(result_random2.x - np.array([1.0, 1.0])):.6f}")

# Demonstrate step size adaptation with different parameters
print("\n=== Step Size Adaptation Strategies ===")

# Aggressive expansion/contraction
print("\n--- Aggressive Step Size Adaptation ---")
result_aggressive = solve_directsearch(
    f=rosenbrock,
    x0=x0,
    alpha0=2.0,         # Larger initial step
    gamma_inc=3.0,      # Aggressive expansion
    gamma_dec=0.25,     # Aggressive contraction
    alpha_max=1000.0,
    alpha_min=1e-6,
    maxevals=2000,
    verbose=False
)

print(f"Final value: {result_aggressive.f:.8f}")
print(f"Function evaluations: {result_aggressive.nf}")

# Conservative expansion/contraction
print("\n--- Conservative Step Size Adaptation ---")
result_conservative = solve_directsearch(
    f=rosenbrock,
    x0=x0,
    alpha0=0.5,         # Smaller initial step
    gamma_inc=1.5,      # Conservative expansion
    gamma_dec=0.75,     # Conservative contraction
    alpha_max=1000.0,
    alpha_min=1e-6,
    maxevals=2000,
    verbose=False
)

print(f"Final value: {result_conservative.f:.8f}")
print(f"Function evaluations: {result_conservative.nf}")

# Compare results
print("\n=== Comparison Summary ===")
results_table = [
    ("2n Polling", result_2n.f, result_2n.nf),
    ("n+1 Polling", result_nplus1.f, result_nplus1.nf),
    ("Random 2 Directions", result_random2.f, result_random2.nf),
    ("Aggressive Adaptation", result_aggressive.f, result_aggressive.nf),
    ("Conservative Adaptation", result_conservative.f, result_conservative.nf)
]

for name, f_val, n_evals in results_table:
    print(f"{name:30s}: f = {f_val:12.8f}, evals = {n_evals:5d}")

print("\n✓ Polling strategies and step size adaptation demonstrated!")
```

## Example 3: Subspace direct search for high-dimensional problems

Use subspace direct search (`solve` with a sketch dimension) on 20- and 50-dimensional problems and compare sketching choices with standard direct search.

```python
# Example 3: Subspace Direct Search for High-Dimensional Problems

import numpy as np
from directsearch import solve, solve_directsearch

# Define high-dimensional test function
def sphere_function(x):
    """High-dimensional sphere function: f(x) = sum(x_i^2)"""
    return np.sum(x**2)

def shifted_ellipsoid(x):
    """Shifted ellipsoid function with varying condition number"""
    n = len(x)
    weights = np.arange(1, n+1)
    shifted = x - 1.0  # Optimal point at (1, 1, ..., 1)
    return np.sum(weights * shifted**2)

print("=== Subspace Direct Search for High-Dimensional Problems ===")

# Test 1: High-dimensional sphere function
n_dim = 20
np.random.seed(42)  # For reproducibility
x0_sphere = np.random.randn(n_dim)

print(f"Problem dimension: {n_dim}")
print(f"Test function: Sphere function (sum of squares)")
print(f"Expected minimum: origin with value 0")
print(f"Starting point: random")
print(f"Initial value: {sphere_function(x0_sphere):.6f}")

# Standard direct search (for comparison)
print("\n--- Standard Direct Search ---")
result_standard = solve_directsearch(
    f=sphere_function,
    x0=x0_sphere.copy(),
    alpha0=1.0,
    alpha_min=1e-4,
    maxevals=500,  # Limited evaluations
    verbose=False
)

print(f"Final value: {result_standard.f:.8f}")
print(f"Function evaluations: {result_standard.nf}")
print(f"Distance to optimum: {np.linalg.norm(result_standard.x):.6f}")

# Subspace direct search with sketching
print("\n--- Subspace Direct Search (Gaussian Sketching) ---")
result_subspace = solve(
    f=sphere_function,
    x0=x0_sphere.copy(),
    sketch_dim=5,              # Reduced dimension for polling
    sketch_type='gaussian',    # Random Gaussian projection
    alpha0=1.0,
    alpha_min=1e-4,
    maxevals=500,
    verbose=False
)

print(f"Final value: {result_subspace.f:.8f}")
print(f"Function evaluations: {result_subspace.nf}")
print(f"Distance to optimum: {np.linalg.norm(result_subspace.x):.6f}")
print(f"Sketch dimension used: 5 (vs full dimension {n_dim})")

# Test 2: Different sketching techniques
print("\n=== Comparing Different Sketching Techniques ===")

np.random.seed(43)
x0_ellipsoid = np.random.randn(n_dim)
print(f"\nTest function: Shifted ellipsoid (ill-conditioned)")
print(f"Starting value: {shifted_ellipsoid(x0_ellipsoid):.6f}")

# Gaussian sketching
print("\n--- Gaussian Sketching ---")
result_gaussian = solve(
    f=shifted_ellipsoid,
    x0=x0_ellipsoid.copy(),
    sketch_dim=8,
    sketch_type='gaussian',
    alpha0=1.0,
    alpha_min=1e-4,
    maxevals=1000,
    verbose=False
)

print(f"Final value: {result_gaussian.f:.8f}")
print(f"Function evaluations: {result_gaussian.nf}")

# Orthogonal sketching (random orthogonal projection)
print("\n--- Orthogonal Sketching ---")
result_ortho = solve(
    f=shifted_ellipsoid,
    x0=x0_ellipsoid.copy(),
    sketch_dim=8,
    sketch_type='orthogonal',
    alpha0=1.0,
    alpha_min=1e-4,
    maxevals=1000,
    verbose=False
)

print(f"Final value: {result_ortho.f:.8f}")
print(f"Function evaluations: {result_ortho.nf}")

print(f"\nComparison:")
print(f"  Gaussian sketching:    f = {result_gaussian.f:.8f}, evals = {result_gaussian.nf}")
print(f"  Orthogonal sketching:  f = {result_ortho.f:.8f}, evals = {result_ortho.nf}")

# Test 3: Effect of sketch dimension
print("\n=== Effect of Sketch Dimension ===")

sketch_dims = [3, 5, 8, 12]
np.random.seed(44)
x0_test = np.random.randn(n_dim)

print(f"Full dimension: {n_dim}")
print(f"Testing sketch dimensions: {sketch_dims}")

results_sketch = []
for sketch_dim in sketch_dims:
    result = solve(
        f=sphere_function,
        x0=x0_test.copy(),
        sketch_dim=sketch_dim,
        sketch_type='gaussian',
        alpha0=1.0,
        alpha_min=1e-4,
        maxevals=500,
        verbose=False
    )
    results_sketch.append((sketch_dim, result.f, result.nf))
    print(f"Sketch dim {sketch_dim:2d}: f = {result.f:12.8f}, evals = {result.nf:4d}")

# Analyze efficiency vs sketch dimension
print("\nEfficiency Analysis:")
best_result = min(results_sketch, key=lambda x: x[1])
print(f"Best sketch dimension: {best_result[0]} with f = {best_result[1]:.8f}")

# Test 4: Very high-dimensional problem
print("\n=== Very High-Dimensional Problem ===")

n_very_high = 50
np.random.seed(45)
x0_very_high = np.ones(n_very_high) + 0.1 * np.random.randn(n_very_high)

def high_dim_quadratic(x):
    """Simple quadratic with optimum at origin"""
    return 0.5 * np.sum(x**2) + np.sum(x)

print(f"Problem dimension: {n_very_high}")
print(f"Sketch dimension: 10")
print(f"Initial value: {high_dim_quadratic(x0_very_high):.6f}")

result_very_high = solve(
    f=high_dim_quadratic,
    x0=x0_very_high.copy(),
    sketch_dim=10,
    sketch_type='gaussian',
    alpha0=1.0,
    alpha_min=1e-3,
    maxevals=1000,
    verbose=False
)

print(f"Final value: {result_very_high.f:.8f}")
print(f"Function evaluations: {result_very_high.nf}")
print(f"Efficiency: Reduced {n_very_high} dimensions to {10} for polling")
print(f"Dimensionality reduction: {10/n_very_high*100:.1f}% of original dimension")

# Test 5: Comparison on different problem scales
print("\n=== Scaling Comparison ===")

dimensions = [10, 20, 30, 40]
print("Comparing standard vs subspace direct search across dimensions:")

for dim in dimensions:
    np.random.seed(46)
    x0_scale = np.random.randn(dim)
    sketch_d = max(3, dim // 4)  # Use 25% of dimension
    
    # Standard DS
    result_std = solve_directsearch(
        f=sphere_function,
        x0=x0_scale.copy(),
        alpha0=1.0,
        alpha_min=1e-4,
        maxevals=300,
        verbose=False
    )
    
    # Subspace DS
    result_sub = solve(
        f=sphere_function,
        x0=x0_scale.copy(),
        sketch_dim=sketch_d,
        sketch_type='gaussian',
        alpha0=1.0,
        alpha_min=1e-4,
        maxevals=300,
        verbose=False
    )
    
    print(f"Dim {dim:2d}: Standard f={result_std.f:8.4f} ({result_std.nf:3d} evals) | "
          f"Subspace f={result_sub.f:8.4f} ({result_sub.nf:3d} evals, d={sketch_d})")

print("\n✓ Subspace direct search for high-dimensional problems demonstrated!")
```

## Example 4: Probabilistic direct search and noisy optimization

Run `solve_probabilistic_directsearch` on noisy and stochastic objectives with default and custom forcing functions.

```python
# Example 4: Probabilistic Direct Search and Noisy Optimization

import numpy as np
from directsearch import solve_probabilistic_directsearch

# Define noisy objective functions
def noisy_quadratic(x):
    """Quadratic function with additive noise"""
    true_value = (x[0] - 2.0)**2 + (x[1] - 3.0)**2
    noise = np.random.normal(0, 0.1)  # 10% noise level
    return true_value + noise

def noisy_rosenbrock(x):
    """Rosenbrock function with multiplicative noise"""
    true_value = 100.0 * (x[1] - x[0]**2)**2 + (1.0 - x[0])**2
    noise = np.random.normal(1.0, 0.05)  # 5% multiplicative noise
    return true_value * noise

def stochastic_simulation(x):
    """Simulate stochastic system (e.g., queueing system, Monte Carlo simulation)"""
    # Simulate a simple stochastic process
    # True optimum around (1.5, 2.0)
    base_cost = (x[0] - 1.5)**2 + (x[1] - 2.0)**2
    
    # Add random fluctuations to simulate stochastic system
    n_samples = 10  # Small sample size creates high variance
    samples = [base_cost + np.random.exponential(0.3) for _ in range(n_samples)]
    
    return np.mean(samples)

print("=== Probabilistic Direct Search for Noisy Problems ===")

# Test 1: Noisy quadratic function with default rho
print("\n--- Test 1: Noisy Quadratic Function ---")

x0 = np.array([0.0, 0.0])
print(f"Starting point: {x0}")
print(f"Expected optimum: (2, 3)")

# Use default rho (None) which handles noise well
print("\nUsing default rho (automatic forcing function):")
result_default = solve_probabilistic_directsearch(
    f=noisy_quadratic,
    x0=x0,
    rho=None,  # Use default forcing function
    alpha0=1.0,
    alpha_min=1e-4,
    maxevals=1000,
    verbose=False
)

print(f"Final point: {result_default.x}")
print(f"Final value: {result_default.f:.6f}")
print(f"Function evaluations: {result_default.nf}")
print(f"Distance to true optimum: {np.linalg.norm(result_default.x - np.array([2.0, 3.0])):.6f}")

# Test 2: Custom forcing functions (rho_uses_normd=False for simple signature)
print("\n--- Test 2: Custom Forcing Functions ---")

# Aggressive forcing function (rho depends only on alpha)
print("\nAggressive forcing function: rho(alpha) = 0.8 * alpha^2")
result_aggressive = solve_probabilistic_directsearch(
    f=noisy_quadratic,
    x0=x0,
    rho=lambda alpha: 0.8 * alpha**2,
    rho_uses_normd=False,  # Simple signature: rho(alpha)
    alpha0=1.0,
    alpha_min=1e-4,
    maxevals=1000,
    verbose=False
)

print(f"Final value: {result_aggressive.f:.6f}")
print(f"Function evaluations: {result_aggressive.nf}")
print(f"Distance to optimum: {np.linalg.norm(result_aggressive.x - np.array([2.0, 3.0])):.6f}")

# Conservative forcing function
print("\nConservative forcing function: rho(alpha) = 0.2 * alpha^2")
result_conservative = solve_probabilistic_directsearch(
    f=noisy_quadratic,
    x0=x0,
    rho=lambda alpha: 0.2 * alpha**2,
    rho_uses_normd=False,  # Simple signature: rho(alpha)
    alpha0=1.0,
    alpha_min=1e-4,
    maxevals=1000,
    verbose=False
)

print(f"Final value: {result_conservative.f:.6f}")
print(f"Function evaluations: {result_conservative.nf}")
print(f"Distance to optimum: {np.linalg.norm(result_conservative.x - np.array([2.0, 3.0])):.6f}")

# Test 3: Forcing function with direction norm (rho_uses_normd=True)
print("\n--- Test 3: Direction-Aware Forcing Function ---")

print("Using rho(alpha, norm_d) = 0.5 * alpha^2 * norm_d")
result_norm_aware = solve_probabilistic_directsearch(
    f=noisy_quadratic,
    x0=x0,
    rho=lambda alpha, norm_d: 0.5 * alpha**2 * norm_d,
    rho_uses_normd=True,  # Full signature: rho(alpha, norm_d)
    alpha0=1.0,
    alpha_min=1e-4,
    maxevals=1000,
    verbose=False
)

print(f"Final value: {result_norm_aware.f:.6f}")
print(f"Function evaluations: {result_norm_aware.nf}")
print(f"Distance to optimum: {np.linalg.norm(result_norm_aware.x - np.array([2.0, 3.0])):.6f}")

# Test 4: Noisy Rosenbrock function
print("\n--- Test 4: Noisy Rosenbrock Function ---")

x0_rosen = np.array([-1.2, 1.0])
print(f"Starting point: {x0_rosen}")
print(f"Expected optimum: (1, 1)")

result_noisy_rosen = solve_probabilistic_directsearch(
    f=noisy_rosenbrock,
    x0=x0_rosen,
    rho=lambda alpha: 0.4 * alpha**2,
    rho_uses_normd=False,
    alpha0=1.0,
    alpha_min=1e-4,
    gamma_inc=1.5,
    gamma_dec=0.6,
    maxevals=2000,
    verbose=False
)

print(f"Final point: {result_noisy_rosen.x}")
print(f"Final value: {result_noisy_rosen.f:.6f}")
print(f"Function evaluations: {result_noisy_rosen.nf}")
print(f"Distance to true optimum: {np.linalg.norm(result_noisy_rosen.x - np.array([1.0, 1.0])):.6f}")

# Test 5: Stochastic simulation optimization
print("\n--- Test 5: Stochastic Simulation Optimization ---")

x0_sim = np.array([0.0, 0.0])
print(f"Simulating stochastic system (e.g., queueing, Monte Carlo)")
print(f"Starting point: {x0_sim}")
print(f"True optimum: around (1.5, 2.0)")

result_stochastic = solve_probabilistic_directsearch(
    f=stochastic_simulation,
    x0=x0_sim,
    rho=None,  # Use default, good for high variance
    alpha0=1.0,
    alpha_min=1e-3,
    gamma_inc=1.5,
    gamma_dec=0.7,
    maxevals=1500,
    verbose=False
)

print(f"Final point: {result_stochastic.x}")
print(f"Final value: {result_stochastic.f:.6f}")
print(f"Function evaluations: {result_stochastic.nf}")
print(f"Distance to true optimum: {np.linalg.norm(result_stochastic.x - np.array([1.5, 2.0])):.6f}")

# Test 6: Multiple runs to assess robustness
print("\n--- Test 6: Robustness Analysis (Multiple Runs) ---")

n_runs = 5
final_points = []
final_values = []

print(f"Running {n_runs} independent optimizations on noisy quadratic...")

np.random.seed(100)  # For reproducibility across runs
for run in range(n_runs):
    result = solve_probabilistic_directsearch(
        f=noisy_quadratic,
        x0=np.array([0.0, 0.0]),
        rho=None,
        alpha0=1.0,
        alpha_min=1e-4,
        maxevals=800,
        verbose=False
    )
    final_points.append(result.x)
    final_values.append(result.f)

final_points = np.array(final_points)
final_values = np.array(final_values)
true_optimum = np.array([2.0, 3.0])

print(f"\nResults from {n_runs} runs:")
for i, (point, value) in enumerate(zip(final_points, final_values)):
    distance = np.linalg.norm(point - true_optimum)
    print(f"  Run {i+1}: x = [{point[0]:.4f}, {point[1]:.4f}], f = {value:.6f}, dist = {distance:.6f}")

print(f"\nStatistics:")
mean_point = np.mean(final_points, axis=0)
std_point = np.std(final_points, axis=0)
mean_value = np.mean(final_values)
std_value = np.std(final_values)

print(f"  Mean final point: [{mean_point[0]:.4f}, {mean_point[1]:.4f}]")
print(f"  Std dev of points: [{std_point[0]:.4f}, {std_point[1]:.4f}]")
print(f"  Mean final value: {mean_value:.6f}")
print(f"  Std dev of values: {std_value:.6f}")
print(f"  Mean distance to optimum: {np.mean([np.linalg.norm(p - true_optimum) for p in final_points]):.6f}")

# Comparison summary
print("\n=== Forcing Function Comparison ===")
comparison_results = [
    ("Default rho", result_default.f, result_default.nf),
    ("Aggressive (0.8*α²)", result_aggressive.f, result_aggressive.nf),
    ("Conservative (0.2*α²)", result_conservative.f, result_conservative.nf),
    ("Direction-aware", result_norm_aware.f, result_norm_aware.nf)
]

for name, f_val, n_evals in comparison_results:
    print(f"{name:25s}: f = {f_val:10.6f}, evals = {n_evals:4d}")

print("\n✓ Probabilistic direct search for noisy optimization demonstrated!")
```

## Example 5: Stochastic Three Points (STP) and comparison of methods

Run the Stochastic Three Points method (`solve_stp`) and compare it with the other direct-search solvers on multimodal test functions.

```python
# Example 5: Advanced Features - STP and Comparison of Methods

import numpy as np
import time
from directsearch import solve_stp, solve_directsearch, solve, solve_probabilistic_directsearch

# Define test functions with different characteristics
def rastrigin(x):
    """Rastrigin function - highly multimodal"""
    n = len(x)
    A = 10
    return A * n + np.sum(x**2 - A * np.cos(2 * np.pi * x))

def ackley(x):
    """Ackley function - many local minima"""
    n = len(x)
    sum_sq = np.sum(x**2)
    sum_cos = np.sum(np.cos(2 * np.pi * x))
    return -20 * np.exp(-0.2 * np.sqrt(sum_sq / n)) - np.exp(sum_cos / n) + 20 + np.e

def levy(x):
    """Levy function - challenging optimization landscape"""
    w = 1 + (x - 1) / 4
    term1 = np.sin(np.pi * w[0])**2
    term3 = (w[-1] - 1)**2 * (1 + np.sin(2 * np.pi * w[-1])**2)
    wi = w[:-1]
    term2 = np.sum((wi - 1)**2 * (1 + 10 * np.sin(np.pi * wi + 1)**2))
    return term1 + term2 + term3

print("=== Advanced Direct Search Methods Comparison ===")

# Test 1: Stochastic Three Points (STP) Method
print("\n--- Test 1: Stochastic Three Points (STP) Method ---")
print("STP uses a predefined stepsize sequence: alpha0/(k+1)")
print("It polls Gaussian directions and uses simple decrease")

x0 = np.array([0.5, 0.5])
print(f"\nTest function: Rastrigin (highly multimodal)")
print(f"Starting point: {x0}")
print(f"Initial value: {rastrigin(x0):.6f}")

result_stp = solve_stp(
    f=rastrigin,
    x0=x0,
    alpha0=1.0,
    alpha_min=1e-4,
    maxevals=1000,
    verbose=False
)

print(f"\nSTP Results:")
print(f"Final point: {result_stp.x}")
print(f"Final value: {result_stp.f:.8f}")
print(f"Function evaluations: {result_stp.nf}")
print(f"Exit flag: {result_stp.flag}")

# Test 2: Comprehensive comparison of all methods
print("\n--- Test 2: Comprehensive Method Comparison ---")

test_functions = [
    ("Rastrigin", rastrigin, np.array([0.5, 0.5])),
    ("Ackley", ackley, np.array([2.0, 2.0])),
    ("Levy", levy, np.array([1.5, 1.5]))
]

methods = [
    ("Standard DS", lambda f, x0: solve_directsearch(f, x0, alpha0=1.0, alpha_min=1e-4, maxevals=1000, verbose=False)),
    ("Probabilistic DS", lambda f, x0: solve_probabilistic_directsearch(f, x0, rho=None, alpha0=1.0, alpha_min=1e-4, maxevals=1000, verbose=False)),
    ("STP", lambda f, x0: solve_stp(f, x0, alpha0=1.0, alpha_min=1e-4, maxevals=1000, verbose=False)),
    # sketch_dim must be <= the problem dimension (directsearch asserts it); these test problems are 2-D,
    # so a genuine subspace method polls a 1-D random subspace per iteration.
    ("Subspace DS (d=1)", lambda f, x0: solve(f, x0, sketch_dim=1, sketch_type='gaussian', alpha0=1.0, alpha_min=1e-4, maxevals=1000, verbose=False))
]

for func_name, func, x0_func in test_functions:
    print(f"\n{'='*60}")
    print(f"Test Function: {func_name}")
    print(f"Starting point: {x0_func}")
    print(f"Initial value: {func(x0_func):.6f}")
    print(f"{'='*60}")
    
    results = []
    for method_name, method in methods:
        start_time = time.time()
        result = method(func, x0_func.copy())
        elapsed = time.time() - start_time
        # Direct search only ever accepts a decrease, and each solver stops at its evaluation budget.
        assert result.f <= func(x0_func) + 1e-12, (method_name, func_name, result.f, func(x0_func))
        assert 0 < result.nf <= 1000, (method_name, func_name, result.nf)
        
        results.append({
            'method': method_name,
            'f': result.f,
            'nf': result.nf,
            'time': elapsed,
            'success': True
        })
        
        print(f"{method_name:20s}: f = {result.f:12.8f}, evals = {result.nf:4d}, time = {elapsed:.3f}s")
    
    # Find best result
    successful_results = [r for r in results if r['success']]
    if successful_results:
        best = min(successful_results, key=lambda x: x['f'])
        print(f"\nBest method for {func_name}: {best['method']} with f = {best['f']:.8f}")

# Test 3: Scaling with dimension
print("\n--- Test 3: Scaling with Problem Dimension ---")

def high_dim_sphere(x):
    """Simple sphere function for scaling tests"""
    return np.sum(x**2)

dimensions = [5, 10, 20, 30]
print(f"Testing how methods scale with dimension: {dimensions}")

scaling_results = {}
for dim in dimensions:
    np.random.seed(50)
    x0_dim = np.ones(dim)
    
    print(f"\nDimension: {dim}")
    
    # Standard direct search
    start = time.time()
    result_ds = solve_directsearch(high_dim_sphere, x0_dim, alpha0=1.0, alpha_min=1e-3, maxevals=500, verbose=False)
    time_ds = time.time() - start
    
    # Subspace direct search
    sketch_dim = max(3, dim // 4)  # Use 25% of dimension
    start = time.time()
    result_sub = solve(high_dim_sphere, x0_dim, sketch_dim=sketch_dim, alpha0=1.0, alpha_min=1e-3, maxevals=500, verbose=False)
    time_sub = time.time() - start
    
    print(f"  Standard DS:  f = {result_ds.f:10.6f}, time = {time_ds:.3f}s")
    print(f"  Subspace DS:  f = {result_sub.f:10.6f}, time = {time_sub:.3f}s (sketch_dim={sketch_dim})")
    
    scaling_results[dim] = {
        'standard': (result_ds.f, time_ds),
        'subspace': (result_sub.f, time_sub)
    }

# Test 4: Parameter sensitivity analysis
print("\n--- Test 4: Parameter Sensitivity Analysis ---")

def simple_quadratic(x):
    """Simple quadratic for parameter testing"""
    return (x[0] - 1)**2 + (x[1] - 2)**2

x0_param = np.array([0.0, 0.0])

print("Testing sensitivity to initial step size (alpha0):")
alpha0_values = [0.1, 0.5, 1.0, 2.0, 5.0]

for alpha0 in alpha0_values:
    result = solve_directsearch(
        simple_quadratic,
        x0_param,
        alpha0=alpha0,
        alpha_min=1e-6,
        maxevals=500,
        verbose=False
    )
    print(f"  alpha0 = {alpha0:4.1f}: f = {result.f:12.8f}, evals = {result.nf:4d}")

print("\nTesting sensitivity to expansion/contraction factors:")
factor_configs = [
    (1.5, 0.75, "Conservative"),
    (2.0, 0.5, "Standard"),
    (3.0, 0.25, "Aggressive")
]

for gamma_inc, gamma_dec, name in factor_configs:
    result = solve_directsearch(
        simple_quadratic,
        x0_param,
        alpha0=1.0,
        gamma_inc=gamma_inc,
        gamma_dec=gamma_dec,
        alpha_min=1e-6,
        maxevals=500,
        verbose=False
    )
    print(f"  {name:15s} (γ_inc={gamma_inc:.1f}, γ_dec={gamma_dec:.2f}): f = {result.f:12.8f}, evals = {result.nf:4d}")

# Test 5: Robustness to starting point
print("\n--- Test 5: Robustness to Starting Point ---")

def beale(x):
    """Beale function - multiple local minima"""
    term1 = (1.5 - x[0] + x[0]*x[1])**2
    term2 = (2.25 - x[0] + x[0]*x[1]**2)**2
    term3 = (2.625 - x[0] + x[0]*x[1]**3)**2
    return term1 + term2 + term3

starting_points = [
    np.array([0.0, 0.0]),
    np.array([1.0, 1.0]),
    np.array([2.0, 2.0]),
    np.array([-1.0, -1.0]),
    np.array([4.0, 4.0])
]

print("Testing Beale function from different starting points:")
print("True global minimum: (3, 0.5) with f = 0")

for i, x0_start in enumerate(starting_points):
    result = solve_directsearch(
        beale,
        x0_start,
        alpha0=1.0,
        alpha_min=1e-5,
        maxevals=1500,
        verbose=False
    )
    distance_to_optimum = np.linalg.norm(result.x - np.array([3.0, 0.5]))
    print(f"  Start {i+1} {x0_start}: f = {result.f:12.6f}, dist = {distance_to_optimum:.4f}")

# Test 6: STP vs Probabilistic DS on noisy problems
print("\n--- Test 6: STP vs Probabilistic DS on Noisy Problems ---")

def noisy_sphere(x):
    """Sphere function with noise"""
    true_val = np.sum(x**2)
    noise = np.random.normal(0, 0.1 * np.sqrt(true_val + 1e-8))
    return true_val + noise

np.random.seed(60)
x0_noisy = np.array([2.0, 2.0])

print(f"Noisy sphere function")
print(f"Starting point: {x0_noisy}")

# STP method
np.random.seed(61)
result_stp_noisy = solve_stp(
    noisy_sphere,
    x0_noisy.copy(),
    alpha0=1.0,
    alpha_min=1e-4,
    maxevals=1000,
    verbose=False
)

print(f"\nSTP method:")
print(f"  Final value: {result_stp_noisy.f:.8f}")
print(f"  Final point: {result_stp_noisy.x}")
print(f"  Evaluations: {result_stp_noisy.nf}")

# Probabilistic DS
np.random.seed(61)
result_prob_noisy = solve_probabilistic_directsearch(
    noisy_sphere,
    x0_noisy.copy(),
    rho=None,
    alpha0=1.0,
    alpha_min=1e-4,
    maxevals=1000,
    verbose=False
)

print(f"\nProbabilistic DS:")
print(f"  Final value: {result_prob_noisy.f:.8f}")
print(f"  Final point: {result_prob_noisy.x}")
print(f"  Evaluations: {result_prob_noisy.nf}")

print(f"\nComparison: STP is specialized for noisy functions with predetermined stepsize schedule")

print("\n✓ Advanced direct search methods and comprehensive analysis demonstrated!")
```
