## Example 1: Basic Union-Find Operations and Path Compression

Build a `DisjointSet`, merge elements with `union`, query roots with `find` and `connected`, and list the sets with `itersets`.

```python
# Example 1: Basic Union-Find Operations and Path Compression

from disjoint_set import DisjointSet

# Create disjoint set with basic operations
print("=== Basic Union-Find Operations ===")

# Initialize elements using from_iterable
elements = ['A', 'B', 'C', 'D', 'E', 'F']
ds = DisjointSet.from_iterable(elements)

print(f"Initial elements: {elements}")
print(f"Initial disjoint set: {ds}")
print(f"Initial number of sets: {len(ds)}")

# Demonstrate find operation
print(f"\nFind operations (each element is its own parent initially):")
for element in elements:
    root = ds.find(element)
    print(f"  find('{element}') = '{root}'")

# Union operations
print(f"\nUnion operations:")
unions = [('A', 'B'), ('C', 'D'), ('A', 'C'), ('E', 'F')]

for x, y in unions:
    print(f"  Union('{x}', '{y}')")
    ds.union(x, y)
    root_x = ds.find(x)
    root_y = ds.find(y)
    print(f"    After union: find('{x}') = '{root_x}', find('{y}') = '{root_y}'")
    print(f"    Sets count: {len(ds)}")

# Final connectivity check
print(f"\nFinal connectivity:")
for i, elem1 in enumerate(elements):
    for elem2 in elements[i+1:]:
        connected = ds.connected(elem1, elem2)
        print(f"  '{elem1}' connected to '{elem2}': {connected}")

print(f"\nFinal number of disjoint sets: {len(ds)}")
print(f"Final disjoint set representation: {ds}")

# Demonstrate itersets - iterate over all sets
print(f"\nAll disjoint sets:")
for i, s in enumerate(ds.itersets(), 1):
    print(f"  Set {i}: {sorted(s)}")

print("\n✓ Basic Union-Find operations demonstrated with path compression!")
```

## Example 2: Connected Components in Graphs

Demonstrate finding connected components in undirected graphs using Union-Find.

```python
# Example 2: Connected components in graphs

from disjoint_set import DisjointSet

# Graph representation and connected components
class GraphConnectedComponents:
    def __init__(self):
        self.disjoint_set = DisjointSet()
        self.edges = []
        self.vertices = set()
    
    def add_vertex(self, vertex):
        """Add vertex to graph"""
        self.vertices.add(vertex)
        # Don't need to call .add() - union will add automatically
    
    def add_edge(self, u, v):
        """Add edge between vertices u and v"""
        # Ensure vertices are tracked
        self.vertices.add(u)
        self.vertices.add(v)
        
        self.edges.append((u, v))
        self.disjoint_set.union(u, v)
    
    def get_connected_components(self):
        """Get all connected components"""
        components = {}
        for vertex in self.vertices:
            root = self.disjoint_set.find(vertex)
            if root not in components:
                components[root] = []
            components[root].append(vertex)
        
        return list(components.values())
    
    def are_connected(self, u, v):
        """Check if two vertices are in the same component"""
        return self.disjoint_set.connected(u, v)

print("=== Graph Connected Components Analysis ===")

# Create graph and analyze connected components
graph = GraphConnectedComponents()

# Add vertices and edges to create multiple connected components
edges = [
    (1, 2), (2, 3),           # Component 1: {1, 2, 3}
    (4, 5),                   # Component 2: {4, 5}
    (6, 7), (7, 8), (8, 9),   # Component 3: {6, 7, 8, 9}
    (10,)                     # Component 4: {10} (isolated)
]

print("Building graph with edges:")
for edge in edges:
    if len(edge) == 2:
        u, v = edge
        graph.add_edge(u, v)
        print(f"  Added edge: {u} -- {v}")
    else:
        # Isolated vertex
        graph.add_vertex(edge[0])
        print(f"  Added isolated vertex: {edge[0]}")

print(f"\nTotal vertices: {len(graph.vertices)}")
print(f"Total edges: {len(graph.edges)}")

# Find connected components
components = graph.get_connected_components()
print(f"\nConnected Components ({len(components)} total):")
for i, component in enumerate(components, 1):
    sorted_component = sorted(component)
    print(f"  Component {i}: {sorted_component} (size: {len(component)})")

# Test connectivity queries
print(f"\nConnectivity queries:")
test_pairs = [(1, 3), (1, 4), (6, 9), (4, 10), (2, 8)]
for u, v in test_pairs:
    if u in graph.vertices and v in graph.vertices:
        connected = graph.are_connected(u, v)
        print(f"  Vertices {u} and {v} connected: {connected}")

# Social network example
print(f"\n=== Social Network Friendship Groups ===")
social_network = GraphConnectedComponents()

# Add friendship connections
friendships = [
    ("Alice", "Bob"), ("Bob", "Charlie"), 
    ("David", "Eve"), ("David", "Frank"),
    ("Grace", "Henry"), ("Henry", "Isabel"), ("Isabel", "Grace"),
    ("Jack",)  # Jack has no friends (isolated)
]

print("Building social network:")
for friendship in friendships:
    if len(friendship) == 2:
        person1, person2 = friendship
        social_network.add_edge(person1, person2)
        print(f"  {person1} is friends with {person2}")
    else:
        social_network.add_vertex(friendship[0])
        print(f"  {friendship[0]} joined the network")

# Analyze friendship groups
friend_groups = social_network.get_connected_components()
print(f"\nFriendship Groups ({len(friend_groups)} total):")
for i, group in enumerate(friend_groups, 1):
    sorted_group = sorted(group)
    print(f"  Group {i}: {sorted_group}")

# Test mutual connections
print(f"\nMutual connection queries:")
queries = [("Alice", "Charlie"), ("David", "Frank"), ("Grace", "Isabel"), ("Bob", "Jack")]
for person1, person2 in queries:
    if person1 in social_network.vertices and person2 in social_network.vertices:
        connected = social_network.are_connected(person1, person2)
        connection_type = "mutually connected" if connected else "not connected"
        print(f"  {person1} and {person2}: {connection_type}")

# Demonstrate itersets to show all components
print(f"\nUsing itersets() to show components:")
for i, component_set in enumerate(social_network.disjoint_set.itersets(), 1):
    print(f"  Component {i}: {sorted(component_set)}")

print("\n✓ Connected components analysis demonstrated!")
```

## Example 3: Cycle Detection in Undirected Graphs

Demonstrate cycle detection using Union-Find for graph algorithms.

```python
# Example 3: Cycle detection in undirected graphs

from disjoint_set import DisjointSet

class CycleDetector:
    def __init__(self):
        self.disjoint_set = DisjointSet()
        self.vertices = set()
        self.edges = []
        self.cycle_edges = []
    
    def add_edge_with_cycle_check(self, u, v):
        """Add edge and check for cycle formation"""
        # Track vertices
        self.vertices.add(u)
        self.vertices.add(v)
        
        # Check if adding this edge creates a cycle
        if self.disjoint_set.connected(u, v):
            # Cycle detected! u and v are already connected
            self.cycle_edges.append((u, v))
            print(f"  🔄 CYCLE DETECTED: Adding edge ({u}, {v}) creates a cycle!")
            return True  # Cycle found
        else:
            # No cycle, safe to add edge
            self.disjoint_set.union(u, v)
            self.edges.append((u, v))
            print(f"  ✓ Added edge ({u}, {v}) - No cycle")
            return False  # No cycle
    
    def has_cycle(self):
        """Check if graph has any cycles"""
        return len(self.cycle_edges) > 0
    
    def get_forest_edges(self):
        """Get edges that form a forest (no cycles)"""
        return self.edges.copy()

print("=== Cycle Detection in Graphs ===")

# Example 1: Building a tree (no cycles)
print("Example 1: Building a tree (should have no cycles)")
detector1 = CycleDetector()

tree_edges = [(1, 2), (2, 3), (2, 4), (4, 5), (4, 6)]
print("Adding edges to form a tree:")

for u, v in tree_edges:
    detector1.add_edge_with_cycle_check(u, v)

print(f"Tree has cycles: {detector1.has_cycle()}")
print(f"Forest edges: {detector1.get_forest_edges()}")
print(f"Number of components: {len(detector1.disjoint_set)}")

# Example 2: Adding a cycle-creating edge
print(f"\nExample 2: Adding edge that creates a cycle")
print("Adding edge (3, 6) to the tree:")
detector1.add_edge_with_cycle_check(3, 6)  # This should create a cycle

print(f"Graph now has cycles: {detector1.has_cycle()}")
print(f"Cycle-creating edges: {detector1.cycle_edges}")

# Example 3: Kruskal's MST algorithm simulation
print(f"\n=== Kruskal's MST Algorithm Simulation ===")

class KruskalMST:
    def __init__(self):
        self.disjoint_set = DisjointSet()
        self.mst_edges = []
        self.rejected_edges = []
        self.total_weight = 0
    
    def find_mst(self, edges_with_weights):
        """Find Minimum Spanning Tree using Kruskal's algorithm"""
        # Sort edges by weight
        sorted_edges = sorted(edges_with_weights, key=lambda x: x[2])
        
        # Initialize vertices by doing a union of each with itself
        vertices = set()
        for u, v, _ in sorted_edges:
            vertices.update([u, v])
        
        # No need to explicitly add - union will handle it
        
        print("Processing edges in order of increasing weight:")
        for u, v, weight in sorted_edges:
            if not self.disjoint_set.connected(u, v):
                # No cycle, add to MST
                self.disjoint_set.union(u, v)
                self.mst_edges.append((u, v, weight))
                self.total_weight += weight
                print(f"  ✓ Added edge ({u}, {v}, {weight}) to MST")
            else:
                # Cycle would be formed, reject edge
                self.rejected_edges.append((u, v, weight))
                print(f"  ✗ Rejected edge ({u}, {v}, {weight}) - would create cycle")
        
        return self.mst_edges, self.total_weight

# Graph with weighted edges
weighted_edges = [
    ('A', 'B', 4),
    ('A', 'H', 8),
    ('B', 'C', 8),
    ('B', 'H', 11),
    ('C', 'D', 7),
    ('C', 'F', 4),
    ('C', 'I', 2),
    ('D', 'E', 9),
    ('D', 'F', 14),
    ('E', 'F', 10),
    ('F', 'G', 2),
    ('G', 'H', 1),
    ('G', 'I', 6),
    ('H', 'I', 7)
]

kruskal = KruskalMST()
mst_edges, total_weight = kruskal.find_mst(weighted_edges)

print(f"\nMinimum Spanning Tree:")
print(f"  MST edges: {mst_edges}")
print(f"  Total weight: {total_weight}")
print(f"  Rejected edges: {kruskal.rejected_edges}")

# Network connectivity example
print(f"\n=== Network Connectivity Monitor ===")

class NetworkMonitor:
    def __init__(self):
        self.disjoint_set = DisjointSet()
        self.connections = []
        self.servers = set()
    
    def add_server(self, server_id):
        """Add server to network"""
        self.servers.add(server_id)
        print(f"  Added server {server_id} to network")
    
    def establish_connection(self, server1, server2):
        """Establish connection between servers"""
        if server1 not in self.servers or server2 not in self.servers:
            print(f"  ⚠️  Cannot connect {server1} and {server2}: One or both servers not in network")
            return False
        
        if self.disjoint_set.connected(server1, server2):
            print(f"  ℹ️  {server1} and {server2} already connected (redundant connection)")
        else:
            self.disjoint_set.union(server1, server2)
            print(f"  ✓ Established connection between {server1} and {server2}")
        
        self.connections.append((server1, server2))
        return True
    
    def check_connectivity(self, server1, server2):
        """Check if two servers can communicate"""
        if server1 in self.disjoint_set and server2 in self.disjoint_set:
            connected = self.disjoint_set.connected(server1, server2)
            status = "can communicate" if connected else "cannot communicate"
            print(f"  {server1} and {server2}: {status}")
            return connected
        else:
            print(f"  {server1} and/or {server2}: not in network")
            return False
    
    def get_network_components(self):
        """Get isolated network components"""
        components = {}
        for server in self.servers:
            if server in self.disjoint_set:
                root = self.disjoint_set.find(server)
                if root not in components:
                    components[root] = []
                components[root].append(server)
            else:
                # Isolated server not yet in disjoint set
                components[server] = [server]
        
        return list(components.values())

# Network setup
network = NetworkMonitor()

print("Setting up network infrastructure:")
servers = ['web1', 'web2', 'db1', 'db2', 'cache1', 'cache2', 'backup1']
for server in servers:
    network.add_server(server)

print(f"\nEstablishing connections:")
connections = [
    ('web1', 'web2'),      # Web servers connected
    ('db1', 'db2'),        # Database replication
    ('web1', 'db1'),       # Web to database
    ('cache1', 'cache2'),  # Cache cluster
    ('web2', 'cache1'),    # Web to cache
    # backup1 remains isolated initially
]

for server1, server2 in connections:
    network.establish_connection(server1, server2)

print(f"\nConnectivity tests:")
tests = [('web1', 'db2'), ('web2', 'cache2'), ('db1', 'backup1'), ('cache1', 'db1')]
for server1, server2 in tests:
    network.check_connectivity(server1, server2)

print(f"\nNetwork components:")
components = network.get_network_components()
for i, component in enumerate(components, 1):
    print(f"  Network {i}: {sorted(component)}")

print("\n✓ Cycle detection and network connectivity demonstrated!")
```

## Example 4: Dynamic Connectivity and Union by Rank

Demonstrate advanced Union-Find optimizations and dynamic connectivity queries.

```python
# Example 4: Dynamic connectivity and Union by Rank

from disjoint_set import DisjointSet

# Advanced Union-Find with performance monitoring
class OptimizedUnionFind:
    def __init__(self):
        self.disjoint_set = DisjointSet()
        self.union_operations = 0
        self.find_operations = 0
        self.rank_info = {}
        
    def add_element(self, element):
        """Track element (no need to explicitly add to disjoint set)"""
        self.rank_info[element] = 0  # Initial rank is 0
        
    def union_with_monitoring(self, x, y):
        """Union operation with performance monitoring"""
        self.union_operations += 1
        
        root_x = self.find_with_monitoring(x)
        root_y = self.find_with_monitoring(y)
        
        if root_x == root_y:
            print(f"    Union({x}, {y}): Already in same set (root: {root_x})")
            return root_x
        
        # Union by rank optimization simulation
        rank_x = self.rank_info.get(root_x, 0)
        rank_y = self.rank_info.get(root_y, 0)
        
        if rank_x < rank_y:
            # Attach smaller rank tree under larger rank tree
            self.disjoint_set.union(x, y)
            new_root = self.disjoint_set.find(y)
            print(f"    Union({x}, {y}): Attached rank-{rank_x} tree to rank-{rank_y} tree (new root: {new_root})")
        elif rank_x > rank_y:
            self.disjoint_set.union(x, y)
            new_root = self.disjoint_set.find(x)
            print(f"    Union({x}, {y}): Attached rank-{rank_y} tree to rank-{rank_x} tree (new root: {new_root})")
        else:
            # Equal ranks, increment rank of result
            self.disjoint_set.union(x, y)
            new_root = self.disjoint_set.find(x)
            self.rank_info[new_root] = rank_x + 1
            print(f"    Union({x}, {y}): Equal ranks, new rank = {rank_x + 1} (new root: {new_root})")
        
        return new_root
    
    def find_with_monitoring(self, x):
        """Find operation with path compression monitoring"""
        self.find_operations += 1
        return self.disjoint_set.find(x)
    
    def get_performance_stats(self):
        """Get performance statistics"""
        return {
            'union_operations': self.union_operations,
            'find_operations': self.find_operations,
            'total_operations': self.union_operations + self.find_operations,
            'num_sets': len(self.disjoint_set)
        }

print("=== Dynamic Connectivity with Optimizations ===")

# Performance comparison demo
optimized_uf = OptimizedUnionFind()

print("Building dynamic connectivity structure:")
elements = list(range(1, 9))
for element in elements:
    optimized_uf.add_element(element)

print(f"Initial elements: {elements}")
print(f"Initial sets: {len(optimized_uf.disjoint_set) if len(optimized_uf.disjoint_set) > 0 else len(elements)}")

# Sequence of union operations showing rank optimization
union_sequence = [
    (1, 2), (3, 4), (5, 6), (7, 8),  # Create pairs
    (1, 3), (5, 7),                   # Union pairs
    (1, 5)                            # Final union
]

print(f"\nPerforming union operations with rank optimization:")
for x, y in union_sequence:
    print(f"  Union({x}, {y}):")
    optimized_uf.union_with_monitoring(x, y)

print(f"\nFinal connectivity tests:")
test_pairs = [(1, 8), (2, 7), (3, 6), (4, 5)]
for x, y in test_pairs:
    root_x = optimized_uf.find_with_monitoring(x)
    root_y = optimized_uf.find_with_monitoring(y)
    connected = (root_x == root_y)
    print(f"  {x} and {y}: {'Connected' if connected else 'Not connected'} (roots: {root_x}, {root_y})")

# Performance statistics
stats = optimized_uf.get_performance_stats()
print(f"\nPerformance Statistics:")
print(f"  Union operations: {stats['union_operations']}")
print(f"  Find operations: {stats['find_operations']}")
print(f"  Total operations: {stats['total_operations']}")
print(f"  Final number of sets: {stats['num_sets']}")

# Real-world application: Image segmentation simulation
print(f"\n=== Image Segmentation Simulation ===")

class ImageSegmentation:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.disjoint_set = DisjointSet()
        self.pixel_values = {}
        self.segments = {}
        
        # Initialize each pixel (will be added to disjoint set on first union)
        for row in range(height):
            for col in range(width):
                pixel = (row, col)
                self.pixel_values[pixel] = 0  # Default value
    
    def set_pixel_value(self, row, col, value):
        """Set pixel intensity value"""
        self.pixel_values[(row, col)] = value
    
    def merge_similar_pixels(self, threshold=10):
        """Merge adjacent pixels with similar values"""
        merged_count = 0
        
        for row in range(self.height):
            for col in range(self.width):
                current_pixel = (row, col)
                current_value = self.pixel_values[current_pixel]
                
                # Check adjacent pixels (4-connectivity)
                neighbors = [
                    (row-1, col), (row+1, col),  # Up, Down
                    (row, col-1), (row, col+1)   # Left, Right
                ]
                
                for nr, nc in neighbors:
                    if (0 <= nr < self.height and 0 <= nc < self.width):
                        neighbor_pixel = (nr, nc)
                        neighbor_value = self.pixel_values[neighbor_pixel]
                        
                        # Merge if similar and not already connected
                        if (abs(current_value - neighbor_value) <= threshold and
                            not self.disjoint_set.connected(current_pixel, neighbor_pixel)):
                            self.disjoint_set.union(current_pixel, neighbor_pixel)
                            merged_count += 1
        
        return merged_count
    
    def get_segments(self):
        """Get all image segments"""
        segments = {}
        for row in range(self.height):
            for col in range(self.width):
                pixel = (row, col)
                if pixel in self.disjoint_set:
                    root = self.disjoint_set.find(pixel)
                    
                    if root not in segments:
                        segments[root] = []
                    segments[root].append(pixel)
                else:
                    # Isolated pixel
                    segments[pixel] = [pixel]
        
        return segments
    
    def segment_statistics(self):
        """Get segmentation statistics"""
        segments = self.get_segments()
        sizes = [len(segment) for segment in segments.values()]
        
        return {
            'num_segments': len(segments),
            'largest_segment': max(sizes) if sizes else 0,
            'smallest_segment': min(sizes) if sizes else 0,
            'average_segment_size': sum(sizes) / len(sizes) if sizes else 0
        }

# Create small image and simulate segmentation
print("Creating 4x4 sample image:")
image = ImageSegmentation(4, 4)

# Set pixel values to create regions
pixel_data = [
    [100, 105, 200, 195],
    [95,  110, 205, 190],
    [50,  45,  210, 200],
    [55,  40,  50,  55]
]

for row in range(4):
    for col in range(4):
        image.set_pixel_value(row, col, pixel_data[row][col])

print("Pixel values:")
for row in pixel_data:
    print(f"  {row}")

# Count initial segments (all pixels separate initially)
initial_segments = 4 * 4
print(f"\nBefore segmentation: {initial_segments} segments")

# Perform segmentation
merged = image.merge_similar_pixels(threshold=15)
print(f"Merged {merged} pixel pairs")

print(f"After segmentation: {len(image.disjoint_set)} pixels in disjoint set")

# Show segments
segments = image.get_segments()
print(f"\nSegments found: {len(segments)}")
for i, (root, pixels) in enumerate(segments.items(), 1):
    values = [image.pixel_values[p] for p in pixels]
    avg_value = sum(values) / len(values)
    print(f"  Segment {i} (root: {root}): {len(pixels)} pixels, avg value: {avg_value:.1f}")
    print(f"    Pixels: {sorted(pixels)[:5]}{'...' if len(pixels) > 5 else ''}")

# Statistics
stats = image.segment_statistics()
print(f"\nSegmentation Statistics:")
for key, value in stats.items():
    if isinstance(value, float):
        print(f"  {key}: {value:.2f}")
    else:
        print(f"  {key}: {value}")

print("\n✓ Dynamic connectivity and optimization techniques demonstrated!")
```

## Example 5: Production Applications and Performance Analysis

Demonstrate production-ready Union-Find applications with comprehensive performance analysis.

```python
# Example 5: Production applications and performance analysis

from disjoint_set import DisjointSet
import time
import random

# Production-grade Union-Find system
class ProductionUnionFind:
    def __init__(self, name="UnionFind"):
        self.name = name
        self.disjoint_set = DisjointSet()
        self.operation_log = []
        self.performance_metrics = {
            'union_count': 0,
            'find_count': 0,
            'union_time': 0.0,
            'find_time': 0.0
        }
        
    def timed_union(self, x, y):
        """Union operation with timing"""
        start_time = time.time()
        
        # Elements are added automatically by union
        result = self.disjoint_set.union(x, y)
        
        elapsed = time.time() - start_time
        self.performance_metrics['union_count'] += 1
        self.performance_metrics['union_time'] += elapsed
        
        self.operation_log.append({
            'operation': 'union',
            'args': (x, y),
            'time': elapsed,
            'timestamp': time.time(),
            'sets_after': len(self.disjoint_set)
        })
        
        return result
    
    def timed_find(self, x):
        """Find operation with timing"""
        start_time = time.time()
        result = self.disjoint_set.find(x)
        elapsed = time.time() - start_time
        
        self.performance_metrics['find_count'] += 1
        self.performance_metrics['find_time'] += elapsed
        
        return result
    
    def get_performance_report(self):
        """Generate comprehensive performance report"""
        union_count = self.performance_metrics['union_count']
        find_count = self.performance_metrics['find_count']
        union_time = self.performance_metrics['union_time']
        find_time = self.performance_metrics['find_time']
        
        report = {
            'system_name': self.name,
            'total_operations': union_count + find_count,
            'union_operations': union_count,
            'find_operations': find_count,
            'total_time': union_time + find_time,
            'avg_union_time': union_time / union_count if union_count > 0 else 0,
            'avg_find_time': find_time / find_count if find_count > 0 else 0,
            'current_sets': len(self.disjoint_set),
            'operations_per_second': (union_count + find_count) / (union_time + find_time) if (union_time + find_time) > 0 else 0
        }
        
        return report

# Application 1: Distributed System Cluster Management
print("=== Production Application: Cluster Management ===")

class ClusterManager:
    def __init__(self):
        self.union_find = ProductionUnionFind("ClusterManager")
        self.node_metadata = {}
        self.cluster_events = []
        
    def add_node(self, node_id, metadata=None):
        """Add node to cluster management"""
        # No need to call .add() - will be added on first union/find
        self.node_metadata[node_id] = metadata or {}
        
        event = {
            'type': 'node_added',
            'node_id': node_id,
            'timestamp': time.time()
        }
        self.cluster_events.append(event)
        
    def join_clusters(self, node1, node2, reason="manual"):
        """Join two cluster partitions"""
        # Check if already in same cluster
        root1 = self.union_find.timed_find(node1)
        root2 = self.union_find.timed_find(node2)
        
        if root1 == root2:
            return f"Nodes {node1} and {node2} already in same cluster"
        
        # Perform cluster merge
        self.union_find.timed_union(node1, node2)
        
        event = {
            'type': 'clusters_merged',
            'nodes': (node1, node2),
            'reason': reason,
            'timestamp': time.time()
        }
        self.cluster_events.append(event)
        
        return f"Merged clusters containing {node1} and {node2}"
    
    def get_cluster_info(self, node_id):
        """Get cluster information for a node"""
        cluster_root = self.union_find.timed_find(node_id)
        
        # Find all nodes in same cluster
        cluster_nodes = []
        for node in self.node_metadata:
            if node in self.union_find.disjoint_set:
                if self.union_find.timed_find(node) == cluster_root:
                    cluster_nodes.append(node)
        
        return {
            'cluster_root': cluster_root,
            'cluster_size': len(cluster_nodes),
            'cluster_nodes': sorted(cluster_nodes)
        }
    
    def get_all_clusters(self):
        """Get information about all clusters"""
        clusters = {}
        for node in self.node_metadata:
            if node in self.union_find.disjoint_set:
                root = self.union_find.timed_find(node)
                if root not in clusters:
                    clusters[root] = []
                clusters[root].append(node)
        
        return clusters

# Simulate cluster management scenario
cluster_mgr = ClusterManager()

# Add nodes to cluster
nodes = [f"node-{i:03d}" for i in range(1, 13)]
print(f"Adding {len(nodes)} nodes to cluster management:")

for node in nodes:
    metadata = {
        'region': random.choice(['us-east', 'us-west', 'eu-central']),
        'instance_type': random.choice(['t3.medium', 't3.large', 'c5.xlarge']),
        'availability_zone': random.choice(['a', 'b', 'c'])
    }
    cluster_mgr.add_node(node, metadata)

print(f"Initial state: {len(nodes)} separate nodes")

# Simulate cluster formation based on regions
print(f"\nForming regional clusters:")
region_groups = {}
for node in nodes:
    region = cluster_mgr.node_metadata[node]['region']
    if region not in region_groups:
        region_groups[region] = []
    region_groups[region].append(node)

for region, region_nodes in region_groups.items():
    print(f"  Forming {region} cluster with {len(region_nodes)} nodes")
    for i in range(1, len(region_nodes)):
        result = cluster_mgr.join_clusters(region_nodes[0], region_nodes[i], f"regional_clustering_{region}")

# Show final cluster state
clusters = cluster_mgr.get_all_clusters()
print(f"\nFinal cluster state: {len(clusters)} clusters")
for i, (root, cluster_nodes) in enumerate(clusters.items(), 1):
    # Get common region
    regions = [cluster_mgr.node_metadata[node]['region'] for node in cluster_nodes]
    common_region = regions[0] if all(r == regions[0] for r in regions) else "mixed"
    
    print(f"  Cluster {i} ({root}): {len(cluster_nodes)} nodes in {common_region}")
    print(f"    Nodes: {sorted(cluster_nodes)}")

# Application 2: Data Deduplication System
print(f"\n=== Production Application: Data Deduplication ===")

class DataDeduplicationSystem:
    def __init__(self):
        self.union_find = ProductionUnionFind("DataDeduplication")
        self.file_hashes = {}
        self.file_metadata = {}
        self.duplicate_groups = {}
        
    def add_file(self, file_path, content_hash, size, timestamp):
        """Add file to deduplication system"""
        # No need to call .add() - will be added on first union/find
        self.file_hashes[file_path] = content_hash
        self.file_metadata[file_path] = {
            'size': size,
            'timestamp': timestamp,
            'content_hash': content_hash
        }
    
    def find_and_merge_duplicates(self):
        """Find and group duplicate files"""
        hash_to_files = {}
        
        # Group files by content hash
        for file_path, content_hash in self.file_hashes.items():
            if content_hash not in hash_to_files:
                hash_to_files[content_hash] = []
            hash_to_files[content_hash].append(file_path)
        
        # Merge files with same hash
        duplicates_found = 0
        for content_hash, files in hash_to_files.items():
            if len(files) > 1:
                # Union all files with same hash
                primary_file = files[0]
                for duplicate_file in files[1:]:
                    self.union_find.timed_union(primary_file, duplicate_file)
                    duplicates_found += 1
        
        return duplicates_found
    
    def get_duplicate_groups(self):
        """Get groups of duplicate files"""
        groups = {}
        for file_path in self.file_metadata:
            if file_path in self.union_find.disjoint_set:
                root = self.union_find.timed_find(file_path)
                if root not in groups:
                    groups[root] = []
                groups[root].append(file_path)
        
        # Filter to only groups with multiple files (duplicates)
        duplicate_groups = {root: files for root, files in groups.items() if len(files) > 1}
        return duplicate_groups
    
    def calculate_space_savings(self):
        """Calculate potential space savings from deduplication"""
        duplicate_groups = self.get_duplicate_groups()
        total_savings = 0
        
        savings_details = []
        for root, files in duplicate_groups.items():
            # Keep the oldest file (smallest timestamp), remove others
            files_with_meta = [(f, self.file_metadata[f]) for f in files]
            files_with_meta.sort(key=lambda x: x[1]['timestamp'])  # Sort by timestamp
            
            primary_file = files_with_meta[0]
            duplicate_files = files_with_meta[1:]
            
            group_savings = sum(meta['size'] for _, meta in duplicate_files)
            total_savings += group_savings
            
            savings_details.append({
                'primary_file': primary_file[0],
                'duplicates': [f for f, _ in duplicate_files],
                'savings_bytes': group_savings
            })
        
        return total_savings, savings_details

# Simulate deduplication scenario
print("Simulating file deduplication system:")
dedup_system = DataDeduplicationSystem()

# Sample files with some duplicates
sample_files = [
    ("/home/docs/report1.pdf", "hash_abc123", 1024000, 1609459200),
    ("/backup/reports/report1.pdf", "hash_abc123", 1024000, 1609459300),  # Duplicate
    ("/home/photos/vacation.jpg", "hash_def456", 2048000, 1609459400),
    ("/backup/photos/vacation_copy.jpg", "hash_def456", 2048000, 1609459500),  # Duplicate
    ("/home/docs/manual.txt", "hash_ghi789", 8192, 1609459600),
    ("/temp/manual_backup.txt", "hash_ghi789", 8192, 1609459700),  # Duplicate
    ("/home/videos/movie.mp4", "hash_jkl012", 50331648, 1609459800),  # Unique
    ("/home/docs/report2.pdf", "hash_mno345", 1536000, 1609459900),  # Unique
]

print(f"Adding {len(sample_files)} files to deduplication system:")
for file_path, content_hash, size, timestamp in sample_files:
    dedup_system.add_file(file_path, content_hash, size, timestamp)
    print(f"  Added: {file_path} (size: {size:,} bytes)")

print(f"\nBefore deduplication: {len(sample_files)} files")

# Find and merge duplicates
duplicates_found = dedup_system.find_and_merge_duplicates()
print(f"Found and merged {duplicates_found} duplicate files")
print(f"After deduplication: {len(dedup_system.union_find.disjoint_set)} files in disjoint set")

# Calculate space savings
total_savings, savings_details = dedup_system.calculate_space_savings()
print(f"\nSpace savings analysis:")
print(f"Total space that can be saved: {total_savings:,} bytes ({total_savings/1024/1024:.2f} MB)")

for i, details in enumerate(savings_details, 1):
    print(f"  Duplicate group {i}:")
    print(f"    Keep: {details['primary_file']}")
    print(f"    Remove: {details['duplicates']}")
    print(f"    Savings: {details['savings_bytes']:,} bytes")

# Performance analysis
print(f"\n=== Performance Analysis ===")

# Cluster management performance
cluster_report = cluster_mgr.union_find.get_performance_report()
print(f"\nCluster Management Performance:")
print(f"  System: {cluster_report['system_name']}")
print(f"  Total operations: {cluster_report['total_operations']}")
print(f"  Union operations: {cluster_report['union_operations']}")
print(f"  Find operations: {cluster_report['find_operations']}")
print(f"  Total time: {cluster_report['total_time']:.6f}s")
print(f"  Operations per second: {cluster_report['operations_per_second']:.2f}")
print(f"  Average union time: {cluster_report['avg_union_time']*1000000:.2f}μs")
print(f"  Average find time: {cluster_report['avg_find_time']*1000000:.2f}μs")

# Deduplication performance
dedup_report = dedup_system.union_find.get_performance_report()
print(f"\nDeduplication System Performance:")
print(f"  System: {dedup_report['system_name']}")
print(f"  Total operations: {dedup_report['total_operations']}")
print(f"  Union operations: {dedup_report['union_operations']}")
print(f"  Find operations: {dedup_report['find_operations']}")
print(f"  Total time: {dedup_report['total_time']:.6f}s")
print(f"  Operations per second: {dedup_report['operations_per_second']:.2f}")

print("\n✅ Production applications and performance analysis complete!")
print("Union-Find provides efficient solutions for clustering, deduplication, and connectivity problems.")
```
