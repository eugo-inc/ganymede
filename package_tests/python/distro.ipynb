## Example 1: Basic OS distribution information

## Example 2: Version parsing and comparison


```python
import distro

# Example 2: Version parsing and comparison
print("Example 2: Version parsing and comparison")

try:
    # Get detailed version information
    version_info = distro.version_parts()
    print(f"Version parts: {version_info}")
    
    major_version = distro.major_version()
    minor_version = distro.minor_version()
    build_number = distro.build_number()
    
    print(f"Major version: {major_version}")
    print(f"Minor version: {minor_version}")
    print(f"Build number: {build_number}")
    
    # Get version as different formats
    version_str = distro.version()
    version_pretty = distro.version(pretty=True)
    version_best = distro.version(best=True)
    
    print(f"Version string: {version_str}")
    print(f"Pretty version: {version_pretty}")
    print(f"Best version: {version_best}")
    
    print("Version parsing completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"distro example (cell 4) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished")
```

## Example 3: Distribution family and platform identification


```python
import distro

# Example 3: Distribution family and platform identification
print("Example 3: Distribution family and platform identification")

try:
    # Get distribution family information
    distro_like = distro.like()
    print(f"Distribution like: {distro_like}")
    
    # Get codename
    codename = distro.codename()
    print(f"Codename: {codename}")
    
    # Check for specific distributions
    distro_id = distro.id()
    print(f"Distribution ID: {distro_id}")
    
    # Platform detection examples
    if distro_id in ['ubuntu', 'debian']:
        print("This is a Debian-based system")
    elif distro_id in ['rhel', 'centos', 'fedora']:
        print("This is a Red Hat-based system")
    elif distro_id in ['arch']:
        print("This is an Arch-based system")
    else:
        print(f"This is a {distro_id} system")
    
    # Get full description
    description = distro.name(pretty=True)
    print(f"Full description: {description}")
    
    print("Distribution family identification completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"distro example (cell 6) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished")
```

## Example 4: OS release information and metadata


```python
import distro

# Example 4: OS release information and metadata
print("Example 4: OS release information and metadata")

try:
    # Get comprehensive OS release info
    os_release_info = distro.os_release_info()
    print("OS Release Information:")
    for key, value in os_release_info.items():
        print(f"  {key}: {value}")
    
    # Get LSB release info if available
    try:
        lsb_info = distro.lsb_release_info()
        print("\nLSB Release Information:")
        for key, value in lsb_info.items():
            print(f"  {key}: {value}")
    except:
        print("\nLSB Release information not available")
        raise
    
    # Get distro release info
    try:
        distro_info = distro.distro_release_info()
        print("\nDistro Release Information:")
        for key, value in distro_info.items():
            print(f"  {key}: {value}")
    except:
        print("\nDistro release information not available")
        raise
    
    print("OS release information completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"distro example (cell 8) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished")
```

## Example 5: Cross-platform compatibility and detection


```python
import distro
import platform

# Example 5: Cross-platform compatibility and detection
print("Example 5: Cross-platform compatibility and detection")

try:
    # Compare distro info with platform module
    print("=== Distro module info ===")
    print(f"System: {distro.name()}")
    print(f"Version: {distro.version()}")
    print(f"ID: {distro.id()}")
    
    print("\n=== Platform module info ===")
    print(f"System: {platform.system()}")
    print(f"Release: {platform.release()}")
    print(f"Version: {platform.version()}")
    
    # Demonstrate cross-platform detection
    def get_package_manager():
        distro_id = distro.id()
        if distro_id in ['ubuntu', 'debian']:
            return 'apt'
        elif distro_id in ['rhel', 'centos', 'fedora']:
            return 'yum/dnf'
        elif distro_id in ['arch']:
            return 'pacman'
        elif distro_id in ['opensuse']:
            return 'zypper'
        else:
            return 'unknown'
    
    pkg_manager = get_package_manager()
    print(f"\nLikely package manager: {pkg_manager}")
    
    # Create a system summary
    summary = {
        'os': distro.name(),
        'version': distro.version(),
        'id': distro.id(),
        'codename': distro.codename(),
        'like': distro.like()
    }
    
    print(f"\nSystem summary: {summary}")
    
    print("Cross-platform compatibility detection completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"distro example (cell 10) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished")
```
