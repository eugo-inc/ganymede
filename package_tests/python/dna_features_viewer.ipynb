```python
# Example 1: Create a GraphicFeature and inspect its attributes.
from dna_features_viewer import GraphicFeature

feature = GraphicFeature(
    start=10,
    end=80,
    strand=+1,
    color='#ffcccc',
    label='Gene A',
)

print(f"start={feature.start}, end={feature.end}, strand={feature.strand}")
print(f"label={feature.label}, color={feature.color}")
print(f"length={feature.end - feature.start}")

assert feature.start == 10
assert feature.end == 80
assert feature.strand == 1
assert feature.label == 'Gene A'
```

```python
# Example 2: Build a GraphicRecord with multiple features and render to matplotlib.
import matplotlib
matplotlib.use('Agg')
from dna_features_viewer import GraphicFeature, GraphicRecord

features = [
    GraphicFeature(start=0, end=20, strand=+1, color='#ffd700', label='promoter'),
    GraphicFeature(start=20, end=120, strand=+1, color='#cffccc', label='gene'),
    GraphicFeature(start=120, end=140, strand=+1, color='#ccccff', label='terminator'),
    GraphicFeature(start=145, end=200, strand=-1, color='#ffcccc', label='antisense'),
]
record = GraphicRecord(sequence_length=220, features=features)
ax, _ = record.plot(figure_width=6)

assert len(record.features) == 4
assert record.sequence_length == 220
assert ax is not None
```

```python
# Example 3: CircularGraphicRecord renders plasmid-style maps.
import matplotlib
matplotlib.use('Agg')
from dna_features_viewer import GraphicFeature, CircularGraphicRecord

features = [
    GraphicFeature(start=0, end=300, strand=+1, color='#ffaaaa', label='ori'),
    GraphicFeature(start=400, end=900, strand=+1, color='#aaffaa', label='ampR'),
    GraphicFeature(start=1000, end=1500, strand=-1, color='#aaaaff', label='kanR'),
]
record = CircularGraphicRecord(sequence_length=2000, features=features)
ax, _ = record.plot(figure_width=5)

assert record.sequence_length == 2000
assert len(record.features) == 3
assert ax is not None
```

```python
# Example 4: crop() returns a sub-record with shifted coordinates.
from dna_features_viewer import GraphicFeature, GraphicRecord

features = [
    GraphicFeature(start=10, end=50, strand=+1, color='#fcc', label='a'),
    GraphicFeature(start=60, end=110, strand=-1, color='#cfc', label='b'),
    GraphicFeature(start=120, end=180, strand=+1, color='#ccf', label='c'),
]
record = GraphicRecord(sequence_length=200, features=features)
sub = record.crop((40, 130))

labels = sorted(f.label for f in sub.features if f.label)
print(f"cropped labels: {labels}")
print(f"cropped sequence_length: {sub.sequence_length}")

assert sub.sequence_length == 90
assert 'b' in labels
```

```python
# Example 5: BiopythonTranslator converts a Biopython SeqRecord into a GraphicRecord.
import matplotlib
matplotlib.use('Agg')
from Bio.SeqRecord import SeqRecord
from Bio.Seq import Seq
from Bio.SeqFeature import SeqFeature, FeatureLocation
from dna_features_viewer import BiopythonTranslator

seq = SeqRecord(Seq('A' * 300), id='demo')
seq.features = [
    SeqFeature(FeatureLocation(10, 80, strand=+1), type='CDS', qualifiers={'gene': ['x']}),
    SeqFeature(FeatureLocation(120, 200, strand=-1), type='gene', qualifiers={'gene': ['y']}),
]
translator = BiopythonTranslator()
graphic = translator.translate_record(seq)
ax, _ = graphic.plot(figure_width=6)

assert graphic.sequence_length == 300
assert len(graphic.features) == 2
assert ax is not None
```
