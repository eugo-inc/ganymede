## Example 1: Basic reStructuredText Processing

```python
from docutils.core import publish_doctree, publish_parts

print("Example 1: Basic reStructuredText Processing")

try:
    # Basic reStructuredText content
    rst_content = """
Title
=====

This is a paragraph with **bold** and *italic* text.

- Item 1
- Item 2

Code example::

    print("Hello World")
"""

    print("✓ Processing reStructuredText:")
    
    # Parse to HTML
    html_parts = publish_parts(rst_content, writer_name='html')
    print(f"  HTML body length: {len(html_parts['body'])} chars")
    
    # Parse to document tree
    doctree = publish_doctree(rst_content)
    print(f"  Document tree: {doctree.tagname}")
    print(f"  Child nodes: {len(list(doctree.children))}")
    
    print("✓ Basic reStructuredText processing completed")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"docutils example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished")
```

## Example 2: Document Structure Analysis

```python
from docutils.core import publish_doctree
from docutils import nodes

print("Example 2: Document Structure Analysis")

try:
    # Document with sections
    rst_doc = """
Main Title
==========

Section 1
---------

Paragraph in section 1.

Section 2
---------

Paragraph in section 2.
"""

    print("✓ Analyzing document structure:")
    
    doctree = publish_doctree(rst_doc)
    
    # Count different node types
    sections = list(doctree.traverse(nodes.section))
    paragraphs = list(doctree.traverse(nodes.paragraph))
    titles = list(doctree.traverse(nodes.title))
    
    print(f"  Sections found: {len(sections)}")
    print(f"  Paragraphs found: {len(paragraphs)}")
    print(f"  Titles found: {len(titles)}")
    
    # Show section hierarchy
    for i, section in enumerate(sections):
        title = section[0].astext() if section.children else "Untitled"
        print(f"  Section {i+1}: {title}")
    
    print("✓ Document structure analysis completed")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"docutils example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished")
```

## Example 3: Format Conversion and Writers

```python
from docutils.core import publish_string

print("Example 3: Format Conversion and Writers")

try:
    # Sample document
    rst_text = """
Documentation
=============

This document contains:

1. **Bold text**
2. *Italic text*
3. ``Code snippets``

A code block::

    def hello():
        print("Hello!")
"""

    print("✓ Converting to multiple formats:")
    
    # Convert to HTML
    html_output = publish_string(rst_text, writer_name='html')
    print(f"  HTML output: {len(html_output)} bytes")
    
    # Convert to XML
    xml_output = publish_string(rst_text, writer_name='xml')
    print(f"  XML output: {len(xml_output)} bytes")
    
    print("✓ Format conversion completed")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"docutils example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished")
```

## Example 4: Custom Directives and Extensions

```python
from docutils.core import publish_doctree
from docutils.parsers.rst import directives, Directive
from docutils import nodes

print("Example 4: Custom Directives and Extensions")

try:
    # Simple custom directive
    class HighlightDirective(Directive):
        has_content = True
        
        def run(self):
            text = '\n'.join(self.content)
            node = nodes.paragraph(text=f"[HIGHLIGHT] {text}")
            return [node]
    
    # Register directive
    directives.register_directive('highlight', HighlightDirective)
    
    print("✓ Testing custom directive:")
    
    # Document with custom directive
    rst_with_directive = """
Test Document
=============

.. highlight::
   
   This text should be highlighted!

Regular paragraph.
"""
    
    doctree = publish_doctree(rst_with_directive)
    paragraphs = list(doctree.traverse(nodes.paragraph))
    
    print(f"  Paragraphs processed: {len(paragraphs)}")
    for i, para in enumerate(paragraphs):
        text = para.astext()[:30] + "..." if len(para.astext()) > 30 else para.astext()
        print(f"    {i+1}: {text}")
    
    print("✓ Custom directives completed")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"docutils example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished")
```

## Example 5: Practical Documentation Processing

```python
from docutils.core import publish_parts

print("Example 5: Practical Documentation Processing")

try:
    # API documentation content
    documentation = """
API Reference
=============

Module Overview
---------------

This module provides utilities for data processing.

**process_data(data, options=None)**

  Process input data with optional parameters.
  
  Example usage::
  
      result = process_data([1, 2, 3])
"""

    print("✓ Processing API documentation:")
    
    # Process with HTML writer
    parts = publish_parts(documentation, writer_name='html')
    
    # Extract useful parts
    title = parts.get('title', 'No title')
    body = parts.get('body', '')
    
    print(f"  Document title: {title}")
    print(f"  Body content: {len(body)} characters")
    
    # Check for common elements
    has_sections = '<div class="section"' in body
    has_code = '<pre class="literal-block"' in body
    
    print(f"  Contains sections: {has_sections}")
    print(f"  Contains code blocks: {has_code}")
    
    print("✓ Documentation processing completed")
    
except Exception as e:
    print(f"❌ Error: {e}")
    raise AssertionError(
        f"docutils example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished")
```
