```python
# Example 1: parse a Go-style duration string to a timedelta.
import datetime
import durationpy

td = durationpy.from_str("1h30m")
print(f"1h30m -> {td!r}")
assert isinstance(td, datetime.timedelta)
assert td == datetime.timedelta(hours=1, minutes=30)
```

```python
# Example 2: format a timedelta back into a Go-style duration string.
import datetime
import durationpy

td = datetime.timedelta(seconds=90)
s = durationpy.to_str(td)
print(f"timedelta(seconds=90) -> {s!r}")
# Round-trip parse for confirmation.
assert durationpy.from_str(s) == td
```

```python
# Example 3: parse sub-second units (ms, us, ns).
import datetime
import durationpy

samples = ["500ms", "250us", "1000ns"]
for s in samples:
    td = durationpy.from_str(s)
    print(f"{s:>8} -> {td!r}")
    assert isinstance(td, datetime.timedelta)

# 500ms == 0.5s
assert durationpy.from_str("500ms") == datetime.timedelta(milliseconds=500)
```

```python
# Example 4: parse compound durations with multiple units.
import datetime
import durationpy

td = durationpy.from_str("2h45m30s")
print(f"2h45m30s -> {td!r}")
expected = datetime.timedelta(hours=2, minutes=45, seconds=30)
assert td == expected

# Total seconds sanity.
assert int(td.total_seconds()) == 2 * 3600 + 45 * 60 + 30
```

```python
# Example 5: invalid input raises, and the extended `to_str` form works.
import datetime
import durationpy

try:
    durationpy.from_str("not a duration")
except durationpy.DurationError as e:
    print(f"rejected as expected -> {type(e).__name__} ({e})")
else:
    raise AssertionError("durationpy.from_str should raise DurationError on malformed input")

# `to_str(extended=True)` is supported on recent versions; tolerate either signature.
td = datetime.timedelta(days=2, hours=3)
try:
    s = durationpy.to_str(td, extended=True)
except TypeError:
    s = durationpy.to_str(td)
print(f"timedelta(days=2, hours=3) -> {s!r}")
assert isinstance(s, str) and len(s) > 0
```
