```python
# Example 4: Request validation and local GRIB file handling (offline)

from ecmwf.opendata import Client
import tempfile
import os
from datetime import datetime

# Build a small request and validate it locally (nothing is sent), then handle a locally written GRIB file
print("=== ECMWF Request Validation and Local GRIB File Handling (offline) ===")

try:
    client = Client()
    
    # Temporary directory for the local GRIB file
    with tempfile.TemporaryDirectory() as temp_dir:
        print(f"=== Small Request, Validated Locally ===")
        print(f"Temporary directory: {temp_dir}")
        
        # Define a very small request
        small_request = {
            "type": "fc",           # forecast
            "param": "2t",          # 2-meter temperature only
            "step": 0,              # analysis time (smallest)
            "area": [60, -10, 50, 0],  # Small European region (UK area)
            "grid": [1.0, 1.0],     # Coarse resolution (1 degree)
            "format": "grib"
        }
        
        target_file = os.path.join(temp_dir, "small_sample.grib")
        
        print(f"Request:")
        for param, value in small_request.items():
            print(f"  {param}: {value}")
        print(f"  target: {target_file}")
        
        print(f"\nRequest validated locally -- NOT sent:")
        assert small_request["type"] == "fc" and small_request["param"] == "2t"
        assert len(small_request["area"]) == 4, "area is [north, west, south, east]"
        assert small_request["grid"] == [1.0, 1.0]
        print("  request shape OK (type/param/area/grid)")

        # The file-handling half of this example, against a locally written GRIB stub: a real GRIB2
        # message opens with the ASCII magic `GRIB`, the first thing any reader checks.
        with open(target_file, 'wb') as f:
            f.write(b"GRIB" + b"\x00" * 96)
        file_size = os.path.getsize(target_file)
        print(f"  File size: {file_size:,} bytes ({file_size/1024:.1f} KB)")
        print(f"  File location: {target_file}")
        with open(target_file, 'rb') as f:
            header = f.read(100)  # Read first 100 bytes
            print(f"  File header (first 20 bytes): {header[:20]}")
        assert header[:4] == b"GRIB", "GRIB magic missing from the sample file"
    
    # Demonstrate request optimization for minimal downloads
    print(f"\n=== Request Optimization Strategies ===")
    
    optimization_examples = {
        "Minimal geographic area": {
            "area": [51, -1, 50, 0],  # 1x1 degree around London
            "benefit": "Reduces data size by limiting spatial extent"
        },
        "Single parameter": {
            "param": "2t",  # Just temperature
            "benefit": "Fewer variables = smaller files"
        },
        "Analysis only": {
            "type": "an", "step": 0,
            "benefit": "No forecast steps = immediate data"
        },
        "Coarse resolution": {
            "grid": [2.0, 2.0],  # 2-degree spacing
            "benefit": "Lower resolution = fewer data points"
        }
    }
    
    for strategy, details in optimization_examples.items():
        print(f"\n{strategy}:")
        for key, value in details.items():
            if key != "benefit":
                print(f"  {key}: {value}")
        print(f"  Benefit: {details['benefit']}")
    
    # File format considerations
    print(f"\n=== File Format and Size Considerations ===")
    
    format_comparison = {
        "GRIB": {
            "typical_size": "Compact, efficient compression",
            "pros": "Standard meteorological format",
            "cons": "Requires specialized readers",
            "use_case": "Operational meteorology"
        },
        "NetCDF": {
            "typical_size": "Larger but more accessible", 
            "pros": "Self-describing, widely supported",
            "cons": "Larger file sizes",
            "use_case": "Research and analysis"
        }
    }
    
    for fmt, properties in format_comparison.items():
        print(f"\n{fmt}:")
        for prop, value in properties.items():
            print(f"  {prop}: {value}")
    
    # Download best practices
    print(f"\n=== Download Best Practices ===")
    
    best_practices = [
        "✓ Start with small geographic areas for testing",
        "✓ Use single parameters to minimize file size",
        "✓ Request analysis (step=0) before forecasts",
        "✓ Use temporary directories for test downloads",
        "✓ Implement proper error handling for network issues",
        "✓ Check file existence before processing",
        "✓ Monitor download sizes and times",
        "✓ Clean up temporary files after use",
        "✓ Validate downloaded data integrity",
        "✓ Consider bandwidth and storage limitations"
    ]
    
    for practice in best_practices:
        print(f"  {practice}")
    
    # Error handling patterns
    print(f"\n=== Common Error Scenarios ===")
    
    error_scenarios = {
        "Network connectivity": {
            "error": "Connection timeout or refused",
            "solution": "Check internet connection, retry with backoff"
        },
        "Invalid parameters": {
            "error": "Bad request or parameter validation",
            "solution": "Verify parameter names and valid combinations"
        },
        "Data not available": {
            "error": "Requested data doesn't exist",
            "solution": "Check date/time validity and data availability"
        },
        "Authentication": {
            "error": "Access denied or forbidden",
            "solution": "Verify account access and usage limits"
        }
    }
    
    for scenario, details in error_scenarios.items():
        print(f"\n{scenario}:")
        print(f"  Error: {details['error']}")
        print(f"  Solution: {details['solution']}")

    print("Request validation and local GRIB file handling completed successfully")

except Exception as e:
    print(f"Request validation / file handling failed: {e}")
    raise AssertionError(
        f"ecmwf_opendata example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```
