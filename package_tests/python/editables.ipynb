## Example 1: editables functionality test 1

```python
import editables

# Example 1: Editable package installation support
print("Example 1: Editable package installation support")

try:
    # Create an editable finder for a sample directory
    import tempfile
    import os
    
    # Create a temporary directory to simulate a package
    temp_dir = tempfile.mkdtemp()
    package_dir = os.path.join(temp_dir, 'mypackage')
    os.makedirs(package_dir)
    
    # Create a simple __init__.py
    init_file = os.path.join(package_dir, '__init__.py')
    with open(init_file, 'w') as f:
        f.write('__version__ = "0.1.0"\n')
    
    print(f"Created sample package at: {package_dir}")
    
    # Test editables functionality
    # `EditableProject` maps import names to on-disk paths and emits the .pth/.py shim pair a wheel would ship.
    project = editables.EditableProject('mypackage', temp_dir)
    print(f"EditableProject created: {type(project)}")
    
    # Map the import name onto the package directory
    project.map('mypackage', package_dir)
    print(f"Mapped mypackage -> {package_dir}")
    
    # Show what the project would install
    print(f"Generated files: {[name for name, _ in project.files()]}")
    print(f"Runtime dependencies: {project.dependencies()}")
    
    # Clean up
    import shutil
    shutil.rmtree(temp_dir)
    
    print("editables example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"editables example (cell 2) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```
