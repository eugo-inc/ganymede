```python
# Example 1: basic global edit distance (NW mode) on identical sequences.
import edlib

result = edlib.align("ACGT", "ACGT")
print(result)
assert result["editDistance"] == 0
assert isinstance(result, dict)
assert "editDistance" in result
assert "locations" in result
assert "alphabetLength" in result
```

```python
# Example 2: edit distance with substitutions / insertions / deletions.
import edlib

# 1 substitution: ACGT -> AGGT
r_sub = edlib.align("ACGT", "AGGT")
assert r_sub["editDistance"] == 1

# 1 insertion: ACGT -> ACGTA
r_ins = edlib.align("ACGT", "ACGTA")
assert r_ins["editDistance"] == 1

# 1 deletion: ACGT -> ACG
r_del = edlib.align("ACGT", "ACG")
assert r_del["editDistance"] == 1

# 3 differences
r_three = edlib.align("ACGT", "TGCA")
assert r_three["editDistance"] >= 1
print(f"sub={r_sub['editDistance']} ins={r_ins['editDistance']} del={r_del['editDistance']} mixed={r_three['editDistance']}")
```

```python
# Example 3: infix / HW mode — find query as a substring of a longer target.
import edlib

query = "ACGT"
target = "GGGGACGTGGGG"

result = edlib.align(query, target, mode="HW")
print(result)
assert result["editDistance"] == 0  # query found exactly inside target
assert len(result["locations"]) >= 1

# With 1 mismatch allowed
result2 = edlib.align("ACAT", target, mode="HW")
assert result2["editDistance"] == 1
```

```python
# Example 4: alignment task with CIGAR / path output and getNiceAlignment.
import edlib

query = "ACGTACGT"
target = "ACGTTCGT"

result = edlib.align(query, target, mode="NW", task="path")
print(result)
assert result["editDistance"] == 1
assert result["cigar"] is not None
assert isinstance(result["cigar"], str)

nice = edlib.getNiceAlignment(result, query, target)
print(nice["query_aligned"])
print(nice["matched_aligned"])
print(nice["target_aligned"])
assert "query_aligned" in nice
assert "target_aligned" in nice
```

```python
# Example 5: k-bound + extended CIGAR + additionalEqualities (IUPAC ambiguity).
import edlib

# k-bound: if true distance > k, returns -1
r_kbound = edlib.align("ACGTACGT", "TTTTTTTT", mode="NW", k=2)
assert r_kbound["editDistance"] == -1

# extended CIGAR with task="path"
r_ext = edlib.align("ACGT", "AGGT", mode="NW", task="path")
assert r_ext["cigar"] is not None

# additionalEqualities: treat N as matching any base
r_iupac = edlib.align(
    "ACNT",
    "ACGT",
    mode="NW",
    task="distance",
    additionalEqualities=[("N", "A"), ("N", "C"), ("N", "G"), ("N", "T")],
)
assert r_iupac["editDistance"] == 0
print(f"kbound={r_kbound['editDistance']} iupac_dist={r_iupac['editDistance']}")
print("All edlib smoke tests passed")
```
