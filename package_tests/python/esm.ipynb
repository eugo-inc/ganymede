```python
# Example 1: SDK top-level surface — confirm primary client + model entry points exist.
import esm
from esm import sdk
from esm.sdk import api as sdk_api

# The SDK exposes ESM3 / ESMC client constructors + request/response data classes.
expected_names = ['ESMProtein', 'ESM3InferenceClient', 'GenerationConfig']
missing = [n for n in expected_names if not hasattr(sdk_api, n)]
print(f"esm.sdk module: {sdk}")
print(f"esm.sdk.api exports present: {[n for n in expected_names if hasattr(sdk_api, n)]}")
print(f"missing: {missing}")
assert not missing, f"esm.sdk.api missing expected names: {missing}"
```

```python
# Example 2: ESMProtein data class — construct from a sequence string, round-trip basic attrs.
from esm.sdk.api import ESMProtein

seq = "MKTAYIAKQRQISFVKSHFSRQLEERLGLIEVQAPILSRVGDGTQDNLSGAEK"
protein = ESMProtein(sequence=seq)

print(f"type: {type(protein).__name__}")
print(f"sequence length: {len(protein.sequence)}")
print(f"sequence[:20]: {protein.sequence[:20]}")

# Structure / SASA / function tracks default to None until predicted.
assert protein.sequence == seq
assert isinstance(protein.sequence, str)
```

```python
# Example 3: Tokenizers — esm ships per-track tokenizers (sequence / structure / function / SASA).
from esm.tokenization import get_esm3_model_tokenizers
from esm.utils.constants.models import ESM3_OPEN_SMALL

tokenizers = get_esm3_model_tokenizers(ESM3_OPEN_SMALL)
print(f"tokenizers type: {type(tokenizers).__name__}")

# The TokenizerCollection groups per-track tokenizers; sequence tokenizer is always present.
seq_tok = tokenizers.sequence
print(f"sequence tokenizer: {type(seq_tok).__name__}")
print(f"vocab size: {seq_tok.vocab_size}")

ids = seq_tok.encode("MKTAYIAKQ")
print(f"encoded length: {len(ids)}")
assert len(ids) > 0
```

```python
# Example 4: Generation / sampling config dataclasses — construction + defaults.
from esm.sdk.api import GenerationConfig, SamplingConfig

gen = GenerationConfig(track="sequence", num_steps=8, temperature=1.0)
print(f"GenerationConfig: track={gen.track}, num_steps={gen.num_steps}, temperature={gen.temperature}")

samp = SamplingConfig()
print(f"SamplingConfig type: {type(samp).__name__}")

assert gen.track == "sequence"
assert gen.num_steps == 8
assert gen.temperature == 1.0
```

```python
# Example 5: Constants + utilities — amino-acid alphabet + residue conversion helpers.
from esm.utils.constants import esm3 as esm3_const
from esm.utils import residue_constants

# Sequence vocab contains the 20 canonical amino acids + special tokens.
print(f"SEQUENCE_VOCAB length: {len(esm3_const.SEQUENCE_VOCAB)}")
print(f"SEQUENCE_VOCAB[:10]: {esm3_const.SEQUENCE_VOCAB[:10]}")

# residue_constants holds the standard 20 AAs (single-letter codes).
print(f"restypes: {residue_constants.restypes}")
print(f"num restypes: {len(residue_constants.restypes)}")

assert len(residue_constants.restypes) == 20
assert 'A' in residue_constants.restypes and 'M' in residue_constants.restypes
```
