```python
# Example 1: String Type Annotation Processing

from eval_type_backport import eval_type_backport
from typing import Optional, Union, List, Dict, Any

print("=== Example 1: String Type Annotation Processing ===")

try:
    # Test various string type annotations
    type_strings = [
        "int",
        "str", 
        "Optional[int]",
        "Union[str, int]",
        "List[str]",
        "Dict[str, int]",
        "List[Optional[str]]"
    ]
    
    print("✓ Processing string type annotations:")
    for type_str in type_strings:
        try:
            result = eval_type_backport(type_str, globals(), locals())
            print(f"  '{type_str}' -> {result}")
        except Exception as e:
            print(f"  '{type_str}' -> Error: {e}")
            raise
    
    # Test complex nested types
    complex_type = "Dict[str, List[Optional[Union[int, str]]]]"
    result = eval_type_backport(complex_type, globals(), locals())
    print(f"\n✓ Complex type: {complex_type}")
    print(f"  Result: {result}")
    
    print("✓ String type annotation processing completed")

except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"eval_type_backport example (cell 0) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 completed successfully\n")
```

```python
# Example 2: Forward Reference Handling

from eval_type_backport import eval_type_backport, ForwardRef
from typing import Optional

print("=== Example 2: Forward Reference Handling ===")

try:
    # Create forward references for undefined classes
    forward_refs = [
        "'MyClass'",
        "'User'", 
        "'DataModel'",
        "Optional['Person']",
        "'Node[T]'"
    ]
    
    print("✓ Processing forward references:")
    for ref_str in forward_refs:
        try:
            result = eval_type_backport(ref_str, globals(), locals())
            print(f"  {ref_str} -> {result} (type: {type(result).__name__})")
        except Exception as e:
            print(f"  {ref_str} -> Error: {e}")
            raise
    
    # Create ForwardRef objects directly
    print(f"\n✓ Creating ForwardRef objects:")
    forward_ref = ForwardRef("MyClass")
    print(f"  ForwardRef('MyClass') -> {forward_ref}")
    print(f"  Type: {type(forward_ref)}")
    
    # Test with try_default parameter
    print(f"\n✓ try_default parameter demonstration:")
    try:
        result_with_default = eval_type_backport("UndefinedClass", try_default=True)
        print(f"  With try_default=True: {result_with_default}")
    except Exception as e:
        print(f"  With try_default=True: Error - {e}")
        raise
        
    try:
        result_without_default = eval_type_backport("UndefinedClass", try_default=False)
        print(f"  With try_default=False: {result_without_default}")
    except Exception as e:
        print(f"  With try_default=False: Error - {e}")
        raise

except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"eval_type_backport example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed successfully\n")
```

```python
# Example 3: Generic Types and Custom Namespace

from eval_type_backport import eval_type_backport
from typing import Generic, TypeVar, Callable, Tuple

print("=== Example 3: Generic Types and Custom Namespace ===")

try:
    # Define some custom types for namespace testing
    T = TypeVar('T')
    CustomType = str
    
    # Create custom namespace
    custom_globals = {
        'T': T,
        'CustomType': CustomType,
        'List': list,
        'Dict': dict
    }
    
    # Test generic and custom types
    type_expressions = [
        "Callable[[int, str], bool]",
        "Tuple[int, str, bool]", 
        "CustomType",
        "List[CustomType]",
        "Callable[[T], T]"
    ]
    
    print("✓ Processing generic types with custom namespace:")
    for expr in type_expressions:
        try:
            result = eval_type_backport(expr, custom_globals, locals())
            print(f"  '{expr}' -> {result}")
        except Exception as e:
            print(f"  '{expr}' -> Error: {e}")
            raise
    
    # Compare with and without custom namespace
    test_expr = "CustomType"
    result_custom = eval_type_backport(test_expr, custom_globals)
    try:
        result_default = eval_type_backport(test_expr, globals())
    except:
        result_default = "NameError (expected)"
    
    print(f"\n✓ Namespace comparison for '{test_expr}':")
    print(f"  Custom namespace: {result_custom}")
    print(f"  Default namespace: {result_default}")

except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"eval_type_backport example (cell 2) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 completed successfully\n")
```

```python
# Example 4: Practical Type Validation and Error Handling

from eval_type_backport import eval_type_backport
from typing import Union, Optional, Any
import sys

print("=== Example 4: Practical Type Validation ===")

try:
    # Function to safely evaluate type strings
    def safe_eval_type(type_str, context="unknown"):
        try:
            result = eval_type_backport(type_str, globals(), locals())
            return f"✓ {result}"
        except Exception as e:
            return f"✗ {type(e).__name__}: {e}"
    
    # Test various scenarios including edge cases
    test_cases = [
        ("int", "basic type"),
        ("Optional[str]", "optional type"),
        ("Union[int, str, bool]", "union type"),
        ("'NonExistentClass'", "forward reference"),
        ("InvalidSyntax[", "syntax error"),
        ("List[", "incomplete type"),
        ("42", "non-type value")
    ]
    
    print("✓ Type validation test results:")
    for type_expr, description in test_cases:
        result = safe_eval_type(type_expr, description)
        print(f"  {description:15} | '{type_expr}' -> {result}")
    
    # Test Python version compatibility
    print(f"\n✓ Python version: {sys.version_info}")
    print(f"✓ eval_type_backport helps with type annotation compatibility")
    
    # Performance comparison (basic)
    import time
    test_type = "Optional[Union[str, int]]"
    
    start = time.time()
    for _ in range(100):
        eval_type_backport(test_type)
    duration = time.time() - start
    
    print(f"\n✓ Performance: 100 evaluations in {duration:.4f}s")

except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"eval_type_backport example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed successfully\n")
```
