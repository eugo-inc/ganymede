## Example 1: executing functionality test 1

```python
# Example 1: Basic Source Code Analysis

import executing
import sys

print("=== Basic Source Code Analysis ===")

try:
    def analyze_current_execution():
        """Analyze what's currently executing."""
        frame = sys._getframe()
        source = executing.Source.executing(frame)
        
        print(f"✓ Current function: {frame.f_code.co_name}")
        print(f"✓ Current line: {frame.f_lineno}")
        print(f"✓ Source available: {source is not None}")
        
        if source and hasattr(source, 'node') and source.node:
            print(f"✓ AST node type: {type(source.node).__name__}")
            print(f"✓ Node line: {getattr(source.node, 'lineno', 'N/A')}")
        
        return source
    
    # Test source analysis
    result = analyze_current_execution()
    print(f"✓ Analysis completed: {result is not None}")

except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"executing example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 completed\n")
```

```python
# Example 2: Call Stack Analysis

import executing
import sys

print("=== Call Stack Analysis ===")

try:
    def trace_execution():
        """Trace current execution context."""
        frame = sys._getframe(1)  # Skip this method
        source = executing.Source.executing(frame)
        
        return {
            'function': frame.f_code.co_name,
            'line': frame.f_lineno,
            'source_available': source is not None,
            'locals_count': len(frame.f_locals)
        }
    
    def sample_function_a():
        context = trace_execution()
        print(f"✓ Function A: {context['function']} at line {context['line']}")
        return sample_function_b()
    
    def sample_function_b():
        context = trace_execution()
        print(f"✓ Function B: {context['function']} at line {context['line']}")
        return "completed"
    
    # Execute traced functions
    result = sample_function_a()
    print(f"✓ Result: {result}")

except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"executing example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed\n")
```

## Example 4: executing functionality test 4

```python
# Example 4: Advanced Frame Analysis

import executing
import sys

print("=== Advanced Frame Analysis ===")

try:
    def create_debug_snapshot():
        """Create debug snapshot of call stack."""
        frame = sys._getframe(1)
        source = executing.Source.executing(frame)
        
        return {
            'function': frame.f_code.co_name,
            'filename': frame.f_code.co_filename.split('/')[-1],
            'line': frame.f_lineno,
            'locals': len(frame.f_locals),
            'source_available': source is not None,
            'node_type': type(source.node).__name__ if source and hasattr(source, 'node') and source.node else None
        }
    
    def complex_function(param):
        snapshot = create_debug_snapshot()
        print(f"✓ Function snapshot: {snapshot['function']} at line {snapshot['line']}")
        print(f"✓ Source available: {snapshot['source_available']}")
        print(f"✓ Node type: {snapshot['node_type']}")
        return param * 2
    
    # Test advanced analysis
    result = complex_function(21)
    print(f"✓ Result: {result}")

except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"executing example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed\n")
```

## Example 5: executing functionality test 5

```python
# Example 5: Performance Profiling

import executing
import sys
import time

print("=== Performance Profiling ===")

try:
    class SimpleProfiler:
        def __init__(self):
            self.call_data = {}
        
        def profile_call(self, func_name):
            frame = sys._getframe(1)
            source = executing.Source.executing(frame)
            
            self.call_data[func_name] = {
                'line': frame.f_lineno,
                'source_available': source is not None,
                'node_type': type(source.node).__name__ if source and hasattr(source, 'node') and source.node else None
            }
    
    profiler = SimpleProfiler()
    
    def test_function(x):
        profiler.profile_call('test_function')
        time.sleep(0.001)  # Simulate work
        return x * 2
    
    # Test profiling
    for i in range(3):
        result = test_function(i)
        print(f"✓ Call {i}: result={result}")
    
    print(f"✓ Profiler data: {len(profiler.call_data)} functions tracked")
    for func, data in profiler.call_data.items():
        print(f"  {func}: line {data['line']}, source: {data['source_available']}")

except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"executing example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 completed\n")
```
