## Example 1: expandvars functionality test 1

```python
import expandvars
import os

# Example 1: Basic environment variable expansion
print("Example 1: Basic environment variable expansion")

try:
    # Set up test environment variables
    os.environ['USER_NAME'] = 'Alice'
    os.environ['PROJECT_DIR'] = '/home/alice/projects'
    os.environ['LOG_LEVEL'] = 'INFO'
    
    # Test basic variable expansion
    template = "Hello $USER_NAME! Your project is in $PROJECT_DIR"
    result = expandvars.expand(template)
    print(f"Template: {template}")
    print(f"Expanded: {result}")
    
    # Test with braces
    template_braces = "Config: ${PROJECT_DIR}/config.yml (level=${LOG_LEVEL})"
    result_braces = expandvars.expand(template_braces)
    print(f"Template with braces: {template_braces}")
    print(f"Expanded: {result_braces}")
    
    print("expandvars example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"expandvars example (cell 2) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: expandvars functionality test 2

```python
import expandvars
import os

# Example 2: Advanced variable expansion with defaults and missing variables
print("Example 2: Advanced variable expansion with defaults and missing variables")

try:
    # Set some variables and leave others undefined
    os.environ['DATABASE_HOST'] = 'localhost'
    if 'DATABASE_PORT' in os.environ:
        del os.environ['DATABASE_PORT']  # Ensure it's not set
    
    # Test expansion with missing variables
    template_missing = "Connection: $DATABASE_HOST:$DATABASE_PORT"
    
    print("Testing with missing variable:")
    print(f"Template: {template_missing}")
    try:
        result = expandvars.expand(template_missing)
        print(f"Result: {result}")
    except Exception as e:
        print(f"Error (expected): {e}")
        raise
    
    # Test getenv function for safer access
    print("\nUsing getenv for safer variable access:")
    # `expandvars.getenv` is an internal helper; use the shell `:-` default form through `expand`.
    host = expandvars.expand('${DATABASE_HOST:-default-host}')
    port = expandvars.expand('${DATABASE_PORT:-5432}')
    print(f"Host: {host} (from env)")
    print(f"Port: {port} (default value)")
    
    # Test expandvars alias function
    template_safe = "Safe connection: $DATABASE_HOST:${DATABASE_PORT:-5432}"
    if hasattr(expandvars, 'expandvars'):
        result_alias = expandvars.expandvars(template_safe)
        print(f"Using expandvars function: {result_alias}")
    
    print("expandvars example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"expandvars example (cell 4) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: expandvars functionality test 3

```python
import expandvars
import os

# Example 3: Complex path and configuration expansion
print("Example 3: Complex path and configuration expansion")

try:
    # Set up complex environment scenario
    os.environ['HOME'] = '/home/user'
    os.environ['APP_NAME'] = 'myapp'
    os.environ['VERSION'] = '1.2.3'
    os.environ['ENV'] = 'production'
    
    # Test complex path expansion
    config_template = """
Application: $APP_NAME v$VERSION
Environment: $ENV
Config file: $HOME/.config/$APP_NAME/config.yml
Log file: $HOME/logs/${APP_NAME}_${VERSION}.log
Data dir: $HOME/data/$APP_NAME/$ENV
"""
    
    result = expandvars.expand(config_template.strip())
    print("Configuration template expanded:")
    print(result)
    
    # Test with multiple variables in single path
    complex_path = "${HOME}/projects/${APP_NAME}/releases/v${VERSION}/${ENV}/data"
    expanded_path = expandvars.expand(complex_path)
    print(f"\nComplex path: {complex_path}")
    print(f"Expanded: {expanded_path}")
    
    print("expandvars example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"expandvars example (cell 6) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```
