## Example 1: An icon as inline SVG

```python
import importlib.metadata

import faicons

# Example 1: icon_svg returns an htmltools tag holding the Font Awesome SVG (no network, no web font)
assert importlib.metadata.version("faicons") == "0.2.2"
svg = faicons.icon_svg("github")
html = str(svg)
assert html.startswith("<svg") and "<path" in html and 'class="fa"' in html, html[:200]
print("Example 1 OK:", html[:120])
```

## Example 2: Solid and regular styles

```python
import faicons

# Example 2: many icons come in more than one style; the SVG path differs
solid = str(faicons.icon_svg("star", style="solid"))
regular = str(faicons.icon_svg("star", style="regular"))
assert solid != regular and "<path" in regular
print("Example 2 OK")
```

## Example 3: Colour and size

```python
import faicons

# Example 3: fill, height and width become inline CSS on the <svg>
html = str(faicons.icon_svg("circle-check", fill="green", height="2em", width="2em"))
assert "fill:green" in html and "height:2em" in html and "width:2em" in html, html[:300]
print("Example 3 OK")
```

## Example 4: An accessible title

```python
import faicons

# Example 4: a title makes the icon announce itself to screen readers instead of being hidden
html = str(faicons.icon_svg("bell", title="Notifications"))
assert "<title>Notifications</title>" in html and 'role="img"' in html, html[:300]
print("Example 4 OK")
```

## Example 5: An unknown icon is refused

```python
import faicons

# Example 5: a name that Font Awesome Free 6 does not have raises a ValueError naming it
try:
    faicons.icon_svg("no-such-icon-xyz")
except ValueError as exc:
    assert "no-such-icon-xyz" in str(exc), exc
else:
    raise AssertionError("unknown icon accepted")
print("Example 5 OK")
```
