## Example 1: faker functionality test 1

```python
from faker import Faker

# Example 1: Basic Fake Data Generation
print("Example 1: Basic Fake Data Generation")

try:
    # Create Faker instance
    fake = Faker()
    print("✓ Faker instance created")
    
    # Generate basic personal information
    print("✓ Personal information:")
    print(f"   Name: {fake.name()}")
    print(f"   Email: {fake.email()}")
    print(f"   Phone: {fake.phone_number()}")
    print(f"   Address: {fake.address().replace(chr(10), ', ')}")
    
    # Generate various data types
    print("✓ Various data types:")
    print(f"   Company: {fake.company()}")
    print(f"   Job Title: {fake.job()}")
    print(f"   Website: {fake.url()}")
    print(f"   Credit Card: {fake.credit_card_number()}")
    
    # Test reproducibility with seed
    fake.seed_instance(42)
    name1 = fake.name()
    fake.seed_instance(42)
    name2 = fake.name()
    print(f"✓ Reproducibility test: {name1 == name2}")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"faker example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 2: faker functionality test 2

```python
from faker import Faker

# Example 2: Localized Data Generation and Multiple Providers
print("Example 2: Localized Data Generation and Multiple Providers")

try:
    # Test different locales
    locales = ['en_US', 'fr_FR', 'de_DE', 'ja_JP']
    
    print("✓ Testing different locales:")
    for locale in locales:
        try:
            fake_locale = Faker(locale)
            print(f"   {locale}: {fake_locale.name()}")
        except Exception as e:
            print(f"   {locale}: Error - {str(e)[:30]}")
            raise
    
    # Multi-locale faker
    print("✓ Multi-locale faker:")
    multi_fake = Faker(['en_US', 'fr_FR', 'de_DE'])
    for i in range(3):
        print(f"   Name {i+1}: {multi_fake.name()}")
    
    # Generate comprehensive profiles
    print("✓ Profile generation:")
    fake_profile = Faker('en_US')
    profile = fake_profile.profile()
    print(f"   Name: {profile['name']}")
    print(f"   Email: {profile['mail']}")
    print(f"   Job: {profile['job']}")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"faker example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: faker functionality test 3

```python
from faker import Faker
import json

# Example 3: Custom Data Generation and Structured Data Creation
print("Example 3: Custom Data Generation and Structured Data Creation")

try:
    fake = Faker('en_US')
    fake.seed_instance(123)
    
    # Generate customer dataset
    customers = []
    for i in range(3):
        customer = {
            'customer_id': f"CUST_{i+1:04d}",
            'name': fake.name(),
            'email': fake.email(),
            'balance': round(fake.pyfloat(left_digits=4, right_digits=2, positive=True), 2),
            'is_premium': fake.boolean(chance_of_getting_true=30)
        }
        customers.append(customer)
    
    print("✓ Generated customers:")
    for customer in customers:
        print(f"   {customer['name']}: ${customer['balance']}")
    
    # Export to JSON
    json_output = json.dumps(customers, indent=2)
    print(f"✓ JSON export: {len(json_output)} characters")
    
    print("✓ Dataset complete")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"faker example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: faker functionality test 4

```python
from faker import Faker
from faker.providers import BaseProvider

# Example 4: Custom Providers and Advanced Faker Features
print("Example 4: Custom Providers and Advanced Faker Features")

try:
    # Create custom provider for gaming data
    class GamingProvider(BaseProvider):
        game_genres = ['Action', 'Adventure', 'RPG', 'Strategy', 'Sports']
        
        def game_title(self):
            adjectives = ['Epic', 'Legendary', 'Shadow', 'Crystal']
            nouns = ['Quest', 'Warriors', 'Empire', 'Adventures']
            return f"{self.random_element(adjectives)} {self.random_element(nouns)}"
        
        def game_genre(self):
            return self.random_element(self.game_genres)
    
    # Test custom provider
    fake = Faker()
    fake.add_provider(GamingProvider)
    
    print("✓ Custom gaming provider:")
    for i in range(3):
        print(f"   Game {i+1}: {fake.game_title()} ({fake.game_genre()})")
    
    # Advanced features - weighted choices and patterns
    print("✓ Advanced features:")
    print(f"   Product Code: {fake.bothify(text='PRD-###-??')}")
    print(f"   Version: {fake.numerify(text='#.#.##')}")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"faker example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 5: faker functionality test 5

```python
from faker import Faker
import time

# Example 5: Performance Testing and Production Usage Patterns
print("Example 5: Performance Testing and Production Usage Patterns")

try:
    # Performance test
    fake = Faker()
    fake.seed_instance(42)
    
    start_time = time.time()
    results = []
    for i in range(1000):
        data = {
            'name': fake.name(),
            'email': fake.email(),
            'company': fake.company()
        }
        results.append(data)
    
    duration = time.time() - start_time
    ops_per_sec = 1000 / duration
    
    print(f"✓ Performance test:")
    print(f"   Generated 1000 records in {duration:.3f} seconds")
    print(f"   Operations per second: {ops_per_sec:.1f}")
    
    # Production-like service
    class DataService:
        def __init__(self):
            self.fake = Faker()
        
        def generate_users(self, count=5):
            return [{'id': self.fake.uuid4(), 'name': self.fake.name()} for _ in range(count)]
    
    service = DataService()
    users = service.generate_users()
    print(f"✓ Generated {len(users)} users via service")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"faker example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
```
