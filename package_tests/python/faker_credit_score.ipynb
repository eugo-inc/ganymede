## Example 2: faker_credit_score functionality test 2

```python
# Example 2: Credit Profile Generation

from faker import Faker
from faker_credit_score import CreditScore

print("=== Credit Profile Generation ===")

try:
    fake = Faker()
    fake.add_provider(CreditScore)
    
    # Generate comprehensive credit profiles
    for i in range(3):
        profile = {
            'customer_id': fake.uuid4()[:8],
            'name': fake.name(),
            'credit_scores': {
                'fico_score': fake.credit_score(),
                'vantage3_score': fake.random_int(501, 990),
                'score_range': fake.credit_score_name()
            },
            'financial_info': {
                'annual_income': fake.random_int(25000, 150000),
                'employment': fake.random_element(['Full-time', 'Part-time', 'Self-employed']),
                'debt_to_income': round(fake.random_int(10, 45) / 100, 2)
            }
        }
        
        print(f"\nProfile {i+1}: {profile['customer_id']}")
        print(f"  Name: {profile['name']}")
        print(f"  FICO Score: {profile['credit_scores']['fico_score']}")
        print(f"  Income: ${profile['financial_info']['annual_income']:,}")
        print(f"  DTI: {profile['financial_info']['debt_to_income']}")
    
    print("\n✓ Credit profile generation completed")

except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_credit_score example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed\n")
```

## Example 3: faker_credit_score functionality test 3

```python
# Example 3: Credit Application Simulation

from faker import Faker
from faker_credit_score import CreditScore

print("=== Credit Application Simulation ===")

try:
    fake = Faker()
    fake.add_provider(CreditScore)
    
    # Simulate credit applications
    applications = []
    decision_summary = {'Approved': 0, 'Denied': 0}
    
    for i in range(5):
        app = {
            'app_id': fake.uuid4()[:8],
            'applicant': fake.name(),
            'credit_score': fake.credit_score(),
            'requested_amount': fake.random_int(5000, 50000),
            'annual_income': fake.random_int(30000, 120000)
        }
        
        # Simple approval logic
        if app['credit_score'] >= 650 and app['annual_income'] >= 40000:
            app['decision'] = 'Approved'
            app['interest_rate'] = round(fake.random_int(599, 1299) / 100, 2)
        else:
            app['decision'] = 'Denied'
            app['interest_rate'] = None
        
        decision_summary[app['decision']] += 1
        applications.append(app)
        
        print(f"App {i+1}: {app['applicant']} | Score: {app['credit_score']} | {app['decision']}")
    
    print(f"\nApproval rate: {decision_summary['Approved']}/{len(applications)}")

except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_credit_score example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 completed\n")
```

## Example 4: faker_credit_score functionality test 4

```python
# Example 4: Credit Score Analytics

from faker import Faker
from faker_credit_score import CreditScore

print("=== Credit Score Analytics ===")

try:
    fake = Faker()
    fake.add_provider(CreditScore)
    
    # Generate credit score distribution
    scores = [fake.credit_score() for _ in range(100)]
    
    # Analyze distribution
    score_ranges = {
        'Poor (300-579)': len([s for s in scores if s < 580]),
        'Fair (580-669)': len([s for s in scores if 580 <= s < 670]),
        'Good (670-739)': len([s for s in scores if 670 <= s < 740]),
        'Very Good (740-799)': len([s for s in scores if 740 <= s < 800]),
        'Excellent (800-850)': len([s for s in scores if s >= 800])
    }
    
    print("Credit Score Distribution (100 samples):")
    for range_name, count in score_ranges.items():
        percentage = (count / 100) * 100
        print(f"  {range_name}: {count} ({percentage}%)")
    
    # Statistical summary
    avg_score = sum(scores) / len(scores)
    min_score = min(scores)
    max_score = max(scores)
    
    print(f"\nStatistics:")
    print(f"  Average: {avg_score:.1f}")
    print(f"  Range: {min_score} - {max_score}")

except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_credit_score example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed\n")
```
