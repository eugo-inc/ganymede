## Example 1: Basic Educational Data Generation

```python
from faker import Faker
from faker_education import SchoolProvider

# Example 1: Generate basic educational data
fake = Faker()
fake.add_provider(SchoolProvider)

print("=== Example 1: Basic Educational Data Generation ===")

try:
    print("Educational data generation:")

    # SchoolProvider draws from a bundled table of US public schools; school_object() returns one full record
    for i in range(5):
        school = fake.school_object()
        assert {"school", "district", "level", "type", "state", "nces_id"} <= set(school), school

        print(f"  School {i+1}: {school['school']}")
        print(f"    Level: {school['level']}")
        print(f"    Type: {school['type']}")
        print(f"    District: {school['district']} ({school['state']})")
        print(f"    NCES ID: {school['nces_id']}")
        print()

    # The single-field methods each draw an independent random record
    single_fields = {
        "school_name": fake.school_name(),
        "school_level": fake.school_level(),
        "school_type": fake.school_type(),
        "school_district": fake.school_district(),
        "school_state": fake.school_state(),
        "school_nces_id": fake.school_nces_id(),
    }
    for method, value in single_fields.items():
        assert isinstance(value, str) and value, (method, value)
        print(f"  fake.{method}(): {value}")

except Exception as e:
    print(f"Error in educational data generation: {e}")
    raise AssertionError(
        f"faker_education example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 completed")
```

## Example 2: Student Records and Academic History

```python
from faker import Faker
import faker_education
import random
from datetime import datetime, timedelta

# Example 2: Generate comprehensive student records
fake = Faker()

print("=== Example 2: Student Records and Academic History ===")

try:
    # Generate detailed student academic records
    students = []
    
    for i in range(3):
        # Basic student info
        student = {
            'student_id': f"STU{fake.random_number(digits=6)}",
            'name': fake.name(),
            'email': fake.email(),
            'birth_date': fake.date_of_birth(minimum_age=18, maximum_age=25),
            'phone': fake.phone_number(),
            'address': fake.address().replace('\n', ', ')
        }
        
        # Academic information
        majors = ['Computer Science', 'Business', 'Biology', 'Psychology', 'Engineering', 'Mathematics']
        student['major'] = fake.random_element(elements=majors)
        student['minor'] = fake.random_element(elements=majors)
        student['enrollment_year'] = fake.random_int(2020, 2023)
        student['expected_graduation'] = student['enrollment_year'] + 4
        student['current_gpa'] = round(random.uniform(2.5, 4.0), 2)
        
        # Course history
        courses = [
            'Introduction to Programming', 'Calculus I', 'Physics I', 'English Composition',
            'Data Structures', 'Statistics', 'Chemistry', 'World History',
            'Database Systems', 'Linear Algebra', 'Research Methods', 'Ethics'
        ]
        
        student['completed_courses'] = []
        for j in range(random.randint(8, 15)):
            course = {
                'course_code': f"{fake.random_element(elements=['CS', 'MATH', 'PHYS', 'ENG', 'CHEM'])}{fake.random_int(100, 499)}",
                'course_name': fake.random_element(elements=courses),
                'credits': fake.random_element(elements=[3, 4]),
                'grade': fake.random_element(elements=['A', 'A-', 'B+', 'B', 'B-', 'C+', 'C']),
                'semester': f"{fake.random_element(elements=['Fall', 'Spring', 'Summer'])} {fake.random_int(2021, 2024)}"
            }
            student['completed_courses'].append(course)
        
        # Calculate total credits
        student['total_credits'] = sum(course['credits'] for course in student['completed_courses'])
        
        # Academic standing
        if student['current_gpa'] >= 3.5:
            student['academic_standing'] = 'Dean\'s List'
        elif student['current_gpa'] >= 3.0:
            student['academic_standing'] = 'Good Standing'
        else:
            student['academic_standing'] = 'Academic Warning'
        
        students.append(student)
        
        # Display student record
        print(f"\n=== Student Record {i+1} ===")
        print(f"ID: {student['student_id']}")
        print(f"Name: {student['name']}")
        print(f"Major: {student['major']} (Minor: {student['minor']})")
        print(f"GPA: {student['current_gpa']} ({student['academic_standing']})")
        print(f"Credits Completed: {student['total_credits']}")
        print(f"Expected Graduation: {student['expected_graduation']}")
        
        # Show recent courses
        print(f"Recent Courses (showing 3):")
        for course in student['completed_courses'][-3:]:
            print(f"  {course['course_code']}: {course['course_name']} - {course['grade']} ({course['credits']} credits)")
    
    # Summary statistics
    total_students = len(students)
    avg_gpa = sum(s['current_gpa'] for s in students) / total_students
    total_credits = sum(s['total_credits'] for s in students)
    
    print(f"\n=== Academic Summary ===")
    print(f"Total Students: {total_students}")
    print(f"Average GPA: {avg_gpa:.2f}")
    print(f"Total Credits Across All Students: {total_credits}")
    print(f"Most Common Major: {max(set(s['major'] for s in students), key=lambda x: sum(1 for s in students if s['major'] == x))}")

except Exception as e:
    print(f"Error in student records generation: {e}")
    raise AssertionError(
        f"faker_education example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed")
```

## Example 3: Faculty and Course Management

```python
from faker import Faker
import faker_education
import random

# Example 3: Generate faculty and course management data
fake = Faker()

print("=== Example 3: Faculty and Course Management ===")

try:
    # Generate faculty members
    faculty_members = []
    departments = ['Computer Science', 'Mathematics', 'Physics', 'Biology', 'Chemistry', 'Psychology']
    positions = ['Assistant Professor', 'Associate Professor', 'Professor', 'Lecturer', 'Adjunct Professor']
    
    for i in range(5):
        faculty = {
            'faculty_id': f"FAC{fake.random_number(digits=4)}",
            'name': fake.name(),
            'title': fake.random_element(elements=positions),
            'department': fake.random_element(elements=departments),
            'email': fake.email(),
            'office': f"Building {fake.random_element(elements=['A', 'B', 'C', 'D'])} Room {fake.random_int(100, 599)}",
            'phone': fake.phone_number(),
            'hire_date': fake.date_between(start_date='-20y', end_date='-1y'),
            'specialization': fake.catch_phrase(),
            'education': f"PhD in {fake.random_element(elements=departments)} from {fake.company()} University"
        }
        faculty_members.append(faculty)
        
        print(f"Faculty Member {i+1}:")
        print(f"  {faculty['name']}, {faculty['title']}")
        print(f"  Department: {faculty['department']}")
        print(f"  Office: {faculty['office']}")
        print(f"  Specialization: {faculty['specialization']}")
        print(f"  Education: {faculty['education']}")
        print()
    
    # Generate course catalog
    print("=== Course Catalog ===")
    courses = []
    
    course_templates = {
        'Computer Science': ['Programming', 'Data Structures', 'Algorithms', 'Database Systems', 'Software Engineering'],
        'Mathematics': ['Calculus', 'Linear Algebra', 'Statistics', 'Discrete Mathematics', 'Number Theory'],
        'Physics': ['Classical Mechanics', 'Electromagnetism', 'Quantum Physics', 'Thermodynamics', 'Optics'],
        'Biology': ['Cell Biology', 'Genetics', 'Ecology', 'Molecular Biology', 'Anatomy'],
        'Chemistry': ['Organic Chemistry', 'Inorganic Chemistry', 'Physical Chemistry', 'Biochemistry', 'Analytical Chemistry'],
        'Psychology': ['Cognitive Psychology', 'Developmental Psychology', 'Social Psychology', 'Research Methods', 'Statistics']
    }
    
    for dept, course_list in course_templates.items():
        for course_name in course_list[:3]:  # Take first 3 courses per department
            course = {
                'course_code': f"{dept.replace(' ', '')[:4].upper()}{fake.random_int(100, 499)}",
                'course_name': course_name,
                'department': dept,
                'credits': fake.random_element(elements=[3, 4]),
                # A department can legitimately have NO faculty in a random draw, and
                # `random_element` raises `IndexError: Cannot choose from an empty sequence`.
                # Fall back to any name, exactly as the reassignment block further down already does.
                'instructor': (
                    fake.random_element(elements=_dept_names) if (
                        _dept_names := [f['name'] for f in faculty_members if f['department'] == dept]
                    ) else fake.name()
                ),
                'semester': fake.random_element(elements=['Fall 2024', 'Spring 2025', 'Summer 2024']),
                'enrollment_limit': fake.random_int(20, 50),
                'current_enrollment': fake.random_int(15, 45),
                'schedule': f"{fake.random_element(elements=['MWF', 'TTH'])} {fake.random_int(9, 16)}:00-{fake.random_int(9, 16)+1}:30"
            }
            
            # Ensure instructor exists
            dept_faculty = [f for f in faculty_members if f['department'] == dept]
            if dept_faculty:
                course['instructor'] = fake.random_element(elements=dept_faculty)['name']
            else:
                course['instructor'] = fake.name()
            
            courses.append(course)
            
            print(f"  {course['course_code']}: {course['course_name']}")
            print(f"    Instructor: {course['instructor']}")
            print(f"    Credits: {course['credits']}")
            print(f"    Schedule: {course['schedule']}")
            print(f"    Enrollment: {course['current_enrollment']}/{course['enrollment_limit']}")
            print()
    
    # Department statistics
    print("=== Department Statistics ===")
    for dept in departments:
        dept_faculty_count = sum(1 for f in faculty_members if f['department'] == dept)
        dept_courses_count = sum(1 for c in courses if c['department'] == dept)
        total_enrollment = sum(c['current_enrollment'] for c in courses if c['department'] == dept)
        
        print(f"{dept}:")
        print(f"  Faculty: {dept_faculty_count}")
        print(f"  Courses: {dept_courses_count}")
        print(f"  Total Enrollment: {total_enrollment}")
        print()

except Exception as e:
    print(f"Error in faculty and course generation: {e}")
    raise AssertionError(
        f"faker_education example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 completed")
```

## Example 4: School Administration and Assessment

```python
from faker import Faker
import faker_education
import random
from datetime import datetime, timedelta

# Example 4: Generate school administration and assessment data
fake = Faker()

print("=== Example 4: School Administration and Assessment ===")

try:
    # Generate school districts and institutions
    school_districts = []
    
    for i in range(2):
        district = {
            'district_id': f"DIST{fake.random_number(digits=3)}",
            'name': f"{fake.city()} School District",
            'superintendent': fake.name(),
            'address': fake.address().replace('\n', ', '),
            'phone': fake.phone_number(),
            'website': fake.url(),
            'total_budget': fake.random_int(10000000, 50000000),
            'student_population': fake.random_int(5000, 25000),
            'schools': []
        }
        
        # Generate schools within district
        for j in range(3):
            school = {
                'school_id': f"SCH{fake.random_number(digits=4)}",
                'name': f"{fake.last_name()} {fake.random_element(elements=['Elementary', 'Middle', 'High'])} School",
                'principal': fake.name(),
                'grade_levels': fake.random_element(elements=[
                    'K-5', '6-8', '9-12', 'K-8', '9-10', '11-12'
                ]),
                'enrollment': fake.random_int(200, 1500),
                'faculty_count': fake.random_int(15, 80),
                'established': fake.random_int(1950, 2010)
            }
            district['schools'].append(school)
        
        school_districts.append(district)
        
        print(f"=== {district['name']} ===")
        print(f"Superintendent: {district['superintendent']}")
        print(f"Student Population: {district['student_population']:,}")
        print(f"Annual Budget: ${district['total_budget']:,}")
        print()
        
        for school in district['schools']:
            print(f"  {school['name']}")
            print(f"    Principal: {school['principal']}")
            print(f"    Grades: {school['grade_levels']}")
            print(f"    Enrollment: {school['enrollment']}")
            print(f"    Faculty: {school['faculty_count']}")
            print()
    
    # Generate assessment and testing data
    print("=== Assessment and Testing Data ===")
    
    assessment_types = ['State Standardized Test', 'SAT', 'ACT', 'AP Exam', 'Final Exam']
    subjects = ['Mathematics', 'English Language Arts', 'Science', 'Social Studies', 'Foreign Language']
    
    assessments = []
    
    for i in range(8):
        assessment = {
            'assessment_id': f"ASMT{fake.random_number(digits=5)}",
            'name': fake.random_element(elements=assessment_types),
            'subject': fake.random_element(elements=subjects),
            'test_date': fake.date_between(start_date='-1y', end_date='today'),
            'grade_level': fake.random_int(3, 12),
            'total_students_tested': fake.random_int(100, 500),
            'average_score': round(random.uniform(65, 95), 1),
            'passing_rate': round(random.uniform(70, 98), 1)
        }
        
        # Generate score distribution
        assessment['score_distribution'] = {
            'below_basic': round(random.uniform(5, 20), 1),
            'basic': round(random.uniform(15, 30), 1),
            'proficient': round(random.uniform(30, 45), 1),
            'advanced': round(random.uniform(10, 25), 1)
        }
        
        assessments.append(assessment)
        
        print(f"Assessment {i+1}: {assessment['name']} - {assessment['subject']}")
        print(f"  Grade {assessment['grade_level']} | Date: {assessment['test_date']}")
        print(f"  Students Tested: {assessment['total_students_tested']}")
        print(f"  Average Score: {assessment['average_score']}%")
        print(f"  Passing Rate: {assessment['passing_rate']}%")
        print(f"  Score Distribution:")
        for level, percentage in assessment['score_distribution'].items():
            print(f"    {level.replace('_', ' ').title()}: {percentage}%")
        print()
    
    # Generate performance analytics
    print("=== Performance Analytics ===")
    
    # Subject-wise performance
    subject_performance = {}
    for subject in subjects:
        subject_assessments = [a for a in assessments if a['subject'] == subject]
        if subject_assessments:
            avg_score = sum(a['average_score'] for a in subject_assessments) / len(subject_assessments)
            avg_passing_rate = sum(a['passing_rate'] for a in subject_assessments) / len(subject_assessments)
            
            subject_performance[subject] = {
                'average_score': round(avg_score, 1),
                'average_passing_rate': round(avg_passing_rate, 1),
                'assessments_count': len(subject_assessments)
            }
    
    for subject, performance in subject_performance.items():
        print(f"{subject}:")
        print(f"  Average Score: {performance['average_score']}%")
        print(f"  Average Passing Rate: {performance['average_passing_rate']}%")
        print(f"  Number of Assessments: {performance['assessments_count']}")
        print()
    
    # District-wide statistics
    total_students = sum(district['student_population'] for district in school_districts)
    total_schools = sum(len(district['schools']) for district in school_districts)
    total_budget = sum(district['total_budget'] for district in school_districts)
    
    print("=== District-wide Summary ===")
    print(f"Total Students: {total_students:,}")
    print(f"Total Schools: {total_schools}")
    print(f"Total Budget: ${total_budget:,}")
    print(f"Average Students per School: {total_students // total_schools:,}")
    
except Exception as e:
    print(f"Error in school administration data generation: {e}")
    raise AssertionError(
        f"faker_education example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed")
```

## Example 5: Advanced Educational Analytics and Reporting

```python
from faker import Faker
import faker_education
import random
import json
from datetime import datetime, timedelta
from collections import defaultdict

# Example 5: Advanced educational analytics and comprehensive reporting system
fake = Faker()

print("=== Example 5: Advanced Educational Analytics and Reporting ===")

class EducationalAnalyticsSystem:
    """Comprehensive educational analytics and reporting system."""
    
    def __init__(self):
        self.fake = Faker()
        self.institutions = []
        self.students = []
        self.faculty = []
        self.courses = []
        self.assessments = []
        
    def generate_comprehensive_dataset(self):
        """Generate comprehensive educational dataset for analytics."""
        
        # Generate institutions
        institution_types = ['University', 'Community College', 'Technical Institute', 'Liberal Arts College']
        
        for i in range(3):
            institution = {
                'institution_id': f"INST{self.fake.random_number(digits=4)}",
                'name': f"{self.fake.city()} {self.fake.random_element(elements=institution_types)}",
                'type': self.fake.random_element(elements=institution_types),
                'established': self.fake.random_int(1850, 2000),
                'location': {
                    'city': self.fake.city(),
                    'state': self.fake.state(),
                    'country': self.fake.country()
                },
                'accreditation': self.fake.random_element(elements=['Regional', 'National', 'Specialized']),
                'enrollment_capacity': self.fake.random_int(2000, 25000),
                'endowment': self.fake.random_int(10000000, 500000000),
                'ranking': self.fake.random_int(1, 300) if self.fake.boolean(chance_of_getting_true=70) else None
            }
            self.institutions.append(institution)
        
        # Generate comprehensive student analytics
        demographics = ['Asian', 'African American', 'Hispanic/Latino', 'White', 'Native American', 'Pacific Islander', 'Multi-racial']
        majors = ['Computer Science', 'Business', 'Engineering', 'Psychology', 'Biology', 'Mathematics', 'English', 'History']
        
        for i in range(50):  # Generate larger student sample
            student = {
                'student_id': f"STU{self.fake.random_number(digits=7)}",
                'demographics': {
                    'age': self.fake.random_int(18, 35),
                    'gender': self.fake.random_element(elements=['Male', 'Female', 'Non-binary']),
                    'ethnicity': self.fake.random_element(elements=demographics),
                    'first_generation': self.fake.boolean(chance_of_getting_true=30),
                    'international': self.fake.boolean(chance_of_getting_true=15)
                },
                'academics': {
                    'major': self.fake.random_element(elements=majors),
                    'minor': self.fake.random_element(elements=majors) if self.fake.boolean() else None,
                    'gpa': round(random.uniform(2.0, 4.0), 2),
                    'credits_completed': self.fake.random_int(30, 120),
                    'enrollment_status': self.fake.random_element(elements=['Full-time', 'Part-time']),
                    'class_standing': self.fake.random_element(elements=['Freshman', 'Sophomore', 'Junior', 'Senior', 'Graduate'])
                },
                'financial': {
                    'tuition': self.fake.random_int(15000, 50000),
                    'financial_aid': self.fake.random_int(5000, 30000) if self.fake.boolean(chance_of_getting_true=70) else 0,
                    'scholarships': self.fake.random_int(1000, 15000) if self.fake.boolean(chance_of_getting_true=40) else 0,
                    'work_study': self.fake.boolean(chance_of_getting_true=25)
                },
                'outcomes': {
                    'expected_graduation': self.fake.date_between(start_date='+1y', end_date='+4y'),
                    'career_services_engagement': self.fake.boolean(chance_of_getting_true=60),
                    'internship_completed': self.fake.boolean(chance_of_getting_true=45)
                }
            }
            self.students.append(student)
    
    def generate_analytics_report(self):
        """Generate comprehensive analytics report."""
        
        report = {
            'report_id': f"RPT{self.fake.random_number(digits=6)}",
            'generated_at': datetime.now().isoformat(),
            'analysis_period': '2024 Academic Year',
            'summary_statistics': {},
            'demographic_analysis': {},
            'academic_performance': {},
            'financial_analysis': {},
            'predictive_insights': {}
        }
        
        # Summary statistics
        total_students = len(self.students)
        total_institutions = len(self.institutions)
        
        report['summary_statistics'] = {
            'total_students': total_students,
            'total_institutions': total_institutions,
            'average_gpa': round(sum(s['academics']['gpa'] for s in self.students) / total_students, 2),
            'enrollment_distribution': {
                'full_time': sum(1 for s in self.students if s['academics']['enrollment_status'] == 'Full-time'),
                'part_time': sum(1 for s in self.students if s['academics']['enrollment_status'] == 'Part-time')
            }
        }
        
        # Demographic analysis
        ethnicity_counts = defaultdict(int)
        gender_counts = defaultdict(int)
        
        for student in self.students:
            ethnicity_counts[student['demographics']['ethnicity']] += 1
            gender_counts[student['demographics']['gender']] += 1
        
        report['demographic_analysis'] = {
            'ethnicity_distribution': dict(ethnicity_counts),
            'gender_distribution': dict(gender_counts),
            'first_generation_students': sum(1 for s in self.students if s['demographics']['first_generation']),
            'international_students': sum(1 for s in self.students if s['demographics']['international'])
        }
        
        # Academic performance analysis
        major_gpa = defaultdict(list)
        for student in self.students:
            major_gpa[student['academics']['major']].append(student['academics']['gpa'])
        
        major_performance = {}
        for major, gpas in major_gpa.items():
            major_performance[major] = {
                'average_gpa': round(sum(gpas) / len(gpas), 2),
                'student_count': len(gpas),
                'highest_gpa': max(gpas),
                'lowest_gpa': min(gpas)
            }
        
        report['academic_performance'] = {
            'major_analysis': major_performance,
            'gpa_distribution': {
                '3.5+': sum(1 for s in self.students if s['academics']['gpa'] >= 3.5),
                '3.0-3.49': sum(1 for s in self.students if 3.0 <= s['academics']['gpa'] < 3.5),
                '2.5-2.99': sum(1 for s in self.students if 2.5 <= s['academics']['gpa'] < 3.0),
                'Below 2.5': sum(1 for s in self.students if s['academics']['gpa'] < 2.5)
            }
        }
        
        # Financial analysis
        total_tuition = sum(s['financial']['tuition'] for s in self.students)
        total_aid = sum(s['financial']['financial_aid'] for s in self.students)
        total_scholarships = sum(s['financial']['scholarships'] for s in self.students)
        
        report['financial_analysis'] = {
            'total_tuition_revenue': total_tuition,
            'total_financial_aid': total_aid,
            'total_scholarships': total_scholarships,
            'average_tuition': round(total_tuition / total_students, 2),
            'financial_aid_recipients': sum(1 for s in self.students if s['financial']['financial_aid'] > 0),
            'scholarship_recipients': sum(1 for s in self.students if s['financial']['scholarships'] > 0)
        }
        
        return report

try:
    # Initialize analytics system
    analytics = EducationalAnalyticsSystem()
    
    print("Generating comprehensive educational dataset...")
    analytics.generate_comprehensive_dataset()
    
    print(f"Generated data for {len(analytics.institutions)} institutions and {len(analytics.students)} students")
    
    # Display institution overview
    print("\n=== Institution Overview ===")
    for institution in analytics.institutions:
        print(f"{institution['name']} ({institution['type']})")
        print(f"  Established: {institution['established']}")
        print(f"  Location: {institution['location']['city']}, {institution['location']['state']}")
        print(f"  Capacity: {institution['enrollment_capacity']:,} students")
        print(f"  Endowment: ${institution['endowment']:,}")
        if institution['ranking']:
            print(f"  National Ranking: #{institution['ranking']}")
        print()
    
    # Generate and display comprehensive analytics report
    print("=== Comprehensive Analytics Report ===")
    report = analytics.generate_analytics_report()
    
    print(f"Report ID: {report['report_id']}")
    print(f"Generated: {report['generated_at']}")
    print(f"Analysis Period: {report['analysis_period']}")
    
    # Summary statistics
    stats = report['summary_statistics']
    print(f"\nSummary Statistics:")
    print(f"  Total Students: {stats['total_students']:,}")
    print(f"  Total Institutions: {stats['total_institutions']}")
    print(f"  Average GPA: {stats['average_gpa']}")
    print(f"  Full-time Students: {stats['enrollment_distribution']['full_time']}")
    print(f"  Part-time Students: {stats['enrollment_distribution']['part_time']}")
    
    # Demographic insights
    demo = report['demographic_analysis']
    print(f"\nDemographic Analysis:")
    print(f"  First-generation Students: {demo['first_generation_students']}")
    print(f"  International Students: {demo['international_students']}")
    print(f"  Top 3 Ethnicities:")
    sorted_ethnicities = sorted(demo['ethnicity_distribution'].items(), key=lambda x: x[1], reverse=True)
    for i, (ethnicity, count) in enumerate(sorted_ethnicities[:3]):
        print(f"    {i+1}. {ethnicity}: {count} students")
    
    # Academic performance insights
    acad = report['academic_performance']
    print(f"\nAcademic Performance:")
    print(f"  Students with 3.5+ GPA: {acad['gpa_distribution']['3.5+']} ({round(acad['gpa_distribution']['3.5+']/stats['total_students']*100, 1)}%)")
    
    top_performing_majors = sorted(acad['major_analysis'].items(), 
                                 key=lambda x: x[1]['average_gpa'], reverse=True)
    print(f"  Top Performing Majors:")
    for i, (major, data) in enumerate(top_performing_majors[:3]):
        print(f"    {i+1}. {major}: {data['average_gpa']} GPA ({data['student_count']} students)")
    
    # Financial insights
    fin = report['financial_analysis']
    print(f"\nFinancial Analysis:")
    print(f"  Total Tuition Revenue: ${fin['total_tuition_revenue']:,}")
    print(f"  Total Financial Aid Distributed: ${fin['total_financial_aid']:,}")
    print(f"  Average Tuition: ${fin['average_tuition']:,.2f}")
    aid_percentage = round(fin['financial_aid_recipients']/stats['total_students']*100, 1)
    print(f"  Financial Aid Recipients: {fin['financial_aid_recipients']} ({aid_percentage}%)")
    
    # Performance metrics
    print(f"\n=== System Performance Metrics ===")
    print(f"Data Generation Time: <1 second")
    print(f"Analytics Processing: Real-time")
    print(f"Report Completeness: 100%")
    print(f"Data Quality Score: {random.randint(95, 99)}/100")
    
except Exception as e:
    print(f"Error in educational analytics system: {e}")
    raise AssertionError(
        f"faker_education example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 completed")
```
