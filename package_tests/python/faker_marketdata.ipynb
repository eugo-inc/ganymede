## Example 1: Basic Market Identifiers

```python
from faker import Faker
from faker_marketdata import MarketDataProvider

# Example 1: Basic Market Identifiers
fake = Faker()
fake.add_provider(MarketDataProvider)

print("=== Example 1: Basic Market Identifiers ===")

try:
    # Generate various market identifiers using faker_marketdata
    for i in range(5):
        ticker = fake.ticker()
        cusip = fake.cusip()
        isin = fake.isin()
        
        print(f"Security {i+1}:")
        print(f"  Ticker: {ticker}")
        print(f"  CUSIP: {cusip}")
        print(f"  ISIN: {isin}")
        print()
    
    print("✓ Market identifiers generation completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_marketdata example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 completed")
```

## Example 4: Trading Portfolio Simulation

```python
from faker import Faker
from faker_marketdata import MarketDataProvider

# Example 4: Trading Portfolio Simulation
fake = Faker()
fake.add_provider(MarketDataProvider)

print("=== Example 4: Trading Portfolio Simulation ===")

try:
    # Generate a sample trading portfolio
    print("Portfolio Holdings:")
    
    portfolio_value = 0
    for i in range(5):
        ticker = fake.ticker()
        isin = fake.isin()
        market_type = fake.marketType()
        shares = fake.random_int(100, 1000)
        price = fake.pyfloat(left_digits=2, right_digits=2, positive=True)
        value = shares * price
        portfolio_value += value
        
        print(f"{ticker} ({isin}) - {market_type}")
        print(f"  Shares: {shares:,}, Price: ${price:.2f}, Value: ${value:,.2f}")
        print()
    
    print(f"Total Portfolio Value: ${portfolio_value:,.2f}")
    print("✓ Portfolio simulation completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_marketdata example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed")
```
