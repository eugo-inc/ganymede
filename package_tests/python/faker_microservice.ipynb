## Example 1: Basic Microservice Names

```python
from faker import Faker
from faker_microservice import Provider

# Example 1: Basic Microservice Names
fake = Faker()
fake.add_provider(Provider)

print("=== Example 1: Basic Microservice Names ===")

try:
    # Generate microservice names using faker_microservice
    print("Generated Microservice Names:")
    for i in range(8):
        service_name = fake.microservice()
        print(f"  {i+1}. {service_name}")
    
    print("\n✓ Microservice names generation completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise

print("Example 1 completed")
```

## Example 2: Microservice Registry Simulation

```python
from faker import Faker
from faker_microservice import Provider

# Example 2: Microservice Registry Simulation
fake = Faker()
fake.add_provider(Provider)

print("=== Example 2: Microservice Registry ===")

try:
    # Simulate a microservice registry with fake services
    print("Service Registry:")
    
    for i in range(5):
        service_name = fake.microservice()
        host = fake.ipv4_private()
        port = fake.random_int(8000, 9999)
        status = fake.random_element(['running', 'stopped', 'healthy', 'unhealthy'])
        
        print(f"Service: {service_name}")
        print(f"  Host: {host}:{port}")
        print(f"  Status: {status}")
        print()
    
    print("✓ Service registry simulation completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_microservice example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed")
```

## Example 3: API Gateway Routes

```python
from faker import Faker
from faker_microservice import Provider

# Example 3: API Gateway Routes
fake = Faker()
fake.add_provider(Provider)

print("=== Example 3: API Gateway Routes ===")

try:
    # Generate API routes for microservices
    print("API Gateway Routes:")
    
    for i in range(6):
        service_name = fake.microservice()
        method = fake.random_element(['GET', 'POST', 'PUT', 'DELETE'])
        path = f"/api/v1/{service_name.replace('_', '-')}"
        
        print(f"{method} {path}")
        print(f"  -> {service_name} microservice")
        print()
    
    print("✓ API gateway routes generation completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_microservice example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 completed")
```

## Example 4: Microservice Health Dashboard

```python
from faker import Faker
from faker_microservice import Provider

# Example 4: Microservice Health Dashboard
fake = Faker()
fake.add_provider(Provider)

print("=== Example 4: Service Health Dashboard ===")

try:
    # Generate health status for multiple microservices
    print("Microservice Health Status:")
    
    healthy_count = 0
    total_services = 8
    
    for i in range(total_services):
        service_name = fake.microservice()
        cpu_usage = fake.random_int(10, 95)
        memory_usage = fake.random_int(20, 90)
        response_time = fake.random_int(50, 500)
        
        # Determine health status
        if cpu_usage < 80 and memory_usage < 85 and response_time < 200:
            status = "🟢 Healthy"
            healthy_count += 1
        else:
            status = "🔴 Unhealthy"
        
        print(f"{service_name}: {status}")
        print(f"  CPU: {cpu_usage}%, Memory: {memory_usage}%, Response: {response_time}ms")
    
    print(f"\nSystem Health: {healthy_count}/{total_services} services healthy")
    print("✓ Health dashboard generated")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_microservice example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed")
```

## Example 5: Distributed Tracing Simulation

```python
from faker import Faker
from faker_microservice import Provider

# Example 5: Distributed Tracing Simulation
fake = Faker()
fake.add_provider(Provider)

print("=== Example 5: Distributed Tracing ===")

try:
    # Simulate a distributed trace across microservices
    trace_id = fake.uuid4()[:8]
    print(f"Trace ID: {trace_id}")
    print("\nDistributed Trace:")
    
    # Generate a chain of service calls
    services_in_trace = [fake.microservice() for _ in range(4)]
    total_duration = 0
    
    for i, service_name in enumerate(services_in_trace):
        duration = fake.random_int(10, 150)
        total_duration += duration
        
        span_id = fake.uuid4()[:8]
        operation = fake.random_element(['process_request', 'query_database', 'call_external_api', 'validate_data'])
        
        print(f"  {i+1}. {service_name}")
        print(f"     Span: {span_id}, Operation: {operation}")
        print(f"     Duration: {duration}ms")
        print()
    
    print(f"Total Trace Duration: {total_duration}ms")
    print("✓ Distributed trace simulation completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_microservice example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 completed")
```
