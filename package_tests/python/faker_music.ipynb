## Example 1: Basic Music Data Generation

```python
from faker import Faker
from faker_music import MusicProvider

# Example 1: Basic Music Data Generation
fake = Faker()
fake.add_provider(MusicProvider)

print("=== Example 1: Basic Music Data Generation ===")

try:
    # Generate music-related data using faker_music
    print("Music Elements:")
    for i in range(6):
        genre = fake.music_genre()
        subgenre = fake.music_subgenre()
        instrument = fake.music_instrument()
        category = fake.music_instrument_category()
        
        print(f"{i+1}. Genre: {genre} | Subgenre: {subgenre}")
        print(f"   Instrument: {instrument} ({category})")
        print()
    
    print("✓ Music data generation completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_music example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 completed")
```

## Example 2: Band Formation Simulation

```python
from faker import Faker
from faker_music import MusicProvider

# Example 2: Band Formation Simulation
fake = Faker()
fake.add_provider(MusicProvider)

print("=== Example 2: Band Formation ===")

try:
    # Generate band information using faker_music
    band_name = fake.company().replace('Inc', 'Band').replace('LLC', '')
    genre = fake.music_genre()
    subgenre = fake.music_subgenre()
    
    print(f"Band: {band_name}")
    print(f"Genre: {genre} ({subgenre})")
    print(f"Formed: {fake.year()}")
    print()
    
    print("Band Members:")
    for i in range(4):
        member = fake.name()
        instrument = fake.music_instrument()
        category = fake.music_instrument_category()
        
        print(f"  {member} - {instrument} ({category})")
    
    print(f"\n✓ Band formation completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_music example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed")
```

## Example 3: Album and Song Generation

```python
from faker import Faker
from faker_music import MusicProvider

# Example 3: Album and Song Generation
fake = Faker()
fake.add_provider(MusicProvider)

print("=== Example 3: Album and Song Generation ===")

try:
    # Generate album with songs using faker_music
    album_title = fake.catch_phrase().replace(',', '').title()
    album_genre = fake.music_genre()
    album_subgenre = fake.music_subgenre()
    
    print(f"Album: '{album_title}'")
    print(f"Genre: {album_genre} - {album_subgenre}")
    print(f"Release Year: {fake.year()}")
    print()
    
    print("Track Listing:")
    for track in range(5):
        song_title = fake.catch_phrase().replace(',', '').title()
        duration = f"{fake.random_int(2, 6)}:{fake.random_int(10, 59):02d}"
        
        print(f"  {track+1}. {song_title} ({duration})")
    
    print(f"\n✓ Album generation completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_music example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 completed")
```

## Example 4: Music Festival Lineup

```python
from faker import Faker
from faker_music import MusicProvider

# Example 4: Music Festival Lineup
fake = Faker()
fake.add_provider(MusicProvider)

print("=== Example 4: Music Festival Lineup ===")

try:
    # Generate music festival with diverse genres
    festival_name = fake.catch_phrase().replace(',', '') + ' Music Festival'
    
    print(f"Festival: {festival_name}")
    print(f"Location: {fake.city()}, {fake.country()}")
    print(f"Date: {fake.date_this_year()}")
    print()
    
    print("Lineup by Genre:")
    for stage in range(3):
        genre = fake.music_genre()
        subgenre = fake.music_subgenre()
        
        print(f"  {genre} Stage ({subgenre}):")
        for act in range(3):
            artist = fake.name() if fake.boolean() else fake.company().replace('Inc', '').strip()
            print(f"    • {artist}")
    
    print(f"\n✓ Festival lineup completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_music example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed")
```

## Example 5: Music Store Inventory

```python
from faker import Faker
from faker_music import MusicProvider

# Example 5: Music Store Inventory
fake = Faker()
fake.add_provider(MusicProvider)

print("=== Example 5: Music Store Inventory ===")

try:
    # Generate music store inventory using faker_music
    store_name = fake.company().replace('Inc', 'Music').replace('LLC', 'Instruments')
    
    print(f"Store: {store_name}")
    print(f"Location: {fake.address()}")
    print()
    
    print("Instrument Inventory:")
    categories = set()
    
    for item in range(8):
        instrument = fake.music_instrument()
        category = fake.music_instrument_category()
        price = fake.random_int(50, 2000)
        stock = fake.random_int(1, 15)
        
        categories.add(category)
        print(f"  {instrument} (${price}) - {stock} in stock")
    
    print(f"\nCategories Available: {', '.join(sorted(categories))}")
    print(f"Total Items: 8")
    print(f"\n✓ Music store inventory completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_music example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 completed")
```
