## Example 1: Basic Vehicle Information

```python
from faker import Faker
from faker_vehicle import VehicleProvider

# Example 1: Basic Vehicle Information
fake = Faker()
fake.add_provider(VehicleProvider)

print("=== Example 1: Basic Vehicle Information ===")

try:
    # Generate basic vehicle data using faker_vehicle
    print("Vehicle Inventory:")
    for i in range(8):
        make = fake.vehicle_make()
        model = fake.vehicle_model()
        year = fake.vehicle_year()
        category = fake.vehicle_category()
        
        print(f"{i+1}. {year} {make} {model} ({category})")
    
    print("\n✓ Vehicle information generation completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_vehicle example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 completed")
```

## Example 2: Vehicle Dealership Inventory

```python
from faker import Faker
from faker_vehicle import VehicleProvider

# Example 2: Vehicle Dealership Inventory
fake = Faker()
fake.add_provider(VehicleProvider)

print("=== Example 2: Vehicle Dealership Inventory ===")

try:
    # Generate complete vehicle records using faker_vehicle
    print("Dealership Stock:")
    
    for i in range(5):
        vehicle = fake.vehicle_year_make_model()
        category = fake.vehicle_category()
        price = fake.random_int(15000, 65000)
        mileage = fake.random_int(0, 50000)
        
        print(f"Stock #{i+1}: {vehicle}")
        print(f"  Category: {category} | Price: ${price:,} | Miles: {mileage:,}")
        print()
    
    print("✓ Dealership inventory completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_vehicle example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed")
```

## Example 3: Fleet Management

```python
from faker import Faker
from faker_vehicle import VehicleProvider

# Example 3: Fleet Management
fake = Faker()
fake.add_provider(VehicleProvider)

print("=== Example 3: Fleet Management ===")

try:
    # Generate fleet vehicles using faker_vehicle
    print("Company Fleet:")
    
    for i in range(6):
        vehicle_info = fake.vehicle_year_make_model_cat()
        fleet_id = f"FLEET-{fake.random_int(100, 999)}"
        driver = fake.name()
        
        print(f"{fleet_id}: {vehicle_info}")
        print(f"  Assigned Driver: {driver}")
        print(f"  Mileage: {fake.random_int(25000, 150000):,}")
        print()
    
    print("✓ Fleet management data completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_vehicle example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 completed")
```

## Example 4: Insurance Claims

```python
from faker import Faker
from faker_vehicle import VehicleProvider

# Example 4: Insurance Claims
fake = Faker()
fake.add_provider(VehicleProvider)

print("=== Example 4: Insurance Claims ===")

try:
    # Generate insurance claims using faker_vehicle
    print("Recent Insurance Claims:")
    
    for i in range(4):
        claim_id = fake.bothify('CLM-####-???')
        vehicle = fake.vehicle_make_model()
        year = fake.vehicle_year()
        category = fake.vehicle_category()
        
        print(f"Claim {claim_id}:")
        print(f"  Vehicle: {year} {vehicle} ({category})")
        print(f"  Claim Date: {fake.date_this_year()}")
        print(f"  Estimated Damage: ${fake.random_int(500, 15000):,}")
        print()
    
    print("✓ Insurance claims data completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"faker_vehicle example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed")
```

## Example 5: Vehicle Registration Database
