## Example 1: Web URLs and Domain Generation

```python
from faker import Faker
import random

# Example 1: Web URLs and Domain Generation
fake = Faker()

try:
    # Add faker_web's provider -- the package under test, imported directly
    from faker_web import WebProvider
    fake.add_provider(WebProvider)
    print("faker_web provider loaded successfully")
    server_token = fake.server_token()
    content_type = fake.content_type_popular()
    print(f"Sample server token: {server_token}; popular content type: {content_type}")
    # measured, faker-web 0.3.1: tokens come from the Apache / nginx / Microsoft-IIS / Varnish families
    assert server_token.split('/')[0] in {'Apache', 'nginx', 'Microsoft-IIS', 'Varnish'}, server_token
    assert content_type.split('/')[0] in {'application', 'image', 'text'} and '/' in content_type, content_type
    
    # Generate web domains and URLs
    web_data = []
    tlds = ['.com', '.org', '.net', '.io', '.co', '.dev', '.tech', '.app']
    protocols = ['http', 'https']
    
    for i in range(6):
        domain_name = fake.word().lower()
        tld = random.choice(tlds)
        full_domain = domain_name + tld
        
        web_entry = {
            'domain': full_domain,
            'url': f"{random.choice(protocols)}://{full_domain}",
            'subdomain': f"{random.choice(['www', 'api', 'blog', 'shop'])}.{full_domain}",
            'email': f"{fake.user_name()}@{full_domain}",
            'ip_address': fake.ipv4(),
            'port': random.choice([80, 443, 8080, 3000]),
            'path': f"/{fake.word()}/{fake.word()}",
            'query_params': {
                'id': fake.random_number(digits=6),
                'token': fake.uuid4()[:8]
            }
        }
        
        query_string = '&'.join([f"{k}={v}" for k, v in web_entry['query_params'].items()])
        web_entry['full_url'] = f"{web_entry['url']}{web_entry['path']}?{query_string}"
        web_data.append(web_entry)
    
    print("Generated Web URLs and Domains:")
    for entry in web_data:
        print(f"\n{entry['domain']} ({entry['ip_address']}:{entry['port']})")
        print(f"  Main: {entry['url']}")
        print(f"  API: https://{entry['subdomain']}")
        print(f"  Contact: {entry['email']}")
        print(f"  Resource: {entry['full_url'][:60]}...")
    
    https_count = len([w for w in web_data if w['url'].startswith('https')])
    print(f"\nSSL Adoption: {https_count}/{len(web_data)} sites use HTTPS")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"faker_web example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 completed")
```

## Example 2: HTTP Request/Response Data

```python
from faker import Faker
import random

# Example 2: HTTP Request/Response Data
fake = Faker()

try:
    http_requests = []
    methods = ['GET', 'POST', 'PUT', 'DELETE', 'PATCH']
    status_codes = [200, 201, 301, 400, 401, 403, 404, 500, 502]
    
    for i in range(8):
        request = {
            'method': random.choice(methods),
            'url': f"https://{fake.domain_name()}/{fake.word()}/{fake.word()}",
            'headers': {
                'User-Agent': fake.user_agent(),
                'Accept': 'application/json',
                'Content-Type': 'application/json',
                'Authorization': f"Bearer {fake.uuid4()}"
            },
            'status_code': random.choice(status_codes),
            'response_time_ms': random.randint(50, 2000),
            'content_length': random.randint(100, 50000),
            'ip_address': fake.ipv4(),
            'timestamp': fake.date_time_this_month()
        }
        http_requests.append(request)
    
    print("HTTP Request/Response Log:")
    for req in http_requests:
        status_color = "✓" if req['status_code'] < 400 else "✗"
        print(f"\n{status_color} {req['method']} {req['url'][:50]}...")
        print(f"  Status: {req['status_code']} | Time: {req['response_time_ms']}ms | Size: {req['content_length']} bytes")
        print(f"  IP: {req['ip_address']} | User-Agent: {req['headers']['User-Agent'][:40]}...")
    
    success_rate = len([r for r in http_requests if r['status_code'] < 400]) / len(http_requests)
    avg_response_time = sum(r['response_time_ms'] for r in http_requests) / len(http_requests)
    
    print(f"\nAPI Performance:")
    print(f"  Success Rate: {success_rate:.1%}")
    print(f"  Average Response Time: {avg_response_time:.0f}ms")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"faker_web example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed")
```

## Example 3: Web Analytics and User Sessions

```python
from faker import Faker
import random

# Example 3: Web Analytics and User Sessions  
fake = Faker()

try:
    sessions = []
    for i in range(10):
        session = {
            'session_id': fake.uuid4(),
            'user_id': fake.uuid4() if random.random() < 0.7 else None,
            'ip_address': fake.ipv4(),
            'user_agent': fake.user_agent(),
            'referrer': f"https://{fake.domain_name()}" if random.random() < 0.6 else None,
            'landing_page': f"/{fake.word()}",
            'page_views': random.randint(1, 15),
            'session_duration_seconds': random.randint(30, 3600),
            'bounce': random.choice([True, False]),
            'conversion': random.choice([True, False]) if random.random() < 0.3 else False,
            'device_type': random.choice(['desktop', 'mobile', 'tablet']),
            'browser': random.choice(['Chrome', 'Firefox', 'Safari', 'Edge']),
            'location': {'country': fake.country(), 'city': fake.city()}
        }
        sessions.append(session)
    
    print("Web Analytics Summary:")
    print(f"Total Sessions: {len(sessions)}")
    print(f"Unique Users: {len(set(s['user_id'] for s in sessions if s['user_id']))}")
    
    avg_duration = sum(s['session_duration_seconds'] for s in sessions) / len(sessions)
    bounce_rate = len([s for s in sessions if s['bounce']]) / len(sessions)
    conversion_rate = len([s for s in sessions if s['conversion']]) / len(sessions)
    
    print(f"Average Session Duration: {avg_duration/60:.1f} minutes")
    print(f"Bounce Rate: {bounce_rate:.1%}")
    print(f"Conversion Rate: {conversion_rate:.1%}")
    
    device_stats = {}
    for session in sessions:
        device = session['device_type']
        device_stats[device] = device_stats.get(device, 0) + 1
    
    print(f"\nDevice Distribution:")
    for device, count in device_stats.items():
        print(f"  {device.title()}: {count} ({count/len(sessions):.1%})")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"faker_web example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 completed")
```

## Example 4: Web API and JSON Data

```python
from faker import Faker
import random
import json

# Example 4: Web API and JSON Data
fake = Faker()

try:
    api_responses = []
    for i in range(5):
        response = {
            'endpoint': f"/api/v{random.randint(1,3)}/{fake.word()}",
            'method': random.choice(['GET', 'POST', 'PUT']),
            'status_code': random.choice([200, 201, 400, 404]),
            'data': {
                'id': fake.random_number(digits=6),
                'name': fake.name(),
                'email': fake.email(),
                'created_at': fake.iso8601(),
                'active': random.choice([True, False]),
                'metadata': {
                    'source': fake.domain_name(),
                    'tags': [fake.word() for _ in range(random.randint(1, 4))]
                }
            },
            'pagination': {
                'page': random.randint(1, 10),
                'total_pages': random.randint(5, 50),
                'per_page': random.choice([10, 25, 50, 100])
            } if random.random() < 0.6 else None
        }
        api_responses.append(response)
    
    print("API Response Examples:")
    for resp in api_responses:
        print(f"\n{resp['method']} {resp['endpoint']} - Status: {resp['status_code']}")
        print(f"  Data: {json.dumps(resp['data'], indent=2)[:100]}...")
        if resp['pagination']:
            print(f"  Pagination: Page {resp['pagination']['page']} of {resp['pagination']['total_pages']}")
    
    success_count = len([r for r in api_responses if r['status_code'] < 400])
    print(f"\nAPI Health: {success_count}/{len(api_responses)} successful responses")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"faker_web example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed")
```

## Example 5: Web Security and Performance Monitoring

```python
from faker import Faker
import random

# Example 5: Web Security and Performance Monitoring
fake = Faker()

try:
    security_events = []
    for i in range(8):
        event = {
            'timestamp': fake.date_time_this_month(),
            'event_type': random.choice(['failed_login', 'sql_injection', 'xss_attempt', 'ddos', 'brute_force']),
            'ip_address': fake.ipv4(),
            'user_agent': fake.user_agent(),
            'severity': random.choice(['low', 'medium', 'high', 'critical']),
            'blocked': random.choice([True, False]),
            'country': fake.country_code(),
            'request_details': {
                'url': f"https://{fake.domain_name()}/{fake.word()}",
                'method': random.choice(['GET', 'POST', 'PUT']),
                'payload_size': random.randint(100, 10000)
            }
        }
        security_events.append(event)
    
    # Performance metrics
    performance_data = {
        'load_time_ms': random.randint(200, 5000),
        'ttfb_ms': random.randint(50, 500),  
        'dom_ready_ms': random.randint(300, 2000),
        'page_size_kb': random.randint(500, 3000),
        'requests_count': random.randint(15, 80),
        'cache_hit_rate': round(random.uniform(0.6, 0.95), 2),
        'cdn_usage': random.choice([True, False]),
        'ssl_grade': random.choice(['A+', 'A', 'B', 'C'])
    }
    
    print("Web Security Events:")
    for event in security_events[-5:]:  # Show last 5
        blocked_status = "🛡️ BLOCKED" if event['blocked'] else "⚠️ ALLOWED"
        print(f"{event['event_type'].replace('_', ' ').title()} from {event['ip_address']} ({event['country']})")
        print(f"  {blocked_status} | Severity: {event['severity'].upper()}")
    
    critical_events = len([e for e in security_events if e['severity'] == 'critical'])
    blocked_events = len([e for e in security_events if e['blocked']])
    
    print(f"\nSecurity Summary:")
    print(f"  Total Events: {len(security_events)}")
    print(f"  Critical Events: {critical_events}")
    print(f"  Block Rate: {blocked_events}/{len(security_events)} ({blocked_events/len(security_events):.1%})")
    
    print(f"\nPerformance Metrics:")
    print(f"  Load Time: {performance_data['load_time_ms']}ms")
    print(f"  TTFB: {performance_data['ttfb_ms']}ms") 
    print(f"  Page Size: {performance_data['page_size_kb']}KB")
    print(f"  Cache Hit Rate: {performance_data['cache_hit_rate']:.1%}")
    print(f"  SSL Grade: {performance_data['ssl_grade']}")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"faker_web example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 completed")
```
