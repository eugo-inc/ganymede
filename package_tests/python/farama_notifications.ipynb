## Example 2: Version-Specific Notifications

```python
import farama_notifications

# Example 2: Version-Specific Notifications
print("farama_notifications - Version-Specific Notifications")

try:
    # Set up notifications for a machine learning project
    project_name = "deep_rl_toolkit"
    farama_notifications.notifications[project_name] = {}
    
    # Add version notifications with different types of messages
    versions_and_messages = [
        ("0.1.0", "Initial alpha release - basic DQN implementation"),
        ("0.2.0", "Added PPO and A2C algorithms"),
        ("0.3.0", "BREAKING: Changed API for environment wrappers"),
        ("0.4.0", "Performance improvements - 2x faster training"),
        ("1.0.0", "Stable release with full documentation")
    ]
    
    for version, message in versions_and_messages:
        farama_notifications.notifications[project_name][version] = message
    
    # Display version history
    print(f"Notification history for {project_name}:")
    for version, message in farama_notifications.notifications[project_name].items():
        icon = "🔴" if "BREAKING" in message else "✅"
        print(f"  {icon} v{version}: {message}")
    
    # Check for breaking changes
    breaking_versions = [v for v, msg in farama_notifications.notifications[project_name].items() 
                        if "BREAKING" in msg]
    print(f"\nBreaking changes in versions: {breaking_versions}")
    
    print("✓ Version tracking complete")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"farama_notifications example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: Notification Management Utilities

```python
import farama_notifications
import json

# Example 3: Notification Management Utilities
print("farama_notifications - Notification Management")

try:
    # Helper function to add notifications
    def add_notification(project, version, message):
        if project not in farama_notifications.notifications:
            farama_notifications.notifications[project] = {}
        farama_notifications.notifications[project][version] = message
        print(f"Added: {project} v{version}")
    
    # Add multiple project notifications
    projects = [
        ("gymnasium", "1.0.0", "Major environment API update"),
        ("gymnasium", "1.1.0", "Added new Atari environments"),
        ("pettingzoo", "1.24.0", "Multi-agent environment improvements"),
        ("stable_baselines3", "2.1.0", "New PPO implementation"),
    ]
    
    for project, version, message in projects:
        add_notification(project, version, message)
    
    # Export notifications to JSON
    notifications_json = json.dumps(farama_notifications.notifications, indent=2)
    print(f"\nExported {len(farama_notifications.notifications)} projects")
    
    # Summary statistics
    total_notifications = sum(len(versions) for versions in farama_notifications.notifications.values())
    print(f"Total notifications: {total_notifications}")
    print(f"Projects registered: {list(farama_notifications.notifications.keys())}")
    
    print("✓ Management utilities working")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"farama_notifications example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: Notification Filtering and Search

```python
import farama_notifications

# Example 4: Notification Filtering and Search
print("farama_notifications - Filtering and Search")

try:
    # Set up sample notifications
    sample_data = {
        "openai_gym": {"0.26.0": "Final legacy version"},
        "gymnasium": {"0.28.0": "Migration from gym", "1.0.0": "Stable API release"},
        "stable_baselines3": {"1.8.0": "Bug fixes", "2.0.0": "BREAKING: New API"},
        "ray": {"2.5.0": "Performance improvements"},
        "minari": {"0.4.0": "Dataset management updates"}
    }
    
    # Load sample data
    farama_notifications.notifications.update(sample_data)
    
    # Search functions
    def find_breaking_changes():
        breaking = []
        for project, versions in farama_notifications.notifications.items():
            for version, message in versions.items():
                if "BREAKING" in message.upper():
                    breaking.append(f"{project} v{version}")
        return breaking
    
    def get_latest_versions():
        latest = {}
        for project, versions in farama_notifications.notifications.items():
            if versions:
                latest_version = max(versions.keys())
                latest[project] = latest_version
        return latest
    
    # Run searches
    breaking_changes = find_breaking_changes()
    latest_versions = get_latest_versions()
    
    print(f"Breaking changes found: {breaking_changes}")
    print(f"Latest versions: {latest_versions}")
    print("✓ Search functionality working")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"farama_notifications example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 5: Advanced Notification Operations

```python
import farama_notifications
import copy

# Example 5: Advanced Notification Operations
print("farama_notifications - Advanced Operations")

try:
    # Create a backup of the current state
    backup = copy.deepcopy(farama_notifications.notifications)
    
    # Bulk operations
    farama_notifications.notifications.clear()
    
    # Batch add notifications
    batch_notifications = {
        "farama_ecosystem": {
            "gymnasium": {"1.0.0": "Core RL environment framework"},
            "pettingzoo": {"1.24.0": "Multi-agent RL environments"},
            "minari": {"0.4.0": "RL dataset management"},
            "metaworld": {"2.0.0": "Robotic manipulation tasks"}
        }
    }
    
    # Flatten and add to notifications
    for category, projects in batch_notifications.items():
        for project, versions in projects.items():
            farama_notifications.notifications[project] = versions
    
    # Analysis
    total_projects = len(farama_notifications.notifications)
    total_versions = sum(len(v) for v in farama_notifications.notifications.values())
    
    print(f"Loaded {total_projects} projects with {total_versions} version notifications")
    
    # Restore from backup
    farama_notifications.notifications.clear()
    farama_notifications.notifications.update(backup)
    
    print("✓ Advanced operations complete")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"farama_notifications example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
```
