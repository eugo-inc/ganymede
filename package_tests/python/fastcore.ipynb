## Example 1: Collections and Data Structures

```python
from fastcore.basics import *
from fastcore.foundation import *

# Example 1: Collections and data structures
print("Example 1: Collections and Data Structures")

# L class - enhanced list with functional programming capabilities
data = L([1, 2, 3, 4, 5])
print(f"Original list: {data}")

# Functional operations
doubled = data.map(lambda x: x * 2)
print(f"Doubled: {doubled}")

filtered = data.filter(lambda x: x % 2 == 0)
print(f"Even numbers: {filtered}")

# AttrDict - dictionary with attribute access
config = AttrDict({
    'model': 'bert',
    'learning_rate': 0.001,
    'batch_size': 32
})

print(f"Config model: {config.model}")
print(f"Config learning rate: {config.learning_rate}")

# Chaining operations
result = L(range(10)).filter(lambda x: x > 5).map(lambda x: x ** 2)
print(f"Chained operations result: {result}")
```

## Example 2: Functional Programming Utilities

```python
from fastcore.basics import *
from fastcore.foundation import *
from functools import partial  # fastcore.basics imports it but leaves it out of `__all__`, so `import *` misses it

# Example 2: Functional programming utilities
print("Example 2: Functional Programming Utilities")

# compose - function composition
add_one = lambda x: x + 1
multiply_by_two = lambda x: x * 2

# Compose functions
composed_func = compose(multiply_by_two, add_one)
result = composed_func(5)
print(f"Composed function (add 1, then multiply by 2): {result}")

# partial - partial function application
def power(base, exponent):
    return base ** exponent

square = partial(power, exponent=2)
cube = partial(power, exponent=3)

print(f"Square of 4: {square(4)}")
print(f"Cube of 3: {cube(3)}")

# listify - convert various inputs to lists
single_item = listify(42)
string_item = listify("hello")
list_item = listify([1, 2, 3])

print(f"Listified single item: {single_item}")
print(f"Listified string: {string_item}")
print(f"Listified list: {list_item}")

# flatten - flatten nested structures
nested = [[1, 2], [3, [4, 5]], 6]
flattened = flatten(nested)
print(f"Flattened nested structure: {list(flattened)}")
```

## Example 3: Type System and Validation

```python
from fastcore.basics import *
from fastcore.foundation import *

# Example 3: Type system and validation
print("Example 3: Type System and Validation")

# typed - check argument and return types at runtime (fastcore's `typed` validates; it has no
# `.register`). typedispatch left fastcore for the separate `fasttransform` package, so it is not here.
@typed
def process_data(x: int, factor: float = 2.0) -> float:
    return x * factor

print(f"typed accepted an int: {process_data(21)}")
try:
    process_data("21")
except TypeError as type_error:  # the rejection IS the behaviour under test
    print(f"typed rejected a str: {type_error}")
else:
    raise AssertionError("typed accepted a str for an `int` parameter")

# type_hints - the resolved annotations that `typed` checks against
hints = type_hints(process_data)
print(f"Type hints: {hints}")
assert hints == {"x": int, "factor": float, "return": float}, hints
```

## Example 4: Testing and Debugging Utilities

```python
from fastcore.basics import *
from fastcore.foundation import *
from fastcore.test import test_eq, test_ne, test_close, test_fail  # not re-exported by basics/foundation

# Example 4: Testing and debugging utilities
print("Example 4: Testing and Debugging Utilities")

# test_eq - testing equality with nice output
def test_calculation():
    result = 2 + 2
    test_eq(result, 4)
    print("✓ Addition test passed")
    
    # Test with lists
    numbers = L([1, 2, 3]).map(lambda x: x * 2)
    test_eq(numbers, [2, 4, 6])
    print("✓ List mapping test passed")
    
test_calculation()

# test_ne - testing inequality
def test_inequality():
    test_ne(5, 3)
    print("✓ Inequality test passed")
    
test_inequality()

# test_close - testing floating point equality
def test_floating_point():
    import math
    result = math.sin(math.pi/2)
    test_close(result, 1.0, eps=1e-10)
    print("✓ Floating point test passed")
    
test_floating_point()

# test_fail - assert that a call raises, and that its message says why
test_fail(lambda: 1 / 0, contains="division by zero")
print("✓ Expected-failure test passed")

# dict2obj - attribute access over nested dicts, handy for poking at a config while debugging
from fastcore.xtras import dict2obj
data = dict2obj({"model": "transformer", "params": 175000000, "optim": {"lr": 3e-4}})
test_eq(data.optim.lr, 3e-4)
print(f"Debug view: model={data.model}, lr={data.optim.lr}")

result = sum(range(100000))
test_eq(result, 4999950000)
print(f"Work completed with result: {result}")
```

## Example 5: Advanced Data Processing and Utilities

```python
from fastcore.basics import *
from fastcore.foundation import *

# Example 5: Advanced data processing and utilities
print("Example 5: Advanced Data Processing and Utilities")

# Pipeline processing with L and functional operations
class DataProcessor:
    def __init__(self, data):
        self.data = L(data)
        
    def filter_valid(self):
        """Filter out None and negative values"""
        self.data = self.data.filter(lambda x: x is not None and x >= 0)
        return self
        
    def transform(self, func):
        """Apply transformation function"""
        self.data = self.data.map(func)
        return self
        
    def aggregate(self, func):
        """Apply aggregation function"""
        return func(self.data)

# Process data pipeline
raw_data = [1, -2, 3, None, 5, 8, -1, 13, 21]

processor = DataProcessor(raw_data)
result = (processor
          .filter_valid()
          .transform(lambda x: x ** 2)  # Square values
          .aggregate(lambda data: sum(data) / len(data)))  # Average

print(f"Raw data: {raw_data}")
print(f"Pipeline result (average of squares of positive values): {result:.2f}")

# Using store_attr for class attribute initialization
class ModelConfig:
    def __init__(self, model_name, learning_rate, batch_size, epochs=10):
        store_attr()  # Automatically stores all parameters as attributes
        
    def __repr__(self):
        return f"ModelConfig(model={self.model_name}, lr={self.learning_rate}, batch={self.batch_size}, epochs={self.epochs})"

config = ModelConfig("bert-base", 0.001, 32)
print(f"\nModel configuration: {config}")

# Path utilities and string processing
from fastcore.utils import *

# Creating temporary paths
import tempfile
with tempfile.TemporaryDirectory() as temp_dir:
    temp_path = Path(temp_dir)
    
    # Create and process files
    data_file = temp_path / "data.txt"
    data_file.write_text("Hello from fastcore utilities!")
    
    content = data_file.read_text()
    print(f"File content: {content}")
    print(f"File exists: {data_file.exists()}")
    
print("Advanced data processing completed successfully!")
```
