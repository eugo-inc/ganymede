## Example 1: Basic progress bar functionality

```python
from fastprogress import progress_bar
import time

# Example 1: Basic progress bar functionality  
print("Example 1: Basic progress bar")

# Create a simple progress bar
items = list(range(10))
print(f"Processing {len(items)} items...")

# Use progress_bar for iteration
for i in progress_bar(items):
    # Simulate some work
    time.sleep(0.05)
    
    # Show progress
    if i % 3 == 0:
        print(f"  Processing item {i}")

print(f"Basic progress bar completed")
print(f"Total items processed: {len(items)}")

# Test with different sizes
small_list = list(range(5))
large_list = list(range(20))

print(f"\nProgress bar sizes:")
print(f"Small list: {len(small_list)} items")
print(f"Large list: {len(large_list)} items")

print("Basic progress bar functionality completed")
```

## Example 2: Master bar with nested progress bars

```python
from fastprogress import master_bar, progress_bar
import time

# Example 2: Master bar with nested progress bars
print("Example 2: Master bar and nested progress")

# Create master progress bar for epochs
epochs = list(range(3))
mb = master_bar(epochs)

for epoch in mb:
    mb.write(f"Epoch {epoch + 1}")
    
    # Nested progress for batches within each epoch  
    batches = list(range(8))
    for batch in progress_bar(batches, parent=mb):
        # Simulate training step
        time.sleep(0.02)
        
        # Log some metrics occasionally
        if batch % 3 == 0:
            mb.write(f"  Batch {batch}: Processing...")
    
    mb.write(f"Epoch {epoch + 1} completed")

print("\nMaster bar with nested progress completed")

# Demonstrate master bar features
print(f"Master bar features demonstrated:")
print("- Multi-level progress tracking")
print("- Text logging within progress")
print("- Nested progress bar coordination")
print("- Real-time progress updates")

print("Master bar and nested progress completed")
```

## Example 3: Progress bar with custom display and statistics

```python
from fastprogress import progress_bar
import time
import random

# Example 3: Progress bar with custom display and statistics
print("Example 3: Custom progress display and statistics")

# Create data with random processing times
data_items = [f"file_{i:03d}.txt" for i in range(15)]
print(f"Processing {len(data_items)} files...")

# Track statistics
total_size = 0
errors = 0
start_time = time.time()

# Progress bar with custom display
pbar = progress_bar(data_items)
pbar.comment = "Initializing..."

for i, filename in enumerate(pbar):
    # Simulate variable processing time
    processing_time = random.uniform(0.01, 0.05)
    time.sleep(processing_time)
    
    # Simulate file processing
    file_size = random.randint(100, 1000)
    total_size += file_size
    
    # Occasionally simulate errors
    if random.random() < 0.1:
        errors += 1
        error_msg = f"Warning: {filename}"
        pbar.comment = error_msg
    else:
        pbar.comment = f"Processing {filename} ({file_size} KB)"

# Final statistics
end_time = time.time()
elapsed = end_time - start_time

print(f"\nProcessing completed:")
print(f"  Total files: {len(data_items)}")
print(f"  Total size: {total_size} KB")
print(f"  Errors: {errors}")
print(f"  Time elapsed: {elapsed:.2f} seconds")
print(f"  Average speed: {len(data_items)/elapsed:.1f} files/sec")

print("Custom progress display and statistics completed")
```

## Example 4: Time formatting and advanced progress features

```python
from fastprogress import progress_bar
from fastprogress.fastprogress import format_time
import time

# Example 4: Time formatting and advanced progress features
print("Example 4: Time formatting and advanced progress")

# Test time formatting utilities
print("Time formatting examples:")
time_values = [0.5, 5.2, 65.3, 3661.7]
for t in time_values:
    formatted = format_time(t)
    print(f"  {t:6.1f}s -> {formatted}")

# Create progress bar with custom formatting
total_steps = 12
data = list(range(total_steps))
print(f"\nProcessing {total_steps} steps with timing...")

start_time = time.time()
for i, step in enumerate(progress_bar(data)):
    # Simulate work with variable time
    time.sleep(0.02 + (i % 3) * 0.01)
    
    # Show milestone updates with formatted time
    if (i + 1) % 4 == 0:
        elapsed = time.time() - start_time
        formatted_time = format_time(elapsed)
        print(f"  Milestone: {i + 1}/{total_steps} steps in {formatted_time}")

# Final timing
total_time = time.time() - start_time
print(f"\nCompleted in {format_time(total_time)}")
print(f"Average: {format_time(total_time/total_steps)} per step")

print("Time formatting and advanced progress completed")
```

## Example 5: FastProgress capabilities and best practices

```python
import fastprogress
from fastprogress import progress_bar, master_bar
import time

# Example 5: FastProgress capabilities and best practices
print("Example 5: FastProgress capabilities summary")

# Package information
print(f"FastProgress package information:")
print(f"  Version: {getattr(fastprogress, '__version__', 'Unknown')}")

# Available progress bar types
print(f"\nAvailable progress bar types:")
print(f"  progress_bar: Basic progress tracking")
print(f"  master_bar: Nested progress with logging")

# Demonstrate different use cases
use_cases = [
    "Simple iteration progress",
    "Machine learning training loops", 
    "File processing operations",
    "Data transformation pipelines",
    "Jupyter notebook workflows"
]

print(f"\nCommon use cases ({len(use_cases)} total):")
for i, use_case in enumerate(use_cases, 1):
    print(f"  {i}. {use_case}")

# Performance demonstration
print(f"\nPerformance demonstration:")
data_sizes = [10, 25, 50]

for size in data_sizes:
    start = time.time()
    
    # Quick progress bar test
    for i in progress_bar(range(size)):
        time.sleep(0.001)  # Minimal delay
    
    elapsed = time.time() - start
    speed = size/elapsed if elapsed > 0 else 0
    print(f"  {size:2d} items: {elapsed:.3f}s ({speed:.0f} items/sec)")

print(f"\nFastProgress capabilities summary completed")
```
