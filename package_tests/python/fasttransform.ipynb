```python
# Example 1: Basic Transform class - apply a function with encode/decode
from fasttransform import Transform

class AddOne(Transform):
    def encodes(self, x: int): return x + 1
    def decodes(self, x: int): return x - 1

t = AddOne()
encoded = t(5)
decoded = t.decode(encoded)
print(f"encode(5) = {encoded}")
print(f"decode({encoded}) = {decoded}")
assert encoded == 6
assert decoded == 5
```

```python
# Example 2: Pipeline - compose multiple Transforms
from fasttransform import Transform, Pipeline

class Double(Transform):
    def encodes(self, x: int): return x * 2
    def decodes(self, x: int): return x // 2

class AddTen(Transform):
    def encodes(self, x: int): return x + 10
    def decodes(self, x: int): return x - 10

pipe = Pipeline([Double(), AddTen()])
out = pipe(3)
back = pipe.decode(out)
print(f"pipeline(3) = {out}")
print(f"pipeline.decode({out}) = {back}")
assert out == 16
assert back == 3
```

```python
# Example 3: Type dispatch - encodes acts on specific types only
from fasttransform import Transform

class StrOrIntTransform(Transform):
    def encodes(self, x: int): return x + 100
    def encodes(self, x: str): return x.upper()

t = StrOrIntTransform()
r_int = t(5)
r_str = t("hello")
print(f"t(5) = {r_int}")
print(f"t('hello') = {r_str}")
assert r_int == 105
assert r_str == "HELLO"
```

```python
# Example 4: Plain function as Transform via constructor
from fasttransform import Transform

def neg(x: int) -> int: return -x

t = Transform(neg)
result = t(7)
print(f"Transform(neg)(7) = {result}")
assert result == -7

# Setup hook should be callable without error
t.setup()
print("setup() ran without error")
```

```python
# Example 5: Pipeline with single Transform + tuple-aware apply on lists
from fasttransform import Transform, Pipeline

class Square(Transform):
    def encodes(self, x: int): return x * x

pipe = Pipeline([Square()])
outputs = [pipe(i) for i in range(5)]
print(f"squares 0..4: {outputs}")
assert outputs == [0, 1, 4, 9, 16]

# Empty pipeline is identity
empty = Pipeline([])
assert empty(42) == 42
print("All fasttransform smoke checks passed")
```
