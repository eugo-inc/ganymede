```python
# Example 1: guess type from a PNG byte signature
import filetype

# Minimal PNG header magic: 89 50 4E 47 0D 0A 1A 0A
png_bytes = b"\x89PNG\r\n\x1a\n" + b"\x00" * 32
kind = filetype.guess(png_bytes)
print(f"kind: {kind}")
print(f"extension: {kind.extension}")
print(f"mime: {kind.mime}")

assert kind is not None
assert kind.extension == 'png'
assert kind.mime == 'image/png'
```

```python
# Example 2: dedicated convenience helpers for mime + extension
import filetype

# JPEG SOI marker: FF D8 FF
jpeg_bytes = b"\xff\xd8\xff\xe0" + b"\x00" * 64

ext = filetype.guess_extension(jpeg_bytes)
mime = filetype.guess_mime(jpeg_bytes)
print(f"guess_extension: {ext}")
print(f"guess_mime: {mime}")

assert ext == 'jpg'
assert mime == 'image/jpeg'
```

```python
# Example 3: type-family predicates (is_image, is_video, is_archive, is_audio)
import filetype

png_bytes = b"\x89PNG\r\n\x1a\n" + b"\x00" * 32
# GIF87a header
gif_bytes = b"GIF87a" + b"\x00" * 32
# ZIP local file header signature: 50 4B 03 04
zip_bytes = b"PK\x03\x04" + b"\x00" * 64

print(f"png is_image: {filetype.is_image(png_bytes)}")
print(f"gif is_image: {filetype.is_image(gif_bytes)}")
print(f"zip is_archive: {filetype.is_archive(zip_bytes)}")
print(f"png is_video: {filetype.is_video(png_bytes)}")

assert filetype.is_image(png_bytes) is True
assert filetype.is_image(gif_bytes) is True
assert filetype.is_archive(zip_bytes) is True
assert filetype.is_video(png_bytes) is False
```

```python
# Example 4: enumerate all known types via filetype.types + look up via get_type.
# The top-level `filetype.types` attribute is the flat LIST of registered type
# instances (it shadows the `filetype.types` module by design).
import filetype

all_types = filetype.types
print(f"total registered types: {len(all_types)}")

# Sample the first handful of extensions/mimes
for t in all_types[:5]:
    print(f"  {t.extension:8s} -> {t.mime}")

# Look up by extension
png_type = filetype.get_type(ext='png')
print(f"\nget_type(ext='png') = {png_type} (mime={png_type.mime})")

assert len(all_types) > 20
assert png_type is not None
assert png_type.mime == 'image/png'
```

```python
# Example 5: unknown bytes return None; guess accepts file-like / bytearray
import io

import filetype

unknown = b"this is just plain text, not any known binary format\n" * 4
print(f"guess(unknown text) -> {filetype.guess(unknown)}")
assert filetype.guess(unknown) is None

# bytearray input
png_ba = bytearray(b"\x89PNG\r\n\x1a\n" + b"\x00" * 32)
kind_ba = filetype.guess(png_ba)
assert kind_ba is not None and kind_ba.extension == 'png'

# file-like (BytesIO) input
buf = io.BytesIO(b"\xff\xd8\xff\xe0" + b"\x00" * 64)
kind_io = filetype.guess(buf)
assert kind_io is not None and kind_io.extension == 'jpg'

print("all filetype guesses behaved as expected")
```
