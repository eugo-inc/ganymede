## Example 1: Basic Flask application creation

```python
from flask import Flask
try:
    from importlib.metadata import version
    flask_version = version("flask")
except ImportError:
    import flask
    flask_version = flask.__version__

# Example 1: Basic Flask application creation
print("Example 1: Basic Flask application creation")

app = Flask(__name__)

# Define simple routes
@app.route('/')
def hello_world():
    return 'Hello, Flask World!'

@app.route('/about')
def about():
    return 'This is the about page'

# Test app configuration
print(f"Flask app name: {app.name}")
print(f"Flask version: {flask_version}")
print(f"App instance class: {type(app).__name__}")
print(f"Number of routes: {len(app.url_map._rules)}")

print("Basic Flask application setup completed")
```

## Example 2: Flask JSON API and request handling

```python
from flask import Flask, jsonify

# Example 2: Flask JSON API and request handling
print("Example 2: Flask JSON API and request handling")

app = Flask(__name__)

# Sample data
users = [
    {'id': 1, 'name': 'Alice', 'email': 'alice@example.com'},
    {'id': 2, 'name': 'Bob', 'email': 'bob@example.com'}
]

@app.route('/api/users', methods=['GET'])
def get_users():
    return jsonify(users)

@app.route('/api/users/<int:user_id>', methods=['GET'])
def get_user(user_id):
    user = next((u for u in users if u['id'] == user_id), None)
    return jsonify(user) if user else jsonify({'error': 'User not found'}), 404

# Test API with test client
with app.test_client() as client:
    response = client.get('/api/users')
    print(f"GET /api/users status: {response.status_code}")
    print(f"Response data: {response.json}")
    
    response = client.get('/api/users/1')
    print(f"GET /api/users/1: {response.json}")

print("Flask JSON API example completed")
```

## Example 3: Flask templates and sessions

```python
from flask import Flask, render_template_string, session

# Example 3: Flask templates and sessions
print("Example 3: Flask templates and sessions")

app = Flask(__name__)
app.secret_key = 'demo-secret-key'

# Simple HTML template
template = """
<html><body>
<h1>{{heading}}</h1>
<p>Welcome, {{username}}!</p>
<p>Visit count: {{visit_count}}</p>
</body></html>
"""

@app.route('/dashboard/<username>')
def dashboard(username):
    if 'visit_count' not in session:
        session['visit_count'] = 0
    session['visit_count'] += 1
    
    return render_template_string(template, 
                                  heading="Dashboard",
                                  username=username,
                                  visit_count=session['visit_count'])

# Test template rendering
with app.test_client() as client:
    response = client.get('/dashboard/testuser')
    print(f"Dashboard status: {response.status_code}")
    print(f"Contains username: {'testuser' in response.data.decode()}")

print("Flask templates and sessions example completed")
```

## Example 4: Flask error handling and middleware

```python
from flask import Flask, request, abort

# Example 4: Flask error handling and middleware
print("Example 4: Flask error handling and middleware")

app = Flask(__name__)

# Custom error handlers
@app.errorhandler(404)
def not_found(error):
    return {'error': 'Resource not found'}, 404

@app.errorhandler(401)
def unauthorized(error):
    return {'error': 'Unauthorized'}, 401

# Before/after request middleware
@app.before_request
def before_request():
    print(f"Before: {request.method} {request.path}")

@app.after_request
def after_request(response):
    response.headers['X-Custom-Header'] = 'Flask-Demo'
    return response

@app.route('/protected')
def protected():
    if request.headers.get('Authorization') != 'Bearer demo-token':
        abort(401)
    return {'message': 'Access granted'}

# Test error handling and middleware
with app.test_client() as client:
    response = client.get('/nonexistent')
    print(f"404 response: {response.json}")
    
    response = client.get('/protected', headers={'Authorization': 'Bearer demo-token'})
    print(f"Protected response: {response.json}")

print("Flask error handling example completed")
```

## Example 5: Flask blueprints and application context

```python
from flask import Flask, Blueprint, current_app
try:
    from importlib.metadata import version
    flask_version = version("flask")
except ImportError:
    import flask
    flask_version = flask.__version__

# Example 5: Flask blueprints and application context
print("Example 5: Flask blueprints and application context")

app = Flask(__name__)
app.config['SECRET_KEY'] = 'demo-key'

# Create a blueprint
api_bp = Blueprint('api', __name__, url_prefix='/api/v1')

@api_bp.route('/status')
def api_status():
    return {
        'status': 'operational',
        'flask_version': flask_version,
        'debug_mode': current_app.config.get('DEBUG', False)
    }

# Register the blueprint
app.register_blueprint(api_bp)

# Test blueprint functionality
with app.test_client() as client:
    response = client.get('/api/v1/status')
    print(f"API status response: {response.json}")

print(f"Flask v{flask_version} with {len(list(app.blueprints.keys()))} blueprints")
print("Flask blueprints example completed")
```
