## Example 1: folium functionality test 1

```python
import folium

# Example 1: Basic Map Creation and Configuration
print("Example 1: Basic Map Creation and Configuration")

try:
    # Check folium version
    print(f"✅ Folium version: {folium.__version__}")
    
    # Create basic map centered on New York City
    nyc_map = folium.Map(
        location=[40.7128, -74.0060],
        zoom_start=12,
        tiles='OpenStreetMap'
    )
    print(f"NYC map created at (40.7128, -74.0060)")
    
    # Test different tile layers
    paris_map = folium.Map(
        location=[48.8566, 2.3522],
        zoom_start=13,
        tiles='CartoDB positron'
    )
    print(f"Paris map with CartoDB positron tiles")
    
    # Test HTML generation
    html_content = nyc_map._repr_html_()
    html_length = len(html_content) if html_content else 0
    print(f"HTML content generated: {html_length} characters")

except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"folium example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 completed")
```

## Example 2: folium functionality test 2

```python
import folium

# Example 2: Markers, Popups, and Interactive Elements
print("Example 2: Markers, Popups, and Interactive Elements")

try:
    # Create base map
    marker_map = folium.Map(location=[40.7128, -74.0060], zoom_start=12)

    # Basic marker
    folium.Marker([40.7128, -74.0060], popup='NYC', tooltip='Basic Marker').add_to(marker_map)

    # Custom icon marker
    custom_icon = folium.Icon(color='red', icon='star')
    folium.Marker([40.7589, -73.9851], popup='Times Square', icon=custom_icon).add_to(marker_map)

    # Circle marker
    folium.CircleMarker([40.6892, -74.0445], radius=15, popup='Liberty Island',
                       color='blue', fill=True).add_to(marker_map)

    # HTML popup
    html_popup = '<b>Central Park</b><br>843 acres<br>Opened: 1857'
    folium.Marker([40.7829, -73.9654], popup=folium.Popup(html_popup, max_width=200)).add_to(marker_map)

    print("✓ Added basic markers, custom icons, and HTML popups")
    print(f"✓ Map contains {len([m for m in marker_map._children.values()])} elements")

except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"folium example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed")
```

## Example 3: folium functionality test 3

```python
import folium
import random

# Example 3: Layer Management and Data Visualization
print("Example 3: Layer Management and Data Visualization")

try:
    # Create map with multiple tile layers
    layer_map = folium.Map(location=[40.7128, -74.0060], zoom_start=10)

    # Add different tile layers
    folium.TileLayer('CartoDB positron', name='Light Theme').add_to(layer_map)
    folium.TileLayer('CartoDB dark_matter', name='Dark Theme').add_to(layer_map)
    folium.LayerControl().add_to(layer_map)

    # Create feature groups
    attractions = folium.FeatureGroup(name='Attractions')
    parks = folium.FeatureGroup(name='Parks')

    # Add markers to groups
    folium.Marker([40.7589, -73.9851], popup='Times Square').add_to(attractions)
    folium.Marker([40.7829, -73.9654], popup='Central Park').add_to(parks)

    attractions.add_to(layer_map)
    parks.add_to(layer_map)

    # Heat map layer
    from folium.plugins import HeatMap
    heat_data = [[40.7128, -74.0060, 0.8], [40.7589, -73.9851, 0.9]]
    HeatMap(heat_data).add_to(layer_map)
    print("✓ HeatMap plugin available")
    layer_html = layer_map.get_root().render()
    assert 'L.heatLayer(' in layer_html and layer_html.count('L.marker(') == 2, 'heat layer / markers missing from the rendered map'

    print("✓ Layer management and visualization features demonstrated")

except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"folium example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 completed")
```

## Example 4: folium functionality test 4

```python
import folium

# Example 4: Advanced Folium Plugins and Custom Features
print("Example 4: Advanced Folium Plugins and Custom Features")

try:
    # Create base map
    advanced_map = folium.Map(location=[40.7128, -74.0060], zoom_start=10)

    # Test available plugins
    plugins_tested = []

    from folium.plugins import MeasureControl
    MeasureControl().add_to(advanced_map)
    plugins_tested.append("MeasureControl")

    from folium.plugins import Fullscreen
    Fullscreen().add_to(advanced_map)
    plugins_tested.append("Fullscreen")

    # Custom HTML icon
    custom_icon = folium.DivIcon(
        icon_size=(30, 30),
        html='<div style="font-size: 20pt; color: red;">📍</div>'
    )
    folium.Marker([40.7128, -74.0060], icon=custom_icon, tooltip='Custom Icon').add_to(advanced_map)

    print(f"✓ Plugins tested: {plugins_tested}")
    assert plugins_tested == ["MeasureControl", "Fullscreen"], plugins_tested
    advanced_html = advanced_map.get_root().render()
    assert 'L.Control.Measure(' in advanced_html and 'L.control.fullscreen' in advanced_html, 'plugin controls missing from the rendered map'
    print("✓ Custom HTML icons and advanced features demonstrated")

except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"folium example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed")
```

## Example 5: folium functionality test 5

```python
import folium
import time
import random

# Example 5: Performance Optimization and Production Usage
print("Example 5: Performance Optimization and Production Usage")

try:
    # Performance testing with markers
    performance_map = folium.Map(location=[40.7128, -74.0060], zoom_start=10)

    random.seed(42)
    start_time = time.time()

    # Create 100 markers for performance test
    for i in range(100):
        lat = 40.7128 + random.uniform(-0.1, 0.1)
        lon = -74.0060 + random.uniform(-0.1, 0.1)
        folium.Marker([lat, lon], popup=f'Marker {i+1}').add_to(performance_map)

    creation_time = time.time() - start_time
    print(f"✓ Created 100 markers in {creation_time:.3f} seconds")

    # Test clustering vs individual markers
    from folium.plugins import MarkerCluster
    cluster_map = folium.Map(location=[40.7128, -74.0060], zoom_start=10)
    marker_cluster = MarkerCluster().add_to(cluster_map)

    for i in range(50):
        lat = 40.7128 + random.uniform(-0.05, 0.05)
        lon = -74.0060 + random.uniform(-0.05, 0.05)
        folium.Marker([lat, lon], popup=f'Clustered {i+1}').add_to(marker_cluster)

    print("✓ MarkerCluster plugin available and tested")
    cluster_html = cluster_map.get_root().render()
    assert cluster_html.count('L.markerClusterGroup(') == 1 and cluster_html.count('L.marker(') == 50, 'clustered markers missing from the rendered map'

    print("✓ Performance optimization techniques demonstrated")

except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"folium example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 completed")
```
