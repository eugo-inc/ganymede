## Example 1: Basic Immutable Dictionary Operations

```python
from frozendict import frozendict

# Example 1: Basic immutable dictionary operations
print("Example 1: Basic Immutable Dictionary Operations")

# Create a frozendict from a regular dictionary
regular_dict = {"name": "Alice", "age": 30, "city": "New York"}
frozen_dict = frozendict(regular_dict)

print(f"Original dict: {regular_dict}")
print(f"Frozen dict: {frozen_dict}")
print(f"Type: {type(frozen_dict).__name__}")

# Test immutability - these operations should raise TypeError
print(f"\nTesting immutability:")
try:
    frozen_dict["age"] = 31  # This should fail
except TypeError as e:
    print(f"✓ Correctly prevented mutation: {type(e).__name__}")

# Test read operations work normally
print(f"\nRead operations:")
print(f"  Name: {frozen_dict['name']}")
print(f"  Keys: {list(frozen_dict.keys())}")

# Test hashability (key advantage of frozendict)
print(f"\nHashability test:")
hash_value = hash(frozen_dict)
print(f"✓ Frozendict is hashable: {hash_value}")
print("✓ Can be used as dict key or set member")
```

## Example 2: Creating New Versions and Functional Updates

```python
from frozendict import frozendict

# Example 2: Creating new versions and functional updates
print("Example 2: Creating New Versions and Functional Updates")

# Start with initial configuration
config_v1 = frozendict({
    "database_url": "postgresql://localhost/app",
    "debug": False,
    "max_connections": 100
})

print(f"Configuration v1: {dict(config_v1)}")

# Create new version with updated values (functional approach)
config_v2 = frozendict({**config_v1, "debug": True, "max_connections": 200})
print(f"Configuration v2: {dict(config_v2)}")

# Add new fields
config_v3 = frozendict({**config_v2, "cache_enabled": True})
print(f"Configuration v3: {dict(config_v3)}")

# Functional update helper
def update_config(config, **updates):
    return frozendict({**config, **updates})

# Use helper
prod_config = update_config(config_v3, debug=False, database_url="postgresql://prod/app")
print(f"Production config: {dict(prod_config)}")
print("✓ Original configurations remain unchanged")
```

## Example 3: Using as Set Members and Dictionary Keys

```python
from frozendict import frozendict

# Example 3: Using as set members and dictionary keys
print("Example 3: Using as Set Members and Dictionary Keys")

# Create user profiles as frozendicts
users = [
    frozendict({"name": "Alice", "role": "admin", "department": "IT"}),
    frozendict({"name": "Bob", "role": "user", "department": "Sales"}),
    frozendict({"name": "Carol", "role": "manager", "department": "HR"}),
]

print(f"Created {len(users)} user profiles")

# Use frozendicts in a set (deduplication)
user_set = set(users)
duplicate_alice = frozendict({"name": "Alice", "role": "admin", "department": "IT"})
user_set.add(duplicate_alice)
print(f"Unique users in set: {len(user_set)} (duplicates automatically removed)")

# Use frozendicts as dictionary keys
user_metadata = {}
for user in users:
    metadata = {"login_count": abs(hash(user['name'])) % 100}
    user_metadata[user] = metadata

print(f"\nUser metadata mapping:")
for user_profile, metadata in user_metadata.items():
    print(f"  {user_profile['name']}: {metadata['login_count']} logins")

print("✓ Frozendicts work perfectly as set members and dict keys")
```

## Example 4: Thread Safety and Concurrent Access

```python
from frozendict import frozendict
import threading
import time

# Example 4: Thread safety and concurrent access
print("Example 4: Thread Safety and Concurrent Access")

# Create shared immutable configuration
shared_config = frozendict({
    "max_workers": 10,
    "timeout": 30,
    "database_url": "postgresql://localhost/app"
})

print(f"Shared config created with {len(shared_config)} settings")

# Simulate multiple threads reading configuration
def worker_task(worker_id, config, results):
    for i in range(3):
        # Multiple threads can safely read from frozendict
        timeout = config["timeout"]
        max_workers = config["max_workers"]
        results.append(f"Worker-{worker_id}: timeout={timeout}, workers={max_workers}")
        time.sleep(0.01)

# Run concurrent access test
results = []
threads = []

for worker_id in range(3):
    thread = threading.Thread(target=worker_task, args=(worker_id, shared_config, results))
    threads.append(thread)

for thread in threads:
    thread.start()
for thread in threads:
    thread.join()

print(f"✓ {len(results)} tasks completed safely across {len(threads)} threads")
print("✓ No locks needed - frozendict is inherently thread-safe")
```

## Example 5: Performance Optimization and Memory Efficiency

```python
from frozendict import frozendict
import time

# Example 5: Performance optimization and memory efficiency
print("Example 5: Performance Optimization and Memory Efficiency")

# Caching performance with frozendicts
print("Caching performance demonstration:")

computation_cache = {}

def expensive_computation(params):
    """Simulate expensive computation with caching"""
    cache_key = frozendict(params)
    
    if cache_key in computation_cache:
        return computation_cache[cache_key], True  # Cache hit
    
    # Simulate expensive work
    time.sleep(0.001)
    result = sum(v for v in params.values() if isinstance(v, (int, float)))
    computation_cache[cache_key] = result
    return result, False  # Cache miss

# Test with parameter combinations
test_params = [
    {"x": 10, "y": 20},
    {"a": 1, "b": 2},
    {"x": 10, "y": 20},  # Duplicate - should hit cache
]

hits = 0
for i, params in enumerate(test_params):
    result, was_hit = expensive_computation(params)
    if was_hit:
        hits += 1
    print(f"  Computation {i+1}: result={result}, cache_hit={was_hit}")

print(f"✓ Cache hits: {hits}/{len(test_params)} - frozendict enables perfect caching")
```
