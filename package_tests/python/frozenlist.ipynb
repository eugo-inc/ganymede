## Example 3: HTTP Headers Simulation and Web Framework Integration

Demonstrate FrozenList usage in web applications, particularly for HTTP headers management.

```python
# Example 3: HTTP headers simulation and web framework integration
from frozenlist import FrozenList
from collections import namedtuple

print("Example 3: HTTP headers simulation and web framework integration")

# Simulate HTTP request headers using FrozenList
Header = namedtuple('Header', ['name', 'value'])

# Create request headers
request_headers = FrozenList([
    Header('Content-Type', 'application/json'),
    Header('Authorization', 'Bearer token123'),
    Header('User-Agent', 'MyApp/1.0'),
    Header('Accept', 'application/json'),
    Header('X-Request-ID', 'req-12345')
])

# Add custom headers before freezing
request_headers.append(Header('X-Custom-Header', 'custom-value'))
request_headers.freeze()  # Make immutable for thread safety

print(f"Request headers ({len(request_headers)} items):")
for i, header in enumerate(request_headers):
    print(f"  {i+1}. {header.name}: {header.value}")

# Demonstrate header lookup functionality
def find_header(headers, name):
    for header in headers:
        if header.name.lower() == name.lower():
            return header.value
    return None

print(f"\nHeader lookup:")
print(f"  Content-Type: {find_header(request_headers, 'content-type')}")
print(f"  Authorization: {find_header(request_headers, 'authorization')}")
print(f"  Missing header: {find_header(request_headers, 'missing')}")

# Simulate response headers with status tracking
response_headers = FrozenList([
    Header('Content-Type', 'application/json; charset=utf-8'),
    Header('Server', 'nginx/1.18'),
    Header('X-Response-Time', '25ms'),
    Header('Cache-Control', 'no-cache')
])
response_headers.freeze()

print(f"\nResponse headers: {len(response_headers)} items")
print(f"Thread-safe operations: ✓ (frozen={response_headers.frozen})")

print("HTTP headers simulation completed successfully!")
```

## Example 4: Thread Safety and Concurrent Access Patterns

Demonstrate thread-safe operations and concurrent access patterns with FrozenList.

```python
# Example 4: Thread safety and concurrent access patterns
from frozenlist import FrozenList
import threading
import time
from concurrent.futures import ThreadPoolExecutor, as_completed

print("Example 4: Thread safety and concurrent access patterns")

# Create a shared FrozenList for concurrent access
shared_data = FrozenList(range(100))
shared_data.freeze()  # Thread-safe once frozen

# Thread-safe read operations
def read_worker(thread_id, fl, num_reads=50):
    results = []
    for i in range(num_reads):
        # Safe concurrent read operations
        index = (thread_id * 10 + i) % len(fl)
        value = fl[index]
        results.append((thread_id, index, value))
        time.sleep(0.001)  # Simulate processing time
    return results

# Test concurrent access with multiple threads
print("Testing concurrent read access...")
start_time = time.time()

with ThreadPoolExecutor(max_workers=4) as executor:
    # Submit read tasks
    future_to_thread = {
        executor.submit(read_worker, tid, shared_data, 25): tid 
        for tid in range(4)
    }
    
    all_results = []
    for future in as_completed(future_to_thread):
        thread_id = future_to_thread[future]
        try:
            results = future.result()
            all_results.extend(results)
            print(f"  Thread {thread_id}: {len(results)} reads completed")
        except Exception as e:
            print(f"  Thread {thread_id} error: {e}")
            raise

end_time = time.time()
print(f"Concurrent access completed in {end_time - start_time:.3f}s")
print(f"Total operations: {len(all_results)}")

# Verify data integrity
unique_values = set(result[2] for result in all_results)
expected_values = set(range(100))
data_integrity = len(unique_values.intersection(expected_values))
print(f"Data integrity: {data_integrity}/{len(expected_values)} values verified")

print("\n✓ Thread-safe immutable access completed successfully!")
print(f"✓ All {len(all_results)} concurrent operations succeeded")

print("Thread safety demonstration completed successfully!")
```

## Example 5: Advanced Patterns and Production Use Cases

Demonstrate advanced usage patterns including serialization, caching, and production scenarios.

```python
# Example 5: Advanced patterns and production use cases
from frozenlist import FrozenList
import pickle
import json
from dataclasses import dataclass, asdict
from typing import List, Optional

print("Example 5: Advanced patterns and production use cases")

# Define data structures for production scenarios
@dataclass
class CacheEntry:
    key: str
    value: str
    ttl: int
    metadata: dict

@dataclass
class APIResponse:
    status_code: int
    headers: FrozenList
    data: dict
    cache_entries: FrozenList

# Create a complex data structure with nested FrozenLists
api_headers = FrozenList([
    ('Content-Type', 'application/json'),
    ('X-API-Version', 'v1.2'),
    ('Cache-Control', 'max-age=3600'),
    ('X-Rate-Limit', '1000')
])
api_headers.freeze()

cache_data = FrozenList([
    CacheEntry('user:123', 'John Doe', 3600, {'region': 'us-east'}),
    CacheEntry('product:456', 'Widget Pro', 1800, {'category': 'tools'}),
    CacheEntry('session:789', 'active_session', 900, {'ip': '192.168.1.1'})
])
cache_data.freeze()

# Create API response object
response = APIResponse(
    status_code=200,
    headers=api_headers,
    data={'users': [{'id': 1, 'name': 'John'}], 'total': 1},
    cache_entries=cache_data
)

print(f"API Response created with {len(response.headers)} headers and {len(response.cache_entries)} cache entries")

print("✓ All production patterns completed successfully!")
print("Advanced patterns and production use cases demonstrated successfully!")
```
