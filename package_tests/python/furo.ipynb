# furo

Furo 2025.12.19 is a Sphinx HTML theme with light and dark modes and a searchable sidebar. Its wheel carries the
compiled CSS and JavaScript, so no Node.js is needed. These cells build a site with it, set theme options, check the
search page and assets, the dark-mode code highlighting, and the theme registration.

```python
# Example 1: a site built with html_theme = "furo" links Furo's stylesheet and script.
import pathlib, tempfile
from sphinx.cmd.build import build_main

src = pathlib.Path(tempfile.mkdtemp())
(src / "conf.py").write_text("project = 'F'\nhtml_theme = 'furo'\n")
(src / "index.rst").write_text("F\n=\n\nHello.\n")
assert build_main(["-q", "-W", "-b", "html", str(src), str(src / "_out")]) == 0
html = (src / "_out" / "index.html").read_text()
assert "_static/styles/furo.css" in html and "_static/scripts/furo.js" in html
assert (src / "_out" / "_static" / "styles" / "furo.css").exists()
```

```python
# Example 3: the search page and its index are written, and the page carries Furo's search box.
import pathlib, tempfile
from sphinx.cmd.build import build_main

src = pathlib.Path(tempfile.mkdtemp())
(src / "conf.py").write_text("project = 'F'\nhtml_theme = 'furo'\n")
(src / "index.rst").write_text("F\n=\n\nSearchable words here.\n")
assert build_main(["-q", "-W", "-b", "html", str(src), str(src / "_out")]) == 0
out = src / "_out"
assert (out / "search.html").exists() and (out / "searchindex.js").exists()
assert 'class="sidebar-search"' in (out / "index.html").read_text()
```

```python
# Example 4: Furo writes a second Pygments stylesheet for dark mode, from pygments_dark_style.
import pathlib, tempfile
from sphinx.cmd.build import build_main

src = pathlib.Path(tempfile.mkdtemp())
(src / "conf.py").write_text("project = 'F'\nhtml_theme = 'furo'\npygments_style = 'friendly'\npygments_dark_style = 'monokai'\n")
(src / "index.rst").write_text("F\n=\n\n.. code-block:: python\n\n   print('hi')\n")
assert build_main(["-q", "-W", "-b", "html", str(src), str(src / "_out")]) == 0
css = (src / "_out" / "_static" / "pygments.css").read_text()
print(css[:200])
assert 'body[data-theme="dark"]' in css
```

```python
import importlib.metadata as md

entry_points = {ep.name: ep.value for ep in md.entry_points(group="sphinx.html_themes")}
print(entry_points)
assert entry_points.get("furo") == "furo"
assert md.version("furo") == "2025.12.19"
import furo
assert callable(furo.setup)
from furo.navigation import get_navigation_tree
tree = get_navigation_tree('<ul><li class="toctree-l1 current"><a class="current reference internal" href="#">A</a></li></ul>')
print(tree)
assert "current-page" in tree  # Furo marks the page the sidebar is showing
```
