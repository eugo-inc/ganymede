```python
# Example 3: Text and Binary Data Handling
from __future__ import unicode_literals
from builtins import str, bytes, open
from future.utils import text_to_native_str, native_str_to_bytes, bytes_to_native_str
import sys

print("Text and binary data handling:")
print("=" * 33)

# String and bytes distinction
print("1. Modern string and bytes handling:")
text_string = "Hello, 世界! 🌍"  # Always unicode
binary_data = b"Binary content"  # Always bytes

print(f"   Text string: '{text_string}' (type: {type(text_string)})")
print(f"   Binary data: {binary_data} (type: {type(binary_data)})")

# Encoding and decoding
encoded_text = text_string.encode('utf-8')
decoded_text = encoded_text.decode('utf-8')
print(f"   Encoded: {encoded_text} ({len(encoded_text)} bytes)")
print(f"   Decoded: '{decoded_text}' (matches original: {text_string == decoded_text})")

# File handling with explicit encoding
print(f"\n2. Modern file handling with encoding:")
import tempfile
import os

# Write text file with explicit encoding
text_content = "Line 1: ASCII text\nLine 2: Unicode 世界\nLine 3: Emoji 🚀"
with tempfile.NamedTemporaryFile(mode='w', encoding='utf-8', delete=False) as f:
    temp_file = f.name
    f.write(text_content)

# Read back with explicit encoding
with open(temp_file, 'r', encoding='utf-8') as f:
    read_content = f.read()

print(f"   Written: {len(text_content)} characters")
print(f"   Read: {len(read_content)} characters")
print(f"   Content matches: {text_content == read_content}")

# Clean up
os.unlink(temp_file)

# Utility functions for conversion
print(f"\n3. Future utility functions:")
native_str = text_to_native_str(text_string)
print(f"   Text to native: '{native_str}' (type: {type(native_str)})")

byte_data = native_str_to_bytes(native_str)
print(f"   Native to bytes: {byte_data[:20]}... (type: {type(byte_data)})")

back_to_str = bytes_to_native_str(byte_data)
print(f"   Bytes to native: '{back_to_str}' (type: {type(back_to_str)})")

# Platform-specific string handling
print(f"\n4. Cross-version string compatibility:")
print(f"   Python version: {sys.version_info.major}.{sys.version_info.minor}")
print(f"   Native str type: {type('')}")
print(f"   Future str type: {type(str('test'))}")
print(f"   Default encoding: {sys.getdefaultencoding()}")

print("\nText and binary data handling completed successfully!")
```

```python
# Example 4: Object-Oriented Features and Metaclasses
from __future__ import unicode_literals
from builtins import super, object
from future.utils import with_metaclass
import sys

print("Object-oriented features and metaclasses:")
print("=" * 40)

# Modern super() usage
print("1. Modern super() without arguments:")

class Animal(object):
    def __init__(self, name, species):
        self.name = name
        self.species = species
    
    def speak(self):
        return f"{self.name} makes a sound"
    
    def info(self):
        return f"{self.name} is a {self.species}"

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name, "dog")  # Modern super() syntax
        self.breed = breed
    
    def speak(self):
        base_sound = super().speak()  # Call parent method
        return f"{base_sound} - Woof!"
    
    def info(self):
        base_info = super().info()
        return f"{base_info} of breed {self.breed}"

# Test modern inheritance
dog = Dog("Buddy", "Golden Retriever")
print(f"   Dog name: {dog.name}")
print(f"   Dog sound: {dog.speak()}")
print(f"   Dog info: {dog.info()}")

# Cross-version metaclass compatibility
print(f"\n2. Cross-version metaclass syntax:")

class MetaExample(type):
    def __new__(mcs, name, bases, attrs):
        # Add automatic documentation
        attrs['__doc__'] = f"Auto-generated class {name}"
        attrs['class_name'] = name
        return super(MetaExample, mcs).__new__(mcs, name, bases, attrs)

# Use with_metaclass for Python 2/3 compatibility
class ExampleClass(with_metaclass(MetaExample, object)):
    def __init__(self, value):
        self.value = value
    
    def get_info(self):
        return f"Class: {self.class_name}, Value: {self.value}"

example = ExampleClass("test_value")
print(f"   Instance info: {example.get_info()}")
print(f"   Auto-generated doc: {example.__doc__}")
print(f"   Class name attribute: {example.class_name}")

# Modern property decorators
print(f"\n3. Property decorators and descriptors:")

class Temperature(object):
    def __init__(self, celsius=0):
        self._celsius = celsius
    
    @property
    def celsius(self):
        return self._celsius
    
    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Temperature cannot be below absolute zero")
        self._celsius = value
    
    @property
    def fahrenheit(self):
        return (self._celsius * 9/5) + 32
    
    @fahrenheit.setter
    def fahrenheit(self, value):
        self.celsius = (value - 32) * 5/9

temp = Temperature(25)
print(f"   Celsius: {temp.celsius}°C")
print(f"   Fahrenheit: {temp.fahrenheit}°F")

temp.fahrenheit = 100
print(f"   After setting 100°F - Celsius: {temp.celsius:.2f}°C")

print("\nObject-oriented features completed successfully!")
```

```python
# Example 5: Production Migration Patterns and Best Practices
from __future__ import absolute_import, division, print_function, unicode_literals
from builtins import *
from future import standard_library
from future.utils import iteritems, itervalues, iterkeys, raise_from
import sys
import traceback

print("Production migration patterns and best practices:")
print("=" * 49)

# Standard library alias installation for production
standard_library.install_aliases()

# Migration pattern 1: Safe dictionary iteration
print("1. Safe dictionary iteration patterns:")
data = {'users': 100, 'products': 50, 'orders': 25}

# Use future utilities for cross-version iteration
print("   Dictionary keys:")
for key in iterkeys(data):
    print(f"     Key: {key}")

print("   Dictionary values:")
for value in itervalues(data):
    print(f"     Value: {value}")

print("   Dictionary items:")
for key, value in iteritems(data):
    print(f"     {key}: {value}")

# Migration pattern 2: Exception chaining
print(f"\n2. Exception chaining (Python 3 style):")

def process_data(data):
    try:
        if not isinstance(data, dict):
            raise TypeError("Data must be a dictionary")
        return data['required_key']
    except KeyError as e:
        raise_from(ValueError("Missing required configuration"), e)

try:
    result = process_data({'wrong_key': 'value'})
except ValueError as e:
    print(f"   Caught exception: {type(e).__name__}: {e}")
    assert str(e) == "Missing required configuration", f"unexpected message: {e}"
    # raise_from must chain the original KeyError as __cause__.
    assert isinstance(e.__cause__, KeyError), f"expected a chained KeyError, got {e.__cause__!r}"
    print(f"   Caused by: {type(e.__cause__).__name__}: {e.__cause__}")
else:
    raise AssertionError(f"expected ValueError from process_data, got a result: {result!r}")

# Migration pattern 3: Modern string formatting
print(f"\n3. Modern string formatting patterns:")
user_data = {
    'name': 'Alice',
    'age': 30,
    'city': 'New York'
}

# f-strings work with future imports
formatted_message = f"User {user_data['name']} is {user_data['age']} years old from {user_data['city']}"
print(f"   Formatted: {formatted_message}")

# Format method for complex cases
template = "Hello, {name}! You are {age} years old and live in {city}."
formatted_template = template.format(**user_data)
print(f"   Template: {formatted_template}")

# Migration pattern 4: Cross-version compatibility utilities
print(f"\n4. Cross-version compatibility check:")
print(f"   Python version: {sys.version_info}")
print(f"   Major version: {sys.version_info.major}")
print(f"   Using future builtins: {'Yes' if 'future' in str(type('')) else 'No'}")

# Check for key differences between Python 2 and 3
compatibility_report = {
    'division_behavior': '/' if 5/2 == 2.5 else '//',
    'default_string_type': type('').__name__,
    'range_type': type(range(1)).__name__,
    'dict_keys_type': type({}.keys()).__name__
}

print("   Compatibility report:")
for feature, value in compatibility_report.items():
    print(f"     {feature}: {value}")

# Migration pattern 5: Production deployment considerations
print(f"\n5. Production deployment considerations:")
print("   ✓ All future imports at module level")
print("   ✓ Standard library aliases installed")
print("   ✓ Unicode strings by default")
print("   ✓ Modern exception handling")
print("   ✓ Cross-version iteration patterns")
print("   ✓ Consistent string formatting")

print("\nProduction migration patterns completed successfully!")
```
