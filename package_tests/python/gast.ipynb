# gast: 5 Basic Usage Examples

```python
# Example 1: Basic gast import and AST conversion
import gast
import ast

# Check gast version and basic info
print(f"gast module: {gast.__name__}")
print(f"gast version: {getattr(gast, '__version__', 'unknown')}")

# Parse code with standard ast
code = "x = [1, 2, 3]; y = x[0]"
std_tree = ast.parse(code)
print(f"Standard AST: {type(std_tree)}")

# Convert to gast AST
gast_tree = gast.ast_to_gast(std_tree)
print(f"GAST AST: {type(gast_tree)}")
print(f"GAST conversion successful: {gast_tree is not None}")

# Test direct parsing
direct_tree = gast.parse(code)
print(f"Direct GAST parse: {type(direct_tree)}")
```

```python
# Example 2: GAST node creation and manipulation
import gast

# Create a simple expression: x + 1
left = gast.Name(id='x', ctx=gast.Load())
right = gast.Constant(value=1)
binop = gast.BinOp(left=left, op=gast.Add(), right=right)

print(f"Created BinOp: {type(binop)}")
print(f"Left operand: {binop.left.id}")
print(f"Right operand: {binop.right.value}")
print(f"Operator: {type(binop.op).__name__}")

# Create assignment: result = x + 1
target = gast.Name(id='result', ctx=gast.Store())
assign = gast.Assign(targets=[target], value=binop)

print(f"Created assignment: {type(assign)}")
print(f"Assignment target: {assign.targets[0].id}")

# Dump the GAST tree
dumped = gast.dump(assign)
print(f"GAST dump: {dumped[:50]}...")
```

```python
# Example 3: GAST function definition
import gast
import ast

# Parse a function with standard ast, then convert to gast
code = "def double(x): return x * 2"
ast_tree = ast.parse(code)
gast_tree = gast.ast_to_gast(ast_tree)

# Get the function definition
func_def = gast_tree.body[0]

print(f"Function: {func_def.name}")
print(f"Args: {len(func_def.args.args)}")
print(f"Body statements: {len(func_def.body)}")
print(f"Function type: {type(func_def)}")
print(f"GAST function definition created successfully")
```

```python
# Example 4: GAST visitor pattern
import gast

class VariableCollector(gast.NodeVisitor):
    def __init__(self):
        self.variables = set()
    
    def visit_Name(self, node):
        if isinstance(node.ctx, gast.Load):
            self.variables.add(node.id)
        self.generic_visit(node)

# Create code to analyze
code = '''
def example(a, b):
    c = a + b
    return c + a
'''

# Parse and visit
tree = gast.parse(code)
visitor = VariableCollector()
visitor.visit(tree)

print(f"Variables found: {sorted(visitor.variables)}")
print(f"Variable count: {len(visitor.variables)}")
```

```python
# Example 5: GAST transformation
import gast

class ConstantDoubler(gast.NodeTransformer):
    def visit_Constant(self, node):
        # Double numeric constants
        if isinstance(node.value, (int, float)):
            return gast.Constant(value=node.value * 2)
        return node

# Create original code
code = 'x = 5; y = 3.14; z = x + y'
tree = gast.parse(code)

print("Original constants:")
for node in gast.walk(tree):
    if isinstance(node, gast.Constant) and isinstance(node.value, (int, float)):
        print(f"  {node.value}")

# Apply transformation
transformer = ConstantDoubler()
new_tree = transformer.visit(tree)

print("Transformed constants:")
for node in gast.walk(new_tree):
    if isinstance(node, gast.Constant) and isinstance(node.value, (int, float)):
        print(f"  {node.value}")

print(f"Transformation completed successfully")
```
