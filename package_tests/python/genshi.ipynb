## Example 1: genshi functionality test 1

```python
from genshi.template import MarkupTemplate
from genshi import HTML

# Example 1: Basic template engine functionality
print("Example 1: Basic template engine functionality")

try:
    # Basic template creation and rendering
    print("Setting up Genshi template examples...")
    
    # Simple template with variables
    template_text = """
    <html>
        <head><title>$title</title></head>
        <body>
            <h1>Welcome $name!</h1>
            <p>Today is $date</p>
        </body>
    </html>
    """
    
    tmpl = MarkupTemplate(template_text)
    print("Template created successfully")
    
    # Template context data
    context = {
        'title': 'Genshi Demo',
        'name': 'Alice',
        'date': '2024-08-27'
    }
    
    # Render template
    stream = tmpl.generate(**context)
    result = stream.render('html', doctype='html')
    
    print(f"Template rendered successfully")
    print(f"Context variables: {list(context.keys())}")
    print(f"Output length: {len(result)} characters")
    
    # Template features
    template_features = [
        "Variable substitution with $variable",
        "Expression evaluation with ${expression}",
        "Control structures (if, for, choose)",
        "Template inheritance and inclusion",
        "Automatic HTML escaping"
    ]
    
    print(f"\nGenshi template features:")
    for feature in template_features:
        print(f"  • {feature}")
    
    # Supported output formats
    output_formats = ['html', 'xhtml', 'xml', 'text']
    
    print(f"\nSupported output formats:")
    for fmt in output_formats:
        print(f"  • {fmt}")
    
    print("Example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"genshi example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: genshi functionality test 2

```python
from genshi.template import MarkupTemplate

# Example 2: Control structures and loops
print("Example 2: Control structures and loops")

try:
    # Template with control structures
    print("Setting up control structure examples...")
    
    # Template with loops and conditionals (single root element)
    control_template = """
    <div xmlns:py="http://genshi.edgewall.org/">
        <ul py:if="items">
            <li py:for="item in items">
                <span py:if="item.priority == 'high'" class="urgent">URGENT:</span>
                ${item.name} - ${item.description}
            </li>
        </ul>
        <p py:if="not items">No items found</p>
    </div>
    """
    
    tmpl = MarkupTemplate(control_template)
    
    # Sample data with different priorities
    items = [
        {'name': 'Task 1', 'description': 'Important task', 'priority': 'high'},
        {'name': 'Task 2', 'description': 'Normal task', 'priority': 'medium'},
        {'name': 'Task 3', 'description': 'Low priority task', 'priority': 'low'}
    ]
    
    # Render with data
    stream = tmpl.generate(items=items)
    result = stream.render('html')
    
    print(f"✓ Control structure template rendered")
    print(f"  Number of items: {len(items)}")
    print(f"  High priority items: {sum(1 for item in items if item['priority'] == 'high')}")
    
    # Control directives
    control_directives = [
        'py:if - Conditional rendering',
        'py:for - Loop iteration', 
        'py:choose/py:when/py:otherwise - Multiple conditions'
    ]
    
    print(f"✓ Control directives:")
    for directive in control_directives:
        print(f"  • {directive}")
    
    print("✓ Example 2 completed successfully")
    
except Exception as e:
    print(f"✗ Runtime error: {e}")
    raise AssertionError(
        f"genshi example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished")
```

## Example 3: genshi functionality test 3

```python
from genshi.template import TextTemplate

# Example 3: Text templates and non-HTML output
print("Example 3: Text templates and non-HTML output")

try:
    # Text template for configuration files
    print("Setting up text template examples...")
    
    # Configuration file template
    config_template = """
# Database Configuration
DB_HOST = $db_host
DB_PORT = $db_port
DB_NAME = $db_name

# Application Settings
#for setting in app_settings
${setting.key} = ${setting.value}
#end

# Feature Flags
#if features
#for feature in features
ENABLE_${feature.upper()} = true
#end
#end
    """
    
    tmpl = TextTemplate(config_template)
    
    # Configuration data
    config_data = {
        'db_host': 'localhost',
        'db_port': 5432,
        'db_name': 'myapp',
        'app_settings': [
            {'key': 'DEBUG', 'value': 'false'},
            {'key': 'SECRET_KEY', 'value': 'abc123'}
        ],
        'features': ['logging', 'caching', 'metrics']
    }
    
    # Generate configuration file
    stream = tmpl.generate(**config_data)
    result = stream.render('text')
    
    print(f"Configuration file generated")
    print(f"Output lines: {len(result.splitlines())}")
    print(f"Features enabled: {len(config_data['features'])}")
    
    # Text template features
    text_features = [
        "Plain text output (no HTML)",
        "Configuration file generation",
        "Email template creation",
        "Script generation",
        "Documentation templates"
    ]
    
    print(f"\nText template use cases:")
    for feature in text_features:
        print(f"  • {feature}")
    
    # Template syntax differences
    syntax_notes = [
        "Uses #if/#for/#end instead of py: attributes",
        "No automatic HTML escaping",
        "Whitespace preservation",
        "Line-based control structures"
    ]
    
    print(f"\nText template syntax:")
    for note in syntax_notes:
        print(f"  • {note}")
    
    print("Example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"genshi example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: genshi functionality test 4

```python
from genshi import HTML
from genshi.filters import HTMLSanitizer
from genshi.template import MarkupTemplate

# Example 4: HTML parsing and filtering
print("Example 4: HTML parsing and filtering")

try:
    # HTML parsing and manipulation
    print("Setting up HTML parsing examples...")
    
    # Parse HTML content
    html_content = """
    <div class="content">
        <h1>Article Title</h1>
        <p>This is a <strong>sample</strong> paragraph with <em>formatting</em>.</p>
        <script>alert('dangerous');</script>
        <a href="http://example.com" onclick="malicious()">Link</a>
    </div>
    """
    
    # Parse HTML
    stream = HTML(html_content)
    print("HTML parsed successfully")
    
    # Apply HTML sanitizer to remove dangerous content
    sanitizer = HTMLSanitizer()
    safe_stream = stream | sanitizer
    
    # Render sanitized HTML
    safe_html = safe_stream.render('html')
    print(f"HTML sanitized successfully")
    print(f"Original length: {len(html_content)}")
    print(f"Sanitized length: {len(safe_html)}")
    
    # HTML filtering capabilities
    filter_features = [
        "Remove dangerous scripts and event handlers",
        "Whitelist allowed HTML tags and attributes", 
        "Remove or escape malicious content",
        "Transform HTML structures",
        "Apply custom filtering rules"
    ]
    
    print(f"\nHTML filtering features:")
    for feature in filter_features:
        print(f"  • {feature}")
    
    # Stream processing chain
    processing_steps = [
        "Parse HTML into stream",
        "Apply filters in sequence",
        "Transform content", 
        "Render final output"
    ]
    
    print(f"\nStream processing workflow:")
    for i, step in enumerate(processing_steps, 1):
        print(f"  {i}. {step}")
    
    # Security considerations
    security_features = [
        "XSS prevention through sanitization",
        "Safe HTML rendering",
        "Script tag removal",
        "Event handler stripping",
        "URL validation"
    ]
    
    print(f"\nSecurity features:")
    for feature in security_features:
        print(f"  • {feature}")
    
    print("Example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"genshi example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: genshi functionality test 5

```python
from genshi.template import MarkupTemplate

# Example 5: Template inheritance and advanced features  
print("Example 5: Template inheritance and advanced features")

try:
    # Template inheritance demonstration
    print("Setting up template inheritance examples...")
    
    # Base template with proper namespace
    base_template = """
    <html xmlns:py="http://genshi.edgewall.org/">
        <head>
            <title py:content="title">Default Title</title>
        </head>
        <body>
            <header><h1>My Website</h1></header>
            <main py:content="content">Default content</main>
            <footer><p>&copy; 2024 My Website</p></footer>
        </body>
    </html>
    """
    
    # Simple template with variable substitution
    page_template = """
    <div xmlns:py="http://genshi.edgewall.org/">
        <h2>${page_title}</h2>
        <p py:for="paragraph in content">
            ${paragraph}
        </p>
        <ul py:if="items">
            <li py:for="item in items">${item}</li>
        </ul>
    </div>
    """
    
    # Create and test templates
    base_tmpl = MarkupTemplate(base_template)
    page_tmpl = MarkupTemplate(page_template)
    
    # Test data
    page_data = {
        'page_title': 'About Us',
        'content': ['Welcome to our site.', 'We provide excellent service.'],
        'items': ['Feature 1', 'Feature 2', 'Feature 3']
    }
    
    # Render page template
    stream = page_tmpl.generate(**page_data)
    result = stream.render('html')
    
    print(f"✓ Template rendered successfully")
    print(f"  Content paragraphs: {len(page_data['content'])}")
    print(f"  Feature items: {len(page_data['items'])}")
    
    # Template features
    template_features = [
        "Variable substitution with ${variable}",
        "Control structures (py:if, py:for)",
        "Template functions and macros",
        "Content blocks and inheritance",
        "Automatic HTML escaping"
    ]
    
    print(f"✓ Template features:")
    for feature in template_features:
        print(f"  • {feature}")
    
    print("✓ Example 5 completed successfully")
    
except Exception as e:
    print(f"✗ Runtime error: {e}")
    raise AssertionError(
        f"genshi example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished")
```
