## Example 2: geodatasets functionality test 2

```python
import geodatasets

# Example 2: Testing dataset access functions
print("Example 2: Testing dataset access functions")

try:
    # Test main access function
    if hasattr(geodatasets, 'get_path'):
        print(f"✓ get_path function available")
        
        # 'naturalearth_lowres' was a geopandas.datasets name, never a geodatasets item: the lookup
        # raises ValueError before any download is attempted.
        try:
            geodatasets.get_path('naturalearth_lowres')
        except ValueError as e:
            assert "No matching item found" in str(e), e
            print(f"  Note: naturalearth_lowres is not a geodatasets dataset: {str(e)[:50]}...")
        else:
            raise AssertionError("get_path('naturalearth_lowres') resolved a name geodatasets does not have")
        
        # A real item resolves OFFLINE to its source URL (get_path() would download the file).
        url = geodatasets.get_url('nybb')
        assert url.endswith(geodatasets.data.query_name('nybb')['filename']), url
        print(f"  ✓ nybb url: {url}")
        print(f"  URL type: {type(url).__name__}")
    
    # Check for other access methods
    if hasattr(geodatasets, 'fetch'):
        print(f"✓ fetch function available")
    
    if hasattr(geodatasets, 'available'):
        try:
            available = geodatasets.available()
            print(f"✓ Available datasets: {len(available) if available else 0}")
        except:
            print(f"✓ available() function exists but may need parameters")
            raise
    
    # Test attribute access for datasets
    dataset_attrs = [attr for attr in dir(geodatasets) 
                    if not attr.startswith('_') and not callable(getattr(geodatasets, attr))]
    
    print(f"✓ Dataset attributes found: {len(dataset_attrs)}")
    if dataset_attrs:
        print(f"  Sample datasets: {dataset_attrs[:3]}")
    
    print("✓ Dataset access testing completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geodatasets example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished\n")
```

## Example 3: geodatasets functionality test 3

```python
import geodatasets
import os

# Example 3: Dataset path handling and file operations
print("Example 3: Dataset path handling and file operations")

try:
    # Test different dataset access patterns
    print(f"✓ Testing dataset path operations")
    
    # Try common dataset names. Resolved OFFLINE: get_url() and data.query_name() read geodatasets'
    # bundled catalogue, while get_path() downloads the file (that is example 5, which declares the
    # network). 'naturalearth_lowres', 'world' and 'countries' were geopandas.datasets names, not
    # geodatasets items, and raise ValueError.
    dataset_names = ['naturalearth_lowres', 'world', 'countries', 'nybb']
    found_datasets = []
    
    for name in dataset_names:
        try:
            url = geodatasets.get_url(name)
        except ValueError as e:
            assert "No matching item found" in str(e), e
            print(f"  - {name}: Not a geodatasets name")
            continue
        dataset = geodatasets.data.query_name(name)
        assert url == dataset['url'] and url.endswith(dataset['filename']), (url, dataset['filename'])
        found_datasets.append(name)
        print(f"  ✓ {name}: Found")
        print(f"    File: {dataset['filename']} ({dataset['nrows']} rows, sha256 {dataset['hash'][:12]}...)")
    
    assert found_datasets == ['nybb'], found_datasets
    print(f"✓ Found {len(found_datasets)} accessible datasets")
    
    # Test direct attribute access
    for attr in dir(geodatasets):
        if not attr.startswith('_') and hasattr(geodatasets, attr):
            obj = getattr(geodatasets, attr)
            if isinstance(obj, str) and (obj.endswith('.zip') or obj.endswith('.json')):
                print(f"  Dataset attribute: {attr} -> {obj[:50]}...")
                break
    
    print("✓ Path handling testing completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geodatasets example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished\n")
```

## Example 4: geodatasets functionality test 4

```python
import geodatasets
import inspect

# Example 4: Package structure and API analysis
print("Example 4: Package structure and API analysis")

try:
    # Analyze geodatasets module structure
    print(f"✓ Analyzing geodatasets package structure")
    
    # Get all public attributes
    all_attrs = dir(geodatasets)
    public_attrs = [attr for attr in all_attrs if not attr.startswith('_')]
    
    # Categorize attributes
    functions = []
    classes = []
    constants = []
    
    for attr in public_attrs:
        obj = getattr(geodatasets, attr)
        if inspect.isfunction(obj):
            functions.append(attr)
        elif inspect.isclass(obj):
            classes.append(attr)
        elif isinstance(obj, (str, int, float, bool)):
            constants.append(attr)
    
    print(f"✓ Package contents:")
    print(f"  Total attributes: {len(public_attrs)}")
    print(f"  Functions: {len(functions)} - {functions[:3]}")
    print(f"  Classes: {len(classes)} - {classes[:3]}")
    print(f"  Constants: {len(constants)} - {constants[:3]}")
    
    # Check main functions
    main_functions = ['get_path', 'fetch', 'available']
    for func in main_functions:
        if func in functions:
            sig = inspect.signature(getattr(geodatasets, func))
            print(f"  {func}{sig}")
    
    # Show module docstring if available
    if geodatasets.__doc__:
        doc_preview = geodatasets.__doc__.strip()[:80]
        print(f"✓ Module doc: {doc_preview}...")
    
    print("✓ Package analysis completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geodatasets example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished\n")
```

## Example 5: geodatasets functionality test 5

```python
import geodatasets
import os
import sys

# Example 5: Practical dataset usage and integration
print("Example 5: Practical dataset usage and integration")

try:
    # Test practical usage patterns
    print(f"✓ Testing practical geodatasets usage")
    
    # Test the most common usage pattern
    try:
        # Try to access a real dataset
        if hasattr(geodatasets, 'get_path'):
            # Test with different possible dataset names. 'naturalearth_lowres' and 'world' were
            # geopandas.datasets names: geodatasets raises ValueError for them before any download.
            test_datasets = ['naturalearth_lowres', 'nybb', 'world']
            success_count = 0
            
            for dataset_name in test_datasets:
                try:
                    path = geodatasets.get_path(dataset_name)
                except ValueError as e:
                    assert "No matching item found" in str(e), e
                    print(f"  - {dataset_name}: not a geodatasets name")
                    continue
                assert os.path.isfile(path), f"not a file: {path}"
                # A zipped item is unpacked beside its archive and get_path() returns the member
                # (measured: .../nybb_16a.zip.unzip/nybb_16a/nybb.shp).
                archive = geodatasets.data.query_name(dataset_name)['filename']
                assert f"{archive}.unzip" in path and path.endswith('.shp'), f"unexpected path for {archive}: {path}"
                print(f"  ✓ Successfully accessed: {dataset_name}")
                print(f"    Path: {path}")
                success_count += 1
                break  # Stop after first success
            assert success_count == 1, "nybb did not download"
        
        # Show integration example
        print(f"✓ Integration pattern:")
        print(f"  import geopandas as gpd")
        print(f"  path = geodatasets.get_path('dataset')")
        print(f"  gdf = gpd.read_file(path)")
        
    except Exception as e:
        print(f"  Dataset access test: {str(e)[:50]}...")
        raise
    
    # Package info
    print(f"✓ Runtime info:")
    print(f"  Python version: {sys.version.split()[0]}")
    print(f"  geodatasets location: {geodatasets.__file__ if hasattr(geodatasets, '__file__') else 'N/A'}")
    
    print("✓ Practical usage testing completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geodatasets example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished\n")
```
