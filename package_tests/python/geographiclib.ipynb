## Example 1: geographiclib functionality test 1

```python
from geographiclib.geodesic import Geodesic

# Example 1: Basic geodesic distance calculations
print("Example 1: Basic geodesic distance calculations")

try:
    # Initialize WGS84 geodesic
    geod = Geodesic.WGS84
    print(f"✓ WGS84 geodesic initialized")
    print(f"  Semi-major axis: {geod.a:.1f} m")
    print(f"  Flattening: {geod.f:.10f}")
    
    # Calculate distance between two points
    # New York City to London
    lat1, lon1 = 40.7128, -74.0060  # NYC
    lat2, lon2 = 51.5074, -0.1278   # London
    
    result = geod.Inverse(lat1, lon1, lat2, lon2)
    distance_km = result['s12'] / 1000  # Convert to kilometers
    azimuth = result['azi1']  # Forward azimuth
    
    print(f"✓ Distance calculation:")
    print(f"  NYC to London: {distance_km:.1f} km")
    print(f"  Initial bearing: {azimuth:.1f}°")
    print(f"  Arc length: {result['a12']:.3f}°")
    
    print("✓ Basic geodesic calculations completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geographiclib example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished\n")
```

## Example 2: geographiclib functionality test 2

```python
from geographiclib.geodesic import Geodesic

# Example 2: Forward geodesic calculations and destination points
print("Example 2: Forward geodesic calculations and destination points")

try:
    # Use WGS84 ellipsoid
    geod = Geodesic.WGS84
    
    # Starting point: San Francisco
    lat1, lon1 = 37.7749, -122.4194
    azimuth = 45.0    # Northeast direction
    distance = 100000  # 100 km
    
    # Calculate destination point
    result = geod.Direct(lat1, lon1, azimuth, distance)
    lat2, lon2 = result['lat2'], result['lon2']
    final_azimuth = result['azi2']
    
    print(f"✓ Forward geodesic calculation:")
    print(f"  Start: {lat1:.4f}°, {lon1:.4f}°")
    print(f"  Direction: {azimuth}° for {distance/1000:.1f} km")
    print(f"  End: {lat2:.4f}°, {lon2:.4f}°")
    print(f"  Final bearing: {final_azimuth:.1f}°")
    
    # Calculate multiple points along the path
    num_points = 5
    for i in range(num_points + 1):
        fraction = i / num_points
        dist = distance * fraction
        point = geod.Direct(lat1, lon1, azimuth, dist)
        print(f"  Point {i}: {point['lat2']:.4f}°, {point['lon2']:.4f}°")
    
    print("✓ Forward calculations completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geographiclib example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished\n")
```

## Example 3: geographiclib functionality test 3

```python
from geographiclib.geodesic import Geodesic
from geographiclib.geodesicline import GeodesicLine

# Example 3: Geodesic line and intermediate points
print("Example 3: Geodesic line and intermediate points")

try:
    # Create geodesic line from Paris to Tokyo
    geod = Geodesic.WGS84
    lat1, lon1 = 48.8566, 2.3522   # Paris
    lat2, lon2 = 35.6762, 139.6503 # Tokyo
    
    # Get the inverse solution first
    inverse = geod.Inverse(lat1, lon1, lat2, lon2)
    azimuth = inverse['azi1']
    distance = inverse['s12']
    
    # Create a geodesic line
    line = geod.Line(lat1, lon1, azimuth)
    
    print(f"✓ Geodesic line Paris to Tokyo:")
    print(f"  Total distance: {distance/1000:.1f} km")
    print(f"  Initial azimuth: {azimuth:.1f}°")
    
    # Calculate points at regular intervals
    num_segments = 4
    print(f"✓ Intermediate points:")
    for i in range(num_segments + 1):
        s = distance * i / num_segments
        pos = line.Position(s)
        print(f"  {i*100//num_segments:3d}%: {pos['lat2']:7.3f}°, {pos['lon2']:8.3f}°")
    
    # Calculate position at specific distance
    halfway = line.Position(distance / 2)
    print(f"✓ Halfway point: {halfway['lat2']:.3f}°, {halfway['lon2']:.3f}°")
    
    print("✓ Geodesic line calculations completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geographiclib example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished\n")
```

## Example 4: geographiclib functionality test 4

```python
from geographiclib.geodesic import Geodesic
from geographiclib.polygonarea import PolygonArea

# Example 4: Polygon area calculations on the ellipsoid
print("Example 4: Polygon area calculations on the ellipsoid")

try:
    # Initialize for area calculations
    geod = Geodesic.WGS84
    poly = PolygonArea(geod, False)  # False = don't compute signed area
    
    # Define a polygon (approximate square around Central Park, NYC)
    # These coordinates form a rough square
    vertices = [
        (40.7829, -73.9731),  # NW corner
        (40.7829, -73.9492),  # NE corner
        (40.7589, -73.9492),  # SE corner
        (40.7589, -73.9731),  # SW corner
    ]
    
    print(f"✓ Calculating polygon area:")
    print(f"  Vertices: {len(vertices)}")
    
    # Add vertices to polygon
    for i, (lat, lon) in enumerate(vertices):
        poly.AddPoint(lat, lon)
        print(f"    {i+1}: {lat:.4f}°, {lon:.4f}°")
    
    # Calculate area and perimeter
    result = poly.Compute()
    area_m2 = abs(result[2])  # Area in square meters
    perimeter_m = result[1]   # Perimeter in meters
    
    area_km2 = area_m2 / 1e6  # Convert to square kilometers
    perimeter_km = perimeter_m / 1000  # Convert to kilometers
    
    print(f"✓ Results:")
    print(f"  Area: {area_km2:.3f} km²")
    print(f"  Perimeter: {perimeter_km:.3f} km")
    
    print("✓ Polygon calculations completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geographiclib example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished\n")
```

## Example 5: geographiclib functionality test 5

```python
from geographiclib.geodesic import Geodesic

# Example 5: Multiple distance calculations and geodesic properties
print("Example 5: Multiple distance calculations and geodesic properties")

try:
    # Use WGS84 ellipsoid
    geod = Geodesic.WGS84
    
    print(f"✓ WGS84 ellipsoid properties:")
    print(f"  Semi-major axis: {geod.a:.1f} m")
    print(f"  Flattening: {geod.f:.10f}")
    print(f"  Semi-minor axis: {geod.a * (1 - geod.f):.1f} m")
    
    # Calculate distances between multiple city pairs
    city_pairs = [
        ("Sydney", (-33.8688, 151.2093), "Melbourne", (-37.8136, 144.9631)),
        ("Los Angeles", (34.0522, -118.2437), "San Francisco", (37.7749, -122.4194)),
        ("Berlin", (52.5200, 13.4050), "Munich", (48.1351, 11.5820)),
    ]
    
    print(f"✓ Distance calculations:")
    for city1, coords1, city2, coords2 in city_pairs:
        result = geod.Inverse(coords1[0], coords1[1], coords2[0], coords2[1])
        distance_km = result['s12'] / 1000
        bearing = result['azi1']
        print(f"  {city1} to {city2}: {distance_km:.1f} km, {bearing:.1f}°")
    
    # Calculate circumference at different latitudes
    latitudes = [0, 30, 60, 90]  # Equator, 30°N, 60°N, North Pole
    print(f"✓ Earth circumferences at different latitudes:")
    for lat in latitudes:
        if lat == 90:
            print(f"  {lat:2d}°: 0.0 km (North Pole)")
        else:
            east_point = geod.Direct(lat, 0, 90, 111320)  # ~1 degree east
            circumference = east_point['lon2'] * 360 * 111.32  # Approximate
            print(f"  {lat:2d}°: ~{circumference:.0f} km")
    
    print("✓ Geodesic properties completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geographiclib example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished\n")
```
