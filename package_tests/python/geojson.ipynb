## Example 1: geojson functionality test 1

```python
import geojson

# Example 1: Creating and working with Point geometries
print("Example 1: Creating and working with Point geometries")

try:
    # Create a Point geometry
    point = geojson.Point((-122.4194, 37.7749))  # San Francisco coordinates
    print(f"✓ Point created: {point}")
    print(f"  Type: {point['type']}")
    print(f"  Coordinates: {point['coordinates']}")
    
    # Create Point from coordinates list
    coords = [-74.0060, 40.7128]  # New York coordinates
    ny_point = geojson.Point(coords)
    print(f"✓ NY Point: {ny_point['coordinates']}")
    
    # Access point properties
    longitude = point['coordinates'][0]
    latitude = point['coordinates'][1]
    print(f"✓ Extracted coordinates:")
    print(f"  Longitude: {longitude}")
    print(f"  Latitude: {latitude}")
    
    # Serialize to JSON string
    json_str = geojson.dumps(point)
    print(f"✓ JSON serialization: {json_str[:50]}...")
    
    # Parse from JSON string
    parsed_point = geojson.loads(json_str)
    print(f"✓ Parsed back: {parsed_point['type']}")
    
    print("✓ Point geometry operations completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geojson example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished\n")
```

## Example 2: geojson functionality test 2

```python
import geojson

# Example 2: Creating Features with properties
print("Example 2: Creating Features with properties")

try:
    # Create a Point geometry
    location = geojson.Point((-122.4194, 37.7749))
    
    # Create properties for the feature
    properties = {
        "name": "San Francisco",
        "population": 873965,
        "state": "California",
        "country": "USA"
    }
    
    # Create a Feature with geometry and properties
    feature = geojson.Feature(geometry=location, properties=properties)
    print(f"✓ Feature created: {feature['type']}")
    print(f"  Geometry type: {feature['geometry']['type']}")
    print(f"  Properties: {len(feature['properties'])} items")
    
    # Access feature components
    print(f"✓ Feature details:")
    print(f"  City: {feature['properties']['name']}")
    print(f"  Population: {feature['properties']['population']:,}")
    print(f"  Coordinates: {feature['geometry']['coordinates']}")
    
    # Create feature with ID
    feature_with_id = geojson.Feature(
        geometry=location,
        properties=properties,
        id="sf-city-001"
    )
    print(f"✓ Feature with ID: {feature_with_id['id']}")
    
    # Serialize feature
    feature_json = geojson.dumps(feature, indent=2)
    print(f"✓ Feature JSON length: {len(feature_json)} characters")
    
    print("✓ Feature creation completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geojson example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished\n")
```

## Example 3: geojson functionality test 3

```python
import geojson

# Example 3: Creating FeatureCollections
print("Example 3: Creating FeatureCollections")

try:
    # Create multiple city features
    cities = [
        {"name": "San Francisco", "coords": (-122.4194, 37.7749), "pop": 873965},
        {"name": "Los Angeles", "coords": (-118.2437, 34.0522), "pop": 3979576},
        {"name": "Seattle", "coords": (-122.3321, 47.6062), "pop": 753675}
    ]
    
    # Create features for each city
    features = []
    for city in cities:
        point = geojson.Point(city["coords"])
        properties = {"name": city["name"], "population": city["pop"]}
        feature = geojson.Feature(geometry=point, properties=properties)
        features.append(feature)
    
    print(f"✓ Created {len(features)} city features")
    
    # Create FeatureCollection
    feature_collection = geojson.FeatureCollection(features)
    print(f"✓ FeatureCollection created: {feature_collection['type']}")
    print(f"  Features count: {len(feature_collection['features'])}")
    
    # Access features in collection
    print(f"✓ Cities in collection:")
    for i, feature in enumerate(feature_collection['features']):
        name = feature['properties']['name']
        pop = feature['properties']['population']
        coords = feature['geometry']['coordinates']
        print(f"  {i+1}. {name}: {pop:,} people at {coords}")
    
    # Serialize collection
    collection_json = geojson.dumps(feature_collection)
    print(f"✓ Collection JSON size: {len(collection_json)} characters")
    
    print("✓ FeatureCollection operations completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geojson example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished\n")
```

## Example 4: geojson functionality test 4

```python
import geojson

# Example 4: Working with LineString and Polygon geometries
print("Example 4: Working with LineString and Polygon geometries")

try:
    # Create a LineString (e.g., a simple route)
    route_coords = [
        (-122.4194, 37.7749),  # San Francisco
        (-122.0838, 37.4220),  # Palo Alto
        (-121.8863, 37.3382)   # San Jose
    ]
    linestring = geojson.LineString(route_coords)
    print(f"✓ LineString created: {linestring['type']}")
    print(f"  Points: {len(linestring['coordinates'])}")
    
    # Create a Polygon (e.g., a park boundary)
    polygon_coords = [[
        (-122.5, 37.7), (-122.3, 37.7),
        (-122.3, 37.8), (-122.5, 37.8),
        (-122.5, 37.7)  # Close the polygon
    ]]
    polygon = geojson.Polygon(polygon_coords)
    print(f"✓ Polygon created: {polygon['type']}")
    print(f"  Rings: {len(polygon['coordinates'])}")
    print(f"  Points in exterior ring: {len(polygon['coordinates'][0])}")
    
    # Create features with different geometries
    route_feature = geojson.Feature(
        geometry=linestring,
        properties={"name": "SF Bay Route", "type": "highway"}
    )
    
    park_feature = geojson.Feature(
        geometry=polygon,
        properties={"name": "Golden Gate Park", "type": "park"}
    )
    
    print(f"✓ Created features:")
    print(f"  Route: {route_feature['properties']['name']}")
    print(f"  Park: {park_feature['properties']['name']}")
    
    # Combine into collection
    collection = geojson.FeatureCollection([route_feature, park_feature])
    print(f"✓ Collection with mixed geometries: {len(collection['features'])} features")
    
    print("✓ Complex geometry operations completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geojson example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished\n")
```

## Example 5: geojson functionality test 5

```python
import geojson
import json

# Example 5: Validation and JSON parsing
print("Example 5: Validation and JSON parsing")

try:
    # Create a valid GeoJSON object
    valid_point = geojson.Point((-122.4194, 37.7749))
    print(f"✓ Valid Point created: {valid_point['type']}")
    
    # Test validation (if available)
    try:
        is_valid = geojson.is_valid(valid_point)
        print(f"✓ Point validation: {is_valid}")
    except AttributeError:
        print("✓ Validation method not available")
    
    # Parse from JSON string
    json_string = '{"type": "Point", "coordinates": [-74.0060, 40.7128]}'
    parsed_geojson = geojson.loads(json_string)
    print(f"✓ Parsed from JSON: {parsed_geojson['type']}")
    print(f"  Coordinates: {parsed_geojson['coordinates']}")
    
    # Convert to regular Python dict and back
    geojson_dict = dict(valid_point)
    print(f"✓ Converted to dict: {type(geojson_dict).__name__}")
    
    # Use different serialization options
    compact_json = geojson.dumps(valid_point)
    formatted_json = geojson.dumps(valid_point, indent=2)
    
    print(f"✓ Serialization options:")
    print(f"  Compact JSON: {len(compact_json)} chars")
    print(f"  Formatted JSON: {len(formatted_json)} chars")
    
    # Parse complex GeoJSON
    complex_json = '''{"type": "Feature", "properties": {"name": "Test"}, 
                      "geometry": {"type": "Point", "coordinates": [0, 0]}}'''
    complex_obj = geojson.loads(complex_json)
    print(f"✓ Complex object: {complex_obj['type']}")
    print(f"  Property: {complex_obj['properties']['name']}")
    
    print("✓ Validation and parsing completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geojson example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished\n")
```
