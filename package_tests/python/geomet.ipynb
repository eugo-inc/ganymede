## Example 1: geomet functionality test 1

```python
from geomet import wkt

# Example 1: Converting GeoJSON to WKT format
print("Example 1: Converting GeoJSON to WKT format")

try:
    # GeoJSON Point geometry
    geojson_point = {
        "type": "Point",
        "coordinates": [-122.4194, 37.7749]  # San Francisco
    }
    
    # Convert to WKT
    wkt_point = wkt.dumps(geojson_point)
    print(f"✓ GeoJSON Point converted to WKT:")
    print(f"  GeoJSON: {geojson_point}")
    print(f"  WKT: {wkt_point}")
    
    # GeoJSON LineString geometry
    geojson_line = {
        "type": "LineString",
        "coordinates": [
            [-122.4194, 37.7749],
            [-122.0838, 37.4220],
            [-121.8863, 37.3382]
        ]
    }
    
    wkt_line = wkt.dumps(geojson_line)
    print(f"✓ LineString conversion:")
    print(f"  Points: {len(geojson_line['coordinates'])}")
    print(f"  WKT: {wkt_line}")
    
    # Test with different precision
    wkt_precise = wkt.dumps(geojson_point, decimals=6)
    print(f"✓ High precision WKT: {wkt_precise}")
    
    print("✓ GeoJSON to WKT conversion completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geomet example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished\n")
```

## Example 2: geomet functionality test 2

```python
from geomet import wkt

# Example 2: Converting WKT to GeoJSON format
print("Example 2: Converting WKT to GeoJSON format")

try:
    # WKT Point string
    wkt_point = "POINT (-122.4194 37.7749)"
    
    # Convert to GeoJSON
    geojson_point = wkt.loads(wkt_point)
    print(f"✓ WKT Point converted to GeoJSON:")
    print(f"  WKT: {wkt_point}")
    print(f"  GeoJSON: {geojson_point}")
    print(f"  Type: {geojson_point['type']}")
    print(f"  Coordinates: {geojson_point['coordinates']}")
    
    # WKT Polygon string
    wkt_polygon = "POLYGON ((-122.5 37.7, -122.3 37.7, -122.3 37.8, -122.5 37.8, -122.5 37.7))"
    geojson_polygon = wkt.loads(wkt_polygon)
    print(f"✓ Polygon conversion:")
    print(f"  WKT rings: 1")
    print(f"  GeoJSON type: {geojson_polygon['type']}")
    print(f"  Coordinate rings: {len(geojson_polygon['coordinates'])}")
    print(f"  Points in ring: {len(geojson_polygon['coordinates'][0])}")
    
    # WKT MultiPoint
    wkt_multipoint = "MULTIPOINT ((-122.4 37.7), (-121.9 37.4), (-121.5 37.2))"
    geojson_multipoint = wkt.loads(wkt_multipoint)
    print(f"✓ MultiPoint: {len(geojson_multipoint['coordinates'])} points")
    
    print("✓ WKT to GeoJSON conversion completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geomet example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished\n")
```

## Example 3: geomet functionality test 3

```python
from geomet import wkt

# Example 3: Working with complex polygon geometries
print("Example 3: Working with complex polygon geometries")

try:
    # GeoJSON Polygon with hole (donut shape)
    geojson_polygon_with_hole = {
        "type": "Polygon",
        "coordinates": [
            # Outer ring
            [[-122.5, 37.7], [-122.3, 37.7], [-122.3, 37.8], [-122.5, 37.8], [-122.5, 37.7]],
            # Inner ring (hole)
            [[-122.45, 37.72], [-122.35, 37.72], [-122.35, 37.78], [-122.45, 37.78], [-122.45, 37.72]]
        ]
    }
    
    # Convert to WKT
    wkt_polygon_hole = wkt.dumps(geojson_polygon_with_hole)
    print(f"✓ Polygon with hole conversion:")
    print(f"  Rings: {len(geojson_polygon_with_hole['coordinates'])}")
    print(f"  WKT: {wkt_polygon_hole}")
    
    # Convert back to GeoJSON
    back_to_geojson = wkt.loads(wkt_polygon_hole)
    print(f"✓ Round-trip conversion:")
    print(f"  Original rings: {len(geojson_polygon_with_hole['coordinates'])}")
    print(f"  Converted rings: {len(back_to_geojson['coordinates'])}")
    
    # Multi-polygon WKT
    wkt_multipolygon = "MULTIPOLYGON (((-122.5 37.7, -122.3 37.7, -122.3 37.8, -122.5 37.8, -122.5 37.7)), ((-121.9 37.4, -121.7 37.4, -121.7 37.5, -121.9 37.5, -121.9 37.4)))"
    geojson_multipolygon = wkt.loads(wkt_multipolygon)
    print(f"✓ MultiPolygon conversion:")
    print(f"  Type: {geojson_multipolygon['type']}")
    print(f"  Polygons: {len(geojson_multipolygon['coordinates'])}")
    
    print("✓ Complex polygon operations completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geomet example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished\n")
```

## Example 4: geomet functionality test 4

```python
from geomet import wkb
import binascii

# Example 4: Working with Well-Known Binary (WKB) format
print("Example 4: Working with Well-Known Binary (WKB) format")

try:
    # GeoJSON Point for WKB conversion
    geojson_point = {
        "type": "Point",
        "coordinates": [-122.4194, 37.7749]
    }
    
    # Convert GeoJSON to WKB
    wkb_data = wkb.dumps(geojson_point)
    print(f"✓ GeoJSON to WKB conversion:")
    print(f"  GeoJSON: {geojson_point}")
    print(f"  WKB length: {len(wkb_data)} bytes")
    print(f"  WKB hex: {binascii.hexlify(wkb_data)[:32]}...")
    
    # Convert WKB back to GeoJSON
    back_to_geojson = wkb.loads(wkb_data)
    print(f"✓ WKB to GeoJSON conversion:")
    print(f"  Type: {back_to_geojson['type']}")
    print(f"  Coordinates: {back_to_geojson['coordinates']}")
    
    # Test with LineString
    geojson_line = {
        "type": "LineString",
        "coordinates": [[-122.4, 37.7], [-122.3, 37.8], [-122.2, 37.9]]
    }
    
    wkb_line = wkb.dumps(geojson_line)
    back_to_line = wkb.loads(wkb_line)
    print(f"✓ LineString WKB round-trip:")
    print(f"  Original points: {len(geojson_line['coordinates'])}")
    print(f"  Converted points: {len(back_to_line['coordinates'])}")
    print(f"  WKB size: {len(wkb_line)} bytes")
    
    print("✓ WKB format operations completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geomet example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished\n")
```

## Example 5: geomet functionality test 5

```python
from geomet import wkt, wkb
import geomet

# Example 5: Format validation and error handling
print("Example 5: Format validation and error handling")

try:
    # Test valid conversions
    valid_geojson = {"type": "Point", "coordinates": [0, 0]}
    valid_wkt = "POINT (0 0)"
    
    print(f"✓ Valid format tests:")
    
    # WKT round-trip
    wkt_result = wkt.dumps(valid_geojson)
    wkt_back = wkt.loads(wkt_result)
    print(f"  WKT round-trip: {wkt_back['type']}")
    
    # WKB round-trip  
    wkb_result = wkb.dumps(valid_geojson)
    wkb_back = wkb.loads(wkb_result)
    print(f"  WKB round-trip: {wkb_back['type']}")
    
    # Test error handling
    print(f"✓ Error handling tests:")
    
    try:
        # Invalid WKT
        wkt.loads("INVALID WKT STRING")
    except ValueError as e:
        assert str(e) == "Unsupported geometry type 'INVALID'", e
        print(f"  Invalid WKT handled: {type(e).__name__}")
    else:
        raise AssertionError("expected ValueError for invalid WKT")
    
    try:
        # Invalid GeoJSON structure
        wkt.dumps({"type": "InvalidType", "coordinates": []})
    except ValueError as e:
        assert str(e) == "Unsupported geometry type 'InvalidType'", e
        print(f"  Invalid GeoJSON handled: {type(e).__name__}")
    else:
        raise AssertionError("expected ValueError for an unknown GeoJSON type")
    
    # Test different coordinate precisions
    high_precision = {"type": "Point", "coordinates": [-122.41941234, 37.77494567]}
    wkt_precise = wkt.dumps(high_precision, decimals=8)
    wkt_rounded = wkt.dumps(high_precision, decimals=2)
    
    print(f"✓ Precision control:")
    print(f"  High: {wkt_precise}")
    print(f"  Low:  {wkt_rounded}")
    
    print("✓ Validation and error handling completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geomet example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished\n")
```
