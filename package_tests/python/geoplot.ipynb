## Example 1: geoplot functionality test 1

```python
import geoplot as gplt
import geopandas as gpd
from shapely.geometry import Point
import matplotlib.pyplot as plt

# Example 1: Creating point plots with geoplot
print("Example 1: Creating point plots with geoplot")

try:
    # Create sample city data
    cities = gpd.GeoDataFrame([
        {'name': 'San Francisco', 'population': 873965, 'geometry': Point(-122.4194, 37.7749)},
        {'name': 'Los Angeles', 'population': 3979576, 'geometry': Point(-118.2437, 34.0522)},
        {'name': 'Seattle', 'population': 753675, 'geometry': Point(-122.3321, 47.6062)},
        {'name': 'Portland', 'population': 652503, 'geometry': Point(-122.6784, 45.5152)},
        {'name': 'San Diego', 'population': 1423851, 'geometry': Point(-117.1611, 32.7157)}
    ], crs='EPSG:4326')
    
    print(f"✓ Created dataset with {len(cities)} cities")
    print(f"  Population range: {cities['population'].min():,} - {cities['population'].max():,}")
    
    # Create a basic point plot
    fig, ax = plt.subplots(1, 1, figsize=(8, 6))
    gplt.pointplot(cities, ax=ax)
    ax.set_title("West Coast Cities - Basic Point Plot")
    plt.close()  # Close to avoid display in headless environment
    
    print("✓ Basic pointplot created successfully")
    
    # Create a scaled point plot based on population
    fig, ax = plt.subplots(1, 1, figsize=(8, 6))
    gplt.pointplot(cities, scale='population', limits=(10, 100), ax=ax)
    ax.set_title("Cities Scaled by Population")
    plt.close()
    
    print("✓ Scaled pointplot created successfully")
    print("✓ Point plotting functionality demonstrated")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geoplot example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished\n")
```

## Example 2: geoplot functionality test 2

```python
import geoplot as gplt
import geopandas as gpd
from shapely.geometry import Polygon
import matplotlib.pyplot as plt

# Example 2: Creating choropleth maps
print("Example 2: Creating choropleth maps")

try:
    # Create sample state-like regions with data
    regions = gpd.GeoDataFrame([
        {'name': 'Region A', 'unemployment': 5.2, 'geometry': 
         Polygon([(-125, 35), (-120, 35), (-120, 40), (-125, 40), (-125, 35)])},
        {'name': 'Region B', 'unemployment': 3.8, 'geometry': 
         Polygon([(-120, 35), (-115, 35), (-115, 40), (-120, 40), (-120, 35)])},
        {'name': 'Region C', 'unemployment': 7.1, 'geometry': 
         Polygon([(-125, 40), (-120, 40), (-120, 45), (-125, 45), (-125, 40)])},
        {'name': 'Region D', 'unemployment': 4.5, 'geometry': 
         Polygon([(-120, 40), (-115, 40), (-115, 45), (-120, 45), (-120, 40)])}
    ], crs='EPSG:4326')
    
    print(f"✓ Created {len(regions)} regions for choropleth")
    print(f"  Unemployment range: {regions['unemployment'].min():.1f}% - {regions['unemployment'].max():.1f}%")
    
    # Create choropleth map
    fig, ax = plt.subplots(1, 1, figsize=(10, 8))
    gplt.choropleth(regions, hue='unemployment', cmap='Reds', ax=ax)
    ax.set_title("Regional Unemployment Rates")
    plt.close()
    
    print("✓ Basic choropleth created successfully")
    
    # Create choropleth with different color scheme
    fig, ax = plt.subplots(1, 1, figsize=(10, 8))
    gplt.choropleth(regions, hue='unemployment', scheme='quantiles', cmap='viridis', ax=ax)
    ax.set_title("Unemployment - Quantile Classification")
    plt.close()
    
    print("✓ Classified choropleth created successfully")
    print("✓ Choropleth mapping functionality demonstrated")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geoplot example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished\n")
```

## Example 3: geoplot functionality test 3

```python
import geoplot as gplt
import geopandas as gpd
from shapely.geometry import Point, Polygon
import matplotlib.pyplot as plt

# Example 3: KDE and density mapping
print("Example 3: KDE and density mapping")

try:
    # Create sample economic data
    economic_centers = gpd.GeoDataFrame([
        {'city': 'Tech Hub', 'gdp': 150000, 'geometry': Point(-122.0, 37.5)},
        {'city': 'Finance Center', 'gdp': 200000, 'geometry': Point(-118.0, 34.0)},
        {'city': 'Manufacturing', 'gdp': 75000, 'geometry': Point(-121.0, 45.5)},
        {'city': 'Agriculture', 'gdp': 45000, 'geometry': Point(-119.0, 36.0)},
        {'city': 'Tourism', 'gdp': 60000, 'geometry': Point(-117.0, 32.7)}
    ], crs='EPSG:4326')
    
    print(f"✓ Created economic centers dataset")
    print(f"  GDP range: ${economic_centers['gdp'].min():,} - ${economic_centers['gdp'].max():,}M")
    
    # Create KDE plot (kernel density estimation)
    fig, ax = plt.subplots(1, 1, figsize=(10, 8))
    gplt.kdeplot(economic_centers, ax=ax)
    gplt.pointplot(economic_centers, ax=ax, color='red', s=50)
    ax.set_title("Economic Centers - Density Distribution")
    plt.close()
    
    print("✓ KDE plot created successfully")
    
    # Create cartogram (area-based visualization)
    regions_for_cartogram = gpd.GeoDataFrame([
        {'name': 'Area 1', 'value': 100, 'geometry': 
         Polygon([(-123, 36), (-121, 36), (-121, 38), (-123, 38), (-123, 36)])},
        {'name': 'Area 2', 'value': 200, 'geometry': 
         Polygon([(-119, 33), (-117, 33), (-117, 35), (-119, 35), (-119, 33)])},
        {'name': 'Area 3', 'value': 50, 'geometry': 
         Polygon([(-122, 44), (-120, 44), (-120, 46), (-122, 46), (-122, 44)])}
    ], crs='EPSG:4326')
    
    fig, ax = plt.subplots(1, 1, figsize=(10, 8))
    gplt.cartogram(regions_for_cartogram, scale='value', ax=ax)
    ax.set_title("Value-Scaled Cartogram")
    plt.close()
    
    print("✓ Cartogram created successfully")
    print("✓ Density and cartogram visualizations demonstrated")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geoplot example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished\n")
```

## Example 4: geoplot functionality test 4

```python
import geoplot as gplt
import geopandas as gpd
from shapely.geometry import Point, LineString
import matplotlib.pyplot as plt

# Example 4: Sankey diagrams and flow maps
print("Example 4: Sankey diagrams and flow maps")

try:
    # Create origin points
    origins = gpd.GeoDataFrame([
        {'city': 'City A', 'population': 100000, 'geometry': Point(-122, 37)},
        {'city': 'City B', 'population': 150000, 'geometry': Point(-118, 34)},
        {'city': 'City C', 'population': 80000, 'geometry': Point(-121, 45)}
    ], crs='EPSG:4326')
    
    # Create destination points
    destinations = gpd.GeoDataFrame([
        {'destination': 'Hub 1', 'capacity': 200000, 'geometry': Point(-119, 36)},
        {'destination': 'Hub 2', 'capacity': 180000, 'geometry': Point(-120, 39)}
    ], crs='EPSG:4326')
    
    print(f"✓ Created {len(origins)} origins and {len(destinations)} destinations")
    
    # Create flow lines
    flows = gpd.GeoDataFrame([
        {'origin': 'City A', 'dest': 'Hub 1', 'flow': 50000,
         'geometry': LineString([(-122, 37), (-119, 36)])},
        {'origin': 'City B', 'dest': 'Hub 1', 'flow': 75000,
         'geometry': LineString([(-118, 34), (-119, 36)])},
        {'origin': 'City C', 'dest': 'Hub 2', 'flow': 40000,
         'geometry': LineString([(-121, 45), (-120, 39)])}
    ], crs='EPSG:4326')
    
    # Create Sankey diagram
    fig, ax = plt.subplots(1, 1, figsize=(12, 8))
    
    # Plot base map with destinations
    gplt.polyplot(destinations.geometry.buffer(0.5), facecolor='lightblue', ax=ax)
    gplt.pointplot(destinations, ax=ax, color='blue', s=100)
    
    # Plot flow lines scaled by flow amount
    gplt.sankey(flows, scale='flow', limits=(1, 10), ax=ax)
    gplt.pointplot(origins, ax=ax, color='red', s=50)
    
    ax.set_title("Migration Flows - Sankey Diagram")
    plt.close()
    
    print("✓ Sankey diagram created successfully")
    print(f"  Total flow volume: {flows['flow'].sum():,}")
    print("✓ Flow mapping functionality demonstrated")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geoplot example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished\n")
```

## Example 5: geoplot functionality test 5

```python
import geoplot as gplt
import geopandas as gpd
from shapely.geometry import Point, Polygon
import matplotlib.pyplot as plt

# Example 5: Multiple plot types and styling
print("Example 5: Multiple plot types and styling")

try:
    # Create sample regions
    sample_regions = gpd.GeoDataFrame([
        {'region': 'North', 'population': 50000000, 'geometry': 
         Polygon([(-130, 40), (-110, 40), (-110, 50), (-130, 50), (-130, 40)])},
        {'region': 'Central', 'population': 75000000, 'geometry': 
         Polygon([(-110, 30), (-90, 30), (-90, 40), (-110, 40), (-110, 30)])},
        {'region': 'South', 'population': 30000000, 'geometry': 
         Polygon([(-90, 20), (-70, 20), (-70, 30), (-90, 30), (-90, 20)])}
    ], crs='EPSG:4326')
    
    print(f"✓ Created sample regions with {len(sample_regions)} areas")
    print(f"  Population range: {sample_regions['population'].min()/1e6:.0f}M - {sample_regions['population'].max()/1e6:.0f}M")
    
    # Create styled polyplot
    fig, ax = plt.subplots(1, 1, figsize=(12, 8))
    gplt.polyplot(sample_regions, facecolor='lightblue', edgecolor='navy', 
                 linewidth=2, ax=ax)
    ax.set_title("Regional Boundaries - Styled Polygons")
    plt.close()
    
    print("✓ Styled polygon plot created successfully")
    
    # Create combined visualization with points and polygons
    fig, ax = plt.subplots(1, 1, figsize=(12, 8))
    
    # Plot regions as base
    gplt.polyplot(sample_regions, facecolor='lightgray', edgecolor='black', 
                 linewidth=1, ax=ax)
    
    # Add major cities as points
    major_cities = gpd.GeoDataFrame([
        {'city': 'City A', 'population': 2000000, 'geometry': Point(-120, 45)},
        {'city': 'City B', 'population': 1500000, 'geometry': Point(-100, 35)},
        {'city': 'City C', 'population': 1800000, 'geometry': Point(-80, 25)}
    ], crs='EPSG:4326')
    
    gplt.pointplot(major_cities, scale='population', limits=(50, 200), 
                  color='red', ax=ax)
    
    ax.set_title("Regions with Major Cities")
    ax.set_facecolor('lightblue')
    plt.close()
    
    print("✓ Combined visualization created successfully")
    print("✓ Styling and multi-layer plotting demonstrated")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geoplot example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished\n")
```
