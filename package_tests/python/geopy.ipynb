## Example 1: geopy functionality test 1

```python
from geopy.geocoders import Nominatim
from geopy import Point

# Example 1: Basic geocoding with Nominatim
print("Example 1: Basic geocoding with Nominatim")

try:
    # Initialize geocoder
    geolocator = Nominatim(user_agent="geopy-example")
    print(f"✓ Nominatim geocoder initialized")
    
    # Geocode an address
    location = geolocator.geocode("1600 Amphitheatre Parkway, Mountain View, CA")
    if location:
        print(f"✓ Geocoding successful:")
        print(f"  Address: {location.address}")
        print(f"  Coordinates: {location.latitude:.4f}, {location.longitude:.4f}")
        print(f"  Raw data keys: {len(location.raw)}")
    else:
        print("✓ Geocoding returned no results (expected for demo)")
    
    # Test with a simple city name
    simple_location = geolocator.geocode("San Francisco, CA")
    if simple_location:
        print(f"✓ Simple geocoding:")
        print(f"  City: San Francisco")
        print(f"  Coordinates: {simple_location.latitude:.4f}, {simple_location.longitude:.4f}")
    
    # Create Point objects
    point = Point(37.7749, -122.4194)
    print(f"✓ Point created: {point}")
    print(f"  Latitude: {point.latitude}")
    print(f"  Longitude: {point.longitude}")
    
    print("✓ Basic geocoding functionality demonstrated")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geopy example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished\n")
```

## Example 2: geopy functionality test 2

```python
from geopy.geocoders import Nominatim
from geopy import Point
import time

# Example 2: Reverse geocoding
print("Example 2: Reverse geocoding")

try:
    # Initialize geocoder
    geolocator = Nominatim(user_agent="geopy-example")
    
    # Known coordinates for testing
    test_coordinates = [
        (37.7749, -122.4194),  # San Francisco
        (40.7128, -74.0060),   # New York
        (34.0522, -118.2437)   # Los Angeles
    ]
    
    print(f"✓ Testing reverse geocoding for {len(test_coordinates)} locations")
    
    expected_cities = ["San Francisco", "New York", "Los Angeles"]
    for i, (lat, lon) in enumerate(test_coordinates):
        time.sleep(1)  # Nominatim's usage policy: at most one request a second
        # Reverse geocode coordinates; a failed lookup raises and fails this cell
        location = geolocator.reverse(f"{lat}, {lon}")
        assert location is not None, f"Nominatim found no address for {lat}, {lon}"
        address = location.address
        print(f"✓ Location {i+1}:")
        print(f"  Coordinates: {lat:.4f}, {lon:.4f}")
        print(f"  Address: {address[:60]}...")
        assert expected_cities[i] in address, (expected_cities[i], address)
    
    # Test with Point objects
    point = Point(37.7749, -122.4194)
    time.sleep(1)
    reverse_result = geolocator.reverse(point)
    assert reverse_result is not None and "San Francisco" in reverse_result.address, reverse_result
    print(f"✓ Point reverse geocoding successful")
    
    print("✓ Reverse geocoding functionality demonstrated")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geopy example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished\n")
```

## Example 3: geopy functionality test 3

```python
from geopy.distance import geodesic, great_circle
from geopy import Point

# Example 3: Distance calculations
print("Example 3: Distance calculations")

try:
    # Define some major cities
    cities = {
        'San Francisco': (37.7749, -122.4194),
        'New York': (40.7128, -74.0060),
        'Los Angeles': (34.0522, -118.2437),
        'Chicago': (41.8781, -87.6298)
    }
    
    print(f"✓ Calculating distances between {len(cities)} cities")
    
    # Calculate distances using geodesic (more accurate)
    sf = cities['San Francisco']
    ny = cities['New York']
    la = cities['Los Angeles']
    
    # Geodesic distance (accounts for Earth's curvature)
    distance_sf_ny = geodesic(sf, ny)
    distance_sf_la = geodesic(sf, la)
    
    print(f"✓ Geodesic distances from San Francisco:")
    print(f"  To New York: {distance_sf_ny.kilometers:.0f} km ({distance_sf_ny.miles:.0f} miles)")
    print(f"  To Los Angeles: {distance_sf_la.kilometers:.0f} km ({distance_sf_la.miles:.0f} miles)")
    
    # Great circle distance (spherical approximation)
    gc_distance = great_circle(sf, ny)
    print(f"  Great circle SF-NY: {gc_distance.kilometers:.0f} km")
    
    # Distance with Point objects
    point1 = Point(sf)
    point2 = Point(ny)
    point_distance = geodesic(point1, point2)
    print(f"✓ Point-based distance: {point_distance.kilometers:.0f} km")
    
    # Calculate bearing
    # API drift: `geodesic` exposes distances only — there is no `.bearing`. The initial bearing
    # comes from the underlying geographiclib solver, which geopy already carries as `.geod`.
    bearing = distance_sf_ny.geod.Inverse(
        sf[0], sf[1], ny[0], ny[1]
    )['azi1']
    print(f"✓ Initial bearing SF to NY: {bearing:.1f}°")
    
    print("✓ Distance calculation functionality demonstrated")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geopy example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished\n")
```

## Example 4: geopy functionality test 4

```python
from geopy.distance import geodesic
from geopy import Point

# Example 4: Advanced distance operations and Point manipulation
print("Example 4: Advanced distance operations and Point manipulation")

try:
    # Starting point
    start_point = Point(37.7749, -122.4194)  # San Francisco
    print(f"✓ Starting point: {start_point}")
    
    # Calculate destination point given distance and bearing
    distance = geodesic(kilometers=100)  # 100 km
    bearing = 45  # Northeast
    
    destination = distance.destination(start_point, bearing)
    print(f"✓ Destination point (100km at 45°):")
    print(f"  Coordinates: {destination.latitude:.4f}, {destination.longitude:.4f}")
    
    # Verify the distance
    calculated_distance = geodesic(start_point, destination)
    print(f"  Verification: {calculated_distance.kilometers:.1f} km")
    
    # Calculate multiple points along a path
    waypoints = []
    for i in range(0, 101, 25):  # 0, 25, 50, 75, 100 km
        partial_distance = geodesic(kilometers=i)
        waypoint = partial_distance.destination(start_point, bearing)
        waypoints.append(waypoint)
    
    print(f"✓ Generated {len(waypoints)} waypoints:")
    for i, point in enumerate(waypoints):
        km = i * 25
        print(f"  {km:3d}km: {point.latitude:.3f}, {point.longitude:.3f}")
    
    # Calculate area of a simple triangle
    p1 = Point(37.0, -122.0)
    p2 = Point(38.0, -122.0)
    p3 = Point(37.5, -121.0)
    
    # Perimeter calculation
    side1 = geodesic(p1, p2).kilometers
    side2 = geodesic(p2, p3).kilometers
    side3 = geodesic(p3, p1).kilometers
    perimeter = side1 + side2 + side3
    
    print(f"✓ Triangle perimeter: {perimeter:.1f} km")
    
    print("✓ Advanced distance operations demonstrated")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geopy example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished\n")
```

## Example 5: geopy functionality test 5

```python
from geopy.geocoders import Nominatim
from geopy.exc import GeocoderTimedOut, GeocoderServiceError
from geopy import Point
import time

# Example 5: Error handling and geocoder configuration
print("Example 5: Error handling and geocoder configuration")

try:
    # Initialize geocoder with custom settings
    geolocator = Nominatim(
        user_agent="geopy-test-app",
        timeout=10,
        scheme='https'
    )
    
    print(f"✓ Geocoder configured with custom settings")
    print(f"  User agent: geopy-test-app")
    print(f"  Timeout: 10 seconds")
    
    # Test addresses that might fail or succeed
    test_addresses = [
        "1600 Pennsylvania Avenue, Washington DC",
        "Times Square, New York",
        "Invalid Address That Doesn't Exist 12345"
    ]
    
    print(f"✓ Testing {len(test_addresses)} addresses with error handling:")
    
    for i, address in enumerate(test_addresses):
        try:
            # Add small delay to be respectful to the service
            time.sleep(0.1)
            
            location = geolocator.geocode(address, timeout=5)
            
            if location:
                print(f"  {i+1}. ✓ '{address[:30]}...'")
                print(f"     → {location.latitude:.3f}, {location.longitude:.3f}")
            else:
                print(f"  {i+1}. ✓ '{address[:30]}...' → No results found")
                
        except GeocoderTimedOut:
            print(f"  {i+1}. ⏱ '{address[:30]}...' → Timeout")
        except GeocoderServiceError as e:
            print(f"  {i+1}. ✗ '{address[:30]}...' → Service error")
        except Exception as e:
            print(f"  {i+1}. ✗ '{address[:30]}...' → {type(e).__name__}")
            raise
    
    # Test valid Point creation and validation
    valid_points = [
        (37.7749, -122.4194),  # San Francisco
        (0, 0),                # Equator/Prime Meridian
        (89.9, 179.9)          # Near North Pole
    ]
    
    print(f"✓ Point validation tests:")
    for lat, lon in valid_points:
        try:
            point = Point(lat, lon)
            is_valid = (-90 <= lat <= 90) and (-180 <= lon <= 180)
            print(f"  ({lat}, {lon}) → Valid Point: {is_valid}")
        except Exception as e:
            print(f"  ({lat}, {lon}) → Error: {type(e).__name__}")
            raise
    
    print("✓ Error handling and configuration demonstrated")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geopy example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished\n")
```
