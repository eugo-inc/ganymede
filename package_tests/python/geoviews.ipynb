## Example 1: geoviews functionality test 1

```python
import geoviews as gv
import numpy as np
import pandas as pd

# Example 1: Creating basic geographic points visualization
print("Example 1: Creating basic geographic points visualization")

try:
    # Enable extension for display
    gv.extension('matplotlib')
    
    # Create sample city data
    cities_data = {
        'lon': [-122.4194, -118.2437, -122.3321, -121.4694],
        'lat': [37.7749, 34.0522, 47.6062, 36.7783],
        'city': ['San Francisco', 'Los Angeles', 'Seattle', 'Fresno'],
        'population': [873965, 3979576, 753675, 542107]
    }
    
    df = pd.DataFrame(cities_data)
    print(f"✓ Created dataset with {len(df)} cities")
    
    # Create Points element
    points = gv.Points(df, kdims=['lon', 'lat'], vdims=['city', 'population'])
    print(f"✓ Points element created: {type(points).__name__}")
    print(f"  Dimensions: {len(points.kdims)} key, {len(points.vdims)} value")
    print(f"  Data shape: {len(points.data)} points")
    
    # Add basic styling options (using valid matplotlib options)
    styled_points = points.opts(color='blue', alpha=0.7, fig_size=300)
    print(f"✓ Applied styling with matplotlib backend")
    
    print("✓ Basic geographic points visualization created")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geoviews example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished\n")
```

## Example 2: geoviews functionality test 2

```python
import geoviews as gv
import numpy as np
import pandas as pd

# Example 2: Working with polygons and geographic shapes
print("Example 2: Working with polygons and geographic shapes")

try:
    # Create sample polygon data (simple rectangles representing regions)
    # Region 1: Northern California
    coords1 = [(-125, 39), (-120, 39), (-120, 42), (-125, 42), (-125, 39)]
    poly1 = gv.Polygons([coords1])
    
    # Region 2: Central California  
    coords2 = [(-122, 35), (-118, 35), (-118, 38), (-122, 38), (-122, 35)]
    poly2 = gv.Polygons([coords2])
    
    print(f"✓ Created 2 polygon regions")
    print(f"  Region 1 coordinates: {len(coords1)} points")
    print(f"  Region 2 coordinates: {len(coords2)} points")
    
    # Combine polygons into overlay
    all_polygons = poly1 * poly2  # Using * operator for overlay
    print(f"✓ Polygons combined: {type(all_polygons).__name__}")
    
    # Apply styling to individual polygons
    styled_poly1 = poly1.opts(alpha=0.6, color='lightblue')
    styled_poly2 = poly2.opts(alpha=0.6, color='lightgreen')
    print(f"✓ Applied different colors to each polygon")
    
    # Create combined styled visualization
    styled_combined = styled_poly1 * styled_poly2
    print(f"✓ Created styled combined visualization")
    
    # Check data structure instead of bounds
    print(f"  Poly1 data shape: {len(poly1.data)}")
    print(f"  Poly2 data shape: {len(poly2.data)}")
    print(f"  Combined elements: {len(all_polygons)}")
    
    print("✓ Geographic polygon visualization completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geoviews example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished\n")
```

## Example 3: geoviews functionality test 3

```python
import geoviews as gv
import numpy as np
import pandas as pd

# Example 3: Creating image/raster visualizations
print("Example 3: Creating image/raster visualizations")

try:
    # Create sample raster data (elevation-like)
    lon_range = np.linspace(-125, -115, 20)
    lat_range = np.linspace(32, 42, 15)
    lon_grid, lat_grid = np.meshgrid(lon_range, lat_range)
    
    # Generate elevation-like data with some patterns
    elevation = (
        1000 * np.sin(lon_grid * 0.1) * np.cos(lat_grid * 0.1) + 
        500 * np.random.random(lon_grid.shape) + 1500
    )
    
    print(f"✓ Generated raster data:")
    print(f"  Grid shape: {elevation.shape}")
    print(f"  Longitude range: {lon_range.min():.1f} to {lon_range.max():.1f}")
    print(f"  Latitude range: {lat_range.min():.1f} to {lat_range.max():.1f}")
    print(f"  Elevation range: {elevation.min():.0f} to {elevation.max():.0f}")
    
    # Create Image element
    bounds = (lon_range.min(), lat_range.min(), lon_range.max(), lat_range.max())
    image = gv.Image(elevation, bounds=bounds, kdims=['lon', 'lat'], vdims='elevation')
    
    print(f"✓ Image element created:")
    print(f"  Type: {type(image).__name__}")
    print(f"  Bounds: {image.bounds}")
    print(f"  Data dimensions: {image.data.shape}")
    
    # Add styling
    styled_image = image.opts(cmap='terrain', alpha=0.8)
    print(f"✓ Applied terrain colormap styling")
    
    print("✓ Raster visualization completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geoviews example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished\n")
```

## Example 4: geoviews functionality test 4

```python
import geoviews as gv
import numpy as np
import pandas as pd

# Example 4: Working with paths and line features
print("Example 4: Working with paths and line features")

try:
    # Create sample path data (like roads or routes)
    # Route 1: Highway 101 (simplified)
    coords1 = [(-122.4, 37.8), (-122.2, 37.6), (-122.0, 37.4), (-121.8, 37.2), (-121.6, 37.0)]
    path1 = gv.Path([coords1])
    
    # Route 2: Interstate 5 (simplified)
    coords2 = [(-122.3, 37.9), (-122.3, 37.7), (-122.2, 37.5), (-122.1, 37.3), (-122.0, 37.1)]
    path2 = gv.Path([coords2])
    
    print(f"✓ Created 2 route paths")
    print(f"  Route 1 points: {len(coords1)}")
    print(f"  Route 2 points: {len(coords2)}")
    
    # Combine paths into overlay
    all_paths = path1 * path2
    print(f"✓ Path elements combined: {type(all_paths).__name__}")
    
    # Apply styling to paths (using correct matplotlib option names)
    styled_path1 = path1.opts(color='red', alpha=0.8, linewidth=3)
    styled_path2 = path2.opts(color='blue', alpha=0.8, linewidth=3)
    print(f"✓ Applied different styling to each path")
    
    # Add some points along the routes
    route_points = pd.DataFrame({
        'lon': [-122.3, -122.1, -122.2],
        'lat': [37.7, 37.3, 37.5],
        'type': ['junction', 'exit', 'rest_area']
    })
    
    points = gv.Points(route_points, kdims=['lon', 'lat'], vdims=['type'])
    print(f"✓ Added {len(route_points)} route points")
    
    # Combine paths and points
    combined_viz = styled_path1 * styled_path2 * points
    print(f"✓ Combined paths with points overlay")
    
    print("✓ Path and line visualization completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geoviews example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished\n")
```

## Example 5: geoviews functionality test 5

```python
import geoviews as gv
import numpy as np
import pandas as pd

# Example 5: Combining multiple elements and projections
print("Example 5: Combining multiple elements and projections")

try:
    # Create base map elements
    base_cities = pd.DataFrame({
        'lon': [-122.4, -118.2, -122.3],
        'lat': [37.8, 34.1, 47.6],
        'city': ['San Francisco', 'Los Angeles', 'Seattle'],
        'population': [100, 150, 80]
    })
    
    # Create different GeoViews elements
    cities = gv.Points(base_cities, kdims=['lon', 'lat'], vdims=['city', 'population'])
    
    # Create a simple contour-like feature
    lon_vals = np.linspace(-125, -115, 10)
    lat_vals = np.linspace(32, 48, 10)
    
    contour_data = []
    for i, (lon, lat) in enumerate(zip(lon_vals, lat_vals)):
        contour_data.append([lon, lat, i * 10])
    
    contours_df = pd.DataFrame(contour_data, columns=['lon', 'lat', 'value'])
    contours = gv.Points(contours_df, kdims=['lon', 'lat'], vdims=['value'])
    
    print(f"✓ Created visualization elements:")
    print(f"  Cities: {len(base_cities)} points")
    print(f"  Contours: {len(contour_data)} points")
    
    # Combine elements using overlay
    combined = cities * contours
    print(f"✓ Combined elements: {type(combined).__name__}")
    
    # Apply different styling to each layer (using valid matplotlib options)
    styled_cities = cities.opts(color='red', alpha=0.8, fig_size=300)
    styled_contours = contours.opts(color='blue', alpha=0.6, fig_size=300)
    
    # Final combined visualization
    final_viz = styled_cities * styled_contours
    print(f"✓ Applied styling to combined visualization")
    print(f"  Final type: {type(final_viz).__name__}")
    print(f"  Number of overlaid elements: {len(final_viz)}")
    
    print("✓ Multi-element geographic visualization completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"geoviews example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished\n")
```
