## Example 1: giddy functionality test 1

```python
import giddy
import numpy as np

# Example 1: Markov chain analysis for regional transitions
print("Example 1: Markov chain analysis for regional transitions")

try:
    # Create sample data representing income levels over time for regions
    np.random.seed(42)
    regions = 20
    time_periods = 5
    
    # Generate sample income data (regions x time periods)
    income_data = np.random.lognormal(mean=10, sigma=0.5, size=(regions, time_periods))
    income_data = income_data + np.random.normal(0, 0.1, income_data.shape)  # Add noise
    
    print(f"✓ Generated income data:")
    print(f"  Regions: {regions}")
    print(f"  Time periods: {time_periods}")
    print(f"  Income range: ${income_data.min():.0f} - ${income_data.max():.0f}")
    
    # Create Markov chain using giddy
    markov = giddy.markov.Markov(income_data)
    
    print(f"✓ Markov chain analysis:")
    print(f"  Classes: {markov.classes}")
    print(f"  Transition matrix shape: {markov.p.shape}")
    print(f"  Steady state: {markov.steady_state[:3]}")  # First 3 values
    
    # Calculate basic mobility statistics
    transitions = markov.transitions
    total_transitions = transitions.sum()
    diagonal_sum = np.diag(transitions).sum()
    mobility_ratio = 1 - (diagonal_sum / total_transitions)
    
    print(f"✓ Mobility measures:")
    print(f"  Total transitions: {total_transitions}")
    print(f"  Mobility ratio: {mobility_ratio:.3f}")
    print(f"  Off-diagonal transitions: {total_transitions - diagonal_sum}")
    
    print("✓ Regional transition analysis completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"giddy example (cell 2) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished\n")
```

## Example 2: giddy functionality test 2

```python
import giddy
import numpy as np

# Example 2: Comparing Markov transition matrices
print("Example 2: Comparing Markov transition matrices")

try:
    # Generate sample economic data over time
    np.random.seed(123)
    n_regions = 18
    n_periods = 7
    
    # Create economic indicator data (e.g., productivity index)
    base_productivity = np.random.uniform(0.8, 1.4, n_regions)
    productivity_data = np.zeros((n_regions, n_periods))
    
    for t in range(n_periods):
        if t == 0:
            productivity_data[:, t] = base_productivity
        else:
            # Add random walk with slight mean reversion
            changes = np.random.normal(0, 0.08, n_regions)
            productivity_data[:, t] = productivity_data[:, t-1] + changes
    
    print(f"✓ Generated productivity data:")
    print(f"  Regions: {n_regions}")
    print(f"  Periods: {n_periods}")
    print(f"  Range: {productivity_data.min():.2f} - {productivity_data.max():.2f}")
    
    # Create basic Markov chain
    markov_basic = giddy.markov.Markov(productivity_data)
    
    print(f"✓ Basic Markov analysis:")
    print(f"  Number of classes: {markov_basic.classes}")
    print(f"  Transition matrix:")
    print(f"    Shape: {markov_basic.p.shape}")
    print(f"    Diagonal mean: {np.diag(markov_basic.p).mean():.3f}")
    
    # Analyze transition patterns
    upper_triangular = np.triu(markov_basic.p, k=1).sum()
    lower_triangular = np.tril(markov_basic.p, k=-1).sum()
    
    print(f"✓ Transition analysis:")
    print(f"  Upward mobility: {upper_triangular:.3f}")
    print(f"  Downward mobility: {lower_triangular:.3f}")
    print(f"  Net upward flow: {upper_triangular - lower_triangular:.3f}")
    
    print("✓ Markov transition analysis completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"giddy example (cell 4) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished\n")
```

## Example 3: giddy functionality test 3

```python
import giddy
import numpy as np
from scipy.stats import rankdata

# Example 3: Rank-based mobility analysis
print("Example 3: Rank-based mobility analysis")

try:
    # Create sample data for rank analysis (e.g., city populations over time)
    np.random.seed(789)
    n_cities = 12
    n_years = 8
    
    # Generate population data with different growth patterns
    initial_pop = np.random.uniform(50000, 2000000, n_cities)
    population_data = np.zeros((n_cities, n_years))
    
    for year in range(n_years):
        if year == 0:
            population_data[:, year] = initial_pop
        else:
            # Different cities have different growth rates
            growth_rates = np.random.normal(0.015, 0.02, n_cities)  # 1.5% avg growth
            population_data[:, year] = population_data[:, year-1] * (1 + growth_rates)
    
    print(f"✓ Generated population data:")
    print(f"  Cities: {n_cities}")
    print(f"  Years: {n_years}")
    print(f"  Population range: {population_data.min():.0f} - {population_data.max():.0f}")
    
    # Convert to ranks for each time period
    rank_data = np.zeros_like(population_data)
    for t in range(n_years):
        rank_data[:, t] = rankdata(-population_data[:, t])  # Negative for descending ranks
    
    # Create Markov chain on ranks
    rank_markov = giddy.markov.Markov(rank_data)
    
    print(f"✓ Rank-based Markov analysis:")
    print(f"  Rank classes: {rank_markov.classes}")
    print(f"  Transition matrix shape: {rank_markov.p.shape}")
    print(f"  Persistence (diagonal): {np.diag(rank_markov.p)[:3]}")  # First 3
    
    # Analyze rank mobility manually
    initial_ranks = rank_data[:, 0].astype(int)
    final_ranks = rank_data[:, -1].astype(int)
    rank_changes = np.abs(final_ranks - initial_ranks)
    
    print(f"✓ Rank mobility analysis:")
    print(f"  Average rank change: {rank_changes.mean():.1f}")
    print(f"  Max rank change: {rank_changes.max()}")
    print(f"  Cities with no change: {np.sum(rank_changes == 0)}")
    
    print("✓ Rank-based mobility analysis completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"giddy example (cell 6) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished\n")
```

## Example 4: giddy functionality test 4

```python
import giddy
import numpy as np

# Example 4: Time series decomposition of regional data
print("Example 4: Time series decomposition of regional data")

try:
    # Create sample data with spatial and temporal patterns
    np.random.seed(456)
    n_locations = 25
    n_periods = 6
    
    # Generate employment rate data with trend and seasonality
    employment_data = np.zeros((n_locations, n_periods))
    base_employment = np.random.uniform(0.65, 0.85, n_locations)  # 65-85% employment
    
    for t in range(n_periods):
        if t == 0:
            employment_data[:, t] = base_employment
        else:
            # Add trend and cyclical variation
            trend = 0.005 * t  # Slight upward trend
            cycle = 0.02 * np.sin(2 * np.pi * t / 4)  # 4-period cycle
            noise = np.random.normal(0, 0.02, n_locations)
            employment_data[:, t] = employment_data[:, t-1] + trend + cycle + noise
    
    # Clip to realistic employment range
    employment_data = np.clip(employment_data, 0.4, 0.95)
    
    print(f"✓ Generated employment time series:")
    print(f"  Locations: {n_locations}")
    print(f"  Time periods: {n_periods}")
    print(f"  Employment range: {employment_data.min():.2f} - {employment_data.max():.2f}")
    
    # Analyze temporal patterns
    period_means = employment_data.mean(axis=0)
    location_means = employment_data.mean(axis=1)
    
    print(f"✓ Temporal analysis:")
    print(f"  Period averages: {period_means}")
    print(f"  Trend over time: {np.polyfit(range(n_periods), period_means, 1)[0]:.4f}")
    
    # Create Markov chain to analyze employment transitions
    employment_markov = giddy.markov.Markov(employment_data)
    
    print(f"✓ Employment transition analysis:")
    print(f"  Employment classes: {employment_markov.classes}")
    print(f"  Transition matrix diagonal: {np.diag(employment_markov.p)}")
    
    # Calculate regional stability (variance over time)
    regional_stability = employment_data.var(axis=1)
    stable_regions = np.sum(regional_stability < 0.001)
    
    print(f"✓ Regional stability:")
    print(f"  Stable regions (low variance): {stable_regions}/{n_locations}")
    print(f"  Mean regional variance: {regional_stability.mean():.4f}")
    
    print("✓ Time series analysis completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"giddy example (cell 8) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished\n")
```

## Example 5: giddy functionality test 5

```python
import giddy
from giddy.ergodic import mfpt, steady_state
import numpy as np

# Example 5: Ergodic analysis and long-term behavior
print("Example 5: Ergodic analysis and long-term behavior")

try:
    # Generate regional income data for ergodic analysis
    np.random.seed(101)
    n_regions = 18
    n_years = 10
    
    # Create income data with different growth patterns and convergence
    initial_income = np.random.exponential(scale=40000, size=n_regions)
    income_data = np.zeros((n_regions, n_years))
    
    for year in range(n_years):
        if year == 0:
            income_data[:, year] = initial_income
        else:
            # Add mean reversion (convergence) and random shocks
            mean_income = income_data[:, year-1].mean()
            convergence = 0.05 * (mean_income - income_data[:, year-1])
            shocks = np.random.normal(0, 0.02, n_regions)
            income_data[:, year] = income_data[:, year-1] * (1 + convergence/income_data[:, year-1] + shocks)
    
    print(f"✓ Generated income convergence data:")
    print(f"  Regions: {n_regions}")
    print(f"  Years: {n_years}")
    print(f"  Initial income range: ${income_data[:, 0].min():.0f} - ${income_data[:, 0].max():.0f}")
    print(f"  Final income range: ${income_data[:, -1].min():.0f} - ${income_data[:, -1].max():.0f}")
    
    # Ergodic analysis
    markov = giddy.markov.Markov(income_data)
    
    print(f"✓ Markov ergodic properties:")
    print(f"  Classes: {markov.classes}")
    print(f"  Steady state distribution: {markov.steady_state}")
    
    # Ergodic properties. giddy.ergodic has no `ergodic_index` (its API is steady_state / mfpt / var_mfpt_ergodic),
    # and it needs an ERGODIC chain: the synthetic panel above barely changes class between years (its Markov chain
    # is reducible), so use Kemeny & Snell's Land of Oz chain, the example in giddy's own docstring.
    oz = np.array([[.5, .25, .25], [.5, 0, .5], [.25, .25, .5]])
    pi = steady_state(oz)
    first_passage = mfpt(oz)
    print(f"  Land of Oz steady state: {pi}")
    assert np.allclose(pi, [0.4, 0.2, 0.4]), pi
    # the diagonal of the mean first passage matrix holds the mean recurrence times, 1 / steady state
    assert np.allclose(np.diag(first_passage), 1 / pi), first_passage
    
    # Analyze convergence
    initial_cv = np.std(income_data[:, 0]) / np.mean(income_data[:, 0])
    final_cv = np.std(income_data[:, -1]) / np.mean(income_data[:, -1])
    
    print(f"✓ Convergence analysis:")
    print(f"  Initial CV: {initial_cv:.3f}")
    print(f"  Final CV: {final_cv:.3f}")
    print(f"  Convergence: {'Yes' if final_cv < initial_cv else 'No'}")
    
    print("✓ Ergodic analysis and convergence study completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"giddy example (cell 10) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished\n")
```
