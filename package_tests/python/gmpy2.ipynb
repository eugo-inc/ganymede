# gmpy2

## Example 1: big integers on GMP

```python
# Example 1: mpz arithmetic, primality, integer square roots and gcd on numbers far past 64 bits
import gmpy2
from gmpy2 import mpz

n = mpz(2) ** 521 - 1  # a Mersenne prime
assert gmpy2.is_prime(n) and not gmpy2.is_prime(n + 2)
root, rem = gmpy2.isqrt_rem(mpz(10) ** 40 + 7)
assert root == mpz(10) ** 20 and rem == 7
assert gmpy2.gcd(mpz(2) ** 100 * 3, mpz(2) ** 50 * 9) == mpz(2) ** 50 * 3
assert gmpy2.fac(30) == 265252859812191058636308480000000
print("Example 1:", n.bit_length(), "bit prime;", gmpy2.version(), gmpy2.mp_version())
```

## Example 2: MPFR floats at a chosen precision and rounding

```python
# Example 2: 200-bit sqrt(2), and rounding toward -inf vs +inf bracketing the exact value
import gmpy2
from gmpy2 import mpfr

with gmpy2.context(gmpy2.get_context(), precision=200):
    root2 = gmpy2.sqrt(mpfr(2))
    assert str(root2).startswith("1.414213562373095048801688724209698078569671875376948073")
with gmpy2.context(gmpy2.get_context(), precision=24, round=gmpy2.RoundDown):
    low = mpfr(1) / 3
with gmpy2.context(gmpy2.get_context(), precision=24, round=gmpy2.RoundUp):
    high = mpfr(1) / 3
assert low < high and low.precision == 24, (low, high)
print("Example 2:", str(root2)[:30], "|", low, "<", high, "|", gmpy2.mpfr_version())
```

## Example 3: complex numbers on MPC

```python
# Example 3: exp(i*pi) + 1 is zero to working precision; mpc keeps real and imaginary parts as MPFR numbers
import gmpy2
from gmpy2 import mpc

with gmpy2.context(gmpy2.get_context(), precision=113):
    z = gmpy2.exp(mpc(0, 1) * gmpy2.const_pi()) + 1
    assert abs(z) < gmpy2.mpfr("1e-30"), z
    w = gmpy2.sqrt(mpc(-4, 0))
assert w == mpc(0, 2), w
print("Example 3:", abs(z) < 1e-30, w, gmpy2.mpc_version())
```

## Example 4: exact rationals

```python
# Example 4: mpq keeps fractions exact; the harmonic number H(20) has a known denominator
from gmpy2 import mpq

h20 = sum(mpq(1, k) for k in range(1, 21))
assert h20.denominator == 15519504 and h20.numerator == 55835135, h20
assert mpq(3, 4) + mpq(1, 4) == 1 and mpq(1, 3) * 3 == 1
print("Example 4: H(20) =", h20, "~", float(h20))
```

## Example 5: mpmath runs on gmpy2

```python
# Example 5: with gmpy2 installed, mpmath (and so sympy) uses it as its integer backend
import mpmath
import mpmath.libmp

assert mpmath.libmp.BACKEND == "gmpy", mpmath.libmp.BACKEND
mpmath.mp.dps = 60
pi60 = mpmath.nstr(mpmath.pi, 60)
assert pi60.startswith("3.14159265358979323846264338327950288419716939937510582097494"), pi60
print("Example 5: mpmath backend", mpmath.libmp.BACKEND, "| pi =", pi60[:40], "...")
```
