# google-cloud-resource-manager smoke test

Exercises the primary client classes and request/response types from `google.cloud.resourcemanager_v3`.

```python
# Example 1: top-level client classes are importable from the package root.
from google.cloud import resourcemanager_v3

expected = [
    'ProjectsClient',
    'FoldersClient',
    'OrganizationsClient',
    'TagBindingsClient',
    'TagKeysClient',
    'TagValuesClient',
]
for cls_name in expected:
    cls = getattr(resourcemanager_v3, cls_name)
    print(f"{cls_name}: {cls.__module__}.{cls.__qualname__}")
    assert isinstance(cls, type), f"{cls_name} is not a class"

assert hasattr(resourcemanager_v3, 'types'), 'types module should be exposed'
```

```python
# Example 2: construct a Project message (proto-plus) and round-trip its fields.
from google.cloud.resourcemanager_v3 import Project

p = Project(
    name='projects/123456789',
    parent='organizations/987654321',
    project_id='my-test-project',
    display_name='My Test Project',
)

assert p.project_id == 'my-test-project'
assert p.display_name == 'My Test Project'
assert p.parent == 'organizations/987654321'

# proto-plus message has a `State` enum nested type
assert hasattr(Project, 'State')
print('Project.State members:', [s.name for s in Project.State])
print('Project:', p)
```

```python
# Example 3: build request message types used by the Projects API.
from google.cloud.resourcemanager_v3 import (
    GetProjectRequest,
    ListProjectsRequest,
    SearchProjectsRequest,
)

get_req = GetProjectRequest(name='projects/my-test-project')
assert get_req.name == 'projects/my-test-project'

list_req = ListProjectsRequest(parent='folders/123', page_size=50)
assert list_req.parent == 'folders/123'
assert list_req.page_size == 50

search_req = SearchProjectsRequest(query='displayName:Test*')
assert search_req.query == 'displayName:Test*'

print('GetProjectRequest:', get_req)
print('ListProjectsRequest:', list_req)
print('SearchProjectsRequest:', search_req)
```

```python
# Example 4: Folder, Organization, TagKey, TagValue, TagBinding message types.
from google.cloud.resourcemanager_v3 import (
    Folder,
    Organization,
    TagBinding,
    TagKey,
    TagValue,
)

folder = Folder(name='folders/111', parent='organizations/222', display_name='Eng')
org = Organization(name='organizations/222', display_name='example.com')
tag_key = TagKey(name='tagKeys/333', short_name='env', parent='organizations/222')
tag_value = TagValue(name='tagValues/444', short_name='prod', parent='tagKeys/333')
tag_binding = TagBinding(
    name='tagBindings/abc',
    parent='//cloudresourcemanager.googleapis.com/projects/123',
    tag_value='tagValues/444',
)

assert folder.display_name == 'Eng'
assert org.name == 'organizations/222'
assert tag_key.short_name == 'env'
assert tag_value.short_name == 'prod'
assert tag_binding.tag_value == 'tagValues/444'

print('Folder.State members:', [s.name for s in Folder.State])
```

```python
# Example 5: client classes have the expected RPC methods + transport plumbing.
from google.cloud.resourcemanager_v3 import (
    FoldersClient,
    ProjectsClient,
    TagKeysClient,
)
from google.cloud.resourcemanager_v3.services.projects import transports

projects_methods = {'get_project', 'list_projects', 'search_projects', 'create_project'}
assert projects_methods.issubset(set(dir(ProjectsClient))), 'Projects client missing core RPCs'

folders_methods = {'get_folder', 'list_folders', 'search_folders', 'create_folder'}
assert folders_methods.issubset(set(dir(FoldersClient))), 'Folders client missing core RPCs'

tagkeys_methods = {'get_tag_key', 'list_tag_keys', 'create_tag_key'}
assert tagkeys_methods.issubset(set(dir(TagKeysClient))), 'TagKeys client missing core RPCs'

assert hasattr(transports, 'ProjectsGrpcTransport')
assert hasattr(transports, 'ProjectsTransport')

print('ProjectsClient transport:', ProjectsClient.get_transport_class().__name__)
print('OK google_cloud_resource_manager')
```
