# google_cloud_storage_control smoke test

Exercises the `google.cloud.storage_control_v2` client surface without contacting GCS.

```python
# Example 1: public API surface — the primary client classes are exposed at the top level.
from google.cloud.storage_control_v2 import StorageControlClient, StorageControlAsyncClient

assert callable(StorageControlClient)
assert callable(StorageControlAsyncClient)

# Both clients should expose a path-builder for buckets (a static helper, no network).
bucket_path = StorageControlClient.bucket_path('my-project', 'my-bucket')
print('bucket_path:', bucket_path)
assert 'my-project' in bucket_path and 'my-bucket' in bucket_path

folder_path = StorageControlClient.folder_path('my-project', 'my-bucket', 'a/b/c/')
print('folder_path:', folder_path)
assert 'a/b/c' in folder_path
```

```python
# Example 2: proto message types are exposed in the .types module.
from google.cloud.storage_control_v2 import types

# Construct a Folder message — pure proto-plus, no client needed.
folder = types.Folder(name='projects/_/buckets/my-bucket/folders/foo/')
print('folder.name =', folder.name)
assert folder.name.endswith('foo/')

# Construct a ManagedFolder message.
managed = types.ManagedFolder(name='projects/_/buckets/my-bucket/managedFolders/bar/')
print('managed.name =', managed.name)
assert 'managedFolders' in managed.name
```

```python
# Example 3: build request messages without sending them.
from google.cloud.storage_control_v2 import types

create_req = types.CreateFolderRequest(
    parent='projects/_/buckets/my-bucket',
    folder_id='new-folder/',
    recursive=True,
)
print('parent:', create_req.parent)
print('folder_id:', create_req.folder_id)
print('recursive:', create_req.recursive)
assert create_req.recursive is True
assert create_req.folder_id == 'new-folder/'

list_req = types.ListFoldersRequest(parent='projects/_/buckets/my-bucket', page_size=10)
assert list_req.page_size == 10
```

```python
# Example 4: transport classes (gRPC + async gRPC) are importable.
from google.cloud.storage_control_v2.services.storage_control import transports

# Both the sync gRPC transport and the async one should be present.
assert hasattr(transports, 'StorageControlGrpcTransport')
assert hasattr(transports, 'StorageControlGrpcAsyncIOTransport')

# The base class should declare the canonical default host.
base = transports.StorageControlTransport
print('DEFAULT_HOST:', base.DEFAULT_HOST)
assert 'googleapis.com' in base.DEFAULT_HOST
```

```python
# Example 5: pagers + service-config are part of the surface; verify they import.
from google.cloud.storage_control_v2.services.storage_control import pagers
from google.cloud.storage_control_v2 import StorageControlClient

# The standard list-pagers should be defined on the pagers module.
assert hasattr(pagers, 'ListFoldersPager')
assert hasattr(pagers, 'ListManagedFoldersPager')

# parse_* inverse-builders round-trip cleanly with the path builders.
path = StorageControlClient.managed_folder_path('p', 'b', 'm/')
parsed = StorageControlClient.parse_managed_folder_path(path)
print('parsed:', parsed)
assert parsed['project'] == 'p'
assert parsed['bucket'] == 'b'
assert parsed['managed_folder'].rstrip('/') == 'm'
print('OK google_cloud_storage_control')
```
