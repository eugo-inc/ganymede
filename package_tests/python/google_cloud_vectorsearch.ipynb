# google-cloud-vectorsearch smoke test

Pure Python client for Google Cloud Vector Search (Vertex AI Matching Engine). Tests import-only behavior; no network calls.

```python
# Example 1: import the top-level vectorsearch module and inspect public attributes.
from google.cloud import vectorsearch

public_names = [n for n in dir(vectorsearch) if not n.startswith('_')]
print(f"Number of public attributes: {len(public_names)}")
print(f"First 20 public attributes: {public_names[:20]}")
assert len(public_names) > 0, "vectorsearch should expose public symbols"
```

```python
# Example 2: import google_api_core and google_auth (declared runtime deps) to ensure they coexist.
import google.api_core
import google.auth

print(f"google.api_core version: {getattr(google.api_core, '__version__', 'unknown')}")
print(f"google.auth version: {getattr(google.auth, '__version__', 'unknown')}")
assert google.api_core is not None
assert google.auth is not None
```

```python
# Example 3: proto-plus and protobuf — the message/protocol layer underlying the gRPC client.
import proto
import google.protobuf

print(f"proto-plus version: {getattr(proto, '__version__', 'unknown')}")
print(f"protobuf version: {google.protobuf.__version__}")

# Verify proto.Message is a class (it's the base class for generated messages)
assert isinstance(proto.Message, type) or callable(proto.Message)
```

```python
# Example 4: grpcio runtime is importable + has the channel constructors the client needs.
import grpc

print(f"grpc version: {grpc.__version__}")
# These are the core constructors a Google Cloud client uses to create transports.
assert callable(grpc.insecure_channel)
assert callable(grpc.secure_channel)
assert hasattr(grpc, 'ssl_channel_credentials')
```

```python
# Example 5: try to locate a client/service class on the package without constructing it (no network).
from google.cloud import vectorsearch

# Look for any *Client or *Service class exposed at the top level or in obvious submodules.
candidates = [n for n in dir(vectorsearch) if ('Client' in n or 'Service' in n) and not n.startswith('_')]
print(f"Client/Service-like names on top-level package: {candidates}")

# Also check for a 'types' or 'services' submodule (canonical google-cloud client layout).
submods_present = [m for m in ('types', 'services', 'gapic_version') if hasattr(vectorsearch, m)]
print(f"Canonical google-cloud submodules present: {submods_present}")

assert vectorsearch.__name__ == 'google.cloud.vectorsearch'
```
