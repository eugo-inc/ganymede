## Example 2: google_crc32c functionality test 2

```python
import google_crc32c
import os
import tempfile

# Example 2: CRC32C for file integrity checking
print("Example 2: CRC32C for file integrity checking")

try:
    # Create temporary files for testing
    temp_dir = tempfile.mkdtemp()
    
    # Sample file contents
    test_files = {
        'small.txt': b'Hello, World!',
        'medium.txt': b'This is a longer text file for testing CRC32C checksums.\n' * 10,
        'binary.dat': bytes(range(256)),  # Binary data
        'empty.txt': b''
    }
    
    print("Creating test files and calculating checksums:")
    
    file_checksums = {}
    
    for filename, content in test_files.items():
        file_path = os.path.join(temp_dir, filename)
        
        # Write file
        with open(file_path, 'wb') as f:
            f.write(content)
        
        # Calculate checksum directly from content
        checksum_direct = google_crc32c.value(content)
        
        # Calculate checksum by reading file
        with open(file_path, 'rb') as f:
            file_content = f.read()
            checksum_file = google_crc32c.value(file_content)
        
        # Store for integrity verification
        file_checksums[filename] = checksum_direct
        
        print(f"  {filename}:")
        print(f"    Size: {len(content)} bytes")
        print(f"    CRC32C: 0x{checksum_direct:08x}")
        print(f"    Direct == File: {checksum_direct == checksum_file}")
        print()
    
    # Demonstrate incremental file reading with large data
    print("Testing incremental file checksum calculation:")
    
    large_data = b"This is a chunk of data. " * 1000
    large_file = os.path.join(temp_dir, 'large.txt')
    
    with open(large_file, 'wb') as f:
        f.write(large_data)
    
    # Method 1: Read entire file
    with open(large_file, 'rb') as f:
        full_checksum = google_crc32c.value(f.read())
    
    # Method 2: Read in chunks
    chunk_size = 1024
    checksum = 0
    
    with open(large_file, 'rb') as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            if checksum == 0:
                checksum = google_crc32c.value(chunk)
            else:
                checksum = google_crc32c.extend(checksum, chunk)
    
    print(f"  Large file size: {len(large_data)} bytes")
    print(f"  Full read CRC32C: 0x{full_checksum:08x}")
    print(f"  Chunked read CRC32C: 0x{checksum:08x}")
    print(f"  Match: {full_checksum == checksum}")
    
    # File integrity verification simulation
    print(f"\nFile integrity verification:")
    for filename, expected_checksum in file_checksums.items():
        file_path = os.path.join(temp_dir, filename)
        
        with open(file_path, 'rb') as f:
            current_checksum = google_crc32c.value(f.read())
        
        integrity_ok = current_checksum == expected_checksum
        status = "✓ VALID" if integrity_ok else "✗ CORRUPTED"
        
        print(f"  {filename}: {status}")
    
    # Cleanup
    import shutil
    shutil.rmtree(temp_dir)

except Exception as e:
    print(f"File integrity testing error: {e}")
    
    # Show file integrity concepts
    print("CRC32C file integrity concepts:")
    print("  - Calculate checksum when file is created")
    print("  - Store checksum metadata separately")
    print("  - Verify checksum periodically or before use")
    print("  - Use incremental calculation for large files")
    raise AssertionError(
        f"google_crc32c example (cell 4) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed")
```

## Example 3: google_crc32c functionality test 3

```python
import google_crc32c
import json
import base64

# Example 3: CRC32C for data validation and Google Cloud integration
print("Example 3: CRC32C for data validation and Google Cloud integration")

try:
    # Simulate Google Cloud Storage object metadata with CRC32C
    print("Simulating Google Cloud Storage CRC32C validation:")
    
    # Sample data objects
    cloud_objects = [
        {
            'name': 'document.pdf',
            'data': b'PDF file content would go here...' * 50,
            'content_type': 'application/pdf'
        },
        {
            'name': 'image.jpg', 
            'data': bytes([0xFF, 0xD8, 0xFF, 0xE0]) + b'JPEG data...' * 100,
            'content_type': 'image/jpeg'
        },
        {
            'name': 'config.json',
            'data': json.dumps({'setting1': 'value1', 'setting2': 42}).encode(),
            'content_type': 'application/json'
        }
    ]
    
    # Calculate CRC32C checksums and create metadata
    object_metadata = []
    
    for obj in cloud_objects:
        # Calculate CRC32C checksum
        crc32c_checksum = google_crc32c.value(obj['data'])
        
        # Convert to base64 for storage (Google Cloud format)
        checksum_bytes = crc32c_checksum.to_bytes(4, byteorder='big')
        checksum_b64 = base64.b64encode(checksum_bytes).decode()
        
        metadata = {
            'name': obj['name'],
            'size': len(obj['data']),
            'contentType': obj['content_type'],
            'crc32c': checksum_b64,
            'crc32c_int': crc32c_checksum
        }
        
        object_metadata.append(metadata)
        
        print(f"  Object: {obj['name']}")
        print(f"    Size: {len(obj['data'])} bytes")
        print(f"    Content-Type: {obj['content_type']}")
        print(f"    CRC32C (int): {crc32c_checksum}")
        print(f"    CRC32C (hex): 0x{crc32c_checksum:08x}")
        print(f"    CRC32C (b64): {checksum_b64}")
        print()
    
    # Simulate download validation
    print("Simulating download validation:")
    
    for i, obj in enumerate(cloud_objects):
        metadata = object_metadata[i]
        
        # Simulate downloaded data
        downloaded_data = obj['data']  # In real scenario, this comes from network
        
        # Calculate checksum of downloaded data
        download_checksum = google_crc32c.value(downloaded_data)
        
        # Compare with stored metadata
        expected_checksum = metadata['crc32c_int']
        is_valid = download_checksum == expected_checksum
        
        status = "✓ VALID" if is_valid else "✗ CORRUPTED"
        print(f"  {metadata['name']}: {status}")
        print(f"    Expected: 0x{expected_checksum:08x}")
        print(f"    Actual: 0x{download_checksum:08x}")
    
    # Test incremental validation for streaming data
    print(f"\nStreaming validation example:")
    
    # Simulate receiving data in chunks
    test_data = b"Streaming data validation test. " * 200
    expected_full_checksum = google_crc32c.value(test_data)
    
    # Process in chunks
    chunk_size = 64
    streaming_checksum = 0
    bytes_processed = 0
    
    for i in range(0, len(test_data), chunk_size):
        chunk = test_data[i:i + chunk_size]
        bytes_processed += len(chunk)
        
        if streaming_checksum == 0:
            streaming_checksum = google_crc32c.value(chunk)
        else:
            streaming_checksum = google_crc32c.extend(streaming_checksum, chunk)
        
        # Show progress every few chunks
        if (i // chunk_size) % 20 == 0:
            progress = (bytes_processed / len(test_data)) * 100
            print(f"    Progress: {progress:.1f}% - CRC32C: 0x{streaming_checksum:08x}")
    
    final_valid = streaming_checksum == expected_full_checksum
    print(f"    Final validation: {'✓ VALID' if final_valid else '✗ INVALID'}")

except Exception as e:
    print(f"Cloud integration testing error: {e}")
    
    # Show Google Cloud CRC32C concepts
    print("Google Cloud CRC32C integration concepts:")
    concepts = [
        "GCS objects store CRC32C checksums in metadata",
        "Checksums are base64-encoded for JSON APIs",
        "Client libraries validate data integrity automatically",
        "Streaming uploads/downloads use incremental validation",
        "CRC32C is faster than MD5 for integrity checking"
    ]
    
    for concept in concepts:
        print(f"  • {concept}")
    raise

print("Example 3 completed")
```

## Example 4: google_crc32c functionality test 4

```python
import google_crc32c
import hashlib
import time

# Example 4: CRC32C performance comparison and benchmarking
print("Example 4: CRC32C performance comparison and benchmarking")

try:
    # Create test data of various sizes
    test_sizes = [
        (1024, "1 KB"),
        (10 * 1024, "10 KB"), 
        (100 * 1024, "100 KB"),
        (1024 * 1024, "1 MB")
    ]
    
    print("Performance comparison: CRC32C vs other checksums")
    print("=" * 60)
    
    for size_bytes, size_label in test_sizes:
        print(f"\nTesting with {size_label} of data:")
        
        # Generate test data
        test_data = b"A" * size_bytes
        
        start_time = time.perf_counter()
        crc32c_result = google_crc32c.value(test_data)
        crc32c_time = time.perf_counter() - start_time
        
        # Test standard library CRC32 for comparison
        import zlib
        start_time = time.perf_counter()
        crc32_result = zlib.crc32(test_data)
        crc32_time = time.perf_counter() - start_time
        
        # Test MD5 for comparison
        start_time = time.perf_counter()
        md5_result = hashlib.md5(test_data).hexdigest()
        md5_time = time.perf_counter() - start_time
        
        # Test SHA1 for comparison
        start_time = time.perf_counter()
        sha1_result = hashlib.sha1(test_data).hexdigest()
        sha1_time = time.perf_counter() - start_time
        
        print(f"  CRC32C:  {crc32c_time:.6f}s - 0x{crc32c_result:08x}")
        print(f"  CRC32:   {crc32_time:.6f}s - 0x{crc32_result & 0xffffffff:08x}")
        print(f"  MD5:     {md5_time:.6f}s - {md5_result[:16]}...")
        print(f"  SHA1:    {sha1_time:.6f}s - {sha1_result[:16]}...")
        
        # Calculate throughput
        throughput_mb_s = (size_bytes / (1024 * 1024)) / crc32c_time
        print(f"  CRC32C Throughput: {throughput_mb_s:.1f} MB/s")
    
    # Test incremental vs full calculation performance
    print(f"\nIncremental vs Full calculation comparison:")
    
    large_data = b"Performance test data chunk. " * 10000
    chunk_size = 1024
    
    # Full calculation
    start_time = time.perf_counter()
    full_checksum = google_crc32c.value(large_data)
    full_time = time.perf_counter() - start_time
    
    # Incremental calculation
    start_time = time.perf_counter()
    incremental_checksum = 0
    
    for i in range(0, len(large_data), chunk_size):
        chunk = large_data[i:i + chunk_size]
        if incremental_checksum == 0:
            incremental_checksum = google_crc32c.value(chunk)
        else:
            incremental_checksum = google_crc32c.extend(incremental_checksum, chunk)
    
    incremental_time = time.perf_counter() - start_time
    
    print(f"  Data size: {len(large_data):,} bytes")
    print(f"  Full calculation: {full_time:.6f}s")
    print(f"  Incremental: {incremental_time:.6f}s")
    print(f"  Results match: {full_checksum == incremental_checksum}")
    print(f"  Overhead: {((incremental_time - full_time) / full_time * 100):.1f}%")
    
    # Memory efficiency demonstration
    print(f"\nMemory efficient processing simulation:")
    
    def process_large_data_stream(data, chunk_size=8192):
        """Simulate processing a large data stream with CRC32C"""
        checksum = 0
        total_bytes = 0
        
        # Simulate reading data in chunks
        for i in range(0, len(data), chunk_size):
            chunk = data[i:i + chunk_size]
            total_bytes += len(chunk)
            
            if checksum == 0:
                checksum = google_crc32c.value(chunk)
            else:
                checksum = google_crc32c.extend(checksum, chunk)
            
            # Simulate processing delay
            if i % (chunk_size * 100) == 0:  # Progress every ~100 chunks
                progress = (total_bytes / len(data)) * 100
                yield progress, checksum
        
        yield 100.0, checksum
    
    # Test with streaming data
    stream_data = b"Stream processing test. " * 50000
    
    print(f"  Processing {len(stream_data):,} bytes in {chunk_size:,} byte chunks:")
    
    last_progress = 0
    for progress, current_checksum in process_large_data_stream(stream_data):
        if progress >= last_progress + 20 or progress >= 100:  # Show every 20%
            print(f"    {progress:5.1f}% - CRC32C: 0x{current_checksum:08x}")
            last_progress = progress
    
    # Verify final result
    final_checksum = google_crc32c.value(stream_data)
    print(f"  Streaming result matches direct: {current_checksum == final_checksum}")

except Exception as e:
    print(f"Performance testing error: {e}")
    
    # Show performance concepts
    print("CRC32C performance advantages:")
    advantages = [
        "Hardware acceleration on modern CPUs",
        "Faster than MD5/SHA for integrity checking",
        "Low memory footprint",
        "Efficient incremental calculation",
        "Optimized for Google Cloud workloads"
    ]
    
    for advantage in advantages:
        print(f"  • {advantage}")
    raise

print("Example 4 completed")
```
