# Google APIs Common Protos - Common Protocol Buffer Types

Google APIs Common Protos contains common protocol buffer types used across Google APIs, including standard types for dates, money, addresses, and RPC status codes.

## Example 1: Basic Protocol Buffer Types

```python
# Import common Google API protocol buffer types
from google.api import field_behavior_pb2
from google.api import resource_pb2
from google.type import date_pb2
from google.type import datetime_pb2
from google.rpc import status_pb2
from google.rpc import code_pb2

print("Google API Common Protobuf Types:")
print("✓ field_behavior_pb2 - Field behavior annotations")
print("✓ resource_pb2 - Resource pattern definitions")
print("✓ date_pb2 - Date representations")
print("✓ datetime_pb2 - DateTime representations") 
print("✓ status_pb2 - RPC status messages")
print("✓ code_pb2 - Standard gRPC error codes")


# Test RPC status codes
print(f"\nStandard gRPC Status Codes:")
status_codes = [
    (code_pb2.OK, "OK"),
    (code_pb2.CANCELLED, "CANCELLED"),
    (code_pb2.UNKNOWN, "UNKNOWN"),
    (code_pb2.INVALID_ARGUMENT, "INVALID_ARGUMENT"),
    (code_pb2.DEADLINE_EXCEEDED, "DEADLINE_EXCEEDED"),
    (code_pb2.NOT_FOUND, "NOT_FOUND"),
    (code_pb2.ALREADY_EXISTS, "ALREADY_EXISTS"),
    (code_pb2.PERMISSION_DENIED, "PERMISSION_DENIED"),
    (code_pb2.INTERNAL, "INTERNAL"),
    (code_pb2.UNAVAILABLE, "UNAVAILABLE")
]

for code_value, code_name in status_codes:
    print(f"  {code_value}: {code_name}")
    assert code_pb2.Code.Name(code_value) == code_name, (code_value, code_name)
assert [code_value for code_value, _ in status_codes] == [0, 1, 2, 3, 4, 5, 6, 7, 13, 14], status_codes
```

## Example 2: Date and Time Protobuf Messages

```python
from google.type import date_pb2, datetime_pb2, dayofweek_pb2
from google.type import timeofday_pb2
import datetime as py_datetime

print("Date and Time Protocol Buffer Messages")

# Create a date protobuf message
today = py_datetime.date.today()
date_proto = date_pb2.Date()
date_proto.year = today.year
date_proto.month = today.month
date_proto.day = today.day

print(f"\nDate Protobuf:")
print(f"Year: {date_proto.year}")
print(f"Month: {date_proto.month}")
print(f"Day: {date_proto.day}")
print(f"Serialized size: {date_proto.ByteSize()} bytes")

# Create a time of day protobuf message
now = py_datetime.datetime.now().time()
time_proto = timeofday_pb2.TimeOfDay()
time_proto.hours = now.hour
time_proto.minutes = now.minute
time_proto.seconds = now.second
time_proto.nanos = now.microsecond * 1000  # Convert microseconds to nanoseconds

print(f"\nTime of Day Protobuf:")
print(f"Hours: {time_proto.hours}")
print(f"Minutes: {time_proto.minutes}")
print(f"Seconds: {time_proto.seconds}")
print(f"Nanoseconds: {time_proto.nanos}")

# Test day of week enumeration
print(f"\nDay of Week Values:")
days = [
    (dayofweek_pb2.MONDAY, "MONDAY"),
    (dayofweek_pb2.TUESDAY, "TUESDAY"),
    (dayofweek_pb2.WEDNESDAY, "WEDNESDAY"),
    (dayofweek_pb2.THURSDAY, "THURSDAY"),
    (dayofweek_pb2.FRIDAY, "FRIDAY"),
    (dayofweek_pb2.SATURDAY, "SATURDAY"),
    (dayofweek_pb2.SUNDAY, "SUNDAY")
]

for day_value, day_name in days:
    print(f"  {day_value}: {day_name}")
```

## Example 3: RPC Status and Error Handling

```python
from google.rpc import status_pb2, code_pb2, error_details_pb2
from google.protobuf import any_pb2

print("RPC Status and Error Handling")

# Create various status messages
statuses = [
    {
        "code": code_pb2.OK,
        "message": "Request completed successfully",
        "name": "SUCCESS"
    },
    {
        "code": code_pb2.INVALID_ARGUMENT,
        "message": "The request contains invalid parameters",
        "name": "INVALID_ARGUMENT"
    },
    {
        "code": code_pb2.NOT_FOUND,
        "message": "The requested resource was not found",
        "name": "NOT_FOUND"
    },
    {
        "code": code_pb2.PERMISSION_DENIED,
        "message": "The caller does not have permission to execute the specified operation",
        "name": "PERMISSION_DENIED"
    },
    {
        "code": code_pb2.INTERNAL,
        "message": "Internal server error occurred",
        "name": "INTERNAL_ERROR"
    }
]

print("Status Message Examples:")
for status_info in statuses:
    status_msg = status_pb2.Status()
    status_msg.code = status_info["code"]
    status_msg.message = status_info["message"]
    
    print(f"\n{status_info['name']}:")
    print(f"  Code: {status_msg.code}")
    print(f"  Message: {status_msg.message}")
    print(f"  Serialized size: {status_msg.ByteSize()} bytes")

# Test error details
print(f"\nError Details Capabilities:")
try:
    # Create a basic error info
    error_info = error_details_pb2.ErrorInfo()
    error_info.reason = "RESOURCE_EXHAUSTED"
    error_info.domain = "googleapis.com"
    
    print(f"✓ ErrorInfo - Reason: {error_info.reason}, Domain: {error_info.domain}")
except:
    print("ErrorInfo details not available")
    raise

print(f"\nStatus protobuf validation completed!")
```

## Example 5: Protocol Buffer Serialization and Validation

```python
from google.type import money_pb2, postal_address_pb2
from google.rpc import status_pb2, code_pb2
from google.protobuf import message

print("Protocol Buffer Serialization and Validation")

# Create a money protobuf message
money = money_pb2.Money()
money.currency_code = "USD"
money.units = 123
money.nanos = 450000000  # $123.45

print(f"Money Protobuf:")
print(f"Currency: {money.currency_code}")
print(f"Amount: ${money.units}.{money.nanos//1000000:02d}")
print(f"Serialized: {money.SerializeToString()[:20]}... ({money.ByteSize()} bytes)")

# Create postal address
address = postal_address_pb2.PostalAddress()
address.region_code = "US"
address.postal_code = "94043"
address.administrative_area = "CA"
address.locality = "Mountain View"
address.address_lines.append("1600 Amphitheatre Parkway")

print(f"\nPostal Address Protobuf:")
print(f"Country: {address.region_code}")
print(f"State: {address.administrative_area}")
print(f"City: {address.locality}")
print(f"ZIP: {address.postal_code}")
print(f"Address: {address.address_lines[0] if address.address_lines else 'N/A'}")

# Test protobuf validation
print(f"\nProtobuf Validation:")
print(f"Money message valid: {money.IsInitialized()}")
print(f"Address message valid: {address.IsInitialized()}")

# Test serialization roundtrip
money_bytes = money.SerializeToString()
money_copy = money_pb2.Money()
money_copy.ParseFromString(money_bytes)

print(f"\nSerialization roundtrip test:")
print(f"Original amount: ${money.units}.{money.nanos//1000000:02d}")
print(f"Deserialized amount: ${money_copy.units}.{money_copy.nanos//1000000:02d}")
print(f"Roundtrip successful: {money.units == money_copy.units and money.nanos == money_copy.nanos}")

print(f"\nProtobuf operations completed successfully!")
```
