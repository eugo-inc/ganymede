## Example 1: gpustat functionality test 1

```python
import gpustat

# Example 1: Basic GPU detection and counting
print("Example 1: Basic GPU detection and counting")

try:
    # Check if GPUs are available
    gpu_count = gpustat.gpu_count()
    print(f"✓ GPU detection:")
    print(f"  Number of GPUs detected: {gpu_count}")
    
    if gpu_count > 0:
        print(f"  GPU(s) available for monitoring")
        
        # Try to get basic GPU stats
        try:
            gpu_stats = gpustat.GPUStatCollection.new_query()
            print(f"  GPU stats collection created: {type(gpu_stats).__name__}")
            print(f"  Total GPUs in collection: {len(gpu_stats)}")
        except Exception as e:
            print(f"  GPU stats query info: {str(e)[:50]}...")
            raise
            
    else:
        print(f"  No GPUs detected (running in CPU-only mode)")
        print(f"  This is normal on systems without NVIDIA GPUs")
    
    # Check available functionality
    core_functions = [attr for attr in dir(gpustat.core) if not attr.startswith('_')]
    print(f"✓ Core functionality: {len(core_functions)} functions")
    print(f"  Key functions: {core_functions[:3]}")
    
    print("✓ GPU detection and counting completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"gpustat example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished\n")
```

## Example 2: gpustat functionality test 2

```python
import gpustat
import time

# Example 2: GPU statistics collection and monitoring
print("Example 2: GPU statistics collection and monitoring")

try:
    # The marker above skips this cell where there is no NVIDIA driver, so the query must succeed here
    stats_collection = gpustat.GPUStatCollection.new_query()
    print(f"✓ GPU statistics collection created")
    print(f"  Collection type: {type(stats_collection).__name__}")
    print(f"  Number of GPUs: {len(stats_collection)}")
    assert len(stats_collection) == gpustat.gpu_count() >= 1, (len(stats_collection), gpustat.gpu_count())

    for i, gpu_stat in enumerate(stats_collection):
        print(f"  GPU {i}:")
        print(f"    Type: {type(gpu_stat).__name__}")
        print(f"    Name: {gpu_stat.name}")
        assert gpu_stat.uuid.startswith("GPU-") and gpu_stat.name, gpu_stat.entry
        if gpu_stat.entry["memory.used"] is None:
            print(f"    Memory: unified (NVML reports no dedicated framebuffer)")
        else:
            print(f"    Memory used: {gpu_stat.memory_used} / {gpu_stat.memory_total} MB")
            assert 0 <= gpu_stat.memory_used <= gpu_stat.memory_total, gpu_stat.entry
    
    # Test collection refresh functionality
    print(f"✓ Collection refresh capability:")
    if hasattr(gpustat.GPUStatCollection, 'new_query'):
        print(f"  new_query method available")
    
    # Check for timestamp functionality
    current_time = time.time()
    print(f"✓ Monitoring timestamp: {current_time:.0f}")
    
    print("✓ GPU statistics monitoring demonstrated")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"gpustat example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished\n")
```

## Example 3: gpustat functionality test 3

```python
import gpustat
import json

# Example 3: GPU statistics formatting and output
print("Example 3: GPU statistics formatting and output")

try:
    # Test different output formats
    print(f"✓ Testing output formatting:")
    
    # Check for JSON output capability
    try:
        stats = gpustat.GPUStatCollection.new_query()
        
        # Try JSON serialization if stats are available
        if hasattr(stats, 'jsonify'):
            json_output = stats.jsonify()
            print(f"  JSON output available: {type(json_output)}")
            print(f"  JSON length: {len(json_output) if json_output else 0}")
        else:
            print(f"  JSON output method not found")
            
        # Try string representation
        if len(stats) > 0:
            str_output = str(stats)
            print(f"  String output length: {len(str_output)}")
            print(f"  String preview: {str_output[:50]}...")
        else:
            print(f"  No GPU data for string formatting")
            
    except Exception as format_error:
        print(f"  Format test: {str(format_error)[:50]}...")
        raise
    
    # Test CLI formatting options
    cli_module = gpustat.cli
    print(f"✓ CLI module available: {type(cli_module)}")
    
    # Check for formatting functions
    core_module = gpustat.core
    core_attrs = [attr for attr in dir(core_module) if 'format' in attr.lower()]
    print(f"✓ Core formatting functions: {len(core_attrs)}")
    if core_attrs:
        print(f"  Available: {core_attrs}")
    
    # Test basic statistics structure
    print(f"✓ GPUStat class inspection:")
    gpu_stat_attrs = [attr for attr in dir(gpustat.GPUStat) if not attr.startswith('_')]
    key_attrs = [attr for attr in gpu_stat_attrs if any(word in attr for word in 
                ['memory', 'temp', 'util', 'name', 'uuid'])]
    print(f"  Key attributes: {key_attrs[:5]}")
    
    print("✓ GPU statistics formatting demonstrated")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"gpustat example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished\n")
```

## Example 5: gpustat functionality test 5

```python
import gpustat
import os

# Example 5: GPU monitoring utilities and system integration
print("Example 5: GPU monitoring utilities and system integration")

try:
    # Test GPU detection methods
    print(f"✓ GPU detection methods:")
    
    # Primary detection
    gpu_count = gpustat.gpu_count()
    print(f"  gpu_count(): {gpu_count}")
    
    # Alternative detection through collection
    try:
        collection = gpustat.GPUStatCollection.new_query()
        collection_count = len(collection)
        print(f"  Collection count: {collection_count}")
        print(f"  Detection consistency: {gpu_count == collection_count}")
    except Exception as e:
        print(f"  Collection query: {str(e)[:40]}...")
        raise
    
    # Check environment variables
    nvidia_vars = [var for var in os.environ if 'NVIDIA' in var or 'CUDA' in var]
    print(f"✓ NVIDIA environment: {len(nvidia_vars)} variables")
    if nvidia_vars:
        print(f"  Sample: {nvidia_vars[0] if nvidia_vars else 'None'}")
    
    # Test utility functions
    core_utils = [attr for attr in dir(gpustat.core) if not attr.startswith('_')]
    utility_funcs = [func for func in core_utils if any(word in func for word in 
                    ['check', 'get', 'find', 'detect'])]
    print(f"✓ Utility functions: {len(utility_funcs)}")
    if utility_funcs:
        print(f"  Available: {utility_funcs[:3]}")
    
    # System integration check
    print(f"✓ System integration:")
    print(f"  Module path: {gpustat.__file__ if hasattr(gpustat, '__file__') else 'Unknown'}")
    
    # Check for daemon/service capabilities
    daemon_attrs = [attr for attr in dir(gpustat) if any(word in attr.lower() for word in 
                   ['daemon', 'service', 'watch', 'monitor'])]
    print(f"✓ Service capabilities: {len(daemon_attrs)} attributes")
    
    print("✓ GPU monitoring utilities demonstrated")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"gpustat example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished\n")
```
