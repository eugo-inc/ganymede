```python
# Example 1: Basic GPU Detection and Information Retrieval
import GPUtil

print("Example 1: Basic GPU Detection and Information")
print("=" * 50)

# Get all available GPUs
try:
    gpus = GPUtil.getGPUs()
    
    if len(gpus) == 0:
        print("No NVIDIA GPUs detected on this system")
        print("This is expected on systems without NVIDIA GPUs (e.g., macOS, CPU-only machines)")
        print("\nGPUtil-fix successfully imported and attempted GPU detection")
    else:
        print(f"Found {len(gpus)} GPU(s):\n")
        
        for i, gpu in enumerate(gpus):
            print(f"GPU {i}: {gpu.name}")
            print(f"  ID: {gpu.id}")
            print(f"  UUID: {gpu.uuid}")
            print(f"  Memory Total: {gpu.memoryTotal} MB")
            print(f"  Memory Used: {gpu.memoryUsed} MB")
            print(f"  Memory Free: {gpu.memoryFree} MB")
            print(f"  Memory Util: {gpu.memoryUtil * 100:.1f}%")
            print(f"  GPU Load: {gpu.load * 100:.1f}%")
            print(f"  Temperature: {gpu.temperature}°C")
            print(f"  Driver: {gpu.driver}")
            print()
        
        print("✅ GPU detection and information retrieval successful")
    
except Exception as e:
    print(f"Note: {type(e).__name__}: {e}")
    print("This is expected on systems without NVIDIA GPUs or nvidia-smi")
    print("GPUtil-fix package is working correctly")
    raise AssertionError(
        f"gputil_fix example (cell 0) failed: {type(e).__name__}: {e}"
    ) from e

print("\nExample 1 complete")
```

```python
# Example 2: GPU Utilization Display and Monitoring
import GPUtil
import time

print("Example 2: GPU Utilization Display and Monitoring")
print("=" * 50)

try:
    # Display current GPU utilization in a formatted table
    print("\nCurrent GPU Utilization:")
    GPUtil.showUtilization()
    
    # Simulate monitoring over time (3 snapshots)
    print("\nMonitoring GPU utilization (3 snapshots):")
    for i in range(3):
        print(f"\nSnapshot {i+1}:")
        gpus = GPUtil.getGPUs()
        
        if len(gpus) > 0:
            for gpu in gpus:
                print(f"  GPU {gpu.id}: Load={gpu.load*100:.1f}%, "
                      f"Memory={gpu.memoryUtil*100:.1f}%, "
                      f"Temp={gpu.temperature}°C")
        else:
            print("  No GPUs available for monitoring")
        
        if i < 2:  # Don't sleep after last iteration
            time.sleep(0.5)
    
    print("\n✅ GPU monitoring demonstration complete")
    
except Exception as e:
    print(f"\nNote: GPU monitoring not available - {type(e).__name__}")
    print("This is expected on systems without NVIDIA GPUs")
    print("On systems with NVIDIA GPUs, this would display real-time utilization")
    raise AssertionError(
        f"gputil_fix example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("\nExample 2 complete")
```

```python
# Example 3: Finding Available GPUs with Filters
import GPUtil

print("Example 3: Finding Available GPUs with Filters")
print("=" * 50)

try:
    # Get available GPUs with different criteria
    print("\nSearching for available GPUs with various criteria:\n")
    
    # Default criteria (load < 0.5, memory < 0.5)
    available_default = GPUtil.getAvailable(order='first', limit=999)
    print(f"1. Default criteria (load<50%, mem<50%): {available_default if available_default else 'None'}")
    
    # Strict criteria (load < 0.1, memory < 0.1)
    available_strict = GPUtil.getAvailable(
        order='first',
        limit=999,
        maxLoad=0.1,
        maxMemory=0.1
    )
    print(f"2. Strict criteria (load<10%, mem<10%): {available_strict if available_strict else 'None'}")
    
    # Lenient criteria (load < 0.9, memory < 0.9)
    available_lenient = GPUtil.getAvailable(
        order='first',
        limit=999,
        maxLoad=0.9,
        maxMemory=0.9
    )
    print(f"3. Lenient criteria (load<90%, mem<90%): {available_lenient if available_lenient else 'None'}")
    
    # Different ordering strategies
    print("\nDifferent ordering strategies:")
    
    order_strategies = ['first', 'last', 'random', 'load', 'memory']
    for strategy in order_strategies:
        try:
            available = GPUtil.getAvailable(order=strategy, limit=2, maxLoad=0.9, maxMemory=0.9)
            print(f"  {strategy:8s}: {available if available else 'None'}")
        except Exception as e:
            print(f"  {strategy:8s}: (error - {type(e).__name__})")
            raise
    
    # Get availability as binary list
    print("\nGPU availability as binary list:")
    availability = GPUtil.getAvailability(GPUtil.getGPUs(), maxLoad=0.5, maxMemory=0.5)
    print(f"  Availability: {availability if availability else 'No GPUs detected'}")
    
    if len(available_default) > 0:
        print("\n✅ Found available GPUs successfully")
    else:
        print("\n✅ GPU availability check complete (no available GPUs with current criteria)")
    
except Exception as e:
    print(f"\nNote: GPU availability check not possible - {type(e).__name__}")
    print("This is expected on systems without NVIDIA GPUs")
    print("On GPU systems, this would identify available GPUs based on load/memory thresholds")
    raise AssertionError(
        f"gputil_fix example (cell 2) failed: {type(e).__name__}: {e}"
    ) from e

print("\nExample 3 complete")
```

```python
# Example 4: First Available GPU Selection with Retry Logic
import GPUtil
import time

print("Example 4: First Available GPU Selection with Retry Logic")
print("=" * 50)

try:
    print("\nAttempting to get first available GPU with retry logic...\n")
    
    # Try to get first available GPU with multiple attempts
    # In production, this is useful when waiting for GPUs to become free
    start_time = time.time()
    
    # Attempt 1: With lenient criteria
    print("Attempt 1: Lenient criteria (load<90%, mem<90%)")
    try:
        device_ids = GPUtil.getFirstAvailable(
            order='first',
            maxLoad=0.9,
            maxMemory=0.9,
            attempts=3,
            interval=0.5,  # Wait 0.5 seconds between attempts
            verbose=False
        )
        print(f"  Found GPU(s): {device_ids}")
        lenient_success = True
    except RuntimeError as e:
        print(f"  No GPU available: {e}")
        lenient_success = False
    
    # Attempt 2: With strict criteria
    print("\nAttempt 2: Strict criteria (load<10%, mem<10%)")
    try:
        device_ids_strict = GPUtil.getFirstAvailable(
            order='first',
            maxLoad=0.1,
            maxMemory=0.1,
            attempts=2,
            interval=0.3,
            verbose=False
        )
        print(f"  Found GPU(s): {device_ids_strict}")
        strict_success = True
    except RuntimeError as e:
        print(f"  No GPU available: {e}")
        strict_success = False
    
    elapsed_time = time.time() - start_time
    print(f"\nTotal time for GPU selection attempts: {elapsed_time:.2f}s")
    
    # Demonstrate GPU exclusion
    print("\nDemonstrating GPU exclusion:")
    gpus = GPUtil.getGPUs()
    if len(gpus) > 0:
        # Exclude first GPU by ID
        excluded_by_id = GPUtil.getAvailable(
            order='first',
            limit=999,
            maxLoad=0.9,
            maxMemory=0.9,
            excludeID=[0] if len(gpus) > 0 else []
        )
        print(f"  Available GPUs (excluding ID 0): {excluded_by_id if excluded_by_id else 'None'}")
        
        # Exclude by UUID
        if len(gpus) > 0:
            first_uuid = gpus[0].uuid
            excluded_by_uuid = GPUtil.getAvailable(
                order='first',
                limit=999,
                maxLoad=0.9,
                maxMemory=0.9,
                excludeUUID=[first_uuid]
            )
            print(f"  Available GPUs (excluding UUID {first_uuid[:16]}...): {excluded_by_uuid if excluded_by_uuid else 'None'}")
    else:
        print("  No GPUs to exclude from")
    
    if lenient_success or strict_success:
        print("\n✅ GPU selection with retry logic successful")
    else:
        print("\n✅ GPU selection logic tested (no available GPUs found)")
    
except Exception as e:
    print(f"\nNote: GPU selection not available - {type(e).__name__}: {e}")
    print("This is expected on systems without NVIDIA GPUs")
    print("On GPU systems, this would select the first available GPU with retry logic")
    raise AssertionError(
        f"gputil_fix example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("\nExample 4 complete")
```

```python
# Example 5: Production Deep Learning GPU Setup
import GPUtil
import os

print("Example 5: Production Deep Learning GPU Setup")
print("=" * 50)

try:
    # Step 1: Configure GPU ordering for frameworks
    print("\nStep 1: Configure GPU device ordering")
    os.environ["CUDA_DEVICE_ORDER"] = "PCI_BUS_ID"
    print("  Set CUDA_DEVICE_ORDER=PCI_BUS_ID")
    
    # Step 2: Find best available GPU
    print("\nStep 2: Select optimal GPU for training")
    
    gpus = GPUtil.getGPUs()
    
    if len(gpus) == 0:
        print("  No NVIDIA GPUs detected")
        print("  Would use CPU for training")
        selected_gpu = None
    else:
        # Get GPU with lowest memory usage
        available_gpus = GPUtil.getAvailable(
            order='memory',  # Order by lowest memory usage
            limit=1,         # Get only the best one
            maxLoad=0.8,     # Accept GPUs with load < 80%
            maxMemory=0.8    # Accept GPUs with memory < 80%
        )
        
        if len(available_gpus) > 0:
            selected_gpu = available_gpus[0]
            print(f"  Selected GPU {selected_gpu}")
            
            # Get detailed info about selected GPU
            gpu_info = gpus[selected_gpu]
            print(f"  GPU Name: {gpu_info.name}")
            print(f"  GPU Load: {gpu_info.load * 100:.1f}%")
            print(f"  Memory Available: {gpu_info.memoryFree} MB / {gpu_info.memoryTotal} MB")
            print(f"  Temperature: {gpu_info.temperature}°C")
            
            # Step 3: Set environment variable for frameworks
            print("\nStep 3: Configure CUDA_VISIBLE_DEVICES")
            os.environ["CUDA_VISIBLE_DEVICES"] = str(selected_gpu)
            print(f"  Set CUDA_VISIBLE_DEVICES={selected_gpu}")
            print("  TensorFlow/PyTorch will now use only this GPU")
        else:
            print("  No available GPUs meeting criteria (load<80%, mem<80%)")
            print("  Attempting with more lenient criteria...")
            
            # Fallback: try with very lenient criteria
            fallback_gpus = GPUtil.getAvailable(
                order='memory',
                limit=1,
                maxLoad=0.95,
                maxMemory=0.95
            )
            
            if len(fallback_gpus) > 0:
                selected_gpu = fallback_gpus[0]
                print(f"  Found GPU {selected_gpu} with lenient criteria")
                os.environ["CUDA_VISIBLE_DEVICES"] = str(selected_gpu)
            else:
                print("  All GPUs are heavily utilized, would use CPU or wait")
                selected_gpu = None
    
    # Step 4: Display final configuration
    print("\nStep 4: Final Configuration Summary")
    print("  Environment Variables:")
    print(f"    CUDA_DEVICE_ORDER: {os.environ.get('CUDA_DEVICE_ORDER', 'not set')}")
    print(f"    CUDA_VISIBLE_DEVICES: {os.environ.get('CUDA_VISIBLE_DEVICES', 'not set')}")
    
    if selected_gpu is not None:
        print(f"\n  Ready to start training on GPU {selected_gpu}")
        print("  Example framework usage:")
        print("    # TensorFlow: tf.config.list_physical_devices('GPU')")
        print("    # PyTorch: torch.cuda.is_available() and torch.cuda.current_device()")
    else:
        print("\n  Would proceed with CPU training or implement wait-and-retry logic")
    
    # Step 5: Create a monitoring configuration
    print("\nStep 5: Monitoring Configuration")
    
    if len(gpus) > 0:
        print("  During training, you would monitor:")
        for gpu in gpus:
            print(f"    GPU {gpu.id}: Check load, memory, and temperature periodically")
        print("  Use GPUtil.showUtilization() or GPUtil.getGPUs() in a separate thread")
    else:
        print("  No GPUs to monitor (CPU training mode)")
    
    print("\n✅ Production GPU setup complete")
    print("Ready for deep learning framework initialization!")
    
except Exception as e:
    print(f"\nNote: GPU setup encountered issue - {type(e).__name__}: {e}")
    print("This is expected on systems without NVIDIA GPUs")
    print("On GPU systems, this would configure optimal GPU selection for training")
    print("\nFor CPU-only systems, training would proceed without GPU acceleration")
    raise AssertionError(
        f"gputil_fix example (cell 4) failed: {type(e).__name__}: {e}"
    ) from e

print("\nExample 5 complete")
```
