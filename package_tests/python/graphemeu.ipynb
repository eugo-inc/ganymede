## Example 1: Length as a reader sees it

```python
import grapheme

flag = "🇫🇷"  # one flag made of 2 regional-indicator code points
assert len(flag) == 2 and grapheme.length(flag) == 1
word = "e\u0301t\u00e9"  # é written two ways: e + combining accent, and the single code point
assert len(word) == 4 and grapheme.length(word) == 3, grapheme.length(word)
thumb = "👍🏽"  # an emoji with a skin-tone modifier
assert grapheme.length(thumb) == 1
print(len(flag), "code points ->", grapheme.length(flag), "grapheme")
```

## Example 2: Split into graphemes

```python
import grapheme

text = "🇫🇷 ok"
parts = list(grapheme.graphemes(text))
assert parts == ["🇫🇷", " ", "o", "k"], parts
print(parts)
```

## Example 3: Slice without breaking a character

```python
import grapheme

text = "a👍🏽b"
assert grapheme.slice(text, 1, 2) == "👍🏽"
assert grapheme.slice(text, 0, 2) == "a👍🏽"
print(repr(grapheme.slice(text, 1, 2)))
```

## Example 4: Search by graphemes

```python
import grapheme

text = "noe\u0308l"  # "noël" with a combining diaeresis
assert grapheme.contains(text, "e\u0308")
assert not grapheme.contains(text, "e")  # the bare e is part of a larger character
assert grapheme.startswith(text, "no") and grapheme.endswith(text, "l")
print("contains/startswith/endswith ok")
```

## Example 5: Safe split index and Unicode version

```python
import grapheme

text = "ab🇫🇷cd"
index = grapheme.safe_split_index(text, 3)  # never splits the two code points of the flag
assert index == 2, index
print("safe index", index, "| Unicode", grapheme.UNICODE_VERSION)
assert grapheme.UNICODE_VERSION
```
