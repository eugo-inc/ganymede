## Example 1: Basic Import and Inspection\n
\n
Basic import test and inspection of graphviz package structure.\n

```python
# Example 1: Basic Directed Graph Creation
import graphviz

print("Example 1: Basic Directed Graph (Digraph)")
print("=" * 40)

# Create a simple directed graph
dot = graphviz.Digraph(comment='Organization Chart')

# Add nodes
dot.node('CEO', 'Chief Executive Officer')
dot.node('CTO', 'Chief Technology Officer')
dot.node('CFO', 'Chief Financial Officer')
dot.node('Dev1', 'Developer 1')
dot.node('Dev2', 'Developer 2')
dot.node('Acc1', 'Accountant')

# Add edges (relationships)
dot.edge('CEO', 'CTO')
dot.edge('CEO', 'CFO')
dot.edge('CTO', 'Dev1')
dot.edge('CTO', 'Dev2')
dot.edge('CFO', 'Acc1')

print("Graph created successfully!")
print(f"Graph format: {dot.format}")
print(f"Graph engine: {dot.engine}")
print(f"\nDOT source code:")
print(dot.source)

# Verify the graph structure
print(f"\nGraph has {dot.source.count('->') + dot.source.count('--')} edges")
print(f"Comment: {dot.comment}")

print("\n✅ Example 1 completed successfully!")
```

## Example 2: Core Functionality Test\n
\n
Test core functionality of graphviz package.\n

```python
# Example 2: Undirected Graph with Styling
import graphviz

print("Example 2: Undirected Graph (Graph) with Styling")
print("=" * 50)

# Create an undirected graph
graph = graphviz.Graph(comment='Network Topology', format='svg')

# Set graph attributes
graph.attr(rankdir='LR', bgcolor='lightgray')
graph.attr('node', shape='box', style='filled', fillcolor='lightblue')
graph.attr('edge', color='red', style='dashed')

# Add nodes with custom attributes
graph.node('Router', 'Main Router', shape='diamond', fillcolor='orange')
graph.node('Server1', 'Web Server', fillcolor='lightgreen')
graph.node('Server2', 'Database Server', fillcolor='lightgreen')
graph.node('Client1', 'Client PC 1')
graph.node('Client2', 'Client PC 2')

# Add edges in undirected graph
graph.edge('Router', 'Server1', label='1Gbps')
graph.edge('Router', 'Server2', label='1Gbps')
graph.edge('Router', 'Client1', label='100Mbps')
graph.edge('Router', 'Client2', label='100Mbps')
graph.edge('Server1', 'Server2', label='10Gbps', color='blue', style='bold')

print("Graph attributes set successfully!")
print(f"Graph format: {graph.format}")
print(f"Graph engine: {graph.engine}")
print(f"\nDOT source code:")
print(graph.source)

# Check graph properties
print(f"\nGraph type: Undirected (uses '--' for edges)")
print(f"Number of connections: {graph.source.count('--')}")

print("\n✅ Example 2 completed successfully!")
```

## Example 3: Error Handling and Edge Cases\n
\n
Test error handling and edge cases for graphviz.\n

```python
# Example 3: Subgraphs and Clusters
import graphviz

print("Example 3: Subgraphs and Clusters")
print("=" * 35)

# Create main digraph
main = graphviz.Digraph(comment='System Architecture', engine='dot')
main.attr(rankdir='TB', compound='true')

# Create cluster for frontend
with main.subgraph(name='cluster_frontend') as frontend:
    frontend.attr(label='Frontend Layer', style='filled', color='lightblue')
    frontend.node('ui1', 'React App')
    frontend.node('ui2', 'Mobile App')
    frontend.node('ui3', 'Admin Portal')

# Create cluster for backend
with main.subgraph(name='cluster_backend') as backend:
    backend.attr(label='Backend Layer', style='filled', color='lightgreen')
    backend.node('api1', 'REST API')
    backend.node('api2', 'GraphQL API')
    backend.node('ws', 'WebSocket Server')

# Create cluster for database
with main.subgraph(name='cluster_db') as db:
    db.attr(label='Data Layer', style='filled', color='lightyellow')
    db.node('db1', 'PostgreSQL')
    db.node('db2', 'Redis Cache')
    db.node('db3', 'MongoDB')

# Add edges between clusters
main.edge('ui1', 'api1', label='HTTP')
main.edge('ui2', 'api2', label='GraphQL')
main.edge('ui3', 'api1')
main.edge('api1', 'db1', label='SQL')
main.edge('api2', 'db3', label='NoSQL')
main.edge('ws', 'db2', label='Cache')

print("Subgraphs created successfully!")
print(f"\nDOT source code:")
print(main.source)

# Count clusters
cluster_count = main.source.count('subgraph cluster_')
print(f"\nNumber of clusters: {cluster_count}")
print(f"Total nodes across all layers: {main.source.count('[label=')}")

print("\n✅ Example 3 completed successfully!")
```

## Example 4: Integration and Usage Patterns\n
\n
Test integration patterns and common usage of graphviz.\n

```python
# Example 4: Using Different Engines and Formats
import graphviz

print("Example 4: Different Engines and Formats")
print("=" * 40)

# Test different layout engines
engines = ['dot', 'neato', 'circo', 'fdp']

for engine_name in engines:
    print(f"\n{engine_name.upper()} Engine:")
    print("-" * 30)
    
    # Create a graph with specific engine
    g = graphviz.Graph(engine=engine_name, format='pdf')
    
    # Add some nodes
    g.node('A', 'Node A')
    g.node('B', 'Node B')
    g.node('C', 'Node C')
    g.node('D', 'Node D')
    
    # Add edges
    g.edges(['AB', 'BC', 'CD', 'DA', 'AC'])
    
    print(f"  Engine: {g.engine}")
    print(f"  Format: {g.format}")
    print(f"  Nodes: {g.source.count('[label=')}")
    print(f"  Edges: {g.source.count('--')}")

# Test different output formats
print("\n\nSupported Output Formats:")
print("-" * 30)
formats = ['pdf', 'png', 'svg', 'ps', 'dot']

for fmt in formats:
    g = graphviz.Digraph(format=fmt)
    g.node('Test', 'Test Node')
    print(f"  Format '{fmt}': {g.format}")

# Create a graph with DOT source directly
print("\n\nCreating graph from DOT source:")
print("-" * 30)
dot_source = """
digraph {
    Start -> Process -> End
    Process -> Decision
    Decision -> Yes -> End
    Decision -> No -> Process
}
"""

from_source = graphviz.Source(dot_source)
print(f"Graph created from source")
print(f"Source length: {len(from_source.source)} characters")
print(f"Engine: {from_source.engine}")

print("\n✅ Example 4 completed successfully!")
```

## Example 5: Comprehensive Feature Test\n
\n
Comprehensive test of graphviz features and capabilities.\n

```python
# Example 5: Advanced Features - HTML-like Labels and Custom Shapes
import graphviz

print("Example 5: Advanced Features")
print("=" * 30)

# Create a graph with HTML-like labels
html_graph = graphviz.Digraph(comment='HTML Labels Demo')

# Use HTML-like labels for rich formatting
html_graph.node('record1', '''<
<TABLE BORDER="0" CELLBORDER="1" CELLSPACING="0">
  <TR><TD COLSPAN="2"><B>Server Info</B></TD></TR>
  <TR><TD>IP</TD><TD>192.168.1.1</TD></TR>
  <TR><TD>Status</TD><TD>Active</TD></TR>
</TABLE>>''', shape='plaintext')

html_graph.node('record2', '''<
<TABLE BORDER="0" CELLBORDER="1" CELLSPACING="0">
  <TR><TD PORT="f0">Field 1</TD></TR>
  <TR><TD PORT="f1">Field 2</TD></TR>
  <TR><TD PORT="f2">Field 3</TD></TR>
</TABLE>>''', shape='plaintext')

html_graph.edge('record1', 'record2:f1')

print("HTML-like labels created successfully!")

# Test various node shapes
print("\n\nNode Shape Demonstrations:")
print("-" * 30)

shapes_graph = graphviz.Digraph(comment='Node Shapes')
shapes_graph.attr(rankdir='LR')

# Various shapes
shapes = [
    ('box', 'Box'),
    ('circle', 'Circle'),
    ('ellipse', 'Ellipse'),
    ('diamond', 'Diamond'),
    ('hexagon', 'Hexagon'),
    ('triangle', 'Triangle'),
    ('parallelogram', 'Parallel'),
    ('trapezium', 'Trapezium')
]

for i, (shape, label) in enumerate(shapes):
    node_id = f'shape{i}'
    shapes_graph.node(node_id, label, shape=shape, style='filled', 
                     fillcolor='lightblue')
    if i > 0:
        shapes_graph.edge(f'shape{i-1}', node_id)

print(f"Created {len(shapes)} different node shapes")

# Edge routing and styling
print("\n\nEdge Routing and Styling:")
print("-" * 30)

edge_graph = graphviz.Digraph(comment='Edge Styles')
edge_graph.attr('node', shape='box')

# Add nodes
edge_graph.node('A', 'Start')
edge_graph.node('B', 'Process 1')
edge_graph.node('C', 'Process 2')
edge_graph.node('D', 'End')

# Different edge styles
edge_graph.edge('A', 'B', style='solid', label='solid', color='black')
edge_graph.edge('A', 'C', style='dashed', label='dashed', color='blue')
edge_graph.edge('B', 'D', style='dotted', label='dotted', color='red')
edge_graph.edge('C', 'D', style='bold', label='bold', color='green')

print(f"Edge styles demonstrated: solid, dashed, dotted, bold")

# Test graph copying and pipe operations
print("\n\nGraph Operations:")
print("-" * 30)

# Create a simple graph
original = graphviz.Graph()
original.node('X', 'Original')
original.node('Y', 'Node')
original.edge('X', 'Y')

# Copy the graph
copied = original.copy()
copied.node('Z', 'Copied Node')
copied.edge('Y', 'Z')

print(f"Original graph nodes: {original.source.count('[label=')}")
print(f"Copied graph nodes: {copied.source.count('[label=')}")

# Test pipe (get rendered output format without saving)
print("\n\nTesting pipe operation:")
try:
    # This would normally render the graph and return bytes
    # We'll just check the method exists
    print(f"  pipe method available: {hasattr(shapes_graph, 'pipe')}")
    print(f"  render method available: {hasattr(shapes_graph, 'render')}")
    print(f"  save method available: {hasattr(shapes_graph, 'save')}")
except Exception as e:
    print(f"  Note: {e}")
    raise

print("\n✅ Example 5 completed successfully!")
```
