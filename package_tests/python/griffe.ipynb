## Example 1: griffe functionality test 1

```python
import griffe

# Example 1: Loading and inspecting a Python module
print("Example 1: Loading and inspecting a Python module")

try:
    # Check what loading functions are available
    load_funcs = [attr for attr in dir(griffe) if 'load' in attr.lower()]
    print(f"✓ Loading functions available: {load_funcs}")
    
    # Try the main load function if available
    if hasattr(griffe, 'load'):
        try:
            # Load a simple built-in module
            module = griffe.load("json")
            print(f"✓ Module loaded: {module.name if hasattr(module, 'name') else 'json'}")
            print(f"  Module type: {type(module).__name__}")
            
            # Check if module has members
            if hasattr(module, 'members'):
                members = list(module.members.keys())
                print(f"  Members found: {len(members)}")
                print(f"  Sample members: {members[:5] if members else []}")
            else:
                print(f"  Module structure: {type(module)}")
                
        except Exception as load_error:
            print(f"✓ Load attempt: {str(load_error)[:50]}...")
            raise
    
    # Check for other module-related classes
    module_classes = [attr for attr in dir(griffe) if any(word in attr for word in 
                     ['Module', 'Package', 'Object'])]
    print(f"✓ Module-related classes: {len(module_classes)}")
    if module_classes:
        print(f"  Available: {module_classes[:5]}")
    
    # Check for collection types
    collection_types = [attr for attr in dir(griffe) if any(word in attr for word in 
                       ['Collection', 'Registry', 'Store'])]
    print(f"✓ Collection types: {len(collection_types)}")
    
    print("✓ Module loading exploration completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"griffe example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished\n")
```

## Example 2: griffe functionality test 2

```python
import griffe
import inspect

# Example 2: Analyzing function signatures and docstrings  
print("Example 2: Analyzing function signatures and docstrings")

try:
    # Create sample function for analysis
    def sample_function(arg1: str, arg2: int = 42, *args, **kwargs) -> bool:
        """Sample function with various parameter types.
        
        Args:
            arg1: A string parameter
            arg2: An integer with default value
            *args: Variable positional arguments
            **kwargs: Variable keyword arguments
            
        Returns:
            A boolean result
        """
        return len(arg1) > arg2
    
    # Inspect using griffe if possible, otherwise use standard inspection
    try:
        # Try to create a Function object from griffe
        if hasattr(griffe, 'Function'):
            func_obj = griffe.Function(
                name="sample_function",
                lineno=1
            )
            print(f"✓ Function object created: {type(func_obj).__name__}")
        else:
            print(f"✓ Function class not directly accessible")
            
    except Exception as func_error:
        print(f"✓ Function creation: {str(func_error)[:40]}...")
        raise
    
    # Analyze signature using Python's inspect
    sig = inspect.signature(sample_function)
    print(f"✓ Function signature analysis:")
    print(f"  Function name: sample_function")
    print(f"  Parameters: {len(sig.parameters)}")
    
    for name, param in sig.parameters.items():
        annotation = param.annotation if param.annotation != inspect.Parameter.empty else "None"
        default = param.default if param.default != inspect.Parameter.empty else "Required"
        print(f"    {name}: {annotation} (default: {default})")
    
    # Check return annotation
    return_annotation = sig.return_annotation if sig.return_annotation != inspect.Parameter.empty else "None"
    print(f"  Return type: {return_annotation}")
    
    # Get docstring
    doc = inspect.getdoc(sample_function)
    print(f"  Docstring length: {len(doc) if doc else 0}")
    
    print("✓ Function signature analysis completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"griffe example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished\n")
```

## Example 3: griffe functionality test 3

```python
import griffe

# Example 3: Working with class and attribute inspection
print("Example 3: Working with class and attribute inspection")

try:
    # Create a sample class for inspection
    class SampleClass:
        """A sample class for griffe inspection."""
        
        class_var: str = "default_value"
        
        def __init__(self, value: int):
            self.instance_var = value
            
        def method(self, param: str) -> str:
            """A sample method."""
            return f"{param}_{self.instance_var}"
            
        @property
        def computed_property(self) -> int:
            """A computed property."""
            return self.instance_var * 2
    
    # Try to create griffe objects
    print(f"✓ Sample class created for inspection")
    
    # Check available griffe classes for inspection
    inspection_classes = [attr for attr in dir(griffe) if attr in 
                         ['Class', 'Attribute', 'Method', 'Property']]
    print(f"✓ Inspection classes available: {inspection_classes}")
    
    # Try creating an Attribute object if available
    if hasattr(griffe, 'Attribute'):
        try:
            attr_obj = griffe.Attribute(name="class_var", lineno=1)
            print(f"  Attribute object: {type(attr_obj).__name__}")
            print(f"  Attribute name: {attr_obj.name}")
        except Exception as attr_error:
            print(f"  Attribute creation: {str(attr_error)[:40]}...")
            raise
    
    # Inspect class structure manually
    print(f"✓ Class structure analysis:")
    print(f"  Class name: SampleClass")
    print(f"  Methods: {len([m for m in dir(SampleClass) if not m.startswith('_')])}")
    print(f"  Has docstring: {bool(SampleClass.__doc__)}")
    
    # Check for annotations
    annotations = getattr(SampleClass, '__annotations__', {})
    print(f"  Class annotations: {len(annotations)}")
    for name, annotation in annotations.items():
        print(f"    {name}: {annotation}")
    
    print("✓ Class and attribute inspection completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"griffe example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished\n")
```

## Example 4: griffe functionality test 4

```python
import griffe

# Example 4: API breakage detection and analysis
print("Example 4: API breakage detection and analysis")

try:
    # Check for breakage detection classes
    breakage_classes = [attr for attr in dir(griffe) if 'Breakage' in attr]
    print(f"✓ Breakage detection classes: {len(breakage_classes)}")
    if breakage_classes:
        print(f"  Available: {breakage_classes[:5]}")
    
    # Check for specific breakage types
    breakage_types = [attr for attr in dir(griffe) if any(word in attr for word in 
                     ['Changed', 'Removed', 'Added', 'Moved'])]
    print(f"✓ Change detection types: {len(breakage_types)}")
    if breakage_types:
        print(f"  Sample types: {breakage_types[:3]}")
    
    # Try to create a breakage object if available
    if hasattr(griffe, 'AttributeChangedTypeBreakage'):
        try:
            # A Breakage is bound to the griffe object it describes: `obj` is its first, required
            # argument. Build the old and new versions of a tiny module in memory to get real ones.
            with griffe.temporary_visited_module("value: str = ''") as old_module, \
                 griffe.temporary_visited_module("value: int = 0") as new_module:
                old_attr, new_attr = old_module["value"], new_module["value"]
                breakage = griffe.AttributeChangedTypeBreakage(
                    new_attr,
                    old_value=old_attr.annotation, 
                    new_value=new_attr.annotation
                )
                # griffe's own comparison of the two versions reports the attribute's changed value
                found = [(type(b).__name__, b.obj.path) for b in griffe.find_breaking_changes(old_module, new_module)]
            assert breakage.kind is griffe.BreakageKind.ATTRIBUTE_CHANGED_TYPE, breakage.kind
            assert (breakage.obj.path, str(breakage.old_value), str(breakage.new_value)) == ("module.value", "str", "int")
            assert found == [("AttributeChangedValueBreakage", "module.value")], found
            print(f"✓ Breakage object created: {type(breakage).__name__}")
        except Exception as breakage_error:
            print(f"✓ Breakage creation: {str(breakage_error)[:50]}...")
            raise
    
    # Simulate API comparison scenario
    old_api = {"function_a": "str", "function_b": "int", "class_c": "type"}
    new_api = {"function_a": "int", "function_b": "int", "class_d": "type"}  # Changed and added
    
    print(f"✓ API comparison simulation:")
    print(f"  Old API items: {len(old_api)}")
    print(f"  New API items: {len(new_api)}")
    
    # Simple breakage detection
    changed = {k: (old_api[k], new_api[k]) for k in old_api if k in new_api and old_api[k] != new_api[k]}
    removed = [k for k in old_api if k not in new_api]
    added = [k for k in new_api if k not in old_api]
    
    print(f"  Changed items: {len(changed)} {list(changed.keys())}")
    print(f"  Removed items: {len(removed)} {removed}")
    print(f"  Added items: {len(added)} {added}")
    
    print("✓ API breakage detection demonstrated")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"griffe example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished\n")
```
