## Example 1: Functionality Test 1

```python
import groq
import os
from unittest import mock

# Example 1: Basic Groq client initialization and configuration
print("Example 1: Basic Groq client initialization and configuration")

try:
    # Check what's available in the groq module
    groq_attrs = [attr for attr in dir(groq) if not attr.startswith('_')]
    print(f"✓ Groq module attributes: {len(groq_attrs)}")
    print(f"  Key attributes: {groq_attrs[:5]}")
    
    # Try to create a Groq client if available
    if hasattr(groq, 'Groq'):
        # Without an API key the client refuses to construct. The key may come from GROQ_API_KEY, so the
        # demo runs with it removed from the environment (restored on exit) to hold on every host.
        with mock.patch.dict(os.environ):
            os.environ.pop("GROQ_API_KEY", None)
            try:
                groq.Groq()
            except groq.GroqError as e:
                assert "GROQ_API_KEY" in str(e), e
                print(f"✓ Client creation without a key is refused (API key configuration required)")
            else:
                raise AssertionError("groq.Groq() constructed without an API key")
        
        # With a (placeholder) key the client constructs offline: no request is sent until an API call.
        client = groq.Groq(api_key="gsk_placeholder_key")
        assert client.api_key == "gsk_placeholder_key", client.api_key
        print(f"✓ Groq client created: {type(client).__name__}")
        
        # Check client methods
        client_methods = [attr for attr in dir(client) if not attr.startswith('_')]
        api_methods = [attr for attr in client_methods if any(word in attr.lower() 
                      for word in ['chat', 'completion', 'generate', 'model'])]
        assert {"chat", "models"} <= set(api_methods), api_methods
        assert callable(client.chat.completions.create) and callable(client.models.list)
        print(f"  Client methods: {len(client_methods)}")
        print(f"  API methods: {api_methods[:3]}")
    
    # Check for other classes
    client_classes = [attr for attr in groq_attrs if 'Client' in attr or attr == 'Groq']
    print(f"✓ Client classes: {client_classes}")
    
    # Check version if available
    if hasattr(groq, '__version__'):
        print(f"✓ Groq version: {groq.__version__}")
    
    print("✓ Groq client exploration completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"groq example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished\n")
```

## Example 2: Functionality Test 2

```python
import groq
import os

# Example 2: Chat completions and model interaction
print("Example 2: Chat completions and model interaction")

try:
    # Try to create a client with configuration
    client = None
    if hasattr(groq, 'Groq'):
        try:
            # Try creating client (may need API key from environment)
            client = groq.Groq(api_key=os.getenv('GROQ_API_KEY', 'demo-key'))
            print(f"✓ Client created with configuration")
            
        except Exception as client_error:
            print(f"✓ Client configuration: {str(client_error)[:40]}...")
            raise
    
    # Check for chat completion functionality
    if client and hasattr(client, 'chat'):
        print(f"✓ Chat functionality available")
        
        # Check chat completions structure
        chat_attr = getattr(client, 'chat')
        if hasattr(chat_attr, 'completions'):
            completions = getattr(chat_attr, 'completions')
            print(f"  Completions endpoint: {type(completions).__name__}")
            
            # Check completion methods
            completion_methods = [attr for attr in dir(completions) if not attr.startswith('_')]
            print(f"  Completion methods: {completion_methods[:3]}")
    
    # Simulate a chat completion request structure
    sample_messages = [
        {"role": "system", "content": "You are a helpful assistant"},
        {"role": "user", "content": "Hello, how are you?"}
    ]
    
    print(f"✓ Sample chat structure:")
    print(f"  Messages: {len(sample_messages)}")
    for msg in sample_messages:
        print(f"    {msg['role']}: {msg['content'][:30]}...")
    
    # Check for model-related functionality
    model_attrs = [attr for attr in dir(groq) if 'model' in attr.lower()]
    print(f"✓ Model-related attributes: {len(model_attrs)}")
    
    print("✓ Chat completions exploration completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"groq example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished\n")
```

## Example 5: Functionality Test 5

```python
import groq
import asyncio

# Example 5: Streaming and async functionality
print("Example 5: Streaming and async functionality")

try:
    # Check for async client
    async_classes = [attr for attr in dir(groq) if 'async' in attr.lower() or 'Async' in attr]
    print(f"✓ Async classes: {len(async_classes)}")
    if async_classes:
        print(f"  Available: {async_classes}")
    
    # Check for streaming support
    if hasattr(groq, 'Groq'):
        try:
            client = groq.Groq(api_key="demo")
            
            # Check for stream-related methods
            client_methods = [attr for attr in dir(client) if not attr.startswith('_')]
            stream_methods = [attr for attr in client_methods if 'stream' in attr.lower()]
            print(f"✓ Streaming methods: {stream_methods}")
            
        except Exception as stream_error:
            print(f"✓ Streaming exploration: {str(stream_error)[:40]}...")
            raise
    
    # Demonstrate streaming parameters
    streaming_params = {
        "stream": True,
        "temperature": 0.5,
        "max_tokens": 512
    }
    
    print(f"✓ Streaming parameters:")
    for param, value in streaming_params.items():
        print(f"  {param}: {value}")
    
    # Check for event loop compatibility
    try:
        # In Jupyter, asyncio loop is usually running
        print(f"✓ Async environment check:")
        print(f"  Event loop running: {asyncio.get_event_loop().is_running()}")
    except RuntimeError:
        print(f"  No event loop running (normal for sync context)")
    
    # Simulate streaming response structure
    stream_chunk = {
        "id": "chatcmpl-123",
        "object": "chat.completion.chunk",
        "choices": [{"delta": {"content": "Hello"}, "finish_reason": None}]
    }
    
    print(f"✓ Stream chunk structure: {len(stream_chunk)} fields")
    print(f"  Content delta: {stream_chunk['choices'][0]['delta'].get('content', 'N/A')}")
    
    print("✓ Streaming and async functionality exploration completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"groq example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished\n")
```
