## Example 2: Functionality Test 2

```python
try:
    from google.iam.v1.policy_pb2 import Policy, Binding  # not iam_policy_pb2, which only imports them
    
    # Example 2: IAM Policy structure and bindings
    print("Example 2: IAM Policy structure and bindings")
    
    # Create a Policy instance
    policy = Policy()
    print(f"✓ Policy instance created: {type(policy).__name__}")
    
    # Create bindings
    binding1 = Binding()
    binding1.role = "roles/viewer"
    binding1.members.extend(["user:alice@example.com", "user:bob@example.com"])
    
    binding2 = Binding()
    binding2.role = "roles/editor"
    binding2.members.extend(["user:charlie@example.com"])
    
    # Add bindings to policy
    policy.bindings.extend([binding1, binding2])
    policy.version = 1
    
    print(f"✓ Policy configured:")
    print(f"  Version: {policy.version}")
    print(f"  Bindings: {len(policy.bindings)}")
    
    for i, binding in enumerate(policy.bindings):
        print(f"    Binding {i+1}: {binding.role} → {len(binding.members)} members")
        for member in binding.members:
            print(f"      - {member}")
    
    # Check serialization capabilities
    assert [(b.role, list(b.members)) for b in policy.bindings] == [
        ("roles/viewer", ["user:alice@example.com", "user:bob@example.com"]),
        ("roles/editor", ["user:charlie@example.com"]),
    ], policy
    serialized = policy.SerializeToString()
    assert Policy.FromString(serialized) == policy, "Policy did not survive a serialize/parse round-trip"
    print(f"✓ Serialized policy: {len(serialized)} bytes")
    
    print("✓ IAM Policy structure demonstration completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"grpc_google_iam_v1 example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished\n")
```

## Example 4: Functionality Test 4

```python
try:
    from google.iam.v1.policy_pb2 import Policy, Binding
    from google.protobuf.json_format import MessageToJson, Parse
    
    # Example 4: JSON serialization
    print("Example 4: JSON serialization")
    
    # Create a version 3 policy (the version that supports conditions)
    policy = Policy()
    policy.version = 3
    
    binding = Binding()
    binding.role = "roles/storage.objectViewer"
    binding.members.extend(["user:alice@example.com"])
    
    policy.bindings.extend([binding])
    
    print(f"✓ Policy created with version {policy.version}")
    
    # Convert to JSON
    json_policy = MessageToJson(policy)
    print(f"✓ JSON serialization:")
    print(f"  JSON length: {len(json_policy)} characters")
    print(f"  JSON preview: {json_policy[:100]}...")
    assert '"roles/storage.objectViewer"' in json_policy and '"user:alice@example.com"' in json_policy, json_policy
    
    # Parse back from JSON
    parsed_policy = Policy()
    Parse(json_policy, parsed_policy)
    assert parsed_policy == policy, f"JSON round-trip changed the policy: {parsed_policy}"
    print(f"✓ Parsed from JSON:")
    print(f"  Version: {parsed_policy.version}")
    print(f"  Bindings: {len(parsed_policy.bindings)}")
    
    print("✓ JSON serialization demonstration completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"grpc_google_iam_v1 example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished\n")
```

## Example 5: Functionality Test 5

```python
try:
    # Policy/Binding live in policy_pb2, GetPolicyOptions in options_pb2, and a binding's condition is a google.type.Expr
    from google.iam.v1.policy_pb2 import Policy, Binding
    from google.iam.v1.options_pb2 import GetPolicyOptions
    from google.type.expr_pb2 import Expr
    
    # Example 5: Advanced IAM features - conditions and expressions
    print("Example 5: Advanced IAM features - conditions and expressions")
    
    # Create policy with conditional binding
    policy = Policy()
    policy.version = 3  # Required for conditions
    
    # Create binding with condition
    binding = Binding()
    binding.role = "roles/storage.objectViewer"
    binding.members.extend(["user:temp-user@example.com"])
    
    # Add condition (IAM condition expression)
    condition = Expr()
    condition.title = "Temporary access"
    condition.description = "Access expires at end of month"
    condition.expression = 'request.time < timestamp("2024-01-31T00:00:00Z")'
    
    binding.condition.CopyFrom(condition)
    policy.bindings.extend([binding])
    
    print(f"✓ Conditional policy created:")
    print(f"  Policy version: {policy.version}")
    print(f"  Binding role: {binding.role}")
    print(f"  Condition title: {condition.title}")
    print(f"  Condition expression: {condition.expression}")
    
    # Create GetPolicyOptions
    options = GetPolicyOptions()
    options.requested_policy_version = 3
    print(f"✓ Policy options: version {options.requested_policy_version}")
    
    # Show etag functionality (for safe updates)
    policy.etag = b"example_etag_12345"
    print(f"✓ Policy etag set: {len(policy.etag)} bytes")
    
    # Serialization test
    serialized = policy.SerializeToString()
    deserialized = Policy()
    deserialized.ParseFromString(serialized)
    assert deserialized == policy, "Policy did not survive a serialize/parse round-trip"
    assert deserialized.bindings[0].condition.title == "Temporary access", deserialized
    
    print(f"✓ Serialization round-trip successful:")
    print(f"  Original bindings: {len(policy.bindings)}")
    print(f"  Deserialized bindings: {len(deserialized.bindings)}")
    
    print("✓ Advanced IAM features demonstration completed")
    
except Exception as e:
    print(f"✗ Error: {e}")
    raise AssertionError(
        f"grpc_google_iam_v1 example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished\n")
```
