## Example 1: gymnasium functionality test 1

```python
import gymnasium as gym

# Example 1: Basic CartPole environment creation and interaction
env = gym.make("CartPole-v1")
print(f"Environment: {env.spec.id}")

# Reset environment and get initial observation
observation, info = env.reset()
print(f"Initial observation: {observation}")
print(f"Observation shape: {observation.shape}")

# Take a few random actions
total_reward = 0
for step in range(5):
    action = env.action_space.sample()
    observation, reward, terminated, truncated, info = env.step(action)
    total_reward += reward
    print(f"Step {step+1}: Action={action}, Reward={reward}, Done={terminated or truncated}")
    
    if terminated or truncated:
        break

print(f"Total reward after {step+1} steps: {total_reward}")
env.close()
print("Example 1 completed successfully")
```

## Example 3: gymnasium functionality test 3

```python
import gymnasium as gym
import numpy as np

# Example 3: Random agent playing multiple episodes
env = gym.make("Acrobot-v1")
print(f"Environment: {env.spec.id}")

episode_rewards = []
num_episodes = 3

for episode in range(num_episodes):
    observation, info = env.reset()
    total_reward = 0
    steps = 0
    
    while steps < 100:  # Limit steps per episode
        action = env.action_space.sample()  # Random action
        observation, reward, terminated, truncated, info = env.step(action)
        total_reward += reward
        steps += 1
        
        if terminated or truncated:
            break
    
    episode_rewards.append(total_reward)
    print(f"Episode {episode+1}: {steps} steps, reward: {total_reward:.2f}")

print(f"Average reward: {np.mean(episode_rewards):.2f}")
env.close()
print("Example 3 completed successfully")
```

## Example 4: gymnasium functionality test 4

```python
import gymnasium as gym
from gymnasium.wrappers import TimeLimit, RecordEpisodeStatistics

# Example 4: Using environment wrappers
base_env = gym.make("CartPole-v1")
print(f"Base environment: {base_env.spec.id}")

# Wrap environment with time limit and episode statistics
env = RecordEpisodeStatistics(TimeLimit(base_env, max_episode_steps=50))
print(f"Wrapped environment: {type(env).__name__}")

# Run one episode
observation, info = env.reset()
total_reward = 0
steps = 0

while True:
    action = env.action_space.sample()
    observation, reward, terminated, truncated, info = env.step(action)
    total_reward += reward
    steps += 1
    
    if terminated or truncated:
        print(f"Episode finished after {steps} steps")
        print(f"Episode info: {info}")
        break

env.close()
print("Example 4 completed successfully")
```
