```python
import h11

# Example 1: Basic HTTP/1.1 connection creation
print("Creating HTTP/1.1 connections")

# Create client connection
client_conn = h11.Connection(h11.CLIENT)
print(f"Client connection created")
print(f"Client state: {client_conn.our_state}")
print(f"Client connection type: {type(client_conn).__name__}")

# Create server connection
server_conn = h11.Connection(h11.SERVER)
print(f"Server connection created")
print(f"Server state: {server_conn.our_state}")
print(f"Server connection type: {type(server_conn).__name__}")

# Check available states
print(f"Available states: {[h11.IDLE, h11.SEND_BODY, h11.DONE, h11.CLOSED]}")
print(f"Client initial state: {client_conn.our_state}")
print(f"Server initial state: {server_conn.our_state}")

print("Example 1 completed successfully")
```

```python
import h11

# Example 2: HTTP request creation and sending
print("Creating and sending HTTP requests")

# Create client connection
conn = h11.Connection(h11.CLIENT)

# Create request
request = h11.Request(
    method="GET",
    target="/api/data",
    headers=[
        ("host", "example.com"),
        ("user-agent", "h11-example/1.0"),
        ("connection", "close")
    ]
)

# Send request
data_to_send = conn.send(request)
print(f"Request created: {request.method} {request.target}")
print(f"Request headers: {list(request.headers)}")
print(f"Data to send length: {len(data_to_send)} bytes")
print(f"Connection state after request: {conn.our_state}")

# Send end of message
end_data = conn.send(h11.EndOfMessage())
print(f"End of message data length: {len(end_data)} bytes")
print(f"Final connection state: {conn.our_state}")

print("Example 2 completed successfully")
```

```python
import h11

# Example 3: HTTP response creation and parsing
print("Creating and working with HTTP responses")

# Create server connection
conn = h11.Connection(h11.SERVER)

# Prepare body data first to get correct length
body = b'{"ok": true}'
content_length = str(len(body))

# Create response with correct content-length
response = h11.Response(
    status_code=200,
    headers=[
        ("content-type", "application/json"),
        ("content-length", content_length),
        ("server", "h11-example/1.0")
    ]
)

# Send response
data_to_send = conn.send(response)
print(f"Response created: {response.status_code}")
print(f"Response headers: {list(response.headers)}")
print(f"Response data length: {len(data_to_send)} bytes")

# Send body data
body_data = conn.send(h11.Data(data=body))
print(f"Body data: {body} ({len(body)} bytes)")
print(f"Body data length: {len(body_data)} bytes")

# End response
end_data = conn.send(h11.EndOfMessage())
print(f"End of message sent, connection state: {conn.our_state}")

print("Example 3 completed successfully")
```

```python
import h11

# Example 4: HTTP events processing and parsing
print("Processing HTTP events and data parsing")

# Create server connection for receiving data
conn = h11.Connection(h11.SERVER)

# Simulate receiving raw HTTP request data
raw_data = b"GET /test HTTP/1.1\r\nHost: example.com\r\nConnection: close\r\n\r\n"

# Receive and process the data
conn.receive_data(raw_data)

# Get next event
event = conn.next_event()
print(f"Received event type: {type(event).__name__}")

if isinstance(event, h11.Request):
    print(f"Method: {event.method}")
    print(f"Target: {event.target}")
    print(f"Headers: {list(event.headers)}")
    print(f"HTTP version: {event.http_version}")

# Check if more events are available
try:
    next_event = conn.next_event()
    print(f"Next event: {type(next_event).__name__}")
except h11.NeedMoreData:
    print("Need more data to parse next event")

print("Example 4 completed successfully")
```

```python
import h11

# Example 5: Connection state management and transitions
print("Managing HTTP connection states")

# Create client connection
conn = h11.Connection(h11.CLIENT)
print(f"Initial state: {conn.our_state}")

# Prepare body data and calculate content-length
body_data = b"Hello, world!"
content_length = str(len(body_data))

# Send request with proper content-length
request = h11.Request(
    method="POST", 
    target="/upload", 
    headers=[
        ("host", "example.com"),
        ("content-length", content_length)
    ]
)
conn.send(request)
print(f"After sending request: {conn.our_state}")

# Send body data
conn.send(h11.Data(data=body_data))
print(f"After sending data ({len(body_data)} bytes): {conn.our_state}")

# End request
conn.send(h11.EndOfMessage())
print(f"After ending message: {conn.our_state}")

# Check connection status
print(f"Connection closed: {conn.our_state == h11.CLOSED}")

print("Example 5 completed successfully")
```
