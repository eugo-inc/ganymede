```python
import h5netcdf
import numpy as np
import tempfile
import os

# Example 1: Basic netCDF file creation and writing
print("Creating a netCDF file with h5netcdf")

# Create a temporary file
temp_file = tempfile.NamedTemporaryFile(delete=False, suffix='.nc')
temp_file.close()

try:
    # Create a new netCDF file
    with h5netcdf.File(temp_file.name, 'w') as f:
        print(f"Created netCDF file: {temp_file.name}")
        print(f"File mode: {f.mode}")
        
        # Add a global attribute
        f.attrs['title'] = 'Test netCDF file'
        f.attrs['created_by'] = 'h5netcdf example'
        
        print(f"File attributes: {dict(f.attrs)}")
        print(f"File dimensions: {list(f.dimensions.keys())}")
        
    print("Example 1 completed successfully")
    
finally:
    # Clean up
    if os.path.exists(temp_file.name):
        os.unlink(temp_file.name)
```

```python
import h5netcdf
import numpy as np
import tempfile
import os

# Example 2: Creating variables and storing data
print("Creating variables and storing data")

temp_file = tempfile.NamedTemporaryFile(delete=False, suffix='.nc')
temp_file.close()

try:
    with h5netcdf.File(temp_file.name, 'w') as f:
        # Create dimensions
        f.dimensions['time'] = 10
        f.dimensions['location'] = 3
        
        # Create variables
        temp_var = f.create_variable('temperature', ('time', 'location'), data=np.random.rand(10, 3))
        time_var = f.create_variable('time', ('time',), data=np.arange(10))
        
        print(f"Created variable 'temperature' with shape: {temp_var.shape}")
        print(f"Created variable 'time' with shape: {time_var.shape}")
        print(f"Temperature data sample: {temp_var[0, :]}")
        print(f"Time data: {time_var[:]}")
        
    print("Example 2 completed successfully")
    
finally:
    if os.path.exists(temp_file.name):
        os.unlink(temp_file.name)
```

```python
import h5netcdf
import numpy as np
import tempfile
import os

# Example 3: Working with dimensions and coordinate variables
print("Working with dimensions and coordinate variables")

temp_file = tempfile.NamedTemporaryFile(delete=False, suffix='.nc')
temp_file.close()

try:
    with h5netcdf.File(temp_file.name, 'w') as f:
        # Create dimensions
        f.dimensions['x'] = 5
        f.dimensions['y'] = 4
        f.dimensions['time'] = None  # Unlimited dimension

        # Create coordinate variables
        x_coord = f.create_variable('x', ('x',), data=np.linspace(0, 10, 5))
        y_coord = f.create_variable('y', ('y',), data=np.linspace(0, 8, 4))

        # An unlimited dimension starts at size 0 — resize it before writing
        # (h5netcdf does not auto-grow on __setitem__).
        data_var = f.create_variable('data', ('time', 'y', 'x'), dtype='f4')
        f.resize_dimension('time', 2)
        data_var[0:2, :, :] = np.random.rand(2, 4, 5)  # Add data to first 2 time steps

        print(f"Dimensions: {dict(f.dimensions)}")
        print(f"Data shape: {data_var.shape}")
        print(f"Data sample: {data_var[0, 0, :]}")
        assert data_var.shape == (2, 4, 5)

    print("Example 3 completed successfully")

finally:
    if os.path.exists(temp_file.name):
        os.unlink(temp_file.name)
```

```python
import h5netcdf
import numpy as np
import tempfile
import os

# Example 4: Working with attributes and metadata
print("Working with attributes and metadata")

temp_file = tempfile.NamedTemporaryFile(delete=False, suffix='.nc')
temp_file.close()

try:
    with h5netcdf.File(temp_file.name, 'w') as f:
        # Create dimension and variable
        f.dimensions['time'] = 5
        temp_var = f.create_variable('temperature', ('time',), data=np.array([20.5, 21.2, 19.8, 22.1, 20.9]))
        
        # Add global attributes
        f.attrs['title'] = 'Temperature measurements'
        f.attrs['institution'] = 'Test Lab'
        f.attrs['created'] = '2024-01-01'
        
        # Add variable attributes
        temp_var.attrs['units'] = 'degrees_C'
        temp_var.attrs['long_name'] = 'Air temperature'
        temp_var.attrs['valid_range'] = [-50.0, 50.0]
        
        print(f"Global attributes: {dict(f.attrs)}")
        print(f"Variable attributes: {dict(temp_var.attrs)}")
        print(f"Temperature data: {temp_var[:]}")
        
    print("Example 4 completed successfully")
    
finally:
    if os.path.exists(temp_file.name):
        os.unlink(temp_file.name)
```

```python
import h5netcdf
import numpy as np
import tempfile
import os

# Example 5: Reading data from netCDF file
print("Reading data from netCDF file")

temp_file = tempfile.NamedTemporaryFile(delete=False, suffix='.nc')
temp_file.close()

try:
    # First, create a file with data
    with h5netcdf.File(temp_file.name, 'w') as f:
        f.dimensions['x'] = 3
        f.dimensions['y'] = 2
        data_var = f.create_variable('pressure', ('y', 'x'), data=np.array([[1013.2, 1012.8, 1014.1], [1011.5, 1013.7, 1012.3]]))
        data_var.attrs['units'] = 'hPa'
    
    # Now read the data back
    with h5netcdf.File(temp_file.name, 'r') as f:
        print(f"File variables: {list(f.variables.keys())}")
        print(f"File dimensions: {dict(f.dimensions)}")
        
        pressure = f.variables['pressure']
        print(f"Pressure shape: {pressure.shape}")
        print(f"Pressure units: {pressure.attrs['units']}")
        print(f"Pressure data:\n{pressure[:]}")
        print(f"Pressure slice [0, :]: {pressure[0, :]}")
        
    print("Example 5 completed successfully")
    
finally:
    if os.path.exists(temp_file.name):
        os.unlink(temp_file.name)
```
