```python
import h5py
import numpy as np
import tempfile
import os

# Example 1: Basic HDF5 file creation and dataset writing
print("Creating HDF5 file and writing datasets")

temp_file = tempfile.NamedTemporaryFile(delete=False, suffix='.h5')
temp_file.close()

try:
    # Create HDF5 file and write data
    with h5py.File(temp_file.name, 'w') as f:
        print(f"Created HDF5 file: {temp_file.name}")
        
        # Create datasets with different data
        data1 = np.array([1, 2, 3, 4, 5])
        data2 = np.random.rand(3, 4)
        
        f.create_dataset('integers', data=data1)
        f.create_dataset('matrix', data=data2)
        
        print(f"Created dataset 'integers' with shape: {f['integers'].shape}")
        print(f"Created dataset 'matrix' with shape: {f['matrix'].shape}")
        print(f"Integer data: {f['integers'][:]}")
        print(f"Matrix sample: {f['matrix'][0, :]}")
        
    print("Example 1 completed successfully")
    
finally:
    if os.path.exists(temp_file.name):
        os.unlink(temp_file.name)
```

```python
import h5py
import numpy as np
import tempfile
import os

# Example 2: Dataset reading and array operations
print("Reading HDF5 datasets and performing operations")

temp_file = tempfile.NamedTemporaryFile(delete=False, suffix='.h5')
temp_file.close()

try:
    # Create file with data
    with h5py.File(temp_file.name, 'w') as f:
        data = np.arange(20).reshape(4, 5)
        f.create_dataset('measurements', data=data)
    
    # Read and manipulate data
    with h5py.File(temp_file.name, 'r') as f:
        dataset = f['measurements']
        print(f"Dataset shape: {dataset.shape}")
        print(f"Dataset dtype: {dataset.dtype}")
        
        # Array slicing and operations
        print(f"First row: {dataset[0, :]}")
        print(f"Last column: {dataset[:, -1]}")
        print(f"2x2 subarray:\n{dataset[1:3, 1:3]}")
        print(f"Dataset sum: {np.sum(dataset[:])}")
        
    print("Example 2 completed successfully")
    
finally:
    if os.path.exists(temp_file.name):
        os.unlink(temp_file.name)
```

```python
import h5py
import numpy as np
import tempfile
import os

# Example 3: HDF5 groups and hierarchical structure
print("Creating HDF5 groups and hierarchical structure")

temp_file = tempfile.NamedTemporaryFile(delete=False, suffix='.h5')
temp_file.close()

try:
    with h5py.File(temp_file.name, 'w') as f:
        # Create groups
        experiment = f.create_group('experiment')
        sensors = experiment.create_group('sensors')
        analysis = f.create_group('analysis')
        
        # Add datasets to groups
        sensors.create_dataset('temperature', data=np.random.rand(10))
        sensors.create_dataset('pressure', data=np.random.rand(10))
        analysis.create_dataset('results', data=np.array([1.23, 4.56, 7.89]))
        
        print(f"Root groups: {list(f.keys())}")
        print(f"Experiment subgroups: {list(experiment.keys())}")
        print(f"Sensor datasets: {list(sensors.keys())}")
        print(f"Temperature data: {sensors['temperature'][:3]}")
        print(f"Full path access: {f['experiment/sensors/pressure'][:3]}")
        
    print("Example 3 completed successfully")
    
finally:
    if os.path.exists(temp_file.name):
        os.unlink(temp_file.name)
```

```python
import h5py
import numpy as np
import tempfile
import os

# Example 4: Attributes and metadata handling
print("Working with HDF5 attributes and metadata")

temp_file = tempfile.NamedTemporaryFile(delete=False, suffix='.h5')
temp_file.close()

try:
    with h5py.File(temp_file.name, 'w') as f:
        # Create dataset with attributes
        data = np.linspace(0, 100, 11)
        dataset = f.create_dataset('measurements', data=data)
        
        # Add file attributes
        f.attrs['title'] = 'Scientific measurements'
        f.attrs['version'] = 1.0
        f.attrs['created_by'] = 'h5py example'
        
        # Add dataset attributes
        dataset.attrs['units'] = 'meters'
        dataset.attrs['description'] = 'Distance measurements'
        dataset.attrs['calibration'] = [1.0, 0.0]  # scale, offset
        
        print(f"File attributes: {dict(f.attrs)}")
        print(f"Dataset attributes: {dict(dataset.attrs)}")
        print(f"Measurement units: {dataset.attrs['units']}")
        print(f"Data sample: {dataset[:3]}")
        
    print("Example 4 completed successfully")
    
finally:
    if os.path.exists(temp_file.name):
        os.unlink(temp_file.name)
```

```python
import h5py
import numpy as np
import tempfile
import os

# Example 5: Data types and compression features
print("Using data types and compression in HDF5")

temp_file = tempfile.NamedTemporaryFile(delete=False, suffix='.h5')
temp_file.close()

try:
    with h5py.File(temp_file.name, 'w') as f:
        # Create datasets with different dtypes and compression
        large_data = np.random.randint(0, 100, size=(100, 50))
        
        # Compressed dataset
        f.create_dataset('compressed', data=large_data, compression='gzip', compression_opts=9)
        
        # Different data types
        f.create_dataset('integers', data=np.array([1, 2, 3]), dtype='int32')
        f.create_dataset('floats', data=np.array([1.1, 2.2, 3.3]), dtype='float64')
        f.create_dataset('strings', data=np.array([b'hello', b'world']), dtype='S10')
        
        print(f"Compressed dataset shape: {f['compressed'].shape}")
        print(f"Compression: {f['compressed'].compression}")
        print(f"Integer dtype: {f['integers'].dtype}")
        print(f"Float dtype: {f['floats'].dtype}")
        print(f"String data: {f['strings'][:]}")
        
    print("Example 5 completed successfully")
    
finally:
    if os.path.exists(temp_file.name):
        os.unlink(temp_file.name)
```
