## Example 3: Functionality Test 3

```python
from hatch_jupyter_builder import utils
import tempfile
import os

# Example 3: Utility functions for Jupyter package building
print("Testing utility functions")

# Check what's available in utils
available_utils = [attr for attr in dir(utils) if not attr.startswith('_')]
print(f"Available utility functions: {available_utils}")

# Test basic string normalization (custom implementation)
def normalize_package_name(name):
    """Simple name normalization for demonstration"""
    return name.lower().replace('-', '_').replace('.', '_')

test_names = [
    "my-jupyter-extension",
    "My_Package_Name", 
    "jupyter.widget.example",
    "CAPITAL-LETTERS"
]

print("Testing package name normalization:")
for name in test_names:
    normalized = normalize_package_name(name)
    print(f"  '{name}' -> '{normalized}'")

# Create a temporary workspace
temp_dir = tempfile.mkdtemp()

try:
    print(f"Created workspace: {temp_dir}")
    
    # Test if run_command is available
    if hasattr(utils, 'run_command'):
        print("run_command utility available")
    else:
        print("Creating directory structure manually")
    
    # Create nested directory structure
    test_subdir = os.path.join(temp_dir, "test", "subdir")
    os.makedirs(test_subdir, exist_ok=True)
    
    print(f"Created nested directory: {os.path.exists(test_subdir)}")
    
    print("Example 3 completed successfully")
    
finally:
    # Cleanup
    import shutil
    if os.path.exists(temp_dir):
        shutil.rmtree(temp_dir)
```

## Example 4: Functionality Test 4

```python
import hatch_jupyter_builder
import json
import tempfile
import os

# Example 4: Build configuration and patterns
print("Testing build configuration and patterns")

# Define common ignore patterns for Jupyter extensions
common_ignore_patterns = [
    "*.pyc",
    "__pycache__/",
    "*.egg-info/",
    ".git/",
    "node_modules/",
    ".DS_Store",
    "*.log"
]

print("Common ignore patterns for Jupyter builds:")
for pattern in common_ignore_patterns:
    print(f"  {pattern}")

# Create a mock Jupyter extension config
temp_dir = tempfile.mkdtemp()
config_path = os.path.join(temp_dir, "build_config.json")

try:
    # Sample build configuration for Jupyter extension
    build_config = {
        "name": "my-jupyter-extension",
        "build_dir": "dist",
        "source_dir": "src", 
        "lab_extension": True,
        "ignore_patterns": common_ignore_patterns,
        "build_commands": [
            "jlpm install",
            "jlpm run build"
        ]
    }
    
    with open(config_path, 'w') as f:
        json.dump(build_config, f, indent=2)
    
    print(f"Created config: {config_path}")
    print(f"Extension name: {build_config['name']}")
    print(f"Lab extension: {build_config['lab_extension']}")
    print(f"Build commands: {len(build_config['build_commands'])}")
    
    print("Example 4 completed successfully")
    
finally:
    if os.path.exists(config_path):
        os.unlink(config_path)
    os.rmdir(temp_dir)
```

## Example 5: Functionality Test 5

```python
from hatch_jupyter_builder import utils, __version__
import tempfile
import os
import subprocess

# Example 5: Package version and basic utilities
print("Testing package version and utilities")

# Test version information
print(f"hatch_jupyter_builder version: {__version__}")

# Check available utilities
available_functions = [attr for attr in dir(utils) if not attr.startswith('_')]
print(f"Available utility functions: {available_functions}")

# Create a temporary directory for testing
temp_dir = tempfile.mkdtemp()
old_cwd = os.getcwd()

try:
    os.chdir(temp_dir)
    
    # Create a simple test file
    test_file = os.path.join(temp_dir, "test.txt")
    with open(test_file, 'w') as f:
        f.write("Hello from hatch_jupyter_builder test")
    
    print(f"Created test file: {os.path.exists(test_file)}")
    
    # Test basic command execution (using subprocess directly)
    try:
        result = subprocess.run(["echo", "Command execution test"], 
                              capture_output=True, text=True, timeout=5)
        print(f"Command output: {result.stdout.strip()}")
    except Exception as e:
        print(f"Command execution note: {e}")
        raise
    
    # Test working with file paths
    print(f"Working directory: {os.getcwd()}")
    print(f"Files in directory: {os.listdir(temp_dir)}")
    
    print("Example 5 completed successfully")
    
finally:
    # Restore the working directory, then clean up
    os.chdir(old_cwd)
    import shutil
    if os.path.exists(temp_dir):
        shutil.rmtree(temp_dir)

assert os.getcwd() == old_cwd, os.getcwd()
```
