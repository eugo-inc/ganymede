```python
# Example 1: parse Hjson features absent from strict JSON: comments,
# unquoted keys, unquoted single-line string values, trailing commas absent.
import hjson

text = (
    '{\n'
    '  // line comment\n'
    '  /* block comment */\n'
    '  unquoted_key: hello world\n'
    '  number: 3.14\n'
    '  null_value: null\n'
    '}\n'
)

parsed = hjson.loads(text)
assert parsed['unquoted_key'] == 'hello world'
assert parsed['number'] == 3.14
assert parsed['null_value'] is None
print(parsed)
```

```python
# Example 2: file-like load/dump via dump() and load() against StringIO.
import io
import hjson

obj = {'list': [1, 2, 3], 'flag': False, 'name': 'test'}

buf = io.StringIO()
hjson.dump(obj, buf)
buf.seek(0)
print('serialized:')
print(buf.getvalue())

buf.seek(0)
loaded = hjson.load(buf)
assert loaded == obj
print('round-trip OK')
```

```python
# Example 3: dumpsJSON emits strict JSON; HjsonDecodeError on malformed input.
import hjson

data = {'a': 1, 'b': [2, 3], 'c': None}
json_text = hjson.dumpsJSON(data)
print('strict JSON:', json_text)
assert json_text.startswith('{') and '"a"' in json_text

raised = False
try:
    hjson.loads('{ key without value ')
except hjson.HjsonDecodeError as e:
    raised = True
    print('expected decode error:', type(e).__name__)

assert raised, 'expected HjsonDecodeError on malformed input'
print('hjson API smoke test passed')
```
