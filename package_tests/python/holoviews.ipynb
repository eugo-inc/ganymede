## Example 1: Basic Curve plotting

```python
import holoviews as hv
import numpy as np

# Example 1: Basic Curve plotting
print("Example 1: Creating basic Curve plots with HoloViews")

# Generate sample data
x_data = np.linspace(0, 10, 50)
y_data = np.sin(x_data)

# Create a Curve element
curve = hv.Curve((x_data, y_data))

# Show curve properties
print(f"Curve type: {type(curve).__name__}")
print(f"Curve dimensions: {curve.dimensions()}")
print(f"Data shape: {curve.data.shape if hasattr(curve.data, 'shape') else 'N/A'}")

# Create another curve with custom styling
y_data2 = np.cos(x_data) 
curve2 = hv.Curve((x_data, y_data2), label='Cosine')

print(f"Created curves: {curve.label}, {curve2.label}")

# Show curve bounds
print(f"X range: {curve.range(0)}")
print(f"Y range: {curve.range(1)}")

# Demonstrate curve attributes
print(f"Curve keys: {list(curve.data.keys()) if hasattr(curve.data, 'keys') else 'N/A'}")
print(f"Number of data points: {len(curve)}")

assert [d.name for d in curve.dimensions()] == ["x", "y"]
assert len(curve) == 50 and curve2.label == "Cosine"
assert curve.range(0) == (0.0, 10.0)
low, high = curve.range(1)
assert -1.0 <= low < -0.99 and 0.99 < high <= 1.0
```

## Example 2: Scatter plot visualization

```python
import holoviews as hv
import numpy as np

# Example 2: Scatter plot visualization
print("Example 2: Creating Scatter plots with HoloViews")

# Generate random scatter data
n_points = 100
x_scatter = np.random.randn(n_points)
y_scatter = np.random.randn(n_points) + x_scatter * 0.5
colors = np.random.randint(0, 5, n_points)

# Create basic scatter plot
scatter = hv.Scatter((x_scatter, y_scatter))

print(f"Scatter plot type: {type(scatter).__name__}")
print(f"Number of points: {len(scatter)}")
print(f"Scatter dimensions: {scatter.dimensions()}")

# Create scatter with color dimension
# the first vdim is the y axis, so both value columns are listed
scatter_colored = hv.Scatter((x_scatter, y_scatter, colors), vdims=['y', 'Color'])

print(f"Colored scatter dimensions: {scatter_colored.dimensions()}")
print(f"Value dimensions: {scatter_colored.vdims}")
print(f"Key dimensions: {scatter_colored.kdims}")

# Show data range
print(f"X data range: {scatter.range(0)}")
print(f"Y data range: {scatter.range(1)}")

# Access scatter data
print(f"Data type: {type(scatter.data)}")
print(f"Has color info: {len(scatter_colored.dimensions()) == 3}")

assert len(scatter) == n_points
assert [d.name for d in scatter.dimensions()] == ["x", "y"]
assert [d.name for d in scatter_colored.dimensions()] == ["x", "y", "Color"]
assert (scatter_colored.dimension_values("Color") == colors).all()
assert scatter.range(0) == (x_scatter.min(), x_scatter.max())
```

## Example 3: HeatMap creation

```python
import holoviews as hv
import numpy as np

# Example 3: HeatMap creation
print("Example 3: Creating HeatMap visualizations")

# Generate 2D grid data for heatmap
x = np.linspace(-2, 2, 20)
y = np.linspace(-2, 2, 15)
X, Y = np.meshgrid(x, y)
Z = np.exp(-(X**2 + Y**2))

# Convert to format suitable for HeatMap (x, y, value)
data = []
for i, yi in enumerate(y):
    for j, xi in enumerate(x):
        data.append((xi, yi, Z[i, j]))

# Create HeatMap
heatmap = hv.HeatMap(data)

print(f"HeatMap type: {type(heatmap).__name__}")
print(f"HeatMap dimensions: {heatmap.dimensions()}")
print(f"Data points: {len(heatmap)}")

# Show dimension properties
print(f"Key dimensions: {heatmap.kdims}")
print(f"Value dimensions: {heatmap.vdims}")

# Check data range
print(f"X range: {heatmap.range(0)}")
print(f"Y range: {heatmap.range(1)}")
print(f"Z range: {heatmap.range(2)}")

# Create a simple categorical heatmap
categories = [('A', '1', 10), ('A', '2', 15), ('B', '1', 8), ('B', '2', 12)]
cat_heatmap = hv.HeatMap(categories, kdims=['Letter', 'Number'], vdims='Value')

print(f"Categorical heatmap dimensions: {len(cat_heatmap.dimensions())}")

assert len(heatmap) == 20 * 15
assert [d.name for d in heatmap.dimensions()] == ["x", "y", "z"]
z_low, z_high = heatmap.range(2)
assert z_low == Z.min() and z_high == Z.max()
assert [d.name for d in cat_heatmap.dimensions()] == ["Letter", "Number", "Value"]
assert cat_heatmap.range("Value") == (8, 15)
```

## Example 4: Layout composition and element combination

```python
import holoviews as hv
import numpy as np

# Example 4: Layout composition and element combination
print("Example 4: Layout composition and element combination")

# Create multiple elements
x = np.linspace(0, 4*np.pi, 100)
curve1 = hv.Curve((x, np.sin(x)), label='Sine')
curve2 = hv.Curve((x, np.cos(x)), label='Cosine')

# Create scatter data
scatter_x = np.random.uniform(0, 4*np.pi, 20)
scatter_y = np.random.uniform(-1, 1, 20)
scatter = hv.Scatter((scatter_x, scatter_y), label='Random Points')

# Combine elements using overlay (+)
overlay = curve1 * curve2 * scatter

print(f"Overlay type: {type(overlay).__name__}")
print(f"Overlay contains: {len(overlay)} elements")

# Create layout using side-by-side (+)
layout = curve1 + curve2

print(f"Layout type: {type(layout).__name__}")
print(f"Layout contains: {len(layout)} elements")

# Create more complex layout
complex_layout = (curve1 + curve2) + (scatter + curve1 * curve2)

print(f"Complex layout type: {type(complex_layout).__name__}")

# Show element properties
print(f"First curve label: {curve1.label}")
print(f"Overlay elements: {[el.label for el in overlay if hasattr(el, 'label')]}")
print(f"Layout dimensions: {layout.shape if hasattr(layout, 'shape') else 'N/A'}")

# * overlays elements in one plot, + places them side by side
assert type(overlay).__name__ == "Overlay" and len(overlay) == 3
assert [el.label for el in overlay] == ["Sine", "Cosine", "Random Points"]
assert type(layout).__name__ == "Layout" and len(layout) == 2 and layout.shape == (1, 2)
assert len(complex_layout) == 4  # nested layouts are flattened
```

## Example 5: Dataset and dimension operations

```python
import holoviews as hv
import numpy as np

# Example 5: Dataset and dimension operations
print("Example 5: Dataset and dimension operations")

# Create sample dataset
data = {
    'x': np.random.randn(100),
    'y': np.random.randn(100),  
    'category': np.random.choice(['A', 'B', 'C'], 100),
    'value': np.random.randint(1, 10, 100)
}

# Create Dataset
dataset = hv.Dataset(data, kdims=['x', 'y', 'category'], vdims=['value'])

print(f"Dataset type: {type(dataset).__name__}")
print(f"Dataset dimensions: {dataset.dimensions()}")
print(f"Key dimensions: {dataset.kdims}")
print(f"Value dimensions: {dataset.vdims}")

# Show dataset properties
print(f"Dataset shape: {dataset.data.shape if hasattr(dataset.data, 'shape') else 'N/A'}")
print(f"Number of datapoints: {len(dataset)}")

# Convert to different element types
points = dataset.to(hv.Points, ['x', 'y'])
scatter = dataset.to(hv.Scatter, 'x', 'value')  # Scatter takes one key dimension

print(f"Converted to Points: {type(points).__name__}")
print(f"Converted to Scatter: {type(scatter).__name__}")

# Select subset of data
subset = dataset.select(category='A')
print(f"Subset with category A: {len(subset)} points")

# Group by dimension
grouped = dataset.groupby('category')
print(f"Grouped by category: {type(grouped).__name__}")

assert len(dataset) == 100
assert [d.name for d in dataset.kdims] == ["x", "y", "category"] and [d.name for d in dataset.vdims] == ["value"]
assert len(subset) == int((data["category"] == "A").sum())
# groupby returns a HoloMap keyed by the grouping dimension
present = sorted(set(data["category"]))
assert type(grouped).__name__ == "HoloMap" and sorted(str(k) for k in grouped.keys()) == present
assert sum(len(grouped[k]) for k in grouped.keys()) == 100
```
