## Example 1: Build HTML from Python objects

```python
import importlib.metadata

from htmltools import tags

# Example 1: tags.<name>(children..., attributes) renders to an HTML string
assert importlib.metadata.version("htmltools") == "0.7.0"
page = tags.div(tags.h1("Results"), tags.p("Two rows", class_="note"), id="main")
html = str(page)
assert html.startswith('<div id="main">') and '<p class="note">Two rows</p>' in html, html
print("Example 1 OK:\n" + html)
```

## Example 2: Text is escaped, HTML() is trusted

```python
from htmltools import HTML, tags

# Example 2: plain strings are escaped; wrap trusted markup in HTML()
assert str(tags.span("<b>x</b>")) == "<span>&lt;b&gt;x&lt;/b&gt;</span>"
assert str(tags.span(HTML("<b>x</b>"))) == "<span><b>x</b></span>"
print("Example 2 OK")
```

## Example 3: css() and TagList

```python
from htmltools import TagList, css, tags

# Example 3: css() builds a style string from keyword arguments; a TagList renders siblings
style = css(color="red", font_size="12px", margin=None)
assert style == "color:red;font-size:12px;", style
items = TagList(tags.li("a"), tags.li("b"))
assert str(tags.ul(items)).count("<li>") == 2
print("Example 3 OK:", style)
```

## Example 4: Tags are objects you can edit

```python
from htmltools import tags

# Example 4: add classes and children after creation, and query attributes
t = tags.table(tags.tr(tags.td("1")))
t.add_class("gt_table").add_style("width:100%;")
t.children.append(tags.tr(tags.td("2")))
assert t.attrs["class"] == "gt_table" and "width:100%" in t.attrs["style"]
assert str(t).count("<tr>") == 2
print("Example 4 OK")
```

## Example 5: Declare a JavaScript/CSS dependency

```python
import tempfile
from pathlib import Path

from htmltools import HTMLDependency, TagList, tags

# Example 5: an HTMLDependency travels with the tag that needs it and renders into <head>
src = Path(tempfile.mkdtemp())
(src / "widget.js").write_text("console.log('hi')")
dep = HTMLDependency("widget", "1.0", source={"subdir": str(src)}, script={"src": "widget.js"})
doc = TagList(tags.div("w", dep)).render()
assert [d.name for d in doc["dependencies"]] == ["widget"], doc["dependencies"]
assert "widget.js" in str(dep.as_html_tags()), dep.as_html_tags()
print("Example 5 OK")
```
