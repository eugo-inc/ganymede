```python
# Example 1: public API surface — verify primary classes/functions are exposed.
import httpcore2

expected_names = [
    "ConnectionPool",
    "AsyncConnectionPool",
    "HTTPConnection",
    "Request",
    "Response",
    "URL",
    "Origin",
    "ConnectError",
    "ReadError",
    "WriteError",
    "TimeoutException",
    "ConnectTimeout",
    "ReadTimeout",
]

for name in expected_names:
    obj = getattr(httpcore2, name)
    print(f"  httpcore2.{name} -> {obj!r}")

assert all(hasattr(httpcore2, n) for n in expected_names)
```

```python
# Example 2: URL / Origin / Request model construction.
import httpcore2

url = httpcore2.URL(scheme=b"https", host=b"example.com", port=443, target=b"/")
print(f"url = {url!r}")
print(f"  scheme={url.scheme!r} host={url.host!r} port={url.port} target={url.target!r}")

origin = httpcore2.Origin(scheme=b"https", host=b"example.com", port=443)
print(f"origin = {origin!r}")

req = httpcore2.Request(method=b"GET", url=url, headers=[(b"host", b"example.com")])
print(f"request = {req!r}")
print(f"  method={req.method!r} url={req.url!r}")

assert url.host == b"example.com"
assert origin.port == 443
assert req.method == b"GET"
```

```python
# Example 3: exception hierarchy — sanity-check error class structure.
import httpcore2

# httpcore2 exceptions inherit from a common base.
for exc_name in ["ConnectError", "ReadError", "WriteError", "ConnectTimeout", "ReadTimeout"]:
    exc_cls = getattr(httpcore2, exc_name)
    print(f"  {exc_name}: {exc_cls.__mro__[:3]}")
    assert issubclass(exc_cls, Exception)

# Verify we can raise + catch via base classes.
try:
    raise httpcore2.ConnectError("simulated")
except httpcore2.ConnectError as e:
    print(f"caught ConnectError: {e}")

# TimeoutException is the umbrella for timeout-y errors.
assert issubclass(httpcore2.ConnectTimeout, httpcore2.TimeoutException)
assert issubclass(httpcore2.ReadTimeout, httpcore2.TimeoutException)
```
