```python
# Example 1: format_size / parse_size — human-readable byte sizes
from humanfriendly import format_size, parse_size

assert format_size(1024) == "1.02 KB"
assert format_size(1024, binary=True) == "1 KiB"
assert format_size(1000 * 1000) == "1 MB"

assert parse_size("1 KB") == 1000
assert parse_size("1 KiB") == 1024
assert parse_size("5 MB") == 5_000_000

print(format_size(1234567890))
print(format_size(1234567890, binary=True))
print(parse_size("2.5 GB"))
```

```python
# Example 2: format_timespan / parse_timespan — human-readable durations
from humanfriendly import format_timespan, parse_timespan

# Sub-minute values are rounded to seconds unless detailed=True.
assert format_timespan(0.001) == "0 seconds"
assert format_timespan(0.001, detailed=True) == "1 millisecond"
assert format_timespan(60) == "1 minute"
assert format_timespan(3600) == "1 hour"

# parse_timespan accepts a single <number><unit> token (not "X and Y" phrases).
assert parse_timespan("5s") == 5.0
assert parse_timespan("2 minutes") == 120.0
assert parse_timespan("1.5h") == 5400.0

print(format_timespan(90061))
print(format_timespan(0.5, detailed=True))
print(parse_timespan("30 minutes"))
```

```python
# Example 3: format_number, round_number, pluralize, concatenate
from humanfriendly import format_number, round_number
from humanfriendly.text import pluralize, concatenate  # canonical home (top-level re-exports are deprecated)

assert format_number(1234567.89) == "1,234,567.89"
assert round_number(1.2345) == "1.23"
assert round_number(2.0) == "2"
assert round_number(2.0, keep_width=True) == "2.00"

assert pluralize(1, "file") == "1 file"
assert pluralize(2, "file") == "2 files"
assert pluralize(2, "mouse", "mice") == "2 mice"

assert concatenate(["a", "b", "c"]) == "a, b and c"
assert concatenate(["only"]) == "only"

print(format_number(9876543.21))
print(concatenate(["apples", "oranges", "pears"]))
```

```python
# Example 4: parse_date, format_length, parse_length
from humanfriendly import parse_date, format_length, parse_length

# parse_date returns a (year, month, day, hour, minute, second) tuple
result = parse_date("2025-06-08 14:30:00")
assert result == (2025, 6, 8, 14, 30, 0)

date_only = parse_date("2025-06-08")
assert date_only == (2025, 6, 8, 0, 0, 0)

# length helpers (meters base unit)
assert parse_length("1 km") == 1000
assert parse_length("5 m") == 5

print(parse_date("2025-12-31 23:59:59"))
print(format_length(1234))
print(format_length(0.05))
```
