## Example 1: Time humanization

```python
import humanize
from datetime import datetime, timedelta

# Example 1: Time humanization
print("Example 1: Time humanization")

# Current time and past times
now = datetime.now()
past_times = [
    now - timedelta(seconds=30),
    now - timedelta(minutes=5),
    now - timedelta(hours=2),
    now - timedelta(days=1),
    now - timedelta(days=7)
]

print(f"Current time: {now}")
print("Humanized past times:")

for past_time in past_times:
    humanized = humanize.naturaltime(past_time)
    print(f"  {past_time.strftime('%Y-%m-%d %H:%M:%S')} -> {humanized}")

# Future times
future_time = now + timedelta(hours=3)
future_humanized = humanize.naturaltime(future_time)
print(f"Future time: {future_humanized}")

# timedelta inputs give deterministic text
assert humanize.naturaltime(timedelta(minutes=5)) == "5 minutes ago"
assert humanize.naturaltime(timedelta(hours=2)) == "2 hours ago"
assert humanize.naturaltime(timedelta(days=1)) == "a day ago"
assert humanize.naturaltime(timedelta(hours=-3)) == "3 hours from now"
assert future_humanized.endswith("from now")
```

## Example 2: File size humanization

```python
import humanize

# Example 2: File size humanization
print("Example 2: File size humanization")

# Different file sizes in bytes
file_sizes = [
    512,
    1024,
    1024 * 1024,
    1024 * 1024 * 1024,
    1024 * 1024 * 1024 * 1024,
    1024 * 1024 * 1024 * 1024 * 1024
]

print("File sizes in human-readable format:")
for size in file_sizes:
    natural_size = humanize.naturalsize(size)
    binary_size = humanize.naturalsize(size, binary=True)
    print(f"  {size:>15} bytes -> {natural_size:>10} ({binary_size:>10} binary)")

# Test with decimal values
decimal_size = 1500000
print(f"\nDecimal example:")
print(f"  {decimal_size} bytes -> {humanize.naturalsize(decimal_size)}")
print(f"  {decimal_size} bytes -> {humanize.naturalsize(decimal_size, binary=True)} (binary)")

assert humanize.naturalsize(512) == "512 Bytes"
assert humanize.naturalsize(1024) == "1.0 kB"  # decimal (SI) units by default
assert humanize.naturalsize(1024, binary=True) == "1.0 KiB"
assert humanize.naturalsize(decimal_size) == "1.5 MB"
assert humanize.naturalsize(1024 ** 3, binary=True) == "1.0 GiB"
```

## Example 3: Number formatting and ordinals

```python
import humanize

# Example 3: Number formatting and ordinals
print("Example 3: Number formatting and ordinals")

# Integer to word conversion
numbers = [0, 1, 12, 42, 100, 1000, 1234567]

print("Numbers as words:")
for num in numbers:
    word_form = humanize.intword(num)
    print(f"  {num:>10} -> {word_form}")

# Ordinal numbers
ordinal_numbers = [1, 2, 3, 4, 21, 22, 23, 101, 102, 103]
print("\nOrdinal numbers:")
for num in ordinal_numbers:
    ordinal = humanize.ordinal(num)
    print(f"  {num:>3} -> {ordinal}")

# AP number style (Associated Press)
ap_numbers = [0, 1, 5, 9, 10, 15]
print("\nAP style numbers:")
for num in ap_numbers:
    ap_num = humanize.apnumber(num)
    print(f"  {num:>2} -> {ap_num}")

assert humanize.intword(1234567) == "1.2 million"
assert humanize.intword(42) == "42"
assert humanize.intcomma(1234567) == "1,234,567"
assert [humanize.ordinal(n) for n in (1, 2, 3, 4, 11, 12, 13, 21, 22, 23, 101)] == [
    "1st", "2nd", "3rd", "4th", "11th", "12th", "13th", "21st", "22nd", "23rd", "101st"]
assert [humanize.apnumber(n) for n in ap_numbers] == ["zero", "one", "five", "nine", "10", "15"]
```

## Example 4: Fractional numbers and scientific notation

```python
import humanize

# Example 4: Fractional numbers and scientific notation
print("Example 4: Fractional and scientific numbers")

# Fractional numbers
fractions = [0.5, 0.25, 0.75, 1.5, 2.25, 3.333]
print("Fractional numbers:")
for frac in fractions:
    fraction_text = humanize.fractional(frac)
    print(f"  {frac:>5} -> {fraction_text}")

# Scientific notation to readable format
scientific_numbers = [1e6, 1.5e9, 2.3e12, 4.7e-3, 8.9e-6]
print("\nScientific notation:")
for sci_num in scientific_numbers:
    readable = humanize.scientific(sci_num)
    intword = humanize.intword(sci_num)
    print(f"  {sci_num:>10} -> {readable:>15} ({intword})")

# Large numbers with metric suffixes
large_numbers = [1000, 1500, 1000000, 2500000, 1000000000]
print("\nMetric suffixes:")
for num in large_numbers:
    metric = humanize.metric(num)
    print(f"  {num:>10} -> {metric}")

assert [humanize.fractional(f) for f in (0.5, 0.25, 1.5, 2.25)] == ["1/2", "1/4", "1 1/2", "2 1/4"]
assert humanize.scientific(1e6) == "1.00 x 10⁶"
assert humanize.scientific(4.7e-3) == "4.70 x 10⁻³"
assert humanize.intword(1.5e9) == "1.5 billion"
assert [humanize.metric(n) for n in (1000, 2500000, 1000000000)] == ["1.00 k", "2.50 M", "1.00 G"]
```

## Example 5: Date formatting and delta calculations

```python
import humanize

# Example 5: Date formatting and delta calculations
print("Example 5: Date formatting and delta")

# Natural date formatting
from datetime import date, timedelta

dates = [
    date(2023, 1, 1),
    date(2023, 7, 4),
    date(2023, 12, 25),
    date.today()
]

print("Natural date formatting:")
for d in dates:
    natural_date = humanize.naturaldate(d)
    print(f"  {d} -> {natural_date}")

# Natural day formatting (day of year)
days_of_year = [1, 32, 100, 200, 365]
print("\nDay of year:")
for day in days_of_year:
    natural_day = humanize.naturalday(date(2023, 1, 1) + timedelta(days=day-1))
    print(f"  Day {day:>3} -> {natural_day}")

# Precise delta (time differences)
deltas = [3661, 7200, 86400, 604800]  # seconds
print("\nPrecise time deltas:")
for delta_seconds in deltas:
    precise = humanize.precisedelta(delta_seconds)
    print(f"  {delta_seconds:>6} seconds -> {precise}")

assert humanize.naturaldate(date(2023, 7, 4)) == "Jul 04 2023"  # dates far from today include the year
assert humanize.naturalday(date.today()) == "today"
assert humanize.naturalday(date.today() - timedelta(days=1)) == "yesterday"
assert humanize.naturalday(date.today() + timedelta(days=1)) == "tomorrow"
assert humanize.naturalday(date(2023, 2, 1)) == "Feb 01"
assert humanize.precisedelta(3661) == "1 hour, 1 minute and 1 second"
assert humanize.precisedelta(604800) == "7 days"
```
