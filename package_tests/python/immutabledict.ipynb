```python
# Example 1: Basic construction from dict + key/value access.
from immutabledict import immutabledict

d = immutabledict({"a": 1, "b": 2, "c": 3})
print(f"d = {d}")
print(f"d['a'] = {d['a']}")
print(f"len(d) = {len(d)}")
print(f"list(d.keys()) = {list(d.keys())}")
print(f"list(d.values()) = {list(d.values())}")

assert d["a"] == 1
assert d["b"] == 2
assert len(d) == 3
assert "a" in d
assert "z" not in d
```

```python
# Example 2: Immutability — mutating operations must raise; update() is functional.
from immutabledict import immutabledict

d = immutabledict({"x": 10, "y": 20})

mutation_errors = 0
for op in [
    lambda: d.__setitem__("x", 99),
    lambda: d.__delitem__("x"),
    lambda: d.pop("x"),
    lambda: d.clear(),
]:
    try:
        op()
    except (TypeError, AttributeError) as e:
        mutation_errors += 1
        print(f"Blocked: {type(e).__name__}: {str(e)[:60]}")

print(f"\nTotal mutation operations blocked: {mutation_errors}")
assert mutation_errors == 4

# Since immutabledict 4.x, update() is non-mutating: it returns a NEW
# immutabledict and leaves the original untouched.
updated = d.update({"z": 30})
assert isinstance(updated, immutabledict) and updated["z"] == 30
assert dict(d) == {"x": 10, "y": 20}
```

```python
# Example 3: Construction from kwargs, iterable of pairs, and copy semantics.
from immutabledict import immutabledict

d1 = immutabledict(a=1, b=2)
d2 = immutabledict([("a", 1), ("b", 2)])
d3 = immutabledict({"a": 1, "b": 2})

print(f"From kwargs: {d1}")
print(f"From pairs:  {d2}")
print(f"From dict:   {d3}")

assert dict(d1) == dict(d2) == dict(d3)

# Equality across construction paths.
assert d1 == d2
assert d2 == d3

# Empty immutabledict.
empty = immutabledict()
assert len(empty) == 0
assert dict(empty) == {}
```

```python
# Example 4: Hashability — immutabledict instances can be set/dict keys.
from immutabledict import immutabledict

a = immutabledict({"k": 1, "v": 2})
b = immutabledict({"k": 1, "v": 2})
c = immutabledict({"k": 1, "v": 3})

print(f"hash(a) = {hash(a)}")
print(f"hash(b) = {hash(b)} (same as a: {hash(a) == hash(b)})")
print(f"hash(c) = {hash(c)} (differs from a: {hash(a) != hash(c)})")

# Use as set member.
s = {a, b, c}
print(f"\nSet of 3 immutabledicts collapsed to size: {len(s)}")

# Use as dict key.
registry = {a: "first", c: "second"}
assert registry[b] == "first"
assert len(s) == 2
```

```python
# Example 5: Functional updates — merge via | operator yields a new immutabledict.
from immutabledict import immutabledict

base = immutabledict({"a": 1, "b": 2})
overlay = {"b": 20, "c": 3}

merged = base | overlay
print(f"base    = {base}")
print(f"overlay = {overlay}")
print(f"merged  = {merged}")
print(f"type(merged) = {type(merged).__name__}")

# Original untouched.
assert base == immutabledict({"a": 1, "b": 2})
assert merged["a"] == 1
assert merged["b"] == 20
assert merged["c"] == 3
assert isinstance(merged, immutabledict)
print("\nAll immutabledict smoke tests passed")
```
