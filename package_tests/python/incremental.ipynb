```python
# Example 1: Construct a Version and inspect its components.
from incremental import Version

v = Version("MyProject", 24, 11, 0)
print(f"package: {v.package}")
print(f"major:   {v.major}")
print(f"minor:   {v.minor}")
print(f"micro:   {v.micro}")
print(f"release_candidate: {v.release_candidate}")
print(f"dev:     {v.dev}")

assert v.package == "MyProject"
assert v.major == 24
assert v.minor == 11
assert v.micro == 0
```

```python
# Example 2: String representations (short, public, base).
from incremental import Version

v = Version("MyProject", 1, 2, 3)
print(f"short():  {v.short()}")
print(f"public(): {v.public()}")
print(f"base():   {v.base()}")
print(f"repr:     {v!r}")

assert v.short() == "1.2.3"
assert v.public() == "1.2.3"
assert v.base() == "1.2.3"
```

```python
# Example 3: Pre-release and dev versions render with PEP 440 suffixes.
from incremental import Version

rc = Version("MyProject", 24, 11, 0, release_candidate=1)
print(f"rc public(): {rc.public()}")

dev = Version("MyProject", 24, 11, 0, dev=2)
print(f"dev public(): {dev.public()}")

post = Version("MyProject", 24, 11, 0, post=1)
print(f"post public(): {post.public()}")

assert "rc1" in rc.public()
assert "dev2" in dev.public()
assert "post1" in post.public()
```

```python
# Example 4: Version ordering and equality.
from incremental import Version

older = Version("MyProject", 1, 0, 0)
newer = Version("MyProject", 2, 0, 0)
same = Version("MyProject", 1, 0, 0)
rc = Version("MyProject", 2, 0, 0, release_candidate=1)

assert older < newer
assert newer > older
assert older == same
assert rc < newer  # rc precedes final

ordered = sorted([newer, older, rc])
print([v.public() for v in ordered])
assert ordered[0] == older
assert ordered[-1] == newer
```

```python
# Example 5: getVersionString convenience + NEXT sentinel handling.
from incremental import Version

v = Version("MyProject", 24, 11, 0)
s = v.public()
print(f"public string: {s}")

# The NEXT sentinel is used during development before release-bumping.
next_v = Version("MyProject", "NEXT", 0, 0)
print(f"NEXT short(): {next_v.short()}")
assert "NEXT" in next_v.short() or next_v.major == "NEXT"

# Tuple comparison fields are exposed via _Version__cmpkey style; just confirm core attrs.
assert v.package == "MyProject"
assert isinstance(v.public(), str)
print("incremental smoke test OK")
```
