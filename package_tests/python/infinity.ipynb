## Example 1: Creating and using Infinity objects

```python
import infinity

# Example 1: Creating and using Infinity objects
print("Example 1: Creating and using Infinity objects")

# Create positive and negative infinity objects
pos_inf = infinity.Infinity()
neg_inf = -infinity.Infinity()

print(f"Positive infinity: {pos_inf}")
print(f"Negative infinity: {neg_inf}")
print(f"Type of infinity object: {type(pos_inf)}")

# Test basic comparisons with infinity
numbers = [1, 100, 1000000, -50]
print(f"\nComparisons with regular numbers:")
for num in numbers:
    print(f"  {num} < {pos_inf}: {num < pos_inf}")
    print(f"  {num} > {neg_inf}: {num > neg_inf}")

# Test infinity comparisons
print(f"\nInfinity comparisons:")
print(f"  {pos_inf} == {pos_inf}: {pos_inf == pos_inf}")
print(f"  {pos_inf} > {neg_inf}: {pos_inf > neg_inf}")

assert str(pos_inf) == "inf" and str(neg_inf) == "-inf"
assert pos_inf.positive and not neg_inf.positive
assert all(neg_inf < n < pos_inf for n in numbers)
assert 10 ** 100 < pos_inf and neg_inf < -(10 ** 100)
assert pos_inf == infinity.Infinity() and pos_inf != neg_inf
```

## Example 2: Using the is_infinite() function

```python
import infinity

# Example 2: Using the is_infinite() function
print("Example 2: Using the is_infinite() function")

# Create various values to test
test_values = [
    infinity.Infinity(),
    -infinity.Infinity(), 
    42,
    0,
    -100,
    3.14159,
    float('inf'),
    float('-inf')
]

print("Testing is_infinite() function:")
for value in test_values:
    is_inf = infinity.is_infinite(value)
    print(f"  is_infinite({value}): {is_inf}")

# Test with collections containing infinity
print(f"\nTesting with data structures:")
infinity_list = [1, 2, infinity.Infinity(), 4, 5]
for i, val in enumerate(infinity_list):
    print(f"  is_infinite(list[{i}] = {val}): {infinity.is_infinite(val)}")

# Infinity objects and float infinities are infinite; ordinary numbers are not
assert [infinity.is_infinite(v) for v in test_values] == [True, True, False, False, False, False, True, True]
assert [i for i, v in enumerate(infinity_list) if infinity.is_infinite(v)] == [2]
```

## Example 3: Using inf shortcut and arithmetic operations

```python
import infinity

# Example 3: Using inf shortcut and arithmetic operations
print("Example 3: Using inf shortcut and arithmetic operations")

# Use the inf shortcut to create infinity
inf = infinity.inf
print(f"Shortcut infinity: {inf}")
print(f"Type: {type(inf)}")

# Test arithmetic with infinity (where mathematically valid)
print(f"\nArithmetic operations with infinity:")
print(f"  {inf} + 100 = {inf + 100}")
print(f"  {inf} - 50 = {inf - 50}")
print(f"  -{inf} = {-inf}")

# Test comparisons in sorting context
numbers = [5, -infinity.inf, 10, infinity.inf, 0, 3]
print(f"\nOriginal list: {numbers}")
sorted_numbers = sorted(numbers)
print(f"Sorted with infinities: {sorted_numbers}")

# Find min and max with infinity
print(f"\nMin/Max with infinity:")
print(f"  min(numbers): {min(numbers)}")
print(f"  max(numbers): {max(numbers)}")

# adding or subtracting a finite number leaves infinity unchanged
assert inf + 100 == inf and inf - 50 == inf
assert -inf == -infinity.Infinity() and -inf < 0 < inf
assert sorted_numbers[0] == -inf and sorted_numbers[-1] == inf
assert sorted_numbers[1:-1] == [0, 3, 5, 10]
assert (min(numbers), max(numbers)) == (-inf, inf)
```

## Example 4: Using total_ordering decorator

```python
import infinity
from infinity import total_ordering

# Example 4: Using total_ordering decorator
print("Example 4: Using total_ordering decorator")

# Create a custom class using total_ordering
@total_ordering
class CustomNumber:
    def __init__(self, value):
        self.value = value
    
    def __eq__(self, other):
        if isinstance(other, CustomNumber):
            return self.value == other.value
        return self.value == other
    
    def __lt__(self, other):
        if isinstance(other, CustomNumber):
            return self.value < other.value
        return self.value < other
    
    def __repr__(self):
        return f"CustomNumber({self.value})"

# Test total_ordering functionality
nums = [CustomNumber(5), CustomNumber(1), CustomNumber(10)]
print(f"Custom numbers: {nums}")

# Test all comparison operations (total_ordering generates these)
a, b = CustomNumber(5), CustomNumber(10)
print(f"\nComparison operations:")
print(f"  {a} < {b}: {a < b}")
print(f"  {a} <= {b}: {a <= b}")
print(f"  {a} > {b}: {a > b}")
print(f"  {a} >= {b}: {a >= b}")
print(f"  {a} == {b}: {a == b}")

import functools
assert total_ordering is functools.total_ordering  # infinity re-exports the standard decorator
# __le__, __gt__ and __ge__ were generated from __eq__ and __lt__
assert (a < b, a <= b, a > b, a >= b, a == b) == (True, True, False, False, False)
assert sorted(nums) == [CustomNumber(1), CustomNumber(5), CustomNumber(10)]
```

## Example 5: Practical use case - finding extremes with infinity bounds

```python
import infinity

# Example 5: Practical use case - finding extremes with infinity bounds
print("Example 5: Practical use case - finding extremes with infinity bounds")

# Simulate a data processing scenario with bounds
def find_in_range(data, min_val=None, max_val=None):
    """Find values within range, using infinity as default bounds"""
    if min_val is None:
        min_val = -infinity.inf
    if max_val is None:
        max_val = infinity.inf
    
    return [x for x in data if min_val <= x <= max_val]

# Test data
temperatures = [-50, -10, 0, 25, 30, 45, 100, 150]
print(f"Temperature data: {temperatures}")

# Find different ranges
print(f"\nFinding values in ranges:")
freezing = find_in_range(temperatures, max_val=0)
print(f"  Freezing (≤ 0°C): {freezing}")

room_temp = find_in_range(temperatures, 20, 30)
print(f"  Room temperature (20-30°C): {room_temp}")

above_boiling = find_in_range(temperatures, min_val=100)
print(f"  Above boiling (≥ 100°C): {above_boiling}")

all_temps = find_in_range(temperatures)  # Uses infinity bounds
print(f"  All temperatures (no bounds): {all_temps}")

assert freezing == [-50, -10, 0]
assert room_temp == [25, 30]
assert above_boiling == [100, 150]
assert all_temps == temperatures
```
