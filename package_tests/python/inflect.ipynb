## Example 1: Pluralization with plural_noun() and plural()

```python
import inflect

# Example 1: Pluralization with plural_noun() and plural()
print("Example 1: Pluralization functions")

# Create inflect engine
p = inflect.engine()

# Test plural_noun() function
singular_words = ['cat', 'mouse', 'child', 'foot', 'goose', 'tooth', 'person']
print("Using plural_noun():")
for word in singular_words:
    plural_form = p.plural_noun(word)
    print(f"  {word} -> {plural_form}")

# plural() also handles verbs and pronouns; plural_verb()/plural_adj() are the specific forms
print(f"\nUsing plural(), plural_verb() and plural_adj():")
for word, plural_form in [("mouse", p.plural("mouse")), ("was", p.plural("was")), ("I", p.plural("I")),
                          ("is", p.plural_verb("is")), ("this", p.plural_adj("this"))]:
    print(f"  '{word}' -> '{plural_form}'")

# Test with numbers
print(f"\nPlurals with different counts:")
for count in [0, 1, 2, 5]:
    result = p.plural_noun('book', count)
    print(f"  {count} {result}")

assert [p.plural_noun(w) for w in singular_words] == ["cats", "mice", "children", "feet", "geese", "teeth", "people"]
assert (p.plural("was"), p.plural("I"), p.plural_verb("is"), p.plural_adj("this")) == ("were", "we", "are", "these")
assert [p.plural_noun("book", c) for c in (0, 1, 2, 5)] == ["books", "book", "books", "books"]
assert (p.no("cat", 0), p.no("cat", 3)) == ("no cats", "3 cats")
```

## Example 2: Number to words conversion with number_to_words()

```python
import inflect

# Example 2: Number to words conversion with number_to_words()
print("Example 2: Number to words conversion")

# Create inflect engine
p = inflect.engine()

# Convert various numbers to words
numbers = [0, 1, 13, 42, 99, 100, 256, 1000, 1234, 1000000]
print("Converting numbers to words:")
for num in numbers:
    words = p.number_to_words(num)
    print(f"  {num} -> '{words}'")

# Test with negative numbers
print(f"\nNegative numbers:")
negative_nums = [-1, -15, -100]
for num in negative_nums:
    words = p.number_to_words(num)
    print(f"  {num} -> '{words}'")

# Test with decimal numbers
print(f"\nDecimal numbers:")
decimal_nums = [3.14, 0.5, 1.25]
for num in decimal_nums:
    words = p.number_to_words(num)
    print(f"  {num} -> '{words}'")

assert p.number_to_words(42) == "forty-two"
assert p.number_to_words(256) == "two hundred and fifty-six"
assert p.number_to_words(1234) == "one thousand, two hundred and thirty-four"
assert p.number_to_words(-15) == "minus fifteen"
assert p.number_to_words(3.14) == "three point one four"
```

## Example 3: Ordinal numbers with ordinal()

```python
import inflect

# Example 3: Ordinal numbers with ordinal()
print("Example 3: Ordinal number conversion")

# Create inflect engine
p = inflect.engine()

# Convert numbers to ordinal form
numbers = [1, 2, 3, 4, 11, 12, 13, 21, 22, 23, 101, 111, 121]
print("Converting numbers to ordinals:")
for num in numbers:
    ordinal_form = p.ordinal(num)
    print(f"  {num} -> {ordinal_form}")

# Test ordinals in context
print(f"\nOrdinals in sentences:")
positions = [1, 2, 3, 5, 10, 25]
for pos in positions:
    ordinal_form = p.ordinal(pos)
    print(f"  'You finished in {ordinal_form} place'")

# Combine with number_to_words for spelled-out ordinals
print(f"\nSpelled-out ordinals:")
for num in [1, 2, 3, 11, 21]:
    words = p.number_to_words(p.ordinal(num))
    print(f"  {num} -> '{words}'")

assert [p.ordinal(n) for n in numbers] == [
    "1st", "2nd", "3rd", "4th", "11th", "12th", "13th", "21st", "22nd", "23rd", "101st", "111th", "121st"]
assert [p.number_to_words(p.ordinal(n)) for n in (1, 2, 3, 11, 21)] == [
    "first", "second", "third", "eleventh", "twenty-first"]
```

## Example 4: Articles with a() and an()

```python
import inflect

# Example 4: Articles with a() and an()
print("Example 4: Article selection with a() and an()")

# Create inflect engine
p = inflect.engine()

# Test article selection with various words
words = ['apple', 'banana', 'car', 'elephant', 'house', 'ice cream', 
         'orange', 'university', 'hour', 'honest person', 'European']
print("Selecting correct articles:")
for word in words:
    article_word = p.a(word)  # Returns "a word" or "an word"
    print(f"  {word} -> '{article_word}'")

# Test with just getting the article
print(f"\nJust the article:")
for word in ['apple', 'car', 'hour', 'university', 'honest']:
    article = p.a(word).split()[0]  # Extract just the article
    print(f"  {word} -> article: '{article}'")

# Test in sentence context
print(f"\nIn sentences:")
items = ['umbrella', 'book', 'computer', 'hour']
for item in items:
    sentence = f"I need {p.a(item)} for the trip."
    print(f"  '{sentence}'")

# The article follows pronunciation, not spelling: "an hour", "a university", "a European"
assert [p.a(w) for w in ("apple", "car", "hour", "university", "honest person", "European")] == [
    "an apple", "a car", "an hour", "a university", "an honest person", "a European"]
assert p.an("umbrella") == "an umbrella"
```

## Example 5: List joining with join() and text comparison

```python
import inflect

# Example 5: List joining with join() and text comparison
print("Example 5: List joining and word comparison")

# Create inflect engine
p = inflect.engine()

# Test join() function for natural language lists
items_lists = [
    ['apples'],
    ['apples', 'bananas'], 
    ['apples', 'bananas', 'oranges'],
    ['red', 'green', 'blue', 'yellow', 'purple']
]

print("Joining lists naturally:")
for items in items_lists:
    joined = p.join(items)
    print(f"  {items} -> '{joined}'")

# Test join with different conjunctions
print(f"\nJoining with different conjunctions:")
colors = ['red', 'green', 'blue']
print(f"  Default: '{p.join(colors)}'")
print(f"  With 'or': '{p.join(colors, conj='or')}'")

# Test singular_noun() to get singular forms
print(f"\nConverting plurals back to singular:")
plural_words = ['cats', 'mice', 'children', 'feet', 'geese']
for word in plural_words:
    singular = p.singular_noun(word)
    print(f"  {word} -> {singular}")

assert [p.join(items) for items in items_lists[:3]] == ["apples", "apples and bananas", "apples, bananas, and oranges"]
assert p.join(colors, conj="or") == "red, green, or blue"
assert [p.singular_noun(w) for w in plural_words] == ["cat", "mouse", "child", "foot", "goose"]
assert p.singular_noun("cat") is False  # already singular
assert p.compare("cat", "cats") == "s:p" and p.compare("cat", "dog") is False
```
