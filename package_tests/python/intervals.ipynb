## Example 1: Creating integer intervals with IntInterval.closed()

```python
import intervals

# Example 1: Creating integer intervals with IntInterval.closed()
print("Example 1: Creating integer intervals")

# Create different types of integer intervals
int_intervals = [
    intervals.IntInterval.closed(1, 10),     # Closed interval [1, 10]
    intervals.IntInterval.open(0, 5),        # Open interval (0, 5)
    intervals.IntInterval.closed_open(1, 8), # Half-open interval [1, 8)
    intervals.IntInterval.open_closed(2, 12) # Half-open interval (2, 12]
]

interval_names = ["Closed [1, 10]", "Open (0, 5)", "Closed-open [1, 8)", "Open-closed (2, 12]"]

print("Integer interval properties:")
for name, interval in zip(interval_names, int_intervals):
    print(f"  {name}:")
    print(f"    Representation: {interval}")
    print(f"    Lower bound: {interval.lower}")
    print(f"    Upper bound: {interval.upper}")
    print(f"    Is closed: {interval.is_closed}")
    print(f"    Centre: {interval.centre}")

assert [str(i) for i in int_intervals] == ["[1, 10]", "(0, 5)", "[1, 8)", "(2, 12]"]
assert [i.is_closed for i in int_intervals] == [True, False, False, False]
assert int_intervals[0].centre == 5.5
# membership respects open and closed ends
assert 1 in int_intervals[0] and 10 in int_intervals[0]
assert 0 not in int_intervals[1] and 5 not in int_intervals[1] and 4 in int_intervals[1]
assert 8 not in int_intervals[2] and 12 in int_intervals[3]
```

## Example 2: Float intervals with FloatInterval operations

```python
import intervals

# Example 2: Float intervals with FloatInterval operations
print("Example 2: Float interval operations")

# Create float intervals
float_interval1 = intervals.FloatInterval.closed(1.5, 5.5)
float_interval2 = intervals.FloatInterval.closed(3.0, 8.0)
float_interval3 = intervals.FloatInterval.open(0.0, 10.0)

print("Float interval properties:")
print(f"Interval 1 [1.5, 5.5]: {float_interval1}")
print(f"  Length: {float_interval1.length}")
print(f"  Radius: {float_interval1.radius}")

print(f"Interval 2 [3.0, 8.0]: {float_interval2}")
print(f"  Centre: {float_interval2.centre}")
print(f"  Is open: {float_interval2.is_open}")

print(f"Interval 3 (0.0, 10.0): {float_interval3}")
print(f"  Lower inclusive: {float_interval3.lower_inc}")
print(f"  Upper inclusive: {float_interval3.upper_inc}")

# Test interval relationships
print(f"\nInterval relationships:")
print(f"Interval 1 connected to 2: {float_interval1.is_connected(float_interval2)}")

assert float_interval1.length == 4.0 and float_interval1.radius == 2.0
assert float_interval2.centre == 5.5 and not float_interval2.is_open
assert not float_interval3.lower_inc and not float_interval3.upper_inc
assert float_interval1.is_connected(float_interval2)
assert str(float_interval1 & float_interval2) == "[3.0, 5.5]"  # intersection
assert str(float_interval1 | float_interval2) == "[1.5, 8.0]"  # union of connected intervals
assert 4.0 in float_interval1 and float_interval1 in float_interval3
```

## Example 3: Date intervals with DateInterval

```python
import intervals
from datetime import date

# Example 3: Date intervals with DateInterval
print("Example 3: Date interval operations")

# Create date intervals
start_date = date(2023, 1, 1)
mid_date = date(2023, 6, 15)
end_date = date(2023, 12, 31)

date_interval1 = intervals.DateInterval.closed(start_date, mid_date)
date_interval2 = intervals.DateInterval.closed(mid_date, end_date)
date_interval3 = intervals.DateInterval.open(start_date, end_date)

print("Date interval properties:")
print(f"Q1-Q2 2023: {date_interval1}")
print(f"  Lower: {date_interval1.lower}")
print(f"  Upper: {date_interval1.upper}")
print(f"  Length: {date_interval1.length}")

print(f"Q3-Q4 2023: {date_interval2}")
# Skip centre() for dates as it causes TypeError

print(f"Year 2023 (open): {date_interval3}")
print(f"  Is closed: {date_interval3.is_closed}")

# Test date operations
print(f"\nDate interval relationships:")
print(f"Q1-Q2 connected to Q3-Q4: {date_interval1.is_connected(date_interval2)}")

from datetime import timedelta
assert date_interval1.length == timedelta(days=165)
assert not date_interval3.is_closed
assert date_interval1.is_connected(date_interval2)
assert date(2023, 3, 1) in date_interval1 and date(2023, 9, 1) not in date_interval1
assert start_date not in date_interval3  # open lower bound
```

## Example 4: Interval factory with IntervalFactory()

```python
import intervals

# Example 4: Interval factory with IntervalFactory()
print("Example 4: Using interval factory")

# Create interval factory
factory = intervals.IntervalFactory()

print("Creating intervals with factory:")

# Create various interval types through factory (callable)
int_interval = factory([5, 15], int)
print(f"Integer interval: {int_interval}")
print(f"  Type: {type(int_interval).__name__}")

float_interval = factory([2.5, 7.8], float)
print(f"Float interval: {float_interval}")
print(f"  Type: {type(float_interval).__name__}")

# Create interval from string using from_string method
string_interval = factory.from_string('[10, 20)')
print(f"From string '[10, 20)': {string_interval}")
print(f"  Lower inclusive: {string_interval.lower_inc}")
print(f"  Upper inclusive: {string_interval.upper_inc}")

# Show available interval classes
print(f"\nAvailable interval classes: {len(factory.interval_classes)}")
for cls in factory.interval_classes[:3]:  # Show first 3
    print(f"  {cls.__name__}")

assert type(int_interval).__name__ == "IntInterval" and str(int_interval) == "[5, 15]"
assert type(float_interval).__name__ == "FloatInterval"
assert str(string_interval) == "[10, 20)"
assert string_interval.lower_inc and not string_interval.upper_inc
```

## Example 5: Decimal intervals and canonicalize()

```python
import intervals
from decimal import Decimal

# Example 5: Decimal intervals and canonicalize()
print("Example 5: Decimal intervals and canonicalization")

# Create decimal intervals for precise arithmetic
decimal1 = Decimal('1.25')
decimal2 = Decimal('3.75')
decimal3 = Decimal('2.50')

decimal_interval1 = intervals.DecimalInterval.closed(decimal1, decimal2)
decimal_interval2 = intervals.DecimalInterval.open(decimal3, Decimal('5.00'))

print("Decimal interval operations:")
print(f"Decimal interval 1: {decimal_interval1}")
print(f"  Lower: {decimal_interval1.lower}")
print(f"  Upper: {decimal_interval1.upper}")
print(f"  Centre: {decimal_interval1.centre}")

print(f"Decimal interval 2: {decimal_interval2}")
print(f"  Length: {decimal_interval2.length}")
print(f"  Is open: {decimal_interval2.is_open}")

# Test canonicalization function
# canonicalize() takes ONE interval and only accepts DISCRETE ones (IntInterval): it rewrites an integer
# interval into its half-open [lower, upper) form. Decimal intervals are continuous, so they have none.
print(f"\nUsing canonicalize() function:")
assert decimal_interval1.discrete is False and decimal_interval2.discrete is False
int_open = intervals.IntInterval.open(1, 5)
int_closed = intervals.IntInterval.closed(1, 5)
canonical_open = intervals.canonicalize(int_open)
canonical_closed = intervals.canonicalize(int_closed)
print(f"canonicalize({int_open}) -> {canonical_open}")
print(f"canonicalize({int_closed}) -> {canonical_closed}")
assert str(canonical_open) == "[2, 5)" and canonical_open == int_open, canonical_open
assert str(canonical_closed) == "[1, 6)" and canonical_closed == int_closed, canonical_closed

assert decimal_interval1.centre == Decimal("2.5")
assert decimal_interval2.length == Decimal("2.50") and decimal_interval2.is_open
assert str(decimal_interval1 & decimal_interval2) == "(2.50, 3.75]"
```
