## Example 1: Basic PathManager Operations and File I/O

Demonstrates core PathManager functions: open, exists, isfile, isdir, mkdirs, ls, and copy.

```python
# Example 1: Basic PathManager operations and file I/O
try:
    from iopath.common.file_io import PathManager
    import tempfile
    import os
    
    print("=== IOPath PathManager Setup ===")
    # Create PathManager instance (unique to iopath)
    path_manager = PathManager()
    print(f"PathManager created: {type(path_manager)}")
    print(f"Number of registered handlers: {len(path_manager._path_handlers)}")
    
    print("\n=== Basic File Operations with PathManager ===")
    with tempfile.TemporaryDirectory() as temp_dir:
        # Test file path operations using PathManager API
        test_file = os.path.join(temp_dir, 'test.txt')
        test_content = "Hello IOPath! This is test content."
        
        # Write file using path_manager.open (unique to iopath)
        with path_manager.open(test_file, 'w') as f:
            f.write(test_content)
        
        print(f"File written: {test_file}")
        
        # Check file existence using path_manager.exists
        print(f"File exists: {path_manager.exists(test_file)}")
        print(f"Is file: {path_manager.isfile(test_file)}")
        print(f"Is directory: {path_manager.isdir(test_file)}")
        
        # Read file back using path_manager.open
        with path_manager.open(test_file, 'r') as f:
            read_content = f.read()
        
        print(f"Content match: {test_content == read_content}")
        print(f"Content length: {len(read_content)} characters")
        
        print("\n=== Directory Operations ===")
        # Create nested directory structure using mkdirs (unique to iopath)
        nested_dir = os.path.join(temp_dir, 'nested', 'deep', 'structure')
        path_manager.mkdirs(nested_dir)
        print(f"Created nested directory: {nested_dir}")
        print(f"Directory exists: {path_manager.exists(nested_dir)}")
        print(f"Is directory: {path_manager.isdir(nested_dir)}")
        
        # List directory contents using ls (unique to iopath)
        contents = path_manager.ls(temp_dir)
        print(f"Directory contents: {contents}")
        print(f"Number of items: {len(contents)}")
        
        # Copy file using path_manager.copy (unique to iopath)
        copy_target = os.path.join(nested_dir, 'copied_test.txt')
        path_manager.copy(test_file, copy_target)
        print(f"File copied to: {copy_target}")
        print(f"Copy exists: {path_manager.exists(copy_target)}")
        
        # Verify copy content
        with path_manager.open(copy_target, 'r') as f:
            copy_content = f.read()
        print(f"Copy content matches: {copy_content == test_content}")
        
        # Create multiple files
        print("\n=== Batch File Operations ===")
        for i in range(5):
            file_path = os.path.join(temp_dir, f'file_{i}.txt')
            with path_manager.open(file_path, 'w') as f:
                f.write(f'Content for file {i}')
        
        # List all files
        all_files = path_manager.ls(temp_dir)
        txt_files = [f for f in all_files if f.endswith('.txt')]
        print(f"Created {len(txt_files)} text files")
        print(f"Files: {txt_files}")
    
    print("Example 1: Basic PathManager operations test passed")
    
except Exception as e:
    print(f"Example 1: Error in file operations test: {e}")
    import traceback
    traceback.print_exc()
    raise
```

## Example 2: Binary File Operations and Mode Support

Demonstrates PathManager with binary/text modes, append mode, and copy operations.

```python
# Example 2: Binary file operations and mode support
try:
    from iopath.common.file_io import PathManager
    import tempfile
    import os
    
    print("=== Binary File Operations ===")
    path_manager = PathManager()
    
    with tempfile.TemporaryDirectory() as temp_dir:
        # Test binary data handling
        binary_file = os.path.join(temp_dir, 'binary_data.bin')
        
        # Create some binary data
        binary_data = bytes([i % 256 for i in range(1000)])  # 1000 bytes of data
        
        # Write binary data using path_manager.open with 'wb' mode
        with path_manager.open(binary_file, 'wb') as f:
            f.write(binary_data)
        
        print(f"Binary file written: {len(binary_data)} bytes")
        print(f"File exists: {path_manager.exists(binary_file)}")
        
        # Read binary data back using path_manager.open with 'rb' mode
        with path_manager.open(binary_file, 'rb') as f:
            read_binary = f.read()
        
        print(f"Binary data integrity check: {binary_data == read_binary}")
        print(f"First 10 bytes: {list(read_binary[:10])}")
        print(f"Last 10 bytes: {list(read_binary[-10:])}")
        
        print("\n=== Text vs Binary Mode Comparison ===")
        # Text file
        text_file = os.path.join(temp_dir, 'text_data.txt')
        text_content = "Line 1\nLine 2\nLine 3\n"
        
        with path_manager.open(text_file, 'w') as f:
            f.write(text_content)
        
        with path_manager.open(text_file, 'r') as f:
            read_text = f.read()
        
        print(f"Text mode - Lines: {len(read_text.splitlines())}")
        print(f"Text content preserved: {text_content == read_text}")
        
        # Same content as binary
        text_as_binary_file = os.path.join(temp_dir, 'text_as_binary.bin')
        
        with path_manager.open(text_as_binary_file, 'wb') as f:
            f.write(text_content.encode('utf-8'))
        
        with path_manager.open(text_as_binary_file, 'rb') as f:
            read_binary_text = f.read()
        
        print(f"Binary mode - Bytes: {len(read_binary_text)}")
        print(f"Decoded matches: {read_binary_text.decode('utf-8') == text_content}")
        
        print("\n=== Append Mode Operations ===")
        # Test append mode
        append_file = os.path.join(temp_dir, 'append_test.txt')
        
        # Initial write
        with path_manager.open(append_file, 'w') as f:
            f.write("Initial content\n")
        
        # Append more content
        with path_manager.open(append_file, 'a') as f:
            f.write("Appended line 1\n")
            f.write("Appended line 2\n")
        
        # Read all content
        with path_manager.open(append_file, 'r') as f:
            all_content = f.read()
        
        lines = all_content.splitlines()
        print(f"Total lines after append: {len(lines)}")
        print(f"Lines: {lines}")
        
        print("\n=== Multiple File Operations ===")
        # Create multiple files with different extensions
        file_types = {
            'text': ('.txt', 'Text content'),
            'binary': ('.bin', bytes([1, 2, 3, 4, 5])),
            'data': ('.dat', b'\x00\x01\x02\x03'),
            'log': ('.log', 'Log entry 1\nLog entry 2\n')
        }
        
        for name, (ext, content) in file_types.items():
            file_path = os.path.join(temp_dir, f'{name}{ext}')
            
            if isinstance(content, bytes):
                mode = 'wb'
            else:
                mode = 'w'
            
            with path_manager.open(file_path, mode) as f:
                f.write(content)
            
            print(f"Created {name}{ext}: exists={path_manager.exists(file_path)}, isfile={path_manager.isfile(file_path)}")
        
        # List all files
        all_files = path_manager.ls(temp_dir)
        print(f"\nTotal files created: {len(all_files)}")
        print(f"File list: {sorted(all_files)}")
        
        print("\n=== Copy Operations with Different Modes ===")
        # Copy binary file
        binary_copy = os.path.join(temp_dir, 'binary_copy.bin')
        path_manager.copy(binary_file, binary_copy)
        
        # Verify copy
        with path_manager.open(binary_copy, 'rb') as f:
            copy_data = f.read()
        
        print(f"Binary copy successful: {copy_data == binary_data}")
        
        # Copy text file
        text_copy = os.path.join(temp_dir, 'text_copy.txt')
        path_manager.copy(text_file, text_copy)
        
        # Verify text copy
        with path_manager.open(text_copy, 'r') as f:
            copy_text = f.read()
        
        print(f"Text copy successful: {copy_text == text_content}")
    
    print("Example 2: Binary file operations test passed")
    
except Exception as e:
    print(f"Example 2: Error in binary operations test: {e}")
    import traceback
    traceback.print_exc()
    raise
```

## Example 3: Serialization and Structured Data Handling

Demonstrates PathManager with pickle serialization, JSON files, and data integrity verification.

```python
# Example 3: Serialization and structured data handling
try:
    from iopath.common.file_io import PathManager
    import tempfile
    import pickle
    import json
    import numpy as np
    import os
    
    print("=== Pickle Serialization with PathManager ===")
    path_manager = PathManager()
    
    with tempfile.TemporaryDirectory() as temp_dir:
        pickle_file = os.path.join(temp_dir, 'data.pkl')
        
        # Create complex Python object
        test_data = {
            'numbers': [1, 2, 3, 4, 5],
            'text': 'Hello IOPath!',
            'nested': {
                'array': np.array([1.0, 2.5, 3.7, 4.2]),
                'tuple': (1, 'two', 3.0),
                'dict': {'a': 1, 'b': 2}
            },
            'metadata': {
                'version': 1.0,
                'created_by': 'iopath_test',
                'features': ['feature1', 'feature2', 'feature3']
            }
        }
        
        # Serialize with pickle using PathManager
        with path_manager.open(pickle_file, 'wb') as f:
            pickle.dump(test_data, f)
        
        print(f"Pickle file created: {pickle_file}")
        print(f"File exists: {path_manager.exists(pickle_file)}")
        
        # Deserialize
        with path_manager.open(pickle_file, 'rb') as f:
            loaded_data = pickle.load(f)
        
        print(f"Loaded keys: {list(loaded_data.keys())}") 
        print(f"Numbers match: {loaded_data['numbers'] == test_data['numbers']}")
        print(f"Array match: {np.array_equal(loaded_data['nested']['array'], test_data['nested']['array'])}")
        print(f"Metadata features: {loaded_data['metadata']['features']}")
        
        print("\n=== JSON Configuration Files ===")
        config_file = os.path.join(temp_dir, 'config.json')
        
        config_data = {
            "model": {
                "name": "test_model",
                "version": "1.0.0",
                "parameters": {
                    "learning_rate": 0.001,
                    "batch_size": 32,
                    "epochs": 100,
                    "optimizer": "adam"
                }
            },
            "data": {
                "train_path": "/path/to/train",
                "val_path": "/path/to/val",
                "test_path": "/path/to/test",
                "preprocessing": {
                    "normalize": True,
                    "augment": False,
                    "resize": [224, 224]
                }
            },
            "output": {
                "save_path": "/path/to/models",
                "checkpoint_freq": 10,
                "log_dir": "/path/to/logs"
            }
        }
        
        # Write JSON config using PathManager
        with path_manager.open(config_file, 'w') as f:
            json.dump(config_data, f, indent=2)
        
        print(f"JSON config saved: {config_file}")
        
        # Read and validate JSON
        with path_manager.open(config_file, 'r') as f:
            loaded_config = json.load(f)
        
        print(f"Config sections: {list(loaded_config.keys())}")
        print(f"Learning rate: {loaded_config['model']['parameters']['learning_rate']}")
        print(f"Batch size: {loaded_config['model']['parameters']['batch_size']}")
        print(f"Preprocessing resize: {loaded_config['data']['preprocessing']['resize']}")
        
        print("\n=== Multiple Serialization Formats ===")
        # Create files in different formats
        formats_data = {
            'pickle': (pickle_file, test_data),
            'json': (config_file, config_data)
        }
        
        print("Files created:")
        for fmt, (path, data) in formats_data.items():
            print(f"  {fmt.upper()}: {os.path.basename(path)}")
            print(f"    Exists: {path_manager.exists(path)}")
            print(f"    Is file: {path_manager.isfile(path)}")
        
        print("\n=== Copy Serialized Files ===")
        # Copy pickle file
        pickle_copy = os.path.join(temp_dir, 'backup', 'data_backup.pkl')
        path_manager.mkdirs(os.path.dirname(pickle_copy))
        path_manager.copy(pickle_file, pickle_copy)
        
        # Verify copy works
        with path_manager.open(pickle_copy, 'rb') as f:
            copy_data = pickle.load(f)
        
        print(f"Pickle copy successful: {copy_data['text'] == test_data['text']}")
        
        # Copy JSON file
        json_copy = os.path.join(temp_dir, 'backup', 'config_backup.json')
        path_manager.copy(config_file, json_copy)
        
        with path_manager.open(json_copy, 'r') as f:
            copy_config = json.load(f)
        
        print(f"JSON copy successful: {copy_config['model']['name'] == config_data['model']['name']}")
        
        print("\n=== Roundtrip Data Integrity ===")
        # Test full roundtrip for different data types
        test_objects = [
            ("list", [1, 2, 3, 4, 5]),
            ("dict", {"a": 1, "b": 2, "c": 3}),
            ("numpy_array", np.array([[1, 2], [3, 4]])),
            ("nested", {"list": [1, 2], "dict": {"x": 10}})
        ]
        
        for name, obj in test_objects:
            file_path = os.path.join(temp_dir, f'{name}.pkl')
            
            # Write
            with path_manager.open(file_path, 'wb') as f:
                pickle.dump(obj, f)
            
            # Read
            with path_manager.open(file_path, 'rb') as f:
                loaded_obj = pickle.load(f)
            
            # Verify
            if isinstance(obj, np.ndarray):
                match = np.array_equal(obj, loaded_obj)
            else:
                match = obj == loaded_obj
            
            print(f"{name}: roundtrip successful = {match}")
        
        # List all files
        all_files = path_manager.ls(temp_dir)
        print(f"\nTotal files created: {len(all_files)}")
    
    print("Example 3: Serialization and structured data test passed")
    
except Exception as e:
    print(f"Example 3: Error in serialization test: {e}")
    import traceback
    traceback.print_exc()
    raise
```

## Example 4: Custom PathHandler Implementation and Registration

Demonstrates creating a custom PathHandler class and registering it with PathManager.

```python
# Example 4: Custom PathHandler implementation and registration
try:
    from iopath.common.file_io import PathManager, PathHandler
    import tempfile
    import os
    import time
    
    print("=== Custom Path Handler Implementation ===")
    
    # Create a custom handler for memory-based storage (unique to iopath)
    class MemoryPathHandler(PathHandler):
        """Custom path handler that stores files in memory"""
        
        def __init__(self):
            self._storage = {}  # In-memory storage
            self._metadata = {}  # File metadata
        
        def _get_supported_prefixes(self):
            """Define supported URI prefixes"""
            return ["memory://"]
        
        def _get_local_path(self, path, **kwargs):
            # Memory handler doesn't have local paths
            return None
        
        def _open(self, path, mode="r", **kwargs):
            # Simplified implementation for demonstration
            import io
            if path in self._storage:
                data = self._storage[path]
                if 'b' in mode:
                    return io.BytesIO(data)
                else:
                    return io.StringIO(data.decode('utf-8'))
            else:
                # Create new file
                if 'w' in mode or 'a' in mode:
                    if 'b' in mode:
                        buffer = io.BytesIO()
                    else:
                        buffer = io.StringIO()
                    
                    # Custom buffer that saves to memory on close
                    class MemoryFile:
                        def __init__(self, path, buffer, storage, metadata):
                            self.path = path
                            self.buffer = buffer
                            self.storage = storage
                            self.metadata = metadata
                        
                        def write(self, data):
                            return self.buffer.write(data)
                        
                        def read(self, size=-1):
                            return self.buffer.read(size)
                        
                        def close(self):
                            content = self.buffer.getvalue()
                            if isinstance(content, str):
                                content = content.encode('utf-8')
                            self.storage[self.path] = content
                            self.metadata[self.path] = {
                                'size': len(content),
                                'created': time.time()
                            }
                            self.buffer.close()
                        
                        def __enter__(self):
                            return self
                        
                        def __exit__(self, *args):
                            self.close()
                    
                    return MemoryFile(path, buffer, self._storage, self._metadata)
                else:
                    raise FileNotFoundError(f"Memory file {path} not found")
        
        def _copy(self, src_path, dst_path, overwrite=False, **kwargs):
            """Copy with overwrite parameter"""
            if src_path in self._storage:
                if not overwrite and dst_path in self._storage:
                    raise FileExistsError(f"Destination {dst_path} already exists")
                self._storage[dst_path] = self._storage[src_path]
                self._metadata[dst_path] = self._metadata[src_path].copy()
                return True
            return False
        
        def _exists(self, path, **kwargs):
            return path in self._storage
        
        def _isfile(self, path, **kwargs):
            return path in self._storage
        
        def _isdir(self, path, **kwargs):
            # Simplified: treat directories as paths with trailing /
            return any(stored_path.startswith(path + '/') for stored_path in self._storage)
        
        def _ls(self, path, **kwargs):
            # List files in memory storage
            if path.endswith('/'):
                prefix = path
            else:
                prefix = path + '/'
            
            return [p for p in self._storage.keys() if p.startswith(prefix)]
        
        def _mkdirs(self, path, **kwargs):
            # Memory doesn't need directory creation
            pass
        
        def _rm(self, path, **kwargs):
            if path in self._storage:
                del self._storage[path]
                del self._metadata[path]
    
    print("Custom MemoryPathHandler created")
    
    print("\n=== Registering Custom Handler ===")
    # Create PathManager and register custom handler (unique to iopath)
    path_manager = PathManager()
    memory_handler = MemoryPathHandler()
    
    path_manager.register_handler(memory_handler)
    
    # Verify registration
    supported_prefixes = memory_handler._get_supported_prefixes()
    print(f"Custom handler supports: {supported_prefixes}")
    print(f"Handler registered successfully")
    
    print("\n=== Testing Custom Handler ===")
    # Test memory-based file operations
    memory_file1 = "memory://test1.txt"
    memory_file2 = "memory://test2.txt"
    
    # Write to memory storage
    with path_manager.open(memory_file1, 'w') as f:
        f.write("Hello from memory storage!")
    
    print(f"File written to memory: {memory_file1}")
    print(f"File exists: {path_manager.exists(memory_file1)}")
    print(f"Is file: {path_manager.isfile(memory_file1)}")
    
    # Read from memory storage
    with path_manager.open(memory_file1, 'r') as f:
        content = f.read()
    
    print(f"Content read: '{content}'")
    
    # Copy within memory storage
    path_manager.copy(memory_file1, memory_file2)
    print(f"File copied to: {memory_file2}")
    print(f"Copy exists: {path_manager.exists(memory_file2)}")
    
    # Verify copy content
    with path_manager.open(memory_file2, 'r') as f:
        copy_content = f.read()
    
    print(f"Copy content matches: {content == copy_content}")
    
    # List memory storage contents
    memory_files = memory_handler._ls("memory://")
    print(f"Memory storage contents: {memory_files}")
    
    print("\n=== Handler Comparison ===")
    # Compare different handlers
    with tempfile.TemporaryDirectory() as temp_dir:
        local_file = os.path.join(temp_dir, 'local_test.txt')
        
        # Create local file
        with path_manager.open(local_file, 'w') as f:
            f.write("Hello from local storage!")
        
        # Show unified interface
        test_files = [
            ("local", local_file),
            ("memory", memory_file1)
        ]
        
        print("Unified interface demonstration:")
        for storage_type, file_path in test_files:
            print(f"\n{storage_type.upper()} storage:")
            print(f"  Path: {file_path}")
            print(f"  Exists: {path_manager.exists(file_path)}")
            print(f"  Is file: {path_manager.isfile(file_path)}")
            
            with path_manager.open(file_path, 'r') as f:
                content = f.read()
            
            print(f"  Content: '{content}' ({len(content)} chars)")
    
    print("\n=== Multiple Memory Files ===")
    # Create multiple files in memory
    for i in range(5):
        mem_path = f"memory://file_{i}.txt"
        with path_manager.open(mem_path, 'w') as f:
            f.write(f"Content for memory file {i}")
    
    # List all memory files
    all_memory_files = memory_handler._ls("memory://")
    print(f"Total memory files: {len(all_memory_files)}")
    print(f"Files: {all_memory_files[:5]}")
    
    print("Example 4: Custom PathHandler implementation test passed")
    
except Exception as e:
    print(f"Example 4: Error in custom handler test: {e}")
    import traceback
    traceback.print_exc()
    raise
```

## Example 5: File operations, config round-trips and error handling

Write, read, copy and list files through `PathManager`, save and reload JSON configs, and check the errors raised for missing files.

```python
# Example 5: File operations, config round-trips and error handling
import json
import os
import tempfile

from iopath.common.file_io import PathManager

pm = PathManager()

print("=== Write, read and copy ===")
with tempfile.TemporaryDirectory() as temp_dir:
    payload = "Test content for PathManager operations"
    test_file = os.path.join(temp_dir, "test.txt")
    with pm.open(test_file, "w") as f:
        f.write(payload)
    with pm.open(test_file, "r") as f:
        assert f.read() == payload

    # copy into a directory that does not exist yet
    copy_path = os.path.join(temp_dir, "nested", "copied_file.txt")
    pm.mkdirs(os.path.dirname(copy_path))
    assert pm.copy(test_file, copy_path)
    assert pm.exists(copy_path) and pm.isfile(copy_path)
    with pm.open(copy_path, "r") as f:
        assert f.read() == payload
    print(f"Copied {test_file} -> {copy_path}")

    print("\n=== Errors for missing files ===")
    missing = os.path.join(temp_dir, "missing.txt")
    assert not pm.exists(missing)
    try:
        pm.open(missing, "r")
    except FileNotFoundError as e:
        print(f"open() on a missing file raises {type(e).__name__}")
    else:
        raise AssertionError("opening a missing file did not raise")

print("\n=== JSON config round-trip ===")
with tempfile.TemporaryDirectory() as temp_dir:
    config_dir = os.path.join(temp_dir, "configs")
    pm.mkdirs(config_dir)
    configs = {
        "database": {"host": "localhost", "port": 5432, "name": "example_db"},
        "api": {"base_url": "https://api.example.com", "timeout": 30, "retries": 3},
        "logging": {"level": "INFO", "format": "%(asctime)s - %(levelname)s - %(message)s"},
    }
    for name, saved_config in configs.items():
        with pm.open(os.path.join(config_dir, f"{name}.json"), "w") as f:
            json.dump(saved_config, f, indent=2)

    available = sorted(n[: -len(".json")] for n in pm.ls(config_dir) if n.endswith(".json"))
    print(f"Available configs: {available}")
    assert available == sorted(configs)

    for name, saved_config in configs.items():
        with pm.open(os.path.join(config_dir, f"{name}.json"), "r") as f:
            loaded = json.load(f)
        assert loaded == saved_config, (name, loaded)

print("\n=== Batch of files ===")
with tempfile.TemporaryDirectory() as temp_dir:
    files_created = []
    for i in range(20):
        file_path = os.path.join(temp_dir, f"batch_file_{i:03d}.txt")
        with pm.open(file_path, "w") as f:
            f.write(f"Batch file {i} with content")
        files_created.append(file_path)

    assert all(pm.exists(f) for f in files_created)
    listing = sorted(pm.ls(temp_dir))
    assert listing == [os.path.basename(f) for f in files_created]
    print(f"Created and listed {len(listing)} files")
```
