## Example 1: Creating maps with ipyleaflet.Map()

```python
import ipyleaflet

# Example 1: Creating maps with ipyleaflet.Map()
print("Example 1: Creating interactive maps")

# Create maps with different configurations
map_configs = [
    {"center": [40.7128, -74.0060], "zoom": 10, "name": "New York"},
    {"center": [51.5074, -0.1278], "zoom": 12, "name": "London"},
    {"center": [35.6762, 139.6503], "zoom": 11, "name": "Tokyo"},
]

print("Creating maps with different centers and zoom levels:")
maps = []

for config in map_configs:
    try:
        # Create map using ipyleaflet.Map()
        m = ipyleaflet.Map(center=config["center"], zoom=config["zoom"])
        maps.append(m)
        
        print(f"  {config['name']}:")
        print(f"    Center: {m.center}")
        print(f"    Zoom level: {m.zoom}")
        print(f"    CRS: {m.crs['name']}")
        print(f"    Default layers: {len(m.layers)}")
        
    except Exception as e:
        print(f"  {config['name']}: Error - {e}")
        raise

print(f"\nSuccessfully created {len(maps)} interactive maps")
for config, m in zip(map_configs, maps):
    assert list(m.center) == config["center"] and m.zoom == config["zoom"]
    # Every map uses Web Mercator and starts with one base tile layer
    assert m.crs["name"] == "EPSG3857"
    assert len(m.layers) == 1 and isinstance(m.layers[0], ipyleaflet.TileLayer) and m.layers[0].base
# Traits are validated: a zoom level must be a number
try:
    maps[0].zoom = "far"
    raise AssertionError("expected TraitError")
except Exception as e:
    assert type(e).__name__ == "TraitError"
```

## Example 2: Adding markers with ipyleaflet.Marker()

```python
import ipyleaflet

# Example 2: Adding markers with ipyleaflet.Marker()
print("Example 2: Adding markers to maps")

# Create a map
map_center = [40.7128, -74.0060]
m = ipyleaflet.Map(center=map_center, zoom=12)

# Define marker locations with descriptions
marker_locations = [
    {"location": [40.7589, -73.9851], "name": "Times Square"},
    {"location": [40.7128, -74.0060], "name": "World Trade Center"},
    {"location": [40.7580, -73.9855], "name": "Broadway Theater"},
    {"location": [40.7614, -73.9776], "name": "Central Park"}
]

print("Creating markers:")
markers = []

for marker_data in marker_locations:
    try:
        # Create marker using ipyleaflet.Marker()
        marker = ipyleaflet.Marker(location=marker_data["location"])
        markers.append(marker)
        
        print(f"  {marker_data['name']}:")
        print(f"    Location: {marker.location}")
        print(f"    Draggable: {marker.draggable}")
        print(f"    Z-index offset: {marker.z_index_offset}")
        
        # Add marker to map
        m.add_layer(marker)
        
    except Exception as e:
        print(f"  {marker_data['name']}: Error - {e}")
        raise

print(f"\nAdded {len(markers)} markers to the map")
print(f"Map now has {len(m.layers)} total layers")
assert len(m.layers) == 1 + len(markers)
assert all(marker in m.layers for marker in markers)
assert list(markers[0].location) == [40.7589, -73.9851]
assert markers[0].draggable is True and markers[0].z_index_offset == 0
# Removing a layer takes it off the map again
m.remove(markers[-1])
assert len(m.layers) == len(markers) and markers[-1] not in m.layers
```

## Example 3: Drawing circles with ipyleaflet.Circle()

```python
import ipyleaflet

# Example 3: Drawing circles with ipyleaflet.Circle()
print("Example 3: Creating circular overlays")

# Create a map centered on San Francisco
m = ipyleaflet.Map(center=[37.7749, -122.4194], zoom=11)

# Define circles with different radii and locations
circle_data = [
    {"location": [37.7749, -122.4194], "radius": 2000, "name": "Downtown SF"},
    {"location": [37.8044, -122.2712], "radius": 1500, "name": "Berkeley"},
    {"location": [37.7849, -122.4094], "radius": 1000, "name": "Mission District"},
    {"location": [37.7849, -122.4594], "radius": 800, "name": "Golden Gate Park"}
]

print("Creating circles:")
circles = []

for circle_info in circle_data:
    try:
        # Create circle using ipyleaflet.Circle()
        circle = ipyleaflet.Circle(
            location=circle_info["location"], 
            radius=circle_info["radius"],
            color="blue",
            fill_opacity=0.3
        )
        circles.append(circle)
        
        print(f"  {circle_info['name']}:")
        print(f"    Center: {circle.location}")
        print(f"    Radius: {circle.radius} meters")
        print(f"    Color: {circle.color}")
        print(f"    Fill opacity: {circle.fill_opacity}")
        
        # Add circle to map
        m.add_layer(circle)
        
    except Exception as e:
        print(f"  {circle_info['name']}: Error - {e}")
        raise

print(f"\nAdded {len(circles)} circles to the map")
assert [c.radius for c in circles] == [2000, 1500, 1000, 800]
assert all(c.color == "blue" and c.fill_opacity == 0.3 for c in circles)
assert len(m.layers) == 1 + len(circles)
```

## Example 4: Creating polygons with ipyleaflet.Polygon()

```python
import ipyleaflet

# Example 4: Creating polygons with ipyleaflet.Polygon()
print("Example 4: Creating polygon shapes")

# Create a map centered on Washington DC
m = ipyleaflet.Map(center=[38.9072, -77.0369], zoom=10)

# Define polygon coordinates for different areas
polygon_data = [
    {
        "name": "National Mall",
        "coordinates": [
            [38.8978, -77.0365],
            [38.8978, -77.0220],
            [38.8814, -77.0220], 
            [38.8814, -77.0365]
        ]
    },
    {
        "name": "Pentagon Area", 
        "coordinates": [
            [38.8710, -77.0600],
            [38.8710, -77.0500],
            [38.8650, -77.0500],
            [38.8650, -77.0600]
        ]
    }
]

print("Creating polygons:")
polygons = []

for poly_info in polygon_data:
    try:
        # Create polygon using ipyleaflet.Polygon()
        polygon = ipyleaflet.Polygon(
            locations=poly_info["coordinates"],
            color="red",
            fill_color="yellow",
            fill_opacity=0.4
        )
        polygons.append(polygon)
        
        print(f"  {poly_info['name']}:")
        print(f"    Vertices: {len(polygon.locations)} points")
        print(f"    Color: {polygon.color}")
        print(f"    Fill color: {polygon.fill_color}")
        print(f"    Fill opacity: {polygon.fill_opacity}")
        
        # Add polygon to map
        m.add_layer(polygon)
        
    except Exception as e:
        print(f"  {poly_info['name']}: Error - {e}")
        raise

print(f"\nAdded {len(polygons)} polygons to the map")
assert [len(p.locations) for p in polygons] == [4, 4]
assert polygons[0].locations[0] == [38.8978, -77.0365]
assert all(p.color == "red" and p.fill_color == "yellow" and p.fill_opacity == 0.4 for p in polygons)
assert len(m.layers) == 1 + len(polygons)
```

## Example 5: Custom tile layers with ipyleaflet.TileLayer()

```python
import ipyleaflet

# Example 5: Custom tile layers with ipyleaflet.TileLayer()
print("Example 5: Adding custom tile layers")

# Create a base map
m = ipyleaflet.Map(center=[48.8566, 2.3522], zoom=10)  # Paris

# Define different tile layer sources
tile_layers = [
    {
        "name": "OpenStreetMap",
        "url": "https://tile.openstreetmap.org/{z}/{x}/{y}.png",
        "attribution": "© OpenStreetMap contributors"
    },
    {
        "name": "CartoDB Positron",
        "url": "https://cartodb-basemaps-{s}.global.ssl.fastly.net/light_all/{z}/{x}/{y}.png",
        "attribution": "© CartoDB"
    },
    {
        "name": "Stamen Terrain",
        "url": "https://stamen-tiles.a.ssl.fastly.net/terrain/{z}/{x}/{y}.png",
        "attribution": "© Stamen Design"
    }
]

print("Creating custom tile layers:")
custom_layers = []

for layer_info in tile_layers:
    try:
        # Create tile layer using ipyleaflet.TileLayer()
        tile_layer = ipyleaflet.TileLayer(
            url=layer_info["url"],
            attribution=layer_info["attribution"],
            name=layer_info["name"]
        )
        custom_layers.append(tile_layer)
        
        print(f"  {layer_info['name']}:")
        print(f"    URL: {tile_layer.url[:50]}...")
        print(f"    Attribution: {tile_layer.attribution[:30]}...")
        print(f"    Name: {tile_layer.name}")
        print(f"    Max zoom: {tile_layer.max_zoom}")
        
    except Exception as e:
        print(f"  {layer_info['name']}: Error - {e}")
        raise

# Add first custom layer to map
if custom_layers:
    m.add_layer(custom_layers[0])
    print(f"\nAdded {custom_layers[0].name} as base layer")

print(f"Created {len(custom_layers)} custom tile layers")
assert [layer.name for layer in custom_layers] == ["OpenStreetMap", "CartoDB Positron", "Stamen Terrain"]
assert custom_layers[0].url == "https://tile.openstreetmap.org/{z}/{x}/{y}.png"
assert custom_layers[0].max_zoom == 18
# The added layer sits on top of the default base layer
assert len(m.layers) == 2 and m.layers[-1] is custom_layers[0]
```
