## Example 1: IPython display system with HTML, Markdown, and JSON

```python
from IPython.display import display, HTML, Markdown, JSON
import IPython

# Example 1: IPython display system with HTML, Markdown, and JSON
print("Example 1: IPython display capabilities")

# Display HTML content
html_content = "<h3>IPython HTML Display</h3><p>This is <em>formatted</em> HTML content.</p>"
print("Displaying HTML:")
display(HTML(html_content))

# Display Markdown content
markdown_content = """
## IPython Markdown
- **Bold text**
- *Italic text*  
- `Code snippet`
"""
print("Displaying Markdown:")
display(Markdown(markdown_content))

# Display JSON data
json_data = {"name": "IPython", "version": IPython.__version__, "features": ["display", "magic", "kernel"]}
print("Displaying JSON:")
display(JSON(json_data))

from IPython.core.formatters import DisplayFormatter

# Each display object hands its content to the matching MIME-type repr
assert HTML(html_content)._repr_html_() == html_content
assert Markdown(markdown_content)._repr_markdown_() == markdown_content
assert JSON(json_data).data == json_data
# The display formatter turns an object into a MIME bundle: text/plain plus the rich format
bundle, metadata = DisplayFormatter().format(HTML(html_content))
assert bundle["text/html"] == html_content and "text/plain" in bundle
```

## Example 2: IPython magic system and get_ipython()

```python
from IPython.core.magic import Magics, magics_class, line_magic, cell_magic
from IPython import get_ipython
import IPython
import time

# Example 2: IPython magic system and get_ipython()
print("Example 2: IPython magic system")

# Get the current IPython instance
ip = get_ipython()
print(f"IPython instance: {type(ip).__name__}")
print(f"IPython version: {IPython.__version__}")

# Create a custom magic class
@magics_class
class CustomMagics(Magics):
    
    @line_magic
    def greet(self, line):
        """Simple greeting magic"""
        return f"Hello, {line or 'World'}!"

# Register the custom magic
ip.register_magic_function(CustomMagics(ip).greet)

# Test the custom magic
result = ip.run_line_magic('greet', 'IPython')
print(f"Custom magic result: {result}")

# List available magic commands through magics_manager
magic_names = sorted(ip.magics_manager.magics['line'].keys())
print(f"Available line magics (first 5): {magic_names[:5]}")

assert result == "Hello, IPython!"
assert ip.run_line_magic('greet', '') == "Hello, World!"
assert 'greet' in magic_names and 'time' in magic_names
```

## Example 3: IPython InteractiveShell functionality

```python
from IPython.core.interactiveshell import InteractiveShell
from IPython import get_ipython

# Example 3: IPython InteractiveShell functionality
print("Example 3: IPython InteractiveShell features")

# Get current shell instance
shell = get_ipython()
print(f"Shell type: {type(shell).__name__}")

# Run code and capture result
code = "x = 10; y = 20; result = x + y"
shell.run_cell(code)
# The cell ran in the shell's user namespace (this notebook's own namespace in a kernel)
computed = shell.user_ns.get('result')
print(f"Executed: {code}")
print(f"Result in namespace: {computed}")

# Test code execution with different result types
test_expressions = ["2 + 2", "len('hello')", "[i**2 for i in range(3)]"]

print("Expression results:")
for expr in test_expressions:
    result = shell.run_cell(expr)
    print(f"  {expr} = {result.result}")

# Show shell configuration
print(f"Shell display formatter: {type(shell.display_formatter).__name__}")
print(f"History manager: {type(shell.history_manager).__name__}")

assert isinstance(shell, InteractiveShell)
assert computed == 30 and shell.user_ns['x'] == 10
# run_cell returns an ExecutionResult holding the value of the last expression
results = {expr: shell.run_cell(expr) for expr in test_expressions}
assert [results[e].result for e in test_expressions] == [4, 5, [0, 1, 4]]
assert all(r.success for r in results.values())
```

## Example 4: IPython embedding and kernel functionality

```python
from IPython.terminal.embed import InteractiveShellEmbed
from IPython import embed_kernel
import IPython

# Example 4: IPython embedding and kernel functionality
print("Example 4: IPython embedding capabilities")

# Create an embedded shell instance  
embedded_shell = InteractiveShellEmbed()
print(f"Embedded shell type: {type(embedded_shell).__name__}")
print(f"Shell banner: {embedded_shell.banner1[:50]}...")

# Test shell configuration
config = embedded_shell.config
print(f"Shell config class: {type(config).__name__}")

# Check available extensions and magic functions through magics_manager
builtin_magics = list(embedded_shell.magics_manager.magics['line'].keys())
print(f"Built-in magics count: {len(builtin_magics)}")
print(f"Sample magics: {sorted(builtin_magics)[:5]}")

# Show embedding features
print(f"IPython version info: {IPython.version_info}")
print(f"Extract module locals available: {hasattr(IPython, 'extract_module_locals')}")

# Test namespace functionality
test_vars = {"demo_var": "embedding test", "demo_num": 42}
embedded_shell.push(test_vars)
print(f"Variables pushed to shell namespace")

from IPython.core.interactiveshell import InteractiveShell

assert isinstance(embedded_shell, InteractiveShell)
# push() puts variables into the embedded shell's namespace
assert embedded_shell.user_ns['demo_var'] == "embedding test" and embedded_shell.user_ns['demo_num'] == 42
assert 'time' in builtin_magics
assert IPython.version_info[0] >= 8
```

## Example 5: IPython lib utilities - pretty printing and capture

```python
from IPython.lib import pretty
from IPython.lib.deepreload import reload as deep_reload
from IPython.utils import capture
import IPython.lib

# Example 5: IPython lib utilities - pretty printing and capture
print("Example 5: IPython lib utilities")

# Test pretty printing functionality
data_structure = {"users": [{"name": "Alice", "age": 30}, {"name": "Bob", "age": 25}], "total": 2}
print("Pretty printed data structure:")
pretty_output = pretty.pretty(data_structure)
print(pretty_output)

# Test output capture functionality
print("Testing output capture:")
with capture.capture_output() as captured:
    print("This output is captured")
    print("Second line of captured output")
    
print(f"Captured stdout: {captured.stdout}")
print(f"Captured stderr length: {len(captured.stderr)}")

# Show lib module structure
lib_modules = [attr for attr in dir(IPython.lib) if not attr.startswith('_')]
print(f"IPython.lib modules count: {len(lib_modules)}")
print(f"Sample lib modules: {lib_modules[:5]}")

# Test pretty printing with different types
test_items = [42, [1, 2, 3], {"key": "value"}]
for item in test_items:
    print(f"Pretty: {pretty.pretty(item)}")

assert captured.stdout == "This output is captured\nSecond line of captured output\n"
assert captured.stderr == ""
assert [pretty.pretty(item) for item in test_items] == ['42', '[1, 2, 3]', "{'key': 'value'}"]
# pretty() wraps long structures at max_width
assert pretty_output == "{'users': [{'name': 'Alice', 'age': 30}, {'name': 'Bob', 'age': 25}],\n 'total': 2}"
assert max(len(line) for line in pretty.pretty(data_structure, max_width=30).splitlines()) <= 30
```
