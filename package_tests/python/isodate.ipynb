## Example 1: isodate datetime parsing functionality test 1

```python
import isodate
from datetime import datetime

# Example 1: Parsing ISO 8601 datetime strings with parse_datetime()
print("Example 1: ISO 8601 datetime parsing")

# Test different ISO datetime formats
iso_datetime_strings = [
    "2023-12-25T15:30:45",
    "2023-12-25T15:30:45.123456",
    "2023-12-25T15:30:45Z",
    "2023-12-25T15:30:45+02:00",
    "2023-12-25T15:30:45-05:00"
]

print("Parsing ISO datetime strings:")
for iso_string in iso_datetime_strings:
    try:
        parsed = isodate.parse_datetime(iso_string)
        print(f"  '{iso_string}' -> {parsed}")
        print(f"    Type: {type(parsed).__name__}")
        print(f"    Timezone: {parsed.tzinfo}")
    except Exception as e:
        print(f"  '{iso_string}' -> Error: {e}")
        raise

print("ISO datetime parsing completed successfully")
```

## Example 2: isodate duration parsing functionality test 2

```python
import isodate
from datetime import timedelta

# Example 2: Parsing ISO 8601 duration strings with parse_duration()
print("Example 2: ISO 8601 duration parsing")

# Test different ISO duration formats
iso_duration_strings = [
    "PT30M",           # 30 minutes
    "PT1H30M45S",      # 1 hour, 30 minutes, 45 seconds
    "P1DT2H",          # 1 day, 2 hours
    "P1W",             # 1 week
    "P1Y2M3DT4H5M6S",  # 1 year, 2 months, 3 days, 4 hours, 5 minutes, 6 seconds
    "PT0.5S"           # 0.5 seconds
]

print("Parsing ISO duration strings:")
for duration_string in iso_duration_strings:
    try:
        parsed = isodate.parse_duration(duration_string)
        print(f"  '{duration_string}' -> {parsed}")
        print(f"    Type: {type(parsed).__name__}")
        print(f"    Total seconds: {parsed.total_seconds()}")
    except Exception as e:
        print(f"  '{duration_string}' -> Error: {e}")
        raise

print("ISO duration parsing completed successfully")
```

## Example 3: isodate date parsing functionality test 3

```python
import isodate
from datetime import date

# Example 3: Parsing ISO 8601 date strings with parse_date()
print("Example 3: ISO 8601 date parsing")

# Test different ISO date formats
iso_date_strings = [
    "2023-12-25",      # Basic date
    "20231225",        # Compact date
    "2023-359",        # Ordinal date (day of year)
    "2023359",         # Compact ordinal date
    "2023-W52-1",      # Week date (year-week-day)
    "2023W521"         # Compact week date
]

print("Parsing ISO date strings:")
for date_string in iso_date_strings:
    try:
        parsed = isodate.parse_date(date_string)
        print(f"  '{date_string}' -> {parsed}")
        print(f"    Type: {type(parsed).__name__}")
        print(f"    Weekday: {parsed.strftime('%A')}")
    except Exception as e:
        print(f"  '{date_string}' -> Error: {e}")
        raise

print("ISO date parsing completed successfully")
```

## Example 4: isodate formatting functionality test 4

```python
import isodate
from datetime import datetime, timedelta, timezone

# Example 4: Formatting datetime and duration to ISO 8601 with isodate functions
print("Example 4: ISO 8601 formatting")

# Create datetime objects to format
dt_naive = datetime(2023, 12, 25, 15, 30, 45, 123456)
dt_utc = datetime(2023, 12, 25, 15, 30, 45, tzinfo=timezone.utc)
dt_offset = datetime(2023, 12, 25, 15, 30, 45, tzinfo=timezone(timedelta(hours=2)))

print("Formatting datetimes to ISO strings:")
datetimes = [
    ("Naive datetime", dt_naive),
    ("UTC datetime", dt_utc),
    ("Offset datetime", dt_offset)
]

for name, dt in datetimes:
    formatted = isodate.datetime_isoformat(dt)
    print(f"  {name}: {formatted}")

# Format durations to ISO strings
durations = [
    timedelta(hours=1, minutes=30, seconds=45),
    timedelta(days=1, hours=2),
    timedelta(weeks=1),
    timedelta(seconds=30.5)
]

print("Formatting durations to ISO strings:")
for duration in durations:
    formatted = isodate.duration_isoformat(duration)
    print(f"  {duration} -> {formatted}")

print("ISO formatting completed successfully")
```

## Example 5: isodate time parsing functionality test 5

```python
import isodate
from datetime import time

# Example 5: Parsing ISO 8601 time strings with parse_time()
print("Example 5: ISO 8601 time parsing")

# Test different ISO time formats
iso_time_strings = [
    "15:30:45",         # Basic time
    "153045",           # Compact time
    "15:30:45.123456",  # Time with microseconds
    "15:30:45Z",        # Time with UTC timezone
    "15:30:45+02:00",   # Time with timezone offset
    "15:30:45-05:30",   # Time with negative offset
]

print("Parsing ISO time strings:")
for time_string in iso_time_strings:
    try:
        parsed = isodate.parse_time(time_string)
        print(f"  '{time_string}' -> {parsed}")
        print(f"    Type: {type(parsed).__name__}")
        print(f"    Hour: {parsed.hour}, Minute: {parsed.minute}, Second: {parsed.second}")
        if parsed.tzinfo:
            print(f"    Timezone: {parsed.tzinfo}")
    except Exception as e:
        print(f"  '{time_string}' -> Error: {e}")
        raise

print("ISO time parsing completed successfully")
```
