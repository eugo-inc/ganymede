## Example 1: isoduration functionality test 1

```python
import isoduration
from isoduration.types import DateDuration, Duration, TimeDuration
from datetime import timedelta

# Example 1: ISO 8601 duration parsing and formatting
print("Example 1: ISO 8601 duration parsing and formatting")

try:
    # Parse various ISO 8601 duration strings
    durations = [
        'PT30M',        # 30 minutes
        'PT1H30M',      # 1 hour 30 minutes
        'P1D',          # 1 day
        'P1DT12H',      # 1 day 12 hours
        'PT45S',        # 45 seconds
        'P1Y2M3DT4H5M6S'  # Complex duration
    ]
    
    print("Parsing ISO 8601 durations:")
    for duration_str in durations:
        try:
            parsed = isoduration.parse_duration(duration_str)
            print(f"  {duration_str:15} -> {parsed} ({type(parsed).__name__})")
        except Exception as e:
            print(f"  {duration_str:15} -> Error: {e}")
            raise
    
    # Test duration formatting
    # format_duration() takes an isoduration Duration, not a datetime.timedelta: build one from the
    # timedelta's components.
    td = timedelta(hours=2, minutes=30, seconds=45)
    hours, remainder = divmod(int(td.total_seconds()), 3600)
    minutes, seconds = divmod(remainder, 60)
    duration = Duration(DateDuration(), TimeDuration(hours=hours, minutes=minutes, seconds=seconds))
    iso_string = isoduration.format_duration(duration)
    assert iso_string == 'PT2H30M45S', iso_string
    print(f"\nTimedelta to ISO: {td} -> {iso_string}")
    
    # Round trip test
    original = 'PT2H30M45S'
    parsed_back = isoduration.parse_duration(original)
    print(f"Round trip: {original} -> {parsed_back}")
    
    print("isoduration example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"isoduration example (cell 2) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```
