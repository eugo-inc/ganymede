## Example 1: isort functionality test 1

```python
import isort

# Example 1: Basic import sorting with isort.code()
print("Example 1: Basic import sorting")

# Sample Python code with unsorted imports
unsorted_code = """import os
from collections import defaultdict
import sys
from typing import Dict, List
import json
from pathlib import Path
import re
from datetime import datetime
"""

print("Original unsorted imports:")
print(unsorted_code)

# Sort the imports using isort.code()
sorted_code = isort.code(unsorted_code)

print("After sorting with isort:")
print(sorted_code)

# Check if code needs sorting with isort.check_code()
needs_sorting = not isort.check_code(unsorted_code)
print(f"Code needed sorting: {needs_sorting}")

print("Basic import sorting completed successfully")
```

## Example 2: isort configuration functionality test 2

```python
import isort

# Example 2: isort Config class and configuration options
print("Example 2: isort configuration")

# Create a configuration object
config = isort.Config()
print(f"Default config type: {type(config).__name__}")

# Test with different profiles
profiles = ['black', 'django', 'google']
test_code = "import sys\nfrom collections import defaultdict\nimport os"

print("Testing different profiles:")
for profile in profiles:
    try:
        config_prof = isort.Config(profile=profile)
        sorted_prof = isort.code(test_code, config=config_prof)
        lines = len([line for line in sorted_prof.split('\n') if line.strip()])
        print(f"  {profile}: {lines} import lines")
    except Exception as e:
        print(f"  {profile}: Error - {e}")
        raise

# Test configuration options
custom_config = isort.Config(
    line_length=80,
    multi_line_output=3,
    include_trailing_comma=True
)
custom_sorted = isort.code(test_code, config=custom_config)
print(f"Custom config result: {len(custom_sorted)} chars")

print("Configuration testing completed successfully")
```

## Example 3: isort module placement functionality test 3

```python
import isort

# Example 3: isort module placement with place_module()
print("Example 3: Module placement detection")

# Test module placement for different types of modules
test_modules = [
    "os",           # Standard library
    "sys",          # Standard library  
    "json",         # Standard library
    "requests",     # Third party
    "numpy",        # Third party
    "django",       # Third party
    "myproject",    # Local project
]

print("Module placement analysis:")
for module in test_modules:
    try:
        placement = isort.place_module(module)
        print(f"  {module}: {placement}")
    except Exception as e:
        print(f"  {module}: Error - {e}")
        raise

# Test with a config that has custom sections
config = isort.Config(known_third_party=["requests", "numpy"])
print(f"\nWith custom config (known_third_party):")
# place_module(name, config=...): the module name comes first, the config by keyword
# (unknown_package lands in config.default_section, which is THIRDPARTY)
for module in ["requests", "numpy", "unknown_package"]:
    try:
        placement = isort.place_module(module, config=config)
        print(f"  {module}: {placement}")
        assert placement == "THIRDPARTY", (module, placement)
    except Exception as e:
        print(f"  {module}: Error - {e}")
        raise

print("Module placement completed successfully")
```

## Example 4: isort file checking functionality test 4

```python
import isort
import tempfile
import os

# Example 4: isort file operations with check_file() and file()
print("Example 4: File-based import sorting")

# Create a temporary file with unsorted imports
unsorted_content = """import sys
from collections import defaultdict
import os
from typing import List
import json
"""

# Create temporary file
with tempfile.NamedTemporaryFile(mode='w', suffix='.py', delete=False) as tmp:
    tmp.write(unsorted_content)
    tmp_path = tmp.name

print(f"Created temporary file: {os.path.basename(tmp_path)}")

# Check if file needs sorting with check_file()
needs_sorting = not isort.check_file(tmp_path)
print(f"File needs sorting: {needs_sorting}")

# Sort the file with file() function
result = isort.file(tmp_path)
print(f"File sorted: {result}")

# Read the sorted content
with open(tmp_path, 'r') as f:
    sorted_content = f.read()

print("Sorted file content:")
print(sorted_content[:100] + "..." if len(sorted_content) > 100 else sorted_content)

# Clean up
os.unlink(tmp_path)
print("File-based sorting completed successfully")
```

## Example 5: isort API functionality test 5

```python
import isort

# Example 5: isort API module and sorting with different options
print("Example 5: Advanced isort API usage")

# Test the isort.api module for more control
from isort import api

# Sample code with mixed imports and comments
code_with_comments = """# Important imports
import sys  # System module
from collections import defaultdict
import os
# Third party imports
import requests
from typing import List, Dict
"""

print("Original code with comments:")
print(code_with_comments)

# Sort while preserving comments structure
sorted_with_comments = isort.code(code_with_comments)
print("Sorted code (comments preserved):")
print(sorted_with_comments)

# Test different sorting options
# (options, expected output); the setting is `combine_as_imports` -- `combine_as` is rejected
# with isort.exceptions.UnsupportedSettings
configs = [
    ({"multi_line_output": 3, "line_length": 60}, "import os\nimport sys\nfrom typing import Dict, List\n"),
    ({"force_single_line": True}, "import os\nimport sys\nfrom typing import Dict\nfrom typing import List\n"),
    ({"combine_as_imports": True, "combine_star": True}, "import os\nimport sys\nfrom typing import Dict, List\n"),
]

simple_code = "from typing import List, Dict\nimport os, sys"
print("Testing different formatting options:")

for i, (config_opts, expected) in enumerate(configs, 1):
    try:
        config = isort.Config(**config_opts)
        result = isort.code(simple_code, config=config)
        print(f"  Config {i}: {len(result.split())} tokens")
        assert result == expected, (config_opts, result)
    except Exception as e:
        print(f"  Config {i}: Error - {e}")
        raise

print("Advanced API usage completed successfully")
```
