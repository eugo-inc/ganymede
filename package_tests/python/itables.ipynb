## Example 1: itables show functionality test 1

```python
import itables
import pandas as pd

# Example 1: Interactive table display with itables.show()
print("Example 1: Interactive table display")

# Create a sample DataFrame
df = pd.DataFrame({
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam'],
    'Price': [1200, 25, 75, 300, 80],
    'Category': ['Electronics', 'Accessories', 'Accessories', 'Electronics', 'Electronics'],
    'Stock': [50, 200, 150, 30, 75],
    'Rating': [4.5, 4.2, 4.7, 4.8, 4.1]
})

print("Original DataFrame:")
print(df)

# Display as interactive table
print("\nDisplaying as interactive table:")
result = itables.show(df)
print(f"Interactive table displayed (result type: {type(result)})")

print("Interactive table display completed successfully")
```

## Example 2: itables notebook initialization functionality test 2

```python
import itables
import pandas as pd

# Example 2: Notebook initialization with init_notebook_mode()
print("Example 2: Notebook mode initialization")

# Initialize notebook mode for itables
try:
    itables.init_notebook_mode(all_interactive=True)
    print("Notebook mode initialized with all_interactive=True")
except Exception as e:
    print(f"Initialization error: {e}")
    raise

# Create test data
df = pd.DataFrame({
    'City': ['Tokyo', 'Delhi', 'Shanghai', 'São Paulo', 'Mexico City'],
    'Population': [37400000, 29000000, 26300000, 22000000, 21700000],
    'Country': ['Japan', 'India', 'China', 'Brazil', 'Mexico'],
    'Continent': ['Asia', 'Asia', 'Asia', 'South America', 'North America']
})

print("Sample cities DataFrame:")
print(df.head())

# After initialization, DataFrames should display interactively
print("DataFrame will display interactively in notebook")
itables.show(df)

print("Notebook initialization completed successfully")
```

## Example 3: itables HTML conversion functionality test 3

```python
import itables
import pandas as pd

# Example 3: HTML DataTable conversion with to_html_datatable()
print("Example 3: HTML DataTable conversion")

# Create sample financial data
df = pd.DataFrame({
    'Symbol': ['AAPL', 'GOOGL', 'MSFT', 'AMZN', 'TSLA'],
    'Price': [150.25, 2800.50, 330.75, 3200.00, 850.30],
    'Change': [2.5, -15.25, 5.80, 45.50, -12.75],
    'Volume': [50000000, 1200000, 25000000, 3000000, 18000000],
    'Market_Cap': ['2.4T', '1.8T', '2.5T', '1.6T', '850B']
})

print("Financial data DataFrame:")
print(df)

# Convert to HTML DataTable
html_table = itables.to_html_datatable(df)
print(f"\nHTML DataTable generated")
print(f"HTML length: {len(html_table)} characters")
print(f"Contains DataTable: {'DataTable' in html_table}")
print(f"Contains table element: {'<table' in html_table}")

# Show first 200 characters of HTML
print(f"HTML preview: {html_table[:200]}...")

print("HTML DataTable conversion completed successfully")
```

## Example 4: itables formatting functionality test 4

```python
import itables
import json
import pandas as pd
from itables.datatables_format import datatables_rows

# Example 4: DataTable formatting with datatables_format()
print("Example 4: DataTable formatting")

# Create sample sales data ('ME' = month-end; the old 'M' alias was removed in pandas 2.2+)
df = pd.DataFrame({
    'Date': pd.date_range('2023-01-01', periods=6, freq='ME'),
    'Sales': [125000.50, 138750.25, 142300.75, 159800.00, 167500.30, 175200.80],
    'Units': [500, 555, 569, 639, 670, 701],
    'Profit_Margin': [0.15, 0.18, 0.16, 0.20, 0.19, 0.21]
})

print("Sales data DataFrame:")
print(df)

# Apply DataTable formatting: `itables.datatables_format` is a MODULE, and its `datatables_rows()`
# renders the rows as the JSON the table embeds.
formatted_result = datatables_rows(df)
print(f"\nDataTable formatting applied")
print(f"Formatted result type: {type(formatted_result)}")
rows = json.loads(formatted_result)
assert len(rows) == len(df) and all(len(row) == len(df.columns) for row in rows), rows
assert rows[0] == ["2023-01-31", 125000.5, 500, 0.15], rows[0]

# Show with custom column definitions
itables.show(df, columnDefs=[{"className": "dt-center", "targets": "_all"}])

print("DataTable formatting completed successfully")
```

## Example 5: itables configuration functionality test 5

```python
import itables
import pandas as pd

# Example 5: Configuration and options with itables.options
print("Example 5: itables configuration and options")

# Create sample data
df = pd.DataFrame({
    'ID': range(1, 101),
    'Name': [f'User_{i}' for i in range(1, 101)],
    'Score': [i * 1.5 + 50 for i in range(1, 101)],
    'Category': ['A', 'B', 'C'] * 33 + ['A']
})

print(f"Large DataFrame created: {df.shape}")

# Access and modify itables options
try:
    # Show available options
    options_attrs = [attr for attr in dir(itables.options) if not attr.startswith('_')]
    print(f"Available options: {options_attrs[:5]}")
    
    # Configure table display options
    itables.show(df, 
                 lengthMenu=[5, 10, 25, 50],
                 pageLength=10,
                 scrollX=True,
                 scrollY="400px")
    
    print("Table displayed with custom options:")
    print("- Page length: 10 rows")  
    print("- Length menu: [5, 10, 25, 50]")
    print("- Scrollable X and Y")
    
except Exception as e:
    print(f"Configuration error: {e}")
    # Fallback to basic display
    itables.show(df.head(10))
    raise AssertionError(
        f"itables example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Configuration and options completed successfully")
```
