## Example 1: jaraco.classes properties functionality test 1

```python
from jaraco.classes.properties import classproperty

# Example 1: Class properties with classproperty decorator
print("Example 1: Class properties")

class DatabaseConfig:
    _host = "localhost"
    _port = 5432
    _database = "myapp"
    
    @classproperty
    def connection_string(cls):
        """Generate database connection string"""
        return f"postgresql://{cls._host}:{cls._port}/{cls._database}"
    
    @classproperty  
    def default_timeout(cls):
        """Default connection timeout"""
        return 30

# Test class properties
print("Class properties demonstration:")
print(f"Connection string: {DatabaseConfig.connection_string}")
print(f"Default timeout: {DatabaseConfig.default_timeout}")

# Properties work on both class and instances
db_config = DatabaseConfig()
print(f"Instance connection string: {db_config.connection_string}")
print(f"Instance timeout: {db_config.default_timeout}")

# Verify it's the same for class and instance
print(f"Same value: {DatabaseConfig.connection_string == db_config.connection_string}")

print("Class properties completed successfully")
```

## Example 2: jaraco.classes ancestry functionality test 2

```python
from jaraco.classes.ancestry import all_bases, iter_subclasses

# Example 2: Class ancestry analysis with all_bases and iter_subclasses  
print("Example 2: Class ancestry analysis")

# Create a class hierarchy
class Animal:
    pass

class Mammal(Animal):
    pass

class Bird(Animal):
    pass

class Dog(Mammal):
    pass

class Cat(Mammal):
    pass

class Eagle(Bird):
    pass

# Test all_bases function
print("Testing all_bases function:")
classes_to_test = [Dog, Cat, Eagle, Mammal]
for cls in classes_to_test:
    bases = list(all_bases(cls))
    print(f"  {cls.__name__} bases: {[b.__name__ for b in bases]}")

# Test iter_subclasses function
print("\nTesting iter_subclasses function:")
base_classes = [Animal, Mammal, Bird]
for base_cls in base_classes:
    subclasses = list(iter_subclasses(base_cls))
    print(f"  {base_cls.__name__} subclasses: {[s.__name__ for s in subclasses]}")

print("Class ancestry analysis completed successfully")
```

## Example 3: jaraco.classes meta functionality test 3

```python
from jaraco.classes.meta import LeafClassesMeta

# Example 3: Metaclass for leaf classes with LeafClassesMeta
print("Example 3: Leaf classes metaclass")

# Create a class hierarchy using LeafClassesMeta
class Shape(metaclass=LeafClassesMeta):
    def area(self):
        raise NotImplementedError("Must implement area")

class Polygon(Shape):
    def perimeter(self):
        raise NotImplementedError("Must implement perimeter")

# These are leaf classes (no subclasses)
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14159 * self.radius ** 2

class Rectangle(Polygon):
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height
    
    def perimeter(self):
        return 2 * (self.width + self.height)

# Test the metaclass functionality
print("Testing LeafClassesMeta:")
print(f"Shape.__class__: {Shape.__class__.__name__}")
print(f"Circle.__class__: {Circle.__class__.__name__}")
print(f"Rectangle.__class__: {Rectangle.__class__.__name__}")

# Create instances
circle = Circle(5)
rectangle = Rectangle(4, 6)

print(f"Circle area: {circle.area()}")
print(f"Rectangle area: {rectangle.area()}")
print(f"Rectangle perimeter: {rectangle.perimeter()}")

print("Leaf classes metaclass completed successfully")
```

## Example 4: jaraco.classes NonDataProperty functionality test 4

```python
from jaraco.classes.properties import NonDataProperty

# Example 4: Non-data properties with NonDataProperty
print("Example 4: Non-data properties")

class APIClient:
    def __init__(self, base_url):
        self.base_url = base_url
        self._session = None
    
    @NonDataProperty
    def session(self):
        """Lazy-loaded session object"""
        if self._session is None:
            print("Creating new session...")
            self._session = {"auth": "token", "timeout": 30}
        return self._session
    
    @NonDataProperty
    def headers(self):
        """Standard API headers"""
        return {
            "Content-Type": "application/json",
            "User-Agent": "MyApp/1.0",
            "Accept": "application/json"
        }

# Test NonDataProperty
print("Testing NonDataProperty:")
client = APIClient("https://api.example.com")

# First access - should create session
print("First session access:")
session1 = client.session
print(f"Session: {session1}")

# Second access - should reuse existing session
print("\nSecond session access:")
session2 = client.session  
print(f"Same session: {session1 is session2}")

# Headers property
print(f"\nHeaders: {client.headers}")

# Property can be overridden since it's non-data
client.session = {"override": True}
print(f"Overridden session: {client.session}")

print("Non-data properties completed successfully")
```

## Example 5: jaraco.classes TagRegistered functionality test 5

```python
from jaraco.classes.ancestry import all_bases

# Example 5: Class hierarchy exploration with all_bases
print("Example 5: Advanced class hierarchy exploration")

# Create a complex class hierarchy
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

class LandVehicle(Vehicle):
    def __init__(self, brand, wheels):
        super().__init__(brand)
        self.wheels = wheels

class WaterVehicle(Vehicle):
    def __init__(self, brand, displacement):
        super().__init__(brand)
        self.displacement = displacement

class Car(LandVehicle):
    def __init__(self, brand, model):
        super().__init__(brand, 4)
        self.model = model

class Boat(WaterVehicle):
    def __init__(self, brand, length):
        super().__init__(brand, 1000)
        self.length = length

# Test all_bases with complex hierarchy
print("Testing complex class hierarchies:")
test_classes = [Car, Boat, LandVehicle, WaterVehicle]

for cls in test_classes:
    bases = list(all_bases(cls))
    base_names = [b.__name__ for b in bases]
    print(f"  {cls.__name__} inheritance chain: {' -> '.join(base_names)}")

# Create instances to verify functionality
car = Car("Toyota", "Camry")
boat = Boat("Yamaha", 25)

print(f"\nInstances created:")
print(f"Car: {car.brand} {car.model} ({car.wheels} wheels)")
print(f"Boat: {boat.brand} ({boat.length}ft, {boat.displacement}lbs)")

print("Class hierarchy exploration completed successfully")
```
