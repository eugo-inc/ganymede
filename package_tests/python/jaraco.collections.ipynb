# jaraco.collections: 5 Basic Usage Examples

```python
# Submodule smoke test: import the package + walk every submodule.
import importlib
import pkgutil

import jaraco.collections as _pkg_under_test

failed = []
for module_info in pkgutil.walk_packages(_pkg_under_test.__path__, prefix=f"{_pkg_under_test.__name__}."):
    try:
        importlib.import_module(module_info.name)
    except Exception as e:
        failed.append((module_info.name, type(e).__name__, str(e)))

if failed:
    for name, etype, msg in failed:
        print(f"FAIL {name}: {etype}: {msg}")
    raise AssertionError(f"{len(failed)} submodule(s) of {_pkg_under_test.__name__} failed to import")
print(f"OK — all submodules of {_pkg_under_test.__name__} imported cleanly")
```

```python
# Example 1: BijectiveMap (two-way mapping)
from jaraco.collections import BijectiveMap
m = BijectiveMap()
m['a'] = 'b'
m['c'] = 'd'
print(m['a'], m['b'])  # 'b', 'a'
print(m['c'], m['d'])  # 'd', 'c'
```

```python
# Example 2: RangeMap (keys are inclusive upper bounds of ranges)
from jaraco.collections import RangeMap

m = RangeMap({
    10: 'low',    # 0-10 maps to 'low'
    20: 'mid',    # 11-20 maps to 'mid'
    30: 'high'    # 21-30 maps to 'high'
})

print(m[5])   # 'low'  (5 <= 10)
print(m[15])  # 'mid'  (15 <= 20, but 15 > 10)
print(m[25])  # 'high' (25 <= 30, but 25 > 20)

assert (m[5], m[15], m[25]) == ('low', 'mid', 'high')
```

```python
# Example 3: Projection (filtering dicts by keys)
from jaraco.collections import Projection
d = {'a': 1, 'b': 2, 'c': 3}
f = Projection(['a', 'c'], d)
print(dict(f))  # {'a': 1, 'c': 3}
```

```python
# Example 4: Projection (exclude keys)
from jaraco.collections import Projection
d_full = {'a': 1, 'b': 2, 'c': 3, 'd': 4, 'e': 5}
exclude_keys = ['b', 'd']

# Create projection that excludes keys by using lambda
exclude_projection = Projection(lambda k: k not in exclude_keys, d_full)
print("Excluding b,d:", dict(exclude_projection))
```

```python
#Using Projection with an iterable
data = {'name': 'John', 'age': 30, 'city': 'NYC', 'country': 'USA', 'email': 'john@example.com'}

# Filter to keep only contact information
contact_filter = Projection(['name', 'email', 'city'], data)
print("Contact info:", dict(contact_filter))
```
