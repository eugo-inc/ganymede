# jaraco.text Package Testing

This notebook tests the jaraco.text package, which provides text processing utilities including normalization, formatting, and advanced string operations.

## Example 1: Indenting

```python
import jaraco.text

# Test indent functionality
multiline = "Line 1\nLine 2\nLine 3"
indented = jaraco.text.indent(multiline, "  ")
print(f"\nIndented text:\n{indented}")

```

## Example 2: Text Splitting

```python
import jaraco.text

# Test splitting by lines
multiline_text = "First line\nSecond line\n\nThird line after blank"
print(multiline_text)
lines = multiline_text.splitlines()
print(f"\nSplit lines: {lines}")
```

## Example 3: Text Formatting and Templates

```python
import jaraco.text
from string import Template

# Test text formatting utilities
template_text = "Hello $name, welcome to $place!"
template = Template(template_text)

# Test substitution
result = template.substitute(name="Alice", place="Python")
print(f"Template result: {result}")

# Test safe substitution
partial_result = template.safe_substitute(name="Bob")
print(f"Partial substitution: {partial_result}")

# Test FoldedCase functionality
folded = jaraco.text.FoldedCase("Hello World!")
print(f"\nFolded case: {folded}")
print(f"Comparison: {'hello world!' == folded}")
print(f"Comparison: {'HELLO WORLD!' == folded}")

# Test case-insensitive operations
text_list = ["Apple", "BANANA", "cherry", "DATE"]
folded_list = [jaraco.text.FoldedCase(item) for item in text_list]
print(f"\nOriginal list: {text_list}")
print(f"Folded list: {folded_list}")
print(f"Contains 'apple': {jaraco.text.FoldedCase('apple') in folded_list}")
```

## Example 4: String Normalization

```python
folded1 = jaraco.text.FoldedCase("Hello World")
folded2 = jaraco.text.FoldedCase("HELLO world")
folded3 = jaraco.text.FoldedCase("hello WORLD")
        
print(f"Original strings:")
print(f"  '{folded1}'")
print(f"  '{folded2}'") 
print(f"  '{folded3}'")
print(f"Are they equal? {folded1 == folded2 == folded3}")
print(f"Hash values: {hash(folded1)}, {hash(folded2)}, {hash(folded3)}")
        
# Use in a set to demonstrate uniqueness
case_set = {folded1, folded2, folded3, jaraco.text.FoldedCase("different")}
print(f"Unique items in set: {len(case_set)}")
```

## Example 5: Advanced Text Operations

```python
import jaraco.text
import re

# Test advanced text operations
sample_data = '''
Name: John Doe
Age: 30
City: New York
Email: john.doe@example.com

Name: Jane Smith  
Age: 25
City: Los Angeles
Email: jane.smith@example.com
'''

print("Testing advanced text operations:")
print(f"Sample data length: {len(sample_data)} characters")

# Test line processing
lines = sample_data.strip().split('\n')
non_empty_lines = [line for line in lines if line.strip()]
print(f"Non-empty lines: {len(non_empty_lines)}")

# Test pattern matching and extraction
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'
emails = re.findall(email_pattern, sample_data)
print(f"Found emails: {emails}")

# Test text statistics
words = jaraco.text.words(sample_data)
print(f"\nWord count: {len(words)}")
print(f"First 10 words: {words[:10]}")

# Test levenshtein distance if available
try:
    distance = jaraco.text.levenshtein("kitten", "sitting")
    print(f"\nLevenshtein distance: {distance}")
except AttributeError:
    print("\nLevenshtein distance not available")

```
