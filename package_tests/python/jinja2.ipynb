## Example 1: jinja2 functionality test 1

```python
from jinja2 import Template
import jinja2

# Example 1: Template class and basic variable substitution
print("Example 1: Template Class and Variable Substitution")

# Create Template instance with variable substitution
template_source = "Hello {{ name }}! You have {{ count }} new {{ item_type }}."
template = Template(template_source)

# Test basic rendering with different data types
result1 = template.render(name="Alice", count=5, item_type="messages")
result2 = template.render(name="Bob", count=1, item_type="notification")
print(f"Result 1: {result1}")
print(f"Result 2: {result2}")

# Template with expressions and method calls
expr_template = Template("User: {{ user.name.title() }} | Score: {{ (score * 100) | round(1) }}%")
user_data = {"name": "john doe"}
result3 = expr_template.render(user=user_data, score=0.856)
print(f"Expression result: {result3}")

# Template metadata and properties
print(f"\nTemplate source: {template_source}")
print(f"Template object type: {type(template)}")
print(f"Jinja2 version: {jinja2.__version__}")
print(f"Template environment: {type(template.environment).__name__}")

# Error handling with Template
try:
    bad_template = Template("{{ missing_var }}")
    result = bad_template.render()
except Exception as e:
    print(f"Template error handled: {type(e).__name__}")
    raise

print("Example 1 completed successfully\n")
```

## Example 2: jinja2 functionality test 2

```python
from jinja2 import Environment, DictLoader, FileSystemLoader
import os
import tempfile

# Example 2: Environment class and template loaders
print("Example 2: Environment Class and Template Loaders")

# Create Environment with DictLoader
templates = {
    'greeting': 'Hello {{ name }}!',
    'user_info': 'User: {{ user.name }} ({{ user.role }})',
    'list': '{% for item in items %}{{ loop.index }}. {{ item }}{% if not loop.last %}, {% endif %}{% endfor %}'
}

env = Environment(loader=DictLoader(templates))

# Load and render templates from dictionary
greeting_tmpl = env.get_template('greeting')
print(f"Dictionary loader: {greeting_tmpl.render(name='World')}")

list_tmpl = env.get_template('list')
print(f"List template: {list_tmpl.render(items=['Apple', 'Banana', 'Cherry'])}")

# Create temporary directory with template files for FileSystemLoader
with tempfile.TemporaryDirectory() as tmpdir:
    # Write template file
    tmpl_path = os.path.join(tmpdir, 'message.txt')
    with open(tmpl_path, 'w') as f:
        f.write('Dear {{ recipient }},\n\nYour order #{{ order_id }} is {{ status }}.\n\nTotal: ${{ total | round(2) }}')
    
    # Create Environment with FileSystemLoader
    fs_env = Environment(loader=FileSystemLoader(tmpdir))
    message_tmpl = fs_env.get_template('message.txt')
    
    result = message_tmpl.render(
        recipient="John Doe",
        order_id="12345",
        status="shipped",
        total=99.99
    )
    print(f"\nFileSystem loader result:\n{result}")

# Environment configuration inspection
print(f"\nEnvironment details:")
print(f"Available templates: {list(env.list_templates())}")
print(f"Loader type: {type(env.loader).__name__}")

print("Example 2 completed successfully\n")
```

## Example 3: jinja2 functionality test 3

```python
from jinja2 import Template, Environment

# Example 3: Built-in filters and tests
print("Example 3: Built-in Filters and Tests")

# Test various built-in filters
filter_template = Template("""
Original: {{ text }}
Upper: {{ text | upper }}
Truncate: {{ text | truncate(20) }}
Default: {{ missing | default('Not provided') }}
Join: {{ items | join(' | ') }}
Length: {{ items | length }}
Sort: {{ numbers | sort | join(', ') }}
Unique: {{ duplicates | unique | list | join(', ') }}
""".strip())

result = filter_template.render(
    text="hello jinja2 world",
    items=["apple", "banana", "cherry"],
    numbers=[3, 1, 4, 1, 5, 9],
    duplicates=[1, 2, 2, 3, 3, 3, 4]
)
print("Filter examples:")
print(result)

# Test conditional expressions with valid built-in tests
test_template = Template("""
{% if name is defined %}Name: {{ name }}{% endif %}
{% if age is number %}Age is numeric: {{ age }}{% endif %}
{% if items is iterable %}Items count: {{ items | length }}{% endif %}
{% if text is string %}Text length: {{ text | length }}{% endif %}
{% if value is divisibleby(2) %}{{ value }} is even{% else %}{{ value }} is odd{% endif %}
{% if empty_list is sameas([]) %}Empty list detected{% endif %}
""".strip())

result2 = test_template.render(
    name="Alice",
    age=25,
    items=[1, 2, 3],
    text="Hello",
    value=42,
    empty_list=[]
)
print("\nTest examples:")
print(result2)

# Chaining filters
chain_template = Template("{{ sentence | lower | replace(' ', '_') | truncate(15) }}")
result3 = chain_template.render(sentence="Hello Beautiful World")
print(f"\nChained filters: {result3}")

print("Example 3 completed successfully\n")
```

## Example 4: jinja2 functionality test 4

```python
from jinja2 import Environment, DictLoader

# Example 4: Template inheritance and macros
print("Example 4: Template Inheritance and Macros")

# Define templates with inheritance and macros
templates = {
    'base.html': '''
{%- macro render_nav(items) -%}
<nav>{% for item in items %}<a href="{{ item.url }}">{{ item.title }}</a>{% endfor %}</nav>
{%- endmacro -%}
<!DOCTYPE html>
<html>
<head><title>{% block title %}Default{% endblock %}</title></head>
<body>
{{ render_nav(nav_items) }}
<main>{% block content %}{% endblock %}</main>
</body>
</html>''',
    
    'page.html': '''
{%- extends "base.html" -%}
{%- from "base.html" import render_nav -%}

{% block title %}{{ page_title }}{% endblock %}

{% block content %}
<h1>{{ heading }}</h1>
<p>{{ content }}</p>
{% if show_nav %}{{ render_nav(sub_nav) }}{% endif %}
{% endblock %}'''
}

env = Environment(loader=DictLoader(templates))

# Render child template with inheritance
page_tmpl = env.get_template('page.html')
result = page_tmpl.render(
    page_title="About Us",
    heading="Welcome to Our Site",
    content="This page demonstrates Jinja2 template inheritance and macros.",
    nav_items=[
        {"url": "/home", "title": "Home"},
        {"url": "/about", "title": "About"}
    ],
    show_nav=True,
    sub_nav=[
        {"url": "/team", "title": "Team"},
        {"url": "/contact", "title": "Contact"}
    ]
)

print("Inheritance and macro result:")
print(result[:200] + "...")

# Macro-only template
macro_template = env.from_string('''
{%- macro format_user(user) -%}
{{ user.name }} ({{ user.role }})
{%- endmacro -%}
Users: {% for u in users %}{{ format_user(u) }}{% if not loop.last %}, {% endif %}{% endfor %}
''')

users_result = macro_template.render(users=[
    {"name": "Alice", "role": "Admin"},
    {"name": "Bob", "role": "User"}
])
print(f"\nMacro result: {users_result.strip()}")

print("Example 4 completed successfully\n")
```

## Example 5: jinja2 functionality test 5

```python
from jinja2 import Environment, select_autoescape
from markupsafe import Markup
import jinja2.exceptions

# Example 5: Auto-escaping and security features
print("Example 5: Auto-escaping and Security Features")

# Create environment with auto-escaping enabled
secure_env = Environment(autoescape=select_autoescape(['html', 'xml']))

# Test auto-escaping behavior
escape_template = secure_env.from_string("""
Safe content: {{ safe_content }}
Unsafe content: {{ unsafe_content }}
Manual safe: {{ unsafe_content | safe }}
Force escape: {{ safe_content | forceescape }}
""".strip())

result = escape_template.render(
    safe_content="Hello World",
    unsafe_content="<script>alert('XSS')</script>"
)
print("Auto-escaping demonstration:")
print(result)

# Using Markup for pre-escaped content
markup_template = secure_env.from_string("Pre-marked safe: {{ safe_html }}")
safe_html = Markup("<b>Bold Text</b>")
markup_result = markup_template.render(safe_html=safe_html)
print(f"\nMarkup result: {markup_result}")

# Security: Template sandbox behavior
try:
    dangerous_template = secure_env.from_string("{{ ''.__class__.__mro__[1].__subclasses__() }}")
    result = dangerous_template.render()
    print(f"Dangerous access result: {result}")
except Exception as e:
    print(f"Security protection: {type(e).__name__}")
    raise

# Custom filter with security consideration
def safe_url_filter(url):
    if url.startswith(('http://', 'https://')):
        return Markup(f'<a href="{url}">{url}</a>')
    return url

secure_env.filters['safe_url'] = safe_url_filter

url_template = secure_env.from_string("Link: {{ url | safe_url }}")
url_result = url_template.render(url="https://example.com")
print(f"\nSafe URL filter: {url_result}")

# Environment security settings
print(f"\nSecurity settings:")
print(f"Autoescape enabled: {secure_env.autoescape}")
print(f"Available extensions: {len(secure_env.extensions)}")
print(f"Finalize function: {secure_env.finalize}")

print("Example 5 completed successfully\n")
```
