## Example 1: Functionality Test 1

```python
from jinja2 import Environment, Template
from datetime import datetime, date
import os

# Example 1: File size formatting and date humanization with naturalsize and naturaldate
print("Example 1: File Size and Date Humanization")

# Setup jinja2 environment with humanize extension
env = Environment(extensions=["jinja2_humanize_extension.HumanizeExtension"])

# Test naturalsize filter with various file sizes
file_sizes = [1024, 1048576, 1073741824, 5368709120]
size_template = env.from_string("{{ size|humanize_naturalsize() }} ({{ size|humanize_naturalsize(binary=True) }} binary)")

print("File size formatting:")
for size in file_sizes:
    result = size_template.render(size=size)
    print(f"  {size} bytes -> {result}")

# Test naturaldate filter with dates
dates = [
    date(2024, 1, 1),
    date(2024, 6, 15), 
    date(2024, 12, 25),
    date.today()
]
date_template = env.from_string("{{ date_val|humanize_naturaldate() }}")

print("\nDate formatting:")
for date_val in dates:
    result = date_template.render(date_val=date_val)
    print(f"  {date_val} -> {result}")

# Combined template example
combined_template = env.from_string("""
File: {{ filename }}
Size: {{ filesize|humanize_naturalsize() }}
Modified: {{ modified_date|humanize_naturaldate() }}
""".strip())

result = combined_template.render(
    filename="document.pdf",
    filesize=2097152,
    modified_date=date(2024, 11, 1)
)
print(f"\nCombined example:\n{result}")

print("Example 1 completed successfully\n")
```

## Example 2: Functionality Test 2

```python
from jinja2 import Environment
from datetime import datetime, timedelta
import time

# Example 2: Time and delta humanization with naturaltime and naturaldelta
print("Example 2: Time and Delta Humanization")

# Setup jinja2 environment with humanize extension
env = Environment(extensions=["jinja2_humanize_extension.HumanizeExtension"])

# Test naturaltime filter with various timestamps
now = datetime.now()
timestamps = [
    now - timedelta(seconds=30),
    now - timedelta(minutes=5),
    now - timedelta(hours=2),
    now - timedelta(days=1),
    now - timedelta(days=7),
    now + timedelta(hours=3)
]

time_template = env.from_string("{{ timestamp|humanize_naturaltime() }}")

print("Natural time formatting:")
for timestamp in timestamps:
    result = time_template.render(timestamp=timestamp)
    print(f"  {timestamp.strftime('%Y-%m-%d %H:%M:%S')} -> {result}")

# Test naturaldelta filter with time deltas
deltas = [
    timedelta(seconds=45),
    timedelta(minutes=10, seconds=30),
    timedelta(hours=3, minutes=45),
    timedelta(days=2, hours=8),
    timedelta(weeks=1, days=3)
]

delta_template = env.from_string("{{ delta|humanize_naturaldelta() }}")

print("\nNatural delta formatting:")
for delta in deltas:
    result = delta_template.render(delta=delta)
    print(f"  {delta} -> {result}")

# Activity timeline template
activity_template = env.from_string("""
Activity: {{ activity }}
Duration: {{ duration|humanize_naturaldelta() }}
Started: {{ start_time|humanize_naturaltime() }}
""".strip())

result = activity_template.render(
    activity="Data processing",
    duration=timedelta(hours=2, minutes=30),
    start_time=now - timedelta(hours=1)
)
print(f"\nActivity timeline:\n{result}")

print("Example 2 completed successfully\n")
```

## Example 3: Functionality Test 3

```python
from jinja2 import Environment

# Example 3: Number formatting with ordinal and intcomma
print("Example 3: Number Formatting")

# Setup jinja2 environment with humanize extension
env = Environment(extensions=["jinja2_humanize_extension.HumanizeExtension"])

# Test ordinal filter for rankings and positions
ordinal_numbers = [1, 2, 3, 4, 11, 21, 22, 23, 101, 111, 1001]
ordinal_template = env.from_string("{{ number|humanize_ordinal() }}")

print("Ordinal formatting:")
for number in ordinal_numbers:
    result = ordinal_template.render(number=number)
    print(f"  {number} -> {result}")

# Test intcomma filter for large numbers
large_numbers = [1000, 10000, 100000, 1000000, 1234567890]
comma_template = env.from_string("{{ number|humanize_intcomma() }}")

print("\nComma formatting:")
for number in large_numbers:
    result = comma_template.render(number=number)
    print(f"  {number} -> {result}")

# Competition results template
results_template = env.from_string("""
🏆 Competition Results:
{{- competitor.name }} finished {{ competitor.position|humanize_ordinal() }}
Score: {{ competitor.score|humanize_intcomma() }} points
Prize: ${{ competitor.prize|humanize_intcomma() }}
""".strip())

competitors = [
    {"name": "Alice", "position": 1, "score": 1250000, "prize": 50000},
    {"name": "Bob", "position": 2, "score": 980000, "prize": 25000},
    {"name": "Charlie", "position": 3, "score": 875000, "prize": 10000}
]

print("\nCompetition results:")
for competitor in competitors:
    result = results_template.render(competitor=competitor)
    print(result)

print("Example 3 completed successfully\n")
```

## Example 4: Functionality Test 4

```python
from jinja2 import Environment
from fractions import Fraction

# Example 4: Word numbers and fractions with intword and fractional
print("Example 4: Word Numbers and Fractions")

# Setup jinja2 environment with humanize extension
env = Environment(extensions=["jinja2_humanize_extension.HumanizeExtension"])

# Test intword filter for converting large numbers to words
large_numbers = [
    1000000, 2500000, 1000000000, 
    1200000000, 1500000000000, 
    7800000000000000
]
word_template = env.from_string("{{ number|humanize_intword() }}")

print("Integer to words formatting:")
for number in large_numbers:
    result = word_template.render(number=number)
    print(f"  {number:,} -> {result}")

# Test fractional filter for decimal to fraction conversion
decimals = [0.5, 0.25, 0.125, 0.333, 0.666, 0.1875, 0.875]
fractional_template = env.from_string("{{ decimal|humanize_fractional() }}")

print("\nDecimal to fraction formatting:")
for decimal in decimals:
    result = fractional_template.render(decimal=decimal)
    print(f"  {decimal} -> {result}")

# Financial report template
report_template = env.from_string("""
📊 Financial Summary:
Total Revenue: {{ revenue|humanize_intword() }}
Market Share: {{ market_share|humanize_fractional() }} ({{ (market_share * 100)|round(1) }}%)
Users Served: {{ users|humanize_intword() }}
Growth Rate: {{ growth_rate|humanize_fractional() }} increase
""".strip())

financial_data = {
    "revenue": 2500000000,
    "market_share": 0.125,
    "users": 45000000,
    "growth_rate": 0.0625
}

result = report_template.render(**financial_data)
print(f"\nFinancial report:\n{result}")

# Population statistics
stats_template = env.from_string("{{ city }} has {{ population|humanize_intword() }} residents")

cities = [
    {"city": "Tokyo", "population": 14000000},
    {"city": "New York", "population": 8400000},
    {"city": "London", "population": 9500000}
]

print("\nPopulation statistics:")
for city_data in cities:
    result = stats_template.render(**city_data)
    print(f"  {result}")

print("Example 4 completed successfully\n")
```

## Example 5: Functionality Test 5

```python
from jinja2 import Environment
from datetime import timedelta
import decimal

# Example 5: Precise time deltas and scientific notation with precisedelta and scientific
print("Example 5: Precise Deltas and Scientific Notation")

# Setup jinja2 environment with humanize extension
env = Environment(extensions=["jinja2_humanize_extension.HumanizeExtension"])

# Test precisedelta filter for detailed time formatting
precise_deltas = [
    timedelta(seconds=3661),  # 1 hour, 1 minute, 1 second
    timedelta(days=365, hours=12, minutes=30, seconds=45),
    timedelta(weeks=2, days=3, hours=4, minutes=15),
    timedelta(milliseconds=1500),
    timedelta(hours=25, minutes=75, seconds=90)
]

precise_template = env.from_string("{{ delta|humanize_precisedelta() }}")
minimum_unit_template = env.from_string("{{ delta|humanize_precisedelta(minimum_unit='minutes') }}")

print("Precise delta formatting:")
for delta in precise_deltas:
    result = precise_template.render(delta=delta)
    min_result = minimum_unit_template.render(delta=delta)
    print(f"  {delta} -> {result} (min unit: {min_result})")

# Test scientific filter for scientific notation
scientific_numbers = [
    0.000000123,
    123000000,
    0.00456,
    9876543210,
    0.000000000987,
    123.456789
]

scientific_template = env.from_string("{{ number|humanize_scientific(precision=2) }}")

print("\nScientific notation formatting:")
for number in scientific_numbers:
    result = scientific_template.render(number=number)
    print(f"  {number} -> {result}")

# Laboratory data template
lab_template = env.from_string("""
🔬 Lab Analysis Results:
Sample ID: {{ sample_id }}
Processing Time: {{ processing_time|humanize_precisedelta() }}
Concentration: {{ concentration|humanize_scientific(precision=3) }} mol/L
Mass: {{ mass|humanize_scientific(precision=2) }} grams
Accuracy: {{ accuracy|humanize_scientific(precision=4) }}%
""".strip())

lab_data = {
    "sample_id": "LAB-001",
    "processing_time": timedelta(hours=2, minutes=34, seconds=12),
    "concentration": 0.000000456,
    "mass": 0.00123456,
    "accuracy": 99.9876
}

result = lab_template.render(**lab_data)
print(f"\nLab analysis:\n{result}")

# Performance benchmark template
benchmark_template = env.from_string("""
⚡ Performance Benchmark:
Test Duration: {{ duration|humanize_precisedelta(minimum_unit='seconds') }}
Response Time: {{ response_time|humanize_scientific(precision=1) }} seconds
Error Rate: {{ error_rate|humanize_scientific(precision=2) }}%
""".strip())

benchmark_data = {
    "duration": timedelta(minutes=5, seconds=23, milliseconds=750),
    "response_time": 0.00234,
    "error_rate": 0.0000123
}

result = benchmark_template.render(**benchmark_data)
print(f"\nBenchmark results:\n{result}")

print("Example 5 completed successfully\n")
```
