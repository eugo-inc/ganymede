# jinxed smoke test

Pure-Python implementation of a subset of the ncurses terminfo library. Runtime dep of `blessed`.

```python
# Example 1: setupterm() initializes the terminal database for a given term type.
import jinxed

# xterm-256color is one of the bundled terminfo entries jinxed ships with.
jinxed.setupterm('xterm-256color', 1)

# After setupterm, the module-level state should be initialized.
print(f"setupterm('xterm-256color') completed without error")
assert hasattr(jinxed, 'setupterm')
```

```python
# Example 2: tigetnum() retrieves numeric capabilities (e.g., colors, cols, lines).
import jinxed

jinxed.setupterm('xterm-256color', 1)

colors = jinxed.tigetnum('colors')
print(f"tigetnum('colors') = {colors}")

# xterm-256color should report 256 colors.
assert colors == 256, f"expected 256 colors for xterm-256color, got {colors}"
```

```python
# Example 3: tigetstr() retrieves string capabilities (escape sequences).
import jinxed

jinxed.setupterm('xterm-256color', 1)

# 'bold' is a standard terminfo capability for the bold-text escape sequence.
bold_seq = jinxed.tigetstr('bold')
print(f"tigetstr('bold') = {bold_seq!r}")

# Should return a non-empty bytes sequence (typically b'\x1b[1m').
assert bold_seq is not None
assert isinstance(bold_seq, bytes)
assert len(bold_seq) > 0
```

```python
# Example 4: tparm() parameterizes a capability string (e.g., setaf for foreground color).
import jinxed

jinxed.setupterm('xterm-256color', 1)

setaf = jinxed.tigetstr('setaf')  # set foreground ANSI color
print(f"setaf raw = {setaf!r}")

# Parameterize with color index 2 (green).
green = jinxed.tparm(setaf, 2)
print(f"tparm(setaf, 2) = {green!r}")

assert green is not None
assert isinstance(green, bytes)
assert len(green) > 0
```

```python
# Example 5: tigetflag() retrieves boolean capabilities + has_colors helper.
import jinxed

jinxed.setupterm('xterm-256color', 1)

# 'am' (auto_right_margin) is a standard boolean capability — true for most modern terminals.
am = jinxed.tigetflag('am')
print(f"tigetflag('am') = {am}")

# tigetflag returns an int (bool-like): 1 if set, 0 if not, -1 if absent.
assert am in (0, 1, -1)

# Sanity: tigetstr on a known-present cap should return bytes.
clear_seq = jinxed.tigetstr('clear')
assert clear_seq is not None and isinstance(clear_seq, bytes)
print(f"jinxed terminfo lookup works end-to-end")
```
