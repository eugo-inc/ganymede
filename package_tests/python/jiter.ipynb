## Example 1: jiter functionality test 1

```python
import jiter
import json

# Example 1: from_json function - Fast JSON parsing
print("Example 1: jiter.from_json() - Fast JSON Parsing")

# Test basic JSON parsing with from_json (using bytes)
json_bytes = b'{"name": "Alice", "age": 30, "scores": [95, 87, 92], "active": true}'
result = jiter.from_json(json_bytes)
print(f"Parsed JSON: {result}")
print(f"Result type: {type(result)}")

# Parse complex nested JSON (convert string to bytes)
complex_json = b'''
{
    "users": [
        {"id": 1, "name": "John", "metadata": {"role": "admin", "last_login": null}},
        {"id": 2, "name": "Jane", "metadata": {"role": "user", "last_login": "2024-01-15"}}
    ],
    "total_count": 2,
    "page_info": {"has_next": false, "cursor": "abc123"}
}
'''
complex_result = jiter.from_json(complex_json)
print(f"\nComplex JSON parsed successfully")
print(f"Users count: {len(complex_result['users'])}")
print(f"First user: {complex_result['users'][0]['name']}")

# Parse JSON with different data types
types_json = b'{"string": "text", "number": 42, "float": 3.14, "bool": true, "null": null, "array": [1,2,3]}'
types_result = jiter.from_json(types_json)
print(f"\nData types test:")
for key, value in types_result.items():
    print(f"  {key}: {value} ({type(value).__name__})")

# Convert string to bytes for parsing
json_string = '{"message": "Hello from converted string"}'
json_bytes = json_string.encode('utf-8')
bytes_result = jiter.from_json(json_bytes)
print(f"\nString-to-bytes parsing result: {bytes_result}")

print("Example 1 completed successfully\n")
```

## Example 2: jiter functionality test 2

```python
import jiter
import json

# Example 2: LosslessFloat class - Preserve float precision
print("Example 2: jiter.LosslessFloat - Lossless Float Precision")

# Parse JSON with precise floats that would lose precision as regular floats
json_with_precise_numbers = b'{"pi": 3.141592653589793238462643383279, "e": 2.718281828459045235360287471353, "large": 1.7976931348623157e+308}'

# Parse normally (precision may be lost)
normal_result = jiter.from_json(json_with_precise_numbers)
print(f"Normal parsing:")
print(f"  Pi: {normal_result['pi']}")
print(f"  E: {normal_result['e']}")
print(f"  Pi type: {type(normal_result['pi'])}")

# Parse with LosslessFloat to preserve precision
# Note: LosslessFloat is used internally by jiter for precise float handling
try:
    # Test if we can access LosslessFloat functionality
    lossless_float = jiter.LosslessFloat
    print(f"\nLosslessFloat class available: {lossless_float}")
    
    # Parse the same JSON - jiter may automatically use LosslessFloat internally
    precise_result = jiter.from_json(json_with_precise_numbers)
    print(f"Precise parsing result:")
    print(f"  Pi: {precise_result['pi']}")
    print(f"  E: {precise_result['e']}")
    
except Exception as e:
    print(f"LosslessFloat usage error: {e}")
    raise

# Test with very precise decimal numbers
decimal_json = b'{"precise": 0.123456789012345678901234567890}'
decimal_result = jiter.from_json(decimal_json)
print(f"\nDecimal precision test:")
print(f"Precise decimal: {decimal_result['precise']}")
print(f"Type: {type(decimal_result['precise'])}")

# Compare with standard json library
standard_result = json.loads(decimal_json.decode())
print(f"Standard json result: {standard_result['precise']}")
print(f"Results equal: {decimal_result['precise'] == standard_result['precise']}")

# Test edge cases with floating point numbers
edge_cases = [
    b'{"tiny": 1e-100}',
    b'{"huge": 1e100}',
    b'{"negative": -3.141592653589793}'
]

print(f"\nEdge cases with precise floats:")
for i, case in enumerate(edge_cases):
    result = jiter.from_json(case)
    print(f"  Case {i+1}: {result}")

print("Example 2 completed successfully\n")
```

## Example 3: jiter functionality test 3

```python
import jiter

# Example 3: cache_clear and cache_usage - JSON string caching management
print("Example 3: jiter Cache Management")

# Check initial cache usage
initial_usage = jiter.cache_usage()
print(f"Initial cache usage: {initial_usage}")

# Parse several JSON bytes to populate cache
json_data = [
    b'{"user": "alice", "id": 1}',
    b'{"user": "bob", "id": 2}',
    b'{"product": "laptop", "price": 999}',
    b'{"status": "active", "count": 42}',
    b'{"user": "alice", "id": 1}'  # Repeat to test caching
]

print(f"\nParsing {len(json_data)} JSON byte strings:")
for i, json_bytes in enumerate(json_data):
    result = jiter.from_json(json_bytes)
    usage = jiter.cache_usage()
    print(f"  {i+1}. Parsed: {result} | Cache usage: {usage}")

# Check cache usage after parsing
final_usage = jiter.cache_usage()
print(f"\nFinal cache usage: {final_usage}")
print(f"Cache size increased by: {final_usage - initial_usage}")

# Clear the cache
print(f"\nClearing cache...")
jiter.cache_clear()
cleared_usage = jiter.cache_usage()
print(f"Cache usage after clear: {cleared_usage}")

# Parse again to show cache rebuilding
print(f"\nParsing after cache clear:")
result = jiter.from_json(b'{"test": "cache rebuild"}')
rebuilt_usage = jiter.cache_usage()
print(f"Parsed: {result}")
print(f"Cache usage after rebuild: {rebuilt_usage}")

# Demonstrate cache efficiency
print(f"\nCache efficiency test:")
same_json = b'{"repeated": "content", "value": 123}'
for i in range(3):
    jiter.from_json(same_json)
    usage = jiter.cache_usage()
    print(f"  Parse {i+1}: cache usage = {usage}")

print("Example 3 completed successfully\n")
```

## Example 4: jiter functionality test 4

```python
import jiter
import io

# Example 4: Advanced from_json usage with different data sources
print("Example 4: Advanced jiter.from_json Usage")

# Test parsing from different JSON structures
json_structures = [
    b'{"simple": "value"}',
    b'[1, 2, 3, 4, 5]',
    b'{"nested": {"deep": {"value": 42}}}',
    b'{"array": ["a", "b", "c"], "number": 123.45}',
    b'{"boolean": true, "null": null, "empty": ""}'
]

print("Testing jiter.from_json with different JSON structures:")
for i, json_bytes in enumerate(json_structures):
    try:
        result = jiter.from_json(json_bytes)
        result_type = type(result).__name__
        print(f"  Test {i+1}: {json_bytes.decode()[:30]}... -> {result} ({result_type})")
    except Exception as e:
        print(f"  Test {i+1}: Error - {e}")
        raise

# Test with BytesIO stream content
print(f"\nTesting with BytesIO stream:")
stream_data = b'{"stream": "data", "items": [1, 2, 3], "count": 999}'
stream = io.BytesIO(stream_data)
stream_content = stream.read()
stream.seek(0)  # Reset stream position

result = jiter.from_json(stream_content)
print(f"Stream data: {stream_data.decode()}")
print(f"Parsed result: {result}")
print(f"Result type: {type(result)}")

# Test parsing large JSON arrays
print(f"\nTesting with larger JSON structures:")
large_array = b'[' + b','.join([f'{{"id": {i}, "value": "item_{i}"}}'.encode() for i in range(10)]) + b']'
large_result = jiter.from_json(large_array)
print(f"Large array parsed: {len(large_result)} items")
print(f"First item: {large_result[0]}")
print(f"Last item: {large_result[-1]}")

# Test unicode handling
unicode_json = b'{"message": "\u4f60\u597d\u4e16\u754c", "emoji": "\ud83d\ude00\ud83c\udf0d"}'
unicode_result = jiter.from_json(unicode_json)
print(f"\nUnicode handling:")
print(f"Raw bytes: {unicode_json}")
print(f"Parsed result: {unicode_result}")

print("Example 4 completed successfully\n")
```

## Example 5: jiter functionality test 5

```python
import jiter
import json
import time

# Example 5: Performance comparison and error handling
print("Example 5: Performance Comparison and Error Handling")

# Performance comparison between jiter and standard json
test_json = b'{"users": [{"id": 1, "name": "Alice", "scores": [95, 87, 92]}, {"id": 2, "name": "Bob", "scores": [88, 91, 85]}], "total": 2}'
test_json_str = test_json.decode('utf-8')  # String version for json.loads

# Measure jiter performance
start_time = time.perf_counter()
for _ in range(1000):
    jiter_result = jiter.from_json(test_json)
jiter_time = time.perf_counter() - start_time

# Measure standard json performance
start_time = time.perf_counter()
for _ in range(1000):
    json_result = json.loads(test_json_str)
json_time = time.perf_counter() - start_time

print(f"Performance test (1000 iterations):")
print(f"  jiter.from_json: {jiter_time:.4f} seconds")
print(f"  json.loads: {json_time:.4f} seconds")
print(f"  Speed ratio: {json_time/jiter_time:.2f}x faster" if jiter_time > 0 else "  jiter much faster")

# Error handling with invalid JSON
print(f"\nError handling tests:")
invalid_jsons = [
    b'{"incomplete": ',
    b'{"invalid": "quotes\'}',
    b'[1, 2, 3,]',
    b'{"trailing": "comma",}',
    b''
]

for i, invalid_json in enumerate(invalid_jsons):
    # Every input is invalid JSON: jiter must reject it with a ValueError that names the position.
    try:
        result = jiter.from_json(invalid_json)
    except ValueError as e:
        assert "at line 1 column" in str(e), f"Test {i+1}: unexpected ValueError message: {e}"
        print(f"  Test {i+1}: {type(e).__name__} - {str(e)[:50]}...")
    else:
        raise AssertionError(f"Test {i+1}: jiter accepted invalid JSON {invalid_json!r} -> {result!r}")

# Test with very large numbers and edge cases
edge_cases = [
    b'{"large_int": 9223372036854775807}',
    b'{"small_float": 1e-100}',
    b'{"unicode": "Hello \ud83c\udf0d World"}',
    b'{"empty_string": ""}',
    b'{"null_value": null}'
]

print(f"\nEdge case handling:")
for i, edge_json in enumerate(edge_cases):
    try:
        result = jiter.from_json(edge_json)
        print(f"  Test {i+1}: Success - {result}")
    except Exception as e:
        print(f"  Test {i+1}: Error - {e}")
        raise

print("Example 5 completed successfully\n")
```
