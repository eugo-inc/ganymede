## Example 1: jmespath functionality test 1

```python
import jmespath

# Example 1: jmespath.search() - Basic JSON querying
print("Example 1: jmespath.search() - Basic JSON Queries")

# Sample JSON data
data = {
    "users": [
        {"name": "Alice", "age": 30, "city": "New York"},
        {"name": "Bob", "age": 25, "city": "San Francisco"},
        {"name": "Charlie", "age": 35, "city": "Chicago"}
    ],
    "company": {
        "name": "Tech Corp",
        "founded": 2010,
        "employees": 150
    }
}

# Basic field access
result1 = jmespath.search("company.name", data)
print(f"Company name: {result1}")

# Array access by index
result2 = jmespath.search("users[0].name", data)
print(f"First user: {result2}")

# All names from users array
result3 = jmespath.search("users[*].name", data)
print(f"All user names: {result3}")

# Nested field access
result4 = jmespath.search("company.founded", data)
print(f"Company founded: {result4}")

# Array slicing
result5 = jmespath.search("users[:2].city", data)
print(f"First two cities: {result5}")

# Length function
result6 = jmespath.search("length(users)", data)
print(f"Number of users: {result6}")

print("Example 1 completed successfully\n")
```

## Example 2: jmespath functionality test 2

```python
import jmespath

# Example 2: jmespath.compile() - Compiled expressions for performance
print("Example 2: jmespath.compile() - Compiled Expressions")

# Sample data for testing compiled expressions
inventory = {
    "products": [
        {"id": 1, "name": "Laptop", "price": 999.99, "stock": 15},
        {"id": 2, "name": "Mouse", "price": 29.99, "stock": 100},
        {"id": 3, "name": "Keyboard", "price": 79.99, "stock": 50},
        {"id": 4, "name": "Monitor", "price": 299.99, "stock": 8}
    ],
    "categories": ["Electronics", "Accessories"]
}

# Compile frequently used expressions for better performance
compiled_names = jmespath.compile("products[*].name")
compiled_prices = jmespath.compile("products[*].price")
compiled_expensive = jmespath.compile("products[?price > `100`]")

# Use compiled expressions
product_names = compiled_names.search(inventory)
print(f"Product names: {product_names}")

product_prices = compiled_prices.search(inventory)
print(f"All prices: {product_prices}")

expensive_products = compiled_expensive.search(inventory)
print(f"Expensive products: {[p['name'] for p in expensive_products]}")

# Compare compiled vs direct search
direct_result = jmespath.search("products[*].stock", inventory)
compiled_stock = jmespath.compile("products[*].stock")
compiled_result = compiled_stock.search(inventory)

print(f"Direct search stock: {direct_result}")
print(f"Compiled search stock: {compiled_result}")
print(f"Results match: {direct_result == compiled_result}")

# Reuse compiled expression with different data
new_data = {"products": [{"name": "Phone", "price": 699}, {"name": "Tablet", "price": 399}]}
new_names = compiled_names.search(new_data)
print(f"New product names: {new_names}")

print("Example 2 completed successfully\n")
```

## Example 3: jmespath functionality test 3

```python
import jmespath

# Example 3: Advanced filtering and projections
print("Example 3: Advanced Filtering and Projections")

# E-commerce order data
orders = {
    "orders": [
        {"id": "A001", "customer": "Alice", "total": 150.00, "status": "completed", "items": 3},
        {"id": "B002", "customer": "Bob", "total": 89.99, "status": "pending", "items": 2},
        {"id": "C003", "customer": "Charlie", "total": 250.75, "status": "completed", "items": 5},
        {"id": "D004", "customer": "Diana", "total": 45.50, "status": "cancelled", "items": 1},
        {"id": "E005", "customer": "Eve", "total": 199.99, "status": "completed", "items": 4}
    ]
}

# Filter orders by status
completed_orders = jmespath.search("orders[?status == 'completed']", orders)
print(f"Completed orders: {len(completed_orders)} orders")
print(f"Completed order IDs: {[order['id'] for order in completed_orders]}")

# Filter by numeric conditions
high_value_orders = jmespath.search("orders[?total > `150`]", orders)
print(f"High-value orders (>$150): {[order['id'] for order in high_value_orders]}")

# Combined filtering conditions
large_completed_orders = jmespath.search("orders[?status == 'completed' && total > `150`]", orders)
print(f"Large completed orders: {[order['customer'] for order in large_completed_orders]}")

# Projection - select specific fields
customer_totals = jmespath.search("orders[*].{customer: customer, total: total}", orders)
print(f"Customer totals: {customer_totals[:2]}...")  # Show first 2

# Conditional projection
order_summary = jmespath.search("orders[?status != 'cancelled'].{id: id, customer: customer}", orders)
print(f"Active order summary: {order_summary}")

# Nested filtering with projections
detailed_summary = jmespath.search("orders[?items >= `3`].{order_id: id, customer_name: customer, order_total: total}", orders)
print(f"Detailed summary (3+ items): {detailed_summary}")

print("Example 3 completed successfully\n")
```

## Example 4: jmespath functionality test 4

```python
import jmespath

# Example 4: Built-in functions - sort_by, max_by, min_by
print("Example 4: Built-in Functions - sort_by, max_by, min_by")

# Student grade data
students = {
    "students": [
        {"name": "Alice", "grade": 92, "age": 20, "subject": "Math"},
        {"name": "Bob", "grade": 85, "age": 19, "subject": "Physics"},
        {"name": "Charlie", "grade": 98, "age": 21, "subject": "Chemistry"},
        {"name": "Diana", "grade": 88, "age": 20, "subject": "Biology"},
        {"name": "Eve", "grade": 95, "age": 22, "subject": "Math"}
    ]
}

# Sort students by grade (ascending)
sorted_by_grade = jmespath.search("sort_by(students, &grade)", students)
print("Students sorted by grade:")
for student in sorted_by_grade:
    print(f"  {student['name']}: {student['grade']}")

# Sort by age (descending) - use reverse
sorted_by_age_desc = jmespath.search("reverse(sort_by(students, &age))", students)
print(f"\nStudents by age (oldest first): {[s['name'] for s in sorted_by_age_desc]}")

# Find student with highest grade
top_student = jmespath.search("max_by(students, &grade)", students)
print(f"\nTop student: {top_student['name']} with grade {top_student['grade']}")

# Find youngest student
youngest_student = jmespath.search("min_by(students, &age)", students)
print(f"Youngest student: {youngest_student['name']} (age {youngest_student['age']})")

# Combine functions - sort Math students by grade
math_students_sorted = jmespath.search("sort_by(students[?subject == 'Math'], &grade)", students)
print(f"\nMath students by grade: {[s['name'] for s in math_students_sorted]}")

# Get max and min grades
max_grade = jmespath.search("max(students[*].grade)", students)
min_grade = jmespath.search("min(students[*].grade)", students)
print(f"Grade range: {min_grade} - {max_grade}")

# Average using sum and length
total_grade = jmespath.search("sum(students[*].grade)", students)
student_count = jmespath.search("length(students)", students)
average_grade = total_grade / student_count
print(f"Average grade: {average_grade:.1f}")

print("Example 4 completed successfully\n")
```

## Example 5: jmespath functionality test 5

```python
import jmespath

# Example 5: Complex nested queries and multi-select
print("Example 5: Complex Nested Queries and Multi-select")

# Complex API response structure
api_response = {
    "data": {
        "users": [
            {
                "profile": {"name": "Alice", "email": "alice@example.com"},
                "posts": [
                    {"id": 1, "title": "Hello World", "likes": 15, "tags": ["intro", "welcome"]},
                    {"id": 2, "title": "Python Tips", "likes": 42, "tags": ["python", "coding"]}
                ]
            },
            {
                "profile": {"name": "Bob", "email": "bob@example.com"},
                "posts": [
                    {"id": 3, "title": "Data Science", "likes": 28, "tags": ["data", "science"]},
                    {"id": 4, "title": "Machine Learning", "likes": 67, "tags": ["ml", "ai"]}
                ]
            }
        ]
    },
    "metadata": {"total_users": 2, "api_version": "v1.2"}
}

# Multi-select hash - create custom structure
user_summary = jmespath.search("data.users[*].{name: profile.name, email: profile.email, post_count: length(posts)}", api_response)
print("User summary:")
for user in user_summary:
    print(f"  {user['name']} ({user['email']}): {user['post_count']} posts")

# Nested array flattening using pipe expression
all_posts = jmespath.search("data.users[*].posts[] | [*]", api_response)
print(f"\nAll posts flattened: {len(all_posts)} total posts")

# Complex filtering across nested structures - flatten first then filter
popular_posts = jmespath.search("data.users[*].posts[] | [?likes > `30`]", api_response)
print(f"Popular posts (>30 likes): {len(popular_posts)} posts")
for post in popular_posts:
    print(f"  '{post['title']}' - {post['likes']} likes")

# Multi-level projection with filtering
user_popular_posts = jmespath.search("data.users[*].{name: profile.name, popular_posts: posts[?likes > `25`].title}", api_response)
print(f"\nUsers with popular posts:")
for user in user_popular_posts:
    if user['popular_posts']:
        print(f"  {user['name']}: {user['popular_posts']}")

# Pipe expression - chain operations to find most liked post
top_post = jmespath.search("data.users[*].posts[] | max_by(@, &likes)", api_response)
print(f"\nMost liked post overall: '{top_post['title']}' ({top_post['likes']} likes)")

# Complex conditional with metadata
result = jmespath.search("{user_count: metadata.total_users, has_users: length(data.users) > `0`, version: metadata.api_version}", api_response)
print(f"\nAPI info: {result}")

print("Example 5 completed successfully\n")
```
