## Example 1: json5 functionality test 1

```python
import json5

# Example 1: json5.loads() - Parse JSON5 string with extended syntax
print("Example 1: json5.loads() - Parse JSON5 String")

# JSON5 string with extended syntax features
json5_string = """{
    // This is a comment
    name: "John Doe",        // Unquoted keys
    age: 30,
    'address': {             // Single quotes allowed
        street: "123 Main St",
        city: 'New York',    // Mixed quotes
    },
    hobbies: [
        "reading",
        "coding",            // Trailing comma allowed
    ],
    isActive: true,
    salary: 1234.56,         // Regular number (no separators)
    startDate: +Infinity,    // Special numeric values
}"""

# Parse JSON5 string using loads()
try:
    data = json5.loads(json5_string)
    print(f"Parsed data type: {type(data)}")
    print(f"Name: {data['name']}")
    print(f"Age: {data['age']}")
    print(f"City: {data['address']['city']}")
    print(f"Hobbies: {data['hobbies']}")
    print(f"Salary: {data['salary']}")
    print(f"Start Date: {data['startDate']}")
    
    # Test with malformed JSON5
    malformed_json5 = "{name: 'test', age: }"
    try:
        json5.loads(malformed_json5)
    except ValueError as e:
        assert "Unexpected" in str(e), f"unexpected message: {e!r}"
        print(f"Malformed JSON5 correctly rejected: {type(e).__name__}")
    else:
        raise AssertionError(f"malformed JSON5 was accepted: {malformed_json5!r}")
        
except Exception as e:
    print(f"Parse error: {e}")
    raise AssertionError(
        f"json5 example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 completed successfully\n")
```

## Example 2: json5 functionality test 2

```python
import json5
import math

# Example 2: json5.dumps() - Serialize Python objects to JSON5 string
print("Example 2: json5.dumps() - Serialize to JSON5 String")

# Python data structure with various types
data = {
    "user_info": {
        "name": "Alice Johnson",
        "email": "alice@example.com",
        "age": 28
    },
    "preferences": {
        "theme": "dark",
        "notifications": True,
        "language": "en-US"
    },
    "scores": [95.5, 87.2, 92.8, 98.1],
    "metadata": {
        "created_at": "2023-01-15T10:30:00Z",
        "updated_at": None,
        "version": 2.1,
        "pi_value": math.pi,
        "infinity": float('inf')
    }
}

# Serialize using dumps() with different formatting options
json5_output = json5.dumps(data, indent=2)
print(f"Pretty-printed JSON5:")
print(json5_output)

# Compact serialization
compact_output = json5.dumps(data)
print(f"\nCompact JSON5 (first 100 chars): {compact_output[:100]}...")

# Test with special values
special_data = {"nan": float('nan'), "inf": float('inf'), "neg_inf": float('-inf')}
special_output = json5.dumps(special_data)
print(f"\nSpecial values: {special_output}")

# Verify round-trip (dumps -> loads)
reconstructed = json5.loads(json5_output)
print(f"Round-trip successful: {data['user_info']['name'] == reconstructed['user_info']['name']}")

print("Example 2 completed successfully\n")
```

## Example 3: json5 functionality test 3

```python
import json5
import tempfile
import os

# Example 3: json5.load() - Parse JSON5 from file
print("Example 3: json5.load() - Parse JSON5 from File")

# Create a sample JSON5 configuration file
config_content = """{
    // Application configuration
    app: {
        name: "MyApplication",
        version: "1.2.3",
        debug: true,
    },
    database: {
        host: "localhost",
        port: 5432,
        name: "myapp_db",
        ssl: false,
    },
    features: [
        "authentication",
        "logging",
        "caching",        // Trailing comma
    ],
    limits: {
        maxUsers: 1000,      // Regular number (no separators)
        timeout: 30.0,
        retries: +3,         // Plus sign allowed
    }
}"""

# Write to temporary file and read using json5.load()
try:
    with tempfile.NamedTemporaryFile(mode='w', suffix='.json5', delete=False) as f:
        f.write(config_content)
        temp_filename = f.name
    
    # Load JSON5 from file using load()
    with open(temp_filename, 'r') as f:
        config = json5.load(f)
    
    print(f"Loaded config type: {type(config)}")
    print(f"App name: {config['app']['name']}")
    print(f"App version: {config['app']['version']}")
    print(f"Database host: {config['database']['host']}")
    print(f"Features: {config['features']}")
    print(f"Max users: {config['limits']['maxUsers']}")
    print(f"Debug mode: {config['app']['debug']}")
    
    # Clean up
    os.unlink(temp_filename)
    print("Temporary file cleaned up")
    
except Exception as e:
    print(f"File operation error: {e}")
    # Clean up on error
    if 'temp_filename' in locals():
        try:
            os.unlink(temp_filename)
        except:
            pass
            raise
    raise

print("Example 3 completed successfully\n")
```

## Example 4: json5 functionality test 4

```python
import json5
import tempfile
import os
from datetime import datetime

# Example 4: json5.dump() - Write Python objects to JSON5 file
print("Example 4: json5.dump() - Write to JSON5 File")

# Create application settings data
settings_data = {
    "application": {
        "name": "TaskManager",
        "version": "2.1.0",
        "author": "Development Team"
    },
    "ui_settings": {
        "theme": "dark",
        "font_size": 14,
        "show_notifications": True,
        "auto_save": True
    },
    "user_preferences": {
        "default_view": "list",
        "items_per_page": 25,
        "sort_order": "desc",
        "filters": ["active", "priority"]
    },
    "system": {
        "log_level": "INFO",
        "backup_interval": 3600,
        "temp_cleanup": True,
        "created": datetime.now().isoformat(),
        "max_memory": 512.5
    }
}

# Write to JSON5 file using dump()
try:
    with tempfile.NamedTemporaryFile(mode='w', suffix='.json5', delete=False) as f:
        json5.dump(settings_data, f, indent=2)
        temp_filename = f.name
    
    print(f"Data written to: {temp_filename}")
    
    # Read back the file to verify
    with open(temp_filename, 'r') as f:
        file_contents = f.read()
    
    print(f"File size: {len(file_contents)} characters")
    print(f"First 200 characters:\n{file_contents[:200]}...")
    
    # Verify by loading the file back
    with open(temp_filename, 'r') as f:
        loaded_data = json5.load(f)
    
    print(f"Verification: App name = {loaded_data['application']['name']}")
    print(f"Verification: Font size = {loaded_data['ui_settings']['font_size']}")
    
    # Clean up
    os.unlink(temp_filename)
    print("File operations completed successfully")
    
except Exception as e:
    print(f"File write error: {e}")
    raise AssertionError(
        f"json5 example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed successfully\n")
```

## Example 5: json5 functionality test 5

```python
import json5
from decimal import Decimal
from uuid import uuid4

# Example 5: JSON5Encoder class - Custom serialization with encoder
print("Example 5: JSON5Encoder Class - Custom Serialization")

# Create custom encoder for special data types
class CustomJSON5Encoder(json5.JSON5Encoder):
    def default(self, obj):
        if isinstance(obj, Decimal):
            return float(obj)
        elif hasattr(obj, 'isoformat'):  # datetime objects
            return obj.isoformat()
        elif isinstance(obj, set):
            return list(obj)
        return super().default(obj)

# Test data with custom types
test_data = {
    "transaction_id": str(uuid4()),
    "amount": Decimal('123.45'),
    "currency": "USD", 
    "tags": {"urgent", "payment", "verified"},  # set
    "metadata": {
        "precision": Decimal('0.01'),
        "created_by": "system",
        "flags": {"active", "processed"}  # another set
    }
}

# Use custom encoder
try:
    # Serialize with custom encoder
    custom_json = json5.dumps(test_data, cls=CustomJSON5Encoder, indent=2)
    print("Custom encoded JSON5:")
    print(custom_json)
    
    # Test default encoder behavior: it must refuse the Decimal and the sets (measured: the first one it meets is the Decimal)
    try:
        json5.dumps(test_data)
    except TypeError as e:
        assert "is not JSON5 serializable" in str(e), f"unexpected TypeError message: {e!r}"
        print(f"Default encoder correctly failed: {type(e).__name__}")
    else:
        raise AssertionError("json5.dumps with the default encoder serialized a Decimal and sets")
    
    # Verify the custom encoded data can be parsed back
    parsed_data = json5.loads(custom_json)
    print(f"Round-trip verification: {type(parsed_data['amount'])} = {parsed_data['amount']}")
    print(f"Tags converted to: {type(parsed_data['tags'])} = {parsed_data['tags']}")
    
except Exception as e:
    print(f"Custom encoder error: {e}")
    raise AssertionError(
        f"json5 example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 completed successfully\n")
```
