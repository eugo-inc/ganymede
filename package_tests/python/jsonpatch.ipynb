## Example 1: jsonpatch functionality test 1

```python
import jsonpatch

# Example 1: jsonpatch.apply_patch() - Apply JSON patch operations
print("Example 1: jsonpatch.apply_patch() - Apply JSON Patch Operations")

# Original document
original_doc = {
    "user": {
        "name": "Alice",
        "email": "alice@old-domain.com",
        "age": 25
    },
    "settings": {
        "theme": "light",
        "notifications": True
    },
    "tags": ["user", "active"]
}

# JSON Patch operations to modify the document
patch_ops = [
    {"op": "replace", "path": "/user/email", "value": "alice@new-domain.com"},
    {"op": "replace", "path": "/user/age", "value": 26},
    {"op": "add", "path": "/user/department", "value": "Engineering"},
    {"op": "replace", "path": "/settings/theme", "value": "dark"},
    {"op": "add", "path": "/tags/-", "value": "premium"},
    {"op": "remove", "path": "/settings/notifications"}
]

print("Original document:")
print(f"Email: {original_doc['user']['email']}")
print(f"Age: {original_doc['user']['age']}")
print(f"Tags: {original_doc['tags']}")

# Apply patch using apply_patch()
try:
    result = jsonpatch.apply_patch(original_doc, patch_ops)
    
    print("\nAfter applying patch:")
    print(f"Email: {result['user']['email']}")
    print(f"Age: {result['user']['age']}")
    print(f"Department: {result['user']['department']}")
    print(f"Theme: {result['settings']['theme']}")
    print(f"Tags: {result['tags']}")
    print(f"Notifications removed: {'notifications' not in result['settings']}")
    
except Exception as e:
    print(f"Patch application error: {e}")
    raise

print("Example 1 completed successfully\n")
```

## Example 2: jsonpatch functionality test 2

```python
import jsonpatch

# Example 2: jsonpatch.make_patch() - Generate patch from two documents
print("Example 2: jsonpatch.make_patch() - Generate JSON Patch")

# Original document state
original_state = {
    "product": {
        "id": "prod-123",
        "name": "Laptop",
        "price": 999.99,
        "specs": {
            "cpu": "Intel i5",
            "ram": "8GB",
            "storage": "256GB SSD"
        }
    },
    "inventory": {
        "stock": 50,
        "warehouse": "A1"
    }
}

# Updated document state
updated_state = {
    "product": {
        "id": "prod-123",
        "name": "Gaming Laptop",  # Changed
        "price": 1299.99,        # Changed
        "specs": {
            "cpu": "Intel i7",    # Changed
            "ram": "16GB",        # Changed
            "storage": "512GB SSD",  # Changed
            "graphics": "RTX 3060"   # Added
        }
    },
    "inventory": {
        "stock": 25,             # Changed
        "warehouse": "B2",       # Changed
        "reserved": 5            # Added
    }
}

# Generate patch using make_patch()
try:
    patch = jsonpatch.make_patch(original_state, updated_state)
    
    print(f"Generated patch operations: {len(patch.patch)}")
    print("Patch operations:")
    for i, operation in enumerate(patch.patch):
        print(f"  {i+1}. {operation['op']} at '{operation['path']}'")
    
    # Verify patch by applying it to original
    result = jsonpatch.apply_patch(original_state, patch.patch)
    
    print(f"\nVerification:")
    print(f"Names match: {result['product']['name'] == updated_state['product']['name']}")
    print(f"Prices match: {result['product']['price'] == updated_state['product']['price']}")
    print(f"Stock matches: {result['inventory']['stock'] == updated_state['inventory']['stock']}")
    
except Exception as e:
    print(f"Patch generation error: {e}")
    raise AssertionError(
        f"jsonpatch example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed successfully\n")
```

## Example 3: jsonpatch functionality test 3

```python
import jsonpatch

# Example 3: JsonPatch class - Object-oriented patch handling
print("Example 3: JsonPatch Class - Object-oriented Patch Handling")

# Document to be patched
document = {
    "employees": [
        {"id": 1, "name": "John", "department": "IT", "salary": 70000},
        {"id": 2, "name": "Jane", "department": "HR", "salary": 65000}
    ],
    "company": {
        "name": "TechCorp",
        "founded": 2010
    }
}

# Create JsonPatch object with operations
patch_operations = [
    {"op": "add", "path": "/employees/0/skills", "value": ["Python", "Docker"]},
    {"op": "replace", "path": "/employees/1/salary", "value": 68000},
    {"op": "add", "path": "/employees/-", "value": {"id": 3, "name": "Bob", "department": "DevOps", "salary": 75000}},
    {"op": "replace", "path": "/company/name", "value": "TechCorp Solutions"},
    {"op": "add", "path": "/company/employees_count", "value": 3}
]

try:
    # Create JsonPatch instance
    patch = jsonpatch.JsonPatch(patch_operations)
    
    print(f"JsonPatch object created with {len(patch.patch)} operations")
    print(f"Patch type: {type(patch)}")
    
    # Apply patch to document
    result = patch.apply(document)
    
    print(f"\nResults after applying patch:")
    print(f"Company name: {result['company']['name']}")
    print(f"Employee count: {result['company']['employees_count']}")
    print(f"Total employees in array: {len(result['employees'])}")
    print(f"John's skills: {result['employees'][0].get('skills', 'None')}")
    print(f"Jane's new salary: {result['employees'][1]['salary']}")
    print(f"New employee: {result['employees'][2]['name']} ({result['employees'][2]['department']})")
    
    # Get patch as JSON string
    patch_json = patch.to_string()
    print(f"\nPatch as JSON string length: {len(patch_json)} characters")
    
except Exception as e:
    print(f"JsonPatch error: {e}")
    raise AssertionError(
        f"jsonpatch example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 completed successfully\n")
```

## Example 4: jsonpatch functionality test 4

```python
import jsonpatch
import json

# Example 4: JsonPatch.from_string() - Create patch from JSON string
print("Example 4: JsonPatch.from_string() - Create Patch from JSON String")

# Configuration document
config = {
    "api": {
        "version": "v1",
        "endpoint": "https://api.example.com",
        "timeout": 30
    },
    "features": {
        "caching": True,
        "logging": False,
        "debug": True
    },
    "limits": {
        "requests_per_minute": 100
    }
}

# JSON patch string (could come from API, file, etc.)
patch_json_string = json.dumps([
    {"op": "replace", "path": "/api/version", "value": "v2"},
    {"op": "replace", "path": "/api/endpoint", "value": "https://api-v2.example.com"},
    {"op": "replace", "path": "/api/timeout", "value": 45},
    {"op": "replace", "path": "/features/logging", "value": True},
    {"op": "replace", "path": "/features/debug", "value": False},
    {"op": "add", "path": "/features/monitoring", "value": True},
    {"op": "replace", "path": "/limits/requests_per_minute", "value": 200},
    {"op": "add", "path": "/limits/max_payload_size", "value": "10MB"}
])

print("Original configuration:")
print(f"API version: {config['api']['version']}")
print(f"Logging enabled: {config['features']['logging']}")
print(f"Rate limit: {config['limits']['requests_per_minute']}")

try:
    # Create JsonPatch object from JSON string using JsonPatch.from_string()
    patch = jsonpatch.JsonPatch.from_string(patch_json_string)
    
    print(f"\nPatch created from string: {type(patch)}")
    print(f"Number of operations: {len(patch.patch)}")
    
    # Apply the patch
    updated_config = patch.apply(config)
    
    print(f"\nUpdated configuration:")
    print(f"API version: {updated_config['api']['version']}")
    print(f"API endpoint: {updated_config['api']['endpoint']}")
    print(f"Logging enabled: {updated_config['features']['logging']}")
    print(f"Debug mode: {updated_config['features']['debug']}")
    print(f"Monitoring: {updated_config['features']['monitoring']}")
    print(f"Rate limit: {updated_config['limits']['requests_per_minute']}")
    
except Exception as e:
    print(f"from_string error: {e}")
    raise AssertionError(
        f"jsonpatch example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed successfully\n")
```

## Example 5: jsonpatch functionality test 5

```python
import jsonpatch

# Example 5: Individual patch operations - AddOperation, RemoveOperation, etc.
print("Example 5: Individual Patch Operations - Direct Operation Classes")

# Shopping cart document
cart = {
    "user_id": "user-456",
    "items": [
        {"id": "item-1", "name": "Book", "price": 15.99, "quantity": 2},
        {"id": "item-2", "name": "Pen", "price": 2.50, "quantity": 5}
    ],
    "total": 41.48,
    "status": "active"
}

print("Original cart:")
print(f"Items count: {len(cart['items'])}")
print(f"Total: ${cart['total']}")
print(f"Status: {cart['status']}")

try:
    # Create individual operation instances
    add_op = jsonpatch.AddOperation({"path": "/items/-", "value": {"id": "item-3", "name": "Notebook", "price": 8.99, "quantity": 1}})
    replace_op = jsonpatch.ReplaceOperation({"path": "/total", "value": 50.47})
    copy_op = jsonpatch.CopyOperation({"path": "/backup_status", "from": "/status"})
    move_op = jsonpatch.MoveOperation({"path": "/cart_status", "from": "/status"})
    test_op = jsonpatch.TestOperation({"path": "/user_id", "value": "user-456"})
    
    print(f"\nCreated operations:")
    print(f"Add operation: {type(add_op).__name__}")
    print(f"Replace operation: {type(replace_op).__name__}")
    print(f"Copy operation: {type(copy_op).__name__}")
    print(f"Move operation: {type(move_op).__name__}")
    print(f"Test operation: {type(test_op).__name__}")
    
    # Apply operations sequentially
    cart = add_op.apply(cart)
    cart = replace_op.apply(cart)
    cart = copy_op.apply(cart)
    cart = test_op.apply(cart)  # Test operation (validates value)
    cart = move_op.apply(cart)
    
    print(f"\nAfter applying individual operations:")
    print(f"Items count: {len(cart['items'])}")
    print(f"New item: {cart['items'][2]['name']}")
    print(f"Updated total: ${cart['total']}")
    print(f"Backup status: {cart.get('backup_status', 'None')}")
    print(f"Cart status: {cart.get('cart_status', 'None')}")
    print(f"Original status field exists: {'status' in cart}")
    
except Exception as e:
    print(f"Individual operations error: {e}")
    raise AssertionError(
        f"jsonpatch example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 completed successfully\n")
```
