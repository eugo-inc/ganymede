## Example 1: jsonpath_ng functionality test 1

```python
import jsonpath_ng

# Example 1: jsonpath_ng.parse() - Parse JSONPath expressions
print("Example 1: jsonpath_ng.parse() - Parse JSONPath Expressions")

# Sample JSON data structure
data = {
    "store": {
        "book": [
            {"title": "The Great Gatsby", "author": "F. Scott Fitzgerald", "price": 10.99, "category": "fiction"},
            {"title": "To Kill a Mockingbird", "author": "Harper Lee", "price": 12.99, "category": "fiction"},
            {"title": "1984", "author": "George Orwell", "price": 13.99, "category": "dystopian"},
            {"title": "Pride and Prejudice", "author": "Jane Austen", "price": 9.99, "category": "romance"}
        ],
        "bicycle": {
            "color": "red",
            "price": 19.95
        }
    }
}

# Parse different JSONPath expressions using parse()
try:
    # Basic path parsing
    book_path = jsonpath_ng.parse("$.store.book[*].title")
    print(f"Parsed JSONPath type: {type(book_path)}")
    
    # Extract all book titles
    matches = book_path.find(data)
    titles = [match.value for match in matches]
    print(f"Book titles: {titles}")
    
    # Parse and use different expressions
    author_path = jsonpath_ng.parse("$.store.book[0].author")
    first_author = author_path.find(data)[0].value
    print(f"First author: {first_author}")
    
    # Parse complex expression
    price_path = jsonpath_ng.parse("$.store..price")
    prices = [match.value for match in price_path.find(data)]
    print(f"All prices: {prices}")
    
except Exception as e:
    print(f"Parse error: {e}")
    raise AssertionError(
        f"jsonpath_ng example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 completed successfully\n")
```

## Example 2: jsonpath_ng functionality test 2

```python
import jsonpath_ng

# Example 2: JSONPath.find() - Find matching elements with context
print("Example 2: JSONPath.find() - Find with Context Information")

# Employee data structure
employees = {
    "company": "TechCorp",
    "departments": {
        "engineering": {
            "staff": [
                {"id": 101, "name": "Alice Johnson", "role": "Senior Developer", "salary": 95000},
                {"id": 102, "name": "Bob Smith", "role": "DevOps Engineer", "salary": 90000}
            ]
        },
        "marketing": {
            "staff": [
                {"id": 201, "name": "Carol Davis", "role": "Marketing Manager", "salary": 85000},
                {"id": 202, "name": "David Wilson", "role": "Content Writer", "salary": 65000}
            ]
        }
    }
}

try:
    # Parse JSONPath expression for finding staff
    staff_path = jsonpath_ng.parse("$.departments.*.staff[*]")
    
    # Use find() to get matches with context information
    matches = staff_path.find(employees)
    
    print(f"Found {len(matches)} staff members:")
    for i, match in enumerate(matches):
        print(f"  {i+1}. Name: {match.value['name']}")
        print(f"     Role: {match.value['role']}")
        print(f"     Path: {match.full_path}")
        print(f"     Context: {type(match.context)}")
    
    # Find managers by filtering results manually
    all_staff_matches = staff_path.find(employees)
    managers = [match for match in all_staff_matches if "Manager" in match.value['role']]
    
    if managers:
        manager = managers[0]
        print(f"\nFound manager: {manager.value['name']} (${manager.value['salary']})")
        print(f"Manager path: {manager.full_path}")
        print(f"Manager department: {manager.context.value}")
    
    # Find specific employee by name
    alice_matches = [match for match in all_staff_matches if match.value['name'] == "Alice Johnson"]
    if alice_matches:
        alice = alice_matches[0]
        print(f"\nAlice's details:")
        print(f"  Role: {alice.value['role']}")
        print(f"  Salary: ${alice.value['salary']}")
    
except Exception as e:
    print(f"Find error: {e}")
    raise AssertionError(
        f"jsonpath_ng example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed successfully\n")
```

## Example 3: jsonpath_ng functionality test 3

```python
import jsonpath_ng

# Example 3: Extract all matching values using find() 
print("Example 3: Extract All Matching Values Using find()")

# API response data
api_response = {
    "status": "success",
    "data": {
        "users": [
            {
                "profile": {"name": "John Doe", "email": "john@example.com"},
                "settings": {"notifications": True, "theme": "dark"},
                "posts": [
                    {"id": 1, "title": "Hello World", "likes": 15, "tags": ["intro", "welcome"]},
                    {"id": 2, "title": "Getting Started", "likes": 32, "tags": ["tutorial", "guide"]}
                ]
            },
            {
                "profile": {"name": "Jane Smith", "email": "jane@example.com"},  
                "settings": {"notifications": False, "theme": "light"},
                "posts": [
                    {"id": 3, "title": "Advanced Tips", "likes": 45, "tags": ["advanced", "tips"]},
                    {"id": 4, "title": "Best Practices", "likes": 67, "tags": ["best-practices", "guide"]}
                ]
            }
        ]
    }
}

try:
    # Extract all user names using find() and list comprehension
    names_path = jsonpath_ng.parse("$.data.users[*].profile.name")
    all_names = [match.value for match in names_path.find(api_response)]
    print(f"All user names: {all_names}")
    
    # Extract all post titles
    titles_path = jsonpath_ng.parse("$.data.users[*].posts[*].title")
    all_titles = [match.value for match in titles_path.find(api_response)]
    print(f"All post titles: {all_titles}")
    
    # Extract all tags (nested arrays)
    tags_path = jsonpath_ng.parse("$.data.users[*].posts[*].tags[*]")
    all_tags = [match.value for match in tags_path.find(api_response)]
    unique_tags = list(set(all_tags))
    print(f"All unique tags: {unique_tags}")
    
    # Extract notification settings
    notifications_path = jsonpath_ng.parse("$.data.users[*].settings.notifications")
    notification_settings = [match.value for match in notifications_path.find(api_response)]
    print(f"Notification settings: {notification_settings}")
    
    # Extract and process likes data
    likes_path = jsonpath_ng.parse("$.data.users[*].posts[*].likes")
    likes_values = [match.value for match in likes_path.find(api_response)]
    print(f"All likes: {likes_values}")
    print(f"Total likes: {sum(likes_values)}")
    print(f"Average likes: {sum(likes_values)/len(likes_values):.1f}")
    
except Exception as e:
    print(f"Value extraction error: {e}")
    raise AssertionError(
        f"jsonpath_ng example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 completed successfully\n")
```

## Example 4: jsonpath_ng functionality test 4

```python
import jsonpath_ng

# Example 4: JSONPath.update() - Update values in JSON structure
print("Example 4: JSONPath.update() - Update Values in JSON Structure")

# Configuration data that needs updating
config = {
    "application": {
        "name": "MyApp",
        "version": "1.0.0",
        "settings": {
            "debug": True,
            "log_level": "INFO",
            "features": {
                "caching": False,
                "monitoring": False
            }
        }
    },
    "servers": [
        {"name": "web-1", "status": "active", "port": 8080},
        {"name": "web-2", "status": "maintenance", "port": 8081},
        {"name": "api-1", "status": "active", "port": 3000}
    ]
}

print("Original configuration:")
print(f"Debug mode: {config['application']['settings']['debug']}")
print(f"Caching enabled: {config['application']['settings']['features']['caching']}")
print(f"Server statuses: {[s['status'] for s in config['servers']]}")

try:
    # Update application version
    version_path = jsonpath_ng.parse("$.application.version")
    updated_config = version_path.update(config, "2.1.0")
    
    # Update debug mode
    debug_path = jsonpath_ng.parse("$.application.settings.debug")
    updated_config = debug_path.update(updated_config, False)
    
    # Enable caching feature
    caching_path = jsonpath_ng.parse("$.application.settings.features.caching")
    updated_config = caching_path.update(updated_config, True)
    
    # Update all server statuses to active
    server_status_path = jsonpath_ng.parse("$.servers[*].status")
    updated_config = server_status_path.update(updated_config, "active")
    
    # Update specific server port by index
    api_port_path = jsonpath_ng.parse("$.servers[2].port")  # Third server (api-1)
    updated_config = api_port_path.update(updated_config, 4000)
    
    # Update server names
    first_server_name = jsonpath_ng.parse("$.servers[0].name")
    updated_config = first_server_name.update(updated_config, "web-primary")
    
    print(f"\nAfter updates:")
    print(f"Version: {updated_config['application']['version']}")
    print(f"Debug mode: {updated_config['application']['settings']['debug']}")
    print(f"Caching enabled: {updated_config['application']['settings']['features']['caching']}")
    print(f"Server statuses: {[s['status'] for s in updated_config['servers']]}")
    print(f"API server port: {updated_config['servers'][2]['port']}")
    print(f"First server name: {updated_config['servers'][0]['name']}")
    
except Exception as e:
    print(f"Update error: {e}")
    raise AssertionError(
        f"jsonpath_ng example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed successfully\n")
```

## Example 5: jsonpath_ng functionality test 5

```python
import jsonpath_ng
from jsonpath_ng.ext import parse as ext_parse

# Example 5: Extended JSONPath expressions - Advanced filtering and operations  
print("Example 5: Extended JSONPath Expressions - Advanced Filtering")

# E-commerce order data
orders = {
    "orders": [
        {"id": "ORD-001", "customer": "Alice", "amount": 150.50, "status": "completed", "items": 3, "date": "2023-01-15"},
        {"id": "ORD-002", "customer": "Bob", "amount": 89.99, "status": "pending", "items": 2, "date": "2023-01-16"},
        {"id": "ORD-003", "customer": "Charlie", "amount": 320.00, "status": "completed", "items": 5, "date": "2023-01-17"},
        {"id": "ORD-004", "customer": "Diana", "amount": 45.25, "status": "cancelled", "items": 1, "date": "2023-01-18"},
        {"id": "ORD-005", "customer": "Eve", "amount": 199.99, "status": "completed", "items": 4, "date": "2023-01-19"}
    ]
}

try:
    # Use extended parser for advanced filtering
    # Filter orders by amount greater than 100
    high_value_path = ext_parse("$.orders[?@.amount > 100]")
    high_value_orders = [match.value for match in high_value_path.find(orders)]
    print(f"High-value orders (>$100): {len(high_value_orders)}")
    for order in high_value_orders:
        print(f"  {order['id']}: ${order['amount']} ({order['customer']})")
    
    # Filter completed orders with multiple items
    multi_item_completed = ext_parse("$.orders[?@.status == 'completed' & @.items >= 3]")
    filtered_orders = [match.value for match in multi_item_completed.find(orders)]
    print(f"\nCompleted orders with 3+ items: {len(filtered_orders)}")
    
    # Get customer names from pending orders
    pending_customers = ext_parse("$.orders[?@.status == 'pending'].customer")
    pending_names = [match.value for match in pending_customers.find(orders)]
    print(f"Pending order customers: {pending_names}")
    
    # Calculate total amount for completed orders
    completed_amounts = ext_parse("$.orders[?@.status == 'completed'].amount")
    amounts = [match.value for match in completed_amounts.find(orders)]
    total_completed = sum(amounts)
    print(f"Total completed order value: ${total_completed:.2f}")
    
    # Find orders by customer name pattern (simple string matching)
    alice_orders = ext_parse("$.orders[?@.customer == 'Alice'].id")
    alice_order_ids = [match.value for match in alice_orders.find(orders)]
    print(f"Alice's order IDs: {alice_order_ids}")
    
    # Complex filter: high-value completed orders
    premium_orders = ext_parse("$.orders[?@.amount > 150 & @.status == 'completed']")
    premium_results = [match.value for match in premium_orders.find(orders)]
    print(f"Premium completed orders: {[o['id'] for o in premium_results]}")
    
except Exception as e:
    print(f"Extended JSONPath error: {e}")
    raise AssertionError(
        f"jsonpath_ng example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 completed successfully\n")
```
