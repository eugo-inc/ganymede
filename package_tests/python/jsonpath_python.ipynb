## Example 1: Basic JSONPath Queries and Path Expressions

Basic usage of JSONPath expressions for querying JSON data structures.

```python
# Example 1: Basic JSONPath queries and path expressions
from jsonpath import JSONPath

try:
    print("=== Basic JSONPath Query Operations ===")

    # Sample bookstore data for testing
    sample_data = {
        "store": {
            "book": [
                {"category": "reference", "author": "Nigel Rees", "title": "Sayings of the Century", "price": 8.95},
                {"category": "fiction", "author": "Evelyn Waugh", "title": "Sword of Honour", "price": 12.99},
                {"category": "fiction", "author": "Herman Melville", "title": "Moby Dick", "isbn": "0-553-21311-3", "price": 8.99},
                {"category": "fiction", "author": "J. R. R. Tolkien", "title": "The Lord of the Rings", "isbn": "0-395-19395-8", "price": 22.99}
            ],
            "bicycle": {"color": "red", "price": 19.95}
        }
    }

    # Basic path expressions using JSONPath class
    queries = {
        "Root": "$",
        "Store object": "$.store",
        "All books": "$.store.book[*]",
        "First book": "$.store.book[0]",
        "Last book": "$.store.book[-1]",
        "Book titles": "$.store.book[*].title",
        "All authors": "$.store.book[*].author",
        "Bicycle info": "$.store.bicycle"
    }

    for query_name, expression in queries.items():
        try:
            # Use JSONPath class with parse() method
            results = JSONPath(expression).parse(sample_data)
            print(f"\n{query_name} ({expression}):")
            if results:
                if len(results) == 1:
                    print(f"  Result: {results[0]}")
                else:
                    print(f"  Found {len(results)} items:")
                    for i, result in enumerate(results[:3]):  # Show first 3
                        print(f"    [{i}]: {result}")
                    if len(results) > 3:
                        print(f"    ... and {len(results) - 3} more")
            else:
                print("  No results found")

        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== Array Index and Slice Operations ===")

    # Array operations using slicing
    array_queries = {
        "First two books": "$.store.book[0:2]",
        "Books from index 1": "$.store.book[1:]",
        "Every other book": "$.store.book[::2]",
        "Book at index 1": "$.store.book[1]",
        "Last two books": "$.store.book[-2:]",
        "Reverse order": "$.store.book[::-1]"
    }

    for query_name, expression in array_queries.items():
        try:
            results = JSONPath(expression).parse(sample_data)
            print(f"\n{query_name} ({expression}):")
            if results:
                print(f"  Found {len(results)} items")
                for result in results:
                    if isinstance(result, dict) and 'title' in result:
                        print(f"    - {result['title']}")
                    else:
                        print(f"    - {result}")
            else:
                print("  No results found")

        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== Wildcard and Multi-field Selection ===")

    # Test wildcard and multi-field extraction
    wildcard_tests = [
        ("All store items", "$.store.*"),
        ("All books wildcard", "$.store.book[*]"),
        ("Multiple fields", "$[store,bicycle]"),
        ("Book title and author", "$.store.book[*].(title,author)")
    ]

    for test_name, expression in wildcard_tests:
        try:
            result = JSONPath(expression).parse(sample_data)
            print(f"\n{test_name} ({expression}):")
            print(f"  Found {len(result) if result else 0} results")
            if result and len(result) <= 3:
                for item in result:
                    if isinstance(item, dict):
                        print(f"    - Dict with {len(item)} keys")
                    elif isinstance(item, list):
                        print(f"    - List with {len(item)} items")
                    else:
                        print(f"    - {item}")
        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\nExample 1: Basic JSONPath queries test passed")

except Exception as e:
    print(f"Example 1: Error in basic queries test: {e}")
    import traceback
    traceback.print_exc()
    raise
```

## Example 2: Filtering and Conditional Queries

Demonstrate advanced filtering with conditions and comparison operators.

```python
# Example 2: Filtering and conditional queries
from jsonpath import JSONPath

try:
    print("=== Conditional Filtering Queries ===")

    # Sample data for filtering
    sample_data = {
        "store": {
            "book": [
                {"category": "reference", "author": "Nigel Rees", "title": "Sayings of the Century", "price": 8.95},
                {"category": "fiction", "author": "Evelyn Waugh", "title": "Sword of Honour", "price": 12.99},
                {"category": "fiction", "author": "Herman Melville", "title": "Moby Dick", "isbn": "0-553-21311-3", "price": 8.99},
                {"category": "fiction", "author": "J. R. R. Tolkien", "title": "The Lord of the Rings", "isbn": "0-395-19395-8", "price": 22.99}
            ],
            "bicycle": {"color": "red", "price": 19.95}
        }
    }

    # Filtering with conditions using ?() operator - note: use double quotes for string values
    filter_queries = {
        "Expensive books": '$.store.book[?(@.price>10)]',
        "Fiction books": '$.store.book[?(@.category=="fiction")]',
        "Cheap books": '$.store.book[?(@.price<10)]',
        "Books by Melville": '$.store.book[?(@.author=="Herman Melville")]',
        "Mid-priced books": '$.store.book[?(@.price>8 and @.price<15)]'
    }

    for query_name, expression in filter_queries.items():
        try:
            results = JSONPath(expression).parse(sample_data)
            print(f"\n{query_name} ({expression}):")
            if results:
                print(f"  Found {len(results)} matching items:")
                for result in results:
                    if isinstance(result, dict):
                        title = result.get('title', 'Unknown')
                        price = result.get('price', 'N/A')
                        category = result.get('category', 'N/A')
                        print(f"    - '{title}' (${price}, {category})")
                    else:
                        print(f"    - {result}")
            else:
                print("  No matching items found")

        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== Complex Multi-condition Filters ===")

    # Create more complex test data
    complex_data = {
        "products": [
            {"name": "Laptop", "price": 999, "category": "electronics", "in_stock": True, "rating": 4.5},
            {"name": "Phone", "price": 699, "category": "electronics", "in_stock": False, "rating": 4.2},
            {"name": "Book", "price": 15, "category": "media", "in_stock": True, "rating": 4.8},
            {"name": "Headphones", "price": 199, "category": "electronics", "in_stock": True, "rating": 4.0},
            {"name": "Tablet", "price": 449, "category": "electronics", "in_stock": True, "rating": 4.3}
        ],
        "users": [
            {"name": "Alice", "age": 25, "premium": True, "orders": 15},
            {"name": "Bob", "age": 30, "premium": False, "orders": 3},
            {"name": "Charlie", "age": 35, "premium": True, "orders": 22}
        ]
    }

    complex_filters = {
        "High-rated electronics": '$.products[?(@.category=="electronics" and @.rating>4.2)]',
        "Expensive items": '$.products[?(@.price>500)]',
        "Affordable products": '$.products[?(@.price<200)]',
        "Active users": '$.users[?(@.orders>10)]'
    }

    for query_name, expression in complex_filters.items():
        try:
            results = JSONPath(expression).parse(complex_data)
            print(f"\n{query_name}:")
            if results:
                print(f"  Found {len(results)} items:")
                for result in results:
                    if 'name' in result:
                        if 'price' in result:  # Product
                            print(f"    - {result['name']}: ${result['price']}")
                        else:  # User
                            print(f"    - {result['name']}: {result['orders']} orders")
            else:
                print("  No matches found")

        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== Comparison Operators ===")

    # Test various comparison operators
    comparison_queries = {
        "Greater than 500": '$.products[?(@.price>500)]',
        "Less than or equal 4.3": '$.products[?(@.rating<=4.3)]',
        "Not equal to media": '$.products[?(@.category!="media")]',
        "Price between 100 and 800": '$.products[?(@.price>100 and @.price<800)]',
        "High rating or cheap": '$.products[?(@.rating>4.5 or @.price<50)]'
    }

    for query_name, expression in comparison_queries.items():
        try:
            results = JSONPath(expression).parse(complex_data)
            print(f"\n{query_name}:")
            if results:
                print(f"  Matched {len(results)} items:")
                for result in results[:2]:  # Show first 2
                    if isinstance(result, dict) and 'name' in result:
                        price = result.get('price', 'N/A')
                        rating = result.get('rating', 'N/A')
                        print(f"    - {result['name']}: ${price}, rating {rating}")
            else:
                print("  No matches")
        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== OR Operator for Multiple Values ===")

    # Test OR operator for selecting multiple specific values
    or_queries = {
        "Fiction or reference books": '$.store.book[?(@.category=="fiction" or @.category=="reference")]',
        "Books by Melville or Tolkien": '$.store.book[?(@.author=="Herman Melville" or @.author=="J. R. R. Tolkien")]',
        "Very cheap or very expensive": '$.store.book[?(@.price<9 or @.price>20)]'
    }

    for query_name, expression in or_queries.items():
        try:
            results = JSONPath(expression).parse(sample_data)
            print(f"\n{query_name}:")
            if results:
                print(f"  Found {len(results)} matches")
                for result in results:
                    if isinstance(result, dict):
                        print(f"    - {result.get('title', result.get('name', 'Unknown'))}")
            else:
                print("  No matches")
        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== Filtering Summary ===")

    # Summary of filter capabilities
    filter_features = [
        ("?(@.field>value)", "Greater than comparison"),
        ("?(@.field<value)", "Less than comparison"),
        ('?(@.field=="value")', "String equality (use double quotes!)"),
        ('?(@.field!="value")', "String inequality"),
        ("?(@.a>x and @.b<y)", "AND logical operator"),
        ("?(@.a>x or @.b<y)", "OR logical operator"),
        ("?(@.field)", "Field existence check")
    ]

    print("Filter operators demonstrated:")
    for syntax, description in filter_features:
        print(f"  {syntax:25} - {description}")

    print("\nExample 2: Filtering and conditional queries test passed")

except Exception as e:
    print(f"Example 2: Error in filtering test: {e}")
    import traceback
    traceback.print_exc()
    raise
```

## Example 3: Recursive Descent and Wildcard Patterns

Demonstrate recursive descent operators and wildcard matching patterns.

```python
# Example 3: Recursive descent and wildcard patterns
from jsonpath import JSONPath

try:
    print("=== Recursive Descent Operators ===")

    # Create nested test data for recursive operations
    nested_data = {
        "company": {
            "name": "TechCorp",
            "departments": {
                "engineering": {
                    "employees": [
                        {"name": "Alice", "salary": 90000, "skills": ["Python", "JavaScript"]},
                        {"name": "Bob", "salary": 85000, "skills": ["Java", "C++"]}
                    ],
                    "manager": {"name": "Charlie", "salary": 120000, "skills": ["Leadership", "Python"]}
                },
                "marketing": {
                    "employees": [
                        {"name": "Diana", "salary": 75000, "skills": ["Marketing", "Analytics"]},
                        {"name": "Eve", "salary": 70000, "skills": ["Content", "Design"]}
                    ],
                    "manager": {"name": "Frank", "salary": 95000, "skills": ["Strategy", "Leadership"]}
                }
            },
            "metadata": {
                "founded": 2010,
                "locations": ["New York", "San Francisco"],
                "stats": {
                    "employees_count": 50,
                    "revenue": 10000000
                }
            }
        }
    }

    # Recursive descent queries using .. operator
    recursive_queries = {
        "All names anywhere": "$..name",
        "All salaries": "$..salary",
        "All skills": "$..skills",
        "All employees objects": "$..employees",
        "All manager objects": "$..manager",
        "All locations": "$..locations"
    }

    for query_name, expression in recursive_queries.items():
        try:
            results = JSONPath(expression).parse(nested_data)
            print(f"\n{query_name} ({expression}):")
            if results:
                print(f"  Found {len(results)} items:")
                for i, result in enumerate(results[:4]):  # Show first 4
                    if isinstance(result, list) and len(result) > 0:
                        if len(result) <= 2:
                            print(f"    [{i}]: {result}")
                        else:
                            print(f"    [{i}]: List with {len(result)} items: {result[:2]}...")
                    elif isinstance(result, dict):
                        print(f"    [{i}]: Dict with keys: {list(result.keys())[:3]}")
                    else:
                        print(f"    [{i}]: {result}")
                if len(results) > 4:
                    print(f"    ... and {len(results) - 4} more")
            else:
                print("  No results found")

        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== Wildcard (*) Operator ===")

    # Wildcard patterns using * operator
    wildcard_queries = {
        "All department objects": "$.company.departments.*",
        "All direct children of company": "$.company.*",
        "All employee arrays": "$.company.departments.*.employees",
        "Manager names in all departments": "$.company.departments.*.manager.name"
    }

    for query_name, expression in wildcard_queries.items():
        try:
            results = JSONPath(expression).parse(nested_data)
            print(f"\n{query_name} ({expression}):")
            if results:
                print(f"  Found {len(results)} items:")
                for result in results[:3]:  # Show first 3
                    if isinstance(result, dict):
                        if 'name' in result:
                            print(f"    - Name: {result['name']}")
                        else:
                            print(f"    - Dict with {len(result)} keys")
                    elif isinstance(result, list):
                        print(f"    - List with {len(result)} items")
                    else:
                        print(f"    - {result}")
            else:
                print("  No results")
        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== Combining Recursive Descent with Array Access ===")

    # Complex navigation patterns combining recursive and array access
    navigation_tests = {
        "Skills of all employees (recursive)": "$..employees[*].skills",
        "First employee in each department": "$.company.departments.*.employees[0]",
        "All manager salaries": "$.company.departments.*.manager.salary",
        "All names recursively": "$..name"
    }

    for query_name, expression in navigation_tests.items():
        try:
            results = JSONPath(expression).parse(nested_data)
            print(f"\n{query_name}:")
            if results:
                print(f"  Found {len(results)} results:")
                for result in results[:5]:  # Show first 5
                    if isinstance(result, dict):
                        print(f"    - {result.get('name', result)}")
                    elif isinstance(result, list):
                        print(f"    - {result}")
                    else:
                        print(f"    - {result}")
            else:
                print("  No results found")
        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== Simple Filtering on Direct Paths ===")

    # Demonstrate filtering on direct (non-recursive) paths only
    filter_queries = {
        "Engineering employees with high salary": "$.company.departments.engineering.employees[?(@.salary>80000)]",
        "Marketing employees": "$.company.departments.marketing.employees[*]"
    }

    for query_name, expression in filter_queries.items():
        try:
            results = JSONPath(expression).parse(nested_data)
            print(f"\n{query_name}:")
            if results:
                print(f"  Matched {len(results)} items:")
                for result in results[:3]:
                    if isinstance(result, dict):
                        name = result.get('name', 'Unknown')
                        salary = result.get('salary', 'N/A')
                        print(f"    - {name}: ${salary}")
                    else:
                        print(f"    - {result}")
            else:
                print("  No matches")
        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== Pattern Matching Summary ===")

    # Summary of pattern types
    patterns_tested = [
        ("$..key", "Recursive descent - find 'key' anywhere"),
        ("$.level1.*", "Wildcard - all direct children of level1"),
        ("$..arr[*]", "All elements in any array named 'arr'"),
        ("$.path.arr[?(@.field>val)]", "Filter on direct path (not recursive)"),
        ("$.parent.*.child", "Multi-level wildcard navigation")
    ]

    print("Pattern types demonstrated:")
    for pattern, description in patterns_tested:
        print(f"  {pattern:30} - {description}")

    # Demonstrate wildcard combinations
    wildcard_examples = [
        ("$.*", "All properties of root"),
        ("$.company.*", "All properties of company"),
        ("$..*", "Everything recursively"),
        ("$..employees[*]", "All elements in all employees arrays")
    ]

    print("\nWildcard examples:")
    for expr, desc in wildcard_examples:
        try:
            result = JSONPath(expr).parse(nested_data)
            count = len(result) if result else 0
            print(f"  {expr:30} -> {count:3} results ({desc})")
        except Exception as e:
            print(f"  {expr:30} -> Error: {type(e).__name__}")
            raise

    print("\n=== Practical Use Cases ===")

    # Real-world examples - simple recursive descent only
    practical = {
        "Extract all name fields": "$..name",
        "Find all salary values": "$..salary",
        "Get all employee arrays": "$..employees",
        "All skill lists": "$..skills"
    }

    print("Practical recursive descent examples:")
    for desc, expr in practical.items():
        try:
            result = JSONPath(expr).parse(nested_data)
            count = len(result) if result else 0
            print(f"  {desc:30} -> {count} found")
        except:
            print(f"  {desc:30} -> 0 found")
            raise

    print("\n=== Array Indexing ===")
    
    # Show array indexing capabilities
    indexing_tests = {
        "First engineering employee": "$.company.departments.engineering.employees[0]",
        "Last engineering employee": "$.company.departments.engineering.employees[-1]",
        "First two employees": "$.company.departments.engineering.employees[0:2]"
    }
    
    print("Array indexing examples:")
    for desc, expr in indexing_tests.items():
        try:
            result = JSONPath(expr).parse(nested_data)
            if result:
                if isinstance(result, list) and len(result) == 1:
                    print(f"  {desc}: {result[0].get('name', 'N/A')}")
                elif isinstance(result, list):
                    names = [r.get('name', 'N/A') for r in result if isinstance(r, dict)]
                    print(f"  {desc}: {names}")
        except Exception as e:
            print(f"  {desc}: Error - {type(e).__name__}")
            raise

    print("\nExample 3: Recursive descent and wildcard patterns test passed")

except Exception as e:
    print(f"Example 3: Error in recursive/wildcard test: {e}")
    import traceback
    traceback.print_exc()
    raise
```

## Example 4: Advanced Functions and Expression Evaluation

Demonstrate advanced JSONPath functions and expression evaluation capabilities.

```python
# Example 4: Sorting and field extraction features
from jsonpath import JSONPath
import json

try:
    print("=== Sorting with /() Operator ===")

    # Sample data for sorting
    books_data = {
        "store": {
            "book": [
                {"title": "Moby Dick", "author": "Herman Melville", "price": 8.99, "year": 1851},
                {"title": "The Lord of the Rings", "author": "J. R. R. Tolkien", "price": 22.99, "year": 1954},
                {"title": "Sayings of the Century", "author": "Nigel Rees", "price": 8.95, "year": 1999},
                {"title": "Sword of Honour", "author": "Evelyn Waugh", "price": 12.99, "year": 1965}
            ]
        }
    }

    # Sorting queries using /() operator
    sorting_queries = {
        "Sort by price (ascending)": "$.store.book[/(price)]",
        "Sort by year": "$.store.book[/(year)]",
        "Sort by title": "$.store.book[/(title)]",
        "Sort by author": "$.store.book[/(author)]"
    }

    for query_name, expression in sorting_queries.items():
        try:
            results = JSONPath(expression).parse(books_data)
            print(f"\n{query_name}:")
            if results:
                print(f"  Found {len(results)} books:")
                for result in results:
                    if isinstance(result, dict):
                        title = result.get('title', 'Unknown')
                        price = result.get('price', 'N/A')
                        year = result.get('year', 'N/A')
                        print(f"    - '{title}' (${price}, {year})")
            else:
                print("  No results")
        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== Field Extraction with .() Operator ===")

    # Field extraction queries using .() operator
    extraction_queries = {
        "Extract title and price": "$.store.book[*].(title,price)",
        "Extract author and year": "$.store.book[*].(author,year)",
        "Extract single field": "$.store.book[*].(title)",
        "Extract all fields": "$.store.book[*].(title,author,price,year)"
    }

    for query_name, expression in extraction_queries.items():
        try:
            results = JSONPath(expression).parse(books_data)
            print(f"\n{query_name}:")
            if results:
                print(f"  Extracted {len(results)} items:")
                for result in results[:3]:  # Show first 3
                    print(f"    - {result}")
            else:
                print("  No results")
        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== Combining Sorting with Field Extraction ===")

    # Combine sorting and field extraction
    combined_queries = {
        "Sort by price, extract title": "$.store.book[/(price)].(title,price)",
        "Sort by year, extract info": "$.store.book[/(year)].(author,year)",
        "Cheapest books' titles": "$.store.book[/(price)].(title)",
        "Newest books info": "$.store.book[/(year)].(title,year)"
    }

    for query_name, expression in combined_queries.items():
        try:
            results = JSONPath(expression).parse(books_data)
            print(f"\n{query_name}:")
            if results:
                print(f"  Results ({len(results)} items):")
                for result in results:
                    print(f"    - {result}")
            else:
                print("  No results")
        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== Complex Sorting Scenarios ===")

    # Create data with more fields for complex sorting
    products_data = {
        "products": [
            {"name": "Laptop", "type": "electronics", "price": 999, "rating": 4.5, "stock": 10},
            {"name": "Phone", "type": "electronics", "price": 699, "rating": 4.7, "stock": 25},
            {"name": "Book", "type": "media", "price": 15, "rating": 4.2, "stock": 100},
            {"name": "Tablet", "type": "electronics", "price": 449, "rating": 4.4, "stock": 15},
            {"name": "Headphones", "type": "electronics", "price": 199, "rating": 4.6, "stock": 50}
        ]
    }

    complex_sorting = {
        "By stock (low to high)": "$.products[/(stock)]",
        "By name alphabetically": "$.products[/(name)]",
        "By price": "$.products[/(price)]"
    }

    for query_name, expression in complex_sorting.items():
        try:
            results = JSONPath(expression).parse(products_data)
            print(f"\n{query_name}:")
            if results:
                for result in results[:3]:  # Show first 3
                    if isinstance(result, dict):
                        name = result.get('name', 'Unknown')
                        price = result.get('price', 'N/A')
                        rating = result.get('rating', 'N/A')
                        stock = result.get('stock', 'N/A')
                        print(f"    - {name}: ${price}, rating {rating}, stock {stock}")
            else:
                print("  No results")
        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== Filtering with Sorting ===")

    # Combine filtering and sorting: sort FIRST, then filter. In jsonpath-python 1.1.6 a sort placed
    # AFTER a filter (`[?(...)][/(price)]`) raises JSONPathTypeError, so the sort comes first.
    filter_sort_queries = {
        "Expensive items sorted": ("$.products[/(price)][?(@.price>400)]", ["Tablet", "Phone", "Laptop"]),
        "Low stock sorted": ("$.products[/(stock)][?(@.stock<50)]", ["Laptop", "Tablet", "Phone"])
    }

    for query_name, (expression, expected_names) in filter_sort_queries.items():
        results = JSONPath(expression).parse(products_data)
        print(f"\n{query_name}:")
        print(f"  Results ({len(results)} items):")
        for result in results:
            print(f"    - {result.get('name')}: ${result.get('price')}, stock {result.get('stock')}")
        assert [result["name"] for result in results] == expected_names, results

    print("\n=== Multi-field Selection ===")

    # Using [] for multiple field selection at root level
    multi_field_data = {
        "users": [{"name": "Alice", "age": 25}],
        "products": [{"name": "Laptop", "price": 999}],
        "orders": [{"id": 1, "total": 100}]
    }

    multi_field_queries = {
        "Select users and products": "$[users,products]",
        "Select users and orders": "$[users,orders]",
        "Select all three": "$[users,products,orders]"
    }

    for query_name, expression in multi_field_queries.items():
        try:
            results = JSONPath(expression).parse(multi_field_data)
            print(f"\n{query_name}:")
            if results:
                print(f"  Found {len(results)} field groups:")
                for result in results:
                    if isinstance(result, list):
                        print(f"    - List with {len(result)} items")
                    else:
                        print(f"    - {result}")
            else:
                print("  No results")
        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== Array Slicing with Sorting ===")

    # Array slicing combined with sorting
    slice_sort_queries = {
        "Top 3 cheapest books": "$.store.book[/(price)][0:3]",
        "Bottom 2 by year": "$.store.book[/(year)][0:2]",
        "Middle books by price": "$.store.book[/(price)][1:3]"
    }

    for query_name, expression in slice_sort_queries.items():
        try:
            results = JSONPath(expression).parse(books_data)
            print(f"\n{query_name}:")
            if results:
                for result in results:
                    if isinstance(result, dict):
                        print(f"    - {result.get('title')}: ${result.get('price')}")
            else:
                print("  No results")
        except Exception as e:
            print(f"  Error: {e}")
            raise

    print("\n=== Feature Summary ===")

    features = [
        ("/():", "Sort operator - sorts results by field"),
        ("/(field):", "Sort ascending by field"),
        (".():", "Field extractor - selects specific fields"),
        (".(a,b,c):", "Extract multiple fields"),
        ("[a,b,c]:", "Select multiple keys from object"),
        ("[/(f)][0:n]:", "Sort then slice - get top N items"),
        ("[/(f)][?(@.f>v)]:", "Sort then filter")
    ]

    print("Unique jsonpath-python features demonstrated:")
    for operator, description in features:
        print(f"  {operator:20} {description}")

    print("\nExample 4: Sorting and field extraction test passed")

except Exception as e:
    print(f"Example 4: Error in sorting/extraction test: {e}")
    import traceback
    traceback.print_exc()
    raise
```

## Example 5: Production Data Processing and Integration Patterns

Demonstrate production-ready JSONPath usage with real-world data processing scenarios.

```python
# Example 5: Production data processing and integration patterns
from jsonpath import JSONPath
import json
from typing import Dict, List, Any, Optional

try:
    print("=== Production JSONPath Data Processor ===")

    class JSONPathProcessor:
        """Production-ready JSONPath data processor with caching"""

        def __init__(self):
            self.compiled_queries = {}

        def compile_query(self, name: str, expression: str) -> bool:
            """Pre-compile JSONPath expressions for reuse"""
            # Store the JSONPath object for reuse
            self.compiled_queries[name] = JSONPath(expression)
            return True

        def extract_data(self, query_name: str, data: Dict) -> Optional[List]:
            """Extract data using compiled query"""
            if query_name not in self.compiled_queries:
                return None

            return self.compiled_queries[query_name].parse(data)

        def bulk_extract(self, data: Dict, query_names: List[str]) -> Dict[str, Any]:
            """Extract multiple queries efficiently"""
            results = {}
            for name in query_names:
                results[name] = self.extract_data(name, data)
            return results

    # Create processor and compile common queries
    processor = JSONPathProcessor()

    print("Initializing production JSONPath processor...")

    # E-commerce data
    ecommerce_api = {
        "status": "success",
        "timestamp": "2024-01-16T15:30:00Z",
        "data": {
            "users": [
                {
                    "id": 101,
                    "profile": {
                        "email": "alice@company.com",
                        "tier": "premium",
                        "created": "2023-01-15"
                    },
                    "activity": {
                        "last_login": "2024-01-16",
                        "sessions": [
                            {"id": "sess_1", "duration": 4500, "revenue": 299.99},
                            {"id": "sess_2", "duration": 1200, "revenue": 0}
                        ]
                    }
                },
                {
                    "id": 102,
                    "profile": {
                        "email": "bob@startup.com",
                        "tier": "basic",
                        "created": "2023-06-20"
                    },
                    "activity": {
                        "last_login": "2024-01-10",
                        "sessions": [
                            {"id": "sess_3", "duration": 800, "revenue": 0}
                        ]
                    }
                }
            ],
            "products": [
                {"id": "prod_1", "name": "Laptop", "price": 999.99, "type": "electronics"},
                {"id": "prod_2", "name": "Mouse", "price": 29.99, "type": "electronics"}
            ]
        },
        "metadata": {
            "total_users": 2,
            "total_revenue": 299.99,
            "processing_time_ms": 150
        }
    }

    # Production queries for API response processing
    production_queries = {
        "user_ids": "$.data.users[*].id",
        "user_emails": "$.data.users[*].profile.email",
        "all_sessions": "$.data.users[*].activity.sessions[*]",
        "all_products": "$.data.products[*]"
    }

    compiled_count = 0
    for name, expr in production_queries.items():
        if processor.compile_query(name, expr):
            compiled_count += 1

    print(f"Compiled {compiled_count}/{len(production_queries)} queries successfully")
    assert compiled_count == len(production_queries), compiled_count

    print("\n=== Real-world E-commerce API Processing ===")

    # Business intelligence queries
    bi_queries = {
        "all_user_ids": "$.data.users[*].id",
        "all_emails": "$.data.users[*].profile.email",
        "total_session_revenue": "$.data.users[*].activity.sessions[*].revenue",
        "session_durations": "$.data.users[*].activity.sessions[*].duration",
        "product_info": "$.data.products[*].(name,price)",
        "all_product_names": "$.data.products[*].name"
    }

    # Compile BI queries
    for name, expr in bi_queries.items():
        processor.compile_query(name, expr)

    # Process business intelligence data
    # bulk_extract takes the query NAMES registered with compile_query(), not (name, expression) pairs
    bi_results = processor.bulk_extract(ecommerce_api, list(bi_queries))

    print("Business Intelligence Results:")
    for name, results in bi_results.items():
        if results:
            if name == "total_session_revenue":
                total_revenue = sum(r for r in results if isinstance(r, (int, float)))
                print(f"  {name}: Total ${total_revenue:.2f} from {len(results)} sessions")
            elif name == "session_durations":
                if results:
                    avg_duration = sum(results) / len(results) if results else 0
                    print(f"  {name}: Average {avg_duration:.0f} seconds ({len(results)} sessions)")
            elif name == "product_info":
                print(f"  {name}: {len(results)} products extracted")
                for item in results[:2]:
                    print(f"    - {item}")
            else:
                print(f"  {name}: {len(results)} results")
                if results and len(results) <= 3:
                    for item in results:
                        print(f"    - {item}")
        else:
            print(f"  {name}: No data")

    assert bi_results == {
        "all_user_ids": [101, 102],
        "all_emails": ["alice@company.com", "bob@startup.com"],
        "total_session_revenue": [299.99, 0, 0],
        "session_durations": [4500, 1200, 800],
        "product_info": [{"name": "Laptop", "price": 999.99}, {"name": "Mouse", "price": 29.99}],
        "all_product_names": ["Laptop", "Mouse"],
    }, bi_results

    print("\n=== Advanced Query Patterns ===")

    # Demonstrate sorting and extraction
    advanced_patterns = {
        "Sort products by price": "$.data.products[/(price)]",
        "Product names only": "$.data.products[*].(name)",
        "High revenue sessions": "$.data.users[*].activity.sessions[?(@.revenue>100)]",
        "Long sessions": "$.data.users[*].activity.sessions[?(@.duration>1000)]"
    }

    users_sessions = [user["activity"]["sessions"] for user in ecommerce_api["data"]["users"]]
    expected_advanced = {
        "Sort products by price": sorted(ecommerce_api["data"]["products"], key=lambda product: product["price"]),
        "Product names only": [{"name": "Laptop"}, {"name": "Mouse"}],
        "High revenue sessions": [users_sessions[0][0]],
        "Long sessions": users_sessions[0],
    }

    print("Advanced pattern results:")
    for pattern_name, expression in advanced_patterns.items():
        result = JSONPath(expression).parse(ecommerce_api)
        assert result == expected_advanced[pattern_name], (pattern_name, result)
        print(f"\n{pattern_name}:")
        if result:
            print(f"  Found {len(result)} results:")
            for item in result[:3]:
                if isinstance(item, dict):
                    # Show relevant fields
                    if 'name' in item:
                        print(f"    - Name: {item['name']}")
                    elif 'revenue' in item:
                        print(f"    - Revenue: ${item.get('revenue')}, Duration: {item.get('duration')}s")
                    else:
                        print(f"    - {item}")
                else:
                    print(f"    - {item}")
        else:
            print("  No results")

    print("\n=== ETL Pipeline with JSONPath ===")

    def extract_transform_load(data: Dict, config: Dict) -> Dict:
        """ETL pipeline using JSONPath for extraction"""
        results = {}

        for stage_name, stage_config in config.items():
            expression = stage_config.get("query")
            transform = stage_config.get("transform")

            try:
                # Extract
                extracted = JSONPath(expression).parse(data)

                # Transform
                if transform == "sum":
                    results[stage_name] = sum(extracted) if extracted else 0
                elif transform == "count":
                    results[stage_name] = len(extracted) if extracted else 0
                elif transform == "average":
                    results[stage_name] = sum(extracted) / len(extracted) if extracted else 0
                elif transform == "first":
                    results[stage_name] = extracted[0] if extracted else None
                elif transform == "last":
                    results[stage_name] = extracted[-1] if extracted else None
                else:
                    results[stage_name] = extracted
            except Exception as e:
                results[stage_name] = f"Error: {str(e)[:50]}"

        return results

    # ETL configuration
    etl_config = {
        "total_revenue": {
            "query": "$..sessions[*].revenue",
            "transform": "sum"
        },
        "user_count": {
            "query": "$.data.users[*]",
            "transform": "count"
        },
        "avg_session_duration": {
            "query": "$..sessions[*].duration",
            "transform": "average"
        },
        "first_user_email": {
            "query": "$.data.users[*].profile.email",
            "transform": "first"
        },
        "all_user_emails": {
            "query": "$.data.users[*].profile.email",
            "transform": "list"
        }
    }

    # Run ETL pipeline
    etl_results = extract_transform_load(ecommerce_api, etl_config)
    assert etl_results == {"total_revenue": 299.99, "user_count": 2, "avg_session_duration": 6500 / 3,
                           "first_user_email": "alice@company.com", "all_user_emails": ["alice@company.com", "bob@startup.com"]}, etl_results

    print("ETL Pipeline Results:")
    for metric, value in etl_results.items():
        if isinstance(value, (int, float)):
            print(f"  {metric}: {value:.2f}")
        elif isinstance(value, list) and len(value) <= 3:
            print(f"  {metric}: {value}")
        else:
            print(f"  {metric}: {value}")

    print("\n=== Error Handling and Validation ===")

    # Test error handling
    error_test_cases = [
        ("Valid simple query", "$.data.users[*].id"),
        ("Non-existent path", "$.nonexistent.path.here"),
        ("Wildcard query", "$.data.*"),
        ("Recursive query", "$..name")
    ]

    print("Error handling tests:")
    for test_name, query in error_test_cases:
        try:
            result = JSONPath(query).parse(ecommerce_api)
            status = f"Success: {len(result) if result else 0} results"
        except Exception as e:
            status = f"Error: {type(e).__name__}"

        print(f"  {test_name}: {status}")
        assert status.startswith("Success"), (test_name, status)

    print("\n=== Performance Optimization ===")

    # Demonstrate query reuse
    print("Query reuse optimization:")
    reused_query = JSONPath("$.data.users[*].profile.email")

    # Simulate multiple executions with same query
    for i in range(3):
        result = reused_query.parse(ecommerce_api)
        print(f"  Execution {i+1}: Found {len(result)} emails")

    print(f"\nProcessor has {len(processor.compiled_queries)} compiled queries cached")

    print("\n=== Data Aggregation ===")

    # Aggregate data from multiple paths
    aggregations = {
        "All IDs": "$..id",
        "All prices": "$..price",
        "All revenues": "$..revenue",
        "All durations": "$..duration"
    }

    print("Aggregated data:")
    for name, expr in aggregations.items():
        result = JSONPath(expr).parse(ecommerce_api)
        assert len(result) == {"All IDs": 7, "All prices": 2, "All revenues": 3, "All durations": 3}[name], (name, result)
        count = len(result) if result else 0
        if count > 0 and all(isinstance(x, (int, float)) for x in result):
            total = sum(result)
            print(f"  {name:15} -> {count} items, total: {total:.2f}")
        else:
            print(f"  {name:15} -> {count} items")

    print("\n=== Integration Summary ===")

    integration_features = [
        "Query compilation and caching",
        "Bulk extraction operations",
        "ETL pipeline integration",
        "Sorting with /(field) operator",
        "Field extraction with .(field) operator",
        "Error handling and validation",
        "Query reuse for performance",
        "Data aggregation patterns"
    ]

    print("Production features demonstrated:")
    for i, feature in enumerate(integration_features, 1):
        print(f"  {i}. {feature}")

    print("\nExample 5: Production data processing patterns test passed")

except Exception as e:
    print(f"Example 5: Error in production patterns test: {e}")
    import traceback
    traceback.print_exc()
    raise
```
