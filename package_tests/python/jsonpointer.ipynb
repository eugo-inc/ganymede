# jsonpointer Package Testing

This notebook tests the jsonpointer package - JSON Pointer implementation.

## Example 1: Basic JSON navigation and access


```python
import jsonpointer

# Example 1: Basic JSON navigation and access
print("Example 1: Basic JSON navigation and access")

try:
    # Sample JSON data
    data = {
        "name": "John Doe",
        "age": 30,
        "address": {
            "street": "123 Main St",
            "city": "Anytown",
            "zipcode": "12345"
        },
        "hobbies": ["reading", "swimming", "coding"]
    }
    
    # Create JSON pointers for different paths
    name_pointer = jsonpointer.JsonPointer("/name")
    age_pointer = jsonpointer.JsonPointer("/age")
    street_pointer = jsonpointer.JsonPointer("/address/street")
    first_hobby_pointer = jsonpointer.JsonPointer("/hobbies/0")
    
    # Access values using pointers
    print(f"Name: {name_pointer.resolve(data)}")
    print(f"Age: {age_pointer.resolve(data)}")
    print(f"Street: {street_pointer.resolve(data)}")
    print(f"First hobby: {first_hobby_pointer.resolve(data)}")
    
    # Alternative syntax using resolve_pointer function
    city = jsonpointer.resolve_pointer(data, "/address/city")
    print(f"City: {city}")
    
    print("Basic JSON navigation completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"jsonpointer example (cell 2) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished")
```

## Example 2: Array access and complex structures


```python
import jsonpointer

# Example 2: Array access and complex structures
print("Example 2: Array access and complex structures")

try:
    # JSON data with arrays
    data = {
        "users": [
            {"id": 1, "name": "Alice", "roles": ["admin", "user"]},
            {"id": 2, "name": "Bob", "roles": ["user"]},
            {"id": 3, "name": "Charlie", "roles": ["moderator", "user"]}
        ]
    }
    
    # Access array elements
    first_user = jsonpointer.resolve_pointer(data, "/users/0")
    print(f"First user: {first_user}")
    
    # Access nested array elements
    alice_first_role = jsonpointer.resolve_pointer(data, "/users/0/roles/0")
    print(f"Alice's first role: {alice_first_role}")
    
    # Iterate through array using pointers
    for i in range(len(data["users"])):
        user_pointer = f"/users/{i}/name"
        name = jsonpointer.resolve_pointer(data, user_pointer)
        print(f"User {i}: {name}")
    
    print("Array access and complex structures completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"jsonpointer example (cell 4) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished")
```

## Example 3: Error handling and validation


```python
import jsonpointer

# Example 3: Error handling and validation
print("Example 3: Error handling and validation")

try:
    data = {"api": {"version": "1.0", "endpoints": {"users": "/api/users"}}}
    
    # Test various pointer paths: (path, the value it resolves to, or the exception it must raise)
    test_paths = [
        ("/api/version", "1.0"),                                      # Valid path
        ("/api/endpoints/users", "/api/users"),                       # Valid nested path
        ("/nonexistent/path", jsonpointer.JsonPointerException),      # Invalid path
        ("/api/endpoints/invalid", jsonpointer.JsonPointerException), # Invalid nested path
    ]
    
    for path, expected in test_paths:
        try:
            result = jsonpointer.resolve_pointer(data, path)
        except jsonpointer.JsonPointerException as e:
            assert expected is jsonpointer.JsonPointerException, f"{path} must resolve to {expected!r}, raised: {e}"
            print(f"✓ {path}: rejected as expected ({e})")
        else:
            assert expected is not jsonpointer.JsonPointerException, f"{path} resolved to {result!r}; an invalid path must raise"
            assert result == expected, (path, result)
            print(f"✓ {path}: {result}")
    
    print("Error handling and validation completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"jsonpointer example (cell 6) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished")
```

## Example 4: Special characters and escaping


```python
import jsonpointer

# Example 4: Special characters and escaping
print("Example 4: Special characters and escaping")

try:
    # Data with special characters in keys
    data = {
        "": "empty_key_value",
        "key with spaces": "space_value",
        "key/with/slashes": "slash_value",
        "key~with~tildes": "tilde_value"
    }
    
    # Test accessing keys with special characters
    empty_key = jsonpointer.resolve_pointer(data, "/")
    print(f"Empty key: {empty_key}")
    
    spaces_key = jsonpointer.resolve_pointer(data, "/key with spaces")
    print(f"Spaces key: {spaces_key}")
    
    # Test escaped characters (slashes and tildes)
    slash_pointer = jsonpointer.JsonPointer("/key~1with~1slashes")
    tilde_pointer = jsonpointer.JsonPointer("/key~0with~0tildes")
    
    print(f"Slash pointer result: {slash_pointer.resolve(data)}")
    print(f"Tilde pointer result: {tilde_pointer.resolve(data)}")
    
    print("Special characters and escaping completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"jsonpointer example (cell 8) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished")
```

## Example 5: Practical JSON API navigation


```python
import jsonpointer

# Example 5: Practical JSON API navigation
print("Example 5: Practical JSON API navigation")

try:
    # Simulate a complex API response
    api_response = {
        "status": "success",
        "data": {
            "user": {
                "id": 12345,
                "profile": {
                    "name": "John Doe",
                    "email": "john@example.com",
                    "settings": {
                        "notifications": {"email": True, "sms": False},
                        "privacy": "public"
                    }
                },
                "posts": [
                    {"id": 1, "title": "First Post", "likes": 42},
                    {"id": 2, "title": "Second Post", "likes": 18}
                ]
            }
        },
        "metadata": {
            "timestamp": "2023-12-25T10:00:00Z",
            "version": "1.2.3"
        }
    }
    
    # Navigate through the API response
    user_name = jsonpointer.resolve_pointer(api_response, "/data/user/profile/name")
    user_email = jsonpointer.resolve_pointer(api_response, "/data/user/profile/email")
    email_notifications = jsonpointer.resolve_pointer(api_response, "/data/user/profile/settings/notifications/email")
    first_post_title = jsonpointer.resolve_pointer(api_response, "/data/user/posts/0/title")
    api_version = jsonpointer.resolve_pointer(api_response, "/metadata/version")
    
    print(f"User: {user_name} ({user_email})")
    print(f"Email notifications: {email_notifications}")
    print(f"First post: {first_post_title}")
    print(f"API version: {api_version}")
    
    # Extract all post titles
    posts = jsonpointer.resolve_pointer(api_response, "/data/user/posts")
    print("All posts:")
    for i, post in enumerate(posts):
        title_pointer = f"/data/user/posts/{i}/title"
        likes_pointer = f"/data/user/posts/{i}/likes"
        title = jsonpointer.resolve_pointer(api_response, title_pointer)
        likes = jsonpointer.resolve_pointer(api_response, likes_pointer)
        print(f"  - {title} ({likes} likes)")
    
    print("Practical JSON API navigation completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"jsonpointer example (cell 10) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished")
```
