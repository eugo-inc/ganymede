## Example 1: Functionality Test 1

```python
from jupyterlab_nvdashboard.apps.gpu import ngpus, gpu_handles
import jupyterlab_nvdashboard

# Example 1: GPU discovery and system resource monitoring setup
print("Example 1: GPU discovery and system resource monitoring setup")

# Get package version
try:
    version = jupyterlab_nvdashboard.__version__
    print(f"JupyterLab nvdashboard version: {version}")
except:
    print("JupyterLab nvdashboard installed successfully")
    raise

# Discover available GPUs: `ngpus` is an int and `gpu_handles` a list, both computed once at import
# (the NVML device count and one handle per device; 0 and [] when NVML finds no driver).
num_gpus = ngpus
print(f"Number of GPUs detected: {num_gpus}")
assert isinstance(num_gpus, int) and num_gpus >= 0, repr(num_gpus)
assert len(gpu_handles) == num_gpus, (len(gpu_handles), num_gpus)

if num_gpus > 0:
    handles = gpu_handles
    print(f"GPU handles initialized: {len(handles)} handles")
    for i, handle in enumerate(handles):
        print(f"  GPU {i}: Handle available")
else:
    print("No NVIDIA GPUs detected - CPU monitoring available")

# Show dashboard capabilities
capabilities = ["GPU utilization", "Memory usage", "PCIe throughput", "CPU resources"]
print(f"\nDashboard capabilities: {len(capabilities)} monitoring types")
for cap in capabilities:
    print(f"  ✓ {cap}")

print("Example 1 completed successfully")
```

## Example 2: Functionality Test 2

```python
from jupyterlab_nvdashboard.apps.gpu import GPUResourceWebSocketHandler, GPUUsageWebSocketHandler

# Example 2: GPU monitoring WebSocket handlers configuration
print("Example 2: GPU monitoring WebSocket handlers configuration")

# Configure GPU resource monitoring handlers
gpu_handlers = {
    "resource": GPUResourceWebSocketHandler,
    "usage": GPUUsageWebSocketHandler
}

print(f"GPU monitoring handlers: {len(gpu_handlers)}")
for handler_name, handler_class in gpu_handlers.items():
    print(f"  {handler_name.title()}: {handler_class.__name__}")

# Show handler capabilities
handler_capabilities = {
    "GPUResourceWebSocketHandler": ["Memory allocation", "Process tracking", "Device info"],
    "GPUUsageWebSocketHandler": ["Real-time utilization", "Power consumption", "Temperature"]
}

print(f"\nHandler capabilities:")
for handler, caps in handler_capabilities.items():
    print(f"  {handler}:")
    for cap in caps:
        print(f"    - {cap}")

# WebSocket endpoint configuration
endpoints = {
    "/nvdashboard/gpu-resource": "GPU resource monitoring",
    "/nvdashboard/gpu-usage": "GPU usage statistics"
}

print(f"\nWebSocket endpoints: {len(endpoints)}")
for endpoint, description in endpoints.items():
    print(f"  {endpoint}: {description}")

print("Example 2 completed successfully")
```

## Example 4: Functionality Test 4

```python
from jupyterlab_nvdashboard import setup_handlers, _load_jupyter_server_extension

# Example 4: JupyterLab server extension integration
print("Example 4: JupyterLab server extension integration")

# Create mock web application for handler setup
class MockWebApp:
    def __init__(self):
        self.handlers = []
        self.settings = {'base_url': '/'}  # Add required settings attribute
    
    def add_handlers(self, pattern, handlers):
        self.handlers.extend(handlers)
        print(f"Registered {len(handlers)} handlers for pattern: {pattern}")

# Create mock server app
class MockServerApp:
    def __init__(self):
        self.web_app = MockWebApp()
        self.log = type('Log', (), {'info': lambda self, msg: print(f"[INFO] {msg}")})()

# Setup nvdashboard handlers
server_app = MockServerApp()
setup_handlers(server_app.web_app)

print(f"Dashboard handlers registered: {len(server_app.web_app.handlers)}")

# Load the server extension
_load_jupyter_server_extension(server_app)

# Show dashboard routes
dashboard_routes = [
    "/nvdashboard/gpu-usage",
    "/nvdashboard/gpu-resource", 
    "/nvdashboard/cpu-resource",
    "/nvdashboard/pci-stats"
]

print(f"\nDashboard routes: {len(dashboard_routes)}")
for route in dashboard_routes:
    print(f"  {route}")

print("Example 4 completed successfully")
```

## Example 5: Functionality Test 5

```python
import jupyterlab_nvdashboard
from jupyterlab_nvdashboard.apps.utils import CustomWebSocketHandler

# Example 5: Complete nvdashboard deployment configuration
print("Example 5: Complete nvdashboard deployment configuration")

# Get extension configuration
extension_paths = jupyterlab_nvdashboard._jupyter_labextension_paths()
server_extensions = jupyterlab_nvdashboard._jupyter_server_extension_points()

print(f"Extension configuration:")
for path_config in extension_paths:
    print(f"  Source: {path_config['src']}")
    print(f"  Destination: {path_config['dest']}")

print(f"\nServer extension points: {len(server_extensions)}")
for ext in server_extensions:
    print(f"  Module: {ext['module']}")

# Custom WebSocket handler base class
base_handler = CustomWebSocketHandler
print(f"\nBase handler: {base_handler.__name__}")

# Complete monitoring stack
monitoring_stack = {
    "frontend": "JupyterLab extension with real-time charts",
    "backend": "WebSocket handlers for data streaming",
    "gpu_monitoring": "NVIDIA pynvml integration",
    "cpu_monitoring": "psutil system metrics",
    "communication": "Tornado WebSocket protocol"
}

print(f"\nMonitoring stack components:")
for component, description in monitoring_stack.items():
    print(f"  {component.replace('_', ' ').title()}: {description}")

# Deployment validation
validation = {
    "extension_paths": len(extension_paths) > 0,
    "server_points": len(server_extensions) > 0,
    "handlers_available": hasattr(jupyterlab_nvdashboard, 'setup_handlers')
}

print(f"\nDeployment validation:")
for check, status in validation.items():
    symbol = "✓" if status else "✗"
    print(f"  {symbol} {check.replace('_', ' ').title()}")

print("Example 5 completed successfully")
```
