## Example 2: jupyter_ai functionality test 2

```python
import jupyter_ai

# Example 2: AI provider classes - Different AI service integrations
print("Example 2: AI Provider Classes - Different AI Service Integrations")

try:
    # Collect all provider classes
    all_attrs = dir(jupyter_ai)
    provider_classes = []
    
    for attr_name in all_attrs:
        if not attr_name.startswith('_'):
            attr = getattr(jupyter_ai, attr_name)
            # Look for provider classes
            if (isinstance(attr, type) and 
                (attr_name.endswith('Provider') or attr_name in ['AI21', 'OpenAI', 'Anthropic'])):
                provider_classes.append((attr_name, attr))
    
    print(f"Found {len(provider_classes)} AI provider classes:")
    
    # Examine each provider class
    for name, provider_class in provider_classes[:8]:  # Show first 8
        print(f"\n{name}:")
        print(f"  Class: {provider_class}")
        
        # Check for provider-specific attributes
        if hasattr(provider_class, 'id'):
            print(f"  ID: {provider_class.id}")
        if hasattr(provider_class, 'name'):
            print(f"  Name: {provider_class.name}")
        if hasattr(provider_class, 'model_id_key'):
            print(f"  Model key: {provider_class.model_id_key}")
            
        # Check for models property/method
        if hasattr(provider_class, 'models'):
            try:
                models = provider_class.models
                if isinstance(models, (list, tuple)):
                    print(f"  Models: {models[:3]}{'...' if len(models) > 3 else ''}")
                else:
                    print(f"  Models: {type(models)}")
            except:
                print(f"  Models: (property/method)")
                raise
    
    # Create a simple provider registry simulation
    class ProviderRegistry:
        def __init__(self):
            self.providers = {}
            
        def register(self, provider_class):
            provider_id = getattr(provider_class, 'id', provider_class.__name__.lower())
            self.providers[provider_id] = provider_class
            
        def get_provider_names(self):
            return list(self.providers.keys())
    
    # Register available providers
    registry = ProviderRegistry()
    for name, provider_class in provider_classes[:5]:  # Register first 5
        registry.register(provider_class)
    
    print(f"\nProvider registry simulation:")
    print(f"Registered providers: {registry.get_provider_names()}")
    
except Exception as e:
    print(f"Provider exploration error: {type(e).__name__}: {e}")
    raise AssertionError(
        f"jupyter_ai example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed successfully\n")
```

## Example 3: jupyter_ai functionality test 3

```python
import jupyter_ai
from enum import Enum

# Example 3: AuthStrategy configuration and authentication handling
print("Example 3: AuthStrategy Configuration and Authentication Handling")

try:
    auth_strategy = None
    provider_auth_names = sorted(
        name for name in dir(jupyter_ai) if 'auth' in name.lower() or 'strategy' in name.lower()
    )
    print(f"Auth-related exports on jupyter_ai: {provider_auth_names or '(none — moved to providers)'}")
    print(f"jupyter_ai public surface: {[n for n in dir(jupyter_ai) if not n.startswith('_')]}")
    
    # `auth_strategy` is None in this jupyter_ai version (jupyter_ai no longer exports an AuthStrategy enum),
    # so the enum walk is guarded rather than assumed — `issubclass(None, Enum)` is a TypeError, not a graceful miss.
    if auth_strategy is not None and isinstance(auth_strategy, type) and issubclass(auth_strategy, Enum):
        strategies = list(auth_strategy)
        print(f"Available authentication strategies: {len(strategies)}")
        for strategy in strategies:
            print(f"  - {strategy.name}: {strategy.value}")
    else:
        print("No AuthStrategy enum in this jupyter_ai version — auth strategy is a provider-side concept now.")
    
    # Create a mock authentication configuration
    class AuthConfig:
        """Mock authentication configuration"""
        
        def __init__(self, strategy, **kwargs):
            self.strategy = strategy
            self.config = kwargs
            self.credentials = {}
            
        def set_credential(self, key, value):
            """Set authentication credential"""
            self.credentials[key] = value
            
        def get_credential(self, key):
            """Get authentication credential"""
            return self.credentials.get(key)
            
        def is_configured(self):
            """Check if authentication is properly configured"""
            return bool(self.credentials)
            
        def __repr__(self):
            return f"AuthConfig({self.strategy}, {len(self.credentials)} credentials)"
    
    # Test different authentication configurations
    auth_configs = []
    
    # API Key based authentication
    api_key_auth = AuthConfig("api_key")
    api_key_auth.set_credential("api_key", "sk-test-key-123")
    api_key_auth.set_credential("org_id", "org-456")
    auth_configs.append(api_key_auth)
    
    # OAuth based authentication
    oauth_auth = AuthConfig("oauth")
    oauth_auth.set_credential("client_id", "client_123")
    oauth_auth.set_credential("client_secret", "secret_456")
    oauth_auth.set_credential("redirect_uri", "http://localhost:8888")
    auth_configs.append(oauth_auth)
    
    # Environment based authentication
    env_auth = AuthConfig("environment")
    env_auth.set_credential("env_var", "JUPYTER_AI_API_KEY")
    auth_configs.append(env_auth)
    
    print(f"\nAuthentication configurations:")
    for i, config in enumerate(auth_configs, 1):
        print(f"{i}. {config}")
        print(f"   Configured: {config.is_configured()}")
        print(f"   Credentials: {list(config.credentials.keys())}")
    
except Exception as e:
    print(f"AuthStrategy exploration error: {type(e).__name__}: {e}")
    raise AssertionError(
        f"jupyter_ai example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 completed successfully\n")
```

## Example 4: jupyter_ai functionality test 4

```python
import jupyter_ai
from datetime import datetime
from typing import Dict, List

# Example 4: Chat message handling and conversation management
print("Example 4: Chat Message Handling and Conversation Management")

try:
    # Look for message-related classes in jupyter_ai
    message_classes = []
    all_attrs = dir(jupyter_ai)
    
    for attr_name in all_attrs:
        if not attr_name.startswith('_'):
            attr = getattr(jupyter_ai, attr_name)
            if isinstance(attr, type) and any(keyword in attr_name.lower() 
                                            for keyword in ['message', 'chat', 'conversation']):
                message_classes.append((attr_name, attr))
    
    if message_classes:
        print(f"Found message-related classes: {[name for name, _ in message_classes]}")
    
    # Create a mock chat message system
    class ChatMessage:
        """Mock chat message for Jupyter AI conversations"""
        
        def __init__(self, content: str, role: str = "user", timestamp=None):
            self.content = content
            self.role = role  # user, assistant, system
            self.timestamp = timestamp or datetime.now()
            self.id = hash(f"{content}{timestamp}")
            
        def to_dict(self):
            return {
                "id": self.id,
                "content": self.content,
                "role": self.role,
                "timestamp": self.timestamp.isoformat()
            }
            
        def __repr__(self):
            return f"ChatMessage({self.role}: {self.content[:30]}...)"
    
    class ConversationManager:
        """Mock conversation manager for chat sessions"""
        
        def __init__(self):
            self.messages: List[ChatMessage] = []
            self.session_id = f"session_{datetime.now().strftime('%Y%m%d_%H%M%S')}"
            
        def add_message(self, content: str, role: str = "user"):
            """Add a message to the conversation"""
            message = ChatMessage(content, role)
            self.messages.append(message)
            return message
            
        def get_history(self, limit: int = None):
            """Get conversation history"""
            messages = self.messages[-limit:] if limit else self.messages
            return [msg.to_dict() for msg in messages]
            
        def clear_history(self):
            """Clear conversation history"""
            self.messages.clear()
            
        def get_conversation_stats(self):
            """Get conversation statistics"""
            user_msgs = sum(1 for msg in self.messages if msg.role == "user")
            ai_msgs = sum(1 for msg in self.messages if msg.role == "assistant")
            return {"user_messages": user_msgs, "ai_messages": ai_msgs, "total": len(self.messages)}
    
    # Test the conversation system
    chat = ConversationManager()
    print(f"Created conversation: {chat.session_id}")
    
    # Simulate a conversation
    chat.add_message("Hello, can you help me with Python?", "user")
    chat.add_message("Of course! I'd be happy to help you with Python. What specific topic would you like to learn about?", "assistant")
    chat.add_message("I want to learn about data structures", "user")
    chat.add_message("Great choice! Python has several built-in data structures like lists, dictionaries, sets, and tuples.", "assistant")
    
    # Show conversation statistics
    stats = chat.get_conversation_stats()
    print(f"Conversation stats: {stats}")
    
    # Show recent history
    recent_history = chat.get_history(limit=2)
    print(f"Recent messages: {len(recent_history)}")
    for msg in recent_history:
        print(f"  {msg['role']}: {msg['content'][:50]}...")
    
except Exception as e:
    print(f"Chat message handling error: {type(e).__name__}: {e}")
    raise AssertionError(
        f"jupyter_ai example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 completed successfully\n")
```

## Example 5: jupyter_ai functionality test 5

```python
import jupyter_ai
import json

# Example 5: Configuration and settings management
print("Example 5: Configuration and Settings Management")

try:
    # Look for configuration-related classes
    config_classes = []
    all_attrs = dir(jupyter_ai)
    
    for attr_name in all_attrs:
        if not attr_name.startswith('_'):
            attr = getattr(jupyter_ai, attr_name)
            if isinstance(attr, type) and any(keyword in attr_name.lower() 
                                            for keyword in ['config', 'setting', 'trait']):
                config_classes.append((attr_name, attr))
    
    if config_classes:
        print(f"Configuration classes: {[name for name, _ in config_classes]}")
    else:
        print("No explicit configuration classes found in module")
    
    # Create a mock configuration system
    class JupyterAiConfig:
        """Mock Jupyter AI configuration manager"""
        
        def __init__(self):
            self.providers = {}
            self.global_settings = {
                "default_provider": "openai",
                "max_tokens": 1000,
                "temperature": 0.7,
                "stream_responses": True,
                "save_conversations": True
            }
            self.user_preferences = {
                "theme": "auto",
                "show_timestamps": True,
                "auto_scroll": True,
                "markdown_rendering": True
            }
            
        def add_provider_config(self, provider_id: str, **config):
            """Add configuration for an AI provider"""
            self.providers[provider_id] = config
            
        def get_provider_config(self, provider_id: str):
            """Get configuration for a specific provider"""
            return self.providers.get(provider_id, {})
            
        def update_global_setting(self, key: str, value):
            """Update a global setting"""
            if key in self.global_settings:
                self.global_settings[key] = value
                return True
            return False
            
        def get_global_settings(self):
            """Get all global settings"""
            return self.global_settings.copy()
            
        def set_user_preference(self, key: str, value):
            """Set user preference"""
            self.user_preferences[key] = value
            
        def export_config(self):
            """Export configuration as JSON"""
            return {
                "providers": self.providers,
                "global_settings": self.global_settings,
                "user_preferences": self.user_preferences
            }
            
        def get_config_summary(self):
            """Get configuration summary"""
            return {
                "num_providers": len(self.providers),
                "default_provider": self.global_settings.get("default_provider"),
                "max_tokens": self.global_settings.get("max_tokens"),
                "streaming_enabled": self.global_settings.get("stream_responses")
            }
    
    # Test the configuration system
    config = JupyterAiConfig()
    
    # Add provider configurations
    config.add_provider_config("openai", api_key="sk-...", model="gpt-3.5-turbo", org_id="org-123")
    config.add_provider_config("anthropic", api_key="...", model="claude-3-sonnet")
    config.add_provider_config("local", endpoint="http://localhost:8080", model="llama2")
    
    # Update settings
    config.update_global_setting("max_tokens", 1500)
    config.update_global_setting("temperature", 0.5)
    config.set_user_preference("theme", "dark")
    
    # Show configuration summary
    summary = config.get_config_summary()
    print(f"Configuration summary: {summary}")
    
    # Show provider configurations
    print(f"Configured providers: {list(config.providers.keys())}")
    for provider_id in config.providers.keys():
        provider_config = config.get_provider_config(provider_id)
        print(f"  {provider_id}: {len(provider_config)} settings")
    
    # Export configuration (showing structure)
    full_config = config.export_config()
    print(f"Full config keys: {list(full_config.keys())}")
    
except Exception as e:
    print(f"Configuration management error: {type(e).__name__}: {e}")
    raise AssertionError(
        f"jupyter_ai example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 completed successfully\n")
```
