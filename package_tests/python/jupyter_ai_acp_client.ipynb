# jupyter_ai_acp_client smoke test

Tests that the package imports cleanly and exposes its public API.

```python
# Example 1: inspect top-level public API surface.
import jupyter_ai_acp_client

public_names = [n for n in dir(jupyter_ai_acp_client) if not n.startswith('_')]
print(f"Public names ({len(public_names)}):")
for name in public_names:
    obj = getattr(jupyter_ai_acp_client, name)
    print(f"  {name}: {type(obj).__name__}")

assert isinstance(public_names, list)
```

```python
# Example 2: verify package metadata via importlib.metadata.
from importlib import metadata

dist = metadata.distribution('jupyter-ai-acp-client')
print(f"Name: {dist.metadata['Name']}")
print(f"Version: {dist.version}")
print(f"Summary: {dist.metadata.get('Summary', 'n/a')}")

requires = dist.requires or []
print(f"\nRequires ({len(requires)}):")
for req in requires[:10]:
    print(f"  {req}")

assert dist.version
```

```python
# Example 3: enumerate immediate submodules (one level deep).
import pkgutil

import jupyter_ai_acp_client

submodules = list(pkgutil.iter_modules(jupyter_ai_acp_client.__path__))
print(f"Immediate submodules of jupyter_ai_acp_client ({len(submodules)}):")
for mod in submodules:
    kind = 'package' if mod.ispkg else 'module'
    print(f"  {mod.name} ({kind})")

assert isinstance(submodules, list)
```

```python
# Example 4: verify required runtime dependencies are importable.
import importlib

deps = [
    'acp',
    'jupyter_ai_persona_manager',
    'jupyter_server',
    'jupyterlab_chat',
    'pydantic',
]

missing = []
for dep in deps:
    try:
        importlib.import_module(dep)
        print(f"  OK   {dep}")
    except ImportError as e:
        missing.append((dep, str(e)[:120]))
        print(f"  MISS {dep}: {e}")

print(f"\n{len(deps) - len(missing)} / {len(deps)} runtime deps importable")
```

```python
# Example 5: look for jupyter_server extension entry points / handlers.
import importlib
import pkgutil

import jupyter_ai_acp_client

interesting = ('extension', 'handler', 'app', 'client', 'persona', 'acp')
found = []
for module_info in pkgutil.walk_packages(jupyter_ai_acp_client.__path__, prefix='jupyter_ai_acp_client.'):
    short = module_info.name.rsplit('.', 1)[-1].lower()
    if any(token in short for token in interesting):
        found.append(module_info.name)

print(f"Modules matching interesting tokens ({len(found)}):")
for name in found[:15]:
    print(f"  {name}")

assert hasattr(jupyter_ai_acp_client, '__path__'), 'jupyter_ai_acp_client should be a package'
```
