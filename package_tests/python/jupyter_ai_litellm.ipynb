```python
# Example 1: Inspect the top-level package surface.
import jupyter_ai_litellm

public_names = [n for n in dir(jupyter_ai_litellm) if not n.startswith('_')]
print(f"Public names ({len(public_names)}):")
for name in public_names:
    obj = getattr(jupyter_ai_litellm, name)
    print(f"  {name}: {type(obj).__name__}")

assert hasattr(jupyter_ai_litellm, '__path__'), "package missing __path__"
```

```python
# Example 2: Verify package metadata via importlib.metadata.
from importlib import metadata

dist = metadata.distribution('jupyter-ai-litellm')
print(f"Name: {dist.metadata['Name']}")
print(f"Version: {dist.version}")
summary = dist.metadata.get('Summary', '')
print(f"Summary: {summary}")

assert dist.version, "distribution must have a version"
```

```python
# Example 3: Confirm declared runtime dependencies are importable.
import jupyter_server
import litellm

print(f"jupyter_server version: {getattr(jupyter_server, '__version__', 'unknown')}")
print(f"litellm version: {getattr(litellm, '__version__', 'unknown')}")

assert jupyter_server.__name__ == 'jupyter_server'
assert litellm.__name__ == 'litellm'
```

```python
# Example 4: Enumerate entry points the package contributes (jupyter_ai providers).
from importlib import metadata

dist = metadata.distribution('jupyter-ai-litellm')
eps = list(dist.entry_points)
print(f"Entry points ({len(eps)}):")
for ep in eps:
    print(f"  group={ep.group!r} name={ep.name!r} value={ep.value!r}")

# A jupyter-ai provider plugin typically advertises one or more entry points.
# We don't require a specific count; the lookup itself must succeed.
```

```python
# Example 5: Locate package files on disk and confirm a non-empty install.
import pathlib

import jupyter_ai_litellm

pkg_root = pathlib.Path(jupyter_ai_litellm.__path__[0])
print(f"Package root: {pkg_root}")

py_files = sorted(pkg_root.rglob('*.py'))
print(f"Python files in package: {len(py_files)}")
for f in py_files[:10]:
    print(f"  {f.relative_to(pkg_root)}")

assert pkg_root.exists() and pkg_root.is_dir()
assert len(py_files) >= 1, "expected at least one .py file in the package"
```
