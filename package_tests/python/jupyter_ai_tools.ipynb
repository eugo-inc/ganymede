```python
# Example 4: Probe submodules and inspect any callable / class entry points.
import importlib
import inspect
import pkgutil

import jupyter_ai_tools

submodule_names = []
for module_info in pkgutil.iter_modules(jupyter_ai_tools.__path__, prefix='jupyter_ai_tools.'):
    submodule_names.append(module_info.name)

print(f"Direct submodules: {submodule_names}")

for name in submodule_names[:5]:
    mod = importlib.import_module(name)
    classes = [n for n, v in inspect.getmembers(mod, inspect.isclass) if v.__module__.startswith('jupyter_ai_tools')]
    funcs = [n for n, v in inspect.getmembers(mod, inspect.isfunction) if v.__module__.startswith('jupyter_ai_tools')]
    print(f"  {name}: classes={classes[:5]}, functions={funcs[:5]}")
```

```python
# Example 5: Round-trip sanity — re-import, confirm module identity stability + path on disk.
import importlib
import os

import jupyter_ai_tools

first_id = id(jupyter_ai_tools)
mod = importlib.import_module('jupyter_ai_tools')
second_id = id(mod)

assert first_id == second_id, "Re-import must yield the same module object"

pkg_dir = os.path.dirname(jupyter_ai_tools.__file__) if jupyter_ai_tools.__file__ else jupyter_ai_tools.__path__[0]
assert os.path.isdir(pkg_dir), f"Package dir {pkg_dir} should exist on disk"

print(f"Package located at: {pkg_dir}")
print(f"Module identity stable across re-import: True")
print("jupyter_ai_tools smoke test PASSED")
```
