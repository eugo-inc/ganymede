## Example 2: jupyter_collaboration functionality test 2

```python
import jupyter_collaboration
from pycrdt import Doc, Text, Array, Map

# Example 2: Create collaborative document with CRDT data structures
print("Example 2: Creating collaborative documents with CRDTs")

# Create a new collaborative document
doc = Doc()

# Add collaborative text for real-time editing
doc["notebook_title"] = Text()
doc["notebook_title"].insert(0, "Collaborative Notebook")

# Add collaborative array for cells
doc["cells"] = Array()
doc["cells"].append({"cell_type": "code", "source": "print('Hello')"})
doc["cells"].append({"cell_type": "markdown", "source": "# Title"})

# Add collaborative map for metadata
doc["metadata"] = Map()
doc["metadata"]["kernel"] = "python3"
doc["metadata"]["collaboration_enabled"] = True

print(f"Document ID: {doc.client_id}")
print(f"Title: {str(doc['notebook_title'])}")
print(f"Number of cells: {len(doc['cells'])}")
print(f"Metadata: {dict(doc['metadata'])}")
```

## Example 3: jupyter_collaboration functionality test 3

```python
import jupyter_collaboration
from pycrdt import Doc, Text
import base64

# Example 3: Document synchronization and updates
print("Example 3: Document sync and collaborative updates")

# Create two documents to simulate collaboration
doc1 = Doc()
doc2 = Doc()

# User 1 creates content
doc1["content"] = Text()
doc1["content"].insert(0, "Initial text by User 1")

# Get update from doc1
update = doc1.get_update()
print(f"Update size: {len(update)} bytes")

# Apply update to doc2 (simulating network sync)
doc2.apply_update(update)

# Get the synced Text object with proper type
content2 = doc2.get("content", type=Text)
print(f"Doc2 received: {str(content2)}")

# User 2 makes changes - append to existing text
content2.insert(len(str(content2)), " - edited by User 2")

# Sync back to doc1
update2 = doc2.get_update(doc1.get_state())
doc1.apply_update(update2)

print(f"Final synced text: {str(doc1['content'])}")
print(f"Collaboration active: {jupyter_collaboration.__version__ is not None}")
```

## Example 4: jupyter_collaboration functionality test 4

```python
import jupyter_collaboration
from pycrdt import Doc, Array, Map
import json

# Example 4: Collaborative notebook structure with cells
print("Example 4: Building collaborative notebook structure")

# Create notebook document
notebook = Doc()

# Initialize notebook structure
notebook["cells"] = Array()
notebook["metadata"] = Map()

# Add multiple cells collaboratively
code_cell = {
    "id": "cell-1", 
    "cell_type": "code",
    "source": "import numpy as np\ndata = np.array([1, 2, 3])"
}
notebook["cells"].append(code_cell)

markdown_cell = {
    "id": "cell-2",
    "cell_type": "markdown", 
    "source": "## Data Analysis\nUsing collaborative editing"
}
notebook["cells"].append(markdown_cell)

# Track collaboration metadata
notebook["metadata"]["last_modified"] = "2024-01-01T12:00:00"
notebook["metadata"]["contributors"] = ["user1", "user2"]

print(f"Total cells: {len(notebook['cells'])}")
for i, cell in enumerate(notebook["cells"]):
    print(f"  Cell {i}: {cell['cell_type']} - {cell['id']}")
```

## Example 5: jupyter_collaboration functionality test 5

```python
import jupyter_collaboration
from pycrdt import Doc, Text
import hashlib

# Example 5: Track collaborative changes with observers
print("Example 5: Tracking collaborative document changes")

# Create collaborative document
doc = Doc()
changes_count = 0

# Define simple change observer
def on_change(event):
    global changes_count
    changes_count += 1

# Subscribe to document changes
doc.observe(on_change)

# Make collaborative edits
doc["editor1"] = Text()
doc["editor1"].insert(0, "User 1 typing...")

doc["editor2"] = Text() 
doc["editor2"].insert(0, "User 2 typing simultaneously...")

# Generate document hash for version tracking
doc_state = doc.get_state()
doc_hash = hashlib.md5(doc_state).hexdigest()[:8]

print(f"Document version hash: {doc_hash}")
print(f"Total changes tracked: {changes_count}")
print(f"Collaboration enabled: {jupyter_collaboration.__version__}")
print(f"Editor 1 content: {str(doc['editor1'])[:20]}...")
print(f"Editor 2 content: {str(doc['editor2'])[:20]}...")
```
