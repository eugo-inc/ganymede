## Example 1: jupyter_core functionality test 1

```python
import jupyter_core
from jupyter_core.paths import jupyter_path, jupyter_data_dir

# Example 1: Get Jupyter paths and data directories
print("Example 1: Working with Jupyter paths")

# Get all Jupyter paths
paths = jupyter_path()
print(f"Jupyter paths ({len(paths)} found):")
for i, path in enumerate(paths[:3], 1):
    print(f"  {i}. {path}")

# Get the Jupyter data directory
data_dir = jupyter_data_dir()
print(f"\nJupyter data directory: {data_dir}")

# Check version
print(f"jupyter_core version: {jupyter_core.__version__}")

# Get specific path types
from jupyter_core.paths import SYSTEM_JUPYTER_PATH
print(f"\nSystem Jupyter paths: {len(SYSTEM_JUPYTER_PATH)}")
for path in SYSTEM_JUPYTER_PATH[:2]:
    print(f"  - {path}")
```

## Example 2: jupyter_core functionality test 2

```python
import jupyter_core
from jupyter_core.paths import jupyter_config_dir, jupyter_runtime_dir
import os

# Example 2: Configuration and runtime directories
print("Example 2: Jupyter config and runtime directories")

# Get configuration directory
config_dir = jupyter_config_dir()
print(f"Configuration directory: {config_dir}")
print(f"Config dir exists: {os.path.exists(config_dir)}")

# Get runtime directory for temporary files
runtime_dir = jupyter_runtime_dir()
print(f"\nRuntime directory: {runtime_dir}")
print(f"Runtime dir exists: {os.path.exists(runtime_dir)}")

# Check for environment variable overrides
env_config = os.environ.get('JUPYTER_CONFIG_DIR')
env_runtime = os.environ.get('JUPYTER_RUNTIME_DIR')
print(f"\nEnvironment overrides:")
print(f"  JUPYTER_CONFIG_DIR: {env_config or 'Not set'}")
print(f"  JUPYTER_RUNTIME_DIR: {env_runtime or 'Not set'}")

# Platform info
import sys
print(f"\nPlatform: {sys.platform}")
```

## Example 3: jupyter_core functionality test 3

```python
import jupyter_core
from jupyter_core.application import JupyterApp
from traitlets import Unicode

# Example 3: Create a custom Jupyter application
print("Example 3: Using JupyterApp base class")

class MyJupyterApp(JupyterApp):
    name = "myapp"
    description = "Custom Jupyter Application Example"
    
    # Add a custom trait
    custom_option = Unicode('default', help="A custom option").tag(config=True)
    
    def start(self):
        print(f"Starting {self.name}...")
        print(f"Config dir: {self.config_dir}")
        print(f"Data dir: {self.data_dir}")
        print(f"Runtime dir: {self.runtime_dir}")

# Create and configure the app
app = MyJupyterApp()
print(f"Application: {app.name}")
print(f"Description: {app.description}")
print(f"Version: {app.version}")
print(f"Config file name: {app.config_file_name}")
```

## Example 4: jupyter_core functionality test 4

```python
import jupyter_core
from jupyter_core.paths import secure_write
import tempfile
import os

# Example 4: Securely write configuration files
print("Example 4: Secure file writing with jupyter_core")

# Create a temporary file path
temp_dir = tempfile.gettempdir()
test_file = os.path.join(temp_dir, "jupyter_test_config.json")

# Sample configuration content
config_content = """{
    "NotebookApp": {
        "port": 8888,
        "open_browser": false
    }
}"""

# Use secure_write to safely write the file
try:
    with secure_write(test_file) as f:
        f.write(config_content)
    print(f"Securely wrote config to: {test_file}")
    print(f"File size: {os.path.getsize(test_file)} bytes")
    
    # Clean up
    os.remove(test_file)
    print("Test file cleaned up")
except Exception as e:
    print(f"Write failed: {e}")
    raise
```
