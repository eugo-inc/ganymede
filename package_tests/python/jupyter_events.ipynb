## Example 2: jupyter_events functionality test 2

```python
import jupyter_events
from jupyter_events import EventSchema

# Example 2: Create and work with EventSchema
print("Example 2: EventSchema definition and validation")

# Define a properly formatted event schema with required $id
schema_data = {
    "$id": "https://events.jupyter.org/user_action",
    "version": 1,
    "type": "object",
    "properties": {
        "action": {"type": "string"},
        "user": {"type": "string"},
        "timestamp": {"type": "string"}
    },
    "required": ["action", "user"]
}

# Create an EventSchema with the required schema parameter
event_schema = EventSchema(schema=schema_data)

# Check schema structure
print(f"Schema type: {type(event_schema).__name__}")
print(f"Schema created: {event_schema is not None}")
print(f"Schema ID: {schema_data['$id']}")

# Access the schema data
if hasattr(event_schema, 'schema'):
    print(f"Schema has properties: {'properties' in event_schema.schema}")
    
print(f"Required fields: {schema_data['required']}")
```

## Example 3: jupyter_events functionality test 3

```python
import jupyter_events
from jupyter_events import EventLogger
import logging

# Example 3: Register event handlers and listeners
print("Example 3: Event handler registration")

# Create logger
logger = EventLogger()

# Create a proper logging handler (not event handler)
log_handler = logging.StreamHandler()
log_handler.setLevel(logging.INFO)

# Register logging handler
try:
    logger.register_handler(log_handler)
    print("Logging handler registered successfully")
except Exception as e:
    print(f"Handler registration: {type(e).__name__}")
    raise

# Create async event listener function
async def my_event_listener(logger, schema_id, event_data):
    print(f"Event received - ID: {schema_id}, Data: {event_data}")

# Add event listener with proper signature
try:
    logger.add_listener(schema_id="test.event", listener=my_event_listener)
    print("Event listener added successfully")
except Exception as e:
    print(f"Listener addition: {type(e).__name__}")
    raise

# Test gathering listeners
print(f"Logger configured for event handling")
```

## Example 4: jupyter_events functionality test 4

```python
import jupyter_events
from jupyter_events import EventLogger, EventSchema

# Example 4: Event emission and schema registration
print("Example 4: Event emission and schema registration")

# Create logger
logger = EventLogger()

# Define a schema for notebook events
notebook_schema_data = {
    "$id": "https://events.jupyter.org/notebook_open",
    "version": 1,
    "type": "object", 
    "properties": {
        "notebook_name": {"type": "string"},
        "cell_count": {"type": "integer"},
        "kernel": {"type": "string"}
    },
    "required": ["notebook_name"]
}

# Create EventSchema object
notebook_schema = EventSchema(schema=notebook_schema_data)

# Register schema (takes EventSchema object, not schema_id + data)
try:
    logger.register_event_schema(notebook_schema)
    print("Event schema registered successfully")
except Exception as e:
    print(f"Schema registration: {type(e).__name__}")
    raise

# Create sample event data
event_data = {
    "notebook_name": "example.ipynb",
    "cell_count": 5,
    "kernel": "python3"
}

# Emit event with keyword arguments
try:
    result = logger.emit(schema_id="https://events.jupyter.org/notebook_open", data=event_data)
    print("Event emitted successfully")
    print(f"Result: {type(result).__name__ if result else 'None'}")
except Exception as e:
    print(f"Event emission: {type(e).__name__}")
    raise
```

## Example 5: jupyter_events functionality test 5

```python
import jupyter_events
from jupyter_events import EventLogger
import logging

# Example 5: Event modifiers and handler management
print("Example 5: Event modifiers and handler cleanup")

# Create logger
logger = EventLogger()

# Event modifier function (takes schema_id and event_data)
def event_modifier(schema_id, event_data):
    event_data['modified'] = True
    event_data['timestamp'] = '2024-01-01T12:00:00Z'
    event_data['modifier_applied'] = schema_id
    return event_data

# Add modifier with keyword arguments
try:
    logger.add_modifier(schema_id="test.event", modifier=event_modifier)
    print("Event modifier added successfully")
except Exception as e:
    print(f"Modifier addition: {type(e).__name__}")
    raise

# Create a logging handler for removal test
test_handler = logging.StreamHandler()

# Remove handler test (requires logging.Handler object)
try:
    logger.remove_handler(test_handler)
    print("Handler removal attempted")
except Exception as e:
    print(f"Handler removal: {type(e).__name__}")
    raise

# Check available event handling methods
event_methods = ['emit', 'register_event_schema', 'add_listener', 'add_modifier']
available_methods = [m for m in event_methods if hasattr(logger, m)]
print(f"Available event methods: {available_methods}")
```
