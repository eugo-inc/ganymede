# Smoke test: jupyter_leaflet

`jupyter-leaflet` ships the browser-side files of the ipyleaflet map widget. These examples drive the maps
through ipyleaflet; run `pip install ipyleaflet` as well.

## Example 1: Functionality Test 1

```python
import jupyter_leaflet
from ipyleaflet import Map, TileLayer

# Example 1: Create basic interactive map with custom tile layer
print("Example 1: Creating interactive map with custom tiles")

# Create a map centered on New York City
nyc_map = Map(
    center=(40.7128, -74.0060),
    zoom=12,
    scroll_wheel_zoom=True
)

# Add a custom tile layer
custom_tiles = TileLayer(
    url='https://{s}.tile.openstreetmap.org/{z}/{x}/{y}.png',
    name='OpenStreetMap',
    attribution='© OpenStreetMap contributors'
)

# Add the tile layer to the map
nyc_map.add_layer(custom_tiles)

print(f"Map created with center: {nyc_map.center}")
print(f"Map zoom level: {nyc_map.zoom}")
print(f"Scroll wheel zoom enabled: {nyc_map.scroll_wheel_zoom}")
print(f"Map layers: {len(nyc_map.layers)}")
print("Interactive map ready for display")
```

## Example 2: Functionality Test 2

```python
import jupyter_leaflet
from ipyleaflet import Map, Marker, Icon

# Example 2: Add interactive markers to map
print("Example 2: Adding markers with custom icons")

# Create map centered on San Francisco
sf_map = Map(center=(37.7749, -122.4194), zoom=13)

# Create markers for different locations
markers = [
    Marker(location=(37.7749, -122.4194), 
           title="San Francisco City Hall",
           alt="City Hall"),
    
    Marker(location=(37.8024, -122.4058), 
           title="Alcatraz Island",
           alt="Alcatraz"),
           
    Marker(location=(37.7559, -122.4015),
           title="Mission District", 
           alt="Mission")
]

# Add all markers to the map
for marker in markers:
    sf_map.add_layer(marker)

print(f"Map center: {sf_map.center}")
print(f"Total markers added: {len(markers)}")
print(f"Map has {len(sf_map.layers)} layers")
for i, marker in enumerate(markers, 1):
    print(f"  Marker {i}: {marker.title} at {marker.location}")
```

## Example 3: Functionality Test 3

```python
import jupyter_leaflet  
from ipyleaflet import Map, Circle, Rectangle

# Example 3: Add geometric overlays (circles and rectangles)
print("Example 3: Creating geometric overlays")

# Create map centered on London
london_map = Map(center=(51.5074, -0.1278), zoom=11)

# Create circular overlay around London center
london_circle = Circle(
    location=(51.5074, -0.1278),
    radius=5000,  # 5km radius
    color="blue",
    fill_color="lightblue",
    fill_opacity=0.3
)

# Create rectangular overlay for Thames area
thames_rectangle = Rectangle(
    bounds=((51.4975, -0.1500), (51.5175, -0.1050)),
    color="red",
    fill_color="pink", 
    fill_opacity=0.2
)

# Add overlays to map
london_map.add_layer(london_circle)
london_map.add_layer(thames_rectangle)

print(f"Circle center: {london_circle.location}")
print(f"Circle radius: {london_circle.radius}m")
print(f"Rectangle bounds: {thames_rectangle.bounds}")
print(f"Total overlays: {len(london_map.layers) - 1}")  # -1 for base layer
```

## Example 4: Functionality Test 4

```python
import jupyter_leaflet
from ipyleaflet import Map, Polygon, Polyline

# Example 4: Create polygon and polyline overlays
print("Example 4: Drawing polygons and polylines")

# Create map centered on Tokyo
tokyo_map = Map(center=(35.6762, 139.6503), zoom=12)

# Create polygon for Tokyo Imperial Palace area
palace_polygon = Polygon(
    locations=[
        (35.6825, 139.7521),
        (35.6796, 139.7580), 
        (35.6751, 139.7531),
        (35.6780, 139.7472),
        (35.6825, 139.7521)  # Close the polygon
    ],
    color="green",
    fill_color="lightgreen",
    fill_opacity=0.4
)

# Create polyline for a route through Tokyo
route_polyline = Polyline(
    locations=[
        (35.6762, 139.6503),  # Shinjuku
        (35.6586, 139.7454),  # Tokyo Station  
        (35.6284, 139.7384),  # Ginza
        (35.6618, 139.7038)   # Shibuya
    ],
    color="purple",
    weight=4,
    opacity=0.8
)

# Add overlays to map
tokyo_map.add_layer(palace_polygon)
tokyo_map.add_layer(route_polyline)

print(f"Polygon vertices: {len(palace_polygon.locations)}")
print(f"Polyline points: {len(route_polyline.locations)}")
print(f"Map layers: {len(tokyo_map.layers)}")
```

## Example 5: Functionality Test 5

```python
import jupyter_leaflet
from ipyleaflet import Map, basemaps

# Example 5: Working with different basemap providers
print("Example 5: Using different basemap providers")

# Create map with satellite basemap
satellite_map = Map(
    center=(34.0522, -118.2437),  # Los Angeles
    zoom=10,
    basemap=basemaps.Esri.WorldImagery
)

# Get available basemap information
available_basemaps = [
    ("OpenStreetMap", basemaps.OpenStreetMap.Mapnik),
    ("Satellite", basemaps.Esri.WorldImagery), 
    ("Terrain", basemaps.OpenTopoMap),
    ("CartoDB", basemaps.CartoDB.Positron)
]

print(f"Map center: Los Angeles {satellite_map.center}")
print(f"Current basemap: Satellite imagery")
print(f"Map zoom: {satellite_map.zoom}")

print("\nAvailable basemap providers:")
for name, basemap in available_basemaps:
    if hasattr(basemap, 'url'):
        print(f"  {name}: {basemap.url[:50]}...")
    else:
        print(f"  {name}: Available")

# Demonstrate basemap switching capability
print(f"\nMap configured with {len(available_basemaps)} basemap options")
print("Basemap switching ready for interactive use")
```
