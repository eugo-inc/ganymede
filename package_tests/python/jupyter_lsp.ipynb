## Example 1: jupyter_lsp functionality test 1

```python
import jupyter_lsp
from jupyter_lsp import LanguageServerManager

# Example 1: Create and configure LanguageServerManager
print("Example 1: LanguageServerManager setup")

# Create a language server manager instance
lsm = LanguageServerManager()

# Check manager configuration
print(f"Manager type: {type(lsm).__name__}")
print(f"Manager created: {lsm is not None}")

# Get available language server specs
all_specs = lsm.all_language_servers
print(f"Available language servers: {len(all_specs)}")

# Show some configured servers if any
if all_specs:
    server_names = list(all_specs.keys())[:3]
    print(f"Sample servers: {server_names}")
else:
    print("No language servers currently configured")

# Check if manager is initialized
print(f"Manager initialized: {hasattr(lsm, 'language_servers')}")
print("Language server manager ready")
```

## Example 2: jupyter_lsp functionality test 2

```python
import jupyter_lsp
from jupyter_lsp.specs.utils import NodeModuleSpec
from jupyter_lsp import LanguageServerManager

# Example 2: Work with NodeModuleSpec for language servers
print("Example 2: NodeModuleSpec configuration")

# Create a Node module specification and manager
python_lsp_spec = NodeModuleSpec()
lsp_manager = LanguageServerManager()

# Check the default specification attributes
print(f"Spec type: {type(python_lsp_spec).__name__}")
print(f"Default key: '{python_lsp_spec.key}'")
print(f"Default languages: {python_lsp_spec.languages}")
print(f"Default script: {python_lsp_spec.script}")
print(f"Node module: '{python_lsp_spec.node_module}'")

# Check if the spec can be used
try:
    spec_dict = python_lsp_spec.spec
    print(f"Spec dictionary: {type(spec_dict).__name__}")
    print(f"Spec keys: {list(spec_dict.keys()) if spec_dict else 'Empty'}")
except Exception as e:
    print(f"Spec access: {e}")
    raise

# Test installation check with manager
try:
    is_installed = python_lsp_spec.is_installed(lsp_manager)
    print(f"LSP module installed: {is_installed}")
except Exception as e:
    print(f"Installation check: {type(e).__name__}")
    raise

print("NodeModuleSpec inspection complete")
```

## Example 3: jupyter_lsp functionality test 3

```python
import jupyter_lsp
from jupyter_lsp import LanguageServerManager

# Example 3: Language server session management
print("Example 3: Language server session management")

# Create manager and initialize
lsm = LanguageServerManager()

# Check for available properties
sessions = getattr(lsm, 'sessions', {})
print(f"Active sessions: {len(sessions)}")

# Initialize language servers
try:
    lsm.init_language_servers()
    print("Language servers initialized")
except Exception as e:
    print(f"Server init: {type(e).__name__}")
    raise

# Check available language servers
available_servers = lsm.all_language_servers
print(f"Configured servers: {len(available_servers)}")

# Initialize sessions
try:
    lsm.init_sessions()
    print("Sessions initialized")
except Exception as e:
    print(f"Session init: {type(e).__name__}")
    raise

# Check node roots for server detection
node_roots = lsm.node_roots
print(f"Node roots configured: {len(node_roots) if node_roots else 0}")

# Check autodetect setting
print(f"Autodetect enabled: {lsm.autodetect}")

print("Session management ready")
```

## Example 4: jupyter_lsp functionality test 4

```python
import jupyter_lsp
from jupyter_lsp.specs import pyright, jedi, bash, yaml

# Example 4: Access specific language server specifications
print("Example 4: Language server specification access")

# Check available Python language servers
python_servers = [
    ("Pyright", pyright),
    ("Jedi", jedi),
]

print("Python language servers:")
for name, server_spec in python_servers:
    print(f"  {name}:")
    print(f"    Type: {type(server_spec).__name__}")
    print(f"    Key: {server_spec.key}")
    print(f"    Languages: {server_spec.languages}")

# Check other language servers
other_servers = [
    ("Bash", bash),
    ("YAML", yaml),
]

print("\nOther language servers:")
for name, server_spec in other_servers:
    print(f"  {name}:")
    print(f"    Key: {server_spec.key}")
    print(f"    Languages: {server_spec.languages}")

# Check if servers are installed
from jupyter_lsp import LanguageServerManager
lsm = LanguageServerManager()

print(f"\nChecking installation status:")
for name, server_spec in python_servers + other_servers:
    installed = server_spec.is_installed(lsm)
    assert isinstance(installed, bool), (name, installed)
    print(f"  {name}: {'Installed' if installed else 'Not installed'}")

print("Language server specs accessed")
```

## Example 5: jupyter_lsp functionality test 5

```python
import jupyter_lsp
from jupyter_lsp import LanguageServerManager

# Example 5: Language server configuration and diagnostics
print("Example 5: LSP configuration and diagnostics")

# Create manager with logging
lsm = LanguageServerManager()

# Check configuration options
config_traits = lsm.trait_names()
lsp_traits = [t for t in config_traits if 'language' in t.lower() or 'server' in t.lower()]
print(f"LSP-related config traits: {len(lsp_traits)}")

# Show available diagnostic information
if hasattr(lsm, 'log'):
    print("Logging configured: Yes")
    
# Check server autodetection
autodetect_methods = [m for m in dir(lsm) if 'autodetect' in m.lower()]
if autodetect_methods:
    print(f"Autodetect methods: {autodetect_methods}")

# Get current configuration state
current_config = {}
for trait in ['language_servers', 'node_roots']:
    if hasattr(lsm, trait):
        value = getattr(lsm, trait, None)
        current_config[trait] = type(value).__name__
        
print(f"Configuration state: {current_config}")

# Check if manager is ready for LSP operations
print(f"Ready for LSP: {lsm.all_language_servers is not None}")
print("LSP diagnostics complete")
```
