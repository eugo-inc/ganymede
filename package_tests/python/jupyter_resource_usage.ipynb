## Example 1: Functionality Test 1

```python
import jupyter_resource_usage
from jupyter_resource_usage.config import ResourceUseDisplay

# Example 1: Configure resource usage display settings
print("Example 1: ResourceUseDisplay configuration")

# Create a resource usage display configuration
display_config = ResourceUseDisplay()

# Check default configuration
print(f"Config type: {type(display_config).__name__}")
print(f"Mem limit: {display_config.mem_limit}")
print(f"Track CPU: {display_config.track_cpu_percent}")
print(f"CPU limit: {display_config.cpu_limit}")

# Check available traits
traits = display_config.trait_names()
resource_traits = [t for t in traits if 'mem' in t or 'cpu' in t]
print(f"Resource-related traits: {len(resource_traits)}")
if resource_traits:
    print(f"Sample traits: {resource_traits[:3]}")

# Show current configuration
print(f"Memory warning threshold: {display_config.mem_warning_threshold}")
print(f"CPU warning threshold: {display_config.cpu_warning_threshold}")

print("Resource display configuration loaded")
```

## Example 2: Functionality Test 2

```python
import jupyter_resource_usage
from jupyter_resource_usage import metrics
import psutil

# Example 2: Work with system metrics using psutil directly
print("Example 2: System metrics collection")

# Access psutil module from jupyter_resource_usage
psutil_module = metrics.psutil
print(f"PSUtil module: {type(psutil_module).__name__}")

# Get basic system metrics directly
try:
    # Memory information
    memory_info = psutil_module.virtual_memory()
    print(f"Total memory: {memory_info.total // (1024**3)} GB")
    print(f"Available memory: {memory_info.available // (1024**3)} GB")
    print(f"Memory usage: {memory_info.percent}%")
    
    # CPU information
    cpu_percent = psutil_module.cpu_percent(interval=1)
    cpu_count = psutil_module.cpu_count()
    print(f"CPU usage: {cpu_percent}%")
    print(f"CPU cores: {cpu_count}")
    
except Exception as e:
    print(f"Metrics collection: {type(e).__name__}")
    raise

# Check PSUtilMetricsLoader class availability
PSUtilMetricsLoader = metrics.PSUtilMetricsLoader
print(f"PSUtilMetricsLoader available: {PSUtilMetricsLoader is not None}")

print("System metrics demonstration complete")
```

## Example 3: Functionality Test 3

```python
import jupyter_resource_usage
from jupyter_resource_usage.config import PSUtilMetric, ResourceUseDisplay

# Example 3: Work with PSUtilMetric as a configuration trait
print("Example 3: PSUtilMetric trait configuration")

# Create a PSUtilMetric trait with default values
cpu_metric = PSUtilMetric(
    default_value="cpu_percent",
    help="CPU usage percentage metric"
)

memory_metric = PSUtilMetric(
    default_value="virtual_memory", 
    help="Memory usage metric"
)

# Check metric trait properties
print(f"CPU metric type: {type(cpu_metric).__name__}")
print(f"CPU metric help: {cpu_metric.help}")
print(f"CPU metric default: {cpu_metric.default_value}")

print(f"Memory metric help: {memory_metric.help}")
print(f"Memory metric default: {memory_metric.default_value}")

# Show how these would be used in ResourceUseDisplay
display_config = ResourceUseDisplay()
print(f"\nResourceUseDisplay type: {type(display_config).__name__}")

# Check if display config has metric-related traits
trait_names = display_config.trait_names()
metric_traits = [t for t in trait_names if 'metric' in t.lower()]
print(f"Metric-related traits: {metric_traits}")

print("PSUtilMetric trait configuration complete")
```

## Example 4: Functionality Test 4

```python
import jupyter_resource_usage
from jupyter_resource_usage.api import KernelUsageHandler, APIHandler

# Example 4: Inspect KernelUsageHandler class
print("Example 4: KernelUsageHandler API inspection")

# Inspect the handler class without instantiating
print(f"Handler class: {KernelUsageHandler}")
print(f"Handler type: {type(KernelUsageHandler).__name__}")

# Check class hierarchy
mro = KernelUsageHandler.__mro__
print(f"Class hierarchy: {[cls.__name__ for cls in mro[:3]]}")

# Check class methods and attributes
class_methods = [m for m in dir(KernelUsageHandler) if not m.startswith('_')]
http_methods = [m for m in class_methods if m.lower() in ['get', 'post', 'put', 'delete']]
print(f"HTTP methods: {http_methods}")

# Check APIHandler as well
print(f"\nAPIHandler class: {APIHandler}")
api_methods = [m for m in dir(APIHandler) if not m.startswith('_') and m not in class_methods]
print(f"Additional API methods: {len(api_methods)}")

# Check for URL patterns or routing info
if hasattr(KernelUsageHandler, 'get'):
    print("Handler supports GET requests")
if hasattr(KernelUsageHandler, 'post'):
    print("Handler supports POST requests")

print("KernelUsageHandler inspection complete")
```

## Example 5: Functionality Test 5

```python
import jupyter_resource_usage
from jupyter_resource_usage import load_jupyter_server_extension

# Example 5: Explore server extension functionality
print("Example 5: Server extension exploration")

# Check extension loading function
print(f"Extension loader: {load_jupyter_server_extension}")
print(f"Function type: {type(load_jupyter_server_extension).__name__}")

# Instead of loading, explore what the extension provides
from jupyter_resource_usage import server_extension
print("Server extension module: Available")

# Check what's in the server extension
ext_attrs = [attr for attr in dir(server_extension) if not attr.startswith('_')]
print(f"Extension attributes: {ext_attrs}")
assert server_extension.load_jupyter_server_extension is load_jupyter_server_extension
assert {'ApiHandler', 'PSUtilMetricsLoader', 'PrometheusHandler'} <= set(ext_attrs), ext_attrs


# Check prometheus module availability
from jupyter_resource_usage import prometheus
print("Prometheus metrics: Available")

# Check prometheus module contents
prom_attrs = [attr for attr in dir(prometheus) if not attr.startswith('_')]
print(f"Prometheus functions: {len(prom_attrs)}")
assert {'PSUtilMetricsLoader', 'PrometheusHandler'} <= set(prom_attrs), prom_attrs


# Check package version and info
if hasattr(jupyter_resource_usage, '__version__'):
    print(f"Package version: {jupyter_resource_usage.__version__}")

print("Server extension exploration complete")
```
