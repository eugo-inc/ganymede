## Example 1: Package Version and Information

```python
import jupyter_secrets_manager
import warnings

# Example 1: Get package version and basic information
print("Example 1: Package version and information")

# Get package version
try:
    version = jupyter_secrets_manager.__version__
    print(f"Package version: {version}")
except AttributeError:
    print("Version information not available")

# Check if package imports correctly
print(f"Package type: {type(jupyter_secrets_manager).__name__}")
print(f"Package location: {jupyter_secrets_manager.__file__}")

# Show package attributes
attrs = [attr for attr in dir(jupyter_secrets_manager) if not attr.startswith('_')]
print(f"Public attributes: {len(attrs)}")
if attrs:
    print(f"Available: {attrs}")

print("Package information retrieved")
```

## Example 2: JupyterLab Extension Paths

```python
import jupyter_secrets_manager
import os

# Example 2: Access JupyterLab extension configuration
print("Example 2: JupyterLab extension paths")

# Get extension paths using the function
try:
    extension_paths = jupyter_secrets_manager._jupyter_labextension_paths()
    print(f"Extension paths type: {type(extension_paths).__name__}")
    print(f"Number of extension paths: {len(extension_paths)}")
    
    # Show each extension path configuration
    for i, path_config in enumerate(extension_paths):
        print(f"Extension {i + 1}:")
        print(f"  Source: {path_config.get('src', 'Not specified')}")
        print(f"  Destination: {path_config.get('dest', 'Not specified')}")
        
        # Check if source path exists relative to package
        package_dir = os.path.dirname(jupyter_secrets_manager.__file__)
        src_path = os.path.join(package_dir, path_config.get('src', ''))
        print(f"  Source exists: {os.path.exists(src_path)}")

except Exception as e:
    print(f"Extension path access: {type(e).__name__}")
    raise AssertionError(
        f"jupyter_secrets_manager example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Extension paths accessed")
```

## Example 3: Extension Asset Validation

```python
import jupyter_secrets_manager
import os
import json

# Example 3: Validate extension assets and structure
print("Example 3: Extension asset validation")

# Get package directory
package_dir = os.path.dirname(jupyter_secrets_manager.__file__)
print(f"Package directory: {os.path.basename(package_dir)}")

# Check for extension assets
extension_paths = jupyter_secrets_manager._jupyter_labextension_paths()

for path_config in extension_paths:
    src_dir = os.path.join(package_dir, path_config['src'])
    print(f"Checking source: {path_config['src']}")
    
    if os.path.exists(src_dir):
        # List contents of extension directory
        contents = os.listdir(src_dir)
        print(f"  Contents: {len(contents)} items")
        
        # Look for package.json
        package_json = os.path.join(src_dir, 'package.json')
        if os.path.exists(package_json):
            print("  package.json: Found")
        else:
            print("  package.json: Not found")
    else:
        print(f"  Directory not found: {src_dir}")

print("Asset validation complete")
```

## Example 4: Module Import Validation

```python
import jupyter_secrets_manager
import importlib.util
import warnings
import os

# Example 4: Validate module import and catch warnings
print("Example 4: Module import validation")

# Capture warnings during import
with warnings.catch_warnings(record=True) as warning_list:
    warnings.simplefilter("always")
    
    # Test reimport to catch any warnings
    spec = importlib.util.find_spec('jupyter_secrets_manager')
    # the package was imported above, so its spec must resolve to a real __init__.py
    assert spec is not None and spec.origin and os.path.basename(spec.origin) == '__init__.py', spec
    print(f"Module spec found: {spec.name}")
    print(f"Module origin: {os.path.basename(spec.origin)}")
    
    # Check for import warnings
    if warning_list:
        print(f"Import warnings: {len(warning_list)}")
        for warning in warning_list:
            print(f"  {warning.category.__name__}: {warning.message}")
    else:
        print("No import warnings detected")

# Test version access pattern
# the package exports __version__, and it is the installed distribution's version (measured: '0.5.1' == '0.5.1')
from importlib.metadata import version as dist_version
from jupyter_secrets_manager import __version__
assert __version__ == dist_version("jupyter_secrets_manager"), (__version__, dist_version("jupyter_secrets_manager"))
print(f"Version import successful: {__version__}")

print("Import validation complete")
```

## Example 5: Extension Compatibility Check

```python
import jupyter_secrets_manager
import sys
import os

# Example 5: Check extension compatibility and environment
print("Example 5: Extension compatibility check")

# Check Python version compatibility
python_version = f"{sys.version_info.major}.{sys.version_info.minor}"
print(f"Python version: {python_version}")

# Check if this is a proper installation
package_file = jupyter_secrets_manager.__file__
is_dev_install = 'site-packages' not in package_file
print(f"Development install: {is_dev_install}")
print(f"Install location: {'Development' if is_dev_install else 'Site packages'}")

# Check extension function availability
extension_func = getattr(jupyter_secrets_manager, '_jupyter_labextension_paths', None)
print(f"Extension function available: {extension_func is not None}")

if extension_func:
    # Verify extension configuration
    ext_config = extension_func()
    print(f"Extension configuration valid: {isinstance(ext_config, list)}")
    
    if ext_config:
        config = ext_config[0]
        required_keys = ['src', 'dest']
        has_required = all(key in config for key in required_keys)
        print(f"Required config keys present: {has_required}")

print("Compatibility check complete")
```
