## Example 1: jupyter_server_mathjax functionality test 1

```python
import jupyter_server_mathjax
from jupyter_server_mathjax import MathJaxExtension

# Example 1: Basic MathJaxExtension class instantiation and inspection
print("=== Example 1: MathJaxExtension Class Usage ===")

# Create an instance of MathJaxExtension
extension = MathJaxExtension()
print(f"Extension name: {extension.name}")
print(f"Extension class: {type(extension).__name__}")

# Check key attributes
print(f"Extension URL: {getattr(extension, 'extension_url', 'Not set')}")
print(f"File URL prefix: {getattr(extension, 'file_url_prefix', 'Not set')}")

# Show extension description
if hasattr(extension, 'description'):
    print(f"Description: {extension.description}")

# Test trait system
print(f"Has trait system: {hasattr(extension, 'trait_names')}")
if hasattr(extension, 'trait_names'):
    trait_count = len(extension.trait_names())
    print(f"Number of traits: {trait_count}")

print("Example 1 completed successfully")
```

## Example 2: jupyter_server_mathjax functionality test 2

```python
import jupyter_server_mathjax
from jupyter_server_mathjax import MathJaxExtension

# Example 2: Extension configuration and MathJax settings
print("=== Example 2: Extension Configuration and Settings ===")

extension = MathJaxExtension()

# Check MathJax configuration
if hasattr(extension, 'mathjax_config'):
    mathjax_config = extension.mathjax_config
    print(f"MathJax config type: {type(mathjax_config)}")
    print(f"MathJax config value: {mathjax_config}")

# Examine static paths
if hasattr(extension, 'static_paths'):
    static_paths = extension.static_paths
    print(f"Static paths: {static_paths}")

# Check template paths
if hasattr(extension, 'template_paths'):
    template_paths = extension.template_paths
    print(f"Template paths: {template_paths}")

# Show default URL configuration
default_url = getattr(extension, 'default_url', None)
print(f"Default URL: {default_url}")

# Display configuration-related traits
config_traits = [attr for attr in dir(extension) if 'config' in attr.lower() and not attr.startswith('_')]
print(f"Configuration-related attributes: {config_traits[:5]}")

print("Example 2 completed successfully")
```

## Example 3: jupyter_server_mathjax functionality test 3

```python
import jupyter_server_mathjax
from jupyter_server_mathjax.app import MathJaxExtension

# Example 3: Server application integration and initialization
print("=== Example 3: Server Application Integration ===")

extension = MathJaxExtension()

# Test initialization methods
print("Testing initialization methods...")

try:
    # Initialize settings (safe to call)
    extension.initialize_settings()
    print("✓ initialize_settings() completed")
except Exception as e:
    print(f"initialize_settings() error: {e}")
    raise

# Check server app integration
print(f"Server app class: {getattr(extension, 'serverapp_class', 'Not set')}")
print(f"Server app: {getattr(extension, 'serverapp', 'Not initialized')}")

# Show extension package information
extension_package = extension.get_extension_package()
print(f"Extension package: {extension_package}")

# Display launch-related methods
launch_methods = [method for method in dir(extension) 
                 if any(keyword in method.lower() for keyword in ['launch', 'start', 'stop', 'initialize'])]
print(f"Launch/control methods: {launch_methods[:8]}")

print("Example 3 completed successfully")
```

## Example 4: jupyter_server_mathjax functionality test 4

```python
import jupyter_server_mathjax
from jupyter_server_mathjax.app import STATIC_ASSETS_PATH
from pathlib import Path

# Example 4: MathJax static file serving and assets
print("=== Example 4: MathJax Static File Serving ===")

# Check static assets path
print(f"Static assets path: {STATIC_ASSETS_PATH}")
print(f"Path exists: {Path(STATIC_ASSETS_PATH).exists()}")

# List contents of static assets if path exists
if Path(STATIC_ASSETS_PATH).exists():
    assets = list(Path(STATIC_ASSETS_PATH).iterdir())[:5]  # Show first 5 items
    print(f"Asset files (first 5): {[asset.name for asset in assets]}")

# Check extension's static configuration
extension = jupyter_server_mathjax.MathJaxExtension()
if hasattr(extension, 'static_url_prefix'):
    print(f"Static URL prefix: {extension.static_url_prefix}")

# Test URL path joining utility
from jupyter_server_mathjax.app import url_path_join
sample_url = url_path_join("base", "path", "file.js")
print(f"Sample URL join: {sample_url}")

# Show static-related attributes
static_attrs = [attr for attr in dir(extension) if 'static' in attr.lower()]
print(f"Static-related attributes: {static_attrs}")

print("Example 4 completed successfully")
```

## Example 5: jupyter_server_mathjax functionality test 5

```python
import jupyter_server_mathjax
from jupyter_server_mathjax.app import RedirectHandler, DeprecatedRedirectHandler

# Example 5: Extension handlers and URL routing
print("=== Example 5: Extension Handlers and URL Routing ===")

extension = jupyter_server_mathjax.MathJaxExtension()

# Check handlers initialization
try:
    # Get handlers (may be empty without server context)
    handlers = getattr(extension, 'handlers', [])
    print(f"Current handlers: {len(handlers)} registered")
    
    # Show handler classes available
    print(f"RedirectHandler available: {RedirectHandler}")
    print(f"DeprecatedRedirectHandler available: {DeprecatedRedirectHandler}")
    
except Exception as e:
    print(f"Handler access error: {e}")
    raise

# Check handler-related methods
handler_methods = [method for method in dir(extension) 
                  if 'handler' in method.lower()]
print(f"Handler-related methods: {handler_methods}")

# Test extension point
extension_point = extension.get_extension_point()
print(f"Extension point: {extension_point}")

# Show URL-related attributes
url_attrs = [attr for attr in dir(extension) if 'url' in attr.lower()]
print(f"URL-related attributes: {url_attrs}")

# Test close handlers method (safe to call)
try:
    extension.close_handlers()
    print("✓ close_handlers() completed")
except Exception as e:
    print(f"close_handlers() error: {e}")
    raise

print("Example 5 completed successfully")
```
