## Example 1: jupyter_server_proxy functionality test 1

```python
import jupyter_server_proxy
from jupyter_server_proxy import ServerProxyConfig
from jupyter_server_proxy.config import ServerProxy, LauncherEntry

# Example 1: ServerProxy configuration and setup
print("=== Example 1: ServerProxy Configuration ===")

# Create a ServerProxyConfig instance (from main module)
proxy_config = ServerProxyConfig()
print(f"ServerProxyConfig class: {type(proxy_config).__name__}")

# Create a ServerProxy instance (from config submodule)
server_proxy = ServerProxy()
print(f"ServerProxy class: {type(server_proxy).__name__}")

# Check ServerProxy configuration traits
trait_names = server_proxy.class_trait_names()
print(f"ServerProxy traits available: {len(trait_names)}")
print(f"Sample traits: {list(trait_names)[:5]}")

# LauncherEntry is a traitlets HasTraits — construct with keyword arguments
launcher = LauncherEntry(
    enabled=True,
    icon_path="/path/to/icon.svg",
    title="Test Server",
    path_info="",
    category="Development",
)
print(f"LauncherEntry created: '{launcher.title}' (enabled: {launcher.enabled}, category: {launcher.category})")

# Show LauncherEntry traits (HasTraits — no NamedTuple _fields)
print(f"LauncherEntry traits: {sorted(launcher.trait_names())}")
assert launcher.title == "Test Server"

# Show ServerProxy configurable attributes count
config_attrs = [attr for attr in dir(server_proxy) if not attr.startswith('_')]
print(f"ServerProxy has {len(config_attrs)} attributes/methods")

print("Example 1 completed successfully")
```

## Example 4: jupyter_server_proxy functionality test 4

```python
import jupyter_server_proxy
from jupyter_server_proxy import ujoin
from jupyter_server_proxy.utils import call_with_asked_args

# Example 4: URL path handling and utility functions
print("=== Example 4: URL Path Handling ===")

# Test ujoin function for URL path joining
test_paths = [
    ("base", "path"),
    ("http://localhost:8080", "api", "v1"),
    ("/proxy", "service", "endpoint"),
    ("", "relative", "path")
]

for paths in test_paths:
    joined = ujoin(*paths)
    print(f"ujoin{paths} -> '{joined}'")

# Test call_with_asked_args utility function
def sample_function(arg1, arg2="default", arg3=None):
    return f"Called with: {arg1}, {arg2}, {arg3}"

# call_with_asked_args passes ONLY the arguments the callback's signature names and drops the rest. It does
# not honour defaults (its own source: `FIXME: support default args`) — every named argument must be supplied.
result1 = call_with_asked_args(sample_function, {"arg1": "value1", "arg2": "value2", "arg3": "value3", "unused": 0})
print(f"call_with_asked_args result 1: {result1}")
assert result1 == "Called with: value1, value2, value3", result1

try:
    result2 = call_with_asked_args(sample_function, {"arg1": "value1", "arg2": "value2"})
except TypeError as e:
    assert "missing required positional argument: arg3" in str(e), e
    print(f"call_with_asked_args refused a missing argument as expected: {e}")
else:
    raise AssertionError(f"call_with_asked_args ran without a required argument: {result2}")

# Show URL joining edge cases
edge_cases = ujoin("//double//slash//", "path")
print(f"Edge case handling: '{edge_cases}'")

print("Example 4 completed successfully")
```
