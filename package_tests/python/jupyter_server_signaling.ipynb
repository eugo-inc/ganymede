## Example 1: SignalingExtension class instantiation and inspection

```python
import jupyter_server_signaling
from jupyter_server_signaling import SignalingExtension

# Example 1: SignalingExtension class instantiation and inspection
print("=== Example 1: SignalingExtension Class Usage ===")

# Create a SignalingExtension instance
extension = SignalingExtension()
print(f"Extension class: {type(extension).__name__}")

# Check extension name and version
if hasattr(extension, 'app_name'):
    print(f"App name: {extension.app_name}")

if hasattr(extension, 'name'):
    print(f"Extension name: {extension.name}")

# Show extension traits/configuration
if hasattr(extension, 'class_trait_names'):
    trait_names = extension.class_trait_names()
    print(f"Available traits: {len(trait_names)} total")
    print(f"Sample traits: {list(trait_names)[:5]}")

# Display extension attributes
ext_attrs = [attr for attr in dir(extension) if not attr.startswith('_')]
print(f"Extension methods/attributes: {len(ext_attrs)} available")

# Check if extension has configuration methods
config_methods = [method for method in ext_attrs if 'config' in method.lower()]
print(f"Configuration methods: {config_methods[:5]}")

print("Example 1 completed successfully")
```

## Example 2: SignalingWebSocketHandler class inspection

```python
import jupyter_server_signaling
from jupyter_server_signaling.handlers import SignalingWebSocketHandler

# Example 2: SignalingWebSocketHandler class inspection
print("=== Example 2: SignalingWebSocketHandler Functionality ===")

# Inspect the WebSocket handler class
handler_class = SignalingWebSocketHandler
print(f"Handler class: {handler_class.__name__}")
print(f"Handler type: {type(handler_class)}")

# Check handler methods
handler_methods = [method for method in dir(handler_class) if not method.startswith('_')]
print(f"Handler methods available: {len(handler_methods)}")
print(f"Sample methods: {handler_methods[:8]}")

# Check for WebSocket-specific methods
ws_methods = [method for method in handler_methods if any(ws_term in method.lower() 
              for ws_term in ['open', 'close', 'message', 'write', 'ping', 'pong'])]
print(f"WebSocket methods: {ws_methods}")

# Check for async methods
async_methods = [method for method in handler_methods if method.startswith('on_')]
print(f"Event handler methods: {async_methods}")

# Check handler inheritance
mro = handler_class.__mro__
print(f"Handler inheritance chain: {[cls.__name__ for cls in mro[:4]]}")

# Show handler module location
print(f"Handler module: {handler_class.__module__}")

print("Example 2 completed successfully")
```

## Example 4: WebSocket message handling and communication patterns

```python
import jupyter_server_signaling
from jupyter_server_signaling.handlers import SignalingWebSocketHandler
import json
import inspect

# Example 4: WebSocket message handling and communication patterns
print("=== Example 4: WebSocket Message Handling ===")

# Examine message handling capabilities
handler_class = SignalingWebSocketHandler
print(f"WebSocket handler: {handler_class.__name__}")

# Check for message processing methods
message_methods = [method for method in dir(handler_class) 
                  if any(term in method.lower() for term in ['message', 'send', 'receive'])]
print(f"Message handling methods: {message_methods}")

# Check for JSON-related functionality
json_methods = [method for method in dir(handler_class) if 'json' in method.lower()]
print(f"JSON handling methods: {json_methods}")

# Examine async method signatures
async_methods = ['on_message', 'open', 'on_close'] 
for method_name in async_methods:
    if hasattr(handler_class, method_name):
        method = getattr(handler_class, method_name)
        try:
            sig = inspect.signature(method)
            print(f"{method_name}{sig}")
        except:
            print(f"{method_name}: signature unavailable")
            raise

# Test JSON serialization (common in WebSocket communication)
sample_message = {"type": "signal", "data": {"status": "connected"}}
serialized = json.dumps(sample_message)
print(f"Sample message: {serialized}")

# Show handler's supported operations
print(f"Handler supports WebSocket signaling communication")

print("Example 4 completed successfully")
```

## Example 5: Package structure and handler setup exploration

```python
import jupyter_server_signaling
from jupyter_server_signaling import handlers, app

# Example 5: Package structure and handler setup exploration
print("=== Example 5: Handler Setup and Package Configuration ===")

# Examine the handlers module
handlers_module = handlers
print(f"Handlers module: {handlers_module.__name__}")

# List all handler components
handler_attrs = [attr for attr in dir(handlers_module) if not attr.startswith('_')]
print(f"Handler components: {handler_attrs}")

# Check app module components
app_module = app
app_attrs = [attr for attr in dir(app_module) if not attr.startswith('_')]
print(f"App module components: {app_attrs}")

# Show package version and metadata
print(f"Package version: {getattr(jupyter_server_signaling, '__version__', 'unknown')}")

# Examine typing components used in the package
typing_components = ['Any', 'Dict', 'List']
for component in typing_components:
    if hasattr(jupyter_server_signaling, component):
        item = getattr(jupyter_server_signaling, component)
        print(f"{component}: {type(item)}")

# Check for extension setup patterns
extension_methods = ['load_jupyter_server_extension', 'unload_jupyter_server_extension']
for method in extension_methods:
    if hasattr(jupyter_server_signaling, method):
        print(f"Extension method available: {method}")

# Show package structure summary
print(f"Main package exports: {[attr for attr in dir(jupyter_server_signaling) if not attr.startswith('_')]}")

print("Example 5 completed successfully")
```
