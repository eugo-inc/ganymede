## Example 1: YDocExtension configuration and initialization

```python
import jupyter_server_ydoc
from jupyter_server_ydoc import YDocExtension

# Example 1: YDocExtension configuration and initialization
print("=== Example 1: YDocExtension Configuration ===")

# Create and configure YDocExtension
extension = YDocExtension()
print(f"Created YDocExtension: {type(extension).__name__}")

# Test extension initialization
try:
    # Initialize extension settings
    extension.initialize_settings()
    print("✓ Extension settings initialized successfully")
except Exception as e:
    print(f"Settings initialization: {e}")
    raise

# Configure extension with custom settings
extension.log_level = "INFO"
print(f"Set log level: {extension.log_level}")

# Test extension configuration
if hasattr(extension, 'config'):
    print(f"Extension config type: {type(extension.config)}")

# Show extension is ready for server integration
print(f"Extension ready for Jupyter server integration")
print(f"Package version: {jupyter_server_ydoc.__version__}")

print("Example 1 completed successfully")
```

## Example 2: YDoc store creation and document persistence

```python
import jupyter_server_ydoc
from jupyter_server_ydoc.stores import TempFileYStore
import tempfile
import os

# Example 2: YDoc store creation and document persistence
print("=== Example 2: Document Store and Persistence ===")

# Create a temporary file store
with tempfile.TemporaryDirectory() as temp_dir:
    store_path = os.path.join(temp_dir, "ydoc_store.db")
    
    # Initialize TempFileYStore
    store = TempFileYStore(path=store_path)
    print(f"Created TempFileYStore at: {store_path}")
    
    # Test store configuration
    print(f"Store path configured: {store.path}")
    print(f"Store class: {type(store).__name__}")
    
    # Demonstrate store is ready for document storage
    print("✓ Store ready for document persistence")
    
    # Show store exists and is accessible
    if hasattr(store, 'path') and store.path:
        print(f"✓ Store path validation successful")
    
    # Test store cleanup (automatic with context manager)
    print("✓ Store will be cleaned up automatically")

print("Document store demonstration completed")
print("Example 2 completed successfully")
```

## Example 5: SQLite document store operations

```python
import jupyter_server_ydoc
from jupyter_server_ydoc.stores import SQLiteYStore
import sqlite3
import tempfile
import os

# Example 5: SQLite document store operations
print("=== Example 5: SQLite Document Store Operations ===")

# Create temporary database for testing
with tempfile.NamedTemporaryFile(suffix='.db', delete=False) as tmp_db:
    db_path = tmp_db.name

try:
    # Initialize SQLiteYStore with custom database path
    sqlite_store = SQLiteYStore(path=db_path)
    print(f"Created SQLiteYStore with database: {db_path}")
    
    # Test store configuration
    print(f"Database path: {sqlite_store.path}")
    print(f"✓ SQLite store configured successfully")
    
    # Verify database file creation
    if os.path.exists(db_path):
        print(f"✓ Database file created at: {db_path}")
    
    # Test database connectivity
    try:
        conn = sqlite3.connect(db_path)
        conn.close()
        print("✓ Database connection test successful")
    except Exception as e:
        print(f"Database connection test: {e}")
        raise
    
    # Show store is ready for document persistence
    print("✓ SQLite store ready for Y document operations")
    
finally:
    # Clean up test database
    if os.path.exists(db_path):
        os.unlink(db_path)
        print("✓ Test database cleaned up")

print("Example 5 completed successfully")
```
