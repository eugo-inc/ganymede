## Example 1: jupyter_telemetry functionality test 1

```python
import jupyter_telemetry
from jupyter_telemetry import EventLog

# Example 1: EventLog creation and basic event recording
print("=== Example 1: EventLog Event Recording ===")

# Create EventLog instance
event_logger = EventLog()
print(f"✓ EventLog created: {type(event_logger).__name__}")

# Register a simple schema for user events
user_schema = {
    "$id": "https://example.com/user-action.json",
    "version": 1,
    "type": "object",
    "properties": {
        "action": {"type": "string"},
        "user_id": {"type": "string"},
        "timestamp": {"type": "string"}
    }
}

# Register the schema
event_logger.register_schema(user_schema)
print("✓ Schema registered successfully")

# Record an event using the registered schema
event_data = {
    "action": "login",
    "user_id": "test_user_123",
    "timestamp": "2023-10-01T10:00:00Z"
}

event_logger.record_event("https://example.com/user-action.json", 1, event_data)
print("✓ Event recorded successfully")
print(f"✓ Event data: {event_data}")

print("Example 1 completed successfully")
```

## Example 2: jupyter_telemetry functionality test 2

```python
import jupyter_telemetry
from jupyter_telemetry import EventLog
import logging

# Example 2: EventLog configuration with logging handlers
print("=== Example 2: EventLog Handler Configuration ===")

# Create EventLog with custom logging handlers
event_logger = EventLog()

# Configure StreamHandler for console output
console_handler = logging.StreamHandler()
console_handler.setFormatter(logging.Formatter('%(levelname)s - %(message)s'))

# Configure event logger with handlers
event_logger.handlers = [console_handler]
print(f"✓ EventLog configured with {len(event_logger.handlers)} handler(s)")

# Register schema for performance events
perf_schema = {
    "$id": "https://example.com/performance.json",
    "version": 1,
    "type": "object",
    "properties": {
        "operation": {"type": "string"},
        "duration_ms": {"type": "number"},
        "status": {"type": "string", "enum": ["success", "error"]}
    }
}

event_logger.register_schema(perf_schema)
print("✓ Performance schema registered")

# Record performance event
perf_event = {
    "operation": "notebook_save",
    "duration_ms": 150.5,
    "status": "success"
}

event_logger.record_event("https://example.com/performance.json", 1, perf_event)
print("✓ Performance event recorded with handler output")

print("Example 2 completed successfully")
```

## Example 3: jupyter_telemetry functionality test 3

```python
import jupyter_telemetry
from jupyter_telemetry import EventLog, TELEMETRY_METADATA_VERSION
import tempfile
import json

# Example 3: Schema validation and file-based schema registration
print("=== Example 3: Schema Validation and File Registration ===")

# Create EventLog instance
event_logger = EventLog()
print(f"✓ EventLog created with metadata version: {TELEMETRY_METADATA_VERSION}")

# Create a schema file for testing register_schema_file method
schema_data = {
    "$id": "https://example.com/error-event.json",
    "version": 1,
    "type": "object",
    "properties": {
        "error_type": {"type": "string"},
        "message": {"type": "string"},
        "severity": {"type": "string", "enum": ["low", "medium", "high"]}
    },
    "required": ["error_type", "message"]
}

# Write schema to temporary file
with tempfile.NamedTemporaryFile(mode='w', suffix='.json', delete=False) as f:
    json.dump(schema_data, f)
    schema_file = f.name

# Use register_schema_file method
event_logger.register_schema_file(schema_file)
print("✓ Schema registered from file successfully")

# Test schema validation with valid event
valid_event = {
    "error_type": "validation_error",
    "message": "Schema validation failed",
    "severity": "medium"
}

event_logger.record_event("https://example.com/error-event.json", 1, valid_event)
print("✓ Valid event recorded successfully")

print("✓ Schema validation and file registration completed")

print("Example 3 completed successfully")
```

## Example 4: jupyter_telemetry functionality test 4

```python
import jupyter_telemetry
from jupyter_telemetry import EventLog, traits
import logging

# Example 4: Telemetry trait configuration and allowed schemas
print("=== Example 4: Trait Configuration and Schema Filtering ===")

# Create EventLog with trait configuration
event_logger = EventLog()

# Configure allowed schemas trait for filtering
event_logger.allowed_schemas = [
    "https://example.com/system-metric.json",
    "https://example.com/user-action.json"
]
print(f"✓ Allowed schemas configured: {len(event_logger.allowed_schemas)}")

# Register schemas (both allowed and not allowed)
system_schema = {
    "$id": "https://example.com/system-metric.json",
    "version": 1,
    "type": "object",
    "properties": {
        "metric_name": {"type": "string"},
        "value": {"type": "number"}
    }
}

event_logger.register_schema(system_schema)
print("✓ System metric schema registered")

# Test handlers trait with multiple logging handlers
console_handler = logging.StreamHandler()
event_logger.handlers = [console_handler]
print(f"✓ Handlers trait configured: {type(event_logger.handlers[0]).__name__}")

# Record event with allowed schema
metric_event = {
    "metric_name": "cpu_usage",
    "value": 75.2
}

event_logger.record_event("https://example.com/system-metric.json", 1, metric_event)
print("✓ Event recorded through trait-filtered system")

print("Example 4 completed successfully")
```

## Example 5: jupyter_telemetry functionality test 5

```python
import jupyter_telemetry
from jupyter_telemetry import EventLog
from datetime import datetime

# Example 5: Advanced event emission with timestamp override
print("=== Example 5: Advanced Event Emission and Timestamp Override ===")

# Create EventLog instance
event_logger = EventLog()

# Register schema for session events
session_schema = {
    "$id": "https://example.com/session.json",
    "version": 1,
    "type": "object",
    "properties": {
        "session_id": {"type": "string"},
        "event_type": {"type": "string"},
        "duration": {"type": "number"}
    }
}

event_logger.register_schema(session_schema)
print("✓ Session schema registered")

# Record event with custom timestamp using timestamp_override parameter
custom_timestamp = datetime.now().isoformat()
session_event = {
    "session_id": "sess_12345",
    "event_type": "kernel_start",
    "duration": 2.5
}

# Use record_event with timestamp_override
event_logger.record_event(
    "https://example.com/session.json", 
    1, 
    session_event,
    timestamp_override=custom_timestamp
)
print(f"✓ Event recorded with custom timestamp: {custom_timestamp}")

# Record another event without timestamp override (uses automatic timestamp)
end_event = {
    "session_id": "sess_12345",
    "event_type": "kernel_shutdown",
    "duration": 0.8
}

event_logger.record_event("https://example.com/session.json", 1, end_event)
print("✓ Event recorded with automatic timestamp")

print("✓ Advanced event emission with timestamp control completed")

print("Example 5 completed successfully")
```
