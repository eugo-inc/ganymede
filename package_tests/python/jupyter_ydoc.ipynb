## Example 1: YNotebook for collaborative notebook editing

```python
from jupyter_ydoc import YNotebook
import json

# Example 1: Create and manipulate a YNotebook document
print("Example 1: YNotebook for collaborative notebook editing")

# Create a new YNotebook instance
notebook = YNotebook()
print(f"Created YNotebook: {type(notebook).__name__}")

# Access notebook structure
print(f"YNotebook source available: {hasattr(notebook, 'source')}")
print(f"YNotebook dirty state: {notebook.dirty if hasattr(notebook, 'dirty') else 'N/A'}")

# Get the document's string representation
notebook_str = str(notebook)
print(f"Notebook string length: {len(notebook_str)}")

# Check for collaborative features
has_observe = hasattr(notebook, 'observe')
print(f"Has observe method for changes: {has_observe}")

# Get notebook version if available
version = notebook.version if hasattr(notebook, 'version') else 4
print(f"Notebook format version: {version}")

print("Example 1 completed successfully")
```

## Example 2: YFile for shared file documents

```python
from jupyter_ydoc import YFile
import base64

# Example 2: Working with YFile for document management
print("Example 2: YFile for shared file documents")

# Create a YFile instance
yfile = YFile()
print(f"Created YFile: {type(yfile).__name__}")

# Check file properties
has_path = hasattr(yfile, 'path')
has_content = hasattr(yfile, 'content')
print(f"Has path attribute: {has_path}")
print(f"Has content attribute: {has_content}")

# Get file source if available
if hasattr(yfile, 'source'):
    source = yfile.source
    print(f"YFile source type: {type(source).__name__}")
else:
    print("YFile source not directly accessible")

# Check for collaborative editing support
supports_collab = hasattr(yfile, 'get') or hasattr(yfile, 'set')
print(f"Supports collaborative operations: {supports_collab}")

print("Example 2 completed successfully")
```

## Example 3: YBlob for binary data handling

```python
from jupyter_ydoc import YBlob
import io

# Example 3: Using YBlob for binary data handling
print("Example 3: YBlob for binary data handling")

# Create a YBlob instance for binary content
blob = YBlob()
print(f"Created YBlob: {type(blob).__name__}")

# Check blob capabilities
can_store_binary = hasattr(blob, 'content') or hasattr(blob, 'data')
print(f"Can store binary data: {can_store_binary}")

# Test blob properties
is_doc = hasattr(blob, 'ydoc')
has_source = hasattr(blob, 'source')
print(f"Is YDoc type: {is_doc}")
print(f"Has source property: {has_source}")

# Check for size/length methods
has_size_info = hasattr(blob, '__len__') or hasattr(blob, 'size')
print(f"Has size information: {has_size_info}")

# Verify blob state
blob_state = str(type(blob))
print(f"Blob class info: {blob_state}")

print("Example 3 completed successfully")
```

## Example 4: YUnicode for text synchronization

```python
from jupyter_ydoc import YUnicode
import sys

# Example 4: Using YUnicode for text synchronization
print("Example 4: YUnicode for text synchronization")

# Create a YUnicode instance for text collaboration
yunicode = YUnicode()
print(f"Created YUnicode: {type(yunicode).__name__}")

# Check text manipulation capabilities
has_text_ops = hasattr(yunicode, 'text') or hasattr(yunicode, 'value')
print(f"Has text operations: {has_text_ops}")

# Test string representation
unicode_str = str(yunicode)
print(f"String representation exists: {unicode_str is not None}")

# Check for encoding support
supports_encoding = hasattr(yunicode, 'encode') or hasattr(yunicode, 'decode')
print(f"Supports encoding operations: {supports_encoding}")

# Verify collaborative text features
has_observe = hasattr(yunicode, 'observe')
has_unobserve = hasattr(yunicode, 'unobserve')
print(f"Has observer pattern: observe={has_observe}, unobserve={has_unobserve}")

print("Example 4 completed successfully")
```

## Example 5: Entry points and package utilities

```python
from jupyter_ydoc import entry_points
import inspect

# Example 5: Exploring entry points and package utilities
print("Example 5: Entry points and package utilities")

# Check entry_points function
if callable(entry_points):
    print(f"entry_points is callable: True")
    
    # Get entry points configuration
    eps = entry_points()
    print(f"Entry points type: {type(eps).__name__}")
    
    # Check if it returns a dictionary
    if isinstance(eps, dict):
        print(f"Number of entry point groups: {len(eps)}")
        for key in list(eps.keys())[:3]:  # Show first 3 keys
            print(f"  - {key}")
    else:
        print(f"Entry points returned: {eps}")
else:
    print(f"entry_points type: {type(entry_points).__name__}")

# Verify all main classes are imported
classes = ['YBlob', 'YFile', 'YNotebook', 'YUnicode']
for cls_name in classes:
    exists = hasattr(__import__('jupyter_ydoc'), cls_name)
    print(f"{cls_name} available: {exists}")

print("Example 5 completed successfully")
```
