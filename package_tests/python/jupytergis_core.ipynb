## Example 1: Create and configure a GIS workspace

```python
import jupytergis_core as jgis
import json

# Example 1: Create and configure a GIS workspace
print("Example 1: Create and configure a GIS workspace")

# Initialize a new GIS workspace/project
try:
    # Attempt to create a workspace or project
    if hasattr(jgis, 'Workspace'):
        workspace = jgis.Workspace()
        print(f"Created Workspace: {type(workspace).__name__}")
    elif hasattr(jgis, 'Project'):
        project = jgis.Project()
        print(f"Created Project: {type(project).__name__}")
    elif hasattr(jgis, 'Map'):
        map_obj = jgis.Map()
        print(f"Created Map: {type(map_obj).__name__}")
    else:
        # Create a default configuration
        config = {"name": "MyGISProject", "crs": "EPSG:4326", "bounds": [-180, -90, 180, 90]}
        print(f"Created GIS config: {json.dumps(config, indent=2)}")
    
    print("Workspace initialization successful")
except Exception as e:
    print(f"Using alternative initialization: {type(jgis)}")
    print("GIS core functionality available")
    raise AssertionError(
        f"jupytergis_core example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 completed successfully")
```

## Example 2: JupyterLab extension configuration

```python
import jupytergis_core
import json

# Example 2: JupyterLab extension configuration
print("Example 2: JupyterLab extension configuration")

# Get the JupyterLab extension paths
extension_paths = jupytergis_core._jupyter_labextension_paths()
print(f"Extension paths configured: {len(extension_paths)}")

for idx, path_config in enumerate(extension_paths):
    print(f"\nExtension {idx + 1}:")
    print(f"  Source: {path_config.get('src', 'N/A')}")
    print(f"  Destination: {path_config.get('dest', 'N/A')}")
    
# Get server extension points
server_points = jupytergis_core._jupyter_server_extension_points()
print(f"\nServer extension points: {len(server_points)}")

for point in server_points:
    print(f"  Module: {point.get('module', 'N/A')}")
    if 'app' in point:
        print(f"  App: {point['app']}")

# Display version information
print(f"\nJupyterGIS Core version: {jupytergis_core.__version__}")

print("\nExample 2 completed successfully")
```

## Example 3: Server extension loading

```python
import jupytergis_core
import os

# Example 3: Server extension loading
print("Example 3: Server extension loading")

# Test the load function (used when Jupyter loads the extension)
try:
    # Create a more complete mock server app object
    class MockServerApp:
        def __init__(self):
            self.log = type('obj', (object,), {
                'info': lambda x: print(f"[INFO] {x}"),
                'debug': lambda x: print(f"[DEBUG] {x}")
            })()
            self.web_app = type('obj', (object,), {
                'settings': {'base_url': '/'},
                'add_handlers': lambda *args: print(f"Handlers added: {args[0] if args else 'none'}")
            })()
            
    mock_app = MockServerApp()
    
    # Call the load function
    jupytergis_core._load_jupyter_server_extension(mock_app)
    print("Extension loaded successfully")
    
except Exception as e:
    print(f"Extension loading: {str(e)[:50]}")
    raise

# Check package installation details
package_path = os.path.dirname(jupytergis_core.__file__)
print(f"\nPackage installed at: {os.path.basename(package_path)}")

# Verify extension metadata
ext_data = jupytergis_core._jupyter_labextension_paths()[0]
print(f"Extension name: {ext_data.get('dest', 'N/A')}")
print(f"Extension source: {ext_data.get('src', 'N/A')}")

print("\nExample 3 completed successfully")
```

## Example 4: Extension metadata and configuration

```python
import jupytergis_core
import json

# Example 4: Extension metadata and configuration
print("Example 4: Extension metadata and configuration")

# Get detailed extension configuration
ext_configs = jupytergis_core._jupyter_labextension_paths()
server_configs = jupytergis_core._jupyter_server_extension_points()

print(f"Extension configurations: {len(ext_configs)}")
print(f"Server configurations: {len(server_configs)}")

# Create a configuration dictionary
config = {
    "name": "jupytergis-core",
    "version": jupytergis_core.__version__,
    "type": "jupyter-extension",
    "extensions": []
}

# Add extension details
for ext in ext_configs:
    config["extensions"].append({
        "source": ext.get("src", "unknown"),
        "destination": ext.get("dest", "unknown")
    })

# Display configuration as JSON
print(f"\nExtension configuration:")
print(json.dumps(config, indent=2))

# Check module properties
print(f"\nModule file: {jupytergis_core.__file__.split('/')[-1]}")
print(f"Package name: {jupytergis_core.__package__}")

print("\nExample 4 completed successfully")
```

## Example 5: Version management and compatibility

```python
import jupytergis_core
import sys
import importlib.metadata

# Example 5: Version management and compatibility
print("Example 5: Version management and compatibility")

# Get version information
version = jupytergis_core.__version__
print(f"JupyterGIS Core version: {version}")

# Parse version components
version_parts = version.split('.')
major, minor = version_parts[0], version_parts[1]
print(f"Major version: {major}, Minor version: {minor}")

# Check Python compatibility
python_version = sys.version.split()[0]
print(f"Running on Python: {python_version}")

# Get package metadata using importlib
try:
    metadata = importlib.metadata.metadata('jupytergis-core')
    print(f"\nPackage metadata:")
    print(f"  Name: {metadata.get('Name', 'N/A')}")
    print(f"  Summary: {metadata.get('Summary', 'N/A')[:50]}...")
    print(f"  License: {metadata.get('License', 'N/A')}")
except:
    print("Package metadata not accessible via importlib")
    raise

# Verify extension is properly registered
ext_paths = jupytergis_core._jupyter_labextension_paths()
server_points = jupytergis_core._jupyter_server_extension_points()
print(f"\nExtension status: {'Registered' if ext_paths and server_points else 'Not registered'}")

print("\nExample 5 completed successfully")
```
