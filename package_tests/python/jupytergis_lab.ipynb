## Example 1: Create and manipulate GIS documents

```python
from jupytergis_lab import GISDocument
import json

# Example 1: Create and manipulate GIS documents
print("Example 1: Create and manipulate GIS documents")

# Create a new GIS document
doc = GISDocument()
print(f"Created GISDocument: {type(doc).__name__}")

# Check document properties and methods
if hasattr(doc, 'layers'):
    print(f"Layers in document: {len(doc.layers) if hasattr(doc.layers, '__len__') else 'available'}")

if hasattr(doc, 'sources'):
    print(f"Sources configured: {len(doc.sources) if hasattr(doc.sources, '__len__') else 'available'}")

# Convert document to dictionary if possible
if hasattr(doc, 'to_dict'):
    doc_dict = doc.to_dict()
    print(f"Document structure keys: {list(doc_dict.keys())[:5]}")
    print(f"Document exported as dict with {len(doc_dict)} properties")
else:
    print("Document object created successfully")

print("Example 1 completed successfully")
```

## Example 2: Interactive GIS exploration

```python
from jupytergis_lab import explore
import json
import tempfile
from pathlib import Path

# Example 2: Interactive GIS exploration
print("Example 2: Interactive GIS exploration")

# Create sample GeoJSON data for exploration
sample_geojson = {
    "type": "FeatureCollection",
    "features": [
        {
            "type": "Feature",
            "geometry": {"type": "Point", "coordinates": [-122.4, 37.8]},
            "properties": {"city": "San Francisco", "population": 884363}
        },
        {
            "type": "Feature",
            "geometry": {"type": "Point", "coordinates": [-118.2, 34.0]},
            "properties": {"city": "Los Angeles", "population": 3990456}
        }
    ]
}

# Use explore function for interactive visualization. explore() takes a GeoDataFrame or the PATH of a
# GeoJSON file (a plain dict is rejected as an unsupported input type), and opens the map in a sidecar.
try:
    with tempfile.TemporaryDirectory() as tmp_dir:
        geojson_path = Path(tmp_dir) / "cities.geojson"
        geojson_path.write_text(json.dumps(sample_geojson))
        map_widget = explore(str(geojson_path), layer_name="Cities")
    print(f"Explore widget created: {type(map_widget).__name__ if map_widget else 'Widget initialized'}")
    print(f"Interactive map ready for display in Jupyter ({len(sample_geojson['features'])} features)")
except Exception as e:
    print(f"Explore function called with data: {len(sample_geojson['features'])} features")
    raise

print("Example 2 completed successfully")
```

## Example 3: Notebook integration for GIS workflows

```python
from jupytergis_lab import notebook
import os

# Example 3: Notebook integration for GIS workflows
print("Example 3: Notebook integration for GIS workflows")

# Use the notebook module for Jupyter integration
print(f"Notebook module type: {type(notebook)}")

# Check notebook capabilities
if hasattr(notebook, '__file__'):
    notebook_path = os.path.dirname(notebook.__file__)
    print(f"Notebook module location: {os.path.basename(notebook_path)}")

# Explore available notebook functions
notebook_attrs = [attr for attr in dir(notebook) if not attr.startswith('_')]
print(f"Notebook functions available: {len(notebook_attrs)}")
if notebook_attrs:
    print(f"Sample functions: {notebook_attrs[:3]}")

# Create a notebook-compatible GIS workflow
workflow_config = {
    "type": "gis-notebook",
    "cells": ["map_view", "data_analysis", "export"],
    "interactive": True
}
print(f"\nWorkflow configured with {len(workflow_config['cells'])} cells")

print("Example 3 completed successfully")
```

## Example 4: JupyterLab extension configuration

```python
from jupytergis_lab import _jupyter_labextension_paths
import json

# Example 4: JupyterLab extension configuration
print("Example 4: JupyterLab extension configuration")

# Get JupyterLab extension paths
extension_paths = _jupyter_labextension_paths()
print(f"Extension configurations found: {len(extension_paths)}")

# Process each extension configuration
for idx, ext_config in enumerate(extension_paths, 1):
    print(f"\nExtension {idx}:")
    print(f"  Source: {ext_config.get('src', 'not specified')}")
    print(f"  Destination: {ext_config.get('dest', 'not specified')}")
    
    # Check for additional configuration
    if 'require' in ext_config:
        print(f"  Requirements: {ext_config['require']}")

# Build extension metadata
metadata = {
    "name": "jupytergis-lab",
    "extensions": len(extension_paths),
    "type": "lab-extension",
    "status": "configured"
}
print(f"\nExtension metadata: {json.dumps(metadata, indent=2)}")

print("\nExample 4 completed successfully")
```
