## Example 1: Lightweight GIS map creation

```python
import jupytergis
from jupytergis import GISDocument

# Example 1: Lightweight GIS map creation
print("Example 1: Lightweight GIS map creation")

# Create a lightweight GIS document
doc = GISDocument()
print(f"Created GISDocument: {type(doc).__name__}")

# Configure basic map settings
map_config = {
    "center": [37.7749, -122.4194],
    "zoom": 10,
    "basemap": "OpenStreetMap"
}

# Check document capabilities
if hasattr(doc, 'layers'):
    print(f"Document has layers: {hasattr(doc.layers, '__len__')}")
if hasattr(doc, 'sources'):
    print(f"Document has sources: {hasattr(doc.sources, '__len__')}")

print(f"Map configured: center={map_config['center']}, zoom={map_config['zoom']}")
print("Example 1 completed successfully")
```

## Example 2: Quick GeoJSON visualization

```python
from jupytergis import explore

# Example 2: Quick GeoJSON visualization
print("Example 2: Quick GeoJSON visualization")

# Create sample GeoJSON data
geojson_data = {
    "type": "Feature",
    "geometry": {
        "type": "Polygon",
        "coordinates": [[
            [-122.5, 37.7],
            [-122.4, 37.7],
            [-122.4, 37.8],
            [-122.5, 37.8],
            [-122.5, 37.7]
        ]]
    },
    "properties": {"name": "Sample Area", "type": "demo"}
}

# Visualize the data using explore function. explore() takes a GeoDataFrame or a PATH to a GeoJSON
# file (a dict raises TypeError), and returns None: it opens the explorer as a sidecar widget.
import json
import os
import tempfile

with tempfile.TemporaryDirectory() as tmpdir:
    geojson_path = os.path.join(tmpdir, "sample_area.geojson")
    with open(geojson_path, "w") as fh:
        json.dump({"type": "FeatureCollection", "features": [geojson_data]}, fh)
    widget = explore(geojson_path)
assert widget is None
print(f"Visualization created for: {geojson_data['properties']['name']}")

print("Example 2 completed successfully")
```

## Example 3: Simple layer management

```python
import jupytergis
from jupytergis import GISDocument

# Example 3: Simple layer management
print("Example 3: Simple layer management")

# Create document with layers
doc = GISDocument()

# Define sample layers
layers_config = [
    {"name": "cities", "type": "point", "visible": True},
    {"name": "roads", "type": "line", "visible": True},
    {"name": "regions", "type": "polygon", "visible": False}
]

# Add layers to document (simulated)
if hasattr(doc, 'layers'):
    print(f"Document layers available: {type(doc.layers).__name__}")
    
print(f"Configured {len(layers_config)} layers:")
for layer in layers_config:
    print(f"  - {layer['name']}: {layer['type']} (visible: {layer['visible']})")

# Check if document can be converted to dict
if hasattr(doc, 'to_dict'):
    doc_dict = doc.to_dict()
    print(f"Document structure: {len(doc_dict)} properties")

print("Example 3 completed successfully")
```

## Example 4: Coordinate transformations

```python
import jupytergis
import math

# Example 4: Coordinate transformations
print("Example 4: Coordinate transformations")

# Define coordinate pairs
coords_wgs84 = [-122.4194, 37.7749]  # San Francisco in WGS84

# Create a GIS document for coordinate work
doc = jupytergis.GISDocument()

# Simple coordinate transformation demonstration
lon, lat = coords_wgs84
# Approximate Web Mercator transformation
x = lon * 111319.5
y = math.log(math.tan((90 + lat) * math.pi / 360)) * 6378137

print(f"Original WGS84: {coords_wgs84}")
print(f"Web Mercator (approx): [{x:.2f}, {y:.2f}]")

# Create point feature with coordinates
point_feature = {
    "type": "Feature",
    "geometry": {"type": "Point", "coordinates": coords_wgs84},
    "properties": {"name": "San Francisco", "crs": "EPSG:4326"}
}
print(f"Created point feature: {point_feature['properties']['name']}")

print("Example 4 completed successfully")
```

## Example 5: Export and version information

```python
import jupytergis
import json

# Example 5: Export and version information
print("Example 5: Export and version information")

# Get package version
version = jupytergis.__version__
print(f"JupyterGIS (lite) version: {version}")

# Create a GIS document with data
doc = jupytergis.GISDocument()

# Create sample data for export
export_data = {
    "type": "FeatureCollection",
    "features": [
        {"type": "Feature", 
         "geometry": {"type": "Point", "coordinates": [-122.4, 37.8]},
         "properties": {"city": "San Francisco"}}
    ]
}

# Export to JSON string
output_json = json.dumps(export_data, indent=2)
print(f"Data prepared for export: {len(output_json)} characters")
print(f"Features count: {len(export_data['features'])}")

# Display available functions
available = [attr for attr in dir(jupytergis) if not attr.startswith('_')]
print(f"Available functions: {', '.join(available)}")

print("Example 5 completed successfully")
```
