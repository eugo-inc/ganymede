## Example 4: Server extension setup and handlers

```python
from jupytergis_qgis import setup_handlers, _jupyter_server_extension_points

# Example 4: Server extension setup and handlers
print("Example 4: Server extension setup and handlers")

# Get server extension points
extension_points = _jupyter_server_extension_points()
print(f"Server extension points: {len(extension_points)}")
for point in extension_points:
    print(f"  Module: {point.get('module', 'N/A')}")

# Create mock server app for handler setup
class MockWebApp:
    def __init__(self):
        self.handlers = []
        self.settings = {'base_url': '/'}
        
    def add_handlers(self, pattern, handlers):
        self.handlers.extend(handlers)
        print(f"Added {len(handlers)} handlers with pattern: {pattern}")

# Setup handlers
web_app = MockWebApp()
setup_handlers(web_app)
print(f"Total handlers registered: {len(web_app.handlers)}")

# Show handler setup completion
print(f"Web app settings configured: {bool(web_app.settings)}")
print(f"Base URL: {web_app.settings.get('base_url', 'N/A')}")

print("Example 4 completed successfully")
```

## Example 5: Extension loading and version management

```python
import jupytergis_qgis
from jupytergis_qgis import _load_jupyter_server_extension, _jupyter_labextension_paths

# Example 5: Extension loading and version management
print("Example 5: Extension loading and version management")

# Get version information
version = jupytergis_qgis.__version__
print(f"JupyterGIS QGIS version: {version}")

# Create mock server app with required attributes
class MockServerApp:
    def __init__(self):
        self.log = MockLogger()
        self.web_app = MockWebApp()
        
class MockLogger:
    def info(self, message):
        print(f"[INFO] {message}")

class MockWebApp:
    def __init__(self):
        self.settings = {'base_url': '/'}
        self.handlers = []
        
    def add_handlers(self, pattern, handlers):
        self.handlers.extend(handlers)
        print(f"[INFO] Added {len(handlers)} handlers")

# Load the server extension
server_app = MockServerApp()
_load_jupyter_server_extension(server_app)

# Get lab extension paths
lab_paths = _jupyter_labextension_paths()
print(f"Lab extension configurations: {len(lab_paths)}")
for path in lab_paths:
    print(f"  Source: {path.get('src', 'N/A')}")
    print(f"  Destination: {path.get('dest', 'N/A')}")

print("Example 5 completed successfully")
```
