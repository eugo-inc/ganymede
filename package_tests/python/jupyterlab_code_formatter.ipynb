## Example 1: jupyterlab_code_formatter functionality test 1

```python
from jupyterlab_code_formatter.formatters import BlackFormatter, BaseFormatter

# Example 1: Code formatting with Black formatter
print("Example 1: Code formatting with Black formatter")

# Create a Black formatter instance
black_formatter = BlackFormatter()
print(f"Black formatter created: {black_formatter.__class__.__name__}")

# Sample Python code to format
unformatted_code = """def hello(name,age):
    if name=='Alice':
        print(f'Hello {name}, you are {age} years old!')
    else:print('Unknown person')"""

# Check if Black is available and format code
if black_formatter.importable:
    # format_code(code, notebook: bool, **options) returns the formatted source as a str
    formatted_result = black_formatter.format_code(unformatted_code, notebook=False)
    print("Original code:")
    print(unformatted_code)
    print("\nFormatted code:")
    print(formatted_result)
    assert formatted_result == (
        'def hello(name, age):\n'
        '    if name == "Alice":\n'
        '        print(f"Hello {name}, you are {age} years old!")\n'
        '    else:\n'
        '        print("Unknown person")\n'
    ), formatted_result
else:
    print("Black not available, simulating formatting process")
    print("Black formatter configured successfully")

print("Example 1 completed successfully")
```

## Example 2: jupyterlab_code_formatter functionality test 2

```python
from jupyterlab_code_formatter.formatters import IsortFormatter, Autopep8Formatter

# Example 2: Import sorting with isort formatter
print("Example 2: Import sorting with isort formatter")

# Create an isort formatter instance
isort_formatter = IsortFormatter()
print(f"Isort formatter created: {isort_formatter.__class__.__name__}")

# Sample Python code with unsorted imports
unsorted_imports = """import os
from typing import List
import sys
from collections import defaultdict
import json
from pathlib import Path"""

# Create autopep8 formatter for comparison
autopep8_formatter = Autopep8Formatter()
print(f"Autopep8 formatter created: {autopep8_formatter.__class__.__name__}")

# Check formatter availability
print(f"Isort available: {isort_formatter.importable}")
print(f"Autopep8 available: {autopep8_formatter.importable}")

# Demonstrate formatting capabilities
print(f"\nOriginal imports:")
print(unsorted_imports)
print(f"\nFormatters ready to organize imports and fix PEP8 issues")

print("Example 2 completed successfully")
```

## Example 5: jupyterlab_code_formatter functionality test 5

```python
from jupyterlab_code_formatter import setup_handlers, load_jupyter_server_extension
from jupyterlab_code_formatter.handlers import setup_handlers as handlers_setup

# Example 5: JupyterLab server extension setup
print("Example 5: JupyterLab server extension setup")

# Create a mock web application for demonstration
class MockWebApp:
    def __init__(self):
        self.handlers = []
        self.settings = {'base_url': '/'}
    
    def add_handlers(self, pattern, handlers):
        self.handlers.extend(handlers)
        print(f"Added {len(handlers)} handlers with pattern: {pattern}")

# Create mock server app
class MockServerApp:
    def __init__(self):
        self.web_app = MockWebApp()
        self.log = type('MockLog', (), {'info': lambda self, msg: print(f"[INFO] {msg}")})()

# Setup the server extension
server_app = MockServerApp()
setup_handlers(server_app.web_app)

print(f"Extension setup completed:")
print(f"  Handlers registered: {len(server_app.web_app.handlers)}")
print(f"  Base URL: {server_app.web_app.settings['base_url']}")

# Load the extension
load_jupyter_server_extension(server_app)

print("Example 5 completed successfully")
```
