```python
# Example 1: __version__ is set and non-empty.
import jupyterlab_eventlistener

version = jupyterlab_eventlistener.__version__
print(f"version = {version!r}")
assert isinstance(version, str)
assert len(version) > 0
assert version != 'dev', "package imported outside of a proper installation"
```

```python
# Example 2: _jupyter_labextension_paths returns the expected labextension descriptor.
import jupyterlab_eventlistener

paths = jupyterlab_eventlistener._jupyter_labextension_paths()
print(f"paths = {paths}")
assert isinstance(paths, list)
assert len(paths) == 1
entry = paths[0]
assert isinstance(entry, dict)
assert entry.get('src') == 'labextension'
assert 'dest' in entry
```

```python
# Example 3: _version submodule is importable and version values agree.
import jupyterlab_eventlistener
from jupyterlab_eventlistener import _version

print(f"_version.__version__ = {_version.__version__!r}")
print(f"_version.VERSION     = {_version.VERSION!r}")
assert _version.__version__ == _version.VERSION
assert jupyterlab_eventlistener.__version__ == _version.__version__
```

```python
# Example 4: package metadata reachable via importlib.metadata matches in-module __version__.
import importlib.metadata

import jupyterlab_eventlistener

meta_version = importlib.metadata.version('jupyterlab_eventlistener')
print(f"importlib.metadata version = {meta_version!r}")
print(f"module __version__          = {jupyterlab_eventlistener.__version__!r}")
assert meta_version == jupyterlab_eventlistener.__version__
```

```python
# Example 5: the package ships a labextension dir on disk at the path declared by
# _jupyter_labextension_paths. JupyterLab discovers the JS bundle from there.
import pathlib

import jupyterlab_eventlistener

pkg_root = pathlib.Path(jupyterlab_eventlistener.__path__[0])
print(f"pkg_root = {pkg_root}")
print(f"contents = {sorted(p.name for p in pkg_root.iterdir())}")

entry = jupyterlab_eventlistener._jupyter_labextension_paths()[0]
labext_dir = pkg_root / entry['src']
print(f"labext_dir = {labext_dir}  exists={labext_dir.exists()}")
# The src key is informational; presence on disk depends on how the wheel was built.
# Either way the descriptor must reference a string-typed relative path.
assert isinstance(entry['src'], str) and not entry['src'].startswith('/')
assert isinstance(entry['dest'], str) and len(entry['dest']) > 0
```
