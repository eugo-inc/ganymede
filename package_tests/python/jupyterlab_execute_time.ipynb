## Example 3: jupyterlab_execute_time functionality test 3

```python
import jupyterlab_execute_time
import json
import os

# Example 3: Extension build and dependency information
print("Example 3: Extension build and dependency information")

# Access the raw JSON data loaded from package.json
with open(jupyterlab_execute_time.fid.name, 'r') if hasattr(jupyterlab_execute_time.fid, 'name') else \
     open(os.path.join(jupyterlab_execute_time.HERE, 'labextension', 'package.json'), 'r') as f:
    package_data = json.load(f)

# Show build configuration
build_config = package_data.get('jupyterlab', {}).get('_build', {})
print(f"Build configuration:")
print(f"  Load script: {build_config.get('load', 'N/A')}")
print(f"  Extension entry: {build_config.get('extension', 'N/A')}")
print(f"  Style entry: {build_config.get('style', 'N/A')}")

# Show key dependencies for JupyterLab integration
deps = package_data.get('dependencies', {})
jupyterlab_deps = {k: v for k, v in deps.items() if 'jupyterlab' in k}
print(f"\nJupyterLab dependencies: {len(jupyterlab_deps)}")
for dep, version in list(jupyterlab_deps.items())[:3]:
    print(f"  {dep}: {version}")

print("Example 3 completed successfully")
```

## Example 4: jupyterlab_execute_time functionality test 4

```python
import jupyterlab_execute_time
import time

# Example 4: Simulating execution time functionality
print("Example 4: Simulating execution time functionality")

# This extension tracks execution time in JupyterLab cells
# We can demonstrate the concept it's designed to measure

# Get extension data for context
ext_data = jupyterlab_execute_time.data
print(f"Extension: {ext_data['name']}")
print(f"Purpose: {ext_data['description']}")

# Simulate what the extension measures - execution time
print(f"\nSimulating cell execution timing:")

start_time = time.time()
# Simulate some work that would happen in a notebook cell
result = sum(i**2 for i in range(1000))
end_time = time.time()

execution_duration = end_time - start_time
print(f"Computation result: {result}")
print(f"Execution time: {execution_duration:.4f} seconds")

# This is the type of timing information the extension displays
print(f"\nThe extension would display this timing in JupyterLab")

print("Example 4 completed successfully")
```
