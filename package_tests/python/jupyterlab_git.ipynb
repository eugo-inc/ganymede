## Example 1: jupyterlab_git functionality test 1

```python
from jupyterlab_git import JupyterLabGit, Git

# Example 1: JupyterLab Git extension configuration
print("Example 1: JupyterLab Git extension configuration")

# Create a JupyterLab Git extension instance
git_extension = JupyterLabGit()
print(f"Git extension created: {git_extension.__class__.__name__}")

# Check configuration traits
traits = git_extension.trait_names()
print(f"Configuration traits available: {len(traits)}")
print(f"Sample traits: {traits[:5]}")

# Show extension class hierarchy
print(f"Extension base classes: {[cls.__name__ for cls in git_extension.__class__.__mro__[:3]]}")

# Get configuration help
help_text = git_extension.class_get_help()
print(f"Help documentation available: {len(help_text) > 0}")
print(f"Extension ready for JupyterLab integration")

print("Example 1 completed successfully")
```

## Example 3: jupyterlab_git functionality test 3

```python
# setup_handlers lives in jupyterlab_git.handlers (not re-exported at top level);
# load_jupyter_server_extension is the top-level extension entry point.
from jupyterlab_git import load_jupyter_server_extension
from jupyterlab_git.handlers import setup_handlers
from traitlets.config import Config

# Example 3: JupyterLab server extension setup
print("Example 3: JupyterLab server extension setup")

# Create mock web application for demonstration
class MockWebApp:
    def __init__(self):
        self.handlers = []
        self.settings = {'base_url': '/'}

    def add_handlers(self, pattern, handlers):
        self.handlers.extend(handlers)
        print(f"Added {len(handlers)} handlers with pattern: {pattern}")

# Create mock server app with required attributes
class MockServerApp:
    def __init__(self):
        self.web_app = MockWebApp()
        self.log = type('Log', (), {'info': lambda self, msg: print(f"[INFO] {msg}")})()
        self.config = Config()  # Use proper Config object

# Setup Git handlers
server_app = MockServerApp()
setup_handlers(server_app.web_app)

print(f"Extension setup completed:")
print(f"  Handlers registered: {len(server_app.web_app.handlers)}")

# Load the Git extension
load_jupyter_server_extension(server_app)

print("Example 3 completed successfully")
```

## Example 5: jupyterlab_git functionality test 5

```python
# The git engine module now ships in the jupyterlab-git-core dist
# (jupyterlab_git re-exports the Git class from jupyterlab_git_core.git).
from jupyterlab_git_core import git
from jupyterlab_git import Git
from jupyterlab_git.handlers import GitLogHandler, GitAddAllUnstagedHandler

# Example 5: Git module and advanced handlers
print("Example 5: Git module and advanced handlers")

# Access the git engine submodule
git_module = git
print(f"Git module available: {git_module.__name__}")
assert git_module.Git is Git, "top-level Git must be the core git.Git class"

# Show advanced Git handlers for complex operations
advanced_handlers = [GitLogHandler, GitAddAllUnstagedHandler]
print(f"\nAdvanced Git handlers: {len(advanced_handlers)}")

for handler in advanced_handlers:
    print(f"  - {handler.__name__}")

# These handlers provide sophisticated Git operations
handler_descriptions = {
    "GitLogHandler": "Retrieves Git commit history",
    "GitAddAllUnstagedHandler": "Adds all unstaged files to Git staging"
}

print(f"\nHandler capabilities:")
for handler_name, description in handler_descriptions.items():
    print(f"  {handler_name}: {description}")

# Show the git module components
git_attrs = [attr for attr in dir(git_module) if not attr.startswith('_')]
print(f"\nGit module attributes: {len(git_attrs)}")
print(f"Key components: {git_attrs[:5]}")

print("Example 5 completed successfully")
```
