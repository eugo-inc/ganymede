## Example 1: jupyterlab_lsp functionality test 1

```python
import jupyterlab_lsp

# Example 1: JupyterLab LSP extension configuration
print("Example 1: JupyterLab LSP extension configuration")

# Get extension version and path information
version = jupyterlab_lsp.__version__
print(f"JupyterLab LSP version: {version}")

# Get extension paths for JupyterLab integration
extension_paths = jupyterlab_lsp._jupyter_labextension_paths()
print(f"Extension configurations: {len(extension_paths)}")

for i, path_config in enumerate(extension_paths, 1):
    print(f"\nExtension {i}:")
    print(f"  Source: {path_config['src']}")
    print(f"  Destination: {path_config['dest']}")

# This extension provides Language Server Protocol integration
print(f"\nExtension purpose: Language Server Protocol integration for JupyterLab")
print(f"Enables code completion, hover info, and diagnostics in notebooks")

print("Example 1 completed successfully")
```

## Example 2: jupyterlab_lsp functionality test 2

```python
from jupyterlab_lsp._paths import LAB_EXTENSION_PATH, MAIN_PACKAGE_PATH
import os

# Example 2: Extension path management and file structure
print("Example 2: Extension path management and file structure")

# Use the package's path constants
lab_ext_path = LAB_EXTENSION_PATH
main_pkg_path = MAIN_PACKAGE_PATH

print(f"Lab extension directory: {lab_ext_path.name}")
print(f"Main package directory: {main_pkg_path.name}")

# Check if paths exist
print(f"\nPath validation:")
print(f"  Lab extension exists: {lab_ext_path.exists()}")
print(f"  Main package exists: {main_pkg_path.exists()}")

# Show directory structure
if lab_ext_path.exists():
    contents = list(lab_ext_path.iterdir())
    print(f"\nLab extension contents: {len(contents)} items")
    for item in contents[:3]:
        print(f"  - {item.name}")

if main_pkg_path.exists():
    main_contents = list(main_pkg_path.iterdir())
    print(f"\nMain package contents: {len(main_contents)} items")
    for item in main_contents:
        print(f"  - {item.name}")

print("Example 2 completed successfully")
```

## Example 3: jupyterlab_lsp functionality test 3

```python
import json
from jupyterlab_lsp._paths import MAIN_PACKAGE_PATH

# Example 3: Extension package metadata and LSP configuration
print("Example 3: Extension package metadata and LSP configuration")

# Read package.json from the extension
package_json_path = MAIN_PACKAGE_PATH / "package.json"
if package_json_path.exists():
    with open(package_json_path, 'r') as f:
        package_data = json.load(f)
    
    print(f"Package metadata:")
    print(f"  Name: {package_data['name']}")
    print(f"  Version: {package_data['version']}")
    print(f"  Description: {package_data['description']}")
    
    # Show LSP-related keywords
    keywords = package_data.get('keywords', [])
    lsp_keywords = [k for k in keywords if 'lsp' in k.lower() or 'language' in k.lower()]
    print(f"\nLSP-related keywords: {lsp_keywords}")
    
    # Show JupyterLab configuration if available
    jupyterlab_config = package_data.get('jupyterlab', {})
    if jupyterlab_config:
        print(f"\nJupyterLab configuration found:")
        print(f"  Extension type: {type(jupyterlab_config).__name__}")
        print(f"  Configuration keys: {list(jupyterlab_config.keys())[:3]}")

print("Example 3 completed successfully")
```

## Example 4: jupyterlab_lsp functionality test 4

```python
import json
from pathlib import Path
from jupyterlab_lsp._paths import MAIN_PACKAGE_PATH

# Example 4: LSP schema configuration and installation info
print("Example 4: LSP schema configuration and installation info")

# Check for schema files that configure LSP behavior. They sit in a scoped subdirectory
# (schemas/@jupyter-lsp/jupyterlab-lsp/*.json), so search recursively.
schemas_path = MAIN_PACKAGE_PATH / "schemas"
assert schemas_path.is_dir(), f"no settings schema directory at {schemas_path}"
schema_files = list(schemas_path.rglob("*.json"))
assert schema_files, f"no LSP settings schemas under {schemas_path}"
print(f"LSP schema files found: {len(schema_files)}")
for schema_file in schema_files[:3]:
    print(f"  - {schema_file.name}")

# Read installation information
install_json_path = Path(MAIN_PACKAGE_PATH).parent.parent.parent / "install.json"
if install_json_path.exists():
    with open(install_json_path, 'r') as f:
        install_data = json.load(f)
    
    print(f"\nInstallation information:")
    print(f"  Package manager: {install_data.get('packageManager', 'N/A')}")
    print(f"  Package name: {install_data.get('packageName', 'N/A')}")
    
    uninstall_info = install_data.get('uninstallInstructions', '')
    print(f"  Uninstall: {uninstall_info[:50]}...")

# Show static assets
static_path = MAIN_PACKAGE_PATH / "static"
if static_path.exists():
    static_files = list(static_path.iterdir())
    print(f"\nStatic assets: {len(static_files)} files")

print("Example 4 completed successfully")
```

## Example 5: jupyterlab_lsp functionality test 5

```python
import jupyterlab_lsp
from jupyterlab_lsp._paths import LAB_EXTENSION_PATH, MAIN_PACKAGE_PATH
from pathlib import Path

# Example 5: Complete LSP extension validation and summary
print("Example 5: Complete LSP extension validation and summary")

# Combine all package functionality for a complete validation
version = jupyterlab_lsp.__version__
extension_paths = jupyterlab_lsp._jupyter_labextension_paths()

print(f"JupyterLab LSP Extension Summary:")
print(f"  Version: {version}")
print(f"  Extension paths configured: {len(extension_paths)}")

# Validate installation integrity
validation_results = {
    "version_accessible": hasattr(jupyterlab_lsp, '__version__'),
    "paths_function_works": callable(jupyterlab_lsp._jupyter_labextension_paths),
    "lab_extension_exists": LAB_EXTENSION_PATH.exists(),
    "main_package_exists": MAIN_PACKAGE_PATH.exists(),
}

print(f"\nInstallation validation:")
for check, result in validation_results.items():
    status = "✓" if result else "✗"
    print(f"  {status} {check.replace('_', ' ').title()}")

# Summary of LSP capabilities
print(f"\nLSP Extension provides:")
print(f"  - Code completion and IntelliSense")
print(f"  - Real-time error diagnostics")
print(f"  - Hover documentation and signatures")

print("Example 5 completed successfully")
```
