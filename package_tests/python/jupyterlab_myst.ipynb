## Example 1: jupyterlab_myst functionality test 1

````python
from jupyterlab_myst import MySTNotebookNotary
import nbformat

# Example 1: Create and configure MyST notebook notary for enhanced security
print("Example 1: Create and configure MyST notebook notary for enhanced security")

# Create a MyST-aware notary for validating notebook trust
myst_notary = MySTNotebookNotary()
print(f"MyST notary created: {myst_notary.__class__.__name__}")

# Create a sample notebook with markdown cells containing MyST syntax
nb = nbformat.v4.new_notebook()
markdown_cell = nbformat.v4.new_markdown_cell(source="# MyST Example\n\n```{note}\nThis is a MyST note directive.\n```")
code_cell = nbformat.v4.new_code_cell(source="print('Hello MyST!')")

nb.cells = [markdown_cell, code_cell]

# Check if the notebook cells pass MyST security validation
try:
    is_trusted = myst_notary.check_cells(nb)
    print(f"Notebook trust validation: {'Trusted' if is_trusted else 'Untrusted'}")
    print(f"MyST notary provides enhanced security for markdown with directives")
except Exception as e:
    print(f"MyST notary validation configured for markdown security")
    raise

print("Example 1 completed successfully")
````

## Example 2: jupyterlab_myst functionality test 2

````python
from jupyterlab_myst.notary import yield_markdown_cells
import nbformat

# Example 2: Extract and process MyST markdown cells from notebooks
print("Example 2: Extract and process MyST markdown cells from notebooks")

# Create a sample notebook with various cell types including MyST markdown
nb = nbformat.v4.new_notebook()

# Add different types of cells
intro_cell = nbformat.v4.new_markdown_cell(source="# MyST Document\n\nIntroduction to MyST features.")
note_cell = nbformat.v4.new_markdown_cell(source="```{note}\nThis is an important note.\n```")
warning_cell = nbformat.v4.new_markdown_cell(source="```{warning}\nBe careful with this!\n```")
code_cell = nbformat.v4.new_code_cell(source="# This is code, not markdown")

nb.cells = [intro_cell, note_cell, code_cell, warning_cell]

# Use MyST's markdown cell extractor to process only markdown cells
markdown_cells = list(yield_markdown_cells(nb))
print(f"Total cells in notebook: {len(nb.cells)}")
print(f"Markdown cells extracted: {len(markdown_cells)}")

# Process each markdown cell
for i, cell in enumerate(markdown_cells, 1):
    source_preview = cell['source'][:50].replace('\n', ' ')
    print(f"  Cell {i}: {source_preview}...")

print("Example 2 completed successfully")
````

## Example 3: jupyterlab_myst functionality test 3

````python
import jupyterlab_myst
from traitlets.config import Configurable

# Example 3: Configure JupyterLab server extension for MyST integration
print("Example 3: Configure JupyterLab server extension for MyST integration")

# Get MyST extension configuration for JupyterLab
extension_paths = jupyterlab_myst._jupyter_labextension_paths()
print(f"MyST extension paths: {len(extension_paths)}")

for path_config in extension_paths:
    print(f"  Source: {path_config['src']}")
    print(f"  Destination: {path_config['dest']}")

# Get server extension points for MyST integration
server_extensions = jupyterlab_myst._jupyter_server_extension_points()
print(f"\nServer extension points: {len(server_extensions)}")

for ext_point in server_extensions:
    print(f"  Module: {ext_point['module']}")

# Create a mock server app to demonstrate extension loading
class MockContentsManager(Configurable):
    def __init__(self):
        super().__init__()
        self.notary = None

class MockServerApp:
    def __init__(self):
        self.contents_manager = MockContentsManager()

# Demonstrate how MyST would integrate with server
mock_app = MockServerApp()
print(f"\nBefore MyST integration: {type(mock_app.contents_manager.notary)}")
jupyterlab_myst._load_jupyter_server_extension(mock_app)
print(f"After MyST integration: {type(mock_app.contents_manager.notary).__name__}")

print("Example 3 completed successfully")
````

## Example 4: jupyterlab_myst functionality test 4

````python
from jupyterlab_myst import MySTNotebookNotary
import nbformat

# Example 4: Validate notebook signatures with MyST-aware security
print("Example 4: Validate notebook signatures with MyST-aware security")

# Create MyST notary with custom configuration
myst_notary = MySTNotebookNotary()

# Create notebooks with different trust scenarios
trusted_nb = nbformat.v4.new_notebook()
trusted_nb.cells = [
    nbformat.v4.new_markdown_cell(source="# Safe Content\n\nRegular markdown text."),
    nbformat.v4.new_code_cell(source="x = 1 + 1")
]

untrusted_nb = nbformat.v4.new_notebook()
untrusted_nb.cells = [
    nbformat.v4.new_markdown_cell(source="# Potentially Unsafe\n\n```{eval-rst}\n.. raw:: html\n   <script>alert('xss')</script>\n```"),
    nbformat.v4.new_code_cell(source="import os; os.system('rm -rf /')")
]

# Check notebook signatures with MyST security awareness
notebooks = [("trusted", trusted_nb), ("untrusted", untrusted_nb)]

for name, nb in notebooks:
    try:
        signature_valid = myst_notary.check_signature(nb)
        print(f"{name.capitalize()} notebook signature: {'Valid' if signature_valid else 'Invalid'}")
    except Exception:
        print(f"{name.capitalize()} notebook: MyST security validation applied")
        raise

print("Example 4 completed successfully")
````

## Example 5: jupyterlab_myst functionality test 5

````python
from jupyterlab_myst import MySTNotebookNotary
from jupyterlab_myst.notary import yield_markdown_cells
import nbformat

# Example 5: Complete MyST workflow - create, validate, and process MyST notebooks
print("Example 5: Complete MyST workflow - create, validate, and process MyST notebooks")

# Create a comprehensive MyST notebook with various directives
myst_nb = nbformat.v4.new_notebook()
myst_nb.metadata.kernelspec = {"display_name": "Python 3", "language": "python", "name": "python3"}

# Add MyST cells with different directive types
cells = [
    nbformat.v4.new_markdown_cell(source="# MyST Features Demo\n\n```{note}\nImportant information here.\n```"),
    nbformat.v4.new_markdown_cell(source="```{warning}\nProceed with caution!\n```"),
    nbformat.v4.new_code_cell(source="import matplotlib.pyplot as plt\nplt.plot([1,2,3], [1,4,9])"),
    nbformat.v4.new_markdown_cell(source="```{math}\ne^{i\\pi} + 1 = 0\n```"),
]
myst_nb.cells = cells

# Initialize MyST notary for processing
notary = MySTNotebookNotary()

# Extract and count MyST markdown content
markdown_cells = list(yield_markdown_cells(myst_nb))
print(f"Created MyST notebook with {len(myst_nb.cells)} total cells")
print(f"Markdown cells containing MyST directives: {len(markdown_cells)}")

# Validate notebook with MyST-aware security
cell_validation = notary.check_cells(myst_nb)
print(f"MyST security validation: {'Passed' if cell_validation else 'Requires review'}")

print("Example 5 completed successfully")
````
