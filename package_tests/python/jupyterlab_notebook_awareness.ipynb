```python
# Example 1: Inspect public API surface.
import jupyterlab_notebook_awareness as jna

public_names = [n for n in dir(jna) if not n.startswith('_')]
print(f"Public names ({len(public_names)}):")
for name in public_names:
    obj = getattr(jna, name)
    print(f"  {name}: {type(obj).__name__}")
```

```python
# Example 2: Module file location + package metadata.
import jupyterlab_notebook_awareness as jna
import os

pkg_dir = jna.__path__[0]
print(f"Package directory: {pkg_dir}")
print(f"Contents:")
for entry in sorted(os.listdir(pkg_dir))[:30]:
    full = os.path.join(pkg_dir, entry)
    kind = 'dir' if os.path.isdir(full) else 'file'
    print(f"  [{kind}] {entry}")
```

```python
# Example 3: Distribution metadata lookup.
from importlib.metadata import distribution, PackageNotFoundError

try:
    dist = distribution('jupyterlab-notebook-awareness')
    print(f"Name: {dist.metadata['Name']}")
    print(f"Version: {dist.version}")
    summary = dist.metadata.get('Summary', '(no summary)')
    print(f"Summary: {summary}")
    requires = dist.requires or []
    print(f"Requires ({len(requires)}):")
    for req in requires[:10]:
        print(f"  {req}")
except PackageNotFoundError as e:
    print(f"Distribution not found: {e}")
```

```python
# Example 4: Look for callable / class entries (likely JupyterLab extension hooks).
import jupyterlab_notebook_awareness as jna
import inspect

callables = []
classes = []
for name in dir(jna):
    if name.startswith('_'):
        continue
    obj = getattr(jna, name)
    if inspect.isclass(obj):
        classes.append(name)
    elif callable(obj):
        callables.append(name)

print(f"Classes: {classes}")
print(f"Callables: {callables}")

# Common JupyterLab extension hooks
for hook in ('_jupyter_labextension_paths', '_jupyter_server_extension_points', '_jupyter_server_extension_paths'):
    if hasattr(jna, hook):
        result = getattr(jna, hook)()
        print(f"{hook}() -> {result}")
```

```python
# Example 5: Final assertion — package importable + has __path__.
import jupyterlab_notebook_awareness as jna

assert jna is not None, "package failed to import"
assert hasattr(jna, '__path__'), "expected a package (with __path__)"
assert len(jna.__path__) >= 1, "package __path__ should be non-empty"

version = getattr(jna, '__version__', None)
print(f"jupyterlab_notebook_awareness imported OK (version={version})")
print("PASS")
```
