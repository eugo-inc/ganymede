## Example 2: jupyterlab_notifications functionality test 2

```python
import time
import jupyterlab_notifications

# Example 2: Simulate long-running cell execution for notification demonstration
print("Example 2: Simulate long-running cell execution for notification demonstration")

# Get notification extension info
version = jupyterlab_notifications.__version__
print(f"Using jupyterlab-notifications v{version}")

# Simulate a long-running computation that would trigger notifications
print("Simulating long-running cell execution...")
start_time = time.time()

# Simulate computational work (like data processing, ML training, etc.)
result = 0
for i in range(1000000):
    result += i ** 0.5

end_time = time.time()
execution_time = end_time - start_time

print(f"Computation completed: {result:.2f}")
print(f"Execution time: {execution_time:.3f} seconds")

# Demonstrate notification threshold logic
min_threshold = 60  # Default minimum time for notifications
if execution_time >= min_threshold:
    print(f"✓ Would trigger notification (≥{min_threshold}s)")
else:
    print(f"✗ Below notification threshold (<{min_threshold}s)")

print("Example 2 completed successfully")
```

## Example 4: jupyterlab_notifications functionality test 4

```python
import jupyterlab_notifications
import time

# Example 4: Multiple execution scenarios demonstrating notification triggers
print("Example 4: Multiple execution scenarios demonstrating notification triggers")

# Extension setup
extension_paths = jupyterlab_notifications._jupyter_labextension_paths()
dest_path = extension_paths[0]['dest']
print(f"Extension destination: {dest_path}")

# Simulate different types of cell executions
execution_scenarios = [
    {"name": "Quick calculation", "duration": 0.1, "status": "success"},
    {"name": "Data processing", "duration": 45.0, "status": "success"},
    {"name": "Model training", "duration": 75.0, "status": "success"},
    {"name": "Failed computation", "duration": 30.0, "status": "error"}
]

notification_threshold = 60  # seconds

print(f"Simulating execution scenarios (threshold: {notification_threshold}s):")
for scenario in execution_scenarios:
    name = scenario["name"]
    duration = scenario["duration"]
    status = scenario["status"]
    
    # Simulate brief execution
    time.sleep(0.01)  # Brief simulation
    
    # Determine if notification would be triggered
    would_notify = duration >= notification_threshold
    status_icon = "✓" if status == "success" else "✗"
    notify_icon = "🔔" if would_notify else "🔕"
    
    print(f"  {notify_icon} {name}: {duration}s {status_icon} ({status})")

print("Example 4 completed successfully")
```

## Example 5: jupyterlab_notifications functionality test 5

```python
import jupyterlab_notifications
import os
import platform

# Example 5: Extension compatibility and deployment verification
print("Example 5: Extension compatibility and deployment verification")

# Check extension version and compatibility
version = jupyterlab_notifications.__version__
print(f"JupyterLab Notifications version: {version}")

# Verify extension paths and installation
extension_paths = jupyterlab_notifications._jupyter_labextension_paths()
for path_info in extension_paths:
    src = path_info['src']
    dest = path_info['dest']
    print(f"Extension mapping: {src} → {dest}")

# Check system compatibility for notification features
system_info = {
    "platform": platform.system(),
    "python_version": platform.python_version(),
    "supports_notifications": True  # Most modern browsers support Notification API
}

print(f"\nSystem compatibility:")
print(f"  Platform: {system_info['platform']}")
print(f"  Python: {system_info['python_version']}")
print(f"  Browser notifications: {system_info['supports_notifications']}")

# Show deployment readiness
deployment_checks = {
    "extension_installed": hasattr(jupyterlab_notifications, '__version__'),
    "paths_configured": len(extension_paths) > 0,
    "compatible_version": version >= "0.3.0"
}

print(f"\nDeployment readiness:")
for check, status in deployment_checks.items():
    status_icon = "✓" if status else "✗"
    print(f"  {status_icon} {check.replace('_', ' ').title()}")

print("Example 5 completed successfully")
```
