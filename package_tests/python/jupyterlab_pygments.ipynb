## Example 1: jupyterlab_pygments functionality test 1

```python
from jupyterlab_pygments import JupyterStyle

# Example 1: Create and configure JupyterLab-specific syntax highlighting style
print("Example 1: Create and configure JupyterLab-specific syntax highlighting style")

# Create a JupyterLab style instance
jupyter_style = JupyterStyle()
print(f"JupyterLab style created: {jupyter_style.__class__.__name__}")

# Explore style properties
print(f"Style name: {jupyter_style.name}")
print(f"Background color: {jupyter_style.background_color}")
print(f"Highlight color: {jupyter_style.highlight_color}")

# Show line number styling
print(f"\nLine number styling:")
print(f"  Color: {jupyter_style.line_number_color}")
print(f"  Background: {jupyter_style.line_number_background_color}")
print(f"  Special color: {jupyter_style.line_number_special_color}")

# Check default style and total number of token styles
print(f"\nStyle configuration:")
print(f"  Default style: {jupyter_style.default_style}")
print(f"  Total token styles: {len(jupyter_style.styles)}")

# Show it uses JupyterLab CSS variables
uses_css_vars = any('var(--jp-' in str(value) for value in jupyter_style.styles.values())
print(f"  Uses JupyterLab CSS variables: {uses_css_vars}")

print("Example 1 completed successfully")
```

## Example 2: jupyterlab_pygments functionality test 2

```python
from jupyterlab_pygments import JupyterStyle
from pygments.token import Token

# Example 2: Explore JupyterLab token styling for different code elements
print("Example 2: Explore JupyterLab token styling for different code elements")

# Create JupyterLab style
style = JupyterStyle()

# Examine different types of token styles
token_examples = [
    (Token.Keyword, "Keywords (def, if, class)"),
    (Token.String, "String literals"),
    (Token.Comment, "Comments"),
    (Token.Name.Function, "Function names"),
    (Token.Number, "Numbers"),
    (Token.Operator, "Operators (+, -, *)"),
    (Token.Error, "Error tokens")
]

print("JupyterLab syntax highlighting styles:")
for token, description in token_examples:
    if token in style.styles:
        token_style = style.styles[token]
        print(f"  {description}: {token_style}")
    else:
        print(f"  {description}: (inherits default)")

# Show how JupyterLab integrates with CSS variables
css_vars_count = sum(1 for value in style.styles.values() if 'var(--jp-' in str(value))
print(f"\nJupyterLab CSS variable integration:")
print(f"  Tokens using CSS variables: {css_vars_count}/{len(style.styles)}")

print("Example 2 completed successfully")
```

## Example 4: jupyterlab_pygments functionality test 4

```python
from jupyterlab_pygments import JupyterStyle
from pygments.token import Token

# Example 4: Advanced token style analysis and CSS variable mapping
print("Example 4: Advanced token style analysis and CSS variable mapping")

# Create JupyterLab style for analysis
style = JupyterStyle()

# Analyze CSS variable usage patterns
css_variables = {}
for token, token_style in style.styles.items():
    if 'var(--jp-' in str(token_style):
        # Extract CSS variable names
        css_var = str(token_style).split('var(')[1].split(')')[0] if 'var(' in str(token_style) else None
        if css_var:
            if css_var not in css_variables:
                css_variables[css_var] = []
            css_variables[css_var].append(token)

print(f"CSS variable analysis:")
print(f"  Total unique CSS variables: {len(css_variables)}")
print(f"  Most common variables:")

# Show top CSS variables
for i, (css_var, tokens) in enumerate(sorted(css_variables.items(), key=lambda x: len(x[1]), reverse=True)[:5], 1):
    print(f"    {i}. {css_var}: {len(tokens)} tokens")

# Show token hierarchy coverage
token_types = set()
for token in style.styles.keys():
    token_types.add(str(token).split('.')[1] if '.' in str(token) else str(token))

print(f"\nToken type coverage: {len(token_types)} types")
print(f"  Sample types: {list(token_types)[:5]}")

print("Example 4 completed successfully")
```

## Example 5: jupyterlab_pygments functionality test 5

```python
import jupyterlab_pygments
from jupyterlab_pygments import JupyterStyle

# Example 5: Complete JupyterLab Pygments integration and style validation
print("Example 5: Complete JupyterLab Pygments integration and style validation")

# Get package version if available
try:
    version = jupyterlab_pygments.__version__
    print(f"JupyterLab Pygments version: {version}")
except:
    print("JupyterLab Pygments package installed successfully")
    raise

# Validate complete style configuration
style = JupyterStyle()

# Comprehensive style validation
validation_results = {
    "style_created": isinstance(style, JupyterStyle),
    "has_background": bool(style.background_color),
    "has_highlight": bool(style.highlight_color),
    "has_styles": len(style.styles) > 0,
    "uses_css_vars": any('var(--jp-' in str(v) for v in style.styles.values()),
    "line_numbers_styled": bool(style.line_number_color)
}

print(f"\nStyle validation results:")
for check, result in validation_results.items():
    status = "✓" if result else "✗"
    print(f"  {status} {check.replace('_', ' ').title()}")

# Extension integration check
extension_paths = jupyterlab_pygments._jupyter_labextension_paths()
integration_ready = len(extension_paths) > 0

print(f"\nIntegration status:")
print(f"  Extension paths configured: {integration_ready}")
print(f"  Style tokens available: {len(style.styles)}")
print(f"  JupyterLab theming compatible: {validation_results['uses_css_vars']}")

# Summary
print(f"\nJupyterLab Pygments ready for syntax highlighting with theme integration")

print("Example 5 completed successfully")
```
