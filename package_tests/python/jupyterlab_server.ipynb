## Example 1: jupyterlab_server functionality test 1

```python
from jupyterlab_server import LabConfig, LabServerApp

# Example 1: JupyterLab server configuration and application setup
print("Example 1: JupyterLab server configuration and application setup")

# Create a JupyterLab configuration instance
lab_config = LabConfig()
print(f"LabConfig created: {lab_config.__class__.__name__}")

# Show configuration capabilities
config_methods = [method for method in dir(lab_config) if not method.startswith('_')]
print(f"Configuration methods available: {len(config_methods)}")
print(f"Sample methods: {config_methods[:5]}")

# Create a JupyterLab server application
try:
    # Configure basic server app without starting it
    lab_app = LabServerApp()
    print(f"LabServerApp created: {lab_app.__class__.__name__}")
    
    # Show app configuration
    print(f"Server app configured with:")
    print(f"  App name: {lab_app.name if hasattr(lab_app, 'name') else 'JupyterLab'}")
    print(f"  Config class: {lab_app.config_class.__name__ if hasattr(lab_app, 'config_class') else 'Default'}")
    
except Exception as e:
    print(f"Server app configuration: {e}")
    print("LabServerApp available for JupyterLab server deployment")
    raise

print("Example 1 completed successfully")
```

## Example 3: jupyterlab_server functionality test 3

```python
from jupyterlab_server import slugify, get_openapi_spec

# Example 3: JupyterLab server utilities and API specification
print("Example 3: JupyterLab server utilities and API specification")

# Demonstrate the slugify function for URL-safe names
test_strings = [
    "My Project Name",
    "Data Science Notebook!",
    "User@domain.com workspace",
    "Complex-Name_with_Symbols#123"
]

print("Slugify utility for URL-safe names:")
for text in test_strings:
    try:
        slug = slugify(text)
        print(f"  '{text}' → '{slug}'")
    except Exception as e:
        # Handle potential syntax warnings gracefully
        print(f"  '{text}' → slugified successfully")
        raise

# Get the OpenAPI specification JupyterLab server ships for its REST API
openapi_spec = get_openapi_spec()
print(f"\nOpenAPI specification retrieved:")
print(f"  Type: {type(openapi_spec).__name__}")

spec_keys = list(openapi_spec.keys())
print(f"  Specification sections: {spec_keys}")
assert {"openapi", "info", "paths", "components"} <= set(spec_keys), spec_keys
assert openapi_spec["openapi"].startswith("3."), openapi_spec["openapi"]
assert openapi_spec["info"]["title"] == "JupyterLab Server", openapi_spec["info"]["title"]

api_paths = list(openapi_spec["paths"].keys())
print(f"  REST paths described: {len(api_paths)}")
assert "/lab/api/settings/{schema_name}" in api_paths, api_paths
assert {"get", "put"} <= set(openapi_spec["paths"]["/lab/api/settings/{schema_name}"].keys())

print("Example 3 completed successfully")
```

## Example 4: jupyterlab_server functionality test 4

```python
from jupyterlab_server import LicensesApp, add_handlers

# Example 4: JupyterLab licenses management and handler setup
print("Example 4: JupyterLab licenses management and handler setup")

# Demonstrate LicensesApp class (without instantiation that requires LicensesManager)
licenses_app_class = LicensesApp
print(f"Licenses app class: {licenses_app_class.__name__}")

# Show licenses app capabilities through class inspection
licenses_methods = [m for m in dir(licenses_app_class) if not m.startswith('_') and callable(getattr(licenses_app_class, m, None))]
print(f"License management methods: {len(licenses_methods)}")
print(f"Sample methods: {licenses_methods[:5]}")

# Demonstrate add_handlers for server setup. It takes the route LIST and a LabServerApp, and APPENDS
# (url_pattern, handler_class[, kwargs]) routes to that list — it is not given a tornado web app.
from jupyterlab_server import LabServerApp
from jupyterlab_server.handlers import LabHandler

lab_app = LabServerApp()
handlers = []
add_handlers(handlers, lab_app)
print(f"JupyterLab handlers added: {len(handlers)} total handlers")
for route in handlers:
    print(f"  {route[0]} -> {route[1].__name__}")
assert any(route[1] is LabHandler for route in handlers), handlers
assert any(route[0].startswith(lab_app.licenses_url) for route in handlers), handlers

# Show license management purpose
print(f"\nLicense management features:")
print(f"  Track dependency licenses")
print(f"  Generate license reports")
print(f"  Ensure compliance for deployments")

print("Example 4 completed successfully")
```
