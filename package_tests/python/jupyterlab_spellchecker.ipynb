## Example 1: jupyterlab_spellchecker functionality test 1

```python
from jupyterlab_spellchecker.dictionaries import discover_dictionaries, _extract_code, _scan_for_dictionaries
from pathlib import Path
import tempfile
import os

# Example 1: Dictionary discovery and language code extraction
print("Example 1: Dictionary discovery and language code extraction")

# Test BCP 47 language code extraction from filenames
test_filenames = [
    "en_US.dic",
    "de-DE.dic",
    "fr_FR.dic",
    "es-419.dic",
    "invalid_filename.dic"
]

print("Language code extraction:")
for filename in test_filenames:
    code = _extract_code(filename)
    print(f"  {filename} → {code if code else 'None (invalid)'}")

# Create a mock logger for dictionary scanning
class MockLogger:
    def warning(self, msg): print(f"[WARNING] {msg}")
    def info(self, msg): print(f"[INFO] {msg}")

# Test dictionary scanning with a temporary directory
with tempfile.TemporaryDirectory() as temp_dir:
    # Create mock dictionary files
    test_dict = Path(temp_dir) / "en_US.dic"
    test_aff = Path(temp_dir) / "en_US.aff"
    test_dict.write_text("# Mock dictionary file")
    test_aff.write_text("# Mock affix file")

    print(f"\nScanning for dictionaries in: {temp_dir}")
    dictionaries = _scan_for_dictionaries(temp_dir, MockLogger())
    print(f"Found {len(dictionaries)} dictionaries")

print("Example 1 completed successfully")
```

## Example 5: jupyterlab_spellchecker functionality test 5

```python
from jupyterlab_spellchecker.dictionaries import discover_dictionaries
from jupyter_core.paths import jupyter_path
import tempfile
from pathlib import Path

# Example 5: Complete spellchecker dictionary discovery simulation
print("Example 5: Complete spellchecker dictionary discovery simulation")

# Create mock server app with logging
class MockServerApp:
    def __init__(self):
        self.log = type('Log', (), {
            'info': lambda self, msg: print(f"[INFO] {msg}"),
            'warning': lambda self, msg: print(f"[WARNING] {msg}")
        })()

# Show Jupyter data paths for dictionaries
data_paths = jupyter_path('dictionaries')
print(f"Jupyter dictionary search paths: {len(data_paths)}")
for i, path in enumerate(data_paths, 1):
    print(f"  {i}. {path}")

# Simulate dictionary discovery
server_app = MockServerApp()
with tempfile.TemporaryDirectory() as temp_dir:
    # Add temp directory to search paths for testing
    import jupyterlab_spellchecker.dictionaries as dict_module
    original_jupyter_path = dict_module.jupyter_path
    dict_module.jupyter_path = lambda x: [temp_dir]

    # Create mock dictionary files
    dict_files = ["en_US", "de_DE", "fr_FR"]
    for lang in dict_files:
        (Path(temp_dir) / f"{lang}.dic").write_text(f"# {lang} dictionary")
        (Path(temp_dir) / f"{lang}.aff").write_text(f"# {lang} affix rules")

    # Discover dictionaries
    dictionaries = discover_dictionaries(server_app)

    # Restore original function
    dict_module.jupyter_path = original_jupyter_path

print(f"\nDictionary discovery complete: {len(dictionaries)} found")
print("Spellchecker ready for JupyterLab integration")
print("Example 5 completed successfully")
```
