## Example 1: jupyterlab_widgets functionality test 1

```python
import jupyterlab_widgets
from pathlib import Path

# Example 1: JupyterLab widgets extension configuration
print("Example 1: JupyterLab widgets extension configuration")

# Get extension paths for JupyterLab widgets manager
extension_paths = jupyterlab_widgets._jupyter_labextension_paths()
print(f"Extension paths configured: {len(extension_paths)}")

for path_config in extension_paths:
    print(f"  Source: {path_config['src']}")
    print(f"  Destination: {path_config['dest']}")
    
    # Verify the extension source exists
    src_path = Path(path_config['src'])
    print(f"  Extension exists: {src_path.exists()}")
    
    if src_path.exists() and src_path.is_dir():
        # Count extension files
        contents = list(src_path.rglob('*'))
        files_count = len([f for f in contents if f.is_file()])
        print(f"  Extension files: {files_count}")

# Show package metadata
print(f"\nPackage version: {jupyterlab_widgets.__version__}")
print(f"Extension target: Jupyter widgets manager for JupyterLab")

print("Example 1 completed successfully")
```

## Example 2: jupyterlab_widgets functionality test 2

```python
import jupyterlab_widgets
import json
from pathlib import Path

# Example 2: Extension package metadata and configuration analysis
print("Example 2: Extension package metadata and configuration analysis")

# Get extension path
extension_paths = jupyterlab_widgets._jupyter_labextension_paths()
extension_path = Path(extension_paths[0]['src'])

# Read package.json to understand extension capabilities
package_json_path = extension_path / "package.json"
if package_json_path.exists():
    with open(package_json_path) as f:
        package_data = json.load(f)
    
    print(f"Widget manager extension: {package_data.get('name', 'Unknown')}")
    print(f"Extension version: {package_data.get('version', 'Unknown')}")
    print(f"Description: {package_data.get('description', 'No description')}")
    
    # Show JupyterLab specific configuration
    if 'jupyterlab' in package_data:
        jlab_config = package_data['jupyterlab']
        print(f"\nJupyterLab configuration:")
        print(f"  Extension type: {'Yes' if jlab_config.get('extension') else 'No'}")
        print(f"  Output directory: {jlab_config.get('outputDir', 'Default')}")
        
        # Show shared packages if available
        if 'sharedPackages' in jlab_config:
            shared_pkgs = jlab_config['sharedPackages']
            print(f"  Shared packages: {len(shared_pkgs)}")

print("Example 2 completed successfully")
```

## Example 3: jupyterlab_widgets functionality test 3

```python
import jupyterlab_widgets
from pathlib import Path
import sys

# Example 3: Extension path resolution and fallback mechanism
print("Example 3: Extension path resolution and fallback mechanism")

# Test the dynamic path resolution logic from _jupyter_labextension_paths
labext_name = '@jupyter-widgets/jupyterlab-manager'
here = Path(jupyterlab_widgets.__file__).parent.resolve()
src_prefix = here.parent / 'labextension'

print(f"Extension name: {labext_name}")
print(f"Package location: {here}")
print(f"Primary source path: {src_prefix}")
print(f"Primary path exists: {src_prefix.exists()}")

# Show fallback path (sys.prefix location)
fallback_prefix = Path(sys.prefix) / f'share/jupyter/labextensions/{labext_name}'
print(f"Fallback source path: {fallback_prefix}")
print(f"Fallback path exists: {fallback_prefix.exists()}")

# Demonstrate the actual path resolution
extension_paths = jupyterlab_widgets._jupyter_labextension_paths()
actual_path = extension_paths[0]['src']
print(f"\nResolved extension path: {actual_path}")

# Show which path was chosen
if str(src_prefix) == actual_path:
    print("Used: Primary path (relative to package)")
elif str(fallback_prefix) == actual_path:
    print("Used: Fallback path (sys.prefix)")
else:
    print("Used: Custom resolution path")

print("Example 3 completed successfully")
```

## Example 4: jupyterlab_widgets functionality test 4

```python
import jupyterlab_widgets
from pathlib import Path
import json

# Example 4: Widget extension installation and schema validation
print("Example 4: Widget extension installation and schema validation")

# Get extension information
extension_paths = jupyterlab_widgets._jupyter_labextension_paths()
extension_path = Path(extension_paths[0]['src'])

# Check for installation metadata
install_json_path = extension_path / "install.json"
if install_json_path.exists():
    with open(install_json_path) as f:
        install_data = json.load(f)
    
    print("Installation metadata found:")
    print(f"  Validation disabled: {install_data.get('validation_disabled', False)}")
    print(f"  Source: {install_data.get('source', 'Unknown')}")

# Check for widget schemas
schemas_path = extension_path / "schemas"
if schemas_path.exists():
    schema_files = list(schemas_path.rglob("*.json"))
    print(f"\nWidget schemas available: {len(schema_files)}")
    
    for schema_file in schema_files[:3]:  # Show first 3 schemas
        relative_path = schema_file.relative_to(schemas_path)
        print(f"  - {relative_path}")
        
        # Show schema size
        try:
            with open(schema_file) as f:
                schema_data = json.load(f)
            print(f"    Properties: {len(schema_data.get('properties', {}))}")
        except:
            print(f"    Size: {schema_file.stat().st_size} bytes")
            raise

# Show extension deployment status
print(f"\nExtension deployment:")
print(f"  Target: {extension_paths[0]['dest']}")
print(f"  Ready for JupyterLab widget management")

print("Example 4 completed successfully")
```

## Example 5: jupyterlab_widgets functionality test 5

```python
import jupyterlab_widgets
from pathlib import Path

# Example 5: Complete JupyterLab widgets extension validation
print("Example 5: Complete JupyterLab widgets extension validation")

# Validate extension configuration
extension_paths = jupyterlab_widgets._jupyter_labextension_paths()
extension_config = extension_paths[0]

print(f"JupyterLab Widgets Extension Validation:")
print(f"  Package version: {jupyterlab_widgets.__version__}")
print(f"  Extension name: {extension_config['dest']}")

# Validate extension structure
extension_path = Path(extension_config['src'])
validation_results = {
    "extension_exists": extension_path.exists(),
    "package_json_exists": (extension_path / "package.json").exists(),
    "schemas_available": (extension_path / "schemas").exists(),
    "static_files_present": (extension_path / "static").exists(),
    "install_metadata": (extension_path / "install.json").exists()
}

print(f"\nExtension structure validation:")
for check, result in validation_results.items():
    status = "✓" if result else "✗"
    print(f"  {status} {check.replace('_', ' ').title()}")

# Check if this is the standard Jupyter widgets manager
is_standard_manager = extension_config['dest'] == '@jupyter-widgets/jupyterlab-manager'
print(f"\nStandard widgets manager: {'Yes' if is_standard_manager else 'No'}")

# Summary
all_valid = all(validation_results.values())
print(f"Extension validation: {'PASSED' if all_valid else 'ISSUES FOUND'}")
print(f"Ready for JupyterLab widget rendering and interaction")

print("Example 5 completed successfully")
```
