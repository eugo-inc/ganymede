## Example 1: jupytext functionality test 1

```python
from jupytext import reads, writes, NOTEBOOK_EXTENSIONS, guess_format
import nbformat

# Example 1: Basic notebook to text conversion
print("Example 1: Basic notebook to text conversion")

# Create a simple notebook programmatically
nb = nbformat.v4.new_notebook()
nb.cells.append(nbformat.v4.new_markdown_cell("# Hello Jupytext"))
nb.cells.append(nbformat.v4.new_code_cell("print('Hello, World!')"))
nb.cells.append(nbformat.v4.new_code_cell("x = 42\nprint(f'The answer is {x}')"))

# Convert notebook to Python script format
python_text = writes(nb, fmt="py:percent")
print(f"Converted to Python script ({len(python_text)} characters):")
print("--- Python Script ---")
print(python_text[:200] + "..." if len(python_text) > 200 else python_text)

# Convert notebook to Markdown format
markdown_text = writes(nb, fmt="md")
print(f"\nConverted to Markdown ({len(markdown_text)} characters):")
print("--- Markdown ---")
print(markdown_text[:200] + "..." if len(markdown_text) > 200 else markdown_text)

# Show supported notebook extensions
print(f"\nSupported notebook extensions: {NOTEBOOK_EXTENSIONS}")

print("Example 1 completed successfully")
```

## Example 2: jupytext functionality test 2

```python
from jupytext import reads, guess_format, get_format_implementation

# Example 2: Text to notebook conversion and format detection
print("Example 2: Text to notebook conversion and format detection")

# Sample Python script with percent formatting
python_script = '''# ---
# jupyter:
#   jupytext:
#     text_representation:
#       extension: .py
#       format_name: percent
# ---

# %% [markdown]
# # Data Analysis Example

# %%
import pandas as pd
import numpy as np

# %%
# Create sample data
data = np.random.randn(100, 3)
df = pd.DataFrame(data, columns=['A', 'B', 'C'])
print(f"Created DataFrame with shape: {df.shape}")

# %% [markdown]
# Analysis complete!
'''

# Detect format automatically
detected_format, format_options = guess_format(python_script, ".py")
print(f"Detected format: {detected_format}")
print(f"Format options: {format_options}")

# Convert Python script back to notebook
notebook = reads(python_script, fmt="py:percent")
print(f"\nConverted to notebook:")
print(f"  Number of cells: {len(notebook.cells)}")
print(f"  Cell types: {[cell.cell_type for cell in notebook.cells]}")

# Show first few cells
for i, cell in enumerate(notebook.cells[:3]):
    print(f"  Cell {i+1} ({cell.cell_type}): {cell.source[:50]}...")

# Get format implementation details
format_impl = get_format_implementation(".py", "percent")
print(f"\nFormat implementation: {format_impl}")

print("Example 2 completed successfully")
```

## Example 3: jupytext functionality test 3

```python
from jupytext import TextFileContentsManager, __version__
from jupytext.formats import NOTEBOOK_EXTENSIONS
import tempfile
import os

# Example 3: Jupytext contents manager for seamless integration
print("Example 3: Jupytext contents manager for seamless integration")

# Show version and supported formats
print(f"Jupytext version: {__version__}")
print(f"Notebook extensions: {NOTEBOOK_EXTENSIONS}")

# Create a text file contents manager (mock demonstration)
contents_manager = TextFileContentsManager()
print(f"Contents manager: {contents_manager.__class__.__name__}")

# Show configuration options
print(f"Contents manager attributes:")
manager_attrs = [attr for attr in dir(contents_manager) if not attr.startswith('_')][:10]
for attr in manager_attrs:
    obj = getattr(contents_manager, attr)
    if callable(obj):
        print(f"  {attr}(): Method")
    else:
        print(f"  {attr}: {type(obj).__name__}")

# Demonstrate file format handling
with tempfile.TemporaryDirectory() as temp_dir:
    # Create a sample Python file
    py_file = os.path.join(temp_dir, "notebook.py")
    sample_content = '''# %%
print("Hello from Jupytext!")

# %% [markdown]
# This is a markdown cell
'''
    
    with open(py_file, 'w') as f:
        f.write(sample_content)
    
    print(f"\nFile created: {os.path.basename(py_file)}")
    print(f"Content preview: {sample_content[:50]}...")

print("Example 3 completed successfully")
```

## Example 4: jupytext functionality test 4

```python
from jupytext.myst import is_myst_available, myst_extensions
from jupytext.pandoc import is_pandoc_available
from jupytext.config import load_jupytext_config
from jupytext.languages import _SCRIPT_EXTENSIONS, _COMMENT_CHARS

# Example 4: Jupytext format support and configuration
print("Example 4: Jupytext format support and configuration")

# Check available format converters
print(f"MyST Markdown available: {is_myst_available()}")
if is_myst_available():
    print(f"MyST extensions: {myst_extensions()}")

print(f"Pandoc available: {is_pandoc_available()}")

# Show supported script extensions and their language info
print(f"\nSupported script extensions:")
for ext, info in list(_SCRIPT_EXTENSIONS.items())[:8]:
    print(f"  {ext}: {info['language']} (comment: '{info['comment']}')")

# Show comment characters used across languages
print(f"\nComment characters used: {set(_COMMENT_CHARS)}")
print(f"Total comment variations: {len(_COMMENT_CHARS)}")

# Demonstrate configuration loading
try:
    config = load_jupytext_config(".")
    print(f"\nJupytext configuration loaded:")
    print(f"  Config type: {type(config).__name__}")
    
    # Show configuration attributes
    config_attrs = [attr for attr in dir(config) if not attr.startswith('_')][:5]
    print(f"  Available settings: {config_attrs}")
    
except Exception as e:
    print(f"\nNo configuration file found (default behavior)")
    print(f"Jupytext will use default settings")
    raise

# Show language support
print(f"\nLanguage support demonstrates Jupytext's versatility:")
print(f"  Python, R, Julia, Scala, and many more languages supported")

print("Example 4 completed successfully")
```

## Example 5: jupytext functionality test 5

```python
from jupytext import read, write, reads, writes
from jupytext.combine import combine_inputs_with_outputs
from jupytext.compare import compare_notebooks, NotebookDifference
import nbformat
import tempfile
import os

# Example 5: Complete Jupytext workflow with pairing and comparison
print("Example 5: Complete Jupytext workflow with pairing and comparison")

# Create two versions of a notebook for comparison
nb1 = nbformat.v4.new_notebook()
nb1.cells = [
    nbformat.v4.new_markdown_cell("# Original Version"),
    nbformat.v4.new_code_cell("result = 2 + 2\nprint(result)")
]

nb2 = nbformat.v4.new_notebook()
nb2.cells = [
    nbformat.v4.new_markdown_cell("# Modified Version"),
    nbformat.v4.new_code_cell("result = 3 + 3\nprint(f'Result: {result}')")
]

# Compare notebooks (raises exception if different)
try:
    compare_notebooks(nb1, nb2, raise_on_first_difference=False)
except NotebookDifference as e:
    assert str(e).startswith("Cells 1,2 differ"), e  # both cells were changed
    print(f"Notebook comparison: Differences found between notebooks")
else:
    raise AssertionError("expected NotebookDifference: nb1 and nb2 differ in both cells")

# Demonstrate file I/O with temporary files
with tempfile.TemporaryDirectory() as temp_dir:
    # Write notebook as Python file
    py_file = os.path.join(temp_dir, "demo.py")
    write(nb1, py_file, fmt="py:light")
    
    # Read it back
    nb_from_file = read(py_file)
    print(f"\nFile round-trip conversion:")
    print(f"  Original cells: {len(nb1.cells)}")
    print(f"  Loaded cells: {len(nb_from_file.cells)}")
    
    # Demonstrate combining with outputs (mock outputs)
    nb_with_outputs = nbformat.v4.new_notebook()
    nb_with_outputs.cells = [
        nbformat.v4.new_code_cell("print('hello')", 
                                 outputs=[{"output_type": "stream", "name": "stdout", "text": ["hello\n"]}])
    ]
    
    # Use combine function to merge inputs and outputs
    combined_nb = combine_inputs_with_outputs(nb_from_file, nb_with_outputs)
    print(f"  Combined notebook cells: {len(combined_nb.cells)}")

print("\nJupytext enables seamless notebook-text conversion workflows")
print("Example 5 completed successfully")
```
