```python
# Public API surface inspection.
import kernels

public_names = [n for n in dir(kernels) if not n.startswith('_')]
print(f"Number of public names exported by kernels: {len(public_names)}")
print("First 30 public names:")
for n in sorted(public_names)[:30]:
    obj = getattr(kernels, n)
    print(f"  {n}: {type(obj).__name__}")

assert len(public_names) > 0, "kernels exposes no public names"
```

```python
# Exercise the lock-file / version-resolution helpers (pure-python, no network).
import kernels

# These names are part of the canonical kernels API surface.
candidate_callables = [
    'get_kernel', 'load_kernel', 'get_locked_kernel',
    'install_kernel', 'install_kernel_all_variants',
    'LayerRepository', 'Mode', 'Device',
]

found = {}
for name in candidate_callables:
    if hasattr(kernels, name):
        found[name] = type(getattr(kernels, name)).__name__

print(f"Found {len(found)} canonical API surfaces:")
for name, kind in found.items():
    print(f"  {name}: {kind}")

# We expect at least one of the canonical entrypoints to exist.
assert len(found) > 0, f"None of the canonical kernels API surfaces are available: {candidate_callables}"
```

```python
# Mode / Device enum surface (CPU-safe; no GPU required to instantiate enums).
import kernels

has_mode = hasattr(kernels, 'Mode')
has_device = hasattr(kernels, 'Device')

if has_mode:
    Mode = kernels.Mode
    print(f"Mode type: {type(Mode).__name__}")
    members = [m for m in dir(Mode) if not m.startswith('_')]
    print(f"Mode members (first 10): {members[:10]}")

if has_device:
    Device = kernels.Device
    print(f"Device type: {type(Device).__name__}")
    members = [m for m in dir(Device) if not m.startswith('_')]
    print(f"Device members (first 10): {members[:10]}")

assert has_mode or has_device, "Neither Mode nor Device enum is available"
```
