## Example 1: kiwisolver functionality test 1

```python
import kiwisolver as kiwi

# Example 1: Basic constraint solving with variables
print("Example 1: Basic constraint solving with variables")

# Create variables for a simple layout problem
x = kiwi.Variable("x")
y = kiwi.Variable("y")
width = kiwi.Variable("width")

# Create a solver
solver = kiwi.Solver()

# Add constraints: x should be 10, y should be 20, width should be 100
constraints = [
    x == 10,
    y == 20,
    width == 100
]

for constraint in constraints:
    solver.addConstraint(constraint)

# Update and solve
solver.updateVariables()

print(f"Variable values after solving:")
print(f"  x = {x.value()}")
print(f"  y = {y.value()}")
print(f"  width = {width.value()}")

print("Example 1 completed successfully")
```

## Example 2: kiwisolver functionality test 2

```python
import kiwisolver as kiwi

# Example 2: Constraint strengths and priorities
print("Example 2: Constraint strengths and priorities")

# Create variables
left = kiwi.Variable("left")
right = kiwi.Variable("right")
center = kiwi.Variable("center")

# Create solver
solver = kiwi.Solver()

# Add constraints with different strengths
# Strong constraint: left must be at 0
solver.addConstraint(left == 0)

# Medium constraint: right should be at 100 (can be violated if needed)
solver.addConstraint((right == 100) | kiwi.strength.medium)

# Weak constraint: center should be in the middle
solver.addConstraint((center == (left + right) / 2) | kiwi.strength.weak)

# Required constraint: center must be at least 30
solver.addConstraint(center >= 30)

solver.updateVariables()

print(f"Layout with constraint priorities:")
print(f"  Left edge: {left.value()}")
print(f"  Center: {center.value()}")
print(f"  Right edge: {right.value()}")

print("Example 2 completed successfully")
```

## Example 3: kiwisolver functionality test 3

```python
import kiwisolver as kiwi

# Example 3: Edit variables for dynamic constraints
print("Example 3: Edit variables for dynamic constraints")

# Create variables for a resizable box
box_width = kiwi.Variable("box_width")
box_height = kiwi.Variable("box_height")
aspect_ratio = 1.5

solver = kiwi.Solver()

# Add base constraints
solver.addConstraint(box_width >= 50)
solver.addConstraint(box_height >= 30)
solver.addConstraint(box_height == box_width / aspect_ratio)

# Add edit variable to allow dynamic width changes
solver.addEditVariable(box_width, kiwi.strength.strong)
solver.suggestValue(box_width, 150)

solver.updateVariables()
print(f"Initial box dimensions:")
print(f"  Width: {box_width.value():.2f}")
print(f"  Height: {box_height.value():.2f}")

# Dynamically change the width
solver.suggestValue(box_width, 200)
solver.updateVariables()
print(f"\nAfter resizing:")
print(f"  Width: {box_width.value():.2f}")
print(f"  Height: {box_height.value():.2f}")

print("Example 3 completed successfully")
```

## Example 4: kiwisolver functionality test 4

```python
import kiwisolver as kiwi

# Example 4: Complex expressions and relationships
print("Example 4: Complex expressions and relationships")

# Create variables for UI layout
button1_x = kiwi.Variable("button1_x")
button2_x = kiwi.Variable("button2_x")
button3_x = kiwi.Variable("button3_x")
spacing = kiwi.Variable("spacing")
container_width = kiwi.Variable("container_width")

solver = kiwi.Solver()

# Container width is fixed at 300
solver.addConstraint(container_width == 300)

# Buttons are evenly spaced
solver.addConstraint(button1_x == spacing)
solver.addConstraint(button2_x == button1_x + 80 + spacing)
solver.addConstraint(button3_x == button2_x + 80 + spacing)

# Total width constraint
solver.addConstraint(button3_x + 80 + spacing <= container_width)

# Prefer equal spacing
solver.addConstraint((spacing == 20) | kiwi.strength.medium)

solver.updateVariables()
print(f"Button layout:")
print(f"  Button 1 X: {button1_x.value():.1f}")
print(f"  Button 2 X: {button2_x.value():.1f}")
print(f"  Button 3 X: {button3_x.value():.1f}")
print(f"  Spacing: {spacing.value():.1f}")

print("Example 4 completed successfully")
```

## Example 5: kiwisolver functionality test 5

```python
import kiwisolver as kiwi

# Example 5: Constraint removal and solver management
print("Example 5: Constraint removal and solver management")

# Create variables
x = kiwi.Variable("x")
y = kiwi.Variable("y")

solver = kiwi.Solver()

# Add initial constraints
c1 = x == 10
c2 = y == x + 5
c3 = y <= 20

solver.addConstraint(c1)
solver.addConstraint(c2)
solver.addConstraint(c3)

solver.updateVariables()
print(f"Initial state:")
print(f"  x: {x.value()}")
print(f"  y: {y.value()}")

# Remove a constraint and add a new one
solver.removeConstraint(c2)
c4 = y == x * 2
solver.addConstraint(c4)

solver.updateVariables()
print(f"\nAfter constraint modification:")
print(f"  x: {x.value()}")
print(f"  y: {y.value()}")

# Check if solver has constraints
print(f"\nSolver state:")
print(f"  Has constraints: {solver.hasConstraint(c1)}")
print(f"  Removed constraint present: {solver.hasConstraint(c2)}")

print("Example 5 completed successfully")
```
