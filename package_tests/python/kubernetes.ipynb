## Example 1: kubernetes functionality test 1

```python
from kubernetes import client, config
from kubernetes.client.rest import ApiException

# Example 1: Kubernetes configuration and client setup
print("Example 1: Kubernetes configuration and client setup")

# Mock configuration for demo purposes (no actual cluster needed)
try:
    # Load in-cluster config (would work in a pod)
    config.load_incluster_config()
    print("Loaded in-cluster configuration")
except config.ConfigException:
    try:
        # Load kubeconfig from default location
        config.load_kube_config()
        print("Loaded kubeconfig from file")
    except config.ConfigException:
        print("No kubeconfig found - creating mock configuration")
        # Create mock configuration for demo
        configuration = client.Configuration()
        configuration.host = "https://mock-cluster:6443"
        client.Configuration.set_default(configuration)

# Create API client instances
v1 = client.CoreV1Api()
apps_v1 = client.AppsV1Api()

print(f"Core API client: {type(v1).__name__}")
print(f"Apps API client: {type(apps_v1).__name__}")
print(f"Default host: {client.Configuration.get_default_copy().host}")

print("Example 1 completed successfully")
```

## Example 2: kubernetes functionality test 2

```python
from kubernetes import client, config
from kubernetes.client.rest import ApiException

# Example 2: Kubernetes resource model creation
print("Example 2: Kubernetes resource model creation")

# Create Kubernetes resource models without cluster connection
pod_spec = client.V1PodSpec(
    containers=[
        client.V1Container(
            name="nginx",
            image="nginx:1.20",
            ports=[client.V1ContainerPort(container_port=80)],
            resources=client.V1ResourceRequirements(
                requests={"cpu": "100m", "memory": "128Mi"},
                limits={"cpu": "500m", "memory": "512Mi"}
            )
        )
    ]
)

pod = client.V1Pod(
    metadata=client.V1ObjectMeta(name="test-pod", labels={"app": "nginx"}),
    spec=pod_spec
)

print(f"Pod name: {pod.metadata.name}")
print(f"Container image: {pod.spec.containers[0].image}")
print(f"Resource requests: {pod.spec.containers[0].resources.requests}")
print(f"Labels: {pod.metadata.labels}")

print("Example 2 completed successfully")
```

## Example 3: kubernetes functionality test 3

```python
from kubernetes import dynamic, client, config
import json

# Example 3: Dynamic client for custom resources
print("Example 3: Dynamic client for custom resources")

# Mock configuration setup
try:
    config.load_kube_config()
except config.ConfigException:
    configuration = client.Configuration()
    configuration.host = "https://mock-cluster:6443"
    client.Configuration.set_default(configuration)

# Demonstrate dynamic client classes without cluster connection
print(f"Dynamic module available: {hasattr(dynamic, 'DynamicClient')}")
print(f"ResourceInstance class: {dynamic.ResourceInstance.__name__}")
print(f"Resource class: {dynamic.Resource.__name__}")

# Create a mock resource using dynamic.Resource with required parameters
mock_resource = dynamic.Resource(
    prefix="apis",
    api_version="example.com/v1",
    kind="MyApp",
    name="myapps",
    namespaced=True
)

print(f"Mock resource prefix: {mock_resource.prefix}")
print(f"Mock resource API version: {mock_resource.api_version}")
print(f"Mock resource kind: {mock_resource.kind}")
print(f"Mock resource namespaced: {mock_resource.namespaced}")

print("Example 3 completed successfully")
```

## Example 4: kubernetes functionality test 4

```python
from kubernetes import watch, client, config
from kubernetes.stream import stream

# Example 4: Watch API and stream functionality  
print("Example 4: Watch API and stream functionality")

# Mock configuration
try:
    config.load_kube_config()
except config.ConfigException:
    configuration = client.Configuration()
    configuration.host = "https://mock-cluster:6443"
    client.Configuration.set_default(configuration)

# Demonstrate watch API structure (without actual cluster)
w = watch.Watch()
v1 = client.CoreV1Api()

print(f"Watch object created: {type(w).__name__}")
print(f"Watch methods: {[m for m in dir(w) if not m.startswith('_')]}")

# Demonstrate stream module functions
print(f"\nStream module functions:")
stream_funcs = ['stream', 'portforward', 'ws_client']
for func in stream_funcs:
    if hasattr(stream, func):
        print(f"  {func}: Available")

# Example stream configuration (for exec/portforward)
exec_command = ['/bin/sh', '-c', 'echo "Hello from pod"']
print(f"\nExample exec command: {exec_command}")
print(f"Stream would execute: {' '.join(exec_command)}")

print("Example 4 completed successfully")
```
