## Example 3: LakeFS utilities and data operations

```python
from lakefs_spec import LakeFSFileSystem
from lakefs_spec.util import depaginate, batched
import lakefs_spec
import tempfile
import os

# Example 3: LakeFS utilities and data operations
print("Example 3: LakeFS utilities and data operations")

# Create a temporary file to demonstrate checksum utility
with tempfile.NamedTemporaryFile(mode='w', delete=False) as tmp_file:
    tmp_file.write("Hello LakeFS data versioning!")
    tmp_path = tmp_file.name

print(f"Created temporary file: {os.path.basename(tmp_path)}")
print(f"File exists: {os.path.exists(tmp_path)}")

# Demonstrate pagination utility for API responses
print(f"\nPagination utility available: {callable(depaginate)}")
print(f"Purpose: Handle paginated API responses from LakeFS")

# Demonstrate batching utility
items = ["file1.csv", "file2.csv", "file3.csv", "file4.csv", "file5.csv"]
print(f"\nBatching utility demonstration:")
for i, batch in enumerate(batched(items, 2)):
    print(f"  Batch {i+1}: {list(batch)}")

# Show LakeFS URL path handling
lakefs_paths = [
    "lakefs://my-repo/main/data/table.parquet",
    "lakefs://analytics/feature-branch/models/model.pkl"
]

print(f"\nLakeFS path examples:")
for path in lakefs_paths:
    print(f"  {path}")

# Clean up
os.unlink(tmp_path)
print("Example 3 completed successfully")
```
