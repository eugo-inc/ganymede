## Example 3: langchain_anthropic functionality test 3

```python
from langchain_anthropic import convert_to_anthropic_tool

# Example 3: Tool conversion for Anthropic function calling
print("Example 3: Tool conversion for Anthropic function calling")

# Define a sample function to convert to Anthropic tool format
def calculate_area(length: float, width: float) -> float:
    """Calculate the area of a rectangle.
    
    Args:
        length: The length of the rectangle
        width: The width of the rectangle
        
    Returns:
        The area of the rectangle
    """
    return length * width

# Convert function to Anthropic tool format
try:
    anthropic_tool = convert_to_anthropic_tool(calculate_area)
    
    print(f"Function converted to Anthropic tool format")
    print(f"Tool type: {type(anthropic_tool)}")
    print(f"Tool structure: {anthropic_tool}")
    
except Exception as e:
    print(f"Tool conversion error: {type(e).__name__}: {e}")
    raise AssertionError(
        f"langchain_anthropic example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 completed successfully")
```

## Example 4: langchain_anthropic functionality test 4

```python
from langchain_anthropic import ChatAnthropic

# Example 4: message-handling surface + tool binding on ChatAnthropic.
# (ChatAnthropicMessages was a deprecated alias of ChatAnthropic, removed in
# langchain-anthropic 1.x — exercise the same message surface on ChatAnthropic.)
print("Example 4: ChatAnthropic message handling + tool binding")

chat = ChatAnthropic(
    model="claude-sonnet-4-5",
    max_tokens=800,
    temperature=0.1,
    api_key="dummy-key-for-offline-test",
)
print(f"Model: {chat.model}")

message_methods = ["invoke", "batch", "stream", "astream", "ainvoke"]
for method in message_methods:
    assert hasattr(chat, method), f"missing method: {method}"
    print(f"  {method}: True")

tool_def = {
    "name": "get_weather",
    "description": "Get the weather for a city.",
    "input_schema": {
        "type": "object",
        "properties": {"city": {"type": "string"}},
        "required": ["city"],
    },
}
bound = chat.bind_tools([tool_def])
print(f"bind_tools -> {type(bound).__name__}")

print("Example 4 completed successfully")
```
