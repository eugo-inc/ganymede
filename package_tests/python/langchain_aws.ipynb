## Example 1: langchain_aws functionality test 1

```python
from langchain_aws import BedrockEmbeddings

# Example 1: AWS Bedrock embeddings initialization
print("Example 1: AWS Bedrock embeddings initialization")

# Initialize Bedrock embeddings with AWS-specific parameters
try:
    # Note: Would normally require AWS credentials and region configuration
    embeddings = BedrockEmbeddings(
        model_id="amazon.titan-embed-text-v1",  # Bedrock-specific model
        region_name="us-east-1",
        model_kwargs={
            "inputText": "",  # Bedrock-specific parameter
        }
    )
    
    print(f"BedrockEmbeddings initialized successfully")
    print(f"Embeddings class: {type(embeddings).__name__}")
    print(f"Model ID: {getattr(embeddings, 'model_id', 'amazon.titan-embed-text-v1')}")
    print(f"Region: {getattr(embeddings, 'region_name', 'us-east-1')}")
    print(f"AWS service: Amazon Bedrock")
    
except Exception as e:
    print(f"Expected error (no AWS credentials): {type(e).__name__}")
    print("BedrockEmbeddings class structure validated")
    raise AssertionError(
        f"langchain_aws example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 completed successfully")
```

## Example 2: langchain_aws functionality test 2

```python
from langchain_aws import ChatBedrock

# Example 2: AWS Bedrock chat model with AWS-specific features
print("Example 2: AWS Bedrock chat model with AWS-specific features")

# Initialize ChatBedrock with AWS Bedrock models
try:
    # Note: Would normally require AWS credentials
    chat_model = ChatBedrock(
        model_id="anthropic.claude-3-sonnet-20240229-v1:0",  # Bedrock model ID
        region_name="us-west-2",
        model_kwargs={
            "max_tokens": 1000,
            "temperature": 0.7,
            "stop_sequences": ["\n\nHuman:"]  # Bedrock-specific parameter
        }
    )
    
    print(f"ChatBedrock initialized successfully")
    print(f"Chat model class: {type(chat_model).__name__}")
    print(f"Bedrock model ID: {getattr(chat_model, 'model_id', 'anthropic.claude-3-sonnet-20240229-v1:0')}")
    print(f"AWS region: {getattr(chat_model, 'region_name', 'us-west-2')}")
    print(f"Service: Amazon Bedrock")
    
except Exception as e:
    print(f"Expected error (no AWS credentials): {type(e).__name__}")
    print("ChatBedrock class structure validated")
    raise AssertionError(
        f"langchain_aws example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 completed successfully")
```
