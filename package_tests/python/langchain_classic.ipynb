```python
# Example 1: PromptTemplate — core templated-string primitive.
from langchain_classic.prompts import PromptTemplate

tmpl = PromptTemplate.from_template("Translate {text} to {language}.")
rendered = tmpl.format(text="hello world", language="French")
print(rendered)

assert "hello world" in rendered
assert "French" in rendered
assert set(tmpl.input_variables) == {"text", "language"}
```

```python
# Example 2: ChatPromptTemplate — message-list templating used by chat models.
from langchain_classic.prompts import ChatPromptTemplate

chat_tmpl = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {role}."),
    ("human", "{question}"),
])
messages = chat_tmpl.format_messages(role="librarian", question="What is a Dewey decimal?")
for m in messages:
    print(type(m).__name__, "->", m.content)

assert len(messages) == 2
assert "librarian" in messages[0].content
assert "Dewey" in messages[1].content
```

```python
# Example 3: Output parsers — pure structured-output handling, no LLM needed.
from langchain_classic.output_parsers import CommaSeparatedListOutputParser

parser = CommaSeparatedListOutputParser()
instr = parser.get_format_instructions()
print("Format instructions:", instr)

parsed = parser.parse("red, green, blue, yellow")
print("Parsed:", parsed)

assert isinstance(parsed, list)
assert parsed == ["red", "green", "blue", "yellow"]
```

```python
# Example 4: FakeListLLM + LLMChain — wire a fake LLM through the classic chain API.
from langchain_classic.llms.fake import FakeListLLM
from langchain_classic.chains import LLMChain
from langchain_classic.prompts import PromptTemplate

fake = FakeListLLM(responses=["42", "hello"])
prompt = PromptTemplate.from_template("Answer: {question}")
chain = LLMChain(llm=fake, prompt=prompt)

out = chain.invoke({"question": "meaning of life?"})
print(out)

assert out["text"] == "42"
out2 = chain.invoke({"question": "greeting?"})
assert out2["text"] == "hello"
```

```python
# Example 5: In-memory chat history (memory primitive, no external store).
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.messages import HumanMessage, AIMessage

history = InMemoryChatMessageHistory()
history.add_message(HumanMessage(content="Hi there"))
history.add_message(AIMessage(content="Hello! How can I help?"))
history.add_message(HumanMessage(content="What's 2+2?"))
history.add_message(AIMessage(content="4"))

for m in history.messages:
    print(type(m).__name__, "->", m.content)

assert len(history.messages) == 4
assert history.messages[-1].content == "4"
print("\nlangchain_classic smoke test PASS")
```
