## Example 1: langchain_core functionality test 1

```python
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage
from langchain_core.prompts import ChatPromptTemplate

# Example 1: Core message types and chat prompt templates
print("Example 1: Core message types and chat prompt templates")

# Create different message types
system_msg = SystemMessage(content="You are a helpful AI assistant specialized in explaining concepts clearly.")
human_msg = HumanMessage(content="Explain quantum computing in simple terms")
ai_msg = AIMessage(content="Quantum computing uses quantum bits that can exist in multiple states simultaneously...")

print(f"System message: {type(system_msg).__name__}")
print(f"Human message: {type(human_msg).__name__}")
print(f"AI message: {type(ai_msg).__name__}")

# Create a chat prompt template
chat_template = ChatPromptTemplate.from_messages([
    ("system", "You are an expert {role}"),
    ("human", "Please explain {topic} to me"),
    ("ai", "I'll be happy to explain {topic}. Let me break it down:")
])

print(f"\nChat template created: {type(chat_template).__name__}")
print(f"Template variables: {chat_template.input_variables}")

# Format the template
formatted = chat_template.format_messages(role="scientist", topic="photosynthesis")
print(f"Formatted messages: {len(formatted)} messages")

print("Example 1 completed successfully")
```

## Example 2: langchain_core functionality test 2

```python
from langchain_core.runnables import RunnableLambda, RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

# Example 2: Runnable chains for data processing pipelines
print("Example 2: Runnable chains for data processing pipelines")

# Create a simple data processing function
def process_text(data):
    if isinstance(data, dict):
        text = data.get("text", "")
    else:
        text = str(data)
    return text.upper().strip()

# Create runnable components
text_processor = RunnableLambda(process_text)
passthrough = RunnablePassthrough()
str_parser = StrOutputParser()

print(f"Text processor: {type(text_processor).__name__}")
print(f"Passthrough: {type(passthrough).__name__}")
print(f"String parser: {type(str_parser).__name__}")

# Chain them together using pipe operator
processing_chain = passthrough | text_processor

# Test the chain
test_input = {"text": "  hello world  "}
result = processing_chain.invoke(test_input)

print(f"\nInput: {test_input}")
print(f"Output: {result}")
print(f"Chain type: {type(processing_chain).__name__}")

print("Example 2 completed successfully")
```

## Example 3: langchain_core functionality test 3

```python
from langchain_core.tools import tool, BaseTool
from langchain_core.documents import Document

# Example 3: Custom tool creation and document handling
print("Example 3: Custom tool creation and document handling")

# Create a custom tool using the @tool decorator
@tool
def calculate_word_count(text: str) -> int:
    """Calculate the number of words in a text string.
    
    Args:
        text: The input text to count words for
        
    Returns:
        The number of words in the text
    """
    return len(text.split())

print(f"Custom tool created: {calculate_word_count.name}")
print(f"Tool description: {calculate_word_count.description}")
print(f"Tool type: {type(calculate_word_count).__name__}")

# Test the tool
sample_text = "This is a sample text for word counting"
word_count = calculate_word_count.invoke(sample_text)
print(f"\nSample text: '{sample_text}'")
print(f"Word count: {word_count}")

# Create documents with metadata
doc1 = Document(page_content="AI is revolutionizing technology", metadata={"source": "tech_article", "date": "2024"})
doc2 = Document(page_content="Machine learning enables automation", metadata={"source": "ml_guide", "date": "2024"})

print(f"\nDocument 1: {len(doc1.page_content)} chars, metadata: {doc1.metadata}")
print(f"Document 2: {len(doc2.page_content)} chars, metadata: {doc2.metadata}")

print("Example 3 completed successfully")
```

## Example 4: langchain_core functionality test 4

```python
from langchain_core.callbacks import BaseCallbackHandler
from langchain_core.tracers import BaseTracer

# Example 4: Custom callbacks and tracing for monitoring
print("Example 4: Custom callbacks and tracing for monitoring")

# Create a custom callback handler
class LoggingCallbackHandler(BaseCallbackHandler):
    """Custom callback handler for logging LLM interactions."""
    
    def __init__(self):
        self.events = []
    
    def on_llm_start(self, serialized, prompts, **kwargs):
        self.events.append(f"LLM started with {len(prompts)} prompts")
    
    def on_llm_end(self, response, **kwargs):
        self.events.append(f"LLM completed with response")
    
    def on_chain_start(self, serialized, inputs, **kwargs):
        self.events.append(f"Chain started with inputs: {list(inputs.keys())}")

# Initialize the callback handler
callback = LoggingCallbackHandler()
print(f"Custom callback handler: {type(callback).__name__}")
print(f"Base class: {BaseCallbackHandler.__name__}")

# Simulate some events
callback.on_chain_start({}, {"input": "test"})
callback.on_llm_start({}, ["prompt1", "prompt2"])
callback.on_llm_end({})

print(f"\nLogged events:")
for i, event in enumerate(callback.events, 1):
    print(f"  {i}. {event}")

print(f"\nBase tracer available: {BaseTracer}")

print("Example 4 completed successfully")
```

## Example 5: langchain_core functionality test 5

```python
from langchain_core.runnables import Runnable, RunnableConfig
from langchain_core.output_parsers import BaseOutputParser

# Example 5: Custom runnable and output parser implementation
print("Example 5: Custom runnable and output parser implementation")

# Create a custom output parser
class JSONOutputParser(BaseOutputParser):
    """Parse JSON strings into Python dictionaries."""
    
    def parse(self, text: str) -> dict:
        import json
        try:
            return json.loads(text.strip())
        except json.JSONDecodeError:
            return {"error": "Invalid JSON", "raw_text": text}

# Create a custom runnable
class TextProcessorRunnable(Runnable):
    """Custom runnable for text processing."""
    
    def invoke(self, input_data, config=None):
        if isinstance(input_data, str):
            return input_data.title()
        return str(input_data).title()

# Initialize components
json_parser = JSONOutputParser()
text_processor = TextProcessorRunnable()

print(f"JSON parser: {type(json_parser).__name__}")
print(f"Text processor: {type(text_processor).__name__}")

# Test the parser
json_text = '{"name": "alice", "age": 30}'
parsed = json_parser.parse(json_text)
print(f"\nParsed JSON: {parsed}")

# Test the runnable
processed_text = text_processor.invoke("hello world")
print(f"Processed text: {processed_text}")

print(f"\nRunnable config type: {RunnableConfig}")

print("Example 5 completed successfully")
```
