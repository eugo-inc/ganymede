# langchain_deepseek smoke test

Tests basic import + primary API surface of `langchain_deepseek` (DeepSeek chat model integration for LangChain).

```python
# Example 1: ChatDeepSeek class is importable from the top-level package.
from langchain_deepseek import ChatDeepSeek

print(f"ChatDeepSeek: {ChatDeepSeek}")
print(f"module: {ChatDeepSeek.__module__}")
print(f"qualname: {ChatDeepSeek.__qualname__}")

assert ChatDeepSeek is not None
assert isinstance(ChatDeepSeek, type)
```

```python
# Example 2: ChatDeepSeek is a BaseChatModel subclass (LangChain integration contract).
from langchain_core.language_models.chat_models import BaseChatModel

from langchain_deepseek import ChatDeepSeek

assert issubclass(ChatDeepSeek, BaseChatModel), (
    f"ChatDeepSeek must subclass BaseChatModel, got mro={ChatDeepSeek.__mro__}"
)
print(f"ChatDeepSeek MRO (first 5):")
for cls in ChatDeepSeek.__mro__[:5]:
    print(f"  {cls.__module__}.{cls.__name__}")
```

```python
# Example 3: Instantiate ChatDeepSeek with a dummy API key (no network call).
import os

from langchain_deepseek import ChatDeepSeek

os.environ.setdefault("DEEPSEEK_API_KEY", "sk-dummy-test-key-not-real")

llm = ChatDeepSeek(model="deepseek-chat", temperature=0.0, max_tokens=64)

print(f"llm type: {type(llm).__name__}")
print(f"model: {llm.model_name}")
print(f"temperature: {llm.temperature}")
print(f"max_tokens: {llm.max_tokens}")

assert llm.model_name == "deepseek-chat"
assert llm.temperature == 0.0
```

```python
# Example 4: chat_models submodule path + ChatDeepSeek re-export consistency.
import langchain_deepseek
from langchain_deepseek import ChatDeepSeek
from langchain_deepseek.chat_models import ChatDeepSeek as ChatDeepSeekFromSubmodule

assert ChatDeepSeek is ChatDeepSeekFromSubmodule, (
    "Top-level ChatDeepSeek must be the same object as chat_models.ChatDeepSeek"
)

exported = [n for n in dir(langchain_deepseek) if not n.startswith("_")]
print(f"Top-level public names: {exported}")
assert "ChatDeepSeek" in exported
```

```python
# Example 5: ChatDeepSeek builds on langchain-openai's BaseChatOpenAI
# (DeepSeek exposes an OpenAI-compatible API; since langchain-deepseek 1.x the
# base class is BaseChatOpenAI, not the concrete ChatOpenAI).
from langchain_openai.chat_models.base import BaseChatOpenAI

from langchain_deepseek import ChatDeepSeek

assert issubclass(ChatDeepSeek, BaseChatOpenAI), (
    f"ChatDeepSeek should subclass BaseChatOpenAI (OpenAI-compatible), got mro={ChatDeepSeek.__mro__}"
)

# Verify standard LangChain methods are available on the class.
for method_name in ("invoke", "stream", "batch", "ainvoke", "bind_tools"):
    assert hasattr(ChatDeepSeek, method_name), f"missing method: {method_name}"

print("ChatDeepSeek extends BaseChatOpenAI and exposes the standard Runnable interface.")
print("All assertions passed.")
```
