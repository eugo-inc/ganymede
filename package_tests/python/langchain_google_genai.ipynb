## Example 1: langchain_google_genai functionality test 1

```python
from langchain_google_genai import ChatGoogleGenerativeAI

# Example 1: Gemini function-calling — bind a Python tool (offline, dummy key).
# (GenAIAqa / AqaInput / AqaOutput were removed in langchain-google-genai 4.x;
# tool binding is the current attributed-answer / grounding entry point.)
print("Example 1: Gemini function-calling tool binding")

chat = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    api_key="dummy-key-for-offline-test",
    temperature=0.2,
)
print(f"Model: {chat.model}")

def get_weather(city: str) -> str:
    """Return the weather for a city."""
    return f"sunny in {city}"

bound = chat.bind_tools([get_weather])
print(f"bind_tools -> {type(bound).__name__}")
assert hasattr(bound, "invoke")

print("Example 1 completed successfully")
```

## Example 2: langchain_google_genai functionality test 2

```python
from langchain_google_genai import ComputerUse, Environment, MediaResolution, Modality, create_context_cache

# Example 2: 4.x configuration surface — computer use, media resolution, modality,
# and context caching. (GoogleVectorStore / DoesNotExistsException were removed in
# langchain-google-genai 4.x; these are the current top-level config exports.)
print("Example 2: 4.x configuration surface")

print(f"Environment: {[e.name for e in Environment]}")
print(f"MediaResolution: {[m.name for m in MediaResolution]}")
print(f"Modality: {[m.name for m in Modality]}")

cu = ComputerUse(environment=Environment.ENVIRONMENT_BROWSER)
print(f"ComputerUse environment: {cu.environment.name}")
assert cu.environment is Environment.ENVIRONMENT_BROWSER

assert callable(create_context_cache)
print(f"create_context_cache: {create_context_cache.__name__}")

print("Example 2 completed successfully")
```
