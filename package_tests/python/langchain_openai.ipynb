## Example 1: Basic ChatOpenAI Model Setup and Configuration

Basic usage of ChatOpenAI model configuration and initialization patterns.

```python
# Example 1: Basic ChatOpenAI model setup and configuration
try:
    from langchain_openai import ChatOpenAI
    
    print("=== ChatOpenAI Model Configuration ===")
    
    # Basic model configuration without API calls
    model_configs = {
        "gpt-3.5-turbo": {
            "temperature": 0.7,
            "max_tokens": 1000,
            "timeout": 60.0
        },
        "gpt-4": {
            "temperature": 0.3,
            "max_tokens": 2000,
            "timeout": 120.0
        },
        "gpt-4-turbo": {
            "temperature": 0.5,
            "max_tokens": 4000,
            "timeout": 180.0
        }
    }
    
    print("Model configuration examples:")
    for model_name, config in model_configs.items():
        print(f"  {model_name}:")
        print(f"    Temperature: {config['temperature']} (creativity level)")
        print(f"    Max tokens: {config['max_tokens']} (response length)")
        print(f"    Timeout: {config['timeout']}s (request timeout)")
    
    print("\n=== Model Initialization Patterns ===")
    
    # Initialize models with different configurations (without API key for testing)
    models = {}
    
    for model_name, config in model_configs.items():
        # Create model instance (will work without API key for inspection)
        model = ChatOpenAI(
            model_name=model_name,
            temperature=config['temperature'],
            max_tokens=config['max_tokens'],
            timeout=config['timeout'],
            api_key="dummy-key-for-testing"  # Dummy key for testing
        )
        models[model_name] = model
        print(f"✓ {model_name} model initialized")
        
        # Inspect model properties
        print(f"  Model class: {type(model).__name__}")
        print(f"  Temperature: {model.temperature}")
        print(f"  Max tokens: {model.max_tokens}")
        assert (model.model_name, model.temperature, model.max_tokens, model.request_timeout) == (
            model_name, config['temperature'], config['max_tokens'], config['timeout']), model
        
    
    print(f"\nSuccessfully configured {len(models)} models")
    
    print("\n=== Configuration Validation ===")
    
    # Validate configuration parameters
    def validate_model_config(model_name, config):
        """Validate model configuration parameters"""
        issues = []
        
        # Temperature validation
        temp = config.get('temperature', 0.7)
        if not 0.0 <= temp <= 2.0:
            issues.append(f"Temperature {temp} outside valid range [0.0, 2.0]")
        
        # Max tokens validation
        max_tokens = config.get('max_tokens', 1000)
        model_limits = {
            'gpt-3.5-turbo': 4096,
            'gpt-4': 8192, 
            'gpt-4-turbo': 128000
        }
        
        if model_name in model_limits:
            if max_tokens > model_limits[model_name]:
                issues.append(f"Max tokens {max_tokens} exceeds model limit {model_limits[model_name]}")
        
        # Timeout validation
        timeout = config.get('timeout', 60.0)
        if timeout < 10.0 or timeout > 600.0:
            issues.append(f"Timeout {timeout}s outside recommended range [10.0, 600.0]")
        
        return issues
    
    print("Configuration validation results:")
    for model_name, config in model_configs.items():
        issues = validate_model_config(model_name, config)
        if issues:
            print(f"  {model_name}: Issues found:")
            for issue in issues:
                print(f"    - {issue}")
        else:
            print(f"  {model_name}: Configuration valid ✓")
    
    print("Example 1: Basic ChatOpenAI model setup test passed")
    
except Exception as e:
    print(f"Example 1: Error in model setup test: {e}")
    import traceback
    traceback.print_exc()
    raise
```

## Example 2: Message Formatting and Chat Interfaces

Demonstrate message structures and chat conversation patterns with LangChain format.

```python
# Example 2: Message formatting and chat interfaces
try:
    from langchain_openai import ChatOpenAI
    from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
    
    print("=== LangChain Message Types ===")
    
    # Different message types
    message_examples = {
        "system": SystemMessage(content="You are a helpful AI assistant specialized in Python programming."),
        "human": HumanMessage(content="Can you explain list comprehensions in Python?"),
        "ai": AIMessage(content="List comprehensions provide a concise way to create lists in Python...")
    }
    
    print("Message type examples:")
    for msg_type, message in message_examples.items():
        print(f"  {msg_type.upper()}: {type(message).__name__}")
        print(f"    Content: {message.content[:60]}...")
        print(f"    Type: {message.type if hasattr(message, 'type') else 'N/A'}")
    
    print("\n=== Conversation Structure Patterns ===")
    
    # Create different conversation patterns
    conversation_patterns = {
        "simple_qa": [
            SystemMessage(content="You are a helpful assistant."),
            HumanMessage(content="What is the capital of France?")
        ],
        
        "multi_turn": [
            SystemMessage(content="You are a Python tutor."),
            HumanMessage(content="How do I create a list?"),
            AIMessage(content="You can create a list using square brackets: my_list = [1, 2, 3]"),
            HumanMessage(content="How do I add items to it?")
        ],
        
        "context_rich": [
            SystemMessage(content="You are a code reviewer. Analyze code quality and suggest improvements."),
            HumanMessage(content="""
def process_data(data):
    result = []
    for item in data:
        if item > 0:
            result.append(item * 2)
    return result
            """)
        ],
        
        "structured_prompt": [
            SystemMessage(content="You are a JSON API that returns structured responses."),
            HumanMessage(content="""
Please analyze this text and return JSON with sentiment, keywords, and summary:
"The new product launch was incredibly successful. Customers love the innovative features and the user-friendly interface. Sales exceeded expectations by 150%."
            """)
        ]
    }
    
    print("Conversation pattern analysis:")
    for pattern_name, messages in conversation_patterns.items():
        print(f"\n{pattern_name.upper()}:")
        print(f"  Total messages: {len(messages)}")
        print(f"  Message types: {[type(msg).__name__ for msg in messages]}")
        
        # Calculate token estimate (rough)
        total_chars = sum(len(msg.content) for msg in messages)
        estimated_tokens = total_chars // 4  # Rough estimate: 1 token ≈ 4 characters
        print(f"  Estimated tokens: ~{estimated_tokens}")
        
        # Show message flow
        for i, msg in enumerate(messages):
            role = type(msg).__name__.replace('Message', '').lower()
            preview = msg.content[:40] + "..." if len(msg.content) > 40 else msg.content
            print(f"    {i+1}. {role}: {preview}")
    
    print("\n=== Message Validation and Processing ===")
    
    # Message validation functions
    def validate_conversation(messages):
        """Validate conversation structure"""
        issues = []
        
        if not messages:
            issues.append("Empty conversation")
            return issues
        
        # Check for system message at start
        if not isinstance(messages[0], SystemMessage):
            issues.append("Conversation should start with SystemMessage")
        
        # Check message alternation (for multi-turn)
        human_ai_pattern = [msg for msg in messages if isinstance(msg, (HumanMessage, AIMessage))]
        if len(human_ai_pattern) > 1:
            for i in range(len(human_ai_pattern) - 1):
                current = type(human_ai_pattern[i])
                next_msg = type(human_ai_pattern[i + 1])
                if current == next_msg and current != HumanMessage:
                    issues.append(f"Consecutive {current.__name__} messages at position {i}")
        
        # Check message content length
        for i, msg in enumerate(messages):
            if not msg.content.strip():
                issues.append(f"Empty content in message {i}")
            elif len(msg.content) > 8000:  # Reasonable limit
                issues.append(f"Very long message at position {i} ({len(msg.content)} chars)")
        
        return issues
    
    # Validate all conversation patterns
    print("Conversation validation results:")
    for pattern_name, messages in conversation_patterns.items():
        issues = validate_conversation(messages)
        if issues:
            print(f"  {pattern_name}: Issues found")
            for issue in issues[:2]:  # Show first 2 issues
                print(f"    - {issue}")
        else:
            print(f"  {pattern_name}: Valid conversation structure ✓")
    
    print("\n=== Mock Chat Interface ===")
    
    # Simulate chat interface without API calls
    class MockChatInterface:
        def __init__(self, model_name="gpt-3.5-turbo"):
            self.model_name = model_name
            self.conversation_history = []
        
        def add_message(self, message):
            """Add message to conversation history"""
            self.conversation_history.append(message)
        
        def get_conversation_stats(self):
            """Get statistics about current conversation"""
            stats = {
                "total_messages": len(self.conversation_history),
                "human_messages": sum(1 for m in self.conversation_history if isinstance(m, HumanMessage)),
                "ai_messages": sum(1 for m in self.conversation_history if isinstance(m, AIMessage)),
                "system_messages": sum(1 for m in self.conversation_history if isinstance(m, SystemMessage)),
                "total_chars": sum(len(m.content) for m in self.conversation_history)
            }
            return stats
        
        def format_conversation(self):
            """Format conversation for display"""
            formatted = []
            for i, msg in enumerate(self.conversation_history):
                role = type(msg).__name__.replace('Message', '').upper()
                formatted.append(f"{i+1}. {role}: {msg.content[:100]}...")
            return formatted
    
    # Test mock interface
    chat = MockChatInterface()
    
    # Add some messages
    test_messages = conversation_patterns["multi_turn"]
    for msg in test_messages:
        chat.add_message(msg)
    
    stats = chat.get_conversation_stats()
    print("Mock chat interface stats:")
    for key, value in stats.items():
        print(f"  {key}: {value}")
    
    print("\nFormatted conversation:")
    for line in chat.format_conversation()[:3]:  # Show first 3
        print(f"  {line}")
    
    print("Example 2: Message formatting and chat interfaces test passed")
    
except Exception as e:
    print(f"Example 2: Error in message formatting test: {e}")
    import traceback
    traceback.print_exc()
    raise
```

## Example 3: OpenAI Embeddings and Vector Operations

Demonstrate OpenAI embeddings functionality and vector similarity operations.

```python
# Example 3: OpenAI embeddings and vector operations
try:
    from langchain_openai import OpenAIEmbeddings
    import numpy as np
    from typing import List
    
    print("=== OpenAI Embeddings Configuration ===")
    
    # Different embedding models and their specifications
    embedding_models = {
        "text-embedding-ada-002": {
            "dimensions": 1536,
            "max_tokens": 8191,
            "cost_per_1k_tokens": 0.0001,
            "use_case": "General purpose, good balance of performance and cost"
        },
        "text-embedding-3-small": {
            "dimensions": 1536,
            "max_tokens": 8191,
            "cost_per_1k_tokens": 0.00002,
            "use_case": "Cost-effective option with good performance"
        },
        "text-embedding-3-large": {
            "dimensions": 3072,
            "max_tokens": 8191,
            "cost_per_1k_tokens": 0.00013,
            "use_case": "Highest performance for complex semantic tasks"
        }
    }
    
    print("Available embedding models:")
    for model, specs in embedding_models.items():
        print(f"  {model}:")
        print(f"    Dimensions: {specs['dimensions']}")
        print(f"    Max tokens: {specs['max_tokens']:,}")
        print(f"    Cost: ${specs['cost_per_1k_tokens']:.5f}/1K tokens")
        print(f"    Use case: {specs['use_case']}")
    
    print("\n=== Embedding Instance Configuration ===")
    
    # Create embeddings instances with different configurations
    embedding_configs = {}
    
    for model_name in embedding_models.keys():
        # Create embeddings instance (without API key for testing)
        embeddings = OpenAIEmbeddings(
            model=model_name,
            api_key="dummy-key-for-testing",
            chunk_size=1000,
            max_retries=3
        )
        embedding_configs[model_name] = embeddings
        print(f"✓ {model_name} embeddings configured")
        
        # Inspect configuration
        print(f"  Model: {embeddings.model}")
        print(f"  Chunk size: {embeddings.chunk_size}")
        print(f"  Max retries: {embeddings.max_retries}")
        assert (embeddings.model, embeddings.chunk_size, embeddings.max_retries) == (model_name, 1000, 3), embeddings
        
    
    print(f"\nConfigured {len(embedding_configs)} embedding models")
    
    print("\n=== Mock Vector Operations ===")
    
    # Simulate embedding operations without API calls
    def simulate_embeddings(texts: List[str], model_name: str) -> List[List[float]]:
        """Simulate embedding generation for testing"""
        dimensions = embedding_models[model_name]["dimensions"]
        embeddings = []
        
        for text in texts:
            # Generate deterministic "embedding" based on text content
            np.random.seed(hash(text) % 2**32)  # Deterministic seed from text
            embedding = np.random.normal(0, 1, dimensions)
            # Normalize to unit vector (like real embeddings)
            embedding = embedding / np.linalg.norm(embedding)
            embeddings.append(embedding.tolist())
        
        return embeddings
    
    # Test texts for embedding
    test_texts = [
        "Python is a powerful programming language",
        "Machine learning algorithms can solve complex problems", 
        "The weather is beautiful today",
        "Natural language processing enables computers to understand text",
        "Data science combines statistics and programming"
    ]
    
    print("Sample texts for embedding:")
    for i, text in enumerate(test_texts):
        print(f"  {i+1}. {text}")
    
    # Generate mock embeddings
    model_name = "text-embedding-3-small"
    mock_embeddings = simulate_embeddings(test_texts, model_name)
    
    print(f"\nGenerated {len(mock_embeddings)} mock embeddings")
    print(f"Embedding dimensions: {len(mock_embeddings[0])}")
    print(f"Sample embedding values: {mock_embeddings[0][:5]}...")
    
    print("\n=== Vector Similarity Operations ===")
    
    def cosine_similarity(vec1: List[float], vec2: List[float]) -> float:
        """Calculate cosine similarity between two vectors"""
        vec1_np = np.array(vec1)
        vec2_np = np.array(vec2)
        
        dot_product = np.dot(vec1_np, vec2_np)
        magnitude1 = np.linalg.norm(vec1_np)
        magnitude2 = np.linalg.norm(vec2_np)
        
        if magnitude1 == 0 or magnitude2 == 0:
            return 0.0
        
        return dot_product / (magnitude1 * magnitude2)
    
    def find_most_similar(query_embedding: List[float], embeddings: List[List[float]], 
                         texts: List[str], top_k: int = 3) -> List[tuple]:
        """Find most similar texts to query"""
        similarities = []
        
        for i, embedding in enumerate(embeddings):
            similarity = cosine_similarity(query_embedding, embedding)
            similarities.append((similarity, i, texts[i]))
        
        # Sort by similarity (descending)
        similarities.sort(key=lambda x: x[0], reverse=True)
        
        return similarities[:top_k]
    
    # Test similarity search
    query_text = "Programming and software development"
    query_embedding = simulate_embeddings([query_text], model_name)[0]
    
    print(f"Query: '{query_text}'")
    print("\nMost similar texts:")
    
    similar_texts = find_most_similar(query_embedding, mock_embeddings, test_texts)
    for similarity, idx, text in similar_texts:
        print(f"  {similarity:.3f}: {text}")
    
    print("\n=== Embedding Analytics ===")
    
    # Analyze embedding properties
    def analyze_embeddings(embeddings: List[List[float]]) -> dict:
        """Analyze embedding statistics"""
        embeddings_np = np.array(embeddings)
        
        stats = {
            "count": len(embeddings),
            "dimensions": embeddings_np.shape[1] if embeddings_np.size > 0 else 0,
            "mean_magnitude": np.mean(np.linalg.norm(embeddings_np, axis=1)),
            "std_magnitude": np.std(np.linalg.norm(embeddings_np, axis=1)),
            "mean_value": np.mean(embeddings_np),
            "std_value": np.std(embeddings_np)
        }
        
        return stats
    
    embedding_stats = analyze_embeddings(mock_embeddings)
    
    print("Embedding statistics:")
    for key, value in embedding_stats.items():
        if isinstance(value, float):
            print(f"  {key}: {value:.4f}")
        else:
            print(f"  {key}: {value}")
    
    # Cross-similarity matrix
    print("\nCross-similarity matrix (first 3x3):")
    for i in range(min(3, len(mock_embeddings))):
        row = []
        for j in range(min(3, len(mock_embeddings))):
            sim = cosine_similarity(mock_embeddings[i], mock_embeddings[j])
            row.append(f"{sim:.3f}")
        print(f"  [{i}]: {' '.join(row)}")
    
    print("Example 3: OpenAI embeddings and vector operations test passed")
    
except Exception as e:
    print(f"Example 3: Error in embeddings test: {e}")
    import traceback
    traceback.print_exc()
    raise
```

## Example 4: Azure OpenAI Integration and Multi-Provider Support

Demonstrate Azure OpenAI integration and multi-provider configuration patterns.

```python
# Example 4: Azure OpenAI integration and multi-provider support
try:
    from langchain_openai import AzureChatOpenAI, AzureOpenAIEmbeddings
    
    print("=== Azure OpenAI Configuration ===")
    
    # Azure OpenAI configuration parameters
    azure_configs = {
        "eastus_deployment": {
            "azure_endpoint": "https://myresource-eastus.openai.azure.com/",
            "deployment_name": "gpt-35-turbo-prod",
            "api_version": "2024-02-15-preview",
            "model_version": "0613",
            "region": "East US"
        },
        "westus_deployment": {
            "azure_endpoint": "https://myresource-westus.openai.azure.com/",
            "deployment_name": "gpt-4-prod",
            "api_version": "2024-02-15-preview", 
            "model_version": "0613",
            "region": "West US"
        },
        "europe_deployment": {
            "azure_endpoint": "https://myresource-europe.openai.azure.com/",
            "deployment_name": "text-embedding-ada-002",
            "api_version": "2024-02-15-preview",
            "model_version": "2",
            "region": "West Europe"
        }
    }
    
    print("Azure OpenAI deployment configurations:")
    for config_name, config in azure_configs.items():
        print(f"  {config_name}:")
        print(f"    Endpoint: {config['azure_endpoint']}")
        print(f"    Deployment: {config['deployment_name']}")
        print(f"    API Version: {config['api_version']}")
        print(f"    Region: {config['region']}")
    
    print("\n=== Multi-Provider Model Factory ===")
    
    class ModelFactory:
        """Factory for creating models from different providers"""
        
        @staticmethod
        def create_chat_model(provider: str, **kwargs):
            """Create chat model based on provider"""
            if provider == "openai":
                from langchain_openai import ChatOpenAI
                return ChatOpenAI(api_key="dummy-key", **kwargs)
            
            elif provider == "azure":
                return AzureChatOpenAI(
                    api_key="dummy-key",
                    azure_endpoint=kwargs.get("azure_endpoint", "https://example.openai.azure.com/"),
                    deployment_name=kwargs.get("deployment_name", "gpt-35-turbo"),
                    api_version=kwargs.get("api_version", "2024-02-15-preview"),
                    **{k: v for k, v in kwargs.items() if k not in ["azure_endpoint", "deployment_name", "api_version"]}
                )
            
            else:
                raise ValueError(f"Unsupported provider: {provider}")
        
        @staticmethod
        def create_embeddings_model(provider: str, **kwargs):
            """Create embeddings model based on provider"""
            if provider == "openai":
                from langchain_openai import OpenAIEmbeddings
                return OpenAIEmbeddings(api_key="dummy-key", **kwargs)
            
            elif provider == "azure":
                return AzureOpenAIEmbeddings(
                    api_key="dummy-key", 
                    azure_endpoint=kwargs.get("azure_endpoint", "https://example.openai.azure.com/"),
                    deployment=kwargs.get("deployment_name", "text-embedding-ada-002"),
                    api_version=kwargs.get("api_version", "2024-02-15-preview"),
                    **{k: v for k, v in kwargs.items() if k not in ["azure_endpoint", "deployment_name", "api_version"]}
                )
            
            else:
                raise ValueError(f"Unsupported provider: {provider}")
    
    # Test model factory
    model_configs = [
        {"provider": "openai", "model_name": "gpt-3.5-turbo", "temperature": 0.7},
        {"provider": "azure", **azure_configs["eastus_deployment"], "temperature": 0.5},
        {"provider": "openai", "model": "text-embedding-ada-002"},
        {"provider": "azure", **azure_configs["europe_deployment"]}
    ]
    
    created_models = []
    
    for i, config in enumerate(model_configs):
        provider = config.pop("provider")
        
        if "model_name" in config or "model" in config:
            # It's a chat model
            if "model" in config:  # embeddings
                model = ModelFactory.create_embeddings_model(provider, **config)
                model_type = "embeddings"
            else:  # chat
                model = ModelFactory.create_chat_model(provider, **config)
                model_type = "chat"
        else:
            # Azure chat model
            model = ModelFactory.create_chat_model(provider, **config)
            model_type = "chat"
        
        created_models.append((provider, model_type, model))
        print(f"✓ Created {provider} {model_type} model")
        
    
    print(f"\nSuccessfully created {len(created_models)} models")
    # measured: the 4th config (an embedding deployment) carries neither `model_name` nor `model`, so the factory
    # routes it to a chat model -- the example's own routing, asserted as it is
    assert [(provider, model_type, type(model).__name__) for provider, model_type, model in created_models] == [
        ("openai", "chat", "ChatOpenAI"), ("azure", "chat", "AzureChatOpenAI"),
        ("openai", "embeddings", "OpenAIEmbeddings"), ("azure", "chat", "AzureChatOpenAI"),
    ], created_models
    
    print("\n=== Configuration Management ===")
    
    class ConfigManager:
        """Manage multi-provider configurations"""
        
        def __init__(self):
            self.providers = {}
        
        def add_provider_config(self, name: str, provider_type: str, config: dict):
            """Add provider configuration"""
            self.providers[name] = {
                "type": provider_type,
                "config": config,
                "created_at": "2024-01-16T10:00:00Z"
            }
        
        def get_provider_config(self, name: str) -> dict:
            """Get provider configuration"""
            return self.providers.get(name, {})
        
        def list_providers(self) -> list:
            """List all configured providers"""
            return list(self.providers.keys())
        
        def validate_config(self, name: str) -> list:
            """Validate provider configuration"""
            issues = []
            
            if name not in self.providers:
                return ["Provider not found"]
            
            config = self.providers[name]
            provider_type = config["type"]
            provider_config = config["config"]
            
            # Common validations
            if provider_type == "azure":
                required_fields = ["azure_endpoint", "deployment_name", "api_version"]
                for field in required_fields:
                    if field not in provider_config:
                        issues.append(f"Missing required field: {field}")
                
                # Endpoint validation
                endpoint = provider_config.get("azure_endpoint", "")
                if not endpoint.startswith("https://"):
                    issues.append("Azure endpoint must use HTTPS")
                if not endpoint.endswith(".openai.azure.com/"):
                    issues.append("Invalid Azure OpenAI endpoint format")
            
            elif provider_type == "openai":
                # OpenAI specific validations
                if "model_name" in provider_config:
                    model = provider_config["model_name"]
                    if model not in ["gpt-3.5-turbo", "gpt-4", "gpt-4-turbo"]:
                        issues.append(f"Unsupported OpenAI model: {model}")
            
            return issues
    
    # Test configuration manager
    config_mgr = ConfigManager()
    
    # Add configurations
    config_mgr.add_provider_config("prod_openai", "openai", {
        "model_name": "gpt-4", 
        "temperature": 0.3
    })
    
    config_mgr.add_provider_config("prod_azure", "azure", azure_configs["eastus_deployment"])
    
    config_mgr.add_provider_config("invalid_azure", "azure", {
        "azure_endpoint": "http://invalid.com",  # Invalid - not HTTPS
        "deployment_name": "test"
        # Missing api_version
    })
    
    print("Configuration validation results:")
    for provider in config_mgr.list_providers():
        issues = config_mgr.validate_config(provider)
        if issues:
            print(f"  {provider}: Issues found")
            for issue in issues:
                print(f"    - {issue}")
        else:
            print(f"  {provider}: Valid configuration ✓")
    
    print("\n=== Provider Fallback Strategy ===")
    
    class ProviderFallback:
        """Implement provider fallback for reliability"""
        
        def __init__(self):
            self.primary_providers = []
            self.fallback_providers = []
        
        def add_primary(self, provider_config: dict):
            """Add primary provider"""
            self.primary_providers.append(provider_config)
        
        def add_fallback(self, provider_config: dict):
            """Add fallback provider"""
            self.fallback_providers.append(provider_config)
        
        def get_provider_order(self) -> list:
            """Get ordered list of providers to try"""
            return self.primary_providers + self.fallback_providers
        
        def simulate_request(self, provider_config: dict) -> dict:
            """Simulate making request to provider"""
            # Simulate various outcomes
            provider_type = provider_config.get("type", "unknown")
            
            if provider_type == "azure" and "eastus" in provider_config.get("config", {}).get("azure_endpoint", ""):
                # Simulate Azure East US success
                return {"status": "success", "provider": provider_type, "region": "eastus"}
            elif provider_type == "openai":
                # Simulate OpenAI success
                return {"status": "success", "provider": provider_type}
            else:
                # Simulate failure
                return {"status": "error", "error": "Service unavailable"}
    
    # Setup fallback strategy
    fallback = ProviderFallback()
    
    fallback.add_primary({"type": "azure", "config": azure_configs["eastus_deployment"]})
    fallback.add_fallback({"type": "openai", "config": {"model_name": "gpt-3.5-turbo"}})
    fallback.add_fallback({"type": "azure", "config": azure_configs["westus_deployment"]})
    
    # Test fallback
    print("Testing provider fallback strategy:")
    providers = fallback.get_provider_order()
    
    for i, provider_config in enumerate(providers):
        result = fallback.simulate_request(provider_config)
        provider_type = provider_config["type"]
        
        print(f"  Provider {i+1} ({provider_type}): {result['status']}")
        
        if result["status"] == "success":
            print(f"    ✓ Successfully using {provider_type}")
            break
        else:
            # one provider failing is the fallback being demonstrated; the whole chain failing is the loop's else below
            print(f"    ✗ Unavailable, trying the next provider: {result.get('error', 'Unknown error')}")
    else:
        raise AssertionError("provider fallback: no provider in the chain succeeded")
    
    print("Example 4: Azure OpenAI integration and multi-provider support test passed")
    
except Exception as e:
    print(f"Example 4: Error in Azure integration test: {e}")
    import traceback
    traceback.print_exc()
    raise
```

## Example 5: Production LLM Pipeline and Error Handling

Demonstrate production-ready LLM pipeline with comprehensive error handling and monitoring.

```python
# Example 5: Production LLM pipeline and error handling
try:
    from langchain_openai import ChatOpenAI
    from langchain_core.messages import HumanMessage, SystemMessage
    import time
    from typing import Dict, List, Optional, Any
    from dataclasses import dataclass
    from enum import Enum
    
    print("=== Production LLM Pipeline Architecture ===")
    
    class RequestStatus(Enum):
        PENDING = "pending"
        SUCCESS = "success"
        FAILED = "failed"
        TIMEOUT = "timeout"
        RATE_LIMITED = "rate_limited"
    
    @dataclass
    class LLMRequest:
        id: str
        messages: List[Any]
        model: str
        temperature: float
        max_tokens: int
        timestamp: float
        status: RequestStatus = RequestStatus.PENDING
        response: Optional[str] = None
        error: Optional[str] = None
        duration_ms: Optional[float] = None
    
    @dataclass 
    class LLMMetrics:
        total_requests: int = 0
        successful_requests: int = 0
        failed_requests: int = 0
        average_latency_ms: float = 0.0
        rate_limit_hits: int = 0
        timeout_count: int = 0
        error_types: Dict[str, int] = None
        
        def __post_init__(self):
            if self.error_types is None:
                self.error_types = {}
    
    class ProductionLLMPipeline:
        """Production-ready LLM pipeline with comprehensive error handling"""
        
        def __init__(self):
            self.models = {}
            self.metrics = LLMMetrics()
            self.request_history = []
            self.circuit_breaker_failures = 0
            self.circuit_breaker_threshold = 5
            self.circuit_breaker_open = False
        
        def register_model(self, name: str, model, config: dict):
            """Register a model with configuration"""
            self.models[name] = {
                "instance": model,
                "config": config,
                "request_count": 0,
                "error_count": 0,
                "last_used": None
            }
        
        def create_request(self, messages: List[Any], model_name: str = "default", 
                          **kwargs) -> LLMRequest:
            """Create standardized LLM request"""
            request_id = f"req_{int(time.time() * 1000000)}"
            
            return LLMRequest(
                id=request_id,
                messages=messages,
                model=model_name,
                temperature=kwargs.get("temperature", 0.7),
                max_tokens=kwargs.get("max_tokens", 1000),
                timestamp=time.time()
            )
        
        def validate_request(self, request: LLMRequest) -> List[str]:
            """Validate LLM request"""
            issues = []
            
            # Model validation
            if request.model not in self.models:
                issues.append(f"Model '{request.model}' not registered")
            
            # Messages validation
            if not request.messages:
                issues.append("Empty messages list")
            
            for i, msg in enumerate(request.messages):
                if not hasattr(msg, 'content') or not msg.content.strip():
                    issues.append(f"Empty message content at index {i}")
            
            # Parameter validation
            if not 0.0 <= request.temperature <= 2.0:
                issues.append(f"Temperature {request.temperature} outside valid range [0.0, 2.0]")
            
            if request.max_tokens <= 0 or request.max_tokens > 4096:
                issues.append(f"Max tokens {request.max_tokens} outside valid range (1, 4096]")
            
            # Circuit breaker check
            if self.circuit_breaker_open:
                issues.append("Circuit breaker is open - service temporarily unavailable")
            
            return issues
        
        def simulate_llm_call(self, request: LLMRequest) -> LLMRequest:
            """Simulate LLM API call with various outcomes"""
            start_time = time.time()
            
            try:
                # Simulate different response scenarios
                import random
                
                # 80% success rate
                if random.random() < 0.8:
                    # Simulate successful response
                    time.sleep(random.uniform(0.1, 0.5))  # Simulate API latency
                    request.response = f"Mock response for request {request.id}"
                    request.status = RequestStatus.SUCCESS
                    self.circuit_breaker_failures = 0  # Reset on success
                
                # 10% timeout
                elif random.random() < 0.9:
                    time.sleep(0.1)  # Simulate timeout detection
                    request.status = RequestStatus.TIMEOUT
                    request.error = "Request timeout after 30 seconds"
                    self.circuit_breaker_failures += 1
                
                # 5% rate limiting
                elif random.random() < 0.95:
                    request.status = RequestStatus.RATE_LIMITED
                    request.error = "Rate limit exceeded - please retry after 60 seconds"
                    self.circuit_breaker_failures += 1
                
                # 5% other errors
                else:
                    request.status = RequestStatus.FAILED
                    request.error = "Internal server error"
                    self.circuit_breaker_failures += 1
                
            except Exception as e:
                request.status = RequestStatus.FAILED
                request.error = str(e)
                self.circuit_breaker_failures += 1
            
            finally:
                request.duration_ms = (time.time() - start_time) * 1000
                
                # Update circuit breaker
                if self.circuit_breaker_failures >= self.circuit_breaker_threshold:
                    self.circuit_breaker_open = True
                    print(f"⚠️  Circuit breaker opened after {self.circuit_breaker_failures} failures")
            
            return request
        
        def process_request(self, request: LLMRequest) -> LLMRequest:
            """Process LLM request with full error handling"""
            # Validate request
            issues = self.validate_request(request)
            if issues:
                request.status = RequestStatus.FAILED
                request.error = "; ".join(issues)
                return request
            
            # Update model usage stats
            if request.model in self.models:
                self.models[request.model]["request_count"] += 1
                self.models[request.model]["last_used"] = time.time()
            
            # Process request
            request = self.simulate_llm_call(request)
            
            # Update metrics
            self.update_metrics(request)
            self.request_history.append(request)
            
            return request
        
        def update_metrics(self, request: LLMRequest):
            """Update pipeline metrics"""
            self.metrics.total_requests += 1
            
            if request.status == RequestStatus.SUCCESS:
                self.metrics.successful_requests += 1
            else:
                self.metrics.failed_requests += 1
                
                # Track error types
                error_type = request.status.value
                self.metrics.error_types[error_type] = self.metrics.error_types.get(error_type, 0) + 1
                
                if request.status == RequestStatus.RATE_LIMITED:
                    self.metrics.rate_limit_hits += 1
                elif request.status == RequestStatus.TIMEOUT:
                    self.metrics.timeout_count += 1
            
            # Update average latency
            if request.duration_ms:
                total_latency = self.metrics.average_latency_ms * (self.metrics.total_requests - 1) + request.duration_ms
                self.metrics.average_latency_ms = total_latency / self.metrics.total_requests
        
        def get_health_status(self) -> Dict[str, Any]:
            """Get pipeline health status"""
            if self.metrics.total_requests == 0:
                success_rate = 0.0
            else:
                success_rate = (self.metrics.successful_requests / self.metrics.total_requests) * 100
            
            return {
                "status": "healthy" if success_rate > 80 and not self.circuit_breaker_open else "degraded",
                "success_rate": success_rate,
                "circuit_breaker_open": self.circuit_breaker_open,
                "total_requests": self.metrics.total_requests,
                "average_latency_ms": round(self.metrics.average_latency_ms, 2),
                "registered_models": len(self.models)
            }
    
    # Initialize production pipeline
    pipeline = ProductionLLMPipeline()
    
    # Register models
    try:
        chat_model = ChatOpenAI(
            model_name="gpt-3.5-turbo",
            temperature=0.7,
            api_key="dummy-key-for-testing"
        )
        
        pipeline.register_model("default", chat_model, {
            "model": "gpt-3.5-turbo",
            "max_tokens": 1000,
            "temperature_range": [0.0, 1.0]
        })
        
        print("✓ Registered default ChatOpenAI model")
    except Exception as e:
        print(f"✗ Failed to register model: {e}")
        raise
    
    print("\n=== Pipeline Testing and Monitoring ===")
    
    # Test different request scenarios
    test_scenarios = [
        {
            "name": "Valid request",
            "messages": [
                SystemMessage(content="You are a helpful assistant."),
                HumanMessage(content="What is machine learning?")
            ],
            "temperature": 0.7,
            "max_tokens": 500
        },
        {
            "name": "Invalid temperature",
            "messages": [HumanMessage(content="Hello")],
            "temperature": 3.0,  # Invalid
            "max_tokens": 100
        },
        {
            "name": "Empty messages",
            "messages": [],
            "temperature": 0.5,
            "max_tokens": 200
        },
        {
            "name": "Very long request",
            "messages": [HumanMessage(content="Test " * 1000)],  # Very long
            "temperature": 0.3,
            "max_tokens": 50
        }
    ]
    
    print("Processing test scenarios:")
    
    expected_rejections = {
        "Invalid temperature": "Temperature 3.0 outside valid range [0.0, 2.0]",
        "Empty messages": "Empty messages list",
    }
    simulated_errors = {None, "Request timeout after 30 seconds",
                        "Rate limit exceeded - please retry after 60 seconds", "Internal server error"}
    results = []
    for i, scenario in enumerate(test_scenarios):
        print(f"\nScenario {i+1}: {scenario['name']}")
        
        # Create and process request
        request = pipeline.create_request(**scenario)
        processed_request = pipeline.process_request(request)
        
        results.append(processed_request)
        
        # Display result
        print(f"  Status: {processed_request.status.value}")
        if processed_request.error:
            print(f"  Outcome: {processed_request.error}")
        if processed_request.duration_ms:
            print(f"  Duration: {processed_request.duration_ms:.1f}ms")
        expected = expected_rejections.get(scenario["name"])
        outcome = (processed_request.status.value, processed_request.error)
        if expected is not None:
            assert outcome == ("failed", expected), f"{scenario['name']}: expected rejection {expected!r}, got {outcome}"
        else:
            assert processed_request.error in simulated_errors, f"{scenario['name']}: not a simulated outcome: {outcome}"
            assert (processed_request.error is None) == (processed_request.status is RequestStatus.SUCCESS), outcome
    
    # Process additional requests to test circuit breaker
    print(f"\n=== Load Testing (10 requests) ===")
    
    load_test_messages = [
        SystemMessage(content="You are a helpful assistant."),
        HumanMessage(content="Tell me about Python programming.")
    ]
    
    for i in range(10):
        request = pipeline.create_request(load_test_messages, temperature=0.5)
        result = pipeline.process_request(request)
        
        if i < 3 or result.status != RequestStatus.SUCCESS:  # Show first 3 and any failures
            print(f"  Request {i+1}: {result.status.value}")
            if result.error:
                print(f"    Outcome: {result.error[:50]}...")
        assert (result.error is None) == (result.status is RequestStatus.SUCCESS), result
    
    print("\n=== Pipeline Health Report ===")
    
    health_status = pipeline.get_health_status()
    print("Health status:")
    for key, value in health_status.items():
        print(f"  {key}: {value}")
    
    print("\nDetailed metrics:")
    metrics = pipeline.metrics
    print(f"  Success rate: {(metrics.successful_requests / metrics.total_requests * 100) if metrics.total_requests > 0 else 0:.1f}%")
    print(f"  Rate limit hits: {metrics.rate_limit_hits}")
    print(f"  Timeouts: {metrics.timeout_count}")
    print(f"  Error breakdown: {dict(metrics.error_types)}")
    
    print("Example 5: Production LLM pipeline and error handling test passed")
    
except Exception as e:
    print(f"Example 5: Error in production pipeline test: {e}")
    import traceback
    traceback.print_exc()
    raise
```
