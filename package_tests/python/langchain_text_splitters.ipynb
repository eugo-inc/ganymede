## Example 1: langchain_text_splitters functionality test 1

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Example 1: Recursive character text splitter with intelligent chunking
print("Example 1: Recursive character text splitter with intelligent chunking")

# Create a recursive splitter that tries multiple separators
splitter = RecursiveCharacterTextSplitter(
    separators=["\n\n", "\n", ". ", " ", ""],  # Try these separators in order
    chunk_size=100,
    chunk_overlap=20,
    length_function=len,
    is_separator_regex=False
)

# Sample text with different separator levels
sample_text = """
Machine learning is transforming industries. It enables computers to learn patterns from data.

Deep learning uses neural networks with multiple layers. These networks can process complex data like images and text.

Natural language processing helps computers understand human language. This technology powers chatbots and translation tools.
"""

# Split the text intelligently
chunks = splitter.split_text(sample_text.strip())

print(f"Original text length: {len(sample_text)} characters")
print(f"Number of chunks: {len(chunks)}")
print(f"Chunk sizes: {[len(chunk) for chunk in chunks]}")

for i, chunk in enumerate(chunks):
    print(f"\nChunk {i+1}: {chunk[:50]}...")

print("Example 1 completed successfully")
```

## Example 2: langchain_text_splitters functionality test 2

```python
from langchain_text_splitters import PythonCodeTextSplitter, Language

# Example 2: Python code text splitter for code analysis
print("Example 2: Python code text splitter for code analysis")

# Create a Python-specific code splitter
python_splitter = PythonCodeTextSplitter(
    chunk_size=200,
    chunk_overlap=30
)

# Sample Python code
python_code = '''
def calculate_fibonacci(n):
    """Calculate the nth Fibonacci number using dynamic programming."""
    if n <= 1:
        return n
    
    dp = [0] * (n + 1)
    dp[1] = 1
    
    for i in range(2, n + 1):
        dp[i] = dp[i-1] + dp[i-2]
    
    return dp[n]

class MathUtils:
    @staticmethod
    def factorial(n):
        if n <= 1:
            return 1
        return n * MathUtils.factorial(n - 1)
'''

# Split the Python code intelligently
code_chunks = python_splitter.split_text(python_code)

print(f"Python code splitter for: {Language.PYTHON.value}")
print(f"Number of code chunks: {len(code_chunks)}")

for i, chunk in enumerate(code_chunks):
    print(f"\nCode chunk {i+1}:")
    print(chunk[:80] + "..." if len(chunk) > 80 else chunk)

print("Example 2 completed successfully")
```

## Example 3: langchain_text_splitters functionality test 3

```python
from langchain_text_splitters import MarkdownHeaderTextSplitter

# Example 3: Markdown header text splitter for structured documents
print("Example 3: Markdown header text splitter for structured documents")

# Create a markdown splitter that splits by headers
markdown_splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=[
        ("#", "Header 1"),
        ("##", "Header 2"),
        ("###", "Header 3"),
    ]
)

# Sample markdown document
markdown_text = """
# Introduction to AI
Artificial Intelligence is revolutionizing technology across industries.

## Machine Learning
Machine learning enables computers to learn from data without explicit programming.

### Supervised Learning
In supervised learning, models learn from labeled training data.

### Unsupervised Learning
Unsupervised learning finds patterns in data without labels.

## Deep Learning
Deep learning uses neural networks with multiple layers.

### Neural Networks
Neural networks are inspired by biological brain structures.
"""

# Split by markdown headers
md_chunks = markdown_splitter.split_text(markdown_text)

print(f"Number of markdown sections: {len(md_chunks)}")

for i, chunk in enumerate(md_chunks):
    print(f"\nSection {i+1}:")
    print(f"Metadata: {chunk.metadata}")
    print(f"Content: {chunk.page_content[:60]}...")

print("Example 3 completed successfully")
```

## Example 4: langchain_text_splitters functionality test 4

```python
from langchain_text_splitters import RecursiveJsonSplitter

# Example 4: Recursive JSON splitter for structured data
print("Example 4: Recursive JSON splitter for structured data")

# Create a JSON splitter for handling nested JSON structures
json_splitter = RecursiveJsonSplitter(
    max_chunk_size=150
)

# Sample nested JSON data
json_data = {
    "company": "TechCorp",
    "departments": {
        "engineering": {
            "teams": [
                {"name": "Backend", "members": 15, "technologies": ["Python", "PostgreSQL", "Docker"]},
                {"name": "Frontend", "members": 12, "technologies": ["React", "TypeScript", "Webpack"]},
                {"name": "DevOps", "members": 8, "technologies": ["Kubernetes", "Terraform", "AWS"]}
            ]
        },
        "product": {
            "roadmap": {
                "q1": "Feature development",
                "q2": "Performance optimization",
                "q3": "Security enhancements",
                "q4": "Mobile platform launch"
            }
        }
    },
    "metrics": {
        "revenue": 5000000,
        "customers": 250000,
        "satisfaction": 4.8
    }
}

# Split the JSON into manageable chunks
json_chunks = json_splitter.split_json(json_data)

print(f"Number of JSON chunks: {len(json_chunks)}")

for i, chunk in enumerate(json_chunks):
    print(f"\nJSON chunk {i+1}: {str(chunk)[:80]}...")

print("Example 4 completed successfully")
```

## Example 5: langchain_text_splitters functionality test 5

```python
from langchain_text_splitters import HTMLHeaderTextSplitter

# Example 5: HTML header text splitter for web content
print("Example 5: HTML header text splitter for web content")

# Create an HTML splitter that preserves document structure
html_splitter = HTMLHeaderTextSplitter(
    headers_to_split_on=[
        ("h1", "Header 1"),
        ("h2", "Header 2"),
        ("h3", "Header 3"),
    ]
)

# Sample HTML document
html_content = """
<html>
<body>
    <h1>Web Development Guide</h1>
    <p>This guide covers modern web development practices.</p>
    
    <h2>Frontend Technologies</h2>
    <p>Frontend development focuses on user interfaces and experience.</p>
    
    <h3>HTML & CSS</h3>
    <p>HTML provides structure while CSS handles styling and layout.</p>
    
    <h3>JavaScript</h3>
    <p>JavaScript adds interactivity and dynamic behavior to web pages.</p>
    
    <h2>Backend Technologies</h2>
    <p>Backend development handles server-side logic and data management.</p>
    
    <h3>Databases</h3>
    <p>Databases store and manage application data efficiently.</p>
</body>
</html>
"""

# Split HTML content by headers
html_chunks = html_splitter.split_text(html_content)

print(f"Number of HTML sections: {len(html_chunks)}")

for i, chunk in enumerate(html_chunks):
    print(f"\nHTML section {i+1}:")
    print(f"Metadata: {chunk.metadata}")
    print(f"Content: {chunk.page_content.strip()[:60]}...")

print("Example 5 completed successfully")
```
