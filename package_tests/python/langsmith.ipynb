## Example 2: langsmith functionality test 2

```python
import langsmith
from langsmith import traceable
import uuid
import time

# Example 2: LangSmith tracing and run management  
print("Example 2: LangSmith tracing and run management")

try:
    # Define a traceable function for demonstration
    @traceable
    def process_text(text: str, operation: str = "analyze") -> dict:
        """Example traceable function for text processing"""
        
        print(f"Processing text with operation: {operation}")
        
        # Simulate text processing
        result = {
            'input_text': text,
            'operation': operation,
            'text_length': len(text),
            'word_count': len(text.split()),
            'processed_at': time.time()
        }
        
        # Simulate processing time
        time.sleep(0.1)
        
        return result
    
    # Test the traceable function
    test_text = "This is a sample text for LangSmith tracing demonstration."
    print(f"\nTesting traceable function with text: '{test_text[:50]}...'")
    
    # Call the traceable function
    result = process_text(test_text, "sentiment_analysis")
    
    print("Function result:")
    for key, value in result.items():
        print(f"  {key}: {value}")
    
    # Test multiple operations
    operations = ["tokenize", "summarize", "translate"]
    print(f"\nTesting multiple operations: {operations}")
    
    for op in operations:
        result = process_text(f"Sample text for {op}", op)
        print(f"  {op}: {result['word_count']} words processed")
    
    # Demonstrate run context creation
    print("\nTesting run context management...")
    
    run_data = {
        'run_id': str(uuid.uuid4()),
        'name': 'text_processing_run',
        'start_time': time.time(),
        'inputs': {'text': test_text, 'operation': 'analyze'},
        'run_type': 'chain'
    }
    
    print("Mock run context:")
    for key, value in run_data.items():
        if key == 'run_id':
            print(f"  {key}: {str(value)[:8]}...")  # Truncate UUID for display
        else:
            print(f"  {key}: {value}")

except Exception as e:
    print(f"LangSmith tracing error: {e}")
    
    # Test basic traceable decorator availability
    try:
        print("Testing traceable decorator availability...")
        print(f"Traceable decorator type: {type(traceable).__name__}")
        
        # Test without decorator
        def simple_function(x):
            return x * 2
        
        result = simple_function(5)
        print(f"Simple function result: {result}")
        
    except Exception as fallback_error:
        print(f"Traceable testing error: {fallback_error}")
        raise
    raise

print("Example 2 completed")
```

## Example 3: langsmith functionality test 3

```python
import langsmith
from langsmith import evaluate
import random
import time

# Example 3: LangSmith evaluation and testing framework
print("Example 3: LangSmith evaluation and testing framework")

try:
    # Define sample dataset for evaluation
    evaluation_dataset = [
        {
            'input': 'What is the capital of France?',
            'expected_output': 'Paris',
            'context': 'geography'
        },
        {
            'input': 'Calculate 15 + 27',
            'expected_output': '42',  
            'context': 'mathematics'
        },
        {
            'input': 'What color is the sky?',
            'expected_output': 'blue',
            'context': 'general knowledge'
        },
        {
            'input': 'Name a programming language',
            'expected_output': 'Python',
            'context': 'technology'
        }
    ]
    
    print(f"Created evaluation dataset with {len(evaluation_dataset)} samples")
    
    # Define a model function to evaluate
    def simple_qa_model(question: str) -> str:
        """Simple Q&A model for evaluation testing"""
        
        # Simulate model processing
        time.sleep(0.05)  # Brief delay to simulate processing
        
        # Mock responses based on keywords
        question_lower = question.lower()
        
        if 'capital' in question_lower and 'france' in question_lower:
            return 'Paris'
        elif 'calculate' in question_lower or '+' in question_lower:
            # Extract numbers and calculate (simplified)
            if '15' in question and '27' in question:
                return '42'
            else:
                return 'Unable to calculate'
        elif 'sky' in question_lower and 'color' in question_lower:
            return 'blue'
        elif 'programming language' in question_lower:
            return random.choice(['Python', 'JavaScript', 'Java', 'C++'])
        else:
            return 'I don\'t know'
    
    # Define evaluation metrics
    def accuracy_evaluator(prediction: str, expected: str) -> dict:
        """Evaluate prediction accuracy"""
        correct = prediction.lower().strip() == expected.lower().strip()
        return {
            'accuracy': 1.0 if correct else 0.0,
            'prediction': prediction,
            'expected': expected,
            'match': correct
        }
    
    # Run evaluation on dataset
    print("\nRunning evaluation on dataset:")
    
    results = []
    for i, sample in enumerate(evaluation_dataset):
        input_text = sample['input']
        expected = sample['expected_output']
        context = sample['context']
        
        # Get model prediction
        prediction = simple_qa_model(input_text)
        
        # Evaluate prediction
        eval_result = accuracy_evaluator(prediction, expected)
        
        # Store results
        result = {
            'sample_id': i,
            'input': input_text,
            'context': context,
            'prediction': prediction,
            'expected': expected,
            'accuracy': eval_result['accuracy'],
            'match': eval_result['match']
        }
        results.append(result)
        
        print(f"  Sample {i+1}: {context}")
        print(f"    Q: {input_text}")
        print(f"    Predicted: {prediction}")
        print(f"    Expected: {expected}")
        print(f"    Accuracy: {eval_result['accuracy']}")
        print()
    
    # Calculate overall metrics
    total_samples = len(results)
    correct_predictions = sum(1 for r in results if r['match'])
    overall_accuracy = correct_predictions / total_samples
    
    print(f"Overall evaluation results:")
    print(f"  Total samples: {total_samples}")
    print(f"  Correct predictions: {correct_predictions}")
    print(f"  Overall accuracy: {overall_accuracy:.2%}")

except Exception as e:
    print(f"LangSmith evaluation error: {e}")
    
    # Test basic evaluation functionality
    try:
        print("Testing evaluation module availability...")
        
        # Check if evaluate function is available
        if hasattr(langsmith, 'evaluate'):
            print(f"Evaluate function available: {type(langsmith.evaluate).__name__}")
        else:
            print("Evaluate function not directly available")
        
        # Show basic evaluation concepts
        print("Basic evaluation workflow demonstrated:")
        print("  1. Define dataset with input/expected pairs")
        print("  2. Create model function to test")
        print("  3. Define evaluation metrics")  
        print("  4. Run evaluation and calculate results")
        
    except Exception as fallback_error:
        print(f"Evaluation testing error: {fallback_error}")
        raise
    raise

print("Example 3 completed")
```

## Example 4: langsmith functionality test 4

```python
import langsmith
from langsmith.schemas import Run, RunTypeEnum
import json
import uuid
from datetime import datetime

# Example 4: LangSmith run schemas and data models
print("Example 4: LangSmith run schemas and data models")

try:
    # Create sample run data using LangSmith schemas
    print("Creating LangSmith run objects...")
    
    # Generate sample run data
    run_id = str(uuid.uuid4())
    start_time = datetime.now()
    
    # Create run data structure
    run_data = {
        'id': run_id,
        'name': 'text_classification_chain',
        'run_type': 'chain',
        'start_time': start_time.isoformat(),
        'inputs': {
            'text': 'This is a sample text for classification',
            'model': 'text-classifier-v1'
        },
        'outputs': {
            'classification': 'positive',
            'confidence': 0.85,
            'processing_time': 1.2
        },
        'metadata': {
            'version': '1.0',
            'environment': 'testing',
            'user_id': 'test-user'
        }
    }
    
    print(f"Run created with ID: {run_id[:8]}...")
    print(f"Run type: {run_data['run_type']}")
    print(f"Run name: {run_data['name']}")
    
    # Display run inputs and outputs
    print("\nRun inputs:")
    for key, value in run_data['inputs'].items():
        print(f"  {key}: {value}")
    
    print("\nRun outputs:")
    for key, value in run_data['outputs'].items():
        print(f"  {key}: {value}")
    
    # Create nested runs (parent-child relationship)
    print("\nCreating nested run structure...")
    
    parent_run_id = str(uuid.uuid4())
    child_runs = []
    
    for i in range(3):
        child_run = {
            'id': str(uuid.uuid4()),
            'parent_run_id': parent_run_id,
            'name': f'preprocessing_step_{i+1}',
            'run_type': 'tool',
            'start_time': start_time.isoformat(),
            'inputs': {'data': f'input_chunk_{i+1}'},
            'outputs': {'processed_data': f'output_chunk_{i+1}'},
            'status': 'completed'
        }
        child_runs.append(child_run)
    
    parent_run = {
        'id': parent_run_id,
        'name': 'document_processing_pipeline',
        'run_type': 'chain',
        'child_runs': child_runs,
        'start_time': start_time.isoformat(),
        'status': 'completed'
    }
    
    print(f"Parent run: {parent_run['name']}")
    print(f"Child runs: {len(child_runs)}")
    
    for i, child in enumerate(child_runs):
        print(f"  Child {i+1}: {child['name']} (status: {child['status']})")
    
    # Test run serialization
    print("\nTesting run data serialization...")
    
    try:
        json_data = json.dumps(run_data, indent=2, default=str)
        print("Run data successfully serialized to JSON")
        print(f"JSON size: {len(json_data)} characters")
        
        # Test deserialization
        parsed_data = json.loads(json_data)
        print("Run data successfully deserialized from JSON")
        print(f"Parsed run ID: {parsed_data['id'][:8]}...")
        
    except Exception as json_error:
        print(f"JSON serialization error: {json_error}")
        raise

except Exception as e:
    print(f"LangSmith schemas error: {e}")
    
    # Fallback to testing basic schema availability
    try:
        print("Testing LangSmith schema availability...")
        
        # Check for schema modules
        schema_attrs = []
        if hasattr(langsmith, 'schemas'):
            schema_attrs = [attr for attr in dir(langsmith.schemas) if not attr.startswith('_')]
            print(f"Schema attributes found: {len(schema_attrs)}")
            if schema_attrs:
                print(f"Sample schema attributes: {schema_attrs[:8]}")
        else:
            print("Schemas module not directly available")
        
        # Show run data structure concepts
        print("\nRun data structure concepts:")
        print("  - Unique run ID for tracking")
        print("  - Run type (chain, tool, llm, etc.)")
        print("  - Input/output data capture")
        print("  - Nested run relationships")
        print("  - Metadata for context")
        
    except Exception as fallback_error:
        print(f"Schema testing error: {fallback_error}")
        raise
    raise

print("Example 4 completed")
```

```python
import langsmith
import sys
import json

# Example 5: LangSmith package comprehensive analysis
print("Example 5: LangSmith package comprehensive analysis")

def analyze_langsmith_package():
    """Comprehensive analysis of LangSmith package"""
    
    try:
        print("LangSmith package analysis:")
        print(f"  Package name: {langsmith.__name__}")
        print(f"  Package type: {type(langsmith).__name__}")
        
        # Version information
        if hasattr(langsmith, '__version__'):
            print(f"  Version: {langsmith.__version__}")
        else:
            print("  Version: Not specified")
        
        # Package location
        if hasattr(langsmith, '__file__'):
            print(f"  Location: {langsmith.__file__}")
        
        # Explore main attributes
        attrs = dir(langsmith)
        public_attrs = [attr for attr in attrs if not attr.startswith('_')]
        private_attrs = [attr for attr in attrs if attr.startswith('_')]
        
        print(f"\nPackage structure:")
        print(f"  Total attributes: {len(attrs)}")
        print(f"  Public attributes: {len(public_attrs)}")  
        print(f"  Private attributes: {len(private_attrs)}")
        
        if public_attrs:
            print(f"  Key public attributes: {public_attrs[:10]}")
            
        # Categorize attributes
        classes = []
        functions = []
        modules = []
        
        for attr_name in public_attrs[:15]:  # Limit for performance
            try:
                attr = getattr(langsmith, attr_name)
                if isinstance(attr, type):
                    classes.append(attr_name)
                elif callable(attr):
                    functions.append(attr_name)
                elif hasattr(attr, '__file__'):
                    modules.append(attr_name)
            except:
                pass
        
        print(f"\nAttribute categorization:")
        print(f"  Classes: {classes[:5]}")
        print(f"  Functions: {functions[:5]}")
        print(f"  Modules: {modules[:5]}")
        
    except Exception as e:
        print(f"Package analysis error: {e}")

def test_langsmith_components():
    """Test key LangSmith components"""
    
    print("\nTesting LangSmith component availability:")
    
    components = [
        ('Client', 'Main client for API interactions'),
        ('traceable', 'Decorator for function tracing'),
        ('evaluate', 'Evaluation framework'),
        ('schemas', 'Data schemas and models'),
        ('utils', 'Utility functions'),
    ]
    
    available_components = []
    
    for component, description in components:
        try:
            if hasattr(langsmith, component):
                attr = getattr(langsmith, component)
                available_components.append((component, description))
                print(f"  ✓ {component}: {description}")
                print(f"    Type: {type(attr).__name__}")
            else:
                try:
                    exec(f"from langsmith import {component}")
                    available_components.append((component, description))
                    print(f"  ✓ {component}: {description} (importable)")
                except ImportError:
                    print(f"  ✗ {component}: Not available")
        except Exception as e:
            print(f"  ? {component}: Error ({e})")
    
    print(f"\nComponent summary: {len(available_components)}/{len(components)} available")

def demonstrate_langsmith_concepts():
    """Demonstrate key LangSmith concepts"""
    
    print("\nKey LangSmith concepts:")
    
    concepts = {
        'Tracing': 'Automatic capture of function calls, inputs, and outputs',
        'Evaluation': 'Framework for testing and benchmarking LLM applications',
        'Datasets': 'Managed datasets for evaluation and testing',
        'Projects': 'Organization unit for related runs and experiments',
        'Runs': 'Individual execution traces with metadata',
        'Feedback': 'Human annotation and scoring of outputs'
    }
    
    for concept, description in concepts.items():
        print(f"  {concept}: {description}")
    
    # Example workflow
    print("\nTypical LangSmith workflow:")
    workflow_steps = [
        "1. Initialize LangSmith client with API key",
        "2. Create or select a project",
        "3. Add @traceable decorator to functions",
        "4. Run your application (traces captured automatically)",
        "5. View traces and runs in LangSmith UI",
        "6. Create evaluation datasets",
        "7. Run evaluations on your models",
        "8. Analyze results and iterate"
    ]
    
    for step in workflow_steps:
        print(f"  {step}")

# Run analysis functions
analyze_langsmith_package()
test_langsmith_components()
demonstrate_langsmith_concepts()

print(f"\nExample 5 completed - LangSmith package analysis")
print("Note: LangSmith is a platform for debugging, testing, and monitoring LLM applications")
print("Full functionality requires API key and active LangSmith account")
```
