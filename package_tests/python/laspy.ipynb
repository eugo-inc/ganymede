## Example 1: Creating and reading basic LAS point cloud data

```python
import laspy
import numpy as np

# Example 1: Creating and reading basic LAS point cloud data
# Create synthetic LiDAR point cloud data
points_count = 1000
x = np.random.uniform(0, 100, points_count)
y = np.random.uniform(0, 100, points_count)
z = np.random.uniform(0, 50, points_count)
intensity = np.random.randint(0, 65535, points_count, dtype=np.uint16)

# Create LAS file with point format 1 and supported version
las = laspy.create(point_format=1, file_version="1.2")
las.x = x
las.y = y
las.z = z
las.intensity = intensity

print(f"Created LAS file with {len(las.points)} points")
print(f"Point format: {las.header.point_format}")
print(f"File version: {las.header.version}")
print(f"X range: {las.header.x_min:.2f} to {las.header.x_max:.2f}")
print(f"Y range: {las.header.y_min:.2f} to {las.header.y_max:.2f}")
print(f"Z range: {las.header.z_min:.2f} to {las.header.z_max:.2f}")

# Access point data
print(f"First 5 points (X, Y, Z):")
for i in range(5):
    print(f"  Point {i}: ({las.x[i]:.2f}, {las.y[i]:.2f}, {las.z[i]:.2f})")
```

## Example 2: Writing and reading LAS files with RGB data

```python
import laspy
import numpy as np
import tempfile

# Example 2: Writing and reading LAS files with RGB data
# Create RGB point cloud data (point format 3)
points_count = 500
x = np.random.uniform(0, 50, points_count)
y = np.random.uniform(0, 50, points_count) 
z = np.random.uniform(0, 25, points_count)

# RGB values (0-65535 range for LAS format)
red = np.random.randint(0, 65535, points_count, dtype=np.uint16)
green = np.random.randint(0, 65535, points_count, dtype=np.uint16)
blue = np.random.randint(0, 65535, points_count, dtype=np.uint16)

# Create LAS file with RGB support (format 3)
las_out = laspy.create(point_format=3, file_version="1.2")
las_out.x = x
las_out.y = y
las_out.z = z
las_out.red = red
las_out.green = green
las_out.blue = blue

# Write to temporary file
with tempfile.NamedTemporaryFile(suffix='.las', delete=False) as tmp:
    las_out.write(tmp.name)
    temp_filename = tmp.name

# Read back the file
las_in = laspy.read(temp_filename)
print(f"Successfully wrote and read LAS file: {temp_filename}")
print(f"Points in file: {len(las_in.points)}")
print(f"RGB values for first 3 points:")
for i in range(3):
    print(f"  Point {i}: R={las_in.red[i]}, G={las_in.green[i]}, B={las_in.blue[i]}")
```

## Example 3: Point cloud filtering and classification

```python
import laspy
import numpy as np

# Example 3: Point cloud filtering and classification
# Create classified point cloud data
points_count = 800
x = np.random.uniform(0, 100, points_count)
y = np.random.uniform(0, 100, points_count)
z = np.random.uniform(0, 30, points_count)

# Assign classifications: 1=Unclassified, 2=Ground, 3=Vegetation, 6=Building
classifications = np.random.choice([1, 2, 3, 6], points_count)

# Create point format 0 with classification (use 1.4 for classification support)
las = laspy.create(point_format=0, file_version="1.4")
las.x = x
las.y = y
las.z = z
las.classification = classifications

print(f"Total points: {len(las.points)}")

# Filter points by classification
ground_points = las.points[las.classification == 2]
vegetation_points = las.points[las.classification == 3]
building_points = las.points[las.classification == 6]

print(f"Ground points: {len(ground_points)}")
print(f"Vegetation points: {len(vegetation_points)}")
print(f"Building points: {len(building_points)}")

# Filter by height (Z coordinate)
high_points = las.points[las.z > 20]
print(f"Points above 20m: {len(high_points)}")

# Classification statistics
unique_classes, counts = np.unique(las.classification, return_counts=True)
print("Classification distribution:")
for cls, count in zip(unique_classes, counts):
    print(f"  Class {cls}: {count} points")
```

## Example 4: Point data manipulation and file operations

```python
import laspy
import numpy as np
import tempfile
import os

# Example 4: Point data manipulation and file operations
# Create point cloud data for demonstration
points_count = 1200
x = np.random.uniform(0, 200, points_count)
y = np.random.uniform(0, 200, points_count)
z = np.random.uniform(0, 100, points_count)
intensity = np.random.randint(0, 65535, points_count, dtype=np.uint16)

# Create LAS file
las = laspy.create(point_format=1, file_version="1.2")
las.x = x
las.y = y  
las.z = z
las.intensity = intensity

# Demonstrate point data access and modification
print(f"Original point count: {len(las.points)}")
print(f"Intensity statistics: min={las.intensity.min()}, max={las.intensity.max()}")

# Modify intensity values for demonstration
las.intensity = las.intensity // 2  # Reduce intensity by half
print(f"Modified intensity: min={las.intensity.min()}, max={las.intensity.max()}")

# Write to file and check size
with tempfile.NamedTemporaryFile(suffix='.las', delete=False) as tmp_las:
    las.write(tmp_las.name)
    file_size = os.path.getsize(tmp_las.name)
    print(f"LAS file size: {file_size:,} bytes")
    
    # Read file back and verify
    las_read = laspy.read(tmp_las.name)
    print(f"Verified: Read {len(las_read.points)} points from file")
    print(f"Data integrity check: intensities match = {np.array_equal(las.intensity, las_read.intensity)}")
```

## Example 5: Header information and metadata management

```python
import laspy
import numpy as np
from datetime import datetime

# Example 5: Header information and metadata management
# Create LAS file with comprehensive header
las = laspy.create(point_format=2, file_version="1.3")

# Generate point data
points_count = 600
las.x = np.random.uniform(500000, 501000, points_count)  # UTM coordinates
las.y = np.random.uniform(4000000, 4001000, points_count)
las.z = np.random.uniform(100, 200, points_count)
las.intensity = np.random.randint(0, 65535, points_count, dtype=np.uint16)
las.red = np.random.randint(0, 65535, points_count, dtype=np.uint16)
las.green = np.random.randint(0, 65535, points_count, dtype=np.uint16)
las.blue = np.random.randint(0, 65535, points_count, dtype=np.uint16)

# Set header metadata
las.header.system_identifier = "laspy Example System"
las.header.generating_software = "Python laspy Demo"
las.header.file_creation_day = datetime.now().timetuple().tm_yday
las.header.file_creation_year = datetime.now().year

print("LAS Header Information:")
print(f"  File version: {las.header.version}")
print(f"  Point format: {las.header.point_format}")
print(f"  Point count: {las.header.point_count}")
print(f"  System ID: {las.header.system_identifier}")
print(f"  Software: {las.header.generating_software}")
print(f"  Creation: Day {las.header.file_creation_day}, {las.header.file_creation_year}")
print(f"  Coordinate bounds:")
print(f"    X: {las.header.x_min:.2f} to {las.header.x_max:.2f}")
print(f"    Y: {las.header.y_min:.2f} to {las.header.y_max:.2f}")
print(f"    Z: {las.header.z_min:.2f} to {las.header.z_max:.2f}")
print(f"  Scale factors: {las.header.x_scale}, {las.header.y_scale}, {las.header.z_scale}")
print(f"  Offsets: {las.header.x_offset}, {las.header.y_offset}, {las.header.z_offset}")
```
