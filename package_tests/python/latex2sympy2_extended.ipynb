# latex2sympy2_extended

## Example 1: fractions, roots and powers become exact SymPy numbers

```python
# Example 1: the parser (the 4.13.2-generated grammar) keeps the expression as written; simplify() evaluates it exactly
import importlib.metadata
import sympy
from latex2sympy2_extended import latex2sympy

assert importlib.metadata.version("antlr4-python3-runtime") == "4.13.2"
total = latex2sympy(r"\frac{1}{2} + \frac{1}{3}")
assert sympy.simplify(total) == sympy.Rational(5, 6), total
assert sympy.simplify(latex2sympy(r"\sqrt{8}") - 2 * sympy.sqrt(2)) == 0
assert sympy.simplify(latex2sympy(r"2^{10}")) == 1024
print("Example 1:", total, "=", sympy.simplify(total), "|", latex2sympy(r"\sqrt{8}"), "=", sympy.simplify(latex2sympy(r"\sqrt{8}")))
```

## Example 2: derivatives and integrals

```python
# Example 2: \frac{d}{dx} and \int parse to unevaluated SymPy objects that .doit() evaluates
import sympy
from latex2sympy2_extended import latex2sympy

x = sympy.Symbol("x")
derivative = latex2sympy(r"\frac{d}{dx} x^{3}")
integral = latex2sympy(r"\int_{0}^{1} x^{2} dx")
assert sympy.simplify(derivative.doit() - 3 * x**2) == 0
assert integral.doit() == sympy.Rational(1, 3)
print("Example 2:", derivative, "=", derivative.doit(), "|", integral, "=", integral.doit())
```

## Example 3: matrices

```python
# Example 3: pmatrix parses to a SymPy Matrix with a determinant and an inverse
import sympy
from latex2sympy2_extended import latex2sympy

m = latex2sympy(r"\begin{pmatrix} 1 & 2 \\ 3 & 4 \end{pmatrix}")
assert isinstance(m, sympy.MatrixBase) and m.shape == (2, 2)
assert m.det() == -2
print("Example 3:", m.tolist(), "det", m.det())
```

## Example 4: sets and intervals

```python
# Example 4: interval notation and set braces parse to SymPy sets
import sympy
from latex2sympy2_extended import latex2sympy

interval = latex2sympy(r"[1, 5)")
finite = latex2sympy(r"\{1, 2, 3\}")
assert interval == sympy.Interval.Ropen(1, 5), interval
assert set(finite) == {1, 2, 3} and len(finite) == 3, finite  # its own FiniteSet subclass, compared by elements
assert interval.contains(1) and not interval.contains(5)
print("Example 4:", interval, finite, type(finite).__module__)
```

## Example 5: normalise a model-style answer string, then parse it

```python
# Example 5: normalize_latex strips \boxed{}, dollar signs and units text so the expression underneath parses
import sympy
from latex2sympy2_extended import NormalizationConfig, latex2sympy, normalize_latex

config = NormalizationConfig(basic_latex=True, units=True, malformed_operators=True, nits=True, boxed="all", equations=False)
text = normalize_latex(r"\boxed{\frac{3}{4}}", config)
assert latex2sympy(text) == sympy.Rational(3, 4), text
print("Example 5:", repr(text), "->", latex2sympy(text))
```
