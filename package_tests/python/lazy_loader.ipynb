## Example 1: Basic lazy module loading

```python
import lazy_loader as lz
import sys

# Example 1: Basic lazy module loading
# Create lazy module proxy for numpy - only loads when first accessed
np = lz.load("numpy")

print("Lazy loader created for numpy")
print(f"Type of np: {type(np)}")
print(f"Is numpy already imported? {'numpy' in sys.modules}")

# Access numpy functionality - this triggers actual import
array = np.array([1, 2, 3, 4, 5])
print(f"Created array: {array}")
print(f"Array dtype: {array.dtype}")

# Now numpy is imported
print(f"Is numpy imported now? {'numpy' in sys.modules}")

# Lazy loader behaves like the real module
print(f"Numpy version: {np.__version__}")
print(f"Array sum: {np.sum(array)}")

# Show that it's transparent - works like normal import
zeros = np.zeros(3)
print(f"Zero array: {zeros}")

print("Lazy loading demonstration completed")
```

## Example 2: Lazy submodule importing with submodules

```python
import lazy_loader as lz

# Example 2: Lazy submodule importing with submodules
# Create lazy loaders for specific submodules
linalg = lz.load("numpy.linalg")
random = lz.load("numpy.random")
fft = lz.load("numpy.fft")

print("Created lazy loaders for numpy submodules:")
print(f"  - linalg: {type(linalg)}")
print(f"  - random: {type(random)}")
print(f"  - fft: {type(fft)}")

# Use submodule functionality
matrix = [[1, 2], [3, 4]]
det = linalg.det(matrix)
print(f"Matrix determinant: {det}")

# Generate random data
rand_nums = random.random(5)
print(f"Random numbers: {rand_nums}")

# FFT operations
signal = [1, 2, 3, 4]
fft_result = fft.fft(signal)
print(f"FFT result: {fft_result}")

print("Submodule lazy loading successful")

# Show that only accessed submodules are imported
import sys
numpy_modules = [mod for mod in sys.modules.keys() if 'numpy' in mod]
print(f"Imported numpy modules: {sorted(numpy_modules)[:5]}...")
```

## Example 3: Performance comparison - lazy vs regular imports

```python
import lazy_loader as lz
import time
import sys

# Example 3: Performance comparison - lazy vs regular imports
print("Performance comparison: Lazy vs Regular imports")

# Measure lazy loading setup time for multiple modules
start_time = time.time()
requests_lazy = lz.load("requests")
urllib_lazy = lz.load("urllib.parse")
json_lazy = lz.load("json")
lazy_setup_time = time.time() - start_time

print(f"Lazy loader setup time for 3 modules: {lazy_setup_time:.6f} seconds")
print(f"Requests imported yet? {'requests' in sys.modules}")
print(f"Urllib imported yet? {'urllib.parse' in sys.modules}")

# Measure first access time (actual import)
start_time = time.time()
data = {"name": "test", "value": 123}
json_str = json_lazy.dumps(data)
parsed_url = urllib_lazy.urlparse("https://example.com/path")
first_access_time = time.time() - start_time

print(f"First access time: {first_access_time:.6f} seconds")
print(f"JSON imported now? {'json' in sys.modules}")
print(f"Parsed URL: {parsed_url.netloc}")

# Measure subsequent access time
start_time = time.time()
data2 = json_lazy.loads(json_str)
encoded = urllib_lazy.quote("hello world")
second_access_time = time.time() - start_time

print(f"Second access time: {second_access_time:.6f} seconds")
print(f"Decoded data: {data2}")
print(f"URL encoded: {encoded}")

print(f"\nBenefit: Lazy loading delays import cost until needed")
print(f"Setup: {lazy_setup_time:.6f}s, First use: {first_access_time:.6f}s")
```

## Example 4: Lazy loading with multiple module types

```python
import lazy_loader as lz
import sys

# Example 4: Lazy loading with multiple module types
# Create lazy loaders for different types of modules
pathlib_lazy = lz.load("pathlib")
collections_lazy = lz.load("collections")
datetime_lazy = lz.load("datetime")

print("Created lazy loaders for standard library modules:")
print(f"  - pathlib: {type(pathlib_lazy)}")
print(f"  - collections: {type(collections_lazy)}")
print(f"  - datetime: {type(datetime_lazy)}")

# Check import status before use
print(f"Pathlib imported? {'pathlib' in sys.modules}")
print(f"Collections imported? {'collections' in sys.modules}")

# Use the lazy loaded modules
path = pathlib_lazy.Path("/tmp/example.txt")
counter = collections_lazy.Counter([1, 2, 2, 3, 3, 3])
now = datetime_lazy.datetime.now()

print(f"Path object: {path}")
print(f"Counter result: {counter}")
print(f"Current time: {now.strftime('%Y-%m-%d %H:%M:%S')}")

# Check import status after use
print(f"Pathlib imported now? {'pathlib' in sys.modules}")
print(f"Collections imported now? {'collections' in sys.modules}")

# Show that lazy loader preserves module attributes
print(f"Available pathlib classes: {len([x for x in dir(pathlib_lazy) if not x.startswith('_')])}")
```

## Example 5: Error handling and requirement checking

```python
import lazy_loader as lz
import sys

# Example 5: Error handling and requirement checking
print("Demonstrating lazy_loader error handling and requirements")

# Test lazy loading on a standard module (hashlib is always available, so an ImportError fails this cell)
hashlib_lazy = lz.load("hashlib")
print(f"✅ Created lazy loader for hashlib: {type(hashlib_lazy)}")

# Use hashlib functionality: md5(b"test data") is a fixed digest
hash_obj = hashlib_lazy.md5(b"test data")
hash_result = hash_obj.hexdigest()
assert hash_result == "eb733a00c0c9d336e65691a37ab54293", hash_result
print(f"Hash result: {hash_result[:16]}...")

# Demonstrate error_on_import parameter with non-existent module
try:
    fake_module = lz.load("nonexistent_module_12345", error_on_import=True)
except ModuleNotFoundError as e:
    assert "nonexistent_module_12345" in str(e), e
    print(f"✅ Caught import error immediately: {type(e).__name__}")
else:
    raise AssertionError(f"lazy_loader.load(error_on_import=True) returned {fake_module!r} for a module that does not exist")

# Test with a module that exists but might have issues
modules_before = set(sys.modules.keys())
base64_lazy = lz.load("base64")
print(f"Base64 lazy loader created: {type(base64_lazy)}")

# Use the lazy loaded module
test_data = b"Hello, lazy loading!"
encoded = base64_lazy.b64encode(test_data)
decoded = base64_lazy.b64decode(encoded)
assert decoded == test_data, decoded
print(f"Encoding test successful: {decoded.decode()}")

modules_after = set(sys.modules.keys())
print(f"New modules loaded: {len(modules_after - modules_before)}")
print("Lazy loading with error handling completed")
```
