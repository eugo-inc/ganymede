## Example 1: legendgram functionality test 1

```python
import legendgram
import numpy as np
import matplotlib.pyplot as plt
import palettable

# Example 1: Basic legendgram histogram legend
# Create sample data for a choropleth-style visualization
np.random.seed(42)
data = np.random.normal(50, 15, 100)

# Create breaks within the data range to avoid indexing errors
data_min, data_max = data.min(), data.max()
breaks = [data_min, 25, 40, 55, 70, data_max]

# Use a color palette where pal.number matches len(breaks)
pal = palettable.colorbrewer.sequential.Blues_6

# Create figure and axis
fig, ax = plt.subplots(figsize=(8, 6))

# Create a simple scatter plot as our "map"
x = np.random.uniform(0, 10, len(data))
y = np.random.uniform(0, 10, len(data))
scatter = ax.scatter(x, y, c=data, cmap=pal.mpl_colormap, s=50)

# Add the legendgram - this is the unique functionality
legend_ax = legendgram.legendgram(fig, ax, data, breaks, pal, bins=30)

ax.set_title("Basic Legendgram Example")
plt.tight_layout()
plt.show()

print(f"Legendgram created with {len(breaks)} breaks")
print(f"Data range: {data.min():.1f} to {data.max():.1f}")
print(f"Breaks: {[f'{b:.1f}' for b in breaks]}")
```

## Example 2: legendgram functionality test 2

```python
import legendgram
import numpy as np
import matplotlib.pyplot as plt
import palettable

# Example 2: Customized legendgram with different color schemes
# Create bimodal data distribution
np.random.seed(123)
data1 = np.random.normal(25, 8, 50)
data2 = np.random.normal(75, 12, 50)
data = np.concatenate([data1, data2])

# Define breaks within data range (5 breaks for RdYlBu_5)
data_min, data_max = data.min(), data.max()
breaks = [data_min, 25, 45, 65, data_max]

# Use a diverging color palette where pal.number = len(breaks)
pal = palettable.colorbrewer.diverging.RdYlBu_5

# Create visualization
fig, ax = plt.subplots(figsize=(10, 6))

# Simulate a geographic scatter plot
x = np.random.uniform(-10, 10, len(data))
y = np.random.uniform(-5, 5, len(data))
scatter = ax.scatter(x, y, c=data, cmap=pal.mpl_colormap, s=80, alpha=0.7)

# Add customized legendgram with specific location and size
legend_ax = legendgram.legendgram(
    fig, ax, data, breaks, pal,
    bins=40,
    loc='upper right',
    legend_size=(0.3, 0.25),
    frameon=True
)

ax.set_title("Customized Legendgram with Diverging Colors")
ax.set_xlabel("Longitude")
ax.set_ylabel("Latitude")
plt.show()

print(f"Bimodal data with {len(breaks)} breaks")
print(f"Peak 1: ~{data1.mean():.1f}, Peak 2: ~{data2.mean():.1f}")
print(f"Breaks: {[f'{b:.1f}' for b in breaks]}")
```

## Example 3: legendgram functionality test 3

```python
import legendgram
import numpy as np
import matplotlib.pyplot as plt
import palettable

# Example 3: Legendgram with clipped data range
# Create skewed data with some extreme values
np.random.seed(456)
data = np.random.exponential(20, 80)
# Add some extreme outliers
outliers = np.random.uniform(150, 200, 10)
data = np.concatenate([data, outliers])

# Define breaks within reasonable range (4 breaks for Oranges_4)
data_min = data.min()
breaks = [data_min, 20, 40, 75]

# Use a sequential color palette where pal.number = len(breaks)
pal = palettable.colorbrewer.sequential.Oranges_4

# Create visualization
fig, ax = plt.subplots(figsize=(10, 6))

# Create heatmap-style visualization
x = np.random.uniform(0, 15, len(data))
y = np.random.uniform(0, 10, len(data))
scatter = ax.scatter(x, y, c=data, cmap=pal.mpl_colormap, s=60, alpha=0.8)

# Add legendgram with clipped range to focus on main distribution
legend_ax = legendgram.legendgram(
    fig, ax, data, breaks, pal,
    bins=35,
    clip=(0, 120),  # Clip histogram to show main distribution
    loc='lower right',
    legend_size=(0.25, 0.3)
)

ax.set_title("Legendgram with Clipped Range (Outliers Excluded)")
ax.set_xlabel("X coordinate")
ax.set_ylabel("Y coordinate")
plt.show()

print(f"Data range: {data.min():.1f} to {data.max():.1f}")
print(f"Clipped histogram range: 0 to 120")
print(f"Outliers beyond clip: {np.sum(data > 120)} values")
print(f"Breaks: {[f'{b:.1f}' for b in breaks]}")
```

## Example 4: legendgram functionality test 4

```python
import legendgram
import numpy as np
import matplotlib.pyplot as plt
import palettable

# Example 4: Advanced legendgram styling with tick parameters
# Create uniform data with clear categories
np.random.seed(789)
categories = [10, 25, 40, 55, 70, 85]
data = []
for cat in categories:
    data.extend(np.random.normal(cat, 3, 15))
data = np.array(data)

# Define breaks within data range (7 breaks for Set3_7)
data_min, data_max = data.min(), data.max()
breaks = [data_min, 15, 30, 45, 60, 75, data_max]

# Use a qualitative color palette where pal.number = len(breaks)
pal = palettable.colorbrewer.qualitative.Set3_7

# Create visualization
fig, ax = plt.subplots(figsize=(12, 7))

# Create grid-like scatter plot
rows, cols = 6, 15
x_coords = np.tile(np.arange(cols), rows)
y_coords = np.repeat(np.arange(rows), cols)
scatter = ax.scatter(x_coords, y_coords, c=data, cmap=pal.mpl_colormap, s=100)

# Add legendgram with custom tick parameters
tick_params = {
    'labelsize': 8,
    'length': 3,
    'width': 1,
    'direction': 'out'
}

legend_ax = legendgram.legendgram(
    fig, ax, data, breaks, pal,
    bins=25,
    loc='center right',
    legend_size=(0.2, 0.4),
    frameon=True,
    tick_params=tick_params
)

ax.set_title("Advanced Legendgram with Custom Tick Styling")
ax.set_aspect('equal')
plt.tight_layout()
plt.show()

print(f"Categories: {len(categories)}")
print(f"Custom tick parameters applied to histogram axis")
print(f"Breaks: {[f'{b:.1f}' for b in breaks]}")
```

## Example 5: legendgram functionality test 5

```python
import legendgram
import legendgram.util as lg_util
import numpy as np
import matplotlib.pyplot as plt
import palettable

# Example 5: Legendgram utility functions demonstration
# Create sample data
np.random.seed(101)
data = np.random.gamma(2, 10, 120)

# Test utility functions
print("Legendgram utility functions:")
print(f"Available utilities: {[attr for attr in dir(lg_util) if not attr.startswith('_')]}")

# Create main plot first to get axis for utility functions
fig, ax = plt.subplots(figsize=(10, 6))
x = np.random.uniform(0, 20, len(data))
y = np.random.uniform(0, 15, len(data))

# Test location utilities with actual axis
locations = ['upper left', 'center', 'lower right', 'best']
print("\nLocation utility examples:")
for loc in locations:
    try:
        result = lg_util.make_location(ax, loc)
        print(f"  {loc} -> bounds: [{result[0]:.3f}, {result[1]:.3f}, {result[2]:.3f}, {result[3]:.3f}]")
    except Exception as e:
        print(f"  {loc} -> Error: {e}")
        raise

# Create breaks within data range (5 breaks for Purples_5)
data_min, data_max = data.min(), data.max()
breaks = [data_min, 15, 30, 60, data_max]
pal = palettable.colorbrewer.sequential.Purples_5

scatter = ax.scatter(x, y, c=data, cmap=pal.mpl_colormap, s=50)

# Use utility to determine location bounds and pass to legendgram
legend_ax = legendgram.legendgram(fig, ax, data, breaks, pal, loc='upper right')

ax.set_title("Legendgram with Utility Functions")
plt.show()

print(f"\nData summary: {len(data)} points, range {data.min():.1f}-{data.max():.1f}")
print(f"Demonstrated location utility function usage")
print(f"Breaks: {[f'{b:.1f}' for b in breaks]}")
```
