## Example 1: Basic Thunder model compilation

```python
import thunder
import torch
import torch.nn as nn
print("✅ Lightning-Thunder imported successfully")

# Example 1: Basic Thunder model compilation
print("Example 1: Basic Thunder Model Compilation")

# Create a simple PyTorch model
model = nn.Sequential(
    nn.Linear(2048, 4096),
    nn.ReLU(),
    nn.Linear(4096, 64)
)

thunder_model = thunder.compile(model)
print("✅ Model compiled with Thunder successfully")

# Test the compiled model
x = torch.randn(64, 2048)
output = thunder_model(x)
print(f"Input shape: {x.shape}")
print(f"Output shape: {output.shape}")
assert output.shape == (64, 64), output.shape

print("\n🚀 Thunder Compilation Benefits:")
benefits = [
    "Up to 40% speedup for PyTorch models",
    "Source-to-source compilation optimization",
    "Kernel fusion and memory optimization",
    "Drop-in replacement for existing models"
]

for benefit in benefits:
    print(f"  • {benefit}")

print("Example 1: Thunder compilation concepts demonstrated")
```

## Example 3: Thunder Transform and Optimization Concepts

```python
# Example 3: Thunder Transform and Optimization Concepts
print("=" * 70)
print(f"🚀 Example 3: Thunder Transform and Optimization")
print("=" * 70)

import thunder
import torch
import torch.nn as nn
print("✅ Thunder imported for transform concepts")

# Example 3: Transform pipeline concepts
print("Example 3: Transform and Optimization Pipeline Concepts")

# Create a complex model with multiple operations
class CustomModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv = nn.Conv2d(3, 64, 3, padding=1)
        self.norm = nn.BatchNorm2d(64)
        self.linear = nn.Linear(64 * 32 * 32, 10)
        
    def forward(self, x):
        x = torch.relu(self.norm(self.conv(x)))
        x = x.flatten(1)
        return self.linear(x)

model = CustomModel()
print("✅ Complex model created for transform demonstration")
print("Note: Full GPU transforms require CUDA environment")

print("\n🔧 Available Transforms:")
transforms = [
    "nvfuser_executor - NVIDIA GPU kernel fusion",
    "torch_compile - PyTorch 2.0 compilation",
    "quantization - FP4/FP6/FP8 precision",
    "cudagraphs - CUDA graph optimization",
    "triton_kernels - Custom Triton kernels"
]

for transform in transforms:
    print(f"  • {transform}")

print("\n⚙️ Optimization Pipeline:")
pipeline = [
    "1. Graph capture and analysis",
    "2. Operator fusion identification", 
    "3. Memory layout optimization",
    "4. Kernel generation and compilation",
    "5. Runtime execution optimization"
]

for step in pipeline:
    print(f"  {step}")

print("\n🔧 Usage Pattern:")
print("  # from thunder.transforms import nvfuser_executor")
print("  # compiled = thunder.compile(model, transforms=[nvfuser_executor])")

print("Example 3: Transform concepts demonstrated")

```
