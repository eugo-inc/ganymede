This notebook tests the linkify_it_py package, which provides URL detection and linkification capabilities for text processing.

## Example 1: Basic URL Detection and Linkification

```python
# Example 1: Using linkify-it-py (safe approach)
from linkify_it import LinkifyIt

print('LinkifyIt imported successfully')

# Create linkify instance
linkify = LinkifyIt()
print(f'LinkifyIt instance created: {linkify}')

# Test with simple text
test_text = 'Check out https://www.python.org for more info'
print(f'\nTesting with: "{test_text}"')

try:
    # Test if text contains links
    if hasattr(linkify, 'test'):
        has_links = linkify.test(test_text)
        print(f'Contains links: {has_links}')
    
    # Try to find matches
    if hasattr(linkify, 'match'):
        matches = linkify.match(test_text)
        print(f'Matches found: {len(matches) if matches else 0}')
        if matches:
            for match in matches[:3]:  # Show first 3
                print(f'  {match}')
                
except Exception as e:
    print(f'LinkifyIt usage note: {e}')
    raise AssertionError(
        f"linkify_it_py example (cell 2) failed: {type(e).__name__}: {e}"
    ) from e

print('LinkifyIt exploration complete')
```

## Example 3: Advanced URL Matching and Validation

```python
import linkify_it
import re

# Test advanced URL matching
linkify = linkify_it.LinkifyIt()

# Complex text with mixed content
complex_text = '''
Here's some complex text with various links:
- Official site: https://python.org/downloads/
- Repository: github.com/python/cpython
- Email contact: support@python.org
- Documentation: https://docs.python.org/3/library/
- FTP server: ftp://ftp.python.org/pub/python/
- IP address: http://192.168.1.1:8080/admin
- Localhost: http://localhost:3000/app
- With query params: https://search.com?q=python&type=docs#results
'''

print("Advanced URL matching test:")
print(f"Text length: {len(complex_text)} characters")

matches = linkify.match(complex_text)
print(f"\nFound {len(matches)} URLs:")

for i, match in enumerate(matches, 1):
    print(f"\n{i}. {match.url}")
    print(f"   Schema: {match.schema}")
    
    # Check available attributes
    start_pos = getattr(match, 'index', 'N/A')
    end_pos = getattr(match, 'end', getattr(match, 'lastIndex', 'N/A'))
    
    print(f"   Position: {start_pos}")
    if hasattr(match, 'index') and hasattr(match, 'end'):
        print(f"   Text: '{complex_text[match.index:match.end]}'")
    elif hasattr(match, 'index'):
        # Calculate end position based on URL length
        end_pos = match.index + len(match.url)
        print(f"   Text: '{complex_text[match.index:end_pos]}'")
    
    # Validate URL format
    is_valid_url = bool(re.match(r'^https?://', match.url))
    print(f"   Valid HTTP(S): {is_valid_url}")

# Test edge cases
edge_cases = [
    "Parentheses: (https://example.com)",
    "Brackets: [https://example.com]", 
    "Quotes: 'https://example.com' and \"https://example.com\"",
    "Punctuation: https://example.com, https://example.com.",
    "Multiple dots: www...example...com",
    "IP with port: http://192.168.1.1:8080"
]

print("\n" + "="*50)
print("Edge case testing:")
for case in edge_cases:
    print(f"\nCase: {case}")
    matches = linkify.match(case)
    if matches:
        for match in matches:
            print(f"  Found: '{match.url}' (schema: {match.schema})")
    else:
        print("  No matches")
```

## Example 4: Text Processing and Normalization

```python
import linkify_it

# Test text processing with URLs
linkify = linkify_it.LinkifyIt()

def process_text_with_links(text):
    """Process text and extract/normalize URLs"""
    print(f"Processing: {text[:100]}...")
    
    matches = linkify.match(text)
    
    if not matches:
        print("  No URLs found")
        return text, []
    
    # Extract URL information
    urls = []
    for match in matches:
        # Calculate end position since lastIndex doesn't exist
        start = match.index
        end = start + len(match.url)
        
        url_info = {
            'original': text[start:end],
            'normalized': match.url,
            'schema': match.schema,
            'position': (start, end)
        }
        urls.append(url_info)
    
    print(f"  Found {len(urls)} URLs:")
    for url in urls:
        print(f"    {url['original']} -> {url['normalized']} ({url['schema']})")
    
    return text, urls

# Test with various text samples
test_samples = [
    "Simple: https://example.com and www.test.org",
    "Email heavy: Contact us at info@company.com, support@help.org, or admin@system.net",
    "Mixed protocols: https://secure.com, http://legacy.com, ftp://files.com",
    "Social media: Follow @user on twitter.com/user or facebook.com/page",
    "Development: Check github.com/repo, stackoverflow.com/questions/123, and docs.python.org"
]

print("Text processing tests:")
processed_results = []

for i, sample in enumerate(test_samples, 1):
    print(f"\n--- Sample {i} ---")
    text, urls = process_text_with_links(sample)
    processed_results.append({'text': text, 'urls': urls})

# Summary statistics
print("\n" + "="*50)
print("PROCESSING SUMMARY")
print("="*50)

total_urls = sum(len(result['urls']) for result in processed_results)
schema_counts = {}

for result in processed_results:
    for url in result['urls']:
        schema = url['schema']
        schema_counts[schema] = schema_counts.get(schema, 0) + 1

print(f"Total URLs processed: {total_urls}")
print(f"Schema distribution:")
for schema, count in sorted(schema_counts.items()):
    print(f"  {schema}: {count}")
```

## Example 5: Performance and Batch Processing

```python
import linkify_it
import time

# Performance testing with batch processing
linkify = linkify_it.LinkifyIt()

# Generate test data
def generate_test_texts(count=100):
    """Generate test texts with various URL patterns"""
    patterns = [
        "Visit https://example{}.com for information",
        "Check out www.site{}.org and email user{}@domain.com", 
        "Multiple links: http://first{}.com, https://second{}.net",
        "Documentation at docs.python{}.org/library/",
        "Repository: github.com/user{}/project and stackoverflow.com/q/{}"
    ]
    
    texts = []
    for i in range(count):
        pattern = patterns[i % len(patterns)]
        text = pattern.format(i, i, i, i, i)
        texts.append(text)
    
    return texts

# Batch processing test
print("Performance testing:")
test_texts = generate_test_texts(50)
print(f"Generated {len(test_texts)} test texts")

# Time the processing
start_time = time.time()
total_matches = 0
total_urls = 0

for text in test_texts:
    matches = linkify.match(text)
    if matches:
        total_matches += 1
        total_urls += len(matches)

end_time = time.time()
processing_time = end_time - start_time

print(f"\nPerformance results:")
print(f"  Processing time: {processing_time:.4f} seconds")
print(f"  Texts processed: {len(test_texts)}")
print(f"  Texts with URLs: {total_matches}")
print(f"  Total URLs found: {total_urls}")
print(f"  Average per text: {processing_time/len(test_texts):.6f} seconds")
print(f"  URLs per second: {total_urls/processing_time:.1f}")

# Memory efficiency test
print("\nMemory efficiency test:")
large_text = " ".join([
    f"Link {i}: https://example{i}.com" for i in range(200)
])

print(f"Large text length: {len(large_text)} characters")

start_time = time.time()
matches = linkify.match(large_text)
end_time = time.time()

print(f"Large text processing:")
print(f"  Time: {end_time - start_time:.4f} seconds")
print(f"  URLs found: {len(matches) if matches else 0}")

# Test memory usage pattern
import sys
if matches:
    match_sizes = [sys.getsizeof(match) for match in matches[:10]]
    avg_match_size = sum(match_sizes) / len(match_sizes)
    print(f"  Average match object size: {avg_match_size:.1f} bytes")
```
