## Example 1: Basic Version Comparison and Validation

```python
# Example 1: Basic Version Comparison and Validation
print("=" * 70)
print(f"🚀 Example 1: Basic LooseVersion Comparison")
print("=" * 70)

from looseversion import LooseVersion

# Create version objects for software dependency checking
current_python = LooseVersion("3.8.5")
required_python = LooseVersion("3.7.0")
latest_python = LooseVersion("3.12.0")

print("Python Version Compatibility Check:")
print(f"Current: {current_python}")
print(f"Required: {required_python}")
print(f"Latest: {latest_python}")

# Version comparison operations
print(f"\nComparison Results:")
print(f"Current >= Required: {current_python >= required_python}")
print(f"Current < Latest: {current_python < latest_python}")
print(f"Current == Required: {current_python == required_python}")

# Package version checking for dependency management
package_versions = [
    ("numpy", LooseVersion("1.21.0"), LooseVersion("1.19.0")),
    ("requests", LooseVersion("2.28.1"), LooseVersion("2.25.0")),
    ("django", LooseVersion("4.1.2"), LooseVersion("3.2.0"))
]

print(f"\nPackage Dependency Validation:")
for name, installed, minimum in package_versions:
    compatible = installed >= minimum
    status = "✅ Compatible" if compatible else "❌ Incompatible"
    print(f"{name}: {installed} (min: {minimum}) - {status}")

print("✅ Version comparison operations completed successfully")
```

## Example 2: Version Sorting and Release Management

```python
# Example 2: Version Sorting and Release Management
print("=" * 70)
print(f"🚀 Example 2: Version Sorting and Release Management")
print("=" * 70)

from looseversion import LooseVersion

# Software release versions in random order
release_versions = [
    "2.1.0", "1.9.3", "2.0.0", "1.10.2", "2.0.1", 
    "1.9.10", "2.1.1", "1.8.5", "2.2.0", "1.9.0"
]

# Convert to LooseVersion objects for proper sorting
version_objects = [LooseVersion(v) for v in release_versions]

print("Unsorted Releases:")
for i, version in enumerate(release_versions, 1):
    print(f"  {i:2d}. v{version}")

# Sort versions using LooseVersion comparison
sorted_versions = sorted(version_objects)

print(f"\nSorted Releases (oldest to newest):")
for i, version in enumerate(sorted_versions, 1):
    print(f"  {i:2d}. v{version}")

# Find latest version and create upgrade path
latest = max(version_objects)
current = LooseVersion("1.9.3")

print(f"\nUpgrade Analysis:")
print(f"Current Version: v{current}")
print(f"Latest Version: v{latest}")

# Find all versions newer than current
newer_versions = [v for v in sorted_versions if v > current]
print(f"Available Upgrades:")
for version in newer_versions:
    print(f"  → v{version}")

print("✅ Version sorting and release management completed")
```

## Example 3: Version Parsing and Component Analysis

```python
# Example 3: Version Parsing and Component Analysis
print("=" * 70)
print(f"🚀 Example 3: Version Parsing and Component Analysis")
print("=" * 70)

from looseversion import LooseVersion

# Various version string formats commonly found in software
version_strings = [
    "3.9.2",
    "2.0.0-alpha.1",
    "1.5.3.final",
    "4.2.1-rc2",
    "1.0.0.dev5",
    "2.3.beta"
]

print("Version String Analysis:")
for version_str in version_strings:
    version = LooseVersion(version_str)
    
    print(f"\nInput: '{version_str}'")
    print(f"Parsed: {version}")
    print(f"String representation: {version.vstring}")
    
    # Access parsed components
    if hasattr(version, 'version') and version.version:
        print(f"Components: {version.version}")
    
    # Show version normalization - how LooseVersion standardizes the input
    normalized = str(version)
    print(f"Normalized: '{normalized}'")

# Demonstrate parsing flexibility
print(f"\n" + "=" * 40)
print("Parsing Flexibility Demonstration:")

# LooseVersion keeps separators and letters as str components: '1-0-0' parses to [1, '-', 0, '-', 0], and
# comparing it with '1.0' ([1, 0]) raises TypeError because the components mix int and str. The list below uses
# dot-separated forms only. They compare component by component: leading zeros drop out ('1.00.00' == '1.0.0'),
# and a shorter version sorts first ('1.0' < '1.0.0').
dotted_versions = ["1.0", "1.0.0", "1.00.00"]

parsed_versions = [LooseVersion(v) for v in dotted_versions]

print("Input formats:")
for orig, parsed in zip(dotted_versions, parsed_versions):
    print(f"  '{orig}' → {parsed.version}")
assert [v.version for v in parsed_versions] == [[1, 0], [1, 0, 0], [1, 0, 0]], [v.version for v in parsed_versions]

# Compare each form with the first one
base_version = parsed_versions[0]
print(f"\nEquality check (all compared to '{base_version}'):")
for i, version in enumerate(parsed_versions[1:], 1):
    equal = base_version == version
    print(f"  {equal} : '{dotted_versions[0]}' == '{dotted_versions[i]}'")
assert parsed_versions[1] == parsed_versions[2] and parsed_versions[0] < parsed_versions[1]

print("✅ Version parsing and analysis completed")
```

## Example 4: Complex Version String Handling and Edge Cases

```python
# Example 4: Complex Version String Handling and Edge Cases
print("=" * 70)
print(f"🚀 Example 4: Complex Version String Handling")
print("=" * 70)

from looseversion import LooseVersion

# Complex and non-standard version strings
complex_versions = [
    "1.0.0-alpha.2+build.123",
    "v2.1.3-stable",
    "3.0-SNAPSHOT",
    "1.2.3.4.5.6",
    "2021.03.15",
    "1.0a1",
    "2.0b3",
    "3.0rc1",
    "1.2.post1",
    "version-1.5.0"
]

print("Complex Version String Processing:")
processed_versions = []

for version_str in complex_versions:
    try:
        version = LooseVersion(version_str)
        processed_versions.append((version_str, version))
        print(f"✅ '{version_str}' → {version}")
    except Exception as e:
        print(f"❌ '{version_str}' → Error: {e}")
        raise

# Sort complex versions to show LooseVersion's flexibility
print(f"\nSorted Complex Versions:")
# LooseVersion does not order arbitrary strings: its component list keeps non-numeric runs as str
# ('v2.1.3-stable' -> ['v', 2, 1, 3, '-', 'stable']), and comparing an int component with a str one raises
# TypeError. Sort the all-numeric versions and list the others separately.
valid_versions = [v[1] for v in processed_versions]
# Partition by index: `v in numeric_versions` would call LooseVersion.__eq__, which compares mixed components.
numeric_indices = {
    i for i, v in enumerate(valid_versions) if all(isinstance(c, int) for c in v.version)
}
numeric_versions = [v for i, v in enumerate(valid_versions) if i in numeric_indices]
mixed_versions = [v for i, v in enumerate(valid_versions) if i not in numeric_indices]

for i, version in enumerate(sorted(numeric_versions), 1):
    print(f"  {i:2d}. {version}")
if mixed_versions:
    print(f"  not mutually comparable (components mix int and str): "
          f"{', '.join(str(v) for v in mixed_versions)}")

# Classify on whole str components, not substrings: 'v2.1.3-stable' contains the letter 'a' but has no 'a'
# component. Components can carry the separator in front of them ('3.0-SNAPSHOT' -> [3, 0, '-SNAPSHOT']).
print(f"\nVersion Classification:")
categories = {}
for original, version in processed_versions:
    tokens = {c.strip("-+").lower() for c in version.version if isinstance(c, str)}
    
    if tokens & {"a", "alpha"}:
        category = "Alpha Release"
    elif tokens & {"b", "beta"}:
        category = "Beta Release"
    elif "rc" in tokens:
        category = "Release Candidate"
    elif "snapshot" in tokens:
        category = "Development Snapshot"
    else:
        category = "Stable Release"
    
    categories[original] = category
    print(f"  {original} → {category}")
assert categories == {
    "1.0.0-alpha.2+build.123": "Alpha Release",
    "v2.1.3-stable": "Stable Release",
    "3.0-SNAPSHOT": "Development Snapshot",
    "1.2.3.4.5.6": "Stable Release",
    "2021.03.15": "Stable Release",
    "1.0a1": "Alpha Release",
    "2.0b3": "Beta Release",
    "3.0rc1": "Release Candidate",
    "1.2.post1": "Stable Release",
    "version-1.5.0": "Stable Release",
}, categories

print("✅ Complex version string handling completed")
```

## Example 5: Version Range Checking and Compatibility Matrix

```python
# Example 5: Version Range Checking and Compatibility Matrix
print("=" * 70)
print(f"🚀 Example 5: Version Range Checking and Compatibility")
print("=" * 70)

from looseversion import LooseVersion

# Define compatibility requirements for a software project
requirements = {
    "python": {"min": LooseVersion("3.7.0"), "max": LooseVersion("3.11.9")},
    "django": {"min": LooseVersion("3.2.0"), "max": LooseVersion("4.2.0")},
    "requests": {"min": LooseVersion("2.25.0"), "max": LooseVersion("3.0.0")},
    "numpy": {"min": LooseVersion("1.19.0"), "max": None}  # No upper limit
}

# Current environment versions
environment = {
    "python": LooseVersion("3.9.2"),
    "django": LooseVersion("4.1.5"),
    "requests": LooseVersion("2.28.1"),
    "numpy": LooseVersion("1.24.1")
}

print("Compatibility Matrix Analysis:")
print("=" * 60)

compatible_count = 0
total_packages = len(requirements)

for package, constraints in requirements.items():
    current_version = environment.get(package)
    min_version = constraints["min"]
    max_version = constraints["max"]
    
    print(f"\n📦 {package.upper()}:")
    print(f"  Current: {current_version}")
    print(f"  Required: >= {min_version}", end="")
    if max_version:
        print(f" and < {max_version}")
    else:
        print(" (no upper limit)")
    
    # Check compatibility
    min_compatible = current_version >= min_version
    max_compatible = max_version is None or current_version < max_version
    fully_compatible = min_compatible and max_compatible
    
    if fully_compatible:
        print(f"  Status: ✅ COMPATIBLE")
        compatible_count += 1
    else:
        print(f"  Status: ❌ INCOMPATIBLE")
        if not min_compatible:
            print(f"    → Version too old (need >= {min_version})")
        if not max_compatible:
            print(f"    → Version too new (need < {max_version})")

# Overall compatibility summary
print(f"\n" + "=" * 60)
print(f"COMPATIBILITY SUMMARY:")
print(f"Compatible packages: {compatible_count}/{total_packages}")

if compatible_count == total_packages:
    print("🎉 All packages are compatible!")
else:
    print("⚠️  Some packages need updating for full compatibility")

print("✅ Version compatibility checking completed")
```
