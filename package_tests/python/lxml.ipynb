## Example 1: XML parsing and tree manipulation

```python
from lxml import etree

# Example 1: XML parsing and tree manipulation
print("Example 1: XML Parsing and Tree Manipulation")

# Sample XML document
xml_data = """<?xml version="1.0" encoding="UTF-8"?>
    <bookstore>
        <book id="1" category="fiction">
            <title>The Great Gatsby</title>
            <author>F. Scott Fitzgerald</author>
            <price currency="USD">12.99</price>
            <year>1925</year>
        </book>
        <book id="2" category="science">
            <title>A Brief History of Time</title>
            <author>Stephen Hawking</author>
            <price currency="USD">15.99</price>
            <year>1988</year>
        </book>
        <book id="3" category="fiction">
            <title>To Kill a Mockingbird</title>
            <author>Harper Lee</author>
            <price currency="USD">13.99</price>
            <year>1960</year>
        </book>
    </bookstore>"""

print("Parsing XML document...")

# Parse the XML
# A `str` may not carry an encoding declaration (lxml raises ValueError) -- parse the encoded bytes.
root = etree.fromstring(xml_data.encode("utf-8"))
print(f"Root element: {root.tag}")
print(f"Number of books: {len(root)}")

# Iterate through books
print("\nBook details:")
for i, book in enumerate(root):
    book_id = book.get('id')
    category = book.get('category')
    title = book.find('title').text
    author = book.find('author').text
    price = book.find('price').text
    currency = book.find('price').get('currency')
    year = book.find('year').text
    
    print(f"  Book {book_id}: '{title}' by {author}")
    print(f"    Category: {category}, Price: {price} {currency}, Year: {year}")

# Add a new book
new_book = etree.SubElement(root, "book", id="4", category="mystery")
etree.SubElement(new_book, "title").text = "The Maltese Falcon"
etree.SubElement(new_book, "author").text = "Dashiell Hammett"
price_elem = etree.SubElement(new_book, "price", currency="USD")
price_elem.text = "11.99"
etree.SubElement(new_book, "year").text = "1930"

print(f"\nAfter adding new book: {len(root)} books total")

# Generate XML string
xml_output = etree.tostring(root, encoding='unicode', pretty_print=True)
print(f"Generated XML length: {len(xml_output)} characters")
print(f"Sample output:\n{xml_output[:200]}...")

print("XML parsing and manipulation completed successfully!")
```

## Example 2: XPath queries and advanced element selection

```python
from lxml import etree

# Example 2: XPath queries and advanced element selection
print("Example 2: XPath Queries and Advanced Element Selection")

# Create a more complex XML document
xml_doc = """<?xml version="1.0"?>
    <library>
        <section name="Fiction">
            <book isbn="978-0-7432-7356-5" available="true">
                <title>The Da Vinci Code</title>
                <author nationality="American">Dan Brown</author>
                <price>14.99</price>
                <ratings>
                    <rating source="GoodReads">4.1</rating>
                    <rating source="Amazon">4.3</rating>
                </ratings>
            </book>
            <book isbn="978-0-553-29698-2" available="false">
                <title>A Game of Thrones</title>
                <author nationality="American">George R.R. Martin</author>
                <price>16.99</price>
                <ratings>
                    <rating source="GoodReads">4.4</rating>
                    <rating source="Amazon">4.5</rating>
                </ratings>
            </book>
        </section>
        <section name="Science">
            <book isbn="978-0-553-10953-5" available="true">
                <title>Cosmos</title>
                <author nationality="American">Carl Sagan</author>
                <price>13.99</price>
                <ratings>
                    <rating source="GoodReads">4.3</rating>
                    <rating source="Amazon">4.6</rating>
                </ratings>
            </book>
        </section>
    </library>"""

# Parse the document
root = etree.fromstring(xml_doc)
print("Complex XML document parsed successfully")

# XPath query examples
print("\nXPath query examples:")

# 1. Find all book titles
titles = root.xpath("//title/text()")
print(f"All book titles ({len(titles)}):")
for title in titles:
    print(f"  - {title}")

# 2. Find available books
available_books = root.xpath("//book[@available='true']")
print(f"\nAvailable books ({len(available_books)}):")
for book in available_books:
    title = book.find('title').text
    isbn = book.get('isbn')
    print(f"  - {title} (ISBN: {isbn})")

# 3. Find books with high ratings (>4.0 on GoodReads)
high_rated = root.xpath("//book[ratings/rating[@source='GoodReads' and text()>4.0]]")
print(f"\nHigh-rated books on GoodReads ({len(high_rated)}):")
for book in high_rated:
    title = book.find('title').text
    rating = book.xpath("ratings/rating[@source='GoodReads']/text()")[0]
    print(f"  - {title}: {rating}/5.0")

# 4. Find American authors
american_authors = root.xpath("//author[@nationality='American']/text()")
print(f"\nAmerican authors ({len(american_authors)}):")
for author in american_authors:
    print(f"  - {author}")

# 5. Complex query: Fiction books under $15
cheap_fiction = root.xpath("//section[@name='Fiction']/book[price<15]/title/text()")
print(f"\nFiction books under $15 ({len(cheap_fiction)}):")
for title in cheap_fiction:
    print(f"  - {title}")

# XPath with namespaces (demonstrate capability)
print(f"\nXPath capabilities:")
capabilities = [
    "Element selection: //element, /path/to/element",
    "Attribute filtering: [@attribute='value']",
    "Text content: /text(), contains(text(),'value')",
    "Position-based: [position()=1], [last()]", 
    "Comparison operators: <, >, =, !=",
    "Logical operators: and, or, not"
]

for capability in capabilities:
    print(f"  • {capability}")

print("XPath queries completed successfully!")
```

## Example 3: HTML parsing and web scraping capabilities

```python
from lxml import html

# Example 3: HTML parsing and web scraping capabilities
print("Example 3: HTML Parsing and Web Scraping Capabilities")

# Sample HTML document
html_content = """<!DOCTYPE html>
    <html>
    <head>
        <title>Sample E-commerce Page</title>
        <meta charset="UTF-8">
    </head>
    <body>
        <header>
            <h1>Online Store</h1>
            <nav>
                <ul>
                    <li><a href="/home">Home</a></li>
                    <li><a href="/products">Products</a></li>
                    <li><a href="/contact">Contact</a></li>
                </ul>
            </nav>
        </header>
        
        <main>
            <section class="products">
                <h2>Featured Products</h2>
                <div class="product" data-id="1">
                    <h3>Laptop Computer</h3>
                    <p class="description">High-performance laptop for work and gaming.</p>
                    <span class="price">$999.99</span>
                    <div class="rating" data-stars="4">★★★★☆</div>
                </div>
                
                <div class="product" data-id="2">
                    <h3>Wireless Headphones</h3>
                    <p class="description">Noise-canceling wireless headphones.</p>
                    <span class="price">$199.99</span>
                    <div class="rating" data-stars="5">★★★★★</div>
                </div>
                
                <div class="product" data-id="3">
                    <h3>Smartphone</h3>
                    <p class="description">Latest model with advanced camera.</p>
                    <span class="price">$699.99</span>
                    <div class="rating" data-stars="4">★★★★☆</div>
                </div>
            </section>
        </main>
        
        <footer>
            <p>&copy; 2024 Online Store. All rights reserved.</p>
            <div class="social">
                <a href="https://twitter.com/store">Twitter</a>
                <a href="https://facebook.com/store">Facebook</a>
            </div>
        </footer>
    </body>
    </html>"""

print("Parsing HTML document...")

# Parse HTML
doc = html.fromstring(html_content)

# Extract page title
title = doc.xpath('//title/text()')[0] if doc.xpath('//title/text()') else "No title"
print(f"Page title: {title}")

# Extract navigation links
nav_links = doc.xpath('//nav//a')
print(f"\nNavigation links ({len(nav_links)}):")
for link in nav_links:
    text = link.text_content()
    href = link.get('href', '#')
    print(f"  - {text}: {href}")

# Extract product information
products = doc.xpath('//div[@class="product"]')
print(f"\nProducts found ({len(products)}):")

for product in products:
    product_id = product.get('data-id')
    name = product.xpath('.//h3/text()')[0]
    description = product.xpath('.//p[@class="description"]/text()')[0]
    price = product.xpath('.//span[@class="price"]/text()')[0]
    rating_stars = product.xpath('.//div[@class="rating"]/@data-stars')[0]
    
    print(f"  Product {product_id}: {name}")
    print(f"    Description: {description}")
    print(f"    Price: {price}")
    print(f"    Rating: {rating_stars} stars")

# Extract all links
all_links = doc.xpath('//a/@href')
print(f"\nAll links found ({len(all_links)}):")
for link in all_links:
    print(f"  - {link}")

# CSS selector-like queries using XPath
print(f"\nCSS-like selections:")

# Find elements by class (equivalent to .price)
price_elements = doc.xpath('//*[@class="price"]')
prices = [elem.text_content() for elem in price_elements]
print(f"  All prices: {prices}")

# Find elements with specific data attributes
rated_products = doc.xpath('//*[@data-stars]')
print(f"  Products with ratings: {len(rated_products)}")

# Text content extraction
main_text = doc.xpath('//main')[0].text_content().strip()
print(f"  Main content length: {len(main_text)} characters")

print("HTML parsing and extraction completed successfully!")
```

## Example 4: XML validation and schema processing

```python
from lxml import etree
from io import StringIO

# Example 4: XML validation and schema processing
print("Example 4: XML Validation and Schema Processing")

# Define a simple XML Schema (XSD)
schema_doc = """<?xml version="1.0" encoding="UTF-8"?>
    <xs:schema xmlns:xs="http://www.w3.org/2001/XMLSchema">
        <xs:element name="catalog">
            <xs:complexType>
                <xs:sequence>
                    <xs:element name="product" maxOccurs="unbounded">
                        <xs:complexType>
                            <xs:sequence>
                                <xs:element name="name" type="xs:string"/>
                                <xs:element name="price" type="xs:decimal"/>
                                <xs:element name="category" type="xs:string"/>
                            </xs:sequence>
                            <xs:attribute name="id" type="xs:positiveInteger" use="required"/>
                        </xs:complexType>
                    </xs:element>
                </xs:sequence>
            </xs:complexType>
        </xs:element>
    </xs:schema>"""

# Valid XML document
valid_xml = """<?xml version="1.0"?>
    <catalog>
        <product id="1">
            <name>Laptop</name>
            <price>999.99</price>
            <category>Electronics</category>
        </product>
        <product id="2">
            <name>Book</name>
            <price>19.99</price>
            <category>Literature</category>
        </product>
    </catalog>"""

# Invalid XML document (missing required attribute)
invalid_xml = """<?xml version="1.0"?>
    <catalog>
        <product>
            <name>Invalid Product</name>
            <price>99.99</price>
            <category>Test</category>
        </product>
    </catalog>"""

print("Creating XML Schema...")

# Parse schema
# A `str` may not carry an encoding declaration (lxml raises ValueError) -- parse the encoded bytes.
schema_root = etree.XML(schema_doc.encode("utf-8"))
schema = etree.XMLSchema(schema_root)

print(f"Schema created with {len(schema_root)} elements")

# Create validator
xmlparser = etree.XMLParser(schema=schema)

print("\nValidating documents:")

# Test valid document: a schema violation raises XMLSyntaxError, which fails this example
etree.fromstring(valid_xml, xmlparser)
print("  ✓ Valid XML document passed validation")

# Test invalid document: the validating parser must reject a <product> without its required id attribute
try:
    etree.fromstring(invalid_xml, xmlparser)
except etree.XMLSyntaxError as e:
    assert "The attribute 'id' is required but missing" in str(e), f"unexpected message: {e!r}"
    print(f"  ✓ Invalid XML correctly rejected: {e}")
else:
    raise AssertionError("the schema-validating parser accepted a <product> without its required id attribute")

# Manual validation approach
print("\nManual validation approach:")

# Parse documents without validation
valid_doc = etree.fromstring(valid_xml)
invalid_doc = etree.fromstring(invalid_xml)

# Validate against schema
valid_result = schema.validate(valid_doc)
invalid_result = schema.validate(invalid_doc)

print(f"  Valid document validation: {valid_result}")
print(f"  Invalid document validation: {invalid_result}")
assert valid_result is True and invalid_result is False, (valid_result, invalid_result)

# Get validation errors
if not invalid_result:
    print(f"  Validation errors for invalid document:")
    for error in schema.error_log:
        print(f"    Line {error.line}: {error.message}")

# DTD validation example
print(f"\nDocument processing features:")

# Pretty print XML
pretty_xml = etree.tostring(valid_doc, pretty_print=True, encoding='unicode')
print(f"  Pretty printing: {len(pretty_xml)} characters")

# CDATA handling
cdata_xml = "<message><![CDATA[This is <b>bold</b> text]]></message>"
cdata_doc = etree.fromstring(cdata_xml)
cdata_text = cdata_doc.text
print(f"  CDATA content: {cdata_text}")

# Namespace handling
ns_xml = '<root xmlns:book="http://example.com/book"><book:title>Test</book:title></root>'
ns_doc = etree.fromstring(ns_xml)
namespaces = ns_doc.nsmap
print(f"  Namespaces found: {namespaces}")

print("XML validation and processing completed successfully!")
```

## Example 5: Advanced XML construction and performance optimization

```python
from lxml import etree
from lxml.builder import E

# Example 5: Advanced XML construction and performance optimization
print("Example 5: Advanced XML Construction and Performance Optimization")

# Method 1: Programmatic XML construction using Element()
print("Method 1: Programmatic construction")

# Create root element
catalog = etree.Element("catalog")
catalog.set("version", "1.0")
catalog.set("xmlns", "http://example.com/catalog")

# Add products programmatically
for i in range(1, 4):
    product = etree.SubElement(catalog, "product")
    product.set("id", str(i))
    product.set("featured", "true" if i == 1 else "false")
    
    # Add product details
    etree.SubElement(product, "name").text = f"Product {i}"
    etree.SubElement(product, "price").text = f"{99.99 + i * 100:.2f}"
    etree.SubElement(product, "stock").text = str(50 - i * 10)
    
    # Add categories
    categories = etree.SubElement(product, "categories")
    for cat in ["Electronics", f"Category{i}"]:
        cat_elem = etree.SubElement(categories, "category")
        cat_elem.text = cat

programmatic_xml = etree.tostring(catalog, pretty_print=True, encoding='unicode')
print(f"Generated XML: {len(programmatic_xml)} characters")

# Method 2: Using lxml.builder (E factory)
print("\nMethod 2: E factory construction")

# Build XML using E factory
e_catalog = E.catalog(
    E.product(
        E.name("Gaming Laptop"),
        E.price("1299.99"),
        E.specifications(
            E.cpu("Intel i7"),
            E.ram("16GB"),
            E.storage("1TB SSD")
        ),
        E.reviews(
            E.review(E.rating("5"), E.comment("Excellent performance!"), user="user1"),
            E.review(E.rating("4"), E.comment("Good value for money"), user="user2")
        ),
        id="laptop-001",
        category="electronics"
    ),
    E.product(
        E.name("Wireless Mouse"),
        E.price("29.99"),
        E.specifications(
            E.connectivity("Bluetooth"),
            E.battery("2 AAA"),
            E.range("10m")
        ),
        id="mouse-001", 
        category="accessories"
    ),
    version="2.0"
)

e_xml = etree.tostring(e_catalog, pretty_print=True, encoding='unicode')
print(f"E factory XML: {len(e_xml)} characters")

# Performance comparison
print("\nPerformance optimization techniques:")

import time

# Large document creation test
def create_large_xml_programmatic(num_items):
    root = etree.Element("data")
    for i in range(num_items):
        item = etree.SubElement(root, "item", id=str(i))
        etree.SubElement(item, "value").text = f"value_{i}"
        etree.SubElement(item, "timestamp").text = f"2024-01-{i%30+1:02d}"
    return root

def create_large_xml_builder(num_items):
    items = []
    for i in range(num_items):
        items.append(
            E.item(
                E.value(f"value_{i}"),
                E.timestamp(f"2024-01-{i%30+1:02d}"),
                id=str(i)
            )
        )
    return E.data(*items)

# Performance test
num_items = 1000

# Programmatic approach
start_time = time.time()
prog_doc = create_large_xml_programmatic(num_items)
prog_time = time.time() - start_time

# Builder approach  
start_time = time.time()
builder_doc = create_large_xml_builder(num_items)
builder_time = time.time() - start_time

print(f"  Programmatic ({num_items} items): {prog_time:.4f} seconds")
print(f"  Builder approach ({num_items} items): {builder_time:.4f} seconds")
print(f"  Speed ratio: {prog_time/builder_time:.2f}x")

# Memory efficiency test
import sys
prog_size = sys.getsizeof(etree.tostring(prog_doc))
builder_size = sys.getsizeof(etree.tostring(builder_doc))

print(f"  Serialized size comparison: {abs(prog_size - builder_size)} bytes difference")

# Advanced features
print(f"\nAdvanced XML features:")

# Custom namespace handling
NSMAP = {
    'cat': 'http://example.com/catalog',
    'prod': 'http://example.com/product'
}

ns_root = etree.Element("{http://example.com/catalog}catalog", nsmap=NSMAP)
ns_product = etree.SubElement(ns_root, "{http://example.com/product}product")
ns_product.text = "Namespaced product"

ns_xml = etree.tostring(ns_root, pretty_print=True, encoding='unicode')
print(f"  Namespace XML length: {len(ns_xml)} characters")

# Processing instructions and comments
pi = etree.ProcessingInstruction("xml-stylesheet", 'type="text/xsl" href="style.xsl"')
comment = etree.Comment("This is a generated XML document")

final_doc = etree.Element("document")
final_doc.insert(0, pi)
final_doc.insert(1, comment)
etree.SubElement(final_doc, "content").text = "Main content here"

final_xml = etree.tostring(final_doc, pretty_print=True, encoding='unicode')
print(f"  Document with PI and comments: {len(final_xml)} characters")

print("Advanced XML construction completed successfully!")
```
