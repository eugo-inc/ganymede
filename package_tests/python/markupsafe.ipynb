## Example 1: markupsafe functionality test 1

```python
from markupsafe import Markup, escape, escape_silent

# Example 1: Basic HTML escaping and safe string handling
print("Example 1: Basic HTML Escaping and Safe String Handling")

try:
    # Demonstrate basic HTML escaping
    print("HTML Escaping Examples:")
    
    # Dangerous user input that could cause XSS
    user_inputs = [
        '<script>alert("XSS attack!")</script>',
        'Hello <b>World</b> & "Friends"',
        "<img src='x' onerror='alert(1)'>",
        'Simple text without HTML',
        'Price: $100 > $50 & tax included',
        '<div onclick="malicious()">Click me</div>'
    ]
    
    print("\nEscaping potentially dangerous input:")
    for i, user_input in enumerate(user_inputs, 1):
        escaped = escape(user_input)
        print(f"{i}. Original: {user_input}")
        print(f"   Escaped:  {escaped}")
        print(f"   Type:     {type(escaped).__name__}")
        print()
    
    # Demonstrate Markup class for safe strings
    print("Markup class for safe strings:")
    
    # Create safe markup
    safe_html = Markup('<p>This is <b>safe</b> HTML content</p>')
    unsafe_text = '<script>alert("danger")</script>'
    
    print(f"Safe HTML: {safe_html}")
    print(f"Type: {type(safe_html).__name__}")
    print(f"Is safe: {hasattr(safe_html, '__html__')}")
    
    # Combining safe and unsafe content
    combined = safe_html + escape(unsafe_text)
    print(f"\nCombined safe + escaped: {combined}")
    print(f"Result type: {type(combined).__name__}")
    
    # String formatting with automatic escaping
    template = Markup('<p>Welcome, <strong>{}</strong>! You have {} messages.</p>')
    username = '<script>evil()</script>'
    message_count = 5
    
    formatted = template.format(escape(username), message_count)
    print(f"\nFormatted template: {formatted}")
    
    # Demonstrate escape_silent for None values
    print("\nHandling None values:")
    values = ["Hello", None, "<script>test</script>", 42]
    
    for value in values:
        escaped_regular = escape(value) if value is not None else "N/A"
        escaped_silent = escape_silent(value)
        
        print(f"Value: {value}")
        print(f"  Regular escape: {escaped_regular}")
        print(f"  Silent escape:  {escaped_silent}")
    
    # Show string operations on Markup
    print("\nMarkup string operations:")
    
    markup_text = Markup('<p>Hello <em>World</em></p>')
    
    # String methods that preserve Markup type
    print(f"Original: {markup_text}")
    print(f"Upper: {markup_text.upper()}")
    print(f"Replace: {markup_text.replace('World', 'Universe')}")
    print(f"Strip tags: {markup_text.striptags()}")
    
    # Show escaping characters reference
    print("\nHTML character escaping reference:")
    special_chars = {
        '<': '&lt;',
        '>': '&gt;',
        '&': '&amp;',
        '"': '&quot;',
        "'": '&#x27;'
    }
    
    for char, escaped_form in special_chars.items():
        actual_escaped = escape(char)
        print(f"'{char}' -> '{escaped_form}' (actual: '{actual_escaped}')")
    
    # Basic security features
    print(f"\nSecurity features:")
    security_features = [
        "HTML escaping: Converts dangerous characters to safe entities",
        "XSS prevention: Neutralizes script injection attempts",
        "Safe marking: Markup class identifies already-safe content",
        "Template safety: Integrates with template engines for automatic escaping",
        "Type preservation: Maintains safety information through operations",
        "Performance: Optimized C implementation for speed"
    ]
    
    for feature in security_features:
        print(f"  • {feature}")
    
    print("Basic HTML escaping and safe string handling completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    # Fallback demonstration
    print("Demonstrating HTML escaping concepts:")
    concepts = [
        "HTML escaping: Convert dangerous HTML characters to safe entities",
        "XSS prevention: Protect against cross-site scripting attacks",
        "Safe strings: Mark trusted content to avoid double-escaping",
        "Template integration: Seamless integration with template engines",
        "Security by default: Automatic protection for web applications"
    ]
    
    for concept in concepts:
        print(f"  - {concept}")
    raise
```

## Example 3: markupsafe functionality test 3

```python
from markupsafe import Markup, escape

# Example 3: Template engine integration and real-world usage
print("Example 3: Template Engine Integration and Real-World Usage")

try:
    # Simulate template engine behavior
    class SimpleTemplateEngine:
        """Simple template engine demonstrating MarkupSafe integration"""
        
        def __init__(self, auto_escape=True):
            self.auto_escape = auto_escape
        
        def render_variable(self, value):
            """Render a variable with optional auto-escaping"""
            if self.auto_escape:
                if isinstance(value, Markup):
                    return value  # Already safe, don't escape
                else:
                    return escape(value)  # Escape unsafe content
            else:
                return str(value)
        
        def render_template(self, template_string, context):
            """Simple template rendering with variable substitution"""
            result = template_string
            
            # Replace variables in template
            for key, value in context.items():
                placeholder = f"{{{key}}}"
                if placeholder in result:
                    rendered_value = self.render_variable(value)
                    result = result.replace(placeholder, str(rendered_value))
            
            return Markup(result) if self.auto_escape else result
    
    # Test template engine with different content types
    print("Template engine with auto-escaping:")
    
    engine = SimpleTemplateEngine(auto_escape=True)
    
    template = '''
    <div class="user-profile">
        <h2>Profile: {username}</h2>
        <div class="bio">{user_bio}</div>
        <div class="status">{status_message}</div>
        <div class="custom-html">{custom_content}</div>
    </div>
    '''
    
    # Mix of safe and unsafe content
    context = {
        'username': '<script>alert("hacked")</script>',  # Unsafe - will be escaped
        'user_bio': 'Developer & <script>Security Expert</script>',  # Unsafe - will be escaped
        'status_message': 'Online since 2024',  # Safe text
        'custom_content': Markup('<p>This is <em>trusted</em> HTML</p>')  # Safe - won't be escaped
    }
    
    rendered = engine.render_template(template, context)
    print("Rendered template with auto-escaping:")
    print(rendered)
    print(f"Result type: {type(rendered).__name__}")
    
    # Compare with non-auto-escaping engine
    print("\n" + "="*50)
    print("Template engine WITHOUT auto-escaping (dangerous):")
    
    unsafe_engine = SimpleTemplateEngine(auto_escape=False)
    unsafe_rendered = unsafe_engine.render_template(template, context)
    print("Rendered template without auto-escaping:")
    print(unsafe_rendered)
    print(f"Result type: {type(unsafe_rendered).__name__}")
    
    # Real-world web form handling
    print("\n" + "="*50)
    print("Real-world web form handling:")
    
    class WebFormHandler:
        """Simulate web form handling with MarkupSafe"""
        
        def __init__(self):
            self.errors = []
        
        def validate_and_escape_input(self, field_name, value, max_length=None):
            """Validate and escape user input"""
            if not value:
                self.errors.append(f"{field_name} is required")
                return ""
            
            if max_length and len(value) > max_length:
                self.errors.append(f"{field_name} is too long (max {max_length} characters)")
                return escape(value[:max_length] + "...")
            
            return escape(value)
        
        def render_form_field(self, field_type, name, value="", **attrs):
            """Render form field with escaped values"""
            escaped_value = escape(value)
            escaped_name = escape(name)
            
            attr_strings = []
            for attr, attr_value in attrs.items():
                escaped_attr = escape(str(attr_value))
                attr_strings.append(f'{attr}="{escaped_attr}"')
            
            attrs_str = ' ' + ' '.join(attr_strings) if attr_strings else ''
            
            if field_type == 'text':
                return Markup(f'<input type="text" name="{escaped_name}" value="{escaped_value}"{attrs_str}>')
            elif field_type == 'textarea':
                return Markup(f'<textarea name="{escaped_name}"{attrs_str}>{escaped_value}</textarea>')
            elif field_type == 'select':
                options = attrs.get('options', [])
                option_html = []
                for option_value, option_text in options:
                    selected = ' selected' if str(option_value) == str(value) else ''
                    option_html.append(f'<option value="{escape(option_value)}"{selected}>{escape(option_text)}</option>')
                options_str = ''.join(option_html)
                return Markup(f'<select name="{escaped_name}"{attrs_str}>{options_str}</select>')
        
        def render_error_messages(self):
            """Render error messages safely"""
            if not self.errors:
                return Markup("")
            
            error_items = [f"<li>{escape(error)}</li>" for error in self.errors]
            return Markup(f'<ul class="errors">{"".join(error_items)}</ul>')
    
    # Test form handling
    form_handler = WebFormHandler()
    
    # Simulate user input with potential XSS
    user_data = {
        'name': 'John <script>alert("xss")</script> Doe',
        'email': 'john@example.com',
        'bio': 'I am a <script>hacker</script> and developer',
        'country': 'US'
    }
    
    print("Form field rendering with escaping:")
    
    # Render form fields
    name_field = form_handler.render_form_field('text', 'name', user_data['name'], 
                                               class_='form-control', maxlength='50')
    print(f"Name field: {name_field}")
    
    email_field = form_handler.render_form_field('text', 'email', user_data['email'], 
                                                class_='form-control', type='email')
    print(f"Email field: {email_field}")
    
    bio_field = form_handler.render_form_field('textarea', 'bio', user_data['bio'], 
                                             class_='form-control', rows='3')
    print(f"Bio field: {bio_field}")
    
    country_options = [('US', 'United States'), ('UK', 'United Kingdom'), ('CA', 'Canada')]
    country_field = form_handler.render_form_field('select', 'country', user_data['country'],
                                                  class_='form-control', options=country_options)
    print(f"Country field: {country_field}")
    
    # Validate input and show error handling
    form_handler.validate_and_escape_input('name', '', max_length=50)  # Empty name
    form_handler.validate_and_escape_input('bio', 'x' * 200, max_length=100)  # Too long
    
    error_html = form_handler.render_error_messages()
    print(f"\nError messages: {error_html}")
    
    # Email template rendering
    print("\n" + "="*50)
    print("Email template rendering:")
    
    def render_email_template(template, recipient_data):
        """Render email template with safe HTML and text versions"""
        
        # HTML version (with escaping for user content)
        html_template = Markup(template['html'])
        html_content = html_template.format(
            name=escape(recipient_data['name']),
            company=escape(recipient_data['company']),
            message=Markup(recipient_data['message']),  # Assume message is pre-approved HTML
            unsubscribe_link=recipient_data['unsubscribe_link']  # URLs don't need HTML escaping
        )
        
        # Text version (no HTML escaping needed)
        text_content = template['text'].format(
            name=recipient_data['name'],
            company=recipient_data['company'],
            message_text=recipient_data['message_text'],
            unsubscribe_link=recipient_data['unsubscribe_link']
        )
        
        return {
            'html': html_content,
            'text': text_content
        }
    
    # Email templates
    email_template = {
        'html': '''
        <html>
        <body>
            <h2>Welcome, {name}!</h2>
            <p>Thank you for joining {company}.</p>
            <div class="message">{message}</div>
            <p><a href="{unsubscribe_link}">Unsubscribe</a></p>
        </body>
        </html>
        ''',
        'text': '''
        Welcome, {name}!
        
        Thank you for joining {company}.
        
        {message_text}
        
        Unsubscribe: {unsubscribe_link}
        '''
    }
    
    # Recipient data with potential XSS in user-provided fields
    recipient = {
        'name': 'Alice <script>Johnson</script>',
        'company': 'Tech & <script>Innovation</script> Corp',
        'message': Markup('<p>Welcome to our <strong>premium</strong> service!</p>'),
        'message_text': 'Welcome to our premium service!',
        'unsubscribe_link': 'https://example.com/unsubscribe?token=abc123'
    }
    
    rendered_email = render_email_template(email_template, recipient)
    
    print("HTML email version:")
    print(rendered_email['html'])
    print("\nText email version:")
    print(rendered_email['text'])
    
    # Show integration benefits
    print(f"\nTemplate integration benefits:")
    integration_benefits = [
        "Automatic escaping: Template engines can auto-escape unsafe content",
        "Safe content preservation: Markup objects bypass escaping when appropriate",
        "Form field safety: User input automatically escaped in form rendering",
        "Mixed content handling: Combine trusted templates with user data safely",
        "Email template security: Prevent HTML injection in email content",
        "Development productivity: Security by default reduces vulnerability risk"
    ]
    
    for benefit in integration_benefits:
        print(f"  • {benefit}")
    
    print("Template integration and real-world usage completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    # Fallback demonstration
    print("Demonstrating integration concepts:")
    concepts = [
        "Template engine integration: Seamless escaping in template systems",
        "Web form handling: Safe rendering of user input in HTML forms",
        "Email templates: Secure HTML email generation with user data",
        "Mixed content rendering: Combining trusted and untrusted content safely",
        "Production web apps: Real-world patterns for secure web development"
    ]
    
    for concept in concepts:
        print(f"  - {concept}")
    raise
```

## Example 4: markupsafe functionality test 4

```python
from markupsafe import Markup, escape
import time

# Example 4: Performance optimization and security testing
print("Example 4: Performance Optimization and Security Testing")

try:
    # Performance benchmarking
    def benchmark_escaping_performance():
        """Benchmark different escaping approaches"""
        
        # Test data with various complexity levels
        test_inputs = [
            'Simple text without special characters',
            '<script>alert("XSS")</script>',
            'Complex HTML: <div class="test" onclick="evil()">Content & "quotes"</div>',
            'Mixed content: ' + '<script>hack()</script>' * 100,  # Large payload
            'Special chars: ' + '&<>"\'`' * 50,  # Many special chars
        ]
        
        results = {}
        
        for i, test_input in enumerate(test_inputs):
            # Benchmark escape() function
            start_time = time.time()
            for _ in range(1000):
                escaped = escape(test_input)
            escape_time = time.time() - start_time
            
            # Benchmark Markup creation
            start_time = time.time()
            for _ in range(1000):
                markup = Markup(test_input)
            markup_time = time.time() - start_time
            
            # Benchmark string operations on Markup
            markup_obj = Markup('<p>Test</p>')
            start_time = time.time()
            for _ in range(1000):
                result = markup_obj + escape(test_input)
            concat_time = time.time() - start_time
            
            results[f'input_{i+1}'] = {
                'length': len(test_input),
                'escape_time': escape_time,
                'markup_time': markup_time,
                'concat_time': concat_time,
                'sample': test_input[:30] + '...' if len(test_input) > 30 else test_input
            }
        
        return results
    
    # Run performance benchmark
    print("Performance benchmarking (1000 iterations each):")
    perf_results = benchmark_escaping_performance()
    
    for test_name, metrics in perf_results.items():
        print(f"\n{test_name} (length: {metrics['length']}):")
        print(f"  Sample: {metrics['sample']}")
        print(f"  escape(): {metrics['escape_time']*1000:.2f}ms")
        print(f"  Markup(): {metrics['markup_time']*1000:.2f}ms")
        print(f"  concatenation: {metrics['concat_time']*1000:.2f}ms")
    
    # Security testing - XSS payload resistance
    print("\n" + "="*50)
    print("XSS payload resistance testing:")
    
    # Common XSS payloads
    xss_payloads = [
        '<script>alert("XSS")</script>',
        '<img src=x onerror=alert("XSS")>',
        '<svg onload=alert("XSS")>',
        'javascript:alert("XSS")',
        '<iframe src="javascript:alert(\'XSS\')"></iframe>',
        '<body onload=alert("XSS")>',
        '<input type="image" src=x onerror=alert("XSS")>',
        '<div onclick="alert(\'XSS\')" style="cursor:pointer">Click me</div>',
        '<style>body{background:url("javascript:alert(\'XSS\')")}</style>',
        '"><script>alert("XSS")</script><"'
    ]
    
    print("Testing XSS payload escaping:")
    for i, payload in enumerate(xss_payloads, 1):
        escaped = escape(payload)
        print(f"{i:2d}. Original: {payload}")
        print(f"    Escaped:  {escaped}")
        
        # Verify no dangerous patterns remain
        dangerous_patterns = ['<script', 'javascript:', 'onerror=', 'onload=', 'onclick=']
        is_safe = not any(pattern.lower() in str(escaped).lower() for pattern in dangerous_patterns)
        print(f"    Safe: {'✓' if is_safe else '✗'}")
        print()
    
    # Memory usage analysis
    print("Memory usage analysis:")
    
    import sys
    
    # Compare memory usage of different string types
    regular_string = "Hello <script>World</script>"
    escaped_string = escape(regular_string)
    markup_string = Markup(regular_string)
    
    print(f"Regular string: {sys.getsizeof(regular_string)} bytes")
    print(f"Escaped string: {sys.getsizeof(escaped_string)} bytes")
    print(f"Markup string:  {sys.getsizeof(markup_string)} bytes")
    
    # Test memory with large content
    large_content = "<script>alert('xss')</script>" * 1000
    large_escaped = escape(large_content)
    large_markup = Markup(large_content)
    
    print(f"\nLarge content ({len(large_content):,} chars):")
    print(f"Original: {sys.getsizeof(large_content):,} bytes")
    print(f"Escaped:  {sys.getsizeof(large_escaped):,} bytes")
    print(f"Markup:   {sys.getsizeof(large_markup):,} bytes")
    
    # Security best practices demonstration
    print("\n" + "="*50)
    print("Security best practices:")
    
    class SecureRenderer:
        """Demonstrates security best practices with MarkupSafe"""
        
        @staticmethod
        def safe_attribute_render(tag, attributes):
            """Safely render HTML attributes"""
            attr_parts = []
            for name, value in attributes.items():
                # Escape both attribute name and value
                safe_name = escape(name)
                safe_value = escape(value)
                attr_parts.append(f'{safe_name}="{safe_value}"')
            
            attrs_str = ' '.join(attr_parts)
            return Markup(f'<{escape(tag)} {attrs_str}>')
        
        @staticmethod
        def safe_url_render(url, text, **attrs):
            """Safely render URLs with validation"""
            # Basic URL validation (in production, use more robust validation)
            if not url.startswith(('http://', 'https://', 'mailto:', '/')):
                url = '#'  # Fallback for suspicious URLs
            
            escaped_url = escape(url)
            escaped_text = escape(text)
            
            attr_parts = []
            for name, value in attrs.items():
                attr_parts.append(f'{escape(name)}="{escape(value)}"')
            
            attrs_str = ' ' + ' '.join(attr_parts) if attr_parts else ''
            
            return Markup(f'<a href="{escaped_url}"{attrs_str}>{escaped_text}</a>')
        
        @staticmethod
        def safe_list_render(items, item_renderer=None):
            """Safely render lists with custom item rendering"""
            if not items:
                return Markup('<ul><li>No items</li></ul>')
            
            item_renderer = item_renderer or (lambda x: escape(str(x)))
            
            list_items = []
            for item in items:
                rendered_item = item_renderer(item)
                if not isinstance(rendered_item, Markup):
                    rendered_item = escape(rendered_item)
                list_items.append(f'<li>{rendered_item}</li>')
            
            return Markup('<ul>' + ''.join(list_items) + '</ul>')
    
    # Test secure rendering
    renderer = SecureRenderer()
    
    # Test attribute rendering
    print("Secure attribute rendering:")
    tag_attrs = {
        'class': 'user-input <script>alert("xss")</script>',
        'data-value': '"><script>alert("xss")</script><"',
        'title': 'Tooltip & "quotes"'
    }
    
    safe_tag = renderer.safe_attribute_render('div', tag_attrs)
    print(f"Safe tag: {safe_tag}")
    
    # Test URL rendering
    print("\nSecure URL rendering:")
    test_urls = [
        ('https://example.com', 'Safe link'),
        ('javascript:alert("XSS")', 'Malicious link'),
        ('http://example.com/page?param=<script>alert("xss")</script>', 'Link with XSS'),
    ]
    
    for url, text in test_urls:
        safe_link = renderer.safe_url_render(url, text, class_='external-link')
        print(f"URL: {url[:40]}... -> {safe_link}")
    
    # Test list rendering
    print("\nSecure list rendering:")
    mixed_items = [
        'Safe item',
        '<script>alert("xss")</script>',
        Markup('<em>Trusted HTML</em>'),
        'Item with "quotes" & ampersands'
    ]
    
    safe_list = renderer.safe_list_render(mixed_items)
    print(f"Safe list: {safe_list}")
    
    # Performance optimization tips
    print(f"\nPerformance optimization patterns:")
    optimization_patterns = [
        "Pre-escape static content: Use Markup() for known-safe static HTML",
        "Batch operations: Combine multiple escaping operations when possible",
        "Cache escaped content: Store escaped versions of frequently used content",
        "Avoid double escaping: Check if content is already Markup before escaping",
        "Use escape_silent: Handle None values efficiently in templates",
        "Memory efficiency: Be aware of memory overhead for large content"
    ]
    
    for pattern in optimization_patterns:
        print(f"  • {pattern}")
    
    print("Performance optimization and security testing completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    # Fallback demonstration
    print("Demonstrating performance and security concepts:")
    concepts = [
        "Performance benchmarking: Measure escaping performance across scenarios",
        "XSS resistance testing: Verify protection against common attack vectors",
        "Memory usage analysis: Understand overhead of different string types",
        "Security best practices: Patterns for safe content rendering",
        "Production optimization: Strategies for high-performance secure apps"
    ]
    
    for concept in concepts:
        print(f"  - {concept}")
    raise
```
