# math_verify

## Example 1: the same number written two ways is verified as equal

```python
# Example 1: parse() reads LaTeX and plain expressions; verify(gold, answer) compares them symbolically and numerically
from math_verify import parse, verify

gold = parse(r"$\frac{1}{2}$")
assert verify(gold, parse("0.5"))
assert verify(gold, parse(r"$\frac{2}{4}$"))
assert verify(parse(r"$\sqrt{2}$"), parse("$2^{1/2}$"))
print("Example 1:", gold[0], "== 0.5 and 2/4; sqrt(2) == 2^(1/2)")
```

## Example 2: a wrong answer is not verified

```python
# Example 2: a different value fails, and so does an empty parse
from math_verify import parse, verify

gold = parse("$42$")
assert not verify(gold, parse("$41$"))
assert not verify(gold, parse("no number here"))
print("Example 2: 41 and an empty answer both rejected against", gold[0])
```

## Example 3: extract the answer from a model's free-text reply

```python
# Example 3: the \boxed{} answer is pulled out of a sentence before comparison
from math_verify import LatexExtractionConfig, parse, verify

reply = r"First we add the terms, so the total is $\boxed{\frac{7}{3}}$, which is our answer."
answer = parse(reply, extraction_config=[LatexExtractionConfig()])
assert answer, reply
assert verify(parse(r"$\frac{7}{3}$"), answer)
print("Example 3: extracted", answer[0], "from the reply")
```

## Example 4: sets and intervals compare as sets

```python
# Example 4: element order inside a set does not matter; an interval must match its bounds and openness
from math_verify import parse, verify

assert verify(parse(r"$\{1, 2, 3\}$"), parse(r"$\{3, 1, 2\}$"))
assert verify(parse(r"$(0, 1]$"), parse(r"$(0, 1]$"))
assert not verify(parse(r"$(0, 1]$"), parse(r"$[0, 1]$"))
print("Example 4: sets compared order-free; (0, 1] differs from [0, 1]")
```

## Example 5: symbolic expressions are compared by value

```python
# Example 5: algebraically equal expressions verify; a different polynomial does not
from math_verify import parse, verify

assert verify(parse(r"$(x+1)^2$"), parse(r"$x^2 + 2x + 1$"))
assert not verify(parse(r"$(x+1)^2$"), parse(r"$x^2 + 1$"))
print("Example 5: (x+1)^2 == x^2 + 2x + 1, != x^2 + 1")
```
