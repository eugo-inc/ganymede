```python
import matplotlib.pyplot as plt
import matplotlib
import numpy as np

# Example 1: Basic plotting and figure management
print("matplotlib module imported successfully")

try:
    # Set appropriate backend for Jupyter
    matplotlib.use('Agg')  
    print("Example 1 - Basic plotting and figure management")
    
    # Create sample data
    x = np.linspace(0, 10, 100)
    y1 = np.sin(x) * np.exp(-x/5)
    y2 = np.cos(x) * np.exp(-x/5)
    
    # Create figure with subplots
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
    
    # Left plot: Multiple line styles
    ax1.plot(x, y1, 'b-', linewidth=2, label='sin(x)·e^(-x/5)')
    ax1.plot(x, y2, 'r--', linewidth=2, label='cos(x)·e^(-x/5)')
    ax1.fill_between(x, y1, alpha=0.3, color='blue')
    ax1.set_xlabel('X values')
    ax1.set_ylabel('Y values')
    ax1.set_title('Damped Oscillations')
    ax1.legend()
    ax1.grid(True, alpha=0.4)
    
    # Right plot: Parametric equations
    t = np.linspace(0, 4*np.pi, 200)
    x_param = t * np.cos(t)
    y_param = t * np.sin(t)
    
    ax2.plot(x_param, y_param, 'g-', linewidth=2)
    ax2.set_xlabel('X coordinate')
    ax2.set_ylabel('Y coordinate')
    ax2.set_title('Spiral Parametric Plot')
    ax2.set_aspect('equal')
    ax2.grid(True, alpha=0.4)
    
    plt.tight_layout()
    
    # Save figure (simulated)
    print(f"Figure size: {fig.get_size_inches()}")
    print(f"Number of axes: {len(fig.axes)}")
    print(f"Matplotlib version: {matplotlib.__version__}")
    print(f"Current backend: {matplotlib.get_backend()}")
    
    # Cleanup
    plt.close(fig)
    
    print("matplotlib example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"matplotlib example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: matplotlib functionality test 2

```python
import matplotlib.pyplot as plt
import numpy as np

# Example 2: Statistical plots and data visualization
print("matplotlib statistical plotting and visualization")

try:
    print("Example 2 - Statistical plots and data visualization")
    
    # Generate sample datasets
    np.random.seed(42)
    
    # Create figure with multiple statistical plots
    fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(14, 10))
    
    # 1. Enhanced Bar Chart with error bars
    categories = ['Method A', 'Method B', 'Method C', 'Method D', 'Method E']
    means = [85, 92, 78, 88, 95]
    errors = [3, 2.5, 4, 2.8, 1.8]
    colors = plt.cm.Set3(np.linspace(0, 1, len(categories)))
    
    bars = ax1.bar(categories, means, yerr=errors, capsize=5, 
                   color=colors, edgecolor='black', linewidth=1)
    ax1.set_ylabel('Performance Score')
    ax1.set_title('Method Comparison with Error Bars')
    ax1.set_ylim(70, 100)
    
    # Add value labels on bars
    for bar, mean, err in zip(bars, means, errors):
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height + err + 1,
                f'{mean}±{err}', ha='center', va='bottom', fontweight='bold')
    
    # 2. Multi-series Histogram
    data1 = np.random.normal(100, 15, 1000)
    data2 = np.random.normal(110, 12, 1000) 
    data3 = np.random.normal(95, 18, 1000)
    
    ax2.hist([data1, data2, data3], bins=30, alpha=0.7, 
             label=['Dataset 1', 'Dataset 2', 'Dataset 3'],
             color=['skyblue', 'lightgreen', 'salmon'])
    ax2.set_xlabel('Values')
    ax2.set_ylabel('Frequency')
    ax2.set_title('Overlapping Distributions')
    ax2.legend()
    ax2.grid(True, alpha=0.3)
    
    # 3. Box and Violin Plot Comparison
    violin_data = [np.random.normal(0, std, 200) for std in [1, 1.5, 2, 2.5]]
    violin_parts = ax3.violinplot(violin_data, positions=range(1, 5), 
                                  showmeans=True, showmedians=True)
    
    # Customize violin plot colors
    for pc, color in zip(violin_parts['bodies'], ['red', 'blue', 'green', 'orange']):
        pc.set_facecolor(color)
        pc.set_alpha(0.6)
    
    ax3.set_xlabel('Group')
    ax3.set_ylabel('Values')
    ax3.set_title('Distribution Shapes (Violin Plot)')
    ax3.set_xticks(range(1, 5))
    ax3.set_xticklabels([f'Group {i}' for i in range(1, 5)])
    
    # 4. Correlation Heatmap
    size = 8
    correlation_matrix = np.random.rand(size, size)
    correlation_matrix = (correlation_matrix + correlation_matrix.T) / 2  # Make symmetric
    np.fill_diagonal(correlation_matrix, 1)  # Set diagonal to 1
    
    im = ax4.imshow(correlation_matrix, cmap='RdBu_r', vmin=-1, vmax=1)
    
    # Add text annotations
    for i in range(size):
        for j in range(size):
            text = ax4.text(j, i, f'{correlation_matrix[i, j]:.2f}',
                           ha="center", va="center", color="black", fontsize=8)
    
    ax4.set_title('Correlation Matrix Heatmap')
    ax4.set_xticks(range(size))
    ax4.set_yticks(range(size))
    ax4.set_xticklabels([f'Var{i+1}' for i in range(size)])
    ax4.set_yticklabels([f'Var{i+1}' for i in range(size)])
    
    # Add colorbar
    plt.colorbar(im, ax=ax4)
    
    plt.tight_layout()
    
    print(f"Bar chart categories: {len(categories)}")
    print(f"Histogram datasets: 3 with {len(data1)} points each")
    print(f"Violin plot groups: {len(violin_data)}")
    print(f"Correlation matrix size: {size}x{size}")
    
    plt.close(fig)
    
    print("matplotlib example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"matplotlib example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: matplotlib functionality test 3

```python
import matplotlib.pyplot as plt
import numpy as np

# Example 3: Advanced scatter plots and 3D-style visualizations
print("matplotlib advanced scatter and 3D-style visualization")

try:
    print("Example 3 - Advanced scatter plots and 3D-style visualizations")
    
    # Generate complex dataset
    np.random.seed(123)
    n_points = 300
    
    # Create figure with multiple advanced visualizations
    fig = plt.figure(figsize=(16, 12))
    
    # 1. Multi-dimensional scatter plot
    ax1 = plt.subplot(2, 3, 1)
    
    # Generate correlated data
    x = np.random.randn(n_points)
    y = 2 * x + np.random.randn(n_points) * 0.5
    z = x + y + np.random.randn(n_points) * 0.3  # Third dimension encoded as color
    sizes = 1000 * np.random.rand(n_points) + 50  # Fourth dimension as size
    
    scatter = ax1.scatter(x, y, c=z, s=sizes, alpha=0.6, cmap='viridis', edgecolors='black')
    ax1.set_xlabel('X values')
    ax1.set_ylabel('Y values')
    ax1.set_title('4D Scatter Plot\n(X, Y, Color=Z, Size=W)')
    
    # Add regression line
    m, b = np.polyfit(x, y, 1)
    ax1.plot(x, m*x + b, 'r--', linewidth=2, alpha=0.8)
    
    plt.colorbar(scatter, ax=ax1, label='Z values')
    
    # 2. Bubble chart with categories
    ax2 = plt.subplot(2, 3, 2)
    
    categories = ['Tech', 'Finance', 'Healthcare', 'Energy', 'Retail']
    x_bubble = np.random.rand(len(categories)) * 100
    y_bubble = np.random.rand(len(categories)) * 100
    bubble_sizes = np.random.rand(len(categories)) * 2000 + 500
    colors = ['red', 'blue', 'green', 'orange', 'purple']
    
    for i, (cat, color) in enumerate(zip(categories, colors)):
        ax2.scatter(x_bubble[i], y_bubble[i], s=bubble_sizes[i], 
                   c=color, alpha=0.6, label=cat, edgecolors='black')
        ax2.text(x_bubble[i], y_bubble[i], cat, ha='center', va='center', fontweight='bold')
    
    ax2.set_xlabel('Market Cap (B$)')
    ax2.set_ylabel('Growth Rate (%)')
    ax2.set_title('Industry Bubble Chart')
    ax2.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    
    # 3. Contour plot (3D-style)
    ax3 = plt.subplot(2, 3, 3)
    
    x_cont = np.linspace(-3, 3, 50)
    y_cont = np.linspace(-3, 3, 50)
    X, Y = np.meshgrid(x_cont, y_cont)
    Z = np.exp(-(X**2 + Y**2)) * np.cos(2*X) * np.sin(2*Y)
    
    contour = ax3.contourf(X, Y, Z, levels=20, cmap='RdBu_r', alpha=0.8)
    contour_lines = ax3.contour(X, Y, Z, levels=10, colors='black', alpha=0.5, linewidths=0.5)
    ax3.clabel(contour_lines, inline=True, fontsize=8)
    ax3.set_xlabel('X coordinate')
    ax3.set_ylabel('Y coordinate')
    ax3.set_title('Contour Plot with Labels')
    plt.colorbar(contour, ax=ax3)
    
    # 4. Polar scatter plot
    ax4 = plt.subplot(2, 3, 4, projection='polar')
    
    theta = np.random.uniform(0, 2*np.pi, 100)
    r = np.random.uniform(0, 1, 100)
    colors_polar = theta
    
    ax4.scatter(theta, r, c=colors_polar, cmap='hsv', alpha=0.7)
    ax4.set_title('Polar Coordinate Scatter')
    
    # 5. Stream plot (vector field)
    ax5 = plt.subplot(2, 3, 5)
    
    x_stream = np.linspace(-2, 2, 20)
    y_stream = np.linspace(-2, 2, 20)
    X_stream, Y_stream = np.meshgrid(x_stream, y_stream)
    U = -Y_stream  # Velocity in X direction
    V = X_stream   # Velocity in Y direction
    
    ax5.streamplot(X_stream, Y_stream, U, V, density=1.5, arrowsize=1.2, 
                   arrowstyle='->', color=np.sqrt(U**2 + V**2), cmap='plasma')
    ax5.set_xlabel('X')
    ax5.set_ylabel('Y')
    ax5.set_title('Vector Field Visualization')
    ax5.set_aspect('equal')
    
    # 6. Hexbin plot (2D histogram)
    ax6 = plt.subplot(2, 3, 6)
    
    x_hex = np.random.normal(0, 1, 2000)
    y_hex = x_hex + np.random.normal(0, 0.5, 2000)
    
    hb = ax6.hexbin(x_hex, y_hex, gridsize=20, cmap='Blues', mincnt=1)
    ax6.set_xlabel('X values')
    ax6.set_ylabel('Y values')
    ax6.set_title('Hexagonal Binning Plot')
    plt.colorbar(hb, ax=ax6, label='Count')
    
    plt.tight_layout()
    
    print(f"Multi-dimensional scatter: {n_points} points with 4 dimensions")
    print(f"Bubble chart industries: {len(categories)}")
    print(f"Contour plot resolution: {len(x_cont)}x{len(y_cont)}")
    print(f"Polar scatter points: 100")
    print(f"Stream plot resolution: {len(x_stream)}x{len(y_stream)}")
    print(f"Hexbin data points: 2000")
    
    plt.close(fig)
    
    print("matplotlib example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"matplotlib example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: matplotlib functionality test 4

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import numpy as np
import datetime

# Example 4: Time series and financial visualization
print("matplotlib time series and financial visualization")

try:
    print("Example 4 - Time series and financial visualization")
    
    # Generate time series data
    start_date = datetime.date(2023, 1, 1)
    end_date = datetime.date(2024, 12, 31) 
    dates = [start_date + datetime.timedelta(days=i) for i in range((end_date - start_date).days)]
    
    # Create comprehensive financial dashboard
    fig = plt.figure(figsize=(16, 12))
    
    # 1. Main price chart with volume
    ax1 = plt.subplot(3, 2, (1, 2))  # Span two columns
    
    # Simulate realistic stock price with trends
    np.random.seed(42)
    price = 100
    prices = []
    volumes = []
    
    for i, date in enumerate(dates):
        # Add seasonal and trend effects
        seasonal = 5 * np.sin(2 * np.pi * i / 365.25)  # Annual cycle
        trend = 0.02 * i  # Long-term upward trend
        volatility = np.random.normal(0, 2)
        
        price = max(price + seasonal + trend + volatility, 10)
        prices.append(price)
        volumes.append(np.random.randint(1000000, 5000000))
    
    # Price chart
    ax1.plot(dates, prices, linewidth=1.5, color='blue', label='Stock Price')
    
    # Moving averages
    window_short = 20
    window_long = 50
    
    if len(prices) >= window_long:
        ma_short = np.convolve(prices, np.ones(window_short)/window_short, mode='same')
        ma_long = np.convolve(prices, np.ones(window_long)/window_long, mode='same')
        
        ax1.plot(dates, ma_short, '--', color='orange', linewidth=2, label=f'{window_short}-day MA')
        ax1.plot(dates, ma_long, '--', color='red', linewidth=2, label=f'{window_long}-day MA')
    
    ax1.set_ylabel('Price ($)')
    ax1.set_title('Stock Price with Moving Averages')
    ax1.legend()
    ax1.grid(True, alpha=0.3)
    ax1.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
    ax1.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
    
    # 2. Volume chart
    ax2 = plt.subplot(3, 2, (3, 4))  # Span two columns
    
    # Color volume bars by price change
    price_changes = np.diff(prices, prepend=prices[0])
    colors = ['green' if change >= 0 else 'red' for change in price_changes]
    
    ax2.bar(dates, volumes, color=colors, alpha=0.6, width=1)
    ax2.set_ylabel('Volume')
    ax2.set_title('Trading Volume (Green=Up, Red=Down)')
    ax2.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
    ax2.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
    
    # 3. Returns distribution
    ax3 = plt.subplot(3, 2, 5)
    
    daily_returns = np.array(price_changes) / np.array(prices) * 100  # Percentage returns
    
    ax3.hist(daily_returns, bins=50, alpha=0.7, color='skyblue', edgecolor='black', density=True)
    ax3.axvline(daily_returns.mean(), color='red', linestyle='--', 
                label=f'Mean: {daily_returns.mean():.2f}%')
    ax3.axvline(0, color='black', linestyle='-', alpha=0.5)
    ax3.set_xlabel('Daily Return (%)')
    ax3.set_ylabel('Density')
    ax3.set_title('Returns Distribution')
    ax3.legend()
    ax3.grid(True, alpha=0.3)
    
    # 4. Rolling volatility
    ax4 = plt.subplot(3, 2, 6)
    
    # Calculate 30-day rolling volatility
    volatility_window = 30
    if len(daily_returns) >= volatility_window:
        rolling_vol = []
        vol_dates = []
        
        for i in range(volatility_window, len(daily_returns)):
            vol = np.std(daily_returns[i-volatility_window:i]) * np.sqrt(252)  # Annualized
            rolling_vol.append(vol)
            vol_dates.append(dates[i])
        
        ax4.plot(vol_dates, rolling_vol, color='purple', linewidth=2)
        ax4.fill_between(vol_dates, rolling_vol, alpha=0.3, color='purple')
        ax4.set_ylabel('Volatility (%)')
        ax4.set_xlabel('Date')
        ax4.set_title(f'{volatility_window}-Day Rolling Volatility')
        ax4.grid(True, alpha=0.3)
        ax4.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
        ax4.xaxis.set_major_locator(mdates.MonthLocator(interval=6))
    
    # Format all x-axis labels
    for ax in [ax1, ax2, ax4]:
        plt.setp(ax.xaxis.get_majorticklabels(), rotation=45)
    
    plt.tight_layout()
    
    print(f"Time series length: {len(dates)} days")
    print(f"Price range: ${min(prices):.2f} - ${max(prices):.2f}")
    print(f"Total return: {((prices[-1] / prices[0]) - 1) * 100:.1f}%")
    print(f"Average daily return: {daily_returns.mean():.3f}%")
    print(f"Daily volatility: {daily_returns.std():.3f}%")
    print(f"Annualized volatility: {daily_returns.std() * np.sqrt(252):.1f}%")
    
    plt.close(fig)
    
    print("matplotlib example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"matplotlib example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

```python
import matplotlib.pyplot as plt
import matplotlib.style as mplstyle
import matplotlib.patches as patches
import numpy as np

# Example 5: Styling, annotations, and publication-quality plots
print("matplotlib styling, annotations, and publication-quality visualization")

try:
    print("Example 5 - Styling, annotations, and publication-quality plots")
    
    # Set publication-quality parameters
    plt.rcParams.update({
        'figure.dpi': 150,
        'savefig.dpi': 300,
        'font.size': 11,
        'axes.titlesize': 14,
        'axes.labelsize': 12,
        'xtick.labelsize': 10,
        'ytick.labelsize': 10,
        'legend.fontsize': 10,
        'lines.linewidth': 2,
        'axes.grid': True,
        'grid.alpha': 0.3
    })
    
    # Create comprehensive publication figure
    fig = plt.figure(figsize=(16, 10))
    
    # 1. Scientific plot with annotations
    ax1 = plt.subplot(2, 3, 1)
    
    # Generate experimental-like data
    x = np.linspace(0, 10, 100)
    y_theory = 2 * np.exp(-x/3) * np.cos(x)
    y_exp = y_theory + 0.3 * np.random.randn(100)
    
    ax1.plot(x, y_theory, 'r-', linewidth=3, label='Theoretical Model')
    ax1.scatter(x[::5], y_exp[::5], alpha=0.7, s=50, color='blue', 
                label='Experimental Data', zorder=5)
    
    # Add error bars
    errors = 0.3 * np.ones_like(x[::5])
    ax1.errorbar(x[::5], y_exp[::5], yerr=errors, fmt='none', 
                 color='blue', alpha=0.3, capsize=3)
    
    # Annotations
    ax1.annotate('Peak amplitude', xy=(1.5, y_theory[15]), xytext=(3, 1.5),
                arrowprops=dict(arrowstyle='->', color='black', lw=2),
                fontsize=12, ha='center',
                bbox=dict(boxstyle="round,pad=0.3", facecolor="yellow", alpha=0.7))
    
    ax1.set_xlabel('Time (s)')
    ax1.set_ylabel('Amplitude (V)')
    ax1.set_title('Damped Oscillation Analysis')
    ax1.legend()
    
    # 2. Multi-axis plot
    ax2 = plt.subplot(2, 3, 2)
    ax2_twin = ax2.twinx()
    
    # Primary axis - Temperature
    days = np.arange(1, 31)
    temperature = 20 + 10 * np.sin(days/30 * 2*np.pi) + 2 * np.random.randn(30)
    rainfall = np.maximum(0, 5 + 3 * np.sin(days/30 * 2*np.pi + np.pi/2) + np.random.randn(30))
    
    line1 = ax2.plot(days, temperature, 'r-', linewidth=2, label='Temperature')
    ax2.set_xlabel('Day of Month')
    ax2.set_ylabel('Temperature (°C)', color='red')
    ax2.tick_params(axis='y', labelcolor='red')
    
    # Secondary axis - Rainfall
    bars = ax2_twin.bar(days, rainfall, alpha=0.6, color='blue', label='Rainfall')
    ax2_twin.set_ylabel('Rainfall (mm)', color='blue')
    ax2_twin.tick_params(axis='y', labelcolor='blue')
    
    ax2.set_title('Weather Data (Dual-Axis)')
    
    # Combined legend
    lines1, labels1 = ax2.get_legend_handles_labels()
    lines2, labels2 = ax2_twin.get_legend_handles_labels()
    ax2.legend(lines1 + lines2, labels1 + labels2, loc='upper right')
    
    # 3. Custom shapes and patches
    ax3 = plt.subplot(2, 3, 3)
    
    # Add various geometric shapes
    circle = patches.Circle((0.3, 0.3), 0.15, color='red', alpha=0.7)
    rectangle = patches.Rectangle((0.1, 0.6), 0.3, 0.2, color='blue', alpha=0.7)
    wedge = patches.Wedge((0.7, 0.3), 0.2, 0, 120, color='green', alpha=0.7)
    triangle = patches.Polygon([(0.6, 0.6), (0.8, 0.6), (0.7, 0.85)], 
                              color='orange', alpha=0.7)
    
    ax3.add_patch(circle)
    ax3.add_patch(rectangle)
    ax3.add_patch(wedge)
    ax3.add_patch(triangle)
    
    # Add text annotations
    ax3.text(0.3, 0.3, 'Circle', ha='center', va='center', fontweight='bold')
    ax3.text(0.25, 0.7, 'Rectangle', ha='center', va='center', fontweight='bold')
    ax3.text(0.7, 0.3, 'Wedge', ha='center', va='center', fontweight='bold')
    ax3.text(0.7, 0.7, 'Triangle', ha='center', va='center', fontweight='bold')
    
    ax3.set_xlim(0, 1)
    ax3.set_ylim(0, 1)
    ax3.set_title('Custom Shapes and Patches')
    ax3.set_aspect('equal')
    
    # 4. Logarithmic scaling
    ax4 = plt.subplot(2, 3, 4)
    
    # Generate exponential data
    x_log = np.logspace(0, 3, 50)  # 10^0 to 10^3
    y1_log = x_log**2
    y2_log = 1000 * np.exp(-x_log/500)
    
    ax4.loglog(x_log, y1_log, 'b-', linewidth=2, label='Power Law (x²)')
    ax4.loglog(x_log, y2_log, 'r--', linewidth=2, label='Exponential Decay')
    
    ax4.set_xlabel('X (log scale)')
    ax4.set_ylabel('Y (log scale)')
    ax4.set_title('Log-Log Plot')
    ax4.legend()
    ax4.grid(True, which="both", ls="-", alpha=0.2)
    
    # 5. Pie chart with customization
    ax5 = plt.subplot(2, 3, 5)
    
    sizes = [30, 25, 20, 15, 10]
    labels = ['Category A', 'Category B', 'Category C', 'Category D', 'Category E']
    colors = plt.cm.Set3(np.linspace(0, 1, len(sizes)))
    explode = (0.1, 0, 0.1, 0, 0)  # Explode certain slices
    
    wedges, texts, autotexts = ax5.pie(sizes, labels=labels, colors=colors, 
                                       autopct='%1.1f%%', startangle=90,
                                       explode=explode, shadow=True)
    
    # Customize text
    for autotext in autotexts:
        autotext.set_color('white')
        autotext.set_fontweight('bold')
    
    ax5.set_title('Enhanced Pie Chart')
    
    # 6. Inset plots
    ax6 = plt.subplot(2, 3, 6)
    
    # Main plot
    x_main = np.linspace(0, 20, 200)
    y_main = np.sin(x_main) * np.exp(-x_main/10)
    
    ax6.plot(x_main, y_main, 'b-', linewidth=2)
    ax6.set_xlabel('X values')
    ax6.set_ylabel('Y values')
    ax6.set_title('Main Plot with Inset')
    
    # Create inset
    from mpl_toolkits.axes_grid1.inset_locator import inset_axes
    axins = inset_axes(ax6, width="40%", height="40%", loc='upper right')
    
    # Zoomed region
    x_zoom = x_main[(x_main >= 2) & (x_main <= 8)]
    y_zoom = y_main[(x_main >= 2) & (x_main <= 8)]
    axins.plot(x_zoom, y_zoom, 'r-', linewidth=2)
    axins.set_xlim(2, 8)
    axins.set_ylim(y_zoom.min(), y_zoom.max())
    axins.grid(True, alpha=0.3)
    
    # Mark the region of zoom
    ax6.indicate_inset_zoom(axins, edgecolor="black")
    
    plt.tight_layout()
    
    # Summary statistics
    print(f"Available styles: {len(plt.style.available)}")
    print(f"Figure contains {len(fig.axes)} main axes")
    print(f"Publication DPI: {plt.rcParams['savefig.dpi']}")
    print(f"Log-log data range: X[{x_log.min():.1f}, {x_log.max():.1f}]")
    print(f"Pie chart categories: {len(labels)}")
    print(f"Geometric shapes: Circle, Rectangle, Wedge, Triangle")
    
    plt.close(fig)
    
    print("matplotlib example 5 completed successfully")
    
    # Reset rcParams
    plt.rcdefaults()
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"matplotlib example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
print(f"\nMatplotlib capabilities demonstrated:")
print("- Basic and parametric plotting with figure management")
print("- Statistical visualization (histograms, violin plots, heatmaps)")
print("- Advanced scatter plots and 3D-style visualizations")
print("- Time series and financial data analysis")
print("- Publication-quality styling with annotations and custom shapes")
```
