```python
import mccabe
import ast
import textwrap

# Example 1: Basic cyclomatic complexity analysis
print("mccabe module imported successfully")

try:
    print("Example 1 - Basic cyclomatic complexity analysis")
    
    # Create sample Python code with different complexity levels
    simple_function = textwrap.dedent("""
    def simple_function(x):
        return x * 2
    """)
    
    moderate_function = textwrap.dedent("""
    def moderate_function(x, y):
        if x > 0:
            if y > 0:
                return x + y
            else:
                return x - y
        else:
            return 0
    """)
    
    complex_function = textwrap.dedent("""
    def complex_function(data):
        result = []
        for item in data:
            if item is None:
                continue
            elif isinstance(item, str):
                if len(item) > 0:
                    result.append(item.upper())
                else:
                    result.append("EMPTY")
            elif isinstance(item, int):
                if item > 0:
                    result.append(item * 2)
                elif item < 0:
                    result.append(abs(item))
                else:
                    result.append(1)
            else:
                result.append(str(item))
        return result
    """)
    
    # Analyze each function
    functions = [
        ("Simple Function", simple_function),
        ("Moderate Function", moderate_function),
        ("Complex Function", complex_function)
    ]
    
    print("Cyclomatic Complexity Analysis:")
    for name, code in functions:
        # Calculate complexity using mccabe.get_code_complexity
        complexity_result = mccabe.get_code_complexity(code, threshold=10)
        
        # Parse to get function name for PathGraph analysis
        tree = ast.parse(code)
        func_node = None
        for node in ast.walk(tree):
            if isinstance(node, ast.FunctionDef):
                func_node = node
                break
        
        if func_node:
            # Use PathGraph for detailed analysis
            pathgraph = mccabe.PathGraph(func_node.name, func_node, func_node.lineno)
            complexity = pathgraph.complexity()
            
            print(f"  {name}: {complexity}")
            
            # Provide interpretation
            if complexity <= 2:
                interpretation = "Low complexity - easy to understand and test"
            elif complexity <= 5:
                interpretation = "Moderate complexity - acceptable for most cases"
            elif complexity <= 10:
                interpretation = "High complexity - consider refactoring"
            else:
                interpretation = "Very high complexity - refactoring strongly recommended"
            
            print(f"    Interpretation: {interpretation}")
    
    print("mccabe example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"mccabe example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: mccabe functionality test 2

```python
import mccabe
import ast
import textwrap

# Example 2: PathGraph analysis and control flow visualization
print("mccabe PathGraph analysis and control flow visualization")

try:
    print("Example 2 - PathGraph analysis and control flow visualization")
    
    # Create a function with complex control flow
    complex_control_flow = textwrap.dedent("""
    def analyze_data(data_list, threshold=10, mode='strict'):
        if not data_list:
            return None
        
        results = []
        errors = 0
        
        for i, item in enumerate(data_list):
            try:
                if isinstance(item, (int, float)):
                    if mode == 'strict':
                        if item > threshold:
                            results.append(item * 2)
                        elif item < -threshold:
                            results.append(abs(item))
                        else:
                            continue
                    elif mode == 'lenient':
                        if item != 0:
                            results.append(item)
                    else:
                        results.append(item)
                elif isinstance(item, str):
                    if item.isdigit():
                        num_val = float(item)
                        if num_val > threshold:
                            results.append(num_val)
                    else:
                        errors += 1
                else:
                    errors += 1
            except Exception as e:
                errors += 1
                continue
        
        if errors > len(data_list) // 2:
            return None
        
        return {
            'results': results,
            'errors': errors,
            'success_rate': (len(data_list) - errors) / len(data_list)
        }
    """)
    
    # Parse the function
    tree = ast.parse(complex_control_flow)
    func_node = None
    for node in ast.walk(tree):
        if isinstance(node, ast.FunctionDef):
            func_node = node
            break
    
    if func_node:
        # Create PathGraph for detailed analysis (corrected constructor)
        pathgraph = mccabe.PathGraph(func_node.name, func_node, func_node.lineno)
        
        print(f"Function: {func_node.name}")
        print(f"Cyclomatic Complexity: {pathgraph.complexity()}")
        
        # Analyze the graph structure
        print(f"PathGraph Analysis:")
        print(f"  Number of nodes: {len(pathgraph.nodes)}")
        
        # Get complexity breakdown
        complexity = pathgraph.complexity()
        print(f"  Base complexity: 1")
        print(f"  Additional decision points: {complexity - 1}")
        
        # Calculate relative complexity metrics
        lines_of_code = len([line for line in complex_control_flow.split('\n') if line.strip()])
        complexity_density = complexity / lines_of_code if lines_of_code > 0 else 0
        
        print(f"  Lines of code: {lines_of_code}")
        print(f"  Complexity density: {complexity_density:.3f} (complexity per line)")
        
        # Provide recommendations based on complexity
        print(f"Code Quality Assessment:")
        if complexity <= 5:
            assessment = "Good - low complexity, easy to test"
            recommendation = "No immediate action needed"
        elif complexity <= 10:
            assessment = "Moderate - acceptable but monitor"
            recommendation = "Consider breaking into smaller functions if possible"
        elif complexity <= 15:
            assessment = "High - difficult to test and maintain"
            recommendation = "Refactor into multiple smaller functions"
        else:
            assessment = "Very High - critical refactoring needed"
            recommendation = "Break into multiple functions immediately"
        
        print(f"  Assessment: {assessment}")
        print(f"  Recommendation: {recommendation}")
        
        # Create a simple graph representation
        print(f"Control Flow Summary:")
        decision_keywords = ['if', 'elif', 'for', 'while', 'try', 'except', 'and', 'or']
        code_lines = complex_control_flow.split('\n')
        decision_lines = []
        
        for i, line in enumerate(code_lines, 1):
            line_stripped = line.strip().lower()
            for keyword in decision_keywords:
                if line_stripped.startswith(keyword + ' ') or ' ' + keyword + ' ' in line_stripped:
                    decision_lines.append(f"  Line {i}: {line.strip()}")
                    break
        
        print(f"  Decision points identified: {len(decision_lines)}")
        for decision in decision_lines[:5]:  # Show first 5
            print(decision)
        if len(decision_lines) > 5:
            print(f"  ... and {len(decision_lines) - 5} more")
    
    print("mccabe example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"mccabe example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: mccabe functionality test 3

```python
import mccabe
import ast
import textwrap

# Example 3: Multiple functions and class complexity analysis
print("mccabe multiple functions and class complexity analysis")

try:
    print("Example 3 - Multiple functions and class complexity analysis")
    
    # Create a module with multiple functions and a class
    module_code = textwrap.dedent("""
    class DataProcessor:
        def __init__(self, config=None):
            self.config = config or {}
            
        def simple_method(self, x):
            return x * 2
        
        def validation_method(self, data):
            if not data:
                return False
            
            if isinstance(data, str):
                return len(data) > 0
            elif isinstance(data, (list, tuple)):
                return len(data) > 0 and all(item is not None for item in data)
            elif isinstance(data, dict):
                return len(data) > 0 and all(v is not None for v in data.values())
            else:
                return data is not None
        
        def complex_processing_method(self, items, filters=None):
            if not items:
                return []
            
            results = []
            filters = filters or {}
            
            for item in items:
                try:
                    if 'type_filter' in filters:
                        if not isinstance(item, filters['type_filter']):
                            continue
                    
                    if 'min_value' in filters:
                        if hasattr(item, '__lt__') and item < filters['min_value']:
                            continue
                    
                    if 'max_value' in filters:
                        if hasattr(item, '__gt__') and item > filters['max_value']:
                            continue
                    
                    if 'custom_filter' in filters:
                        filter_func = filters['custom_filter']
                        if callable(filter_func):
                            if not filter_func(item):
                                continue
                        else:
                            raise ValueError("Custom filter must be callable")
                    
                    # Processing logic
                    if isinstance(item, str):
                        if item.isdigit():
                            processed = float(item) * 2
                        else:
                            processed = item.upper()
                    elif isinstance(item, (int, float)):
                        if item > 0:
                            processed = item ** 0.5
                        elif item < 0:
                            processed = abs(item)
                        else:
                            processed = 1
                    else:
                        processed = str(item)
                    
                    results.append(processed)
                    
                except Exception as e:
                    if self.config.get('strict_mode', False):
                        raise
                    else:
                        continue
            
            return results
    
    def utility_function(a, b, operation='add'):
        if operation == 'add':
            return a + b
        elif operation == 'subtract':
            return a - b
        elif operation == 'multiply':
            return a * b
        elif operation == 'divide':
            if b != 0:
                return a / b
            else:
                raise ValueError("Division by zero")
        else:
            raise ValueError("Unknown operation")
    
    def complex_utility(data_dict, transformations=None):
        if not isinstance(data_dict, dict):
            raise TypeError("Input must be a dictionary")
        
        results = {}
        transformations = transformations or {}
        
        for key, value in data_dict.items():
            try:
                if key in transformations:
                    transform = transformations[key]
                    if callable(transform):
                        results[key] = transform(value)
                    else:
                        if transform == 'upper' and isinstance(value, str):
                            results[key] = value.upper()
                        elif transform == 'double' and isinstance(value, (int, float)):
                            results[key] = value * 2
                        elif transform == 'abs' and isinstance(value, (int, float)):
                            results[key] = abs(value)
                        else:
                            results[key] = value
                else:
                    if isinstance(value, list):
                        results[key] = [item for item in value if item is not None]
                    elif isinstance(value, str):
                        results[key] = value.strip()
                    elif isinstance(value, (int, float)):
                        if value < 0:
                            results[key] = 0
                        else:
                            results[key] = value
                    else:
                        results[key] = value
            except Exception:
                results[key] = None
        
        return results
    """)
    
    # Parse the module
    tree = ast.parse(module_code)
    
    # Find all functions and methods
    functions_and_methods = []
    
    for node in ast.walk(tree):
        if isinstance(node, ast.FunctionDef):
            functions_and_methods.append({
                'name': node.name,
                'node': node,
                'type': 'method' if node.name != node.name or any(isinstance(parent, ast.ClassDef) 
                                                                  for parent in ast.walk(tree)) else 'function'
            })
    
    print("Complexity Analysis Results:")
    print("=" * 50)
    
    total_complexity = 0
    complexity_distribution = {'low': 0, 'moderate': 0, 'high': 0, 'very_high': 0}
    
    for item in functions_and_methods:
        name = item['name']
        node = item['node']
        item_type = item['type']
        
        # Calculate complexity using PathGraph
        pathgraph = mccabe.PathGraph(node.name, node, node.lineno)
        complexity = pathgraph.complexity()
        total_complexity += complexity
        
        # Categorize complexity
        if complexity <= 2:
            category = 'low'
            status = "✓"
        elif complexity <= 5:
            category = 'moderate'
            status = "⚠"
        elif complexity <= 10:
            category = 'high'
            status = "⚠"
        else:
            category = 'very_high'
            status = "✗"
        
        complexity_distribution[category] += 1
        
        print(f"{status} {item_type.title()}: {name}")
        print(f"    Complexity: {complexity}")
        
        # Calculate lines of code for this function (approximate)
        func_lines = len([line for line in module_code.split('\n')[node.lineno-1:] 
                         if line.strip() and not line.strip().startswith('def ') 
                         and not line.strip().startswith('class ')])
        if func_lines > 25:  # Limit to reasonable estimate
            func_lines = 25
        
        print(f"    Estimated lines: {func_lines}")
        if func_lines > 0:
            print(f"    Complexity/LOC ratio: {complexity/func_lines:.3f}")
        print()
    
    # Overall statistics
    print("Module Summary:")
    print("=" * 50)
    print(f"Total functions/methods analyzed: {len(functions_and_methods)}")
    print(f"Total cyclomatic complexity: {total_complexity}")
    print(f"Average complexity: {total_complexity / len(functions_and_methods):.2f}")
    
    print(f"\nComplexity Distribution:")
    print(f"  Low (≤2):       {complexity_distribution['low']} functions")
    print(f"  Moderate (3-5): {complexity_distribution['moderate']} functions")
    print(f"  High (6-10):    {complexity_distribution['high']} functions")
    print(f"  Very High (>10): {complexity_distribution['very_high']} functions")
    
    # Module quality assessment
    high_complexity_count = complexity_distribution['high'] + complexity_distribution['very_high']
    total_count = len(functions_and_methods)
    high_complexity_ratio = high_complexity_count / total_count if total_count > 0 else 0
    
    print(f"\nModule Quality Assessment:")
    if high_complexity_ratio <= 0.1:
        quality = "Excellent"
    elif high_complexity_ratio <= 0.2:
        quality = "Good"
    elif high_complexity_ratio <= 0.3:
        quality = "Fair"
    else:
        quality = "Needs Improvement"
    
    print(f"  Overall Quality: {quality}")
    print(f"  High complexity functions: {high_complexity_ratio:.1%}")
    
    if high_complexity_count > 0:
        print(f"  Recommendation: Focus on refactoring the {high_complexity_count} high-complexity functions")
    else:
        print(f"  Recommendation: Code complexity is well-managed")
    
    print("mccabe example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"mccabe example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: mccabe functionality test 4

```python
import mccabe
import ast
import textwrap
import os

# Example 4: File-based analysis and threshold checking
print("mccabe file-based analysis and threshold checking")

try:
    print("Example 4 - File-based analysis and threshold checking")
    
    # Create multiple sample Python files in memory
    sample_files = {
        'simple_module.py': textwrap.dedent("""
        def add(a, b):
            return a + b
        
        def multiply(a, b):
            return a * b
        
        def greet(name):
            if name:
                return f"Hello, {name}!"
            else:
                return "Hello, stranger!"
        """),
        
        'moderate_module.py': textwrap.dedent("""
        def process_list(items, filters=None):
            if not items:
                return []
            
            result = []
            filters = filters or {}
            
            for item in items:
                if 'min_val' in filters and item < filters['min_val']:
                    continue
                elif 'max_val' in filters and item > filters['max_val']:
                    continue
                else:
                    if isinstance(item, str):
                        result.append(item.upper())
                    else:
                        result.append(item * 2)
            
            return result
        """),
        
        'complex_module.py': textwrap.dedent("""
        def complex_data_processor(data, config=None):
            if not data:
                raise ValueError("Data cannot be empty")
            
            config = config or {}
            results = []
            errors = []
            
            for i, record in enumerate(data):
                try:
                    if not isinstance(record, dict):
                        errors.append(f"Record {i} is not a dictionary")
                        continue
                    
                    # Validate required fields
                    if 'id' not in record:
                        errors.append(f"Record {i} missing 'id' field")
                        continue
                    
                    if 'type' not in record:
                        record['type'] = 'unknown'
                    
                    # Process based on type
                    if record['type'] == 'numeric':
                        if 'value' in record:
                            if isinstance(record['value'], (int, float)):
                                if record['value'] > 0:
                                    processed_value = record['value'] ** 0.5
                                elif record['value'] < 0:
                                    processed_value = abs(record['value'])
                                else:
                                    processed_value = 1
                            else:
                                try:
                                    processed_value = float(record['value'])
                                except ValueError:
                                    errors.append(f"Record {i} has invalid numeric value")
                                    continue
                        else:
                            processed_value = 0
                        
                        result = {
                            'id': record['id'],
                            'type': 'numeric',
                            'value': processed_value,
                            'original': record.get('value', 0)
                        }
                    
                    elif record['type'] == 'text':
                        text_value = record.get('text', '')
                        if isinstance(text_value, str):
                            if config.get('uppercase', False):
                                processed_text = text_value.upper()
                            elif config.get('lowercase', False):
                                processed_text = text_value.lower()
                            else:
                                processed_text = text_value.strip()
                        else:
                            processed_text = str(text_value)
                        
                        result = {
                            'id': record['id'],
                            'type': 'text',
                            'text': processed_text,
                            'length': len(processed_text)
                        }
                    
                    elif record['type'] == 'list':
                        list_value = record.get('items', [])
                        if isinstance(list_value, list):
                            filtered_items = []
                            for item in list_value:
                                if item is not None:
                                    if config.get('filter_empty', True):
                                        if item != '':
                                            filtered_items.append(item)
                                    else:
                                        filtered_items.append(item)
                        else:
                            filtered_items = [list_value] if list_value is not None else []
                        
                        result = {
                            'id': record['id'],
                            'type': 'list',
                            'items': filtered_items,
                            'count': len(filtered_items)
                        }
                    
                    else:
                        result = {
                            'id': record['id'],
                            'type': 'unknown',
                            'raw_data': record
                        }
                    
                    # Apply global filters
                    if config.get('require_id_prefix'):
                        prefix = config['require_id_prefix']
                        if not str(result['id']).startswith(prefix):
                            errors.append(f"Record {i} ID doesn't start with required prefix")
                            continue
                    
                    results.append(result)
                
                except Exception as e:
                    errors.append(f"Record {i} processing error: {str(e)}")
                    continue
            
            return {
                'results': results,
                'errors': errors,
                'success_count': len(results),
                'error_count': len(errors),
                'total_processed': len(data)
            }
        """)
    }
    
    # Analyze each file for complexity
    print("File-by-File Complexity Analysis:")
    print("=" * 60)
    
    file_results = {}
    complexity_threshold = 7  # Standard threshold for moderate complexity
    
    for filename, code in sample_files.items():
        print(f"\nAnalyzing: {filename}")
        print("-" * 40)
        
        # Parse the file
        tree = ast.parse(code, filename=filename)
        
        # Find all functions
        functions = []
        for node in ast.walk(tree):
            if isinstance(node, ast.FunctionDef):
                functions.append(node)
        
        file_complexities = []
        violations = []
        
        for func_node in functions:
            # Use PathGraph for complexity calculation
            pathgraph = mccabe.PathGraph(func_node.name, func_node, func_node.lineno)
            complexity = pathgraph.complexity()
            file_complexities.append(complexity)
            
            status_symbol = "✓" if complexity <= complexity_threshold else "⚠"
            print(f"  {status_symbol} {func_node.name}: {complexity}")
            
            if complexity > complexity_threshold:
                violations.append({
                    'function': func_node.name,
                    'complexity': complexity,
                    'line': func_node.lineno
                })
        
        # File statistics
        total_functions = len(functions)
        avg_complexity = sum(file_complexities) / len(file_complexities) if file_complexities else 0
        max_complexity = max(file_complexities) if file_complexities else 0
        violation_count = len(violations)
        
        file_results[filename] = {
            'total_functions': total_functions,
            'avg_complexity': avg_complexity,
            'max_complexity': max_complexity,
            'violations': violations,
            'violation_count': violation_count
        }
        
        print(f"\n  File Summary:")
        print(f"    Functions: {total_functions}")
        print(f"    Average complexity: {avg_complexity:.2f}")
        print(f"    Maximum complexity: {max_complexity}")
        print(f"    Threshold violations: {violation_count}")
        
        if violations:
            print(f"    Functions exceeding threshold ({complexity_threshold}):")
            for violation in violations:
                print(f"      - {violation['function']}: {violation['complexity']} (line {violation['line']})")
    
    # Project-wide analysis
    print(f"\n\nProject-Wide Analysis:")
    print("=" * 60)
    
    all_functions = sum([r['total_functions'] for r in file_results.values()])
    all_violations = sum([r['violation_count'] for r in file_results.values()])
    violation_rate = (all_violations / all_functions * 100) if all_functions > 0 else 0
    
    project_avg_complexity = sum([r['avg_complexity'] * r['total_functions'] for r in file_results.values()]) / all_functions if all_functions > 0 else 0
    project_max_complexity = max([r['max_complexity'] for r in file_results.values()]) if file_results else 0
    
    print(f"Total functions analyzed: {all_functions}")
    print(f"Project average complexity: {project_avg_complexity:.2f}")
    print(f"Project maximum complexity: {project_max_complexity}")
    print(f"Threshold violations: {all_violations} ({violation_rate:.1f}%)")
    
    # Quality assessment
    print(f"\nProject Quality Assessment:")
    if violation_rate == 0:
        quality_grade = "A"
        assessment = "Excellent - All functions meet complexity standards"
    elif violation_rate <= 10:
        quality_grade = "B"
        assessment = "Good - Most functions meet complexity standards"
    elif violation_rate <= 25:
        quality_grade = "C"
        assessment = "Fair - Some functions need refactoring"
    else:
        quality_grade = "D"
        assessment = "Poor - Many functions exceed complexity thresholds"
    
    print(f"  Quality Grade: {quality_grade}")
    print(f"  Assessment: {assessment}")
    
    # Recommendations
    print(f"\nRecommendations:")
    if all_violations == 0:
        print("  - Code complexity is well-managed")
        print("  - Continue current coding practices")
    else:
        print(f"  - Refactor {all_violations} functions that exceed the complexity threshold")
        print(f"  - Consider breaking complex functions into smaller, focused functions")
        print(f"  - Implement complexity checking in your CI/CD pipeline")
        
        # Prioritize by severity
        high_priority = [f for results in file_results.values() for f in results['violations'] if f['complexity'] > 15]
        if high_priority:
            print(f"  - HIGH PRIORITY: {len(high_priority)} functions with complexity > 15")
    
    print("mccabe example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"mccabe example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

```python
import mccabe
import ast
import textwrap

# Example 5: Integration workflows and automated code quality
print("mccabe integration workflows and automated code quality")

try:
    print("Example 5 - Integration workflows and automated code quality")
    
    # Simulate a code quality workflow for a development team
    print("Automated Code Quality Workflow:")
    print("=" * 50)
    
    # Configuration for complexity analysis
    quality_config = {
        'complexity_threshold': 10,        # Maximum allowed complexity
        'warning_threshold': 7,            # Warning level
        'excellent_threshold': 3,          # Excellent code threshold
        'max_functions_per_file': 20,      # File size recommendation
        'min_complexity_for_review': 5     # Require code review above this
    }
    
    # Sample codebase with different quality levels
    codebase = {
        'utils/helpers.py': textwrap.dedent("""
        def format_name(first, last):
            return f"{first} {last}".strip()
        
        def calculate_percentage(value, total):
            if total == 0:
                return 0
            return (value / total) * 100
        
        def validate_email(email):
            if '@' in email and '.' in email:
                return True
            return False
        """),
        
        'processing/data_handler.py': textwrap.dedent("""
        def process_user_data(users, filters=None):
            if not users:
                return []
            
            filters = filters or {}
            processed = []
            
            for user in users:
                if 'active' in filters and user.get('active') != filters['active']:
                    continue
                
                if 'age_min' in filters and user.get('age', 0) < filters['age_min']:
                    continue
                
                if 'age_max' in filters and user.get('age', 0) > filters['age_max']:
                    continue
                
                processed_user = {
                    'id': user['id'],
                    'name': format_user_name(user),
                    'status': 'active' if user.get('active', False) else 'inactive'
                }
                
                processed.append(processed_user)
            
            return processed
        
        def format_user_name(user):
            first = user.get('first_name', '').strip()
            last = user.get('last_name', '').strip()
            
            if first and last:
                return f"{first} {last}"
            elif first:
                return first
            elif last:
                return f"(last name: {last})"
            else:
                return "Unknown User"
        """),
        
        'business/order_processor.py': textwrap.dedent("""
        def process_order(order_data, inventory, pricing, customer_info):
            if not order_data or not inventory:
                raise ValueError("Invalid input data")
            
            order_id = order_data.get('id')
            if not order_id:
                raise ValueError("Order ID is required")
            
            total_cost = 0
            processed_items = []
            out_of_stock = []
            pricing_issues = []
            
            for item in order_data.get('items', []):
                try:
                    product_id = item['product_id']
                    quantity = item['quantity']
                    
                    # Check inventory
                    if product_id not in inventory:
                        out_of_stock.append(product_id)
                        continue
                    
                    available_qty = inventory[product_id]['quantity']
                    if available_qty < quantity:
                        if available_qty > 0:
                            quantity = available_qty
                            item['adjusted_quantity'] = quantity
                        else:
                            out_of_stock.append(product_id)
                            continue
                    
                    # Calculate pricing
                    if product_id in pricing:
                        unit_price = pricing[product_id]['price']
                        
                        # Apply discounts
                        discount = 0
                        if customer_info.get('vip', False):
                            discount = pricing[product_id].get('vip_discount', 0)
                        elif quantity >= pricing[product_id].get('bulk_threshold', 10):
                            discount = pricing[product_id].get('bulk_discount', 0)
                        
                        item_total = quantity * unit_price * (1 - discount)
                        
                        # Apply taxes
                        tax_rate = pricing[product_id].get('tax_rate', 0.08)
                        item_total_with_tax = item_total * (1 + tax_rate)
                        
                        processed_item = {
                            'product_id': product_id,
                            'quantity': quantity,
                            'unit_price': unit_price,
                            'discount': discount,
                            'tax_rate': tax_rate,
                            'subtotal': item_total,
                            'total': item_total_with_tax
                        }
                        
                        processed_items.append(processed_item)
                        total_cost += item_total_with_tax
                    else:
                        pricing_issues.append(product_id)
                
                except KeyError as e:
                    pricing_issues.append(f"Missing data for {product_id}: {str(e)}")
                except Exception as e:
                    pricing_issues.append(f"Error processing {product_id}: {str(e)}")
            
            # Calculate shipping
            shipping_cost = 0
            if total_cost < 50:
                shipping_cost = 9.99
            elif total_cost < 100:
                shipping_cost = 4.99
            
            final_total = total_cost + shipping_cost
            
            return {
                'order_id': order_id,
                'items': processed_items,
                'subtotal': total_cost,
                'shipping': shipping_cost,
                'total': final_total,
                'out_of_stock': out_of_stock,
                'pricing_issues': pricing_issues,
                'status': 'processed' if not out_of_stock and not pricing_issues else 'issues'
            }
        """)
    }
    
    # Analyze the entire codebase
    codebase_report = {
        'files': {},
        'summary': {
            'total_files': 0,
            'total_functions': 0,
            'total_violations': 0,
            'functions_need_review': 0,
            'excellent_functions': 0
        }
    }
    
    print("Analyzing codebase...")
    
    for file_path, code in codebase.items():
        print(f"\n📁 {file_path}")
        
        # Parse file
        tree = ast.parse(code, filename=file_path)
        functions = [node for node in ast.walk(tree) if isinstance(node, ast.FunctionDef)]
        
        file_analysis = {
            'functions': [],
            'violations': [],
            'warnings': [],
            'needs_review': [],
            'excellent': []
        }
        
        for func_node in functions:
            # Use PathGraph for complexity calculation
            pathgraph = mccabe.PathGraph(func_node.name, func_node, func_node.lineno)
            complexity = pathgraph.complexity()
            
            func_info = {
                'name': func_node.name,
                'complexity': complexity,
                'line': func_node.lineno
            }
            
            file_analysis['functions'].append(func_info)
            
            # Categorize based on complexity
            if complexity > quality_config['complexity_threshold']:
                file_analysis['violations'].append(func_info)
                status = "❌ VIOLATION"
            elif complexity > quality_config['warning_threshold']:
                file_analysis['warnings'].append(func_info)
                status = "⚠️  WARNING"
            elif complexity <= quality_config['excellent_threshold']:
                file_analysis['excellent'].append(func_info)
                status = "✨ EXCELLENT"
            else:
                status = "✅ GOOD"
            
            if complexity >= quality_config['min_complexity_for_review']:
                file_analysis['needs_review'].append(func_info)
            
            print(f"  {status} {func_node.name}() - Complexity: {complexity}")
        
        # File summary
        total_funcs = len(functions)
        avg_complexity = sum(f['complexity'] for f in file_analysis['functions']) / total_funcs if total_funcs > 0 else 0
        
        print(f"  📊 File Summary:")
        print(f"     Functions: {total_funcs}")
        print(f"     Average complexity: {avg_complexity:.2f}")
        print(f"     Violations: {len(file_analysis['violations'])}")
        print(f"     Warnings: {len(file_analysis['warnings'])}")
        print(f"     Need review: {len(file_analysis['needs_review'])}")
        print(f"     Excellent: {len(file_analysis['excellent'])}")
        
        codebase_report['files'][file_path] = file_analysis
        
        # Update summary
        codebase_report['summary']['total_files'] += 1
        codebase_report['summary']['total_functions'] += total_funcs
        codebase_report['summary']['total_violations'] += len(file_analysis['violations'])
        codebase_report['summary']['functions_need_review'] += len(file_analysis['needs_review'])
        codebase_report['summary']['excellent_functions'] += len(file_analysis['excellent'])
    
    # Generate final report
    print(f"\n\n📋 CODEBASE QUALITY REPORT")
    print("=" * 50)
    
    summary = codebase_report['summary']
    
    print(f"Files analyzed: {summary['total_files']}")
    print(f"Functions analyzed: {summary['total_functions']}")
    print(f"")
    
    violation_rate = (summary['total_violations'] / summary['total_functions']) * 100 if summary['total_functions'] > 0 else 0
    excellent_rate = (summary['excellent_functions'] / summary['total_functions']) * 100 if summary['total_functions'] > 0 else 0
    review_rate = (summary['functions_need_review'] / summary['total_functions']) * 100 if summary['total_functions'] > 0 else 0
    
    print(f"🚨 Complexity violations: {summary['total_violations']} ({violation_rate:.1f}%)")
    print(f"🔍 Functions need review: {summary['functions_need_review']} ({review_rate:.1f}%)")
    print(f"✨ Excellent functions: {summary['excellent_functions']} ({excellent_rate:.1f}%)")
    
    # Quality score calculation
    quality_score = max(0, 100 - (violation_rate * 5) - (review_rate * 1) + (excellent_rate * 0.5))
    
    print(f"\n🎯 Overall Quality Score: {quality_score:.1f}/100")
    
    if quality_score >= 90:
        grade = "A"
        recommendation = "Excellent code quality! Keep up the good work."
    elif quality_score >= 80:
        grade = "B"
        recommendation = "Good code quality. Focus on reducing complexity violations."
    elif quality_score >= 70:
        grade = "C"
        recommendation = "Fair code quality. Prioritize refactoring complex functions."
    else:
        grade = "D"
        recommendation = "Poor code quality. Immediate attention needed for complex functions."
    
    print(f"📝 Grade: {grade}")
    print(f"💡 Recommendation: {recommendation}")
    
    # Action items
    if summary['total_violations'] > 0:
        print(f"\n⚡ IMMEDIATE ACTIONS REQUIRED:")
        for file_path, analysis in codebase_report['files'].items():
            if analysis['violations']:
                print(f"  📁 {file_path}:")
                for violation in analysis['violations']:
                    print(f"     - Refactor {violation['name']}() (complexity: {violation['complexity']})")
    
    # CI/CD Integration suggestions
    print(f"\n🔧 CI/CD Integration Recommendations:")
    print(f"  - Set complexity threshold to {quality_config['complexity_threshold']} in pre-commit hooks")
    print(f"  - Require code review for functions with complexity ≥ {quality_config['min_complexity_for_review']}")
    print(f"  - Block merge if complexity violations exist")
    print(f"  - Generate complexity reports on each pull request")
    
    print("mccabe example 5 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"mccabe example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
print(f"\nMccabe capabilities demonstrated:")
print("- Basic cyclomatic complexity calculation for individual functions")
print("- PathGraph analysis and detailed control flow visualization")  
print("- Multi-function and class-level complexity analysis")
print("- File-based analysis with threshold checking and violation reporting")
print("- Production workflow integration with automated code quality assessment")
```
