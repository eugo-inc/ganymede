## Example 2: mdformat functionality test 2

````python
import mdformat
import textwrap

# Example 2: Configuration options and formatting styles
print("mdformat configuration options and formatting styles")

try:
    print("Example 2 - Configuration options and formatting styles")
    
    # Sample Markdown with various formatting elements
    sample_markdown = textwrap.dedent("""
    # Main Title
    
    This document contains various **formatting elements** that can be customized.
    
    ## Code and Links
    
    Here's some `inline code` and a [link](https://example.com "Title").
    
    ```javascript
    function example() {
        return "formatted code";
    }
    ```
    
    ## Lists and emphasis
    
    - First item with *emphasis*
    - Second item with **strong emphasis**  
    - Third item with ~~strikethrough~~
    
    1. Numbered list
    2. Another item
    3. Final item
    """)
    
    # Test different configuration options
    configurations = [
        ("Default", {}),
        ("Wrap at 80", {"wrap": 80}),
        ("No wrap", {"wrap": "no"}),
        ("ATX headers", {"end_of_line": "lf"}),
    ]
    
    print("Testing different formatting configurations:")
    print("=" * 50)
    
    for config_name, config_options in configurations:
        print(f"\n{config_name} Configuration:")
        print("-" * 30)
        
        try:
            # Format with specific configuration
            formatted = mdformat.text(sample_markdown, options=config_options)
            
            # Show first few lines
            lines = formatted.split('\n')
            for i, line in enumerate(lines[:8]):
                print(f"  {i+1:2d}: {repr(line)}")
            
            if len(lines) > 8:
                print(f"     ... ({len(lines) - 8} more lines)")
            
            # Show statistics
            print(f"  Total lines: {len(lines)}")
            print(f"  Total characters: {len(formatted)}")
            
            # Check line lengths if wrap option is specified
            if 'wrap' in config_options and config_options['wrap'] not in ['no', None]:
                max_line_len = max(len(line) for line in lines if line.strip())
                print(f"  Max line length: {max_line_len}")
                if max_line_len <= config_options['wrap']:
                    print(f"  ✓ Wrap constraint satisfied")
                else:
                    print(f"  ⚠ Some lines exceed wrap limit")
        
        except Exception as e:
            print(f"  Error with {config_name}: {str(e)}")
            raise
    
    # Test formatting consistency
    print(f"\nConsistency Test:")
    original = mdformat.text(sample_markdown)
    reformatted = mdformat.text(original)
    
    if original == reformatted:
        print("✓ Formatting is idempotent (stable)")
    else:
        print("⚠ Formatting changes on multiple applications")
        print(f"  Original length: {len(original)}")
        print(f"  Reformatted length: {len(reformatted)}")
    
    print("mdformat example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"mdformat example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
````

````python
import mdformat
import textwrap

# Example 3: Complex document formatting and structure preservation
print("mdformat complex document formatting and structure preservation")

try:
    print("Example 3 - Complex document formatting and structure preservation")
    
    # Complex Markdown document with various elements
    complex_markdown = textwrap.dedent("""
    # Project Documentation
    
    > **Note**: This is an important document that needs consistent formatting.
    
    ## Table of Contents
    
    1. [Introduction](#introduction)
    2. [Installation](#installation)  
    3. [Usage Examples](#usage-examples)
    
    ## Introduction
    
    This project provides a **comprehensive solution** for data processing with the following features:
    
    - High-performance algorithms
    - Memory-efficient data structures  
    - Extensible plugin architecture
    - Cross-platform compatibility
    
    ### Key Benefits
    
    | Feature | Benefit | Impact |
    |---------|---------|--------|
    | Performance | 10x faster | Reduced processing time |
    | Memory | 50% less usage | Better scalability |
    | Plugins | Extensible | Easy customization |
    
    ## Installation
    
    Install using pip:
    
    ```bash
    pip install example-package
    ```
    
    Or using conda:
    
    ```bash
    conda install -c example example-package
    ```
    
    ### Requirements
    
    - Python 3.8+
    - NumPy >= 1.19.0
    - Pandas >= 1.2.0
    
    ## Usage Examples
    
    ### Basic Usage
    
    ```python
    import example_package as ep
    
    # Create a processor
    processor = ep.DataProcessor(
        algorithm='fast',
        memory_limit='1GB'
    )
    
    # Process data
    result = processor.process(data)
    print(f"Processed {len(result)} items")
    ```
    
    ### Advanced Configuration
    
    For advanced users, you can customize the behavior:
    
    ```python
    config = {
        'threads': 4,
        'chunk_size': 1000,
        'cache_enabled': True
    }
    
    processor = ep.DataProcessor(**config)
    ```
    
    ---
    
    ## Footnotes
    
    [^1]: Performance benchmarks conducted on standard datasets.
    [^2]: Memory usage measured with standard profiling tools.
    """)
    
    print("Formatting complex document...")
    
    # Format the complex document
    formatted_doc = mdformat.text(complex_markdown)
    
    # Analyze the formatting results
    original_lines = complex_markdown.split('\n')
    formatted_lines = formatted_doc.split('\n')
    
    print(f"Document Analysis:")
    print(f"  Original lines: {len(original_lines)}")
    print(f"  Formatted lines: {len(formatted_lines)}")
    print(f"  Character count: {len(complex_markdown)} → {len(formatted_doc)}")
    
    # Check preservation of key elements
    preserved_elements = []
    
    # Check headers
    original_headers = [line for line in original_lines if line.startswith('#')]
    formatted_headers = [line for line in formatted_lines if line.startswith('#')]
    if len(original_headers) == len(formatted_headers):
        preserved_elements.append("Headers")
    
    # Check code blocks
    original_code_blocks = complex_markdown.count('```')
    formatted_code_blocks = formatted_doc.count('```')
    if original_code_blocks == formatted_code_blocks:
        preserved_elements.append("Code blocks")
    
    # Check tables
    original_table_rows = len([line for line in original_lines if '|' in line])
    formatted_table_rows = len([line for line in formatted_lines if '|' in line])
    if original_table_rows <= formatted_table_rows:  # Formatting might add separator rows
        preserved_elements.append("Tables")
    
    # Check lists
    original_list_items = len([line for line in original_lines if line.strip().startswith('-')])
    formatted_list_items = len([line for line in formatted_lines if line.strip().startswith('-')])
    if original_list_items == formatted_list_items:
        preserved_elements.append("Lists")
    
    # Check links
    original_links = complex_markdown.count('[')
    formatted_links = formatted_doc.count('[')
    if original_links == formatted_links:
        preserved_elements.append("Links")
    
    print(f"  Preserved elements: {preserved_elements}")
    
    # Show sample of formatted output
    print(f"\nFormatted Document Sample (first 10 lines):")
    print("=" * 50)
    for i, line in enumerate(formatted_lines[:10]):
        print(f"{i+1:2d}: {line}")
    print("    ... (truncated)")
    
    # Check table formatting specifically
    table_lines = [line for line in formatted_lines if '|' in line and line.strip()]
    if table_lines:
        print(f"\nTable Formatting Sample:")
        for line in table_lines[:4]:  # Show first few table rows
            print(f"    {line}")
        if len(table_lines) > 4:
            print(f"    ... ({len(table_lines) - 4} more table rows)")
    
    print("mdformat example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"mdformat example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
````

## Example 4: mdformat functionality test 4

````python
import mdformat
import textwrap
import tempfile
import os

# Example 4: File operations and batch processing
print("mdformat file operations and batch processing")

try:
    print("Example 4 - File operations and batch processing")
    
    # Create sample Markdown files for batch processing
    sample_files = {
        "README.md": textwrap.dedent("""
        # Project Name
        
        A brief description  of the project with   extra spaces.
        
        ## Installation  
        
        ```bash
        pip install   project-name
        ```
        
        ## Usage
        
        - First step
        -Second step (missing space)
        - Third step
        """),
        
        "CONTRIBUTING.md": textwrap.dedent("""
        # Contributing Guidelines    
        
        Thank you for  contributing!
        
        ##   How to contribute:
        
        1.   Fork the repository
        2. Create a branch
        3.Make changes
        4. Submit PR
        
        ### Code Style
        
        -   Use consistent formatting  
        - Follow   PEP 8
        """),
        
        "CHANGELOG.md": textwrap.dedent("""
        # Changelog   
        
        All notable changes   to this project.
        
        ##   [1.0.0] - 2024-01-01
        
        ### Added
        -  New feature X
        - Feature Y   
        -Feature Z (missing space)
        
        ###   Fixed  
        - Bug fix A
        """)
    }
    
    # Create temporary directory and files
    with tempfile.TemporaryDirectory() as temp_dir:
        print(f"Created temporary directory: {temp_dir}")
        
        # Write sample files
        file_paths = []
        for filename, content in sample_files.items():
            file_path = os.path.join(temp_dir, filename)
            with open(file_path, 'w', encoding='utf-8') as f:
                f.write(content)
            file_paths.append(file_path)
            print(f"  Created: {filename}")
        
        # Process each file and analyze results
        print(f"\nProcessing files...")
        processing_results = []
        
        for file_path in file_paths:
            filename = os.path.basename(file_path)
            
            # Read original content
            with open(file_path, 'r', encoding='utf-8') as f:
                original_content = f.read()
            
            # Format the content
            formatted_content = mdformat.text(original_content)
            
            # Write formatted content to new file
            formatted_path = file_path.replace('.md', '_formatted.md')
            with open(formatted_path, 'w', encoding='utf-8') as f:
                f.write(formatted_content)
            
            # Analyze changes
            original_lines = len(original_content.split('\n'))
            formatted_lines = len(formatted_content.split('\n'))
            size_change = len(formatted_content) - len(original_content)
            
            result = {
                'filename': filename,
                'original_lines': original_lines,
                'formatted_lines': formatted_lines,
                'size_change': size_change,
                'size_change_pct': (size_change / len(original_content)) * 100 if original_content else 0
            }
            processing_results.append(result)
            
            print(f"  {filename}:")
            print(f"    Lines: {original_lines} → {formatted_lines}")
            print(f"    Size change: {size_change:+d} chars ({result['size_change_pct']:+.1f}%)")
        
        # Batch processing summary
        print(f"\nBatch Processing Summary:")
        print("=" * 40)
        
        total_files = len(processing_results)
        total_original_lines = sum(r['original_lines'] for r in processing_results)
        total_formatted_lines = sum(r['formatted_lines'] for r in processing_results)
        total_size_change = sum(r['size_change'] for r in processing_results)
        
        print(f"Files processed: {total_files}")
        print(f"Total lines: {total_original_lines} → {total_formatted_lines}")
        print(f"Total size change: {total_size_change:+d} characters")
        print(f"Average size change: {total_size_change/total_files:+.1f} chars per file")
        
        # Quality metrics
        files_reduced = sum(1 for r in processing_results if r['size_change'] < 0)
        files_increased = sum(1 for r in processing_results if r['size_change'] > 0)
        files_unchanged = sum(1 for r in processing_results if r['size_change'] == 0)
        
        print(f"\nQuality Impact:")
        print(f"  Files reduced in size: {files_reduced}")
        print(f"  Files increased in size: {files_increased}")
        print(f"  Files unchanged: {files_unchanged}")
        
        # Verify formatting consistency
        print(f"\nConsistency Verification:")
        consistency_passed = 0
        for file_path in file_paths:
            formatted_path = file_path.replace('.md', '_formatted.md')
            
            with open(formatted_path, 'r', encoding='utf-8') as f:
                once_formatted = f.read()
            
            twice_formatted = mdformat.text(once_formatted)
            
            if once_formatted == twice_formatted:
                consistency_passed += 1
        
        print(f"  Files with consistent formatting: {consistency_passed}/{total_files}")
        # mdformat output is a fixed point: formatting an already-formatted file must not change it
        assert consistency_passed == total_files, f"mdformat changed {total_files - consistency_passed} already-formatted file(s)"
        print("  ✓ All files are idempotent (stable formatting)")
        
        # File existence verification
        print(f"\nGenerated Files:")
        for file_path in file_paths:
            formatted_path = file_path.replace('.md', '_formatted.md')
            assert os.path.exists(formatted_path), f"formatted output {formatted_path} was not created"
            size = os.path.getsize(formatted_path)
            assert size > 0, f"formatted output {formatted_path} is empty"
            print(f"  ✓ {os.path.basename(formatted_path)} ({size} bytes)")
    
    print(f"\nFile processing workflow completed successfully")
    
    print("mdformat example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"mdformat example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
````

## Example 5: mdformat functionality test 5

````python
import mdformat
import textwrap
import re

# Example 5: Integration workflows and automated documentation
print("mdformat integration workflows and automated documentation")

try:
    print("Example 5 - Integration workflows and automated documentation")
    
    # Simulate a documentation workflow for a software project
    print("Automated Documentation Formatting Workflow")
    print("=" * 50)
    
    # Sample documentation files from different sources
    documentation_sources = {
        "api_docs.md": textwrap.dedent("""
        # API   Documentation
        
        ## Authentication   
        
        All API requests require authentication   using API keys.
        
        ### Headers
        
        -   `Authorization: Bearer <token>`  
        - `Content-Type:   application/json`
        
        ## Endpoints
        
        ###   GET /users
        
        Returns list of users.
        
        **Parameters:**
        
        |Parameter|Type|Required|Description|
        |---|---|---|---|
        |limit|integer|No|Maximum results   |
        |offset|integer|No|  Pagination offset|
        
        **Response:**
        
        ```json
        {
          "users": [
            {
              "id":   1,
              "name": "John  Doe"
            }
          ]
        }
        ```
        """),
        
        "user_guide.md": textwrap.dedent("""
        # User Guide    
        
        ## Getting Started   
        
        ### Prerequisites
        
        Before you begin,   ensure you have:
        
        1.   Python 3.8+
        2. Package manager (pip/conda)  
        3.Administrative privileges
        
        ### Installation   Steps  
        
        1.Download the package:
        
           ```bash
           curl -O https://example.com/package.tar.gz  
           ```
        
        2. Extract files:
        
           ```bash
           tar -xzf   package.tar.gz
           ```
        
        3.  Install dependencies:
        
           ```bash
           pip install   -r requirements.txt
           ```
        
        ##   Configuration
        
        Create   a config file:
        
        ```yaml
        server:
          host:   localhost
          port: 8080
        database:
          url:   sqlite:///app.db
        ```
        """)
    }
    
    # Configuration for different document types
    formatting_profiles = {
        "api_docs": {
            "wrap": "no",  # Keep code samples intact
            "end_of_line": "lf"
        },
        "user_guide": {
            "wrap": 80,  # Wrap for readability
            "end_of_line": "lf"
        },
        "default": {
            "wrap": "no",
            "end_of_line": "lf"
        }
    }
    
    # Process documentation with appropriate profiles
    print("Processing documentation files...")
    
    formatted_docs = {}
    quality_metrics = {
        "total_files": 0,
        "total_improvements": 0,
        "formatting_issues_fixed": 0,
        "consistency_achieved": 0
    }
    
    for doc_name, content in documentation_sources.items():
        print(f"\n📄 Processing {doc_name}")
        print("-" * 40)
        
        # Determine profile based on document type
        if "api" in doc_name.lower():
            profile = formatting_profiles["api_docs"]
        elif "guide" in doc_name.lower():
            profile = formatting_profiles["user_guide"]
        else:
            profile = formatting_profiles["default"]
        
        # Format with appropriate profile
        formatted_content = mdformat.text(content, options=profile)
        formatted_docs[doc_name] = formatted_content
        
        # Analyze improvements
        original_issues = []
        
        # Check for common formatting issues
        if re.search(r'#\s{2,}', content):  # Multiple spaces after headers
            original_issues.append("Header spacing")
        if re.search(r'\n\s*\n\s*\n', content):  # Multiple blank lines
            original_issues.append("Excessive blank lines")
        if re.search(r'-\s*[^\s]', content):  # List items without space
            original_issues.append("List formatting")
        if re.search(r'\|\s*[^|]*\s*[^|]*\|', content):  # Table formatting
            if content.count('|') != formatted_content.count('|'):
                original_issues.append("Table alignment")
        
        # Check if issues were fixed
        fixed_issues = []
        if "Header spacing" in original_issues and not re.search(r'#\s{2,}', formatted_content):
            fixed_issues.append("Header spacing")
        if "Excessive blank lines" in original_issues and not re.search(r'\n\s*\n\s*\n', formatted_content):
            fixed_issues.append("Excessive blank lines")
        if "List formatting" in original_issues and not re.search(r'-\s*[^\s]', formatted_content):
            fixed_issues.append("List formatting")
        
        print(f"  Profile applied: {[k for k, v in formatting_profiles.items() if v == profile][0]}")
        print(f"  Original issues detected: {len(original_issues)}")
        print(f"  Issues fixed: {len(fixed_issues)}")
        print(f"  Character count: {len(content)} → {len(formatted_content)}")
        
        if fixed_issues:
            print(f"  Improvements: {', '.join(fixed_issues)}")
        
        quality_metrics["total_files"] += 1
        quality_metrics["formatting_issues_fixed"] += len(fixed_issues)
        
        # Test idempotency
        re_formatted = mdformat.text(formatted_content, options=profile)
        if formatted_content == re_formatted:
            quality_metrics["consistency_achieved"] += 1
            print(f"  ✓ Formatting is stable")
        else:
            print(f"  ⚠ Formatting may be unstable")
    
    # Generate quality report
    print(f"\n📊 DOCUMENTATION QUALITY REPORT")
    print("=" * 50)
    
    print(f"Files processed: {quality_metrics['total_files']}")
    print(f"Formatting issues fixed: {quality_metrics['formatting_issues_fixed']}")
    print(f"Files with stable formatting: {quality_metrics['consistency_achieved']}/{quality_metrics['total_files']}")
    
    # Calculate quality score
    total_possible_score = quality_metrics['total_files'] * 3  # Max 3 points per file
    actual_score = quality_metrics['consistency_achieved'] + (quality_metrics['formatting_issues_fixed'] // 2)
    quality_percentage = min(100, (actual_score / total_possible_score) * 100)
    
    print(f"Overall quality score: {quality_percentage:.1f}%")
    
    # Integration recommendations
    print(f"\n🔧 Integration Recommendations:")
    print(f"  - Set up pre-commit hooks to format Markdown files")
    print(f"  - Configure CI/CD to check formatting consistency")
    print(f"  - Use profile-based formatting for different document types")
    print(f"  - Implement automated formatting in documentation builds")
    
    if quality_metrics['consistency_achieved'] == quality_metrics['total_files']:
        print(f"  ✅ All documents have stable formatting - ready for automation")
    else:
        print(f"  ⚠ Some documents need manual review before automation")
    
    # Show sample of formatted output
    print(f"\n📝 Sample Formatted Content (API Docs):")
    print("-" * 40)
    sample_lines = formatted_docs["api_docs.md"].split('\n')[:8]
    for i, line in enumerate(sample_lines):
        print(f"{i+1:2d}: {line}")
    print("    ... (content continues)")
    
    print("mdformat example 5 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"mdformat example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
print(f"\nMdformat capabilities demonstrated:")
print("- Basic Markdown formatting and standardization")
print("- Configuration options and customizable formatting styles")
print("- Complex document structure preservation and formatting")
print("- File operations with batch processing workflows")
print("- Integration workflows for automated documentation quality")
````
