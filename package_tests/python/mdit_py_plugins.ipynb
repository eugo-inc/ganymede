# mdit_py_plugins: 5 Basic Usage Examples

```python
# Example 1: Using footnotes plugin
from markdown_it import MarkdownIt
from mdit_py_plugins.footnote import footnote_plugin
md = MarkdownIt().use(footnote_plugin)
text = 'Here is a footnote[^1].\n\n[^1]: This is the footnote content.'
print(md.render(text))
```

```python
# Example 2: Using task lists plugin
from mdit_py_plugins.tasklists import tasklists_plugin
md = MarkdownIt().use(tasklists_plugin)
text = '- [x] Completed task\n- [ ] Incomplete task\n- [x] Another completed task'
print(md.render(text))
```

```python
# Example 3: Using front matter plugin
from mdit_py_plugins.front_matter import front_matter_plugin
md = MarkdownIt().use(front_matter_plugin)
text = '---\ntitle: My Document\nauthor: John Doe\n---\n\n# Content\n\nThis is the body.'
result = md.render(text)
print(result)
```

```python
# Example 4: Using definition lists plugin
from mdit_py_plugins.deflist import deflist_plugin
md = MarkdownIt().use(deflist_plugin)
text = 'Term 1\n:   Definition 1\n\nTerm 2\n:   Definition 2a\n:   Definition 2b'
print(md.render(text))
```

```python
# Example 5: Combining multiple plugins
md = MarkdownIt().use(footnote_plugin).use(tasklists_plugin)
text = '# Todo List\n\n- [x] Task with footnote[^1]\n- [ ] Pending task\n\n[^1]: Footnote for completed task'
print(md.render(text))
```
