# Mdurl Package Examples

This notebook demonstrates 5 basic, versatile, and comprehensive examples of the `mdurl` package.

Mdurl provides URL utilities for markdown processing.


## Example 1: Basic URL encoding and decoding

```python
from mdurl import encode, decode

# Basic URL encoding
test_strings = [
    "hello world",
    "special chars: @#$%^&*()",
    "unicode: café münücherì",
    "spaces and symbols: a b c!@#",
    "path/with/slashes"
]

print("URL Encoding examples:")
for text in test_strings:
    encoded = encode(text)
    print(f"'{text}' -> '{encoded}'")

# Basic URL decoding
encoded_strings = [
    "hello%20world",
    "caf%C3%A9",
    "test%2Fpath%2Fhere",
    "query%3Dparam%26another%3Dvalue"
]

print("\nURL Decoding examples:")
for encoded in encoded_strings:
    try:
        decoded = decode(encoded)
        print(f"'{encoded}' -> '{decoded}'")
    except Exception as e:
        print(f"'{encoded}' -> Error: {e}")
        raise

# Round-trip test
print("\nRound-trip test:")
original = "Hello, 世界! @#$%"
encoded = encode(original)
decoded = decode(encoded)
print(f"Original: '{original}'")
print(f"Encoded:  '{encoded}'")
print(f"Decoded:  '{decoded}'")
print(f"Round-trip successful: {original == decoded}")

# Verify mdurl functionality
print(f"\nMdurl module functions available:")
print(f"encode function: {callable(encode)}")
print(f"decode function: {callable(decode)}")
```

## Example 2: URL parsing and validation

```python
from mdurl import parse, format as fmt

# Test URLs for parsing
test_urls = [
    "https://example.com/path/to/page?query=value#section",
    "http://user:pass@domain.com:8080/path",
    "ftp://files.example.org/file.txt",
    "mailto:user@example.com",
    "/relative/path?param=value",
    "//example.com/protocol-relative",
    "file:///local/path/file.txt"
]

print("URL Parsing examples:")
for url in test_urls:
    try:
        parsed = parse(url)
        print(f"\nURL: {url}")
        print(f"  Protocol: {getattr(parsed, 'protocol', 'N/A')}")
        print(f"  Hostname: {getattr(parsed, 'hostname', 'N/A')}")
        print(f"  Pathname: {getattr(parsed, 'pathname', 'N/A')}")
        print(f"  Search: {getattr(parsed, 'search', 'N/A')}")
        print(f"  Hash: {getattr(parsed, 'hash', 'N/A')}")
    except Exception as e:
        print(f"\nURL: {url} -> Parse error: {e}")
        raise

# Test URL formatting
print("\n\nURL Formatting test:")
try:
    # Create a URL object to format
    sample_url = "https://example.com/path?param=value#section"
    parsed_url = parse(sample_url)
    
    # Try to format it back
    formatted = fmt(parsed_url)
    print(f"Original:  {sample_url}")
    print(f"Formatted: {formatted}")
    print(f"Format successful: {isinstance(formatted, str)}")
    
except Exception as e:
    print(f"Format function error: {e}")
    raise

# Test edge cases
edge_cases = [
    "",  # Empty string
    "not-a-url",  # Invalid URL
    "://no-protocol",  # Malformed
    "https://",  # Incomplete
]

print("\nEdge cases:")
for case in edge_cases:
    try:
        result = parse(case)
        print(f"'{case}' -> Parsed successfully")
    except Exception as e:
        print(f"'{case}' -> Error: {type(e).__name__}")
        raise

# Verify parsing functions
print(f"\nParsing functions available:")
print(f"parse function: {callable(parse)}")
print(f"format function: {callable(fmt)}")
```

## Example 3: URL manipulation and components

```python
import mdurl

# Test various mdurl functions
print("Available mdurl functions:")
functions = [attr for attr in dir(mdurl) if not attr.startswith('_') and callable(getattr(mdurl, attr))]
print(f"Functions: {functions}")

# Test URL component extraction
test_url = "https://user:pass@example.com:8080/path/to/resource?query=value&another=param#section1"
print(f"\nAnalyzing URL: {test_url}")

try:
    parsed = mdurl.parse(test_url)
    
    # Extract all available components
    components = {
        'protocol': getattr(parsed, 'protocol', None),
        'hostname': getattr(parsed, 'hostname', None),
        'port': getattr(parsed, 'port', None),
        'pathname': getattr(parsed, 'pathname', None),
        'search': getattr(parsed, 'search', None),
        'hash': getattr(parsed, 'hash', None),
        'auth': getattr(parsed, 'auth', None),
        'host': getattr(parsed, 'host', None)
    }
    
    print("Components:")
    for component, value in components.items():
        print(f"  {component}: {value}")
        
except Exception as e:
    print(f"Parsing error: {e}")
    raise

# Test URL building/reconstruction
print("\nURL reconstruction test:")
url_parts = {
    'protocol': 'https:',
    'hostname': 'api.example.com',
    'port': '443',
    'pathname': '/v1/users',
    'search': '?limit=10&offset=0',
    'hash': '#results'
}

# Try different ways to construct URLs
constructed_parts = [
    f"{url_parts['protocol']}//{url_parts['hostname']}{url_parts['pathname']}",
    f"{url_parts['protocol']}//{url_parts['hostname']}:{url_parts['port']}{url_parts['pathname']}{url_parts['search']}",
    f"{url_parts['protocol']}//{url_parts['hostname']}{url_parts['pathname']}{url_parts['search']}{url_parts['hash']}"
]

for i, constructed in enumerate(constructed_parts, 1):
    print(f"Constructed URL {i}: {constructed}")
    try:
        # Validate by parsing
        parsed_constructed = mdurl.parse(constructed)
        print(f"  Valid: ✓")
    except Exception as e:
        print(f"  Valid: ✗ ({e})")
        raise

# Test special characters handling
special_urls = [
    "https://example.com/path with spaces",
    "https://example.com/path?query=special chars @#$%",
    "https://café.example.com/",
    "https://example.com/path/файл.html"
]

print("\nSpecial characters handling:")
for url in special_urls:
    try:
        encoded = mdurl.encode(url)
        parsed = mdurl.parse(encoded)
        print(f"✓ '{url}' -> handled successfully")
    except Exception as e:
        print(f"✗ '{url}' -> error: {e}")
        raise

print(f"\nMdurl URL manipulation completed!")
```

## Example 4: Integration with markdown processing

```python
import mdurl
import re

# Simulate markdown link processing
def process_markdown_links(markdown_text):
    """Process URLs in markdown text"""
    # Simple regex for markdown links: [text](url)
    link_pattern = r'\[([^\]]+)\]\(([^\)]+)\)'
    
    results = []
    for match in re.finditer(link_pattern, markdown_text):
        text = match.group(1)
        url = match.group(2)
        
        # Process URL with mdurl
        try:
            # Encode URL if needed
            encoded_url = mdurl.encode(url)
            
            # Parse URL
            parsed = mdurl.parse(encoded_url)
            
            results.append({
                'text': text,
                'original_url': url,
                'encoded_url': encoded_url,
                'hostname': getattr(parsed, 'hostname', 'N/A'),
                'pathname': getattr(parsed, 'pathname', 'N/A'),
                'valid': True
            })
        except Exception as e:
            results.append({
                'text': text,
                'original_url': url,
                'error': str(e),
                'valid': False
            })
    
    return results

# Test markdown content
markdown_content = """
# Sample Markdown

Here are some links:

- [Google](https://www.google.com)
- [Local file](./path/to/file.html)
- [API endpoint](https://api.example.com/v1/users?limit=10)
- [Email](mailto:contact@example.com)
- [Special chars](https://example.com/path with spaces)
- [Unicode domain](https://café.example.com/path)
- [Fragment](https://example.com/page#section-1)
"""

print("Processing markdown links:")
links = process_markdown_links(markdown_content)
assert len(links) == 7, f"expected the 7 markdown links above, found {len(links)}"

for i, link in enumerate(links, 1):
    print(f"\nLink {i}: {link['text']}")
    print(f"  Original: {link['original_url']}")
    
    if link['valid']:
        print(f"  Encoded:  {link['encoded_url']}")
        print(f"  Hostname: {link['hostname']}")
        print(f"  Path:     {link['pathname']}")
    else:
        # every link above is a well-formed URL that mdurl encodes and parses; a recorded error is a library failure
        raise AssertionError(f"mdurl failed on {link['original_url']!r}: {link['error']}")

# URL normalization for markdown
def normalize_markdown_url(url):
    """Normalize URLs for consistent markdown output"""
    try:
        # Parse the URL
        parsed = mdurl.parse(url)
        
        # Reconstruct with normalization
        normalized_parts = []
        
        if hasattr(parsed, 'protocol') and parsed.protocol:
            normalized_parts.append(parsed.protocol)
            normalized_parts.append('//')
        
        if hasattr(parsed, 'hostname') and parsed.hostname:
            normalized_parts.append(parsed.hostname.lower())  # Lowercase hostname
        
        if hasattr(parsed, 'port') and parsed.port:
            normalized_parts.append(f":{parsed.port}")
        
        if hasattr(parsed, 'pathname') and parsed.pathname:
            normalized_parts.append(parsed.pathname)
        
        if hasattr(parsed, 'search') and parsed.search:
            normalized_parts.append(parsed.search)
        
        if hasattr(parsed, 'hash') and parsed.hash:
            normalized_parts.append(parsed.hash)
        
        return ''.join(normalized_parts)
    
    except Exception:
        return url  # Return original if normalization fails

print("\n\nURL normalization examples:")
test_urls = [
    "HTTPS://EXAMPLE.COM/PATH",
    "https://example.com:443/path",
    "http://example.com:80/path",
    "https://example.com/./path/../other"
]

for url in test_urls:
    normalized = normalize_markdown_url(url)
    print(f"'{url}' -> '{normalized}'")

print(f"\nMarkdown URL processing completed!")
```

## Example 5: Error handling and edge cases

```python
import mdurl

# Test malformed URLs
malformed_urls = [
    "",                    # Empty
    "   ",                 # Whitespace only
    "not-a-url",          # No protocol
    "://missing-protocol", # Missing protocol
    "http://",            # Incomplete
    "http:///path",       # Missing hostname
    "http://host:abc/",   # Invalid port
    "http://host:99999/", # Port out of range
]

print("Malformed URL handling:")
for url in malformed_urls:
    print(f"\nTesting: '{url}'")
    
    # Test parsing
    try:
        parsed = mdurl.parse(url)
        print(f"  Parse: ✓ (hostname: {getattr(parsed, 'hostname', 'None')})")
    except Exception as e:
        print(f"  Parse: ✗ {type(e).__name__}: {e}")
        raise
    
    # Test encoding
    try:
        encoded = mdurl.encode(url)
        print(f"  Encode: ✓ (length: {len(encoded)})")
    except Exception as e:
        print(f"  Encode: ✗ {type(e).__name__}: {e}")
        raise

# Test extreme cases
extreme_cases = [
    "a" * 1000,  # Very long URL
    "http://" + "x" * 50 + ".com",  # Long hostname
    "http://example.com/" + "path/" * 20,  # Deep path
    "http://example.com?" + "&".join([f"param{i}=value{i}" for i in range(10)]),  # Many parameters
]

print("\n\nExtreme cases:")
for i, case in enumerate(extreme_cases, 1):
    case_preview = case[:50] + "..." if len(case) > 50 else case
    print(f"\nCase {i}: {case_preview} (length: {len(case)})")
    
    try:
        start_time = __import__('time').time()
        parsed = mdurl.parse(case)
        duration = __import__('time').time() - start_time
        print(f"  Parsed in {duration:.4f}s")
        
        # Test if we can access components
        hostname = getattr(parsed, 'hostname', None)
        pathname = getattr(parsed, 'pathname', None)
        print(f"  Hostname length: {len(hostname) if hostname else 0}")
        print(f"  Pathname length: {len(pathname) if pathname else 0}")
        
    except Exception as e:
        print(f"  Error: {type(e).__name__}: {str(e)[:100]}")
        raise

# Test character encoding edge cases
encoding_cases = [
    "http://example.com/\tTab",         # Tab character
    "http://example.com/\nNewline",     # Newline
    "http://example.com/🌟🎉📱",        # Emoji
    "http://example.com/测试中文",         # Chinese characters
]

print("\n\nCharacter encoding cases:")
for case in encoding_cases:
    try:
        encoded = mdurl.encode(case)
        decoded = mdurl.decode(encoded)
        roundtrip_ok = case == decoded
        print(f"✓ Encoding case handled: {roundtrip_ok}")
    except Exception as e:
        print(f"✗ Encoding error: {type(e).__name__}")
        raise

# Final verification
print(f"\nFinal verification:")
core_functions = ['encode', 'decode', 'parse']
all_functions_exist = all(hasattr(mdurl, func) for func in core_functions)
print(f"All core functions available: {all_functions_exist}")

# Test basic functionality still works
try:
    test_encode = mdurl.encode("test string")
    test_decode = mdurl.decode("test%20string")
    test_parse = mdurl.parse("https://example.com")
    basic_functionality = True
except Exception:
    basic_functionality = False

print(f"Basic functionality works: {basic_functionality}")
print(f"\nMdurl error handling testing completed!")
```
