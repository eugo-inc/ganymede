## Example 1: Meson CLI Commands and Version

Demonstrate meson command-line interface and version management.

```python
# Example 1: Meson CLI commands and version
import subprocess
import sys

print('Testing meson CLI commands...')

# List of meson commands to test
meson_commands = [
    ['--version'],
    ['--help'],
    ['setup', '--help'],
    ['compile', '--help'],
    ['test', '--help']
]

# Test each command
for cmd in meson_commands:
    try:
        # Try direct meson command first
        result = subprocess.run(['meson'] + cmd, capture_output=True, text=True, timeout=5)
        
        if result.returncode == 0:
            output_lines = result.stdout.split('\n')[:5]
            print(f'\nmeson {" ".join(cmd)}:')
            for line in output_lines:
                if line.strip():
                    print(f'  {line}')
        else:
            # Try Python module approach
            module_result = subprocess.run(
                [sys.executable, '-m', 'mesonbuild.mesonmain'] + cmd,
                capture_output=True, text=True, timeout=5
            )
            
            if module_result.returncode == 0:
                output_lines = module_result.stdout.split('\n')[:3]
                print(f'\nmeson {" ".join(cmd)} (via Python module):')
                for line in output_lines:
                    if line.strip():
                        print(f'  {line}')
            else:
                print(f'\nmeson {" ".join(cmd)}: Not available')
                
    except (subprocess.TimeoutExpired, FileNotFoundError) as e:
        print(f'\nmeson {" ".join(cmd)}: {type(e).__name__}')
    except Exception as e:
        print(f'\nmeson {" ".join(cmd)}: Error - {str(e)[:50]}')
        raise

print('Example 1: Meson CLI commands test completed successfully')
```

## Example 2: Meson Build File Structure

Demonstrate meson build file structure and syntax.

```python
# Example 2: Meson build file structure
import tempfile
import os

print('Creating sample meson build files...')

# Create temporary directory for meson project
with tempfile.TemporaryDirectory() as temp_dir:
    print(f'Working in temporary directory: {temp_dir}')
    
    # Sample meson.build file content
    meson_build_content = '''project('sample_project', 'c',
  version : '1.0.0',
  default_options : ['warning_level=2'])

# Define a simple executable
executable('hello',
  'src/hello.c',
  install : true)

# Define a library
lib = static_library('mylib',
  'src/lib.c',
  install : true)

# Define tests
test_exe = executable('test_hello',
  'tests/test_hello.c',
  link_with : lib)

test('hello test', test_exe)

# Subdirectories
subdir('src')
subdir('tests')
'''

    # Sample source file
    hello_c_content = '''#include <stdio.h>

int main() {
    printf("Hello from Meson!\\n");
    return 0;
}
'''

    # Sample library file
    lib_c_content = '''int add_numbers(int a, int b) {
    return a + b;
}
'''

    # Write files
    files_to_create = {
        'meson.build': meson_build_content,
        'src/hello.c': hello_c_content,
        'src/lib.c': lib_c_content,
        'src/meson.build': 'sources = files("hello.c", "lib.c")',
        'tests/meson.build': 'test_sources = files("test_hello.c")',
        'tests/test_hello.c': '#include <assert.h>\nint main() { return 0; }'
    }
    
    for filepath, content in files_to_create.items():
        full_path = os.path.join(temp_dir, filepath)
        os.makedirs(os.path.dirname(full_path), exist_ok=True)
        
        with open(full_path, 'w') as f:
            f.write(content)
        print(f'Created: {filepath}')
    
    # Display project structure
    print(f'\nMeson project structure:')
    for root, dirs, files in os.walk(temp_dir):
        level = root.replace(temp_dir, '').count(os.sep)
        indent = ' ' * 2 * level
        print(f'{indent}{os.path.basename(root)}/')
        sub_indent = ' ' * 2 * (level + 1)
        for file in files:
            print(f'{sub_indent}{file}')

print('Example 2: Meson build file structure completed successfully')
```

## Example 3: Meson Python Integration

Demonstrate meson's Python integration capabilities and features.

```python
# Example 3: Meson Python integration
import sys

print('Demonstrating Meson Python integration...')

# Sample meson.build with Python integration
python_meson_build = '''project('python_project', 'c',
  version : '1.0.0')

# Find Python installation
python_installation = import('python').find_installation('python3')

# Python extension module
python_installation.extension_module('mymodule',
  'src/mymodule.c',
  install : true)

# Python dependency
pydep = dependency('python3-dev')

# Custom target using Python
python_installation.install_sources(
  'python/mymodule.py',
  subdir : 'mymodule'
)
'''

# Sample Python extension C code
extension_c_code = '''#include <Python.h>

static PyObject* add_numbers(PyObject* self, PyObject* args) {
    int a, b;
    if (!PyArg_ParseTuple(args, "ii", &a, &b)) {
        return NULL;
    }
    return PyLong_FromLong(a + b);
}

static PyMethodDef module_methods[] = {
    {"add_numbers", add_numbers, METH_VARARGS, "Add two numbers"},
    {NULL, NULL, 0, NULL}
};

static struct PyModuleDef module_definition = {
    PyModuleDef_HEAD_INIT,
    "mymodule",
    "A simple example module",
    -1,
    module_methods
};

PyMODINIT_FUNC PyInit_mymodule(void) {
    return PyModule_Create(&module_definition);
}
'''

# Display Python integration examples
print('Python-specific meson.build content:')
for line in python_meson_build.split('\n')[:15]:
    if line.strip():
        print(f'  {line}')

print('\nPython extension patterns:')
patterns = {
    'python_installation': 'import("python").find_installation()',
    'extension_module': 'python_installation.extension_module()',
    'install_sources': 'python_installation.install_sources()',
    'python_dependency': 'dependency("python3-dev")'
}

for pattern, description in patterns.items():
    print(f'  {pattern}: {description}')

# Check Python development headers availability
try:
    import sysconfig
    python_include = sysconfig.get_path('include')
    print(f'\nPython include path: {python_include}')
    print(f'Python library path: {sysconfig.get_config_var("LIBDIR")}')
except Exception as e:
    print(f'\nPython config error: {e}')
    raise

print('Example 3: Meson Python integration completed successfully')
```

## Example 4: Meson Configuration and Options

Demonstrate meson configuration management and build options.

```python
# Example 4: Meson configuration and options
print('Demonstrating Meson configuration and build options...')

# Sample meson_options.txt file
meson_options_content = '''option('enable_debug', type : 'boolean', value : false, description : 'Enable debug build')
option('optimization', type : 'combo', choices : ['0', '1', '2', '3', 's'], value : '2', description : 'Optimization level')
option('enable_tests', type : 'boolean', value : true, description : 'Build tests')
option('prefix_path', type : 'string', value : '/usr/local', description : 'Installation prefix')
option('enable_docs', type : 'feature', value : 'auto', description : 'Build documentation')
'''

# Sample configuration usage in meson.build
config_meson_build = '''project('configurable_project', 'c', version : '1.0.0')

# Get options
debug_build = get_option('enable_debug')
opt_level = get_option('optimization')
build_tests = get_option('enable_tests')

# Configure based on options
if debug_build
  add_project_arguments('-DDEBUG', language : 'c')
endif

# Set compiler options
if opt_level == '3'
  add_project_arguments('-O3', '-march=native', language : 'c')
endif

# Conditional compilation
if build_tests
  subdir('tests')
endif

# Configuration data
conf_data = configuration_data()
conf_data.set('VERSION', meson.project_version())
conf_data.set('DEBUG', debug_build)

configure_file(
  input : 'config.h.in',
  output : 'config.h',
  configuration : conf_data
)
'''

# Display configuration examples
print('meson_options.txt content:')
for line in meson_options_content.split('\n')[:5]:
    if line.strip():
        print(f'  {line}')

print('\nConfiguration usage in meson.build:')
for line in config_meson_build.split('\n')[:12]:
    if line.strip():
        print(f'  {line}')

# Common meson configuration commands
config_commands = {
    'setup': 'meson setup builddir -Denable_debug=true',
    'configure': 'meson configure builddir -Doptimization=3',
    'reconfigure': 'meson setup --reconfigure builddir',
    'wipe': 'meson setup --wipe builddir'
}

print('\nCommon configuration commands:')
for cmd_type, command in config_commands.items():
    print(f'  {cmd_type}: {command}')

# Build types and options
build_options = {
    'buildtype': ['plain', 'debug', 'debugoptimized', 'release', 'minsize'],
    'optimization': ['0', '1', '2', '3', 's'],
    'debug': ['true', 'false'],
    'warning_level': ['0', '1', '2', '3']
}

print('\nBuilt-in build options:')
for option, values in build_options.items():
    print(f'  {option}: {", ".join(values)}')

print('Example 4: Meson configuration and options completed successfully')
```
