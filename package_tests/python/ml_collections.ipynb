## Example 1: Basic ConfigDict creation and manipulation

```python
import ml_collections
from ml_collections import ConfigDict

# Example 1: Basic ConfigDict creation and manipulation
print("ml_collections basic ConfigDict operations")

try:
    print("Example 1 - Basic ConfigDict creation and manipulation")
    
    # Create a basic configuration
    config = ConfigDict()
    
    # Add configuration parameters
    config.model_name = 'ResNet50'
    config.num_epochs = 100
    config.batch_size = 32
    config.learning_rate = 0.001
    config.optimizer = 'Adam'
    
    print(f"Basic configuration created:")
    print(f"  Model: {config.model_name}")
    print(f"  Epochs: {config.num_epochs}")
    print(f"  Batch size: {config.batch_size}")
    print(f"  Learning rate: {config.learning_rate}")
    print(f"  Optimizer: {config.optimizer}")
    
    # Nested configuration
    config.model_config = ConfigDict()
    config.model_config.num_layers = 50
    config.model_config.dropout_rate = 0.1
    config.model_config.activation = 'relu'
    
    config.training_config = ConfigDict()
    config.training_config.early_stopping = True
    config.training_config.patience = 10
    config.training_config.validation_split = 0.2
    
    config.data_config = ConfigDict()
    config.data_config.dataset_path = '/data/imagenet'
    config.data_config.augmentation = True
    config.data_config.normalization = {'mean': [0.485, 0.456, 0.406], 'std': [0.229, 0.224, 0.225]}
    
    print(f"\nNested configuration:")
    print(f"  Model layers: {config.model_config.num_layers}")
    print(f"  Dropout rate: {config.model_config.dropout_rate}")
    print(f"  Early stopping: {config.training_config.early_stopping}")
    print(f"  Validation split: {config.training_config.validation_split}")
    print(f"  Dataset path: {config.data_config.dataset_path}")
    print(f"  Augmentation: {config.data_config.augmentation}")
    
    # Dictionary-like operations
    print(f"\nDictionary-like operations:")
    
    # Access via keys
    print(f"  Access via key ['model_name']: {config['model_name']}")
    print(f"  Access via attr model_name: {config.model_name}")
    
    # Update configurations
    config.update({
        'seed': 42,
        'device': 'cuda',
        'num_workers': 4
    })
    
    print(f"  After update - Seed: {config.seed}, Device: {config.device}")
    
    # Check if keys exist
    has_model = 'model_name' in config
    has_missing = 'missing_key' in config
    print(f"  Has 'model_name': {has_model}")
    print(f"  Has 'missing_key': {has_missing}")
    assert config['model_name'] == config.model_name == 'ResNet50'
    assert config.seed == 42 and config.device == 'cuda'
    assert has_model and not has_missing
    
    # Get all keys
    all_keys = list(config.keys())
    print(f"  All top-level keys: {all_keys}")
    
    # Convert to regular dict
    regular_dict = config.to_dict()
    print(f"  Converted to dict type: {type(regular_dict).__name__}")
    print(f"  Dict keys: {list(regular_dict.keys())}")
    assert type(regular_dict) is dict
    assert regular_dict['model_config'] == {'num_layers': 50, 'dropout_rate': 0.1, 'activation': 'relu'}
    
    # Create from existing dict
    existing_dict = {
        'experiment_name': 'image_classification',
        'output_dir': '/experiments/run_001',
        'metrics': ['accuracy', 'precision', 'recall'],
        'hyperparams': {
            'momentum': 0.9,
            'weight_decay': 1e-4
        }
    }
    
    new_config = ConfigDict(existing_dict)
    print(f"\nCreated from existing dict:")
    print(f"  Experiment: {new_config.experiment_name}")
    print(f"  Metrics: {new_config.metrics}")
    print(f"  Momentum: {new_config.hyperparams.momentum}")
    # Nested dicts become ConfigDicts with attribute access
    assert isinstance(new_config.hyperparams, ConfigDict) and new_config.hyperparams.weight_decay == 1e-4
    
    # Show ConfigDict features
    print(f"\nConfigDict features:")
    config_features = [
        "Attribute access: Use dot notation like config.model_name",
        "Dictionary access: Use bracket notation like config['model_name']", 
        "Nested structures: Automatic ConfigDict creation for nested dicts",
        "JSON serializable: Convert to/from regular dictionaries",
        "Dynamic modification: Add/update/delete configuration parameters",
        "Type preservation: Maintains original data types"
    ]
    
    for feature in config_features:
        print(f"  • {feature}")
    
    # Test type handling
    print(f"\nType handling:")
    config.test_types = ConfigDict()
    config.test_types.integer = 42
    config.test_types.float_val = 3.14159
    config.test_types.string = "test_string"
    config.test_types.boolean = True
    config.test_types.list_val = [1, 2, 3, 4]
    config.test_types.none_val = None
    
    for key, value in config.test_types.items():
        print(f"  {key}: {value} (type: {type(value).__name__})")
    assert config.test_types.to_dict() == {
        'integer': 42, 'float_val': 3.14159, 'string': 'test_string',
        'boolean': True, 'list_val': [1, 2, 3, 4], 'none_val': None,
    }

    # ConfigDict is type-safe: a field keeps the type of its first value
    try:
        config.num_epochs = 'many'
    except TypeError:
        print("  Assigning a str to the int field num_epochs raised TypeError")
    else:
        raise AssertionError("ConfigDict accepted a str for an int field")
    
    print("ml_collections example 1 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"ml_collections example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: FrozenConfigDict and immutable configurations

```python
from ml_collections import ConfigDict, FrozenConfigDict

# Example 2: FrozenConfigDict and immutable configurations
print("ml_collections FrozenConfigDict and immutable configurations")

try:
    print("Example 2 - FrozenConfigDict and immutable configurations")
    
    # Create a mutable configuration first
    mutable_config = ConfigDict()
    mutable_config.model = 'transformer'
    mutable_config.hidden_size = 768
    mutable_config.num_heads = 12
    mutable_config.num_layers = 12
    mutable_config.vocab_size = 50257
    
    # Add optimizer configuration
    mutable_config.optimizer = ConfigDict()
    mutable_config.optimizer.name = 'AdamW'
    mutable_config.optimizer.lr = 1e-4
    mutable_config.optimizer.beta1 = 0.9
    mutable_config.optimizer.beta2 = 0.999
    mutable_config.optimizer.eps = 1e-8
    mutable_config.optimizer.weight_decay = 0.01
    
    # Add training configuration
    mutable_config.training = ConfigDict()
    mutable_config.training.batch_size = 16
    mutable_config.training.gradient_accumulation_steps = 4
    mutable_config.training.max_steps = 10000
    mutable_config.training.warmup_steps = 1000
    mutable_config.training.save_every = 1000
    
    print(f"Mutable configuration:")
    print(f"  Model: {mutable_config.model}")
    print(f"  Hidden size: {mutable_config.hidden_size}")
    print(f"  Optimizer: {mutable_config.optimizer.name}")
    print(f"  Learning rate: {mutable_config.optimizer.lr}")
    print(f"  Batch size: {mutable_config.training.batch_size}")
    
    # Test mutability
    print(f"\nTesting mutability:")
    original_lr = mutable_config.optimizer.lr
    mutable_config.optimizer.lr = 2e-4
    print(f"  Changed learning rate from {original_lr} to {mutable_config.optimizer.lr}")
    
    # Add new parameter
    mutable_config.new_param = "dynamically_added"
    print(f"  Added new parameter: {mutable_config.new_param}")
    
    # Create frozen version
    print(f"\nCreating frozen configuration:")
    frozen_config = FrozenConfigDict(mutable_config)
    
    print(f"Frozen configuration:")
    print(f"  Type: {type(frozen_config).__name__}")
    print(f"  Model: {frozen_config.model}")
    print(f"  Hidden size: {frozen_config.hidden_size}")
    print(f"  Optimizer type: {type(frozen_config.optimizer).__name__}")
    print(f"  Training type: {type(frozen_config.training).__name__}")
    
    # Test immutability
    print(f"\nTesting immutability:")
    
    # Try to modify frozen config (should fail)
    try:
        frozen_config.model = 'bert'
    except AttributeError as e:  # FrozenConfigDict refuses every __setattr__ with AttributeError
        assert "immutable" in str(e), f"unexpected message: {e!r}"
        print(f"  ✓ Cannot modify model: {type(e).__name__}")
    else:
        raise AssertionError("FrozenConfigDict accepted an assignment to model")
    
    # Try to modify nested frozen config (should fail)
    try:
        frozen_config.optimizer.lr = 3e-4
    except AttributeError as e:  # FrozenConfigDict refuses every __setattr__ with AttributeError
        assert "immutable" in str(e), f"unexpected message: {e!r}"
        print(f"  ✓ Cannot modify optimizer.lr: {type(e).__name__}")
    else:
        raise AssertionError("FrozenConfigDict accepted an assignment to the nested optimizer.lr")
    
    # Try to add new parameter (should fail)
    try:
        frozen_config.another_param = "should_fail"
    except AttributeError as e:  # FrozenConfigDict refuses every __setattr__ with AttributeError
        assert "immutable" in str(e), f"unexpected message: {e!r}"
        print(f"  ✓ Cannot add new parameter: {type(e).__name__}")
    else:
        raise AssertionError("FrozenConfigDict accepted a new key (another_param)")
    
    # Reading from frozen config should work
    print(f"\nReading from frozen config (should work):")
    print(f"  Model: {frozen_config.model}")
    print(f"  Learning rate: {frozen_config.optimizer.lr}")
    print(f"  Max steps: {frozen_config.training.max_steps}")
    
    # Convert back to mutable
    print(f"\nConverting back to mutable:")
    unfrozen_config = ConfigDict(frozen_config)
    print(f"  Type after unfreezing: {type(unfrozen_config).__name__}")
    
    # Test that unfrozen version is mutable
    unfrozen_config.test_modification = "this_should_work"
    unfrozen_config.optimizer.lr = 5e-4
    print(f"  Added test_modification: {unfrozen_config.test_modification}")
    print(f"  Modified lr: {unfrozen_config.optimizer.lr}")
    
    # Dictionary conversion
    print(f"\nDictionary conversions:")
    
    frozen_dict = frozen_config.to_dict()
    print(f"  Frozen to dict: {type(frozen_dict).__name__}")
    print(f"  Dict model: {frozen_dict['model']}")
    print(f"  Nested dict type: {type(frozen_dict['optimizer']).__name__}")
    
    # Create frozen from dict directly
    config_dict = {
        'architecture': 'CNN',
        'layers': {
            'conv1': {'filters': 32, 'kernel_size': 3},
            'conv2': {'filters': 64, 'kernel_size': 3},
            'dense': {'units': 128}
        },
        'training_params': {
            'epochs': 50,
            'batch_size': 64
        }
    }
    
    direct_frozen = FrozenConfigDict(config_dict)
    print(f"\nCreated frozen directly from dict:")
    print(f"  Architecture: {direct_frozen.architecture}")
    print(f"  Conv1 filters: {direct_frozen.layers.conv1.filters}")
    print(f"  Training epochs: {direct_frozen.training_params.epochs}")
    
    # Comparison and equality
    print(f"\nComparison and equality:")
    
    # Create identical configurations
    config1 = ConfigDict({'a': 1, 'b': {'c': 2}})
    config2 = ConfigDict({'a': 1, 'b': {'c': 2}})
    frozen1 = FrozenConfigDict(config1)
    frozen2 = FrozenConfigDict(config2)
    
    print(f"  ConfigDict equality: {config1 == config2}")
    print(f"  FrozenConfigDict equality: {frozen1 == frozen2}")
    print(f"  Mixed equality: {config1 == frozen1}")
    
    # Use cases for frozen configs
    print(f"\nUse cases for frozen configurations:")
    use_cases = [
        "Hyperparameter storage: Prevent accidental modification during training",
        "Model architecture: Ensure model structure remains constant", 
        "Experiment reproducibility: Lock configuration for consistent results",
        "Production deployment: Prevent runtime configuration changes",
        "Configuration validation: Ensure required parameters are not modified",
        "Multi-process safety: Share configurations safely across processes"
    ]
    
    for use_case in use_cases:
        print(f"  • {use_case}")
    
    # Performance and memory considerations
    print(f"\nConfiguration management patterns:")
    patterns = [
        "Development: Use mutable ConfigDict for experimentation",
        "Training start: Freeze configuration to prevent accidental changes",
        "Checkpointing: Save frozen config with model checkpoints",
        "Model serving: Use frozen config for inference deployment",
        "A/B testing: Compare frozen configurations for different experiments",
        "Config inheritance: Create variants from frozen base configurations"
    ]
    
    for pattern in patterns:
        print(f"  • {pattern}")
    
    print("ml_collections example 2 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"ml_collections example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: FieldReference and configuration templating

```python
from ml_collections import ConfigDict, FieldReference

# Example 3: FieldReference and configuration templating
print("ml_collections FieldReference and configuration templating")

try:
    print("Example 3 - FieldReference and configuration templating")

    # Create base configuration with field references
    config = ConfigDict()

    # Define base parameters
    config.base_lr = 1e-3
    config.num_epochs = 100
    config.batch_size = 32
    config.model_dim = 512

    # get_ref() returns a FieldReference to a field; arithmetic on it builds a lazy reference,
    # so each derived field is computed from the current value of its source when it is read.
    config.warmup_steps = config.get_ref('num_epochs') * 10
    config.total_steps = config.get_ref('num_epochs') * 1000
    config.lr_schedule = ConfigDict()
    config.lr_schedule.initial_lr = config.get_ref('base_lr') * 0.1
    config.lr_schedule.peak_lr = config.get_ref('base_lr')
    config.lr_schedule.final_lr = config.get_ref('base_lr') * 0.01

    # Model architecture with references
    config.model = ConfigDict()
    config.model.hidden_dim = config.get_ref('model_dim')
    config.model.ff_dim = config.get_ref('model_dim') * 4
    config.model.num_heads = config.get_ref('model_dim') // 64

    # Data processing with references
    config.data = ConfigDict()
    config.data.train_batch_size = config.get_ref('batch_size')
    config.data.eval_batch_size = config.get_ref('batch_size') * 2
    config.data.sequence_length = config.get_ref('model_dim') // 4

    print(f"Base configuration with references:")
    print(f"  Base LR: {config.base_lr}")
    print(f"  Epochs: {config.num_epochs}")
    print(f"  Batch size: {config.batch_size}")
    print(f"  Model dim: {config.model_dim}")

    # Reading a referenced field returns its computed value
    print(f"\nResolved field references:")
    print(f"  Warmup steps: {config.warmup_steps}")
    print(f"  Total steps: {config.total_steps}")
    print(f"  Initial LR: {config.lr_schedule.initial_lr}")
    print(f"  FF dimension: {config.model.ff_dim}")
    print(f"  Number of heads: {config.model.num_heads}")
    assert (config.warmup_steps, config.total_steps, config.model.ff_dim, config.model.num_heads) == (1000, 100000, 2048, 8)
    assert (config.model.hidden_dim, config.data.eval_batch_size, config.data.sequence_length) == (512, 64, 128)
    assert abs(config.lr_schedule.initial_lr - 1e-4) < 1e-12

    # Show impact of changing base parameters
    print(f"\nDemonstrating parameter coupling:")

    original_model_dim = config.model_dim
    config.model_dim = 768  # Change base parameter

    print(f"  Changed model_dim from {original_model_dim} to {config.model_dim}")
    print(f"  New FF dimension: {config.model.ff_dim} (4x model_dim)")
    print(f"  New number of heads: {config.model.num_heads} (model_dim // 64)")
    # Every field derived from model_dim follows the new value
    assert (config.model.hidden_dim, config.model.ff_dim, config.model.num_heads) == (768, 3072, 12)
    assert config.data.sequence_length == 192

    # A FieldReference can also be created directly and shared by several configurations
    shared_seed = FieldReference(0)
    train_cfg, eval_cfg = ConfigDict(), ConfigDict()
    train_cfg.seed = shared_seed
    eval_cfg.seed = shared_seed
    shared_seed.set(42)
    print(f"  Shared seed after set(42): train={train_cfg.seed}, eval={eval_cfg.seed}")
    assert train_cfg.seed == eval_cfg.seed == 42

    # Complex field references with multiple dependencies
    print(f"\nComplex field references:")

    config.optimization = ConfigDict()
    model_dim_ref = config.get_ref('model_dim')
    batch_size_ref = config.get_ref('batch_size')

    # Memory usage estimation based on multiple parameters
    config.optimization.estimated_memory_gb = model_dim_ref * model_dim_ref * batch_size_ref * 4 / (1024**3)

    # Gradient accumulation to reach an effective batch of 64
    config.optimization.grad_accumulation_steps = 64 // batch_size_ref

    # Effective batch size calculation, chained on another derived field
    config.optimization.effective_batch_size = batch_size_ref * config.optimization.get_ref('grad_accumulation_steps')

    print(f"  Estimated memory: {config.optimization.estimated_memory_gb:.4f} GB")
    print(f"  Gradient accumulation steps: {config.optimization.grad_accumulation_steps}")
    print(f"  Effective batch size: {config.optimization.effective_batch_size}")
    assert config.optimization.estimated_memory_gb == 768 * 768 * 32 * 4 / 1024**3
    assert (config.optimization.grad_accumulation_steps, config.optimization.effective_batch_size) == (2, 64)

    # Halving the batch size doubles the accumulation steps and keeps the effective batch
    config.batch_size = 16
    print(f"  After batch_size = 16: {config.optimization.grad_accumulation_steps} steps, "
          f"effective batch {config.optimization.effective_batch_size}")
    assert (config.optimization.grad_accumulation_steps, config.optimization.effective_batch_size) == (4, 64)

    # Configuration templates and inheritance
    print(f"\nConfiguration templates:")

    # Create base template
    def create_base_config():
        base = ConfigDict()
        base.model_type = "transformer"
        base.vocab_size = 50257
        base.max_seq_length = 1024
        return base

    def create_model_config(model_dim, num_layers):
        config = create_base_config()
        config.model_dim = model_dim
        config.num_layers = num_layers
        config.num_heads = config.get_ref('model_dim') // 64
        config.ff_dim = config.get_ref('model_dim') * 4
        return config

    # Generate different model configurations
    small_config = create_model_config(256, 6)
    large_config = create_model_config(1024, 24)

    for label, model_config in (("Small", small_config), ("Large", large_config)):
        print(f"  {label} model:")
        print(f"    Model dim: {model_config.model_dim}")
        print(f"    Num layers: {model_config.num_layers}")
        print(f"    Num heads: {model_config.num_heads}")
        print(f"    FF dim: {model_config.ff_dim}")
    assert (small_config.num_heads, small_config.ff_dim) == (4, 1024)
    assert (large_config.num_heads, large_config.ff_dim) == (16, 4096)

    # Field reference benefits
    print(f"\nField reference benefits:")
    benefits = [
        "Parameter coupling: Derived parameters automatically update with base changes",
        "Configuration consistency: Ensures related parameters stay synchronized",
        "Template reuse: Create configuration families with shared parameter relationships",
        "Lazy computation: Derived values are computed from the current values when read",
        "Experiment scaling: Easy to scale entire configuration by changing base parameters"
    ]

    for benefit in benefits:
        print(f"  • {benefit}")

    # Common field reference patterns
    print(f"\nCommon field reference patterns:")
    patterns = [
        "Scaling relationships: ff_dim = model_dim * 4",
        "Batch size calculations: eval_batch = train_batch * 2",
        "Learning rate schedules: peak_lr = base_lr, warmup_lr = base_lr * 0.1",
        "Memory optimization: grad_acc_steps = target_batch // actual_batch",
        "Architecture ratios: num_heads = model_dim // head_dim"
    ]

    for pattern in patterns:
        print(f"  • {pattern}")

    print("ml_collections example 3 completed successfully")

except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"ml_collections example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished")
```

## Example 4: Real-world ML experiment configuration management

```python
from ml_collections import ConfigDict, FrozenConfigDict
import json

# Example 4: Real-world ML experiment configuration management
print("ml_collections real-world ML experiment configuration")

try:
    print("Example 4 - Real-world ML experiment configuration management")
    
    # Create comprehensive experiment configuration
    def create_experiment_config(experiment_name="image_classification_v1"):
        """Create a comprehensive ML experiment configuration"""
        
        config = ConfigDict()
        
        # Experiment metadata
        config.experiment = ConfigDict()
        config.experiment.name = experiment_name
        config.experiment.version = "1.0"
        config.experiment.description = "Image classification with ResNet on ImageNet"
        config.experiment.tags = ["computer_vision", "classification", "resnet"]
        config.experiment.author = "ml_team"
        
        # Dataset configuration
        config.dataset = ConfigDict()
        config.dataset.name = "imagenet"
        config.dataset.data_dir = "/data/imagenet"
        config.dataset.num_classes = 1000
        config.dataset.image_size = 224
        config.dataset.train_split = 0.9
        config.dataset.val_split = 0.1
        config.dataset.augmentation = ConfigDict()
        config.dataset.augmentation.random_crop = True
        config.dataset.augmentation.random_flip = True
        config.dataset.augmentation.color_jitter = 0.1
        config.dataset.augmentation.normalize = True
        
        # Model architecture
        config.model = ConfigDict()
        config.model.architecture = "resnet50"
        config.model.pretrained = True
        config.model.num_classes = config.dataset.num_classes
        config.model.dropout_rate = 0.1
        config.model.batch_norm = True
        
        # Training configuration
        config.training = ConfigDict()
        config.training.batch_size = 128
        config.training.num_epochs = 90
        config.training.learning_rate = 0.1
        config.training.momentum = 0.9
        config.training.weight_decay = 1e-4
        config.training.label_smoothing = 0.1
        
        # Learning rate schedule
        config.training.lr_schedule = ConfigDict()
        config.training.lr_schedule.type = "cosine_annealing"
        config.training.lr_schedule.warmup_epochs = 5
        config.training.lr_schedule.min_lr = 1e-6
        
        # Optimization
        config.optimizer = ConfigDict()
        config.optimizer.name = "SGD"
        config.optimizer.nesterov = True
        config.optimizer.clip_grad_norm = 5.0
        
        # Evaluation and logging
        config.evaluation = ConfigDict()
        config.evaluation.metrics = ["accuracy", "top5_accuracy", "loss"]
        config.evaluation.eval_frequency = 5  # epochs
        config.evaluation.save_frequency = 10  # epochs
        
        config.logging = ConfigDict()
        config.logging.log_frequency = 100  # steps
        config.logging.tensorboard = True
        config.logging.wandb = ConfigDict()
        config.logging.wandb.enabled = True
        config.logging.wandb.project = "image_classification"
        config.logging.wandb.entity = "ml_team"
        
        # Infrastructure
        config.infrastructure = ConfigDict()
        config.infrastructure.device = "cuda"
        config.infrastructure.num_gpus = 4
        config.infrastructure.mixed_precision = True
        config.infrastructure.distributed = True
        config.infrastructure.num_workers = 8
        
        # Checkpointing
        config.checkpointing = ConfigDict()
        config.checkpointing.save_dir = f"/checkpoints/{experiment_name}"
        config.checkpointing.save_best = True
        config.checkpointing.save_last = True
        config.checkpointing.resume_from = None
        
        return config
    
    # Create experiment configurations
    print("Creating experiment configurations:")
    
    base_config = create_experiment_config("resnet50_baseline")
    print(f"  Base experiment: {base_config.experiment.name}")
    print(f"  Model: {base_config.model.architecture}")
    print(f"  Batch size: {base_config.training.batch_size}")
    print(f"  Learning rate: {base_config.training.learning_rate}")
    
    # Create variant configurations
    configs = {}
    configs['baseline'] = base_config
    
    # High learning rate variant
    configs['high_lr'] = ConfigDict(base_config)
    configs['high_lr'].experiment.name = "resnet50_high_lr"
    configs['high_lr'].training.learning_rate = 0.2
    configs['high_lr'].training.lr_schedule.warmup_epochs = 10
    
    # Large batch variant
    configs['large_batch'] = ConfigDict(base_config)
    configs['large_batch'].experiment.name = "resnet50_large_batch"
    configs['large_batch'].training.batch_size = 256
    configs['large_batch'].training.learning_rate = 0.2  # Scale with batch size
    configs['large_batch'].infrastructure.num_gpus = 8
    
    # Data augmentation variant
    configs['heavy_aug'] = ConfigDict(base_config)
    configs['heavy_aug'].experiment.name = "resnet50_heavy_aug"
    configs['heavy_aug'].dataset.augmentation.cutmix = True
    configs['heavy_aug'].dataset.augmentation.mixup = True
    configs['heavy_aug'].dataset.augmentation.randaugment = ConfigDict()
    configs['heavy_aug'].dataset.augmentation.randaugment.num_ops = 2
    configs['heavy_aug'].dataset.augmentation.randaugment.magnitude = 9
    
    print(f"\nCreated {len(configs)} experiment variants:")
    for name, config in configs.items():
        print(f"  {name}: {config.experiment.name}")
        print(f"    LR: {config.training.learning_rate}, Batch: {config.training.batch_size}")
    
    # Configuration validation
    print(f"\nConfiguration validation:")
    
    def validate_config(config, experiment_name):
        """Validate experiment configuration"""
        errors = []
        warnings = []
        
        # Required fields validation
        required_fields = [
            'experiment.name',
            'model.architecture', 
            'training.batch_size',
            'training.learning_rate',
            'dataset.num_classes'
        ]
        
        for field in required_fields:
            parts = field.split('.')
            current = config
            try:
                for part in parts:
                    current = current[part]
                if current is None:
                    errors.append(f"Required field {field} is None")
            except KeyError:
                errors.append(f"Missing required field: {field}")
        
        # Validation rules
        if config.training.learning_rate > 1.0:
            warnings.append("Learning rate > 1.0 is unusually high")
        
        if config.training.batch_size < 16:
            warnings.append("Very small batch size may cause training instability")
        
        if config.model.dropout_rate > 0.5:
            warnings.append("High dropout rate may hurt model capacity")
        
        # Infrastructure validation
        if config.infrastructure.num_gpus > 1 and not config.infrastructure.distributed:
            warnings.append("Multiple GPUs without distributed training")
        
        if config.training.batch_size % config.infrastructure.num_gpus != 0:
            warnings.append("Batch size not divisible by number of GPUs")
        
        return errors, warnings
    
    # Validate all configurations
    for name, config in configs.items():
        errors, warnings = validate_config(config, name)
        print(f"  {name}:")
        print(f"    Errors: {len(errors)}")
        print(f"    Warnings: {len(warnings)}")
        
        for error in errors[:2]:  # Show first 2 errors
            print(f"      Error: {error}")
        for warning in warnings[:2]:  # Show first 2 warnings  
            print(f"      Warning: {warning}")
    
    # Configuration serialization and persistence
    print(f"\nConfiguration serialization:")
    
    # Convert to regular dict for JSON serialization
    baseline_dict = configs['baseline'].to_dict()
    
    # Serialize to JSON
    json_config = json.dumps(baseline_dict, indent=2, default=str)
    print(f"  JSON serialization: {len(json_config)} characters")
    
    # Deserialize back to ConfigDict
    loaded_dict = json.loads(json_config)
    loaded_config = ConfigDict(loaded_dict)
    
    print(f"  Deserialized config type: {type(loaded_config).__name__}")
    print(f"  Config preserved: {loaded_config.experiment.name}")
    print(f"  Nested access: {loaded_config.model.architecture}")
    
    # Configuration comparison
    print(f"\nConfiguration comparison:")
    
    def compare_configs(config1, config2, name1="Config1", name2="Config2"):
        """Compare two configurations and show differences"""
        dict1 = config1.to_dict() if hasattr(config1, 'to_dict') else config1
        dict2 = config2.to_dict() if hasattr(config2, 'to_dict') else config2
        
        def flatten_dict(d, prefix=""):
            items = []
            for k, v in d.items():
                new_key = f"{prefix}.{k}" if prefix else k
                if isinstance(v, dict):
                    items.extend(flatten_dict(v, new_key))
                else:
                    items.append((new_key, v))
            return items
        
        flat1 = dict(flatten_dict(dict1))
        flat2 = dict(flatten_dict(dict2))
        
        all_keys = set(flat1.keys()) | set(flat2.keys())
        differences = []
        
        for key in sorted(all_keys):
            val1 = flat1.get(key, "<missing>")
            val2 = flat2.get(key, "<missing>")
            if val1 != val2:
                differences.append((key, val1, val2))
        
        return differences
    
    # Compare baseline with variants
    for variant_name in ['high_lr', 'large_batch']:
        differences = compare_configs(
            configs['baseline'], 
            configs[variant_name],
            'baseline', 
            variant_name
        )
        
        print(f"  {variant_name} differences from baseline:")
        for key, baseline_val, variant_val in differences[:5]:  # Show first 5
            print(f"    {key}: {baseline_val} → {variant_val}")
    
    # Configuration freezing for production
    print(f"\nProduction configuration management:")
    
    # Freeze configuration for training
    production_config = FrozenConfigDict(configs['baseline'])
    
    print(f"  Frozen production config: {type(production_config).__name__}")
    print(f"  Experiment name: {production_config.experiment.name}")
    
    # Test immutability
    try:
        production_config.training.learning_rate = 0.05
    except AttributeError as e:
        assert "FrozenConfigDict is immutable" in str(e), e
        print(f"  ✓ Production config is immutable: {type(e).__name__}")
    else:
        raise AssertionError("assigning into a FrozenConfigDict should raise AttributeError")
    
    # Configuration management best practices
    print(f"\nConfiguration management best practices:")
    best_practices = [
        "Version control: Store configs in version control with code",
        "Validation: Validate configurations before training starts",
        "Templating: Use templates for common experiment patterns",
        "Serialization: Save configurations with model checkpoints",
        "Comparison: Track differences between experiment variants",
        "Freezing: Lock configurations during production training",
        "Documentation: Include experiment descriptions and metadata",
        "Reproducibility: Fix all random seeds and configuration parameters"
    ]
    
    for practice in best_practices:
        print(f"  • {practice}")
    
    print("ml_collections example 4 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"ml_collections example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

```python
import copy
import ml_collections
from ml_collections import ConfigDict, FrozenConfigDict

# Example 5: ml_collections advanced features and production workflow
print("ml_collections advanced features and production workflow")

try:
    print("Example 5 - Advanced ml_collections features and production workflow")
    
    # Demonstrate package structure and capabilities
    print(f"ml_collections Package Overview:")
    print(f"  Module type: {type(ml_collections).__name__}")
    print(f"  Package location: {ml_collections.__file__ if hasattr(ml_collections, '__file__') else 'Built-in'}")
    
    # Package version and metadata
    if hasattr(ml_collections, '__version__'):
        print(f"  Version: {ml_collections.__version__}")
    
    # Key components
    key_components = {
        'ConfigDict': 'Mutable configuration dictionary with attribute access',
        'FrozenConfigDict': 'Immutable configuration dictionary',
        'FieldReference': 'Dynamic field references for computed parameters'
    }
    
    print(f"  Key Components:")
    for component, description in key_components.items():
        try:
            comp = getattr(ml_collections, component)
            print(f"    ✓ {component}: {description}")
        except AttributeError:
            print(f"    ○ {component}: Not directly accessible")
    assert all(hasattr(ml_collections, component) for component in key_components)
    
    # Production workflow demonstration
    print(f"\nProduction Workflow Pattern:")
    
    # Step 1: Configuration hierarchy and inheritance
    print(f"  1. Configuration Hierarchy:")
    
    class MLConfigurationManager:
        """Production configuration management system"""
        
        def __init__(self):
            self.base_configs = {}
            self.experiment_configs = {}
        
        def create_base_config(self, config_name):
            """Create base configuration template"""
            base = ConfigDict()
            
            # Common base settings
            base.seed = 42
            base.experiment_group = "default"
            base.logging_level = "INFO"
            base.output_dir = "/experiments"
            
            # Model defaults
            base.model = ConfigDict()
            base.model.type = "neural_network" 
            base.model.activation = "relu"
            base.model.dropout = 0.1
            
            # Training defaults
            base.training = ConfigDict()
            base.training.epochs = 100
            base.training.batch_size = 32
            base.training.learning_rate = 1e-3
            base.training.patience = 10
            
            # Data defaults
            base.data = ConfigDict()
            base.data.validation_split = 0.2
            base.data.shuffle = True
            base.data.preprocessing = ConfigDict()
            base.data.preprocessing.normalize = True
            base.data.preprocessing.standardize = False
            
            self.base_configs[config_name] = base
            return base
        
        def create_task_specific_config(self, base_name, task_name, overrides=None):
            """Create task-specific configuration from base"""
            if base_name not in self.base_configs:
                raise ValueError(f"Base config '{base_name}' not found")
            
            # Start with a deep copy: ConfigDict(other) shares the nested ConfigDicts
            task_config = copy.deepcopy(self.base_configs[base_name])
            task_config.task_name = task_name
            
            # Apply task-specific overrides
            if overrides:
                self._deep_update(task_config, overrides)
            
            self.experiment_configs[task_name] = task_config
            return task_config
        
        def _deep_update(self, base_config, updates):
            """Deep update configuration with new values"""
            for key, value in updates.items():
                if key in base_config and isinstance(base_config[key], ConfigDict) and isinstance(value, dict):
                    self._deep_update(base_config[key], value)
                else:
                    base_config[key] = value
    
    # Test configuration management
    config_manager = MLConfigurationManager()
    
    # Create base configurations
    base_config = config_manager.create_base_config("ml_base")
    print(f"     Created base config with {len(base_config.keys())} sections")
    base_snapshot = base_config.to_dict()
    
    # Create task-specific configurations
    classification_overrides = {
        'model': {
            'type': 'classifier',
            'num_classes': 10,
            'final_activation': 'softmax'
        },
        'training': {
            'loss_function': 'categorical_crossentropy',
            'metrics': ['accuracy', 'precision', 'recall']
        },
        'data': {
            'augmentation': True,
            'class_weights': 'balanced'
        }
    }
    
    regression_overrides = {
        'model': {
            'type': 'regressor', 
            'output_dim': 1,
            'final_activation': 'linear'
        },
        'training': {
            'loss_function': 'mse',
            'metrics': ['mae', 'rmse']
        },
        'data': {
            'target_scaling': 'standard'
        }
    }
    
    classification_config = config_manager.create_task_specific_config(
        "ml_base", "image_classification", classification_overrides
    )
    
    regression_config = config_manager.create_task_specific_config(
        "ml_base", "house_price_prediction", regression_overrides
    )
    
    print(f"     Classification model type: {classification_config.model.type}")
    print(f"     Classification metrics: {classification_config.training.metrics}")
    print(f"     Regression model type: {regression_config.model.type}")
    print(f"     Regression loss: {regression_config.training.loss_function}")
    assert classification_config.model.type == 'classifier' and classification_config.model.num_classes == 10
    assert regression_config.model.type == 'regressor' and regression_config.training.loss_function == 'mse'
    assert 'num_classes' not in regression_config.model
    assert base_config.to_dict() == base_snapshot  # the template is left untouched
    
    # Step 2: Configuration validation and type checking
    print(f"  2. Configuration Validation:")
    
    class ConfigValidator:
        """Validate ML configurations"""
        
        def __init__(self):
            self.validation_rules = {
                'training.learning_rate': self._validate_learning_rate,
                'training.batch_size': self._validate_batch_size,
                'model.dropout': self._validate_dropout,
                'data.validation_split': self._validate_split,
            }
        
        def validate_config(self, config, config_name):
            """Comprehensive configuration validation"""
            errors = []
            warnings = []
            
            # Type validation
            try:
                if not isinstance(config.training.epochs, int):
                    errors.append("training.epochs must be integer")
                    
                if not isinstance(config.training.learning_rate, (int, float)):
                    errors.append("training.learning_rate must be numeric")
                    
            except AttributeError as e:
                errors.append(f"Missing required field: {e}")
            
            # Custom rule validation
            for field_path, validator in self.validation_rules.items():
                try:
                    value = self._get_nested_value(config, field_path)
                    field_errors, field_warnings = validator(value, field_path)
                    errors.extend(field_errors)
                    warnings.extend(field_warnings)
                except Exception as e:
                    errors.append(f"Validation error for {field_path}: {e}")
            
            return errors, warnings
        
        def _get_nested_value(self, config, path):
            """Get nested configuration value by path"""
            keys = path.split('.')
            value = config
            for key in keys:
                value = value[key]
            return value
        
        def _validate_learning_rate(self, lr, path):
            errors, warnings = [], []
            if lr <= 0:
                errors.append(f"{path}: must be positive")
            elif lr > 1.0:
                warnings.append(f"{path}: unusually high ({lr})")
            elif lr < 1e-6:
                warnings.append(f"{path}: unusually low ({lr})")
            return errors, warnings
        
        def _validate_batch_size(self, batch_size, path):
            errors, warnings = [], []
            if batch_size <= 0:
                errors.append(f"{path}: must be positive")
            elif batch_size < 8:
                warnings.append(f"{path}: very small batch size may cause instability")
            elif batch_size > 512:
                warnings.append(f"{path}: large batch size may require memory optimization")
            return errors, warnings
        
        def _validate_dropout(self, dropout, path):
            errors, warnings = [], []
            if not 0 <= dropout <= 1:
                errors.append(f"{path}: must be between 0 and 1")
            elif dropout > 0.5:
                warnings.append(f"{path}: high dropout may reduce model capacity")
            return errors, warnings
        
        def _validate_split(self, split, path):
            errors, warnings = [], []
            if not 0 < split < 1:
                errors.append(f"{path}: must be between 0 and 1")
            elif split < 0.1:
                warnings.append(f"{path}: very small validation split")
            elif split > 0.3:
                warnings.append(f"{path}: large validation split reduces training data")
            return errors, warnings
    
    # Test validation
    validator = ConfigValidator()
    
    for config_name, config in [("classification", classification_config), ("regression", regression_config)]:
        errors, warnings = validator.validate_config(config, config_name)
        print(f"     {config_name}: {len(errors)} errors, {len(warnings)} warnings")
        assert not errors, errors
        
        for error in errors[:2]:
            print(f"       Error: {error}")
        for warning in warnings[:2]:
            print(f"       Warning: {warning}")
    
    # Step 3: Environment-specific configurations
    print(f"  3. Environment-Specific Configurations:")
    
    def create_environment_config(base_config, environment):
        """Create environment-specific configuration"""
        env_config = copy.deepcopy(base_config)
        env_config.environment = environment
        
        if environment == "development":
            env_config.training.epochs = 5  # Fast iteration
            env_config.model.dropout = 0.2  # Higher for debugging
            env_config.logging_level = "DEBUG"
            env_config.save_frequency = 1
            
        elif environment == "staging":  
            env_config.training.epochs = 20  # Medium run
            env_config.model.dropout = 0.1  # Moderate
            env_config.logging_level = "INFO"
            env_config.save_frequency = 5
            env_config.validation_enabled = True
            
        elif environment == "production":
            env_config.training.epochs = 100  # Full training
            env_config.model.dropout = 0.05  # Conservative
            env_config.logging_level = "WARNING"
            env_config.save_frequency = 10
            env_config.monitoring_enabled = True
            env_config.auto_scaling = True
            
        return FrozenConfigDict(env_config)  # Freeze for immutability
    
    environments = ["development", "staging", "production"]
    env_configs = {}
    
    for env in environments:
        env_config = create_environment_config(classification_config, env)
        env_configs[env] = env_config
        print(f"     {env}: {env_config.training.epochs} epochs, {env_config.logging_level} logging")
    assert [env_configs[env].training.epochs for env in environments] == [5, 20, 100]
    assert all(isinstance(cfg, FrozenConfigDict) for cfg in env_configs.values())
    assert classification_config.training.epochs == 100  # the source config is not modified
    try:
        env_configs['production'].training.epochs = 1
    except AttributeError:
        pass
    else:
        raise AssertionError("FrozenConfigDict accepted an assignment")
    
    # Step 4: Configuration management utilities
    print(f"  4. Configuration Management Utilities:")
    
    class ConfigurationUtils:
        """Utilities for configuration management"""
        
        @staticmethod
        def merge_configs(base_config, override_config):
            """Merge two configurations with override precedence"""
            merged = copy.deepcopy(base_config)
            
            def deep_merge(base, override):
                for key, value in override.items():
                    if key in base and isinstance(base[key], ConfigDict) and isinstance(value, dict):
                        deep_merge(base[key], value)
                    else:
                        base[key] = value
            
            deep_merge(merged, override_config)
            return merged
        
        @staticmethod
        def config_summary(config):
            """Generate configuration summary"""
            def count_params(cfg, prefix=""):
                count = 0
                for key, value in cfg.items():
                    if isinstance(value, ConfigDict):
                        count += count_params(value, f"{prefix}.{key}" if prefix else key)
                    else:
                        count += 1
                return count
            
            total_params = count_params(config)
            sections = list(config.keys())
            
            return {
                'total_parameters': total_params,
                'sections': sections,
                'is_frozen': isinstance(config, FrozenConfigDict)
            }
        
        @staticmethod
        def export_for_logging(config):
            """Export config in format suitable for experiment logging"""
            flat_config = {}
            
            def flatten(cfg, prefix=""):
                for key, value in cfg.items():
                    full_key = f"{prefix}.{key}" if prefix else key
                    if isinstance(value, ConfigDict):
                        flatten(value, full_key)
                    else:
                        flat_config[full_key] = value
            
            flatten(config)
            return flat_config
    
    utils = ConfigurationUtils()
    
    # Test utilities
    for env_name, env_config in list(env_configs.items())[:2]:
        summary = utils.config_summary(env_config)
        print(f"     {env_name}: {summary['total_parameters']} parameters, frozen: {summary['is_frozen']}")
        
        flat_config = utils.export_for_logging(env_config)
        print(f"       Flattened: {len(flat_config)} key-value pairs")
        assert summary['is_frozen'] and summary['total_parameters'] == len(flat_config)
        assert flat_config['environment'] == env_name
    
    # Step 5: Best practices and patterns
    print(f"  5. Best Practices:")
    
    best_practices = [
        "Hierarchical organization: Use nested ConfigDicts for logical grouping",
        "Environment separation: Different configs for dev/staging/production",
        "Validation integration: Validate configs before experiment execution",
        "Immutable production: Use FrozenConfigDict for production deployments",
        "Version tracking: Include config versioning in experiment metadata",
        "Template inheritance: Build specialized configs from common templates",
        "Serialization support: Ensure configs are JSON/YAML serializable",
        "Documentation: Include descriptions and metadata in configuration"
    ]
    
    for i, practice in enumerate(best_practices, 1):
        print(f"     {i}. {practice}")
    
    # Package ecosystem integration
    print(f"\nEcosystem Integration:")
    integrations = [
        "ML Frameworks: TensorFlow, PyTorch, JAX configuration management",
        "Experiment tracking: Weights & Biases, MLflow, Neptune integration",
        "Hyperparameter tuning: Optuna, Ray Tune parameter space definition",
        "Model serving: Configuration for inference pipelines and deployment",
        "Data pipelines: ETL and preprocessing configuration management",
        "Infrastructure: Kubernetes, Docker container configuration"
    ]
    
    for integration in integrations:
        print(f"  • {integration}")
    
    # Production deployment patterns
    print(f"\nProduction Deployment Patterns:")
    deployment_patterns = [
        "Configuration validation: Strict validation before training/inference",
        "Environment isolation: Separate configurations for different stages",
        "A/B testing: Variant configurations for controlled experiments",
        "Rollback support: Version-controlled configuration management",
        "Monitoring integration: Configuration-driven monitoring and alerting",
        "Auto-scaling: Dynamic configuration based on resource utilization"
    ]
    
    for pattern in deployment_patterns:
        print(f"  • {pattern}")
    
    print("ml_collections example 5 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"ml_collections example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
