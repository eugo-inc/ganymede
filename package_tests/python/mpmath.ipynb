# mpmath: 5 Basic Usage Examples

```python
# Example 1: Arbitrary-precision arithmetic
import mpmath
mpmath.mp.dps = 50  # set decimal places
root2 = mpmath.mpf('2') ** mpmath.mpf('0.5')
print(root2)
assert mpmath.nstr(root2, 15, strip_zeros=False) == '1.41421356237310'
assert mpmath.almosteq(root2 * root2, 2)
```

```python
# Example 2: Special functions
import mpmath
g = mpmath.gamma(0.5)
print(g)
assert mpmath.almosteq(g, mpmath.sqrt(mpmath.pi))
```

```python
# Example 3: Numerical integration
import mpmath
integral = mpmath.quad(lambda x: mpmath.exp(-x**2), [0, mpmath.inf])
print(integral)
assert mpmath.almosteq(integral, mpmath.sqrt(mpmath.pi) / 2)
```

```python
# Example 4: Matrix operations
import mpmath
A = mpmath.matrix([[1, 2], [3, 4]])
print(A**2)
assert (A**2).tolist() == [[7, 10], [15, 22]]
```

```python
# Example 5: Show mpmath version
import mpmath
print(mpmath.__version__)
assert mpmath.__version__.split('.')[0].isdigit()
```
