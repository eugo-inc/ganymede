## Example 1: msgspec functionality test 1

```python
import msgspec
from datetime import datetime

# Example 1: Basic struct definition and JSON serialization
print("Example 1: Basic Struct Definition and JSON Serialization")

try:
    # Define a structured data class using msgspec
    class User(msgspec.Struct):
        id: int
        name: str
        email: str
        is_active: bool = True
        created_at: datetime = msgspec.field(default_factory=datetime.now)
    
    print("User struct defined with typed fields")
    
    # Create user instances
    users = [
        User(id=1, name="Alice Johnson", email="alice@example.com"),
        User(id=2, name="Bob Smith", email="bob@example.com", is_active=False),
        User(id=3, name="Carol Davis", email="carol@company.org")
    ]
    
    print(f"Created {len(users)} user instances")
    
    # Serialize to JSON
    print("\nJSON Serialization:")
    for user in users:
        json_data = msgspec.json.encode(user)
        print(f"  User {user.id}: {json_data.decode()}")
    
    # Bulk serialization
    all_users_json = msgspec.json.encode(users)
    print(f"Bulk JSON: {all_users_json.decode()[:100]}...")
    
    # Deserialize from JSON
    print("\nJSON Deserialization:")
    json_input = '{"id": 99, "name": "Test User", "email": "test@example.com", "is_active": true}'
    
    decoded_user = msgspec.json.decode(json_input.encode(), type=User)
    print(f"Decoded: {decoded_user}")
    print(f"Type: {type(decoded_user).__name__}")
    print(f"Active: {decoded_user.is_active}")
    
    # Show struct benefits
    print(f"\nStruct capabilities:")
    struct_features = [
        "Type safety: Compile-time type checking with annotations",
        "Default values: Support for default and factory defaults", 
        "Fast serialization: Optimized JSON encoding/decoding",
        "Memory efficiency: Compact representation in memory",
        "Schema validation: Automatic validation during deserialization"
    ]
    
    for feature in struct_features:
        print(f"  • {feature}")
    
    # Performance comparison
    import json
    import time
    
    # Test serialization performance
    test_user = users[0]
    
    # msgspec timing
    start = time.time()
    for _ in range(1000):
        msgspec.json.encode(test_user)
    msgspec_time = time.time() - start
    
    # Standard json timing (with dict conversion)
    user_dict = {"id": test_user.id, "name": test_user.name, "email": test_user.email, 
                "is_active": test_user.is_active, "created_at": test_user.created_at.isoformat()}
    
    start = time.time()
    for _ in range(1000):
        json.dumps(user_dict)
    json_time = time.time() - start
    
    print(f"\nPerformance comparison (1000 iterations):")
    print(f"  msgspec: {msgspec_time:.4f}s")
    print(f"  stdlib json: {json_time:.4f}s")
    print(f"  msgspec is {json_time/msgspec_time:.1f}x faster")
    
    print("Basic struct and JSON serialization completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    # Fallback demonstration
    print("Demonstrating msgspec concepts:")
    concepts = [
        "Struct definition: Typed data classes with automatic serialization",
        "JSON performance: High-speed encoding/decoding operations", 
        "Type safety: Compile-time and runtime type validation",
        "Schema validation: Automatic data validation during processing",
        "Memory efficiency: Optimized memory usage and representation"
    ]
    
    for concept in concepts:
        print(f"  - {concept}")
    raise
```

## Example 2: msgspec functionality test 2

```python
import msgspec
from typing import List, Optional, Dict, Any

# Example 2: MessagePack serialization and complex types
print("Example 2: MessagePack Serialization and Complex Types")

try:
    # Define complex nested structures
    class Address(msgspec.Struct):
        street: str
        city: str
        state: str
        zip_code: str
        country: str = "USA"
    
    class Contact(msgspec.Struct):
        phone: Optional[str] = None
        email: Optional[str] = None
        preferred: str = "email"
    
    class Person(msgspec.Struct):
        id: int
        name: str
        age: int
        addresses: List[Address]
        contact: Contact
        tags: List[str] = msgspec.field(default_factory=list)
        metadata: Dict[str, Any] = msgspec.field(default_factory=dict)
    
    print("Complex nested struct hierarchy defined")
    
    # Create complex data
    person = Person(
        id=1001,
        name="John Doe",
        age=35,
        addresses=[
            Address(
                street="123 Main St",
                city="San Francisco", 
                state="CA",
                zip_code="94105"
            ),
            Address(
                street="456 Oak Ave",
                city="Berkeley",
                state="CA", 
                zip_code="94701"
            )
        ],
        contact=Contact(
            phone="+1-555-123-4567",
            email="john.doe@example.com"
        ),
        tags=["developer", "python", "data"],
        metadata={"department": "engineering", "level": "senior", "remote": True}
    )
    
    print(f"Created complex person object: {person.name}")
    print(f"  Addresses: {len(person.addresses)}")
    print(f"  Tags: {person.tags}")
    print(f"  Metadata: {person.metadata}")
    
    # MessagePack serialization
    print(f"\nMessagePack Serialization:")
    
    # Encode to MessagePack
    msgpack_data = msgspec.msgpack.encode(person)
    print(f"  Encoded size: {len(msgpack_data)} bytes")
    print(f"  First 50 bytes: {msgpack_data[:50]}")
    
    # Decode from MessagePack
    decoded_person = msgspec.msgpack.decode(msgpack_data, type=Person)
    print(f"  Decoded: {decoded_person.name}")
    print(f"  Age: {decoded_person.age}")
    print(f"  Primary address: {decoded_person.addresses[0].city}")
    
    # Compare with JSON
    json_data = msgspec.json.encode(person)
    print(f"\nSize comparison:")
    print(f"  MessagePack: {len(msgpack_data)} bytes")
    print(f"  JSON: {len(json_data)} bytes")
    print(f"  Space savings: {(len(json_data) - len(msgpack_data))/len(json_data)*100:.1f}%")
    
    # Performance comparison
    import time
    
    iterations = 1000
    
    # MessagePack performance
    start = time.time()
    for _ in range(iterations):
        msgspec.msgpack.encode(person)
    msgpack_encode_time = time.time() - start
    
    start = time.time()
    for _ in range(iterations):
        msgspec.msgpack.decode(msgpack_data, type=Person)
    msgpack_decode_time = time.time() - start
    
    # JSON performance
    start = time.time()
    for _ in range(iterations):
        msgspec.json.encode(person)
    json_encode_time = time.time() - start
    
    start = time.time()  
    for _ in range(iterations):
        msgspec.json.decode(json_data, type=Person)
    json_decode_time = time.time() - start
    
    print(f"\nPerformance comparison ({iterations} iterations):")
    print(f"  MessagePack encode: {msgpack_encode_time:.4f}s")
    print(f"  JSON encode: {json_encode_time:.4f}s")
    print(f"  MessagePack decode: {msgpack_decode_time:.4f}s")
    print(f"  JSON decode: {json_decode_time:.4f}s")
    
    # Show complex type features
    print(f"\nComplex type features:")
    type_features = [
        "Nested structures: Support for deeply nested object hierarchies",
        "Optional fields: Handle missing or null values gracefully",
        "Generic types: List, Dict, Optional, Any with full type safety",
        "Default factories: Dynamic default value creation",
        "Binary formats: Efficient MessagePack encoding/decoding",
        "Type preservation: Maintain exact type information during serialization"
    ]
    
    for feature in type_features:
        print(f"  • {feature}")
    
    # Test round-trip integrity
    print(f"\nRound-trip integrity test:")
    
    # Serialize and deserialize multiple times
    current_data = person
    for i in range(3):
        # Encode to MessagePack and back
        msgpack_bytes = msgspec.msgpack.encode(current_data)
        current_data = msgspec.msgpack.decode(msgpack_bytes, type=Person)
        
        # Verify data integrity
        assert current_data.name == person.name
        assert len(current_data.addresses) == len(person.addresses)
        assert current_data.metadata == person.metadata
    
    print(f"  ✓ Data integrity maintained through multiple round-trips")
    
    print("MessagePack and complex types completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    # Fallback demonstration
    print("Demonstrating MessagePack concepts:")
    concepts = [
        "Binary serialization: Efficient MessagePack format support",
        "Complex nesting: Handle deeply nested object structures",
        "Type preservation: Maintain type safety in binary format",
        "Performance optimization: Fast encoding/decoding operations",
        "Space efficiency: Compact binary representation of data"
    ]
    
    for concept in concepts:
        print(f"  - {concept}")
    raise
```

## Example 3: msgspec functionality test 3

```python
import msgspec
from typing import Annotated, Union, Literal
from enum import Enum

# Example 3: Schema validation and error handling
print("Example 3: Schema Validation and Error Handling")

try:
    # Define enums and literal types for validation
    class Status(Enum):
        PENDING = "pending"
        ACTIVE = "active" 
        INACTIVE = "inactive"
        SUSPENDED = "suspended"
    
    class Order(msgspec.Struct, kw_only=True):  # kw_only: the required `items` may follow defaulted fields
        order_id: str
        customer_id: int
        total: float
        status: Status
        priority: Literal["low", "medium", "high", "urgent"] = "medium"
        items: Annotated[int, msgspec.Meta(ge=1)]  # >= 1; constraints live in Meta, msgspec.field() takes none
        discount: Annotated[float, msgspec.Meta(ge=0.0, le=1.0)] = 0.0  # 0-100%
    
    print("Order struct with validation constraints defined")
    
    # Test valid orders
    valid_orders = [
        {
            "order_id": "ORD-001",
            "customer_id": 12345,
            "total": 299.99,
            "status": "active",
            "priority": "high", 
            "items": 3,
            "discount": 0.15
        },
        {
            "order_id": "ORD-002", 
            "customer_id": 67890,
            "total": 89.50,
            "status": "pending",
            "items": 1
        }
    ]
    
    print(f"\nValidating {len(valid_orders)} valid orders:")
    
    processed_orders = []
    for i, order_data in enumerate(valid_orders):
        try:
            # Serialize to JSON and deserialize with validation
            json_bytes = msgspec.json.encode(order_data)
            order = msgspec.json.decode(json_bytes, type=Order)
            processed_orders.append(order)
            
            print(f"  ✓ Order {i+1}: {order.order_id} - {order.status.value}")
            print(f"    Total: ${order.total}, Items: {order.items}, Priority: {order.priority}")
            
        except msgspec.ValidationError as e:
            print(f"  ✗ Order {i+1} validation failed: {e}")
        except Exception as e:
            print(f"  ✗ Order {i+1} error: {e}")
            raise
    
    assert len(processed_orders) == len(valid_orders), processed_orders

    # Test invalid orders to demonstrate validation
    invalid_orders = [
        {
            "order_id": "ORD-003",
            "customer_id": "invalid_id",  # Should be int
            "total": 150.0,
            "status": "active",
            "items": 2
        },
        {
            "order_id": "ORD-004", 
            "customer_id": 11111,
            "total": -50.0,  # Negative total (no constraint but unusual)
            "status": "invalid_status",  # Invalid enum value
            "items": 1
        },
        {
            "order_id": "ORD-005",
            "customer_id": 22222,
            "total": 75.0,
            "status": "pending",
            "priority": "invalid_priority",  # Invalid literal
            "items": 0,  # Violates ge=1 constraint
            "discount": 1.5  # Violates le=1.0 constraint
        }
    ]
    
    print(f"\nTesting {len(invalid_orders)} invalid orders:")
    
    expected_error_paths = ["$.customer_id", "$.status", "$.priority"]
    for i, (order_data, expected_path) in enumerate(zip(invalid_orders, expected_error_paths, strict=True)):
        json_bytes = msgspec.json.encode(order_data)
        try:
            order = msgspec.json.decode(json_bytes, type=Order)
        except msgspec.ValidationError as e:
            assert f"at `{expected_path}`" in str(e), f"Order {i+1}: expected a failure at {expected_path}, got {e}"
            print(f"  ✓ Order {i+1} correctly rejected:")
            print(f"    Reason: {str(e)[:100]}...")
        else:
            raise AssertionError(f"Order {i+1} should have failed validation, decoded as {order!r}")
    
    # Demonstrate custom validation with pre/post processing
    class Product(msgspec.Struct):
        sku: str
        name: str
        price: Annotated[float, msgspec.Meta(gt=0.0)]  # Greater than 0
        category: Union[str, None] = None
        
        def __post_init__(self):
            # Custom validation logic
            if len(self.sku) < 3:
                raise ValueError("SKU must be at least 3 characters")
            if self.price > 10000:
                raise ValueError("Price cannot exceed $10,000")
            if self.name.strip() != self.name:
                raise ValueError("Product name cannot have leading/trailing whitespace")
    
    print(f"\nCustom validation with post-init:")
    
    # (data, the message it must be rejected with, or None if valid). msgspec re-raises a ValueError from
    # __post_init__ as msgspec.ValidationError carrying the same message (measured, msgspec 0.21.1)
    test_products = [
        ({"sku": "ABC123", "name": "Laptop", "price": 999.99, "category": "electronics"}, None),
        ({"sku": "XY", "name": "Short SKU", "price": 25.0}, "SKU must be at least 3 characters"),  # Invalid: SKU too short
        ({"sku": "DEF456", "name": "  Tablet  ", "price": 399.99}, "Product name cannot have leading/trailing whitespace"),  # Invalid: whitespace
        ({"sku": "GHI789", "name": "Expensive Item", "price": 15000.0}, "Price cannot exceed $10,000")  # Invalid: price too high
    ]
    
    for i, (product_data, expected_error) in enumerate(test_products):
        json_bytes = msgspec.json.encode(product_data)
        try:
            product = msgspec.json.decode(json_bytes, type=Product)
        except msgspec.ValidationError as e:
            assert str(e) == expected_error, f"Product {i+1}: expected {expected_error!r}, got {e}"
            print(f"  ✓ Product {i+1} correctly rejected: {e}")
        else:
            assert expected_error is None, f"Product {i+1} was not rejected ({expected_error}): {product!r}"
            print(f"  ✓ Product {i+1}: {product.sku} - ${product.price}")
    
    # Show validation features
    print(f"\nValidation capabilities:")
    validation_features = [
        "Type validation: Automatic type checking and conversion",
        "Enum validation: Ensure values match predefined enum options",
        "Literal validation: Restrict string values to specific literals",
        "Numeric constraints: Support for ge, le, gt, lt field constraints",
        "Custom validation: Post-init hooks for complex business rules",
        "Clear error messages: Detailed validation error reporting"
    ]
    
    for feature in validation_features:
        print(f"  • {feature}")
    
    # Error handling patterns
    print(f"\nError handling patterns:")
    error_patterns = [
        "Catch ValidationError: Handle schema validation failures gracefully",
        "Field-level errors: Identify specific fields that failed validation",
        "Custom exceptions: Raise domain-specific errors in post-init",
        "Batch validation: Process multiple items with error collection",
        "Fallback values: Provide defaults for optional fields"
    ]
    
    for pattern in error_patterns:
        print(f"  • {pattern}")
    
    print("Schema validation and error handling completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    # Fallback demonstration
    print("Demonstrating validation concepts:")
    concepts = [
        "Schema enforcement: Automatic validation of data structure and types",
        "Constraint validation: Numeric bounds and string literal restrictions",
        "Error reporting: Clear messages for validation failures",
        "Custom validation: Post-processing hooks for business rules",
        "Type safety: Compile-time and runtime type checking"
    ]
    
    for concept in concepts:
        print(f"  - {concept}")
    raise
```

## Example 4: msgspec functionality test 4

```python
import msgspec
from typing import Union, Generic, TypeVar
import dataclasses

# Example 4: Advanced features and interoperability
print("Example 4: Advanced Features and Interoperability")

try:
    # Generic types and inheritance
    T = TypeVar('T')
    
    class Response(msgspec.Struct, Generic[T]):
        success: bool
        data: T
        message: str = ""
        timestamp: float = msgspec.field(default_factory=lambda: __import__('time').time())
    
    class UserInfo(msgspec.Struct):
        user_id: int
        username: str
        email: str
    
    class ErrorInfo(msgspec.Struct):
        error_code: int
        error_type: str
        details: str
    
    print("Generic response structure defined")
    
    # Create typed responses
    user_response = Response[UserInfo](
        success=True,
        data=UserInfo(user_id=123, username="alice_dev", email="alice@dev.com"),
        message="User retrieved successfully"
    )
    
    error_response = Response[ErrorInfo](
        success=False,
        data=ErrorInfo(error_code=404, error_type="NotFound", details="User not found"),
        message="Operation failed"
    )
    
    print(f"Created typed responses:")
    print(f"  User response: {user_response.success}, {user_response.data.username}")
    print(f"  Error response: {error_response.data.error_code}")
    
    # Interoperability with dataclasses
    @dataclasses.dataclass
    class DataclassUser:
        id: int
        name: str
        active: bool = True
    
    # Convert between dataclass and msgspec struct
    dc_user = DataclassUser(id=456, name="Bob Builder")
    
    # Create msgspec struct from dataclass data
    class MsgspecUser(msgspec.Struct):
        id: int  
        name: str
        active: bool = True
    
    # Convert dataclass to dict, then to msgspec struct
    user_dict = dataclasses.asdict(dc_user)
    ms_user = msgspec.convert(user_dict, type=MsgspecUser)
    
    print(f"\nDataclass interoperability:")
    print(f"  Dataclass: {dc_user}")
    print(f"  Msgspec: {ms_user}")
    print(f"  Types: {type(dc_user).__name__} -> {type(ms_user).__name__}")
    
    # Union types for polymorphic data
    class TextMessage(msgspec.Struct, kw_only=True):  # kw_only: the defaulted `type` may precede required fields
        type: str = "text"
        content: str
        sender: str
    
    class ImageMessage(msgspec.Struct, kw_only=True):
        type: str = "image"
        url: str
        alt_text: str
        sender: str
    
    class AudioMessage(msgspec.Struct, kw_only=True):
        type: str = "audio"
        url: str
        duration: float
        sender: str
    
    # Define union type for different message types
    Message = Union[TextMessage, ImageMessage, AudioMessage]
    
    # Create polymorphic messages
    messages = [
        TextMessage(content="Hello, world!", sender="alice"),
        ImageMessage(url="https://example.com/image.jpg", alt_text="Sunset", sender="bob"),
        AudioMessage(url="https://example.com/audio.mp3", duration=45.2, sender="carol")
    ]
    
    print(f"\nPolymorphic messages:")
    for i, msg in enumerate(messages):
        json_data = msgspec.json.encode(msg)
        print(f"  Message {i+1} ({msg.type}): {len(json_data)} bytes")
        
        # Decode back to specific type
        decoded = msgspec.json.decode(json_data, type=type(msg))
        print(f"    Decoded: {decoded.sender} sent {decoded.type}")
    
    # Custom encoders/decoders for special types
    from datetime import datetime, timezone
    
    class TimestampedEvent(msgspec.Struct):
        event_type: str
        timestamp: datetime
        data: dict
        
        # Custom encoding for datetime
        def __msgspec_encode__(self):
            return {
                "event_type": self.event_type,
                "timestamp": self.timestamp.isoformat(),
                "data": self.data
            }
    
    # Create timestamped event
    event = TimestampedEvent(
        event_type="user_login",
        timestamp=datetime.now(timezone.utc),
        data={"user_id": 789, "ip": "192.168.1.100"}
    )
    
    print(f"\nCustom encoding:")
    print(f"  Event: {event.event_type} at {event.timestamp}")
    
    # Multiple format support
    print(f"\nMultiple format serialization:")
    
    test_data = user_response
    
    # JSON
    json_bytes = msgspec.json.encode(test_data)
    print(f"  JSON: {len(json_bytes)} bytes")
    
    # MessagePack
    msgpack_bytes = msgspec.msgpack.encode(test_data)
    print(f"  MessagePack: {len(msgpack_bytes)} bytes")
    
    # Verify round-trip for both formats
    json_decoded = msgspec.json.decode(json_bytes, type=type(test_data))
    msgpack_decoded = msgspec.msgpack.decode(msgpack_bytes, type=type(test_data))
    
    print(f"  JSON round-trip: {json_decoded.success}")
    print(f"  MessagePack round-trip: {msgpack_decoded.success}")
    
    # Schema evolution and version handling
    class V1Document(msgspec.Struct):
        id: str
        title: str
        content: str
    
    class V2Document(msgspec.Struct):
        id: str
        title: str
        content: str
        tags: list[str] = msgspec.field(default_factory=list)  # New field
        version: int = 2
    
    # Simulate schema evolution
    v1_data = {"id": "doc123", "title": "Test Doc", "content": "Content here"}
    
    # Serialize as V1
    v1_json = msgspec.json.encode(v1_data)
    
    # Deserialize as V2 (with defaults for new fields)
    v2_doc = msgspec.json.decode(v1_json, type=V2Document)
    print(f"\nSchema evolution:")
    print(f"  V1 -> V2: {v2_doc.title}, tags: {v2_doc.tags}, version: {v2_doc.version}")
    
    # Show advanced features
    print(f"\nAdvanced capabilities:")
    advanced_features = [
        "Generic types: Support for parameterized types with full type safety",
        "Union types: Handle polymorphic data structures elegantly",
        "Dataclass interop: Convert between msgspec and dataclass objects",
        "Custom encoding: Override serialization for special types",
        "Multiple formats: JSON, MessagePack, and other format support",
        "Schema evolution: Handle versioning and backward compatibility"
    ]
    
    for feature in advanced_features:
        print(f"  • {feature}")
    
    print("Advanced features and interoperability completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    # Fallback demonstration
    print("Demonstrating advanced concepts:")
    concepts = [
        "Generic types: Parameterized structures with type safety",
        "Polymorphic data: Union types for different data structures",
        "Framework interop: Integration with dataclasses and other libraries",
        "Custom serialization: Override encoding/decoding behavior",
        "Schema versioning: Handle data structure evolution over time"
    ]
    
    for concept in concepts:
        print(f"  - {concept}")
    raise
```

## Example 5: msgspec functionality test 5

```python
import msgspec
import time
import json
import sys
from concurrent.futures import ThreadPoolExecutor

# Example 5: Performance benchmarks and production optimization
print("Example 5: Performance Benchmarks and Production Optimization")

try:
    # Define production-ready data structures
    class Transaction(msgspec.Struct):
        transaction_id: str
        user_id: int
        amount: float
        currency: str = "USD"
        timestamp: float = msgspec.field(default_factory=time.time)
        metadata: dict = msgspec.field(default_factory=dict)
    
    class TransactionBatch(msgspec.Struct):
        batch_id: str
        transactions: list[Transaction]
        processed_at: float = msgspec.field(default_factory=time.time)
        total_count: int = 0  # msgspec.field() has no init=False; __post_init__ fills it in
        
        def __post_init__(self):
            self.total_count = len(self.transactions)
    
    print("Production transaction structures defined")
    
    # Generate test data for performance testing
    def generate_transactions(count: int) -> list[Transaction]:
        transactions = []
        for i in range(count):
            tx = Transaction(
                transaction_id=f"tx_{i:06d}",
                user_id=i % 1000,
                amount=round(10.0 + (i % 1000) * 0.5, 2),
                currency="USD" if i % 2 == 0 else "EUR",
                metadata={"source": "api", "batch": i // 100}
            )
            transactions.append(tx)
        return transactions
    
    # Performance testing with different dataset sizes
    test_sizes = [100, 1000, 5000]
    
    print(f"\nPerformance benchmarks:")
    
    for size in test_sizes:
        print(f"\n--- Dataset size: {size} transactions ---")
        
        # Generate test data
        transactions = generate_transactions(size)
        batch = TransactionBatch(
            batch_id=f"batch_{size}",
            transactions=transactions
        )
        
        # JSON serialization benchmark
        json_times = []
        for _ in range(5):  # Multiple runs for average
            start = time.time()
            json_data = msgspec.json.encode(batch)
            json_times.append(time.time() - start)
        
        avg_json_time = sum(json_times) / len(json_times)
        json_size = len(json_data)
        
        # MessagePack serialization benchmark
        msgpack_times = []
        for _ in range(5):
            start = time.time()
            msgpack_data = msgspec.msgpack.encode(batch)
            msgpack_times.append(time.time() - start)
        
        avg_msgpack_time = sum(msgpack_times) / len(msgpack_times)
        msgpack_size = len(msgpack_data)
        
        # Deserialization benchmarks
        json_decode_times = []
        for _ in range(5):
            start = time.time()
            decoded = msgspec.json.decode(json_data, type=TransactionBatch)
            json_decode_times.append(time.time() - start)
        
        avg_json_decode_time = sum(json_decode_times) / len(json_decode_times)
        
        msgpack_decode_times = []
        for _ in range(5):
            start = time.time()
            decoded = msgspec.msgpack.decode(msgpack_data, type=TransactionBatch)
            msgpack_decode_times.append(time.time() - start)
        
        avg_msgpack_decode_time = sum(msgpack_decode_times) / len(msgpack_decode_times)
        
        # Results
        print(f"JSON encode:    {avg_json_time*1000:.2f}ms ({json_size:,} bytes)")
        print(f"JSON decode:    {avg_json_decode_time*1000:.2f}ms")
        print(f"MsgPack encode: {avg_msgpack_time*1000:.2f}ms ({msgpack_size:,} bytes)")
        print(f"MsgPack decode: {avg_msgpack_decode_time*1000:.2f}ms")
        print(f"Size savings:   {(json_size-msgpack_size)/json_size*100:.1f}%")
        print(f"Throughput:     {size/(avg_json_time*1000):.0f} tx/ms (JSON)")
    
    # Memory usage analysis
    print(f"\nMemory usage analysis:")
    
    test_tx = transactions[0]
    
    # Memory usage for different representations
    struct_size = sys.getsizeof(test_tx)
    json_bytes = msgspec.json.encode(test_tx)
    msgpack_bytes = msgspec.msgpack.encode(test_tx)
    
    print(f"  Struct object: {struct_size} bytes")
    print(f"  JSON bytes: {len(json_bytes)} bytes")
    print(f"  MessagePack bytes: {len(msgpack_bytes)} bytes")
    
    # Compare with stdlib json and dict
    tx_dict = {
        "transaction_id": test_tx.transaction_id,
        "user_id": test_tx.user_id,
        "amount": test_tx.amount,
        "currency": test_tx.currency,
        "timestamp": test_tx.timestamp,
        "metadata": test_tx.metadata
    }
    
    dict_size = sys.getsizeof(tx_dict)
    stdlib_json = json.dumps(tx_dict).encode()
    
    print(f"  Dict object: {dict_size} bytes")
    print(f"  Stdlib JSON: {len(stdlib_json)} bytes")
    print(f"  Memory efficiency: {dict_size/struct_size:.2f}x (dict/struct)")
    
    # Concurrent processing benchmark
    print(f"\nConcurrent processing test:")
    
    def process_batch(batch_data):
        # Simulate processing by serializing and deserializing
        json_data = msgspec.json.encode(batch_data)
        decoded = msgspec.json.decode(json_data, type=TransactionBatch)
        return len(decoded.transactions)
    
    # Create multiple batches for concurrent processing
    batches = []
    for i in range(10):
        batch_txs = generate_transactions(200)
        batch = TransactionBatch(batch_id=f"concurrent_{i}", transactions=batch_txs)
        batches.append(batch)
    
    # Sequential processing
    start = time.time()
    sequential_results = [process_batch(batch) for batch in batches]
    sequential_time = time.time() - start
    
    # Concurrent processing
    start = time.time()
    with ThreadPoolExecutor(max_workers=4) as executor:
        concurrent_results = list(executor.map(process_batch, batches))
    concurrent_time = time.time() - start
    
    print(f"  Sequential: {sequential_time:.4f}s")
    print(f"  Concurrent: {concurrent_time:.4f}s")
    print(f"  Speedup: {sequential_time/concurrent_time:.2f}x")
    
    # Production optimization patterns
    print(f"\nProduction optimization patterns:")
    
    class OptimizedSerializer:
        def __init__(self):
            # Pre-compile encoders for better performance
            self.json_encoder = msgspec.json.Encoder()
            self.msgpack_encoder = msgspec.msgpack.Encoder()
            
            # Pre-compile decoders
            self.json_decoder = msgspec.json.Decoder(type=TransactionBatch)
            self.msgpack_decoder = msgspec.msgpack.Decoder(type=TransactionBatch)
        
        def json_encode(self, obj):
            return self.json_encoder.encode(obj)
        
        def json_decode(self, data):
            return self.json_decoder.decode(data)
        
        def msgpack_encode(self, obj):
            return self.msgpack_encoder.encode(obj)
        
        def msgpack_decode(self, data):
            return self.msgpack_decoder.decode(data)
    
    # Test optimized serializer
    optimizer = OptimizedSerializer()
    test_batch = batches[0]
    
    # Performance comparison
    iterations = 1000
    
    # Regular encode/decode
    start = time.time()
    for _ in range(iterations):
        data = msgspec.json.encode(test_batch)
        decoded = msgspec.json.decode(data, type=TransactionBatch)
    regular_time = time.time() - start
    
    # Optimized encode/decode
    start = time.time()
    for _ in range(iterations):
        data = optimizer.json_encode(test_batch)
        decoded = optimizer.json_decode(data)
    optimized_time = time.time() - start
    
    print(f"  Regular serializer: {regular_time:.4f}s")
    print(f"  Optimized serializer: {optimized_time:.4f}s")
    print(f"  Performance gain: {regular_time/optimized_time:.2f}x")
    
    # Show production best practices
    print(f"\nProduction best practices:")
    best_practices = [
        "Pre-compile encoders/decoders: Create reusable instances for better performance",
        "Choose format wisely: MessagePack for size, JSON for debugging/compatibility",
        "Batch processing: Process multiple items together for efficiency",
        "Memory management: Monitor memory usage with large datasets",
        "Concurrent processing: Use thread pools for CPU-intensive serialization",
        "Schema versioning: Plan for data structure evolution in production",
        "Error handling: Implement robust validation and error recovery"
    ]
    
    for practice in best_practices:
        print(f"  • {practice}")
    
    print("Performance benchmarks and optimization completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    # Fallback demonstration
    print("Demonstrating performance concepts:")
    concepts = [
        "Benchmarking: Measure serialization performance across different scenarios",
        "Memory efficiency: Analyze memory usage and optimization opportunities",
        "Concurrent processing: Scale serialization with concurrent execution",
        "Production optimization: Pre-compile encoders for better performance",
        "Format selection: Choose appropriate serialization format for use case"
    ]
    
    for concept in concepts:
        print(f"  - {concept}")
    raise
```
