## Example 2: Case-Insensitive MultiDict and HTTP Headers

Demonstrate case-insensitive operations for HTTP headers and web data processing.

```python
# Example 2: Case-insensitive MultiDict and HTTP headers
try:
    from multidict import CIMultiDict, MultiDict
    
    print("=== Case-Insensitive MultiDict for HTTP Headers ===")
    # Create case-insensitive MultiDict (common for HTTP headers)
    headers = CIMultiDict()
    
    # Add HTTP headers with different case variations
    headers.add('Content-Type', 'text/html')
    headers.add('content-type', 'application/json')  # Same key, different case
    headers.add('CONTENT-TYPE', 'text/plain')
    headers.add('Accept', 'text/html,application/xhtml+xml')
    headers.add('accept', 'application/json')
    headers.add('User-Agent', 'Mozilla/5.0')
    
    print(f"Headers created: {len(headers)} total entries")
    print(f"Content-Type entries: {len(headers.getall('content-type'))}")
    
    # Case-insensitive access
    print(f"\nCase-insensitive access:")
    print(f"'Content-Type': {headers.get('Content-Type')}")
    print(f"'content-type': {headers.get('content-type')}")
    print(f"'CONTENT-TYPE': {headers.get('CONTENT-TYPE')}")
    
    # Get all variations of Content-Type
    content_types = headers.getall('Content-Type')
    print(f"All Content-Type values: {content_types}")
    
    print("\n=== HTTP Response Headers Simulation ===")
    response_headers = CIMultiDict()
    response_headers['Content-Type'] = 'application/json'
    response_headers['Content-Length'] = '1024'
    response_headers.add('Set-Cookie', 'session_id=abc123; Path=/')
    response_headers.add('Set-Cookie', 'user_pref=theme_dark; Path=/')
    response_headers.add('Set-Cookie', 'lang=en; Path=/')
    response_headers['Cache-Control'] = 'no-cache'
    response_headers.add('Vary', 'Accept-Encoding')
    response_headers.add('Vary', 'User-Agent')
    
    print("Response headers:")
    for key, value in response_headers.items():
        print(f"  {key}: {value}")
    
    # Cookies handling (multiple Set-Cookie headers)
    cookies = response_headers.getall('Set-Cookie')
    print(f"\nCookies to set: {len(cookies)}")
    for i, cookie in enumerate(cookies):
        print(f"  Cookie {i+1}: {cookie}")
    
    # Vary header handling (for caching)
    vary_headers = response_headers.getall('Vary')
    print(f"Vary headers: {vary_headers}")
    
    print("\n=== Case Sensitivity Comparison ===")
    # Compare regular MultiDict vs CIMultiDict
    regular_md = MultiDict()
    case_insensitive_md = CIMultiDict()
    
    # Add same keys with different cases
    regular_md.add('Header', 'value1')
    regular_md.add('header', 'value2')
    regular_md.add('HEADER', 'value3')
    
    case_insensitive_md.add('Header', 'value1')
    case_insensitive_md.add('header', 'value2')
    case_insensitive_md.add('HEADER', 'value3')
    
    print(f"Regular MultiDict entries: {len(regular_md)}")
    print(f"Case-insensitive entries: {len(case_insensitive_md)}")
    print(f"Regular keys: {list(regular_md.keys())}")
    print(f"CI keys: {list(case_insensitive_md.keys())}")
    
    print("Example 2: Case-insensitive MultiDict test passed")
    
except Exception as e:
    print(f"Example 2: Error in case-insensitive test: {e}")
    import traceback
    traceback.print_exc()
    raise
```

## Example 3: Form Data Processing and Query Parameters

Demonstrate handling web form data and URL query parameters with multiple values.

```python
# Example 3: Form data processing and query parameters
try:
    from multidict import MultiDict
    from urllib.parse import parse_qsl
    
    print("=== URL Query Parameters Processing ===")
    # Simulate URL query string with multiple values
    query_string = "color=red&color=blue&color=green&size=large&size=medium&category=electronics&page=1"
    
    # Parse query string into MultiDict
    query_params = MultiDict(parse_qsl(query_string, keep_blank_values=True))
    
    print(f"Query string: {query_string}")
    print(f"Parsed parameters: {len(query_params)} total")
    print(f"Unique parameter names: {len(set(query_params.keys()))}")
    
    # Access multi-value parameters
    colors = query_params.getall('color')
    sizes = query_params.getall('size')
    print(f"Selected colors: {colors}")
    print(f"Selected sizes: {sizes}")
    print(f"Category: {query_params.get('category')}")
    print(f"Page: {query_params.get('page', '1')}")
    
    print("\n=== Form Data Simulation ===")
    # Simulate HTML form with checkboxes and multi-select
    form_data = MultiDict()
    
    # Checkbox selections (multiple values for same field)
    form_data.extend([
        ('interests', 'programming'),
        ('interests', 'music'),
        ('interests', 'travel'),
        ('skills', 'python'),
        ('skills', 'javascript'),
        ('skills', 'sql'),
        ('experience', 'junior'),
        ('newsletter', 'yes'),
        ('notifications', 'email'),
        ('notifications', 'sms')
    ])
    
    print("Form data submitted:")
    print(f"  Interests: {form_data.getall('interests')}")
    print(f"  Skills: {form_data.getall('skills')}")
    print(f"  Experience: {form_data.get('experience')}")
    print(f"  Newsletter: {form_data.get('newsletter') == 'yes'}")
    print(f"  Notifications: {form_data.getall('notifications')}")
    
    print("\n=== Search Filters Processing ===")
    # E-commerce search filters
    search_filters = MultiDict()
    search_filters.add('brand', 'apple')
    search_filters.add('brand', 'samsung') 
    search_filters.add('brand', 'google')
    search_filters.add('price_range', '100-500')
    search_filters.add('price_range', '500-1000')
    search_filters.add('rating', '4+')
    search_filters.add('features', 'waterproof')
    search_filters.add('features', 'wireless')
    search_filters.add('category', 'smartphones')
    
    print("Search filters:")
    for key in set(search_filters.keys()):
        values = search_filters.getall(key)
        if len(values) > 1:
            print(f"  {key}: {values} (multiple)")
        else:
            print(f"  {key}: {values[0]}")
    
    print("\n=== Filter Combinations and Logic ===")
    # Process filters for database query
    filter_logic = {}
    
    for key in set(search_filters.keys()):
        values = search_filters.getall(key)
        if key == 'brand':
            filter_logic[key] = {'$in': values}  # MongoDB-style OR logic
        elif key == 'price_range':
            # Convert ranges to min/max values
            ranges = []
            for range_str in values:
                min_val, max_val = map(int, range_str.split('-'))
                ranges.append({'min': min_val, 'max': max_val})
            filter_logic[key] = ranges
        elif key == 'features':
            filter_logic[key] = {'$all': values}  # MongoDB-style AND logic
        else:
            filter_logic[key] = values[0]
    
    print("Generated filter logic:")
    for key, logic in filter_logic.items():
        print(f"  {key}: {logic}")
    
    # Count filter selections
    total_filters = sum(len(search_filters.getall(key)) for key in set(search_filters.keys()))
    print(f"\nTotal filter selections: {total_filters}")
    print(f"Filter categories: {len(set(search_filters.keys()))}")
    
    print("Example 3: Form data and query parameters test passed")
    
except Exception as e:
    print(f"Example 3: Error in form data test: {e}")
    import traceback
    traceback.print_exc()
    raise
```

## Example 4: Immutable MultiDict and Data Validation

Demonstrate immutable variants and data validation patterns for web applications.

```python
# Example 4: Immutable MultiDict and data validation
try:
    from multidict import MultiDict, CIMultiDict, MultiDictProxy, CIMultiDictProxy
    
    print("=== Immutable MultiDict Variants ===")
    # Create mutable MultiDict
    mutable_md = MultiDict()
    mutable_md.add('key1', 'value1')
    mutable_md.add('key1', 'value2')
    mutable_md.add('key2', 'value3')
    
    print(f"Mutable MultiDict: {mutable_md}")
    
    # Create immutable proxy
    immutable_md = MultiDictProxy(mutable_md)
    print(f"Immutable proxy created: {type(immutable_md)}")
    print(f"Immutable content: {immutable_md}")
    print(f"Same values: {immutable_md.getall('key1') == mutable_md.getall('key1')}")
    
    # Try to modify immutable (should raise exception)
    try:
        immutable_md.add('new_key', 'new_value')
    except AttributeError as e:
        # MultiDictProxy is a read-only view: it has no mutating methods at all
        assert "has no attribute 'add'" in str(e), f"unexpected message: {e!r}"
        print(f"Immutable proxy refused .add() as expected ({type(e).__name__})")
    else:
        raise AssertionError("Immutable modification succeeded (unexpected)")
    assert 'new_key' not in mutable_md
    
    print("\n=== Case-Insensitive Immutable Headers ===")
    # HTTP headers that should be immutable after processing
    headers = CIMultiDict()
    headers['Content-Type'] = 'application/json'
    headers.add('Accept', 'application/json')
    headers.add('Accept', 'text/html')
    headers['Authorization'] = 'Bearer token123'
    
    # Create immutable headers for processing
    readonly_headers = CIMultiDictProxy(headers)
    
    print(f"Readonly headers: {len(readonly_headers)} entries")
    print(f"Content-Type: {readonly_headers.get('content-type')}")
    print(f"Accept values: {readonly_headers.getall('ACCEPT')}")
    
    # Headers remain accessible but cannot be modified
    try:
        readonly_headers['X-Custom'] = 'custom-value'
    except TypeError as e:
        assert "does not support item assignment" in str(e), f"unexpected message: {e!r}"
        print(f"Readonly headers refused item assignment as expected ({type(e).__name__})")
    else:
        raise AssertionError("Readonly headers modified (unexpected)")
    assert 'X-Custom' not in headers
    
    print("\n=== Data Validation Patterns ===")
    # Validation helper functions
    def validate_form_data(form_data):
        """Validate web form data with MultiDict"""
        errors = MultiDict()
        
        # Required fields validation
        required_fields = ['email', 'password', 'name']
        for field in required_fields:
            if not form_data.get(field):
                errors.add(field, f'{field} is required')
        
        # Email validation
        email = form_data.get('email', '')
        if email and '@' not in email:
            errors.add('email', 'Invalid email format')
        
        # Password validation
        password = form_data.get('password', '')
        if password and len(password) < 6:
            errors.add('password', 'Password must be at least 6 characters')
        
        # Multi-select validation
        interests = form_data.getall('interests')
        if len(interests) > 5:
            errors.add('interests', 'Maximum 5 interests allowed')
        
        # Age validation
        age = form_data.get('age')
        if age:
            try:
                age_int = int(age)
                if age_int < 13 or age_int > 120:
                    errors.add('age', 'Age must be between 13 and 120')
            except ValueError:
                errors.add('age', 'Age must be a valid number')
        
        return errors
    
    # Test validation with valid data
    valid_form = MultiDict([
        ('email', 'user@example.com'),
        ('password', 'securepass'),
        ('name', 'John Doe'),
        ('age', '25'),
        ('interests', 'reading'),
        ('interests', 'gaming'),
    ])
    
    errors = validate_form_data(valid_form)
    print(f"Valid form errors: {len(errors)} (should be 0)")
    
    # Test validation with invalid data
    invalid_form = MultiDict([
        ('email', 'invalid-email'),
        ('password', '123'),  # Too short
        # Missing required 'name' field
        ('age', 'not-a-number'),
        ('interests', 'reading'),
        ('interests', 'gaming'),
        ('interests', 'music'),
        ('interests', 'sports'),
        ('interests', 'cooking'),
        ('interests', 'travel'),  # Too many interests
    ])
    
    errors = validate_form_data(invalid_form)
    print(f"Invalid form errors: {len(errors)}")
    
    # Group errors by field
    error_summary = {}
    for field, error in errors.items():
        if field not in error_summary:
            error_summary[field] = []
        error_summary[field].append(error)
    
    print("Error summary:")
    for field, field_errors in error_summary.items():
        print(f"  {field}: {field_errors}")
    
    print("\n=== Configuration with Defaults ===")
    # Configuration management with MultiDict
    config = MultiDict()
    config.add('database.host', 'localhost')
    config.add('database.port', '5432')
    config.add('allowed_origins', 'https://example.com')
    config.add('allowed_origins', 'https://app.example.com')
    config.add('allowed_origins', 'https://admin.example.com')
    config.add('feature_flags', 'new_ui')
    config.add('feature_flags', 'beta_features')
    
    # Create immutable config for runtime
    runtime_config = MultiDictProxy(config)
    
    print(f"Database host: {runtime_config.get('database.host')}")
    print(f"Database port: {runtime_config.get('database.port', '5432')}")
    print(f"Allowed origins: {runtime_config.getall('allowed_origins')}")
    print(f"Feature flags: {runtime_config.getall('feature_flags')}")
    
    print("Example 4: Immutable MultiDict and validation test passed")
    
except Exception as e:
    print(f"Example 4: Error in immutable/validation test: {e}")
    import traceback
    traceback.print_exc()
    raise
```

## Example 5: Performance and Integration Patterns

Demonstrate performance optimization and integration with web frameworks and serialization.

```python
# Example 5: Performance and integration patterns
try:
    from multidict import MultiDict, CIMultiDict
    import time
    import json
    
    print("=== Performance Comparison ===")
    # Compare performance with regular dict vs MultiDict
    n_operations = 1000
    
    # Regular dictionary performance
    start_time = time.time()
    regular_dict = {}
    for i in range(n_operations):
        key = f'key_{i % 100}'  # Create duplicate keys
        if key in regular_dict:
            if not isinstance(regular_dict[key], list):
                regular_dict[key] = [regular_dict[key]]
            regular_dict[key].append(f'value_{i}')
        else:
            regular_dict[key] = f'value_{i}'
    
    regular_time = time.time() - start_time
    
    # MultiDict performance
    start_time = time.time()
    multi_dict = MultiDict()
    for i in range(n_operations):
        key = f'key_{i % 100}'
        multi_dict.add(key, f'value_{i}')
    
    multidict_time = time.time() - start_time
    
    print(f"Regular dict simulation: {regular_time*1000:.2f}ms")
    print(f"MultiDict operations: {multidict_time*1000:.2f}ms")
    print(f"Performance ratio: {regular_time/multidict_time:.2f}x")
    
    # Memory usage comparison
    import sys
    regular_size = sys.getsizeof(regular_dict)
    multidict_size = sys.getsizeof(multi_dict)
    
    print(f"Regular dict size: {regular_size} bytes")
    print(f"MultiDict size: {multidict_size} bytes")
    print(f"Memory ratio: {multidict_size/regular_size:.2f}x")
    
    print("\n=== Bulk Operations Performance ===")
    # Test bulk operations
    test_data = [(f'key_{i % 50}', f'value_{i}') for i in range(500)]
    
    # extend() method
    start_time = time.time()
    bulk_md = MultiDict()
    bulk_md.extend(test_data)
    extend_time = time.time() - start_time
    
    # Individual add() calls
    start_time = time.time()
    individual_md = MultiDict()
    for key, value in test_data:
        individual_md.add(key, value)
    individual_time = time.time() - start_time
    
    print(f"Bulk extend(): {extend_time*1000:.2f}ms")
    print(f"Individual add(): {individual_time*1000:.2f}ms")
    print(f"Bulk vs individual: {individual_time/extend_time:.2f}x faster")
    
    print("\n=== Serialization Integration ===")
    # Convert MultiDict to JSON-serializable format
    def multidict_to_json_compatible(md):
        """Convert MultiDict to JSON-compatible format"""
        result = {}
        for key in set(md.keys()):
            values = md.getall(key)
            if len(values) == 1:
                result[key] = values[0]
            else:
                result[key] = values
        return result
    
    def json_to_multidict(data):
        """Convert JSON data back to MultiDict"""
        md = MultiDict()
        for key, value in data.items():
            if isinstance(value, list):
                for v in value:
                    md.add(key, v)
            else:
                md.add(key, value)
        return md
    
    # Test serialization round-trip
    original_md = MultiDict()
    original_md.extend([
        ('colors', 'red'),
        ('colors', 'blue'),
        ('colors', 'green'),
        ('size', 'large'),
        ('category', 'electronics'),
        ('tags', 'popular'),
        ('tags', 'featured')
    ])
    
    # Serialize to JSON
    json_data = multidict_to_json_compatible(original_md)
    json_string = json.dumps(json_data, indent=2)
    
    print("Serialized to JSON:")
    print(json_string)
    
    # Deserialize back to MultiDict
    loaded_data = json.loads(json_string)
    restored_md = json_to_multidict(loaded_data)
    
    print(f"\nRound-trip verification:")
    print(f"Original keys: {sorted(set(original_md.keys()))}")
    print(f"Restored keys: {sorted(set(restored_md.keys()))}")
    
    # Verify all values are preserved
    keys_match = set(original_md.keys()) == set(restored_md.keys())
    values_match = all(
        original_md.getall(key) == restored_md.getall(key)
        for key in set(original_md.keys())
    )
    
    print(f"Keys match: {keys_match}")
    print(f"Values match: {values_match}")
    
    print("\n=== Web Framework Integration Pattern ===")
    # Simulate web framework request processing
    class WebRequest:
        def __init__(self):
            self.headers = CIMultiDict()
            self.query_params = MultiDict()
            self.form_data = MultiDict()
    
    # Simulate incoming request
    request = WebRequest()
    
    # Add headers
    request.headers.extend([
        ('Accept', 'text/html'),
        ('Accept', 'application/json'),
        ('Accept-Language', 'en-US'),
        ('Accept-Language', 'en'),
        ('User-Agent', 'Mozilla/5.0'),
        ('Cookie', 'session=abc123'),
        ('Cookie', 'theme=dark')
    ])
    
    # Add query parameters
    request.query_params.extend([
        ('filter', 'active'),
        ('filter', 'recent'),
        ('sort', 'date'),
        ('limit', '20')
    ])
    
    # Process request
    print("Request processing:")
    print(f"  Accept types: {request.headers.getall('Accept')}")
    print(f"  Languages: {request.headers.getall('Accept-Language')}")
    print(f"  Filters: {request.query_params.getall('filter')}")
    print(f"  Sort: {request.query_params.get('sort', 'default')}")
    print(f"  Limit: {int(request.query_params.get('limit', '10'))}")
    
    # Response headers
    response_headers = CIMultiDict()
    response_headers['Content-Type'] = 'application/json'
    response_headers.add('Set-Cookie', 'last_visit=2024-01-01; Path=/')
    response_headers.add('Set-Cookie', 'preferences=updated; Path=/')
    response_headers['Cache-Control'] = 'no-cache'
    
    print(f"  Response cookies: {response_headers.getall('Set-Cookie')}")
    
    print("Example 5: Performance and integration patterns test passed")
    
except Exception as e:
    print(f"Example 5: Error in performance/integration test: {e}")
    import traceback
    traceback.print_exc()
    raise
```
