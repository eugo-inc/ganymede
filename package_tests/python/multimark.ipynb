```python
import multimark

html = multimark.markdown_to_html("# Title\n\nSome **bold** text with a [link](https://example.com).\n")
assert "<h1>Title</h1>" in html and "<strong>bold</strong>" in html and 'href="https://example.com"' in html
print("Example 1 OK:", html.splitlines()[0])
```

## Example 2: GitHub Flavored tables, strikethrough and task lists

```python
import multimark

# Example 2: the GFM extensions, the part only cmark-gfm's extensions library provides
text = "| a | b |\n|---|---|\n| 1 | ~~2~~ |\n\n- [x] done\n"
html = multimark.markdown_to_html(text, extensions=["table", "strikethrough", "tasklist"])
assert "<table>" in html and "<td>1</td>" in html and "<del>2</del>" in html and 'type="checkbox"' in html
print("Example 2 OK:", html.count("<td>"), "table cells")
```

## Example 3: Other output formats

```python
import multimark

# Example 3: LaTeX, man page, CommonMark and XML renderers
text = "**x** and *y*\n"
assert "\\textbf{x}" in multimark.markdown_to_latex(text)
assert ".PP" in multimark.markdown_to_man(text) or "x" in multimark.markdown_to_man(text)
assert multimark.markdown_to_commonmark(text).strip() == text.strip()
assert "<strong>" in multimark.markdown_to_xml(text) or "strong" in multimark.markdown_to_xml(text)
print("Example 3 OK: latex, man, commonmark, xml")
```

## Example 4: Batch rendering releases the GIL once for the whole list

```python
import multimark

# Example 4: markdown_to_html_batch renders many documents in one call; results stay in order
docs = [f"**doc {i}**" for i in range(100)]
results = multimark.markdown_to_html_batch(docs)
assert len(results) == 100 and all("<strong>doc " in r for r in results)
assert results[7] == multimark.markdown_to_html(docs[7]) and "doc 7" in results[7]
print("Example 4 OK:", len(results), "documents rendered")
```

## Example 5: Parse tree access and the library version

```python
import multimark

doc = multimark.parse("# H\n\ntext\n")
assert doc.type == multimark.NodeType.DOCUMENT
kinds = [child.type for child in doc.children]
assert multimark.NodeType.HEADING in kinds and multimark.NodeType.PARAGRAPH in kinds
assert multimark.cmark_version() == "0.29.0.gfm.13"
print("Example 5 OK:", multimark.cmark_version(), [k.name for k in kinds])
```
