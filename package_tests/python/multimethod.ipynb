## Example 1

```python
# Example 1: Basic multiple dispatch by type
from multimethod import multimethod

print("Example 1 - Type-based dispatch")

# Define overloaded functions
@multimethod
def process(data: int) -> str:
    return f"Integer: {data} * 2 = {data * 2}"

@multimethod
def process(data: str) -> str:
    return f"String: '{data}' -> '{data.upper()}'"

@multimethod
def process(data: list) -> str:
    return f"List: {len(data)} items"

# Test different types
print(process(42))
print(process("hello"))
print(process([1, 2, 3]))

print("Example 1 completed")
```

## Example 2

```python
# Example 2: Multiple argument dispatch
from multimethod import multimethod

print("Example 2 - Multiple argument dispatch")

@multimethod
def combine(a: int, b: int) -> str:
    return f"Int + Int: {a + b}"

@multimethod
def combine(a: str, b: str) -> str:
    return f"Str + Str: '{a + b}'"

@multimethod
def combine(a: int, b: str) -> str:
    return f"Int + Str: '{b * a}'"

# Test combinations
print(combine(5, 7))
print(combine("Hello", " World"))
print(combine(3, "Python"))

print("Example 2 completed")
```

## Example 3

```python
# Example 3: Class-based dispatch
from multimethod import multimethod

print("Example 3 - Class-based dispatch")

class Shape:
    pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

@multimethod
def area(shape: Circle) -> float:
    return 3.14 * shape.radius ** 2

@multimethod
def area(shape: Rectangle) -> float:
    return shape.width * shape.height

# Test
c = Circle(5)
r = Rectangle(4, 6)
print(f"Circle area: {area(c):.2f}")
print(f"Rectangle area: {area(r):.2f}")

print("Example 3 completed")
```

## Example 4

```python
# Example 4: Numeric type hierarchy
from multimethod import multimethod
import numbers

print("Example 4 - Numeric type hierarchy")

@multimethod
def calc(x: int, y: int) -> str:
    return f"Int: {x} * {y} = {x * y}"

@multimethod
def calc(x: float, y: float) -> str:
    return f"Float: {x:.2f} / {y:.2f} = {x/y:.2f}"

@multimethod
def calc(x: numbers.Number, y: numbers.Number) -> str:
    return f"Mixed: {x} + {y} = {x + y}"

# Test different numeric combinations
print(calc(4, 5))
print(calc(10.0, 2.5))
print(calc(7, 3.14))

print("Example 4 completed")
```

## Example 5

```python
# Example 5: Method overloading in classes
from multimethod import multimethod

print("Example 5 - Method overloading")

class DataProcessor:
    @multimethod
    def add(self, data: int):
        return f"Added int: {data}"
    
    @multimethod
    def add(self, data: str):
        return f"Added string: '{data}'"
    
    @multimethod
    def add(self, data: list):
        return f"Added list: {len(data)} items"

# Test overloaded methods
processor = DataProcessor()
print(processor.add(42))
print(processor.add("test"))
print(processor.add([1, 2, 3, 4]))

print("Example 5 completed")
```
