## Example 3: multiurl functionality test 3

```python
import multiurl
import tempfile
import os

# Example 3: Advanced downloading with Downloader class and configuration
print("multiurl advanced downloading with Downloader class and configuration")

try:
    print("Example 3 - Advanced downloading with Downloader class and configuration")
    
    # Create multiple test files of different sizes
    test_files = []
    for i, size in enumerate([100, 500, 1000], 1):
        with tempfile.NamedTemporaryFile(mode='w', delete=False, suffix=f'_test{i}.txt') as temp_file:
            content = f"Test file {i} content.\n" * size
            temp_file.write(content)
            test_files.append((temp_file.name, len(content)))
    
    print(f"Created {len(test_files)} test files:")
    for i, (path, size) in enumerate(test_files, 1):
        print(f"  File {i}: {os.path.basename(path)} ({size} bytes)")
    
    # Test Downloader class functionality
    for i, (source_path, expected_size) in enumerate(test_files, 1):
        print(f"\n--- Testing Downloader {i} ---")
        
        file_url = f"file://{source_path}"
        target_path = tempfile.mktemp(suffix=f'_downloaded{i}.txt')
        
        try:
            # Create Downloader instance
            downloader = multiurl.Downloader(file_url)
            print(f"Created downloader for file {i}")
            
            # Perform download
            result = downloader.download(target_path)
            
            # Verify download
            if os.path.exists(target_path):
                actual_size = os.path.getsize(target_path)
                print(f"Download successful: {actual_size} bytes")
                print(f"Size matches: {actual_size == expected_size}")
                assert result == actual_size == expected_size, (result, actual_size, expected_size)
                
                # Read a sample of content
                with open(target_path, 'r') as f:
                    sample = f.read(50)
                print(f"Content sample: {sample}...")
                
                os.unlink(target_path)
            else:
                raise AssertionError(f"Downloader {i} did not create {target_path}")
        
        except Exception as e:
            print(f"Error downloading file {i}: {e}")
            raise
    
    # multiurl.robust(call, maximum_tries, retry_after) wraps a requests-style call (url -> response)
    # and retries it; it is not a downloader object and has no .download(). Serve the files over a
    # loopback HTTP server so the robust path runs for real, offline.
    import functools, http.server, threading, requests

    class _QuietHandler(http.server.SimpleHTTPRequestHandler):
        def log_message(self, *args):
            pass

    server = http.server.ThreadingHTTPServer(
        ("127.0.0.1", 0), functools.partial(_QuietHandler, directory=tempfile.gettempdir()))
    threading.Thread(target=server.serve_forever, daemon=True).start()
    session = requests.Session()
    session.trust_env = False  # loopback only, never through an ambient proxy
    robust_get = multiurl.robust(session.get, maximum_tries=3, retry_after=0)

    def robust_download(url, target):
        # the file:// URL's file, fetched from the loopback server through multiurl.robust
        response = robust_get(f"http://127.0.0.1:{server.server_port}/{os.path.basename(url)}")
        response.raise_for_status()
        with open(target, 'wb') as out:
            out.write(response.content)
        return len(response.content)

    # Test different download strategies
    print(f"\n--- Download Strategy Testing ---")
    test_file = test_files[1]  # Use medium-sized file
    source_path, _ = test_file
    file_url = f"file://{source_path}"
    
    strategies = ['download', 'robust']
    for strategy in strategies:
        try:
            target_path = tempfile.mktemp(suffix=f'_{strategy}.txt')
            
            if strategy == 'download':
                result = multiurl.download(file_url, target_path)
            elif strategy == 'robust':
                result = robust_download(file_url, target_path)
            
            if os.path.exists(target_path):
                size = os.path.getsize(target_path)
                print(f"{strategy.capitalize()} strategy: {size} bytes downloaded")
                assert result == size == os.path.getsize(source_path), (strategy, result, size)
                os.unlink(target_path)
            else:
                raise AssertionError(f"{strategy} strategy did not create {target_path}")
        
        except Exception as e:
            print(f"{strategy.capitalize()} strategy error: {e}")
            raise
    
    # Performance comparison
    print(f"\n--- Performance Testing ---")
    largest_file = test_files[-1]  # Use largest file
    source_path, file_size = largest_file
    file_url = f"file://{source_path}"
    
    import time
    
    methods = [
        ('Basic download', lambda url, target: multiurl.download(url, target)),
        ('Downloader class', lambda url, target: multiurl.Downloader(url).download(target)),
        ('Robust download', robust_download)
    ]
    
    for method_name, method_func in methods:
        target_path = tempfile.mktemp(suffix='_perf.txt')
        
        try:
            start_time = time.time()
            result = method_func(file_url, target_path)
            end_time = time.time()
            
            duration = end_time - start_time
            
            if os.path.exists(target_path):
                size = os.path.getsize(target_path)
                speed = size / duration if duration > 0 else 0
                print(f"{method_name}: {duration:.4f}s ({speed:.0f} bytes/s)")
                assert size == file_size, (method_name, size, file_size)
                os.unlink(target_path)
            else:
                raise AssertionError(f"{method_name} did not create {target_path}")
        
        except Exception as e:
            print(f"{method_name} error: {e}")
            raise
    
    # Cleanup test files
    for source_path, _ in test_files:
        try:
            os.unlink(source_path)
        except:
            pass
            raise
    server.shutdown()
    server.server_close()
    
    print("multiurl example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"multiurl example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: multiurl functionality test 4

```python
import multiurl
import tempfile
import os

# Example 4: Error handling and resilience features
print("multiurl error handling and resilience features")

try:
    print("Example 4 - Error handling and resilience features")
    
    # Test error handling with invalid URLs
    print("--- Error Handling Tests ---")
    
    # Each invalid URL fails with a specific error in the installed multiurl: a missing file is
    # FileNotFoundError; an unknown scheme is a KeyError on its (scheme, has_parts) downloader lookup;
    # a bare word is taken as a relative file path, so it is FileNotFoundError too.
    invalid_scenarios = [
        ("Non-existent file", "file:///this/path/does/not/exist.txt", FileNotFoundError),
        ("Invalid protocol", "invalid://example.com/file.txt", KeyError),
        ("Malformed URL", "not-a-url-at-all", FileNotFoundError),
    ]
    
    for scenario_name, invalid_url, expected_error in invalid_scenarios:
        print(f"\nTesting {scenario_name}: {invalid_url}")
        target_path = tempfile.mktemp(suffix='_error_test.txt')
        
        try:
            result = multiurl.download(invalid_url, target_path)
        except expected_error as e:
            if expected_error is KeyError:
                assert e.args == (("invalid", False),), e.args
            print(f"  Expected error caught: {type(e).__name__}: {str(e)[:80]}...")
        else:
            raise AssertionError(f"{scenario_name}: expected {expected_error.__name__}, download returned {result!r}")
        
        # Cleanup if file was created -- it must not have been: the download failed before writing
        try:
            os.unlink(target_path)
        except FileNotFoundError:
            pass
        else:
            raise AssertionError(f"{scenario_name}: the failed download left {target_path} behind")
    
    # Test resilience with different downloader types
    print(f"\n--- Resilience Testing ---")
    
    # Create a valid test file for resilience testing
    with tempfile.NamedTemporaryFile(mode='w', delete=False, suffix='.txt') as temp_file:
        temp_file.write("Resilience test content\nMultiple lines for testing\nThird line of content")
        test_file_path = temp_file.name
    
    file_url = f"file://{test_file_path}"
    print(f"Testing resilience with: {os.path.basename(test_file_path)}")
    
    # multiurl.robust(call, maximum_tries, retry_after) wraps a requests-style call (url -> response)
    # and retries it; it is not a downloader object and has no .download(). Serve the files over a
    # loopback HTTP server so the robust path runs for real, offline.
    import functools, http.server, threading, requests

    class _QuietHandler(http.server.SimpleHTTPRequestHandler):
        def log_message(self, *args):
            pass

    server = http.server.ThreadingHTTPServer(
        ("127.0.0.1", 0), functools.partial(_QuietHandler, directory=tempfile.gettempdir()))
    threading.Thread(target=server.serve_forever, daemon=True).start()
    session = requests.Session()
    session.trust_env = False  # loopback only, never through an ambient proxy
    robust_get = multiurl.robust(session.get, maximum_tries=3, retry_after=0)

    def robust_download(url, target):
        # the file:// URL's file, fetched from the loopback server through multiurl.robust
        response = robust_get(f"http://127.0.0.1:{server.server_port}/{os.path.basename(url)}")
        response.raise_for_status()
        with open(target, 'wb') as out:
            out.write(response.content)
        return len(response.content)

    # Test different approaches to downloading
    approaches = [
        ("Standard download", lambda url, target: multiurl.download(url, target)),
        ("Downloader class", lambda url, target: multiurl.Downloader(url).download(target)),
        ("Robust downloader", robust_download),
    ]
    
    for approach_name, download_func in approaches:
        print(f"\n{approach_name}:")
        target_path = tempfile.mktemp(suffix='_resilience.txt')
        
        try:
            result = download_func(file_url, target_path)
            
            if os.path.exists(target_path):
                size = os.path.getsize(target_path)
                print(f"  Success: {size} bytes downloaded")
                
                # Verify content integrity
                with open(target_path, 'r') as f:
                    content = f.read()
                
                lines = content.split('\n')
                print(f"  Content verification: {len(lines)} lines")
                print(f"  First line: {lines[0] if lines else 'Empty'}")
                assert lines == ["Resilience test content", "Multiple lines for testing", "Third line of content"], lines
                
                os.unlink(target_path)
            else:
                raise AssertionError(f"{approach_name} did not create {target_path}")
        
        except Exception as e:
            print(f"  Error: {type(e).__name__}: {e}")
            raise
    
    # Test boundary conditions
    print(f"\n--- Boundary Condition Tests ---")
    
    boundary_tests = [
        ("Empty file", ""),
        ("Single character", "A"),
        ("Long line", "X" * 10000),
        ("Many short lines", "\n".join(f"Line {i}" for i in range(1000))),
    ]
    
    for test_name, content in boundary_tests:
        print(f"\n{test_name}:")
        
        # Create test file
        with tempfile.NamedTemporaryFile(mode='w', delete=False, suffix='.txt') as temp_file:
            temp_file.write(content)
            boundary_file_path = temp_file.name
        
        boundary_url = f"file://{boundary_file_path}"
        target_path = tempfile.mktemp(suffix='_boundary.txt')
        
        try:
            # Test with robust downloader
            result = robust_download(boundary_url, target_path)
            
            if os.path.exists(target_path):
                downloaded_size = os.path.getsize(target_path)
                original_size = len(content.encode('utf-8'))
                
                print(f"  Original: {original_size} bytes")
                print(f"  Downloaded: {downloaded_size} bytes")
                print(f"  Size match: {original_size == downloaded_size}")
                assert result == downloaded_size == original_size, (test_name, result, downloaded_size, original_size)
                
                os.unlink(target_path)
            else:
                raise AssertionError(f"{test_name}: robust download did not create {target_path}")
        
        except Exception as e:
            print(f"  Error: {e}")
            raise
        
        # Cleanup
        try:
            os.unlink(boundary_file_path)
        except:
            pass
            raise
    
    # Test concurrent downloads simulation
    print(f"\n--- Download Management ---")
    
    # Create multiple test files
    test_files = []
    for i in range(3):
        with tempfile.NamedTemporaryFile(mode='w', delete=False, suffix=f'_concurrent{i}.txt') as temp_file:
            temp_file.write(f"Concurrent test file {i+1}\n" * 100)
            test_files.append(temp_file.name)
    
    print(f"Created {len(test_files)} files for concurrent testing")
    
    # Download all files sequentially (simulating management)
    successful_downloads = 0
    total_size = 0
    
    for i, source_path in enumerate(test_files):
        file_url = f"file://{source_path}"
        target_path = tempfile.mktemp(suffix=f'_managed{i}.txt')
        
        try:
            downloader = multiurl.Downloader(file_url)
            result = downloader.download(target_path)
            
            if os.path.exists(target_path):
                size = os.path.getsize(target_path)
                total_size += size
                successful_downloads += 1
                print(f"  File {i+1}: {size} bytes downloaded")
                os.unlink(target_path)
            
        except Exception as e:
            print(f"  File {i+1}: Error - {e}")
            raise
    
    print(f"Download summary: {successful_downloads}/{len(test_files)} successful, {total_size} total bytes")
    
    # Cleanup
    for source_path in [test_file_path] + test_files:
        try:
            os.unlink(source_path)
        except:
            pass
            raise
    server.shutdown()
    server.server_close()
    
    print("multiurl example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"multiurl example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: multiurl functionality test 5

```python
import multiurl
import tempfile
import os
import time

# Example 5: Production workflows and advanced download management
print("multiurl production workflows and advanced download management")

try:
    print("Example 5 - Production workflows and advanced download management")
    
    # Simulate a production download management system
    print("--- Production Download Manager Simulation ---")
    
    class DownloadManager:
        def __init__(self):
            self.downloads = []
            self.stats = {'successful': 0, 'failed': 0, 'total_bytes': 0}
        
        def add_download(self, url, target_path, strategy='standard'):
            self.downloads.append({
                'url': url,
                'target': target_path,
                'strategy': strategy,
                'status': 'pending'
            })
        
        def execute_downloads(self):
            print(f"Executing {len(self.downloads)} downloads...")
            
            for i, download in enumerate(self.downloads):
                print(f"\nDownload {i+1}/{len(self.downloads)}: {os.path.basename(download['target'])}")
                
                try:
                    start_time = time.time()
                    
                    # Choose download strategy
                    if download['strategy'] == 'standard':
                        result = multiurl.download(download['url'], download['target'])
                    elif download['strategy'] == 'robust':
                        result = robust_download(download['url'], download['target'])
                    else:  # downloader class
                        downloader = multiurl.Downloader(download['url'])
                        result = downloader.download(download['target'])
                    
                    end_time = time.time()
                    duration = end_time - start_time
                    
                    if os.path.exists(download['target']):
                        size = os.path.getsize(download['target'])
                        speed = size / duration if duration > 0 else 0
                        
                        download['status'] = 'completed'
                        download['size'] = size
                        download['duration'] = duration
                        download['speed'] = speed
                        
                        self.stats['successful'] += 1
                        self.stats['total_bytes'] += size
                        
                        print(f"  Success: {size} bytes in {duration:.3f}s ({speed:.0f} B/s)")
                        assert size == os.path.getsize(download['url'][len("file://"):]), (download, size)
                    else:
                        download['status'] = 'failed'
                        self.stats['failed'] += 1
                        raise AssertionError(f"{download['strategy']} download did not create {download['target']}")
                
                except Exception as e:
                    download['status'] = 'error'
                    download['error'] = str(e)
                    self.stats['failed'] += 1
                    print(f"  Error: {e}")
                    raise
        
        def generate_report(self):
            print(f"\n--- Download Manager Report ---")
            print(f"Total downloads: {len(self.downloads)}")
            print(f"Successful: {self.stats['successful']}")
            print(f"Failed: {self.stats['failed']}")
            print(f"Success rate: {(self.stats['successful']/len(self.downloads)*100) if self.downloads else 0:.1f}%")
            print(f"Total bytes downloaded: {self.stats['total_bytes']:,}")
            
            if self.stats['successful'] > 0:
                avg_size = self.stats['total_bytes'] / self.stats['successful']
                print(f"Average file size: {avg_size:.0f} bytes")
                
                successful_downloads = [d for d in self.downloads if d['status'] == 'completed']
                if successful_downloads:
                    avg_duration = sum(d['duration'] for d in successful_downloads) / len(successful_downloads)
                    avg_speed = sum(d['speed'] for d in successful_downloads) / len(successful_downloads)
                    print(f"Average download time: {avg_duration:.3f}s")
                    print(f"Average download speed: {avg_speed:.0f} B/s")
    
    # Create test files for the download manager
    test_files = []
    file_sizes = [500, 1500, 3000, 1000, 2000]  # Different sizes
    
    for i, size in enumerate(file_sizes, 1):
        with tempfile.NamedTemporaryFile(mode='w', delete=False, suffix=f'_prod{i}.txt') as temp_file:
            content = f"Production test file {i}\n" + "Content line\n" * size
            temp_file.write(content)
            test_files.append(temp_file.name)
    
    print(f"Created {len(test_files)} test files for production simulation")
    
    # multiurl.robust(call, maximum_tries, retry_after) wraps a requests-style call (url -> response)
    # and retries it; it is not a downloader object and has no .download(). Serve the files over a
    # loopback HTTP server so the robust path runs for real, offline.
    import functools, http.server, threading, requests

    class _QuietHandler(http.server.SimpleHTTPRequestHandler):
        def log_message(self, *args):
            pass

    server = http.server.ThreadingHTTPServer(
        ("127.0.0.1", 0), functools.partial(_QuietHandler, directory=tempfile.gettempdir()))
    threading.Thread(target=server.serve_forever, daemon=True).start()
    session = requests.Session()
    session.trust_env = False  # loopback only, never through an ambient proxy
    robust_get = multiurl.robust(session.get, maximum_tries=3, retry_after=0)

    def robust_download(url, target):
        # the file:// URL's file, fetched from the loopback server through multiurl.robust
        response = robust_get(f"http://127.0.0.1:{server.server_port}/{os.path.basename(url)}")
        response.raise_for_status()
        with open(target, 'wb') as out:
            out.write(response.content)
        return len(response.content)

    # Initialize download manager
    manager = DownloadManager()
    
    # Add downloads with different strategies
    strategies = ['standard', 'robust', 'class']
    for i, source_path in enumerate(test_files):
        file_url = f"file://{source_path}"
        target_path = tempfile.mktemp(suffix=f'_prod_download{i}.txt')
        strategy = strategies[i % len(strategies)]
        
        manager.add_download(file_url, target_path, strategy)
        print(f"Added download {i+1}: {strategy} strategy")
    
    # Execute all downloads
    manager.execute_downloads()
    
    # Generate comprehensive report
    manager.generate_report()
    
    # Advanced analysis
    print(f"\n--- Advanced Analysis ---")
    
    # Strategy performance comparison
    strategy_stats = {}
    for download in manager.downloads:
        strategy = download['strategy']
        if strategy not in strategy_stats:
            strategy_stats[strategy] = {'count': 0, 'successful': 0, 'total_time': 0, 'total_bytes': 0}
        
        strategy_stats[strategy]['count'] += 1
        if download['status'] == 'completed':
            strategy_stats[strategy]['successful'] += 1
            strategy_stats[strategy]['total_time'] += download['duration']
            strategy_stats[strategy]['total_bytes'] += download['size']
    
    print("Strategy Performance:")
    for strategy, stats in strategy_stats.items():
        success_rate = (stats['successful'] / stats['count'] * 100) if stats['count'] > 0 else 0
        avg_speed = (stats['total_bytes'] / stats['total_time']) if stats['total_time'] > 0 else 0
        
        print(f"  {strategy.capitalize()}: {success_rate:.1f}% success, {avg_speed:.0f} B/s avg")
    
    # File size analysis
    print(f"\nFile Size Analysis:")
    completed_downloads = [d for d in manager.downloads if d['status'] == 'completed']
    if completed_downloads:
        sizes = [d['size'] for d in completed_downloads]
        print(f"  Smallest file: {min(sizes):,} bytes")
        print(f"  Largest file: {max(sizes):,} bytes")
        print(f"  Size range: {max(sizes) - min(sizes):,} bytes")
    
    # Cleanup
    print(f"\n--- Cleanup ---")
    cleaned_files = 0
    
    # Clean up source files
    for source_path in test_files:
        try:
            os.unlink(source_path)
            cleaned_files += 1
        except:
            pass
            raise
    
    # Clean up downloaded files
    for download in manager.downloads:
        try:
            if os.path.exists(download['target']):
                os.unlink(download['target'])
                cleaned_files += 1
        except:
            pass
            raise
    
    print(f"Cleaned up {cleaned_files} files")
    server.shutdown()
    server.server_close()
    
    # Production recommendations
    print(f"\n--- Production Recommendations ---")
    print(f"✓ Use robust downloader for critical files")
    print(f"✓ Implement retry logic for failed downloads")
    print(f"✓ Monitor download speeds and adjust timeouts")
    print(f"✓ Log all download attempts for troubleshooting")
    print(f"✓ Validate downloaded file integrity")
    print(f"✓ Implement concurrent downloads for better throughput")
    
    assert manager.stats['failed'] == 0, f"{manager.stats['failed']} downloads failed"
    assert manager.stats['successful'] == len(manager.downloads) == len(file_sizes), manager.stats
    if manager.stats['successful'] == len(manager.downloads):
        print(f"✅ All downloads completed successfully - system is production-ready")
    
    print("multiurl example 5 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"multiurl example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
print(f"\nMultiurl capabilities demonstrated:")
print("- Basic file downloading and multiurl setup")
print("- Protocol support and downloader modules exploration")
print("- Advanced downloading with Downloader class and configuration")
print("- Error handling and resilience features")
print("- Production workflows and advanced download management")
```
