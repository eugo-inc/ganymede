## Example 3: mypy_boto3_ce functionality test 3

```python
from mypy_boto3_ce import CostExplorerClient
import inspect

# Example 3: Inspect CostExplorerClient method signatures
print("CostExplorerClient method signatures:")
print("-" * 50)

# Get key methods and their signatures
methods_to_inspect = ['get_cost_and_usage', 'get_cost_forecast', 'get_anomalies', 'get_tags']

for method_name in methods_to_inspect:
    method = getattr(CostExplorerClient, method_name)
    sig = inspect.signature(method)
    print(f"\n{method_name}:")
    print(f"  Parameters: {list(sig.parameters.keys())[:5]}")
    
# Count all available methods
all_methods = [m for m in dir(CostExplorerClient) if not m.startswith('_')]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 3 completed successfully")
```
