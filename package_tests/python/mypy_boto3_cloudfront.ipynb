```python
from mypy_boto3_cloudfront import CloudFrontClient
from mypy_boto3_cloudfront.type_defs import DistributionConfigTypeDef

# Example 1: Explore CloudFrontClient type definitions
print("CloudFront Client type information:")
print("-" * 50)

# Show client class information
print(f"Client class: {CloudFrontClient.__name__}")
print(f"Client module: {CloudFrontClient.__module__}")

# Explore DistributionConfigTypeDef structure
print(f"\nDistributionConfigTypeDef structure:")
if hasattr(DistributionConfigTypeDef, '__required_keys__'):
    required = sorted(DistributionConfigTypeDef.__required_keys__)
    print(f"  Required keys ({len(required)}): {required[:5]}...")
    optional = DistributionConfigTypeDef.__optional_keys__
    print(f"  Optional keys ({len(optional)}): {sorted(optional)[:5]}...")

print("\nExample 1 completed successfully")
```
