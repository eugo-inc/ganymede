## Example 5: mypy_boto3_codebuild functionality test 5

```python
from mypy_boto3_codebuild import CodeBuildClient
import inspect

# Example 5: Inspect CodeBuildClient method signatures
print("CodeBuildClient method signatures:")
print("-" * 50)

# Get key methods and their signatures
methods_to_inspect = ['start_build', 'create_project', 'batch_get_builds', 'update_project']

for method_name in methods_to_inspect:
    method = getattr(CodeBuildClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(CodeBuildClient) if not m.startswith('_') and callable(getattr(CodeBuildClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
