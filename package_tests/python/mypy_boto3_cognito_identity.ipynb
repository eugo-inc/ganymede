## Example 5: mypy_boto3_cognito_identity functionality test 5

```python
from mypy_boto3_cognito_identity import CognitoIdentityClient
import inspect

# Example 5: Inspect CognitoIdentityClient method signatures
print("CognitoIdentityClient method signatures:")
print("-" * 50)

# Get key methods and their signatures
methods_to_inspect = ['get_id', 'get_credentials_for_identity', 'create_identity_pool', 'describe_identity_pool']

for method_name in methods_to_inspect:
    method = getattr(CognitoIdentityClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(CognitoIdentityClient) if not m.startswith('_') and callable(getattr(CognitoIdentityClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
