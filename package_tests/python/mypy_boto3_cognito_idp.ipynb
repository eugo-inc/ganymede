## Example 5: mypy_boto3_cognito_idp functionality test 5

```python
from mypy_boto3_cognito_idp import CognitoIdentityProviderClient
import inspect

# Example 5: Inspect CognitoIdentityProviderClient method signatures
print("CognitoIdentityProviderClient method signatures:")
print("-" * 50)

# Get key methods and their signatures
methods_to_inspect = ['admin_create_user', 'sign_up', 'initiate_auth', 'confirm_sign_up']

for method_name in methods_to_inspect:
    method = getattr(CognitoIdentityProviderClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(CognitoIdentityProviderClient) if not m.startswith('_') and callable(getattr(CognitoIdentityProviderClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
