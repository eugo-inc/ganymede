## Example 5: mypy_boto3_datasync functionality test 5

```python
from mypy_boto3_datasync import DataSyncClient
import inspect

# Example 5: Inspect DataSyncClient method signatures
print("DataSyncClient method signatures:")
print("-" * 50)

# Get key methods and their signatures
methods_to_inspect = ['create_task', 'start_task_execution', 'create_location_s3', 'describe_task']

for method_name in methods_to_inspect:
    method = getattr(DataSyncClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(DataSyncClient) if not m.startswith('_') and callable(getattr(DataSyncClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
