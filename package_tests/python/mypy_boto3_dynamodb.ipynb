## Example 5: mypy_boto3_dynamodb functionality test 5

```python
from mypy_boto3_dynamodb import DynamoDBClient
import inspect

# Example 5: Inspect DynamoDBClient method signatures
print("DynamoDBClient method signatures:")
print("-" * 50)

# Get key methods and their signatures
methods_to_inspect = ['put_item', 'get_item', 'query', 'scan', 'batch_write_item']

for method_name in methods_to_inspect:
    method = getattr(DynamoDBClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(DynamoDBClient) if not m.startswith('_') and callable(getattr(DynamoDBClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
