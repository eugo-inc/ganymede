## Example 5: mypy_boto3_ec2 functionality test 5

```python
from mypy_boto3_ec2 import EC2Client
import inspect

# Example 5: Inspect EC2Client method signatures
print("EC2Client method signatures:")
print("-" * 50)

# Get key methods and their signatures
methods_to_inspect = ['run_instances', 'describe_instances', 'create_security_group', 'create_vpc']

for method_name in methods_to_inspect:
    method = getattr(EC2Client, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(EC2Client) if not m.startswith('_') and callable(getattr(EC2Client, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
