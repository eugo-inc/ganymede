## Example 5: mypy_boto3_ecs functionality test 5

```python
from mypy_boto3_ecs import ECSClient
import inspect

# Example 5: Inspect ECSClient method signatures
print("ECSClient method signatures:")
print("-" * 50)

# Get key ECS methods and their signatures
methods_to_inspect = ['run_task', 'create_service', 'register_task_definition', 'describe_tasks']

for method_name in methods_to_inspect:
    method = getattr(ECSClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(ECSClient) if not m.startswith('_') and callable(getattr(ECSClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
