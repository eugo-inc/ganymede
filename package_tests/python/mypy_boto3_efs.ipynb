## Example 5: mypy_boto3_efs functionality test 5

```python
from mypy_boto3_efs import EFSClient
import inspect

# Example 5: Inspect EFSClient method signatures
print("EFSClient method signatures:")
print("-" * 50)

# Get key EFS methods and their signatures
methods_to_inspect = ['create_file_system', 'create_access_point', 'create_mount_target', 'describe_file_systems']

for method_name in methods_to_inspect:
    method = getattr(EFSClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(EFSClient) if not m.startswith('_') and callable(getattr(EFSClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
