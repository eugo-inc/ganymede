## Example 5: mypy_boto3_elbv2 functionality test 5

```python
from mypy_boto3_elbv2 import ElasticLoadBalancingv2Client
import inspect

# Example 5: Inspect ELBv2 client method signatures
print("ElasticLoadBalancingv2Client method signatures:")
print("-" * 50)

# Get key ELBv2 methods and their signatures
methods_to_inspect = ['create_load_balancer', 'create_target_group', 'create_listener', 'register_targets']

for method_name in methods_to_inspect:
    method = getattr(ElasticLoadBalancingv2Client, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(ElasticLoadBalancingv2Client) if not m.startswith('_') and callable(getattr(ElasticLoadBalancingv2Client, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
