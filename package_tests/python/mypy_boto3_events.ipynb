## Example 5: mypy_boto3_events functionality test 5

```python
from mypy_boto3_events import EventBridgeClient
import inspect

# Example 5: Inspect EventBridgeClient method signatures
print("EventBridgeClient method signatures:")
print("-" * 50)

# Get key EventBridge methods and their signatures
methods_to_inspect = ['put_rule', 'put_targets', 'put_events', 'create_event_bus']

for method_name in methods_to_inspect:
    method = getattr(EventBridgeClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(EventBridgeClient) if not m.startswith('_') and callable(getattr(EventBridgeClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
