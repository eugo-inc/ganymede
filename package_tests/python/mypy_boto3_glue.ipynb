## Example 5: mypy_boto3_glue functionality test 5

```python
from mypy_boto3_glue import GlueClient
import inspect

# Example 5: Inspect GlueClient method signatures
print("GlueClient method signatures:")
print("-" * 50)

# Get key Glue methods and their signatures
methods_to_inspect = ['create_job', 'create_crawler', 'create_database', 'start_crawler']

for method_name in methods_to_inspect:
    method = getattr(GlueClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(GlueClient) if not m.startswith('_') and callable(getattr(GlueClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
