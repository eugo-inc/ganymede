## Example 5: mypy_boto3_iam functionality test 5

```python
from mypy_boto3_iam import IAMClient
import inspect

# Example 5: Inspect IAMClient method signatures
print("IAMClient method signatures:")
print("-" * 50)

# Get key IAM methods and their signatures
methods_to_inspect = ['create_user', 'create_role', 'attach_role_policy', 'create_access_key']

for method_name in methods_to_inspect:
    method = getattr(IAMClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(IAMClient) if not m.startswith('_') and callable(getattr(IAMClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
