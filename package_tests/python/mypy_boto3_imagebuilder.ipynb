## Example 5: mypy_boto3_imagebuilder functionality test 5

```python
from mypy_boto3_imagebuilder import ImagebuilderClient
import inspect

# Example 5: Inspect ImagebuilderClient method signatures
print("ImagebuilderClient method signatures:")
print("-" * 50)

# Get key Image Builder methods and their signatures
methods_to_inspect = ['create_image', 'create_component', 'create_image_pipeline', 'create_image_recipe']

for method_name in methods_to_inspect:
    method = getattr(ImagebuilderClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(ImagebuilderClient) if not m.startswith('_') and callable(getattr(ImagebuilderClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
