## Example 5: mypy_boto3_kinesis functionality test 5

```python
from mypy_boto3_kinesis import KinesisClient
import importlib.resources
import inspect

# Example 5: Inspect KinesisClient method signatures
print("KinesisClient method signatures:")
print("-" * 50)

# Get key Kinesis methods and their signatures
methods_to_inspect = ['create_stream', 'put_record', 'get_records', 'get_shard_iterator']

for method_name in methods_to_inspect:
    method = getattr(KinesisClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(KinesisClient) if not m.startswith('_') and callable(getattr(KinesisClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")
assert set(methods_to_inspect) <= set(all_methods), sorted(set(methods_to_inspect) - set(all_methods))

# What a type checker reads: the PEP 561 marker and the .pyi stubs must ship with the package
package_files = importlib.resources.files('mypy_boto3_kinesis')
for stub_file in ['py.typed', 'client.pyi', 'type_defs.pyi']:
    assert package_files.joinpath(stub_file).is_file(), f"mypy_boto3_kinesis ships no {stub_file}"

print("\nExample 5 completed successfully")
```
