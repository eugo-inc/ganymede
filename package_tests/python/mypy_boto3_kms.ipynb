## Example 5: mypy_boto3_kms functionality test 5

```python
from mypy_boto3_kms import KMSClient
import inspect

# Example 5: Inspect KMSClient method signatures
print("KMSClient method signatures:")
print("-" * 50)

# Get key KMS methods and their signatures
methods_to_inspect = ['create_key', 'encrypt', 'decrypt', 'generate_data_key']

for method_name in methods_to_inspect:
    method = getattr(KMSClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(KMSClient) if not m.startswith('_') and callable(getattr(KMSClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
