## Example 5: mypy_boto3_organizations functionality test 5

```python
from mypy_boto3_organizations import OrganizationsClient
import inspect

# Example 5: Inspect OrganizationsClient method signatures
print("OrganizationsClient method signatures:")
print("-" * 50)

# Get key Organizations methods and their signatures
methods_to_inspect = ['create_account', 'create_organizational_unit', 'list_accounts', 'attach_policy']

for method_name in methods_to_inspect:
    method = getattr(OrganizationsClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(OrganizationsClient) if not m.startswith('_') and callable(getattr(OrganizationsClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
