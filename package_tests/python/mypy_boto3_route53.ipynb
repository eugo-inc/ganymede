## Example 5: mypy_boto3_route53 functionality test 5

```python
from mypy_boto3_route53 import Route53Client
import inspect

# Example 5: Inspect Route53Client method signatures
print("Route53Client method signatures:")
print("-" * 50)

# Get key Route 53 methods and their signatures
methods_to_inspect = ['create_hosted_zone', 'change_resource_record_sets', 'create_health_check', 'list_hosted_zones']

for method_name in methods_to_inspect:
    method = getattr(Route53Client, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(Route53Client) if not m.startswith('_') and callable(getattr(Route53Client, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
