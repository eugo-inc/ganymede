## Example 5: mypy_boto3_s3 functionality test 5

```python
from mypy_boto3_s3 import S3Client
import inspect

# Example 5: Inspect S3Client method signatures
print("S3Client method signatures:")
print("-" * 50)

# Get key S3 methods and their signatures
methods_to_inspect = ['create_bucket', 'put_object', 'get_object', 'list_objects_v2']

for method_name in methods_to_inspect:
    method = getattr(S3Client, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(S3Client) if not m.startswith('_') and callable(getattr(S3Client, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
