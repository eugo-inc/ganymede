## Example 5: mypy_boto3_secretsmanager functionality test 5

```python
from mypy_boto3_secretsmanager import SecretsManagerClient
import inspect

# Example 5: Inspect SecretsManagerClient method signatures
print("SecretsManagerClient method signatures:")
print("-" * 50)

# Get key Secrets Manager methods and their signatures
methods_to_inspect = ['create_secret', 'get_secret_value', 'rotate_secret', 'delete_secret']

for method_name in methods_to_inspect:
    method = getattr(SecretsManagerClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(SecretsManagerClient) if not m.startswith('_') and callable(getattr(SecretsManagerClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
