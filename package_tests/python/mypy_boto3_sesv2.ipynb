## Example 5: mypy_boto3_sesv2 functionality test 5

```python
from mypy_boto3_sesv2 import SESV2Client
import inspect

# Example 5: Inspect SESV2Client method signatures
print("SESV2Client method signatures:")
print("-" * 50)

# Get key SES v2 methods and their signatures
methods_to_inspect = ['send_email', 'create_contact', 'create_email_template', 'put_account_sending_attributes']

for method_name in methods_to_inspect:
    method = getattr(SESV2Client, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(SESV2Client) if not m.startswith('_') and callable(getattr(SESV2Client, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
