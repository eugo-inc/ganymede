## Example 5: mypy_boto3_ssm functionality test 5

```python
from mypy_boto3_ssm import SSMClient
import inspect

# Example 5: Inspect SSMClient method signatures
print("SSMClient method signatures:")
print("-" * 50)

# Get key SSM methods and their signatures
methods_to_inspect = ['put_parameter', 'get_parameter', 'send_command', 'describe_instance_information']

for method_name in methods_to_inspect:
    method = getattr(SSMClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(SSMClient) if not m.startswith('_') and callable(getattr(SSMClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
