## Example 5: mypy_boto3_sts functionality test 5

```python
from mypy_boto3_sts import STSClient
import inspect

# Example 5: Inspect STSClient method signatures
print("STSClient method signatures:")
print("-" * 50)

# Get key STS methods and their signatures
methods_to_inspect = ['assume_role', 'get_caller_identity', 'get_session_token', 'assume_role_with_web_identity']

for method_name in methods_to_inspect:
    method = getattr(STSClient, method_name)
    sig = inspect.signature(method)
    print(f"{method_name}: {list(sig.parameters.keys())[:3]}")

# Count all available methods
all_methods = [m for m in dir(STSClient) if not m.startswith('_') and callable(getattr(STSClient, m))]
print(f"\nTotal public methods: {len(all_methods)}")

print("\nExample 5 completed successfully")
```
