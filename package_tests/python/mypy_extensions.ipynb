# mypy_extensions Package Testing

This notebook tests the mypy_extensions package, which provides additional type system constructs for Python type checking.

## Example 1: TypedDict usage

```python
from mypy_extensions import TypedDict
class Person(TypedDict):
    name: str
    age: int
person = {'name': 'Alice', 'age': 30}
print(f"Person: {person}")
```

## Example 2: Arg annotations

```python
from mypy_extensions import Arg
def func(x: Arg(str, 'x')):
    return x.upper()
result = func('hello')
print(f"Result: {result}")
```

## Example 3: VarArg usage

```python
from mypy_extensions import VarArg
def func(*args: VarArg(int)):
    return sum(args)
result = func(1, 2, 3)
print(f"Sum: {result}")
```

## Example 4: NoReturn annotation

```python
from mypy_extensions import NoReturn
def never_returns() -> NoReturn:
    raise ValueError('Never returns')
print(f"NoReturn defined")
```
