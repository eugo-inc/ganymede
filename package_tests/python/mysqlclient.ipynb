## Example 1: mysqlclient functionality test 1

```python
import MySQLdb

# Example 1: Basic MySQL module information and constants
print("MySQLdb (mysqlclient) Module Information:")
print("-" * 50)

# Display module information
print(f"API Level: {MySQLdb.apilevel}")
print(f"Thread safety: {MySQLdb.threadsafety}")
print(f"Parameter style: {MySQLdb.paramstyle}")

# Version information
if hasattr(MySQLdb, 'version_info'):
    print(f"Version info: {MySQLdb.version_info}")

# Get client info
client_info = MySQLdb.get_client_info()
print(f"MySQL client info: {client_info}")

# Data type constants
print(f"\nData Type Constants:")
data_types = ['STRING', 'BINARY', 'NUMBER', 'DATE', 'DATETIME', 'TIMESTAMP']
for dtype in data_types:
    if hasattr(MySQLdb, dtype):
        print(f"  {dtype}: {getattr(MySQLdb, dtype)}")

print("\nExample 1 completed successfully")
```

## Example 2: mysqlclient functionality test 2

```python
import MySQLdb
import datetime

# Example 2: Date/Time and binary data handling
print("MySQLdb Date/Time and Binary Data Handling:")
print("-" * 50)

# Date and time utilities
print("Date/Time Utilities:")
mysql_date = MySQLdb.Date(2024, 1, 15)
print(f"  Date: {mysql_date}")

mysql_time = MySQLdb.Time(14, 30, 45)
print(f"  Time: {mysql_time}")

mysql_timestamp = MySQLdb.Timestamp(2024, 1, 15, 14, 30, 45)
print(f"  Timestamp: {mysql_timestamp}")

# Binary data handling
print(f"\nBinary Data Handling:")
test_data = b"Hello, MySQL Binary Data!"
mysql_binary = MySQLdb.Binary(test_data)
print(f"  Original: {test_data}")
print(f"  Binary object type: {type(mysql_binary)}")

# Field type constants
if hasattr(MySQLdb, 'FIELD_TYPE'):
    field_types = [attr for attr in dir(MySQLdb.FIELD_TYPE) if not attr.startswith('_')]
    print(f"\nField type constants available: {len(field_types)}")

print("\nExample 2 completed successfully")
```

## Example 3: mysqlclient functionality test 3

```python
import MySQLdb

# Example 3: Exception hierarchy and error handling
print("MySQLdb Exception Hierarchy:")
print("-" * 50)

# Get all exception classes
exception_classes = {
    'Warning': MySQLdb.Warning,
    'Error': MySQLdb.Error,
    'InterfaceError': MySQLdb.InterfaceError,
    'DatabaseError': MySQLdb.DatabaseError,
    'DataError': MySQLdb.DataError,
    'OperationalError': MySQLdb.OperationalError,
    'IntegrityError': MySQLdb.IntegrityError,
    'InternalError': MySQLdb.InternalError,
    'ProgrammingError': MySQLdb.ProgrammingError,
    'NotSupportedError': MySQLdb.NotSupportedError
}

# Show exception hierarchy
for exc_name, exc_class in exception_classes.items():
    base_names = [base.__name__ for base in exc_class.__bases__]
    print(f"{exc_name}: {exc_class}")
    print(f"  Base classes: {base_names}")

print("\nExample 3 completed successfully")
```

## Example 4: mysqlclient functionality test 4

```python
import MySQLdb

# Example 4: SQL parameterization and query safety
print("MySQLdb SQL Parameterization:")
print("-" * 50)

print(f"Parameter style: {MySQLdb.paramstyle}")

# Safe query examples with parameters
safe_queries = [
    ('SELECT * FROM users WHERE id = %s', (123,)),
    ('SELECT * FROM users WHERE username = %s AND active = %s', ('john_doe', True)),
    ('INSERT INTO products (name, price, quantity) VALUES (%s, %s, %s)', ('Widget', 29.99, 100))
]

print("\nSafe parameterized queries:")
for i, (query, params) in enumerate(safe_queries, 1):
    print(f"\n  Example {i}:")
    print(f"    Query: {query}")
    print(f"    Params: {params}")

print("\nExample 4 completed successfully")
```
