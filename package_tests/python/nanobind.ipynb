```python
# Example 1: package metadata + version sanity check
import nanobind

version = nanobind.__version__
print(f"nanobind version: {version}")

# Version is a dotted string like '2.12.0'
parts = version.split('.')
assert len(parts) >= 2, f"unexpected version format: {version}"
assert all(p.split('+')[0].split('rc')[0].split('a')[0].split('b')[0].isdigit() or p[0].isdigit() for p in parts[:2]), f"non-numeric version parts: {parts}"

# Major version should be at least 1 for a real release
major = int(parts[0])
assert major >= 1, f"unexpected major version: {major}"
print(f"Major version: {major}")
```

```python
# Example 2: locate the C++ headers that nanobind ships
import os
import nanobind

# nanobind ships a `include/` directory with its C++ headers, used by
# build systems that compile native extension modules.
pkg_dir = os.path.dirname(nanobind.__file__)
print(f"nanobind package dir: {pkg_dir}")

include_dir = os.path.join(pkg_dir, 'include')
print(f"include dir: {include_dir}")
assert os.path.isdir(include_dir), f"missing include dir: {include_dir}"

# Primary header is nanobind/nanobind.h
main_header = os.path.join(include_dir, 'nanobind', 'nanobind.h')
assert os.path.isfile(main_header), f"missing main header: {main_header}"
print(f"Found main header: {main_header}")
```

```python
# Example 3: cmake_dir() / include_dir() helpers used by build systems
import os
import nanobind

# nanobind exposes helper functions that build systems (CMake, scikit-build)
# call to locate its files.
cmake_dir = nanobind.cmake_dir()
print(f"cmake_dir: {cmake_dir}")
assert os.path.isdir(cmake_dir), f"cmake_dir not a directory: {cmake_dir}"

include_dir = nanobind.include_dir()
print(f"include_dir: {include_dir}")
assert os.path.isdir(include_dir), f"include_dir not a directory: {include_dir}"

# CMake config files should live under cmake_dir
cmake_config = os.path.join(cmake_dir, 'nanobind-config.cmake')
assert os.path.isfile(cmake_config), f"missing nanobind-config.cmake at {cmake_config}"
print(f"Found nanobind-config.cmake")
```

```python
# Example 4: bundled C++ source files for the nanobind library
import os
import nanobind

# nanobind ships a `src/` directory with the C++ source files that the
# downstream user's build compiles into libnanobind.
pkg_dir = os.path.dirname(nanobind.__file__)
src_dir = os.path.join(pkg_dir, 'src')
print(f"src dir: {src_dir}")
assert os.path.isdir(src_dir), f"missing src dir: {src_dir}"

cpp_files = [f for f in os.listdir(src_dir) if f.endswith(('.cpp', '.h'))]
print(f"Found {len(cpp_files)} C++ source/header files")
assert len(cpp_files) > 0, "expected at least one C++ source file"
for f in sorted(cpp_files)[:10]:
    print(f"  {f}")
```

```python
# Example 5: stubgen module exists (nanobind's stub generator)
import nanobind
import nanobind.stubgen

# stubgen is nanobind's CLI tool for generating .pyi stub files for
# bindings. We just check that it imports and exposes its core entry.
print(f"nanobind.stubgen module: {nanobind.stubgen}")
print(f"file: {nanobind.stubgen.__file__}")

# The module should define StubGen (the class that drives stub generation).
assert hasattr(nanobind.stubgen, 'StubGen'), "nanobind.stubgen.StubGen missing"
print(f"StubGen class: {nanobind.stubgen.StubGen}")

# Sanity: it should be a class (not a stub).
import inspect
assert inspect.isclass(nanobind.stubgen.StubGen), "StubGen is not a class"
print("nanobind.stubgen.StubGen is a class - OK")
```
