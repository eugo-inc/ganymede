## Example 1: Basic NATS client setup and configuration

```python
import nats

# Example 1: NATS module structure and components
print("NATS Module Structure:")
print("-" * 50)

# Show available components
components = [attr for attr in dir(nats) if not attr.startswith('_')]
print(f"Available components: {components}")

# Main client class
print(f"\nMain NATS client class: {nats.NATS}")

# Error types
error_types = [attr for attr in dir(nats.errors) if 'Error' in attr and not attr.startswith('_')]
print(f"\nError types ({len(error_types)}): {error_types[:5]}...")

# Protocol components
if hasattr(nats, 'protocol'):
    protocol_items = [attr for attr in dir(nats.protocol) if not attr.startswith('_')]
    print(f"Protocol components: {protocol_items}")

# NUID (unique ID generation)
if hasattr(nats, 'nuid'):
    print(f"\nNUID module: {nats.nuid}")

print("\nExample 1 completed successfully")
```

## Example 2: Message publishing and subscription patterns

```python
import nats

# Example 2: NATS subject patterns and messaging concepts
print("NATS Subject Patterns:")
print("-" * 50)

# Subject pattern examples
subject_patterns = {
    "Simple": "user.login",
    "Hierarchical": "orders.us.west.item123",
    "Wildcard (*)": "user.* (matches user.login, user.logout)",
    "Full wildcard (>)": "orders.> (matches orders.us, orders.us.west, etc.)",
    "Multiple levels": "metrics.*.cpu (matches metrics.server1.cpu, metrics.server2.cpu)"
}

print("Subject pattern types:")
for pattern_type, example in subject_patterns.items():
    print(f"  {pattern_type}: {example}")

# Message structure
print("\nMessage structure fields:")
message_fields = ["subject", "data", "reply", "headers"]
for field in message_fields:
    print(f"  - {field}")

# Queue groups concept
print("\nQueue groups: Load balancing messages among subscribers")
print("  Example: subscribe('work.tasks', queue='workers')")

print("\nExample 2 completed successfully")
```

## Example 3: Request-reply patterns and service communication

```python
import nats.errors as errors

# Example 3: NATS error handling and exception types
print("NATS Exception Types:")
print("-" * 50)

# All error classes
exception_classes = [
    errors.Error,
    errors.ConnectionClosedError,
    errors.TimeoutError,
    errors.NoServersError,
    errors.MaxPayloadError
]

print("Core exception types:")
for exc_class in exception_classes:
    print(f"  {exc_class.__name__}: {exc_class}")

# Show all available errors
all_errors = [attr for attr in dir(errors) if 'Error' in attr and not attr.startswith('_')]
print(f"\nTotal error types available: {len(all_errors)}")
print(f"Sample errors: {all_errors[:10]}")

# Error hierarchy
print(f"\nBase error class: {errors.Error}")
print(f"  Derived errors handle specific failure scenarios")

print("\nExample 3 completed successfully")
```

## Example 4: Stream processing and JetStream functionality

```python
import nats.js as js

# Example 4: JetStream components and configuration
print("NATS JetStream Components:")
print("-" * 50)

# JetStream components
js_components = [attr for attr in dir(js) if not attr.startswith('_')]
print(f"JetStream components: {js_components}")

# Stream configuration options
print("\nStream configuration options:")
stream_options = {
    "name": "Stream identifier",
    "subjects": "List of subjects to capture",
    "retention": "limits, interest, or workqueue",
    "max_age": "Maximum message age in nanoseconds",
    "max_msgs": "Maximum number of messages",
    "replicas": "Number of replicas for HA"
}

for option, description in stream_options.items():
    print(f"  {option}: {description}")

# Consumer configuration
print("\nConsumer configuration options:")
consumer_options = ["durable_name", "deliver_policy", "ack_policy", "max_deliver", "ack_wait"]
for option in consumer_options:
    print(f"  - {option}")

print("\nExample 4 completed successfully")
```

## Example 5: Production deployment and monitoring patterns

```python
import nats

# Example 5: NATS connection configuration options
print("NATS Connection Configuration:")
print("-" * 50)

# Connection options
connection_options = {
    "servers": ["nats://localhost:4222", "nats://demo.nats.io:4222"],
    "name": "My NATS Client",
    "reconnect_time_wait": 2,
    "max_reconnect_attempts": 10,
    "connect_timeout": 10,
    "drain_timeout": 30
}

print("Connection options:")
for option, value in connection_options.items():
    print(f"  {option}: {value}")

# Client instance
nc = nats.NATS()
print(f"\nNATS client instance: {type(nc).__name__}")

# Available client methods
client_methods = [m for m in dir(nc) if not m.startswith('_') and callable(getattr(nc, m))]
print(f"Client methods ({len(client_methods)}): {client_methods}")

print("\nExample 5 completed successfully")
```
