## Example 1: Basic natural sorting and comparison with standard sort

```python
# Example 1: Basic natural sorting comparison
import natsort

print("Example 1: Basic Natural Sorting")
print(f"natsort version: {getattr(natsort, '__version__', 'unknown')}")

# Demonstrate the problem with standard sorting
file_list = ["file1.txt", "file10.txt", "file2.txt", "file20.txt", "file100.txt", "file3.txt"]

print(f"\nOriginal: {file_list}")

# Standard Python sort (lexicographic)
standard_sorted = sorted(file_list)
print(f"Standard: {standard_sorted}")

# Natural sort (human-friendly)
natural_sorted = natsort.natsorted(file_list)
print(f"Natural:  {natural_sorted}")

# Version numbers
versions = ["v1.2.3", "v1.10.1", "v1.2.10", "v2.0.0", "v1.1.0"]
print(f"\nVersions: {versions}")
print(f"Natural:  {natsort.natsorted(versions)}")

print("\nExample 1 completed successfully")
```

## Example 2: Advanced sorting options and customization

```python
# Example 2: Advanced sorting options and customization
from natsort import natsorted, ns

print("Example 2: Advanced Sorting Options and Customization")

# Test data with various challenges
mixed_data = ["item2", "ITEM10", "Item1", "item11", "ITEM3", "item100", "Item20"]

print(f"Original: {mixed_data}")

# Case-insensitive sorting
case_insensitive = natsorted(mixed_data, alg=ns.IGNORECASE)
print(f"Case-insensitive: {case_insensitive}")

# Numeric data with different formats
numeric_data = ["1.5", "10.2", "2.1", "1.10", "10.10", "2.2"]
real_numeric = natsorted(numeric_data, alg=ns.REAL)
print(f"\nNumeric: {numeric_data}")
print(f"Real numeric sort: {real_numeric}")

# Signed number handling
signed_data = ["item-2", "item+10", "item1", "item-11", "item+3"]
signed_sort = natsorted(signed_data, alg=ns.SIGNED)
print(f"\nSigned: {signed_data}")
print(f"Signed sort: {signed_sort}")

print("\nExample 2 completed successfully")
```

## Example 3: Locale-aware sorting and internationalization

```python
# Example 3: Locale-aware and human-friendly sorting
from natsort import natsorted, humansorted, ns

print("Example 3: Locale-aware and Human-friendly Sorting")

# International characters
international_data = ["café10", "cafe2", "café1", "cafe20", "naïve5", "naive10"]

print(f"International: {international_data}")

# Standard natural sort
standard_natural = natsorted(international_data)
print(f"Standard natural: {standard_natural}")

# Human-friendly sorting
human_sorted = humansorted(international_data)
print(f"Human-sorted: {human_sorted}")

# Case handling with international characters
case_intl = ["Café10", "CAFÉ1", "café2", "Cafe20", "NAÏVE5", "naïve1"]
case_insensitive = natsorted(case_intl, alg=ns.IGNORECASE)
print(f"\nCase-insensitive: {case_insensitive}")

print("\nExample 3 completed successfully")
```

## Example 4: Real-world applications and performance optimization

```python
# Example 4: Index-based operations and os_sorted
from natsort import natsorted, os_sorted, index_natsorted, order_by_index

print("Example 4: Index Operations and OS-style Sorting")

# Index-based sorting
file_listing = ["document1.pdf", "document10.pdf", "document2.pdf", "image1.jpg", "image10.jpg"]

print(f"Files: {file_listing}")

# OS-style sorting (like file manager)
os_style = os_sorted(file_listing)
print(f"OS-style: {os_style}")

# Get sorting indices
sort_indices = index_natsorted(file_listing)
print(f"\nSort indices: {sort_indices}")

# Apply indices to parallel arrays
file_sizes = [1024, 10240, 2048, 512, 5120]
sorted_files = order_by_index(file_listing, sort_indices)
sorted_sizes = order_by_index(file_sizes, sort_indices)

print(f"Sorted files: {sorted_files}")
print(f"Sorted sizes: {sorted_sizes}")

print("\nExample 4 completed successfully")
```

## Example 5: Advanced features and integration patterns

```python
# Example 5: Key generation and realsorted
from natsort import natsorted, realsorted, natsort_keygen

print("Example 5: Key Generation and Real Number Sorting")

# Real number sorting
mixed_numeric = ["price10.5", "price2.25", "price1.0", "price100.75", "price2.5"]

print(f"Mixed numeric: {mixed_numeric}")

# Standard natural sort
standard_nat = natsorted(mixed_numeric)
print(f"Standard: {standard_nat}")

# Real number aware sorting
real_sorted = realsorted(mixed_numeric)
print(f"Real sorted: {real_sorted}")

# Key generation for repeated sorting
keyfunc = natsort_keygen()
data = ["item10", "item2", "item1", "item20"]
sorted_data = sorted(data, key=keyfunc)
print(f"\nWith key function: {sorted_data}")

# Sort complex data structures
products = [{"name": "Product10"}, {"name": "Product2"}, {"name": "Product1"}]
name_sorted = natsorted(products, key=lambda x: x["name"])
print(f"Products: {[p['name'] for p in name_sorted]}")

print("\nExample 5 completed successfully")
```
