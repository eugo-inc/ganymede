## Example 1: Basic nbconvert exporters and format support

```python
import nbconvert

print("Example 1: Basic nbconvert exporters")
print(f"nbconvert version: {nbconvert.__version__}")

# Get available export formats
export_names = nbconvert.get_export_names()
print(f"\nSupported formats ({len(export_names)}): {export_names}")

# Test exporter instantiation
html_exporter = nbconvert.HTMLExporter()
print(f"\nHTML exporter: {type(html_exporter).__name__}")
print(f"  Template: {html_exporter.template_file}")
print(f"  MIME type: {html_exporter.output_mimetype}")

md_exporter = nbconvert.MarkdownExporter()
print(f"\nMarkdown exporter: {type(md_exporter).__name__}")

py_exporter = nbconvert.PythonExporter()
print(f"Python exporter: {type(py_exporter).__name__}")

print("\nExample 1 completed successfully")
```

## Example 2: Notebook conversion and export workflow

```python
import nbconvert
import nbformat

print("Example 2: Notebook conversion workflow")

# Create a simple notebook
nb = nbformat.v4.new_notebook()
nb.cells.append(nbformat.v4.new_markdown_cell("# Test Notebook"))
nb.cells.append(nbformat.v4.new_code_cell("print('Hello, nbconvert!')"))

print(f"Created notebook with {len(nb.cells)} cells")

# Test HTML conversion
html_exporter = nbconvert.HTMLExporter()
body, resources = html_exporter.from_notebook_node(nb)
print(f"\nHTML conversion: {len(body)} characters")
print(f"Resources keys: {list(resources.keys())[:5]}...")

# Test Markdown conversion
md_exporter = nbconvert.MarkdownExporter()
md_body, md_resources = md_exporter.from_notebook_node(nb)
print(f"\nMarkdown conversion: {len(md_body)} characters")

print("\nExample 2 completed successfully")
```

## Example 3: Preprocessors and postprocessors customization

```python
import nbconvert

print("Example 3: Preprocessors and postprocessors")

# Explore preprocessors
print("Preprocessors module:")
preprocessor_attrs = [attr for attr in dir(nbconvert.preprocessors) if not attr.startswith('_')]
print(f"  Components: {len(preprocessor_attrs)}")
print(f"  Key preprocessors: {preprocessor_attrs[:5]}")

# Explore postprocessors
print("\nPostprocessors module:")
postprocessor_attrs = [attr for attr in dir(nbconvert.postprocessors) if not attr.startswith('_')]
print(f"  Components: {len(postprocessor_attrs)}")
print(f"  Key postprocessors: {postprocessor_attrs}")

# Exporter configuration
html_exporter = nbconvert.HTMLExporter()
print(f"\nHTML exporter configuration:")
print(f"  Exclude input: {html_exporter.exclude_input}")
print(f"  Exclude output: {html_exporter.exclude_output}")

print("\nExample 3 completed successfully")
```

## Example 4: Template system and custom output formats

```python
import nbconvert

print("Example 4: Template system and filters")

# Explore template-based exporters
print("Template-based exporters:")
exporters = ["HTMLExporter", "LatexExporter", "SlidesExporter", "MarkdownExporter"]
for exp_name in exporters:
    if hasattr(nbconvert, exp_name):
        print(f"  {exp_name}: Available")

# Template configuration
html_exporter = nbconvert.HTMLExporter()
print(f"\nHTML template:")
print(f"  Template file: {html_exporter.template_file}")
print(f"  Template paths: Available for customization")

# Filters module
print(f"\nFilters module: {nbconvert.filters}")
filter_attrs = [attr for attr in dir(nbconvert.filters) if not attr.startswith('_')]
print(f"  Filter components: {len(filter_attrs)}")

print("\nExample 4 completed successfully")
```

## Example 5: Command-line interface and production workflows

```python
import nbconvert

print("Example 5: Export function and writers")

# Test export function
print("Using export() function:")
html_exporter = nbconvert.HTMLExporter()
print(f"  Exporter created: {type(html_exporter).__name__}")

# Writers module
if hasattr(nbconvert, 'writers'):
    print(f"\nWriters module: {nbconvert.writers}")
    writer_attrs = [attr for attr in dir(nbconvert.writers) if not attr.startswith('_')]
    print(f"  Writer components: {len(writer_attrs)}")
    print(f"  Key writers: {writer_attrs[:5]}")

# Exporter classes
print(f"\nAvailable exporter classes:")
exporter_classes = ['HTMLExporter', 'PDFExporter', 'LatexExporter', 'MarkdownExporter', 'PythonExporter']
for exp_class in exporter_classes:
    status = "Available" if hasattr(nbconvert, exp_class) else "Not available"
    print(f"  {exp_class}: {status}")

print("\nExample 5 completed successfully")
```
