## Example 1: Basic notebook diffing and comparison

```python
import nbdime
import nbformat

print("Example 1: Basic notebook diffing")
print(f"nbdime version: {nbdime.__version__}")

# Create two sample notebooks
nb1 = nbformat.v4.new_notebook()
nb1.cells.append(nbformat.v4.new_markdown_cell("# Analysis v1"))
nb1.cells.append(nbformat.v4.new_code_cell("data = [1, 2, 3]"))

nb2 = nbformat.v4.new_notebook()
nb2.cells.append(nbformat.v4.new_markdown_cell("# Analysis v2"))
nb2.cells.append(nbformat.v4.new_code_cell("data = [1, 2, 3, 4, 5]"))
nb2.cells.append(nbformat.v4.new_code_cell("print(data)"))

print(f"\nNotebook 1: {len(nb1.cells)} cells")
print(f"Notebook 2: {len(nb2.cells)} cells")

# Perform diff
diff_result = nbdime.diff_notebooks(nb1, nb2)
print(f"\nDiff result: {type(diff_result)}")
print(f"Changes detected: {len(diff_result)} operations")

print("\nExample 1 completed successfully")
```

## Example 2: Notebook merging and conflict resolution

```python
import nbdime
import nbformat

print("Example 2: Notebook merging")

# Create base, local, and remote notebooks
base_nb = nbformat.v4.new_notebook()
base_nb.cells.append(nbformat.v4.new_markdown_cell("# Project"))
base_nb.cells.append(nbformat.v4.new_code_cell("x = 1"))

local_nb = nbformat.v4.new_notebook()
local_nb.cells.append(nbformat.v4.new_markdown_cell("# Enhanced Project"))
local_nb.cells.append(nbformat.v4.new_code_cell("x = 1\ny = 2"))

remote_nb = nbformat.v4.new_notebook()
remote_nb.cells.append(nbformat.v4.new_markdown_cell("# Project with Stats"))
remote_nb.cells.append(nbformat.v4.new_code_cell("x = 1\nz = 3"))

print(f"Base: {len(base_nb.cells)} cells")
print(f"Local: {len(local_nb.cells)} cells")
print(f"Remote: {len(remote_nb.cells)} cells")

# Perform merge
merged_nb, conflicts = nbdime.merge_notebooks(base_nb, local_nb, remote_nb)
print(f"\nMerged: {len(merged_nb.cells)} cells")
print(f"Conflicts: {len(conflicts)}")

print("\nExample 2 completed successfully")
```

## Example 3: Patch operations and applying changes

```python
import nbdime
import nbformat
import copy

print("Example 3: Patch operations")

# Create original notebook
original_nb = nbformat.v4.new_notebook()
original_nb.metadata.title = "Original"
original_nb.cells.append(nbformat.v4.new_code_cell("x = 1"))

# Create modified version
modified_nb = copy.deepcopy(original_nb)
modified_nb.metadata.title = "Modified"
modified_nb.cells.append(nbformat.v4.new_code_cell("y = 2"))

print(f"Original: {len(original_nb.cells)} cells")
print(f"Modified: {len(modified_nb.cells)} cells")

# Generate patch
patch = nbdime.diff_notebooks(original_nb, modified_nb)
print(f"\nPatch operations: {len(patch)}")

# Apply patch
patched_nb = nbdime.patch_notebook(original_nb, patch)
print(f"Patched: {len(patched_nb.cells)} cells")
print(f"Title match: {patched_nb.metadata.title == modified_nb.metadata.title}")

print("\nExample 3 completed successfully")
```

## Example 5: Git integration and version control workflows

```python
import nbdime
import nbformat

print("Example 5: Diff format and pretty printing")

# Create notebooks for comparison
nb1 = nbformat.v4.new_notebook()
nb1.cells.append(nbformat.v4.new_code_cell("import numpy as np\nx = np.array([1, 2, 3])"))

nb2 = nbformat.v4.new_notebook()
nb2.cells.append(nbformat.v4.new_code_cell("import numpy as np\nimport pandas as pd\nx = np.array([1, 2, 3, 4, 5])"))

print("Comparing two notebooks:")
print(f"  Notebook 1: {len(nb1.cells)} cells")
print(f"  Notebook 2: {len(nb2.cells)} cells")

# Generate diff
diff = nbdime.diff_notebooks(nb1, nb2)
print(f"\nDiff result: {len(diff)} operations")

# Check prettyprint module
if hasattr(nbdime, 'prettyprint'):
    print(f"Prettyprint module: Available")

# Check diff_format module
if hasattr(nbdime, 'diff_format'):
    print(f"Diff format module: Available")

print("\nExample 5 completed successfully")
```
