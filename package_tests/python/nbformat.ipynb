## Example 1: Creating and manipulating notebook structures

```python
# Example 1: Creating and manipulating notebook structures
import nbformat
import json

print("Example 1: Creating and Manipulating Notebook Structures")

try:
    print(f"nbformat version: {getattr(nbformat, '__version__', 'unknown')}")
    
    # Create a new notebook
    nb = nbformat.v4.new_notebook()
    print(f"Created new notebook: {type(nb)}")
    print(f"Format version: {nb.nbformat}.{nb.nbformat_minor}")
    
    # Add metadata
    nb.metadata.title = "Sample Data Analysis"
    nb.metadata.authors = [{"name": "Data Scientist"}]
    nb.metadata.kernelspec = {
        "display_name": "Python 3",
        "language": "python", 
        "name": "python3"
    }
    
    # Create different types of cells
    print(f"\nAdding various cell types:")
    
    # Markdown cell with formatting
    intro_md = nbformat.v4.new_markdown_cell("""
# Data Analysis Notebook

This notebook demonstrates:
- Data loading and preprocessing
- Statistical analysis
- Visualization techniques

**Dataset**: Sample customer data
""")
    nb.cells.append(intro_md)
    print(f"Added markdown cell")
    
    # Code cell with imports
    imports_code = nbformat.v4.new_code_cell("""
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load data
data = pd.DataFrame({
    'customer_id': range(1, 101),
    'age': np.random.randint(18, 80, 100),
    'spend': np.random.normal(500, 150, 100)
})
print(f"Data loaded: {data.shape}")
""")
    nb.cells.append(imports_code)
    print(f"Added code cell with imports")
    
    # Code cell with analysis
    analysis_code = nbformat.v4.new_code_cell("""
# Basic statistics
print("Customer Demographics:")
print(f"Average age: {data['age'].mean():.1f}")
print(f"Average spend: ${data['spend'].mean():.2f}")

# Age groups
data['age_group'] = pd.cut(data['age'], bins=[0, 25, 45, 65, 100], 
                          labels=['Young', 'Adult', 'Middle-age', 'Senior'])
print("\nSpend by age group:")
print(data.groupby('age_group')['spend'].mean())
""")
    nb.cells.append(analysis_code)
    print(f"Added analysis code cell")
    
    # Raw cell for notes
    raw_cell = nbformat.v4.new_raw_cell("""
Raw notes:
- Data collected from Q1 2024
- Sample size: 100 customers
- Currency: USD
""")
    nb.cells.append(raw_cell)
    print(f"Added raw cell")
    
    # Show notebook structure
    print(f"\nNotebook structure:")
    print(f"Total cells: {len(nb.cells)}")
    cell_types = [cell.cell_type for cell in nb.cells]
    for i, (cell, cell_type) in enumerate(zip(nb.cells, cell_types)):
        print(f"  Cell {i+1}: {cell_type} ({len(cell.source)} chars)")
    
    # Access cell content
    print(f"\nCell content examples:")
    print(f"First markdown cell title:")
    first_line = nb.cells[0].source.split('\n')[1]
    print(f"  {first_line}")
    
    print(f"Code cells summary:")
    code_cells = [cell for cell in nb.cells if cell.cell_type == 'code']
    for i, cell in enumerate(code_cells):
        lines = len(cell.source.split('\n'))
        print(f"  Code cell {i+1}: {lines} lines")
    
    # Demonstrate cell manipulation
    print(f"\nCell manipulation:")
    
    # Add execution count to code cells
    for i, cell in enumerate(nb.cells):
        if cell.cell_type == 'code':
            cell.execution_count = i + 1
            print(f"Set execution count {i+1} for code cell")
    
    # Add tags to cells
    nb.cells[1].metadata.tags = ['data-loading']
    nb.cells[2].metadata.tags = ['analysis', 'statistics']
    print(f"Added tags to cells")
    
    # Show metadata
    print(f"\nNotebook metadata:")
    print(f"Title: {nb.metadata.get('title', 'N/A')}")
    print(f"Kernel: {nb.metadata.kernelspec.get('display_name', 'N/A')}")
    print(f"Authors: {len(nb.metadata.get('authors', []))}")
    
    print("Creating and manipulating notebooks completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    print("Demonstrating notebook structure concepts:")
    concepts = [
        "Notebook creation: Create new notebooks programmatically",
        "Cell types: Support for markdown, code, and raw cells",
        "Metadata management: Notebook and cell-level metadata",
        "Cell manipulation: Add, modify, and organize notebook cells",
        "Structure access: Programmatic access to notebook components"
    ]
    
    for concept in concepts:
        print(f"  - {concept}")
    raise  # re-raise: a broken package must fail this example, not print past it
```

## Example 2: Reading, writing and validation

```python
# Example 2: Reading, writing and validation
import nbformat
import tempfile
import os

print("Example 2: Reading, Writing and Validation")

try:
    # Create a sample notebook
    nb = nbformat.v4.new_notebook()
    nb.cells.append(nbformat.v4.new_markdown_cell("# Test Notebook"))
    nb.cells.append(nbformat.v4.new_code_cell("print('Hello World')"))
    
    print(f"Created notebook with {len(nb.cells)} cells")
    
    # Write to string
    nb_string = nbformat.writes(nb)
    print(f"Notebook as string: {len(nb_string)} characters")
    
    # Read from string
    nb_from_string = nbformat.reads(nb_string, as_version=4)
    assert nb_from_string == nb, "writes/reads round trip changed the notebook"
    print(f"Read notebook: {len(nb_from_string.cells)} cells")
    
    # File operations
    with tempfile.NamedTemporaryFile(mode='w', suffix='.ipynb', delete=False) as f:
        nbformat.write(nb, f)
        temp_file = f.name
    
    print(f"Written to file: {os.path.basename(temp_file)}")
    
    # Read from file
    with open(temp_file, 'r') as f:
        nb_from_file = nbformat.read(f, as_version=4)
    assert nb_from_file == nb, "write/read round trip changed the notebook"
    
    print(f"Read from file: {len(nb_from_file.cells)} cells")
    
    # Validation (the notebook was just built with nbformat's own constructors: a ValidationError here fails the cell)
    nbformat.validate(nb)
    print("✓ Notebook validation passed")
    
    # Clean up
    os.unlink(temp_file)
    print("File operations completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    print("Demonstrating I/O concepts:")
    concepts = [
        "String I/O: Read/write notebooks as strings",
        "File I/O: Save and load notebook files",
        "Validation: Verify notebook format compliance",
        "Version handling: Support multiple notebook formats"
    ]
    for concept in concepts:
        print(f"  - {concept}")
    raise  # re-raise: a broken package must fail this example, not print past it
```

## Example 3: nbformat functionality test 3

```python
# Example 3: Version conversion and metadata handling
import nbformat

print("Example 3: Version Conversion and Metadata Handling")

try:
    # Create notebook with rich metadata
    nb = nbformat.v4.new_notebook()
    nb.metadata.update({
        'kernelspec': {'display_name': 'Python 3', 'language': 'python', 'name': 'python3'},
        'language_info': {'name': 'python', 'version': '3.8.0'}
    })
    
    # Add cell with outputs
    code_cell = nbformat.v4.new_code_cell("print('Test')")
    code_cell.outputs = [nbformat.v4.new_output('stream', name='stdout', text='Test\n')]
    nb.cells.append(code_cell)
    
    print(f"Version conversion capabilities:")
    print(f"Current format: {nb.nbformat}.{nb.nbformat_minor}")
    print(f"Available versions: {list(nbformat.versions.keys())}")
    
    assert nbformat.validate(nb) is None
    print("✓ Validation passed")
    
    print("Version handling completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    print("Demonstrating version concepts")
    raise AssertionError(
        f"nbformat example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: nbformat functionality test 4

```python
# Example 4: Cell outputs and execution results
import nbformat

print("Example 4: Cell Outputs and Execution Results")

try:
    # Create notebook with various output types
    nb = nbformat.v4.new_notebook()
    
    # Code cell with multiple outputs
    code_cell = nbformat.v4.new_code_cell("x = 5\nprint(x)\nx")
    
    # Add different output types
    code_cell.outputs = [
        nbformat.v4.new_output('stream', name='stdout', text='5\n'),
        nbformat.v4.new_output('execute_result', execution_count=1, data={'text/plain': '5'})
    ]
    code_cell.execution_count = 1
    nb.cells.append(code_cell)
    
    print(f"Created cell with {len(code_cell.outputs)} outputs")
    print(f"Output types: {[out.output_type for out in code_cell.outputs]}")
    
    # Show output structure
    for i, output in enumerate(code_cell.outputs):
        print(f"Output {i+1}: {output.output_type}")
        if hasattr(output, 'text'):
            print(f"  Text: {repr(output.text[:20])}")
        if hasattr(output, 'data'):
            print(f"  Data: {list(output.data.keys())}")
    
    print("Output handling completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    print("Demonstrating output concepts")
    raise AssertionError(
        f"nbformat example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: nbformat functionality test 5

```python
# Example 5: Notebook analysis and processing utilities
import nbformat

print("Example 5: Notebook Analysis and Processing Utilities")

try:
    # Create sample notebook for analysis
    nb = nbformat.v4.new_notebook()
    nb.cells.extend([
        nbformat.v4.new_markdown_cell("# Analysis"),
        nbformat.v4.new_code_cell("import pandas as pd"),
        nbformat.v4.new_code_cell("data = pd.read_csv('data.csv')"),
        nbformat.v4.new_markdown_cell("## Results"),
        nbformat.v4.new_code_cell("print(data.head())")
    ])
    
    # Analyze notebook structure
    print(f"Notebook analysis:")
    print(f"Total cells: {len(nb.cells)}")
    
    cell_counts = {}
    for cell in nb.cells:
        cell_type = cell.cell_type
        cell_counts[cell_type] = cell_counts.get(cell_type, 0) + 1
    
    for cell_type, count in cell_counts.items():
        print(f"  {cell_type} cells: {count}")
    
    # Find code cells with imports
    import_cells = []
    for i, cell in enumerate(nb.cells):
        if cell.cell_type == 'code' and 'import' in cell.source:
            import_cells.append(i)
    
    print(f"Cells with imports: {import_cells}")
    
    # Calculate complexity metrics
    total_lines = sum(len(cell.source.split('\n')) for cell in nb.cells if cell.cell_type == 'code')
    print(f"Total code lines: {total_lines}")
    
    print("Notebook analysis completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    print("Demonstrating analysis concepts")
    raise AssertionError(
        f"nbformat example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
