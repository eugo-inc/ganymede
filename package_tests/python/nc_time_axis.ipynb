## Example 3: Climate data visualization and long-term time series

```python
import nc_time_axis
import matplotlib.pyplot as plt
import numpy as np
import cftime
from datetime import timedelta

print("Example 3: Climate data visualization and long-term time series")

try:
    print("Creating climate data scenarios:")
    
    # Create long-term climate datasets
    print(f"\nGenerating multi-decade climate time series:")
    
    # Historical climate data (1980-2023)
    start_historical = cftime.datetime(1980, 1, 1, calendar='standard')
    historical_times = []
    current_time = start_historical
    
    # Generate monthly data points
    while current_time.year <= 2023:
        historical_times.append(current_time)
        # Add one month (approximate)
        if current_time.month == 12:
            current_time = cftime.datetime(current_time.year + 1, 1, 1, calendar='standard')
        else:
            current_time = cftime.datetime(current_time.year, current_time.month + 1, 1, calendar='standard')
    
    print(f"Historical data: {len(historical_times)} monthly points ({historical_times[0].year}-{historical_times[-1].year})")
    
    # Generate realistic temperature anomalies
    # Base trend + seasonal cycle + random variation
    n_points = len(historical_times)
    years = np.array([t.year for t in historical_times])
    months = np.array([t.month for t in historical_times])
    
    # Long-term warming trend
    base_trend = (years - 1980) * 0.02  # 0.2°C per decade
    
    # Seasonal cycle
    seasonal = 2.0 * np.sin(2 * np.pi * (months - 1) / 12)
    
    # Inter-annual variability
    interannual = np.random.normal(0, 0.8, n_points)
    
    # Climate oscillations (simplified ENSO-like pattern)
    enso_phase = np.sin(2 * np.pi * years / 3.5) * 0.5  # ~3.5 year cycle
    
    temperature_anomaly = base_trend + seasonal + interannual + enso_phase
    
    # Create comprehensive climate visualization
    print(f"\nCreating comprehensive climate visualization:")
    
    fig, axes = plt.subplots(3, 1, figsize=(16, 12))
    fig.suptitle('Long-term Climate Data Analysis with NetCDF Time Axis', fontsize=16, fontweight='bold')
    
    # Plot 1: Full time series
    ax1 = axes[0]
    ax1.plot(historical_times, temperature_anomaly, 'b-', linewidth=1, alpha=0.7, label='Monthly Anomaly')
    
    # Add 5-year running mean
    window = 60  # 5 years * 12 months
    if len(temperature_anomaly) >= window:
        running_mean = np.convolve(temperature_anomaly, np.ones(window)/window, mode='valid')
        running_times = historical_times[window//2:-window//2+1]
        ax1.plot(running_times, running_mean, 'r-', linewidth=3, label='5-year Running Mean')
    
    ax1.set_ylabel('Temperature Anomaly (°C)')
    ax1.set_title('Global Temperature Anomaly (1980-2023)')
    ax1.grid(True, alpha=0.3)
    ax1.legend()
    ax1.axhline(y=0, color='k', linestyle='--', alpha=0.5)
    
    # Plot 2: Decadal comparison
    ax2 = axes[1]
    
    decades = [(1980, 1989), (1990, 1999), (2000, 2009), (2010, 2019), (2020, 2023)]
    colors = ['blue', 'green', 'orange', 'red', 'purple']
    
    for i, ((start_year, end_year), color) in enumerate(zip(decades, colors)):
        # Filter data for this decade
        decade_mask = [(t.year >= start_year and t.year <= end_year) for t in historical_times]
        decade_times = [t for t, mask in zip(historical_times, decade_mask) if mask]
        decade_temps = temperature_anomaly[np.array(decade_mask)]
        
        if len(decade_times) > 0:
            ax2.scatter([t.year + (t.month-1)/12 for t in decade_times], decade_temps, 
                       c=color, alpha=0.6, s=8, label=f'{start_year}s')
    
    ax2.set_ylabel('Temperature Anomaly (°C)')
    ax2.set_title('Decadal Temperature Distribution')
    ax2.grid(True, alpha=0.3)
    ax2.legend()
    
    # Plot 3: Seasonal patterns by decade
    ax3 = axes[2]
    
    # Calculate seasonal averages for each decade
    seasonal_data = {}
    for start_year, end_year in [(1980, 1989), (2000, 2009), (2020, 2023)]:
        decade_mask = [(t.year >= start_year and t.year <= end_year) for t in historical_times]
        decade_months = [t.month for t, mask in zip(historical_times, decade_mask) if mask]
        decade_temps = temperature_anomaly[np.array(decade_mask)]
        
        monthly_means = []
        for month in range(1, 13):
            month_temps = [temp for m, temp in zip(decade_months, decade_temps) if m == month]
            monthly_means.append(np.mean(month_temps) if month_temps else 0)
        
        seasonal_data[f'{start_year}s'] = monthly_means
    
    months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
             'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
    
    for decade, temps in seasonal_data.items():
        ax3.plot(months, temps, marker='o', linewidth=2, label=decade)
    
    ax3.set_ylabel('Temperature Anomaly (°C)')
    ax3.set_title('Seasonal Temperature Patterns by Decade')
    ax3.grid(True, alpha=0.3)
    ax3.legend()
    plt.setp(ax3.get_xticklabels(), rotation=45)
    
    plt.tight_layout()
    plt.show()
    
    # Climate analysis insights
    print(f"\nClimate analysis insights:")
    
    insights = [
        f"Warming trend: {(temperature_anomaly[-60:].mean() - temperature_anomaly[:60].mean()):.2f}°C over 40+ years",
        f"Seasonal amplitude: {np.max(seasonal):.1f}°C peak-to-peak variation",
        f"Interannual variability: ±{np.std(interannual):.1f}°C standard deviation",
        f"Recent temperatures: {temperature_anomaly[-12:].mean():.2f}°C average anomaly (last year)",
        f"Maximum anomaly: {np.max(temperature_anomaly):.2f}°C in {historical_times[np.argmax(temperature_anomaly)].year}"
    ]
    
    for insight in insights:
        print(f"  - {insight}")
    
    # NetCDF time axis advantages for climate data
    print(f"\nNetCDF time axis advantages for climate data:")
    
    advantages = [
        "Calendar consistency: Handles leap years and calendar changes accurately",
        "Long-term stability: Maintains precision over multi-decade time spans",
        "Seasonal alignment: Proper monthly/seasonal boundary handling",
        "Timezone neutrality: UTC-based time coordinates for global datasets",
        "Metadata preservation: Retains time coordinate attributes and units"
    ]
    
    for advantage in advantages:
        print(f"  - {advantage}")
    
    print("\nClimate data visualization and long-term time series completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    print("Demonstrating climate visualization concepts:")
    concepts = [
        "Long-term trends: Multi-decade time series analysis",
        "Seasonal patterns: Annual cycle visualization and comparison",
        "Climate variability: Interannual and decadal variations"
    ]
    for concept in concepts:
        print(f"  - {concept}")
    raise

print("Example 3 finished")
```

## Example 4: Multi-calendar comparison and coordinate system integration

```python
import nc_time_axis
import matplotlib.pyplot as plt
import numpy as np
import cftime
from datetime import timedelta

print("Example 4: Multi-calendar comparison and coordinate system integration")

try:
    print("Exploring multi-calendar coordinate systems:")
    
    # Create equivalent time periods in different calendar systems
    print(f"\nCreating equivalent time series across calendar systems:")
    
    calendars = {
        'standard': {
            'description': 'Gregorian calendar with leap years',
            'start': cftime.datetime(2020, 1, 1, calendar='standard'),
            'color': 'blue'
        },
        '360_day': {
            'description': '12 months of exactly 30 days each',
            'start': cftime.datetime(2020, 1, 1, calendar='360_day'),
            'color': 'red'
        },
        'noleap': {
            'description': 'Gregorian calendar without leap years',
            'start': cftime.datetime(2020, 1, 1, calendar='noleap'),
            'color': 'green'
        },
        'julian': {
            'description': 'Julian calendar with leap years',
            'start': cftime.datetime(2020, 1, 1, calendar='julian'),
            'color': 'orange'
        }
    }
    
    # Generate time coordinates for each calendar
    calendar_data = {}
    for cal_name, cal_info in calendars.items():
        try:
            times = []
            start_time = cal_info['start']
            
            # Generate 3 years of monthly data
            current = start_time
            for i in range(36):  # 3 years * 12 months
                times.append(current)
                
                # Advance by one month
                if current.month == 12:
                    current = cftime.datetime(current.year + 1, 1, 1, calendar=cal_name)
                else:
                    current = cftime.datetime(current.year, current.month + 1, 1, calendar=cal_name)
            
            # Generate synthetic climate data
            months_since_start = [(t.year - start_time.year) * 12 + (t.month - start_time.month) for t in times]
            seasonal_cycle = np.sin(2 * np.pi * np.array(months_since_start) / 12) * 15 + 20
            trend = np.array(months_since_start) * 0.05
            noise = np.random.normal(0, 2, len(times))
            
            calendar_data[cal_name] = {
                'times': times,
                'data': seasonal_cycle + trend + noise,
                'description': cal_info['description'],
                'color': cal_info['color']
            }
            
            print(f"  {cal_name}: {len(times)} points, {times[0]} to {times[-1]}")
            
        except Exception as cal_error:
            print(f"  {cal_name}: Calendar not supported - {cal_error}")
            raise
    
    # Create comprehensive multi-calendar comparison
    print(f"\nCreating multi-calendar comparison visualization:")
    
    fig, axes = plt.subplots(2, 2, figsize=(16, 12))
    fig.suptitle('NetCDF Time Axis: Multi-Calendar Coordinate Systems', fontsize=16, fontweight='bold')
    
    # Individual calendar plots
    calendar_names = list(calendar_data.keys())
    for i, cal_name in enumerate(calendar_names[:4]):  # Show up to 4 calendars
        row, col = i // 2, i % 2
        ax = axes[row, col]
        
        data = calendar_data[cal_name]
        ax.plot(data['times'], data['data'], color=data['color'], 
                linewidth=2, marker='o', markersize=3, alpha=0.8)
        
        ax.set_title(f"{cal_name.title()} Calendar\n{data['description']}")
        ax.set_ylabel('Temperature (°C)')
        ax.grid(True, alpha=0.3)
        
        # Rotate x-axis labels
        plt.setp(ax.get_xticklabels(), rotation=45, ha='right')
        
        # Add calendar-specific annotations
        if cal_name == '360_day':
            ax.text(0.02, 0.98, 'Exactly 360 days/year', transform=ax.transAxes, 
                   verticalalignment='top', fontsize=10, bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
        elif cal_name == 'noleap':
            ax.text(0.02, 0.98, 'No leap years', transform=ax.transAxes,
                   verticalalignment='top', fontsize=10, bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.8))
        elif cal_name == 'standard':
            ax.text(0.02, 0.98, 'Modern Gregorian', transform=ax.transAxes,
                   verticalalignment='top', fontsize=10, bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.8))
        elif cal_name == 'julian':
            ax.text(0.02, 0.98, 'Historical Julian', transform=ax.transAxes,
                   verticalalignment='top', fontsize=10, bbox=dict(boxstyle='round', facecolor='orange', alpha=0.3))
    
    plt.tight_layout()
    plt.show()
    
    # Calendar comparison analysis
    print(f"\nCalendar comparison analysis:")
    
    # Compare year lengths
    print(f"\\nYear lengths in different calendars:")
    year_lengths = {
        'standard': '365/366 days (leap years)',
        '360_day': '360 days (constant)',
        'noleap': '365 days (constant)', 
        'julian': '365/366 days (different leap rule)'
    }
    
    for cal_name, length in year_lengths.items():
        if cal_name in calendar_data:
            print(f"  {cal_name}: {length}")
    
    # Calendar use cases
    print(f"\nCalendar system use cases:")
    
    use_cases = {
        'standard': [
            'Modern climate observations',
            'Real-time weather data',
            'Contemporary analysis and forecasting'
        ],
        '360_day': [
            'Simplified climate models',
            'Theoretical climate studies',
            'Educational climate demonstrations'
        ],
        'noleap': [
            'Idealized climate experiments',
            'Seasonal analysis without leap year complications',
            'Simplified long-term climate projections'
        ],
        'julian': [
            'Historical climate reconstructions',
            'Archaeological climate studies',
            'Pre-1582 historical analysis'
        ]
    }
    
    for cal_name, cases in use_cases.items():
        if cal_name in calendar_data:
            print(f"  {cal_name}:")
            for case in cases:
                print(f"    - {case}")
    
    # Coordinate system integration features
    print(f"\nCoordinate system integration features:")
    
    integration_features = [
        "Automatic axis registration: Seamless integration with matplotlib coordinate systems",
        "Multi-dimensional support: Compatible with 2D/3D coordinate arrays and grids",
        "Projection compatibility: Works with map projections and geographic coordinate systems",
        "Unit preservation: Maintains time unit metadata throughout coordinate transformations",
        "Precision handling: Maintains temporal precision across coordinate conversions"
    ]
    
    for feature in integration_features:
        print(f"  - {feature}")
    
    # Technical advantages
    print(f"\nTechnical advantages of nc_time_axis:")
    
    technical_advantages = [
        "Native cftime support: Direct integration with cftime datetime objects",
        "Matplotlib compatibility: Full compatibility with matplotlib plotting functions",
        "Performance optimization: Efficient handling of large time coordinate arrays",
        "Memory efficiency: Minimal memory overhead for time coordinate processing",
        "Thread safety: Safe for use in multi-threaded applications"
    ]
    
    for advantage in technical_advantages:
        print(f"  - {advantage}")
    
    print("\\nMulti-calendar comparison and coordinate system integration completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    print("Demonstrating multi-calendar concepts:")
    concepts = [
        "Calendar systems: Support for multiple calendar standards simultaneously",
        "Coordinate integration: Seamless integration with matplotlib coordinate systems",
        "Comparative analysis: Side-by-side analysis of different calendar approaches"
    ]
    for concept in concepts:
        print(f"  - {concept}")
    raise

print("Example 4 finished")
```

## Example 5: Advanced time axis customization and scientific applications

```python
import nc_time_axis
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import numpy as np
import cftime
from datetime import timedelta, datetime

print("Example 5: Advanced time axis customization and scientific applications")

try:
    print("Demonstrating advanced time axis customization:")
    
    # Create high-resolution scientific time series
    print(f"\nCreating high-resolution scientific datasets:")
    
    # Atmospheric measurement time series (hourly data)
    start_atmos = cftime.datetime(2023, 7, 1, 0, 0, 0, calendar='standard')
    atmos_times = [start_atmos + timedelta(hours=i) for i in range(168)]  # 1 week of hourly data
    
    # Simulate atmospheric CO2 measurements with diurnal cycle
    hours = np.array([t.hour for t in atmos_times])
    days = np.array([(t - start_atmos).total_seconds() / 86400 for t in atmos_times])
    
    # Base CO2 level with diurnal variation and weather patterns
    base_co2 = 420.0
    diurnal_cycle = -5.0 * np.cos(2 * np.pi * hours / 24)  # Daily CO2 cycle
    weather_pattern = 2.0 * np.sin(2 * np.pi * days / 3.5) + 1.0 * np.sin(2 * np.pi * days / 1.2)
    measurement_noise = np.random.normal(0, 0.5, len(atmos_times))
    
    co2_data = base_co2 + diurnal_cycle + weather_pattern + measurement_noise
    
    print(f"Atmospheric data: {len(atmos_times)} hourly measurements")
    print(f"Time span: {atmos_times[0]} to {atmos_times[-1]}")
    
    # Ocean temperature profile (6-hour intervals)
    start_ocean = cftime.datetime(2023, 8, 15, 0, 0, 0, calendar='standard')
    ocean_times = [start_ocean + timedelta(hours=i*6) for i in range(120)]  # 30 days of 6-hourly data
    
    # Simulate ocean temperature with tidal and thermal cycles
    hours_ocean = np.array([t.hour for t in ocean_times])
    days_ocean = np.array([(t - start_ocean).total_seconds() / 86400 for t in ocean_times])
    
    base_temp = 18.5
    thermal_cycle = 2.0 * np.sin(2 * np.pi * days_ocean) * np.exp(-days_ocean/15)  # Seasonal cooling
    tidal_influence = 0.5 * np.sin(2 * np.pi * hours_ocean / 12.42)  # ~12.4 hour tidal cycle
    thermal_inertia = np.cumsum(np.random.normal(0, 0.1, len(ocean_times))) * 0.02
    
    ocean_temp_data = base_temp + thermal_cycle + tidal_influence + thermal_inertia
    
    print(f"Ocean data: {len(ocean_times)} measurements every 6 hours")
    print(f"Time span: {ocean_times[0]} to {ocean_times[-1]}")
    
    # Create advanced visualization with custom formatting
    print(f"\nCreating advanced scientific visualization:")
    
    fig, axes = plt.subplots(3, 1, figsize=(16, 14))
    fig.suptitle('Advanced Time Axis Customization for Scientific Data', fontsize=16, fontweight='bold')
    
    # Plot 1: High-resolution atmospheric CO2 with custom time formatting
    ax1 = axes[0]
    ax1.plot(atmos_times, co2_data, 'g-', linewidth=1, alpha=0.8, label='Hourly CO2 Measurements')
    
    # Add daily averages
    daily_co2 = []
    daily_times = []
    for day in range(7):
        day_start = start_atmos + timedelta(days=day)
        day_end = start_atmos + timedelta(days=day+1)
        day_indices = [i for i, t in enumerate(atmos_times) if day_start <= t < day_end]
        if day_indices:
            daily_co2.append(np.mean(co2_data[day_indices]))
            daily_times.append(day_start + timedelta(hours=12))  # Midday
    
    ax1.plot(daily_times, daily_co2, 'ro-', linewidth=3, markersize=8, label='Daily Averages')
    
    ax1.set_ylabel('CO2 Concentration (ppm)')
    ax1.set_title('Atmospheric CO2 Measurements - Diurnal Cycle Analysis')
    ax1.grid(True, alpha=0.3)
    ax1.legend()
    
    # Custom time formatting for hourly data
    plt.setp(ax1.get_xticklabels(), rotation=30, ha='right')
    
    # Plot 2: Ocean temperature with tidal patterns
    ax2 = axes[1]
    ax2.plot(ocean_times, ocean_temp_data, 'b-', linewidth=1.5, alpha=0.8, label='6-hourly Ocean Temperature')
    
    # Add trend line
    z = np.polyfit(range(len(ocean_times)), ocean_temp_data, 1)
    trend_line = np.poly1d(z)(range(len(ocean_times)))
    ax2.plot(ocean_times, trend_line, 'r--', linewidth=2, label=f'Trend ({z[0]*24:.3f}°C/day)')
    
    ax2.set_ylabel('Temperature (°C)')
    ax2.set_title('Ocean Temperature Profile - Tidal and Thermal Cycles')
    ax2.grid(True, alpha=0.3)
    ax2.legend()
    
    plt.setp(ax2.get_xticklabels(), rotation=30, ha='right')
    
    # Plot 3: Combined frequency analysis
    ax3 = axes[2]
    
    # Create frequency spectrum visualization (simplified)
    # Show dominant periods in the time series
    periods = {
        'CO2 Diurnal': {'period': 24, 'amplitude': 5.0, 'color': 'green'},
        'Ocean Thermal': {'period': 24*30, 'amplitude': 2.0, 'color': 'blue'},
        'Tidal': {'period': 12.42, 'amplitude': 0.5, 'color': 'cyan'},
        'Weather': {'period': 24*3.5, 'amplitude': 2.0, 'color': 'orange'}
    }
    
    period_values = [info['period'] for info in periods.values()]
    amplitude_values = [info['amplitude'] for info in periods.values()]
    colors = [info['color'] for info in periods.values()]
    labels = list(periods.keys())
    
    bars = ax3.bar(range(len(periods)), amplitude_values, color=colors, alpha=0.7)
    ax3.set_ylabel('Signal Amplitude')
    ax3.set_xlabel('Dominant Periods')
    ax3.set_title('Spectral Analysis - Dominant Periodicities')
    ax3.set_xticks(range(len(periods)))
    ax3.set_xticklabels([f'{label}\\n({period:.1f}h)' for label, period in zip(labels, period_values)])
    ax3.grid(True, alpha=0.3, axis='y')
    
    # Add period annotations
    for i, (bar, period) in enumerate(zip(bars, period_values)):
        height = bar.get_height()
        ax3.text(bar.get_x() + bar.get_width()/2., height + 0.1,
                f'{period:.1f}h', ha='center', va='bottom', fontweight='bold')
    
    plt.tight_layout()
    plt.show()
    
    # Advanced customization features
    print(f"\nAdvanced time axis customization features:")
    
    customization_features = {
        "Precision control": "Sub-second precision for high-frequency measurements",
        "Custom tick spacing": "User-defined major/minor tick intervals",
        "Multi-scale axes": "Simultaneous display of different time scales",
        "Timezone handling": "Automatic conversion between time zones",
        "Format templates": "Scientific notation and custom date formats",
        "Annotation support": "Time-based annotations and event markers"
    }
    
    for feature, description in customization_features.items():
        print(f"  {feature}: {description}")
    
    # Scientific application domains
    print(f"\nScientific application domains:")
    
    application_domains = {
        "Climate science": [
            "Long-term climate trend analysis",
            "Paleoclimate reconstruction",
            "Climate model validation and comparison"
        ],
        "Atmospheric science": [
            "Air quality monitoring and analysis",
            "Greenhouse gas concentration tracking",
            "Weather pattern analysis and forecasting"
        ],
        "Oceanography": [
            "Sea surface temperature analysis",
            "Ocean current measurement and modeling",
            "Marine ecosystem monitoring"
        ],
        "Geophysics": [
            "Seismic data analysis and earthquake monitoring",
            "Magnetic field variation studies",
            "Gravitational wave data processing"
        ],
        "Environmental monitoring": [
            "Pollution level tracking and trend analysis",
            "Ecosystem health assessment",
            "Resource usage pattern analysis"
        ]
    }
    
    for domain, applications in application_domains.items():
        print(f"  {domain}:")
        for application in applications:
            print(f"    - {application}")
    
    # Performance and integration benefits
    print(f"\nPerformance and integration benefits:")
    
    benefits = [
        "Efficient rendering: Optimized for large time series datasets",
        "Memory management: Smart caching and lazy evaluation of time coordinates",
        "Interactive compatibility: Seamless integration with interactive matplotlib backends",
        "Export quality: High-quality output for scientific publications",
        "Cross-platform: Consistent behavior across operating systems and Python versions"
    ]
    
    for benefit in benefits:
        print(f"  - {benefit}")
    
    print("\\nAdvanced time axis customization and scientific applications completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    print("Demonstrating advanced customization concepts:")
    concepts = [
        "High-resolution data: Handling of sub-daily and high-frequency time series",
        "Scientific formatting: Specialized time formatting for scientific applications",
        "Multi-scale analysis: Simultaneous analysis of different temporal scales"
    ]
    for concept in concepts:
        print(f"  - {concept}")
    raise

print("Example 5 finished")
```
