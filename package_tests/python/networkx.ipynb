## Example 1: Basic graph creation and manipulation

```python
import networkx as nx
import numpy as np

print("Example 1: Basic graph creation and manipulation")

try:
    print(f"NetworkX version: {nx.__version__}")
    
    # Create different types of graphs
    print(f"\nCreating different graph types:")
    
    # Undirected graph
    G = nx.Graph()
    print(f"Undirected graph created: {type(G).__name__}")
    
    # Add nodes
    G.add_node(1)
    G.add_node("A")
    G.add_node("station_1", type="sensor", location="north")
    
    # Add multiple nodes
    G.add_nodes_from([2, 3, 4, "B", "C"])
    G.add_nodes_from([(5, {"color": "red"}), (6, {"color": "blue"})])
    
    print(f"Nodes added: {list(G.nodes())}")
    print(f"Total nodes: {G.number_of_nodes()}")
    
    # Add edges
    G.add_edge(1, 2)
    G.add_edge("A", "B")
    G.add_edge(1, "A", weight=0.5)
    
    # Add multiple edges
    G.add_edges_from([(2, 3), (3, 4), (4, 1)])
    G.add_edges_from([("B", "C", {"weight": 1.2}), (5, 6, {"weight": 2.1})])
    
    print(f"Edges added: {list(G.edges())}")
    print(f"Total edges: {G.number_of_edges()}")
    
    # Directed graph
    print(f"\nDirected graph example:")
    DG = nx.DiGraph()
    DG.add_edges_from([(1, 2), (2, 3), (3, 1), (2, 4)])
    
    print(f"Directed edges: {list(DG.edges())}")
    print(f"In-degrees: {dict(DG.in_degree())}")
    print(f"Out-degrees: {dict(DG.out_degree())}")
    
    # Multigraph (multiple edges between same nodes)
    print(f"\nMultiGraph example:")
    MG = nx.MultiGraph()
    MG.add_edge(1, 2, key="road")
    MG.add_edge(1, 2, key="rail")
    MG.add_edge(1, 2, key="air")
    
    print(f"Multiple edges between 1-2: {MG.number_of_edges(1, 2)}")
    print(f"Edge keys: {list(MG[1][2].keys())}")
    
    # Graph properties and information
    print(f"\nGraph properties:")
    print(f"  Nodes: {G.number_of_nodes()}")
    print(f"  Edges: {G.number_of_edges()}")
    print(f"  Density: {nx.density(G):.3f}")
    print(f"  Is connected: {nx.is_connected(G)}")
    
    # Node and edge attributes
    print(f"\nNode and edge attributes:")
    
    # Access node attributes
    print(f"  Node 'station_1' attributes: {G.nodes['station_1']}")
    print(f"  Node 5 attributes: {G.nodes[5]}")
    
    # Access edge attributes
    if G.has_edge(1, "A"):
        print(f"  Edge (1, 'A') attributes: {G.edges[1, 'A']}")
    
    # Modify attributes
    G.nodes[1]["importance"] = "high"
    G.edges[2, 3]["traffic"] = "heavy"
    
    print(f"  Updated node 1: {G.nodes[1]}")
    print(f"  Updated edge (2, 3): {G.edges[2, 3]}")
    
    # Graph operations
    print(f"\nBasic graph operations:")
    
    # Neighbors
    neighbors_1 = list(G.neighbors(1))
    print(f"  Neighbors of node 1: {neighbors_1}")
    
    # Degree
    degrees = dict(G.degree())
    print(f"  Node degrees: {degrees}")
    
    # Check connectivity
    print(f"  Has edge (1, 2): {G.has_edge(1, 2)}")
    print(f"  Has node 'Z': {G.has_node('Z')}")
    
    # Subgraphs
    subgraph_nodes = [1, 2, 3, 4]
    subgraph = G.subgraph(subgraph_nodes)
    print(f"  Subgraph nodes: {list(subgraph.nodes())}")
    print(f"  Subgraph edges: {list(subgraph.edges())}")
    
    # Graph copying
    G_copy = G.copy()
    print(f"  Original graph edges: {G.number_of_edges()}")
    print(f"  Copied graph edges: {G_copy.number_of_edges()}")
    
    # Remove elements
    G_copy.remove_node("C")
    G_copy.remove_edge(1, 2)
    
    print(f"  After removal - nodes: {G_copy.number_of_nodes()}, edges: {G_copy.number_of_edges()}")
    
    # Graph types comparison
    print(f"\nGraph types comparison:")
    
    graph_types = [
        ("Graph", nx.Graph(), "Undirected, simple edges"),
        ("DiGraph", nx.DiGraph(), "Directed, simple edges"),
        ("MultiGraph", nx.MultiGraph(), "Undirected, multiple edges allowed"),
        ("MultiDiGraph", nx.MultiDiGraph(), "Directed, multiple edges allowed")
    ]
    
    for name, graph_obj, description in graph_types:
        print(f"  {name}: {description}")
        print(f"    Directed: {graph_obj.is_directed()}")
        print(f"    Multigraph: {graph_obj.is_multigraph()}")
    
    print("\nBasic graph creation and manipulation completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    print("Demonstrating NetworkX concepts:")
    concepts = [
        "Graph creation: Different types of graphs (directed, undirected, multi)",
        "Node and edge management: Adding, removing, and accessing graph elements",
        "Attributes: Storing metadata on nodes and edges"
    ]
    for concept in concepts:
        print(f"  - {concept}")
    raise

print("Example 1 finished")
```

## Example 2: Graph algorithms - paths, centrality, and clustering

```python
import networkx as nx
import numpy as np

print("Example 2: Graph algorithms - paths, centrality, and clustering")

try:
    print("Creating sample network for algorithm demonstration:")
    
    # Create a more complex network for analysis
    G = nx.Graph()
    
    # Add nodes representing cities
    cities = {
        'NYC': {'population': 8400000, 'region': 'northeast'},
        'LA': {'population': 4000000, 'region': 'west'},
        'Chicago': {'population': 2700000, 'region': 'midwest'},
        'Houston': {'population': 2300000, 'region': 'south'},
        'Phoenix': {'population': 1700000, 'region': 'west'},
        'Philadelphia': {'population': 1600000, 'region': 'northeast'},
        'San Antonio': {'population': 1500000, 'region': 'south'},
        'San Diego': {'population': 1400000, 'region': 'west'},
        'Dallas': {'population': 1300000, 'region': 'south'},
        'San Jose': {'population': 1000000, 'region': 'west'}
    }
    
    for city, attrs in cities.items():
        G.add_node(city, **attrs)
    
    # Add edges representing connections (simplified distances)
    connections = [
        ('NYC', 'Philadelphia', {'distance': 95, 'type': 'highway'}),
        ('NYC', 'Chicago', {'distance': 790, 'type': 'highway'}),
        ('LA', 'San Diego', {'distance': 120, 'type': 'highway'}),
        ('LA', 'Phoenix', {'distance': 370, 'type': 'highway'}),
        ('LA', 'San Jose', {'distance': 350, 'type': 'highway'}),
        ('Chicago', 'Houston', {'distance': 1090, 'type': 'highway'}),
        ('Houston', 'Dallas', {'distance': 240, 'type': 'highway'}),
        ('Houston', 'San Antonio', {'distance': 200, 'type': 'highway'}),
        ('Phoenix', 'San Diego', {'distance': 360, 'type': 'highway'}),
        ('Dallas', 'San Antonio', {'distance': 275, 'type': 'highway'}),
        ('Philadelphia', 'Chicago', {'distance': 760, 'type': 'highway'}),
        ('San Jose', 'San Diego', {'distance': 500, 'type': 'highway'})
    ]
    
    G.add_edges_from([(u, v, attrs) for u, v, attrs in connections])
    
    print(f"Created network with {G.number_of_nodes()} cities and {G.number_of_edges()} connections")
    
    # Path algorithms
    print(f"\nPath algorithms:")
    
    # Shortest path
    try:
        shortest_path = nx.shortest_path(G, 'NYC', 'LA')
        path_length = nx.shortest_path_length(G, 'NYC', 'LA')
        print(f"  Shortest path NYC -> LA: {' -> '.join(shortest_path)}")
        print(f"  Path length (hops): {path_length}")
    except nx.NetworkXNoPath:
        print(f"  No path found between NYC and LA")
    
    # Weighted shortest path (using distance)
    try:
        weighted_path = nx.shortest_path(G, 'NYC', 'LA', weight='distance')
        weighted_length = nx.shortest_path_length(G, 'NYC', 'LA', weight='distance')
        print(f"  Weighted shortest path NYC -> LA: {' -> '.join(weighted_path)}")
        print(f"  Total distance: {weighted_length} miles")
    except nx.NetworkXNoPath:
        print(f"  No weighted path found")
    
    # All pairs shortest paths (sample)
    all_paths = dict(nx.all_pairs_shortest_path_length(G))
    print(f"  Distance from NYC to all cities:")
    for city, distance in sorted(all_paths['NYC'].items()):
        if city != 'NYC':
            print(f"    {city}: {distance} hops")
    
    # Centrality measures
    print(f"\nCentrality measures:")
    
    # Degree centrality
    degree_centrality = nx.degree_centrality(G)
    print(f"  Degree centrality (top 3):")
    sorted_degree = sorted(degree_centrality.items(), key=lambda x: x[1], reverse=True)
    for city, centrality in sorted_degree[:3]:
        print(f"    {city}: {centrality:.3f}")
    
    # Betweenness centrality
    betweenness_centrality = nx.betweenness_centrality(G)
    print(f"  Betweenness centrality (top 3):")
    sorted_betweenness = sorted(betweenness_centrality.items(), key=lambda x: x[1], reverse=True)
    for city, centrality in sorted_betweenness[:3]:
        print(f"    {city}: {centrality:.3f}")
    
    # Closeness centrality
    closeness_centrality = nx.closeness_centrality(G)
    print(f"  Closeness centrality (top 3):")
    sorted_closeness = sorted(closeness_centrality.items(), key=lambda x: x[1], reverse=True)
    for city, centrality in sorted_closeness[:3]:
        print(f"    {city}: {centrality:.3f}")
    
    # Eigenvector centrality
    try:
        eigenvector_centrality = nx.eigenvector_centrality(G, max_iter=1000)
        print(f"  Eigenvector centrality (top 3):")
        sorted_eigenvector = sorted(eigenvector_centrality.items(), key=lambda x: x[1], reverse=True)
        for city, centrality in sorted_eigenvector[:3]:
            print(f"    {city}: {centrality:.3f}")
    except:
        print(f"  Eigenvector centrality: Could not converge")
        raise
    
    # Clustering analysis
    print(f"\nClustering analysis:")
    
    # Clustering coefficient
    clustering_coeff = nx.clustering(G)
    avg_clustering = nx.average_clustering(G)
    print(f"  Average clustering coefficient: {avg_clustering:.3f}")
    print(f"  Individual clustering coefficients:")
    for city, coeff in sorted(clustering_coeff.items(), key=lambda x: x[1], reverse=True)[:5]:
        print(f"    {city}: {coeff:.3f}")
    
    # Triangles
    triangles = nx.triangles(G)
    total_triangles = sum(triangles.values()) // 3  # Each triangle counted 3 times
    print(f"  Total triangles in network: {total_triangles}")
    print(f"  Triangles per node (top 3):")
    sorted_triangles = sorted(triangles.items(), key=lambda x: x[1], reverse=True)
    for city, count in sorted_triangles[:3]:
        print(f"    {city}: {count}")
    
    # Connected components
    print(f"\nConnectivity analysis:")
    
    print(f"  Is connected: {nx.is_connected(G)}")
    print(f"  Number of connected components: {nx.number_connected_components(G)}")
    
    if nx.is_connected(G):
        diameter = nx.diameter(G)
        radius = nx.radius(G)
        print(f"  Network diameter: {diameter}")
        print(f"  Network radius: {radius}")
        
        # Center and periphery
        center = nx.center(G)
        periphery = nx.periphery(G)
        print(f"  Network center: {center}")
        print(f"  Network periphery: {periphery}")
    
    # Community detection (simple approach)
    print(f"\nCommunity detection:")
    
    # Find communities using simple greedy modularity optimization
    communities = nx.community.greedy_modularity_communities(G)
    print(f"  Number of communities found: {len(communities)}")
    
    for i, community in enumerate(communities):
        print(f"  Community {i+1}: {list(community)}")
    
    # Modularity
    modularity = nx.community.modularity(G, communities)
    print(f"  Modularity: {modularity:.3f}")
    
    # Advanced algorithms
    print(f"\nAdvanced algorithms:")
    
    # Maximum flow (if we had a flow network)
    # Convert to directed graph for flow example
    DG = G.to_directed()
    
    # Add capacity attributes
    for u, v, data in DG.edges(data=True):
        # Use inverse of distance as capacity (closer = higher capacity)
        if 'distance' in data:
            DG[u][v]['capacity'] = 1000 / data['distance']
        else:
            DG[u][v]['capacity'] = 1.0
    
    try:
        max_flow_value = nx.maximum_flow_value(DG, 'NYC', 'LA', capacity='capacity')
        print(f"  Maximum flow from NYC to LA: {max_flow_value:.2f}")
    except:
        print(f"  Maximum flow: Could not compute")
        raise
    
    # Minimum spanning tree
    mst = nx.minimum_spanning_tree(G, weight='distance')
    mst_weight = sum(data['distance'] for u, v, data in mst.edges(data=True))
    print(f"  Minimum spanning tree total weight: {mst_weight:.0f} miles")
    print(f"  MST edges: {mst.number_of_edges()}")
    
    # Graph metrics summary
    print(f"\nGraph metrics summary:")
    
    metrics = {
        'Nodes': G.number_of_nodes(),
        'Edges': G.number_of_edges(),
        'Density': nx.density(G),
        'Average degree': sum(dict(G.degree()).values()) / G.number_of_nodes(),
        'Average clustering': avg_clustering,
        'Diameter': nx.diameter(G) if nx.is_connected(G) else 'N/A'
    }
    
    for metric, value in metrics.items():
        if isinstance(value, float):
            print(f"  {metric}: {value:.3f}")
        else:
            print(f"  {metric}: {value}")
    
    print("\nGraph algorithms - paths, centrality, and clustering completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    print("Demonstrating graph algorithm concepts:")
    concepts = [
        "Path algorithms: Finding shortest paths and analyzing connectivity",
        "Centrality measures: Identifying important nodes in networks",
        "Clustering analysis: Understanding local network structure"
    ]
    for concept in concepts:
        print(f"  - {concept}")
    raise

print("Example 2 finished")
```

## Example 3: Graph generators and classic graph types

```python
import networkx as nx
import numpy as np

print("Example 3: Graph generators and classic graph types")

try:
    print("Demonstrating NetworkX graph generators:")
    
    # Classic graph types
    print(f"\nClassic graph types:")
    
    # Complete graph
    complete_graph = nx.complete_graph(6)
    print(f"  Complete graph K6: {complete_graph.number_of_nodes()} nodes, {complete_graph.number_of_edges()} edges")
    print(f"    Density: {nx.density(complete_graph):.3f}")
    
    # Path graph
    path_graph = nx.path_graph(8)
    print(f"  Path graph P8: {path_graph.number_of_nodes()} nodes, {path_graph.number_of_edges()} edges")
    print(f"    Diameter: {nx.diameter(path_graph)}")
    
    # Cycle graph
    cycle_graph = nx.cycle_graph(10)
    print(f"  Cycle graph C10: {cycle_graph.number_of_nodes()} nodes, {cycle_graph.number_of_edges()} edges")
    print(f"    Is bipartite: {nx.is_bipartite(cycle_graph)}")
    
    # Star graph
    star_graph = nx.star_graph(7)  # Central node + 7 leaf nodes
    print(f"  Star graph S7: {star_graph.number_of_nodes()} nodes, {star_graph.number_of_edges()} edges")
    center_node = [n for n, d in star_graph.degree() if d == star_graph.number_of_nodes() - 1][0]
    print(f"    Center node: {center_node} with degree {star_graph.degree(center_node)}")
    
    # Wheel graph
    wheel_graph = nx.wheel_graph(8)  # Hub + 7 nodes in a cycle
    print(f"  Wheel graph W8: {wheel_graph.number_of_nodes()} nodes, {wheel_graph.number_of_edges()} edges")
    
    # Regular graphs
    print(f"\nRegular graphs:")
    
    # k-regular graph (all nodes have degree k)
    petersen = nx.petersen_graph()
    print(f"  Petersen graph: {petersen.number_of_nodes()} nodes, {petersen.number_of_edges()} edges")
    degrees = list(dict(petersen.degree()).values())
    print(f"    All degrees: {set(degrees)} (3-regular)")
    print(f"    Diameter: {nx.diameter(petersen)}")
    
    # Random regular graph
    random_regular = nx.random_regular_graph(4, 12)  # 4-regular graph with 12 nodes
    print(f"  Random 4-regular graph: {random_regular.number_of_nodes()} nodes, {random_regular.number_of_edges()} edges")
    
    # Bipartite graphs
    print(f"\nBipartite graphs:")
    
    # Complete bipartite graph
    complete_bipartite = nx.complete_bipartite_graph(4, 6)
    print(f"  Complete bipartite K4,6: {complete_bipartite.number_of_nodes()} nodes, {complete_bipartite.number_of_edges()} edges")
    print(f"    Is bipartite: {nx.is_bipartite(complete_bipartite)}")
    
    if nx.is_bipartite(complete_bipartite):
        partite_sets = nx.bipartite.sets(complete_bipartite)
        print(f"    Partition sizes: {[len(s) for s in partite_sets]}")
    
    # Random bipartite graph
    random_bipartite = nx.bipartite.random_graph(8, 12, 0.3)
    print(f"  Random bipartite graph: {random_bipartite.number_of_nodes()} nodes, {random_bipartite.number_of_edges()} edges")
    
    # Random graph models
    print(f"\nRandom graph models:")
    
    # Erdős–Rényi random graph
    erdos_renyi = nx.erdos_renyi_graph(20, 0.2)
    print(f"  Erdős–Rényi G(20, 0.2): {erdos_renyi.number_of_nodes()} nodes, {erdos_renyi.number_of_edges()} edges")
    print(f"    Expected edges: {20 * 19 / 2 * 0.2:.1f}")
    print(f"    Is connected: {nx.is_connected(erdos_renyi)}")
    
    # Watts-Strogatz small-world graph
    small_world = nx.watts_strogatz_graph(30, 6, 0.3)
    print(f"  Watts-Strogatz small-world: {small_world.number_of_nodes()} nodes, {small_world.number_of_edges()} edges")
    print(f"    Clustering coefficient: {nx.average_clustering(small_world):.3f}")
    print(f"    Average shortest path: {nx.average_shortest_path_length(small_world):.2f}")
    
    # Barabási–Albert preferential attachment
    preferential = nx.barabasi_albert_graph(50, 3)
    print(f"  Barabási–Albert preferential: {preferential.number_of_nodes()} nodes, {preferential.number_of_edges()} edges")
    degrees = list(dict(preferential.degree()).values())
    print(f"    Degree range: {min(degrees)} to {max(degrees)}")
    print(f"    Average degree: {np.mean(degrees):.2f}")
    
    # Scale-free network properties
    print(f"\nScale-free network analysis:")
    
    # Degree distribution analysis
    degree_sequence = sorted([d for n, d in preferential.degree()], reverse=True)
    degree_counts = {}
    for degree in degree_sequence:
        degree_counts[degree] = degree_counts.get(degree, 0) + 1
    
    print(f"  Degree distribution (Barabási–Albert):")
    for degree in sorted(degree_counts.keys(), reverse=True)[:5]:
        print(f"    Degree {degree}: {degree_counts[degree]} nodes")
    
    # Power law fitting (simplified check)
    high_degree_nodes = [n for n, d in preferential.degree() if d >= np.percentile(degrees, 90)]
    print(f"  High-degree nodes (top 10%): {len(high_degree_nodes)}")
    
    # Lattice and grid graphs
    print(f"\nLattice and grid graphs:")
    
    # 2D grid graph
    grid_2d = nx.grid_2d_graph(5, 6)
    print(f"  2D Grid (5×6): {grid_2d.number_of_nodes()} nodes, {grid_2d.number_of_edges()} edges")
    print(f"    Diameter: {nx.diameter(grid_2d)}")
    
    # Triangular lattice
    triangular = nx.triangular_lattice_graph(4, 5)
    print(f"  Triangular lattice: {triangular.number_of_nodes()} nodes, {triangular.number_of_edges()} edges")
    
    # Hexagonal lattice
    hexagonal = nx.hexagonal_lattice_graph(3, 4)
    print(f"  Hexagonal lattice: {hexagonal.number_of_nodes()} nodes, {hexagonal.number_of_edges()} edges")
    
    # Tree structures
    print(f"\nTree structures:")
    
    # Random tree
    # `nx.random_tree` was removed in networkx 3.4; `random_labeled_tree` is its replacement.
    random_tree = nx.random_labeled_tree(15, seed=42)
    print(f"  Random tree: {random_tree.number_of_nodes()} nodes, {random_tree.number_of_edges()} edges")
    print(f"    Is tree: {nx.is_tree(random_tree)}")
    print(f"    Tree diameter: {nx.diameter(random_tree)}")
    
    # Balanced tree
    balanced_tree = nx.balanced_tree(3, 3)  # Branching factor 3, height 3
    print(f"  Balanced tree (r=3, h=3): {balanced_tree.number_of_nodes()} nodes, {balanced_tree.number_of_edges()} edges")
    
    # Social network models
    print(f"\nSocial network models:")
    
    # Karate club graph (famous social network)
    karate = nx.karate_club_graph()
    print(f"  Karate club network: {karate.number_of_nodes()} nodes, {karate.number_of_edges()} edges")
    print(f"    Clustering coefficient: {nx.average_clustering(karate):.3f}")
    print(f"    Transitivity: {nx.transitivity(karate):.3f}")
    
    # Community structure in karate club
    karate_communities = nx.community.greedy_modularity_communities(karate)
    print(f"    Communities detected: {len(karate_communities)}")
    
    # Configuration model (preserves degree sequence)
    degree_sequence_karate = [d for n, d in karate.degree()]
    config_model = nx.configuration_model(degree_sequence_karate)
    config_model = nx.Graph(config_model)  # Remove multi-edges and self-loops
    config_model.remove_edges_from(nx.selfloop_edges(config_model))
    
    print(f"  Configuration model: {config_model.number_of_nodes()} nodes, {config_model.number_of_edges()} edges")
    
    # Geometric and spatial graphs
    print(f"\nGeometric and spatial graphs:")
    
    # Random geometric graph
    geometric = nx.random_geometric_graph(25, 0.3)
    print(f"  Random geometric graph: {geometric.number_of_nodes()} nodes, {geometric.number_of_edges()} edges")
    print(f"    Is connected: {nx.is_connected(geometric)}")
    
    # Graph generator comparison
    print(f"\nGraph generator comparison:")
    
    generators = [
        ("Complete K10", nx.complete_graph(10)),
        ("Path P20", nx.path_graph(20)),
        ("Cycle C15", nx.cycle_graph(15)),
        ("Star S12", nx.star_graph(12)),
        ("ER G(20,0.2)", nx.erdos_renyi_graph(20, 0.2)),
        ("WS Small-world", nx.watts_strogatz_graph(20, 4, 0.3)),
        ("BA Scale-free", nx.barabasi_albert_graph(20, 2))
    ]
    
    print(f"  {'Graph Type':<15} {'Nodes':<6} {'Edges':<6} {'Density':<8} {'Clustering':<10}")
    print(f"  {'-' * 60}")
    
    for name, graph in generators:
        density = nx.density(graph)
        clustering = nx.average_clustering(graph) if nx.is_connected(graph) or len(graph) > 0 else 0
        print(f"  {name:<15} {graph.number_of_nodes():<6} {graph.number_of_edges():<6} {density:<8.3f} {clustering:<10.3f}")
    
    print("\nGraph generators and classic graph types completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    print("Demonstrating graph generator concepts:")
    concepts = [
        "Classic graphs: Complete, path, cycle, star, and wheel graphs",
        "Random models: Erdős–Rényi, Watts-Strogatz, and Barabási–Albert graphs", 
        "Structured graphs: Lattices, trees, and geometric graphs"
    ]
    for concept in concepts:
        print(f"  - {concept}")
    raise

print("Example 3 finished")
```

## Example 4: Graph I/O, visualization, and data interchange

```python
import networkx as nx
import tempfile
import os
import json

print("Example 4: Graph I/O, visualization, and data interchange")

try:
    print("Creating sample graph for I/O demonstration:")
    
    # Create a sample social network graph
    G = nx.Graph()
    
    # Add nodes with attributes
    people = {
        'Alice': {'age': 28, 'occupation': 'engineer', 'city': 'SF'},
        'Bob': {'age': 34, 'occupation': 'teacher', 'city': 'NYC'},
        'Charlie': {'age': 31, 'occupation': 'designer', 'city': 'LA'},
        'Diana': {'age': 29, 'occupation': 'scientist', 'city': 'SF'},
        'Eve': {'age': 26, 'occupation': 'artist', 'city': 'NYC'},
        'Frank': {'age': 33, 'occupation': 'engineer', 'city': 'LA'}
    }
    
    for person, attrs in people.items():
        G.add_node(person, **attrs)
    
    # Add edges with relationship attributes
    relationships = [
        ('Alice', 'Bob', {'relationship': 'colleagues', 'years_known': 3}),
        ('Alice', 'Diana', {'relationship': 'friends', 'years_known': 5}),
        ('Bob', 'Charlie', {'relationship': 'college_friends', 'years_known': 12}),
        ('Charlie', 'Eve', {'relationship': 'roommates', 'years_known': 2}),
        ('Diana', 'Frank', {'relationship': 'siblings', 'years_known': 25}),
        ('Eve', 'Frank', {'relationship': 'dating', 'years_known': 1}),
        ('Alice', 'Frank', {'relationship': 'friends', 'years_known': 4})
    ]
    
    for u, v, attrs in relationships:
        G.add_edge(u, v, **attrs)
    
    print(f"Sample social network: {G.number_of_nodes()} people, {G.number_of_edges()} relationships")
    
    # File format demonstrations
    print(f"\nGraph I/O formats:")
    
    with tempfile.TemporaryDirectory() as temp_dir:
        # GraphML format (XML-based, preserves attributes)
        graphml_file = os.path.join(temp_dir, 'social_network.graphml')
        nx.write_graphml(G, graphml_file)
        
        # Read back and verify
        G_graphml = nx.read_graphml(graphml_file)
        print(f"  GraphML: Saved and loaded {G_graphml.number_of_nodes()} nodes, {G_graphml.number_of_edges()} edges")
        print(f"    Attributes preserved: {list(G_graphml.nodes['Alice'].keys())}")
        
        # GML format (simple text format)
        gml_file = os.path.join(temp_dir, 'social_network.gml')
        nx.write_gml(G, gml_file)
        
        G_gml = nx.read_gml(gml_file)
        print(f"  GML: Saved and loaded {G_gml.number_of_nodes()} nodes, {G_gml.number_of_edges()} edges")
        
        # Pajek NET format
        pajek_file = os.path.join(temp_dir, 'social_network.net')
        nx.write_pajek(G, pajek_file)
        
        G_pajek = nx.read_pajek(pajek_file)
        print(f"  Pajek: Saved and loaded {G_pajek.number_of_nodes()} nodes, {G_pajek.number_of_edges()} edges")
        
        # Edge list format (simple, no attributes)
        edgelist_file = os.path.join(temp_dir, 'social_network.edgelist')
        nx.write_edgelist(G, edgelist_file)
        
        G_edgelist = nx.read_edgelist(edgelist_file)
        print(f"  Edge list: Saved and loaded {G_edgelist.number_of_nodes()} nodes, {G_edgelist.number_of_edges()} edges")
        
        # Adjacency list format
        adjlist_file = os.path.join(temp_dir, 'social_network.adjlist')
        nx.write_adjlist(G, adjlist_file)
        
        G_adjlist = nx.read_adjlist(adjlist_file)
        print(f"  Adjacency list: Saved and loaded {G_adjlist.number_of_nodes()} nodes, {G_adjlist.number_of_edges()} edges")
        
        # JSON format (node-link data)
        print(f"\nJSON serialization:")
        
        # Convert to JSON-serializable format
        node_link_data = nx.node_link_data(G)
        
        json_file = os.path.join(temp_dir, 'social_network.json')
        with open(json_file, 'w') as f:
            json.dump(node_link_data, f, indent=2)
        
        # Read back from JSON
        with open(json_file, 'r') as f:
            loaded_data = json.load(f)
        
        G_json = nx.node_link_graph(loaded_data)
        print(f"  JSON node-link: Saved and loaded {G_json.number_of_nodes()} nodes, {G_json.number_of_edges()} edges")
        # `nodes(data=True)` yields `(node, attrs)` pairs.
        sample_node, sample_attrs = next(iter(G_json.nodes(data=True)))
        print(f"    Sample node data: {sample_node}: {sample_attrs}")
        
        # Adjacency data format
        adj_data = nx.adjacency_data(G)
        G_adj_data = nx.adjacency_graph(adj_data)
        print(f"  JSON adjacency: {G_adj_data.number_of_nodes()} nodes, {G_adj_data.number_of_edges()} edges")
    
    # Matrix representations
    print(f"\nMatrix representations:")
    
    # Adjacency matrix
    adj_matrix = nx.adjacency_matrix(G)
    print(f"  Adjacency matrix shape: {adj_matrix.shape}")
    print(f"    Matrix density: {adj_matrix.nnz / (adj_matrix.shape[0] * adj_matrix.shape[1]):.3f}")
    
    # Convert back from adjacency matrix
    node_list = list(G.nodes())
    G_from_adj = nx.from_numpy_array(adj_matrix.toarray())
    G_from_adj = nx.relabel_nodes(G_from_adj, dict(enumerate(node_list)))
    print(f"    Reconstructed from matrix: {G_from_adj.number_of_nodes()} nodes, {G_from_adj.number_of_edges()} edges")
    
    # Incidence matrix (for bipartite representation)
    inc_matrix = nx.incidence_matrix(G)
    print(f"  Incidence matrix shape: {inc_matrix.shape}")
    
    # Laplacian matrix
    laplacian = nx.laplacian_matrix(G)
    print(f"  Laplacian matrix shape: {laplacian.shape}")
    eigenvalues = sorted([float(x) for x in laplacian.toarray().diagonal()])
    print(f"    Diagonal values range: {min(eigenvalues):.1f} to {max(eigenvalues):.1f}")
    
    # Data interchange with pandas
    print(f"\nPandas integration:")
    
    try:
        import pandas as pd
        
        # Convert edges to DataFrame
        edges_df = nx.to_pandas_edgelist(G)
        print(f"  Edges DataFrame shape: {edges_df.shape}")
        print(f"    Columns: {list(edges_df.columns)}")
        print(f"    Sample edge data:")
        print(f"      {edges_df.iloc[0].to_dict()}")
        
        # Convert nodes to DataFrame
        nodes_data = []
        for node, attrs in G.nodes(data=True):
            node_dict = {'node': node}
            node_dict.update(attrs)
            nodes_data.append(node_dict)
        
        nodes_df = pd.DataFrame(nodes_data)
        print(f"  Nodes DataFrame shape: {nodes_df.shape}")
        print(f"    Columns: {list(nodes_df.columns)}")
        
        # Reconstruct graph from pandas
        G_from_pandas = nx.from_pandas_edgelist(edges_df, 
                                               source='source', 
                                               target='target', 
                                               edge_attr=True)
        print(f"  Reconstructed from pandas: {G_from_pandas.number_of_nodes()} nodes, {G_from_pandas.number_of_edges()} edges")
        
    except ImportError:
        print("  Pandas not available for integration demo")
    
    # Visualization preparation (layout algorithms)
    print(f"\nLayout algorithms for visualization:")
    
    # Different layout algorithms
    layouts = {
        'spring': nx.spring_layout(G, k=1, iterations=50),
        'circular': nx.circular_layout(G),
        'shell': nx.shell_layout(G),
        'kamada_kawai': nx.kamada_kawai_layout(G),
        'planar': None  # Will try planar layout
    }
    
    # Try planar layout if graph is planar
    if nx.is_planar(G):
        layouts['planar'] = nx.planar_layout(G)
    else:
        del layouts['planar']
        print("  Graph is not planar, skipping planar layout")
    
    print(f"  Available layouts: {list(layouts.keys())}")
    
    # Layout properties
    for layout_name, pos in layouts.items():
        if pos:
            x_coords = [coord[0] for coord in pos.values()]
            y_coords = [coord[1] for coord in pos.values()]
            print(f"    {layout_name}: x-range [{min(x_coords):.2f}, {max(x_coords):.2f}], y-range [{min(y_coords):.2f}, {max(y_coords):.2f}]")
    
    # Graph properties for visualization
    print(f"\nVisualization-relevant properties:")
    
    # Node properties for styling
    degree_centrality = nx.degree_centrality(G)
    betweenness_centrality = nx.betweenness_centrality(G)
    
    print(f"  Degree centrality range: {min(degree_centrality.values()):.3f} to {max(degree_centrality.values()):.3f}")
    print(f"  Betweenness centrality range: {min(betweenness_centrality.values()):.3f} to {max(betweenness_centrality.values()):.3f}")
    
    # Edge properties for styling
    edge_betweenness = nx.edge_betweenness_centrality(G)
    print(f"  Edge betweenness range: {min(edge_betweenness.values()):.3f} to {max(edge_betweenness.values()):.3f}")
    
    # Export for external visualization tools
    print(f"\nExport for external tools:")
    
    export_formats = [
        "GraphML: Compatible with Gephi, yEd, Cytoscape",
        "GML: Compatible with graph analysis software",
        "Pajek NET: Compatible with Pajek network analysis",
        "JSON: Web-based visualization (D3.js, vis.js)",
        "CSV/TSV: Spreadsheet applications and R",
        "Matrix formats: MATLAB, NumPy, scientific computing"
    ]
    
    for format_desc in export_formats:
        print(f"  - {format_desc}")
    
    # Best practices for graph I/O
    print(f"\nBest practices for graph I/O:")
    
    best_practices = [
        "Choose format based on target application and attribute complexity",
        "Use GraphML for full attribute preservation and tool compatibility",
        "Use JSON for web applications and API integration",
        "Use edge lists for simple networks without attributes",
        "Validate data integrity after loading from external sources",
        "Consider compression for large networks",
        "Document attribute meanings and data provenance"
    ]
    
    for practice in best_practices:
        print(f"  - {practice}")
    
    print("\nGraph I/O, visualization, and data interchange completed successfully!")
    
except Exception as e:
    print(f"Error: {e}")
    print("Demonstrating graph I/O concepts:")
    concepts = [
        "File formats: GraphML, GML, Pajek, edge lists, and JSON serialization",
        "Matrix representations: Adjacency, incidence, and Laplacian matrices",
        "Visualization preparation: Layout algorithms and styling properties"
    ]
    for concept in concepts:
        print(f"  - {concept}")
    raise

print("Example 4 finished")
```

## Example 5: Network Analysis and Graph Properties

```python
import networkx as nx
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

# Create a social network for analysis
G = nx.karate_club_graph()
print(f"Karate Club Network: {G.number_of_nodes()} nodes, {G.number_of_edges()} edges")

# Basic network properties
print(f"Density: {nx.density(G):.3f}")
print(f"Average clustering: {nx.average_clustering(G):.3f}")
print(f"Transitivity: {nx.transitivity(G):.3f}")

# Degree distribution analysis
degrees = [G.degree(n) for n in G.nodes()]
degree_count = Counter(degrees)
print(f"Degree distribution: {dict(sorted(degree_count.items()))}")
print(f"Average degree: {np.mean(degrees):.2f}")

# Community detection using modularity
communities = nx.community.greedy_modularity_communities(G)
print(f"Communities found: {len(communities)}")
for i, community in enumerate(communities):
    print(f"  Community {i+1}: {len(community)} nodes")

# Network efficiency and robustness
print(f"Global efficiency: {nx.global_efficiency(G):.3f}")
print(f"Local efficiency: {nx.local_efficiency(G):.3f}")
print(f"Network diameter: {nx.diameter(G)}")
print(f"Average path length: {nx.average_shortest_path_length(G):.3f}")
```
