## Example 1: newversion functionality test 1

```python
import newversion

# Example 1: Basic version creation and manipulation
print("Example 1: Basic version creation and manipulation")

try:
    # Create version objects from strings
    versions = ['1.0.0', '2.1.3', '0.5.2', '10.20.30']
    
    print("Creating version objects:")
    for v_str in versions:
        v = newversion.Version(v_str)
        print(f"  '{v_str}' -> {v} (major={v.major}, minor={v.minor}, micro={v.micro})")
    
    # Test version comparison
    v1 = newversion.Version('1.2.3')
    v2 = newversion.Version('1.2.4')
    v3 = newversion.Version('2.0.0')
    
    print(f"\nVersion comparisons:")
    print(f"  {v1} < {v2}: {v1 < v2}")
    print(f"  {v2} < {v3}: {v2 < v3}")
    print(f"  {v1} == {newversion.Version('1.2.3')}: {v1 == newversion.Version('1.2.3')}")
    
    print("newversion example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"newversion example (cell 2) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: newversion functionality test 2

```python
import newversion

# Example 2: Version bumping operations
print("Example 2: Version bumping operations")

try:
    # Start with a base version
    base_version = newversion.Version('1.2.3')
    print(f"Base version: {base_version}")
    
    # Test all bump operations
    bumps = [
        ('major', base_version.bump_major()),
        ('minor', base_version.bump_minor()), 
        ('micro', base_version.bump_micro()),
        ('release', base_version.bump_release())
    ]
    
    print("\nBump operations:")
    for bump_type, bumped in bumps:
        print(f"  {bump_type:8}: {base_version} -> {bumped}")
    
    # Test prerelease versions
    v_pre = newversion.Version('1.0.0')
    # bump_prerelease takes the label as the keyword `release_type`; the first positional parameter is the increment `inc`.
    pre_alpha = v_pre.bump_prerelease(release_type='alpha')
    pre_beta = v_pre.bump_prerelease(release_type='beta')
    pre_rc = v_pre.bump_prerelease(release_type='rc')
    
    print(f"\nPrerelease versions:")
    print(f"  alpha: {v_pre} -> {pre_alpha}")  
    print(f"  beta:  {v_pre} -> {pre_beta}")
    print(f"  rc:    {v_pre} -> {pre_rc}")
    
    print("newversion example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"newversion example (cell 4) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: newversion functionality test 3

```python
import newversion

# Example 3: Version parsing and validation
print("Example 3: Version parsing and validation")

try:
    # Test various version formats
    test_versions = [
        '1.0.0',
        '2.1.0a1',      # alpha
        '1.5.0b2',      # beta  
        '3.0.0rc1',     # release candidate
        '1.0.0.post1',  # post release
        '2.0.0.dev3'    # development
    ]
    
    print("Parsing different version formats:")
    for v_str in test_versions:
        try:
            v = newversion.Version(v_str)
            status_info = []
            if v.is_prerelease:
                status_info.append("prerelease")
            if v.is_postrelease:
                status_info.append("postrelease") 
            if v.is_devrelease:
                status_info.append("dev")
            if v.is_stable:
                status_info.append("stable")
                
            status = ", ".join(status_info) if status_info else "normal"
            print(f"  {v_str:12} -> {v} ({status})")
        except Exception as e:
            print(f"  {v_str:12} -> Error: {e}")
            raise
    
    print("newversion example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"newversion example (cell 6) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```
