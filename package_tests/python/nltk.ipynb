## Example 1: Basic Text Tokenization and Processing

```python
import nltk
from nltk.tokenize import LineTokenizer, RegexpTokenizer, TreebankWordTokenizer, wordpunct_tokenize

# Example 1: Basic Text Tokenization and Processing
print("Demonstrating NLTK text tokenization and basic processing")

try:
    # These tokenizers need no downloaded data (word_tokenize and sent_tokenize need the punkt models)
    sample_text = """
    Natural Language Processing (NLP) is a fascinating field! 
    It combines computational linguistics with machine learning.
    NLTK is one of the most popular Python libraries for NLP tasks.
    Let's explore tokenization, parsing, and text analysis.
    """
    
    print("Original text:")
    print(sample_text.strip())
    
    # Line tokenization: one token per non-blank line
    lines = LineTokenizer().tokenize(sample_text)
    print(f"\nLines: {len(lines)}")
    assert len(lines) == 4, lines
    
    # Penn Treebank word tokenization splits off punctuation and contractions
    treebank = TreebankWordTokenizer()
    words = treebank.tokenize(sample_text)
    print(f"\nTreebank tokens ({len(words)}): {words[:12]}...")
    assert words[:6] == ['Natural', 'Language', 'Processing', '(', 'NLP', ')'], words[:6]
    assert treebank.tokenize("They'll save and invest more.") == ['They', "'ll", 'save', 'and', 'invest', 'more', '.']
    assert wordpunct_tokenize('Hello, world!') == ['Hello', ',', 'world', '!']
    
    # Word-only tokens and their frequency distribution
    word_tokens = RegexpTokenizer(r"\w+").tokenize(sample_text.lower())
    freq = nltk.FreqDist(word_tokens)
    
    print(f"\nText Statistics:")
    print(f"  Characters: {len(sample_text)}")
    print(f"  Words: {freq.N()}")
    print(f"  Unique words: {freq.B()}")
    print(f"  Lexical diversity: {freq.B() / freq.N():.2f}")
    assert (freq.N(), freq.B()) == (35, 33), (freq.N(), freq.B())
    
    print(f"\nMost common words:")
    for word, count in freq.most_common(5):
        print(f"  '{word}': {count}")
    assert freq['nlp'] == 2 and freq['is'] == 2, freq.most_common(5)
    
    print("\nExample 1: Basic text tokenization completed successfully")
    
except Exception as e:
    raise AssertionError(
        f"nltk example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 2: Text Preprocessing and Cleaning

```python
import nltk
import os
import re
import string
import tempfile

# Example 2: Text Preprocessing and Cleaning
print("Demonstrating text preprocessing and cleaning techniques")

try:
    from nltk.corpus import stopwords
    from nltk.stem import PorterStemmer, SnowballStemmer
    from nltk.tokenize import RegexpTokenizer
    
    # Sample messy text for cleaning
    messy_text = """
    Hey there!!! This is a SAMPLE text with lots of 
    unnecessary punctuation???... and UPPERCASE letters.
    We need to clean this up: remove stopwords, normalize case,
    and maybe do some stemming. #NLP @python 123 testing!
    """
    
    print("Original messy text:")
    print(messy_text)
    
    # Step 1: Basic cleaning
    # Convert to lowercase
    text_lower = messy_text.lower()
    print(f"\n1. Lowercase conversion:")
    print(text_lower)
    
    # Remove punctuation and special characters
    text_no_punct = re.sub(r'[^\w\s]', ' ', text_lower)
    text_no_punct = re.sub(r'\s+', ' ', text_no_punct).strip()
    print(f"\n2. Punctuation removed:")
    print(text_no_punct)
    
    # Remove numbers (optional)
    text_no_nums = re.sub(r'\d+', '', text_no_punct)
    text_no_nums = re.sub(r'\s+', ' ', text_no_nums).strip()
    print(f"\n3. Numbers removed:")
    print(text_no_nums)
    
    # Tokenization
    tokens = RegexpTokenizer(r'\w+').tokenize(text_no_nums)
    print(f"\n4. Tokenized ({len(tokens)} tokens):")
    print(tokens)
    
    basic_stop_words = {
        'i', 'me', 'my', 'myself', 'we', 'our', 'ours', 'ourselves', 'you', 'your', 'yours',
        'yourself', 'yourselves', 'he', 'him', 'his', 'himself', 'she', 'her', 'hers',
        'herself', 'it', 'its', 'itself', 'they', 'them', 'their', 'theirs', 'themselves',
        'what', 'which', 'who', 'whom', 'this', 'that', 'these', 'those', 'am', 'is', 'are',
        'was', 'were', 'be', 'been', 'being', 'have', 'has', 'had', 'having', 'do', 'does',
        'did', 'doing', 'a', 'an', 'the', 'and', 'but', 'if', 'or', 'because', 'as', 'until',
        'while', 'of', 'at', 'by', 'for', 'with', 'through', 'during', 'before', 'after',
        'above', 'below', 'up', 'down', 'in', 'out', 'on', 'off', 'over', 'under', 'again',
        'further', 'then', 'once', 'to'
    }
    # A small local stopwords corpus, so this cell needs no download.
    # nltk ignores data paths that other accounts can write to, so keep them private.
    nltk_data_dir = tempfile.mkdtemp(prefix='nltk_data_')
    stopwords_dir = os.path.join(nltk_data_dir, 'corpora', 'stopwords')
    os.makedirs(stopwords_dir, mode=0o700)
    os.chmod(os.path.join(nltk_data_dir, 'corpora'), 0o700)
    stopwords_file = os.path.join(stopwords_dir, 'english')
    with open(stopwords_file, 'w', encoding='utf-8') as f:
        f.write('\n'.join(sorted(basic_stop_words)) + '\n')
    os.chmod(stopwords_file, 0o600)
    nltk.data.path.insert(0, nltk_data_dir)
    stop_words = set(stopwords.words('english'))
    assert stop_words == basic_stop_words, stop_words ^ basic_stop_words
    print(f"Using NLTK stopwords: {len(stop_words)} words")
    
    filtered_tokens = [word for word in tokens if word not in stop_words and len(word) > 2]
    print(f"\n5. Stopwords removed ({len(filtered_tokens)} tokens):")
    print(filtered_tokens)
    assert filtered_tokens == [
        'hey', 'there', 'sample', 'text', 'lots', 'unnecessary', 'punctuation', 'uppercase', 'letters', 'need',
        'clean', 'remove', 'stopwords', 'normalize', 'case', 'maybe', 'some', 'stemming', 'nlp', 'python', 'testing',
    ], filtered_tokens
    
    # Stemming
    stemmer = PorterStemmer()
    stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]
    print(f"\n6. Porter stemming:")
    print(stemmed_tokens)
    assert [stemmer.stem(w) for w in ('running', 'letters', 'testing', 'unnecessary')] == ['run', 'letter', 'test', 'unnecessari']
    assert SnowballStemmer('english').stem('generously') == 'generous'
    
    # Show stemming examples
    print(f"\nStemming examples:")
    for orig, stem in zip(filtered_tokens[:5], stemmed_tokens[:5]):
        if orig != stem:
            print(f"  '{orig}' -> '{stem}'")
    
    # Final processed text
    processed_text = ' '.join(stemmed_tokens)
    print(f"\n7. Final processed text:")
    print(processed_text)
    
    # Comparison statistics
    print(f"\nProcessing Statistics:")
    print(f"  Original length: {len(messy_text)} characters")
    print(f"  Final length: {len(processed_text)} characters")
    print(f"  Original tokens: {len(messy_text.split())}")
    print(f"  Final tokens: {len(stemmed_tokens)}")
    print(f"  Reduction: {(1 - len(stemmed_tokens)/len(messy_text.split()))*100:.1f}%")
    
    print("\nExample 2: Text preprocessing and cleaning completed successfully")
    
except Exception as e:
    raise AssertionError(
        f"nltk example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: Part-of-Speech Tagging and Named Entity Recognition

```python
import nltk
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
# ne_chunk needs the named-entity chunker tables and the English word list;
# its classifier also imports numpy (the nltk[machine-learning] extra)
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')

# Example 3: Part-of-Speech Tagging and Named Entity Recognition
print("Demonstrating POS tagging and named entity recognition")

try:
    from collections import Counter

    from nltk import ne_chunk, pos_tag
    from nltk.tokenize import sent_tokenize, word_tokenize
    
    # Sample text with entities
    sample_text = """
    Apple Inc. is headquartered in Cupertino, California.
    The company was founded by Steve Jobs, Steve Wozniak, and Ronald Wayne in 1976.
    Tim Cook is the current CEO of Apple.
    They develop the iPhone, iPad, and Mac computers.
    """
    
    print("Sample text:")
    print(sample_text.strip())
    
    # Sentence and word tokenization with the punkt models
    sentences = sent_tokenize(sample_text)
    print(f"\nSentences: {len(sentences)}")
    assert len(sentences) == 4, sentences
    tokens = word_tokenize(sample_text)
    print(f"Tokens: {len(tokens)} words")
    print(tokens[:15], "..." if len(tokens) > 15 else "")
    assert tokens[:3] == ['Apple', 'Inc.', 'is'], tokens[:3]
    
    # POS tagging with the averaged perceptron tagger
    pos_tags = pos_tag(tokens)
    print(f"\nPOS Tags (first 10):")
    for word, tag in pos_tags[:10]:
        print(f"  {word:12} -> {tag}")
    tag_counts = Counter(tag for word, tag in pos_tags)
    print(f"\nPOS tag distribution:")
    for tag, count in tag_counts.most_common(5):
        print(f"  {tag}: {count}")
    tags = dict(pos_tags)
    assert (tags['Cupertino'], tags['founded'], tags['1976']) == ('NNP', 'VBN', 'CD'), tags
    
    # Named-entity chunking on the tagged tokens
    chunks = ne_chunk(pos_tags)
    named_entities = [(' '.join(token for token, pos in chunk.leaves()), chunk.label())
                      for chunk in chunks if hasattr(chunk, 'label')]
    print(f"\nNLTK Named Entity Chunks:")
    for entity, label in named_entities:
        print(f"  {entity} -> {label}")
    entity_labels = dict(named_entities)
    assert entity_labels.get('Tim Cook') == 'PERSON', named_entities
    assert entity_labels.get('Cupertino') == 'GPE', named_entities
    
    print("\nExample 3: POS tagging and NER completed successfully")
    
except Exception as e:
    raise AssertionError(
        f"nltk example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: Text Classification and Sentiment Analysis

```python
import nltk

nltk.download('vader_lexicon')

# Example 4: Text Classification and Sentiment Analysis
print("Demonstrating text classification and sentiment analysis")

try:
    from nltk.classify import NaiveBayesClassifier, accuracy
    from nltk.sentiment import SentimentIntensityAnalyzer
    from nltk.tokenize import RegexpTokenizer
    
    # Sample texts for sentiment analysis
    sample_texts = [
        "I love this new smartphone! The camera quality is amazing.",
        "This movie was terrible. I wasted my time watching it.",
        "The weather today is okay, nothing special.",
        "Fantastic product! Highly recommended to everyone.",
        "Poor customer service. Very disappointing experience."
    ]
    
    print("Sample texts for sentiment analysis:")
    for i, text in enumerate(sample_texts, 1):
        print(f"  {i}. {text}")
    
    # NLTK's VADER sentiment analyzer
    sia = SentimentIntensityAnalyzer()
    print(f"\nVADER Sentiment Analysis:")
    compound = []
    for i, text in enumerate(sample_texts, 1):
        scores = sia.polarity_scores(text)
        compound.append(scores['compound'])
        sentiment = "Positive" if scores['compound'] > 0.1 else "Negative" if scores['compound'] < -0.1 else "Neutral"
        print(f"  Text {i}: {sentiment} (compound: {scores['compound']:.2f})")
        print(f"    Details: pos={scores['pos']:.2f}, neu={scores['neu']:.2f}, neg={scores['neg']:.2f}")
        assert abs(scores['pos'] + scores['neu'] + scores['neg'] - 1) < 0.01, scores
    # The two clear recommendations score positive, the two complaints negative
    assert compound[0] > 0.5 and compound[3] > 0.5, compound
    assert compound[1] < -0.3 and compound[4] < -0.3, compound
    
    # Naive Bayes classifier on bag-of-words features (needs no downloaded data)
    tokenizer = RegexpTokenizer(r"\w+")
    
    def features(text):
        return {word: True for word in tokenizer.tokenize(text.lower())}
    
    training = [(features(text), label) for text, label in [
        ("I love this phone, the camera is amazing", "pos"),
        ("Fantastic product, highly recommended", "pos"),
        ("Great service and a wonderful experience", "pos"),
        ("This movie was terrible and boring", "neg"),
        ("Poor quality, very disappointing", "neg"),
        ("Awful support, I wasted my time", "neg"),
    ]]
    classifier = NaiveBayesClassifier.train(training)
    print(f"\nNaive Bayes training accuracy: {accuracy(classifier, training):.2f}")
    classifier.show_most_informative_features(5)
    assert accuracy(classifier, training) == 1.0
    assert classifier.classify(features("amazing camera, I love it")) == "pos"
    assert classifier.classify(features("terrible and disappointing")) == "neg"
    
    print("\nExample 4: Text classification and sentiment analysis completed successfully")
    
except Exception as e:
    raise AssertionError(
        f"nltk example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 5: Text Similarity and Language Processing

```python
import math

import nltk

# Example 5: Text Similarity and Language Processing
print("Demonstrating text similarity and advanced language processing")

try:
    from nltk.metrics import edit_distance, jaccard_distance
    from nltk.text import TextCollection
    from nltk.tokenize import RegexpTokenizer
    from nltk.util import ngrams
    
    # Sample documents for similarity analysis
    documents = [
        "Natural language processing is a field of artificial intelligence.",
        "NLP combines computational linguistics with machine learning techniques.",
        "Text mining and information retrieval are related to NLP.",
        "Machine learning algorithms are used in natural language understanding.",
        "Artificial intelligence includes computer vision and natural language processing."
    ]
    
    print("Sample documents:")
    for i, doc in enumerate(documents, 1):
        print(f"  {i}. {doc}")
    
    tokenizer = RegexpTokenizer(r"\w+")
    processed_docs = [tokenizer.tokenize(doc.lower()) for doc in documents]
    
    # Jaccard similarity (1 - nltk's Jaccard distance) between the word sets
    print(f"\nJaccard Similarity Analysis:")
    for i in range(len(documents)):
        for j in range(i + 1, len(documents)):
            similarity = 1 - jaccard_distance(set(processed_docs[i]), set(processed_docs[j]))
            if similarity > 0.1:  # Only show significant similarities
                print(f"  Doc {i+1} vs Doc {j+1}: {similarity:.3f}")
    assert jaccard_distance({'a', 'b', 'c'}, {'a', 'b', 'd'}) == 0.5
    
    # N-gram analysis
    print(f"\nN-gram Analysis:")
    doc1_bigrams = list(ngrams(processed_docs[0], 2))
    print(f"Bigrams from document 1:")
    for bigram in doc1_bigrams[:5]:
        print(f"  {' '.join(bigram)}")
    assert doc1_bigrams[:2] == [('natural', 'language'), ('language', 'processing')], doc1_bigrams[:2]
    assert len(doc1_bigrams) == len(processed_docs[0]) - 1
    
    # TF-IDF with nltk's TextCollection
    print(f"\nTF-IDF (TextCollection):")
    collection = TextCollection(processed_docs)
    for word in ('natural', 'learning', 'nlp'):
        scores = [collection.tf_idf(word, doc) for doc in processed_docs]
        print(f"  {word!r}: idf={collection.idf(word):.3f}, tf-idf per doc={[round(s, 3) for s in scores]}")
    # 'natural' appears in 3 of the 5 documents; document 1 has 9 words
    assert math.isclose(collection.idf('natural'), math.log(5 / 3)), collection.idf('natural')
    assert math.isclose(collection.tf('natural', processed_docs[0]), 1 / 9)
    
    # Edit distance for string similarity
    print(f"\nEdit Distance Analysis:")
    sample_words = ["natural", "language", "processing", "artificial", "intelligence"]
    variations = ["natura", "languag", "procesing", "artifical", "inteligence"]
    print("Word similarity using edit distance:")
    distances = []
    for orig, var in zip(sample_words, variations):
        distance = edit_distance(orig, var)
        distances.append(distance)
        similarity = 1 - (distance / max(len(orig), len(var)))
        print(f"  '{orig}' vs '{var}': distance={distance}, similarity={similarity:.3f}")
    assert distances == [1, 1, 1, 1, 1], distances
    assert edit_distance('kitten', 'sitting') == 3
    
    print("\nExample 5: Text similarity and language processing completed successfully")
    
except Exception as e:
    raise AssertionError(
        f"nltk example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
```
