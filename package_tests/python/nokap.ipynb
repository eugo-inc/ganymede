## Example 1: The API and its errors

```python
import importlib.metadata

import nokap

# Example 1: webshot / from_html drive Chrome; every failure is a NokapError subclass
assert importlib.metadata.version("nokap") == "0.1.0"
assert all(callable(f) for f in (nokap.webshot, nokap.from_html, nokap.close, nokap.find_chrome))
assert issubclass(nokap.ChromeNotFoundError, nokap.NokapError) and issubclass(nokap.CDPError, nokap.NokapError)
print("Example 1 OK:", sorted(nokap.__all__))
```

```python
import os

import nokap

os.environ.pop("CHROME_PATH", None)
try:
    path = nokap.find_chrome()
    print("Example 2: a browser is installed here:", path)
except nokap.ChromeNotFoundError as exc:
    assert "CHROME_PATH" in str(exc), exc
    print("Example 2 OK:", exc)
```

## Example 3: CHROME_PATH must point at a real program

```python
import os

import nokap

# Example 3: a CHROME_PATH that does not exist is not used
os.environ["CHROME_PATH"] = "/nonexistent/chrome"
try:
    nokap.find_chrome()
except nokap.ChromeNotFoundError as exc:
    print("Example 3 OK:", exc)
else:
    print("Example 3: find_chrome accepted the path (it does not check it)")
finally:
    os.environ.pop("CHROME_PATH", None)
```

## Example 4: The command line

```python
import shutil
import subprocess

# Example 4: the nokap console script (click) lists its commands without needing a browser
exe = shutil.which("nokap")
assert exe, "nokap is not on PATH"
r = subprocess.run([exe, "--help"], capture_output=True, text=True)
assert r.returncode == 0 and "Usage" in r.stdout, (r.stdout, r.stderr)
print("Example 4 OK:\n" + r.stdout[:400])
```

## Example 5: Paper sizes for PDF output

```python
import typing

import nokap

# Example 5: PaperSize names the page sizes a PDF export takes (webshot(..., page_size=...))
sizes = typing.get_args(nokap.PaperSize)
assert "letter" in sizes and "a4" in sizes, sizes
assert "page_size" in nokap.webshot.__code__.co_varnames
print("Example 5 OK:", sizes)
```
