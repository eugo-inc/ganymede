## Example 1: Notebook Module and Application

```python
import notebook
from notebook.app import JupyterNotebookApp

print("Example 1: Notebook Module and Application")

# Show notebook version
print(f"\nNotebook version: {notebook.__version__}")
print(f"Version info: {notebook.version_info}")
assert notebook.__version__.split(".")[0] == str(notebook.version_info[0]), notebook.__version__

# Show main application class
print(f"\nJupyterNotebookApp: {JupyterNotebookApp}")

# Show app attributes
app_attrs = [x for x in dir(JupyterNotebookApp) if not x.startswith('_')]
print(f"Total app attributes: {len(app_attrs)}")
print(f"Sample attributes: {app_attrs[:10]}")

print("\nExample 1 completed successfully")
```

## Example 2: Application Configuration

```python
from notebook.app import JupyterNotebookApp

print("Example 2: Application Configuration")

# Create app instance
app = JupyterNotebookApp()

# Show configuration attributes
print(f"\nApp configuration:")
print(f"  App name: {app.app_name}")
print(f"  App version: {app.app_version}")
print(f"  App URL: {app.app_url}")
assert app.app_name == "Jupyter Notebook", app.app_name

# Show configurable traits
print(f"\nConfigurable attributes:")
config_attrs = ['port', 'browser', 'notebook_dir', 'base_url']
for attr in config_attrs:
    if hasattr(app, attr):
        print(f"  {attr}: {getattr(app, attr)}")

print("\nExample 2 completed successfully")
```

## Example 4: Handlers and Extensions

```python
from notebook import app

print("Example 4: Handlers and Extensions")

# Show handler classes
print(f"\nHandler classes:")
print(f"  NotebookBaseHandler: {app.NotebookBaseHandler}")
print(f"  NotebookHandler: {app.NotebookHandler}")
print(f"  TreeHandler: {app.TreeHandler}")
print(f"  TerminalHandler: {app.TerminalHandler}")
for handler in (app.NotebookHandler, app.TreeHandler):
    assert issubclass(handler, app.NotebookBaseHandler), handler

# Show extension mixins
print(f"\nExtension components:")
print(f"  ExtensionHandlerMixin: {app.ExtensionHandlerMixin}")
print(f"  ExtensionHandlerJinjaMixin: {app.ExtensionHandlerJinjaMixin}")

print("\nExample 4 completed successfully")
```

## Example 5: Configuration and Flags

```python
from notebook import app

print("Example 5: Configuration and Flags")

# Show aliases
print(f"\nCommand-line aliases:")
aliases = app.aliases
print(f"  Total aliases: {len(aliases)}")
print(f"  Sample aliases: {list(aliases.keys())[:5]}")

# Show flags
print(f"\nCommand-line flags:")
flags = app.flags
print(f"  Total flags: {len(flags)}")
print(f"  Sample flags: {list(flags.keys())[:5]}")

# Show URL utilities
print(f"\nURL utilities:")
print(f"  url_escape: {app.url_escape}")
print(f"  is_url: {app.is_url}")
assert app.url_escape("my notebook.ipynb") == "my%20notebook.ipynb"
assert app.is_url("https://example.com/") and not app.is_url("not a url")

print("\nExample 5 completed successfully")
```
