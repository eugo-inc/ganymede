## Example 1: Available OAuth Authenticators

```python
import oauthenticator

print("Example 1: Available OAuth Authenticators")

# Show version
print(f"Version: {oauthenticator.__version__}")

# Show main authenticator classes
print(f"\nMain authenticators:")
print(f"  OAuthenticator: {oauthenticator.OAuthenticator}")
print(f"  GitHubOAuthenticator: {oauthenticator.GitHubOAuthenticator}")
print(f"  GoogleOAuthenticator: {oauthenticator.GoogleOAuthenticator}")
print(f"  BitbucketOAuthenticator: {oauthenticator.BitbucketOAuthenticator}")

# Show local variants
print(f"\nLocal authenticators:")
print(f"  LocalGitHubOAuthenticator: {oauthenticator.LocalGitHubOAuthenticator}")
print(f"  LocalGoogleOAuthenticator: {oauthenticator.LocalGoogleOAuthenticator}")
for provider_class in (oauthenticator.GitHubOAuthenticator, oauthenticator.GoogleOAuthenticator,
                       oauthenticator.BitbucketOAuthenticator):
    assert issubclass(provider_class, oauthenticator.OAuthenticator), provider_class
assert issubclass(oauthenticator.LocalGitHubOAuthenticator, oauthenticator.GitHubOAuthenticator)

print("\nExample 1 completed successfully")
```

## Example 2: GitHub OAuth Authenticator

```python
import oauthenticator

print("Example 2: GitHub OAuth Authenticator")

# Create GitHub authenticator instance
auth = oauthenticator.GitHubOAuthenticator()
print(f"Authenticator: {type(auth).__name__}")

# Show configurable attributes
attrs = ['client_id', 'client_secret', 'oauth_callback_url', 'scope']
print(f"\nConfigurable attributes:")
for attr in attrs:
    if hasattr(auth, attr):
        print(f"  {attr}: {getattr(auth, attr)}")
assert auth.login_service == "GitHub", auth.login_service
assert auth.authorize_url.startswith("https://github.com/"), auth.authorize_url

print("\nExample 2 completed successfully")
```

## Example 3: Google OAuth Authenticator

```python
import oauthenticator

print("Example 3: Google OAuth Authenticator")

# Create Google authenticator
auth = oauthenticator.GoogleOAuthenticator()
print(f"Authenticator: {type(auth).__name__}")

# Show Google-specific configuration
print(f"\nGoogle OAuth configuration:")
print(f"  hosted_domain: {getattr(auth, 'hosted_domain', None)}")
print(f"  login_service: {getattr(auth, 'login_service', None)}")
assert auth.login_service == "Google", auth.login_service
assert "openid" in auth.scope, auth.scope

print("\nExample 3 completed successfully")
```
