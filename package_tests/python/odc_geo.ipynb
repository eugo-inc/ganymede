## Example 1: Coordinate Reference Systems

```python
from odc.geo import CRS

print("Example 1: Coordinate Reference Systems")

# Create CRS objects
crs_wgs84 = CRS("EPSG:4326")
crs_mercator = CRS("EPSG:3857")
crs_utm = CRS("EPSG:32633")

print(f"WGS84: {crs_wgs84}")
print(f"Web Mercator: {crs_mercator}")
print(f"UTM 33N: {crs_utm}")

# Show CRS properties
print(f"\nWGS84 EPSG: {crs_wgs84.epsg}")
assert (crs_wgs84.epsg, crs_mercator.epsg, crs_utm.epsg) == (4326, 3857, 32633)
assert crs_wgs84.geographic and not crs_mercator.geographic

print("\nExample 1 completed successfully")
```

## Example 2: BoundingBox

```python
from odc.geo import BoundingBox, CRS

print("Example 2: BoundingBox")

# Create bounding box
crs = CRS("EPSG:4326")
bbox = BoundingBox(left=10, bottom=50, right=20, top=60, crs=crs)

print(f"BoundingBox: {bbox}")
print(f"Width: {bbox.width}")
print(f"Height: {bbox.height}")
print(f"Aspect: {bbox.aspect}")

# Transform to different CRS
mercator = CRS("EPSG:3857")
bbox_merc = bbox.to_crs(mercator)
print(f"\nTransformed CRS: {bbox_merc.crs}")
assert (bbox.width, bbox.height) == (10, 10)
assert bbox_merc.crs.epsg == 3857 and bbox_merc.left < bbox_merc.right

print("\nExample 2 completed successfully")
```

## Example 3: GeoBox

```python
from odc.geo.geobox import GeoBox
from odc.geo import CRS
from affine import Affine

print("Example 3: GeoBox")

# Create GeoBox
crs = CRS("EPSG:4326")
transform = Affine.translation(10, 50) * Affine.scale(0.1, -0.1)
geobox = GeoBox((100, 100), transform, crs)

print(f"Shape: {geobox.shape}")
print(f"CRS: {geobox.crs}")
print(f"Resolution: {geobox.resolution}")
print(f"Bounds: {geobox.boundingbox}")
assert geobox.shape == (100, 100)
bounds = geobox.boundingbox
assert (bounds.left, bounds.bottom, bounds.right, bounds.top) == (10.0, 40.0, 20.0, 50.0), bounds

print("\nExample 3 completed successfully")
```

## Example 4: Geometry

```python
from odc.geo import Geometry, CRS
from shapely.wkt import loads

print("Example 4: Geometry")

# Create geometry from WKT
crs = CRS("EPSG:4326")
wkt = "POLYGON ((10 50, 20 50, 20 60, 10 60, 10 50))"
shapely_geom = loads(wkt)
geom = Geometry(shapely_geom, crs)

print(f"Type: {geom.geom_type}")
print(f"CRS: {geom.crs}")
print(f"Bounds: {geom.boundingbox}")

# Transform
mercator = CRS("EPSG:3857")
geom_merc = geom.to_crs(mercator)
print(f"\nTransformed CRS: {geom_merc.crs}")
assert geom.geom_type == "Polygon" and geom.area == 100.0
assert geom_merc.crs.epsg == 3857

print("\nExample 4 completed successfully")
```
