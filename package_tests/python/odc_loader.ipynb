## Example 3: Resolve Functions

```python
from odc.loader import resolve_dst_dtype, resolve_dst_nodata, resolve_src_nodata, RasterLoadParams
import numpy as np

print("Example 3: Resolve Functions")

# Create load params
cfg = RasterLoadParams(dtype="uint8")

# Test resolve_dst_dtype
dtype1 = resolve_dst_dtype("uint8", cfg)
print(f"Resolved dtype: {dtype1}")

dtype2 = resolve_dst_dtype("float32", RasterLoadParams(dtype="float32"))
print(f"Float dtype: {dtype2}")

# Test resolve_dst_nodata
nodata1 = resolve_dst_nodata(np.dtype('float32'), cfg)
print(f"\nNodata for float32: {nodata1}")

# Test resolve_src_nodata
cfg2 = RasterLoadParams(src_nodata_override=-9999.0)
src_nodata = resolve_src_nodata(None, cfg2)
print(f"Source nodata: {src_nodata}")
assert np.dtype(dtype1) == np.dtype("uint8") and np.dtype(dtype2) == np.dtype("float32")
assert np.isnan(nodata1)
assert src_nodata == -9999.0

print("\nExample 3 completed successfully")
```

## Example 4: RasterLoadParams

```python
from odc.loader import RasterLoadParams
import numpy as np

print("Example 4: RasterLoadParams")

# Create raster load params
params1 = RasterLoadParams(dtype="uint16")
print(f"Params 1: {params1}")
print(f"  dtype: {params1.dtype}")

params2 = RasterLoadParams(dtype="float32", fill_value=np.nan)
print(f"\nParams 2: {params2}")
print(f"  dtype: {params2.dtype}")
print(f"  fill_value: {params2.fill_value}")

params3 = RasterLoadParams(dtype="uint8", fill_value=0, resampling="bilinear")
print(f"\nParams 3: {params3}")
print(f"  dtype: {params3.dtype}")
print(f"  resampling: {params3.resampling}")
assert (params3.dtype, params3.fill_value, params3.resampling) == ("uint8", 0, "bilinear")
assert np.isnan(params2.fill_value)

print("\nExample 4 completed successfully")
```
