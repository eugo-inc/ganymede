## Example 2: RasterBandMetadata

```python
from odc.stac import RasterBandMetadata

print("Example 2: RasterBandMetadata")

# Create band metadata
band1 = RasterBandMetadata(data_type="uint16", nodata=0)
print(f"Band 1: {band1}")
print(f"  data_type: {band1.data_type}")
print(f"  nodata: {band1.nodata}")

# Create band with units
band2 = RasterBandMetadata(data_type="float32", nodata=-9999.0, units="reflectance")
print(f"\nBand 2: {band2}")
print(f"  units: {band2.units}")
assert (band1.data_type, band1.nodata) == ("uint16", 0)
assert (band2.data_type, band2.nodata, band2.units) == ("float32", -9999.0, "reflectance")

print("\nExample 2 completed successfully")
```

## Example 3: RasterLoadParams

```python
from odc.stac import RasterLoadParams
import numpy as np

print("Example 3: RasterLoadParams")

# Create load params
params1 = RasterLoadParams(dtype="uint16")
print(f"Params 1: {params1}")
print(f"  dtype: {params1.dtype}")

params2 = RasterLoadParams(dtype="float32", fill_value=np.nan, resampling="bilinear")
print(f"\nParams 2: {params2}")
print(f"  fill_value: {params2.fill_value}")
print(f"  resampling: {params2.resampling}")

params3 = RasterLoadParams(dtype="uint8", use_overviews=False)
print(f"\nParams 3: use_overviews={params3.use_overviews}")
assert params2.resampling == "bilinear" and params3.use_overviews is False

print("\nExample 3 completed successfully")
```

## Example 4: RasterSource

```python
from odc.stac import RasterSource, RasterBandMetadata

print("Example 4: RasterSource")

# Create raster source
src1 = RasterSource(uri="s3://bucket/image.tif")
print(f"Source 1: {src1}")
print(f"  uri: {src1.uri}")
print(f"  band: {src1.band}")

# Create source with metadata
meta = RasterBandMetadata(data_type="uint16", nodata=0)
src2 = RasterSource(uri="s3://bucket/red.tif", band=1, meta=meta)
print(f"\nSource 2: {src2}")
print(f"  meta: {src2.meta}")

# Source with subdataset
src3 = RasterSource(uri="file.hdf", subdataset="temperature")
print(f"\nSource 3 subdataset: {src3.subdataset}")
assert (src2.uri, src2.band, src2.meta.data_type) == ("s3://bucket/red.tif", 1, "uint16")
assert src3.subdataset == "temperature"

print("\nExample 4 completed successfully")
```

## Example 5: Configuration Functions

```python
from odc.stac import configure_rio, configure_s3_access

print("Example 5: Configuration Functions")

# Show configure_rio function
print(f"configure_rio: {configure_rio}")

# Configure rasterio with cloud defaults
configure_rio(cloud_defaults=True)
print(f"Configured rasterio with cloud defaults")

# Show configure_s3_access function
print(f"\nconfigure_s3_access: {configure_s3_access}")

# Configure S3 access (with default settings)
configure_s3_access(aws_unsigned=True)
print(f"Configured S3 access with unsigned requests")

print("\nExample 5 completed successfully")
```
