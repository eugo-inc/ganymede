# opencensus smoke test

```python
# Example 1: Exploring opencensus
import opencensus

print('opencensus imported successfully')
attrs = [attr for attr in dir(opencensus) if not attr.startswith('_')]
print(f'Available attributes: {attrs}')

# The common opencensus modules are this package's own subpackages, not optional extras: import them directly
import opencensus.trace
import opencensus.stats
import opencensus.tags

for module in (opencensus.trace, opencensus.stats, opencensus.tags):
    print(f'\n✅ {module.__name__} available')
    module_attrs = [attr for attr in dir(module) if not attr.startswith('_')][:5]
    print(f'   Attributes: {module_attrs}...')

# trace and tags export their API at package level (stats exports nothing public from its __init__)
assert {'Span', 'span_context', 'tracers'} <= set(dir(opencensus.trace)), dir(opencensus.trace)
assert {'Tag', 'TagKey', 'TagMap', 'TagValue'} <= set(dir(opencensus.tags)), dir(opencensus.tags)

print('\nOpencensus exploration complete')
```

```python
# Example 2: Start a span
from opencensus.trace.tracer import Tracer
tracer = Tracer()
with tracer.span(name='test-span') as span:
    print(span)
```

```python
# Example 3: Add attributes to a span
from opencensus.trace.tracer import Tracer
tracer = Tracer()
with tracer.span(name='span-with-attrs') as span:
    span.add_attribute('key', 'value')
    print(span.attributes)
```

```python
# Example 4: Use a custom sampler
from opencensus.trace.tracer import Tracer
from opencensus.trace.samplers import AlwaysOnSampler
tracer = Tracer(sampler=AlwaysOnSampler())
print(tracer.sampler)
```

```python
# Example 5: stats recorder and view manager
from opencensus.stats import stats as stats_module
from opencensus.stats.stats_recorder import StatsRecorder
from opencensus.stats.view_manager import ViewManager

stats = stats_module.stats
view_manager = stats.view_manager
stats_recorder = stats.stats_recorder
print(stats_recorder)
print(view_manager)
assert isinstance(stats_recorder, StatsRecorder), type(stats_recorder)
assert isinstance(view_manager, ViewManager), type(view_manager)
```
