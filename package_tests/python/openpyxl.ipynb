## Example 1: Basic workbook and worksheet operations

```python
# Example 1: Basic workbook and worksheet operations
import openpyxl
from openpyxl import Workbook, load_workbook
from openpyxl.utils import get_column_letter, column_index_from_string
from io import BytesIO

print("Example 1 - Basic workbook and worksheet operations")

try:
    # Test basic import and version
    print(f"✓ openpyxl module imported successfully")
    print(f"  Version: {openpyxl.__version__}")
    
    # Create a new workbook
    wb = Workbook()
    ws = wb.active
    print(f"✓ Created workbook with active worksheet: '{ws.title}'")
    
    # Test basic cell operations
    print(f"\n✓ Testing basic cell operations:")
    
    # Write headers
    headers = ['Name', 'Age', 'City', 'Department', 'Salary']
    for col, header in enumerate(headers, start=1):
        ws.cell(row=1, column=col, value=header)
    
    print(f"  Added {len(headers)} headers to row 1")
    
    # Add employee data using different cell access methods
    employee_data = [
        ['Alice Johnson', 28, 'New York', 'Engineering', 85000],
        ['Bob Smith', 32, 'San Francisco', 'Marketing', 72000],
        ['Charlie Brown', 29, 'Chicago', 'Sales', 68000],
        ['Diana Ross', 35, 'Boston', 'HR', 75000],
        ['Edward Wong', 31, 'Seattle', 'Engineering', 90000]
    ]
    
    # Method 1: Using cell coordinates
    for row_idx, employee in enumerate(employee_data[:2], start=2):
        for col_idx, value in enumerate(employee, start=1):
            ws.cell(row=row_idx, column=col_idx, value=value)
    
    # Method 2: Using cell references
    ws['A4'] = employee_data[2][0]  # Charlie Brown
    ws['B4'] = employee_data[2][1]  # 29
    ws['C4'] = employee_data[2][2]  # Chicago
    ws['D4'] = employee_data[2][3]  # Sales
    ws['E4'] = employee_data[2][4]  # 68000
    
    # Method 3: Using append
    for employee in employee_data[3:]:
        ws.append(employee)
    
    print(f"  Added {len(employee_data)} employee records using different methods")
    print(f"  Final worksheet dimensions: {ws.max_row} rows × {ws.max_column} columns")
    
    # Test cell access and data types
    print(f"\n✓ Testing cell access methods:")
    
    # Access by coordinate
    print(f"  Cell A1 (coordinate): '{ws['A1'].value}'")
    print(f"  Cell B2 (coordinate): {ws['B2'].value} ({type(ws['B2'].value).__name__})")
    
    # Access by row/column
    print(f"  Cell (1,3) row/col: '{ws.cell(row=1, column=3).value}'")
    print(f"  Cell (3,5) row/col: ${ws.cell(row=3, column=5).value:,}")
    
    # Test column utilities
    print(f"\n✓ Testing column utilities:")
    print(f"  Column 1 letter: {get_column_letter(1)}")
    print(f"  Column 5 letter: {get_column_letter(5)}")
    print(f"  Column 'E' index: {column_index_from_string('E')}")
    
    # Test row/column iteration
    print(f"\n✓ Testing data iteration:")
    
    # Iterate over specific range
    salary_sum = 0
    high_earners = 0
    
    for row in ws.iter_rows(min_row=2, max_row=ws.max_row, min_col=5, max_col=5, values_only=True):
        salary = row[0]
        if isinstance(salary, (int, float)):
            salary_sum += salary
            if salary > 75000:
                high_earners += 1
    
    print(f"  Total salary budget: ${salary_sum:,}")
    print(f"  High earners (>$75k): {high_earners}")
    assert (salary_sum, high_earners) == (390000, 2), (salary_sum, high_earners)
    assert (get_column_letter(5), column_index_from_string('E')) == ('E', 5)
    
    # Test worksheet properties
    print(f"\n✓ Worksheet properties:")
    print(f"  Title: '{ws.title}'")
    print(f"  Max row: {ws.max_row}")
    print(f"  Max column: {ws.max_column}")
    print(f"  Dimensions: {ws.dimensions}")
    
    # Test in-memory save/load
    print(f"\n✓ Testing in-memory operations:")
    buffer = BytesIO()
    wb.save(buffer)
    buffer.seek(0)
    
    # Load from buffer
    wb_loaded = load_workbook(buffer)
    ws_loaded = wb_loaded.active
    
    print(f"  Saved workbook size: {len(buffer.getvalue())} bytes")
    print(f"  Loaded worksheet: '{ws_loaded.title}'")
    print(f"  Data integrity check: {ws_loaded['A2'].value == employee_data[0][0]}")
    # Every value written above comes back from the saved file unchanged
    assert [c.value for c in ws_loaded[1]] == headers
    assert [list(row) for row in ws_loaded.iter_rows(min_row=2, values_only=True)] == employee_data
    
    print("Example 1 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"openpyxl example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished")
```

## Example 2: Advanced cell formatting and styling

```python
# Example 2: Advanced cell formatting and styling
import openpyxl
from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side, NamedStyle
from openpyxl.styles.numbers import FORMAT_CURRENCY_USD_SIMPLE, FORMAT_PERCENTAGE

print("Example 2 - Advanced cell formatting and styling")

try:
    # Create workbook with comprehensive formatting
    wb = Workbook()
    ws = wb.active
    ws.title = "Sales Dashboard"
    
    print(f"✓ Created workbook: '{ws.title}'")
    
    # Define styles
    print(f"\n✓ Creating custom styles:")
    
    # Header style
    header_style = NamedStyle(name="header_style")
    header_style.font = Font(bold=True, color='FFFFFF', size=12)
    header_style.fill = PatternFill(start_color='1f4e79', end_color='1f4e79', fill_type='solid')
    header_style.alignment = Alignment(horizontal='center', vertical='center')
    header_style.border = Border(
        left=Side(style='thin', color='000000'),
        right=Side(style='thin', color='000000'),
        top=Side(style='thin', color='000000'),
        bottom=Side(style='thin', color='000000')
    )
    wb.add_named_style(header_style)
    
    # Data style
    data_style = NamedStyle(name="data_style")
    data_style.alignment = Alignment(horizontal='left', vertical='center')
    data_style.border = Border(
        left=Side(style='thin', color='CCCCCC'),
        right=Side(style='thin', color='CCCCCC'),
        top=Side(style='thin', color='CCCCCC'),
        bottom=Side(style='thin', color='CCCCCC')
    )
    wb.add_named_style(data_style)
    
    print(f"  Created named styles: header_style, data_style")
    
    # Create sales report data
    print(f"\n✓ Creating sales report with formatting:")
    
    headers = ['Product', 'Category', 'Units Sold', 'Unit Price', 'Total Revenue', 'Profit Margin']
    products_data = [
        ['Laptop Pro', 'Electronics', 150, 1299.99, 194998.50, 0.25],
        ['Wireless Mouse', 'Electronics', 500, 29.99, 14995.00, 0.40],
        ['Office Chair', 'Furniture', 75, 249.99, 18749.25, 0.30],
        ['Coffee Maker', 'Appliances', 120, 89.99, 10798.80, 0.35],
        ['Notebook Set', 'Office Supplies', 300, 12.99, 3897.00, 0.60]
    ]
    
    # Apply header formatting
    for col, header in enumerate(headers, start=1):
        cell = ws.cell(row=1, column=col, value=header)
        cell.style = header_style
    
    # Apply data formatting with conditional styling
    for row_idx, product in enumerate(products_data, start=2):
        for col_idx, value in enumerate(product, start=1):
            cell = ws.cell(row=row_idx, column=col_idx, value=value)
            
            # Apply base data style
            cell.style = data_style
            
            # Column-specific formatting
            if col_idx == 3:  # Units Sold
                cell.number_format = '#,##0'
            elif col_idx == 4:  # Unit Price
                cell.number_format = FORMAT_CURRENCY_USD_SIMPLE
            elif col_idx == 5:  # Total Revenue
                cell.number_format = FORMAT_CURRENCY_USD_SIMPLE
                # Highlight high revenue (>$15,000)
                if value > 15000:
                    cell.fill = PatternFill(start_color='C6EFCE', end_color='C6EFCE', fill_type='solid')
            elif col_idx == 6:  # Profit Margin
                cell.number_format = FORMAT_PERCENTAGE
                # Color code profit margins
                if value >= 0.4:
                    cell.fill = PatternFill(start_color='C6EFCE', end_color='C6EFCE', fill_type='solid')  # Green
                elif value >= 0.3:
                    cell.fill = PatternFill(start_color='FFEB9C', end_color='FFEB9C', fill_type='solid')  # Yellow
                else:
                    cell.fill = PatternFill(start_color='FFC7CE', end_color='FFC7CE', fill_type='solid')  # Red
    
    print(f"  Applied formatting to {len(products_data)} product rows")
    
    # Adjust column widths
    column_widths = [15, 12, 12, 12, 15, 12]
    for col, width in enumerate(column_widths, start=1):
        column_letter = ws.cell(row=1, column=col).column_letter
        ws.column_dimensions[column_letter].width = width
    
    print(f"  Set column widths: {column_widths}")
    
    # Add summary section
    print(f"\n✓ Adding summary section:")
    
    summary_row = len(products_data) + 3
    
    # Summary labels and values
    summaries = [
        ('Total Products:', len(products_data)),
        ('Total Units Sold:', sum(p[2] for p in products_data)),
        ('Total Revenue:', sum(p[4] for p in products_data)),
        ('Average Profit Margin:', sum(p[5] for p in products_data) / len(products_data))
    ]
    
    for i, (label, value) in enumerate(summaries):
        label_cell = ws.cell(row=summary_row + i, column=1, value=label)
        value_cell = ws.cell(row=summary_row + i, column=2, value=value)
        
        # Format summary labels
        label_cell.font = Font(bold=True)
        
        # Format summary values
        if 'Revenue' in label:
            value_cell.number_format = FORMAT_CURRENCY_USD_SIMPLE
        elif 'Margin' in label:
            value_cell.number_format = FORMAT_PERCENTAGE
        elif 'Units' in label:
            value_cell.number_format = '#,##0'
    
    print(f"  Added {len(summaries)} summary metrics")
    
    # Test different fill patterns and colors
    print(f"\n✓ Testing additional formatting options:")
    
    # Create a small color palette demonstration
    palette_start_row = summary_row + len(summaries) + 2
    colors = ['FF0000', '00FF00', '0000FF', 'FFFF00', 'FF00FF', '00FFFF']
    fill_types = ['solid', 'darkDown', 'darkUp', 'darkGrid', 'lightGrid']
    
    ws.cell(row=palette_start_row, column=1, value="Color Palette Demo:")
    ws.cell(row=palette_start_row, column=1).font = Font(bold=True)
    
    for i, color in enumerate(colors[:3]):  # Show first 3 colors
        cell = ws.cell(row=palette_start_row + 1, column=i + 1, value=f"#{color}")
        cell.fill = PatternFill(start_color=color, end_color=color, fill_type='solid')
        cell.alignment = Alignment(horizontal='center')
    
    print(f"  Added color palette demonstration")
    
    # Test text alignment variations
    align_row = palette_start_row + 3
    alignments = [
        ('Left', Alignment(horizontal='left')),
        ('Center', Alignment(horizontal='center')),
        ('Right', Alignment(horizontal='right')),
        ('Justify', Alignment(horizontal='justify'))
    ]
    
    ws.cell(row=align_row, column=1, value="Text Alignment Demo:")
    ws.cell(row=align_row, column=1).font = Font(bold=True)
    
    for i, (align_name, alignment) in enumerate(alignments):
        cell = ws.cell(row=align_row + 1, column=i + 1, value=align_name)
        cell.alignment = alignment
        cell.border = Border(bottom=Side(style='thin'))
    
    print(f"  Added text alignment demonstration")
    
    # Summary of formatting features
    print(f"\n✓ Applied formatting features:")
    features = [
        "Named styles for consistency",
        "Conditional formatting based on values",
        "Number formatting (currency, percentage, thousands)",
        "Cell borders and alignment",
        "Column width adjustment",
        "Color coding for data visualization",
        "Font styling (bold, color, size)",
        "Fill patterns and background colors"
    ]
    
    for feature in features:
        print(f"    • {feature}")
    
    print("Example 2 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"openpyxl example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished")
```

## Example 3: Formulas, calculations, and data validation

```python
# Example 3: Formulas, calculations, and data validation
import openpyxl
from openpyxl import Workbook
from openpyxl.utils import get_column_letter
from openpyxl.worksheet.datavalidation import DataValidation
import datetime
import random

print("Example 3 - Formulas, calculations, and data validation")

try:
    # Create workbook with formulas and validation
    wb = Workbook()
    ws = wb.active
    ws.title = "Financial Analysis"
    
    print(f"✓ Created workbook: '{ws.title}'")
    
    # Set up financial data with headers
    print(f"\n✓ Creating financial analysis with formulas:")
    
    headers = ['Month', 'Revenue', 'Fixed Costs', 'Variable Costs', 'Total Costs', 'Gross Profit', 'Profit Margin', 'ROI']
    for col, header in enumerate(headers, start=1):
        ws.cell(row=1, column=col, value=header)
    
    # Add monthly data
    months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
    base_revenue = 50000
    monthly_data = []
    
    for i, month in enumerate(months):
        # Generate realistic financial data
        revenue = base_revenue + random.randint(-5000, 10000)
        fixed_costs = 20000
        variable_costs = int(revenue * random.uniform(0.3, 0.5))
        
        row_num = i + 2
        
        # Input raw data
        ws.cell(row=row_num, column=1, value=month)  # Month
        ws.cell(row=row_num, column=2, value=revenue)  # Revenue
        ws.cell(row=row_num, column=3, value=fixed_costs)  # Fixed Costs
        ws.cell(row=row_num, column=4, value=variable_costs)  # Variable Costs
        
        # Add formulas for calculated fields
        ws.cell(row=row_num, column=5, value=f'=C{row_num}+D{row_num}')  # Total Costs = Fixed + Variable
        ws.cell(row=row_num, column=6, value=f'=B{row_num}-E{row_num}')  # Gross Profit = Revenue - Total Costs
        ws.cell(row=row_num, column=7, value=f'=F{row_num}/B{row_num}')  # Profit Margin = Profit / Revenue
        ws.cell(row=row_num, column=8, value=f'=F{row_num}/E{row_num}')  # ROI = Profit / Total Costs
        
        monthly_data.append([month, revenue, fixed_costs, variable_costs])
    
    print(f"  Added {len(months)} months of financial data with formulas")
    
    # Add summary formulas
    print(f"\n✓ Adding summary calculations:")
    
    summary_row = len(months) + 3
    summary_formulas = [
        ('TOTAL REVENUE:', f'=SUM(B2:B{len(months)+1})'),
        ('TOTAL COSTS:', f'=SUM(E2:E{len(months)+1})'),
        ('TOTAL PROFIT:', f'=SUM(F2:F{len(months)+1})'),
        ('AVG PROFIT MARGIN:', f'=AVERAGE(G2:G{len(months)+1})'),
        ('MAX MONTHLY PROFIT:', f'=MAX(F2:F{len(months)+1})'),
        ('MIN MONTHLY PROFIT:', f'=MIN(F2:F{len(months)+1})')
    ]
    
    for i, (label, formula) in enumerate(summary_formulas):
        ws.cell(row=summary_row + i, column=1, value=label)
        ws.cell(row=summary_row + i, column=2, value=formula)
    
    print(f"  Added {len(summary_formulas)} summary formulas")
    
    # Test different formula types
    print(f"\n✓ Testing various formula types:")
    
    formula_demo_row = summary_row + len(summary_formulas) + 2
    
    # Mathematical formulas
    ws.cell(row=formula_demo_row, column=1, value="Math Formulas Demo:")
    math_formulas = [
        ('Square Root:', '=SQRT(100)'),
        ('Power:', '=POWER(5,2)'),
        ('Absolute:', '=ABS(-15)'),
        ('Round:', '=ROUND(3.14159,2)'),
        ('Random:', '=RAND()'),
        ('PI:', '=PI()')
    ]
    
    for i, (label, formula) in enumerate(math_formulas):
        ws.cell(row=formula_demo_row + i + 1, column=1, value=label)
        ws.cell(row=formula_demo_row + i + 1, column=2, value=formula)
    
    print(f"  Added {len(math_formulas)} mathematical formulas")
    
    # Date and time formulas
    date_demo_row = formula_demo_row + len(math_formulas) + 3
    ws.cell(row=date_demo_row, column=1, value="Date/Time Formulas:")
    
    date_formulas = [
        ('Today:', '=TODAY()'),
        ('Now:', '=NOW()'),
        ('Year:', '=YEAR(TODAY())'),
        ('Month:', '=MONTH(TODAY())'),
        ('Day:', '=DAY(TODAY())'),
        ('Weekday:', '=WEEKDAY(TODAY())')
    ]
    
    for i, (label, formula) in enumerate(date_formulas):
        ws.cell(row=date_demo_row + i + 1, column=1, value=label)
        ws.cell(row=date_demo_row + i + 1, column=2, value=formula)
    
    print(f"  Added {len(date_formulas)} date/time formulas")
    
    # Add data validation
    print(f"\n✓ Adding data validation:")
    
    # Create a new sheet for validation demo
    validation_ws = wb.create_sheet("Data Validation Demo")
    
    # Number validation
    validation_ws['A1'] = 'Age (18-65):'
    validation_ws['B1'] = ''  # Cell for user input
    
    age_validation = DataValidation(
        type="whole",
        operator="between",
        formula1=18,
        formula2=65,
        showErrorMessage=True,
        errorTitle="Invalid Age",
        error="Age must be between 18 and 65"
    )
    validation_ws.add_data_validation(age_validation)
    age_validation.add('B1')
    
    # List validation
    validation_ws['A2'] = 'Department:'
    validation_ws['B2'] = ''
    
    dept_validation = DataValidation(
        type="list",
        formula1='"Engineering,Marketing,Sales,HR,Finance"',
        showErrorMessage=True,
        errorTitle="Invalid Department",
        error="Please select from the dropdown list"
    )
    validation_ws.add_data_validation(dept_validation)
    dept_validation.add('B2')
    
    # Date validation
    validation_ws['A3'] = 'Start Date (Future):'
    validation_ws['B3'] = ''
    
    date_validation = DataValidation(
        type="date",
        operator="greaterThan",
        formula1=datetime.date.today(),
        showErrorMessage=True,
        errorTitle="Invalid Date",
        error="Start date must be in the future"
    )
    validation_ws.add_data_validation(date_validation)
    date_validation.add('B3')
    
    # Decimal validation
    validation_ws['A4'] = 'Score (0.0-100.0):'
    validation_ws['B4'] = ''
    
    score_validation = DataValidation(
        type="decimal",
        operator="between",
        formula1=0.0,
        formula2=100.0,
        showErrorMessage=True,
        errorTitle="Invalid Score",
        error="Score must be between 0.0 and 100.0"
    )
    validation_ws.add_data_validation(score_validation)
    score_validation.add('B4')
    
    print(f"  Added data validation to '{validation_ws.title}' sheet:")
    validation_types = ['Age (whole number range)', 'Department (list)', 'Start Date (future dates)', 'Score (decimal range)']
    for vtype in validation_types:
        print(f"    • {vtype}")
    
    # Test conditional formulas
    print(f"\n✓ Testing conditional formulas:")
    
    # Go back to main sheet and add conditional formulas
    conditional_row = date_demo_row + len(date_formulas) + 3
    ws.cell(row=conditional_row, column=1, value="Conditional Formulas:")
    
    # Add sample data for conditional formulas
    ws.cell(row=conditional_row + 1, column=1, value="Grade:")
    ws.cell(row=conditional_row + 1, column=2, value=85)  # Sample score
    ws.cell(row=conditional_row + 1, column=3, value='=IF(B' + str(conditional_row + 1) + '>=90,"A",IF(B' + str(conditional_row + 1) + '>=80,"B",IF(B' + str(conditional_row + 1) + '>=70,"C","F")))')
    
    ws.cell(row=conditional_row + 2, column=1, value="Status:")
    ws.cell(row=conditional_row + 2, column=2, value=75)  # Sample value
    ws.cell(row=conditional_row + 2, column=3, value='=IF(B' + str(conditional_row + 2) + '>90,"Excellent",IF(B' + str(conditional_row + 2) + '>70,"Good","Needs Improvement"))')
    
    # VLOOKUP example (create lookup table first)
    lookup_start = conditional_row + 5
    ws.cell(row=lookup_start, column=1, value="VLOOKUP Demo:")
    
    # Create lookup table
    lookup_table = [
        ['Product ID', 'Product Name', 'Price'],
        [101, 'Laptop', 999.99],
        [102, 'Mouse', 29.99],
        [103, 'Keyboard', 79.99],
        [104, 'Monitor', 299.99]
    ]
    
    for i, row in enumerate(lookup_table):
        for j, value in enumerate(row):
            ws.cell(row=lookup_start + 1 + i, column=j + 1, value=value)
    
    # VLOOKUP formula
    ws.cell(row=lookup_start + len(lookup_table) + 1, column=1, value="Lookup ID 102:")
    ws.cell(row=lookup_start + len(lookup_table) + 1, column=2, value=f'=VLOOKUP(102,A{lookup_start + 2}:C{lookup_start + len(lookup_table)},2,FALSE)')
    
    print(f"  Added conditional formulas: IF statements, VLOOKUP")
    
    # Number formatting for formula results
    print(f"\n✓ Applying number formatting to formula results:")
    
    # Format percentage columns
    for row in range(2, len(months) + 2):
        ws.cell(row=row, column=7).number_format = '0.00%'  # Profit Margin
        ws.cell(row=row, column=8).number_format = '0.00%'  # ROI
    
    # Format currency columns
    for row in range(2, len(months) + 2):
        for col in [2, 3, 4, 5, 6]:  # Revenue, costs, profit columns
            ws.cell(row=row, column=col).number_format = '$#,##0.00'
    
    # Format summary formulas
    for i in range(len(summary_formulas)):
        if 'MARGIN' in summary_formulas[i][0]:
            ws.cell(row=summary_row + i, column=2).number_format = '0.00%'
        else:
            ws.cell(row=summary_row + i, column=2).number_format = '$#,##0.00'
    
    print(f"  Applied currency and percentage formatting to calculated fields")
    
    # Summary of formula capabilities
    print(f"\n✓ Formula and calculation features demonstrated:")
    features = [
        "Basic arithmetic formulas (SUM, subtraction, division)",
        "Statistical functions (AVERAGE, MAX, MIN)",
        "Mathematical functions (SQRT, POWER, ABS, ROUND)",
        "Date and time functions (TODAY, NOW, YEAR, MONTH)",
        "Conditional logic (IF statements, nested conditions)",
        "Lookup functions (VLOOKUP)",
        "Data validation (numbers, lists, dates, decimals)",
        "Number formatting for formula results",
        "Cross-sheet references and calculations"
    ]
    
    for feature in features:
        print(f"    • {feature}")
    
    print(f"\nWorkbook structure: {len(wb.worksheets)} sheets")
    for sheet in wb.worksheets:
        print(f"  • {sheet.title}")
    
    print("Example 3 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"openpyxl example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished")
```

## Example 4: Charts, Excel tables, and advanced worksheet features

```python
# Example 4: Charts, Excel tables, and advanced worksheet features
import openpyxl
from openpyxl import Workbook
from openpyxl.chart import BarChart, LineChart, PieChart, Reference, Series
from openpyxl.chart.axis import DateAxis, ChartLines
from openpyxl.chart.data_source import NumData, NumVal
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side, NamedStyle
from openpyxl.worksheet.table import Table, TableStyleInfo
from openpyxl.utils import get_column_letter
from openpyxl.workbook.protection import WorkbookProtection
from openpyxl.worksheet.protection import SheetProtection
import random

print("Example 4 - Charts, Excel tables, and advanced worksheet features")

try:
    # Create comprehensive business dashboard
    wb = Workbook()
    ws = wb.active
    ws.title = "Sales Dashboard"
    
    print(f"✓ Created workbook: '{ws.title}'")
    
    # Create comprehensive sales data
    print(f"\n✓ Building comprehensive sales data:")
    
    # Headers for sales data
    headers = ['Month', 'Product', 'Category', 'Sales Rep', 'Units Sold', 'Unit Price', 'Total Revenue', 'Cost', 'Profit']
    for col, header in enumerate(headers, start=1):
        ws.cell(row=1, column=col, value=header)
    
    # Generate comprehensive sales data
    months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
    products = [
        ('Laptop Pro', 'Electronics', 1299.99, 899.99),
        ('Wireless Mouse', 'Electronics', 29.99, 18.99),
        ('Office Chair', 'Furniture', 249.99, 174.99),
        ('Standing Desk', 'Furniture', 599.99, 419.99),
        ('Coffee Maker', 'Appliances', 89.99, 62.99),
        ('Tablet', 'Electronics', 399.99, 279.99)
    ]
    sales_reps = ['Alice Johnson', 'Bob Smith', 'Carol Davis', 'David Wilson']
    
    sales_data = []
    row_num = 2
    
    for month in months:
        for product_name, category, unit_price, cost in products:
            # Random sales rep and units
            sales_rep = random.choice(sales_reps)
            units_sold = random.randint(5, 50)
            total_revenue = units_sold * unit_price
            total_cost = units_sold * cost
            profit = total_revenue - total_cost
            
            # Write data to worksheet
            row_data = [month, product_name, category, sales_rep, units_sold, unit_price, total_revenue, total_cost, profit]
            for col, value in enumerate(row_data, start=1):
                ws.cell(row=row_num, column=col, value=value)
            
            sales_data.append(row_data)
            row_num += 1
    
    print(f"  Generated {len(sales_data)} sales records across {len(months)} months")
    
    # Create Excel Table for better data management
    print(f"\n✓ Creating Excel table:")
    
    table_range = f"A1:{get_column_letter(len(headers))}{len(sales_data) + 1}"
    sales_table = Table(displayName="SalesData", ref=table_range)
    
    # Add table style
    table_style = TableStyleInfo(
        name="TableStyleMedium9",
        showFirstColumn=False,
        showLastColumn=False,
        showRowStripes=True,
        showColumnStripes=True
    )
    sales_table.tableStyleInfo = table_style
    ws.add_table(sales_table)
    
    print(f"  Created Excel table 'SalesData' with range {table_range}")
    
    # Format currency and number columns
    for row in range(2, len(sales_data) + 2):
        ws.cell(row=row, column=6).number_format = '$#,##0.00'  # Unit Price
        ws.cell(row=row, column=7).number_format = '$#,##0.00'  # Total Revenue
        ws.cell(row=row, column=8).number_format = '$#,##0.00'  # Cost
        ws.cell(row=row, column=9).number_format = '$#,##0.00'  # Profit
    
    # Create multiple chart types
    print(f"\n✓ Creating multiple chart types:")
    
    # 1. Monthly Revenue Bar Chart
    chart_ws = wb.create_sheet("Charts")
    
    # Aggregate monthly revenue data for chart
    monthly_totals = {}
    for data_row in sales_data:
        month = data_row[0]
        revenue = data_row[6]
        if month not in monthly_totals:
            monthly_totals[month] = 0
        monthly_totals[month] += revenue
    
    # Write monthly data to chart sheet
    chart_ws['A1'] = 'Month'
    chart_ws['B1'] = 'Total Revenue'
    for i, (month, revenue) in enumerate(monthly_totals.items(), start=2):
        chart_ws[f'A{i}'] = month
        chart_ws[f'B{i}'] = revenue
    
    # Create bar chart
    bar_chart = BarChart()
    bar_chart.title = "Monthly Revenue Trend"
    bar_chart.y_axis.title = "Revenue ($)"
    bar_chart.x_axis.title = "Month"
    bar_chart.width = 15
    bar_chart.height = 10
    
    data_ref = Reference(chart_ws, min_col=2, min_row=1, max_row=len(monthly_totals) + 1)
    cats_ref = Reference(chart_ws, min_col=1, min_row=2, max_row=len(monthly_totals) + 1)
    
    bar_chart.add_data(data_ref, titles_from_data=True)
    bar_chart.set_categories(cats_ref)
    chart_ws.add_chart(bar_chart, "D2")
    
    print(f"  Created bar chart showing {len(monthly_totals)} months of revenue data")
    
    # 2. Category Distribution Pie Chart
    # Aggregate category data
    category_totals = {}
    for data_row in sales_data:
        category = data_row[2]
        revenue = data_row[6]
        if category not in category_totals:
            category_totals[category] = 0
        category_totals[category] += revenue
    
    # Write category data
    chart_ws['A10'] = 'Category'
    chart_ws['B10'] = 'Revenue'
    for i, (category, revenue) in enumerate(category_totals.items(), start=11):
        chart_ws[f'A{i}'] = category
        chart_ws[f'B{i}'] = revenue
    
    # Create pie chart
    pie_chart = PieChart()
    pie_chart.title = "Revenue by Category"
    pie_chart.width = 12
    pie_chart.height = 10
    
    pie_data = Reference(chart_ws, min_col=2, min_row=10, max_row=10 + len(category_totals))
    pie_cats = Reference(chart_ws, min_col=1, min_row=11, max_row=10 + len(category_totals))
    
    pie_chart.add_data(pie_data, titles_from_data=True)
    pie_chart.set_categories(pie_cats)
    chart_ws.add_chart(pie_chart, "D20")
    
    print(f"  Created pie chart showing {len(category_totals)} product categories")
    
    # 3. Sales Rep Performance Line Chart
    # Aggregate sales rep data
    rep_totals = {}
    for data_row in sales_data:
        rep = data_row[3]
        profit = data_row[8]
        if rep not in rep_totals:
            rep_totals[rep] = 0
        rep_totals[rep] += profit
    
    # Write sales rep data
    chart_ws['A20'] = 'Sales Rep'
    chart_ws['B20'] = 'Total Profit'
    for i, (rep, profit) in enumerate(rep_totals.items(), start=21):
        chart_ws[f'A{i}'] = rep
        chart_ws[f'B{i}'] = profit
    
    # Create line chart
    line_chart = LineChart()
    line_chart.title = "Sales Rep Performance"
    line_chart.y_axis.title = "Profit ($)"
    line_chart.x_axis.title = "Sales Representative"
    line_chart.width = 12
    line_chart.height = 8
    
    line_data = Reference(chart_ws, min_col=2, min_row=20, max_row=20 + len(rep_totals))
    line_cats = Reference(chart_ws, min_col=1, min_row=21, max_row=20 + len(rep_totals))
    
    line_chart.add_data(line_data, titles_from_data=True)
    line_chart.set_categories(line_cats)
    chart_ws.add_chart(line_chart, "Q2")
    
    print(f"  Created line chart showing {len(rep_totals)} sales representatives")
    
    # Create summary and analysis sheet
    print(f"\n✓ Creating summary and analysis sheet:")
    
    summary_ws = wb.create_sheet("Executive Summary")
    
    # Executive summary with formulas
    summary_ws['A1'] = 'EXECUTIVE SUMMARY'
    summary_ws['A1'].font = Font(bold=True, size=16)
    
    summary_metrics = [
        ('Total Revenue:', f"=SUM('{ws.title}'!G:G)"),
        ('Total Profit:', f"=SUM('{ws.title}'!I:I)"),
        ('Average Order Value:', f"=AVERAGE('{ws.title}'!G:G)"),
        ('Total Units Sold:', f"=SUM('{ws.title}'!E:E)"),
        ('Profit Margin:', f"=(SUM('{ws.title}'!I:I)/SUM('{ws.title}'!G:G))"),
        ('Number of Transactions:', f"=COUNTA('{ws.title}'!A:A)-1"),
        ('Best Month (Revenue):', 'Manual Analysis Required'),
        ('Top Category:', 'Manual Analysis Required')
    ]
    
    for i, (metric, formula) in enumerate(summary_metrics, start=3):
        summary_ws[f'A{i}'] = metric
        summary_ws[f'B{i}'] = formula if not metric.startswith(('Best', 'Top')) else formula
        
        # Format cells
        summary_ws[f'A{i}'].font = Font(bold=True)
        if 'Revenue' in metric or 'Profit' in metric or 'Value' in metric:
            summary_ws[f'B{i}'].number_format = '$#,##0.00'
        elif 'Margin' in metric:
            summary_ws[f'B{i}'].number_format = '0.00%'
        elif 'Units' in metric or 'Transactions' in metric:
            summary_ws[f'B{i}'].number_format = '#,##0'
    
    print(f"  Added {len(summary_metrics)} executive summary metrics")
    
    # Advanced worksheet features
    print(f"\n✓ Implementing advanced worksheet features:")
    
    # Freeze panes on main sheet
    ws.freeze_panes = 'B2'  # Freeze first column and first row
    print(f"  Applied freeze panes to main data sheet")
    
    # Auto-filter on table
    ws.auto_filter.ref = table_range
    print(f"  Applied auto-filter to data range")
    
    # Column width optimization
    for col in range(1, len(headers) + 1):
        column_letter = get_column_letter(col)
        if col in [2, 3, 4]:  # Text columns
            ws.column_dimensions[column_letter].width = 15
        elif col in [6, 7, 8, 9]:  # Currency columns
            ws.column_dimensions[column_letter].width = 12
        else:  # Other columns
            ws.column_dimensions[column_letter].width = 10
    
    print(f"  Optimized column widths for readability")
    
    # Create data validation sheet
    print(f"\n✓ Adding advanced data validation:")
    
    validation_ws = wb.create_sheet("Data Entry Form")
    
    # Create input form with validations
    form_fields = [
        ('Product Name:', 'Text input'),
        ('Category:', 'Electronics,Furniture,Appliances'),
        ('Unit Price:', 'Decimal (0-10000)'),
        ('Sales Rep:', 'Alice Johnson,Bob Smith,Carol Davis,David Wilson'),
        ('Month:', 'Jan,Feb,Mar,Apr,May,Jun,Jul,Aug,Sep,Oct,Nov,Dec')
    ]
    
    from openpyxl.worksheet.datavalidation import DataValidation
    
    for i, (label, validation_info) in enumerate(form_fields, start=1):
        validation_ws[f'A{i}'] = label
        validation_ws[f'A{i}'].font = Font(bold=True)
        
        # Apply appropriate validation
        if 'Text input' in validation_info:
            # No specific validation for text
            pass
        elif 'Decimal' in validation_info:
            # Decimal validation
            decimal_validation = DataValidation(
                type="decimal",
                operator="between",
                formula1=0,
                formula2=10000
            )
            validation_ws.add_data_validation(decimal_validation)
            decimal_validation.add(f'B{i}')
        else:
            # List validation
            list_validation = DataValidation(
                type="list",
                formula1=f'"{validation_info}"'
            )
            validation_ws.add_data_validation(list_validation)
            list_validation.add(f'B{i}')
    
    print(f"  Created data entry form with {len(form_fields)} validated fields")
    
    # Summary of advanced features
    print(f"\n✓ Advanced worksheet features implemented:")
    features = [
        f"Excel Tables with formatting ({len(sales_data)} records)",
        f"Multiple chart types (Bar, Pie, Line)",
        f"Cross-sheet formulas and references",
        f"Executive summary with calculated metrics",
        f"Freeze panes and auto-filtering",
        f"Column width optimization",
        f"Data validation forms",
        f"Professional styling and formatting",
        f"Multi-sheet workbook organization ({len(wb.worksheets)} sheets)"
    ]
    
    for feature in features:
        print(f"    • {feature}")
    
    print(f"\nWorkbook structure:")
    for i, sheet in enumerate(wb.worksheets, 1):
        print(f"  {i}. {sheet.title}")
        if hasattr(sheet, 'max_row') and sheet.max_row > 1:
            print(f"     - Data rows: {sheet.max_row - 1}")
        if hasattr(sheet, '_charts') and sheet._charts:
            print(f"     - Charts: {len(sheet._charts)}")
    
    print("Example 4 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"openpyxl example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished")
```

## Example 5: Advanced file I/O operations and data manipulation

```python
# Example 5: Advanced file I/O operations and data manipulation
import openpyxl
from openpyxl import Workbook, load_workbook
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.utils.dataframe import dataframe_to_rows
from openpyxl.utils import get_column_letter
from io import BytesIO
import tempfile
import os
import random
import datetime

print("Example 5 - Advanced file I/O operations and data manipulation")

try:
    # Create comprehensive data processing workflow
    print(f"✓ Creating comprehensive data processing workflow")
    
    # Step 1: Generate large dataset
    print(f"\n✓ Step 1: Generating large dataset for processing:")
    
    # Create master workbook with multiple data sources
    master_wb = Workbook()
    
    # Create customer data sheet
    customers_ws = master_wb.active
    customers_ws.title = "Customers"
    
    customer_headers = ['Customer ID', 'Name', 'Email', 'City', 'State', 'Join Date', 'Status']
    for col, header in enumerate(customer_headers, start=1):
        customers_ws.cell(row=1, column=col, value=header)
    
    # Generate customer data
    cities = ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix', 'Philadelphia']
    states = ['NY', 'CA', 'IL', 'TX', 'AZ', 'PA']
    statuses = ['Active', 'Inactive', 'Premium', 'Trial']
    
    customer_data = []
    for i in range(1, 101):  # 100 customers
        customer = [
            f"CUST{i:04d}",
            f"Customer {i}",
            f"customer{i}@example.com",
            random.choice(cities),
            random.choice(states),
            datetime.date(2023, random.randint(1, 12), random.randint(1, 28)),
            random.choice(statuses)
        ]
        customer_data.append(customer)
        
        for col, value in enumerate(customer, start=1):
            customers_ws.cell(row=i+1, column=col, value=value)
    
    print(f"  Generated {len(customer_data)} customer records")
    
    # Create orders data sheet
    orders_ws = master_wb.create_sheet("Orders")
    
    order_headers = ['Order ID', 'Customer ID', 'Product', 'Quantity', 'Unit Price', 'Order Date', 'Status']
    for col, header in enumerate(order_headers, start=1):
        orders_ws.cell(row=1, column=col, value=header)
    
    # Generate order data
    products = ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam', 'Speakers']
    prices = [999.99, 29.99, 79.99, 299.99, 59.99, 149.99]
    order_statuses = ['Pending', 'Shipped', 'Delivered', 'Cancelled']
    
    order_data = []
    for i in range(1, 251):  # 250 orders
        order = [
            f"ORD{i:04d}",
            f"CUST{random.randint(1, 100):04d}",
            random.choice(products),
            random.randint(1, 5),
            random.choice(prices),
            datetime.date(2024, random.randint(1, 8), random.randint(1, 28)),
            random.choice(order_statuses)
        ]
        order_data.append(order)
        
        for col, value in enumerate(order, start=1):
            orders_ws.cell(row=i+1, column=col, value=value)
    
    print(f"  Generated {len(order_data)} order records")
    
    # Create product inventory sheet
    inventory_ws = master_wb.create_sheet("Inventory")
    
    inventory_headers = ['Product', 'SKU', 'Category', 'Stock', 'Reorder Level', 'Supplier', 'Cost']
    for col, header in enumerate(inventory_headers, start=1):
        inventory_ws.cell(row=1, column=col, value=header)
    
    # Generate inventory data
    categories = ['Electronics', 'Accessories', 'Office Equipment']
    suppliers = ['TechCorp', 'OfficeSupply Inc', 'Digital Solutions', 'Hardware Plus']
    
    inventory_data = []
    for i, (product, price) in enumerate(zip(products, prices), start=1):
        inventory = [
            product,
            f"SKU{i:03d}",
            random.choice(categories),
            random.randint(10, 100),
            random.randint(5, 20),
            random.choice(suppliers),
            price * 0.7  # Cost is 70% of selling price
        ]
        inventory_data.append(inventory)
        
        for col, value in enumerate(inventory, start=1):
            inventory_ws.cell(row=i+1, column=col, value=value)
    
    print(f"  Generated {len(inventory_data)} inventory records")
    
    # Step 2: Save and load operations with different formats
    print(f"\n✓ Step 2: Testing multiple file I/O operations:")
    
    # Save to memory buffer
    buffer = BytesIO()
    master_wb.save(buffer)
    buffer.seek(0)
    
    print(f"  Saved master workbook to memory buffer: {len(buffer.getvalue())} bytes")
    
    # Load from buffer and verify data integrity
    loaded_wb = load_workbook(buffer)
    
    print(f"  Loaded workbook with {len(loaded_wb.worksheets)} worksheets:")
    for sheet in loaded_wb.worksheets:
        print(f"    • {sheet.title}: {sheet.max_row - 1} data rows")
    
    # Test data integrity
    customers_loaded = loaded_wb['Customers']
    orders_loaded = loaded_wb['Orders']
    inventory_loaded = loaded_wb['Inventory']
    
    # Verify specific data points
    customer_check = customers_loaded['A2'].value == customer_data[0][0]
    order_check = orders_loaded['A2'].value == order_data[0][0]
    inventory_check = inventory_loaded['A2'].value == inventory_data[0][0]
    
    print(f"  Data integrity checks: Customer={customer_check}, Orders={order_check}, Inventory={inventory_check}")
    assert customer_check and order_check and inventory_check
    
    # Step 3: Data manipulation and analysis
    print(f"\n✓ Step 3: Advanced data manipulation and analysis:")
    
    # Create analysis workbook
    analysis_wb = Workbook()
    analysis_ws = analysis_wb.active
    analysis_ws.title = "Data Analysis"
    
    # Customer Analysis
    print(f"  Performing customer analysis:")
    
    # Count customers by status
    status_counts = {}
    for i in range(2, customers_loaded.max_row + 1):
        status = customers_loaded.cell(row=i, column=7).value
        if status in status_counts:
            status_counts[status] += 1
        else:
            status_counts[status] = 1
    
    # Write customer analysis
    analysis_ws['A1'] = 'CUSTOMER ANALYSIS'
    analysis_ws['A1'].font = Font(bold=True, size=14)
    
    analysis_ws['A3'] = 'Customer Status Distribution:'
    analysis_ws['A3'].font = Font(bold=True)
    
    row = 4
    for status, count in status_counts.items():
        analysis_ws[f'A{row}'] = status
        analysis_ws[f'B{row}'] = count
        analysis_ws[f'C{row}'] = count / len(customer_data)  # Percentage
        analysis_ws[f'C{row}'].number_format = '0.00%'
        row += 1
    
    print(f"    Customer status distribution: {status_counts}")
    
    # Order Analysis
    print(f"  Performing order analysis:")
    
    # Calculate order metrics
    total_orders = len(order_data)
    total_revenue = 0
    orders_by_status = {}
    orders_by_product = {}
    
    for i in range(2, orders_loaded.max_row + 1):
        quantity = orders_loaded.cell(row=i, column=4).value or 0
        unit_price = orders_loaded.cell(row=i, column=5).value or 0
        status = orders_loaded.cell(row=i, column=7).value
        product = orders_loaded.cell(row=i, column=3).value
        
        total_revenue += quantity * unit_price
        
        # Count by status
        if status in orders_by_status:
            orders_by_status[status] += 1
        else:
            orders_by_status[status] = 1
        
        # Count by product
        if product in orders_by_product:
            orders_by_product[product] += quantity
        else:
            orders_by_product[product] = quantity
    
    # Write order analysis
    analysis_ws['A10'] = 'ORDER ANALYSIS'
    analysis_ws['A10'].font = Font(bold=True, size=14)
    
    metrics = [
        ('Total Orders:', total_orders),
        ('Total Revenue:', total_revenue),
        ('Average Order Value:', total_revenue / total_orders),
        ('Top Product:', max(orders_by_product, key=orders_by_product.get)),
        ('Top Product Sales:', max(orders_by_product.values()))
    ]
    
    row = 12
    for metric, value in metrics:
        analysis_ws[f'A{row}'] = metric
        analysis_ws[f'B{row}'] = value
        analysis_ws[f'A{row}'].font = Font(bold=True)
        
        if 'Revenue' in metric or 'Value' in metric:
            analysis_ws[f'B{row}'].number_format = '$#,##0.00'
        elif 'Orders' in metric or 'Sales' in metric:
            analysis_ws[f'B{row}'].number_format = '#,##0'
        
        row += 1
    
    print(f"    Total revenue: ${total_revenue:,.2f} from {total_orders} orders")
    print(f"    Top selling product: {max(orders_by_product, key=orders_by_product.get)}")
    
    # Inventory Analysis
    print(f"  Performing inventory analysis:")
    
    low_stock_items = []
    total_inventory_value = 0
    
    for i in range(2, inventory_loaded.max_row + 1):
        product = inventory_loaded.cell(row=i, column=1).value
        stock = inventory_loaded.cell(row=i, column=4).value or 0
        reorder_level = inventory_loaded.cell(row=i, column=5).value or 0
        cost = inventory_loaded.cell(row=i, column=7).value or 0
        
        total_inventory_value += stock * cost
        
        if stock <= reorder_level:
            low_stock_items.append((product, stock, reorder_level))
    
    # Write inventory analysis
    analysis_ws['A20'] = 'INVENTORY ANALYSIS'
    analysis_ws['A20'].font = Font(bold=True, size=14)
    
    analysis_ws['A22'] = f'Total Inventory Value: ${total_inventory_value:,.2f}'
    analysis_ws['A22'].font = Font(bold=True)
    
    analysis_ws['A24'] = f'Low Stock Alerts ({len(low_stock_items)} items):'
    analysis_ws['A24'].font = Font(bold=True)
    
    if low_stock_items:
        analysis_ws['A25'] = 'Product'
        analysis_ws['B25'] = 'Current Stock'
        analysis_ws['C25'] = 'Reorder Level'
        
        for i, (product, stock, reorder) in enumerate(low_stock_items, start=26):
            analysis_ws[f'A{i}'] = product
            analysis_ws[f'B{i}'] = stock
            analysis_ws[f'C{i}'] = reorder
            # Highlight low stock items
            analysis_ws[f'A{i}'].fill = PatternFill(start_color='FFCCCC', end_color='FFCCCC', fill_type='solid')
            analysis_ws[f'B{i}'].fill = PatternFill(start_color='FFCCCC', end_color='FFCCCC', fill_type='solid')
            analysis_ws[f'C{i}'].fill = PatternFill(start_color='FFCCCC', end_color='FFCCCC', fill_type='solid')
    
    print(f"    Inventory value: ${total_inventory_value:,.2f}")
    print(f"    Low stock alerts: {len(low_stock_items)} items need reordering")
    
    # Step 4: Data export and consolidation
    print(f"\n✓ Step 4: Data consolidation and export operations:")
    
    # Create consolidated report
    report_ws = analysis_wb.create_sheet("Consolidated Report")
    
    # Executive summary
    report_ws['A1'] = 'EXECUTIVE DASHBOARD'
    report_ws['A1'].font = Font(bold=True, size=16)
    report_ws['A1'].fill = PatternFill(start_color='366092', end_color='366092', fill_type='solid')
    report_ws['A1'].font = Font(bold=True, size=16, color='FFFFFF')
    
    # Key metrics
    key_metrics = [
        ('Total Customers:', len(customer_data)),
        ('Active Customers:', status_counts.get('Active', 0)),
        ('Total Orders:', total_orders),
        ('Total Revenue:', total_revenue),
        ('Inventory Value:', total_inventory_value),
        ('Items Needing Reorder:', len(low_stock_items))
    ]
    
    row = 3
    for metric, value in key_metrics:
        report_ws[f'A{row}'] = metric
        report_ws[f'B{row}'] = value
        report_ws[f'A{row}'].font = Font(bold=True)
        
        if 'Revenue' in metric or 'Value' in metric:
            report_ws[f'B{row}'].number_format = '$#,##0.00'
        else:
            report_ws[f'B{row}'].number_format = '#,##0'
        
        row += 1
    
    # Test multiple save/load cycles
    print(f"  Testing multiple save/load cycles:")
    
    # Save analysis workbook
    analysis_buffer = BytesIO()
    analysis_wb.save(analysis_buffer)
    analysis_buffer.seek(0)
    
    # Reload and verify
    reloaded_analysis = load_workbook(analysis_buffer)
    
    # Verify all sheets are present
    expected_sheets = ['Data Analysis', 'Consolidated Report']
    loaded_sheets = [sheet.title for sheet in reloaded_analysis.worksheets]
    
    print(f"    Expected sheets: {expected_sheets}")
    print(f"    Loaded sheets: {loaded_sheets}")
    print(f"    Sheet integrity: {set(expected_sheets) == set(loaded_sheets)}")
    
    # Test data filtering and selection
    print(f"\n✓ Step 5: Advanced data filtering and manipulation:")
    
    # Filter high-value customers (Premium status)
    premium_customers = []
    for i in range(2, customers_loaded.max_row + 1):
        if customers_loaded.cell(row=i, column=7).value == 'Premium':
            customer_id = customers_loaded.cell(row=i, column=1).value
            customer_name = customers_loaded.cell(row=i, column=2).value
            premium_customers.append((customer_id, customer_name))
    
    print(f"  Premium customers: {len(premium_customers)}")
    
    # Filter high-value orders (>$500)
    high_value_orders = []
    for i in range(2, orders_loaded.max_row + 1):
        quantity = orders_loaded.cell(row=i, column=4).value or 0
        unit_price = orders_loaded.cell(row=i, column=5).value or 0
        total_value = quantity * unit_price
        
        if total_value > 500:
            order_id = orders_loaded.cell(row=i, column=1).value
            customer_id = orders_loaded.cell(row=i, column=2).value
            high_value_orders.append((order_id, customer_id, total_value))
    
    print(f"  High-value orders (>$500): {len(high_value_orders)}")
    
    # Create filtered data worksheet
    filtered_ws = analysis_wb.create_sheet("Filtered Data")
    
    # Write premium customers
    filtered_ws['A1'] = 'Premium Customers'
    filtered_ws['A1'].font = Font(bold=True)
    filtered_ws['A2'] = 'Customer ID'
    filtered_ws['B2'] = 'Customer Name'
    
    for i, (cust_id, name) in enumerate(premium_customers[:10], start=3):  # Show first 10
        filtered_ws[f'A{i}'] = cust_id
        filtered_ws[f'B{i}'] = name
    
    # Write high-value orders
    filtered_ws['D1'] = 'High-Value Orders'
    filtered_ws['D1'].font = Font(bold=True)
    filtered_ws['D2'] = 'Order ID'
    filtered_ws['E2'] = 'Customer ID'
    filtered_ws['F2'] = 'Order Value'
    
    for i, (order_id, cust_id, value) in enumerate(high_value_orders[:10], start=3):
        filtered_ws[f'D{i}'] = order_id
        filtered_ws[f'E{i}'] = cust_id
        filtered_ws[f'F{i}'] = value
        filtered_ws[f'F{i}'].number_format = '$#,##0.00'
    
    print(f"  Created filtered data worksheet with premium customers and high-value orders")
    
    # Final statistics
    print(f"\n✓ Final processing statistics:")
    
    total_records_processed = len(customer_data) + len(order_data) + len(inventory_data)
    total_sheets_created = len(analysis_wb.worksheets)
    
    processing_stats = [
        f"Total records processed: {total_records_processed:,}",
        f"Total worksheets created: {total_sheets_created}",
        f"Data integrity checks: All passed",
        f"Memory efficiency: Buffer-based operations",
        f"Analysis features: Status counts, revenue analysis, inventory alerts",
        f"Export capabilities: Multiple sheet formats, formatted reports"
    ]
    
    for stat in processing_stats:
        print(f"    • {stat}")
    
    print(f"\nFinal workbook structure ({total_sheets_created} sheets):")
    for i, sheet in enumerate(analysis_wb.worksheets, 1):
        print(f"  {i}. {sheet.title}")
        if hasattr(sheet, 'max_row') and sheet.max_row > 1:
            print(f"     - Rows with data: {sheet.max_row}")
    
    print("Example 5 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"openpyxl example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished")
```
