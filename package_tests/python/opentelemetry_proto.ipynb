## Example 1: Common protobuf data types

```python
from opentelemetry.proto.common.v1 import common_pb2

print("Example 1 - Common protobuf data types")

# Create a KeyValue pair
kv = common_pb2.KeyValue()
kv.key = "service.name"
kv.value.string_value = "my-service"

print(f"Created KeyValue: {kv.key} = {kv.value.string_value}")
# A serialize/parse round trip returns an equal message
assert common_pb2.KeyValue.FromString(kv.SerializeToString()) == kv
assert kv.value.WhichOneof("value") == "string_value"

# Create an AnyValue with different types
int_value = common_pb2.AnyValue()
int_value.int_value = 42

print(f"Integer value: {int_value.int_value}")

# Create an InstrumentationScope
scope = common_pb2.InstrumentationScope()
scope.name = "my-library"
scope.version = "1.0.0"

print(f"Scope: {scope.name} v{scope.version}")
print("Example 1 completed successfully")
```

## Example 2: Trace protobuf structures

```python
from opentelemetry.proto.trace.v1 import trace_pb2

print("Example 2 - Trace protobuf structures")

# Create a Span
span = trace_pb2.Span()
span.name = "my-operation"
span.trace_id = b'\x01\x02\x03\x04\x05\x06\x07\x08\x09\x0a\x0b\x0c\x0d\x0e\x0f\x10'
span.span_id = b'\x01\x02\x03\x04\x05\x06\x07\x08'

print(f"Span name: {span.name}")
print(f"Trace ID length: {len(span.trace_id)} bytes")

# Create a Status
status = trace_pb2.Status()
status.code = trace_pb2.Status.STATUS_CODE_OK
status.message = "Success"

print(f"Status: {status.code} - {status.message}")

# Create ResourceSpans
resource_spans = trace_pb2.ResourceSpans()
print(f"Created ResourceSpans: {type(resource_spans).__name__}")
print("Example 2 completed successfully")
```

## Example 3: Metrics protobuf structures

```python
from opentelemetry.proto.metrics.v1 import metrics_pb2

print("Example 3 - Metrics protobuf structures")

# Create a Metric
metric = metrics_pb2.Metric()
metric.name = "http_requests_total"
metric.description = "Total HTTP requests"
metric.unit = "1"

print(f"Metric: {metric.name}")
print(f"Description: {metric.description}")

# Create an Exemplar
exemplar = metrics_pb2.Exemplar()
exemplar.time_unix_nano = 1234567890000000000

print(f"Exemplar timestamp: {exemplar.time_unix_nano}")

# Test aggregation temporality
temporality = metrics_pb2.AGGREGATION_TEMPORALITY_CUMULATIVE
print(f"Temporality: {temporality}")

print("Example 3 completed successfully")
```

## Example 4: Resource protobuf structures

```python
from opentelemetry.proto.resource.v1 import resource_pb2
from opentelemetry.proto.common.v1 import common_pb2

print("Example 4 - Resource protobuf structures")

# Create a Resource with attributes
resource = resource_pb2.Resource()

# Add service name attribute
kv1 = common_pb2.KeyValue()
kv1.key = "service.name"
kv1.value.string_value = "payment-service"
resource.attributes.append(kv1)

# Add service version attribute
kv2 = common_pb2.KeyValue()
kv2.key = "service.version"
kv2.value.string_value = "1.2.3"
resource.attributes.append(kv2)

print(f"Resource attributes count: {len(resource.attributes)}")
for attr in resource.attributes:
    print(f"  {attr.key}: {attr.value.string_value}")

print("Example 4 completed successfully")
```

## Example 5: Combined protobuf structures

```python
from opentelemetry.proto.trace.v1 import trace_pb2
from opentelemetry.proto.common.v1 import common_pb2
from opentelemetry.proto.resource.v1 import resource_pb2

print("Example 5 - Combined protobuf structures")

# Create span with attributes
span = trace_pb2.Span()
span.name = "database-query"

# Add span attribute
attr = common_pb2.KeyValue()
attr.key = "db.system"
attr.value.string_value = "postgresql"
span.attributes.append(attr)

print(f"Span: {span.name}")
print(f"Attributes: {span.attributes[0].key} = {span.attributes[0].value.string_value}")
parsed = trace_pb2.Span.FromString(span.SerializeToString())
assert parsed.name == "database-query"
assert parsed.attributes[0].value.string_value == "postgresql"

# Create resource
resource = resource_pb2.Resource()
resource.dropped_attributes_count = 0

print(f"Resource dropped count: {resource.dropped_attributes_count}")
print("Example 5 completed successfully")
```
