```python
# Example 1: Import the primary detector classes and verify they are ResourceDetector subclasses.
from opentelemetry.sdk.resources import ResourceDetector
from opentelemetry.resource.detector.azure.app_service import AzureAppServiceResourceDetector
from opentelemetry.resource.detector.azure.vm import AzureVMResourceDetector

assert issubclass(AzureAppServiceResourceDetector, ResourceDetector)
assert issubclass(AzureVMResourceDetector, ResourceDetector)

print(f"AzureAppServiceResourceDetector: {AzureAppServiceResourceDetector}")
print(f"AzureVMResourceDetector: {AzureVMResourceDetector}")
```

```python
# Example 2: Instantiate AzureAppServiceResourceDetector and call detect() without Azure env vars.
# Outside an App Service environment, expect an empty (or near-empty) resource attribute map.
import os
from opentelemetry.resource.detector.azure.app_service import AzureAppServiceResourceDetector

# Ensure App Service env vars are NOT set so detect() returns an empty/default Resource.
for key in ('WEBSITE_SITE_NAME', 'WEBSITE_OWNER_NAME', 'WEBSITE_RESOURCE_GROUP'):
    os.environ.pop(key, None)

detector = AzureAppServiceResourceDetector()
resource = detector.detect()
print(f"Resource type: {type(resource).__name__}")
print(f"Attributes (no Azure env): {dict(resource.attributes)}")
assert hasattr(resource, 'attributes')
```

```python
# Example 3: AzureVMResourceDetector.detect() outside a VM environment.
# Without IMDS reachable, detect() should return a Resource (possibly empty) rather than raise.
from opentelemetry.resource.detector.azure.vm import AzureVMResourceDetector
from opentelemetry.sdk.resources import Resource

detector = AzureVMResourceDetector()
resource = detector.detect()
print(f"VM detector returned: {type(resource).__name__}")
print(f"Attributes: {dict(resource.attributes)}")
assert isinstance(resource, Resource)
```
