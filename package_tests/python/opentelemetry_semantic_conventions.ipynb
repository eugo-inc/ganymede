## Example 1: HTTP Semantic Conventions

```python
from opentelemetry.semconv.trace import SpanAttributes
from opentelemetry.sdk.trace import TracerProvider
from opentelemetry import trace

print("Example 1 - HTTP Semantic Conventions")

# Setup tracer
trace.set_tracer_provider(TracerProvider())
tracer = trace.get_tracer(__name__)

# Create HTTP span with semantic conventions
with tracer.start_as_current_span("http-request") as span:
    span.set_attribute(SpanAttributes.HTTP_METHOD, "GET")
    span.set_attribute(SpanAttributes.HTTP_URL, "https://api.example.com/users")
    span.set_attribute(SpanAttributes.HTTP_STATUS_CODE, 200)
    span.set_attribute(SpanAttributes.HTTP_HOST, "api.example.com")
    
    print(f"HTTP Method: {span.attributes.get(SpanAttributes.HTTP_METHOD)}")
    print(f"Status Code: {span.attributes.get(SpanAttributes.HTTP_STATUS_CODE)}")
    assert span.attributes[SpanAttributes.HTTP_METHOD] == "GET"
    assert span.attributes[SpanAttributes.HTTP_STATUS_CODE] == 200

print("Example 1 completed successfully")
```

## Example 2: Resource Semantic Conventions

```python
from opentelemetry.semconv.resource import ResourceAttributes
from opentelemetry.sdk.resources import Resource
from opentelemetry.sdk.trace import TracerProvider

print("Example 2 - Resource Semantic Conventions")

# Create resource with semantic conventions
resource = Resource.create({
    ResourceAttributes.SERVICE_NAME: "my-api-service",
    ResourceAttributes.SERVICE_VERSION: "1.2.3",
    ResourceAttributes.SERVICE_NAMESPACE: "production",
    ResourceAttributes.SERVICE_INSTANCE_ID: "instance-001"
})

print(f"Service: {resource.attributes.get(ResourceAttributes.SERVICE_NAME)}")
print(f"Version: {resource.attributes.get(ResourceAttributes.SERVICE_VERSION)}")
print(f"Namespace: {resource.attributes.get(ResourceAttributes.SERVICE_NAMESPACE)}")
assert resource.attributes[ResourceAttributes.SERVICE_NAME] == "my-api-service"

provider = TracerProvider(resource=resource)
print("TracerProvider created with resource")
print("Example 2 completed successfully")
```

## Example 3: Database Semantic Conventions

```python
from opentelemetry.semconv.trace import SpanAttributes
from opentelemetry import trace

print("Example 3 - Database Semantic Conventions")

tracer = trace.get_tracer(__name__)

# Create database span with semantic conventions
with tracer.start_as_current_span("db-query") as span:
    span.set_attribute(SpanAttributes.DB_SYSTEM, "postgresql")
    span.set_attribute(SpanAttributes.DB_NAME, "users_db")
    span.set_attribute(SpanAttributes.DB_STATEMENT, "SELECT * FROM users WHERE id = $1")
    span.set_attribute(SpanAttributes.DB_OPERATION, "SELECT")
    span.set_attribute(SpanAttributes.DB_SQL_TABLE, "users")
    
    print(f"DB System: {span.attributes.get(SpanAttributes.DB_SYSTEM)}")
    print(f"Operation: {span.attributes.get(SpanAttributes.DB_OPERATION)}")
    print(f"Table: {span.attributes.get(SpanAttributes.DB_SQL_TABLE)}")

print("Example 3 completed successfully")
```

## Example 4: RPC Semantic Conventions

```python
from opentelemetry.semconv.trace import SpanAttributes
from opentelemetry import trace

print("Example 4 - RPC Semantic Conventions")

tracer = trace.get_tracer(__name__)

# Create RPC span with semantic conventions
with tracer.start_as_current_span("rpc-call") as span:
    span.set_attribute(SpanAttributes.RPC_SYSTEM, "grpc")
    span.set_attribute(SpanAttributes.RPC_SERVICE, "user.UserService")
    span.set_attribute(SpanAttributes.RPC_METHOD, "GetUser")
    span.set_attribute(SpanAttributes.RPC_GRPC_STATUS_CODE, 0)
    
    print(f"RPC System: {span.attributes.get(SpanAttributes.RPC_SYSTEM)}")
    print(f"Service: {span.attributes.get(SpanAttributes.RPC_SERVICE)}")
    print(f"Method: {span.attributes.get(SpanAttributes.RPC_METHOD)}")
    print(f"Status: {span.attributes.get(SpanAttributes.RPC_GRPC_STATUS_CODE)}")

print("Example 4 completed successfully")
```

## Example 5: Network Semantic Conventions

```python
from opentelemetry.semconv.trace import SpanAttributes
from opentelemetry import trace

print("Example 5 - Network Semantic Conventions")

tracer = trace.get_tracer(__name__)

# Create network span with semantic conventions
with tracer.start_as_current_span("network-request") as span:
    span.set_attribute(SpanAttributes.NET_PEER_NAME, "api.service.com")
    span.set_attribute(SpanAttributes.NET_PEER_PORT, 443)
    span.set_attribute(SpanAttributes.NET_TRANSPORT, "ip_tcp")
    span.set_attribute(SpanAttributes.NET_HOST_NAME, "localhost")
    span.set_attribute(SpanAttributes.NET_HOST_PORT, 8080)
    
    print(f"Peer: {span.attributes.get(SpanAttributes.NET_PEER_NAME)}:{span.attributes.get(SpanAttributes.NET_PEER_PORT)}")
    print(f"Transport: {span.attributes.get(SpanAttributes.NET_TRANSPORT)}")

print("Example 5 completed successfully")
```
