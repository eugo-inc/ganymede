## Example 1: Basic AI Semantic Conventions

```python
from opentelemetry.semconv_ai import SpanAttributes
from opentelemetry.sdk.trace import TracerProvider
from opentelemetry import trace

print("Example 1 - Basic AI Semantic Conventions")

# Setup tracer
trace.set_tracer_provider(TracerProvider())
tracer = trace.get_tracer(__name__)

# Create span with LLM attributes
with tracer.start_as_current_span("llm-completion") as span:
    span.set_attribute(SpanAttributes.LLM_REQUEST_MODEL, "gpt-4")
    span.set_attribute(SpanAttributes.LLM_REQUEST_MAX_TOKENS, 100)
    span.set_attribute(SpanAttributes.LLM_REQUEST_TEMPERATURE, 0.7)
    span.set_attribute(SpanAttributes.LLM_USAGE_PROMPT_TOKENS, 50)
    span.set_attribute(SpanAttributes.LLM_USAGE_COMPLETION_TOKENS, 30)
    
    print("LLM span with semantic conventions created")

print("Example 1 completed successfully")
```

## Example 2: AI Model Monitoring Conventions

```python
from opentelemetry.semconv_ai import SpanAttributes
from opentelemetry import trace

print("Example 2 - AI Model Monitoring Conventions")

tracer = trace.get_tracer(__name__)

# Track streaming response
with tracer.start_as_current_span("streaming-response") as span:
    span.set_attribute(SpanAttributes.LLM_IS_STREAMING, True)
    span.set_attribute(SpanAttributes.LLM_REQUEST_MODEL, "claude-3")
    span.set_attribute(SpanAttributes.LLM_REQUEST_TOP_P, 0.9)
    span.set_attribute(SpanAttributes.LLM_FREQUENCY_PENALTY, 0.5)
    span.set_attribute(SpanAttributes.LLM_PRESENCE_PENALTY, 0.3)
    
    print(f"Streaming: {span.attributes.get(SpanAttributes.LLM_IS_STREAMING)}")
    print(f"Model: {span.attributes.get(SpanAttributes.LLM_REQUEST_MODEL)}")

print("Example 2 completed successfully")
```

## Example 3: Vector Database Observability

```python
from opentelemetry.semconv_ai import SpanAttributes
from opentelemetry import trace

print("Example 3 - Vector Database Observability")

tracer = trace.get_tracer(__name__)

# Track ChromaDB operations
with tracer.start_as_current_span("chromadb-add") as span:
    span.set_attribute(SpanAttributes.CHROMADB_ADD_DOCUMENTS_COUNT, 10)
    span.set_attribute(SpanAttributes.CHROMADB_ADD_EMBEDDINGS_COUNT, 10)
    span.set_attribute(SpanAttributes.CHROMADB_ADD_IDS_COUNT, 10)
    
    print(f"Documents added: {span.attributes.get(SpanAttributes.CHROMADB_ADD_DOCUMENTS_COUNT)}")

# Track query operations
with tracer.start_as_current_span("chromadb-query") as span:
    span.set_attribute(SpanAttributes.CHROMADB_QUERY_N_RESULTS, 5)
    print(f"Query n_results: {span.attributes.get(SpanAttributes.CHROMADB_QUERY_N_RESULTS)}")

print("Example 3 completed successfully")
```

## Example 4: LLM Response Tracking

```python
from opentelemetry.semconv_ai import SpanAttributes
from opentelemetry import trace

print("Example 4 - LLM Response Tracking")

tracer = trace.get_tracer(__name__)

# Track completion with usage metrics
with tracer.start_as_current_span("llm-tracking") as span:
    span.set_attribute(SpanAttributes.LLM_REQUEST_MODEL, "text-davinci-003")
    span.set_attribute(SpanAttributes.LLM_USAGE_TOTAL_TOKENS, 150)
    span.set_attribute(SpanAttributes.LLM_USAGE_PROMPT_TOKENS, 100)
    span.set_attribute(SpanAttributes.LLM_USAGE_COMPLETION_TOKENS, 50)
    span.set_attribute(SpanAttributes.LLM_RESPONSE_FINISH_REASON, "stop")
    
    total_tokens = span.attributes.get(SpanAttributes.LLM_USAGE_TOTAL_TOKENS)
    print(f"Total tokens used: {total_tokens}")
    assert total_tokens == 150
    print(f"Finish reason: {span.attributes.get(SpanAttributes.LLM_RESPONSE_FINISH_REASON)}")

print("Example 4 completed successfully")
```

## Example 5: Advanced LLM Configuration

```python
from opentelemetry.semconv_ai import SpanAttributes
from opentelemetry import trace

print("Example 5 - Advanced LLM Configuration")

tracer = trace.get_tracer(__name__)

# Track advanced LLM parameters
with tracer.start_as_current_span("advanced-llm") as span:
    span.set_attribute(SpanAttributes.LLM_REQUEST_MODEL, "gpt-4-turbo")
    span.set_attribute(SpanAttributes.LLM_REQUEST_MAX_TOKENS, 4096)
    span.set_attribute(SpanAttributes.LLM_REQUEST_TEMPERATURE, 0.8)
    span.set_attribute(SpanAttributes.LLM_REQUEST_TOP_P, 0.95)
    span.set_attribute(SpanAttributes.LLM_TOP_K, 40)
    span.set_attribute(SpanAttributes.LLM_SYSTEM, "openai")  # the model vendor (gen_ai.system)
    
    print(f"Model: {span.attributes.get(SpanAttributes.LLM_REQUEST_MODEL)}")
    print(f"Max tokens: {span.attributes.get(SpanAttributes.LLM_REQUEST_MAX_TOKENS)}")
    print(f"Temperature: {span.attributes.get(SpanAttributes.LLM_REQUEST_TEMPERATURE)}")
    assert span.attributes.get(SpanAttributes.LLM_SYSTEM) == "openai"

print("Example 5 completed successfully")
```
