## Example 1: Basic optimized einsum operations

```python
import opt_einsum as oe
import numpy as np

print("Example 1 - Basic optimized einsum operations")

# Create test arrays
a = np.random.rand(10, 10)
b = np.random.rand(10, 15)
c = np.random.rand(15, 20)

print(f"Array shapes: a{a.shape}, b{b.shape}, c{c.shape}")

# Matrix multiplication using contract
result1 = oe.contract('ij,jk->ik', a, b)
print(f"Matrix multiply: {result1.shape}")

# Chain multiplication
result2 = oe.contract('ij,jk,kl->il', a, b, c)
print(f"Chain multiply: {result2.shape}")
assert np.allclose(result2, a @ b @ c)

# Get optimization path
path_info = oe.contract_path('ij,jk,kl->il', a, b, c)
print(f"Optimization path: {path_info[0]}")

print("Example 1 completed successfully")
```

## Example 2: Contraction path optimization

```python
import opt_einsum as oe
import numpy as np

print("Example 2 - Contraction path optimization")

# Create tensors
A = np.random.rand(50, 30)
B = np.random.rand(30, 40)
C = np.random.rand(40, 25)

# Define contraction
expression = 'ij,jk,kl->il'

# Test different optimization strategies
for strategy in ['greedy', 'optimal']:
    path, info = oe.contract_path(expression, A, B, C, optimize=strategy)
    print(f"{strategy}: path={path}, cost={info.opt_cost}")

# Perform optimized contraction
result = oe.contract(expression, A, B, C, optimize='optimal')
print(f"Result shape: {result.shape}")

print("Example 2 completed successfully")
```

## Example 3: Tensor operations

```python
import opt_einsum as oe
import numpy as np

print("Example 3 - Tensor operations")

# Batch matrix multiplication
batch_a = np.random.rand(3, 10, 8)
batch_b = np.random.rand(3, 8, 12)
result1 = oe.contract('bij,bjk->bik', batch_a, batch_b)
print(f"Batch matmul: {result1.shape}")

# Outer product
vec1 = np.random.rand(4)
vec2 = np.random.rand(6)
result2 = oe.contract('i,j->ij', vec1, vec2)
print(f"Outer product: {result2.shape}")

# Trace
matrix = np.random.rand(10, 10)
trace = oe.contract('ii->', matrix)
print(f"Trace: {trace}")
assert np.isclose(trace, np.trace(matrix))

# Tensor dot
t1 = np.random.rand(3, 4, 5)
t2 = np.random.rand(5, 6, 7)
result3 = oe.contract('ijk,klm->ijlm', t1, t2)
print(f"Tensor dot: {result3.shape}")

print("Example 3 completed successfully")
```

## Example 4: Performance comparison

```python
import opt_einsum as oe
import numpy as np
import time

print("Example 4 - Performance comparison")

# Create test tensors
n = 20
A = np.random.rand(n, n, n)
B = np.random.rand(n, n, n)
C = np.random.rand(n, n, n)

expression = 'ijk,jkl,lmi->m'

# Compare with numpy
start = time.time()
numpy_result = np.einsum(expression, A, B, C)
numpy_time = time.time() - start

# Optimized version
start = time.time()
opt_result = oe.contract(expression, A, B, C, optimize='optimal')
opt_time = time.time() - start

print(f"NumPy time: {numpy_time:.4f}s")
print(f"Optimized time: {opt_time:.4f}s")
print(f"Results match: {np.allclose(numpy_result, opt_result)}")
assert np.allclose(numpy_result, opt_result)

print("Example 4 completed successfully")
```

## Example 5: Advanced tensor operations

```python
import opt_einsum as oe
import numpy as np

print("Example 5 - Advanced tensor operations")

# Attention mechanism
queries = np.random.rand(10, 64)
keys = np.random.rand(20, 64)
values = np.random.rand(20, 32)

attention_weights = oe.contract('qd,kd->qk', queries, keys)
attention_weights = attention_weights / np.sum(attention_weights, axis=1, keepdims=True)
attended = oe.contract('qk,kv->qv', attention_weights, values)
print(f"Attention output: {attended.shape}")

# Graph message passing
adj = np.random.rand(10, 10)
features = np.random.rand(10, 16)
weight = np.random.rand(16, 32)
messages = oe.contract('ij,jk,kl->il', adj, features, weight)
print(f"Graph messages: {messages.shape}")

# Contract expression for reuse
expr = oe.contract_expression('ij,jk->ik', (10, 10), (10, 15))
print(f"Expression created: {type(expr)}")

print("Example 5 completed successfully")
```
