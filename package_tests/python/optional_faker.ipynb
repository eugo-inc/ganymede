## Example 1: Basic none_or with direct values

```python
from faker import Faker
import optional_faker

# Initialize Faker with optional_faker wrapper
fake = Faker()
Faker.seed(42)

# Use none_or with pre-generated values - may return None or the value
print("Testing none_or with direct values:")
for i in range(10):
    # Each call randomly returns either the value or None
    name = fake.none_or(fake.name())
    email = fake.none_or(fake.email())
    print(f"  Person {i+1}: {name} - {email}")

# Verify that some values are None and some are not
results = [fake.none_or(fake.pystr()) for _ in range(100)]
none_count = sum(1 for r in results if r is None)
value_count = sum(1 for r in results if r is not None)
print(f"\nOut of 100 calls: {none_count} None, {value_count} values")
assert none_count + value_count == 100
assert 0 < none_count < 100  # none_or returns None for some calls and the value for others
```

## Example 2: Using none_or with callables and arguments

```python
from faker import Faker
import optional_faker

# Initialize Faker
fake = Faker()
Faker.seed(123)

# Use none_or with callables and their arguments
print("Testing none_or with callables:")

# Passing method as callable with positional and keyword arguments
for i in range(8):
    # pystr can take min_chars and max_chars arguments
    result = fake.none_or(fake.pystr, min_chars=5, max_chars=15)
    print(f"  Random string {i+1}: {result}")

# Works with various Faker methods
print("\nDifferent data types:")
phone = fake.none_or(fake.phone_number)
address = fake.none_or(fake.address)
date = fake.none_or(fake.date_between, start_date='-30y', end_date='today')
print(f"  Phone: {phone}")
print(f"  Address: {address}")
print(f"  Date: {date}")
```

## Example 3: Generating optional data for realistic datasets

```python
from faker import Faker
import optional_faker

# Create realistic user profiles with optional fields
fake = Faker()
Faker.seed(999)

print("Generating user profiles with optional fields:")
users = []
for i in range(5):
    user = {
        'id': i + 1,
        'name': fake.name(),  # Required field
        'email': fake.email(),  # Required field
        'phone': fake.none_or(fake.phone_number),  # Optional
        'company': fake.none_or(fake.company),  # Optional
        'job_title': fake.none_or(fake.job),  # Optional
        'website': fake.none_or(fake.url),  # Optional
    }
    users.append(user)
    print(f"\nUser {user['id']}:")
    for key, value in user.items():
        print(f"  {key}: {value}")
```

## Example 4: none_or with custom callables (non-Faker)

```python
from faker import Faker
import optional_faker
import random

# none_or works with any callable, not just Faker methods
fake = Faker()
Faker.seed(555)

# Define custom functions
def custom_id():
    return f"ID-{random.randint(1000, 9999)}"

def format_percentage(value):
    return f"{value}%"

print("Testing none_or with custom callables:")

# Use none_or with custom functions
for i in range(8):
    # Custom ID generator
    user_id = fake.none_or(custom_id)
    # Custom formatter with arguments
    discount = fake.none_or(format_percentage, random.randint(5, 50))
    # Mix with Faker methods
    product = fake.none_or(fake.word)
    
    print(f"  Entry {i+1}: ID={user_id}, Product={product}, Discount={discount}")
```

## Example 5: Testing Provider classes and Generator

```python
from faker import Faker
import optional_faker
from optional_faker import Provider, Generator

# Verify Provider and Generator classes are properly exposed
fake = Faker()
Faker.seed(777)

print("Testing optional_faker Provider classes:")
print(f"  Provider class available: {Provider is not None}")
print(f"  Generator class available: {Generator is not None}")

# Create a custom provider using optional_faker's Provider
class CustomOptionalProvider(Provider):
    def optional_color(self):
        colors = ['red', 'blue', 'green', 'yellow', 'purple']
        return self.random_element(colors)

# Add custom provider and use none_or with it
fake.add_provider(CustomOptionalProvider)
print("\nUsing custom provider with none_or:")
for i in range(10):
    color = fake.none_or(fake.optional_color)
    assert color in (None, "red", "blue", "green", "yellow", "purple")
    print(f"  Color {i+1}: {color}")
```
