## Example 1: Basic tree flattening and unflattening

```python
# Example 1: Basic tree flattening and unflattening
import optree

print("Example 1: OpTree - Basic Flatten/Unflatten\n")

# Create a nested PyTree structure
print("Creating nested PyTree:")
tree = {
    'a': 1,
    'b': (2, [3, 4]),
    'c': 5,
    'd': {'e': 6, 'f': [7, 8, 9]}
}
print(f"  Original tree: {tree}\n")

# Flatten the tree
print("Flattening tree:")
leaves, treespec = optree.tree_flatten(tree)
print(f"  Leaves: {leaves}")
print(f"  TreeSpec: {treespec}")
print(f"  Number of leaves: {len(leaves)}\n")

# Unflatten back to original structure
print("Unflattening tree:")
reconstructed = optree.tree_unflatten(treespec, leaves)
print(f"  Reconstructed: {reconstructed}")
print(f"  Match original: {reconstructed == tree}\n")
assert reconstructed == tree
assert leaves == [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Flatten with different structures
print("Flattening different structures:")
list_tree = [1, [2, 3], [[4, 5], 6]]
list_leaves, list_spec = optree.tree_flatten(list_tree)
print(f"  List tree: {list_tree}")
print(f"  Flattened: {list_leaves}")

tuple_tree = (1, (2, 3), ((4, 5), 6))
tuple_leaves, tuple_spec = optree.tree_flatten(tuple_tree)
print(f"\n  Tuple tree: {tuple_tree}")
print(f"  Flattened: {tuple_leaves}")

print("\nOpTree efficiently flattens nested container structures")
```

## Example 2: Tree mapping operations

```python
# Example 2: Tree mapping operations
import optree

print("Example 2: Tree Mapping Operations\n")

# Create a tree with numerical values
print("Creating numerical tree:")
tree = {
    'x': [1, 2, 3],
    'y': (4, 5),
    'z': {'a': 6, 'b': [7, 8]}
}
print(f"  Original: {tree}\n")

# Map a function over the tree
print("Applying lambda x: x * 2:")
doubled = optree.tree_map(lambda x: x * 2, tree)
print(f"  Result: {doubled}\n")

# Map with square function
print("Applying lambda x: x ** 2:")
squared = optree.tree_map(lambda x: x ** 2, tree)
print(f"  Result: {squared}\n")

# Map over multiple trees
print("Mapping over two trees:")
tree1 = {'a': [1, 2], 'b': 3}
tree2 = {'a': [10, 20], 'b': 30}
print(f"  Tree1: {tree1}")
print(f"  Tree2: {tree2}")

result = optree.tree_map(lambda x, y: x + y, tree1, tree2)
print(f"  Sum: {result}\n")

# Map with custom function
print("Applying custom function (x * 10 + 5):")
custom = optree.tree_map(lambda x: x * 10 + 5, tree)
print(f"  Result: {custom}")

print("\nOpTree tree_map applies functions to all leaves")
```

## Example 3: Tree leaves extraction and structure operations

```python
# Example 3: Tree leaves extraction and structure operations
import optree

print("Example 3: Tree Leaves and Structure Operations\n")

# Create a complex tree
print("Creating complex tree:")
tree = {
    'layer1': {
        'weights': [1.0, 2.0, 3.0],
        'bias': [0.1, 0.2]
    },
    'layer2': {
        'weights': [[4.0, 5.0], [6.0, 7.0]],
        'bias': [0.3]
    }
}
print(f"  Tree: {tree}\n")

# Extract all leaves
print("Extracting leaves:")
leaves = optree.tree_leaves(tree)
print(f"  All leaves: {leaves}")
print(f"  Number of leaves: {len(leaves)}\n")

# Get tree structure
print("Getting tree structure:")
structure = optree.tree_structure(tree)
print(f"  Structure: {structure}")
print(f"  Num leaves: {structure.num_leaves}")
print(f"  Num nodes: {structure.num_nodes}\n")

# Flatten with paths
print("Flattening with paths:")
paths, values, spec = optree.tree_flatten_with_path(tree)
print(f"  Number of paths: {len(paths)}")
for path, value in zip(paths[:5], values[:5]):
    print(f"    {path}: {value}")

# Count leaves in different trees
print("\nCounting leaves in different structures:")
trees = [
    [1, 2, 3],
    {'a': 1, 'b': [2, 3]},
    ((1, 2), (3, (4, 5)))
]
for t in trees:
    count = len(optree.tree_leaves(t))
    print(f"  {t}: {count} leaves")

print("\nOpTree provides utilities for tree introspection")
```

## Example 4: Tree reduction and aggregation operations

```python
# Example 4: Tree reduction and aggregation operations
import optree
from functools import reduce
import operator

print("Example 4: Tree Reduction and Aggregation\n")

# Create a numerical tree
print("Creating numerical tree:")
tree = {
    'group1': [10, 20, 30],
    'group2': (40, 50),
    'group3': {'a': 60, 'b': [70, 80, 90]}
}
print(f"  Tree: {tree}\n")

# Sum all leaves using tree_reduce
print("Computing sum of all leaves:")
total_sum = optree.tree_reduce(operator.add, tree)
print(f"  Sum: {total_sum}")

# Alternative: flatten then reduce
leaves = optree.tree_leaves(tree)
manual_sum = sum(leaves)
print(f"  Manual sum: {manual_sum}")
print(f"  Match: {total_sum == manual_sum}\n")

# Find maximum value
print("Finding maximum value:")
max_value = optree.tree_reduce(max, tree)
print(f"  Maximum: {max_value}\n")
assert total_sum == manual_sum == 450
assert max_value == 90

# Product of all leaves
print("Computing product of all leaves:")
product = optree.tree_reduce(operator.mul, tree)
print(f"  Product: {product:,}\n")

# Count non-zero values
print("Counting elements:")
tree_with_zeros = {'a': [1, 0, 2], 'b': (0, 3), 'c': {'d': 0, 'e': [4, 5]}}
all_leaves = optree.tree_leaves(tree_with_zeros)
non_zero_count = sum(1 for x in all_leaves if x != 0)
print(f"  Tree: {tree_with_zeros}")
print(f"  Total elements: {len(all_leaves)}")
print(f"  Non-zero elements: {non_zero_count}")

print("\nOpTree enables efficient tree aggregation operations")
```

## Example 5: Tree manipulation with filtering and transformation

```python
# Example 5: Tree manipulation with filtering and transformation
import optree

print("Example 5: Tree Filtering and Advanced Transformations\n")

# Create a mixed tree
print("Creating mixed tree:")
tree = {
    'positive': [1, 2, 3, 4, 5],
    'negative': [-1, -2, -3],
    'mixed': [10, -5, 8, -3, 15],
    'nested': {'a': [6, -7], 'b': -8}
}
print(f"  Tree: {tree}\n")

# Filter positive values only
print("Filtering positive values:")
all_values = optree.tree_leaves(tree)
positive_values = [x for x in all_values if x > 0]
print(f"  All values: {all_values}")
print(f"  Positive only: {positive_values}")
print(f"  Count: {len(positive_values)} out of {len(all_values)}\n")

# Transform: clip values to range [-5, 5]
print("Clipping values to [-5, 5]:")
clipped = optree.tree_map(lambda x: max(-5, min(5, x)), tree)
print(f"  Original: {tree}")
print(f"  Clipped: {clipped}\n")

# Transform: take absolute values
print("Taking absolute values:")
abs_tree = optree.tree_map(abs, tree)
print(f"  Absolute: {abs_tree}\n")

# Conditional transformation
print("Conditional transformation (square if positive, zero if negative):")
conditional = optree.tree_map(lambda x: x**2 if x > 0 else 0, tree)
print(f"  Result: {conditional}\n")

# Complex transformation combining multiple operations
print("Complex transformation (normalize to [0, 1]):")
leaves_list = optree.tree_leaves(tree)
min_val = min(leaves_list)
max_val = max(leaves_list)
range_val = max_val - min_val
normalized = optree.tree_map(lambda x: (x - min_val) / range_val if range_val != 0 else 0, tree)
print(f"  Min: {min_val}, Max: {max_val}")
print(f"  Normalized: {normalized}")
norm_leaves = optree.tree_leaves(normalized)
print(f"  New range: [{min(norm_leaves):.3f}, {max(norm_leaves):.3f}]")

print("\nOpTree enables complex tree transformations and filtering")
```
