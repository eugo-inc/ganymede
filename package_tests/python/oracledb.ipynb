## Regression guard: repr() of a default ConnectParams

```python
# Guards against oracle/python-oracledb#604: in oracledb 26.0.0, repr() of any ConnectParams raised
# ValueError because transaction_priority defaulted to None. Fixed in 26.0.1.
import oracledb

params = oracledb.ConnectParams()
text = repr(params)
assert text.startswith("ConnectParams(") and "transaction_priority=" in text, text

priority = params.transaction_priority
assert priority is oracledb.TransactionPriority.DEFAULT, repr(priority)
# TransactionPriority is a str-mixed Enum whose DEFAULT member's value is the empty string.
assert priority == "", repr(priority)
print(f"repr(ConnectParams()) OK ({len(text)} chars); transaction_priority={priority!r}")
```

## Example 1: Connection Parameters and DSN Configuration

```python
import oracledb

# Test ConnectParams and DSN creation (core oracledb functionality)
params = oracledb.ConnectParams(
    user="testuser",
    host="localhost",
    port=1521,
    service_name="XEPDB1"
)

print(f"Connect params created: {params}")
print(f"User: {params.user}")
print(f"Host: {params.host}")
print(f"Port: {params.port}")
print(f"Service name: {params.service_name}")

# Test DSN creation
dsn = oracledb.makedsn("localhost", 1521, service_name="XEPDB1")
print(f"\nDSN created: {dsn}")

# Verify thin mode is the default (nothing in this notebook calls init_oracle_client())
print(f"\nThin mode active: {oracledb.is_thin_mode()}")
assert oracledb.is_thin_mode() is True
```
