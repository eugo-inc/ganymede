# ordered_set: 5 Basic Usage Examples

```python
# Example 1: Create an OrderedSet
from ordered_set import OrderedSet
s = OrderedSet([1, 2, 3])
print(s)
assert list(s) == [1, 2, 3]
```

```python
# Example 2: Add and remove elements
from ordered_set import OrderedSet
s = OrderedSet()
s.add(1)
s.add(2)
s.discard(1)
print(s)
assert list(s) == [2]
```

```python
# Example 3: Set operations
from ordered_set import OrderedSet
a = OrderedSet([1, 2, 3])
b = OrderedSet([3, 4, 5])
print(a | b)  # Union
print(a & b)  # Intersection
assert list(a | b) == [1, 2, 3, 4, 5]
assert list(a & b) == [3]
```

```python
# Example 4: Indexing and slicing
from ordered_set import OrderedSet
s = OrderedSet([10, 20, 30, 40])
print(s[1])
print(s[1:3])
assert s[1] == 20
assert list(s[1:3]) == [20, 30]
```
