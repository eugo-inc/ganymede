## Example 1: Basic omdict Creation and Multi-Value Storage

```python
from orderedmultidict import omdict

# Create an ordered multidict and add multiple values per key
omd = omdict()
omd['color'] = 'red'
omd.add('color', 'blue')  # Add another value for same key
omd.add('color', 'green')
omd['size'] = 'large'

print("Standard dict access (gets first value):")
print(f"  omd['color'] = {omd['color']}")

print("\nGet all values for a key:")
print(f"  omd.getlist('color') = {omd.getlist('color')}")
assert omd['color'] == 'red'
assert omd.getlist('color') == ['red', 'blue', 'green']

print("\nStandard items (shows only first value):")
print(f"  list(omd.items()) = {list(omd.items())}")

print("\nAll items (shows all values including duplicates):")
print(f"  list(omd.allitems()) = {list(omd.allitems())}")
```

## Example 2: List Operations with addlist and setlist

```python
from orderedmultidict import omdict

# Test addlist and setlist methods (unique to omdict)
omd = omdict()
omd.addlist('tags', ['python', 'data-structures'])
omd.addlist('tags', ['multidict'])

print("After addlist operations:")
print(f"  omd.getlist('tags') = {omd.getlist('tags')}")

# setlist replaces all values for a key
omd.setlist('tags', ['new', 'replacement', 'tags'])
print("\nAfter setlist (replaces all values):")
print(f"  omd.getlist('tags') = {omd.getlist('tags')}")

# Test setdefault with multi-values
omd.setdefault('priority', 'high')
omd.add('priority', 'medium')
print("\nAfter setdefault + add:")
print(f"  omd.getlist('priority') = {omd.getlist('priority')}")
```

## Example 3: Initialization and Order Preservation

```python
from orderedmultidict import omdict

# Test various initialization methods
omd1 = omdict([('a', 1), ('a', 2), ('b', 3)])
print("From list of tuples:")
print(f"  allitems: {list(omd1.allitems())}")

omd2 = omdict({'x': 10, 'y': 20})
omd2.add('x', 11)
print("\nFrom dict with added value:")
print(f"  allitems: {list(omd2.allitems())}")

# Order is preserved
omd3 = omdict()
for i in range(5):
    omd3[f'key{i}'] = i
print("\nOrder preservation check:")
print(f"  keys: {list(omd3.keys())}")
print(f"  values: {list(omd3.values())}")
```

## Example 4: Item Access and Manipulation

```python
from orderedmultidict import omdict

# Test allkeys, allvalues with duplicates
omd = omdict([('fruit', 'apple'), ('fruit', 'banana'), ('veggie', 'carrot'), ('fruit', 'cherry')])

print("Unique keys vs all keys:")
print(f"  list(omd.keys()) = {list(omd.keys())}")
print(f"  list(omd.allkeys()) = {list(omd.allkeys())}")

print("\nUnique values vs all values:")
print(f"  list(omd.values()) = {list(omd.values())}")
print(f"  list(omd.allvalues()) = {list(omd.allvalues())}")

# Test pop and poplist
print("\nPop operations:")
print(f"  omd.pop('veggie') = {omd.pop('veggie')}")
print(f"  omd.poplist('fruit') = {omd.poplist('fruit')}")
print(f"  Remaining items: {list(omd.allitems())}")
```

## Example 5: Advanced Features - lists, size, and load

```python
from orderedmultidict import omdict

# Test size (total number of items including duplicates)
omd = omdict([('a', 1), ('a', 2), ('b', 3), ('c', 4), ('c', 5), ('c', 6)])

print("Unique keys vs total items:")
print(f"  len(omd) = {len(omd)} (unique keys)")
print(f"  omd.size() = {omd.size()} (total items)")
assert (len(omd), omd.size()) == (3, 6)

print("\nAll key-value pairs:")
print(f"  allitems: {list(omd.allitems())}")

# Test getlist for each key
print("\nValues per key:")
for key in omd.keys():
    print(f"  {key}: {omd.getlist(key)}")

# Test load method
omd2 = omdict()
omd2.load([('x', 1), ('x', 2), ('y', 3)])
print(f"\nAfter load(): {list(omd2.allitems())}")
```
