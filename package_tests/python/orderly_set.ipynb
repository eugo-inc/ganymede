## Example 1: Basic OrderedSet operations

```python
import orderly_set

# Example 1: Basic OrderedSet operations
print("Basic OrderedSet operations:")

# Create an OrderedSet
os = orderly_set.OrderedSet(['apple', 'banana', 'cherry'])
print(f"Created OrderedSet: {os}")

# Add elements (maintains order, prevents duplicates)
os.add('date')
os.add('banana')  # Won't add duplicate
print(f"After adding elements: {os}")

# Check membership
print(f"'banana' in set: {'banana' in os}")
print(f"'grape' in set: {'grape' in os}")

# Length and iteration
print(f"Set length: {len(os)}")
print("Iterating in order:")
for item in os:
    print(f"  - {item}")

# Indexing (ordered sets support indexing)
print(f"First element: {os[0]}")
print(f"Last element: {os[-1]}")

# Remove elements
os.remove('cherry')
print(f"After removing 'cherry': {os}")

# Discard (no error if element doesn't exist)
os.discard('grape')  # Won't raise error
print(f"After discarding 'grape': {os}")
assert list(os) == ['apple', 'banana', 'date']

print("Basic operations completed")
```

## Example 2: Set operations and mathematics

```python
import orderly_set

# Example 2: Set operations and mathematics
print("Set operations and mathematics:")

# Create two ordered sets
set1 = orderly_set.OrderedSet(['a', 'b', 'c', 'd'])
set2 = orderly_set.OrderedSet(['c', 'd', 'e', 'f'])

print(f"Set 1: {set1}")
print(f"Set 2: {set2}")

# Union (combines sets, maintains order)
union_result = set1 | set2
print(f"Union (set1 | set2): {union_result}")

# Intersection (common elements, maintains order from first set)
intersection_result = set1 & set2
print(f"Intersection (set1 & set2): {intersection_result}")

# Difference (elements in first set but not second)
difference_result = set1 - set2
print(f"Difference (set1 - set2): {difference_result}")
assert list(union_result) == ['a', 'b', 'c', 'd', 'e', 'f']
assert list(intersection_result) == ['c', 'd'] and list(difference_result) == ['a', 'b']

# Symmetric difference (elements in either set but not both)
sym_diff_result = set1 ^ set2
print(f"Symmetric difference (set1 ^ set2): {sym_diff_result}")

# Update operations (modify in place)
set3 = orderly_set.OrderedSet(['x', 'y', 'z'])
print(f"\nOriginal set3: {set3}")

set3.update(['a', 'b'])
print(f"After update with ['a', 'b']: {set3}")

# Subset and superset checks
set4 = orderly_set.OrderedSet(['a', 'b'])
set5 = orderly_set.OrderedSet(['a', 'b', 'c', 'd'])

print(f"\nset4: {set4}")
print(f"set5: {set5}")
print(f"set4 is subset of set5: {set4 <= set5}")
print(f"set5 is superset of set4: {set5 >= set4}")

# Disjoint check
set6 = orderly_set.OrderedSet(['x', 'y'])
print(f"set4 and set6 are disjoint: {set4.isdisjoint(set6)}")

print("Set operations completed")
```

## Example 3: Advanced indexing and slicing

```python
import orderly_set

# Example 3: Advanced indexing and slicing
print("Advanced indexing and slicing:")

# Create ordered set with numbers
numbers = orderly_set.OrderedSet(range(10))
print(f"Numbers set: {numbers}")

# Positive indexing
print(f"Element at index 0: {numbers[0]}")
print(f"Element at index 3: {numbers[3]}")
print(f"Element at index 7: {numbers[7]}")

# Negative indexing
print(f"Last element: {numbers[-1]}")
print(f"Second to last: {numbers[-2]}")
print(f"Third from end: {numbers[-3]}")

# Slicing operations
print(f"First 3 elements: {list(numbers[:3])}")
print(f"Last 3 elements: {list(numbers[-3:])}")
print(f"Elements 2-5: {list(numbers[2:6])}")
print(f"Every second element: {list(numbers[::2])}")

print("Advanced indexing completed")
```

## Example 4: Performance comparisons and use cases

```python
import orderly_set
import time

# Example 4: Performance comparisons and use cases
print("Performance comparisons and use cases:")

# Compare OrderedSet vs regular set vs list
data = list(range(1000))

# Test insertion performance
print("Insertion performance comparison:")

# Regular set
start_time = time.time()
regular_set = set()
for item in data:
    regular_set.add(item)
set_time = time.time() - start_time

# List (with uniqueness check)
start_time = time.time()
unique_list = []
for item in data:
    if item not in unique_list:
        unique_list.append(item)
list_time = time.time() - start_time

# OrderedSet
start_time = time.time()
ordered_set = orderly_set.OrderedSet()
for item in data:
    ordered_set.add(item)
ordered_set_time = time.time() - start_time

print(f"Regular set: {set_time:.4f}s")
print(f"Unique list: {list_time:.4f}s") 
print(f"OrderedSet: {ordered_set_time:.4f}s")

# Test lookup performance
lookup_items = [100, 500, 999, 1500]  # Some exist, some don't

print(f"\nLookup performance (checking {len(lookup_items)} items):")

# Regular set lookup
start_time = time.time()
for item in lookup_items:
    _ = item in regular_set
set_lookup_time = time.time() - start_time

# List lookup
start_time = time.time()
for item in lookup_items:
    _ = item in unique_list
list_lookup_time = time.time() - start_time

# OrderedSet lookup
start_time = time.time()
for item in lookup_items:
    _ = item in ordered_set
ordered_lookup_time = time.time() - start_time

print(f"Regular set lookup: {set_lookup_time:.6f}s")
print(f"List lookup: {list_lookup_time:.6f}s")
print(f"OrderedSet lookup: {ordered_lookup_time:.6f}s")

# Use cases demonstration
print(f"\nPractical use cases:")

# Use case 1: Preserving order of unique items
print("1. Preserving order of unique items:")
items_with_duplicates = ['apple', 'banana', 'apple', 'cherry', 'banana', 'date']
unique_ordered = orderly_set.OrderedSet(items_with_duplicates)
print(f"   Original: {items_with_duplicates}")
print(f"   Unique ordered: {list(unique_ordered)}")

# Use case 2: Set operations with order preservation
print("2. Set operations maintaining order:")
tags1 = orderly_set.OrderedSet(['python', 'programming', 'data'])
tags2 = orderly_set.OrderedSet(['data', 'science', 'analysis'])
combined_tags = tags1 | tags2
print(f"   Tags 1: {list(tags1)}")
print(f"   Tags 2: {list(tags2)}")
print(f"   Combined: {list(combined_tags)}")

# Use case 3: Order-sensitive queue with uniqueness
print("3. Priority queue with unique items:")
priority_queue = orderly_set.OrderedSet()
priority_queue.add('critical_task')
priority_queue.add('normal_task')
priority_queue.add('low_task')
priority_queue.add('critical_task')  # Won't duplicate
print(f"   Queue: {list(priority_queue)}")

# Process items in order
while priority_queue:
    task = priority_queue.pop(0)  # Get first item
    print(f"   Processing: {task}")

print("Performance comparisons completed")
```

## Example 5: Integration with other data structures

```python
import orderly_set

# Example 5: Integration with other data structures
print("Integration with other data structures:")

# Integration with dictionaries
print("1. Dictionary key ordering:")
data = {'z': 1, 'a': 2, 'c': 3, 'b': 4}
ordered_keys = orderly_set.OrderedSet(data.keys())
print(f"   Original dict keys: {list(data.keys())}")
print(f"   Ordered keys: {list(ordered_keys)}")

# Create ordered dict-like structure
ordered_dict = {key: data[key] for key in ordered_keys}
print(f"   Reordered dict: {ordered_dict}")

# Integration with lists
print("2. List deduplication with order preservation:")
list_with_dupes = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]
deduplicated = orderly_set.OrderedSet(list_with_dupes)
print(f"   Original list: {list_with_dupes}")
print(f"   Deduplicated: {list(deduplicated)}")

# Integration with itertools
print("3. Working with itertools:")
import itertools

# Chain multiple sequences maintaining order
seq1 = ['a', 'b', 'c']
seq2 = ['b', 'c', 'd']  
seq3 = ['c', 'd', 'e']

chained = orderly_set.OrderedSet(itertools.chain(seq1, seq2, seq3))
print(f"   Chained sequences: {list(chained)}")

# Integration with JSON-like data
print("4. Processing JSON-like structures:")
json_data = {
    "tags": ["python", "data", "python", "analysis", "data"],
    "categories": ["tech", "programming", "tech", "science"]
}

# Clean up duplicate tags while preserving order
json_data["unique_tags"] = list(orderly_set.OrderedSet(json_data["tags"]))
json_data["unique_categories"] = list(orderly_set.OrderedSet(json_data["categories"]))

print(f"   Original tags: {json_data['tags']}")
print(f"   Unique tags: {json_data['unique_tags']}")
print(f"   Original categories: {json_data['categories']}")
print(f"   Unique categories: {json_data['unique_categories']}")

# Integration with pandas-like operations
print("5. Simulating pandas-like unique operations:")
data_series = [10, 20, 10, 30, 20, 40, 10]
unique_values = orderly_set.OrderedSet(data_series)
value_counts = {val: data_series.count(val) for val in unique_values}

print(f"   Data series: {data_series}")
print(f"   Unique values (ordered): {list(unique_values)}")
print(f"   Value counts: {value_counts}")

print("Integration examples completed")
```
