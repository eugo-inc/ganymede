## Example 1: Basic Override Decorator

```python
from overrides import override

# Example 1: Basic @override usage
print("Example 1 - Basic override")

class Animal:
    def speak(self):
        return "Some sound"
    
    def move(self):
        return "Moving"

class Dog(Animal):
    @override
    def speak(self):
        return "Woof!"
    
    @override
    def move(self):
        return "Running"

dog = Dog()
print(f"Dog speaks: {dog.speak()}")
print(f"Dog moves: {dog.move()}")
assert (dog.speak(), dog.move()) == ("Woof!", "Running")

# @override rejects a method that has no base-class counterpart
try:
    class Cat(Animal):
        @override
        def purr(self):
            return "Purr"
except TypeError as e:
    print(f"Rejected as expected: {e}")
else:
    raise AssertionError("@override accepted a method that overrides nothing")

print("Example 1 completed")
```

## Example 2: Multiple Inheritance with Override

```python
from overrides import override

# Example 2: Multiple inheritance
print("Example 2 - Multiple inheritance")

class Flyable:
    def fly(self):
        return "Flying"

class Swimmable:
    def swim(self):
        return "Swimming"

class Duck(Flyable, Swimmable):
    @override
    def fly(self):
        return "Duck flying"
    
    @override
    def swim(self):
        return "Duck swimming"

duck = Duck()
print(f"Duck: {duck.fly()}, {duck.swim()}")

print("Example 2 completed")
```

## Example 3: Abstract Method Pattern

```python
from overrides import override

# Example 3: Abstract pattern with override
print("Example 3 - Abstract pattern")

class Processor:
    def process(self, data):
        raise NotImplementedError("Must override process")

class TextProcessor(Processor):
    @override
    def process(self, data):
        return data.upper()

class NumberProcessor(Processor):
    @override
    def process(self, data):
        return data * 2

text_proc = TextProcessor()
num_proc = NumberProcessor()

print(f"Text: {text_proc.process('hello')}")
print(f"Number: {num_proc.process(42)}")

print("Example 3 completed")
```

## Example 4: Property Override

```python
from overrides import override

# Example 4: Override properties
print("Example 4 - Property override")

class BaseService:
    @property
    def name(self):
        return "BaseService"
    
    def get_info(self):
        return f"Service: {self.name}"

class DatabaseService(BaseService):
    @property
    @override
    def name(self):
        return "DatabaseService"

base = BaseService()
db = DatabaseService()

print(f"Base: {base.get_info()}")
print(f"DB: {db.get_info()}")

print("Example 4 completed")
```

## Example 5: Repository Pattern with Override

```python
from overrides import override

# Example 5: Repository pattern
print("Example 5 - Repository pattern")

class BaseRepository:
    def find_by_id(self, id):
        raise NotImplementedError("Must override")
    
    def save(self, entity):
        raise NotImplementedError("Must override")

class UserRepository(BaseRepository):
    def __init__(self):
        self.users = {1: "Alice", 2: "Bob"}
    
    @override
    def find_by_id(self, id):
        return self.users.get(id, "Not found")
    
    @override
    def save(self, entity):
        self.users[entity["id"]] = entity["name"]
        return True

repo = UserRepository()
print(f"User 1: {repo.find_by_id(1)}")
repo.save({"id": 3, "name": "Charlie"})
print(f"User 3: {repo.find_by_id(3)}")
assert repo.find_by_id(3) == "Charlie"
assert repo.find_by_id(99) == "Not found"

print("Example 5 completed")
```
