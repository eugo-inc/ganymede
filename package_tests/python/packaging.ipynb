```python
# Example 1: Version Parsing and Comparison
# Demonstrate version parsing, comparison, and validation

from packaging import version
from packaging.version import Version, InvalidVersion

def demonstrate_version_parsing():
    """Show various version formats and parsing capabilities"""
    
    version_strings = [
        "1.0.0",
        "2.1.3",
        "1.0.0-alpha",
        "1.0.0-alpha.1", 
        "1.0.0-beta.2",
        "1.0.0-rc.1",
        "1.0.0+build.1",
        "1.2.3.dev1",
        "2.0.0a1",
        "3.0.0b2",
        "4.0.0rc3"
    ]
    
    parsed_versions = []
    
    print("=== Version Parsing ===")
    for v_str in version_strings:
        try:
            v = version.parse(v_str)
            parsed_versions.append((v_str, v))
            print(f"'{v_str}' -> {v} (type: {type(v).__name__})")
            print(f"  Public: {v.public}")
            print(f"  Base: {v.base_version}")
            if hasattr(v, 'is_prerelease'):
                print(f"  Pre-release: {v.is_prerelease}")
        except InvalidVersion as e:
            print(f"'{v_str}' -> INVALID: {e}")
    
    return parsed_versions

def demonstrate_version_comparison(versions):
    """Show version comparison operations"""
    
    print(f"\n=== Version Comparison ===")
    
    # Sort versions
    version_objects = [v[1] for v in versions]
    sorted_versions = sorted(version_objects)
    
    print("Versions in sorted order:")
    for v in sorted_versions:
        print(f"  {v}")
    
    # Compare specific versions
    v1 = version.parse("1.0.0")
    v2 = version.parse("1.0.1")
    v3 = version.parse("1.0.0-alpha")
    
    print(f"\nComparison examples:")
    print(f"{v1} < {v2}: {v1 < v2}")
    print(f"{v1} == {v2}: {v1 == v2}")
    print(f"{v3} < {v1}: {v3 < v1}")  # Pre-release < release
    print(f"{v1} >= {v3}: {v1 >= v3}")

# Run version demonstrations
print("=== Packaging Version Utilities ===")

versions = demonstrate_version_parsing()
demonstrate_version_comparison(versions)

# Practical version filtering
print(f"\n=== Practical Version Filtering ===")
test_versions = ["1.0.0", "1.1.0", "2.0.0-alpha", "2.0.0", "2.1.0"]
minimum_version = version.parse("1.1.0")

compatible_versions = []
for v_str in test_versions:
    v = version.parse(v_str)
    if v >= minimum_version and not v.is_prerelease:
        compatible_versions.append(v_str)

print(f"Minimum version: {minimum_version}")
print(f"Compatible versions: {compatible_versions}")
assert compatible_versions == ["1.1.0", "2.0.0", "2.1.0"]
assert version.parse("1.0.0-alpha") < version.parse("1.0.0")

print("Example 1: Version parsing and comparison completed successfully!")
```

```python
# Example 2: Requirement Specification Parsing and Validation
# Demonstrate requirement parsing, dependency resolution, and constraint handling

from packaging import requirements
from packaging.requirements import Requirement, InvalidRequirement
from packaging import specifiers

def parse_requirement_strings():
    """Parse various requirement string formats"""
    
    req_strings = [
        "requests>=2.25.0",
        "Django>=3.0,<4.0",
        "numpy==1.21.0",
        "Flask[async]>=1.1.0",
        "pytest>=6.0; python_version>='3.8'",
        "scipy>=1.5.0,!=1.6.0",
        "Pillow>=8.0.0,<10.0.0",
        "gunicorn[gthread]>=20.0.0",
        "SQLAlchemy>=1.4.0; extra=='database'",
        "cryptography>=3.0; sys_platform!='win32'"
    ]
    
    parsed_requirements = []
    
    print("=== Requirement String Parsing ===")
    for req_str in req_strings:
        try:
            req = Requirement(req_str)
            parsed_requirements.append(req)
            
            print(f"'{req_str}':")
            print(f"  Name: {req.name}")
            print(f"  Specifier: {req.specifier}")
            print(f"  Extras: {req.extras}")
            if req.marker:
                print(f"  Marker: {req.marker}")
            if req.url:
                print(f"  URL: {req.url}")
            print()
        except InvalidRequirement as e:
            print(f"'{req_str}' -> INVALID: {e}")
    
    return parsed_requirements

def check_version_compatibility(requirements_list):
    """Check version compatibility against requirements"""
    
    print("=== Version Compatibility Checking ===")
    
    # Test versions for common packages
    test_cases = [
        ("requests", "2.25.1"),
        ("requests", "2.24.0"),
        ("Django", "3.2.0"),
        ("Django", "4.1.0"), 
        ("numpy", "1.21.0"),
        ("numpy", "1.20.0"),
        ("Flask", "1.1.4"),
        ("Flask", "2.0.1")
    ]
    
    for req in requirements_list:
        if not req.extras and not req.marker:  # Skip complex requirements for simplicity
            print(f"Checking requirement: {req}")
            
            for pkg_name, pkg_version in test_cases:
                if req.name.lower() == pkg_name.lower():
                    version_obj = version.parse(pkg_version)
                    is_satisfied = version_obj in req.specifier
                    status = "✅ SATISFIED" if is_satisfied else "❌ NOT SATISFIED"
                    print(f"  {pkg_name} {pkg_version}: {status}")
            print()

# Run requirement parsing demonstration
print("=== Packaging Requirements and Specifiers ===")

requirements_list = parse_requirement_strings()
check_version_compatibility(requirements_list)

# Advanced specifier operations
print("=== Advanced Specifier Operations ===")

spec1 = specifiers.SpecifierSet(">=1.0.0,<2.0.0")
spec2 = specifiers.SpecifierSet(">=1.5.0,<3.0.0")

print(f"Specifier 1: {spec1}")
print(f"Specifier 2: {spec2}")

# Test version compatibility
test_versions = ["0.9.0", "1.0.0", "1.4.0", "1.6.0", "2.0.0", "2.5.0"]
print(f"\nVersion compatibility:")
for v in test_versions:
    version_obj = version.parse(v)
    in_spec1 = version_obj in spec1
    in_spec2 = version_obj in spec2
    in_both = in_spec1 and in_spec2
    print(f"  {v}: Spec1({in_spec1}) Spec2({in_spec2}) Both({in_both})")
assert [v for v in test_versions if version.parse(v) in spec1 and version.parse(v) in spec2] == ["1.6.0"]

print("Example 2: Requirement specification parsing completed successfully!")
```

```python
# Example 3: Environment Markers and Conditional Dependencies
# Demonstrate environment marker evaluation and conditional requirements

from packaging import markers
from packaging.markers import Marker, InvalidMarker
import platform
import sys

def demonstrate_environment_markers():
    """Show environment marker parsing and evaluation"""
    
    marker_strings = [
        "python_version >= '3.8'",
        "sys_platform == 'win32'", 
        "platform_machine == 'x86_64'",
        "python_version >= '3.9' and sys_platform != 'win32'",
        "extra == 'dev'",
        "os_name == 'posix'",
        "implementation_name == 'cpython'",
        "python_full_version >= '3.8.5'",
        "platform_python_implementation == 'PyPy'",
        "sys_platform == 'darwin' and platform_machine == 'arm64'"
    ]
    
    print("=== Environment Marker Evaluation ===")
    print(f"Current environment:")
    print(f"  Python version: {sys.version_info}")
    print(f"  Platform: {platform.system()}")
    print(f"  Machine: {platform.machine()}")
    print(f"  Implementation: {platform.python_implementation()}")
    print()
    
    valid_markers = []
    
    for marker_str in marker_strings:
        try:
            marker = Marker(marker_str)
            valid_markers.append((marker_str, marker))
            
            # Evaluate marker in current environment
            is_satisfied = marker.evaluate()
            status = "✅ TRUE" if is_satisfied else "❌ FALSE"
            print(f"'{marker_str}': {status}")
            
        except InvalidMarker as e:
            print(f"'{marker_str}': INVALID - {e}")
    
    return valid_markers

def demonstrate_custom_environment_evaluation(markers_list):
    """Show marker evaluation with custom environment"""
    
    print(f"\n=== Custom Environment Evaluation ===")
    
    # Define custom environments to test
    environments = [
        {
            "python_version": "3.9",
            "sys_platform": "linux",
            "platform_machine": "x86_64",
            "implementation_name": "cpython",
            "extra": "dev"
        },
        {
            "python_version": "3.7",
            "sys_platform": "win32", 
            "platform_machine": "AMD64",
            "implementation_name": "cpython",
            "extra": None
        },
        {
            "python_version": "3.10",
            "sys_platform": "darwin",
            "platform_machine": "arm64",
            "implementation_name": "cpython",
            "extra": "test"
        }
    ]
    
    for i, env in enumerate(environments, 1):
        print(f"Environment {i}: {env}")
        
        for marker_str, marker in markers_list:
            try:
                is_satisfied = marker.evaluate(environment=env)
                status = "✅ TRUE" if is_satisfied else "❌ FALSE"
                print(f"  '{marker_str}': {status}")
            except Exception as e:
                print(f"  '{marker_str}': ERROR - {e}")
        print()

def practical_dependency_resolution():
    """Show practical dependency resolution with markers"""
    
    print("=== Practical Dependency Resolution ===")
    
    # Example requirements with markers from a real project
    requirements_with_markers = [
        "colorama >= 0.4.3 ; sys_platform == 'win32'",
        "pytest >= 6.0 ; extra == 'dev'",
        "black >= 21.0 ; extra == 'dev'",
        "uvloop >= 0.15.0 ; sys_platform != 'win32'",
        "dataclasses >= 0.6 ; python_version < '3.7'",
        "importlib-metadata >= 1.0 ; python_version < '3.8'"
    ]
    
    current_env = {
        "python_version": f"{sys.version_info.major}.{sys.version_info.minor}",
        "sys_platform": sys.platform,
        "extra": "dev"  # Assuming we want dev dependencies
    }
    
    print(f"Resolving dependencies for environment: {current_env}")
    print()
    
    applicable_requirements = []
    
    for req_str in requirements_with_markers:
        try:
            req = Requirement(req_str)
            
            if req.marker:
                is_applicable = req.marker.evaluate(environment=current_env)
            else:
                is_applicable = True  # No marker means always applicable
            
            status = "REQUIRED" if is_applicable else "SKIPPED"
            print(f"{req.name}: {status}")
            if req.marker:
                print(f"  Condition: {req.marker}")
            print(f"  Version: {req.specifier}")
            
            if is_applicable:
                applicable_requirements.append(req.name)
            
        except Exception as e:
            print(f"Error processing '{req_str}': {e}")
        print()
    
    print(f"Final dependency list: {applicable_requirements}")

# Run marker demonstrations
print("=== Packaging Environment Markers ===")

markers_list = demonstrate_environment_markers()
demonstrate_custom_environment_evaluation(markers_list)
practical_dependency_resolution()

print("Example 3: Environment markers and conditional dependencies completed successfully!")
```

```python
# Example 4: Package Name Normalization and Utilities
# Demonstrate package name normalization, canonicalization, and utility functions

from packaging.utils import canonicalize_name, parse_wheel_filename, parse_sdist_filename
from packaging.utils import InvalidWheelFilename, InvalidSdistFilename

def demonstrate_name_normalization():
    """Show package name normalization and canonicalization"""
    
    package_names = [
        "Django",
        "django",
        "DJANGO",
        "flask-restful",
        "Flask_RESTful",
        "Flask.RESTful",
        "requests-oauthlib",
        "requests_oauthlib",
        "requests.oauthlib",
        "Pillow",
        "PIL",
        "scikit-learn",
        "scikit_learn"
    ]
    
    print("=== Package Name Normalization ===")
    canonical_names = {}
    
    for name in package_names:
        canonical = canonicalize_name(name)
        canonical_names[name] = canonical
        print(f"'{name}' -> '{canonical}'")
    
    # Group by canonical name
    print(f"\n=== Grouping by Canonical Name ===")
    groups = {}
    for original, canonical in canonical_names.items():
        if canonical not in groups:
            groups[canonical] = []
        groups[canonical].append(original)
    
    for canonical, variants in groups.items():
        if len(variants) > 1:
            print(f"'{canonical}': {variants}")
    
    return canonical_names

def parse_distribution_filenames():
    """Parse wheel and source distribution filenames"""
    
    wheel_files = [
        "Django-3.2.13-py3-none-any.whl",
        "numpy-1.21.5-cp39-cp39-win_amd64.whl",
        "Pillow-9.1.0-cp38-cp38-macosx_10_10_x86_64.whl",
        "requests-2.27.1-py2.py3-none-any.whl",
        "scikit_learn-1.0.2-cp39-cp39-linux_x86_64.whl"
    ]
    
    sdist_files = [
        "Django-3.2.13.tar.gz",
        "requests-2.27.1.tar.gz", 
        "numpy-1.21.5.zip",
        "flask-2.1.0.tar.gz",
        "Pillow-9.1.0.tar.gz"
    ]
    
    print(f"\n=== Wheel Filename Parsing ===")
    for wheel_file in wheel_files:
        try:
            parsed = parse_wheel_filename(wheel_file)
            name, version, build_tag, tag = parsed
            print(f"'{wheel_file}':")
            print(f"  Name: {name}")
            print(f"  Version: {version}")
            print(f"  Build: {build_tag}")
            print(f"  Tag: {tag}")
            print()
        except InvalidWheelFilename as e:
            print(f"'{wheel_file}': INVALID - {e}")
    
    print(f"=== Source Distribution Filename Parsing ===")
    for sdist_file in sdist_files:
        try:
            parsed = parse_sdist_filename(sdist_file)
            name, version = parsed
            print(f"'{sdist_file}': {name} v{version}")
        except InvalidSdistFilename as e:
            print(f"'{sdist_file}': INVALID - {e}")

def demonstrate_practical_package_management():
    """Show practical package management scenarios"""
    
    print(f"\n=== Practical Package Management ===")
    
    # Simulate package index with various name formats
    package_index = [
        ("Django", "3.2.13"),
        ("django", "3.1.0"),  # Same package, different case
        ("Flask-RESTful", "0.3.9"),
        ("flask_restful", "0.3.8"),  # Same package, different separators
        ("requests", "2.27.1"),
        ("Requests", "2.26.0"),  # Same package, different case
    ]
    
    # Normalize and find latest versions
    canonical_packages = {}
    for name, version_str in package_index:
        canonical = canonicalize_name(name)
        version_obj = version.parse(version_str)
        
        if canonical not in canonical_packages:
            canonical_packages[canonical] = []
        canonical_packages[canonical].append((name, version_obj))
    
    print("Latest versions by canonical name:")
    for canonical, versions in canonical_packages.items():
        # Sort by version and get the latest
        latest = max(versions, key=lambda x: x[1])
        all_names = [v[0] for v in versions]
        print(f"  {canonical}: {latest[1]} (names: {all_names})")
    
    # Package compatibility checking
    print(f"\n=== Package Compatibility ===")
    user_requirements = ["django>=3.0", "flask-restful==0.3.9", "requests>=2.25.0"]
    
    for req_str in user_requirements:
        req = Requirement(req_str)
        canonical = canonicalize_name(req.name)
        
        if canonical in canonical_packages:
            available_versions = canonical_packages[canonical]
            compatible_versions = []
            
            for name, version_obj in available_versions:
                if version_obj in req.specifier:
                    compatible_versions.append((name, version_obj))
            
            print(f"Requirement: {req_str}")
            if compatible_versions:
                best = max(compatible_versions, key=lambda x: x[1])
                print(f"  ✅ Compatible: {best[0]} {best[1]}")
            else:
                print(f"  ❌ No compatible versions found")
        else:
            print(f"Requirement: {req_str}")
            print(f"  ❌ Package not found in index")

# Run package utility demonstrations
print("=== Packaging Utilities and Name Normalization ===")

canonical_names = demonstrate_name_normalization()
parse_distribution_filenames()
demonstrate_practical_package_management()

print("Example 4: Package name normalization and utilities completed successfully!")
```

```python
# Example 5: Production Dependency Management and Version Resolution
# Demonstrate complete dependency management pipeline for production use

from packaging import version, requirements, markers, specifiers
from packaging.utils import canonicalize_name
from collections import defaultdict, deque
import sys

class DependencyResolver:
    """Production-grade dependency resolver using packaging utilities"""
    
    def __init__(self):
        self.package_index = {}  # name -> [versions]
        self.dependency_graph = defaultdict(list)  # package -> dependencies
        self.conflicts = []
        self.resolution_log = []
    
    def add_package_version(self, name, version_str, dependencies=None):
        """Add a package version to the index"""
        canonical_name = canonicalize_name(name)
        version_obj = version.parse(version_str)
        
        if canonical_name not in self.package_index:
            self.package_index[canonical_name] = []
        
        self.package_index[canonical_name].append({
            'version': version_obj,
            'original_name': name,
            'dependencies': dependencies or []
        })
    
    def resolve_dependencies(self, requirement_strings, environment=None):
        """Resolve dependencies for given requirements"""
        
        if environment is None:
            environment = {
                'python_version': f"{sys.version_info.major}.{sys.version_info.minor}",
                'sys_platform': sys.platform,
                'implementation_name': 'cpython'
            }
        
        self.resolution_log.append(f"Starting resolution for: {requirement_strings}")
        self.resolution_log.append(f"Environment: {environment}")
        
        # Parse requirements
        parsed_requirements = []
        for req_str in requirement_strings:
            try:
                req = requirements.Requirement(req_str)
                
                # Check if requirement applies to this environment
                if req.marker and not req.marker.evaluate(environment=environment):
                    self.resolution_log.append(f"Skipping {req.name}: marker not satisfied")
                    continue
                
                parsed_requirements.append(req)
            except Exception as e:
                self.resolution_log.append(f"Failed to parse requirement '{req_str}': {e}")
        
        # Find compatible versions
        resolution = {}
        for req in parsed_requirements:
            canonical_name = canonicalize_name(req.name)
            
            if canonical_name not in self.package_index:
                self.conflicts.append(f"Package not found: {req.name}")
                continue
            
            # Find compatible versions
            compatible_versions = []
            for pkg_info in self.package_index[canonical_name]:
                if pkg_info['version'] in req.specifier:
                    compatible_versions.append(pkg_info)
            
            if not compatible_versions:
                self.conflicts.append(f"No compatible version for {req.name} {req.specifier}")
                continue
            
            # Select highest compatible version
            best_version = max(compatible_versions, key=lambda x: x['version'])
            resolution[canonical_name] = best_version
            
            self.resolution_log.append(
                f"Resolved {req.name} to {best_version['original_name']} {best_version['version']}"
            )
        
        return resolution
    
    def generate_resolution_report(self, resolution):
        """Generate comprehensive resolution report"""
        
        report = ["=== Dependency Resolution Report ===", ""]
        
        if resolution:
            report.append("Resolved packages:")
            total_size_estimate = 0
            
            for canonical_name, pkg_info in sorted(resolution.items()):
                report.append(f"  {pkg_info['original_name']} == {pkg_info['version']}")
                
                # Estimate package size (mock calculation)
                size_estimate = len(canonical_name) * 1000 + int(str(pkg_info['version']).replace('.', '')) * 100
                total_size_estimate += size_estimate
                
                if pkg_info['dependencies']:
                    report.append(f"    Dependencies: {pkg_info['dependencies']}")
            
            report.extend(["", f"Estimated total size: {total_size_estimate:,} bytes", ""])
        
        if self.conflicts:
            report.append("Conflicts found:")
            for conflict in self.conflicts:
                report.append(f"  ❌ {conflict}")
            report.append("")
        
        if self.resolution_log:
            report.append("Resolution log:")
            for log_entry in self.resolution_log[-10:]:  # Show last 10 entries
                report.append(f"  {log_entry}")
        
        return "\n".join(report)

# Create production dependency resolver
print("=== Production Dependency Management Pipeline ===")

resolver = DependencyResolver()

# Populate mock package index
packages_data = [
    ("Django", "3.2.13", ["pytz>=2021.1", "sqlparse>=0.2.2"]),
    ("Django", "3.1.7", ["pytz>=2020.1", "sqlparse>=0.2.2"]), 
    ("django", "4.0.4", ["pytz>=2021.1", "sqlparse>=0.3.0"]),
    ("requests", "2.27.1", ["urllib3>=1.21.1,<1.27", "certifi>=2017.4.17"]),
    ("requests", "2.25.1", ["urllib3>=1.21.1,<1.27", "certifi>=2017.4.17"]),
    ("Flask", "2.1.0", ["Werkzeug>=2.0", "Jinja2>=3.0"]),
    ("flask", "1.1.4", ["Werkzeug>=0.15", "Jinja2>=2.10.1"]),
    ("numpy", "1.22.3", []),
    ("numpy", "1.21.5", []),
    ("pytest", "7.1.1", ["py>=1.8.2", "pluggy>=0.12,<2.0"]),
    ("black", "22.3.0", ["click>=8.0.0", "platformdirs>=2"])
]

for name, version_str, deps in packages_data:
    resolver.add_package_version(name, version_str, deps)

print(f"Package index populated with {len(resolver.package_index)} unique packages")

# Test different resolution scenarios
test_scenarios = [
    {
        "name": "Basic Web Stack",
        "requirements": ["Django>=3.2", "requests>=2.25.0"],
        "environment": {"python_version": "3.9", "sys_platform": "linux"}
    },
    {
        "name": "Development Environment", 
        "requirements": ["Flask>=2.0", "pytest>=6.0", "black>=20.0; extra=='dev'"],
        "environment": {"python_version": "3.10", "sys_platform": "darwin", "extra": "dev"}
    },
    {
        "name": "Data Science Stack",
        "requirements": ["numpy>=1.20.0", "requests>=2.20.0"],
        "environment": {"python_version": "3.8", "sys_platform": "win32"}
    },
    {
        "name": "Version Conflict Test",
        "requirements": ["Django==3.1.7", "django>=4.0"],  # Intentional conflict
        "environment": {"python_version": "3.9", "sys_platform": "linux"}
    }
]

for scenario in test_scenarios:
    print(f"\n--- {scenario['name']} ---")
    
    resolver.conflicts.clear()
    resolver.resolution_log.clear()
    
    resolution = resolver.resolve_dependencies(
        scenario['requirements'], 
        scenario['environment']
    )
    
    report = resolver.generate_resolution_report(resolution)
    print(report)

# Performance and validation summary
print(f"\n=== Pipeline Summary ===")
print(f"Total packages in index: {sum(len(versions) for versions in resolver.package_index.values())}")
print(f"Unique package names: {len(resolver.package_index)}")

# Validate all packages in index
validation_results = {"valid": 0, "invalid": 0}
for canonical_name, versions in resolver.package_index.items():
    for pkg_info in versions:
        try:
            # Validate version format
            str(pkg_info['version'])
            validation_results["valid"] += 1
        except Exception:
            validation_results["invalid"] += 1

print(f"Package validation: {validation_results['valid']} valid, {validation_results['invalid']} invalid")

print("Example 5: Production dependency management completed successfully!")
```
