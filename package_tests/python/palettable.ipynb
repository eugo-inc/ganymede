## Example 1: Basic ColorBrewer Palettes

```python
import palettable
from palettable.colorbrewer.qualitative import Set1_9
from palettable.colorbrewer.sequential import Blues_9
from palettable.colorbrewer.diverging import RdBu_11

# Example 1: Basic ColorBrewer palettes
print("Example 1 - ColorBrewer palettes")

palettes = [("Set1_9", Set1_9), ("Blues_9", Blues_9), ("RdBu_11", RdBu_11)]

for name, palette in palettes:
    print(f"\n{name}:")
    print(f"  Type: {palette.type}")
    print(f"  Colors: {palette.number}")
    print(f"  Hex colors: {palette.hex_colors[:3]}...")
    print(f"  RGB colors: {palette.colors[:2]}")

# Test color format conversions
print(f"\nColor format examples (Set1_9):")
print(f"  Hex: {Set1_9.hex_colors[0]}")
print(f"  RGB: {Set1_9.colors[0]}")
# Convert RGB to normalized (0-1)
r, g, b = Set1_9.colors[0]
print(f"  RGB normalized: ({r/255:.3f}, {g/255:.3f}, {b/255:.3f})")
assert Set1_9.number == len(Set1_9.hex_colors) == len(Set1_9.colors) == 9
assert Set1_9.hex_colors[0].upper() == "#E41A1C" and list(Set1_9.colors[0]) == [228, 26, 28]

print("Example 1 completed")
```

## Example 2: Scientific Color Maps

```python
from palettable.scientific.sequential import Batlow_10
from palettable.matplotlib import Viridis_10, Plasma_10

# Example 2: Scientific palettes
print("Example 2 - Scientific color maps")

scientific = [
    ("Batlow_10", Batlow_10, "Perceptually uniform"),
    ("Viridis_10", Viridis_10, "Matplotlib default"),
    ("Plasma_10", Plasma_10, "High contrast")
]

for name, palette, description in scientific:
    print(f"\n{name} - {description}:")
    print(f"  Colors: {palette.number}")
    print(f"  Range: {palette.hex_colors[0]} → {palette.hex_colors[-1]}")

# Test brightness progression
brightness = [(0.299*r + 0.587*g + 0.114*b) for r,g,b in Viridis_10.colors]
print(f"\nViridis brightness: {min(brightness):.0f} to {max(brightness):.0f}")
print(f"Monotonic: {brightness == sorted(brightness)}")
assert brightness == sorted(brightness)  # viridis gets steadily lighter

print("Example 2 completed")
```

## Example 3: Tableau and CartoDB Palettes

```python
from palettable.tableau import Tableau_10
from palettable.cartocolors.qualitative import Bold_10
from palettable.cartocolors.sequential import BluYl_7

# Example 3: Business visualization palettes
print("Example 3 - Business palettes")

business = [
    ("Tableau_10", Tableau_10, "Business dashboards"),
    ("Bold_10", Bold_10, "Web-friendly"),
    ("BluYl_7", BluYl_7, "Sequential data")
]

for name, palette, use_case in business:
    print(f"\n{name} - {use_case}:")
    print(f"  Colors: {palette.number}")
    print(f"  Sample: {palette.hex_colors[:4]}")

# Color distinctiveness analysis
colors = Bold_10.colors
avg_brightness = sum((0.299*r + 0.587*g + 0.114*b) for r,g,b in colors) / len(colors)
print(f"\nBold_10 average brightness: {avg_brightness:.1f}")

print("Example 3 completed")
```

## Example 4: Custom Gradient Creation

```python
# Example 4: Custom color gradients
print("Example 4 - Custom gradients")

def create_gradient(start_rgb, end_rgb, steps=8):
    """Create a color gradient."""
    sr, sg, sb = start_rgb
    er, eg, eb = end_rgb
    
    colors = []
    for i in range(steps):
        ratio = i / (steps - 1)
        r = int(sr + (er - sr) * ratio)
        g = int(sg + (eg - sg) * ratio)
        b = int(sb + (eb - sb) * ratio)
        colors.append(f'#{r:02x}{g:02x}{b:02x}')
    return colors

gradients = [
    ("Blue→Red", (0,0,255), (255,0,0)),
    ("Green→Yellow", (0,128,0), (255,255,0))
]

for name, start, end in gradients:
    gradient = create_gradient(start, end)
    print(f"\n{name}: {gradient[:4]}...{gradient[-1]}")

print("Example 4 completed")
```

## Example 5: Palette Quality Assessment

```python
from palettable.colorbrewer.qualitative import Pastel1_9

# Example 5: Palette quality metrics
print("Example 5 - Palette quality")

def assess_quality(palette):
    """Calculate palette quality metrics."""
    colors = palette.colors
    brightness = [(0.299*r + 0.587*g + 0.114*b) for r,g,b in colors]
    
    return {
        "brightness_range": max(brightness) - min(brightness),
        "avg_brightness": sum(brightness) / len(brightness)
    }

test_palettes = [
    ("Set1_9", Set1_9),
    ("Pastel1_9", Pastel1_9),
    ("Viridis_10", Viridis_10)
]

for name, palette in test_palettes:
    quality = assess_quality(palette)
    print(f"\n{name}:")
    print(f"  Brightness range: {quality['brightness_range']:.1f}")
    print(f"  Average: {quality['avg_brightness']:.1f}")

print("Example 5 completed")
```
