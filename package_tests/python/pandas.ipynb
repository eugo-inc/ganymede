## Example 1: Basic DataFrame creation and inspection

```python
import pandas as pd
import numpy as np

# Example 1: Basic pandas DataFrame creation and inspection
print("Example 1: Basic pandas DataFrame operations")

# Create sample data
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'age': [25, 30, 35, 28, 32],
    'city': ['New York', 'London', 'Paris', 'Tokyo', 'Sydney'],
    'salary': [50000, 60000, 70000, 55000, 65000]
}

# Create DataFrame
df = pd.DataFrame(data)

print(f"DataFrame created with shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
print(f"Data types:\n{df.dtypes}")
print(f"\nFirst 3 rows:")
print(df.head(3))

# Basic statistics
print(f"\nBasic statistics:")
print(df.describe())

assert df.shape == (5, 4)
assert df["salary"].sum() == 300000
print(f"Pandas version: {pd.__version__}")
print("Basic DataFrame operations completed")
```

## Example 2: Data filtering, sorting, and selection

```python
import pandas as pd
import numpy as np

# Example 2: Data filtering, sorting, and selection
print("Example 2: Data filtering and selection")

# Create sample dataset
np.random.seed(42)
dates = pd.date_range('2024-01-01', periods=100, freq='D')
df = pd.DataFrame({
    'date': dates,
    'sales': np.random.randint(1000, 5000, 100),
    'region': np.random.choice(['North', 'South', 'East', 'West'], 100),
    'product': np.random.choice(['A', 'B', 'C'], 100)
})

print(f"Dataset shape: {df.shape}")
print(f"Sample data:")
print(df.head())

# Filtering operations
high_sales = df[df['sales'] > 3000]
print(f"\nHigh sales records (>3000): {len(high_sales)}")

# Multiple conditions
north_high_sales = df[(df['region'] == 'North') & (df['sales'] > 2500)]
print(f"North region high sales: {len(north_high_sales)}")

# Sorting
sorted_df = df.sort_values(['region', 'sales'], ascending=[True, False])
print(f"\nTop sales by region:")
print(sorted_df.groupby('region')['sales'].first())

# Column selection and operations
print(f"\nAverage sales by region:")
region_avg = df.groupby('region')['sales'].mean()
print(region_avg.round(2))

print("Data filtering and selection completed")
```

## Example 3: Data aggregation and groupby operations

```python
import pandas as pd
import numpy as np

# Example 3: Data aggregation and groupby operations
print("Example 3: Data aggregation and groupby")

# Create sample data
np.random.seed(42)
df = pd.DataFrame({
    'department': np.random.choice(['Sales', 'Marketing', 'Engineering', 'HR'], 200),
    'employee_id': range(1, 201),
    'salary': np.random.randint(40000, 120000, 200),
    'experience': np.random.randint(1, 20, 200),
    'performance_score': np.random.uniform(3.0, 5.0, 200)
})

print(f"Employee dataset shape: {df.shape}")
print(f"Departments: {df['department'].unique()}")

# Basic groupby operations
dept_stats = df.groupby('department').agg({
    'salary': ['mean', 'median', 'count'],
    'experience': 'mean',
    'performance_score': 'mean'
})

print(f"\nDepartment statistics:")
print(dept_stats.round(2))

# Multiple groupby with custom aggregations
experience_bins = pd.cut(df['experience'], bins=[0, 5, 10, 20], labels=['Junior', 'Mid', 'Senior'])
df['experience_level'] = experience_bins

cross_stats = df.groupby(['department', 'experience_level']).agg({
    'salary': 'mean',
    'performance_score': 'mean'
}).round(2)

print(f"\nDepartment vs Experience Level:")
print(cross_stats)

# Apply custom functions
def salary_category(x):
    if x < 60000:
        return 'Low'
    elif x < 90000:
        return 'Medium'
    else:
        return 'High'

df['salary_category'] = df['salary'].apply(salary_category)
print(f"\nSalary categories distribution:")
print(df['salary_category'].value_counts())

print("Data aggregation completed")
```

## Example 4: Data merging, joining, and reshaping

```python
import pandas as pd
import numpy as np
from io import StringIO

# Example 4: Data merging, joining, and reshaping
print("Example 4: Data merging and reshaping")

# Create sample datasets
customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 5],
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'city': ['NYC', 'LA', 'Chicago', 'Miami', 'Seattle']
})

orders = pd.DataFrame({
    'order_id': [101, 102, 103, 104, 105, 106],
    'customer_id': [1, 2, 1, 3, 4, 2],
    'product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Laptop', 'Tablet'],
    'amount': [1200, 25, 75, 300, 1200, 400]
})

print("Customers data:")
print(customers)
print(f"\nOrders data:")
print(orders)

# Inner join
merged_data = pd.merge(customers, orders, on='customer_id', how='inner')
print(f"\nMerged data (inner join):")
print(merged_data)

# Aggregation after merge
customer_totals = merged_data.groupby(['customer_id', 'name']).agg({
    'amount': 'sum',
    'order_id': 'count'
}).rename(columns={'order_id': 'order_count'})

assert len(merged_data) == 6
assert customer_totals.loc[(1, "Alice"), "amount"] == 1275
print(f"\nCustomer order totals:")
print(customer_totals)

# Pivot table
pivot_table = merged_data.pivot_table(
    index='name', 
    columns='product', 
    values='amount', 
    aggfunc='sum', 
    fill_value=0
)

print(f"\nPivot table - Customer vs Product:")
print(pivot_table)

# Reshaping with melt
melted = pivot_table.reset_index().melt(
    id_vars='name', 
    var_name='product', 
    value_name='total_spent'
)
print(f"\nMelted data:")
print(melted[melted['total_spent'] > 0].head())

print("Data merging and reshaping completed")
```

## Example 5: Time series analysis and advanced pandas features

```python
import pandas as pd
import numpy as np

# Example 5: Time series analysis and advanced pandas features
print("Example 5: Time series and advanced features")

# Create time series data
dates = pd.date_range('2024-01-01', periods=365, freq='D')
np.random.seed(42)

# Simulate stock-like data with trend and seasonality
trend = np.linspace(100, 150, 365)
seasonal = 10 * np.sin(2 * np.pi * np.arange(365) / 365.25 * 4)  # Quarterly seasonality
noise = np.random.normal(0, 5, 365)
values = trend + seasonal + noise

ts_df = pd.DataFrame({
    'date': dates,
    'price': values,
    'volume': np.random.randint(1000, 10000, 365)
})

ts_df.set_index('date', inplace=True)

print(f"Time series data shape: {ts_df.shape}")
print(f"Date range: {ts_df.index.min()} to {ts_df.index.max()}")
print(f"Sample data:")
print(ts_df.head())

# Time-based operations
print(f"\nMonthly averages:")
# 'ME' == month-end (pandas >= 2.2; the old 'M' alias was deprecated in 2.2 and removed in 3.0)
monthly_avg = ts_df.resample('ME').mean()
print(monthly_avg.head())

# Rolling statistics
ts_df['price_ma_7'] = ts_df['price'].rolling(window=7).mean()
ts_df['price_ma_30'] = ts_df['price'].rolling(window=30).mean()

print(f"\nRolling averages sample:")
print(ts_df[['price', 'price_ma_7', 'price_ma_30']].head(10))

# Percentage change and volatility
ts_df['daily_return'] = ts_df['price'].pct_change()
ts_df['volatility_30d'] = ts_df['daily_return'].rolling(window=30).std()

print(f"\nPrice statistics:")
print(f"Average price: ${ts_df['price'].mean():.2f}")
print(f"Price volatility (30-day): {ts_df['volatility_30d'].mean():.4f}")
print(f"Best day return: {ts_df['daily_return'].max():.4f}")
print(f"Worst day return: {ts_df['daily_return'].min():.4f}")

# Advanced indexing and selection
q4_data = ts_df['2024-10':'2024-12']
print(f"\nQ4 data points: {len(q4_data)}")
print(f"Q4 average price: ${q4_data['price'].mean():.2f}")

print("Time series analysis completed")
print(f"\nPandas capabilities demonstrated:")
print("- DataFrame creation and manipulation")
print("- Data filtering, sorting, and selection") 
print("- Groupby operations and aggregations")
print("- Data merging and reshaping")
print("- Time series analysis and rolling statistics")
```
