## Example 1: Type Annotations for DataFrame Operations

```python
import pandas as pd
from typing import reveal_type

# Example 1: Type annotations from pandas-stubs
print("Example 1 - DataFrame type annotations")

# pandas-stubs provides type hints for DataFrame creation
df: pd.DataFrame = pd.DataFrame({'A': [1, 2, 3], 'B': [4.0, 5.0, 6.0]})
print(f"Created DataFrame: {df.shape}")

# Type stubs ensure proper return types
subset: pd.DataFrame = df.head(2)
print(f"head() returns DataFrame: {type(subset).__name__}")

# Series type annotation
series: pd.Series = df['A']
print(f"Column access returns Series: {type(series).__name__}")

# Aggregation returns Series
result: pd.Series = df.mean()
print(f"mean() returns Series: {type(result).__name__}")

# Show that types match expectations
assert isinstance(subset, pd.DataFrame)
assert isinstance(series, pd.Series)

print("Type annotations verified")
```

## Example 2: Series Type Inference

```python
import pandas as pd
import numpy as np

# Example 2: Series type annotations
print("Example 2 - Series type inference")

# pandas-stubs provides generic Series types
int_series: pd.Series = pd.Series([1, 2, 3, 4, 5])
float_series: pd.Series = pd.Series([1.1, 2.2, 3.3])
str_series: pd.Series = pd.Series(['a', 'b', 'c'])

print(f"Int series dtype: {int_series.dtype}")
print(f"Float series dtype: {float_series.dtype}")
print(f"String series dtype: {str_series.dtype}")

# Operations maintain type information
doubled: pd.Series = int_series * 2
print(f"\nDoubled series: {doubled.tolist()}")

# Boolean indexing returns Series
filtered: pd.Series = int_series[int_series > 2]
print(f"Filtered series: {filtered.tolist()}")

print("Series type inference working")
```

## Example 3: GroupBy Type Safety

```python
import pandas as pd

# Example 3: GroupBy operations with type hints
print("Example 3 - GroupBy type safety")

# Create DataFrame for grouping
df = pd.DataFrame({
    'category': ['A', 'B', 'A', 'B', 'A'],
    'value': [10, 20, 30, 40, 50]
})

# pandas-stubs provides DataFrameGroupBy type
grouped = df.groupby('category')
print(f"GroupBy object: {type(grouped).__name__}")

# Aggregations return DataFrame
result: pd.DataFrame = grouped.sum()
print(f"\nGrouped sum:\n{result}")

# Different aggregations
mean_result: pd.DataFrame = grouped.mean()
count_result: pd.DataFrame = grouped.count()

print(f"\nMean by category:\n{mean_result}")

print("GroupBy types verified")
```

## Example 4: IO Operations with Type Hints

```python
import pandas as pd
from io import StringIO

# Example 4: IO operations type safety
print("Example 4 - IO type annotations")

df = pd.DataFrame({'x': [1, 2, 3], 'y': [4, 5, 6]})

# pandas-stubs ensures correct return types
json_str: str = df.to_json()
print(f"to_json() returns str: {type(json_str).__name__}")

csv_str: str = df.to_csv(index=False)
print(f"to_csv() returns str: {type(csv_str).__name__}")

# Reading returns DataFrame
csv_data = StringIO(csv_str)
df_read: pd.DataFrame = pd.read_csv(csv_data)
print(f"\nread_csv() returns DataFrame: {type(df_read).__name__}")
print(f"Shape: {df_read.shape}")

# JSON reading
df_json: pd.DataFrame = pd.read_json(StringIO(json_str))
print(f"read_json() returns DataFrame: {df_json.shape}")

print("IO type safety verified")
```

## Example 5: Index and Indexing Type Annotations

```python
import pandas as pd

# Example 5: Index type safety
print("Example 5 - Index type annotations")

# pandas-stubs provides Index types
df = pd.DataFrame(
    {'A': [1, 2, 3], 'B': [4, 5, 6]},
    index=['x', 'y', 'z']
)

# Index access with type hints
idx: pd.Index = df.index
print(f"Index type: {type(idx).__name__}")
print(f"Index values: {idx.tolist()}")

# loc returns DataFrame or Series
row: pd.Series = df.loc['x']
print(f"\nloc['x'] returns Series: {type(row).__name__}")

subset_df: pd.DataFrame = df.loc[['x', 'y']]
print(f"loc[['x','y']] returns DataFrame: {type(subset_df).__name__}")

# Column access maintains type
col: pd.Series = df.A
print(f"\nAttribute access returns Series: {type(col).__name__}")

print("Index types verified")
```
