## Example 1: Basic AST Elements

```python
import pandocfilters as pf

# Example 1: Basic AST elements
print("Example 1 - Basic elements")

# Text elements
text = pf.Str("Hello")
emph = pf.Emph([pf.Str("emphasized")])
strong = pf.Strong([pf.Str("bold")])

print(f"Str: {text}")
print(f"Emph: {emph}")
print(f"Strong: {strong}")

# Paragraph
para = pf.Para([pf.Str("Text "), pf.Emph([pf.Str("here")])])
print(f"Para: {para['t']}")

# Link and Image
link = pf.Link(['',[],[]], [pf.Str("Click")], ["http://example.com", ""])
img = pf.Image(['',[],[]], [pf.Str("Alt")], ["img.png", ""])

print(f"Link: {link['t']}")
print(f"Image: {img['t']}")

print("Example 1 completed")
```

## Example 2: Filter Functions

```python
import pandocfilters as pf

# Example 2: Filter functions
print("Example 2 - Filters")

def caps_filter(key, value, format, meta):
    if key == 'Str':
        return pf.Str(value.upper())

doc = [pf.Para([pf.Str("hello world")])]
print(f"Original: {doc[0]}")

filtered = pf.walk(doc, caps_filter, format='html', meta={})
print(f"Filtered: {filtered[0]}")

# Stringify function
text = pf.stringify([pf.Str("Hello "), pf.Emph([pf.Str("world")])])
print(f"Stringified: '{text}'")
assert filtered[0]["c"][0] == pf.Str("HELLO WORLD")
assert text == "Hello world"

# Convert emphasis to strong
def emph_to_strong(key, value, format, meta):
    if key == 'Emph':
        return pf.Strong(value)

print("Filter functions working")
print("Example 2 completed")
```

## Example 3: Block Elements

```python
import pandocfilters as pf

# Example 3: Block elements
print("Example 3 - Block elements")

# Headers
h1 = pf.Header(1, ('', [], []), [pf.Str("Title")])
h2 = pf.Header(2, ('', [], []), [pf.Str("Subtitle")])
print(f"Header 1: {h1['t']}")
print(f"Header 2: {h2['t']}")

# Lists
bullet = pf.BulletList([[pf.Para([pf.Str("Item 1")])], [pf.Para([pf.Str("Item 2")])]])
print(f"BulletList: {len(bullet['c'])} items")

# Code block
code = pf.CodeBlock(('', ['python'], []), "print('hello')")
print(f"CodeBlock: {code['t']}")

# Block quote
quote = pf.BlockQuote([pf.Para([pf.Str("Quoted text")])])
print(f"BlockQuote: {quote['t']}")

# Horizontal rule
hr = pf.HorizontalRule()
print(f"HorizontalRule: {hr}")

print("Example 3 completed")
```

## Example 4: JSON Processing

```python
import pandocfilters as pf
import json

# Example 4: JSON processing
print("Example 4 - JSON processing")

# Create document
doc = {
    "pandoc-api-version": [1, 17, 5, 4],
    "meta": {},
    "blocks": [pf.Para([pf.Str("test")])]
}

json_str = json.dumps(doc)
print(f"JSON length: {len(json_str)}")

# Filter to uppercase
def upper_filter(key, value, format, meta):
    if key == 'Str':
        return pf.Str(value.upper())

filtered = pf.walk(doc['blocks'], upper_filter, 'html', {})
print(f"Filtered: {filtered[0]}")
assert json.loads(json_str)["blocks"][0] == pf.Para([pf.Str("test")])
assert filtered[0] == pf.Para([pf.Str("TEST")])

print("Example 4 completed")
```

## Example 5: Advanced Elements

```python
import pandocfilters as pf

# Example 5: Advanced elements
print("Example 5 - Advanced elements")

# Div with attributes
div = pf.Div(('id', ['class1'], [['key', 'val']]), [pf.Para([pf.Str("Content")])])
print(f"Div: {div['t']}")

# Math elements
inline_math = pf.Math('InlineMath', r'x^2')
display_math = pf.Math('DisplayMath', r'\int_0^1 x dx')
print(f"InlineMath: {inline_math['t']}")
print(f"DisplayMath: {display_math['t']}")

# RawBlock
raw = pf.RawBlock('html', '<div>HTML</div>')
print(f"RawBlock: {raw['t']}")

# LineBreak and Space
lb = pf.LineBreak()
space = pf.Space()
print(f"LineBreak: {lb}")
print(f"Space: {space}")

print("Example 5 completed")
```
