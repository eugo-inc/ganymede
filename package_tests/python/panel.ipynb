## Example 1: Basic Panel components and layout

```python
import panel as pn

# Example 1: Basic Panel components and layout
print("Example 1: Testing basic Panel components and layout")

try:
    # Enable Panel extension (required for Jupyter)
    pn.extension()
    
    print(f"Panel version: {pn.__version__}")
    print("Panel extension enabled")
    
    # Create basic panes
    html_pane = pn.pane.HTML("<h2>Welcome to Panel</h2><p>This is HTML content.</p>")
    markdown_pane = pn.pane.Markdown("# Markdown Title\nThis is **bold** text.")
    
    print(f"HTML pane created: {type(html_pane).__name__}")
    print(f"Markdown pane created: {type(markdown_pane).__name__}")
    
    # Create layout components
    column_layout = pn.Column(
        html_pane,
        markdown_pane,
        pn.pane.HTML("<hr>"),
        width=400
    )
    
    row_layout = pn.Row(
        pn.pane.HTML("Left content"),
        pn.HSpacer(),
        pn.pane.HTML("Right content")
    )
    
    print(f"Column layout created with {len(column_layout)} objects")
    print(f"Row layout created with {len(row_layout)} objects")
    
    # Test layout properties
    print(f"Column width: {column_layout.width}")
    print(f"Layout objects available: Row, Column, Tabs, GridSpec")
    
    print("panel example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"panel example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Interactive widgets and controls

```python
import panel as pn

# Example 2: Interactive widgets
print("Example 2 - Interactive widgets")

pn.extension()

# Create widgets
slider = pn.widgets.FloatSlider(name="Float Slider", start=0, end=100, value=50)
text = pn.widgets.TextInput(name="Text", value="Hello")
select = pn.widgets.Select(name="Select", options=["A", "B", "C"], value="A")
button = pn.widgets.Button(name="Click Me", button_type="primary")

print(f"Slider: {type(slider).__name__} (value={slider.value})")
print(f"Text: {type(text).__name__}")
print(f"Select: {type(select).__name__}")
print(f"Button: {type(button).__name__}")

# Checkbox and toggle
checkbox = pn.widgets.Checkbox(name="Enable", value=True)
toggle = pn.widgets.Toggle(name="Toggle", value=False)

print(f"Checkbox: {checkbox.value}")
print(f"Toggle: {toggle.value}")

# Number input
number = pn.widgets.IntInput(name="Number", value=42, start=0, end=100)
print(f"Number: {type(number).__name__} (value={number.value})")

print("Example 2 completed")
```

## Example 3: Data visualization integration

```python
import panel as pn
import pandas as pd
import numpy as np

# Example 3: Data visualization integration
print("Example 3: Testing data visualization integration")

try:
    # Create sample data
    np.random.seed(42)
    data = pd.DataFrame({
        'x': range(20),
        'y': np.random.randn(20).cumsum(),
        'category': np.random.choice(['A', 'B', 'C'], 20)
    })
    
    print(f"Sample data created: {data.shape}")
    
    # Create DataFrame pane
    df_pane = pn.pane.DataFrame(
        data.head(10), 
        name="Data Table",
        width=400,
        height=300
    )
    
    print(f"DataFrame pane created: {type(df_pane).__name__}")
    
    # Create plot using matplotlib
    try:
        import matplotlib.pyplot as plt
        
        fig, ax = plt.subplots(figsize=(6, 4))
        ax.plot(data['x'], data['y'], 'b-o')
        ax.set_title('Sample Line Plot')
        ax.set_xlabel('X values')
        ax.set_ylabel('Y values')
        plt.tight_layout()
        
        # Wrap matplotlib figure in Panel
        plot_pane = pn.pane.Matplotlib(fig, dpi=144)
        print(f"Matplotlib pane created: {type(plot_pane).__name__}")
        
        plt.close(fig)  # Clean up
        
    except ImportError:
        print("Matplotlib not available for plotting")
        plot_pane = pn.pane.HTML("<p>Plot would be displayed here</p>")
    
    # Create bokeh plot if available
    try:
        from bokeh.plotting import figure
        from bokeh.models import HoverTool
        
        p = figure(title="Interactive Bokeh Plot", width=400, height=300)
        p.circle(data['x'], data['y'], size=8, alpha=0.7)
        p.line(data['x'], data['y'], line_width=2)
        
        bokeh_pane = pn.pane.Bokeh(p)
        print(f"Bokeh pane created: {type(bokeh_pane).__name__}")
        
    except ImportError:
        print("Bokeh not available for interactive plotting")
        bokeh_pane = pn.pane.HTML("<p>Interactive plot would be displayed here</p>")
    
    # Create combined dashboard layout
    dashboard = pn.template.FastListTemplate(
        title="Data Dashboard",
        sidebar=[
            pn.pane.Markdown("## Controls"),
            pn.widgets.Select(name="Dataset", options=["Sample Data", "Test Data"]),
            pn.widgets.Checkbox(name="Show Grid", value=True)
        ],
        main=[
            pn.Row(df_pane, plot_pane),
            bokeh_pane
        ]
    )
    
    print(f"Dashboard template created: {type(dashboard).__name__}")
    print(f"Dashboard components: {len(dashboard.sidebar)} sidebar + {len(dashboard.main)} main")
    
    print("panel example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"panel example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: Reactive programming and callbacks

```python
import panel as pn
import param

# Example 4: Reactive programming and callbacks
print("Example 4: Testing reactive programming and callbacks")

try:
    # Create reactive widgets
    number_input = pn.widgets.NumberInput(
        name="Input Value", 
        value=10, 
        start=0, 
        end=100
    )
    
    multiplier_slider = pn.widgets.FloatSlider(
        name="Multiplier",
        value=2.0,
        start=0.1,
        end=10.0,
        step=0.1
    )
    
    print(f"Number input: {type(number_input).__name__} (value={number_input.value})")
    print(f"Multiplier slider: {type(multiplier_slider).__name__} (value={multiplier_slider.value})")
    
    # Create reactive function using pn.bind
    @pn.depends(number_input.param.value, multiplier_slider.param.value)
    def calculate_result(number, multiplier):
        result = number * multiplier
        return pn.pane.HTML(f"""
        <div style="padding: 10px; background: #f0f0f0; border-radius: 5px;">
            <h3>Calculation Result</h3>
            <p><strong>{number}</strong> × <strong>{multiplier}</strong> = <strong>{result}</strong></p>
        </div>
        """)
    
    print("Reactive calculation function defined with pn.depends")
    
    # Test the reactive function
    result_pane = calculate_result
    print(f"Result pane function created: {callable(result_pane)}")
    
    # Create param-based reactive class
    class ReactiveCalculator(param.Parameterized):
        value = param.Number(default=5, bounds=(0, 100))
        operation = param.ObjectSelector(default="square", objects=["square", "cube", "sqrt"])
        
        @param.depends('value', 'operation')
        def calculate(self):
            if self.operation == "square":
                result = self.value ** 2
                op_text = "squared"
            elif self.operation == "cube":
                result = self.value ** 3
                op_text = "cubed"
            else:  # sqrt
                result = self.value ** 0.5
                op_text = "square root"
            
            return f"{self.value} {op_text} = {result:.2f}"
    
    calculator = ReactiveCalculator()
    print(f"Reactive calculator created: {type(calculator).__name__}")
    print(f"Initial calculation: {calculator.calculate()}")
    
    # Test parameter updates
    calculator.value = 8
    calculator.operation = "cube"
    print(f"Updated calculation: {calculator.calculate()}")
    assert calculator.calculate() == "8 cubed = 512.00"
    
    # Create interactive dashboard with callbacks
    def create_interactive_app():
        # Status display
        status_pane = pn.pane.HTML("Ready")
        
        # Button with callback
        action_button = pn.widgets.Button(name="Process Data", button_type="primary")
        
        def button_callback(event):
            status_pane.object = f"<b>Processing...</b> Button clicked {event.clicks} times"
        
        action_button.on_click(button_callback)
        
        return pn.Column(
            pn.pane.Markdown("## Interactive Demo"),
            number_input,
            multiplier_slider,
            result_pane,
            pn.Spacer(height=20),
            action_button,
            status_pane
        )
    
    app = create_interactive_app()
    print(f"Interactive app created: {type(app).__name__} with {len(app)} components")
    
    print("panel example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"panel example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: Templates and deployment patterns

```python
import panel as pn

# Example 5: Templates and deployment
print("Example 5 - Templates and deployment")

pn.extension()

# Create FastListTemplate
template = pn.template.FastListTemplate(
    title="Dashboard",
    sidebar_width=300,
    header_background='#2596be'
)

print(f"Template: {type(template).__name__}")

# Add sidebar
template.sidebar.extend([
    pn.pane.Markdown("## Controls"),
    pn.widgets.Select(name="Source", options=["DB A", "DB B"], value="DB A"),
    pn.widgets.MultiChoice(name="Metrics", options=["Revenue", "Users"], value=["Revenue"])
])

# Add main content
template.main.extend([
    pn.pane.Markdown("# Dashboard"),
    pn.Row(pn.pane.HTML("<div>Chart 1</div>"), pn.pane.HTML("<div>Chart 2</div>"))
])

print(f"Sidebar: {len(template.sidebar)} items")
print(f"Main: {len(template.main)} items")
assert (len(template.sidebar), len(template.main)) == (3, 2)

print("Example 5 completed")
```
