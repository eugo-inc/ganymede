# panel_material_ui smoke test

Tests basic import + a handful of Material UI components built on Panel.

```python
# Example 1: Button widget construction.
from panel_material_ui import Button

btn = Button(label="Click me", variant="contained", color="primary")
assert btn.label == "Click me"
assert btn.variant == "contained"
assert btn.color == "primary"
print(f"Button: label={btn.label!r}, variant={btn.variant!r}, color={btn.color!r}")
```

```python
# Example 2: TextInput value round-trip.
from panel_material_ui import TextInput

ti = TextInput(label="Name", value="Ada")
assert ti.label == "Name"
assert ti.value == "Ada"

ti.value = "Grace"
assert ti.value == "Grace"
print(f"TextInput value updated to {ti.value!r}")
```

```python
# Example 3: Slider numeric widget.
from panel_material_ui import IntSlider

slider = IntSlider(name="Count", start=0, end=10, value=3)
assert slider.start == 0
assert slider.end == 10
assert slider.value == 3

slider.value = 7
assert slider.value == 7
print(f"IntSlider: range=[{slider.start},{slider.end}], value={slider.value}")
```

```python
# Example 4: Select widget options.
from panel_material_ui import Select

sel = Select(label="Pick one", options=["a", "b", "c"], value="b")
assert sel.options == ["a", "b", "c"]
assert sel.value == "b"

sel.value = "c"
assert sel.value == "c"
print(f"Select: options={sel.options}, value={sel.value!r}")
```

```python
# Example 5: Layout container (Column / Row) composition.
from panel_material_ui import Button, Column, Row, TextInput

btn = Button(label="Submit")
ti = TextInput(label="Email")
row = Row(ti, btn)
col = Column(row)

assert len(row) == 2
assert len(col) == 1
assert col[0] is row
print(f"Layout: Column with {len(col)} child(ren), Row with {len(row)} child(ren)")
```
