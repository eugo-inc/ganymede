```python
# Example 1: CSS selectors on an HTML document.
from parsel import Selector

html = """
<html>
  <body>
    <h1 class="title">Hello Parsel</h1>
    <ul id="items">
      <li class="item">alpha</li>
      <li class="item">beta</li>
      <li class="item">gamma</li>
    </ul>
  </body>
</html>
"""

sel = Selector(text=html)
title = sel.css('h1.title::text').get()
items = sel.css('ul#items li.item::text').getall()
print(f"title = {title!r}")
print(f"items = {items}")

assert title == 'Hello Parsel'
assert items == ['alpha', 'beta', 'gamma']
```

```python
# Example 2: XPath selectors + attribute extraction.
from parsel import Selector

html = """
<html>
  <body>
    <a href="https://example.com/a" class="link">A</a>
    <a href="https://example.com/b" class="link">B</a>
    <a href="https://example.com/c" class="other">C</a>
  </body>
</html>
"""

sel = Selector(text=html)
hrefs = sel.xpath('//a[@class="link"]/@href').getall()
texts = sel.xpath('//a[@class="link"]/text()').getall()
print(f"hrefs = {hrefs}")
print(f"texts = {texts}")

assert hrefs == ['https://example.com/a', 'https://example.com/b']
assert texts == ['A', 'B']
```

```python
# Example 3: SelectorList chaining + re() regex extraction.
from parsel import Selector

html = """
<html><body>
  <div class="product">
    <span class="name">Widget</span><span class="price">$19.99</span>
  </div>
  <div class="product">
    <span class="name">Gadget</span><span class="price">$42.50</span>
  </div>
</body></html>
"""

sel = Selector(text=html)
products = sel.css('div.product')
names = products.css('span.name::text').getall()
prices = products.css('span.price::text').re(r'\$(\d+\.\d+)')
print(f"names = {names}")
print(f"prices = {prices}")

assert names == ['Widget', 'Gadget']
assert prices == ['19.99', '42.50']
```

```python
# Example 4: XML parsing with namespace handling.
from parsel import Selector

xml = """<?xml version="1.0"?>
<rss xmlns:atom="http://www.w3.org/2005/Atom" version="2.0">
  <channel>
    <title>Example Feed</title>
    <item><title>Post 1</title><link>https://ex.com/1</link></item>
    <item><title>Post 2</title><link>https://ex.com/2</link></item>
  </channel>
</rss>
"""

sel = Selector(text=xml, type='xml')
channel_title = sel.xpath('//channel/title/text()').get()
item_titles = sel.xpath('//item/title/text()').getall()
item_links = sel.xpath('//item/link/text()').getall()
print(f"channel = {channel_title!r}")
print(f"item_titles = {item_titles}")
print(f"item_links = {item_links}")

assert channel_title == 'Example Feed'
assert len(item_titles) == 2
```

```python
# Example 5: JMESPath on JSON + Selector.jmespath().
from parsel import Selector

json_text = """
{
  "users": [
    {"name": "Alice", "age": 30, "role": "admin"},
    {"name": "Bob", "age": 25, "role": "user"},
    {"name": "Carol", "age": 35, "role": "admin"}
  ]
}
"""

sel = Selector(text=json_text)
admin_names = sel.jmespath('users[?role==`admin`].name').getall()
all_ages = sel.jmespath('users[*].age').getall()
print(f"admin_names = {admin_names}")
print(f"all_ages = {all_ages}")

assert admin_names == ['Alice', 'Carol']
assert all_ages == [30, 25, 35]
print("\nAll parsel smoke tests passed.")
```
