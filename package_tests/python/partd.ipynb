## Example 1: Dict Storage

```python
import partd

# Example 1: Dict storage
print("Example 1 - Dict storage")

# Create in-memory dict storage
storage = partd.Dict()
print(f"Created: {type(storage).__name__}")

# Append data
storage.append({'x': b'hello', 'y': b'world'})
storage.append({'x': b' there', 'z': b'new'})

# Read data
x_data = storage.get('x')
y_data = storage.get('y')
print(f"x: {x_data}")
print(f"y: {y_data}")

# Get multiple keys
all_data = storage.get(['x', 'y', 'z'])
print(f"All: {all_data}")
assert x_data == b'hello there'
assert all_data == [b'hello there', b'world', b'new']

print("Example 1 completed")
```

## Example 2: File Storage

```python
import partd
import tempfile
import shutil

# Example 2: File storage
print("Example 2 - File storage")

temp_dir = tempfile.mkdtemp()
storage = partd.File(temp_dir)
print(f"Created file storage: {temp_dir}")

# Store data
storage.append({'user_1': b'Alice', 'user_2': b'Bob'})
print("Data stored")

# Read data
user1 = storage.get('user_1')
print(f"User 1: {user1}")

# Append more data
storage.append({'user_1': b' - updated'})
updated = storage.get('user_1')
print(f"Updated: {updated}")
assert updated == b'Alice - updated'

shutil.rmtree(temp_dir)
print("Example 2 completed")
```

## Example 3: Pickle Storage

```python
import partd
import tempfile
import shutil

# Example 3: Pickle for Python objects
print("Example 3 - Pickle storage")

temp_dir = tempfile.mkdtemp()
storage = partd.Pickle(partd.File(temp_dir))
print(f"Created pickle storage")

# Store Python objects (must be lists)
storage.append({
    'config': [{'host': 'localhost'}],
    'metrics': [[1.5, 2.3, 3.7]]
})

# Read back
config = storage.get('config')
metrics = storage.get('metrics')
print(f"Config: {config}")
print(f"Metrics: {metrics}")

shutil.rmtree(temp_dir)
print("Example 3 completed")
```

## Example 4: Buffer with File Backend

```python
import partd
import tempfile
import shutil

# Example 4: Buffer with file backend
print("Example 4 - Buffer storage")

temp_dir = tempfile.mkdtemp()
storage = partd.Buffer(partd.Dict(), partd.File(temp_dir))
print(f"Created buffer with file backend")

# Add data to buffer
for i in range(3):
    storage.append({f'key_{i}': f'value_{i}'.encode()})
print("Added data to buffer")

# Read data
data = storage.get('key_0')
print(f"Key 0: {data}")

# Flush to disk
storage.flush()
print("Flushed to disk")

shutil.rmtree(temp_dir)
print("Example 4 completed")
```

## Example 5: PandasBlocks Storage

```python
import partd
import pandas as pd
import tempfile
import shutil

# Example 5: PandasBlocks for DataFrames
print("Example 5 - PandasBlocks")

temp_dir = tempfile.mkdtemp()
storage = partd.PandasBlocks(partd.File(temp_dir))
print("Created PandasBlocks storage")

# Store DataFrames
df1 = pd.DataFrame({'a': [1, 2], 'b': [3, 4]})
df2 = pd.DataFrame({'a': [5, 6], 'b': [7, 8]})

storage.append({'data1': df1, 'data2': df2})
print("Stored DataFrames")

# Read back
result = storage.get('data1')
print(f"Retrieved DataFrame:\n{result}")

shutil.rmtree(temp_dir)
print("Example 5 completed")
```
