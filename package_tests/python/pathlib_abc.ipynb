## Example 1: JoinablePath Abstract Base

```python
from pathlib_abc import JoinablePath
import pathlib

# Example 1: JoinablePath abstract base
print("Example 1 - JoinablePath")

# Show abstract base class
print(f"JoinablePath: {JoinablePath}")

# Check abstract methods
abstract_methods = getattr(JoinablePath, '__abstractmethods__', set())
print(f"Abstract methods: {sorted(abstract_methods)}")
assert abstract_methods, "JoinablePath should declare abstract methods"

# Standard Path compatibility
std_path = pathlib.Path("/home/user/file.txt")
print(f"\nStandard Path: {std_path}")
print(f"Is JoinablePath: {isinstance(std_path, JoinablePath)}")

# Show available methods
methods = [m for m in dir(JoinablePath) if not m.startswith('_')]
print(f"Public methods: {len(methods)}")
print(f"Sample: {methods[:5]}")

print("Example 1 completed")
```

## Example 4: Path Class Hierarchy

```python
from pathlib_abc import JoinablePath, ReadablePath, WritablePath
import pathlib

# Example 4: Path hierarchy
print("Example 4 - Path hierarchy")

# Test standard path against all ABC types
std_path = pathlib.Path("/tmp/test.txt")

tests = [
    ("JoinablePath", JoinablePath),
    ("ReadablePath", ReadablePath),
    ("WritablePath", WritablePath)
]

for name, abc_class in tests:
    result = isinstance(std_path, abc_class)
    print(f"{name}: {result}")

# Show MRO
# Readable and writable paths both build on JoinablePath
assert issubclass(ReadablePath, JoinablePath)
assert issubclass(WritablePath, JoinablePath)

mro = [c.__name__ for c in pathlib.Path.__mro__[:5]]
print(f"\nPath MRO: {mro}")

print("Example 4 completed")
```

## Example 5: Module Information

```python
import pathlib_abc
import pathlib

# Example 5: Module info
print("Example 5 - Module info")

# Show module version
version = getattr(pathlib_abc, '__version__', 'Not available')
print(f"Version: {version}")

# Show public exports
exports = [name for name in dir(pathlib_abc) if not name.startswith('_')]
print(f"\nPublic exports: {exports}")

# Show ABC classes
from pathlib_abc import JoinablePath, ReadablePath, WritablePath
classes = [JoinablePath, ReadablePath, WritablePath]

for cls in classes:
    abstract_count = len(getattr(cls, '__abstractmethods__', set()))
    print(f"\n{cls.__name__}:")
    print(f"  Abstract methods: {abstract_count}")

print("\nExample 5 completed")
```
