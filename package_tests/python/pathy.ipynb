## Example 1: Basic Pathy Usage

```python
from pathy import Pathy, use_fs

# Example 1: Basic Pathy usage
print("Example 1 - Basic Pathy")

# Enable filesystem adapter for testing
use_fs()

# Create cloud paths
gs_path = Pathy("gs://my-bucket/data/file.csv")
s3_path = Pathy("s3://my-storage/logs/app.log")

print(f"GCS path: {gs_path}")
print(f"S3 path: {s3_path}")

# Path properties
print(f"\nGCS properties:")
print(f"  Scheme: {gs_path.scheme}")
print(f"  Parts: {gs_path.parts}")
print(f"  Name: {gs_path.name}")
print(f"  Suffix: {gs_path.suffix}")

print("Example 1 completed")
```

## Example 2: File Operations

```python
from pathy import Pathy, use_fs

# Example 2: File operations
print("Example 2 - File operations")

use_fs()

# Create file path
file_path = Pathy("gs://test-bucket/data/sample.txt")
file_path.parent.mkdir(parents=True, exist_ok=True)

# Write content
content = "Hello, Pathy!\nTest content."
file_path.write_text(content)
print(f"Wrote to: {file_path}")

# Read content
read_content = file_path.read_text()
print(f"Read: {read_content[:30]}...")

# Check existence
print(f"Exists: {file_path.exists()}")
assert read_content == content
assert file_path.exists()

print("Example 2 completed")
```

## Example 3: Path Operations

```python
from pathy import Pathy, use_fs

# Example 3: Path operations
print("Example 3 - Path operations")

use_fs()

base_path = Pathy("gs://bucket/data/file.txt")

# Path joining
nested = base_path.parent / "nested" / "deep.json"
print(f"Nested: {nested}")

# Path manipulation
backup = base_path.with_suffix('.bak')
renamed = base_path.with_name('new_file.txt')
print(f"Backup: {backup}")
print(f"Renamed: {renamed}")
assert str(nested) == "gs://bucket/data/nested/deep.json"
assert str(renamed) == "gs://bucket/data/new_file.txt"

# Parent navigation
print(f"Parent: {base_path.parent}")
print(f"Parents: {list(base_path.parents)[:2]}")

print("Example 3 completed")
```

## Example 4: Pattern Matching

```python
from pathy import Pathy, use_fs

# Example 4: Pattern matching
print("Example 4 - Pattern matching")

use_fs()

# Create test files
base = Pathy("gs://test-bucket/data/")
for i in range(3):
    file = base / f"file_{i}.txt"
    file.parent.mkdir(parents=True, exist_ok=True)
    file.write_text(f"Content {i}")

# Glob patterns
txt_files = list(base.glob("*.txt"))
print(f"Found {len(txt_files)} .txt files")

# Recursive glob
all_files = list(Pathy("gs://test-bucket/").rglob("*.txt"))
print(f"Total .txt files: {len(all_files)}")

print("Example 4 completed")
```

## Example 5: Pathlib Compatibility

```python
from pathy import Pathy, use_fs
import pathlib

# Example 5: Pathlib compatibility
print("Example 5 - Pathlib compatibility")

use_fs()

# Compare interfaces
local_path = pathlib.Path("/tmp/file.txt")
cloud_path = Pathy("gs://bucket/file.txt")

methods = ['parts', 'name', 'suffix', 'parent', 'with_suffix', 'is_absolute']

print("Method compatibility:")
for method in methods:
    local_has = hasattr(local_path, method)
    cloud_has = hasattr(cloud_path, method)
    print(f"  {method}: local={local_has}, cloud={cloud_has}")

# Show both work the same way
print(f"\nCloud suffix: {cloud_path.suffix}")
print(f"Local suffix: {local_path.suffix}")

print("Example 5 completed")
```
