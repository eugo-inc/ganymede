# pip: version, installed distributions, help and the python -m pip command line

```python
# Example 1: Check pip version
import re
from importlib.metadata import version
import pip

print(pip.__version__)
assert re.match(r"\d+\.\d+", pip.__version__), pip.__version__
assert pip.__version__ == version("pip"), (pip.__version__, version("pip"))
```

```python
# Example 2: List installed packages via importlib.metadata
# importlib.metadata lists the installed distributions
from importlib.metadata import distributions

dists = sorted(
    (dist.metadata["Name"], dist.version)
    for dist in distributions()
    if dist.metadata["Name"] is not None
)
print(f"{len(dists)} installed distributions")
for name, version in dists[:20]:
    print(name, version)
names = {name.lower() for name, _ in dists}
assert len(dists) >= 1
assert "pip" in names, sorted(names)[:20]
```

```python
# Example 3: Ask pip for help on a command
import os
import subprocess
import sys

result = subprocess.run(
    [sys.executable, "-m", "pip", "help", "install"],
    capture_output=True, text=True, check=True, timeout=120,
    env={**os.environ, "PIP_DISABLE_PIP_VERSION_CHECK": "1"},
)
print(result.stdout[:400])
assert "Usage:" in result.stdout, result.stdout[:400]
assert "pip install" in result.stdout, result.stdout[:400]
```

```python
# Example 4: Show pip's location
import importlib.util
import os
import pip

print(pip.__file__)
assert os.path.isfile(pip.__file__), pip.__file__
assert os.path.basename(os.path.dirname(pip.__file__)) == "pip", pip.__file__
assert importlib.util.find_spec("pip").origin == pip.__file__
```

```python
# Example 5: pip CLI via `python -m pip` (the supported invocation surface)
import subprocess
import sys

import pip

result = subprocess.run(
    [sys.executable, "-m", "pip", "--version"],
    capture_output=True, text=True, check=True, timeout=120,
)
print(result.stdout.strip())
assert pip.__version__ in result.stdout
```
