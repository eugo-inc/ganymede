# platformdirs: per-user and site-wide data, config and cache directories

```python
# Example 1: Get user data directory
import os
from pathlib import Path
from platformdirs import user_data_dir, user_data_path

data_dir = user_data_dir('MyApp')
print(data_dir)
assert isinstance(data_dir, str) and 'MyApp' in data_dir, data_dir
assert os.path.isabs(data_dir), data_dir
assert user_data_path('MyApp') == Path(data_dir)  # the *_path variants return the same location as a Path
```

```python
# Example 2: Get user config directory
import os
from platformdirs import user_config_dir

config_dir = user_config_dir('MyApp')
print(config_dir)
assert isinstance(config_dir, str) and 'MyApp' in config_dir, config_dir
assert os.path.isabs(config_dir), config_dir
```

```python
# Example 3: Get user cache directory
import os
from platformdirs import PlatformDirs, user_cache_dir

cache_dir = user_cache_dir('MyApp')
print(cache_dir)
assert isinstance(cache_dir, str) and 'MyApp' in cache_dir, cache_dir
assert os.path.isabs(cache_dir), cache_dir

# The PlatformDirs object gives the same answers as the module-level functions
dirs = PlatformDirs('MyApp')
assert dirs.user_cache_dir == cache_dir, (dirs.user_cache_dir, cache_dir)
# A version is appended as the last path component
versioned = PlatformDirs('MyApp', version='1.0').user_cache_dir
print(versioned)
assert os.path.basename(versioned) == '1.0', versioned
```

```python
# Example 4: Get site data directory
import os
from platformdirs import site_data_dir

site_dir = site_data_dir('MyApp')
print(site_dir)
assert isinstance(site_dir, str) and 'MyApp' in site_dir, site_dir
assert os.path.isabs(site_dir), site_dir
```

```python
# Example 5: Show platformdirs version
from importlib.metadata import version
import platformdirs

print(platformdirs.__version__)
assert platformdirs.__version__ == version('platformdirs'), (platformdirs.__version__, version('platformdirs'))
```
