## Example 1: A line plot rendered to text

```python
import plotext as plt

# Example 1: build() returns the chart as a string instead of drawing it on the terminal.
plt.clear_figure()
plt.plotsize(60, 15)
plt.plot([x * x for x in range(20)], label="x squared")
plt.title("line")
chart = plt.build()
print(chart)
assert "line" in chart and "x squared" in chart
assert len(chart.splitlines()) == 15, len(chart.splitlines())
```

## Example 2: Scatter plot with two series and axis labels

```python
import plotext as plt

# Example 2: two scatter series, axis labels and limits.
plt.clear_figure()
plt.plotsize(60, 18)
plt.scatter([1, 2, 3, 4], [4, 3, 2, 1], label="down", marker="dot")
plt.scatter([1, 2, 3, 4], [1, 2, 3, 4], label="up", marker="x")
plt.xlabel("step")
plt.ylabel("value")
plt.xlim(0, 5)
chart = plt.build()
print(chart)
assert "step" in chart and "down" in chart and "up" in chart
```

## Example 3: Bar chart (what aiconfigurator's terminal report draws)

```python
import plotext as plt

# Example 3: a horizontal bar chart of named values.
plt.clear_figure()
plt.plotsize(60, 12)
plt.bar(["tp1", "tp2", "tp4", "tp8"], [120.0, 70.5, 41.2, 30.9], orientation="horizontal")
plt.title("latency by parallelism")
chart = plt.build()
print(chart)
for name in ("tp1", "tp2", "tp4", "tp8"):
    assert name in chart, name
```

## Example 4: Histogram of two samples

```python
import random
import plotext as plt

# Example 4: hist() bins raw samples itself; two series share one chart, each with a label.
rng = random.Random(7)
fast = [rng.gauss(10, 2) for _ in range(500)]
slow = [rng.gauss(16, 3) for _ in range(500)]
plt.clear_figure()
plt.plotsize(70, 16)
plt.hist(fast, bins=20, label="fast")
plt.hist(slow, bins=20, label="slow")
plt.title("latency histogram")
chart = plt.uncolorize(plt.build())
print(chart)
assert "latency histogram" in chart and "fast" in chart and "slow" in chart
```

## Example 5: Saving a plot as plain text

```python
import os
import tempfile
import plotext as plt

# Example 5: save_fig writes the last built chart (without colour codes) to a file; build first.
plt.clear_figure()
plt.plotsize(50, 10)
plt.plot([0, 1, 0, 1, 0])
plt.title("saved")
plt.build()
with tempfile.TemporaryDirectory() as folder:
    path = os.path.join(folder, "chart.txt")
    plt.save_fig(path)
    text = open(path, encoding="utf-8").read()
print(text)
assert "saved" in text and "\x1b[" not in text
```
