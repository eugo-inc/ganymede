## Example 1: Basic setup and simple charts

```python
import plotly.graph_objects as go
import plotly.express as px
import plotly
import plotly.colors
import plotly.figure_factory
import plotly.subplots
import numpy as np

# Example 1: Basic Plotly Setup and Simple Charts
print("Example 1: Basic Plotly setup and simple charts")

# Show plotly version and setup
print(f"Plotly version: {plotly.__version__}")
print(f"Plotly location: {plotly.__file__}")

# The submodules imported above; a missing one fails the import itself.
key_modules = ['graph_objects', 'express', 'subplots', 'figure_factory', 'colors']
for module in key_modules:
    assert hasattr(plotly, module), module
    print(f"✓ plotly.{module}")

# Create sample data
print("\n--- Sample Data Creation ---")

np.random.seed(42)
x_data = np.arange(10)
y_data = np.random.randint(10, 100, 10)
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 78, 32]

print(f"X data: {x_data}")
print(f"Y data: {y_data}")
print(f"Categories: {categories}")
print(f"Values: {values}")

# Basic line chart with graph_objects
print("\n--- Line Chart with Graph Objects ---")

try:
    fig_line = go.Figure()
    fig_line.add_trace(go.Scatter(
        x=x_data,
        y=y_data,
        mode='lines+markers',
        name='Sample Data',
        line=dict(color='blue', width=2),
        marker=dict(size=8, color='red')
    ))
    
    fig_line.update_layout(
        title='Sample Line Chart',
        xaxis_title='X Axis',
        yaxis_title='Y Axis',
        showlegend=True
    )
    
    # Get figure as dict to verify structure
    fig_dict = fig_line.to_dict()
    print(f"Line chart created: {len(fig_dict['data'])} traces")
    print(f"Chart type: {fig_dict['data'][0]['type']}")
    print(f"Layout keys: {list(fig_dict['layout'].keys())[:5]}")
    assert len(fig_dict['data']) == 1
    assert fig_dict['data'][0]['type'] == 'scatter'
    assert fig_line.data[0].mode == 'lines+markers'
    assert list(fig_line.data[0].x) == list(range(10))
    assert list(fig_line.data[0].y) == y_data.tolist()
    assert fig_line.layout.title.text == 'Sample Line Chart'
    assert fig_line.layout.xaxis.title.text == 'X Axis'
    
except Exception as e:
    print(f"Line chart error: {e}")
    raise

# Basic bar chart with Plotly Express
print("\n--- Bar Chart with Plotly Express ---")

try:
    fig_bar = px.bar(
        x=categories,
        y=values,
        title='Sample Bar Chart',
        labels={'x': 'Categories', 'y': 'Values'},
        color=values,
        color_continuous_scale='viridis'
    )
    
    # Verify bar chart
    bar_dict = fig_bar.to_dict()
    print(f"Bar chart created: {len(bar_dict['data'])} traces")
    print(f"Chart type: {bar_dict['data'][0]['type']}")
    print(f"Number of bars: {len(bar_dict['data'][0]['x'])}")
    assert bar_dict['data'][0]['type'] == 'bar'
    assert len(bar_dict['data'][0]['x']) == 5
    assert list(fig_bar.data[0].x) == categories
    assert list(fig_bar.data[0].y) == values
    
except Exception as e:
    print(f"Bar chart error: {e}")
    raise

print("Basic Plotly setup and simple charts completed")
```

## Example 2: Plotly Express charts from a DataFrame

```python
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd
import numpy as np

# Example 2: Data Visualization with Plotly Express
print("Example 2: Data visualization with Plotly Express")

# Create sample dataset
print("\n--- Sample Dataset Creation ---")

np.random.seed(42)
n_points = 100

# Create synthetic dataset
data = {
    'x': np.random.normal(0, 1, n_points),
    'y': np.random.normal(0, 1, n_points),
    'category': np.random.choice(['Group A', 'Group B', 'Group C'], n_points),
    'size': np.random.uniform(10, 50, n_points),
    'value': np.random.uniform(0, 100, n_points)
}

df = pd.DataFrame(data)
print(f"Dataset created: {df.shape}")
print(f"Columns: {list(df.columns)}")
print(f"Categories: {df['category'].unique()}")

# Scatter plot
print("\n--- Scatter Plot ---")

try:
    fig_scatter = px.scatter(
        df,
        x='x',
        y='y',
        color='category',
        size='size',
        hover_data=['value'],
        title='Interactive Scatter Plot',
        width=600,
        height=400
    )
    
    scatter_dict = fig_scatter.to_dict()
    print(f"Scatter plot created: {len(scatter_dict['data'])} traces")
    print(f"Categories in plot: {len(set([trace['name'] for trace in scatter_dict['data']]))}")
    # One trace per category, holding all 100 points between them
    assert {trace.name for trace in fig_scatter.data} == {'Group A', 'Group B', 'Group C'}
    assert all(trace.type == 'scatter' for trace in fig_scatter.data)
    assert sum(len(trace.x) for trace in fig_scatter.data) == n_points
    
except Exception as e:
    print(f"Scatter plot error: {e}")
    raise

# Box plot
print("\n--- Box Plot ---")

try:
    fig_box = px.box(
        df,
        x='category',
        y='value',
        title='Box Plot by Category',
        color='category'
    )
    
    box_dict = fig_box.to_dict()
    print(f"Box plot created: {len(box_dict['data'])} traces")
    print(f"Box plot type: {box_dict['data'][0]['type']}")
    assert len(fig_box.data) == 3 and all(trace.type == 'box' for trace in fig_box.data)
    assert sum(len(trace.y) for trace in fig_box.data) == n_points
    
except Exception as e:
    print(f"Box plot error: {e}")
    raise

# Histogram
print("\n--- Histogram ---")

try:
    fig_hist = px.histogram(
        df,
        x='value',
        color='category',
        title='Value Distribution by Category',
        nbins=20,
        opacity=0.7
    )
    
    hist_dict = fig_hist.to_dict()
    print(f"Histogram created: {len(hist_dict['data'])} traces")
    print(f"Values in the first trace: {len(fig_hist.data[0].x)}")
    assert len(fig_hist.data) == 3 and all(trace.type == 'histogram' for trace in fig_hist.data)
    assert sum(len(trace.x) for trace in fig_hist.data) == n_points
    assert fig_hist.data[0].marker.opacity == 0.7
    
except Exception as e:
    print(f"Histogram error: {e}")
    raise

# Time series simulation
print("\n--- Time Series Plot ---")

try:
    # Create time series data
    dates = pd.date_range('2024-01-01', periods=30, freq='D')
    ts_data = {
        'date': dates,
        'value1': np.cumsum(np.random.normal(0, 1, 30)) + 100,
        'value2': np.cumsum(np.random.normal(0, 0.5, 30)) + 50
    }
    ts_df = pd.DataFrame(ts_data)
    
    fig_ts = px.line(
        ts_df,
        x='date',
        y=['value1', 'value2'],
        title='Time Series Example',
        labels={'value': 'Measurement', 'date': 'Date'}
    )
    
    ts_dict = fig_ts.to_dict()
    print(f"Time series created: {len(ts_dict['data'])} traces")
    print(f"Date range: {len(dates)} days")
    assert [trace.name for trace in fig_ts.data] == ['value1', 'value2']
    assert all(len(trace.x) == 30 and len(trace.y) == 30 for trace in fig_ts.data)
    
except Exception as e:
    print(f"Time series error: {e}")
    raise

# Correlation heatmap
print("\n--- Correlation Heatmap ---")

try:
    # Create correlation matrix
    numeric_cols = ['x', 'y', 'size', 'value']
    correlation_matrix = df[numeric_cols].corr()
    
    fig_heatmap = px.imshow(
        correlation_matrix,
        title='Correlation Heatmap',
        color_continuous_scale='RdBu',
        aspect='auto'
    )
    
    heatmap_dict = fig_heatmap.to_dict()
    print(f"Heatmap created: {len(heatmap_dict['data'])} traces")
    print(f"Matrix size: {correlation_matrix.shape}")
    assert heatmap_dict['data'][0]['type'] == 'heatmap'
    z = np.asarray(fig_heatmap.data[0].z)
    assert z.shape == (4, 4), z.shape
    assert np.allclose(np.diag(z), 1.0)  # each column correlates perfectly with itself
    
except Exception as e:
    print(f"Heatmap error: {e}")
    raise

print("Data visualization with Plotly Express completed")
```

## Example 3: Surface, polar, subplot and candlestick charts

```python
import plotly.graph_objects as go
import plotly.subplots as sp
import numpy as np

# Example 3: Advanced Charts with Graph Objects
print("Example 3: Advanced charts with Graph Objects")

# Create sample data for advanced charts
print("\n--- Advanced Data Preparation ---")

np.random.seed(42)

# 3D surface data
x_3d = np.linspace(-5, 5, 20)
y_3d = np.linspace(-5, 5, 20)
X, Y = np.meshgrid(x_3d, y_3d)
Z = np.sin(np.sqrt(X**2 + Y**2))

print(f"3D surface data: {Z.shape}")
print(f"Z range: {Z.min():.2f} to {Z.max():.2f}")

# Polar data
theta = np.linspace(0, 2*np.pi, 100)
r1 = 1 + 0.5 * np.sin(5*theta)
r2 = 1 + 0.3 * np.cos(7*theta)

print(f"Polar data points: {len(theta)}")

# 3D Surface Plot
print("\n--- 3D Surface Plot ---")

try:
    fig_3d = go.Figure(data=[go.Surface(
        x=X,
        y=Y,
        z=Z,
        colorscale='viridis',
        opacity=0.8
    )])
    
    fig_3d.update_layout(
        title='3D Surface Plot',
        scene=dict(
            xaxis_title='X Axis',
            yaxis_title='Y Axis',
            zaxis_title='Z Axis'
        ),
        width=600,
        height=500
    )
    
    surface_dict = fig_3d.to_dict()
    surface_z = np.asarray(fig_3d.data[0].z)
    print(f"3D surface created: {surface_dict['data'][0]['type']}")
    print(f"Surface dimensions: {surface_z.shape[0]} x {surface_z.shape[1]}")
    assert surface_dict['data'][0]['type'] == 'surface'
    assert surface_z.shape == (20, 20), surface_z.shape
    assert np.allclose(surface_z, Z)
    assert fig_3d.layout.scene.zaxis.title.text == 'Z Axis'
    
except Exception as e:
    print(f"3D surface error: {e}")
    raise

# Polar Plot
print("\n--- Polar Plot ---")

try:
    fig_polar = go.Figure()
    
    # Add multiple polar traces
    fig_polar.add_trace(go.Scatterpolar(
        r=r1,
        theta=theta * 180/np.pi,  # Convert to degrees
        mode='lines',
        name='Pattern 1',
        line_color='blue'
    ))
    
    fig_polar.add_trace(go.Scatterpolar(
        r=r2,
        theta=theta * 180/np.pi,
        mode='lines',
        name='Pattern 2',
        line_color='red'
    ))
    
    fig_polar.update_layout(
        title='Polar Plot Example',
        polar=dict(
            radialaxis=dict(visible=True, range=[0, 2]),
            angularaxis=dict(direction='counterclockwise', period=360)
        ),
        showlegend=True
    )
    
    polar_dict = fig_polar.to_dict()
    print(f"Polar plot created: {len(polar_dict['data'])} traces")
    print(f"Polar plot type: {polar_dict['data'][0]['type']}")
    assert [trace.type for trace in fig_polar.data] == ['scatterpolar', 'scatterpolar']
    assert [trace.name for trace in fig_polar.data] == ['Pattern 1', 'Pattern 2']
    assert all(len(trace.r) == 100 for trace in fig_polar.data)
    assert tuple(fig_polar.layout.polar.radialaxis.range) == (0, 2)
    
except Exception as e:
    print(f"Polar plot error: {e}")
    raise

# Subplots example
print("\n--- Subplots ---")

try:
    # Create subplots with different chart types
    fig_subplots = sp.make_subplots(
        rows=2, cols=2,
        subplot_titles=('Line Plot', 'Bar Chart', 'Scatter Plot', 'Histogram'),
        specs=[[{"secondary_y": False}, {"secondary_y": False}],
               [{"secondary_y": False}, {"secondary_y": False}]]
    )
    
    # Data for subplots
    x_data = np.arange(10)
    y_data = np.random.randint(10, 50, 10)
    
    # Add line plot
    fig_subplots.add_trace(
        go.Scatter(x=x_data, y=y_data, mode='lines+markers', name='Line'),
        row=1, col=1
    )
    
    # Add bar chart  
    fig_subplots.add_trace(
        go.Bar(x=['A', 'B', 'C', 'D'], y=[20, 35, 30, 25], name='Bars'),
        row=1, col=2
    )
    
    # Add scatter plot
    fig_subplots.add_trace(
        go.Scatter(x=np.random.randn(20), y=np.random.randn(20), 
                  mode='markers', name='Scatter'),
        row=2, col=1
    )
    
    # Add histogram
    fig_subplots.add_trace(
        go.Histogram(x=np.random.normal(0, 1, 100), name='Histogram'),
        row=2, col=2
    )
    
    fig_subplots.update_layout(
        title_text="Subplots Example",
        showlegend=False,
        height=600
    )
    
    subplots_dict = fig_subplots.to_dict()
    print(f"Subplots created: {len(subplots_dict['data'])} traces")
    print(f"Subplot grid: 2x2")
    assert [trace.type for trace in fig_subplots.data] == ['scatter', 'bar', 'scatter', 'histogram']
    # Each trace sits on its own cell's axes, in row-major order.
    assert [(trace.xaxis, trace.yaxis) for trace in fig_subplots.data] == [
        ('x', 'y'), ('x2', 'y2'), ('x3', 'y3'), ('x4', 'y4')
    ]
    assert [a.text for a in fig_subplots.layout.annotations] == [
        'Line Plot', 'Bar Chart', 'Scatter Plot', 'Histogram'
    ]
    
except Exception as e:
    print(f"Subplots error: {e}")
    raise

# Candlestick chart simulation
print("\n--- Candlestick Chart ---")

try:
    # Simulate stock price data
    dates = [f"2024-01-{i:02d}" for i in range(1, 21)]
    opens = np.random.uniform(100, 110, 20)
    closes = opens + np.random.normal(0, 2, 20)  
    highs = np.maximum(opens, closes) + np.random.uniform(0, 3, 20)
    lows = np.minimum(opens, closes) - np.random.uniform(0, 3, 20)
    
    fig_candlestick = go.Figure(data=[go.Candlestick(
        x=dates,
        open=opens,
        high=highs,
        low=lows,
        close=closes,
        name='Stock Price'
    )])
    
    fig_candlestick.update_layout(
        title='Candlestick Chart Example',
        yaxis_title='Price',
        xaxis_title='Date'
    )
    
    candlestick_dict = fig_candlestick.to_dict()
    print(f"Candlestick chart created: {candlestick_dict['data'][0]['type']}")
    print(f"Data points: {len(dates)}")
    candle = fig_candlestick.data[0]
    assert candle.type == 'candlestick'
    assert len(candle.x) == 20 and len(candle.open) == 20
    assert (np.asarray(candle.high) >= np.maximum(candle.open, candle.close)).all()
    assert (np.asarray(candle.low) <= np.minimum(candle.open, candle.close)).all()
    
except Exception as e:
    print(f"Candlestick error: {e}")
    raise

print("Advanced charts with Graph Objects completed")
```

## Example 4: Customization and styling

```python
import plotly.graph_objects as go
import plotly.colors as pc_colors
import json

# Example 4: Customization and Styling
print("Example 4: Customization and styling")

# Explore color palettes and themes
print("\n--- Color Palettes ---")

try:
    # Show available color scales
    qualitative_colors = pc_colors.qualitative
    sequential_colors = pc_colors.sequential
    diverging_colors = pc_colors.diverging
    
    print(f"Qualitative palettes: {len(qualitative_colors.__dict__)} available")
    print(f"Sequential palettes: {len(sequential_colors.__dict__)} available") 
    print(f"Diverging palettes: {len(diverging_colors.__dict__)} available")
    
    # Sample some color palettes
    sample_qual = ['Plotly', 'Set1', 'Set2', 'Pastel']
    available_qual = []
    
    for palette in sample_qual:
        if hasattr(qualitative_colors, palette):
            colors = getattr(qualitative_colors, palette)
            available_qual.append((palette, len(colors)))
            print(f"  {palette}: {len(colors)} colors")
    
except Exception as e:
    print(f"Color palette error: {e}")
    raise

# Custom styled chart
print("\n--- Custom Styling Example ---")

try:
    # Create data for styling demo
    categories = ['Sales', 'Marketing', 'Engineering', 'Support', 'HR']
    q1_values = [85, 75, 90, 65, 55]
    q2_values = [88, 80, 95, 70, 60]
    q3_values = [92, 78, 98, 75, 65]
    q4_values = [95, 82, 102, 80, 68]
    
    fig_custom = go.Figure()
    
    # Add multiple bar traces with custom colors
    quarters = [('Q1', q1_values, '#1f77b4'), ('Q2', q2_values, '#ff7f0e'), 
               ('Q3', q3_values, '#2ca02c'), ('Q4', q4_values, '#d62728')]
    
    for quarter, values, color in quarters:
        fig_custom.add_trace(go.Bar(
            x=categories,
            y=values,
            name=quarter,
            marker_color=color,
            text=values,
            textposition='auto'
        ))
    
    # Custom layout styling
    fig_custom.update_layout(
        title={
            'text': 'Quarterly Performance by Department',
            'x': 0.5,
            'xanchor': 'center',
            'font': {'size': 20, 'family': 'Arial Black'}
        },
        # The axis title font is set under title.font
        xaxis={
            'title': {'text': 'Department', 'font': {'size': 14, 'color': 'darkblue'}},
            'tickfont': {'size': 12}
        },
        yaxis={
            'title': {'text': 'Performance Score', 'font': {'size': 14, 'color': 'darkblue'}},
            'tickfont': {'size': 12},
            'gridcolor': 'lightgray'
        },
        plot_bgcolor='white',
        paper_bgcolor='lightgray',
        barmode='group',
        showlegend=True,
        legend={
            'x': 1.02,
            'y': 1,
            'bgcolor': 'white',
            'bordercolor': 'black',
            'borderwidth': 1
        },
        width=700,
        height=500
    )
    
    custom_dict = fig_custom.to_dict()
    print(f"Custom styled chart created: {len(custom_dict['data'])} traces")
    print(f"Layout customizations applied: {len(custom_dict['layout'])} properties")
    assert len(custom_dict['data']) == 4, custom_dict['data']
    assert custom_dict['layout']['xaxis']['title'] == {'text': 'Department', 'font': {'size': 14, 'color': 'darkblue'}}
    assert custom_dict['layout']['yaxis']['title']['font'] == {'size': 14, 'color': 'darkblue'}
    
except Exception as e:
    print(f"Custom styling error: {e}")
    raise

# Annotations and shapes
print("\n--- Annotations and Shapes ---")

try:
    # Create a simple scatter plot for annotation
    import numpy as np
    np.random.seed(42)
    
    x_scatter = np.random.normal(0, 1, 50)
    y_scatter = np.random.normal(0, 1, 50)
    
    fig_annotated = go.Figure()
    
    fig_annotated.add_trace(go.Scatter(
        x=x_scatter,
        y=y_scatter,
        mode='markers',
        marker=dict(size=8, color='blue', opacity=0.6),
        name='Data Points'
    ))
    
    # Add annotations
    fig_annotated.add_annotation(
        x=max(x_scatter),
        y=max(y_scatter),
        text="Maximum Point",
        showarrow=True,
        arrowhead=2,
        arrowcolor="red",
        font=dict(color="red", size=12)
    )
    
    fig_annotated.add_annotation(
        x=0,
        y=0,
        text="Origin",
        showarrow=True,
        arrowhead=1,
        arrowcolor="green",
        font=dict(color="green", size=12)
    )
    
    # Add shapes
    fig_annotated.add_shape(
        type="circle",
        x0=-1, y0=-1, x1=1, y1=1,
        line=dict(color="orange", width=2),
        fillcolor="orange",
        opacity=0.2
    )
    
    fig_annotated.add_shape(
        type="rect",
        x0=-2, y0=-2, x1=2, y1=2,
        line=dict(color="purple", width=2, dash="dash")
    )
    
    fig_annotated.update_layout(
        title="Annotated Scatter Plot with Shapes",
        showlegend=True,
        width=600,
        height=500
    )
    
    annotated_dict = fig_annotated.to_dict()
    print(f"Annotated chart created with {len(annotated_dict['layout']['annotations'])} annotations")
    print(f"Shapes added: {len(annotated_dict['layout']['shapes'])}")
    
except Exception as e:
    print(f"Annotations error: {e}")
    raise

# Export and serialization
print("\n--- Export and Serialization ---")

try:
    # Convert figure to different formats
    sample_fig = go.Figure(go.Bar(x=['A', 'B', 'C'], y=[1, 2, 3]))
    
    # To dict
    fig_dict = sample_fig.to_dict()
    print(f"Figure as dict: {len(fig_dict)} top-level keys")
    
    # To JSON
    fig_json = sample_fig.to_json()
    json_size = len(fig_json)
    print(f"Figure as JSON: {json_size} characters")
    
    # Verify JSON is valid
    parsed_json = json.loads(fig_json)
    print(f"JSON validation: {'✓ Valid' if 'data' in parsed_json else '✗ Invalid'}")
    
    # HTML string (basic test)
    html_str = sample_fig.to_html()
    html_size = len(html_str)
    print(f"Figure as HTML: {html_size} characters")
    
except Exception as e:
    print(f"Export error: {e}")
    raise

print("Customization and styling completed")
```

## Example 5: Range selectors, animation frames, templates and WebGL traces

```python
import plotly.graph_objects as go
import plotly.express as px
import plotly.io as pio
import pandas as pd
import numpy as np

# Example 5: Advanced Plotly Features and Integration
print("Example 5: Advanced Plotly features and integration")

# Interactive features and widgets
print("\n--- Interactive Features ---")

try:
    # Create data for interactive demo
    np.random.seed(42)
    n_points = 100
    
    # Simulate stock price data over time
    dates = pd.date_range('2024-01-01', periods=n_points, freq='D')
    base_price = 100
    returns = np.random.normal(0.001, 0.02, n_points)  # Daily returns
    prices = [base_price]
    
    for ret in returns[1:]:
        prices.append(prices[-1] * (1 + ret))
    
    # Create interactive time series with range slider
    fig_interactive = go.Figure()
    
    fig_interactive.add_trace(go.Scatter(
        x=dates,
        y=prices,
        mode='lines',
        name='Stock Price',
        line=dict(color='blue', width=2),
        hovertemplate='<b>Date</b>: %{x}<br><b>Price</b>: $%{y:.2f}<extra></extra>'
    ))
    
    # Add moving average
    window = 10
    moving_avg = pd.Series(prices).rolling(window=window).mean()
    
    fig_interactive.add_trace(go.Scatter(
        x=dates,
        y=moving_avg,
        mode='lines',
        name=f'{window}-day MA',
        line=dict(color='red', width=1, dash='dash')
    ))
    
    # Configure interactivity
    fig_interactive.update_layout(
        title='Interactive Stock Price Chart',
        xaxis=dict(
            rangeselector=dict(
                buttons=list([
                    dict(count=7, label="7d", step="day", stepmode="backward"),
                    dict(count=30, label="30d", step="day", stepmode="backward"),
                    dict(count=60, label="60d", step="day", stepmode="backward"),
                    dict(step="all")
                ])
            ),
            rangeslider=dict(visible=True),
            type="date"
        ),
        yaxis_title="Price ($)",
        hovermode='x unified'
    )
    
    interactive_dict = fig_interactive.to_dict()
    print(f"Interactive chart created: {len(interactive_dict['data'])} traces")
    print(f"Range selector buttons: {len(interactive_dict['layout']['xaxis']['rangeselector']['buttons'])}")
    assert [trace.name for trace in fig_interactive.data] == ['Stock Price', '10-day MA']
    assert len(fig_interactive.layout.xaxis.rangeselector.buttons) == 4
    assert fig_interactive.layout.xaxis.rangeslider.visible is True
    # The first window-1 moving-average values are undefined.
    ma = np.asarray(fig_interactive.data[1].y, dtype=float)
    assert np.isnan(ma[:window - 1]).all() and not np.isnan(ma[window - 1:]).any()
    
except Exception as e:
    print(f"Interactive features error: {e}")
    raise

# Animations (basic structure)
print("\n--- Animation Setup ---")

try:
    # Create data for simple animation
    frames_data = []
    x_data = np.arange(10)
    
    # Generate frames (5 frames)
    for frame_idx in range(5):
        y_data = np.random.randint(10, 50, 10) + frame_idx * 5
        frames_data.append({
            'x': x_data.tolist(),
            'y': y_data.tolist(),
            'frame_name': f'Frame {frame_idx + 1}'
        })
    
    # Create initial figure
    fig_anim = go.Figure(
        data=[go.Bar(x=frames_data[0]['x'], y=frames_data[0]['y'])],
        layout=go.Layout(
            title="Animation Example",
            updatemenus=[dict(
                type="buttons",
                buttons=[dict(label="Play", method="animate", 
                             args=[None, {"frame": {"duration": 500}}])]
            )],
            sliders=[dict(
                steps=[dict(args=[[f"frame{i}"]], label=f"Frame {i+1}") 
                      for i in range(len(frames_data))]
            )]
        ),
        frames=[go.Frame(data=[go.Bar(x=frame['x'], y=frame['y'])], 
                        name=f"frame{i}") 
               for i, frame in enumerate(frames_data)]
    )
    
    anim_dict = fig_anim.to_dict()
    print(f"Animation structure created: {len(anim_dict['frames'])} frames")
    print(f"Control buttons: {len(anim_dict['layout']['updatemenus'][0]['buttons'])}")
    assert [frame.name for frame in fig_anim.frames] == [f"frame{i}" for i in range(5)]
    assert list(fig_anim.frames[4].data[0].y) == frames_data[4]['y']
    assert len(fig_anim.layout.sliders[0].steps) == 5
    
except Exception as e:
    print(f"Animation setup error: {e}")
    raise

# Templates and themes
print("\n--- Templates and Themes ---")

try:
    # Show available templates
    available_templates = pio.templates
    template_names = list(available_templates.keys())[:10]  # First 10
    print(f"Available templates: {template_names}")
    
    # Create sample chart with different template
    sample_data = {'x': ['A', 'B', 'C', 'D'], 'y': [10, 15, 13, 17]}
    
    fig_template = px.bar(x=sample_data['x'], y=sample_data['y'], 
                         title="Template Example")
    
    # Apply a built-in template
    assert 'plotly_white' in available_templates, list(available_templates)
    fig_template.update_layout(template='plotly_white')
    print("Applied 'plotly_white' template")
    
    template_dict = fig_template.to_dict()
    print(f"Template chart created: {template_dict['data'][0]['type']}")
    assert template_dict['data'][0]['type'] == 'bar'
    # plotly_white sets a white plot background
    assert fig_template.layout.template.layout.plot_bgcolor == 'white'
    
except Exception as e:
    print(f"Templates error: {e}")
    raise

# Performance considerations
print("\n--- Performance Optimization ---")

try:
    # Large dataset handling simulation
    large_n = 10000
    large_x = np.random.normal(0, 1, large_n)
    large_y = np.random.normal(0, 1, large_n)
    
    # Create scatter plot with sampling for performance
    sample_size = 1000  # Sample for performance
    sample_indices = np.random.choice(large_n, sample_size, replace=False)
    
    fig_performance = go.Figure()
    fig_performance.add_trace(go.Scattergl(  # Use WebGL for better performance
        x=large_x[sample_indices],
        y=large_y[sample_indices],
        mode='markers',
        marker=dict(size=3, opacity=0.6),
        name=f'Sample ({sample_size} of {large_n} points)'
    ))
    
    fig_performance.update_layout(
        title=f"Performance Example: {sample_size} sampled points",
        showlegend=True
    )
    
    perf_dict = fig_performance.to_dict()
    print(f"Performance chart: {len(fig_performance.data[0].x)} points plotted")
    print(f"Chart type: {perf_dict['data'][0]['type']} (WebGL accelerated)")
    assert perf_dict['data'][0]['type'] == 'scattergl'
    assert len(fig_performance.data[0].x) == sample_size
    
except Exception as e:
    print(f"Performance optimization error: {e}")
    raise

print("Advanced Plotly features and integration completed")
```
