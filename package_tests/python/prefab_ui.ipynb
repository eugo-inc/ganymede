# prefab_ui smoke test

Verifies the `prefab_ui` package (Prefab generative UI framework from PrefectHQ) imports and its primary APIs (`Rx` reactive state, components DSL) are reachable.

```python
# Example 1: import the components DSL and confirm the canonical primitives are present.
from prefab_ui import components

expected = ["Card", "CardContent", "CardFooter", "Column", "Row", "Badge", "Input"]
available = {name: getattr(components, name, None) for name in expected}

for name, obj in available.items():
    print(f"{name}: {type(obj).__name__ if obj is not None else 'MISSING'}")

missing = [n for n, o in available.items() if o is None]
assert not missing, f"Missing components: {missing}"
```

```python
# Example 2: exercise the Rx reactive-state class — the README's headline API.
from prefab_ui.rx import Rx

name = Rx("name")
with_default = name.default("world")

print(f"Rx instance: {name!r}")
print(f"Rx type: {type(name).__name__}")
print(f"Rx.default(...) returns: {type(with_default).__name__}")

# Rx should be chainable / fluent; .default() returns an Rx-like
assert with_default is not None
```

```python
# Example 3: build the README's "hello world" card using the context-manager DSL.
from prefab_ui.components import (
    Card, CardContent, CardFooter, Column, Row, H3, Muted, Input, Badge,
)
from prefab_ui.rx import Rx

name = Rx("name").default("world")

with Card() as card:
    with CardContent():
        with Column(gap=3):
            H3(f"Hello, {name}!")
            Muted("Type below and watch this update in real time.")
            Input(name="name", placeholder="Your name...")
    with CardFooter():
        with Row(gap=2):
            Badge(f"Name: {name}", variant="default")
            Badge("Prefab", variant="success")

print(f"Built card: {type(card).__name__}")
assert card is not None
```

```python
# Example 4: a minimal standalone component should serialize / repr cleanly.
from prefab_ui.components import Badge, H3

badge = Badge("hello", variant="default")
heading = H3("Title")

for component in (badge, heading):
    cls = type(component).__name__
    r = repr(component)
    print(f"{cls}: repr length = {len(r)}")
    assert isinstance(r, str) and len(r) > 0

print("Component reprs OK")
```
