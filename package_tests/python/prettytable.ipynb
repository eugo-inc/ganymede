## Example 1: Build a table row by row and print it

```python
from prettytable import PrettyTable

# Example 1: the basic API: field names, rows, and the printed text.
table = PrettyTable()
table.field_names = ["City", "Area", "Population"]
table.add_row(["Adelaide", 1295, 1158259])
table.add_row(["Brisbane", 5905, 1857594])
table.add_row(["Darwin", 112, 120900])
text = table.get_string()
print(text)
assert "Brisbane" in text and text.count("\n") == 6, text
assert len(table.rows) == 3
```

## Example 2: Sorting, slicing and choosing columns

```python
from prettytable import PrettyTable

# Example 2: get_string can sort, slice rows and keep only some columns.
table = PrettyTable(["Model", "GPUs", "Latency ms"])
table.add_rows([["a", 8, 42.7], ["b", 4, 13.1], ["c", 2, 77.0], ["d", 1, 5.5]])
text = table.get_string(sortby="Latency ms", reversesort=True, start=0, end=2, fields=["Model", "Latency ms"])
print(text)
lines = [line for line in text.splitlines() if line.startswith("|")]
assert "c" in lines[1] and "a" in lines[2], text
assert "GPUs" not in text
```

## Example 3: Styles and alignment

```python
from prettytable import PrettyTable, TableStyle

# Example 3: a Markdown table and right-aligned numbers.
table = PrettyTable(["Item", "Count"])
table.add_rows([["apples", 3], ["pears", 12]])
table.align["Count"] = "r"
table.set_style(TableStyle.MARKDOWN)
text = table.get_string()
print(text)
assert text.splitlines()[0].startswith("|") and "Item" in text.splitlines()[0], text
assert set(text.splitlines()[1]) <= set("|-: ") and text.splitlines()[1].count(":") >= 1, text
```

## Example 4: Other output formats: HTML, CSV and JSON

```python
import json
from prettytable import PrettyTable

# Example 4: the same table rendered as HTML, CSV and JSON.
table = PrettyTable(["Name", "Score"])
table.add_rows([["x", 1], ["y", 2]])
html = table.get_html_string()
csv_text = table.get_csv_string()
data = json.loads(table.get_json_string())
print(html, csv_text, data, sep="\n")
assert "<table>" in html and "<th>Score</th>" in html
assert csv_text.splitlines()[0] == "Name,Score"
assert data[0] == ["Name", "Score"] and data[1] == {"Name": "x", "Score": 1}
```

## Example 5: Reading a table from CSV and wide characters

```python
import io
from prettytable import from_csv

# Example 5: from_csv, and East Asian wide characters measured with wcwidth.
table = from_csv(io.StringIO("name,label\nalpha,東京\nbeta,ok\n"))
text = table.get_string()
print(text)
widths = {len(line) for line in text.splitlines() if line.startswith("+")}
assert len(widths) == 1, widths
assert table.field_names == ["name", "label"] and len(table.rows) == 2
```
