## Example 1: Basic Progress Bar Usage

```python
import progressbar
import time

# Example 1: Basic Progress Bar Usage
print("Example 1 - Basic Progress Bar Usage")

# Simple progress bar with percentage
print("\n1. Basic percentage progress bar:")
bar = progressbar.ProgressBar(max_value=100, widgets=[
    progressbar.Percentage(), ' ',
    progressbar.Bar(), ' ',
    progressbar.ETA()
])

# Simulate a task
bar.start()
for i in range(101):
    time.sleep(0.01)
    bar.update(i)
assert bar.value == 100, bar.value
bar.finish()
assert bar.value == bar.max_value == 100, bar.value
assert bar.percentage == 100.0, bar.percentage

print("Basic progress bar completed")

print("\nExample 1 completed successfully")
```

## Example 2: Advanced Progress Bar Customization

```python
import progressbar
import time

# Example 2: Progress Bar with Custom Widgets
print("Example 2 - Progress Bar with Custom Widgets")

# Custom widgets configuration
widgets = [
    'Processing: ', progressbar.Percentage(), ' ',
    progressbar.Bar(marker='#', left='[', right=']'), ' ',
    progressbar.ETA(), ' ',
    progressbar.FileTransferSpeed()
]

print("\nFile processing simulation:")
files_to_process = 50
pbar_files = progressbar.ProgressBar(max_value=files_to_process, widgets=widgets)
pbar_files.start()

for i in range(files_to_process + 1):
    time.sleep(0.02)
    pbar_files.update(i)
assert pbar_files.value == files_to_process, pbar_files.value

pbar_files.finish()
assert pbar_files.value == pbar_files.max_value == files_to_process
assert len(pbar_files.widgets) == len(widgets)
print("File processing completed")

print("\nExample 2 completed successfully")
```

## Example 3: Context Manager and Iterator Support

```python
import progressbar
import time

# Example 3: Context Manager and Iterator Support
print("Example 3 - Context Manager and Iterator Support")

# Using progress bar as context manager
print("\n1. Context manager usage:")
data_to_process = list(range(60))

with progressbar.ProgressBar(max_value=len(data_to_process)) as bar:
    for i, item in enumerate(data_to_process):
        time.sleep(0.02)
        bar.update(i + 1)

# Leaving the with-block finishes the bar.
assert bar.value == bar.max_value == len(data_to_process), bar.value
assert bar.end_time is not None
print("Context manager progress completed")

# Iterator wrapper for automatic progress tracking
print("\n2. Iterator wrapper:")
items = list(range(40))

seen = []
for item in progressbar.progressbar(items, prefix='Processing: '):
    time.sleep(0.02)
    seen.append(item)
assert seen == items  # the wrapper yields every item, in order

print("Iterator wrapper completed")

print("\nExample 3 completed successfully")
```

## Example 4: Nested Progress Bars

```python
import progressbar
import time

# Example 4: Nested Progress Bars
print("Example 4 - Nested Progress Bars")

total_batches = 5
items_per_batch = 10

# Outer progress bar for batches
outer_widgets = [
    'Overall Progress: ', progressbar.Percentage(), ' ',
    progressbar.Bar(marker='█'), ' ',
    'Batch: ', progressbar.Counter(), f'/{total_batches}', ' ',
    progressbar.ETA()
]

print(f"\nProcessing {total_batches} batches with {items_per_batch} items each:")

inner_values = []
with progressbar.ProgressBar(max_value=total_batches, widgets=outer_widgets) as outer_bar:
    for batch_num in range(total_batches):
        # Inner progress bar for the items of this batch, drawn one line below the outer bar
        inner_bar = progressbar.ProgressBar(max_value=items_per_batch, line_offset=1, prefix=f'Batch {batch_num + 1}: ')
        inner_bar.start()
        for item_num in range(items_per_batch):
            time.sleep(0.02)
            inner_bar.update(item_num + 1)
        inner_bar.finish()
        inner_values.append(inner_bar.value)
        outer_bar.update(batch_num + 1)

assert outer_bar.value == total_batches, outer_bar.value
assert inner_values == [items_per_batch] * total_batches, inner_values
print("Nested progress bars completed")

print("\nExample 4 completed successfully")
```

## Example 5: Advanced Features and Unknown Length Progress

```python
import progressbar
import time

# Example 5: Advanced Features and Unknown Length Progress
print("Example 5 - Advanced Features and Unknown Length Progress")

# Progress bar with unknown length (indeterminate)
print("\n1. Unknown length progress bar:")
bar = progressbar.ProgressBar(max_value=progressbar.UnknownLength)
bar.start()

for i in range(30):
    time.sleep(0.05)
    bar.update(i)

bar.finish()
# With no known length there is no maximum to fill up to: the bar keeps the last value.
assert bar.max_value is progressbar.UnknownLength
assert bar.value == 29, bar.value
print("Unknown length progress completed")

# Progress bar with Timer widget
print("\n2. Progress bar with timer:")
widgets = [
    'Processing: ', progressbar.Percentage(), ' ',
    progressbar.Bar(), ' ',
    progressbar.Timer(), ' seconds elapsed'
]

timer_bar = progressbar.ProgressBar(max_value=50, widgets=widgets)
timer_bar.start()

for i in range(51):
    time.sleep(0.03)
    timer_bar.update(i)

timer_bar.finish()
assert timer_bar.value == timer_bar.max_value == 50, timer_bar.value
assert timer_bar.data()["total_seconds_elapsed"] > 0, timer_bar.data()["total_seconds_elapsed"]
print("Timer progress completed")

print("\nExample 5 completed successfully")
```
