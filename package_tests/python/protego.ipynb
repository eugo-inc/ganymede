## Example 1: Basic Robots.txt Parsing

```python
import protego

# Example 1: Basic Robots.txt Parsing
print("Example 1 - Basic Robots.txt Parsing")

# Sample robots.txt content
robots_txt_content = """
User-agent: *
Disallow: /admin/
Disallow: /private/
Allow: /public/
Crawl-delay: 1

User-agent: Googlebot
Disallow: /temp/
Allow: /api/
Crawl-delay: 0.5

User-agent: BadBot
Disallow: /

Sitemap: https://example.com/sitemap.xml
"""

print("\n1. Creating robots.txt parser:")
# Parse the robots.txt content
rp = protego.Protego.parse(robots_txt_content)

print(f"  ✓ Protego parser created")
print(f"  Parser type: {type(rp).__name__}")

# Test URL access permissions
test_urls = [
    "/admin/users",
    "/public/info", 
    "/private/data",
    "/temp/file.txt",
    "/api/v1/users",
    "/sitemap.xml"
]

user_agents = ["*", "Googlebot", "BadBot", "CustomBot"]

print("\n2. Testing URL access permissions:")
for user_agent in user_agents:
    print(f"\n  User-agent: {user_agent}")
    for url in test_urls:
        can_fetch = rp.can_fetch(url, user_agent)
        status = "✓ Allowed" if can_fetch else "✗ Disallowed"
        print(f"    {url:20} → {status}")

# Test crawl delays
print("\n3. Crawl delay settings:")
for user_agent in ["*", "Googlebot", "BadBot", "CustomBot"]:
    delay = rp.crawl_delay(user_agent)
    if delay is not None:
        print(f"  {user_agent:15} → {delay} seconds")
    else:
        print(f"  {user_agent:15} → No delay specified")

# Get sitemaps
sitemaps = rp.sitemaps
print(f"\n4. Sitemap URLs found:")
for sitemap in sitemaps:
    print(f"  ✓ {sitemap}")

print("Example 1 completed successfully")
```

## Example 3: Web Scraping Compliance Checking

```python
import protego
import time

# Example 3: Web Scraping Compliance Checking
print("Example 3 - Web Scraping Compliance Checking")

# Simulate robots.txt from a real e-commerce site
ecommerce_robots = """
User-agent: *
Disallow: /checkout/
Disallow: /cart/
Disallow: /account/
Disallow: /admin/
Allow: /products/
Allow: /categories/
Allow: /search?q=
Disallow: /search?*sort=
Crawl-delay: 5

User-agent: Googlebot
Allow: /
Disallow: /checkout/
Disallow: /cart/
Disallow: /account/
Crawl-delay: 1

User-agent: PriceComparisonBot
Disallow: /

User-agent: SEOBot
Allow: /products/
Allow: /categories/
Disallow: /
Crawl-delay: 10

Sitemap: https://shop.example.com/sitemap.xml
"""

print("\n1. E-commerce site robots.txt analysis:")
rp = protego.Protego.parse(ecommerce_robots)

# Simulate a web scraping scenario
class WebScrapingBot:
    def __init__(self, user_agent, robots_parser):
        self.user_agent = user_agent
        self.robots = robots_parser
        self.last_request_time = 0
    
    def can_scrape(self, url):
        """Check if URL can be scraped according to robots.txt"""
        return self.robots.can_fetch(url, self.user_agent)
    
    def get_crawl_delay(self):
        """Get required crawl delay for this bot"""
        return self.robots.crawl_delay(self.user_agent) or 0
    
    def should_wait(self):
        """Check if bot should wait before next request"""
        delay = self.get_crawl_delay()
        if delay > 0:
            elapsed = time.time() - self.last_request_time
            return max(0, delay - elapsed)
        return 0
    
    def simulate_request(self, url):
        """Simulate making a request with compliance checking"""
        if not self.can_scrape(url):
            return {"status": "blocked", "reason": "Disallowed by robots.txt"}
        
        wait_time = self.should_wait()
        if wait_time > 0:
            return {"status": "delayed", "wait": wait_time}
        
        self.last_request_time = time.time()
        return {"status": "success", "url": url}

# Test different bot scenarios
bots = [
    ("*", "Generic crawler"),
    ("Googlebot", "Google search bot"),
    ("PriceComparisonBot", "Price comparison service"),
    ("SEOBot", "SEO analysis bot")
]

urls_to_test = [
    "/products/laptop-123",
    "/categories/electronics",
    "/checkout/payment",
    "/account/profile", 
    "/search?q=laptop",
    "/search?q=phone&sort=price"
]

print(f"\n2. Bot compliance testing:")
for user_agent, description in bots:
    bot = WebScrapingBot(user_agent, rp)
    print(f"\n  {description} ({user_agent}):")
    print(f"    Crawl delay: {bot.get_crawl_delay()}s")
    
    for url in urls_to_test[:4]:  # Test first 4 URLs
        result = bot.simulate_request(url)
        if result["status"] == "success":
            print(f"    {url:25} → ✓ Allowed")
        elif result["status"] == "blocked":
            print(f"    {url:25} → ✗ Blocked")
        else:
            print(f"    {url:25} → ⏱ Delayed")

# Compliance report
print(f"\n3. Compliance summary:")
compliance_stats = {}
for user_agent, _ in bots:
    allowed = sum(1 for url in urls_to_test if rp.can_fetch(url, user_agent))
    total = len(urls_to_test)
    compliance_stats[user_agent] = (allowed, total)
    
for user_agent, (allowed, total) in compliance_stats.items():
    percentage = (allowed / total) * 100
    print(f"  {user_agent:20} → {allowed}/{total} URLs allowed ({percentage:.1f}%)")

# Best practices for compliant scraping
print(f"\n4. Web scraping best practices:")
best_practices = [
    "Always check robots.txt before scraping",
    "Respect crawl-delay directives",
    "Use appropriate User-Agent strings",
    "Monitor for rate limiting responses",
    "Cache robots.txt to avoid repeated requests",
    "Handle temporary fetch failures gracefully",
    "Consider using official APIs when available",
    "Respect server capacity and avoid overloading"
]

for i, practice in enumerate(best_practices, 1):
    print(f"  {i}. {practice}")

print("Example 3 completed successfully")
```

## Example 5: Production Web Crawler Implementation

```python
import protego
import time
from datetime import datetime, timedelta

# Example 5: Production Web Crawler Implementation
print("Example 5 - Production Web Crawler Implementation")

# Production-grade robots.txt parser with caching and compliance
class RobotsTxtManager:
    """Production robots.txt manager with caching and validation"""
    
    def __init__(self):
        self.cache = {}
        self.cache_ttl = 3600  # 1 hour cache TTL
        self.default_crawl_delay = 1
        
    def get_robots_parser(self, domain, robots_content=None):
        """Get cached robots parser or create new one"""
        cache_key = domain
        current_time = time.time()
        
        # Check cache
        if cache_key in self.cache:
            cached_parser, cached_time = self.cache[cache_key]
            if current_time - cached_time < self.cache_ttl:
                return cached_parser
        
        # Create new parser
        if robots_content is None:
            # In production, would fetch from https://domain/robots.txt
            robots_content = self._get_default_robots()
        
        parser = protego.Protego.parse(robots_content)
        self.cache[cache_key] = (parser, current_time)
        return parser
    
    def _get_default_robots(self):
        """Default robots.txt for testing"""
        return """
User-agent: *
Disallow: /admin/
Disallow: /private/
Allow: /api/public/
Crawl-delay: 1

User-agent: Googlebot
Allow: /
Crawl-delay: 0.5

Sitemap: https://example.com/sitemap.xml
"""

class ProductionWebCrawler:
    """Production web crawler with robots.txt compliance"""
    
    def __init__(self, user_agent, respect_robots=True):
        self.user_agent = user_agent
        self.respect_robots = respect_robots
        self.robots_manager = RobotsTxtManager()
        self.request_history = {}
        self.stats = {
            'requests_made': 0,
            'requests_blocked': 0,
            'requests_delayed': 0,
            'domains_crawled': set()
        }
    
    def can_crawl(self, url, domain):
        """Check if URL can be crawled"""
        if not self.respect_robots:
            return True
            
        robots_parser = self.robots_manager.get_robots_parser(domain)
        return robots_parser.can_fetch(url, self.user_agent)
    
    def get_crawl_delay(self, domain):
        """Get crawl delay for domain"""
        robots_parser = self.robots_manager.get_robots_parser(domain)
        delay = robots_parser.crawl_delay(self.user_agent)
        return delay if delay is not None else self.robots_manager.default_crawl_delay
    
    def should_wait(self, domain):
        """Calculate wait time before next request"""
        if domain not in self.request_history:
            return 0
            
        last_request = self.request_history[domain]
        delay = self.get_crawl_delay(domain)
        elapsed = time.time() - last_request
        return max(0, delay - elapsed)
    
    def crawl_url(self, url, domain):
        """Crawl a single URL with compliance checking"""
        # Check robots.txt permission
        if not self.can_crawl(url, domain):
            self.stats['requests_blocked'] += 1
            return {
                'status': 'blocked',
                'url': url,
                'reason': 'Blocked by robots.txt'
            }
        
        # Check crawl delay
        wait_time = self.should_wait(domain)
        if wait_time > 0:
            self.stats['requests_delayed'] += 1
            return {
                'status': 'delayed',
                'url': url,
                'wait_time': wait_time,
                'message': f'Must wait {wait_time:.2f} seconds'
            }
        
        # Simulate successful crawl
        self.request_history[domain] = time.time()
        self.stats['requests_made'] += 1
        self.stats['domains_crawled'].add(domain)
        
        return {
            'status': 'success',
            'url': url,
            'timestamp': datetime.now().isoformat(),
            'crawl_delay': self.get_crawl_delay(domain)
        }
    
    def get_stats(self):
        """Get crawler statistics"""
        total_attempts = (self.stats['requests_made'] + 
                         self.stats['requests_blocked'] + 
                         self.stats['requests_delayed'])
        
        return {
            **self.stats,
            'total_attempts': total_attempts,
            'success_rate': (self.stats['requests_made'] / max(1, total_attempts)) * 100,
            'domains_count': len(self.stats['domains_crawled'])
        }

print("\n1. Production crawler setup:")
crawler = ProductionWebCrawler("ProductionBot/1.0", respect_robots=True)
print(f"  ✓ Crawler initialized with user-agent: {crawler.user_agent}")
print(f"  ✓ Robots.txt compliance: {crawler.respect_robots}")

# Simulate crawling session
crawl_targets = [
    ("/products/item-123", "shop.example.com"),
    ("/api/public/users", "api.example.com"), 
    ("/admin/dashboard", "shop.example.com"),
    ("/products/item-456", "shop.example.com"),  # Should be delayed
    ("/private/data", "api.example.com"),
    ("/api/public/products", "api.example.com")
]

print(f"\n2. Crawling simulation:")
for url, domain in crawl_targets:
    result = crawler.crawl_url(url, domain)
    status_icon = {
        'success': '✓',
        'blocked': '✗', 
        'delayed': '⏱'
    }.get(result['status'], '?')
    
    print(f"  {status_icon} {url:25} [{domain:15}] → {result['status']}")
    if 'reason' in result:
        print(f"    Reason: {result['reason']}")
    elif 'wait_time' in result:
        print(f"    Wait: {result['wait_time']:.2f}s")

# Crawler statistics
print(f"\n3. Crawler statistics:")
stats = crawler.get_stats()
for key, value in stats.items():
    if key == 'domains_crawled':
        print(f"  {key}: {sorted(value)}")
    elif key == 'success_rate':
        print(f"  {key}: {value:.1f}%")
    else:
        print(f"  {key}: {value}")

# Production features demonstration
print(f"\n4. Production crawler features:")
features = [
    "✓ Robots.txt caching with TTL",
    "✓ Per-domain crawl delay enforcement", 
    "✓ Request rate limiting and timing",
    "✓ Compliance checking and blocking",
    "✓ Detailed statistics and monitoring",
    "✓ Error handling and graceful failures",
    "✓ Configurable user-agent strings",
    "✓ Multi-domain crawling support"
]

for feature in features:
    print(f"  {feature}")

# Best practices summary
print(f"\n5. Robots.txt compliance best practices:")
best_practices = [
    "Cache robots.txt files with appropriate TTL",
    "Respect crawl-delay directives per user-agent",
    "Handle malformed robots.txt gracefully",
    "Monitor and log compliance violations",
    "Use descriptive and identifiable user-agent strings",
    "Implement exponential backoff for blocked requests",
    "Regular validation of robots.txt parsing",
    "Provide opt-out mechanisms for website owners"
]

for i, practice in enumerate(best_practices, 1):
    print(f"  {i}. {practice}")

print("Example 5 completed successfully")
```
