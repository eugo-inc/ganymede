# proto_plus: 5 Basic Usage Examples

```python
import proto

class User(proto.Message):
    name = proto.Field(proto.STRING, number=1)
    score = proto.Field(proto.INT32, number=2)

# Create a user
user = User()
user.name = "Bob"
user.score = 100

# Use the actual proto method to convert to dict
result = type(user).to_dict(user)
print(result)
# Output: {'name': 'Bob', 'score': 100}
```

```python
import proto

class MyMessage(proto.Message):
    field = proto.Field(proto.INT32, number=1)

msg = MyMessage(field=123)
print(msg)
# Output: field: 123
```

```python
# Example 3: Serialize to bytes
import proto

msg = MyMessage(field=123)
serialized_pb = msg._pb.SerializeToString()
print(f"Via protobuf: {serialized_pb}")
```

```python
# Example 4: Parse from bytes
import proto

msg = MyMessage(field=123)
serialized_pb = msg._pb.SerializeToString()
print(f"Serialized: {serialized_pb}")
new_msg = MyMessage()
new_msg._pb.ParseFromString(serialized_pb)
print(f"Deserialized: {new_msg}")
```

```python
import proto

class Status(proto.Enum):
    UNKNOWN = 0
    ACTIVE = 1
    INACTIVE = 2

class Task(proto.Message):
    name = proto.Field(proto.STRING, number=1)
    status = proto.Field(proto.ENUM, enum=Status, number=2)

# Create a task with enum
task = Task()
task.name = "Complete project"
task.status = Status.ACTIVE

print(f"Task: {task.name}")
print(f"Status: {task.status}")
print(f"Status name: {task.status.name}")

# Serialize using the underlying protobuf object
data = task._pb.SerializeToString()
print(f"Serialized: {len(data)} bytes")

# Deserialize
restored_task = Task()
restored_task._pb.ParseFromString(data)
print(f"Restored: {restored_task.name}, {restored_task.status.name}")
```
