## Example 1: psycopg-c C extension availability and performance

```python
import psycopg
import sys

# Example 1: psycopg-c C extension availability and performance
print("psycopg module imported successfully")

try:
    print("Example 1 - psycopg-c C extension detection")
    
    # Check if C extension is available
    import psycopg_c
    print("psycopg-c C extension is available")
    print(f"C extension module: {psycopg_c}")
    
    # Show C extension version if available
    if hasattr(psycopg_c, '__version__'):
        print(f"psycopg-c version: {psycopg_c.__version__}")
        
    # Check psycopg implementation
    print(f"psycopg version: {psycopg.__version__}")
    
    # Connection class info
    conn_class = psycopg.Connection
    print(f"Connection class: {conn_class}")
    print(f"Connection module: {conn_class.__module__}")
    
    # Cursor class info  
    cursor_class = psycopg.Cursor
    print(f"Cursor class: {cursor_class}")
    print(f"Cursor module: {cursor_class.__module__}")
    
    # Which libpq wrapper psycopg loaded: Connection.__module__ is 'psycopg' for every implementation,
    # psycopg.pq.__impl__ names it ('c' = the psycopg_c extension)
    print(f"psycopg libpq implementation: {psycopg.pq.__impl__}")
    assert psycopg.pq.__impl__ == "c", psycopg.pq.__impl__
    assert psycopg_c.__version__ == psycopg.__version__, (psycopg_c.__version__, psycopg.__version__)
    
    print("psycopg-c example 1 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"psycopg_c example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 4: Memory management and C extension benefits

```python
import psycopg
import gc
import sys

# Example 4: Memory management and C extension benefits
print("psycopg module imported successfully")

try:
    print("Example 4 - psycopg-c memory management and efficiency")
    
    # Memory usage demonstration
    initial_objects = len(gc.get_objects())
    print(f"Initial Python objects: {initial_objects}")
    
    # Create multiple connection info objects
    connections = []
    for i in range(100):
        dsn = f"host=localhost port=543{i%10} dbname=db{i} user=user{i}"
        parsed = psycopg.conninfo.conninfo_to_dict(dsn)
        connections.append(parsed)
    
    mid_objects = len(gc.get_objects())
    print(f"Objects after 100 connection infos: {mid_objects}")
    print(f"Object increase: {mid_objects - initial_objects}")
    
    # Create SQL queries with C extension
    from psycopg import sql
    queries = []
    
    for i in range(100):
        query = sql.SQL("SELECT * FROM table_{num} WHERE id = %s").format(
            num=sql.Literal(i)
        )
        queries.append(query.as_string())
    
    final_objects = len(gc.get_objects())
    print(f"Objects after 100 SQL queries: {final_objects}")
    print(f"Total object increase: {final_objects - initial_objects}")
    
    # Cleanup
    del connections, queries
    gc.collect()
    
    after_cleanup = len(gc.get_objects())
    print(f"Objects after cleanup: {after_cleanup}")
    
    # C extension provides:
    print("\nC extension benefits:")
    print("- Lower memory overhead per object")
    print("- Faster garbage collection")
    print("- Reduced Python interpreter overhead")
    print("- Better cache locality")
    print("- Native C data structures")
    
    # Show system info
    print(f"\nPython version: {sys.version}")
    print(f"Reference count enabled: {hasattr(sys, 'getrefcount')}")
    
    print("psycopg-c example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"psycopg_c example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```
