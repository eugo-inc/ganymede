## Example 1: Basic psycopg_pool Import

```python
import psycopg_pool

# Example 1: Basic psycopg_pool Import
print("Example 1 - Basic psycopg_pool Import")

# Show module name and version only
print(f"\nModule name: {psycopg_pool.__name__}")
print(f"Version: {psycopg_pool.__version__}")

# Check if key exports exist without accessing them
has_conn_pool = 'ConnectionPool' in dir(psycopg_pool)
has_async_pool = 'AsyncConnectionPool' in dir(psycopg_pool)
has_null_pool = 'NullConnectionPool' in dir(psycopg_pool)

print(f"\nKey classes present:")
print(f"  ConnectionPool: {has_conn_pool}")
print(f"  AsyncConnectionPool: {has_async_pool}")
print(f"  NullConnectionPool: {has_null_pool}")

print("\nExample 1 completed successfully")
```

## Example 2: Exception Classes Check

```python
import psycopg_pool

# Example 2: Exception Classes Check
print("Example 2 - Exception Classes Check")

# Check for exception classes without accessing them
exc_classes = ['PoolClosed', 'PoolTimeout', 'TooManyRequests']

print("\nException classes in module:")
for exc_name in exc_classes:
    present = exc_name in dir(psycopg_pool)
    print(f"  {exc_name}: {'present' if present else 'not found'}")

# Count total attributes
total_attrs = len(dir(psycopg_pool))
print(f"\nTotal module attributes: {total_attrs}")

print("\nExample 2 completed successfully")
```

## Example 3: Pool Classes Availability

```python
import psycopg_pool

# Example 3: Pool Classes Availability
print("Example 3 - Pool Classes Availability")

# List of expected pool classes
pool_classes = [
    'ConnectionPool',
    'AsyncConnectionPool', 
    'NullConnectionPool',
    'AsyncNullConnectionPool'
]

print("\nPool classes availability:")
available_count = 0
for class_name in pool_classes:
    if class_name in dir(psycopg_pool):
        print(f"  ✓ {class_name}")
        available_count += 1
    else:
        print(f"  ✗ {class_name}")

print(f"\nAvailable: {available_count}/{len(pool_classes)}")

print("\nExample 3 completed successfully")
```
