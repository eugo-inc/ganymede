## Example 1: Basic PuLP linear programming setup and problem creation

```python
import pulp

# Example 1: Basic PuLP linear programming setup and problem creation
print("pulp module imported successfully")

try:
    print("Example 1 - PuLP linear programming problem creation")
    
    # Show PuLP version and main classes
    print(f"PuLP version: {pulp.__version__}")
    print(f"LpProblem class: {pulp.LpProblem}")
    print(f"LpVariable class: {pulp.LpVariable}")
    
    # Create a simple linear programming problem
    # Maximize: 3x + 2y
    # Subject to: x + y <= 4, 2x + y <= 7, x >= 0, y >= 0
    
    prob = pulp.LpProblem("Simple_LP", pulp.LpMaximize)
    print(f"Problem created: {prob.name}")
    print(f"Problem sense: {prob.sense}")
    
    # Create decision variables
    x = pulp.LpVariable("x", lowBound=0)
    y = pulp.LpVariable("y", lowBound=0)
    
    print(f"Variables created: {x.name}, {y.name}")
    print(f"Variable bounds: {x.name} >= {x.lowBound}, {y.name} >= {y.lowBound}")
    
    # Set objective function
    prob += 3*x + 2*y, "Objective_Function"
    
    # Add constraints
    prob += x + y <= 4, "Constraint_1"
    prob += 2*x + y <= 7, "Constraint_2"
    
    print("Problem formulation:")
    print(f"Objective: Maximize {3*x + 2*y}")
    print(f"Constraints: {x + y} <= 4, {2*x + y} <= 7")
    print(f"Variables: {x.name} >= 0, {y.name} >= 0")
    
    # Show problem status constants
    status_constants = [
        pulp.LpStatusOptimal, pulp.LpStatusNotSolved,
        pulp.LpStatusInfeasible, pulp.LpStatusUnbounded
    ]
    
    print("LP status constants:")
    for status in status_constants:
        print(f"  {status}")
    
    print("pulp example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pulp example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Solving optimization problems and accessing results

```python
import pulp

# Example 2: Solving optimization problems and accessing results
print("pulp module imported successfully")

try:
    print("Example 2 - PuLP problem solving and solution analysis")
    
    # Create and solve a production optimization problem
    # A company makes chairs and tables
    # Chair: 1 hour labor, 4 units wood, profit $5
    # Table: 2 hours labor, 6 units wood, profit $8
    # Available: 40 hours labor, 120 units wood
    
    prob = pulp.LpProblem("Production_Planning", pulp.LpMaximize)
    
    # Decision variables
    chairs = pulp.LpVariable("Chairs", lowBound=0, cat='Integer')
    tables = pulp.LpVariable("Tables", lowBound=0, cat='Integer')
    
    print(f"Variables: {chairs.name} (integer), {tables.name} (integer)")
    
    # Objective function: maximize profit
    prob += 5*chairs + 8*tables, "Total_Profit"
    
    # Constraints
    prob += chairs + 2*tables <= 40, "Labor_Constraint"      # Labor hours
    prob += 4*chairs + 6*tables <= 120, "Wood_Constraint"    # Wood units
    
    print("Problem formulation:")
    print("Maximize: 5*Chairs + 8*Tables")
    print("Subject to:")
    print("  Labor: Chairs + 2*Tables <= 40")
    print("  Wood: 4*Chairs + 6*Tables <= 120") 
    print("  Chairs, Tables >= 0 (integer)")
    
    # Solve the problem
    print("\nSolving the problem...")
    prob.solve(pulp.PULP_CBC_CMD(msg=0))  # Use CBC solver silently
    
    # Display results
    print(f"Status: {pulp.LpStatus[prob.status]}")
    
    if prob.status == pulp.LpStatusOptimal:
        print("Optimal solution found:")
        print(f"  Chairs to produce: {chairs.varValue}")
        print(f"  Tables to produce: {tables.varValue}")
        print(f"  Maximum profit: ${prob.objective.value()}")
        
        # Check constraint utilization
        labor_used = chairs.varValue + 2*tables.varValue
        wood_used = 4*chairs.varValue + 6*tables.varValue
        
        print("Resource utilization:")
        print(f"  Labor: {labor_used}/40 hours ({labor_used/40*100:.1f}%)")
        print(f"  Wood: {wood_used}/120 units ({wood_used/120*100:.1f}%)")
    else:
        print("No optimal solution found")
    
    print("pulp example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pulp example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: Variable types and advanced constraint handling

```python
import pulp

# Example 3: Variable types and advanced constraint handling
print("pulp module imported successfully")

try:
    print("Example 3 - PuLP variable types and advanced constraints")
    
    # Demonstrate different variable types
    variable_types = {
        'Continuous': pulp.LpContinuous,
        'Integer': pulp.LpInteger, 
        'Binary': pulp.LpBinary
    }
    
    print("PuLP variable types:")
    for name, var_type in variable_types.items():
        print(f"  {name}: {var_type}")
    
    # Create variables of different types
    cont_var = pulp.LpVariable("Continuous_Var", cat=pulp.LpContinuous)
    int_var = pulp.LpVariable("Integer_Var", lowBound=0, cat=pulp.LpInteger)
    bin_var = pulp.LpVariable("Binary_Var", cat=pulp.LpBinary)
    bounded_var = pulp.LpVariable("Bounded_Var", lowBound=5, upBound=15)
    
    print("Variable examples:")
    print(f"  {cont_var.name}: continuous, unbounded")
    print(f"  {int_var.name}: integer, >= 0") 
    print(f"  {bin_var.name}: binary (0 or 1)")
    print(f"  {bounded_var.name}: continuous, 5 <= x <= 15")
    
    # Diet optimization problem with mixed variable types
    prob = pulp.LpProblem("Diet_Optimization", pulp.LpMinimize)
    
    # Foods and their nutritional content
    foods = ['Bread', 'Milk', 'Cheese', 'Potato', 'Fish']
    costs = {'Bread': 2.0, 'Milk': 3.5, 'Cheese': 8.0, 'Potato': 1.5, 'Fish': 11.0}
    calories = {'Bread': 410, 'Milk': 150, 'Cheese': 420, 'Potato': 220, 'Fish': 560}
    protein = {'Bread': 15, 'Milk': 8, 'Cheese': 32, 'Potato': 6, 'Fish': 25}
    
    # Decision variables - amount of each food (continuous >= 0)
    food_vars = pulp.LpVariable.dicts("Food", foods, lowBound=0, cat=pulp.LpContinuous)
    
    # Binary variables - whether to include each food type
    include_vars = pulp.LpVariable.dicts("Include", foods, cat=pulp.LpBinary)
    
    print("Diet optimization problem:")
    print(f"Foods: {foods}")
    print(f"Costs per unit: {costs}")
    
    # Objective: minimize total cost
    prob += pulp.lpSum([costs[food] * food_vars[food] for food in foods])
    
    # Nutritional constraints
    prob += pulp.lpSum([calories[food] * food_vars[food] for food in foods]) >= 2000, "Calories"
    prob += pulp.lpSum([protein[food] * food_vars[food] for food in foods]) >= 50, "Protein"
    
    # Logical constraints: if food amount > 0, then include = 1
    for food in foods:
        prob += food_vars[food] <= 100 * include_vars[food], f"Include_{food}"
    
    # Maximum variety constraint
    prob += pulp.lpSum(include_vars) <= 3, "Max_Variety"
    
    print("Constraints:")
    print("  Minimum 2000 calories")
    print("  Minimum 50g protein")
    print("  Maximum 3 different food types")
    
    print("pulp example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pulp example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: Solvers and solver configuration options

```python
import pulp

# Example 4: Solvers and solver configuration options
print("pulp module imported successfully")

try:
    print("Example 4 - PuLP solvers and configuration")
    
    # List available solvers
    print("Available solvers in PuLP:")
    solvers = [
        ("PULP_CBC_CMD", "Default CBC solver"),
        ("GLPK_CMD", "GNU Linear Programming Kit"),
        ("CPLEX_CMD", "IBM CPLEX (commercial)"),
        ("GUROBI_CMD", "Gurobi Optimizer (commercial)"),
        ("COIN_CMD", "COIN-OR solvers")
    ]
    
    for solver_name, description in solvers:
        if hasattr(pulp, solver_name):
            solver_class = getattr(pulp, solver_name)
            try:
                solver = solver_class()
                available = solver.available()
                status = "✓ Available" if available else "✗ Not available"
                print(f"  {solver_name}: {description} - {status}")
            except:
                print(f"  {solver_name}: {description} - ✗ Error checking")
                raise
    
    # Solve a problem with different solver configurations
    prob = pulp.LpProblem("Solver_Demo", pulp.LpMaximize)
    
    # Simple knapsack problem
    # Items with (weight, value)
    items = {'A': (10, 60), 'B': (20, 100), 'C': (30, 120)}
    capacity = 50
    
    # Binary variables for each item
    item_vars = pulp.LpVariable.dicts("Item", items.keys(), cat=pulp.LpBinary)
    
    # Objective: maximize value
    prob += pulp.lpSum([items[item][1] * item_vars[item] for item in items])
    
    # Weight constraint
    prob += pulp.lpSum([items[item][0] * item_vars[item] for item in items]) <= capacity
    
    print("\nKnapsack problem:")
    print(f"Items: {items}")
    print(f"Capacity: {capacity}")
    
    # Solve with different solver options
    solver_configs = [
        (pulp.PULP_CBC_CMD(msg=0), "CBC (silent)"),
        (pulp.PULP_CBC_CMD(msg=1, timeLimit=10), "CBC (verbose, 10s limit)"),
    ]
    
    for solver, description in solver_configs:
        try:
            prob.solve(solver)
            print(f"\nSolver: {description}")
            print(f"Status: {pulp.LpStatus[prob.status]}")
            
            if prob.status == pulp.LpStatusOptimal:
                selected_items = [item for item in items if item_vars[item].varValue == 1]
                total_weight = sum(items[item][0] for item in selected_items)
                total_value = sum(items[item][1] for item in selected_items)
                
                print(f"Selected items: {selected_items}")
                print(f"Total weight: {total_weight}/{capacity}")
                print(f"Total value: {total_value}")
        except Exception as e:
            print(f"Solver {description} failed: {e}")
            raise
    
    # Solver parameters
    print("\nCommon solver parameters:")
    print("  msg: 0 (silent) or 1 (verbose)")
    print("  timeLimit: Maximum solving time in seconds")
    print("  gapRel: Relative optimality gap tolerance")
    print("  threads: Number of threads to use")
    
    print("pulp example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pulp example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: Real-world applications and best practices

```python
import pulp

# Example 5: Real-world applications and best practices
print("pulp module imported successfully")

try:
    print("Example 5 - PuLP real-world applications and best practices")
    
    # Transportation/logistics optimization example
    print("Transportation optimization problem:")
    
    # Warehouses and customers with supply/demand
    warehouses = ['W1', 'W2', 'W3']
    customers = ['C1', 'C2', 'C3', 'C4']
    
    # Supply at each warehouse
    supply = {'W1': 100, 'W2': 150, 'W3': 120}
    
    # Demand at each customer  
    demand = {'C1': 80, 'C2': 90, 'C3': 70, 'C4': 60}
    
    # Transportation costs (warehouse -> customer)
    costs = {
        ('W1', 'C1'): 8, ('W1', 'C2'): 6, ('W1', 'C3'): 10, ('W1', 'C4'): 9,
        ('W2', 'C1'): 9, ('W2', 'C2'): 12, ('W2', 'C3'): 13, ('W2', 'C4'): 7,
        ('W3', 'C1'): 14, ('W3', 'C2'): 9, ('W3', 'C3'): 16, ('W3', 'C4'): 5
    }
    
    print(f"Warehouses: {warehouses}")
    print(f"Customers: {customers}")
    print(f"Total supply: {sum(supply.values())}")
    print(f"Total demand: {sum(demand.values())}")
    
    # Create transportation problem
    prob = pulp.LpProblem("Transportation", pulp.LpMinimize)
    
    # Decision variables: amount shipped from warehouse w to customer c
    routes = [(w, c) for w in warehouses for c in customers]
    ship_vars = pulp.LpVariable.dicts("Ship", routes, lowBound=0, cat=pulp.LpContinuous)
    
    # Objective: minimize total transportation cost
    prob += pulp.lpSum([costs[route] * ship_vars[route] for route in routes])
    
    # Supply constraints: don't exceed warehouse capacity
    for w in warehouses:
        prob += pulp.lpSum([ship_vars[(w, c)] for c in customers]) <= supply[w], f"Supply_{w}"
    
    # Demand constraints: satisfy customer demand
    for c in customers:
        prob += pulp.lpSum([ship_vars[(w, c)] for w in warehouses]) >= demand[c], f"Demand_{c}"
    
    # Solve the problem
    prob.solve(pulp.PULP_CBC_CMD(msg=0))
    
    if prob.status == pulp.LpStatusOptimal:
        print("\nOptimal transportation plan:")
        total_cost = 0
        for route in routes:
            quantity = ship_vars[route].varValue
            if quantity > 0:
                cost = costs[route] * quantity
                print(f"  {route[0]} -> {route[1]}: {quantity} units (cost: ${cost})")
                total_cost += cost
        print(f"Total transportation cost: ${total_cost}")
    
    # PuLP best practices
    print("\nPuLP best practices:")
    practices = [
        "Use meaningful variable and constraint names",
        "Check problem status before accessing solution",
        "Handle infeasible and unbounded cases",
        "Use appropriate variable types (continuous/integer/binary)",
        "Scale problem data to avoid numerical issues",
        "Set solver time limits for large problems",
        "Validate solution feasibility",
        "Use lpSum() for efficient constraint building"
    ]
    
    for practice in practices:
        print(f"  ✓ {practice}")
    
    # Common problem types
    print("\nCommon optimization problem types:")
    problem_types = {
        "Production Planning": "Maximize profit subject to resource constraints",
        "Transportation": "Minimize shipping costs with supply/demand",
        "Assignment": "Optimal matching of resources to tasks",
        "Scheduling": "Optimize timing and resource allocation", 
        "Portfolio Optimization": "Balance risk and return in investments",
        "Network Flow": "Optimize flow through networks",
        "Facility Location": "Choose optimal locations for facilities"
    }
    
    for problem, description in problem_types.items():
        print(f"  {problem}: {description}")
    
    print("pulp example 5 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pulp example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
