## Example 1: Basic Evaluator Usage

```python
import pure_eval
import ast

# Example 1: Basic Evaluator Usage
print("Example 1 - Basic Evaluator Usage")

print(f"\npure_eval version: {pure_eval.__version__}")

# Create an evaluator with empty context
evaluator = pure_eval.Evaluator({})

# Evaluate simple expressions
expressions = ["2 + 3", "10 * 5", "2 ** 8", "(5 + 3) * 2"]

print("\nEvaluating expressions:")
for expr in expressions:
    node = ast.parse(expr, mode='eval').body
    try:
        result = evaluator[node]
        print(f"  {expr} = {result}")
    except pure_eval.CannotEval:
        print(f"  {expr} = Cannot evaluate")

print("\nExample 1 completed successfully")
```

## Example 4: is_expression_interesting Function

```python
import pure_eval
import ast

# Example 4: is_expression_interesting Function
print("Example 4 - is_expression_interesting Function")

# Test is_expression_interesting with different nodes
test_cases = [
    ("42", "Constant"),
    ("x + y", "Binary operation"),
    ("[1, 2, 3]", "List"),
]

print("\nTesting expression interestingness:")
for expr, desc in test_cases:
    node = ast.parse(expr, mode='eval').body
    # is_expression_interesting needs a frame and value
    print(f"  {desc} ({expr}): AST node type = {type(node).__name__}")

# Show available functions
print(f"\nAvailable functions:")
print(f"  Evaluator: {pure_eval.Evaluator}")
print(f"  CannotEval: {pure_eval.CannotEval}")
print(f"  is_expression_interesting: {pure_eval.is_expression_interesting}")

print("\nExample 4 completed successfully")
```

## Example 5: Complex Expressions

```python
import pure_eval
import ast

# Example 5: Complex Expressions
print("Example 5 - Complex Expressions")

# Create evaluator with various data types
context = {
    'nums': [1, 2, 3, 4, 5],
    'data': {'a': 10, 'b': 20},
    'flag': True
}
evaluator = pure_eval.Evaluator(context)

# Evaluate complex expressions
expressions = [
    "nums[0]",
    "data['a']",
    "nums[0] + data['a']",
    "flag and True",
]

print("\nEvaluating complex expressions:")
for expr in expressions:
    node = ast.parse(expr, mode='eval').body
    try:
        result = evaluator[node]
        print(f"  {expr} = {result}")
    except pure_eval.CannotEval:
        print(f"  {expr} = Cannot evaluate")

print("\nExample 5 completed successfully")
```
