```python
# Example 1: Construct a basic 3Dmol view and verify its core attributes.
import py3Dmol

view = py3Dmol.view(width=400, height=300)
assert view is not None
assert hasattr(view, 'addModel')
assert hasattr(view, 'setStyle')
assert hasattr(view, 'zoomTo')
assert hasattr(view, 'render')
print(f"view type: {type(view).__name__}")
print(f"view has addModel: {callable(view.addModel)}")
```

```python
# Example 2: Load a small inline XYZ molecule (water) and apply a style.
import py3Dmol

xyz_water = """3
Water molecule
O    0.000    0.000    0.000
H    0.757    0.586    0.000
H   -0.757    0.586    0.000
"""

view = py3Dmol.view(width=300, height=300)
view.addModel(xyz_water, 'xyz')
view.setStyle({}, {'stick': {}})
view.zoomTo()
print("Loaded water XYZ model and applied stick style")
assert view is not None
```

```python
# Example 3: Use chained styling and check that the view produces HTML output.
import py3Dmol

pdb_glycine = """ATOM      1  N   GLY A   1      -0.966   0.493   1.500  1.00  0.00           N
ATOM      2  CA  GLY A   1       0.257   0.418   0.692  1.00  0.00           C
ATOM      3  C   GLY A   1       1.526   0.583   1.500  1.00  0.00           C
ATOM      4  O   GLY A   1       2.665   0.473   0.998  1.00  0.00           O
END
"""

view = py3Dmol.view(width=400, height=300)
view.addModel(pdb_glycine, 'pdb')
view.setStyle({}, {'cartoon': {'color': 'spectrum'}, 'stick': {}})
view.setBackgroundColor('white')
view.zoomTo()
html = view._make_html()
assert isinstance(html, str) and len(html) > 0
print(f"HTML output length: {len(html)} chars")
```

```python
# Example 4: Add shapes (sphere, arrow) to a view and confirm JavaScript output contains them.
import py3Dmol

view = py3Dmol.view(width=400, height=300)
view.addSphere({'center': {'x': 0, 'y': 0, 'z': 0}, 'radius': 1.0, 'color': 'red'})
view.addArrow({
    'start': {'x': 0, 'y': 0, 'z': 0},
    'end': {'x': 5, 'y': 0, 'z': 0},
    'color': 'blue',
})
view.zoomTo()

# Forwarded method calls accumulate into the view's startjs buffer
# (view.js() is not an API method - unknown names become generated JS calls).
js = view.startjs
assert 'addSphere' in js
assert 'addArrow' in js
print("Sphere and arrow shape calls captured in generated JS")
```

```python
# Example 5: Use a grid view (multi-panel) and confirm it has the expected viewergrid attribute.
import py3Dmol

xyz_h2 = """2
Hydrogen
H 0.0 0.0 0.0
H 0.74 0.0 0.0
"""

grid = py3Dmol.view(width=600, height=300, viewergrid=(1, 2))
grid.addModel(xyz_h2, 'xyz', viewer=(0, 0))
grid.addModel(xyz_h2, 'xyz', viewer=(0, 1))
grid.setStyle({}, {'sphere': {}}, viewer=(0, 0))
grid.setStyle({}, {'stick': {}}, viewer=(0, 1))
grid.zoomTo()

html = grid._make_html()
assert isinstance(html, str) and len(html) > 0
assert 'viewergrid' in grid.startjs or 'viewer_' in html
print("Grid view (1x2) constructed and rendered to HTML successfully")
```
