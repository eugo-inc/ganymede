```python
# Example 1: Basic pybind11 Setup and Configuration Discovery
import pybind11
from pybind11 import get_cmake_dir, get_include
import os
import sys

print("=== pybind11 Setup and Configuration ===")

# Basic package information
print(f"pybind11 version: {pybind11.__version__}")
print(f"pybind11 location: {pybind11.__file__}")

# Get important paths for building extensions
try:
    include_dir = get_include()
    print(f"Include directory: {include_dir}")
    print(f"Include directory exists: {os.path.exists(include_dir)}")
    
    cmake_dir = get_cmake_dir()
    print(f"CMake directory: {cmake_dir}")
    print(f"CMake directory exists: {os.path.exists(cmake_dir)}")
    
    # Check for header files in include directory
    if os.path.exists(include_dir):
        pybind_headers = []
        for item in os.listdir(include_dir):
            if item.endswith('.h'):
                pybind_headers.append(item)
        print(f"Header files found: {len(pybind_headers)}")
        print(f"Sample headers: {pybind_headers[:3]}")
        
except Exception as e:
    print(f"Error getting paths: {e}")
    raise

# Platform and Python version compatibility
print(f"Python version: {sys.version_info.major}.{sys.version_info.minor}")
print(f"Platform: {sys.platform}")

# Package attributes
attrs = [attr for attr in dir(pybind11) if not attr.startswith('_')]
print(f"Available public attributes: {len(attrs)}")
print(f"Key utilities: {[attr for attr in attrs if 'get_' in attr or 'build' in attr.lower()]}")
print("Example 1 completed successfully!")
```

```python
# Example 2: Setup Helpers and Build Configuration
from pybind11.setup_helpers import Pybind11Extension, build_ext
from pybind11.setup_helpers import ParallelCompile, naive_recompile
import tempfile
import textwrap

print("=== Setup Helpers and Build Configuration ===")

# Test Pybind11Extension configuration
print("Testing Pybind11Extension configuration...")

# Create a minimal extension configuration
try:
    ext = Pybind11Extension(
        "test_module",
        ["example.cpp"],  # Source files (would exist in real project)
        include_dirs=[pybind11.get_include()],
        language='c++',
        cxx_std=14
    )
    
    print(f"Extension name: {ext.name}")
    print(f"Extension language: {getattr(ext, 'language', 'not specified')}")
    print(f"Include directories: {len(ext.include_dirs)} directories")
    print(f"Sources: {ext.sources}")
    
    # Check extension attributes
    ext_attrs = [attr for attr in dir(ext) if not attr.startswith('_')]
    build_attrs = [attr for attr in ext_attrs if 'compile' in attr or 'link' in attr]
    print(f"Build-related attributes: {build_attrs}")
    
except Exception as e:
    print(f"Extension configuration error: {e}")
    raise AssertionError(
        f"pybind11 example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e

# Test build_ext class
print(f"\nBuild extension class: {build_ext}")
print(f"Build extension base classes: {build_ext.__bases__}")

# Test ParallelCompile
print(f"ParallelCompile utility: {ParallelCompile}")
print(f"Naive recompile function: {naive_recompile}")

# Configuration options
print("Configuration capabilities:")
print("- Automatic C++ standard detection")
print("- Parallel compilation support")
print("- Cross-platform build configuration")
print("- CMake integration helpers")

print("Example 2 completed successfully!")
```

```python
# Example 3: CMake Integration and Build Tools
import os
import subprocess
import tempfile

print("=== CMake Integration and Build Tools ===")

# Access CMake configuration files
cmake_dir = pybind11.get_cmake_dir()
print(f"CMake configuration directory: {cmake_dir}")

if os.path.exists(cmake_dir):
    cmake_files = [f for f in os.listdir(cmake_dir) if f.endswith('.cmake')]
    print(f"CMake files available: {len(cmake_files)}")
    print(f"CMake configuration files: {cmake_files}")
    
    # Check for main pybind11 CMake config
    config_file = os.path.join(cmake_dir, 'pybind11Config.cmake')
    if os.path.exists(config_file):
        print(f"Main config file size: {os.path.getsize(config_file)} bytes")
        with open(config_file, 'r') as f:
            content = f.read()
            if 'pybind11_add_module' in content:
                print("✓ pybind11_add_module function found in CMake config")
            if 'PYBIND11_PYTHON_VERSION' in content:
                print("✓ Python version detection available")

# Create a sample CMakeLists.txt content for demonstration
cmake_content = textwrap.dedent("""
    cmake_minimum_required(VERSION 3.12)
    project(pybind11_example)
    
    find_package(pybind11 REQUIRED)
    
    pybind11_add_module(example_module example.cpp)
    
    target_compile_definitions(example_module PRIVATE VERSION_INFO="dev")
""")

print(f"Sample CMakeLists.txt content preview:")
print(cmake_content[:200] + "...")

# Test if cmake is available (optional)
try:
    cmake_version = subprocess.run(['cmake', '--version'], 
                                   capture_output=True, text=True, timeout=5)
    if cmake_version.returncode == 0:
        version_line = cmake_version.stdout.split('\n')[0]
        print(f"System CMake: {version_line}")
    else:
        print("CMake not available in PATH")
except (subprocess.TimeoutExpired, FileNotFoundError):
    print("CMake not available or timeout")

print("CMake integration provides:")
print("- Automatic Python detection")  
print("- Cross-platform build rules")
print("- Extension module configuration")
print("- Compiler flag management")
print("Example 3 completed successfully!")
```

```python
# Example 4: Header File Inspection and C++ Integration Capabilities
import os

print("=== Header File Inspection and C++ Capabilities ===")

# Inspect pybind11 header files
include_dir = pybind11.get_include()
pybind11_include = os.path.join(include_dir, 'pybind11')

if os.path.exists(pybind11_include):
    headers = [f for f in os.listdir(pybind11_include) if f.endswith('.h')]
    print(f"Core pybind11 headers: {len(headers)}")
    
    # Categorize headers by functionality
    core_headers = [h for h in headers if h in ['pybind11.h', 'cast.h', 'pytypes.h']]
    stl_headers = [h for h in headers if 'stl' in h or 'std' in h]
    eigen_headers = [h for h in headers if 'eigen' in h]
    numpy_headers = [h for h in headers if 'numpy' in h]
    
    print(f"Core binding headers: {core_headers}")
    print(f"STL integration headers: {stl_headers}")
    print(f"Eigen integration headers: {eigen_headers}")
    print(f"NumPy integration headers: {numpy_headers}")
    
    # Sample a few key headers for size analysis
    key_headers = ['pybind11.h', 'cast.h', 'stl.h', 'numpy.h']
    for header in key_headers:
        header_path = os.path.join(pybind11_include, header)
        if os.path.exists(header_path):
            size = os.path.getsize(header_path)
            print(f"{header}: {size} bytes")

# Check for detailed subdirectories
subdirs = [d for d in os.listdir(pybind11_include) if os.path.isdir(os.path.join(pybind11_include, d))]
if subdirs:
    print(f"Specialized subdirectories: {subdirs}")
    
    # Check detail directory for advanced features
    if 'detail' in subdirs:
        detail_dir = os.path.join(pybind11_include, 'detail')
        detail_files = [f for f in os.listdir(detail_dir) if f.endswith('.h')]
        print(f"Implementation detail headers: {len(detail_files)}")

# Show C++ integration capabilities
print("\nC++ Integration Capabilities:")
print("✓ Automatic type conversions (Python ↔ C++)")
print("✓ STL container support (vector, map, set, etc.)")
print("✓ NumPy array integration")
print("✓ Eigen matrix library support")
print("✓ Exception handling and translation")
print("✓ Smart pointer support (shared_ptr, unique_ptr)")
print("✓ Function overloading and default arguments")
print("✓ Class inheritance and virtual functions")

print("Example 4 completed successfully!")
```

```python
# Example 5: Sample C++ Binding Code Generation and Project Template
import textwrap

print("=== Sample C++ Binding Code and Project Template ===")

# Generate sample pybind11 C++ code
cpp_binding_code = textwrap.dedent("""
    #include <pybind11/pybind11.h>
    #include <pybind11/stl.h>
    #include <pybind11/numpy.h>
    
    // Sample C++ class to bind
    class Calculator {
    public:
        Calculator(double initial_value) : value(initial_value) {}
        
        double add(double x) { value += x; return value; }
        double multiply(double x) { value *= x; return value; }
        double get_value() const { return value; }
        
        std::vector<double> process_array(const std::vector<double>& input) {
            std::vector<double> result;
            for (double val : input) {
                result.push_back(val * 2 + 1);
            }
            return result;
        }
        
    private:
        double value;
    };
    
    // Binding code
    namespace py = pybind11;
    
    PYBIND11_MODULE(example_module, m) {
        m.doc() = "pybind11 example plugin";
        
        py::class_<Calculator>(m, "Calculator")
            .def(py::init<double>())
            .def("add", &Calculator::add)
            .def("multiply", &Calculator::multiply) 
            .def("get_value", &Calculator::get_value)
            .def("process_array", &Calculator::process_array);
            
        m.def("add_numbers", [](double a, double b) {
            return a + b;
        }, "A function that adds two numbers");
    }
""")

print("Sample C++ binding code:")
print(cpp_binding_code[:500] + "...\n")

# Generate setup.py template  
setup_py_code = textwrap.dedent(f"""
    from pybind11.setup_helpers import Pybind11Extension, build_ext
    from setuptools import setup
    import pybind11
    
    ext_modules = [
        Pybind11Extension(
            "example_module",
            ["src/main.cpp"],
            include_dirs=[pybind11.get_include()],
            cxx_std=14,
        ),
    ]
    
    setup(
        name="example_extension",
        ext_modules=ext_modules,
        cmdclass={{"build_ext": build_ext}},
        python_requires=">=3.6",
    )
""")

print("Sample setup.py:")
print(setup_py_code[:400] + "...\n")

# Show project structure
project_structure = textwrap.dedent("""
    Project Structure:
    example_project/
    ├── setup.py                 # Build configuration
    ├── src/
    │   └── main.cpp            # C++ source with bindings
    ├── CMakeLists.txt          # Alternative CMake build
    └── tests/
        └── test_module.py      # Python tests
""")

print(project_structure)

print("Building process:")
print("1. pip install pybind11[global]  # Install build dependencies")  
print("2. python setup.py build_ext --inplace  # Build extension")
print("3. python -c 'import example_module; print(example_module.add_numbers(1, 2))'")

print("\nKey pybind11 features demonstrated:")
print("✓ Class binding with constructors and methods")
print("✓ STL container integration (std::vector)")
print("✓ Lambda function binding")
print("✓ Automatic type conversion")
print("✓ Modern C++ standards support (C++14+)")
print("Example 5 completed successfully!")
```
