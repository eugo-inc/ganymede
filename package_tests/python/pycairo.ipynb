## Example 1: Basic Cairo surface creation and simple drawing

```python
import cairo
import math
import io

# Example 1: Basic Cairo surface creation and simple drawing
print("cairo module imported successfully")

try:
    print("Example 1 - Basic Cairo surface and simple drawing operations")
    
    # Create an image surface
    width, height = 400, 300
    surface = cairo.ImageSurface(cairo.FORMAT_ARGB32, width, height)
    ctx = cairo.Context(surface)
    
    print(f"Created surface: {width}x{height}")
    print(f"Surface format: {surface.get_format()}")
    print(f"Surface stride: {surface.get_stride()}")
    
    # Set background color (white)
    ctx.set_source_rgb(1, 1, 1)
    ctx.paint()
    
    # Draw simple shapes
    # Draw a red rectangle
    ctx.set_source_rgb(1, 0, 0)
    ctx.rectangle(50, 50, 100, 80)
    ctx.fill()
    
    # Draw a blue circle
    ctx.set_source_rgb(0, 0, 1)
    ctx.arc(250, 100, 40, 0, 2 * math.pi)
    ctx.fill()
    
    # Draw a green line
    ctx.set_source_rgb(0, 1, 0)
    ctx.set_line_width(3)
    ctx.move_to(50, 200)
    ctx.line_to(350, 200)
    ctx.stroke()
    
    # Get surface data
    buf = surface.get_data()
    print(f"Surface data buffer size: {len(buf)} bytes")
    
    # Surface properties
    print(f"Width: {surface.get_width()}, Height: {surface.get_height()}")
    
    # Context state information
    ctx.save()
    ctx.set_line_width(5)
    print(f"Line width after change: {ctx.get_line_width()}")
    ctx.restore()
    print(f"Line width after restore: {ctx.get_line_width()}")
    
    print("Basic drawing operations completed")
    print("cairo example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pycairo example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Advanced drawing with gradients and patterns

```python
import cairo
import math

# Example 2: Advanced drawing with gradients and patterns
print("cairo module imported successfully")

try:
    print("Example 2 - Advanced Cairo drawing with gradients and patterns")
    
    # Create surface and context
    width, height = 500, 400
    surface = cairo.ImageSurface(cairo.FORMAT_ARGB32, width, height)
    ctx = cairo.Context(surface)
    
    # White background
    ctx.set_source_rgb(1, 1, 1)
    ctx.paint()
    
    # Linear gradient example
    print("Creating linear gradient...")
    gradient = cairo.LinearGradient(50, 50, 200, 150)
    gradient.add_color_stop_rgb(0, 1, 0, 0)    # Red at start
    gradient.add_color_stop_rgb(0.5, 1, 1, 0)  # Yellow in middle  
    gradient.add_color_stop_rgb(1, 0, 0, 1)    # Blue at end
    
    ctx.set_source(gradient)
    ctx.rectangle(50, 50, 150, 100)
    ctx.fill()
    
    # Radial gradient example
    print("Creating radial gradient...")
    radial = cairo.RadialGradient(325, 100, 10, 325, 100, 60)
    radial.add_color_stop_rgba(0, 1, 1, 1, 1)    # White center
    radial.add_color_stop_rgba(0.7, 0.8, 0.8, 0.8, 0.8)  # Light gray
    radial.add_color_stop_rgba(1, 0, 0, 0, 0.5)  # Transparent black edge
    
    ctx.set_source(radial)
    ctx.arc(325, 100, 60, 0, 2 * math.pi)
    ctx.fill()
    
    # Pattern from surface
    print("Creating pattern from surface...")
    pattern_surface = cairo.ImageSurface(cairo.FORMAT_ARGB32, 20, 20)
    pattern_ctx = cairo.Context(pattern_surface)
    
    # Create checkerboard pattern
    pattern_ctx.set_source_rgb(0.8, 0.8, 0.8)
    pattern_ctx.paint()
    pattern_ctx.set_source_rgb(0.2, 0.2, 0.2)
    pattern_ctx.rectangle(0, 0, 10, 10)
    pattern_ctx.rectangle(10, 10, 10, 10)
    pattern_ctx.fill()
    
    # Apply pattern
    pattern = cairo.SurfacePattern(pattern_surface)
    pattern.set_extend(cairo.EXTEND_REPEAT)
    ctx.set_source(pattern)
    ctx.rectangle(50, 200, 400, 100)
    ctx.fill()
    
    # Text with gradient
    print("Drawing text with gradient...")
    text_gradient = cairo.LinearGradient(50, 320, 450, 350)
    text_gradient.add_color_stop_rgb(0, 0.8, 0.2, 0.2)
    text_gradient.add_color_stop_rgb(1, 0.2, 0.2, 0.8)
    
    ctx.select_font_face("Arial", cairo.FONT_SLANT_NORMAL, cairo.FONT_WEIGHT_BOLD)
    ctx.set_font_size(24)
    ctx.move_to(50, 350)
    ctx.set_source(gradient)
    ctx.show_text("Cairo Graphics with Python")
    
    # Pattern and gradient info
    print(f"Linear gradient color stops: 3")
    print(f"Radial gradient color stops: 3") 
    print(f"Pattern surface size: 20x20")
    print(f"Pattern extend mode: {pattern.get_extend()}")
    
    print("Advanced drawing with gradients completed")
    print("cairo example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pycairo example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: Path operations and transformations

```python
import cairo
import math

# Example 3: Path operations and transformations
print("cairo module imported successfully")

try:
    print("Example 3 - Cairo path operations and coordinate transformations")
    
    # Create surface and context
    width, height = 600, 500
    surface = cairo.ImageSurface(cairo.FORMAT_ARGB32, width, height)
    ctx = cairo.Context(surface)
    
    # White background
    ctx.set_source_rgb(1, 1, 1)
    ctx.paint()
    
    # Complex path creation
    print("Creating complex path...")
    ctx.set_source_rgb(0, 0, 0)
    ctx.set_line_width(2)
    
    # Star shape using path operations
    def draw_star(ctx, x, y, inner_radius, outer_radius, points):
        angle_step = 2 * math.pi / points
        ctx.move_to(x + outer_radius, y)
        
        for i in range(1, points * 2 + 1):
            angle = i * angle_step / 2
            if i % 2 == 0:
                radius = outer_radius
            else:
                radius = inner_radius
            px = x + radius * math.cos(angle)
            py = y + radius * math.sin(angle)
            ctx.line_to(px, py)
        ctx.close_path()
    
    # Draw star
    draw_star(ctx, 150, 150, 30, 60, 5)
    ctx.set_source_rgb(1, 0.8, 0)
    ctx.fill_preserve()
    ctx.set_source_rgb(0, 0, 0)
    ctx.stroke()
    
    # Transformation matrix operations
    print("Applying transformations...")
    ctx.save()
    
    # Translation and rotation
    ctx.translate(400, 150)
    ctx.rotate(math.pi / 4)  # 45 degrees
    ctx.scale(1.5, 0.8)
    
    # Draw transformed rectangle
    ctx.set_source_rgb(0, 0.5, 1)
    ctx.rectangle(-40, -25, 80, 50)
    ctx.fill()
    
    ctx.restore()
    
    # Bezier curve paths
    print("Drawing Bezier curves...")
    ctx.set_source_rgb(1, 0, 0)
    ctx.set_line_width(3)
    
    # Cubic Bezier curve
    ctx.move_to(50, 300)
    ctx.curve_to(100, 250, 200, 350, 250, 300)
    ctx.stroke()
    
    # Quadratic-like curve using cubic Bezier
    ctx.move_to(300, 300)
    control_x, control_y = 375, 250
    end_x, end_y = 450, 300
    ctx.curve_to(
        control_x, control_y,
        control_x, control_y,
        end_x, end_y
    )
    ctx.stroke()
    
    # Path information and analysis
    print("Analyzing current path...")
    
    # Create a path for analysis
    ctx.new_path()
    ctx.move_to(100, 400)
    ctx.line_to(200, 420)
    ctx.curve_to(250, 380, 300, 440, 350, 400)
    
    # Get current point
    try:
        current_x, current_y = ctx.get_current_point()
        print(f"Current path point: ({current_x:.1f}, {current_y:.1f})")
    except:
        print("No current path point")
        raise
    
    # Stroke the analysis path
    ctx.set_source_rgb(0, 0.8, 0)
    ctx.stroke()
    
    # Matrix operations
    matrix = ctx.get_matrix()
    print(f"Current transformation matrix:")
    print(f"  xx: {matrix.xx:.3f}, xy: {matrix.xy:.3f}")
    print(f"  yx: {matrix.yx:.3f}, yy: {matrix.yy:.3f}")
    print(f"  x0: {matrix.x0:.3f}, y0: {matrix.y0:.3f}")
    
    # Path extents
    ctx.new_path()
    ctx.rectangle(50, 350, 300, 100)
    x1, y1, x2, y2 = ctx.path_extents()
    print(f"Path extents: ({x1}, {y1}) to ({x2}, {y2})")
    ctx.stroke()
    
    print("Path operations and transformations completed")
    print("cairo example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pycairo example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: Text Rendering and Measurements

```python
import cairo

# Example 4: Text Rendering and Measurements
print("Example 4 - Text Rendering and Measurements")

# Create surface and context
surface = cairo.ImageSurface(cairo.FORMAT_ARGB32, 400, 200)
ctx = cairo.Context(surface)

# White background
ctx.set_source_rgb(1, 1, 1)
ctx.paint()

# Set font and draw text
ctx.select_font_face("Arial", cairo.FONT_SLANT_NORMAL, cairo.FONT_WEIGHT_BOLD)
ctx.set_font_size(24)

text = "Cairo Text Example"
text_extents = ctx.text_extents(text)
font_extents = ctx.font_extents()

# Unpack font extents tuple (ascent, descent, height, max_x_advance, max_y_advance)
ascent, descent, height, max_x_adv, max_y_adv = font_extents

print(f"Text: '{text}'")
print(f"Text width: {text_extents[2]:.2f}, height: {text_extents[3]:.2f}")
print(f"Font ascent: {ascent:.2f}, descent: {descent:.2f}, height: {height:.2f}")

# Draw the text
ctx.set_source_rgb(0, 0, 0.8)
ctx.move_to(50, 100)
ctx.show_text(text)

print("\nExample 4 completed successfully")
```

## Example 5: Surface formats and output operations

```python
import cairo
import io
import math

# Example 5: Surface formats and output operations
print("cairo module imported successfully")

try:
    print("Example 5 - Cairo surface formats and output operations")
    
    # Different surface formats
    formats = [
        (cairo.FORMAT_ARGB32, "ARGB32"),
        (cairo.FORMAT_RGB24, "RGB24"), 
        (cairo.FORMAT_A8, "A8"),
        (cairo.FORMAT_A1, "A1")
    ]
    
    surfaces = {}
    
    print("Creating surfaces with different formats...")
    for format_type, format_name in formats:
        surface = cairo.ImageSurface(format_type, 200, 150)
        surfaces[format_name] = surface
        
        print(f"  {format_name}:")
        print(f"    Format: {surface.get_format()}")
        print(f"    Stride: {surface.get_stride()}")
        print(f"    Width: {surface.get_width()}")
        print(f"    Height: {surface.get_height()}")
    
    # Draw on ARGB32 surface (full color)
    print("\nDrawing on ARGB32 surface...")
    surface = surfaces["ARGB32"]
    ctx = cairo.Context(surface)
    
    # Gradient background
    gradient = cairo.LinearGradient(0, 0, 200, 150)
    gradient.add_color_stop_rgb(0, 0.2, 0.4, 0.8)
    gradient.add_color_stop_rgb(1, 0.8, 0.6, 0.2)
    ctx.set_source(gradient)
    ctx.paint()
    
    # Draw some shapes
    ctx.set_source_rgb(1, 1, 1)
    ctx.arc(100, 75, 30, 0, 2 * math.pi)
    ctx.fill()
    
    ctx.set_source_rgb(0, 0, 0)
    ctx.rectangle(50, 100, 100, 20)
    ctx.fill()
    
    # PNG output to memory buffer
    print("Writing PNG data to memory buffer...")
    
    png_buffer = io.BytesIO()
    surface.write_to_png(png_buffer)
    png_data = png_buffer.getvalue()
    
    print(f"PNG buffer size: {len(png_data)} bytes")
    print(f"PNG magic bytes: {png_data[:8].hex()}")
    
    # Surface data access
    print("Accessing raw surface data...")
    
    # Get pixel data
    buf = surface.get_data()
    print(f"Raw buffer size: {len(buf)} bytes")
    
    # Analyze first few pixels (ARGB32 format)
    stride = surface.get_stride()
    width = surface.get_width()
    
    # Sample pixel analysis
    for y in range(min(3, surface.get_height())):
        for x in range(min(5, width)):
            offset = y * stride + x * 4  # 4 bytes per pixel in ARGB32
            if offset + 3 < len(buf):
                b = buf[offset]
                g = buf[offset + 1]  
                r = buf[offset + 2]
                a = buf[offset + 3]
                print(f"  Pixel ({x},{y}): ARGB({a},{r},{g},{b})")
    
    # RGB24 surface operations
    print("\nWorking with RGB24 format...")
    rgb_surface = surfaces["RGB24"]
    rgb_ctx = cairo.Context(rgb_surface)
    
    # Simple drawing on RGB surface
    rgb_ctx.set_source_rgb(0.8, 0.2, 0.2)
    rgb_ctx.paint()
    
    rgb_ctx.set_source_rgb(1, 1, 1)
    rgb_ctx.move_to(10, 30)
    rgb_ctx.select_font_face("Arial", cairo.FONT_SLANT_NORMAL, cairo.FONT_WEIGHT_BOLD)
    rgb_ctx.set_font_size(16)
    rgb_ctx.show_text("RGB24 Format")
    
    # Surface copying and conversion
    print("Demonstrating surface operations...")
    
    # Create a new surface from existing data
    target_surface = cairo.ImageSurface(cairo.FORMAT_ARGB32, 200, 150)
    target_ctx = cairo.Context(target_surface)
    
    # Paint source surface onto target
    target_ctx.set_source_surface(surface, 0, 0)
    target_ctx.paint()
    
    # Add overlay
    target_ctx.set_source_rgba(1, 1, 1, 0.3)
    target_ctx.rectangle(0, 0, 200, 150)
    target_ctx.fill()
    
    print("Surface copying completed")
    
    # Surface properties summary
    print("\nSurface properties summary:")
    for format_name, surf in surfaces.items():
        buf_size = len(surf.get_data())
        print(f"  {format_name}: {surf.get_width()}x{surf.get_height()}, {buf_size} bytes")
    
    # Memory usage calculation
    total_memory = sum(len(surf.get_data()) for surf in surfaces.values())
    print(f"Total surface memory: {total_memory} bytes ({total_memory/1024:.1f} KB)")
    
    print("Surface formats and output operations completed")
    print("cairo example 5 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pycairo example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
