## Example 1: Basic PEP 8 Style Checking

```python
import pycodestyle
import tempfile
import os

# Example 1: Basic PEP 8 Style Checking
print("Example 1 - Basic PEP 8 Style Checking")

# Create sample code with style issues
sample_code = '''
def badFunction( x,y ):
    if x>y:
        return x+y
'''

# Write to temporary file
with tempfile.NamedTemporaryFile(mode='w', suffix='.py', delete=False) as f:
    f.write(sample_code)
    temp_file = f.name

# Check the file
style_guide = pycodestyle.StyleGuide()
result = style_guide.check_files([temp_file])

print(f"Total errors found: {result.get_count()}")
print(f"Error types: {len(result.counters)}")

# Clean up
os.unlink(temp_file)

print("\nExample 1 completed successfully")
```

## Example 2: Custom Style Guide Configuration

```python
import pycodestyle
import tempfile
import os

# Example 2: Custom Style Guide Configuration
print("Example 2 - Custom Style Guide Configuration")

# Sample code
code = '''import os,sys
def func(x,y):
    return x+y
'''

with tempfile.NamedTemporaryFile(mode='w', suffix='.py', delete=False) as f:
    f.write(code)
    temp_file = f.name

# Custom configuration
custom_guide = pycodestyle.StyleGuide(
    max_line_length=100,
    ignore=['E231']  # Ignore missing whitespace after comma
)

result = custom_guide.check_files([temp_file])
print(f"Errors with custom config: {result.get_count()}")

# Compare with default
default_guide = pycodestyle.StyleGuide()
default_result = default_guide.check_files([temp_file])
print(f"Errors with default config: {default_result.get_count()}")

os.unlink(temp_file)

print("\nExample 2 completed successfully")
```

## Example 3: Error Statistics and Analysis

```python
import pycodestyle
import tempfile
import os

# Example 3: Error Statistics and Analysis
print("Example 3 - Error Statistics and Analysis")

# Create code with various errors
code = '''import os,sys
def bad_func(x,y,z):
    result=x+y*z
    if result>10:return result
    else:return 0
'''

with tempfile.NamedTemporaryFile(mode='w', suffix='.py', delete=False) as f:
    f.write(code)
    temp_file = f.name

# Check with statistics
guide = pycodestyle.StyleGuide(quiet=True)
result = guide.check_files([temp_file])

print(f"Total errors: {result.get_count()}")

# Show error breakdown
print("\nError breakdown:")
for error_code, count in sorted(result.counters.items()):
    if not error_code.startswith(('physical', 'logical', 'files', 'directories')):
        print(f"  {error_code}: {count}")

os.unlink(temp_file)

print("\nExample 3 completed successfully")
```

## Example 4: Selective Error Checking

```python
import pycodestyle
import tempfile
import os

# Example 4: Selective Error Checking
print("Example 4 - Selective Error Checking")

code = '''import os,sys
def func(x,y):
    if x>y:return x+y
    return x-y
'''

with tempfile.NamedTemporaryFile(mode='w', suffix='.py', delete=False) as f:
    f.write(code)
    temp_file = f.name

# Check only specific error types
configs = [
    ('All errors', {}),
    ('Only E2xx', {'select': ['E2']}),
    ('Only E7xx', {'select': ['E7']}),
]

for name, options in configs:
    guide = pycodestyle.StyleGuide(quiet=True, **options)
    result = guide.check_files([temp_file])
    print(f"{name}: {result.get_count()} errors")

os.unlink(temp_file)

print("\nExample 4 completed successfully")
```

## Example 5: Code Quality Comparison

```python
import pycodestyle
import tempfile
import os

# Example 5: Code Quality Comparison
print("Example 5 - Code Quality Comparison")

good_code = '''import os

def add(x, y):
    return x + y
'''

bad_code = '''import os,sys
def add(x,y):
    return x+y
'''

results = {}
for name, code in [('Good', good_code), ('Bad', bad_code)]:
    with tempfile.NamedTemporaryFile(mode='w', suffix='.py', delete=False) as f:
        f.write(code)
        temp_file = f.name
    
    guide = pycodestyle.StyleGuide(quiet=True)
    result = guide.check_files([temp_file])
    results[name] = result.get_count()
    
    os.unlink(temp_file)

print(f"Good code errors: {results['Good']}")
print(f"Bad code errors: {results['Bad']}")
print(f"Improvement: {results['Bad'] - results['Good']} errors")

print("\nExample 5 completed successfully")
```
