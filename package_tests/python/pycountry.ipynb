## Example 1: Basic country lookup and information retrieval

```python
import pycountry

# Example 1: Basic country lookup and information retrieval
print("pycountry module imported successfully")

try:
    print("Example 1 - Basic pycountry country lookup and information")
    
    # Lookup countries by different attributes
    print("Country lookups by different attributes:")
    
    # By alpha-2 code
    us = pycountry.countries.get(alpha_2='US')
    print(f"US by alpha-2: {us.name} ({us.alpha_2}, {us.alpha_3})")
    
    # By alpha-3 code
    can = pycountry.countries.get(alpha_3='CAN')
    print(f"Canada by alpha-3: {can.name} ({can.alpha_2}, {can.alpha_3})")
    
    # By name
    uk = pycountry.countries.get(name='United Kingdom')
    print(f"UK by name: {uk.name} ({uk.alpha_2}, {uk.alpha_3})")
    
    # By numeric code
    de = pycountry.countries.get(numeric='276')
    print(f"Germany by numeric: {de.name} ({de.alpha_2}, numeric: {de.numeric})")
    
    # Show country attributes
    print(f"\nDetailed country information for {us.name}:")
    print(f"  Official name: {getattr(us, 'official_name', 'N/A')}")
    print(f"  Alpha-2 code: {us.alpha_2}")
    print(f"  Alpha-3 code: {us.alpha_3}")
    print(f"  Numeric code: {us.numeric}")
    
    # Count total countries
    total_countries = len(list(pycountry.countries))
    print(f"\nTotal countries in database: {total_countries}")
    
    # Show some example countries
    print("Sample countries:")
    for i, country in enumerate(pycountry.countries):
        if i >= 5:
            break
        print(f"  {country.alpha_2}: {country.name}")
    
    print("Basic country lookup completed")
    print("pycountry example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pycountry example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Language and currency information

```python
import pycountry

# Example 2: Language and currency information
print("pycountry module imported successfully")

try:
    print("Example 2 - Language and currency information lookup")
    
    # Language lookup
    print("Language information:")
    
    # By ISO 639-1 code
    en = pycountry.languages.get(alpha_2='en')
    print(f"English by alpha-2: {en.name} ({en.alpha_2})")
    
    # By ISO 639-3 code  
    fra = pycountry.languages.get(alpha_3='fra')
    print(f"French by alpha-3: {fra.name} ({fra.alpha_2}, {fra.alpha_3})")
    
    # By name
    spanish = pycountry.languages.get(name='Spanish')
    print(f"Spanish by name: {spanish.name} ({spanish.alpha_2}, {spanish.alpha_3})")
    
    # Show language with different scopes
    languages_sample = []
    for lang in pycountry.languages:
        languages_sample.append(lang)
        if len(languages_sample) >= 3:
            break
    
    print(f"\nSample languages:")
    for lang in languages_sample:
        scope = getattr(lang, 'scope', 'individual')
        lang_type = getattr(lang, 'type', 'living')
        print(f"  {lang.name}: scope={scope}, type={lang_type}")
    
    # Currency lookup
    print(f"\nCurrency information:")
    
    # By alphabetic code
    usd = pycountry.currencies.get(alpha_3='USD')
    print(f"USD: {usd.name} (numeric: {usd.numeric})")
    
    eur = pycountry.currencies.get(alpha_3='EUR')
    print(f"EUR: {eur.name} (numeric: {eur.numeric})")
    
    jpy = pycountry.currencies.get(alpha_3='JPY')
    print(f"JPY: {jpy.name} (numeric: {jpy.numeric})")
    
    # Count totals
    total_languages = len(list(pycountry.languages))
    total_currencies = len(list(pycountry.currencies))
    
    print(f"\nDatabase statistics:")
    print(f"  Total languages: {total_languages}")
    print(f"  Total currencies: {total_currencies}")
    
    # Show sample currencies
    currencies_sample = list(pycountry.currencies)[:5]
    print(f"Sample currencies:")
    for curr in currencies_sample:
        print(f"  {curr.alpha_3}: {curr.name}")
    
    print("Language and currency information completed")
    print("pycountry example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pycountry example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: Country subdivisions and advanced lookups

```python
import pycountry

# Example 3: Country subdivisions and advanced lookups
print("pycountry module imported successfully")

try:
    print("Example 3 - Country subdivisions and advanced geographic data")
    
    # Country subdivisions (states, provinces, etc.)
    print("Country subdivision examples:")
    
    # US states
    us_subdivisions = []
    for subdivision in pycountry.subdivisions:
        if subdivision.country_code == 'US':
            us_subdivisions.append(subdivision)
            if len(us_subdivisions) >= 5:
                break
    
    print("US states/territories:")
    for sub in us_subdivisions:
        subdivision_type = getattr(sub, 'type', 'N/A')
        print(f"  {sub.code}: {sub.name} ({subdivision_type})")
    
    # Canadian provinces
    ca_subdivisions = []
    for subdivision in pycountry.subdivisions:
        if subdivision.country_code == 'CA':
            ca_subdivisions.append(subdivision)
            if len(ca_subdivisions) >= 5:
                break
    
    print(f"\nCanadian provinces/territories:")
    for sub in ca_subdivisions:
        subdivision_type = getattr(sub, 'type', 'N/A')
        print(f"  {sub.code}: {sub.name} ({subdivision_type})")
    
    # Specific subdivision lookup
    california = pycountry.subdivisions.get(code='US-CA')
    if california:
        print(f"\nCalifornia details:")
        print(f"  Name: {california.name}")
        print(f"  Code: {california.code}")
        print(f"  Country: {california.country_code}")
        print(f"  Type: {getattr(california, 'type', 'N/A')}")
    
    # Search functionality
    print(f"\nSearch examples:")
    
    # Find countries containing "island"
    island_countries = []
    for country in pycountry.countries:
        if 'island' in country.name.lower():
            island_countries.append(country)
    
    print(f"Countries with 'island' in name ({len(island_countries)}):")
    for country in island_countries[:3]:
        print(f"  {country.alpha_2}: {country.name}")
    
    # Find languages starting with 'A'
    a_languages = []
    for language in pycountry.languages:
        if language.name.startswith('A'):
            a_languages.append(language)
            if len(a_languages) >= 3:
                break
    
    print(f"\nLanguages starting with 'A':")
    for lang in a_languages:
        print(f"  {getattr(lang, 'alpha_2', 'N/A')}: {lang.name}")
    
    # Count subdivisions by country
    subdivision_counts = {}
    for subdivision in pycountry.subdivisions:
        country_code = subdivision.country_code
        subdivision_counts[country_code] = subdivision_counts.get(country_code, 0) + 1
    
    top_countries = sorted(subdivision_counts.items(), key=lambda x: x[1], reverse=True)[:3]
    print(f"\nCountries with most subdivisions:")
    for country_code, count in top_countries:
        country = pycountry.countries.get(alpha_2=country_code)
        country_name = country.name if country else country_code
        print(f"  {country_code} ({country_name}): {count} subdivisions")
    
    print("Country subdivisions and advanced lookups completed")
    print("pycountry example 3 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pycountry example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: Scripts and practical data validation

```python
import pycountry

# Example 4: Scripts and practical data validation
print("pycountry module imported successfully")

try:
    print("Example 4 - Scripts and practical data validation use cases")
    
    # Script information
    print("Script (writing system) information:")
    
    # Common scripts
    scripts_of_interest = ['Latn', 'Arab', 'Cyrl', 'Hans', 'Hira']
    
    for script_code in scripts_of_interest:
        script = pycountry.scripts.get(alpha_4=script_code)
        if script:
            print(f"  {script.alpha_4}: {script.name}")
            if hasattr(script, 'numeric'):
                print(f"    Numeric code: {script.numeric}")
    
    # Practical validation functions
    print(f"\nPractical validation examples:")
    
    def validate_country_code(code):
        """Validate if a country code exists"""
        country = pycountry.countries.get(alpha_2=code.upper())
        return country is not None, country
    
    def validate_currency_code(code):
        """Validate if a currency code exists"""
        currency = pycountry.currencies.get(alpha_3=code.upper())
        return currency is not None, currency
    
    def validate_language_code(code):
        """Validate if a language code exists"""
        language = pycountry.languages.get(alpha_2=code.lower())
        return language is not None, language
    
    # Test validation functions
    test_codes = [
        ('country', 'US'), ('country', 'GB'), ('country', 'XX'),
        ('currency', 'USD'), ('currency', 'EUR'), ('currency', 'XXX'),
        ('language', 'en'), ('language', 'fr'), ('language', 'zz')
    ]
    
    print("Code validation results:")
    for code_type, code in test_codes:
        if code_type == 'country':
            valid, obj = validate_country_code(code)
        elif code_type == 'currency':
            valid, obj = validate_currency_code(code)
        elif code_type == 'language':
            valid, obj = validate_language_code(code)
        
        status = "✓" if valid else "✗"
        name = obj.name if valid and obj else "Invalid"
        print(f"  {status} {code_type.upper()} {code}: {name}")
    
    # Geographic region analysis
    print(f"\nGeographic analysis:")
    
    # Analyze European countries
    european_countries = []
    for country in pycountry.countries:
        # Simple heuristic for European countries (not comprehensive)
        if any(term in country.name.lower() for term in ['iceland', 'norway', 'sweden', 'finland', 'denmark']):
            european_countries.append(country)
    
    print(f"Nordic countries sample:")
    for country in european_countries[:3]:
        print(f"  {country.alpha_2}: {country.name}")
    
    # Create a simple country data structure
    country_data = {
        'US': {'currency': 'USD', 'languages': ['en'], 'subdivisions': []},
        'CA': {'currency': 'CAD', 'languages': ['en', 'fr'], 'subdivisions': []},
        'DE': {'currency': 'EUR', 'languages': ['de'], 'subdivisions': []},
    }
    
    # Enrich with pycountry data
    for country_code, data in country_data.items():
        country = pycountry.countries.get(alpha_2=country_code)
        if country:
            data['official_name'] = country.name
            data['alpha_3'] = country.alpha_3
            data['numeric'] = country.numeric
    
    print(f"\nEnriched country data structure:")
    for code, data in country_data.items():
        print(f"  {code}:")
        for key, value in data.items():
            print(f"    {key}: {value}")
    
    print("Scripts and practical validation completed")
    print("pycountry example 4 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pycountry example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: Comprehensive database analysis and utility functions

```python
import pycountry

# Example 5: Comprehensive database analysis and utility functions
print("pycountry module imported successfully")

try:
    print("Example 5 - Comprehensive pycountry database analysis and utilities")
    
    # Complete database statistics
    print("Complete pycountry database statistics:")
    
    databases = {
        'Countries': pycountry.countries,
        'Languages': pycountry.languages,
        'Currencies': pycountry.currencies,
        'Scripts': pycountry.scripts,
        'Subdivisions': pycountry.subdivisions
    }
    
    for db_name, db in databases.items():
        count = len(list(db))
        print(f"  {db_name}: {count} entries")
    
    # Utility functions for common tasks
    print(f"\nUtility functions demonstration:")
    
    def get_country_info(country_code):
        """Get comprehensive country information"""
        country = pycountry.countries.get(alpha_2=country_code.upper())
        if not country:
            return None
        
        # Get subdivisions for this country
        subdivisions = [s for s in pycountry.subdivisions if s.country_code == country.alpha_2]
        
        return {
            'name': country.name,
            'alpha_2': country.alpha_2,
            'alpha_3': country.alpha_3,
            'numeric': country.numeric,
            'official_name': getattr(country, 'official_name', None),
            'subdivision_count': len(subdivisions)
        }
    
    def find_countries_by_region(region_keywords):
        """Find countries containing region keywords"""
        matching_countries = []
        for country in pycountry.countries:
            if any(keyword.lower() in country.name.lower() for keyword in region_keywords):
                matching_countries.append(country)
        return matching_countries
    
    def get_currency_users(currency_code):
        """Get countries that might use a currency (simplified)"""
        # Note: This is a simplified example - real currency usage requires external data
        common_currency_mappings = {
            'USD': ['US', 'EC', 'BQ', 'SV', 'GU', 'MH', 'FM', 'MP', 'PW', 'PR', 'TC', 'VG', 'VI'],
            'EUR': ['AD', 'AT', 'BE', 'CY', 'DE', 'EE', 'ES', 'FI', 'FR', 'GR', 'IE', 'IT', 'LT', 'LU', 'LV', 'MC', 'MT', 'NL', 'PT', 'SI', 'SK', 'SM', 'VA'],
            'GBP': ['GB', 'GG', 'IM', 'JE']
        }
        
        country_codes = common_currency_mappings.get(currency_code.upper(), [])
        countries = []
        for code in country_codes[:5]:  # Limit to first 5
            country = pycountry.countries.get(alpha_2=code)
            if country:
                countries.append(country)
        return countries
    
    # Test utility functions
    print("Testing utility functions:")
    
    # Test get_country_info
    for code in ['US', 'FR', 'JP']:
        info = get_country_info(code)
        if info:
            print(f"  {code}: {info['name']} ({info['subdivision_count']} subdivisions)")
    
    # Test find_countries_by_region
    island_countries = find_countries_by_region(['Island', 'Islands'])
    print(f"\nCountries with 'Island' in name ({len(island_countries)}):")
    for country in island_countries[:4]:
        print(f"  {country.alpha_2}: {country.name}")
    
    # Test get_currency_users
    for curr_code in ['USD', 'EUR']:
        currency = pycountry.currencies.get(alpha_3=curr_code)
        users = get_currency_users(curr_code)
        print(f"\n{curr_code} ({currency.name}) users (sample):")
        for country in users:
            print(f"  {country.alpha_2}: {country.name}")
    
    # Data quality analysis
    print(f"\nData quality analysis:")
    
    # Check for countries with/without official names
    countries_with_official = sum(1 for c in pycountry.countries if hasattr(c, 'official_name'))
    total_countries = len(list(pycountry.countries))
    
    print(f"  Countries with official names: {countries_with_official}/{total_countries}")
    
    # Check language coverage
    languages_with_alpha2 = sum(1 for l in pycountry.languages if hasattr(l, 'alpha_2'))
    total_languages = len(list(pycountry.languages))
    
    print(f"  Languages with ISO 639-1 codes: {languages_with_alpha2}/{total_languages}")
    
    # Most common subdivision types
    subdivision_types = {}
    for subdivision in pycountry.subdivisions:
        sub_type = getattr(subdivision, 'type', 'unknown')
        subdivision_types[sub_type] = subdivision_types.get(sub_type, 0) + 1
    
    top_subdivision_types = sorted(subdivision_types.items(), key=lambda x: x[1], reverse=True)[:3]
    print(f"  Top subdivision types:")
    for sub_type, count in top_subdivision_types:
        print(f"    {sub_type}: {count}")
    
    print("Comprehensive database analysis completed")
    print("pycountry example 5 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pycountry example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
