# pycparser: 5 Basic Usage Examples

```python
# Example 1: Parse a simple C expression
from pycparser import c_parser
parser = c_parser.CParser()
ast = parser.parse('int x = 1 + 2;')
print(ast)
```

```python
# Example 2: Parse a C function
from pycparser import c_parser
parser = c_parser.CParser()
code = 'int add(int a, int b) { return a + b; }'
ast = parser.parse(code)
print(ast)
```

```python
# Example 3: Walk the AST
from pycparser import c_parser, c_ast
parser = c_parser.CParser()
ast = parser.parse('int x = 1 + 2;')
ast.show()
```

```python
# Example 4: Parse a C struct
from pycparser import c_parser
parser = c_parser.CParser()
code = 'struct Point { int x; int y; };'
ast = parser.parse(code)
ast.show()
```
