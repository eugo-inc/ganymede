## Example 1: Basic CRDT document creation and array operations

```python
import pycrdt

# Example 1: Basic CRDT document creation and array operations
print("pycrdt module imported successfully")

try:
    print("Example 1 - Basic pycrdt CRDT document and data structure operations")
    
    # Create a new Y-doc (Yjs-compatible document)
    doc1 = pycrdt.Doc()
    print(f"Created document 1: {type(doc1).__name__}")
    print(f"Document client ID: {doc1.client_id}")
    
    # Create an array for collaborative data
    array1 = doc1.get("items", type=pycrdt.Array)
    print(f"Created array: {type(array1).__name__}")
    
    # Add items to the array
    array1.append("Hello")
    array1.append("World")
    array1.append("from")
    array1.append("CRDT")
    
    print(f"Array content: {list(array1)}")
    print(f"Array length: {len(array1)}")
    
    # Create a map for key-value data
    map1 = doc1.get("data", type=pycrdt.Map)
    map1["message"] = "Collaborative editing"
    map1["version"] = "1.0"
    map1["active"] = True
    
    print(f"Map content: {dict(map1)}")
    
    # Test synchronization with another document
    doc2 = pycrdt.Doc()
    print(f"Created document 2: {type(doc2).__name__}")
    
    # Get update from first document and apply to second
    update = doc1.get_update()
    print(f"Update size: {len(update)} bytes")
    
    doc2.apply_update(update)
    
    # Access synced data structures
    array2 = doc2.get("items", type=pycrdt.Array)
    map2 = doc2.get("data", type=pycrdt.Map)
    
    print(f"Synced array: {list(array2)}")
    print(f"Synced map: {dict(map2)}")
    print(f"Documents synchronized: {list(array1) == list(array2)}")
    
    # Show available CRDT types
    crdt_types = [attr for attr in dir(pycrdt) if not attr.startswith('_') and attr in ['Array', 'Map', 'Text', 'Doc']]
    print(f"Main CRDT types: {crdt_types}")
    
    print("Basic CRDT document operations completed")
    print("pycrdt example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pycrdt example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Conflict resolution and concurrent operations

```python
import pycrdt

# Example 2: Conflict resolution and concurrent operations
print("pycrdt module imported successfully")

try:
    print("Example 2 - CRDT conflict resolution and concurrent operations")
    
    # Create two independent documents
    doc_alice = pycrdt.Doc()
    doc_bob = pycrdt.Doc()
    
    print("Created documents for Alice and Bob")
    print(f"Alice client ID: {doc_alice.client_id}")
    print(f"Bob client ID: {doc_bob.client_id}")
    
    # Both start with the same initial state
    alice_array = doc_alice.get("shared_list", type=pycrdt.Array)
    bob_array = doc_bob.get("shared_list", type=pycrdt.Array)
    
    # Initial sync
    initial_update = doc_alice.get_update()
    doc_bob.apply_update(initial_update)
    
    # Alice adds items
    alice_array.append("Alice_item_1")
    alice_array.append("Alice_item_2")
    
    # Bob also adds items (concurrently)
    bob_array.append("Bob_item_1") 
    bob_array.append("Bob_item_2")
    
    print(f"Alice's array: {list(alice_array)}")
    print(f"Bob's array: {list(bob_array)}")
    
    # Sync Alice's changes to Bob
    alice_update = doc_alice.get_update(doc_bob.get_state())
    doc_bob.apply_update(alice_update)
    
    # Sync Bob's changes to Alice
    bob_update = doc_bob.get_update(doc_alice.get_state())
    doc_alice.apply_update(bob_update)
    
    print(f"\nAfter synchronization:")
    print(f"Alice's array: {list(alice_array)}")
    print(f"Bob's array: {list(bob_array)}")
    print(f"Arrays match: {list(alice_array) == list(bob_array)}")
    
    # Test with Maps and concurrent updates
    alice_map = doc_alice.get("shared_map", type=pycrdt.Map)
    bob_map = doc_bob.get("shared_map", type=pycrdt.Map)
    
    # Concurrent map updates
    alice_map["author"] = "Alice"
    alice_map["timestamp"] = "2024-01-01"
    
    bob_map["editor"] = "Bob"
    bob_map["timestamp"] = "2024-01-02"  # Conflicting key
    
    print(f"\nBefore map sync:")
    print(f"Alice's map: {dict(alice_map)}")
    print(f"Bob's map: {dict(bob_map)}")
    
    # Sync map changes
    alice_map_update = doc_alice.get_update(doc_bob.get_state())
    doc_bob.apply_update(alice_map_update)
    
    bob_map_update = doc_bob.get_update(doc_alice.get_state())
    doc_alice.apply_update(bob_map_update)
    
    print(f"\nAfter map sync:")
    print(f"Alice's map: {dict(alice_map)}")
    print(f"Bob's map: {dict(bob_map)}")
    print(f"Maps match: {dict(alice_map) == dict(bob_map)}")
    
    # Show conflict resolution (last-write-wins for maps)
    print(f"Timestamp value (last-write-wins): {alice_map.get('timestamp')}")
    
    print("Conflict resolution and concurrent operations completed")
    print("pycrdt example 2 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"pycrdt example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: Text Operations and String Manipulation

```python
import pycrdt

# Example 3: Text Operations and String Manipulation
print("Example 3 - Text Operations and String Manipulation")

# Create document with text
doc = pycrdt.Doc()
text = doc.get("content", type=pycrdt.Text)

# Add text content
text += "Hello CRDT World!"
print(f"Initial text: '{str(text)}'")

# Insert text at position
text.insert(6, "Collaborative ")
print(f"After insert: '{str(text)}'")

# Get text length
print(f"Text length: {len(text)}")

# Create another doc and sync
doc2 = pycrdt.Doc()
doc2.apply_update(doc.get_update())
text2 = doc2.get("content", type=pycrdt.Text)

print(f"Synced text: '{str(text2)}'")
print(f"Texts match: {str(text) == str(text2)}")

print("\nExample 3 completed successfully")
```

## Example 4: Document State and Updates

```python
import pycrdt

# Example 4: Document State and Updates
print("Example 4 - Document State and Updates")

# Create two documents
doc1 = pycrdt.Doc()
doc2 = pycrdt.Doc()

# Add data to doc1
array1 = doc1.get("items", type=pycrdt.Array)
array1.extend(["item1", "item2", "item3"])

# Get state vector before sync
state_before = doc2.get_state()

# Sync to doc2
update = doc1.get_update()
doc2.apply_update(update)

array2 = doc2.get("items", type=pycrdt.Array)
print(f"Doc2 after sync: {list(array2)}")

# Add more items to doc1
array1.append("item4")

# Get incremental update only
incremental = doc1.get_update(state_before)
print(f"Full update: {len(update)} bytes")
print(f"Incremental: {len(incremental)} bytes")

print("\nExample 4 completed successfully")
```

## Example 5: Multi-User Collaboration Simulation

```python
import pycrdt

# Example 5: Multi-User Collaboration Simulation
print("Example 5 - Multi-User Collaboration Simulation")

# Create documents for 3 users
docs = {
    "alice": pycrdt.Doc(),
    "bob": pycrdt.Doc(),
    "charlie": pycrdt.Doc()
}

# Alice creates initial content
alice_text = docs["alice"].get("notes", type=pycrdt.Text)
alice_text += "Project Notes:\n"

# Broadcast Alice's changes
alice_update = docs["alice"].get_update()
docs["bob"].apply_update(alice_update)
docs["charlie"].apply_update(alice_update)

# Bob and Charlie add content
bob_text = docs["bob"].get("notes", type=pycrdt.Text)
bob_text += "- Task 1: Setup\n"

charlie_text = docs["charlie"].get("notes", type=pycrdt.Text)
charlie_text += "- Task 2: Testing\n"

# Final sync
docs["alice"].apply_update(docs["bob"].get_update(docs["alice"].get_state()))
docs["alice"].apply_update(docs["charlie"].get_update(docs["alice"].get_state()))

final_text = docs["alice"].get("notes", type=pycrdt.Text)
print(f"Final collaborative document:\n{str(final_text)}")

print("\nExample 5 completed successfully")
```
