# pydantic: 5 Basic Usage Examples

```python
# Example 1: Basic model
from pydantic import BaseModel
class User(BaseModel):
    id: int
    name: str
user = User(id=1, name='Alice')
print(user)
assert user.id == 1 and user.name == 'Alice', user
assert user.model_dump() == {'id': 1, 'name': 'Alice'}, user.model_dump()
# Lax mode coerces a numeric string to the declared int
assert User(id='7', name='Carol').id == 7
```

```python
# Example 2: Validation error
from pydantic import BaseModel, ValidationError
class User(BaseModel):
    id: int
    name: str
try:
    User(id='not-an-int', name='Bob')
except ValidationError as e:
    print(e)
    assert e.error_count() == 1, e.errors()
    assert e.errors()[0]['loc'] == ('id',) and e.errors()[0]['type'] == 'int_parsing', e.errors()
else:
    raise AssertionError("ValidationError not raised for id='not-an-int'")
```

```python
# Example 3: Model with default values
from pydantic import BaseModel
class Product(BaseModel):
    name: str
    price: float = 9.99
product = Product(name='Widget')
print(product)
assert product.name == 'Widget' and product.price == 9.99, product
assert Product(name='Gadget', price='1.5').price == 1.5
```

```python
# Example 4: Nested models
from pydantic import BaseModel
class Address(BaseModel):
    city: str
    zip: str
class User(BaseModel):
    id: int
    address: Address
user = User(id=2, address=Address(city='NY', zip='10001'))
print(user)
assert user.id == 2 and user.address.city == 'NY' and user.address.zip == '10001', user
# A plain dict is validated into the nested model
from_dict = User(id=3, address={'city': 'SF', 'zip': '94016'})
assert isinstance(from_dict.address, Address) and from_dict.address.city == 'SF', from_dict
assert from_dict.model_dump() == {'id': 3, 'address': {'city': 'SF', 'zip': '94016'}}, from_dict.model_dump()
```
