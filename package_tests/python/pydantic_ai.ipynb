## Example 1: Basic Agent Creation

```python
import pydantic_ai
from pydantic_ai import Agent

# Example 1: Basic Agent Creation
print("Example 1 - Basic Agent Creation")

# Create a simple agent
agent = Agent(
    model='test',
    system_prompt='You are a helpful assistant.'
)

print(f"Agent created: {type(agent).__name__}")
print(f"Model: {agent.model}")

# Check available components
components = [x for x in dir(pydantic_ai) if not x.startswith('_')]
print(f"\nAvailable components: {len(components)}")
print(f"Sample: {components[:5]}")

# Show agent attributes
attrs = [a for a in dir(agent) if not a.startswith('_')]
print(f"\nAgent attributes: {len(attrs)}")
print(f"Sample methods: {attrs[:5]}")

print("\nExample 1 completed successfully")
```

## Example 2: Structured Output Models

```python
import pydantic_ai
from pydantic_ai import Agent
from pydantic import BaseModel

# Example 2: Structured Outputs with Pydantic
print("Example 2 - Structured Outputs with Pydantic")

# Define output structure
class TaskOutput(BaseModel):
    title: str
    priority: str
    estimated_minutes: int

# Create agent with structured output
agent = Agent(
    model='test',
    system_prompt='Create task summaries.',
    output_type=TaskOutput
)

print(f"Agent with output type: {agent.output_type.__name__}")

# Validate structure
sample = TaskOutput(title="Fix bug", priority="high", estimated_minutes=30)
print(f"\nSample output validated:")
print(f"  Title: {sample.title}")
print(f"  Priority: {sample.priority}")
print(f"  Duration: {sample.estimated_minutes}min")

print("\nExample 2 completed successfully")
```

## Example 3: Dependency Injection and Tools

```python
import pydantic_ai
from pydantic_ai import Agent, RunContext

# Example 3: Dependency Injection
print("Example 3 - Dependency Injection")

# Define context dependency
class DatabaseContext:
    def __init__(self, db_name: str):
        self.db_name = db_name
        self.queries = []
    
    def query(self, sql: str) -> dict:
        self.queries.append(sql)
        return {"status": "success", "rows": 5}

# Create agent with dependency
agent = Agent(
    model='test',
    system_prompt='You are a database assistant.',
    deps_type=DatabaseContext
)

print(f"Agent with deps_type: {agent.deps_type.__name__}")

# Test dependency
db = DatabaseContext("users_db")
result = db.query("SELECT * FROM users")
print(f"Query result: {result}")
print(f"Total queries: {len(db.queries)}")

print("\nExample 3 completed successfully")
```

## Example 4: Advanced Validation and Error Handling

```python
import pydantic_ai
from pydantic_ai import Agent
from pydantic import BaseModel, ValidationError

# Example 4: Validation and Error Handling
print("Example 4 - Validation and Error Handling")

# Define model with constraints
class ValidatedOutput(BaseModel):
    score: int
    feedback: str
    
    class Config:
        validate_assignment = True

# Test validation
valid = ValidatedOutput(score=85, feedback="Good work")  # valid input: a ValidationError here fails the cell
assert (valid.score, valid.feedback) == (85, "Good work"), valid
print(f"Valid output: score={valid.score}, feedback={valid.feedback}")

# Test invalid data
try:
    invalid = ValidatedOutput(score="not_a_number", feedback="Test")
except ValidationError as e:
    assert [(err["loc"], err["type"]) for err in e.errors()] == [(("score",), "int_parsing")], e.errors()
    print("Invalid score correctly rejected ✓")
else:
    raise AssertionError(f"ValidatedOutput accepted score='not_a_number': {invalid!r}")

print("\nExample 4 completed successfully")
```

## Example 5: Integration Patterns and Best Practices

```python
import pydantic_ai
from pydantic_ai import Agent
from pydantic import BaseModel

# Example 5: Multiple Agents with Different Output Types
print("Example 5 - Multiple Agents with Different Output Types")

# Define different output types
class SummaryOutput(BaseModel):
    summary: str
    word_count: int

class AnalysisOutput(BaseModel):
    sentiment: str
    confidence: float

# Create specialized agents
agents = {
    'summarizer': Agent(model='test', system_prompt='Summarize text', output_type=SummaryOutput),
    'analyzer': Agent(model='test', system_prompt='Analyze sentiment', output_type=AnalysisOutput)
}

print(f"Created {len(agents)} specialized agents:")
for name, agent in agents.items():
    output_type_name = agent.output_type.__name__ if hasattr(agent.output_type, '__name__') else str(agent.output_type)
    print(f"  {name}: {output_type_name}")

# Show model info
print(f"\nAgent configurations:")
for name in agents:
    print(f"  {name}: model={agents[name].model}")

print("\nExample 5 completed successfully")
```
