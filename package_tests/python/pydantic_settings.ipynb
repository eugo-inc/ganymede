# pydantic_settings: 5 Basic Usage Examples

```python
# Example 1: Basic Settings class
from pydantic_settings import BaseSettings
class Settings(BaseSettings):
    foo: str = 'bar'
settings = Settings()
print(settings.foo)
```

```python
# Example 2: Load from environment variable
import os
os.environ['FOO'] = 'baz'
from pydantic_settings import BaseSettings
class Settings(BaseSettings):
    foo: str
settings = Settings()
print(settings.foo)
```

```python
# Example 3: Load from .env file
with open('.env', 'w') as f:
    f.write('FOO=from_env_file\n')
from pydantic_settings import BaseSettings
class Settings(BaseSettings):
    foo: str
settings = Settings(_env_file='.env')
print(settings.foo)
```

```python
# Example 4: Clean up .env file
import os
os.remove('.env')
print('Removed .env')
```

```python
# Example 5: Show pydantic_settings version
import pydantic_settings
print(pydantic_settings.__version__)
```
