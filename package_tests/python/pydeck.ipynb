## Example 1: pydeck functionality test 1

```python
import pydeck as pdk
import pandas as pd
import numpy as np

# Example 1: Basic pydeck scatterplot layer
print("Creating basic pydeck visualization:")

try:
    # Generate sample geospatial data
    np.random.seed(42)
    data = pd.DataFrame({
        'latitude': np.random.uniform(37.7, 37.8, 50),
        'longitude': np.random.uniform(-122.5, -122.4, 50),
        'size': np.random.uniform(20, 100, 50)
    })
    
    # Create scatterplot layer
    layer = pdk.Layer(
        'ScatterplotLayer',
        data=data,
        get_position='[longitude, latitude]',
        get_color='[200, 30, 0, 160]',
        get_radius='size',
    )
    
    # Create view state
    view_state = pdk.ViewState(
        latitude=37.75,
        longitude=-122.45,
        zoom=11,
        pitch=0
    )
    
    # Create deck
    deck = pdk.Deck(
        layers=[layer],
        initial_view_state=view_state,
        tooltip={"text": "Latitude: {latitude}\nLongitude: {longitude}"}
    )
    
    print(f"✓ Data points: {len(data)}")
    print(f"✓ Layer type: {layer.type}")
    print(f"✓ View state: lat={view_state.latitude}, lon={view_state.longitude}")
    print(f"✓ Deck created with {len(deck.layers)} layer(s)")
    
    print("Example 1 completed successfully")
    
except Exception as e:
    print(f"Error in example 1: {e}")
    raise AssertionError(
        f"pydeck example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 2: pydeck functionality test 2

```python
import pydeck as pdk
import pandas as pd
import numpy as np

# Example 2: HexagonLayer for aggregated visualization
print("Creating hexagon layer visualization:")

try:
    # Generate denser sample data for aggregation
    np.random.seed(123)
    n_points = 200
    
    # Create clustered data points
    cluster_centers = [(37.76, -122.43), (37.78, -122.41), (37.74, -122.47)]
    data_points = []
    
    for center_lat, center_lon in cluster_centers:
        cluster_size = n_points // len(cluster_centers)
        lats = np.random.normal(center_lat, 0.005, cluster_size)
        lons = np.random.normal(center_lon, 0.005, cluster_size)
        
        for lat, lon in zip(lats, lons):
            data_points.append({
                'latitude': lat,
                'longitude': lon,
                'value': np.random.exponential(100)
            })
    
    data = pd.DataFrame(data_points)
    
    # Create hexagon layer
    hexagon_layer = pdk.Layer(
        'HexagonLayer',
        data=data,
        get_position='[longitude, latitude]',
        get_weight='value',
        radius=200,
        elevation_scale=4,
        elevation_range=[0, 1000],
        pickable=True,
        extruded=True,
    )
    
    # Create 3D view state
    view_state = pdk.ViewState(
        latitude=37.76,
        longitude=-122.44,
        zoom=12,
        pitch=45,
        bearing=0
    )
    
    # Create deck with hexagon layer
    deck = pdk.Deck(
        layers=[hexagon_layer],
        initial_view_state=view_state,
        tooltip={"text": "Count: {count}\nValue: {value}"}
    )
    
    print(f"✓ Data points: {len(data)}")
    print(f"✓ Clusters: {len(cluster_centers)}")
    print(f"✓ Layer type: {hexagon_layer.type}")
    print(f"✓ 3D view: pitch={view_state.pitch}°, extruded={hexagon_layer.extruded}")
    
    print("Example 2 completed successfully")
    
except Exception as e:
    print(f"Error in example 2: {e}")
    raise AssertionError(
        f"pydeck example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: pydeck functionality test 3

```python
import pydeck as pdk
import pandas as pd
import numpy as np

# Example 3: PathLayer for route visualization
print("Creating path layer visualization:")

try:
    # Generate sample route data
    np.random.seed(456)
    
    # Create a path from San Francisco to Oakland
    path_coords = []
    start_lat, start_lon = 37.7749, -122.4194  # SF
    end_lat, end_lon = 37.8044, -122.2712      # Oakland
    
    # Generate intermediate points along the path
    n_segments = 20
    for i in range(n_segments + 1):
        t = i / n_segments
        lat = start_lat + t * (end_lat - start_lat) + np.random.normal(0, 0.001)
        lon = start_lon + t * (end_lon - start_lon) + np.random.normal(0, 0.001)
        path_coords.append([lon, lat])
    
    # Create multiple paths with different properties
    paths_data = [
        {
            "name": "Route 1",
            "path": path_coords,
            "color": [255, 0, 0, 200],
            "width": 8
        },
        {
            "name": "Route 2", 
            "path": [[coord[0] + 0.01, coord[1] + 0.005] for coord in path_coords],
            "color": [0, 255, 0, 200],
            "width": 5
        },
        {
            "name": "Route 3",
            "path": [[coord[0] - 0.01, coord[1] - 0.005] for coord in path_coords], 
            "color": [0, 0, 255, 200],
            "width": 3
        }
    ]
    
    data = pd.DataFrame(paths_data)
    
    # Create path layer
    path_layer = pdk.Layer(
        "PathLayer",
        data=data,
        pickable=True,
        get_color="color",
        width_scale=1,
        get_path="path",
        get_width="width"
    )
    
    # Create view state
    view_state = pdk.ViewState(
        latitude=37.79,
        longitude=-122.34,
        zoom=10,
        pitch=30,
        bearing=-20
    )
    
    # Create deck
    deck = pdk.Deck(
        layers=[path_layer],
        initial_view_state=view_state,
        tooltip={"text": "Route: {name}\nWidth: {width}"}
    )
    
    print(f"✓ Number of routes: {len(data)}")
    print(f"✓ Segments per route: {len(path_coords)}")
    print(f"✓ Layer type: {path_layer.type}")
    print(f"✓ View bearing: {view_state.bearing}°")
    
    print("Example 3 completed successfully")
    
except Exception as e:
    print(f"Error in example 3: {e}")
    raise AssertionError(
        f"pydeck example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: pydeck functionality test 4

```python
import pydeck as pdk
import pandas as pd
import numpy as np

# Example 4: Multiple Layers with ColumnLayer
print("Creating multi-layer visualization:")

try:
    # Generate building data
    np.random.seed(789)
    n_buildings = 100
    
    buildings_data = []
    for i in range(n_buildings):
        lat = np.random.uniform(37.77, 37.80)
        lon = np.random.uniform(-122.44, -122.40)
        height = np.random.exponential(200)
        population = np.random.poisson(50)
        
        buildings_data.append({
            'latitude': lat,
            'longitude': lon,
            'height': height,
            'population': population
        })
    
    buildings_df = pd.DataFrame(buildings_data)
    
    # Create column layer
    column_layer = pdk.Layer(
        'ColumnLayer',
        data=buildings_df,
        get_position='[longitude, latitude]',
        get_elevation='height',
        get_fill_color='[200, 100, 255, 200]',
        radius=50,
        pickable=True
    )
    
    # Create view state
    view_state = pdk.ViewState(
        latitude=37.785,
        longitude=-122.42,
        zoom=13,
        pitch=60,
        bearing=45
    )
    
    # Create deck
    deck = pdk.Deck(
        layers=[column_layer],
        initial_view_state=view_state,
        tooltip={"text": "Height: {height:.0f}m\nPopulation: {population}"}
    )
    
    print(f"✓ Buildings: {len(buildings_df)}")
    print(f"✓ Max height: {buildings_df['height'].max():.1f}m")
    print(f"✓ View: pitch={view_state.pitch}°, bearing={view_state.bearing}°")
    
    print("Example 4 completed successfully")
    
except Exception as e:
    print(f"Error in example 4: {e}")
    raise AssertionError(
        f"pydeck example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 5: pydeck functionality test 5

```python
import pydeck as pdk
import pandas as pd
import numpy as np

# Example 5: Advanced visualization with custom map styles and data utilities
print("Creating advanced pydeck visualization:")

try:
    # Generate temporal-spatial data
    np.random.seed(999)
    n_points = 150
    
    # Create data with timestamps
    time_data = []
    base_time = 1640995200  # Jan 1, 2022
    
    for i in range(n_points):
        # Create data points moving over time
        t = i / n_points
        lat = 37.76 + 0.02 * np.sin(t * 4 * np.pi) + np.random.normal(0, 0.002)
        lon = -122.44 + 0.03 * np.cos(t * 3 * np.pi) + np.random.normal(0, 0.002)
        timestamp = base_time + i * 3600  # Hourly data
        
        time_data.append({
            'latitude': lat,
            'longitude': lon,
            'timestamp': timestamp,
            'speed': np.random.exponential(30),
            'direction': np.random.uniform(0, 360),
            'intensity': np.random.gamma(2, 50)
        })
    
    df = pd.DataFrame(time_data)
    
    # Create arc layer to show movement
    arc_data = []
    for i in range(len(df) - 1):
        arc_data.append({
            'source_lat': df.iloc[i]['latitude'],
            'source_lon': df.iloc[i]['longitude'],
            'target_lat': df.iloc[i+1]['latitude'],
            'target_lon': df.iloc[i+1]['longitude'],
            'intensity': df.iloc[i]['intensity']
        })
    
    arc_df = pd.DataFrame(arc_data)
    
    # Create arc layer
    arc_layer = pdk.Layer(
        'ArcLayer',
        data=arc_df,
        get_source_position='[source_lon, source_lat]',
        get_target_position='[target_lon, target_lat]',
        get_source_color='[255, 100, 0, 160]',
        get_target_color='[0, 100, 255, 160]',
        get_width='intensity / 20',
        width_scale=1,
        pickable=True,
    )
    
    # Create heatmap layer
    heatmap_layer = pdk.Layer(
        'HeatmapLayer',
        data=df,
        get_position='[longitude, latitude]',
        get_weight='speed',
        radius_pixels=60,
        intensity=1,
        threshold=0.03,
    )
    
    # Use custom map style
    map_style = pdk.map_styles.DARK
    
    # Create view state with animation-ready settings
    view_state = pdk.ViewState(
        latitude=37.76,
        longitude=-122.44,
        zoom=12,
        pitch=45,
        bearing=30
    )
    
    # Create deck with advanced features
    deck = pdk.Deck(
        layers=[heatmap_layer, arc_layer],
        initial_view_state=view_state,
        map_style=map_style,
        tooltip={
            "text": "Speed: {speed:.1f}\nIntensity: {intensity:.0f}\nDirection: {direction:.0f}°"
        }
    )
    
    # Test data utilities
    print(f"✓ Temporal data points: {len(df)}")
    print(f"✓ Movement arcs: {len(arc_df)}")
    print(f"✓ Time range: {(df['timestamp'].max() - df['timestamp'].min()) / 3600:.1f} hours")
    print(f"✓ Map style: {map_style}")
    print(f"✓ Layer types: {[layer.type for layer in deck.layers]}")
    
    # Test pydeck data utilities
    from pydeck import data_utils
    bounds = data_utils.compute_view(df[['longitude', 'latitude']])
    print(f"✓ Data bounds computed: {type(bounds).__name__}")
    
    print("Example 5 completed successfully")
    
except Exception as e:
    print(f"Error in example 5: {e}")
    raise AssertionError(
        f"pydeck example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
```
