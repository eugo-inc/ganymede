```python
# Example 1: Top-level public API surface inspection.
import docket

public_names = [n for n in dir(docket) if not n.startswith('_')]
print(f"docket exposes {len(public_names)} public names")
for name in public_names[:30]:
    obj = getattr(docket, name)
    print(f"  {name}: {type(obj).__name__}")

assert len(public_names) > 0, "docket should expose at least one public name"
```

```python
# Example 2: Verify the Docket class (the primary user-facing entry point) is importable.
from docket import Docket

print(f"Docket class: {Docket}")
print(f"Docket.__module__: {Docket.__module__}")
print(f"Docket.__qualname__: {Docket.__qualname__}")

# Inspect callable methods on Docket
methods = [m for m in dir(Docket) if not m.startswith('_') and callable(getattr(Docket, m, None))]
print(f"Docket has {len(methods)} public methods/attributes")
for m in methods[:15]:
    print(f"  - {m}")

assert isinstance(Docket, type), "Docket should be a class"
```

```python
# Example 3: Verify the Worker class is importable + inspect.
from docket import Worker

print(f"Worker class: {Worker}")
print(f"Worker.__module__: {Worker.__module__}")

methods = [m for m in dir(Worker) if not m.startswith('_')]
print(f"Worker has {len(methods)} public attrs")
for m in methods[:15]:
    print(f"  - {m}")

assert isinstance(Worker, type), "Worker should be a class"
```

```python
# Example 4: Instantiate a Docket without connecting (constructor-only smoke test).
import inspect

from docket import Docket

sig = inspect.signature(Docket.__init__)
print(f"Docket.__init__ signature: {sig}")
for pname, param in sig.parameters.items():
    if pname == 'self':
        continue
    print(f"  {pname}: default={param.default!r}, kind={param.kind.name}")

# Just construct — do not connect to Redis.
d = Docket(name="smoketest", url="redis://localhost:6379/0")
print(f"\nConstructed Docket: name={d.name!r}, url={d.url!r}")
assert d.name == "smoketest"
```

```python
# Example 5: Verify Typer CLI app is wired up.
from docket import cli as docket_cli

print(f"docket.cli module: {docket_cli}")
cli_names = [n for n in dir(docket_cli) if not n.startswith('_')]
print(f"docket.cli exposes {len(cli_names)} names")
for n in cli_names[:20]:
    obj = getattr(docket_cli, n)
    print(f"  {n}: {type(obj).__name__}")

# Typer apps are typically named `app` by convention
if hasattr(docket_cli, 'app'):
    print(f"\ndocket.cli.app type: {type(docket_cli.app).__name__}")
    assert docket_cli.app is not None
else:
    assert len(cli_names) > 0, "docket.cli should expose at least one name"
```
