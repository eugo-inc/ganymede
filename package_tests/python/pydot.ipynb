## Example 1: pydot functionality test 1

```python
import pydot

# Example 1: Basic graph creation with pydot
print("Creating basic directed graph:")

try:
    # Create a directed graph
    graph = pydot.Dot(graph_type='digraph', rankdir='TB')
    graph.set_name('example_graph')
    
    # Add nodes with attributes
    nodes = [
        ("Start", {"style": "filled", "fillcolor": "lightgreen", "shape": "circle"}),
        ("Process", {"style": "filled", "fillcolor": "lightblue", "shape": "box"}),
        ("Decision", {"style": "filled", "fillcolor": "yellow", "shape": "diamond"}),
        ("End", {"style": "filled", "fillcolor": "lightcoral", "shape": "circle"})
    ]
    
    for name, attrs in nodes:
        node = pydot.Node(name, **attrs)
        graph.add_node(node)
    
    # Add edges with labels
    edges = [
        ("Start", "Process", "begin"),
        ("Process", "Decision", "evaluate"),
        ("Decision", "End", "complete")
    ]
    
    for src, dst, label in edges:
        edge = pydot.Edge(src, dst, label=label)
        graph.add_edge(edge)
    
    print(f"✓ Graph name: {graph.get_name()}")
    print(f"✓ Graph type: {graph.get_graph_type()}")
    print(f"✓ Nodes: {len(graph.get_nodes())}")
    print(f"✓ Edges: {len(graph.get_edges())}")
    print(f"✓ Direction: {graph.get('rankdir')}")
    
    # Generate DOT representation
    dot_string = graph.to_string()
    print(f"✓ DOT string generated: {len(dot_string)} characters")
    
    print("Example 1 completed successfully")
    
except Exception as e:
    print(f"Error in example 1: {e}")
    raise AssertionError(
        f"pydot example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 2: pydot functionality test 2

```python
import pydot

# Example 2: Subgraph and clustering functionality
print("Creating graph with subgraphs and clusters:")

try:
    # Create main graph
    main_graph = pydot.Dot(graph_type='digraph', rankdir='LR')
    main_graph.set_name('clustered_graph')
    
    # Create cluster subgraph
    cluster = pydot.Cluster('cluster_processing', label='Processing Module')
    cluster.set('style', 'filled')
    cluster.set('color', 'lightgray')
    
    # Add nodes to cluster
    cluster_nodes = ['Input', 'Transform', 'Validate', 'Output']
    for node_name in cluster_nodes:
        node = pydot.Node(node_name, 
                         style='filled', 
                         fillcolor='white',
                         shape='box')
        cluster.add_node(node)
    
    # Add edges within cluster
    cluster_edges = [
        ('Input', 'Transform'),
        ('Transform', 'Validate'),
        ('Validate', 'Output')
    ]
    
    for src, dst in cluster_edges:
        edge = pydot.Edge(src, dst, color='blue')
        cluster.add_edge(edge)
    
    # Add cluster to main graph
    main_graph.add_subgraph(cluster)
    
    # Add external nodes
    external_nodes = [
        ('Source', {'fillcolor': 'lightgreen'}),
        ('Sink', {'fillcolor': 'lightcoral'})
    ]
    
    for name, attrs in external_nodes:
        node = pydot.Node(name, style='filled', shape='ellipse', **attrs)
        main_graph.add_node(node)
    
    # Add external connections
    external_edges = [
        ('Source', 'Input'),
        ('Output', 'Sink')
    ]
    
    for src, dst in external_edges:
        edge = pydot.Edge(src, dst, style='dashed', color='red')
        main_graph.add_edge(edge)
    
    print(f"✓ Main graph nodes: {len(main_graph.get_nodes())}")
    print(f"✓ Main graph edges: {len(main_graph.get_edges())}")
    print(f"✓ Subgraphs: {len(main_graph.get_subgraphs())}")
    print(f"✓ Cluster nodes: {len(cluster.get_nodes())}")
    print(f"✓ Cluster edges: {len(cluster.get_edges())}")
    print(f"✓ Cluster label: {cluster.get('label')}")
    
    print("Example 2 completed successfully")
    
except Exception as e:
    print(f"Error in example 2: {e}")
    raise AssertionError(
        f"pydot example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: pydot functionality test 3

```python
import pydot

# Example 3: Graph creation from adjacency matrix and edges
print("Creating graphs from data structures:")

try:
    # Test graph_from_edges
    edge_list = [
        ('A', 'B', {'weight': 5, 'color': 'red'}),
        ('B', 'C', {'weight': 3, 'color': 'blue'}),
        ('C', 'D', {'weight': 2, 'color': 'green'}),
        ('A', 'D', {'weight': 7, 'color': 'purple'})
    ]
    
    # Create graph from edge list
    graph_from_edges = pydot.graph_from_edges(edge_list, directed=True)
    graph_from_edges.set_name('edge_graph')
    
    print(f"✓ Graph from edges:")
    print(f"  Nodes: {len(graph_from_edges.get_nodes())}")
    print(f"  Edges: {len(graph_from_edges.get_edges())}")
    
    # Test graph_from_adjacency_matrix
    adjacency_matrix = [
        [0, 1, 0, 1],  # A connects to B, D
        [0, 0, 1, 0],  # B connects to C  
        [0, 0, 0, 1],  # C connects to D
        [0, 0, 0, 0]   # D connects to nothing
    ]
    
    node_names = ['A', 'B', 'C', 'D']
    
    adj_graph = pydot.graph_from_adjacency_matrix(adjacency_matrix, node_names, directed=True)
    adj_graph.set_name('adjacency_graph')
    
    print(f"✓ Graph from adjacency matrix:")
    print(f"  Nodes: {len(adj_graph.get_nodes())}")
    print(f"  Edges: {len(adj_graph.get_edges())}")
    
    # Test graph_from_incidence_matrix
    # Incidence matrix: rows=nodes, cols=edges
    incidence_matrix = [
        [1, 0, 0, 1],   # A: incident to edge 0 and 3
        [-1, 1, 0, 0],  # B: incident to edge 0 (out) and 1 (in)
        [0, -1, 1, 0],  # C: incident to edge 1 (out) and 2 (in)
        [0, 0, -1, -1]  # D: incident to edge 2 (out) and 3 (out)
    ]
    
    inc_graph = pydot.graph_from_incidence_matrix(incidence_matrix, node_names, directed=True)
    inc_graph.set_name('incidence_graph')
    
    print(f"✓ Graph from incidence matrix:")
    print(f"  Nodes: {len(inc_graph.get_nodes())}")  
    print(f"  Edges: {len(inc_graph.get_edges())}")
    
    # Verify graph structures
    for graph, name in [(graph_from_edges, 'edges'), (adj_graph, 'adjacency'), (inc_graph, 'incidence')]:
        dot_str = graph.to_string()
        print(f"✓ {name.capitalize()} graph DOT: {len(dot_str)} chars")
    
    print("Example 3 completed successfully")
    
except Exception as e:
    print(f"Error in example 3: {e}")
    raise AssertionError(
        f"pydot example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: pydot functionality test 4

```python
import pydot
import tempfile
import os

# Example 4: DOT parsing and graph I/O operations
print("Testing DOT parsing and file I/O:")

try:
    # Create a DOT string representation
    dot_string = """
    digraph test_graph {
        rankdir=TB;
        node [style=filled];
        
        Start [fillcolor=lightgreen, shape=circle];
        Process1 [fillcolor=lightblue, label="Process\\nStep 1"];
        Process2 [fillcolor=lightblue, label="Process\\nStep 2"];
        Decision [fillcolor=yellow, shape=diamond];
        End [fillcolor=lightcoral, shape=circle];
        
        Start -> Process1 [label="begin"];
        Process1 -> Process2 [label="continue", style=bold];
        Process2 -> Decision [label="evaluate"];
        Decision -> End [label="complete", color=green];
        Decision -> Process1 [label="retry", color=red, style=dashed];
    }
    """
    
    # Parse DOT string to create graph
    graphs = pydot.graph_from_dot_data(dot_string)
    parsed_graph = graphs[0]  # graph_from_dot_data returns a list
    
    print(f"✓ Parsed graph from DOT string:")
    print(f"  Graph name: {parsed_graph.get_name()}")
    print(f"  Nodes: {len(parsed_graph.get_nodes())}")
    print(f"  Edges: {len(parsed_graph.get_edges())}")
    print(f"  Rank direction: {parsed_graph.get('rankdir')}")
    
    # Test with temporary file I/O
    with tempfile.TemporaryDirectory() as temp_dir:
        dot_file_path = os.path.join(temp_dir, "test_graph.dot")
        
        # Write DOT to file
        with open(dot_file_path, 'w') as f:
            f.write(dot_string)
        
        # Read graph from DOT file
        file_graphs = pydot.graph_from_dot_file(dot_file_path)
        file_graph = file_graphs[0]
        
        print(f"✓ Read graph from DOT file:")
        print(f"  File exists: {os.path.exists(dot_file_path)}")
        print(f"  File size: {os.path.getsize(dot_file_path)} bytes")
        print(f"  Parsed nodes: {len(file_graph.get_nodes())}")
        print(f"  Parsed edges: {len(file_graph.get_edges())}")
        
        # Verify graph structure is preserved
        original_dot = parsed_graph.to_string()
        file_dot = file_graph.to_string()
        
        print(f"✓ DOT string comparison:")
        print(f"  Original length: {len(original_dot)}")
        print(f"  File parsed length: {len(file_dot)}")
        print(f"  Graphs equivalent: {len(original_dot) == len(file_dot)}")
    
    print("Example 4 completed successfully")
    
except Exception as e:
    print(f"Error in example 4: {e}")
    raise AssertionError(
        f"pydot example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 5: pydot functionality test 5

```python
import pydot

# Example 5: Advanced graph attributes and styling options
print("Testing advanced graph styling and attributes:")

try:
    # Create a styled graph with various node shapes and attributes
    styled_graph = pydot.Dot(graph_type='digraph', rankdir='LR')
    styled_graph.set_name('styled_graph')
    
    # Set global graph attributes
    styled_graph.set('bgcolor', 'white')
    styled_graph.set('fontname', 'Arial')
    styled_graph.set('fontsize', '14')
    styled_graph.set('label', 'Advanced Pydot Styling Example')
    styled_graph.set('labelloc', 'top')
    
    # Define various node styles
    node_styles = [
        ('Database', {'shape': 'cylinder', 'fillcolor': 'lightblue', 'style': 'filled'}),
        ('Server', {'shape': 'box3d', 'fillcolor': 'lightgreen', 'style': 'filled'}),
        ('Client', {'shape': 'house', 'fillcolor': 'lightyellow', 'style': 'filled'}),
        ('Network', {'shape': 'cloud', 'fillcolor': 'lightgray', 'style': 'filled'}),
        ('Security', {'shape': 'octagon', 'fillcolor': 'lightcoral', 'style': 'filled'}),
        ('API', {'shape': 'hexagon', 'fillcolor': 'lightpink', 'style': 'filled'})
    ]
    
    # Add styled nodes
    for name, attrs in node_styles:
        node = pydot.Node(name, **attrs)
        styled_graph.add_node(node)
    
    # Define edges with various styles
    edge_styles = [
        ('Client', 'Security', {'style': 'dashed', 'color': 'red', 'label': 'authenticate'}),
        ('Security', 'API', {'style': 'bold', 'color': 'green', 'label': 'authorized'}),
        ('API', 'Server', {'arrowhead': 'diamond', 'color': 'blue', 'label': 'request'}),
        ('Server', 'Database', {'arrowhead': 'open', 'color': 'purple', 'label': 'query'}),
        ('Server', 'Network', {'style': 'dotted', 'color': 'orange', 'label': 'broadcast'}),
        ('Network', 'Client', {'arrowhead': 'crow', 'color': 'brown', 'label': 'response'})
    ]
    
    # Add styled edges
    for src, dst, attrs in edge_styles:
        edge = pydot.Edge(src, dst, **attrs)
        styled_graph.add_edge(edge)
    
    print(f"✓ Styled graph created:")
    print(f"  Graph background: {styled_graph.get('bgcolor')}")
    print(f"  Graph label: {styled_graph.get('label')}")
    print(f"  Nodes with styles: {len(styled_graph.get_nodes())}")
    print(f"  Edges with styles: {len(styled_graph.get_edges())}")
    
    # Test attribute access and modification
    all_nodes = styled_graph.get_nodes()
    all_edges = styled_graph.get_edges()
    
    print(f"✓ Node shapes: {[node.get('shape') for node in all_nodes if node.get('shape')]}")
    print(f"✓ Edge styles: {[edge.get('style') for edge in all_edges if edge.get('style')]}")
    print(f"✓ Edge colors: {[edge.get('color') for edge in all_edges if edge.get('color')]}")
    
    # Test DOT attributes validation
    valid_graph_attrs = [attr for attr in pydot.GRAPH_ATTRIBUTES if attr in styled_graph.get_attributes()]
    valid_node_attrs = []
    valid_edge_attrs = []
    
    for node in all_nodes:
        valid_node_attrs.extend([attr for attr in pydot.NODE_ATTRIBUTES if attr in node.get_attributes()])
    
    for edge in all_edges:
        valid_edge_attrs.extend([attr for attr in pydot.EDGE_ATTRIBUTES if attr in edge.get_attributes()])
    
    print(f"✓ Valid graph attributes used: {len(set(valid_graph_attrs))}")
    print(f"✓ Valid node attributes used: {len(set(valid_node_attrs))}")  
    print(f"✓ Valid edge attributes used: {len(set(valid_edge_attrs))}")
    
    # Generate final DOT representation
    final_dot = styled_graph.to_string()
    print(f"✓ Final DOT string: {len(final_dot)} characters")
    
    print("Example 5 completed successfully")
    
except Exception as e:
    print(f"Error in example 5: {e}")
    raise AssertionError(
        f"pydot example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
```
