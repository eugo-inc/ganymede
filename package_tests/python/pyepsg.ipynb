## Example 2: pyepsg functionality test 2

```python
import pyepsg
import xml.etree.ElementTree as ET

# Example 2: Mock EPSG coordinate reference system creation
print("Testing EPSG CRS object creation:")

try:
    # Since pyepsg requires XML elements, create mock XML structure
    # for demonstration purposes
    
    # Create sample GML XML for WGS84 geographic CRS
    wgs84_xml = f"""
    <gml:GeodeticCRS xmlns:gml="{pyepsg.GML_NS[1:-1]}" gml:id="epsg-crs-4326">
        <gml:name>WGS 84</gml:name>
        <gml:domainOfValidity>World</gml:domainOfValidity>
        <gml:scope>Horizontal component of 3D system</gml:scope>
    </gml:GeodeticCRS>
    """
    
    # Parse XML element
    root = ET.fromstring(wgs84_xml)
    
    # Create GeodeticCRS object with XML element
    geodetic_crs = pyepsg.GeodeticCRS(root)
    
    print(f"✓ Created GeodeticCRS: {type(geodetic_crs).__name__}")
    print(f"✓ CRS ID: {geodetic_crs.id}")
    print(f"✓ CRS Name: {geodetic_crs.name}")
    print(f"✓ Domain of Validity: {geodetic_crs.domain_of_validity}")
    print(f"✓ Scope: {geodetic_crs.scope}")
    
    # Test inheritance hierarchy
    print(f"✓ Is CRS instance: {isinstance(geodetic_crs, pyepsg.CRS)}")
    print(f"✓ Is GeodeticCRS instance: {isinstance(geodetic_crs, pyepsg.GeodeticCRS)}")
    
    print("Example 2 completed successfully")
    
except Exception as e:
    print(f"Error in CRS creation: {e}")
    print("Note: pyepsg requires valid GML XML elements for object creation")
    raise AssertionError(
        f"pyepsg example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: pyepsg functionality test 3

```python
import pyepsg
import xml.etree.ElementTree as ET

# Example 3: Projected coordinate reference system creation
print("Testing ProjectedCRS and coordinate system components:")

try:
    # Create sample GML XML for UTM projected CRS
    utm_xml = f"""
    <gml:ProjectedCRS xmlns:gml="{pyepsg.GML_NS[1:-1]}" gml:id="epsg-crs-32633">
        <gml:name>WGS 84 / UTM zone 33N</gml:name>
        <gml:domainOfValidity>Europe - 12°E to 18°E</gml:domainOfValidity>
        <gml:scope>Large and medium scale topographic mapping</gml:scope>
        <gml:coordinateSystem>
            <gml:CartesianCS gml:id="epsg-cs-4400">
                <gml:name>Cartesian 2D CS</gml:name>
                <gml:axis>
                    <gml:CoordinateSystemAxis gml:id="epsg-axis-1" uom="m">
                        <gml:name>Easting</gml:name>
                        <gml:axisAbbrev>E</gml:axisAbbrev>
                        <gml:axisDirection>east</gml:axisDirection>
                    </gml:CoordinateSystemAxis>
                </gml:axis>
                <gml:axis>
                    <gml:CoordinateSystemAxis gml:id="epsg-axis-2" uom="m">
                        <gml:name>Northing</gml:name>
                        <gml:axisAbbrev>N</gml:axisAbbrev>
                        <gml:axisDirection>north</gml:axisDirection>
                    </gml:CoordinateSystemAxis>
                </gml:axis>
            </gml:CartesianCS>
        </gml:coordinateSystem>
    </gml:ProjectedCRS>
    """
    
    # Parse XML element
    utm_root = ET.fromstring(utm_xml)
    
    # Create ProjectedCRS object
    projected_crs = pyepsg.ProjectedCRS(utm_root)
    
    print(f"✓ Created ProjectedCRS: {type(projected_crs).__name__}")
    print(f"✓ CRS ID: {projected_crs.id}")
    print(f"✓ CRS Name: {projected_crs.name}")
    print(f"✓ Domain: {projected_crs.domain_of_validity}")
    print(f"✓ Scope: {projected_crs.scope}")
    
    # Test coordinate system access
    if hasattr(projected_crs, 'coordinate_system'):
        cs = projected_crs.coordinate_system
        print(f"✓ Has coordinate system: {cs is not None}")
        if cs and hasattr(cs, 'axes'):
            print(f"✓ Number of axes: {len(cs.axes) if cs.axes else 'undefined'}")
    
    # Test class inheritance
    print(f"✓ Is CRS: {isinstance(projected_crs, pyepsg.CRS)}")
    print(f"✓ Is ProjectedCRS: {isinstance(projected_crs, pyepsg.ProjectedCRS)}")
    
    print("Example 3 completed successfully")
    
except Exception as e:
    print(f"Error in ProjectedCRS creation: {e}")
    print("Note: Complex XML structure required for full CRS definition")
    raise AssertionError(
        f"pyepsg example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: pyepsg functionality test 4

```python
import pyepsg
import xml.etree.ElementTree as ET

# Example 4: Axis and Unit of Measure (UOM) components
print("Testing coordinate system axes and units of measure:")

try:
    # Create sample XML for coordinate system axis
    axis_xml = f"""
    <gml:CoordinateSystemAxis xmlns:gml="{pyepsg.GML_NS[1:-1]}" gml:id="epsg-axis-106" uom="9122">
        <gml:name>Geodetic longitude</gml:name>
        <gml:axisAbbrev>Lon</gml:axisAbbrev>
        <gml:axisDirection>east</gml:axisDirection>
        <gml:minimumValue>-180.00</gml:minimumValue>
        <gml:maximumValue>180.00</gml:maximumValue>
    </gml:CoordinateSystemAxis>
    """
    
    # Parse and create Axis object
    axis_root = ET.fromstring(axis_xml)
    longitude_axis = pyepsg.Axis(axis_root)
    
    print(f"✓ Created Axis: {type(longitude_axis).__name__}")
    print(f"✓ Axis object type verified")
    
    # Create sample XML for Unit of Measure
    uom_xml = f"""
    <gml:BaseUnit xmlns:gml="{pyepsg.GML_NS[1:-1]}" gml:id="epsg-uom-9122">
        <gml:name>degree</gml:name>
        <gml:unitsSystem xmlns:xlink="{pyepsg.XLINK_NS[1:-1]}" xlink:href="epsg.xml#SI"/>
    </gml:BaseUnit>
    """
    
    # Parse and create UOM object  
    uom_root = ET.fromstring(uom_xml)
    degree_uom = pyepsg.UOM(uom_root)
    
    print(f"✓ Created UOM: {type(degree_uom).__name__}")
    
    # Test CartesianCS with multiple axes
    cartesian_xml = f"""
    <gml:CartesianCS xmlns:gml="{pyepsg.GML_NS[1:-1]}" gml:id="epsg-cs-4400">
        <gml:identifier codeSpace="OGP">urn:ogc:def:cs:EPSG::4400</gml:identifier>
        <gml:name>Cartesian 2D CS. Axes: easting, northing (E,N). UoM: m</gml:name>
    </gml:CartesianCS>
    """
    
    cartesian_root = ET.fromstring(cartesian_xml)
    cartesian_cs = pyepsg.CartesianCS(cartesian_root)
    
    print(f"✓ Created CartesianCS: {type(cartesian_cs).__name__}")
    # API drift: the attribute is `identifier`, not `id`.
    print(f"✓ CS ID: {cartesian_cs.identifier}")
    print(f"✓ CS Name: {cartesian_cs.name}")
    
    # Verify object types
    print(f"✓ Axis is Axis instance: {isinstance(longitude_axis, pyepsg.Axis)}")
    print(f"✓ UOM is UOM instance: {isinstance(degree_uom, pyepsg.UOM)}")
    print(f"✓ CartesianCS is CartesianCS instance: {isinstance(cartesian_cs, pyepsg.CartesianCS)}")
    
    print("Example 4 completed successfully")
    
except Exception as e:
    print(f"Error in axis/UOM creation: {e}")
    print("Note: Proper GML XML structure required for component creation")
    raise AssertionError(
        f"pyepsg example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 5: pyepsg functionality test 5

```python
import pyepsg
import xml.etree.ElementTree as ET

# Example 5: Compound CRS and pyepsg.get() function simulation
print("Testing CompoundCRS and complete CRS workflow:")

try:
    # Create compound CRS XML (combining horizontal and vertical CRS)
    compound_xml = f"""
    <gml:CompoundCRS xmlns:gml="{pyepsg.GML_NS[1:-1]}" gml:id="epsg-crs-7415">
        <gml:name>Amersfoort / RD New + NAP height</gml:name>
        <gml:domainOfValidity>Netherlands - onshore</gml:domainOfValidity>
        <gml:scope>Engineering survey, topographic mapping</gml:scope>
        <gml:componentReferenceSystem>
            <gml:ProjectedCRS gml:id="epsg-crs-28992">
                <gml:name>Amersfoort / RD New</gml:name>
            </gml:ProjectedCRS>
        </gml:componentReferenceSystem>
        <gml:componentReferenceSystem>
            <gml:VerticalCRS gml:id="epsg-crs-5709">
                <gml:name>NAP height</gml:name>
            </gml:VerticalCRS>
        </gml:componentReferenceSystem>
    </gml:CompoundCRS>
    """
    
    # Parse and create CompoundCRS
    compound_root = ET.fromstring(compound_xml)
    compound_crs = pyepsg.CompoundCRS(compound_root)
    
    print(f"✓ Created CompoundCRS: {type(compound_crs).__name__}")
    print(f"✓ CRS ID: {compound_crs.id}")
    print(f"✓ CRS Name: {compound_crs.name}")
    print(f"✓ Domain: {compound_crs.domain_of_validity}")
    print(f"✓ Scope: {compound_crs.scope}")
    
    # Test component access if available
    if hasattr(compound_crs, 'component_reference_systems'):
        components = compound_crs.component_reference_systems
        print(f"✓ Component CRS count: {len(components) if components else 0}")
    
    # Demonstrate pyepsg workflow patterns
    common_epsg_codes = {
        4326: "WGS 84 (Geographic)",
        3857: "WGS 84 / Pseudo-Mercator", 
        32633: "WGS 84 / UTM zone 33N",
        2154: "RGF93 / Lambert-93",
        4269: "NAD83 (Geographic)"
    }
    
    print(f"\n✓ Common EPSG codes for reference:")
    for code, description in common_epsg_codes.items():
        print(f"  EPSG:{code} - {description}")
    
    # Test get function pattern (would normally fetch from EPSG.io)
    print(f"\n✓ pyepsg.get() function pattern:")
    print(f"  Usage: crs = pyepsg.get(4326)  # Fetches WGS84 from EPSG.io")
    print(f"  API URL: {pyepsg.EPSG_IO_URL}")
    print(f"  Returns: CRS object with full GML definition")
    
    # Test all CRS types inheritance
    print(f"\n✓ CRS type hierarchy validation:")
    print(f"  CompoundCRS → CRS: {issubclass(pyepsg.CompoundCRS, pyepsg.CRS)}")
    print(f"  ProjectedCRS → CRS: {issubclass(pyepsg.ProjectedCRS, pyepsg.CRS)}")
    print(f"  GeodeticCRS → CRS: {issubclass(pyepsg.GeodeticCRS, pyepsg.CRS)}")
    
    print("Example 5 completed successfully")
    
except Exception as e:
    print(f"Error in compound CRS workflow: {e}")
    print("Note: pyepsg is designed to work with live EPSG.io API data")
    raise AssertionError(
        f"pyepsg example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
```
