```python
# Example 1: build a tiny FASTA file and open it with Fasta.
import os
import tempfile
from pyfaidx import Fasta

tmpdir = tempfile.mkdtemp()
fasta_path = os.path.join(tmpdir, 'tiny.fa')
with open(fasta_path, 'w') as f:
    f.write('>chr1\n')
    f.write('ACGTACGTACGTACGTACGT\n')
    f.write('>chr2\n')
    f.write('TTTTAAAACCCCGGGGTTTT\n')

fa = Fasta(fasta_path)
print('Keys:', list(fa.keys()))
print('chr1 length:', len(fa['chr1']))
assert set(fa.keys()) == {'chr1', 'chr2'}
assert len(fa['chr1']) == 20
```

```python
# Example 2: slice a Sequence via 1-based coordinates and check seq content.
import os
import tempfile
from pyfaidx import Fasta

tmpdir = tempfile.mkdtemp()
fasta_path = os.path.join(tmpdir, 'slice.fa')
with open(fasta_path, 'w') as f:
    f.write('>seqA\n')
    f.write('AAAAACCCCCGGGGGTTTTT\n')

fa = Fasta(fasta_path)
# 1-based inclusive: positions 6..10 are CCCCC
region = fa['seqA'][5:10]
print('Slice [5:10] (0-based py slice):', str(region))
print('repr:', repr(region))
assert str(region) == 'CCCCC'
# Negative indexing too
assert str(fa['seqA'][-5:]) == 'TTTTT'
```

```python
# Example 3: reverse complement / complement via the Sequence object.
import os
import tempfile
from pyfaidx import Fasta

tmpdir = tempfile.mkdtemp()
fasta_path = os.path.join(tmpdir, 'rc.fa')
with open(fasta_path, 'w') as f:
    f.write('>seqA\n')
    f.write('ACGT\n')

fa = Fasta(fasta_path)
seq = fa['seqA'][:]
print('forward:', str(seq))
print('complement:', str(seq.complement))
print('reverse complement:', str((-seq)))
assert str(seq) == 'ACGT'
assert str(seq.complement) == 'TGCA'
assert str(-seq) == 'ACGT'  # reverse complement of ACGT is ACGT
```

```python
# Example 4: use get_seq() with 1-based coordinates for a named region.
import os
import tempfile
from pyfaidx import Fasta

tmpdir = tempfile.mkdtemp()
fasta_path = os.path.join(tmpdir, 'getseq.fa')
with open(fasta_path, 'w') as f:
    f.write('>chrX\n')
    f.write('ACGTACGTACGTACGTACGT\n')

fa = Fasta(fasta_path)
# get_seq is 1-based inclusive on both ends
region = fa.get_seq('chrX', 1, 4)
print('chrX:1-4 =', str(region))
assert str(region) == 'ACGT'
region2 = fa.get_seq('chrX', 5, 8)
assert str(region2) == 'ACGT'
```

```python
# Example 5: index file (.fai) is generated and Faidx low-level API works.
import os
import tempfile
from pyfaidx import Faidx

tmpdir = tempfile.mkdtemp()
fasta_path = os.path.join(tmpdir, 'idx.fa')
with open(fasta_path, 'w') as f:
    f.write('>r1\n')
    f.write('ACGTACGTAC\n')
    f.write('>r2\n')
    f.write('TTTTAAAATT\n')

fai = Faidx(fasta_path)
print('index keys:', list(fai.index.keys()))
fai_path = fasta_path + '.fai'
print('.fai exists:', os.path.exists(fai_path))
fetched = fai.fetch('r1', 1, 4)
print('r1:1-4 =', str(fetched))
assert os.path.exists(fai_path)
assert set(fai.index.keys()) == {'r1', 'r2'}
assert str(fetched) == 'ACGT'
print('OK pyfaidx')
```
