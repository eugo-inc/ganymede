## Example 1: pyflakes functionality test 1

```python
import pyflakes.messages as messages
from pyflakes.checker import Checker
import ast

# Example 3: Pyflakes message types and error classification
print("Testing pyflakes message types and error classification:")

try:
    # Code samples designed to trigger specific message types
    code_with_errors = '''
import os
import sys
import unused_module
from collections import defaultdict
from json import loads, dumps

def test_function():
    # Unused variable
    unused_local = 42
    
    # Undefined variable
    result = undefined_var + 10
    
    # Redefined function
    def inner_func():
        return 1
    
    def inner_func():  # Redefinition
        return 2
    
    # Used imports
    data = loads('{"key": "value"}')
    formatted = dumps(data)
    
    return formatted

# Undefined function call
missing_function()
'''
    
    # Analyze the code
    tree = ast.parse(code_with_errors, filename="error_test.py")
    checker = Checker(tree, filename="error_test.py")
    
    print(f"✓ Total messages: {len(checker.messages)}")
    
    # Classify messages by type
    message_types = {}
    for message in checker.messages:
        msg_type = type(message).__name__
        if msg_type not in message_types:
            message_types[msg_type] = []
        message_types[msg_type].append(str(message))
    
    print(f"✓ Message types found: {len(message_types)}")
    for msg_type, msgs in message_types.items():
        print(f"  {msg_type}: {len(msgs)} instances")
        for msg in msgs[:2]:  # Show first 2 examples
            print(f"    - {msg}")
    
    # Test available message classes
    available_messages = [attr for attr in dir(messages) if attr[0].isupper()]
    print(f"✓ Available message types: {len(available_messages)}")
    print(f"  Examples: {available_messages[:5]}")
    
    print("Example 3 completed successfully")
    
except Exception as e:
    print(f"Error in message classification: {e}")
    raise AssertionError(
        f"pyflakes example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 5: pyflakes functionality test 5

```python
import pyflakes.api as api
import tempfile
import os
import subprocess

# Example 5: Production workflow and integration patterns
print("Testing pyflakes production workflow integration:")

try:
    # Create a sample Python project structure
    project_files = {
        'main.py': '''
import sys
import os
from utils import helper_function
from config import settings

def main():
    print("Starting application...")
    result = helper_function(settings.get('data'))
    return result

if __name__ == "__main__":
    sys.exit(main())
''',
        'utils.py': '''
import json
import re
import datetime  # Unused import

def helper_function(data):
    unused_variable = 100  # Unused local variable
    processed = json.loads(data) if data else {}
    return processed

def deprecated_function():  # Unused function
    """This function is no longer used."""
    return "deprecated"
''',
        'config.py': '''
import os
from pathlib import Path

# Configuration settings
settings = {
    'data': '{"key": "value"}',
    'debug': os.getenv('DEBUG', False),
    'path': Path.home() / 'app_data'
}

# Undefined reference will cause error
missing_config = undefined_variable
'''
    }
    
    # Analyze project files
    with tempfile.TemporaryDirectory() as project_dir:
        analysis_results = {}
        
        # Write project files
        for filename, content in project_files.items():
            file_path = os.path.join(project_dir, filename)
            with open(file_path, 'w') as f:
                f.write(content)
        
        # Analyze each file
        total_issues = 0
        for filename in project_files:
            file_path = os.path.join(project_dir, filename)
            
            # Use pyflakes API for analysis
            with open(file_path, 'r') as f:
                content = f.read()
            
            # Count issues by checking manually (api.check prints to stdout)
            import io
            import contextlib
            
            output_capture = io.StringIO()
            with contextlib.redirect_stdout(output_capture):
                api.check(content, filename)
            
            output = output_capture.getvalue()
            file_issues = len(output.strip().split('\n')) if output.strip() else 0
            
            analysis_results[filename] = {
                'lines': len(content.splitlines()),
                'issues': file_issues
            }
            total_issues += file_issues
        
        print(f"✓ Project analysis completed")
        print(f"✓ Files analyzed: {len(project_files)}")
        print(f"✓ Total issues found: {total_issues}")
        
        # Display per-file results
        for filename, results in analysis_results.items():
            print(f"  {filename}: {results['lines']} lines, ~{results['issues']} issues")
        
        # Test recursive checking capability
        print(f"✓ Recursive analysis available: {callable(api.checkRecursive)}")
        
        # Production integration patterns
        integration_patterns = [
            "Pre-commit hooks: pyflakes --version",
            "CI/CD pipelines: pyflakes *.py",
            "IDE integration: real-time checking",
            "Code review: automated quality gates",
            "Build systems: fail on pyflakes errors"
        ]
        
        print(f"✓ Integration patterns:")
        for pattern in integration_patterns:
            print(f"  - {pattern}")
    
    print("Example 5 completed successfully")
    
except Exception as e:
    print(f"Error in workflow integration: {e}")
    raise AssertionError(
        f"pyflakes example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
```
