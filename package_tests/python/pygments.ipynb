# pygments: 5 Basic Usage Examples

```python
# Example 1: Highlight Python code
from pygments import highlight
from pygments.lexers import PythonLexer
from pygments.formatters import HtmlFormatter

code = 'print("Hello, world!")'
result = highlight(code, PythonLexer(), HtmlFormatter())
print(result[:100])  # Show first 100 chars
```

```python
# Example 2: Different lexers (with proper imports)
from pygments import highlight
from pygments.lexers import JavascriptLexer, CssLexer
from pygments.formatters import HtmlFormatter

js_code = 'function hello() { console.log("Hello!"); }'
js_result = highlight(js_code, JavascriptLexer(), HtmlFormatter())
print("JavaScript highlighted:", len(js_result), "characters")

css_code = 'body { color: red; }'
css_result = highlight(css_code, CssLexer(), HtmlFormatter())
print("CSS highlighted:", len(css_result), "characters")

# Alternative approach - using get_lexer_by_name()
from pygments.lexers import get_lexer_by_name

js_lexer = get_lexer_by_name('javascript')
css_lexer = get_lexer_by_name('css')

js_result2 = highlight(js_code, js_lexer, HtmlFormatter())
css_result2 = highlight(css_code, css_lexer, HtmlFormatter())

print("Alternative approach also works!")
print("JS length:", len(js_result2), "CSS length:", len(css_result2))
```

```python
# Example 3: Terminal formatter
from pygments.formatters import TerminalFormatter

python_code = 'def hello():\n    return "world"'
terminal_result = highlight(python_code, PythonLexer(), TerminalFormatter())
print("Terminal formatted:")
print(terminal_result)
```

```python
# Example 4: List available lexers
from pygments.lexers import get_all_lexers

lexers = list(get_all_lexers())
print(f"Total lexers available: {len(lexers)}")
print("First 5 lexers:")
for name, aliases, filenames, mimetypes in lexers[:5]:
    print(f"  {name}: {aliases}")
```

```python
# Example 5: Guess lexer from content
from pygments.lexers import guess_lexer

sample_codes = [
    'print("Python")',
    'console.log("JavaScript")',
    'echo "Bash"',
    'SELECT * FROM table;'
]

for code in sample_codes:
    try:
        lexer = guess_lexer(code)
        print(f"'{code}' -> {lexer.name}")
    except Exception as e:
        print(f"'{code}' -> Could not guess: {e}")
        raise
```
