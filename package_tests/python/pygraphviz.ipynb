## Example 1: Basic Graph Creation and Operations

```python
import pygraphviz as pgv

# Example 1: Basic graph creation and operations
print("=== Basic Graph Creation and Operations ===")

# Create different types of graphs
print("Creating various graph types:")

# Directed graph
directed_graph = pgv.AGraph(directed=True, strict=True)
directed_graph.add_edges_from([
    ('Start', 'Process'), ('Process', 'Decision'), 
    ('Decision', 'End'), ('Decision', 'Process')
])

print(f"✓ Directed graph: {directed_graph.number_of_nodes()} nodes, {directed_graph.number_of_edges()} edges")
print(f"  Nodes: {list(directed_graph.nodes())}")
print(f"  Edges: {list(directed_graph.edges())}")

# Undirected graph
undirected_graph = pgv.AGraph(directed=False)
undirected_graph.add_nodes_from(['Alice', 'Bob', 'Charlie', 'Diana'])
undirected_graph.add_edges_from([
    ('Alice', 'Bob'), ('Bob', 'Charlie'), 
    ('Charlie', 'Diana'), ('Diana', 'Alice')
])

print(f"✓ Undirected graph: {undirected_graph.number_of_nodes()} nodes, {undirected_graph.number_of_edges()} edges")

# Graph properties
print(f"\nGraph Properties:")
print(f"  Directed graph is_directed: {directed_graph.is_directed()}")
print(f"  Undirected graph is_directed: {undirected_graph.is_directed()}")
print(f"  Directed graph is_strict: {directed_graph.is_strict()}")

# Node and edge operations
print(f"\nNode and Edge Access:")
start_node = directed_graph.get_node('Start')
process_edge = directed_graph.get_edge('Start', 'Process')

print(f"  Node 'Start': {start_node}")
print(f"  Edge 'Start' -> 'Process': {process_edge}")
print(f"  Node 'Start' degree: {directed_graph.degree(start_node)}")

print("✓ Basic graph operations demonstrated")
```

## Example 2: Graph Styling and Visual Attributes

```python
import pygraphviz as pgv

# Example 2: Graph styling and visual attributes
print("=== Graph Styling and Visual Attributes ===")

# Create a styled workflow graph
workflow = pgv.AGraph(directed=True, rankdir='TB')

# Add nodes with different styles
workflow.add_node('Start', shape='ellipse', color='green', style='filled', fillcolor='lightgreen')
workflow.add_node('Input', shape='parallelogram', color='blue', style='filled', fillcolor='lightblue')
workflow.add_node('Process', shape='box', color='orange', style='filled', fillcolor='lightyellow')
workflow.add_node('Decision', shape='diamond', color='red', style='filled', fillcolor='lightcoral')
workflow.add_node('Output', shape='parallelogram', color='blue', style='filled', fillcolor='lightblue')
workflow.add_node('End', shape='ellipse', color='green', style='filled', fillcolor='lightgreen')

# Add styled edges
workflow.add_edge('Start', 'Input', color='darkgreen', style='bold', arrowhead='vee')
workflow.add_edge('Input', 'Process', color='blue', style='solid')
workflow.add_edge('Process', 'Decision', color='orange', style='dashed')
workflow.add_edge('Decision', 'Output', color='red', label='Yes', fontcolor='red')
workflow.add_edge('Decision', 'Process', color='red', label='No', fontcolor='red', style='dotted')
workflow.add_edge('Output', 'End', color='darkgreen', style='bold', arrowhead='vee')

print(f"✓ Created styled workflow with {workflow.number_of_nodes()} nodes and {workflow.number_of_edges()} edges")

# Create a network topology graph
network = pgv.AGraph(directed=False, layout='neato')

# Add network devices with specific styling
network.add_node('Router', shape='box', color='red', style='filled', fillcolor='pink')
network.add_node('Switch1', shape='box', color='blue', style='filled', fillcolor='lightblue')
network.add_node('Switch2', shape='box', color='blue', style='filled', fillcolor='lightblue')
network.add_node('Server', shape='cylinder', color='green', style='filled', fillcolor='lightgreen')
network.add_node('Firewall', shape='octagon', color='orange', style='filled', fillcolor='lightyellow')

# Add connections with different line styles
network.add_edge('Router', 'Firewall', color='red', style='bold', penwidth=3)
network.add_edge('Firewall', 'Switch1', color='orange', style='solid', penwidth=2)
network.add_edge('Firewall', 'Switch2', color='orange', style='solid', penwidth=2)
network.add_edge('Switch1', 'Server', color='green', style='dashed')
network.add_edge('Switch2', 'Server', color='green', style='dashed')

print(f"✓ Created network topology with {network.number_of_nodes()} nodes and {network.number_of_edges()} edges")

# Demonstrate attribute access and modification
print(f"\nAttribute Demonstration:")
start_node = workflow.get_node('Start')
print(f"Start node shape: {start_node.attr['shape']}")
print(f"Start node color: {start_node.attr['color']}")

# Modify attributes
start_node.attr['label'] = 'BEGIN'
start_node.attr['fontsize'] = '14'
print(f"Modified Start node label: {start_node.attr['label']}")

# Graph-level attributes
workflow.graph_attr['label'] = 'Sample Workflow'
workflow.graph_attr['fontsize'] = '16'
workflow.node_attr['fontname'] = 'Arial'
workflow.edge_attr['fontname'] = 'Arial'

print(f"Set graph-level attributes")
print(f"Graph label: {workflow.graph_attr.get('label', 'None')}")

print("✓ Graph styling and visual attributes demonstrated")
```

## Example 3: Layout Engines and Algorithms

```python
import pygraphviz as pgv

# Example 3: Layout engines and algorithms
print("=== Layout Engines and Algorithms ===")

# Create a test graph for layout testing
test_graph = pgv.AGraph(directed=True)
nodes = ['A', 'B', 'C', 'D', 'E', 'F']
test_graph.add_nodes_from(nodes)

edges = [
    ('A', 'B'), ('A', 'C'), ('B', 'D'), ('C', 'D'),
    ('D', 'E'), ('E', 'F'), ('F', 'A'), ('B', 'F')
]
test_graph.add_edges_from(edges)

print(f"Created test graph: {test_graph.number_of_nodes()} nodes, {test_graph.number_of_edges()} edges")

# Test different layout engines
layout_engines = ['dot', 'neato', 'fdp', 'sfdp', 'circo', 'twopi']
print(f"\nTesting layout engines:")

successful_layouts = []
for engine in layout_engines:
    # Create a copy for each layout test
    layout_test = test_graph.copy()
    layout_test.layout(prog=engine)
    assert all(layout_test.get_node(n).attr['pos'] for n in nodes), engine
    successful_layouts.append(engine)
    print(f"  ✓ {engine}: Successfully applied layout")

# Demonstrate layout-specific graphs
print(f"\nLayout-Specific Examples:")

# Hierarchical layout with dot
hierarchical = pgv.AGraph(directed=True, rankdir='TB')
hierarchical.add_edges_from([
    ('CEO', 'CTO'), ('CEO', 'CFO'), ('CEO', 'COO'),
    ('CTO', 'Dev Lead'), ('CTO', 'QA Lead'),
    ('Dev Lead', 'Developer1'), ('Dev Lead', 'Developer2'),
    ('QA Lead', 'Tester1'), ('QA Lead', 'Tester2')
])

hierarchical.layout(prog='dot')
assert hierarchical.get_node('CEO').attr['pos'], 'dot left CEO without a position'
print(f"  ✓ Hierarchical layout (dot): {hierarchical.number_of_nodes()} nodes")

# Circular layout with circo
circular = pgv.AGraph(directed=False)
circular.add_edges_from([
    ('Node1', 'Node2'), ('Node2', 'Node3'), ('Node3', 'Node4'),
    ('Node4', 'Node5'), ('Node5', 'Node6'), ('Node6', 'Node1'),
    ('Node1', 'Center'), ('Node3', 'Center'), ('Node5', 'Center')
])

circular.layout(prog='circo')
assert circular.get_node('Center').attr['pos'], 'circo left Center without a position'
print(f"  ✓ Circular layout (circo): {circular.number_of_nodes()} nodes")

# Force-directed layout with neato
force_directed = pgv.AGraph(directed=False)
import itertools
fd_nodes = ['Hub1', 'Hub2', 'Hub3', 'Spoke1', 'Spoke2', 'Spoke3', 'Spoke4']
force_directed.add_nodes_from(fd_nodes)

# Create hub-and-spoke topology
force_directed.add_edges_from([
    ('Hub1', 'Spoke1'), ('Hub1', 'Spoke2'),
    ('Hub2', 'Spoke2'), ('Hub2', 'Spoke3'),
    ('Hub3', 'Spoke3'), ('Hub3', 'Spoke4'),
    ('Hub1', 'Hub2'), ('Hub2', 'Hub3')
])

force_directed.layout(prog='neato')
assert force_directed.get_node('Hub1').attr['pos'], 'neato left Hub1 without a position'
print(f"  ✓ Force-directed layout (neato): {force_directed.number_of_nodes()} nodes")

# Layout performance comparison
print(f"\nLayout Engine Summary:")
assert successful_layouts == layout_engines, successful_layouts
print(f"  Available engines: {len(successful_layouts)}/{len(layout_engines)}")
print(f"  Working engines: {', '.join(successful_layouts)}")
print(f"  Best for hierarchies: dot")
print(f"  Best for networks: neato, fdp")
print(f"  Best for circular: circo")

print("✓ Layout engines and algorithms demonstrated")
```

## Example 4: DOT Language and Graph I/O

```python
import pygraphviz as pgv
import tempfile
import os

# Example 4: DOT language and graph I/O
print("=== DOT Language and Graph I/O ===")

# Create graph from DOT language string
dot_string = """
digraph sample {
    rankdir=LR;
    
    // Node definitions with attributes
    start [shape=ellipse, color=green, style=filled];
    process [shape=box, color=blue, style=filled];
    decision [shape=diamond, color=orange, style=filled];
    end [shape=ellipse, color=red, style=filled];
    
    // Edge definitions with attributes  
    start -> process [label="begin", color=blue];
    process -> decision [label="evaluate", color=green];
    decision -> end [label="true", color=red];
    decision -> process [label="false", color=orange, style=dashed];
}
"""

print("Creating graph from DOT string:")
dot_graph = pgv.AGraph(string=dot_string)
assert (dot_graph.number_of_nodes(), dot_graph.number_of_edges(), dot_graph.is_directed()) == (4, 4, True)
assert list(dot_graph.nodes()) == ['start', 'process', 'decision', 'end'], list(dot_graph.nodes())
print(f"✓ DOT graph created: {dot_graph.number_of_nodes()} nodes, {dot_graph.number_of_edges()} edges")
print(f"  Nodes: {list(dot_graph.nodes())}")
print(f"  Directed: {dot_graph.is_directed()}")

# Generate DOT language output
print(f"\nDOT Language Generation:")
programmatic_graph = pgv.AGraph(directed=True)
programmatic_graph.add_edges_from([
    ('Input', 'Validate'), ('Validate', 'Process'), 
    ('Process', 'Output'), ('Validate', 'Error')
])

# Add styling
programmatic_graph.get_node('Input').attr.update(shape='parallelogram', color='blue')
programmatic_graph.get_node('Output').attr.update(shape='parallelogram', color='green')
programmatic_graph.get_node('Error').attr.update(shape='ellipse', color='red')

dot_output = str(programmatic_graph)
print(f"Generated DOT string ({len(dot_output)} characters):")
print(dot_output[:300] + "..." if len(dot_output) > 300 else dot_output)

# File I/O operations
print(f"\nFile I/O Operations:")

# Write to file
with tempfile.NamedTemporaryFile(mode='w', suffix='.dot', delete=False) as temp_file:
    temp_filename = temp_file.name
    programmatic_graph.write(temp_filename)

print(f"✓ Graph written to file: {os.path.basename(temp_filename)}")

# Read from file
loaded_graph = pgv.AGraph(temp_filename)
print(f"✓ Graph loaded from file: {loaded_graph.number_of_nodes()} nodes, {loaded_graph.number_of_edges()} edges")

# Verify integrity
original_nodes = set(programmatic_graph.nodes())
loaded_nodes = set(loaded_graph.nodes())
integrity_check = original_nodes == loaded_nodes

assert integrity_check, (original_nodes, loaded_nodes)
print("✓ Data integrity check: PASSED")

# Clean up temporary file
os.unlink(temp_filename)

# Graph copying and manipulation
print(f"\nGraph Copying and Manipulation:")

# Create original graph
original = pgv.AGraph(directed=False)
original.add_edges_from([('A', 'B'), ('B', 'C'), ('C', 'D'), ('D', 'A')])
original.graph_attr['label'] = 'Original Graph'

# Create copies
shallow_copy = original.copy()
print(f"✓ Shallow copy created: {shallow_copy.number_of_nodes()} nodes")

# Modify copy and verify independence
shallow_copy.add_node('E')
shallow_copy.add_edge('D', 'E')
shallow_copy.graph_attr['label'] = 'Modified Copy'

print(f"Original after copy modification: {original.number_of_nodes()} nodes")
print(f"Copy after modification: {shallow_copy.number_of_nodes()} nodes")
assert (original.number_of_nodes(), shallow_copy.number_of_nodes()) == (4, 5), 'copy() must not share nodes with the original'
print(f"Graphs are independent: {original.number_of_nodes() != shallow_copy.number_of_nodes()}")

# String representation comparison
print(f"\nString Representations:")
print(f"Original DOT length: {len(str(original))} characters")
print(f"Modified copy DOT length: {len(str(shallow_copy))} characters")

print("✓ DOT language and graph I/O demonstrated")
```

## Example 5: Real-world Application - Software Architecture Visualization

```python
import pygraphviz as pgv

# Example 5: Real-world application - software architecture visualization
print("=== Software Architecture Visualization ===")

# Create a microservices architecture diagram
architecture = pgv.AGraph(directed=True, rankdir='TB', compound=True)

# Define layers and components
frontend_services = ['Web App', 'Mobile App', 'Admin Panel']
api_services = ['API Gateway', 'Auth Service', 'User Service', 'Order Service']
data_services = ['User DB', 'Order DB', 'Cache', 'Message Queue']
external_services = ['Payment API', 'Email Service', 'Analytics']

print("Building software architecture diagram...")

# Add frontend layer
for service in frontend_services:
    architecture.add_node(service, 
        shape='box', 
        color='lightblue', 
        style='filled',
        layer='frontend'
    )

# Add API layer  
for service in api_services:
    architecture.add_node(service,
        shape='ellipse',
        color='lightgreen', 
        style='filled',
        layer='api'
    )

# Add data layer
for service in data_services:
    architecture.add_node(service,
        shape='cylinder' if 'DB' in service else 'box',
        color='lightyellow',
        style='filled', 
        layer='data'
    )

# Add external services
for service in external_services:
    architecture.add_node(service,
        shape='hexagon',
        color='lightcoral',
        style='filled',
        layer='external'
    )

# Define connections between layers
connections = [
    # Frontend to API Gateway
    ('Web App', 'API Gateway'),
    ('Mobile App', 'API Gateway'), 
    ('Admin Panel', 'API Gateway'),
    
    # API Gateway to services
    ('API Gateway', 'Auth Service'),
    ('API Gateway', 'User Service'),
    ('API Gateway', 'Order Service'),
    
    # Services to data layer
    ('Auth Service', 'User DB'),
    ('User Service', 'User DB'),
    ('User Service', 'Cache'),
    ('Order Service', 'Order DB'),
    ('Order Service', 'Message Queue'),
    
    # Services to external APIs
    ('Order Service', 'Payment API'),
    ('Auth Service', 'Email Service'),
    ('User Service', 'Analytics'),
]

# Add edges with styling
for source, target in connections:
    # Determine edge style based on connection type
    if any(service in data_services for service in [source, target]):
        style, color = 'dashed', 'blue'
    elif any(service in external_services for service in [source, target]):  
        style, color = 'dotted', 'red'
    else:
        style, color = 'solid', 'black'
    
    architecture.add_edge(source, target, style=style, color=color)

print(f"✓ Architecture diagram created:")
print(f"  Total components: {architecture.number_of_nodes()}")
print(f"  Total connections: {architecture.number_of_edges()}")
print(f"  Frontend services: {len(frontend_services)}")
print(f"  API services: {len(api_services)}")
print(f"  Data services: {len(data_services)}")
print(f"  External services: {len(external_services)}")

# Add graph-level styling and metadata
architecture.graph_attr.update({
    'label': 'Microservices Architecture\\n(Generated with PyGraphviz)',
    'fontsize': '16',
    'fontname': 'Arial Bold',
    'rankdir': 'TB',
    'ranksep': '1.0',
    'nodesep': '0.5'
})

# Set default node and edge attributes
architecture.node_attr.update({
    'fontname': 'Arial',
    'fontsize': '10'
})

architecture.edge_attr.update({
    'fontname': 'Arial', 
    'fontsize': '8'
})

# Generate comprehensive system documentation
print(f"\nSystem Architecture Analysis:")

# Analyze connectivity patterns
api_gateway_connections = len([e for e in architecture.edges() if 'API Gateway' in e])
database_connections = len([e for e in architecture.edges() if any('DB' in node for node in e)])
external_connections = len([e for e in architecture.edges() if any(ext in e for ext in external_services for e_node in e)])

print(f"  API Gateway connections: {api_gateway_connections}")
print(f"  Database connections: {database_connections}") 
print(f"  External API connections: {external_connections}")

# Component dependency analysis
print(f"  High-traffic nodes:")
for node in architecture.nodes():
    degree = architecture.degree(node)
    if degree >= 3:
        print(f"    {node}: {degree} connections")

# Generate DOT representation for documentation
dot_representation = str(architecture)
print(f"\nGenerated DOT specification: {len(dot_representation)} characters")

# Architecture validation
validation_results = {
    'has_frontend': len(frontend_services) > 0,
    'has_api_layer': len(api_services) > 0,
    'has_data_persistence': any('DB' in service for service in data_services),
    'has_external_integration': len(external_services) > 0,
    'proper_layering': len(connections) > 0
}

print(f"\nArchitecture Validation:")
for check, passed in validation_results.items():
    status = "✓ PASS" if passed else "✗ FAIL"
    print(f"  {check.replace('_', ' ').title()}: {status}")

overall_score = sum(validation_results.values()) / len(validation_results) * 100
print(f"  Overall Architecture Score: {overall_score:.1f}%")

print("✓ Software architecture visualization completed")
```
