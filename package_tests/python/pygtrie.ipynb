## Example 1: pygtrie functionality test 1

```python
import pygtrie

# Example 1: Basic StringTrie creation and operations
print("Testing pygtrie StringTrie functionality:")

try:
    # Create a string trie for dictionary-like operations
    word_trie = pygtrie.StringTrie()
    
    # Add word definitions
    words = {
        'cat': 'A small domesticated carnivorous mammal',
        'car': 'A road vehicle with four wheels',
        'card': 'A piece of thick, stiff paper',
        'care': 'The provision of what is necessary',
        'careful': 'Making sure of avoiding potential danger',
        'dog': 'A domesticated carnivorous mammal',
        'door': 'A hinged barrier used to close off an entrance'
    }
    
    # Populate the trie
    for word, definition in words.items():
        word_trie[word] = definition
    
    print(f"✓ Created StringTrie with {len(word_trie)} entries")
    print(f"✓ Trie type: {type(word_trie).__name__}")
    
    # Test basic operations
    print(f"✓ Lookup 'car': {word_trie['car']}")
    print(f"✓ Contains 'cat': {'cat' in word_trie}")
    print(f"✓ Contains 'catch': {'catch' in word_trie}")
    
    # List all keys
    all_keys = list(word_trie.keys())
    print(f"✓ All keys: {sorted(all_keys)}")
    
    # Available trie classes
    trie_classes = ['StringTrie', 'CharTrie', 'Trie', 'PrefixSet']
    print(f"✓ Available trie types: {trie_classes}")
    
    print("Example 1 completed successfully")
    
except Exception as e:
    print(f"Error in StringTrie operations: {e}")
    raise AssertionError(
        f"pygtrie example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 2: pygtrie functionality test 2

```python
import pygtrie

# Example 2: Prefix-based operations and searching
print("Testing pygtrie prefix operations:")

try:
    # Create trie for URL routing simulation: with separator='/', each path segment is one step of the trie
    route_trie = pygtrie.StringTrie(separator='/')
    
    # Define URL routes with handlers
    routes = {
        'api/members': 'MemberListHandler',
        'api/members/profile': 'MemberProfileHandler', 
        'api/members/settings': 'MemberSettingsHandler',
        'api/posts': 'PostListHandler',
        'api/posts/comments': 'CommentHandler',
        'admin/dashboard': 'AdminDashboardHandler',
        'admin/members': 'AdminMemberHandler',
        'public/about': 'AboutPageHandler',
        'public/contact': 'ContactPageHandler'
    }
    
    # Populate route trie
    for route, handler in routes.items():
        route_trie[route] = handler
    
    assert len(route_trie) == len(routes)
    print(f"✓ Created routing trie with {len(route_trie)} routes")
    
    # Test prefix-based lookups using items with prefix
    prefixes_to_test = {'api': 5, 'api/members': 3, 'admin': 2, 'public': 2}
    
    for prefix, expected_count in prefixes_to_test.items():
        # Find all routes with this prefix
        matching_routes = list(route_trie.items(prefix=prefix))
        assert len(matching_routes) == expected_count, (prefix, matching_routes)
        print(f"✓ Routes with prefix '{prefix}': {len(matching_routes)}")
        for key, handler in matching_routes[:3]:  # Show first 3
            print(f"    {key} → {handler}")
    
    # Prefixes are whole segments: 'api/member' is not a prefix of 'api/members'
    assert route_trie.has_subtrie('api/members') and not route_trie.has_subtrie('api/member')
    
    # Test longest prefix matching: the longest stored route that is a prefix of the path
    test_paths = {
        'api/members/profile/edit': 'api/members/profile',
        'admin/dashboard': 'admin/dashboard',
        'public/home': None,
    }
    
    print(f"✓ Longest prefix matching:")
    for path, expected_key in test_paths.items():
        step = route_trie.longest_prefix(path)
        if step:  # a miss returns a step that is false, rather than raising
            assert step.key == expected_key, (path, step.key)
            print(f"  '{path}' → matches '{step.key}' ({step.value})")
        else:
            assert expected_key is None, (path, expected_key)
            print(f"  '{path}' → no match found")
    
    print("Example 2 completed successfully")
    
except Exception as e:
    print(f"Error in prefix operations: {e}")
    raise AssertionError(
        f"pygtrie example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: pygtrie functionality test 3

```python
import pygtrie

# Example 3: CharTrie for character-level operations
print("Testing pygtrie CharTrie functionality:")

try:
    # Create character-level trie for DNA sequence analysis
    dna_trie = pygtrie.CharTrie()
    
    # DNA sequences with associated properties
    dna_sequences = {
        'ATCG': {'type': 'start_codon', 'frequency': 0.25},
        'ATCGA': {'type': 'promoter', 'frequency': 0.15},
        'ATCGAT': {'type': 'enhancer', 'frequency': 0.10},
        'GCTA': {'type': 'terminator', 'frequency': 0.20},
        'GCTAG': {'type': 'regulatory', 'frequency': 0.12},
        'TATA': {'type': 'TATA_box', 'frequency': 0.30},
        'TATAAT': {'type': 'strong_promoter', 'frequency': 0.08}
    }
    
    # Populate CharTrie with DNA sequences
    for sequence, properties in dna_sequences.items():
        dna_trie[sequence] = properties
    
    print(f"✓ Created CharTrie with {len(dna_trie)} DNA sequences")
    print(f"✓ Trie type: {type(dna_trie).__name__}")
    
    # Test character-level prefix matching by iterating all items
    test_prefixes = ['AT', 'GC', 'TA']
    
    for prefix in test_prefixes:
        # Find sequences starting with prefix by filtering all items
        matching = [(k, v) for k, v in dna_trie.items() if k.startswith(prefix)]
        print(f"✓ Sequences starting with '{prefix}': {len(matching)}")
        for seq, props in matching:
            print(f"    {seq} ({props['type']})")
    
    # Test iteration and filtering
    high_freq = [(s, p) for s, p in dna_trie.items() if p['frequency'] > 0.15]
    
    print(f"✓ High-frequency sequences (>15%): {len(high_freq)}")
    for seq, props in high_freq:
        print(f"    {seq}: {props['frequency']:.2f} ({props['type']})")
    
    # Test key operations
    print(f"✓ Key operations:")
    print(f"  has_key('ATCG'): {dna_trie.has_key('ATCG')}")
    print(f"  has_key('ZZZZZ'): {dna_trie.has_key('ZZZZZ')}")
    print(f"  'TATA' in trie: {'TATA' in dna_trie}")
    
    print("Example 3 completed successfully")
    
except Exception as e:
    print(f"Error in CharTrie operations: {e}")
    raise AssertionError(
        f"pygtrie example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: pygtrie functionality test 4

```python
import pygtrie

# Example 4: PrefixSet for efficient prefix checking
print("Testing pygtrie PrefixSet functionality:")

try:
    # Create PrefixSet for IP address filtering
    blocked_prefixes = pygtrie.PrefixSet([
        '192.168.',
        '10.',
        '172.16.',
        '172.17.',
        '172.18.',
        '127.',
        '169.254.',
        'fe80::',
        'fc00::',
        '::1'
    ])
    
    print(f"✓ Created PrefixSet with blocked IP prefixes")
    print(f"✓ PrefixSet type: {type(blocked_prefixes).__name__}")
    print(f"✓ Number of prefixes: {len(blocked_prefixes)}")
    
    # Test IP addresses against the prefix set
    test_ips = [
        '192.168.1.1',      # Private network
        '10.0.0.1',         # Private network
        '172.16.0.1',       # Private network
        '8.8.8.8',          # Public DNS
        '1.1.1.1',          # Public DNS
        '127.0.0.1',        # Localhost
        '169.254.1.1',      # Link-local
        '172.20.0.1',       # Not in our blocked range
        'fe80::1',          # IPv6 link-local
        '2001:db8::1'       # IPv6 public
    ]
    
    blocked_count = 0
    allowed_count = 0
    
    print(f"✓ Testing IP addresses:")
    for ip in test_ips:
        is_blocked = ip in blocked_prefixes
        status = "BLOCKED" if is_blocked else "ALLOWED"
        print(f"    {ip:<15} → {status}")
        
        if is_blocked:
            blocked_count += 1
        else:
            allowed_count += 1
    
    print(f"✓ Summary: {blocked_count} blocked, {allowed_count} allowed")
    
    # Test adding new prefixes
    print(f"✓ Testing dynamic prefix addition:")
    
    # Add new prefix
    blocked_prefixes.add('203.0.113.')  # RFC5737 test network
    print(f"  Added '203.0.113.' prefix")
    
    # Test new prefix
    test_new = '203.0.113.5'
    is_blocked_new = test_new in blocked_prefixes
    print(f"  {test_new} is now {'blocked' if is_blocked_new else 'allowed'}")
    print(f"  Updated prefix count: {len(blocked_prefixes)}")
    
    # Test prefix iteration
    all_prefixes = list(blocked_prefixes)
    print(f"✓ Current prefixes: {sorted(all_prefixes)[:5]}... ({len(all_prefixes)} total)")
    
    # Test copy functionality
    backup_set = pygtrie.PrefixSet(blocked_prefixes)
    print(f"✓ Created backup PrefixSet with {len(backup_set)} prefixes")
    
    print("Example 4 completed successfully")
    
except Exception as e:
    print(f"Error in PrefixSet operations: {e}")
    raise AssertionError(
        f"pygtrie example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 5: pygtrie functionality test 5

```python
import pygtrie

# Example 5: Advanced trie operations and performance patterns
print("Testing advanced pygtrie operations and use cases:")

try:
    # Create autocomplete system using StringTrie
    autocomplete_trie = pygtrie.StringTrie()
    
    # Programming languages and their descriptions
    languages = {
        'python': {'type': 'interpreted', 'year': 1991, 'popularity': 95},
        'java': {'type': 'interpreted', 'year': 1995, 'popularity': 85},
        'javascript': {'type': 'interpreted', 'year': 1995, 'popularity': 90},
        'go': {'type': 'compiled', 'year': 2009, 'popularity': 75},
        'rust': {'type': 'compiled', 'year': 2010, 'popularity': 70},
        'ruby': {'type': 'interpreted', 'year': 1995, 'popularity': 60},
        'php': {'type': 'interpreted', 'year': 1995, 'popularity': 55},
        'swift': {'type': 'compiled', 'year': 2014, 'popularity': 65},
        'kotlin': {'type': 'compiled', 'year': 2011, 'popularity': 68},
        'typescript': {'type': 'compiled', 'year': 2012, 'popularity': 80}
    }
    
    # Populate autocomplete trie
    for lang, info in languages.items():
        autocomplete_trie[lang] = info
    
    print(f"✓ Created autocomplete trie with {len(autocomplete_trie)} languages")
    
    # Test autocomplete functionality by filtering items
    def autocomplete(prefix, max_results=5):
        suggestions = [(k, v) for k, v in autocomplete_trie.items() if k.startswith(prefix)]
        return sorted(suggestions, key=lambda x: x[1]['popularity'], reverse=True)[:max_results]
    
    # Test various prefixes
    test_prefixes = ['p', 'j', 'r', 'type', 'go', 'k']
    
    print(f"✓ Autocomplete testing:")
    for prefix in test_prefixes:
        suggestions = autocomplete(prefix)
        print(f"  '{prefix}' → {len(suggestions)} suggestions")
        for lang, info in suggestions[:2]:  # Show top 2
            print(f"    {lang} ({info['type']}, popularity: {info['popularity']})")
    
    # Test trie traversal and statistics
    stats = {'interpreted': 0, 'compiled': 0, 'total_pop': 0}
    years = []
    
    for lang, info in autocomplete_trie.items():
        stats[info['type']] += 1
        stats['total_pop'] += info['popularity']
        years.append(info['year'])
    
    print(f"✓ Statistics:")
    print(f"  Interpreted: {stats['interpreted']}, Compiled: {stats['compiled']}")
    print(f"  Avg year: {sum(years)/len(years):.1f}, Total popularity: {stats['total_pop']}")
    
    # Test key operations
    print(f"✓ Key checks:")
    print(f"  'python' in trie: {'python' in autocomplete_trie}")
    print(f"  'c++' in trie: {'c++' in autocomplete_trie}")
    
    print("Example 5 completed successfully")
    
except Exception as e:
    print(f"Error in advanced operations: {e}")
    raise AssertionError(
        f"pygtrie example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
```
