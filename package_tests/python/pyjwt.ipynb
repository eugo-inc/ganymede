## Example 1: pyjwt functionality test 1

```python
import jwt  # PyJWT imports as 'jwt'
import datetime
from datetime import timezone

# Example 1: Basic JWT token creation and decoding
print("Testing PyJWT basic token operations:")

try:
    # Create payload with user information
    payload = {
        'user_id': 12345,
        'username': 'alice_smith',
        'email': 'alice@example.com',
        'role': 'user',
        'exp': datetime.datetime.now(timezone.utc) + datetime.timedelta(hours=1),
        'iat': datetime.datetime.now(timezone.utc),
        'iss': 'example-app'
    }
    
    # Secret key for HMAC signing
    secret_key = 'your-256-bit-secret-key-here'
    
    # Encode JWT token using HS256 algorithm
    token = jwt.encode(payload, secret_key, algorithm='HS256')
    
    print(f"✓ JWT token created: {type(token).__name__}")
    print(f"✓ Token length: {len(token)} characters")
    print(f"✓ Token preview: {token[:50]}...")
    
    # Decode and verify JWT token
    decoded_payload = jwt.decode(token, secret_key, algorithms=['HS256'])
    
    print(f"✓ Token decoded successfully")
    print(f"✓ User ID: {decoded_payload['user_id']}")
    print(f"✓ Username: {decoded_payload['username']}")
    print(f"✓ Role: {decoded_payload['role']}")
    print(f"✓ Issuer: {decoded_payload['iss']}")
    
    # Available JWT functions
    jwt_functions = [attr for attr in dir(jwt) if not attr.startswith('_')]
    print(f"✓ Available JWT functions: {jwt_functions}")
    
    print("Example 1 completed successfully")
    
except Exception as e:
    print(f"Error in JWT operations: {e}")
    raise AssertionError(
        f"pyjwt example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 2: pyjwt functionality test 2

```python
import jwt
from jwt.exceptions import InvalidTokenError, ExpiredSignatureError, InvalidSignatureError
import datetime
from datetime import timezone

# Example 2: JWT validation and error handling
print("Testing JWT validation and error handling:")

try:
    secret_key = 'secure-secret-key-2023'
    
    # Test scenarios with different token validation cases
    test_scenarios = [
        {
            'name': 'Valid token',
            'payload': {
                'user_id': 100,
                'role': 'admin',
                'exp': datetime.datetime.now(timezone.utc) + datetime.timedelta(minutes=30)
            },
            'decode_key': 'secure-secret-key-2023',
            'should_work': True
        },
        {
            'name': 'Expired token',
            'payload': {
                'user_id': 200,
                'role': 'user',
                'exp': datetime.datetime.now(timezone.utc) - datetime.timedelta(minutes=5)  # Past expiry
            },
            'decode_key': 'secure-secret-key-2023',
            'should_work': False
        },
        {
            'name': 'Wrong secret key',
            'payload': {
                'user_id': 300,
                'role': 'guest',
                'exp': datetime.datetime.now(timezone.utc) + datetime.timedelta(minutes=30)
            },
            'decode_key': 'wrong-secret-key',  # Wrong key for decoding
            'should_work': False
        }
    ]
    
    validation_results = []
    
    for scenario in test_scenarios:
        print(f"\n✓ Testing: {scenario['name']}")
        
        # Create token
        token = jwt.encode(scenario['payload'], secret_key, algorithm='HS256')
        print(f"  Token created: {len(token)} chars")
        
        # Try to decode
        try:
            decoded = jwt.decode(token, scenario['decode_key'], algorithms=['HS256'])
            success = True
            error_type = None
            print(f"  ✓ Decoded successfully: user_id={decoded.get('user_id')}")
            
        except ExpiredSignatureError:
            success = False
            error_type = 'ExpiredSignatureError'
            print(f"  ✗ Token expired")
            
        except InvalidSignatureError:
            success = False
            error_type = 'InvalidSignatureError'
            print(f"  ✗ Invalid signature")
            
        except InvalidTokenError as e:
            success = False
            error_type = 'InvalidTokenError'
            print(f"  ✗ Invalid token: {e}")
        
        # Validate expected outcome
        expected_success = scenario['should_work']
        result_correct = success == expected_success
        
        validation_results.append({
            'scenario': scenario['name'],
            'expected': expected_success,
            'actual': success,
            'error_type': error_type,
            'correct': result_correct
        })
        
        print(f"  Result: {'✓ As expected' if result_correct else '✗ Unexpected'}")
    
    # Summary
    correct_count = sum(1 for r in validation_results if r['correct'])
    total_count = len(validation_results)
    
    print(f"\n✓ Validation summary: {correct_count}/{total_count} scenarios handled correctly")
    
    print("Example 2 completed successfully")
    
except Exception as e:
    print(f"Error in validation testing: {e}")
    raise AssertionError(
        f"pyjwt example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: pyjwt functionality test 3

```python
import jwt
import datetime
from datetime import timezone

# Example 3: Different JWT algorithms and signing methods
print("Testing JWT algorithms and signing methods:")

try:
    # Test payload
    base_payload = {
        'user_id': 42,
        'username': 'demo_user',
        'exp': datetime.datetime.now(timezone.utc) + datetime.timedelta(hours=2),
        'iat': datetime.datetime.now(timezone.utc)
    }
    
    # Test different algorithms
    algorithms_to_test = [
        {
            'name': 'HMAC SHA256',
            'algorithm': 'HS256',
            'key': 'hmac-secret-key-for-testing',
            'decode_key': 'hmac-secret-key-for-testing'
        },
        {
            'name': 'HMAC SHA384',
            'algorithm': 'HS384',
            'key': 'hmac-sha384-secret-key-for-testing',
            'decode_key': 'hmac-sha384-secret-key-for-testing'
        },
        {
            'name': 'HMAC SHA512',
            'algorithm': 'HS512',
            'key': 'hmac-sha512-secret-key-for-testing-purposes',
            'decode_key': 'hmac-sha512-secret-key-for-testing-purposes'
        }
    ]
    
    algorithm_results = []
    
    for alg_config in algorithms_to_test:
        print(f"\n✓ Testing {alg_config['name']} ({alg_config['algorithm']}):")

        # Every HMAC algorithm here must round-trip: an encode/decode failure fails the cell
        token = jwt.encode(
            base_payload,
            alg_config['key'],
            algorithm=alg_config['algorithm']
        )

        print(f"  Encoded token length: {len(token)} characters")
        print(f"  Token header: {token.split('.')[0]}")  # Base64 header
        assert jwt.get_unverified_header(token) == {'alg': alg_config['algorithm'], 'typ': 'JWT'}, token

        # Decode with algorithm
        decoded = jwt.decode(
            token,
            alg_config['decode_key'],
            algorithms=[alg_config['algorithm']]
        )

        assert (decoded['user_id'], decoded['username']) == (42, 'demo_user'), decoded
        print(f"  ✓ Successfully decoded: user_id={decoded['user_id']}")

        algorithm_results.append({
            'algorithm': alg_config['algorithm'],
            'name': alg_config['name'],
            'success': True,
            'token_length': len(token)
        })
    
    # Test algorithm specification validation
    print(f"\n✓ Algorithm validation testing:")
    
    # Create token with HS256
    test_token = jwt.encode(base_payload, 'test-key', algorithm='HS256')
    
    # Try to decode with different algorithm list
    try:
        # This should fail - token was created with HS256 but we only allow HS384. PyJWT rejects the
        # header's `alg` before it checks any signature, so the error is InvalidAlgorithmError.
        decoded = jwt.decode(test_token, 'test-key', algorithms=['HS384'])
    except jwt.InvalidAlgorithmError as e:
        print(f"  ✓ Correctly rejected HS256 token when only HS384 allowed ({type(e).__name__})")
    else:
        raise AssertionError("an HS256 token decoded with only HS384 allowed")
    
    # Summary
    successful_algorithms = [r for r in algorithm_results if r['success']]
    print(f"\n✓ Algorithm summary: {len(successful_algorithms)}/{len(algorithm_results)} algorithms working")
    assert len(successful_algorithms) == len(algorithms_to_test) == 3, algorithm_results
    
    for result in successful_algorithms:
        print(f"  {result['name']}: {result['token_length']} chars")
    
    print("Example 3 completed successfully")
    
except Exception as e:
    print(f"Error in algorithm testing: {e}")
    raise AssertionError(
        f"pyjwt example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: pyjwt functionality test 4

```python
import jwt
from jwt import PyJWTError
import datetime
from datetime import timezone
import json
import base64

# Example 4: JWT header inspection and token structure analysis
print("Testing JWT token structure and header inspection:")

try:
    # Create tokens with different header configurations
    secret_key = 'analysis-secret-key'
    
    payload = {
        'user_id': 1001,
        'username': 'analyst',
        'permissions': ['read', 'write'],
        'department': 'engineering',
        'exp': datetime.datetime.now(timezone.utc) + datetime.timedelta(hours=4),
        'iat': datetime.datetime.now(timezone.utc),
        'nbf': datetime.datetime.now(timezone.utc),  # Not before
        'jti': 'unique-token-id-12345'  # JWT ID
    }
    
    # Create token with additional headers
    additional_headers = {
        'kid': 'key-id-2023',  # Key ID
        'typ': 'JWT',
        'alg': 'HS256'
    }
    
    token = jwt.encode(
        payload,
        secret_key,
        algorithm='HS256',
        headers=additional_headers
    )
    
    print(f"✓ Created token with custom headers")
    print(f"✓ Token length: {len(token)} characters")
    
    # Decode without verification to inspect structure
    unverified_header = jwt.get_unverified_header(token)
    print(f"✓ Unverified header: {unverified_header}")
    
    # Parse token structure manually
    token_parts = token.split('.')
    print(f"✓ Token parts: {len(token_parts)} (header.payload.signature)")
    
    # Decode header (first part)
    header_decoded = base64.urlsafe_b64decode(token_parts[0] + '==')  # Add padding
    header_json = json.loads(header_decoded.decode('utf-8'))
    print(f"✓ Decoded header JSON: {header_json}")
    
    # Decode payload (second part) - without verification
    payload_decoded = base64.urlsafe_b64decode(token_parts[1] + '==')  # Add padding
    payload_json = json.loads(payload_decoded.decode('utf-8'))
    print(f"✓ Decoded payload keys: {list(payload_json.keys())}")
    
    # Verify and decode properly
    verified_payload = jwt.decode(token, secret_key, algorithms=['HS256'])
    
    print(f"✓ Verified token successfully")
    print(f"✓ User: {verified_payload['username']}")
    print(f"✓ Permissions: {verified_payload['permissions']}")
    print(f"✓ JWT ID: {verified_payload['jti']}")
    
    # Test timestamp claims
    current_time = datetime.datetime.now(timezone.utc)
    exp_time = datetime.datetime.fromtimestamp(verified_payload['exp'], timezone.utc)
    iat_time = datetime.datetime.fromtimestamp(verified_payload['iat'], timezone.utc)
    
    print(f"✓ Token issued at: {iat_time.strftime('%Y-%m-%d %H:%M:%S UTC')}")
    print(f"✓ Token expires at: {exp_time.strftime('%Y-%m-%d %H:%M:%S UTC')}")
    print(f"✓ Time until expiry: {(exp_time - current_time).total_seconds() / 3600:.2f} hours")
    
    # Test invalid token structure
    print(f"\n✓ Testing invalid token structures:")
    
    invalid_tokens = [
        'invalid.token',  # Only 2 parts
        'not.a.token.with.too.many.parts',  # Too many parts
        'invalid-base64.invalid-base64.invalid-base64',  # Invalid base64
    ]
    
    for invalid_token in invalid_tokens:
        try:
            accepted = jwt.decode(invalid_token, secret_key, algorithms=['HS256'])
        except jwt.DecodeError as e:
            print(f"  ✓ Correctly rejected: {type(e).__name__}")
        else:
            raise AssertionError(f"jwt.decode accepted a malformed token {invalid_token!r}: {accepted!r}")
    
    print("Example 4 completed successfully")
    
except Exception as e:
    print(f"Error in token structure analysis: {e}")
    raise AssertionError(
        f"pyjwt example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 5: pyjwt functionality test 5

```python
import jwt
from jwt.exceptions import InvalidTokenError, ExpiredSignatureError, InvalidSignatureError, InvalidKeyError
import datetime
from datetime import timezone

# Example 5: Production JWT patterns and security considerations
print("Testing production JWT patterns and security:")

try:
    # Production-like JWT configuration
    class JWTManager:
        def __init__(self, secret_key, algorithm='HS256', expiry_hours=24):
            self.secret_key = secret_key
            self.algorithm = algorithm
            self.expiry_hours = expiry_hours
        
        def generate_token(self, user_data):
            payload = {
                **user_data,
                'exp': datetime.datetime.now(timezone.utc) + datetime.timedelta(hours=self.expiry_hours),
                'iat': datetime.datetime.now(timezone.utc),
                'iss': 'myapp-production'
            }
            return jwt.encode(payload, self.secret_key, algorithm=self.algorithm)
        
        def verify_token(self, token):
            try:
                payload = jwt.decode(
                    token, 
                    self.secret_key, 
                    algorithms=[self.algorithm],
                    options={
                        'verify_exp': True,
                        'verify_iat': True,
                        'verify_signature': True
                    }
                )
                return {'valid': True, 'payload': payload}
            except ExpiredSignatureError:
                return {'valid': False, 'error': 'Token expired'}
            except InvalidSignatureError:
                return {'valid': False, 'error': 'Invalid signature'}
            except InvalidTokenError as e:
                return {'valid': False, 'error': f'Invalid token: {e}'}
    
    # Initialize JWT manager
    jwt_manager = JWTManager(
        secret_key='production-grade-secret-key-2023-secure',
        algorithm='HS256',
        expiry_hours=2
    )
    
    print(f"✓ JWT Manager initialized")
    
    # Test user authentication workflow
    users = [
        {'user_id': 1001, 'username': 'alice', 'role': 'admin', 'department': 'IT'},
        {'user_id': 1002, 'username': 'bob', 'role': 'user', 'department': 'Sales'},
        {'user_id': 1003, 'username': 'charlie', 'role': 'manager', 'department': 'Marketing'}
    ]
    
    user_tokens = {}
    
    print(f"✓ Testing user token generation:")
    for user in users:
        token = jwt_manager.generate_token(user)
        user_tokens[user['username']] = token
        
        # Verify immediately
        verification = jwt_manager.verify_token(token)
        status = "✓ Valid" if verification['valid'] else f"✗ {verification['error']}"
        print(f"  {user['username']}: {status}")
    
    # Test token verification scenarios
    print(f"\n✓ Testing security scenarios:")
    
    # Valid token access
    alice_token = user_tokens['alice']
    alice_verification = jwt_manager.verify_token(alice_token)
    if alice_verification['valid']:
        payload = alice_verification['payload']
        print(f"  ✓ Alice access granted: {payload['role']} in {payload['department']}")
    
    # Tampered token test
    tampered_token = alice_token[:-5] + 'XXXXX'  # Change signature
    tampered_verification = jwt_manager.verify_token(tampered_token)
    print(f"  ✓ Tampered token: {tampered_verification['error']}")
    
    # Wrong secret key test
    wrong_key_manager = JWTManager('wrong-secret-key')
    wrong_key_verification = wrong_key_manager.verify_token(alice_token)
    print(f"  ✓ Wrong secret key: {wrong_key_verification['error']}")
    
    # Test token refresh pattern
    print(f"\n✓ Token refresh pattern:")
    
    # Simulate token nearing expiry
    short_lived_manager = JWTManager(
        'refresh-test-key',
        expiry_hours=0.001  # Very short expiry for testing
    )
    
    refresh_token = short_lived_manager.generate_token(users[0])
    print(f"  Short-lived token generated")
    
    # Wait a moment and test expiry (simulated)
    import time
    time.sleep(0.1)  # Small delay
    
    expired_verification = short_lived_manager.verify_token(refresh_token)
    if not expired_verification['valid'] and 'expired' in expired_verification['error'].lower():
        print(f"  ✓ Token correctly expired: {expired_verification['error']}")
        
        # Generate new token (refresh)
        new_token = short_lived_manager.generate_token(users[0])
        new_verification = short_lived_manager.verify_token(new_token)
        if new_verification['valid']:
            print(f"  ✓ New token generated and verified successfully")
    
    # Security best practices summary
    security_practices = [
        "Use strong, unique secret keys",
        "Set appropriate token expiry times", 
        "Validate all token claims (exp, iat, iss)",
        "Use HTTPS in production",
        "Store tokens securely (httpOnly cookies)",
        "Implement token refresh mechanisms",
        "Log authentication events",
        "Use algorithm whitelist in verification"
    ]
    
    print(f"\n✓ Security best practices:")
    for practice in security_practices[:4]:  # Show first 4
        print(f"  - {practice}")
    
    print("Example 5 completed successfully")
    
except Exception as e:
    print(f"Error in production patterns: {e}")
    raise AssertionError(
        f"pyjwt example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
```
