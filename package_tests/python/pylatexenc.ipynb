## Example 1: Basic LaTeX to Text Conversion

```python
from pylatexenc.latex2text import LatexNodes2Text
import pylatexenc

# Example 1: Basic LaTeX to text conversion
print("=== Basic LaTeX to Text Conversion ===")

print(f"pylatexenc version: {pylatexenc.__version__}")

# Create basic converter
converter = LatexNodes2Text()

# Test various LaTeX constructs
latex_samples = [
    r"Hello \textbf{world}! This is \emph{emphasized} text.",
    r"Mathematical symbols: $\alpha$, $\beta$, $\gamma$, and $\pi$.",
    r"Equations: $E = mc^2$ and $F = ma$.",
    r"Special characters: caf\'e, r\'esum\'e, na\"ive.",
    r"Citations and references: See \cite{smith2020} and \ref{eq:1}.",
    r"Lists: \textbf{Item 1}, \textit{Item 2}, \underline{Item 3}.",
    r"Sectioning: \section{Introduction} \subsection{Overview}",
    r"Accented characters: \'{a}, \`{e}, \^{o}, \~{n}, \"{u}."
]

print("LaTeX to text conversion examples:")
for i, latex_text in enumerate(latex_samples, 1):
    try:
        plain_text = converter.latex_to_text(latex_text)
        print(f"\n{i:2d}. LaTeX: {latex_text}")
        print(f"    Text:  {plain_text}")
    except Exception as e:
        print(f"\n{i:2d}. LaTeX: {latex_text}")
        print(f"    Error: {e}")
        raise

# Test with longer document excerpt
long_latex = r"""
\title{Sample Document}
\author{John Doe}
\section{Introduction}
This document demonstrates \textbf{bold}, \textit{italic}, and \textsc{small caps} text.
We can also include mathematical expressions like $x^2 + y^2 = r^2$ and Greek letters 
such as $\alpha$, $\beta$, $\gamma$, and $\delta$.
\subsection{Special Characters}
Various accents: caf\'e, na\"ive, r\^ole, and pi\~nata.
\section{Conclusion}
This concludes our \emph{brief} example.
"""

print(f"\nLong document conversion:")
print(f"Original LaTeX ({len(long_latex)} characters):")
print(long_latex[:200] + "..." if len(long_latex) > 200 else long_latex)

converted_long = converter.latex_to_text(long_latex)
print(f"\nConverted text ({len(converted_long)} characters):")
print(converted_long)

print("✓ Basic LaTeX to text conversion demonstrated")
```

## Example 2: Unicode to LaTeX Encoding

```python
from pylatexenc.latexencode import unicode_to_latex, UnicodeToLatexEncoder

# Example 2: Unicode to LaTeX encoding
print("=== Unicode to LaTeX Encoding ===")

# Test basic unicode_to_latex function
unicode_samples = [
    "Café résumé naïve Zürich",
    "Mathematical symbols: α β γ δ ε ζ η θ",
    "More math: ∞ ∑ ∫ ∂ ∇ ± × ÷",
    "Arrows and symbols: → ← ↑ ↓ ↔ ≤ ≥ ≠ ≈",
    "Currency and misc: € £ ¥ § © ® ™ ° ′ ″",
    "Fractions: ½ ⅓ ¼ ¾ ⅛ ⅜ ⅝ ⅞",
    "Accented letters: àáâãäåæçèéêë",
    "Eastern European: čđšž ąęłńóśćź"
]

print("Basic Unicode to LaTeX encoding:")
for i, unicode_text in enumerate(unicode_samples, 1):
    try:
        latex_encoded = unicode_to_latex(unicode_text)
        print(f"\n{i:2d}. Unicode: {unicode_text}")
        print(f"    LaTeX:   {latex_encoded}")
    except Exception as e:
        print(f"\n{i:2d}. Unicode: {unicode_text}")
        print(f"    Error:   {e}")
        raise

# Test with custom encoder settings
print(f"\nCustom encoder configurations:")

# Create encoders with different settings
encoders = {
    'default': UnicodeToLatexEncoder(),
    'non_ascii_only': UnicodeToLatexEncoder(non_ascii_only=True),
    'replacement_latex_protection': UnicodeToLatexEncoder(replacement_latex_protection='braces')
}

test_text = "Café with α-particles at 25°C costs €5.50"

for name, encoder in encoders.items():
    encoded = encoder.unicode_to_latex(test_text)
    print(f"  {name:>25}: {encoded}")

# Test encoding of mathematical expressions
math_unicode = [
    "∀x∈ℝ: x² ≥ 0",
    "lim(n→∞) 1/n = 0", 
    "∫₀^∞ e^(-x²) dx = √π/2",
    "∂f/∂x = df/dx",
    "∑ᵢ₌₁^n i² = n(n+1)(2n+1)/6"
]

print(f"\nMathematical Unicode expressions:")
for i, math_text in enumerate(math_unicode, 1):
    encoded = unicode_to_latex(math_text)
    print(f"{i}. {math_text}")
    print(f"   {encoded}")

# Test with different character sets
print(f"\nInternational character sets:")

international_texts = {
    'French': "Les élèves étudient à l'école française",
    'German': "Müller schreibt über Größe und Wärme", 
    'Spanish': "El niño come mañana en España",
    'Polish': "Żółć gęślą jaźń",
    'Czech': "Příliš žluťoučký kůň úpěl ďábelské ódy",
    'Nordic': "Åse spør om røde børn i Köln"
}

for language, text in international_texts.items():
    encoded = unicode_to_latex(text)
    print(f"  {language:>8}: {text}")
    print(f"           → {encoded}")

print("✓ Unicode to LaTeX encoding demonstrated")
```

## Example 3: Advanced Text Conversion with Math Modes

```python
from pylatexenc.latex2text import LatexNodes2Text

# Example 3: Advanced text conversion with math modes
print("=== Advanced Text Conversion with Math Modes ===")

# Test mathematical content with different modes
math_latex = r"""
The fundamental theorem states that $\int_a^b f'(x)dx = f(b) - f(a)$.
For complex numbers, we have $|z|^2 = z \cdot \bar{z}$ where $z = a + bi$.
The quadratic formula gives us $x = \frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$.
Einstein's equation: $E = mc^2$ relates energy and mass.
The wave equation: $\frac{\partial^2 u}{\partial t^2} = c^2 \nabla^2 u$.
"""

# Test different math modes
math_modes = {
    'text': 'Convert math to readable text',
    'verbatim': 'Keep math expressions as-is', 
    'remove': 'Remove math expressions entirely'
}

print("Mathematical content with different modes:")
for mode, description in math_modes.items():
    print(f"\n{mode.upper()} mode ({description}):")
    converter = LatexNodes2Text(math_mode=mode)
    result = converter.latex_to_text(math_latex)
    print(result.strip())

# Test advanced converter options
print(f"\nAdvanced converter configurations:")

# Different converter configurations
configs = [
    {
        'name': 'Default',
        'options': {}
    },
    {
        'name': 'Strict LaTeX Spacing',
        'options': {'strict_latex_spaces': True}
    },
    {
        'name': 'Keep Braced Groups',
        'options': {'keep_braced_groups': True}
    },
    {
        'name': 'Custom Math + No Braces',
        'options': {'math_mode': 'text', 'keep_braced_groups': False}
    }
]

test_latex = r"The function $f(x) = x^2$ has derivative $f'(x) = 2x$. Also, \textbf{bold \emph{nested} text}."

for config in configs:
    converter = LatexNodes2Text(**config['options'])
    result = converter.latex_to_text(test_latex)
    print(f"  {config['name']:>20}: {result}")

# Test handling of complex mathematical expressions
complex_math_examples = [
    r"Matrix notation: $\mathbf{A} = \begin{pmatrix} a & b \\ c & d \end{pmatrix}$",
    r"Summation: $\sum_{n=1}^{\infty} \frac{1}{n^2} = \frac{\pi^2}{6}$",
    r"Integral: $\oint_C \mathbf{F} \cdot d\mathbf{r} = \iint_S (\nabla \times \mathbf{F}) \cdot \mathbf{n} \, dS$",
    r"Limits: $\lim_{x \to \infty} \left(1 + \frac{1}{x}\right)^x = e$",
    r"Greek alphabet: $\Alpha \Beta \Gamma \Delta \Epsilon \Zeta \Eta \Theta$"
]

print(f"\nComplex mathematical expressions:")
text_converter = LatexNodes2Text(math_mode='text')
verbatim_converter = LatexNodes2Text(math_mode='verbatim')

for i, math_expr in enumerate(complex_math_examples, 1):
    print(f"\n{i}. Original: {math_expr}")
    
    text_result = text_converter.latex_to_text(math_expr)
    print(f"   Text:     {text_result}")
    
    verbatim_result = verbatim_converter.latex_to_text(math_expr) 
    print(f"   Verbatim: {verbatim_result}")

# Test with scientific document excerpt
scientific_latex = r"""
\section{Methodology}
Our approach uses the \emph{finite element method} to solve the partial differential equation:
\begin{equation}
\nabla^2 u + k^2 u = 0
\end{equation}
where $k = \frac{2\pi}{\lambda}$ is the wavenumber and $\lambda$ is the wavelength.
The boundary conditions are $u|_{\partial \Omega} = 0$.
Results show convergence rates of $\mathcal{O}(h^2)$ where $h$ is the mesh size.
"""

print(f"\nScientific document processing:")
scientific_converter = LatexNodes2Text(
    math_mode='text',
    strict_latex_spaces=True,
    keep_braced_groups=False
)

scientific_text = scientific_converter.latex_to_text(scientific_latex)
print(scientific_text.strip())

print("✓ Advanced text conversion with math modes demonstrated")
```

## Example 4: LaTeX Document Parsing with LatexWalker

```python
from pylatexenc.latexwalker import LatexWalker, LatexWalkerError
from pylatexenc.latexwalker import LatexMacroNode, LatexEnvironmentNode, LatexCharsNode

# Example 4: LaTeX document parsing with LatexWalker
print("=== LaTeX Document Parsing with LatexWalker ===")

# Sample LaTeX document for parsing
sample_document = r"""
\documentclass{article}
\usepackage{amsmath}

\title{Sample Research Paper}
\author{Dr. Jane Smith}
\date{\today}

\begin{document}

\maketitle

\section{Introduction}
This paper presents a novel approach to solving the equation:
\begin{equation}
\label{eq:main}
f(x) = \sum_{n=1}^{\infty} \frac{(-1)^n}{n!} x^n
\end{equation}

We use the \textbf{Monte Carlo method} with \emph{importance sampling}.

\subsection{Previous Work}
Earlier studies by \cite{johnson2019} and \cite{brown2020} showed different results.

\section{Methodology}
Our approach involves three main steps:
\begin{enumerate}
\item Data collection from $N = 1000$ samples
\item Statistical analysis using $\chi^2$ test
\item Validation with cross-validation
\end{enumerate}

\section{Results}
The results are shown in Table~\ref{tab:results} and Figure~\ref{fig:graph}.

\section{Conclusion}
This work demonstrates the effectiveness of our approach.

\end{document}
"""

print("Parsing LaTeX document structure:")

# Parse the document
walker = LatexWalker(sample_document)
try:
    nodelist, pos, len_parsed = walker.get_latex_nodes()
    
    print(f"✓ Parsed {len(nodelist)} top-level nodes")
    print(f"✓ Parsed {len_parsed} characters out of {len(sample_document)}")
    
    # Analyze the parsed structure
    node_types = {}
    macros_found = []
    environments_found = []
    
    for node in nodelist:
        node_type = type(node).__name__
        node_types[node_type] = node_types.get(node_type, 0) + 1
        
        if hasattr(node, 'macroname'):
            macros_found.append(node.macroname)
        elif hasattr(node, 'environmentname'):
            environments_found.append(node.environmentname)
    
    print(f"\nNode type distribution:")
    for node_type, count in node_types.items():
        print(f"  {node_type}: {count}")
    
    print(f"\nMacros found: {set(macros_found)}")
    print(f"Environments found: {set(environments_found)}")

except LatexWalkerError as e:
    print(f"Parsing error: {e}")

# Demonstrate detailed node inspection
print(f"\nDetailed node inspection (first 10 nodes):")
for i, node in enumerate(nodelist[:10]):
    node_type = type(node).__name__
    
    if hasattr(node, 'macroname'):
        print(f"  {i:2d}. {node_type}: \\{node.macroname}")
        if hasattr(node, 'nodeargd') and node.nodeargd:
            args = []
            if node.nodeargd.argnlist:
                for arg in node.nodeargd.argnlist:
                    if arg is not None:
                        args.append(str(arg)[:30] + "..." if len(str(arg)) > 30 else str(arg))
            if args:
                print(f"      Args: {args}")
                
    elif hasattr(node, 'environmentname'):
        print(f"  {i:2d}. {node_type}: \\begin{{{node.environmentname}}}")
        
    elif hasattr(node, 'chars'):
        chars_preview = node.chars[:50].replace('\n', '\\n')
        print(f"  {i:2d}. {node_type}: '{chars_preview}{'...' if len(node.chars) > 50 else ''}'")
        
    else:
        print(f"  {i:2d}. {node_type}: {str(node)[:50]}...")

# Test parsing of mathematical expressions
math_examples = [
    r"Simple inline: $x^2 + y^2 = z^2$",
    r"Display math: $$\int_0^\infty e^{-x^2} dx = \frac{\sqrt{\pi}}{2}$$",
    r"Align environment: \begin{align} x &= a + b \\ y &= c + d \end{align}",
    r"Matrix: $\begin{pmatrix} 1 & 2 \\ 3 & 4 \end{pmatrix}$"
]

print(f"\nParsing mathematical expressions:")
for i, math_expr in enumerate(math_examples, 1):
    try:
        math_walker = LatexWalker(math_expr)
        math_nodes, _, _ = math_walker.get_latex_nodes()
        
        print(f"{i}. {math_expr}")
        print(f"   Parsed into {len(math_nodes)} nodes")
        
        for j, node in enumerate(math_nodes):
            node_info = type(node).__name__
            if hasattr(node, 'displaytype'):
                node_info += f" (display: {node.displaytype})"
            elif hasattr(node, 'macroname'):
                node_info += f" (macro: {node.macroname})"
            elif hasattr(node, 'environmentname'):
                node_info += f" (env: {node.environmentname})"
            print(f"      Node {j}: {node_info}")
            
    except Exception as e:
        print(f"{i}. {math_expr} -> Error: {e}")
        raise

print("✓ LaTeX document parsing with LatexWalker demonstrated")
```

## Example 5: Real-world Document Processing Pipeline

```python
from pylatexenc.latex2text import LatexNodes2Text
from pylatexenc.latexencode import unicode_to_latex
from pylatexenc.latexwalker import LatexWalker

# Example 5: Real-world document processing pipeline
print("=== Real-world Document Processing Pipeline ===")

# Simulate a document processing workflow
class LaTeXProcessor:
    """A comprehensive LaTeX document processor"""
    
    def __init__(self):
        self.text_converter = LatexNodes2Text(
            math_mode='text',
            strict_latex_spaces=True,
            keep_braced_groups=False
        )
        self.stats = {
            'documents_processed': 0,
            'total_chars_input': 0,
            'total_chars_output': 0,
            'errors': []
        }
    
    def extract_metadata(self, latex_content):
        """Extract document metadata using LatexWalker"""
        metadata = {
            'title': None,
            'author': None,
            'sections': [],
            'citations': [],
            'labels': [],
            'math_expressions': 0
        }
        
        try:
            walker = LatexWalker(latex_content)
            nodelist, _, _ = walker.get_latex_nodes()
            
            for node in nodelist:
                if hasattr(node, 'macroname'):
                    if node.macroname == 'title' and node.nodeargd and node.nodeargd.argnlist:
                        if node.nodeargd.argnlist[0]:
                            metadata['title'] = self.text_converter.latex_to_text(
                                latex_content[node.nodeargd.argnlist[0].pos:node.nodeargd.argnlist[0].pos + node.nodeargd.argnlist[0].len]
                            )
                    elif node.macroname == 'author' and node.nodeargd and node.nodeargd.argnlist:
                        if node.nodeargd.argnlist[0]:
                            metadata['author'] = self.text_converter.latex_to_text(
                                latex_content[node.nodeargd.argnlist[0].pos:node.nodeargd.argnlist[0].pos + node.nodeargd.argnlist[0].len]
                            )
                    elif node.macroname in ['section', 'subsection', 'subsubsection']:
                        if node.nodeargd and node.nodeargd.argnlist and node.nodeargd.argnlist[0]:
                            section_title = self.text_converter.latex_to_text(
                                latex_content[node.nodeargd.argnlist[0].pos:node.nodeargd.argnlist[0].pos + node.nodeargd.argnlist[0].len]
                            )
                            metadata['sections'].append((node.macroname, section_title))
                    elif node.macroname == 'cite':
                        if node.nodeargd and node.nodeargd.argnlist and node.nodeargd.argnlist[0]:
                            citation = latex_content[node.nodeargd.argnlist[0].pos:node.nodeargd.argnlist[0].pos + node.nodeargd.argnlist[0].len]
                            metadata['citations'].append(citation)
                    elif node.macroname == 'label':
                        if node.nodeargd and node.nodeargd.argnlist and node.nodeargd.argnlist[0]:
                            label = latex_content[node.nodeargd.argnlist[0].pos:node.nodeargd.argnlist[0].pos + node.nodeargd.argnlist[0].len]
                            metadata['labels'].append(label)
                
                # Count math expressions
                if hasattr(node, 'displaytype'):
                    metadata['math_expressions'] += 1
            
        except Exception as e:
            self.stats['errors'].append(f"Metadata extraction: {e}")
        
        return metadata
    
    def process_document(self, latex_content, target_format='text'):
        """Process a complete LaTeX document"""
        self.stats['documents_processed'] += 1
        self.stats['total_chars_input'] += len(latex_content)
        
        result = {
            'original_length': len(latex_content),
            'metadata': {},
            'converted_text': '',
            'processing_notes': []
        }
        
        try:
            # Extract metadata
            result['metadata'] = self.extract_metadata(latex_content)
            
            # Convert to text
            if target_format == 'text':
                result['converted_text'] = self.text_converter.latex_to_text(latex_content)
                self.stats['total_chars_output'] += len(result['converted_text'])
            
            result['processing_notes'].append("✓ Document processed successfully")
            
        except Exception as e:
            self.stats['errors'].append(f"Document processing: {e}")
            result['processing_notes'].append(f"✗ Processing error: {e}")
        
        return result
    
    def get_statistics(self):
        """Get processing statistics"""
        if self.stats['total_chars_input'] > 0:
            compression_ratio = self.stats['total_chars_output'] / self.stats['total_chars_input']
        else:
            compression_ratio = 0
            
        return {
            **self.stats,
            'compression_ratio': compression_ratio,
            'avg_input_size': self.stats['total_chars_input'] / max(1, self.stats['documents_processed']),
            'avg_output_size': self.stats['total_chars_output'] / max(1, self.stats['documents_processed'])
        }

# Test the document processor
processor = LaTeXProcessor()

# Sample academic documents for processing
test_documents = [
    {
        'name': 'Research Paper',
        'content': r"""
        \title{Machine Learning Approaches to Climate Modeling}
        \author{Dr. Alice Johnson and Dr. Bob Smith}
        \section{Introduction}
        Climate change prediction requires sophisticated \textbf{machine learning} models.
        The relationship between temperature $T$ and CO$_2$ concentration $C$ can be modeled as:
        $$T = \alpha \log(C/C_0) + T_0$$
        where $\alpha = 3.7$ K and $T_0$ is the pre-industrial temperature.
        \section{Methodology}
        We employed \emph{neural networks} and \textsc{random forests} as described in \cite{jones2021}.
        \label{sec:methods}
        \section{Results}
        Our model achieved 95\% accuracy on the test dataset.
        """
    },
    {
        'name': 'Mathematical Theorem',
        'content': r"""
        \title{A New Proof of the Pythagorean Theorem}
        \author{Prof. Carl Mathematics}
        \section{Theorem Statement}
        For any right triangle with legs of length $a$ and $b$, and hypotenuse of length $c$:
        \begin{equation}
        \label{eq:pythagoras}
        a^2 + b^2 = c^2
        \end{equation}
        \section{Proof}
        Consider a square with side length $(a+b)$. The area can be computed as:
        $$\text{Area} = (a+b)^2 = a^2 + 2ab + b^2$$
        This completes our elegant proof \cite{euclid300bc}.
        """
    },
    {
        'name': 'Technical Report',
        'content': r"""
        \title{Performance Analysis of Database Systems}
        \author{Engineering Team}
        \section{Executive Summary}
        This report analyzes the performance of various \textbf{database management systems}.
        \subsection{Methodology}
        We measured query response times for $N = 10^6$ operations.
        \subsubsection{Test Configuration}
        Hardware: 64GB RAM, SSD storage at 500MB/s throughput.
        \section{Findings}
        Response times follow a log-normal distribution with mean $\mu = 2.3$ms.
        \label{sec:results}
        """
    }
]

print("Processing sample documents:")
for i, doc in enumerate(test_documents, 1):
    print(f"\n{i}. {doc['name']}:")
    result = processor.process_document(doc['content'])
    
    # Display metadata
    metadata = result['metadata']
    print(f"   Title: {metadata.get('title', 'N/A')}")
    print(f"   Author: {metadata.get('author', 'N/A')}")
    print(f"   Sections: {len(metadata.get('sections', []))}")
    print(f"   Citations: {len(metadata.get('citations', []))}")
    print(f"   Math expressions: {metadata.get('math_expressions', 0)}")
    
    # Show text preview
    text_preview = result['converted_text'][:150].replace('\n', ' ')
    print(f"   Text preview: {text_preview}...")
    print(f"   Length: {result['original_length']} → {len(result['converted_text'])} chars")

# Round-trip conversion test
print(f"\nRound-trip conversion test:")
unicode_text = "Mathematical constants: π ≈ 3.14159, e ≈ 2.71828, φ = (1+√5)/2 ≈ 1.618"
latex_encoded = unicode_to_latex(unicode_text)
text_decoded = processor.text_converter.latex_to_text(latex_encoded)

print(f"Original:  {unicode_text}")
print(f"→ LaTeX:   {latex_encoded}")
print(f"→ Text:    {text_decoded}")

# Display final statistics
print(f"\nProcessing Statistics:")
stats = processor.get_statistics()
for key, value in stats.items():
    if isinstance(value, float):
        print(f"  {key}: {value:.3f}")
    elif isinstance(value, list):
        print(f"  {key}: {len(value)} items")
    else:
        print(f"  {key}: {value}")

print("✓ Real-world document processing pipeline demonstrated")
```
