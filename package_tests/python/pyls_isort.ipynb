## Example 3: pyls_isort functionality test 3

```python
import isort

# Example 3: Import sorting with different styles
print("Testing import sorting with different styles:")

try:
    # Test with standard library, third-party, and local imports
    mixed_imports = '''import sys
import requests
from myapp import models
import os
from typing import List
import numpy as np
from .utils import helper'''
    
    print("Original mixed imports:")
    for line in mixed_imports.split('\n'):
        print(f"  {line}")
    
    # Sort with default isort settings
    sorted_default = isort.code(mixed_imports)
    print("\nSorted (default style):")
    for line in sorted_default.split('\n'):
        print(f"  {line}")
    
    # Sort with different profile (black compatible)
    sorted_black = isort.code(mixed_imports, profile="black")
    print("\nSorted (black profile):")
    for line in sorted_black.split('\n'):
        print(f"  {line}")
    
    # Test complex imports
    complex_imports = '''from typing import Dict, List, Optional
import os, sys
from pathlib import Path'''
    
    sorted_complex = isort.code(complex_imports, force_single_line=True)
    print("\nComplex imports (one per line):")
    for line in sorted_complex.split('\n'):
        print(f"  {line}")
    
    print("Example 3 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    print(f"Error type: {type(e).__name__}")
    raise AssertionError(
        f"pyls_isort example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: pyls_isort functionality test 4

```python
from pyls_isort.plugin import Config
import isort

# Example 4: Configuration and isort settings
print("Testing pyls_isort configuration:")

try:
    # Explore Config class from plugin
    config = Config()
    print(f"✓ Config object created: {type(config).__name__}")
    
    # Check isort configuration options
    isort_config = isort.Config()
    config_attrs = [attr for attr in dir(isort_config) if not attr.startswith('_')]
    print(f"✓ isort Config has {len(config_attrs)} attributes")
    print(f"  Sample settings: {config_attrs[:5]}")
    
    # Test different sorting configurations
    test_imports = '''import sys
import os
from typing import List
import json'''
    
    # Default sorting
    sorted_default = isort.code(test_imports)
    print("\nDefault sorting:")
    print(sorted_default)
    
    # With custom line length
    sorted_custom = isort.code(test_imports, line_length=80, force_grid_wrap=0)
    print("Custom line length (80):")
    print(sorted_custom)
    
    # Check isort version
    print(f"✓ isort version: {isort.__version__}")
    print(f"✓ Available profiles: black, django, pycharm, google")
    
    print("Example 4 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    print(f"Error type: {type(e).__name__}")
    raise AssertionError(
        f"pyls_isort example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```
