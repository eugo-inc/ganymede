## Example 3: pylsp_mypy functionality test 3

```python
from pylsp_mypy.plugin import parse_line

# Example 3: MyPy error parsing
print("Testing pylsp_mypy error parsing:")

try:
    # Sample MyPy output lines
    mypy_lines = [
        'test.py:5: error: Incompatible return value type (got "str", expected "int")  [return-value]',
        'test.py:12: error: Argument 1 to "len" has incompatible type "int"  [arg-type]',
        'test.py:8: note: Revealed type is "builtins.str"',
    ]
    
    print(f"✓ Testing {len(mypy_lines)} MyPy output lines")
    
    # Parse each line
    for i, line in enumerate(mypy_lines, 1):
        parsed = parse_line(line)
        if parsed:
            print(f"✓ Line {i} parsed successfully")
            print(f"  Message: {parsed.get('message', 'N/A')[:50]}")
            line_num = parsed.get('range', {}).get('start', {}).get('line', 'N/A')
            print(f"  Line number: {line_num}")
        else:
            print(f"  Line {i}: No parse result")
    
    # Test regex patterns
    from pylsp_mypy.plugin import line_pattern, whole_line_pattern
    print(f"✓ Regex patterns available")
    print(f"  Line pattern has match method: {hasattr(line_pattern, 'match')}")
    
    print("Example 3 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"pylsp_mypy example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: pylsp_mypy functionality test 4

```python
from pylsp_mypy.plugin import get_cmd
import tempfile
import os

# Example 4: MyPy command generation
print("Testing pylsp_mypy command generation:")

try:
    # Create a test Python file
    test_code = '''def add(a: int, b: int) -> int:
    return a + b

result = add("hello", 5)  # Type error
'''
    
    with tempfile.NamedTemporaryFile(mode='w', suffix='.py', delete=False) as f:
        f.write(test_code)
        temp_file = f.name
    
    print(f"✓ Created test file: {os.path.basename(temp_file)}")
    
    # Build mypy command with overrides
    overrides = ['--ignore-missing-imports', '--show-error-codes']
    # API drift: the signature is `get_cmd(settings: dict, cmd: str)` — settings FIRST, then the
    # executable name. The old call passed a file path as the settings mapping, so the plugin's
    # `settings.get(...)` hit a str and raised `AttributeError: 'str' object has no attribute 'get'`.
    cmd = get_cmd({"overrides": overrides}, "mypy")
    
    print(f"✓ MyPy command generated")
    print(f"  Command type: {type(cmd).__name__}")
    print(f"  Command parts: {len(cmd)} arguments")
    print(f"  First arg: {cmd[0] if cmd else 'N/A'}")
    print(f"  Last arg: {os.path.basename(cmd[-1]) if cmd else 'N/A'}")
    
    # Cleanup
    os.unlink(temp_file)
    
    print("Example 4 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"pylsp_mypy example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```
