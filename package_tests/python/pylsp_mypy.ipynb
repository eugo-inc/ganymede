```python
import os, pathlib, tempfile
from pylsp import uris
from pylsp.config.config import Config
from pylsp.workspace import Document, Workspace
from pylsp_mypy.plugin import pylsp_lint

print({k: os.environ.get(k) for k in ("FORCE_COLOR", "MYPY_FORCE_COLOR")})
root = pathlib.Path(tempfile.mkdtemp())
os.chdir(root)  # mypy keeps its cache in the working folder; a fresh one, so no earlier run answers for this file
planted = root / "planted.py"
planted.write_text('def half(x: int) -> int:\n    return x / 2\n\nname: str = 42\n')
workspace = Workspace(uris.from_fs_path(str(root)), endpoint=None)
config = Config(workspace.root_uri, {}, 0, {})
config.update({"plugins": {"pylsp_mypy": {"enabled": True, "live_mode": False}}})
document = Document(uris.from_fs_path(str(planted)), workspace)
found = sorted((d["range"]["start"]["line"] + 1, d["code"]) for d in pylsp_lint(config, workspace, document, True)
               if d.get("source") == "mypy")
print("mypy diagnostics pylsp_mypy parsed:", found)
assert found == [(2, "return-value"), (4, "assignment")], (
    f"expected the 2 planted mypy errors, got {found}: is mypy's colour forced (MYPY_FORCE_COLOR / FORCE_COLOR)?")
print("Example 3 completed successfully")
```

## Example 4: pylsp_mypy functionality test 4

```python
from pylsp_mypy.plugin import get_cmd
import tempfile
import os

# Example 4: MyPy command generation
print("Testing pylsp_mypy command generation:")

try:
    # Create a test Python file
    test_code = '''def add(a: int, b: int) -> int:
    return a + b

result = add("hello", 5)  # Type error
'''
    
    with tempfile.NamedTemporaryFile(mode='w', suffix='.py', delete=False) as f:
        f.write(test_code)
        temp_file = f.name
    
    print(f"✓ Created test file: {os.path.basename(temp_file)}")
    
    # Build mypy command with overrides
    overrides = ['--ignore-missing-imports', '--show-error-codes']
    # API drift: the signature is `get_cmd(settings: dict, cmd: str)` — settings FIRST, then the
    # executable name. The old call passed a file path as the settings mapping, so the plugin's
    # `settings.get(...)` hit a str and raised `AttributeError: 'str' object has no attribute 'get'`.
    cmd = get_cmd({"overrides": overrides}, "mypy")
    
    print(f"✓ MyPy command generated")
    print(f"  Command type: {type(cmd).__name__}")
    print(f"  Command parts: {len(cmd)} arguments")
    print(f"  First arg: {cmd[0] if cmd else 'N/A'}")
    print(f"  Last arg: {os.path.basename(cmd[-1]) if cmd else 'N/A'}")
    
    # Cleanup
    os.unlink(temp_file)
    
    print("Example 4 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"pylsp_mypy example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```
