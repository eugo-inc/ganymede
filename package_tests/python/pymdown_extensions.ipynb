## Example 1: Fenced code blocks with syntax highlighting (superfences + highlight)

````python
import importlib.metadata

import markdown

# Example 1: superfences lets fenced code nest anywhere; highlight colours it with Pygments
assert importlib.metadata.version("pymdown-extensions") == "11.0.2"
source = "- item\n\n    ```python\n    print('hi')\n    ```\n"
html = markdown.markdown(source, extensions=["pymdownx.superfences", "pymdownx.highlight"])
assert "<li>" in html and "highlight" in html and "print" in html, html
print("Example 1 OK:", html[:120].replace("\n", " "))
````

## Example 2: Task lists render as checkboxes

````python
import markdown

# Example 2: GitHub-style [ ] and [x] items become disabled checkboxes
html = markdown.markdown("- [x] build\n- [ ] ship\n", extensions=["pymdownx.tasklist"])
assert html.count('type="checkbox"') == 2 and "checked" in html, html
print("Example 2 OK:", html.replace("\n", " "))
````

## Example 3: Math is kept for MathJax (arithmatex)

````python
import markdown

# Example 3: inline $...$ and block $$...$$ are wrapped so the math engine can find them untouched
html = markdown.markdown("Euler: $e^{i\\pi} + 1 = 0$\n\n$$\\int_0^1 x\\,dx$$", extensions=["pymdownx.arithmatex"],
                         extension_configs={"pymdownx.arithmatex": {"generic": True}})
assert 'class="arithmatex"' in html and "e^{i\\pi}" in html, html
print("Example 3 OK:", html.replace("\n", " ")[:160])
````

## Example 4: Tabs, details, marks and strike-through

````python
import markdown

# Example 4: several small syntax extensions together
source = '=== "A"\n    first\n\n=== "B"\n    second\n\n??? note "More"\n    hidden\n\n==marked== ~~gone~~\n'
html = markdown.markdown(source, extensions=["pymdownx.tabbed", "pymdownx.details", "pymdownx.mark", "pymdownx.tilde"])
assert "tabbed-set" in html and "<details" in html and "<mark>marked</mark>" in html and "<del>gone</del>" in html, html
print("Example 4 OK")
````

## Example 5: Emoji shortcodes and magic links

````python
import markdown
import pymdownx.emoji

# Example 5: :smile: becomes an emoji, a bare URL becomes a link
html = markdown.markdown("Hi :smile: see https://example.com", extensions=["pymdownx.emoji", "pymdownx.magiclink"],
                         extension_configs={"pymdownx.emoji": {"emoji_generator": pymdownx.emoji.to_alt}})
assert "\U0001f604" in html and 'href="https://example.com"' in html, html
print("Example 5 OK:", html)
````
