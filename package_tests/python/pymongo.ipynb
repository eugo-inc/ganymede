## Example 2: pymongo functionality test 2

```python
from pymongo import InsertOne, UpdateOne, DeleteOne, ReplaceOne
from pymongo.errors import ConnectionFailure, ServerSelectionTimeoutError

# Example 2: MongoDB operations and bulk operations
print("Testing PyMongo database operations:")

try:
    # Bulk operation classes
    print(f"✓ Bulk operation classes available:")
    operations = [
        ("InsertOne", InsertOne),
        ("UpdateOne", UpdateOne), 
        ("DeleteOne", DeleteOne),
        ("ReplaceOne", ReplaceOne)
    ]
    
    for op_name, op_class in operations:
        print(f"  {op_name}: {op_class}")
    
    # Create sample bulk operations (without executing)
    bulk_ops = []
    
    # Insert operation
    insert_op = InsertOne({"name": "Alice", "age": 30, "city": "New York"})
    bulk_ops.append(("insert", insert_op))
    
    # Update operation
    update_op = UpdateOne(
        {"name": "Alice"}, 
        {"$set": {"age": 31, "last_updated": "2024-01-01"}}
    )
    bulk_ops.append(("update", update_op))
    
    # Delete operation
    delete_op = DeleteOne({"name": "Alice"})
    bulk_ops.append(("delete", delete_op))
    
    print(f"✓ Created {len(bulk_ops)} bulk operations:")
    for op_type, operation in bulk_ops:
        print(f"  {op_type}: {type(operation).__name__}")
    
    # MongoDB query operators and syntax
    query_examples = [
        {"name": "Alice"},  # Equality
        {"age": {"$gte": 18}},  # Greater than or equal
        {"city": {"$in": ["New York", "Los Angeles"]}},  # In array
        {"$and": [{"age": {"$gte": 18}}, {"city": "New York"}]}  # Compound
    ]
    
    print(f"✓ MongoDB query patterns:")
    for i, query in enumerate(query_examples, 1):
        print(f"  Query {i}: {query}")
    
    print("Example 2 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    print(f"Error type: {type(e).__name__}")
    raise AssertionError(
        f"pymongo example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: pymongo functionality test 3

```python
from pymongo import ASCENDING, DESCENDING, IndexModel
from pymongo import ReadPreference, WriteConcern

# Example 3: Indexing, sorting, and database configuration
print("Testing PyMongo indexing and configuration:")

try:
    # Sort order constants
    print(f"✓ Sort order constants:")
    print(f"  ASCENDING: {ASCENDING}")
    print(f"  DESCENDING: {DESCENDING}")
    
    # Index creation examples
    indexes = [
        IndexModel([("name", ASCENDING)]),  # Single field index
        IndexModel([("age", DESCENDING)]),  # Descending index
        IndexModel([("name", ASCENDING), ("age", DESCENDING)]),  # Compound index
        IndexModel([("location", "2dsphere")]),  # Geospatial index
        IndexModel([("description", "text")])  # Text index
    ]
    
    print(f"✓ Index models created: {len(indexes)} indexes")
    for i, index in enumerate(indexes, 1):
        print(f"  Index {i}: {type(index).__name__}")
    
    # Read preferences for replica sets
    read_prefs = [
        ReadPreference.PRIMARY,
        ReadPreference.PRIMARY_PREFERRED,
        ReadPreference.SECONDARY,
        ReadPreference.SECONDARY_PREFERRED,
        ReadPreference.NEAREST
    ]
    
    print(f"✓ Read preference options:")
    for pref in read_prefs:
        print(f"  {pref.name}: {pref.mode}")
    
    # Write concern levels
    write_concerns = [
        WriteConcern(w=1),  # Acknowledge from primary
        WriteConcern(w="majority"),  # Majority acknowledgment
        WriteConcern(w=1, j=True),  # With journal
        WriteConcern(w=1, fsync=True)  # With fsync
    ]
    
    print(f"✓ Write concern configurations: {len(write_concerns)}")
    for i, wc in enumerate(write_concerns, 1):
        # API drift: `WriteConcern` no longer exposes per-option attributes like `.journal`;
        # `.document` is the authoritative mapping of what was actually set.
        print(f"  Config {i}: w={wc.acknowledged}, j={wc.document.get('j')}")
    
    # Aggregation pipeline examples
    pipeline_stages = [
        {"$match": {"age": {"$gte": 18}}},
        {"$group": {"_id": "$city", "count": {"$sum": 1}, "avg_age": {"$avg": "$age"}}},
        {"$sort": {"count": -1}},
        {"$limit": 10}
    ]
    
    print(f"✓ Aggregation pipeline stages: {len(pipeline_stages)}")
    for i, stage in enumerate(pipeline_stages, 1):
        stage_name = list(stage.keys())[0]
        print(f"  Stage {i}: {stage_name}")
    
    print("Example 3 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    print(f"Error type: {type(e).__name__}")
    raise AssertionError(
        f"pymongo example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: pymongo functionality test 4

```python
from pymongo.errors import *
from pymongo import MongoClient
import pymongo.monitoring

# Example 4: Error handling and monitoring
print("Testing PyMongo error handling and monitoring:")

try:
    # PyMongo exception hierarchy
    exception_classes = [
        ConnectionFailure,
        ServerSelectionTimeoutError,
        NetworkTimeout,
        AutoReconnect,
        ConfigurationError,
        OperationFailure,
        DuplicateKeyError,
        BulkWriteError,
        WriteConcernError,
        WriteError
    ]
    
    print(f"✓ Available exception classes: {len(exception_classes)}")
    for exc_class in exception_classes[:5]:  # Show first 5
        print(f"  {exc_class.__name__}: {exc_class}")
    print(f"  ... and {len(exception_classes) - 5} more")
    
    # Test error scenarios (without actual server). The "no server" case targets a loopback port
    # guaranteed closed (bind an ephemeral port, read it back, close it) rather than a made-up
    # hostname, so it needs no DNS and cannot reach any real server.
    import socket
    with socket.socket() as probe:
        probe.bind(("127.0.0.1", 0))
        closed_port = probe.getsockname()[1]
    error_scenarios = [
        ("Connection timeout", lambda: MongoClient(f"mongodb://127.0.0.1:{closed_port}", serverSelectionTimeoutMS=100).server_info(), ServerSelectionTimeoutError),
        ("Invalid connection string", lambda: MongoClient("invalid://connection"), InvalidURI),
    ]
    
    print(f"✓ Testing error scenarios:")
    for scenario_name, scenario_func, expected_error in error_scenarios:
        try:
            scenario_func()
        except expected_error as e:
            print(f"  {scenario_name}: {type(e).__name__} (expected)")
        else:
            raise AssertionError(f"{scenario_name}: expected {expected_error.__name__}, got success")
    
    # Monitoring and event system
    print(f"✓ Monitoring system available:")
    monitoring_classes = [
        "CommandListener",
        "ServerListener", 
        "ServerHeartbeatListener",
        "TopologyListener",
        "ConnectionPoolListener"
    ]
    
    for listener_name in monitoring_classes:
        if hasattr(pymongo.monitoring, listener_name):
            listener_class = getattr(pymongo.monitoring, listener_name)
            print(f"  {listener_name}: {listener_class}")
    
    # Client options for production deployments
    production_options = {
        'serverSelectionTimeoutMS': 30000,
        'connectTimeoutMS': 20000,
        'socketTimeoutMS': 20000,
        'heartbeatFrequencyMS': 10000,
        'retryWrites': True,
        'retryReads': True,
        'maxPoolSize': 100,
        'minPoolSize': 10
    }
    
    print(f"✓ Production client options: {len(production_options)}")
    for option, value in list(production_options.items())[:3]:
        print(f"  {option}: {value}")
    print(f"  ... and {len(production_options) - 3} more options")
    
    print("Example 4 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    print(f"Error type: {type(e).__name__}")
    raise AssertionError(
        f"pymongo example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```
