## Example 1: pymssql functionality test 1

```python
import pymssql

# Example 1: Basic pymssql package information
print("Testing pymssql SQL Server database driver:")

# Package version and DB-API compliance
print(f"✓ pymssql version: {pymssql.__version__}")
print(f"✓ DB-API level: {pymssql.apilevel}")
print(f"✓ Thread safety: {pymssql.threadsafety}")
print(f"✓ Parameter style: {pymssql.paramstyle}")

# FreeTDS version (underlying C library)
freetds_version = pymssql.get_freetds_version()
print(f"✓ FreeTDS version: {freetds_version}")

# Connection configuration
max_conn = pymssql.get_max_connections()
print(f"✓ Max connections: {max_conn}")

# Set max connections example
pymssql.set_max_connections(8192)
new_max = pymssql.get_max_connections()
print(f"✓ Updated max connections: {new_max}")

# Core classes
print(f"✓ Connection class: {pymssql.Connection}")
print(f"✓ Cursor class: {pymssql.Cursor}")
print(f"✓ Connect function: {type(pymssql.connect).__name__}")
```

## Example 2: pymssql functionality test 2

```python
from pymssql import STRING, BINARY, NUMBER, DATETIME, DECIMAL
from pymssql import Date, Time, Timestamp, Binary
import datetime

# Example 2: Data type objects and constructors
print("Testing pymssql data types:")

# DB-API 2.0 type objects
print(f"✓ STRING: {STRING}")
print(f"✓ BINARY: {BINARY}")
print(f"✓ NUMBER: {NUMBER}")
print(f"✓ DATETIME: {DATETIME}")
print(f"✓ DECIMAL: {DECIMAL}")

# Type constructors
test_date = Date(2024, 12, 25)
print(f"✓ Date(2024, 12, 25): {test_date}")

test_time = Time(14, 30, 45)
print(f"✓ Time(14, 30, 45): {test_time}")

test_timestamp = Timestamp(2024, 12, 25, 14, 30, 45)
print(f"✓ Timestamp: {test_timestamp}")

test_binary = Binary(b"binary data")
print(f"✓ Binary: {len(test_binary)} bytes, type: {type(test_binary).__name__}")
```

## Example 3: pymssql functionality test 3

```python
from pymssql import (Error, Warning, InterfaceError, DatabaseError, 
                     DataError, OperationalError, IntegrityError, 
                     InternalError, ProgrammingError, NotSupportedError)

# Example 3: Exception hierarchy
print("Testing pymssql exception classes:")

# All exception types
exceptions = [
    ("Error", Error),
    ("Warning", Warning),
    ("InterfaceError", InterfaceError),
    ("DatabaseError", DatabaseError),
    ("DataError", DataError),
    ("OperationalError", OperationalError),
    ("IntegrityError", IntegrityError),
    ("InternalError", InternalError),
    ("ProgrammingError", ProgrammingError),
    ("NotSupportedError", NotSupportedError)
]

for name, exc_class in exceptions:
    print(f"✓ {name}: {exc_class}")

# Test exception creation
test_error = Error("Test database error")
print(f"✓ Created error: {test_error}")
print(f"✓ Error base class: {Error.__bases__[0].__name__}")
```

## Example 4: pymssql functionality test 4

```python
from pymssql import output

# Example 4: Stored procedure output parameters
print("Testing pymssql stored procedure support:")

# Output parameter helper
print(f"✓ output function: {output}")
print(f"  Type: {type(output).__name__}")

# Example of using output parameters
param_value = output(int)
print(f"✓ output(int): {param_value}")
print(f"  Type: {type(param_value)}")

param_str = output(str, "default")
print(f"✓ output(str, 'default'): {param_str}")

# Example SQL with output parameters
sql_example = "EXEC GetUserCount @category=%s, @count=%s OUTPUT"
print(f"✓ Example SQL with OUTPUT: {sql_example}")

# Parameter styles
print(f"✓ Parameter styles:")
print(f"  Named: WHERE name = %(name)s")
print(f"  Positional: WHERE id = %s")
print(f"  Output: @result = %s OUTPUT")

# Cursor methods for stored procedures
print(f"✓ Cursor methods:")
print(f"  execute() - Run SQL/stored procedure")
print(f"  callproc() - Call stored procedure")
print(f"  nextset() - Move to next result set")
```
