## Example 2: pynacl functionality test 2

```python
from nacl.public import PrivateKey, Box
from nacl.encoding import Base64Encoder

# Example 2: Public key cryptography with Curve25519
print("Testing PyNaCl public key cryptography:")

# Generate key pairs for Alice and Bob
alice_private = PrivateKey.generate()
alice_public = alice_private.public_key

bob_private = PrivateKey.generate()
bob_public = bob_private.public_key

print(f"✓ Generated key pairs")
print(f"  Alice private key: {len(alice_private.encode())} bytes")
print(f"  Bob public key: {len(bob_public.encode())} bytes")

# Create encryption boxes
alice_box = Box(alice_private, bob_public)
bob_box = Box(bob_private, alice_public)

print(f"✓ Created encryption boxes: {type(alice_box).__name__}")

# Encrypt and decrypt message
message = b"Hello Bob! This is a secret message."
encrypted = alice_box.encrypt(message)
decrypted = bob_box.decrypt(encrypted)
assert decrypted == message, decrypted

print(f"✓ Encrypted {len(message)} bytes to {len(encrypted)} bytes")
print(f"✓ Decrypted: {decrypted.decode()}")
print(f"✓ Decryption successful: {decrypted == message}")
```

## Example 3: pynacl functionality test 3

```python
from nacl.secret import SecretBox
from nacl.utils import random
from nacl.encoding import HexEncoder

# Example 3: Secret key cryptography with XSalsa20-Poly1305
print("Testing PyNaCl secret key cryptography:")

try:
    # Generate a random 32-byte secret key
    secret_key = random(SecretBox.KEY_SIZE)
    print(f"✓ Generated secret key:")
    print(f"  Key size: {len(secret_key)} bytes ({SecretBox.KEY_SIZE} required)")
    print(f"  Key (hex): {secret_key.hex()}")
    
    # Create a SecretBox for symmetric encryption
    box = SecretBox(secret_key)
    print(f"✓ Created SecretBox: {type(box)}")
    
    # Encrypt multiple messages
    messages = [
        b"This is a confidential document.",
        b"Account number: 1234567890",
        b"Password: super_secret_123!",
        b"Credit card: 4111-1111-1111-1111"
    ]
    
    encrypted_messages = []
    
    print(f"✓ Encrypting {len(messages)} messages:")
    for i, message in enumerate(messages):
        encrypted = box.encrypt(message)
        encrypted_messages.append(encrypted)
        
        print(f"  Message {i+1}:")
        print(f"    Original: {len(message)} bytes")
        print(f"    Encrypted: {len(encrypted)} bytes")
        print(f"    Ciphertext: {encrypted.ciphertext[:16].hex()}...")
        print(f"    Nonce: {encrypted.nonce.hex()}")
    
    # Decrypt all messages
    print(f"✓ Decrypting messages:")
    for i, encrypted in enumerate(encrypted_messages):
        decrypted = box.decrypt(encrypted)
        assert decrypted == messages[i], (i, decrypted)
        print(f"  Message {i+1} decrypted: {decrypted.decode()}")
        print(f"    Integrity verified: {decrypted == messages[i]}")
    
    # Demonstrate nonce uniqueness
    message = b"Test message for nonce demonstration"
    encryptions = [box.encrypt(message) for _ in range(3)]
    
    print(f"✓ Nonce uniqueness demonstration:")
    nonces = [enc.nonce for enc in encryptions]
    assert len(set(nonces)) == len(nonces), [nonce.hex() for nonce in nonces]
    print(f"  Three encryptions of same message have different nonces:")
    for i, nonce in enumerate(nonces):
        print(f"    Nonce {i+1}: {nonce.hex()}")
    print(f"  All nonces unique: {len(set(nonces)) == len(nonces)}")
    
    # Security properties
    print(f"✓ Security properties:")
    print(f"  - XSalsa20 stream cipher for confidentiality")
    print(f"  - Poly1305 MAC for authenticity and integrity")
    print(f"  - 192-bit nonces prevent replay attacks")
    print(f"  - Constant-time operations resist timing attacks")
    
    print("Example 3 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    print(f"Error type: {type(e).__name__}")
    raise AssertionError(
        f"pynacl example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: pynacl functionality test 4

```python
from nacl.signing import SigningKey
from nacl.encoding import HexEncoder
from nacl.exceptions import BadSignatureError

# Example 4: Digital signatures with Ed25519
print("Testing PyNaCl digital signatures:")

# Generate signing key pair
signing_key = SigningKey.generate()
verify_key = signing_key.verify_key

print(f"✓ Generated Ed25519 key pair:")
print(f"  Signing key: {len(signing_key.encode())} bytes")
print(f"  Verify key: {len(verify_key.encode())} bytes")

# Sign messages
messages = [b"Important contract", b"Transaction #123"]

for i, message in enumerate(messages, 1):
    signed = signing_key.sign(message)
    print(f"✓ Signed message {i}: {len(signed)} bytes")
    
    # Verify signature
    verify_key.verify(signed)
    print(f"  Verification: Valid ✓")

# Test tampered message detection
try:
    verify_key.verify(b"Tampered message", signed.signature)
except BadSignatureError:
    print(f"✓ Tampering detected correctly")
else:
    raise AssertionError("a tampered message verified against another message's signature")
```

## Example 5: pynacl functionality test 5

```python
from nacl.hash import sha256, blake2b, generichash
from nacl.utils import random
from nacl.pwhash import argon2i

# Example 5: Cryptographic hashing and key derivation
print("Testing PyNaCl hashing and key derivation:")

# Cryptographic hash functions
data = b"Sensitive data that needs integrity"

# SHA-256 hashing
sha256_hash = sha256(data)
print(f"✓ SHA-256: {sha256_hash[:16].hex()}...")

# BLAKE2b hashing
blake2b_hash = blake2b(data, digest_size=32)
print(f"✓ BLAKE2b: {blake2b_hash[:16].hex()}...")

# Generic hash with key
key = random(32)
keyed_hash = generichash(data, key=key, digest_size=64)
print(f"✓ Keyed BLAKE2b: {keyed_hash[:16].hex()}...")

# Password hashing with Argon2
password = b"user_password_123"
salt = random(argon2i.SALTBYTES)

# Hash password
pwhash = argon2i.kdf(32, password, salt)
print(f"✓ Argon2i password hash: {pwhash[:16].hex()}...")
print(f"  Hash length: {len(pwhash)} bytes")

print("Example 5 completed successfully")
```
