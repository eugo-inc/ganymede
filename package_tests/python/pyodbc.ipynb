## Example 1: pyodbc functionality test 1

```python
import pyodbc

# Example 1: Basic PyODBC interface information
print("Testing PyODBC ODBC database interface:")

# Version and DB-API compliance
print(f"✓ PyODBC version: {pyodbc.version}")
print(f"✓ API level: {pyodbc.apilevel}")
print(f"✓ Thread safety: {pyodbc.threadsafety}")
print(f"✓ Parameter style: {pyodbc.paramstyle}")

# Available ODBC drivers
drivers = pyodbc.drivers()
print(f"✓ Available ODBC drivers: {len(drivers)}")
for driver in drivers[:3]:
    print(f"  {driver}")

# Data type constants
print(f"✓ Data types:")
print(f"  STRING: {pyodbc.STRING}")
print(f"  BINARY: {pyodbc.BINARY}")
print(f"  NUMBER: {pyodbc.NUMBER}")
print(f"  DATETIME: {pyodbc.DATETIME}")
print(f"  ROWID: {pyodbc.ROWID}")
```

## Example 2: pyodbc functionality test 2

```python
import pyodbc

# Example 2: Connection string building
print("Testing PyODBC connection string patterns:")

# Connection string examples
conn_strings = {
    "SQL Server": "DRIVER={ODBC Driver 17 for SQL Server};SERVER=localhost;DATABASE=test;",
    "PostgreSQL": "DRIVER={PostgreSQL UNICODE};SERVER=localhost;DATABASE=test;",
    "MySQL": "DRIVER={MySQL ODBC 8.0 Driver};SERVER=localhost;DATABASE=test;",
    "SQLite": "DRIVER={SQLite3 ODBC Driver};DATABASE=test.db;"
}

print(f"✓ Connection string patterns:")
for db, conn_str in conn_strings.items():
    print(f"  {db}: {conn_str[:50]}...")

# Parameter placeholder style
print(f"✓ Parameter style: {pyodbc.paramstyle}")
print(f"  Example: SELECT * FROM users WHERE id = ?")

# Connection and Cursor classes
print(f"✓ Connection class: {pyodbc.Connection}")
print(f"✓ Cursor class: {pyodbc.Cursor}")

# Row class for result rows
print(f"✓ Row class: {pyodbc.Row}")
```

## Example 3: pyodbc functionality test 3

```python
import pyodbc

# Example 3: ODBC SQL data type constants
print("Testing PyODBC ODBC data type constants:")

# SQL data type constants that exist in pyodbc
sql_types = [
    ('SQL_CHAR', pyodbc.SQL_CHAR),
    ('SQL_VARCHAR', pyodbc.SQL_VARCHAR),
    ('SQL_WCHAR', pyodbc.SQL_WCHAR),
    ('SQL_WVARCHAR', pyodbc.SQL_WVARCHAR),
    ('SQL_DECIMAL', pyodbc.SQL_DECIMAL),
    ('SQL_INTEGER', pyodbc.SQL_INTEGER),
    ('SQL_FLOAT', pyodbc.SQL_FLOAT),
    ('SQL_DOUBLE', pyodbc.SQL_DOUBLE),
    ('SQL_TYPE_DATE', pyodbc.SQL_TYPE_DATE),
    ('SQL_TYPE_TIME', pyodbc.SQL_TYPE_TIME),
    ('SQL_TYPE_TIMESTAMP', pyodbc.SQL_TYPE_TIMESTAMP)
]

print(f"✓ SQL data type constants:")
for name, value in sql_types:
    print(f"  {name}: {value}")

# Python type mappings
print(f"✓ Python type mappings:")
print(f"  str → VARCHAR: {pyodbc.STRING}")
print(f"  int → INTEGER: {pyodbc.NUMBER}")
print(f"  bytes → BINARY: {pyodbc.BINARY}")
```

## Example 4: pyodbc functionality test 4

```python
from pyodbc import (Error, InterfaceError, DatabaseError, DataError,
                     OperationalError, IntegrityError, InternalError,
                     ProgrammingError, NotSupportedError)

# Example 4: Exception hierarchy
print("Testing PyODBC exception classes:")

# All exception types
exceptions = [
    ("Error", Error),
    ("InterfaceError", InterfaceError),
    ("DatabaseError", DatabaseError),
    ("DataError", DataError),
    ("OperationalError", OperationalError),
    ("IntegrityError", IntegrityError),
    ("InternalError", InternalError),
    ("ProgrammingError", ProgrammingError),
    ("NotSupportedError", NotSupportedError)
]

print(f"✓ Exception hierarchy:")
for name, exc_class in exceptions:
    print(f"  {name}: {exc_class}")

# Test exception creation
test_error = Error("Test database error")
print(f"✓ Created error: {test_error}")
print(f"✓ Error base: {Error.__bases__[0].__name__}")
```
