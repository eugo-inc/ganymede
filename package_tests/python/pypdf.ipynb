## Example 1: Basic PDF Reading and Information

```python
from pypdf import PdfReader, PdfWriter
from io import BytesIO

# Example 1: Create and read PDF
print("Testing pypdf basic PDF operations:")

# Create a simple PDF
writer = PdfWriter()
writer.add_blank_page(width=200, height=200)
writer.add_blank_page(width=300, height=300)

# Write to BytesIO
pdf_bytes = BytesIO()
writer.write(pdf_bytes)
pdf_bytes.seek(0)

# Read the PDF
reader = PdfReader(pdf_bytes)
print(f"✓ Created PDF with {len(writer.pages)} pages")
print(f"✓ Read PDF with {len(reader.pages)} pages")
print(f"✓ Page 1 dimensions: {reader.pages[0].mediabox.width} x {reader.pages[0].mediabox.height}")
print(f"✓ Page 2 dimensions: {reader.pages[1].mediabox.width} x {reader.pages[1].mediabox.height}")
print(f"✓ PDF metadata: {reader.metadata if reader.metadata else 'None'}")
```

## Example 2: PDF Merging and Splitting

```python
from pypdf import PdfWriter, PdfReader
from io import BytesIO

# Example 2: Merge and split PDFs
print("Testing pypdf merge and split operations:")

# Create two PDFs
pdf1 = PdfWriter()
pdf1.add_blank_page(width=200, height=200)
pdf1.add_blank_page(width=200, height=200)

pdf2 = PdfWriter()
pdf2.add_blank_page(width=300, height=300)

# Write to BytesIO
stream1 = BytesIO()
pdf1.write(stream1)
stream2 = BytesIO()
pdf2.write(stream2)

# Merge PDFs
merger = PdfWriter()
merger.append(BytesIO(stream1.getvalue()))
merger.append(BytesIO(stream2.getvalue()))

print(f"✓ Merged {len(merger.pages)} pages from 2 PDFs")
print(f"✓ Page sizes: {[f'{p.mediabox.width}x{p.mediabox.height}' for p in merger.pages]}")
```

## Example 3: Page Manipulation and Transformations

```python
from pypdf import PdfWriter, PdfReader, Transformation
from io import BytesIO

# Example 3: Page transformations
print("Testing pypdf page transformations:")

# Create PDF with page
writer = PdfWriter()
page = writer.add_blank_page(width=200, height=200)

# Rotate page
page.rotate(90)
print(f"✓ Rotated page 90 degrees")

# Scale page
page.scale(2, 2)
print(f"✓ Scaled page 2x")

# Create transformation
transform = Transformation().translate(tx=10, ty=20).rotate(45)
print(f"✓ Created transformation: translate(10,20) + rotate(45)")

# Merge rotated page
page2 = writer.add_blank_page(width=300, height=300)
page2.merge_transformed_page(page, transform)
print(f"✓ Merged transformed page")
print(f"✓ Total pages: {len(writer.pages)}")
```

## Example 4: Text Extraction and Metadata

```python
from pypdf import PdfReader, PdfWriter
from io import BytesIO

# Example 4: Metadata and text extraction
print("Testing pypdf metadata and text operations:")

# Create PDF with metadata
writer = PdfWriter()
page = writer.add_blank_page(width=200, height=200)

# Add metadata
writer.add_metadata({
    '/Title': 'Test Document',
    '/Author': 'PyPDF Test',
    '/Subject': 'PDF Testing',
    '/Creator': 'pypdf example'
})

# Read back metadata
pdf_bytes = BytesIO()
writer.write(pdf_bytes)
pdf_bytes.seek(0)

reader = PdfReader(pdf_bytes)
print(f"✓ Title: {reader.metadata.title}")
print(f"✓ Author: {reader.metadata.author}")
print(f"✓ Subject: {reader.metadata.subject}")
print(f"✓ Creator: {reader.metadata.creator}")
print(f"✓ Page text: '{reader.pages[0].extract_text()}'")
```

## Example 5: Advanced PDF Operations

```python
from pypdf import PdfWriter, PdfReader
from io import BytesIO

# Example 5: Encryption and page operations
print("Testing pypdf encryption and advanced operations:")

# Create PDF
writer = PdfWriter()
writer.add_blank_page(width=200, height=200)
writer.add_blank_page(width=300, height=300)

# Encrypt PDF
writer.encrypt(user_password="user123", owner_password="owner456")
print(f"✓ Encrypted PDF with user and owner passwords")

# Write and read encrypted PDF
pdf_bytes = BytesIO()
writer.write(pdf_bytes)
pdf_bytes.seek(0)

reader = PdfReader(pdf_bytes, password="user123")
print(f"✓ Opened encrypted PDF: {len(reader.pages)} pages")
print(f"✓ Is encrypted: {reader.is_encrypted}")

# Clone and remove page
writer2 = PdfWriter(clone_from=reader)
writer2.remove_page(1)
print(f"✓ Cloned PDF and removed page: {len(writer2.pages)} pages remaining")
```
