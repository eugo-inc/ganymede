## Example 1: pyproj functionality test 1

```python
from pyproj import CRS, Proj
import pyproj

# Example 1: CRS creation and inspection
print("Testing pyproj CRS operations:")

# Create CRS from different sources
crs_epsg = CRS.from_epsg(4326)
crs_string = CRS.from_string("EPSG:3857")
crs_wkt = CRS.from_user_input("+proj=utm +zone=10 +datum=WGS84")

print(f"✓ Created CRS from EPSG: {crs_epsg.name}")
print(f"✓ Created CRS from string: {crs_string.name}")
print(f"✓ Created CRS from PROJ: {crs_wkt.name}")

# CRS properties
print(f"✓ CRS type: {crs_epsg.type_name}")
print(f"✓ Authority: {crs_epsg.to_authority()}")
print(f"✓ Axes: {[ax.name for ax in crs_epsg.axis_info]}")
print(f"✓ Is geographic: {crs_epsg.is_geographic}")
print(f"✓ Is projected: {crs_string.is_projected}")
print(f"✓ Area of use: {crs_epsg.area_of_use.name if crs_epsg.area_of_use else 'None'}")
```

## Example 2: pyproj functionality test 2

```python
from pyproj import Transformer, CRS

# Example 2: Coordinate transformations
print("Testing pyproj coordinate transformations:")

# Create transformer between WGS84 and Web Mercator
crs_src = CRS.from_epsg(4326)
crs_dst = CRS.from_epsg(3857)
transformer = Transformer.from_crs(crs_src, crs_dst, always_xy=True)

# Transform single point (San Francisco)
lon, lat = -122.4194, 37.7749
x, y = transformer.transform(lon, lat)
print(f"✓ Transformed ({lon}, {lat}) -> ({x:.2f}, {y:.2f})")

# Inverse transformation
lon_back, lat_back = transformer.transform(x, y, direction='INVERSE')
print(f"✓ Inverse ({x:.2f}, {y:.2f}) -> ({lon_back:.6f}, {lat_back:.6f})")

# Batch transformation
lons = [-122.4194, -73.9352, -87.6298]
lats = [37.7749, 40.7306, 41.8781]
xs, ys = transformer.transform(lons, lats)
print(f"✓ Batch transformed {len(lons)} points")
print(f"✓ X range: {min(xs):.0f} to {max(xs):.0f}")
```

## Example 3: pyproj functionality test 3

```python
from pyproj import Geod

# Example 3: Geodesic calculations
print("Testing pyproj geodesic operations:")

# Create geodesic calculator
geod = Geod(ellps='WGS84')
print(f"✓ Created WGS84 geodesic")
print(f"✓ Semi-major axis: {geod.a:.3f} m")
print(f"✓ Flattening: {geod.f:.10f}")

# Calculate distance and azimuth between cities
# San Francisco to New York
lon1, lat1 = -122.4194, 37.7749
lon2, lat2 = -74.0060, 40.7128
forward_az, back_az, distance = geod.inv(lon1, lat1, lon2, lat2)

print(f"✓ SF to NYC distance: {distance/1000:.1f} km")
print(f"✓ Forward azimuth: {forward_az:.1f}°")
print(f"✓ Back azimuth: {back_az:.1f}°")

# Calculate polygon area
lons = [-122.5, -122.0, -122.0, -122.5, -122.5]
lats = [37.5, 37.5, 38.0, 38.0, 37.5]
area, perimeter = geod.polygon_area_perimeter(lons, lats)
print(f"✓ Polygon area: {abs(area)/1e6:.1f} km²")
print(f"✓ Perimeter: {perimeter/1000:.1f} km")
```

## Example 4: pyproj functionality test 4

```python
from pyproj import CRS, Proj

# Example 4: Proj and CRS conversions
print("Testing pyproj Proj and CRS conversions:")

# Create Proj from different definitions
proj_merc = Proj(proj='merc', lat_0=0, lon_0=0)
proj_utm = Proj(proj='utm', zone=10, ellps='WGS84')
proj_lcc = Proj(proj='lcc', lat_1=33, lat_2=45, lat_0=39, lon_0=-96)

print(f"✓ Created Mercator projection: {proj_merc.crs.name}")
print(f"✓ Created UTM projection: {proj_utm.crs.name}")
print(f"✓ Created LCC projection: {proj_lcc.crs.name}")

# Convert CRS to different formats
crs = CRS.from_epsg(4326)
print(f"✓ To WKT: {len(crs.to_wkt())} chars")
print(f"✓ To PROJ4: {crs.to_proj4()}")
print(f"✓ To JSON: {len(crs.to_json())} chars")

# CRS equality and comparison
crs1 = CRS.from_epsg(4326)
crs2 = CRS.from_string("EPSG:4326")
crs3 = CRS.from_epsg(3857)
print(f"✓ EPSG:4326 == 'EPSG:4326': {crs1 == crs2}")
print(f"✓ EPSG:4326 == EPSG:3857: {crs1 == crs3}")
```

## Example 5: pyproj functionality test 5

```python
from pyproj import CRS, Transformer, get_codes

# Example 5: Database queries and transformation pipelines
print("Testing pyproj database and pipeline operations:")

# Query EPSG database
epsg_codes = get_codes('EPSG', 'CRS', allow_deprecated=False)
print(f"✓ EPSG CRS codes: {len(epsg_codes)} available")
print(f"✓ Sample codes: {epsg_codes[:5]}")

# Get transformation operations
crs_src = CRS.from_epsg(4326)
crs_dst = CRS.from_epsg(3857)
ops = Transformer.from_crs(crs_src, crs_dst).operations
print(f"✓ Transformation operations: {ops}")

# Create pipeline transformer
pipeline = "+proj=pipeline +step +proj=unitconvert +xy_in=deg +xy_out=rad +step +proj=merc"
transformer = Transformer.from_pipeline(pipeline)
x, y = transformer.transform(-122.4194, 37.7749)
print(f"✓ Pipeline transform: ({x:.2f}, {y:.2f})")

# CRS coordinate operation
crs = CRS.from_epsg(32610)
print(f"✓ CRS coordinate operation: {crs.coordinate_operation}")
```
