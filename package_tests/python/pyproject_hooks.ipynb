```python
# Example 1: Basic Build Backend Detection and Configuration
import pyproject_hooks
import tempfile
import os
import json

# Create mock pyproject.toml content for demonstration
mock_pyproject_content = """
[build-system]
requires = ["setuptools >= 45", "wheel", "setuptools_scm[toml] >= 6.2"]
build-backend = "setuptools.build_meta"

[project]
name = "example-package"
version = "0.1.0"
description = "Example package for pyproject_hooks demonstration"
authors = [{name = "Test Author", email = "test@example.com"}]
"""

# Create temporary directory with pyproject.toml
temp_dir = tempfile.mkdtemp()
pyproject_path = os.path.join(temp_dir, "pyproject.toml")

with open(pyproject_path, "w") as f:
    f.write(mock_pyproject_content)

print("Build Backend Detection and Configuration:")
print("=" * 45)

try:
    # Create BuildBackendHookCaller for the project
    backend = pyproject_hooks.BuildBackendHookCaller(
        source_dir=temp_dir,
        build_backend="setuptools.build_meta"
    )
    
    print(f"Backend created successfully")
    print(f"Source directory: {temp_dir}")
    print(f"Build backend: setuptools.build_meta")
    
    # Get backend capabilities
    print(f"\nBuild backend capabilities:")
    print(f"  Subprocess runner: {hasattr(backend, '_subprocess_runner')}")
    print(f"  Hook calling support: {hasattr(backend, 'get_requires_for_build_wheel')}")
    
    # Mock configuration info
    build_config = {
        "backend": "setuptools.build_meta", 
        "requires": ["setuptools >= 45", "wheel", "setuptools_scm[toml] >= 6.2"],
        "source_dir": temp_dir,
        "isolation": True
    }
    
    print(f"\nBuild Configuration:")
    for key, value in build_config.items():
        print(f"  {key}: {value}")
        
except Exception as e:
    print(f"Backend detection completed with note: {str(e)[:100]}")
    print("Build backend detection and configuration demonstrated")
    raise

# Cleanup
try:
    os.unlink(pyproject_path)
    os.rmdir(temp_dir)
except:
    pass
    raise

print("\nBuild backend detection and configuration completed")
```
