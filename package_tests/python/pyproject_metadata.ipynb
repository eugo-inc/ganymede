```python
# Example 3: Error Handling and Metadata Validation
from pyproject_metadata import StandardMetadata
from pyproject_metadata.errors import ConfigurationError

print("Error Handling and Metadata Validation:")
print("=" * 40)

# Test various invalid metadata scenarios
invalid_metadata_cases = [
    {
        "name": "Missing required name field",
        "data": {
            "project": {
                "version": "1.0.0",
                "description": "Package without name"
            }
        },
        "expected_error": 'Field "project.name" is required'
    },
    {
        "name": "Invalid version format", 
        "data": {
            "project": {
                "name": "test-package",
                "version": "not-a-version",
                "description": "Package with invalid version"
            }
        },
        "expected_error": "is an invalid PEP 440 version string"
    },
    {
        "name": "Invalid Python version requirement",
        "data": {
            "project": {
                "name": "test-package",
                "version": "1.0.0", 
                "requires-python": "invalid-version",
                "description": "Package with invalid Python requirement"
            }
        },
        "expected_error": "is an invalid Python version specifier string"
    }
]

validation_results = []

for case in invalid_metadata_cases:
    print(f"\nTesting: {case['name']}")
    try:
        StandardMetadata.from_pyproject(case['data'], project_dir=".")
    except ConfigurationError as e:
        error_msg = str(e)
        assert case['expected_error'] in error_msg, f"{case['name']}: unexpected message {error_msg!r}"
        print(f"  ✗ Validation failed as expected: {error_msg[:60]}...")
        validation_results.append(("FAIL", case['name']))
    else:
        raise AssertionError(f"{case['name']}: from_pyproject accepted invalid metadata {case['data']!r}")

# pyproject-metadata does not validate the syntax of author e-mail addresses; the value is written through as given
print(f"\nTesting: Author e-mail is passed through unvalidated")
email_data = {"project": {"name": "test-package", "version": "1.0.0", "authors": [{"name": "Test", "email": "invalid-email"}]}}
email_metadata = StandardMetadata.from_pyproject(email_data, project_dir=".")
assert email_metadata.authors == [("Test", "invalid-email")], email_metadata.authors
assert "Author-Email: Test <invalid-email>" in str(email_metadata.as_rfc822()), str(email_metadata.as_rfc822())
print(f"  ✓ Accepted as written: {email_metadata.authors}")

# Test valid metadata
print(f"\nTesting valid metadata:")
try:
    valid_data = {
        "project": {
            "name": "valid-package",
            "version": "1.0.0",
            "description": "A valid package for testing"
        }
    }
    metadata = StandardMetadata.from_pyproject(valid_data, project_dir=".")
    print(f"  ✓ Valid metadata processed successfully")
    print(f"  Package: {metadata.name} v{metadata.version}")
    validation_results.append(("PASS", "Valid metadata"))
except ConfigurationError as e:
    raise AssertionError(f"valid metadata was rejected: {e}") from e

# Test all_errors parameter
print(f"\nTesting all_errors parameter:")
try:
    invalid_multi_error = {
        "project": {
            "name": "test",
            "version": "bad-version",
            "requires-python": "bad-python-req"
        }
    }
    metadata = StandardMetadata.from_pyproject(invalid_multi_error, project_dir=".", all_errors=True)
except ExceptionGroup as e:
    # all_errors=True collects every field problem into one ExceptionGroup of ConfigurationErrors
    error_type = type(e).__name__
    print(f"  ✓ Multiple errors caught: {error_type}")
    print(f"  Collected problems: {len(e.exceptions)}")
    assert [type(x).__name__ for x in e.exceptions] == ["ConfigurationError", "ConfigurationError"], e.exceptions
    assert "project.version" in str(e.exceptions[0]) and "project.requires-python" in str(e.exceptions[1])
else:
    raise AssertionError("all_errors=True accepted an invalid version and requires-python")

print(f"\nValidation Summary:")
passes = sum(1 for r, _ in validation_results if r == "PASS")
fails = sum(1 for r, _ in validation_results if r == "FAIL")
print(f"  Passed: {passes}, Failed (expected): {fails}")
assert (passes, fails) == (1, len(invalid_metadata_cases)), validation_results
    
print("\nError handling and validation testing completed")
```

```python
# Example 4: Dynamic Metadata Construction and Build System Integration
from pyproject_metadata import StandardMetadata
import tempfile
import os

# Dynamic metadata construction for different project types
project_templates = {
    "library": {
        "project": {
            "name": "data-utils", 
            "version": "1.5.0",
            "description": "Utility library for data processing and analysis",
            "authors": [{"name": "Project Team", "email": "team@example.com"}],
            "requires-python": ">=3.8",
            "dependencies": ["numpy>=1.20.0", "pandas>=1.3.0", "requests>=2.25.0"],
            "optional-dependencies": {
                "dev": ["pytest>=6.0.0", "black>=21.0.0"],
                "plotting": ["matplotlib>=3.3.0", "seaborn>=0.11.0"]
            },
            "classifiers": [
                "Development Status :: 4 - Beta",
                "Programming Language :: Python :: 3",
                "Programming Language :: Python :: 3.8"
            ]
        }
    },
    "cli_tool": {
        "project": {
            "name": "deploy-manager",
            "version": "2.0.0", 
            "description": "Command-line deployment management tool",
            "authors": [{"name": "Project Team", "email": "team@example.com"}],
            "requires-python": ">=3.8",
            "dependencies": ["click>=8.0.0", "pyyaml>=5.4.0", "jinja2>=3.0.0"],
            "scripts": {
                "deploy": "deploy_manager.cli:main",
                "deploy-dev": "deploy_manager.dev:dev_main"
            },
            "classifiers": [
                "Development Status :: 4 - Beta",
                "Programming Language :: Python :: 3",
                "Programming Language :: Python :: 3.8"
            ]
        }
    }
}

print("Dynamic Metadata Construction and Build System Integration:")
print("=" * 58)

for project_type, pyproject_data in project_templates.items():
    print(f"\n{project_type.replace('_', ' ').title()} Project:")
    
    # Create metadata object
    metadata = StandardMetadata.from_pyproject(pyproject_data, project_dir=".")
    
    print(f"  Name: {metadata.name}")
    print(f"  Version: {metadata.version}")
    print(f"  Dependencies: {len(metadata.dependencies) if metadata.dependencies else 0}")
    
    # Show build system integration
    if metadata.optional_dependencies:
        print(f"  Optional dependency groups: {len(metadata.optional_dependencies)}")
        for group in metadata.optional_dependencies.keys():
            print(f"    - {group}")
        
    # Generate core metadata for build tools
    core_metadata = metadata.as_rfc822()
    metadata_lines = len(core_metadata.as_string().split('\n'))
    print(f"  Generated core metadata: {metadata_lines} lines")
    
    # Show entry points if present
    if metadata.entrypoints:
        console_scripts = metadata.entrypoints.get('console_scripts', {})
        if console_scripts:
            print(f"  Console scripts: {len(console_scripts)}")

print("\nDynamic metadata construction for build system integration completed")
```
