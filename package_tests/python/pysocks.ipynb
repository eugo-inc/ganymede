## Example 1: Basic PySocks SOCKS proxy library setup

```python
import socks

# Example 1: Basic PySocks SOCKS proxy library setup
print("Testing PySocks SOCKS proxy library:")

try:
    # PySocks version and basic information
    print(f"✓ PySocks version: {socks.__version__}")
    print(f"✓ Python SOCKS client library for proxy connections")
    
    # Available proxy types
    proxy_types = [
        ('SOCKS4', socks.SOCKS4, 'SOCKS version 4 proxy protocol'),
        ('SOCKS5', socks.SOCKS5, 'SOCKS version 5 proxy protocol (RFC 1928)'),
        ('HTTP', socks.HTTP, 'HTTP CONNECT proxy protocol')
    ]
    
    print(f"✓ Supported proxy types:")
    for name, constant, description in proxy_types:
        print(f"  {name} (value {constant}): {description}")
    
    # Core classes and functions
    core_components = [
        ('socksocket', 'Enhanced socket class with SOCKS proxy support'),
        ('create_connection', 'Create TCP connection through SOCKS proxy'),
        ('set_default_proxy', 'Set default proxy for all socksocket instances'),
        ('get_default_proxy', 'Retrieve current default proxy settings'),
        ('wrap_module', 'Monkey-patch socket module for proxy support')
    ]
    
    print(f"✓ Core components:")
    for component, description in core_components:
        if hasattr(socks, component):
            print(f"  {component}(): {description}")
    
    # Exception hierarchy
    exception_classes = [
        'ProxyError', 'GeneralProxyError', 'ProxyConnectionError',
        'SOCKS4Error', 'SOCKS5Error', 'SOCKS5AuthError', 'HTTPError'
    ]
    
    print(f"✓ Exception classes:")
    available_exceptions = []
    for exc_name in exception_classes:
        if hasattr(socks, exc_name):
            exc_class = getattr(socks, exc_name)
            available_exceptions.append(exc_name)
            print(f"  {exc_name}: {exc_class}")
    
    print(f"  Total exception classes: {len(available_exceptions)}")
    
    # Default ports for different proxy types
    if hasattr(socks, 'DEFAULT_PORTS'):
        print(f"✓ Default proxy ports: {socks.DEFAULT_PORTS}")
    
    print("Example 1 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    print(f"Error type: {type(e).__name__}")
    raise AssertionError(
        f"pysocks example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 2: SOCKS socket creation and proxy configuration

```python
import socks

# Example 2: SOCKS socket creation and proxy configuration
print("Testing PySocks socket creation and proxy configuration:")

try:
    # Create SOCKS socket instances
    print(f"✓ Creating different SOCKS socket types:")
    
    # Basic socksocket creation
    sock = socks.socksocket()
    print(f"  Default socksocket: {type(sock)}")
    print(f"  Socket family: {sock.family}")
    print(f"  Socket type: {sock.type}")
    sock.close()
    
    # Create socket with explicit parameters
    sock_tcp = socks.socksocket(socks.socket.AF_INET, socks.socket.SOCK_STREAM)
    print(f"  TCP socksocket: {type(sock_tcp)}")
    sock_tcp.close()
    
    # Proxy configuration examples
    print(f"✓ Proxy configuration examples:")
    
    proxy_configs = [
        ("SOCKS5 proxy", socks.SOCKS5, "127.0.0.1", 1080, None, None),
        ("SOCKS4 proxy", socks.SOCKS4, "localhost", 1080, None, None),
        ("HTTP proxy", socks.HTTP, "proxy.example.com", 8080, None, None),
        ("Authenticated SOCKS5", socks.SOCKS5, "secure-proxy.com", 1080, "user", "pass"),
        ("No proxy", None, None, None, None, None)
    ]
    
    for config_name, proxy_type, addr, port, username, password in proxy_configs:
        print(f"  {config_name}:")
        
        if proxy_type is not None:
            print(f"    Type: {proxy_type} ({socks.PRINTABLE_PROXY_TYPES.get(proxy_type, 'Unknown')})")
            print(f"    Address: {addr}:{port}")
            if username and password:
                print(f"    Authentication: {username}:{'*' * len(password)}")
            else:
                print(f"    Authentication: None")
        else:
            print(f"    Direct connection (no proxy)")
    
    # Test proxy setting on socket
    print(f"✓ Setting proxy on socket:")
    
    test_sock = socks.socksocket()
    
    # Set SOCKS5 proxy (example configuration)
    test_sock.set_proxy(socks.SOCKS5, "127.0.0.1", 1080)
    print(f"  Proxy set successfully")
    print(f"  Socket ready for proxy connections")
    
    # Get current proxy settings
    try:
        proxy_info = test_sock.get_proxy()
        if proxy_info:
            print(f"  Current proxy: {proxy_info}")
        else:
            print(f"  No proxy configured")
    except AttributeError:
        print(f"  Proxy info not directly accessible")
    
    test_sock.close()
    
    # Default proxy configuration
    print(f"✓ Default proxy configuration:")
    
    # Set a default proxy for all future socksocket instances
    socks.set_default_proxy(socks.SOCKS5, "localhost", 1080)
    print(f"  Default SOCKS5 proxy set to localhost:1080")
    
    # Get current default proxy
    default_proxy = socks.get_default_proxy()
    if default_proxy:
        print(f"  Current default proxy: {default_proxy}")
    else:
        print(f"  No default proxy set")
    
    # Clear default proxy
    socks.set_default_proxy()  # No arguments = clear
    print(f"  Default proxy cleared")
    
    print("Example 2 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    print(f"Error type: {type(e).__name__}")
    raise AssertionError(
        f"pysocks example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: SOCKS proxy connection and error handling

```python
import socks
import socket

# Example 3: SOCKS proxy connection and error handling
print("Testing PySocks connection establishment and error handling:")

try:
    # Connection creation through SOCKS proxy
    print("✓ Testing connection creation through SOCKS proxy:")
    
    # Using create_connection function
    try:
        # Example configuration (would normally connect to real proxy)
        proxy_config = {
            'proxy_type': socks.SOCKS5,
            'addr': '127.0.0.1',
            'port': 1080,
            'rdns': True,  # Remote DNS resolution
            'username': None,
            'password': None
        }
        
        print(f"  Proxy configuration:")
        print(f"    Type: {proxy_config['proxy_type']} (SOCKS5)")
        print(f"    Address: {proxy_config['addr']}:{proxy_config['port']}")
        print(f"    Remote DNS: {proxy_config['rdns']}")
        print(f"    Authentication: {'Yes' if proxy_config['username'] else 'None'}")
        
        # Socket with proxy settings (demo without actual connection)
        sock = socks.socksocket()
        sock.set_proxy(proxy_config['proxy_type'], proxy_config['addr'], proxy_config['port'])
        print(f"  Socket configured with proxy settings")
        sock.close()
        
    except Exception as conn_error:
        print(f"  Connection demo: {type(conn_error).__name__}")
        raise
    
    # Exception handling examples
    print("✓ Exception handling examples:")
    
    exception_scenarios = [
        ('ProxyError', 'Base class for all proxy-related errors'),
        ('GeneralProxyError', 'General SOCKS proxy errors'),
        ('ProxyConnectionError', 'Connection establishment failures'),
        ('SOCKS4Error', 'SOCKS4-specific protocol errors'),
        ('SOCKS5Error', 'SOCKS5-specific protocol errors'),
        ('SOCKS5AuthError', 'SOCKS5 authentication failures'),
        ('HTTPError', 'HTTP proxy-specific errors')
    ]
    
    for exc_name, description in exception_scenarios:
        if hasattr(socks, exc_name):
            exc_class = getattr(socks, exc_name)
            print(f"  {exc_name}: {description}")
            print(f"    Class: {exc_class}")
        else:
            print(f"  {exc_name}: Not available")
    
    # Timeout and socket options
    print("✓ Socket options and timeout handling:")
    
    sock = socks.socksocket()
    
    # Set socket timeout
    sock.settimeout(30.0)  # 30 second timeout
    print(f"  Socket timeout set to 30 seconds")
    
    # Socket options that work with SOCKS
    sock.setsockopt(socket.SOL_SOCKET, socket.SO_REUSEADDR, 1)
    print(f"  SO_REUSEADDR enabled")
    
    # Get socket information
    timeout = sock.gettimeout()
    print(f"  Current timeout: {timeout} seconds")
    
    sock.close()
    
    print("Example 3 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    print(f"Error type: {type(e).__name__}")
    raise AssertionError(
        f"pysocks example (cell 5) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 4: Module wrapping and monkey-patching capabilities

```python
import socks
import socket as std_socket

# Example 4: Module wrapping and monkey-patching capabilities
print("Testing PySocks module wrapping and monkey-patching:")

try:
    # Module wrapping demonstration
    print("✓ Module wrapping capabilities:")
    
    # Save original socket module state
    original_socket = std_socket.socket
    original_create_connection = std_socket.create_connection if hasattr(std_socket, 'create_connection') else None
    
    print(f"  Original socket class: {original_socket}")
    if original_create_connection:
        print(f"  Original create_connection: {original_create_connection}")
    
    # Wrap socket module (monkey-patch)
    print("  Wrapping socket module with SOCKS support...")
    
    # This replaces socket.socket with socksocket
    socks.wrap_module(std_socket)
    print(f"  Module wrapped successfully")
    
    # Check if socket module is now wrapped
    wrapped_socket = std_socket.socket
    wrapped_create_connection = std_socket.create_connection if hasattr(std_socket, 'create_connection') else None
    
    print(f"  Wrapped socket class: {wrapped_socket}")
    if wrapped_create_connection:
        print(f"  Wrapped create_connection: {wrapped_create_connection}")
    
    # Demonstrate that wrapped socket has SOCKS capabilities
    test_socket = std_socket.socket()  # This is now a socksocket
    print(f"  Created socket type: {type(test_socket)}")
    
    # Check for SOCKS-specific methods
    socks_methods = ['set_proxy', 'get_proxy']
    available_methods = []
    
    for method in socks_methods:
        if hasattr(test_socket, method):
            available_methods.append(method)
            print(f"    ✓ {method}() method available")
    
    print(f"  SOCKS methods available: {len(available_methods)}")
    
    # Set proxy on wrapped socket
    if hasattr(test_socket, 'set_proxy'):
        test_socket.set_proxy(socks.SOCKS5, "localhost", 1080)
        print(f"  Proxy configured on wrapped socket")
    
    test_socket.close()
    
    # Restore original socket module (unwrap)
    print("✓ Restoring original socket module:")
    
    # Restore original socket functionality
    std_socket.socket = original_socket
    if original_create_connection:
        std_socket.create_connection = original_create_connection
    
    print(f"  Socket module restored to original state")
    
    # Verify restoration
    restored_socket = std_socket.socket()
    print(f"  Restored socket type: {type(restored_socket)}")
    restored_socket.close()
    
    # Global proxy settings impact
    print("✓ Global proxy settings demonstration:")
    
    # Set global default proxy
    socks.set_default_proxy(socks.SOCKS5, "proxy.example.com", 1080)
    
    # Create socksocket (inherits global proxy)
    global_proxy_socket = socks.socksocket()
    print(f"  Socket with global proxy settings created")
    
    # Get current global proxy
    current_default = socks.get_default_proxy()
    if current_default:
        print(f"  Current global proxy: {current_default}")
    
    global_proxy_socket.close()
    
    # Clear global proxy
    socks.set_default_proxy()  # Clear
    print(f"  Global proxy settings cleared")
    
    print("Example 4 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    print(f"Error type: {type(e).__name__}")
    raise AssertionError(
        f"pysocks example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 5: SSL/TLS wrapping and connection patterns

```python
import socks
import ssl

# Example 5: SSL/TLS wrapping and connection patterns
print("Testing PySocks SSL/TLS integration and production patterns:")

try:
    # SSL/TLS over SOCKS proxy
    print("✓ SSL/TLS integration with SOCKS proxy:")
    
    # Create SOCKS socket for SSL connection
    sock = socks.socksocket()
    
    # Configure SOCKS5 proxy
    sock.set_proxy(socks.SOCKS5, "127.0.0.1", 1080)
    print(f"  SOCKS5 proxy configured for SSL connection")
    
    # SSL context creation
    ssl_context = ssl.create_default_context()
    print(f"  SSL context created: {type(ssl_context)}")
    print(f"  SSL version: {ssl_context.protocol}")
    
    # Demonstrate SSL wrapping capability (without actual connection)
    print(f"  SSL socket wrapping capability available")
    print(f"  Can wrap SOCKS socket with SSL/TLS encryption")
    
    sock.close()
    
    # Connection pooling and reuse patterns
    print("✓ Connection management patterns:")
    
    connection_pool = []
    pool_size = 3
    
    print(f"  Creating connection pool of {pool_size} SOCKS sockets:")
    
    for i in range(pool_size):
        pool_sock = socks.socksocket()
        pool_sock.set_proxy(socks.SOCKS5, "localhost", 1080)
        pool_sock.settimeout(30.0)  # 30-second timeout
        connection_pool.append(pool_sock)
        print(f"    Socket {i+1}: {type(pool_sock)} created")
    
    print(f"  Pool created with {len(connection_pool)} sockets")
    
    # Cleanup pool
    for i, pool_sock in enumerate(connection_pool):
        pool_sock.close()
        print(f"    Socket {i+1} closed")
    
    connection_pool.clear()
    print(f"  Connection pool cleaned up")
    
    # Configuration validation and best practices
    print("✓ Configuration validation and best practices:")
    
    def validate_proxy_config(proxy_type, addr, port, username=None, password=None):
        """Validate proxy configuration parameters"""
        valid_types = [socks.SOCKS4, socks.SOCKS5, socks.HTTP]
        
        if proxy_type not in valid_types:
            return False, f"Invalid proxy type: {proxy_type}"
        
        if not isinstance(addr, str) or not addr.strip():
            return False, "Invalid proxy address"
        
        if not isinstance(port, int) or port <= 0 or port > 65535:
            return False, f"Invalid proxy port: {port}"
        
        if proxy_type == socks.SOCKS4 and (username or password):
            return False, "SOCKS4 does not support authentication"
        
        return True, "Configuration valid"
    
    # Test configurations
    test_configs = [
        (socks.SOCKS5, "proxy.example.com", 1080, None, None),
        (socks.SOCKS4, "localhost", 1080, None, None),
        (socks.HTTP, "http-proxy.com", 8080, "user", "pass"),
        (socks.SOCKS5, "", 1080, None, None),  # Invalid
        (socks.SOCKS4, "proxy.com", 1080, "user", "pass")  # Invalid
    ]
    
    print(f"  Testing {len(test_configs)} proxy configurations:")
    
    for i, config in enumerate(test_configs):
        valid, message = validate_proxy_config(*config)
        status = "✓" if valid else "✗"
        print(f"    Config {i+1}: {status} {message}")
    
    # Error recovery patterns
    print("✓ Error recovery and resilience patterns:")
    
    def create_resilient_socket(proxy_configs, max_retries=3):
        """Create socket with fallback proxy configurations"""
        for attempt in range(max_retries):
            for config in proxy_configs:
                try:
                    sock = socks.socksocket()
                    if config:
                        sock.set_proxy(*config)
                        print(f"    Attempt {attempt+1}: Proxy configured")
                    else:
                        print(f"    Attempt {attempt+1}: Direct connection")
                    return sock
                except Exception as e:
                    print(f"    Attempt {attempt+1} failed: {type(e).__name__}")
                    continue
        return None
    
    # Fallback configurations (proxy -> direct)
    fallback_configs = [
        (socks.SOCKS5, "primary-proxy.com", 1080),
        (socks.SOCKS5, "backup-proxy.com", 1080),
        None  # Direct connection
    ]
    
    # set_proxy() only records the proxy (no I/O), so the FIRST configuration must already yield a socket
    resilient_sock = create_resilient_socket(fallback_configs)
    assert resilient_sock is not None, "create_resilient_socket returned None for a configuration-only fallback chain"
    assert resilient_sock.proxy[1] == "primary-proxy.com", resilient_sock.proxy
    print(f"  Resilient socket created: {type(resilient_sock)}")
    resilient_sock.close()
    
    print("Example 5 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    print(f"Error type: {type(e).__name__}")
    raise AssertionError(
        f"pysocks example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
```
