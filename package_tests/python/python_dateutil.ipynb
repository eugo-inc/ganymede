# python_dateutil Package Testing

This notebook tests the python_dateutil package - Date/time utilities.

## Example 1: Advanced date parsing


```python
from dateutil import parser
from datetime import datetime

# Example 1: Advanced date parsing
print("Example 1: Advanced date parsing")

try:
    # Parse various date formats automatically
    date_strings = [
        "2023-12-25",
        "December 25, 2023",
        "25/12/2023",
        "2023-12-25T15:30:00Z",
        "Mon Dec 25 15:30:00 2023",
        "25 Dec 2023 3:30 PM",
        "2023-12-25 15:30:00.123456"
    ]
    
    expected_dates = [
        datetime(2023, 12, 25), datetime(2023, 12, 25), datetime(2023, 12, 25), datetime(2023, 12, 25, 15, 30),
        datetime(2023, 12, 25, 15, 30), datetime(2023, 12, 25, 15, 30), datetime(2023, 12, 25, 15, 30, 0, 123456),
    ]
    print("Parsing various date formats:")
    for date_str, expected in zip(date_strings, expected_dates, strict=True):
        parsed_date = parser.parse(date_str)
        assert parsed_date.replace(tzinfo=None) == expected, (date_str, parsed_date)
        print(f"'{date_str}' -> {parsed_date}")
    
    # Parse with fuzzy matching (ignore extra text)
    fuzzy_strings = [
        "The meeting is on December 25, 2023 at 3 PM",
        "Born on 1990-05-15 in New York",
        "Event scheduled for 2024-01-01"
    ]
    
    expected_fuzzy = [datetime(2023, 12, 25, 15, 0), datetime(1990, 5, 15), datetime(2024, 1, 1)]
    print("\nFuzzy parsing (ignoring extra text):")
    for fuzzy_str, expected in zip(fuzzy_strings, expected_fuzzy, strict=True):
        parsed_date = parser.parse(fuzzy_str, fuzzy=True)
        assert parsed_date == expected, (fuzzy_str, parsed_date)
        print(f"'{fuzzy_str}' -> {parsed_date}")
    
    print("Advanced date parsing completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"python_dateutil example (cell 2) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 1 finished")
```

## Example 2: Timezone handling and conversion


```python
from dateutil import tz, parser
from datetime import datetime

# Example 2: Timezone handling and conversion
print("Example 2: Timezone handling and conversion")

try:
    # Get various timezones
    utc = tz.UTC
    local_tz = tz.tzlocal()
    eastern = tz.gettz('America/New_York')
    pacific = tz.gettz('America/Los_Angeles')
    
    # Create a datetime in UTC
    utc_time = datetime(2023, 12, 25, 15, 30, 0, tzinfo=utc)
    print(f"UTC time: {utc_time}")
    
    # Convert to different timezones
    local_time = utc_time.astimezone(local_tz)
    eastern_time = utc_time.astimezone(eastern)
    pacific_time = utc_time.astimezone(pacific)
    
    print(f"Local time: {local_time}")
    print(f"Eastern time: {eastern_time}")
    print(f"Pacific time: {pacific_time}")
    
    # Parse timezone-aware strings
    tz_strings = [
        "2023-12-25T15:30:00Z",           # UTC
        "2023-12-25T10:30:00-05:00",     # Eastern
        "2023-12-25T07:30:00-08:00",     # Pacific
    ]
    
    print("\nParsing timezone-aware strings:")
    for tz_str in tz_strings:
        parsed = parser.parse(tz_str)
        print(f"'{tz_str}' -> {parsed} ({parsed.tzinfo})")
    
    print("Timezone handling and conversion completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"python_dateutil example (cell 4) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 2 finished")
```

## Example 3: Relative dates and deltas


```python
from dateutil.relativedelta import relativedelta
from datetime import datetime

# Example 3: Relative dates and deltas
print("Example 3: Relative dates and deltas")

try:
    base_date = datetime(2023, 1, 15)
    print(f"Base date: {base_date}")
    
    # Add/subtract various time periods
    operations = [
        ("+ 3 months", relativedelta(months=3)),
        ("+ 2 years", relativedelta(years=2)),
        ("+ 45 days", relativedelta(days=45)),
        ("- 6 months", relativedelta(months=-6)),
        ("+ 1 year, 2 months, 10 days", relativedelta(years=1, months=2, days=10)),
        ("Next Monday", relativedelta(weekday=0)),  # Monday is 0
        ("Last Friday", relativedelta(weekday=4, weeks=-1)),  # Friday is 4
    ]
    
    print("\nRelative date calculations:")
    for desc, delta in operations:
        result = base_date + delta
        print(f"{desc}: {result}")
    
    # Calculate differences between dates
    date1 = datetime(2023, 1, 15)
    date2 = datetime(2025, 3, 22)
    
    diff = relativedelta(date2, date1)
    print(f"\nDifference between {date2} and {date1}:")
    print(f"Years: {diff.years}, Months: {diff.months}, Days: {diff.days}")
    
    # Age calculation
    birth_date = datetime(1990, 5, 15)
    today = datetime(2023, 12, 25)
    age = relativedelta(today, birth_date)
    print(f"\nAge calculation:")
    print(f"Born: {birth_date}")
    print(f"Today: {today}")
    print(f"Age: {age.years} years, {age.months} months, {age.days} days")
    
    print("Relative dates and deltas completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"python_dateutil example (cell 6) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 3 finished")
```

## Example 4: Recurring dates and rules


```python
from dateutil.rrule import rrule, DAILY, WEEKLY, MONTHLY, YEARLY, MO, TU, WE, TH, FR
from datetime import datetime

# Example 4: Recurring dates and rules
print("Example 4: Recurring dates and rules")

try:
    start_date = datetime(2024, 1, 1)
    
    # Daily recurrence for a week
    daily_rule = rrule(DAILY, dtstart=start_date, count=7)
    print("Daily for 7 days:")
    for date in daily_rule:
        print(f"  {date.strftime('%Y-%m-%d %A')}")
    
    # Weekly on weekdays (Monday to Friday)
    weekday_rule = rrule(WEEKLY, byweekday=(MO, TU, WE, TH, FR), 
                        dtstart=start_date, count=10)
    print("\nWeekdays only (first 10):")
    for date in weekday_rule:
        print(f"  {date.strftime('%Y-%m-%d %A')}")
    
    # Monthly on the 15th
    monthly_rule = rrule(MONTHLY, bymonthday=15, dtstart=start_date, count=6)
    print("\nMonthly on the 15th (6 months):")
    for date in monthly_rule:
        print(f"  {date.strftime('%Y-%m-%d')}")
    
    # Every last Friday of the month
    last_friday = rrule(MONTHLY, byweekday=FR(-1), dtstart=start_date, count=6)
    print("\nLast Friday of each month (6 months):")
    for date in last_friday:
        print(f"  {date.strftime('%Y-%m-%d %A')}")
    
    print("Recurring dates and rules completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"python_dateutil example (cell 8) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 4 finished")
```

## Example 5: Easter calculation and utilities


```python
from dateutil.easter import easter
from dateutil.relativedelta import relativedelta
from datetime import datetime

# Example 5: Easter calculation and utilities
print("Example 5: Easter calculation and utilities")

try:
    # Calculate Easter dates for different years
    years = [2023, 2024, 2025, 2026, 2027]
    
    print("Easter dates:")
    for year in years:
        easter_date = easter(year)
        print(f"  {year}: {easter_date.strftime('%Y-%m-%d %A')}")
    
    # Calculate related Christian holidays based on Easter
    year = 2024
    easter_2024 = easter(year)
    
    # Good Friday (2 days before Easter)
    good_friday = easter_2024 - relativedelta(days=2)
    
    # Palm Sunday (1 week before Easter)  
    palm_sunday = easter_2024 - relativedelta(weeks=1)
    
    # Ascension Day (39 days after Easter)
    ascension_day = easter_2024 + relativedelta(days=39)
    
    # Pentecost (49 days after Easter)
    pentecost = easter_2024 + relativedelta(days=49)
    
    print(f"\nChristian holidays for {year}:")
    print(f"  Palm Sunday: {palm_sunday.strftime('%Y-%m-%d %A')}")
    print(f"  Good Friday: {good_friday.strftime('%Y-%m-%d %A')}")
    print(f"  Easter: {easter_2024.strftime('%Y-%m-%d %A')}")
    print(f"  Ascension Day: {ascension_day.strftime('%Y-%m-%d %A')}")
    print(f"  Pentecost: {pentecost.strftime('%Y-%m-%d %A')}")
    
    # Show how Easter varies between years
    print(f"\nEaster date variations:")
    easter_dates = [easter(y) for y in range(2020, 2030)]
    earliest = min(easter_dates)
    latest = max(easter_dates)
    print(f"  Earliest (2020-2029): {earliest.strftime('%Y-%m-%d')}")
    print(f"  Latest (2020-2029): {latest.strftime('%Y-%m-%d')}")
    
    print("Easter calculation and utilities completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"python_dateutil example (cell 10) failed: {type(e).__name__}: {e}"
    ) from e

print("Example 5 finished")
```
