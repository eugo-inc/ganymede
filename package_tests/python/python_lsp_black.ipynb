## Example 1: Basic Black formatting

```python
import black
from pylsp_black.plugin import get_eol_chars

# Example 1: Basic Black formatting
print("Testing python-lsp-black Black formatting:")

# Unformatted code
code = "def func(a,b,c):\n    if a:x=b+c\n    else:x=b-c\n    return x"

print(f"✓ Original code: {len(code)} chars")

# Format with Black
formatted = black.format_str(code, mode=black.FileMode())
print(f"✓ Formatted code: {len(formatted)} chars")
print(f"✓ Changes made: {code != formatted}")

# Show formatted output
print(f"✓ Formatted output:")
for i, line in enumerate(formatted.strip().split('\n'), 1):
    print(f"  {i}: {line}")

# Test EOL detection
eol = get_eol_chars(code)
print(f"✓ EOL characters: {repr(eol)}")
```

## Example 2: Black configuration modes

```python
import black

# Example 2: Black configuration modes
print("Testing python-lsp-black configuration:")

# Test different line lengths
code = "def process(data, filter_func=None, transform=None, key=None): return sorted([transform(x) if transform else x for x in data], key=key)"

configs = [
    ('Default (88)', black.FileMode()),
    ('PEP 8 (79)', black.FileMode(line_length=79)),
    ('Long (120)', black.FileMode(line_length=120))
]

for name, mode in configs:
    formatted = black.format_str(code, mode=mode)
    lines = formatted.strip().split('\n')
    max_len = max(len(line) for line in lines)
    
    print(f"✓ {name}:")
    print(f"  Lines: {len(lines)}")
    print(f"  Max length: {max_len}")

# Test string normalization
code_str = "text = 'hello'"
norm_on = black.format_str(code_str, mode=black.FileMode(string_normalization=True))
norm_off = black.format_str(code_str, mode=black.FileMode(string_normalization=False))
print(f"✓ String normalization on: {norm_on.strip()}")
print(f"✓ String normalization off: {norm_off.strip()}")
```

## Example 3: Formatting complex code structures

```python
import black

# Example 3: Formatting complex code structures
print("Testing python-lsp-black complex formatting:")

# Complex unformatted code
complex_code = """import os,sys,json
class DataProcessor:
    def __init__(self,config):self.config=config
    def process(self,items):
        result=[]
        for item in items:
            if item:result.append({'value':item,'done':True})
        return result"""

print(f"✓ Original: {len(complex_code.split(chr(10)))} lines")

# Format
formatted = black.format_str(complex_code, mode=black.FileMode())
print(f"✓ Formatted: {len(formatted.split(chr(10)))} lines")

# Show differences
orig_lines = complex_code.split('\n')
form_lines = formatted.split('\n')

print(f"✓ Sample changes:")
for i, (o, f) in enumerate(zip(orig_lines[:3], form_lines[:3]), 1):
    if o != f:
        print(f"  Line {i}: '{o}' -> '{f}'")
```

## Example 4: Target Python versions

```python
import black

# Example 4: Target Python versions
print("Testing python-lsp-black target versions:")

# Code with version-specific features
version_tests = [
    ('F-strings (3.6+)', 'msg = f"Hello {name}"', {black.TargetVersion.PY36}),
    ('Walrus (3.8+)', 'if (n := len(x)) > 5: print(n)', {black.TargetVersion.PY38}),
    ('Type hints (3.9+)', 'def f(x: list[str]) -> dict[str, int]: return {}', {black.TargetVersion.PY39}),
    ('Match (3.10+)', 'match x:\n    case 1: pass', {black.TargetVersion.PY310})
]

for name, code, versions in version_tests:
    try:
        mode = black.FileMode(target_versions=versions)
        formatted = black.format_str(code, mode=mode)
        print(f"✓ {name}:")
        print(f"  Target: {[v.name for v in versions]}")
        print(f"  Success: {len(formatted) > 0}")
    except Exception as e:
        print(f"✗ {name}: {type(e).__name__}")
        raise
```

## Example 5: Performance and error handling

```python
import black
import time

# Example 5: Performance and error handling
print("Testing python-lsp-black performance:")

# Performance test
sizes = [('Small', 10), ('Medium', 100), ('Large', 500)]

for name, lines in sizes:
    code_lines = [f"def func{i}(a,b): return a+b+{i}" for i in range(lines)]
    code = '\n'.join(code_lines)
    
    start = time.time()
    formatted = black.format_str(code, mode=black.FileMode())
    duration = (time.time() - start) * 1000
    
    print(f"✓ {name} ({lines} functions):")
    print(f"  Time: {duration:.2f}ms")
    print(f"  Speed: {len(code) / duration * 1000:.0f} chars/sec")

# Error handling
print(f"✓ Error handling:")
invalid_codes = [
    ('Syntax error', 'def broken(\n'),
    ('Bad indent', 'def f():\nreturn x')
]

for name, code in invalid_codes:
    try:
        black.format_str(code, mode=black.FileMode())
    except black.InvalidInput as e:
        assert "cannot parse" in str(e).lower(), e
        print(f"  {name}: Correctly rejected")
    else:
        raise AssertionError(f"black formatted invalid code ({name}): {code!r}")
```
