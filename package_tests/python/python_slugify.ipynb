## Example 1: Basic text slugification and URL-safe string generation

```python
from slugify import slugify

# Example 1: Basic text slugification and URL-safe string generation
print("Testing python-slugify basic functionality:")

try:
    # Test module import and basic functionality
    print(f"✓ python-slugify imported successfully")
    print(f"    slugify function: {slugify}")
    
    # Basic slugification tests
    print(f"✓ Basic slugification:")
    
    test_strings = [
        "Hello World",
        "My Blog Post Title",
        "UPPERCASE TITLE",
        "Mixed CaSe Article",
        "Special Characters: @#$%^&*()",
        "Punctuation, marks! And more?",
        "Numbers 123 and text",
        "Multiple    Spaces    Here"
    ]
    
    for text in test_strings:
        slug = slugify(text)
        print(f"    '{text}' -> '{slug}'")
    
    # Test separator options
    print(f"✓ Separator customization:")
    
    sample_text = "Hello World Test Article"
    separators = ['-', '_', '.', '']
    
    for sep in separators:
        slug = slugify(sample_text, separator=sep)
        sep_display = f"'{sep}'" if sep else "''"
        print(f"    separator={sep_display}: '{slug}'")
    
    # Test case handling
    print(f"✓ Case handling options:")
    
    test_text = "Hello WORLD Test"
    
    # Lowercase (default)
    slug_lower = slugify(test_text, lowercase=True)
    print(f"    lowercase=True: '{slug_lower}'")
    
    # Preserve case
    slug_preserve = slugify(test_text, lowercase=False)
    print(f"    lowercase=False: '{slug_preserve}'")
    
    # Test max length limitation
    print(f"✓ Length control:")
    
    long_text = "This is a very long title that should be truncated to a reasonable length"
    
    # Default (no limit)
    slug_full = slugify(long_text)
    print(f"    Full length: '{slug_full}' ({len(slug_full)} chars)")
    
    # Limited length
    slug_short = slugify(long_text, max_length=30)
    print(f"    Max 30 chars: '{slug_short}' ({len(slug_short)} chars)")
    
    # Very short
    slug_tiny = slugify(long_text, max_length=15)
    print(f"    Max 15 chars: '{slug_tiny}' ({len(slug_tiny)} chars)")
    
    # Expected slugs for the inputs above
    assert slugify("Hello World") == "hello-world"
    assert slugify("Special Characters: @#$%^&*()") == "special-characters"
    assert slugify("Punctuation, marks! And more?") == "punctuation-marks-and-more"
    assert slugify("Multiple    Spaces    Here") == "multiple-spaces-here"
    assert [slugify(sample_text, separator=sep) for sep in separators] == [
        "hello-world-test-article", "hello_world_test_article", "hello.world.test.article", "helloworldtestarticle"]
    assert slug_lower == "hello-world-test" and slug_preserve == "Hello-WORLD-Test"
    assert slug_short == "this-is-a-very-long-title-that" and slug_tiny == "this-is-a-very"

    print("Example 1 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    print(f"Error type: {type(e).__name__}")
    raise AssertionError(
        f"python_slugify example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 2: Unicode and international text handling

```python
from slugify import slugify

# Example 2: Unicode and international text handling
print("Testing python-slugify unicode and international text:")

try:
    # Test international character support
    print(f"✓ International text slugification:")
    
    international_texts = [
        ("French", "Café, résumé, naïve, François"),
        ("Spanish", "José, niño, mañana, corazón"),
        ("German", "Müller, Größe, weiß, Straße"),
        ("Portuguese", "São Paulo, João, coração"),
        ("Italian", "città, perché, più"),
        ("Polish", "Kraków, Łódź, Gdańsk"),
        ("Czech", "Praha, Brno, Český"),
        ("Turkish", "İstanbul, Ankara, Türkiye")
    ]
    
    for language, text in international_texts:
        slug = slugify(text)
        print(f"    {language}: '{text}' -> '{slug}'")
    
    # Test Cyrillic characters
    print(f"✓ Cyrillic text handling:")
    
    cyrillic_texts = [
        "Привет мир",
        "Москва Россия", 
        "Добро пожаловать",
        "Санкт-Петербург"
    ]
    
    for text in cyrillic_texts:
        slug = slugify(text)
        print(f"    '{text}' -> '{slug}'")
    
    # Test Asian languages
    print(f"✓ Asian language support:")
    
    asian_texts = [
        ("Chinese", "你好世界"),
        ("Chinese", "北京 上海"),
        ("Japanese", "こんにちは世界"),
        ("Korean", "안녕하세요"),
        ("Thai", "สวัสดี")
    ]
    
    for language, text in asian_texts:
        slug = slugify(text)
        print(f"    {language}: '{text}' -> '{slug}'")
    
    # Test Arabic and Hebrew
    print(f"✓ Right-to-left languages:")
    
    rtl_texts = [
        ("Arabic", "مرحبا بالعالم"),
        ("Arabic", "الرياض جدة"),
        ("Hebrew", "שלום עולם"),
        ("Persian", "سلام دنیا")
    ]
    
    for language, text in rtl_texts:
        slug = slugify(text)
        print(f"    {language}: '{text}' -> '{slug}'")
    
    # Test mixed language content
    print(f"✓ Mixed language content:")
    
    mixed_texts = [
        "English with Français",
        "Hello 世界 World",
        "Café in 北京 City",
        "Restaurant Москва-Paris",
        "Hotel São-Paulo-東京"
    ]
    
    for text in mixed_texts:
        slug = slugify(text)
        print(f"    '{text}' -> '{slug}'")
    
    # Test emoji and special unicode
    print(f"✓ Emoji and special unicode:")
    
    emoji_texts = [
        "Hello 👋 World 🌍",
        "Coffee ☕ and Cake 🍰",
        "Travel ✈️ Adventure 🗺️",
        "Music 🎵 and Dance 💃",
        "Tech 💻 Innovation 🚀"
    ]
    
    for text in emoji_texts:
        slug = slugify(text)
        print(f"    '{text}' -> '{slug}'")
    
    # Accented Latin and Cyrillic letters are transliterated to ASCII; emoji are dropped
    assert slugify("Café, résumé, naïve, François") == "cafe-resume-naive-francois"
    assert slugify("Müller, Größe, weiß, Straße") == "muller-grosse-weiss-strasse"
    assert slugify("Kraków, Łódź, Gdańsk") == "krakow-lodz-gdansk"
    assert slugify("Привет мир") == "privet-mir"
    assert slugify("Hello 👋 World 🌍") == "hello-world"
    for text in cyrillic_texts + [t for _, t in asian_texts + rtl_texts] + mixed_texts:
        slug = slugify(text)
        assert slug and slug.isascii() and slug == slug.lower(), (text, slug)

    print("Example 2 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    print(f"Error type: {type(e).__name__}")
    raise AssertionError(
        f"python_slugify example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: Truncation, word boundaries, stopwords and separators

```python
from slugify import slugify, smart_truncate

# Example 3: Truncation, word boundaries, stopwords and separators
print("Testing python-slugify truncation and stopwords:")

# max_length cuts the slug at that many characters; word_boundary=True cuts only between whole words
title = "This is a very long article title that needs smart truncation"
cut = slugify(title, max_length=30)
whole_words = slugify(title, max_length=30, word_boundary=True)
print(f"✓ max_length=30:                     '{cut}'")
print(f"✓ max_length=30, word_boundary=True: '{whole_words}'")
assert cut == "this-is-a-very-long-article-ti"
assert whole_words == "this-is-a-very-long-article"

# smart_truncate is the truncation helper slugify uses; it works on any string
truncated = smart_truncate("one two three four", max_length=10, word_boundary=True, separator=" ")
print(f"✓ smart_truncate('one two three four', 10, word_boundary=True): '{truncated}'")
assert truncated == "one two"
assert smart_truncate("one two three four", max_length=10) == "one two th"

# stopwords removes those words from the slug
stopwords = ['the', 'a', 'an', 'to', 'of']
expected = {
    "The Quick Brown Fox": "quick-brown-fox",
    "A Guide to the Best Practices": "guide-best-practices",
    "An Introduction to Machine Learning": "introduction-machine-learning",
    "The Art of Computer Programming": "art-computer-programming",
}
for text, slug in expected.items():
    result = slugify(text, stopwords=stopwords)
    print(f"✓ '{text}' -> '{result}' (without stopwords: '{slugify(text)}')")
    assert result == slug, result
assert slugify('The Quick Brown Fox', stopwords=['the']) == 'quick-brown-fox'

# Custom separators
separator_tests = [
    ("file name with spaces", "_", "file_name_with_spaces"),
    ("URL path component", "-", "url-path-component"),
    ("Database table name", "_", "database_table_name"),
]
for text, sep, slug in separator_tests:
    result = slugify(text, separator=sep)
    print(f"✓ '{text}' -> '{result}' (sep: '{sep}')")
    assert result == slug, result
```

## Example 4: Replacements, language replacement lists and Unicode output

```python
import slugify as slugify_module
from slugify import slugify

# Example 4: Replacements, language replacement lists and Unicode output
print("Testing python-slugify replacements and Unicode output:")

# replacements is a list of (old, new) pairs applied to the text before it is slugified
custom_replacements = [('&', '-and-'), ('+', '-plus-'), ('@', '-at-'), ('%', '-percent-'), ('$', '-dollar-')]
expected = {
    "Smith & Jones Company": ("smith-jones-company", "smith-and-jones-company"),
    "Price: $29.99 + tax": ("price-29-99-tax", "price-dollar-29-99-plus-tax"),
    "Email: info@example.com": ("email-info-example-com", "email-info-at-example-com"),
    "Discount: 25% off": ("discount-25-off", "discount-25-percent-off"),
}
for text, (standard, custom) in expected.items():
    result = slugify(text, replacements=custom_replacements)
    print(f"✓ '{text}': standard '{slugify(text)}', with replacements '{result}'")
    assert slugify(text) == standard
    assert result == custom, result
assert slugify("Version 2.0 Release", replacements=[('.', '-dot-')]) == "version-2-dot-0-release"

# The module ships replacement lists for some languages; GERMAN spells umlauts out (ü -> ue)
for list_name in ("CYRILLIC", "GERMAN", "GREEK"):
    pairs = getattr(slugify_module, list_name)
    print(f"✓ {list_name}: {len(pairs)} pairs, for example {pairs[:2]}")
    assert isinstance(pairs, list) and pairs and all(len(pair) == 2 for pair in pairs)
german = {
    "Müller und Schmidt": ("muller-und-schmidt", "mueller-und-schmidt"),
    "Österreich und Schweiz": ("osterreich-und-schweiz", "oesterreich-und-schweiz"),
}
for text, (plain, spelled_out) in german.items():
    result = slugify(text, replacements=slugify_module.GERMAN)
    print(f"✓ '{text}': '{slugify(text)}', with GERMAN '{result}'")
    assert slugify(text) == plain
    assert result == spelled_out, result
assert slugify("Καλημέρα κόσμε") == "kalemera-kosme"   # Greek is transliterated too

# allow_unicode=True keeps letters from any script instead of transliterating them to ASCII
unicode_cases = {
    "Café résumé": "café-résumé",
    "北京 上海": "北京-上海",
    "Москва Россия": "москва-россия",
}
for text, unicode_slug in unicode_cases.items():
    ascii_slug = slugify(text)
    result = slugify(text, allow_unicode=True)
    print(f"✓ '{text}': ASCII '{ascii_slug}', Unicode '{result}'")
    assert ascii_slug.isascii()
    assert result == unicode_slug, result
assert slugify('Café résumé') == 'cafe-resume'
assert slugify('Café résumé', allow_unicode=True) == 'café-résumé'

# Underscores, dots and repeated separators all collapse to a single hyphen
patterns = {
    "word1_word2_word3": "word1-word2-word3",
    "item-1-item-2-item-3": "item-1-item-2-item-3",
    "test.case.example": "test-case-example",
    "multi...dots...here": "multi-dots-here",
    "spaces   multiple   here": "spaces-multiple-here",
}
for text, slug in patterns.items():
    assert slugify(text) == slug, (text, slugify(text))
print(f"✓ {len(patterns)} separator patterns normalised")
```

## Example 5: Slugs for URLs, filenames, table names and CSS classes

```python
from slugify import slugify

# Example 5: Slugs for URLs, filenames, table names and CSS classes
print("Testing python-slugify for common naming tasks:")

try:
    # Test blog and CMS use cases
    print(f"✓ Blog and CMS use cases:")
    
    blog_titles = [
        "Getting Started with Python Web Development",
        "10 Tips for Better Code Quality & Performance",
        "Machine Learning: A Beginner's Guide (2024)",
        "Why TypeScript is Taking Over JavaScript",
        "The Future of Web Development: Trends & Predictions"
    ]
    
    for title in blog_titles:
        slug = slugify(title, max_length=50)
        url = f"https://blog.example.com/posts/{slug}"
        print(f"    Title: '{title}'")
        print(f"    URL: {url}")
        print()
    
    # Test e-commerce product names
    print(f"✓ E-commerce product slugs:")
    
    products = [
        ("Wireless Bluetooth Headphones - Premium Quality", 199.99),
        ("Smart Home Security Camera (4K, Night Vision)", 299.99),
        ("Ergonomic Office Chair - Adjustable Height", 449.99),
        ("Gaming Laptop - RTX 4080, 32GB RAM", 2499.99),
        ("Organic Coffee Beans - Fair Trade Certified", 24.99)
    ]
    
    for product_name, price in products:
        slug = slugify(product_name, max_length=40)
        product_url = f"https://shop.example.com/products/{slug}"
        print(f"    Product: '{product_name}' (${price})")
        print(f"    URL: {product_url}")
    
    # Test filename generation
    print(f"✓ File and directory naming:")
    
    file_names = [
        "Project Report - Final Version.pdf",
        "Meeting Notes (2024-01-15).docx",
        "Data Analysis & Visualization.xlsx",
        "User Testing Results - Q1 2024.pptx",
        "API Documentation v2.1.md"
    ]
    
    for original_name in file_names:
        # Extract extension
        if '.' in original_name:
            name_part, extension = original_name.rsplit('.', 1)
            slug = slugify(name_part, max_length=30, separator='_')
            safe_filename = f"{slug}.{extension.lower()}"
        else:
            safe_filename = slugify(original_name, max_length=30, separator='_')
        
        print(f"    Original: '{original_name}'")
        print(f"    Safe: '{safe_filename}'")
    
    # Test database table/column naming
    print(f"✓ Database naming conventions:")
    
    table_names = [
        "User Account Information",
        "Product Categories & Tags",
        "Order Processing History",
        "Customer Support Tickets",
        "Analytics & Reporting Data"
    ]
    
    for table_desc in table_names:
        table_name = slugify(table_desc, separator='_', lowercase=True, max_length=25)
        print(f"    Description: '{table_desc}'")
        print(f"    Table name: {table_name}")
    
    # Test CSS class generation
    print(f"✓ CSS class name generation:")
    
    component_names = [
        "Navigation Menu Bar",
        "User Profile Card",
        "Product Image Gallery",
        "Search Results Container",
        "Footer Social Links"
    ]
    
    for component in component_names:
        css_class = slugify(component, separator='-', lowercase=True)
        bem_base = css_class
        bem_element = f"{bem_base}__item"
        bem_modifier = f"{bem_base}--active"
        
        print(f"    Component: '{component}'")
        print(f"    Base class: .{bem_base}")
        print(f"    BEM element: .{bem_element}")
        print(f"    BEM modifier: .{bem_modifier}")
        print()
    
    # Batch processing
    print(f"✓ Batch processing:")
    
    # Generate test data
    test_texts = []
    for i in range(10):
        text = f"Test Article Title Number {i+1}: Advanced Python Programming"
        test_texts.append(text)
    
    # Process batch
    processed_slugs = []
    for text in test_texts:
        slug = slugify(text, max_length=35)
        processed_slugs.append(slug)
    
    print(f"    Processed {len(test_texts)} texts")
    print(f"    Sample results:")
    for i in range(min(3, len(processed_slugs))):
        print(f"      '{test_texts[i]}' -> '{processed_slugs[i]}'")
    
    # Test edge cases and validation
    print(f"✓ Edge cases and validation:")
    
    edge_cases = [
        "",  # Empty string
        "   ",  # Only spaces
        "!!!@@@###",  # Only special characters
        "123456789",  # Only numbers
        "a",  # Single character
        "A" * 100,  # Very long string
        "Mixed123!@#Content"  # Mixed content
    ]
    
    for text in edge_cases:
        slug = slugify(text, max_length=20)
        is_valid = len(slug) > 0 and slug.replace('-', '').replace('_', '').isalnum()
        print(f"    Input: '{text[:20]}...' -> '{slug}' (Valid: {is_valid})")
    
    assert slugify("Getting Started with Python Web Development", max_length=50) == "getting-started-with-python-web-development"
    assert slugify("Machine Learning: A Beginner's Guide (2024)", max_length=50) == "machine-learning-a-beginner-s-guide-2024"
    assert slugify("Gaming Laptop - RTX 4080, 32GB RAM", max_length=40) == "gaming-laptop-rtx-4080-32gb-ram"
    assert slugify("Meeting Notes (2024-01-15)", max_length=30, separator="_") == "meeting_notes_2024_01_15"
    assert slugify("Product Categories & Tags", separator="_", max_length=25) == "product_categories_tags"
    assert slugify("Navigation Menu Bar") == "navigation-menu-bar"
    assert all(len(s) <= 35 for s in processed_slugs) and len(set(processed_slugs)) == len(test_texts)
    # Inputs without letters or digits give an empty slug
    assert [slugify(t, max_length=20) for t in edge_cases] == ["", "", "", "123456789", "a", "a" * 20, "mixed123-content"]

    print("Example 5 completed successfully")
    
except Exception as e:
    print(f"Error: {e}")
    print(f"Error type: {type(e).__name__}")
    raise AssertionError(
        f"python_slugify example (cell 9) failed: {type(e).__name__}: {e}"
    ) from e
```
