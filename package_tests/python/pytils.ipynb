## Example 1: Russian Text Transliteration

```python
# Example 1: Russian Text Transliteration
import pytils

print("=== Russian Transliteration ===")

# Basic transliteration from Cyrillic to Latin
russian_texts = [
    "Привет, мир!",
    "Москва - столица России", 
    "Программирование на Python",
    "Пётр Великий",
    "Щёлкунчик"
]

print("Cyrillic to Latin transliteration:")
for text in russian_texts:
    latin = pytils.translit.translify(text)
    print(f"  '{text}' -> '{latin}'")

# Reverse transliteration (Latin back to Cyrillic) 
print("\nLatin to Cyrillic transliteration:")
latin_texts = ["Privet, mir!", "Moskva", "Python", "Petr"]
for text in latin_texts:
    try:
        cyrillic = pytils.translit.detranslify(text)
        print(f"  '{text}' -> '{cyrillic}'")
    except Exception as e:
        print(f"  '{text}' -> Error: {e}")
        raise

# URL-safe transliteration
print("\nURL-friendly transliteration:")
for text in ["Пример статьи", "Новости дня"]:
    url_safe = pytils.translit.slugify(text)
    print(f"  '{text}' -> '{url_safe}'")
```

## Example 2: Russian Numeral Word Forms

```python
# Example 2: Russian Numeral Word Forms
import pytils

print("=== Russian Numeral Declension ===")

# Convert numbers to Russian word forms
numbers_to_convert = [1, 2, 5, 11, 21, 100, 1000, 1234567]

for num in numbers_to_convert:
    # Integer to words
    words = pytils.numeral.in_words(num)
    print(f"{num} -> '{words}'")

# Numbers with gender and cases
print("\n=== Gendered Number Forms ===")

# Different genders for Russian numbers
genders = [pytils.numeral.MALE, pytils.numeral.FEMALE, pytils.numeral.NEUTER]
gender_names = ["masculine", "feminine", "neuter"]

test_numbers = [1, 2, 3, 4]

for num in test_numbers:
    print(f"Number {num}:")
    for gender, name in zip(genders, gender_names):
        try:
            words = pytils.numeral.in_words(num, gender=gender)
            print(f"  {name}: '{words}'")
        except Exception as e:
            print(f"  {name}: Error - {e}")
            raise

# Money amounts in Russian
print("\n=== Money Amounts ===")
amounts = [1, 10, 25, 100, 1234.56]

for amount in amounts:
    try:
        money_words = pytils.numeral.rubles(amount)
        print(f"{amount} rubles -> '{money_words}'")
    except Exception as e:
        print(f"{amount} rubles -> Error: {e}")
        raise
```

## Example 3: Russian Date and Time Formatting

```python
# Example 3: Russian Date and Time Formatting
import pytils
from datetime import datetime, timedelta

print("=== Russian Date/Time Formatting ===")

# Current date and time formatting
now = datetime.now()
print(f"Current datetime: {now}")

# Format distance of time
print("\n=== Time Distance in Russian ===")

# Create various time deltas for testing
time_points = [
    now - timedelta(seconds=30),    # 30 seconds ago
    now - timedelta(minutes=5),     # 5 minutes ago  
    now - timedelta(hours=2),       # 2 hours ago
    now - timedelta(days=1),        # 1 day ago
    now - timedelta(days=7),        # 1 week ago
    now - timedelta(days=30),       # 1 month ago
]

for time_point in time_points:
    # signature is (from_time, accuracy=1, to_time=None): `to_time` must be passed by keyword
    distance = pytils.dt.distance_of_time_in_words(time_point, to_time=now)
    delta = now - time_point
    print(f"  {delta} ago -> '{distance}'")
    assert distance.endswith("назад"), distance

# Specific day names and formatting  
print("\n=== Russian Date Names ===")

# Test specific dates
test_dates = [
    datetime(2024, 1, 1),   # New Year
    datetime(2024, 3, 8),   # Women's Day
    datetime(2024, 5, 9),   # Victory Day
    datetime(2024, 12, 31), # New Year's Eve
]

for date in test_dates:
    try:
        formatted = pytils.dt.ru_strftime(u"%A, %d %B %Y года", date, inflected=True)
        print(f"  {date.strftime('%Y-%m-%d')} -> '{formatted}'")
    except Exception as e:
        print(f"  Date formatting error: {e}")
        raise
```

## Example 4: Typography and Text Processing Utilities

```python
# Example 4: Typography and Text Processing Utilities
import pytils

print("=== Russian Typography Utils ===")

# Typography improvements for Russian text
sample_texts = [
    'Это "тест" типографики',
    "Температура была -10 градусов",  
    "Цена: 1000 рублей",
    "Номер телефона: +7 (495) 123-45-67",
    "Web-сайт: www.example.com"
]

print("Typography enhancement:")
for text in sample_texts:
    enhanced = pytils.typo.typography(text)  # pytils.typo's entry point (there is no typo.typo)
    print(f"  '{text}' -> '{enhanced}'")
    assert isinstance(enhanced, str) and enhanced
assert pytils.typo.typography(sample_texts[0]) == 'Это\u202f«тест»\u202fтипографики'  # straight quotes -> guillemets

# Utility functions for text processing
print("\n=== Text Processing Utilities ===")

# Check and process text values
test_values = [123, "test string", "", None, 0, "   ", "ten chars!"]

for value in test_values:
    # Check if value is positive (for numbers): returns None, raises ValueError when negative
    if isinstance(value, (int, float)):
        pytils.utils.check_positive(value, strict=False)
        print(f"  {value} is positive or zero: True")
    
    # Check length constraints: check_length(value, length) returns None on an EXACT match, else raises ValueError
    if value is not None:
        text = str(value)
        try:
            pytils.utils.check_length(text, 10)
            length_ok = True
        except ValueError as e:
            assert str(e) == f"length must be 10, not {len(text)}", e
            length_ok = False
        assert length_ok == (len(text) == 10)
        print(f"  '{value}' length == 10: {length_ok}")

# Split values utility
print("\n=== Split Values Utility ===")
test_strings = ["one,two,three", "a;b;c", "x|y|z", "single"]

for string in test_strings:
    try:
        split_result = pytils.utils.split_values(string)
        print(f"  '{string}' -> {split_result}")
    except Exception as e:
        print(f"  Split error for '{string}': {e}")
        raise
```

## Example 5: Russian localization workflow

```python
# Example 5: Russian localization workflow
import pytils
from datetime import datetime

print("=== Russian Localization Example ===")

# Simulate a Russian e-commerce scenario
print("Russian E-commerce Localization:")

# Product information
product_names = ["Смартфон Apple iPhone", "Ноутбук ASUS", "Наушники Sony"]
quantities = [1, 2, 5]
prices = [89999.99, 45500.0, 12900.0]

for product, qty, price in zip(product_names, quantities, prices):
    # Transliterate product name for URL
    url_slug = pytils.translit.slugify(product)
    
    # Format quantity in words
    qty_words = pytils.numeral.in_words(qty)
    
    # Format price in rubles
    price_words = pytils.numeral.rubles(price)
    
    print(f"\nProduct: {product}")
    print(f"  URL slug: {url_slug}")
    print(f"  Quantity: {qty_words}")
    print(f"  Price: {price_words}")

# Order summary with date
print(f"\n=== Order Summary ===")
order_date = datetime.now()
formatted_date = pytils.dt.ru_strftime("%d %B %Y года", order_date, inflected=True)

total_items = sum(quantities)
total_price = sum(p * q for p, q in zip(prices, quantities))

print(f"Дата заказа: {formatted_date}")
print(f"Общее количество товаров: {pytils.numeral.in_words(total_items)}")
print(f"Общая стоимость: {pytils.numeral.rubles(total_price)}")

# Generate confirmation message with typography
confirmation = f'Ваш заказ на сумму {total_price} рублей успешно оформлен!'
formatted_confirmation = pytils.typo.typography(confirmation)  # pytils.typo's entry point (there is no typo.typo)
print(f"\nПодтверждение: {formatted_confirmation}")
assert formatted_confirmation.startswith("Ваш\u202fзаказ"), formatted_confirmation  # non-breaking glue after a short word

print("\nAll pytils localization examples completed successfully!")
```
