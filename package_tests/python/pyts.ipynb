```python
# Example 1: Generate a synthetic time series dataset and verify shape.
import numpy as np
from pyts.datasets import make_cylinder_bell_funnel

X, y = make_cylinder_bell_funnel(n_samples=30, random_state=42)
print(f"X shape: {X.shape}")
print(f"y shape: {y.shape}")
print(f"unique classes: {np.unique(y)}")

assert X.ndim == 2
assert X.shape[0] == 30
assert y.shape[0] == 30
assert len(np.unique(y)) == 3
```

```python
# Example 2: Symbolic Aggregate Approximation (SAX) transformation.
import numpy as np
from pyts.approximation import SymbolicAggregateApproximation

rng = np.random.RandomState(0)
X = rng.randn(5, 100)

sax = SymbolicAggregateApproximation(n_bins=5, strategy='quantile')
X_sax = sax.fit_transform(X)

print(f"input shape: {X.shape}")
print(f"SAX output shape: {X_sax.shape}")
print(f"first row symbols: {X_sax[0][:20]}")

assert X_sax.shape == X.shape
assert X_sax.dtype.kind in ('U', 'O')
```

```python
# Example 3: Gramian Angular Field image transformation.
import numpy as np
from pyts.image import GramianAngularField

rng = np.random.RandomState(1)
X = rng.randn(3, 64)

gaf = GramianAngularField(image_size=32, method='summation')
X_gaf = gaf.fit_transform(X)

print(f"input shape: {X.shape}")
print(f"GAF output shape: {X_gaf.shape}")
print(f"value range: [{X_gaf.min():.3f}, {X_gaf.max():.3f}]")

assert X_gaf.shape == (3, 32, 32)
assert X_gaf.min() >= -1.0 - 1e-6
assert X_gaf.max() <= 1.0 + 1e-6
```

```python
# Example 4: KNN time series classifier with DTW distance.
import numpy as np
from pyts.classification import KNeighborsClassifier
from pyts.datasets import make_cylinder_bell_funnel

X, y = make_cylinder_bell_funnel(n_samples=40, random_state=7)
X_train, y_train = X[:30], y[:30]
X_test, y_test = X[30:], y[30:]

clf = KNeighborsClassifier(n_neighbors=3, metric='dtw')
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
acc = (y_pred == y_test).mean()

print(f"predictions: {y_pred}")
print(f"accuracy: {acc:.3f}")

assert y_pred.shape == y_test.shape
assert 0.0 <= acc <= 1.0
```

```python
# Example 5: Piecewise Aggregate Approximation (PAA) dimensionality reduction.
import numpy as np
from pyts.approximation import PiecewiseAggregateApproximation

rng = np.random.RandomState(2)
X = rng.randn(4, 100)

paa = PiecewiseAggregateApproximation(window_size=10)
X_paa = paa.fit_transform(X)

print(f"input shape: {X.shape}")
print(f"PAA output shape: {X_paa.shape}")
print(f"first row reduced: {X_paa[0]}")

assert X_paa.shape == (4, 10)
assert np.isfinite(X_paa).all()
print("\npyts smoke test PASSED")
```
