## Example 1: Basic timezone creation and usage

```python
import pytz
from datetime import datetime

# Example 1: Basic timezone creation and usage
print("Example 1: Basic timezone operations")

# Get UTC timezone
utc_tz = pytz.UTC
print(f"UTC timezone: {utc_tz}")

# Get specific timezones
ny_tz = pytz.timezone('America/New_York')
london_tz = pytz.timezone('Europe/London')
tokyo_tz = pytz.timezone('Asia/Tokyo')

print(f"New York timezone: {ny_tz}")
print(f"London timezone: {london_tz}")
print(f"Tokyo timezone: {tokyo_tz}")

# Create timezone-aware datetime
utc_now = datetime.now(utc_tz)
print(f"Current UTC time: {utc_now}")

# Show timezone info
print(f"Timezone zone: {ny_tz.zone}")
print(f"Available timezones: {len(pytz.all_timezones)} total")

assert utc_now.utcoffset().total_seconds() == 0
assert ny_tz.zone == 'America/New_York' and str(tokyo_tz) == 'Asia/Tokyo'
assert {'America/New_York', 'Europe/London', 'Asia/Tokyo'} <= set(pytz.all_timezones)

print("Basic timezone operations completed")
```

## Example 2: Timezone conversion

```python
import pytz
from datetime import datetime, timedelta

# Example 2: Timezone conversion
print("Example 2: Timezone conversion")

# Create a naive datetime
naive_dt = datetime(2024, 6, 15, 14, 30, 0)
print(f"Naive datetime: {naive_dt}")

# Localize to different timezones
ny_tz = pytz.timezone('America/New_York')
london_tz = pytz.timezone('Europe/London')
tokyo_tz = pytz.timezone('Asia/Tokyo')

# Localize naive datetime to NY time
ny_dt = ny_tz.localize(naive_dt)
print(f"NY time: {ny_dt}")

# Convert to other timezones
london_dt = ny_dt.astimezone(london_tz)
tokyo_dt = ny_dt.astimezone(tokyo_tz)
utc_dt = ny_dt.astimezone(pytz.UTC)

print(f"London time: {london_dt}")
print(f"Tokyo time: {tokyo_dt}")
print(f"UTC time: {utc_dt}")

# Show time differences
print(f"NY-London offset: {(london_dt.hour - ny_dt.hour)} hours")
print(f"NY-Tokyo offset: {(tokyo_dt.hour - ny_dt.hour) % 24} hours")

# 14:30 EDT (UTC-4) on 15 June 2024 is 18:30 UTC, 19:30 BST and 03:30 JST the next day
assert ny_dt.utcoffset() == timedelta(hours=-4)
assert utc_dt == pytz.UTC.localize(datetime(2024, 6, 15, 18, 30))
assert (london_dt.hour, london_dt.minute) == (19, 30)
assert (tokyo_dt.day, tokyo_dt.hour) == (16, 3)
assert ny_dt == london_dt == tokyo_dt == utc_dt   # the same instant

print("Timezone conversion completed")
```

## Example 3: Daylight Saving Time handling

```python
import pytz
from datetime import datetime, timedelta

# Example 3: Daylight Saving Time handling
print("Example 3: Daylight Saving Time handling")

# Get timezone that observes DST
eastern_tz = pytz.timezone('US/Eastern')

# Test dates in summer (DST) and winter (standard time)
summer_date = datetime(2024, 7, 15, 12, 0, 0)
winter_date = datetime(2024, 1, 15, 12, 0, 0)

# Localize both dates
summer_eastern = eastern_tz.localize(summer_date)
winter_eastern = eastern_tz.localize(winter_date)

print(f"Summer (DST): {summer_eastern}")
print(f"Winter (STD): {winter_eastern}")

# Convert to UTC to see the difference
summer_utc = summer_eastern.astimezone(pytz.UTC)
winter_utc = winter_eastern.astimezone(pytz.UTC)

print(f"Summer UTC: {summer_utc}")
print(f"Winter UTC: {winter_utc}")

# Show DST offset
print(f"Summer offset: {summer_eastern.strftime('%z')}")
print(f"Winter offset: {winter_eastern.strftime('%z')}")

# dst() is the daylight-saving adjustment in effect at that moment
print(f"Summer dst(): {summer_eastern.dst()}, winter dst(): {winter_eastern.dst()}")
assert summer_eastern.strftime('%z') == '-0400' and winter_eastern.strftime('%z') == '-0500'
assert summer_eastern.dst() == timedelta(hours=1) and winter_eastern.dst() == timedelta(0)
assert (summer_utc.hour, winter_utc.hour) == (16, 17)

# With is_dst=None, localize rejects wall-clock times that DST makes missing or ambiguous
for wall_time, error in [(datetime(2024, 3, 10, 2, 30), pytz.NonExistentTimeError),
                         (datetime(2024, 11, 3, 1, 30), pytz.AmbiguousTimeError)]:
    try:
        eastern_tz.localize(wall_time, is_dst=None)
    except error:
        print(f"{wall_time}: {error.__name__}")
    else:
        raise AssertionError(f"{wall_time} was accepted")

print("Daylight Saving Time handling completed")
```

## Example 4: Common timezone operations

```python
import pytz
from datetime import datetime

# Example 4: Common timezone operations
print("Example 4: Common timezone operations")

# Get common timezone groups
common_timezones = list(pytz.common_timezones)[:10]
print(f"First 10 common timezones: {common_timezones}")

# Get timezone by country
us_timezones = [tz for tz in pytz.all_timezones if tz.startswith('US/')]
print(f"US timezones: {us_timezones[:5]}")

# Get current time in multiple timezones
current_utc = datetime.now(pytz.UTC)
print(f"\nCurrent times around the world:")

timezone_cities = {
    'UTC': pytz.UTC,
    'New York': pytz.timezone('America/New_York'),
    'Los Angeles': pytz.timezone('America/Los_Angeles'),
    'London': pytz.timezone('Europe/London'),
    'Tokyo': pytz.timezone('Asia/Tokyo'),
    'Sydney': pytz.timezone('Australia/Sydney')
}

for city, tz in timezone_cities.items():
    local_time = current_utc.astimezone(tz)
    print(f"{city}: {local_time.strftime('%Y-%m-%d %H:%M:%S %Z')}")

# Timezone abbreviations
print(f"\nTimezone abbreviations:")
for city, tz in list(timezone_cities.items())[:3]:
    local_time = current_utc.astimezone(tz)
    print(f"{city}: {local_time.strftime('%Z')} ({local_time.strftime('%z')})")

assert common_timezones == sorted(common_timezones)
assert 'US/Eastern' in us_timezones and 'Europe/London' in pytz.common_timezones
# Every converted time is the same instant as current_utc
assert all(current_utc.astimezone(tz) == current_utc for tz in timezone_cities.values())

print("Common timezone operations completed")
```

## Example 5: Country lookup, time arithmetic and normalize()

```python
import pytz
from datetime import datetime, timedelta

# Example 5: Country lookup, time arithmetic and normalize()
print("Example 5: Country lookup, time arithmetic and normalize()")

# Timezone lookup and validation
print("Timezone information:")
print(f"pytz version: {pytz.__version__}")
print(f"Total timezones: {len(pytz.all_timezones)}")
print(f"Common timezones: {len(pytz.common_timezones)}")

# Country timezone mapping
country_timezones = pytz.country_timezones.get('US', [])
print(f"US timezones: {len(country_timezones)} ({country_timezones[:3]}...)")

# Test timezone arithmetic
ny_tz = pytz.timezone('America/New_York')
base_time = ny_tz.localize(datetime(2024, 6, 15, 9, 0, 0))

print(f"\nTimezone arithmetic:")
print(f"Base time: {base_time}")

# Add time within June (no DST change in this range)
future_times = []
for hours in [1, 6, 24, 168]:  # 1h, 6h, 1d, 1w
    future_time = base_time + timedelta(hours=hours)
    future_times.append((hours, future_time))
    print(f"+ {hours:3}h: {future_time}")

# Timezone comparison
print(f"\nTimezone features:")
for tz_name in ['UTC', 'US/Eastern', 'Europe/London', 'Asia/Tokyo']:
    tz = pytz.timezone(tz_name) if tz_name != 'UTC' else pytz.UTC
    sample_time = base_time.astimezone(tz)
    print(f"{tz_name:12}: {sample_time.strftime('%H:%M %Z (%z)')}")

assert 'America/New_York' in country_timezones
assert [t.strftime('%Y-%m-%d %H:%M %z') for _, t in future_times] == [
    '2024-06-15 10:00 -0400', '2024-06-15 15:00 -0400', '2024-06-16 09:00 -0400', '2024-06-22 09:00 -0400']
assert base_time.astimezone(pytz.timezone('Europe/London')).strftime('%H:%M %Z') == '14:00 BST'
assert base_time.astimezone(pytz.timezone('Asia/Tokyo')).strftime('%H:%M %Z') == '22:00 JST'

# Across a DST change, adding a timedelta keeps the old UTC offset; normalize() corrects it.
# DST started in New York at 02:00 on 10 March 2024.
before = ny_tz.localize(datetime(2024, 3, 9, 12, 0, 0))
added = before + timedelta(days=1)
normalized = ny_tz.normalize(added)
print(f"\n{before} + 1 day = {added}, normalize() -> {normalized}")
assert added.strftime('%H:%M %z') == '12:00 -0500'
assert normalized.strftime('%H:%M %z') == '13:00 -0400'
assert normalized == added   # the same instant, shown with the correct offset

print("Country lookup, arithmetic and normalize() completed")
```
