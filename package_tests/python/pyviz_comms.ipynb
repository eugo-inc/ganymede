## Example 1: Basic Comm creation

```python
import pyviz_comms

# Example 1: Basic Comm creation
print("Testing pyviz-comms Comm creation:")

# Create communication object with default settings
comm1 = pyviz_comms.Comm()
print(f"✓ Default Comm: id={comm1.id}, name={comm1.name}")

# Create Comm with specific ID
comm2 = pyviz_comms.Comm(id="custom_id")
print(f"✓ Custom ID Comm: id={comm2.id}, name={comm2.name}")

# Create Comm with error callback
def on_error(msg):
    print(f"  Error: {msg}")

comm3 = pyviz_comms.Comm(id="error_comm", on_error=on_error)
print(f"✓ Callback Comm: id={comm3.id}")

# Test close method
comm1.close()
print(f"✓ Comm closed successfully")
assert len(comm1.id) == 32   # a random hex id when none is given
assert (comm2.id, comm3.id) == ("custom_id", "error_comm")
assert comm1.name.startswith("Comm") and len({comm1.name, comm2.name, comm3.name}) == 3   # numbered names
```

## Example 2: JupyterComm and CommManager

```python
import pyviz_comms

# Example 2: JupyterComm and CommManager
print("Testing pyviz-comms JupyterComm:")

# Create JupyterComm instance
jupyter_comm = pyviz_comms.JupyterComm(id="jupyter_test")
print(f"✓ JupyterComm: id={jupyter_comm.id}, name={jupyter_comm.name}")

# Check CommManager
print(f"✓ CommManager class: {pyviz_comms.CommManager}")
print(f"✓ CommManager methods: {[m for m in dir(pyviz_comms.CommManager) if not m.startswith('_')]}")

# Check JupyterCommManager
print(f"✓ JupyterCommManager: {pyviz_comms.JupyterCommManager}")

# Create JupyterCommJS
js_comm = pyviz_comms.JupyterCommJS(id="js_test")
print(f"✓ JupyterCommJS: id={js_comm.id}, name={js_comm.name}")
assert jupyter_comm.id == "jupyter_test" and jupyter_comm.name.startswith("JupyterComm")
assert issubclass(pyviz_comms.JupyterComm, pyviz_comms.Comm)
assert issubclass(pyviz_comms.JupyterCommJS, pyviz_comms.JupyterComm)
assert issubclass(pyviz_comms.JupyterCommManager, pyviz_comms.CommManager)
# The Jupyter manager uses JupyterComm on the kernel side and JupyterCommJS on the browser side
assert pyviz_comms.JupyterCommManager.server_comm is pyviz_comms.JupyterComm
assert pyviz_comms.JupyterCommManager.client_comm is pyviz_comms.JupyterCommJS
```

## Example 3: Comm send and decode

```python
import pyviz_comms
import json

# Example 3: Comm send and decode
print("Testing pyviz-comms send and decode:")

# Create Comm
comm = pyviz_comms.Comm(id="message_test")
print(f"✓ Created Comm: {comm.id}")

# Test send method
test_data = {"type": "update", "value": 42}
comm.send(test_data)
print(f"✓ Sent message: {test_data}")

# Test decode method
encoded_msg = json.dumps({"data": [1, 2, 3]})
decoded = comm.decode(encoded_msg)
print(f"✓ Decoded message: {decoded}")

# Test close method
comm.close()
print(f"✓ Comm closed")
# The base Comm passes a message through unchanged; JupyterComm takes the data field of a Jupyter comm message
assert decoded == encoded_msg
assert pyviz_comms.JupyterComm.decode({"content": {"data": {"value": 42}}}) == {"value": 42}
```

## Example 4: JavaScript templates and Bokeh integration

```python
import pyviz_comms

# Example 4: JavaScript templates and Bokeh integration
print("Testing pyviz-comms JavaScript integration:")

# Check bokeh_msg_handler template
bokeh_handler = pyviz_comms.bokeh_msg_handler
print(f"✓ Bokeh handler type: {type(bokeh_handler).__name__}")
print(f"✓ Bokeh handler length: {len(bokeh_handler)} chars")

# Check JavaScript callback template
js_callback = pyviz_comms.JS_CALLBACK
print(f"✓ JS_CALLBACK type: {type(js_callback).__name__}")
print(f"✓ JS_CALLBACK length: {len(js_callback)} chars")

# Check PyViz proxy template
pyviz_proxy = pyviz_comms.PYVIZ_PROXY
print(f"✓ PYVIZ_PROXY type: {type(pyviz_proxy).__name__}")
print(f"✓ PYVIZ_PROXY length: {len(pyviz_proxy)} chars")

# Check abort JavaScript
abort_js = pyviz_comms.ABORT_JS
print(f"✓ ABORT_JS type: {type(abort_js).__name__}")
assert all(isinstance(t, str) and t for t in (bokeh_handler, js_callback, pyviz_proxy, abort_js))
assert "window.PyViz" in pyviz_proxy   # the proxy sets up the browser-side PyViz registry
```

## Example 5: StandardOutput and callback handling

```python
import pyviz_comms

# Example 5: StandardOutput and callback handling
print("Testing pyviz-comms StandardOutput:")

# Create StandardOutput instance
stdout = pyviz_comms.StandardOutput()
print(f"✓ StandardOutput created: {type(stdout).__name__}")

# Test append and list operations
stdout.append("Line 1")
stdout.append("Line 2")
print(f"✓ Appended {len(stdout)} lines: {stdout}")

# Create Comm with stdout callback
comm = pyviz_comms.Comm(id="stdout_test", on_stdout=lambda msg: print(f"stdout: {msg}"))
print(f"✓ Comm with stdout callback: {comm.id}")

# Check extension support
print(f"✓ Extension available: {pyviz_comms.extension}")
assert isinstance(stdout, list) and list(stdout) == ["Line 1", "Line 2"]
assert comm.id == "stdout_test"
```
