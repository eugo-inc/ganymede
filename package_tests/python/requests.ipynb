# requests: 5 Basic Usage Examples

```python
# Builtin retry for flaky external endpoints (httpbingo.org) — retries transient network errors + 5xx/429.
import time as _t
def _retry(_fn, _tries=5, _delay=1.0):
    for _i in range(_tries):
        try:
            _r = _fn()
            if getattr(_r, 'status_code', 200) in (429, 500, 502, 503, 504) and _i < _tries - 1:
                _t.sleep(_delay * (_i + 1)); continue
            return _r
        except Exception:
            if _i == _tries - 1: raise
            _t.sleep(_delay * (_i + 1))

# Example 1: Basic GET request
import requests
response = _retry(lambda: requests.get('https://httpbingo.org/get'))
print(response.status_code, response.json())
```

```python
# Example 2: POST request with data
import requests
response = _retry(lambda: requests.post('https://httpbingo.org/post', data={'key': 'value'}))
print(response.status_code, response.json())
```

```python
# Example 3: Custom headers
import requests
headers = {'X-Test': 'true'}
response = _retry(lambda: requests.get('https://httpbingo.org/headers', headers=headers))
print(response.status_code, response.json())
```

```python
# Example 4: Timeout and error handling
import requests

# The server waits 2 seconds before answering, so a 1-second timeout must fire
timed_out = False
try:
    response = requests.get('https://httpbingo.org/delay/2', timeout=1)
except requests.exceptions.Timeout:
    timed_out = True
    print('Request timed out!')
assert timed_out, 'a 1-second timeout must interrupt a 2-second delay'
```
