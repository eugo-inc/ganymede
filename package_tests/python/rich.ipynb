# Rich Package Examples

This notebook demonstrates 5 basic, versatile, and comprehensive examples of the `rich` package.

Rich is a Python library for rich text and beautiful formatting in the terminal.


## Example 1: Basic console output with colors and styles

```python
from rich.console import Console
from rich.text import Text

# Create a console instance
console = Console()

# Basic colored text
console.print("Hello", style="bold red")
console.print("World!", style="bold blue")

# Rich markup
console.print("[bold red]Alert![/bold red] [green]Success![/green]")
console.print("[italic]This is italic[/italic] and [underline]this is underlined[/underline]")

# Text objects for more control
text = Text("Rainbow text!", style="bold")
text.stylize("red", 0, 7)
text.stylize("blue", 8, 12)
console.print(text)

# Verify rich is working
console.print(f"Rich console initialized: {type(console).__name__}")
console.print(f"Console size: {console.size}")
```

## Example 2: Tables and structured data

```python
from rich.console import Console
from rich.table import Table

console = Console()

# Create a table
table = Table(title="Sample Data Table")
table.add_column("Name", style="cyan", no_wrap=True)
table.add_column("Age", style="magenta", justify="right")
table.add_column("City", style="green")
table.add_column("Status", style="yellow")

# Add rows
table.add_row("Alice", "25", "New York", "Active")
table.add_row("Bob", "30", "San Francisco", "Inactive")
table.add_row("Charlie", "35", "Chicago", "Active")

console.print(table)

# Simple data structure printing
data = {
    "users": ["alice", "bob", "charlie"],
    "active": True,
    "count": 42
}
console.print("\nData structure:")
console.print(data)
```

## Example 3: Progress bars and status indicators

```python
from rich.console import Console
from rich.progress import Progress, SpinnerColumn, TextColumn, BarColumn, TaskProgressColumn
from rich.status import Status
import time

console = Console()

# Simple progress bar demonstration
console.print("\nProgress Bar Demo:")
with Progress() as progress:
    task = progress.add_task("[green]Processing...", total=100)
    
    for i in range(100):
        # Simulate work
        time.sleep(0.01)
        progress.update(task, advance=1)

# Custom progress bar
console.print("\nCustom Progress Bar:")
with Progress(
    SpinnerColumn(),
    TextColumn("[progress.description]{task.description}"),
    BarColumn(),
    TaskProgressColumn(),
) as progress:
    task = progress.add_task("[cyan]Downloading...", total=50)
    
    for i in range(50):
        time.sleep(0.02)
        progress.update(task, advance=1)

# Status spinner
console.print("\nStatus Demo:")
with Status("[bold green]Working on tasks...") as status:
    time.sleep(1)
    status.update("[bold blue]Almost done...")
    time.sleep(1)

console.print("[bold green]✓[/bold green] All tasks completed!")
```

## Example 4: Panels, syntax highlighting, and markdown

```python
from rich.console import Console
from rich.panel import Panel
from rich.syntax import Syntax
from rich.markdown import Markdown

console = Console()

# Panels for organized content
console.print(Panel("[bold blue]Welcome to Rich![/bold blue]", title="Info", style="cyan"))
console.print(Panel("[red]Warning: This is important![/red]", title="Alert", style="red"))

# Syntax highlighting
python_code = '''
def hello_world():
    """A simple greeting function"""
    name = "World"
    print(f"Hello, {name}!")
    return True

if __name__ == "__main__":
    hello_world()
'''

syntax = Syntax(python_code, "python", theme="monokai", line_numbers=True)
console.print(Panel(syntax, title="Python Code", expand=False))

# Markdown rendering
markdown_text = """
# Rich Markdown Support

Rich can render **markdown** with:

- *Italic text*
- **Bold text**
- `Code snippets`
- [Links](https://github.com/Textualize/rich)

> This is a blockquote

## Features

1. Syntax highlighting
2. Tables
3. Progress bars
"""

markdown = Markdown(markdown_text)
console.print(markdown)
```

## Example 5: Error handling, logging, and advanced features

```python
from rich.console import Console
from rich.logging import RichHandler
from rich.traceback import install
from rich.tree import Tree
from rich.columns import Columns
from rich.panel import Panel
from rich.text import Text
from rich.table import Table
import logging

# Install rich traceback handler
install(show_locals=True)

console = Console()

# Set up rich logging
logging.basicConfig(
    level="INFO", 
    format="%(message)s",
    datefmt="[%X]",
    handlers=[RichHandler(console=console)]
)

logger = logging.getLogger("rich")

# Test logging
logger.info("This is an info message")
logger.warning("This is a warning message")
logger.error("This is an error message")

# Tree structure
tree = Tree("[bold blue]Project Structure")
tree.add("📁 src/")
src_branch = tree.add("📁 components/")
src_branch.add("📄 main.py")
src_branch.add("📄 utils.py")
tree.add("📄 README.md")
tree.add("📄 requirements.txt")

console.print(tree)

# Columns layout
console.print("\n[bold]Column Layout Demo:[/bold]")
panels = [
    Panel("Column 1\nContent", style="red"),
    Panel("Column 2\nMore content", style="green"),
    Panel("Column 3\nEven more", style="blue")
]
console.print(Columns(panels))

# Check the console and the core renderables
# Test that rich imports are working
assert hasattr(console, 'print'), "Console should have print method"
assert hasattr(console, 'size'), "Console should have size property"
console.print(f"✓ Rich console fully functional")

# Test that we can create various rich objects
test_text = Text("Test")
test_table = Table()
assert test_text.plain == "Test" and test_table.row_count == 0, (test_text.plain, test_table.row_count)
console.print(f"✓ Rich components importable")


console.print("\n[bold green]Rich testing completed successfully![/bold green]")
```
