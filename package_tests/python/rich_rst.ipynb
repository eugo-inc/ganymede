```python
# Example 1: Construct the RestructuredText renderable from a simple RST string.
from rich_rst import RestructuredText

rst_source = """
Hello rich_rst
==============

This is a paragraph with **bold** and *italic* text.
"""

rst = RestructuredText(rst_source)
print(type(rst).__name__)
assert isinstance(rst, RestructuredText)
```

```python
# Example 2: Render an RST document to the console via rich.
from rich.console import Console
from rich_rst import RestructuredText

rst_source = """
Section
-------

A bullet list:

- alpha
- beta
- gamma
"""

console = Console(record=True, width=80)
console.print(RestructuredText(rst_source))
output = console.export_text()
print(output)
assert "alpha" in output and "beta" in output and "gamma" in output
```

```python
# Example 3: Render an RST code block with syntax highlighting via pygments.
from rich.console import Console
from rich_rst import RestructuredText

rst_source = """
Code Sample
===========

.. code-block:: python

    def greet(name):
        return f"Hello, {name}!"
"""

console = Console(record=True, width=80)
console.print(RestructuredText(rst_source))
output = console.export_text()
assert "def greet" in output
```

```python
# Example 4: Render an RST table.
from rich.console import Console
from rich_rst import RestructuredText

rst_source = """
+--------+--------+
| Col A  | Col B  |
+========+========+
| 1      | apple  |
+--------+--------+
| 2      | banana |
+--------+--------+
"""

console = Console(record=True, width=80)
console.print(RestructuredText(rst_source))
output = console.export_text()
assert "apple" in output and "banana" in output
```

```python
# Example 5: Confirm RestructuredText implements rich's __rich_console__ protocol.
from rich.console import Console
from rich_rst import RestructuredText

rst = RestructuredText("A simple *emphasized* paragraph.")

# rich renderables expose __rich_console__
assert hasattr(rst, "__rich_console__"), "RestructuredText must be a rich renderable"

console = Console(record=True, width=80)
console.print(rst)
output = console.export_text().strip()
print(repr(output))
assert "emphasized" in output
assert "paragraph" in output
```
