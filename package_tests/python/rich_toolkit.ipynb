```python
# Example 1: Construct a RichToolkit with a theme and verify basic attributes.
from rich_toolkit import RichToolkit, RichToolkitTheme
from rich_toolkit.styles.tagged import TaggedStyle

# RichToolkitTheme takes a BaseStyle instance as its style and a mapping of
# rich style strings as its theme.
theme = RichToolkitTheme(
    style=TaggedStyle(tag_width=11),
    theme={
        "tag.title": "black on bright_blue",
        "tag": "white on bright_blue",
        "placeholder": "grey85",
        "text": "white",
        "selected": "cyan",
        "result": "grey85",
        "progress": "cyan",
        "error": "red",
        "log.info": "cyan",
    },
)

toolkit = RichToolkit(theme=theme)
assert toolkit is not None
print(f"toolkit type: {type(toolkit).__name__}")
print(f"style type: {type(toolkit.style).__name__}")
```

```python
# Example 2: Use the toolkit context manager to print text + a title.
from rich_toolkit import RichToolkit, RichToolkitTheme
from rich_toolkit.styles.tagged import TaggedStyle

# RichToolkitTheme takes a BaseStyle instance as its style and a mapping of
# rich style strings as its theme.
theme = RichToolkitTheme(
    style=TaggedStyle(tag_width=11),
    theme={
        "tag.title": "black on bright_blue",
        "tag": "white on bright_blue",
        "placeholder": "grey85",
        "text": "white",
        "selected": "cyan",
        "result": "grey85",
        "progress": "cyan",
        "error": "red",
        "log.info": "cyan",
    },
)

with RichToolkit(theme=theme) as tk:
    tk.print_title("rich_toolkit smoke test")
    tk.print("Hello from rich_toolkit")
    tk.print_line()

print("context-manager print flow OK")
```

```python
# Example 3: Inspect the styles subpackage and ensure key style classes exist.
import rich_toolkit.styles as styles_pkg
from rich_toolkit.styles.base import BaseStyle
from rich_toolkit.styles.tagged import TaggedStyle

assert issubclass(TaggedStyle, BaseStyle)
tagged = TaggedStyle()
print(f"TaggedStyle instance: {tagged!r}")
print(f"BaseStyle: {BaseStyle.__module__}.{BaseStyle.__name__}")
print(f"styles package path: {styles_pkg.__path__}")
```

```python
# Example 4: Exercise the menu element (constructor + option shape, no interactive input).
# Menu takes `label` as its first parameter; options are Option TypedDicts
# ({"name": ..., "value": ...}).
from rich_toolkit.menu import Menu
from rich_toolkit.styles.tagged import TaggedStyle

style = TaggedStyle()
menu = Menu(
    label="Pick one",
    options=[
        {"name": "Option A", "value": "a"},
        {"name": "Option B", "value": "b"},
        {"name": "Option C", "value": "c"},
    ],
    style=style,
)

# Menu stores the label as a rich.text.Text — compare its plain text.
assert str(menu.label) == "Pick one"
assert len(menu.options) == 3
print(f"Menu label: {menu.label}")  # renders as plain text
print(f"Menu option count: {len(menu.options)}")
```

```python
# Example 5: Public-API surface — confirm the documented top-level exports resolve.
import rich_toolkit

expected_names = ["RichToolkit", "RichToolkitTheme"]
resolved = {}
for name in expected_names:
    obj = getattr(rich_toolkit, name, None)
    resolved[name] = obj is not None
    if obj is not None:
        print(f"  {name}: {obj.__module__}.{obj.__name__}")

missing = [n for n, ok in resolved.items() if not ok]
assert not missing, f"Missing public API names: {missing}"
print("\nAll expected public-API names resolved.")
```
