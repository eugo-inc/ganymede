# roman_numerals

roman-numerals 4.1.0 converts integers to and from Roman numerals; Sphinx uses it to number enumerated lists and LaTeX
front matter. These cells cover conversion both ways, lower case, the range limits and invalid input, a full round
trip over every legal value, and the place Sphinx calls it.

```python
# Example 1: an integer becomes a numeral, and a numeral string parses back to the same value.
from roman_numerals import RomanNumeral

numeral = RomanNumeral(1994)
print(numeral, int(numeral))
assert str(numeral) == "MCMXCIV"
assert RomanNumeral.from_string("MCMXCIV") == numeral
assert int(RomanNumeral.from_string("MMXXVI")) == 2026
```

```python
# Example 2: upper and lower case forms, and parsing accepts either.
from roman_numerals import RomanNumeral

n = RomanNumeral(14)
print(n.to_uppercase(), n.to_lowercase())
assert n.to_uppercase() == "XIV" and n.to_lowercase() == "xiv"
assert RomanNumeral.from_string("xiv") == n
```

```python
# Example 3: the range is 1..3999, and anything outside it is refused with its own error.
from roman_numerals import MAX, MIN, OutOfRangeError, RomanNumeral

print(MIN, MAX)
assert (MIN, MAX) == (1, 3_999)
for bad in (0, 4_000, -5):
    try:
        RomanNumeral(bad)
    except OutOfRangeError as error:
        print(bad, "->", error)
    else:
        raise AssertionError(f"{bad} was accepted")
```

```python
# Example 4: invalid numerals are refused rather than guessed at.
from roman_numerals import InvalidRomanNumeralError, RomanNumeral

for bad in ("IIII", "VX", "IIX", "ABC", ""):
    try:
        RomanNumeral.from_string(bad)
    except InvalidRomanNumeralError as error:
        print(repr(bad), "->", error)
    else:
        raise AssertionError(f"{bad!r} was accepted")
```
