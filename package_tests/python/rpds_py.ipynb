## Example 1: Immutable Lists with Persistent Data Structures

```python
# Example 1: Immutable Lists with Persistent Data Structures
from rpds import List

# Create immutable list
original_list = List([1, 2, 3, 4, 5])
print("Immutable List Operations:")
print(f"Original list: {list(original_list)}")

# Prepend operation (List only supports push_front)
prepended = original_list.push_front(0)
print(f"After push_front(0): {list(prepended)}")
print(f"Original unchanged: {list(original_list)}")

# Multiple operations
extended = original_list.push_front(-2).push_front(-1)
print(f"Chained push_front: {list(extended)}")

# List access and properties
print(f"Length: {len(original_list)}")
print(f"First element: {original_list.first}")
print(f"Rest (tail): {list(original_list.rest)}")

# Drop first element
shorter = original_list.drop_first()
print(f"After drop_first: {list(shorter)}")

# Convert to Python list for full access
python_list = list(original_list)
print(f"As Python list: {python_list}")
print(f"Last element: {python_list[-1]}")

# Demonstrate structural sharing
print(f"\nMemory efficiency - original and new share structure")
print(f"Original list: {list(original_list)}")
print(f"Prepended list: {list(prepended)}")

assert list(original_list) == [1, 2, 3, 4, 5]  # unchanged by every operation above
assert list(prepended) == [0, 1, 2, 3, 4, 5]
assert list(extended) == [-1, -2, 1, 2, 3, 4, 5]
assert original_list.first == 1 and list(original_list.rest) == [2, 3, 4, 5]
assert list(shorter) == [2, 3, 4, 5]
print("✅ Immutable list operations completed")
```

## Example 2: Immutable Dictionaries and Hash Maps

```python
# Example 2: Immutable Dictionaries and Hash Maps
from rpds import HashTrieMap

# Create immutable hash map
config = HashTrieMap({"app": "myapp", "version": "1.0", "debug": False})
print("Immutable Hash Map Operations:")
print(f"Original config: {dict(config)}")

# Insert new key-value pair (returns new map)
updated_config = config.insert("port", 8080)
print(f"After adding port: {dict(updated_config)}")
print(f"Original unchanged: {dict(config)}")

# Update existing value
debug_config = config.insert("debug", True)
print(f"Debug enabled: {dict(debug_config)}")

# Remove key
minimal_config = updated_config.remove("version")
print(f"After removing version: {dict(minimal_config)}")

# Bulk updates using discard (safe removal - no error if key doesn't exist)
safe_removal = config.discard("nonexistent_key")
print(f"Safe removal of non-existent key: {dict(safe_removal)}")

# Dictionary operations
print(f"Contains 'app': {'app' in config}")
print(f"Get value with default: {config.get('timeout', 30)}")
print(f"Keys: {list(config.keys())}")
print(f"Values: {list(config.values())}")
print(f"Items: {list(config.items())}")

# Convert to Python dict
python_dict = dict(config)
print(f"As Python dict: {python_dict}")

# Demonstrate persistent updates
dev_config = config.insert("debug", True).insert("port", 3000)
prod_config = config.insert("debug", False).insert("port", 80)

print(f"\nDevelopment config: {dict(dev_config)}")
print(f"Production config: {dict(prod_config)}")
print(f"Original config unchanged: {dict(config)}")

assert dict(config) == {"app": "myapp", "version": "1.0", "debug": False}  # unchanged
assert dict(updated_config) == {"app": "myapp", "version": "1.0", "debug": False, "port": 8080}
assert dict(debug_config)["debug"] is True
assert dict(minimal_config) == {"app": "myapp", "debug": False, "port": 8080}
assert safe_removal == config
assert config.get("timeout", 30) == 30
assert dict(dev_config)["port"] == 3000 and dict(prod_config)["port"] == 80
try:
    config.remove("nonexistent_key")
except KeyError:
    pass
else:
    raise AssertionError("remove() of a missing key must raise KeyError")
print("✅ Immutable hash map operations completed")
```

## Example 3: Immutable Sets and Queue Operations

```python
# Example 3: Immutable Sets and Queue Operations
from rpds import HashTrieSet, Queue

print("Immutable Set Operations:")
# Create immutable set
permissions = HashTrieSet(["read", "write", "execute"])
print(f"Original permissions: {set(permissions)}")

# Add elements (returns new set)
admin_permissions = permissions.insert("admin")
print(f"Admin permissions: {set(admin_permissions)}")
print(f"Original unchanged: {set(permissions)}")

# Remove elements
limited_permissions = permissions.remove("execute")
print(f"Limited permissions: {set(limited_permissions)}")

# Safe removal with discard (no error if element doesn't exist)
safe_removal = permissions.discard("nonexistent")
print(f"Safe discard result: {set(safe_removal)}")

# Set membership testing
print(f"Has 'read' permission: {'read' in permissions}")
print(f"Has 'admin' permission: {'admin' in permissions}")

# Convert to Python set
python_set = set(permissions)
print(f"As Python set: {python_set}")

print("\nImmutable Queue Operations:")
# Create immutable queue
task_queue = Queue()
print(f"Empty queue length: {len(task_queue)}")

# Enqueue operations
queue_with_tasks = task_queue.enqueue("task1").enqueue("task2").enqueue("task3")
print(f"Queue with tasks: {list(queue_with_tasks)}")
print(f"Queue length: {len(queue_with_tasks)}")

# Peek at the first element (iteration starts at the front of the queue)
print(f"First task (peek): {next(iter(queue_with_tasks))}")

# Dequeue operation (returns new queue without first element)
remaining_queue = queue_with_tasks.dequeue()
print(f"After dequeue, first: {next(iter(remaining_queue))}")
print(f"Remaining queue: {list(remaining_queue)}")

# Process queue simulation
current_queue = queue_with_tasks
processed_tasks = []

print("\nProcessing all tasks:")
while len(current_queue) > 0:
    task = next(iter(current_queue))
    current_queue = current_queue.dequeue()
    processed_tasks.append(task)
    print(f"  Processing: {task}, Queue size: {len(current_queue)}")

print(f"All processed tasks: {processed_tasks}")

assert set(permissions) == {"read", "write", "execute"}  # unchanged
assert set(admin_permissions) == {"read", "write", "execute", "admin"}
assert set(limited_permissions) == {"read", "write"}
assert set(safe_removal) == set(permissions)
assert len(task_queue) == 0 and list(queue_with_tasks) == ["task1", "task2", "task3"]
assert next(iter(remaining_queue)) == "task2" and list(remaining_queue) == ["task2", "task3"]
assert processed_tasks == ["task1", "task2", "task3"]
print("✅ Immutable set and queue operations completed")
```

## Example 4: Advanced Data Structure Operations and Performance

```python
# Example 4: Advanced Data Structure Operations and Performance
from rpds import HashTrieMap, List, HashTrieSet
import time

print("Advanced Data Structure Operations:")

# Create nested immutable structures
user_data = HashTrieMap({
    "users": List([
        HashTrieMap({"id": 1, "name": "Alice", "roles": HashTrieSet(["admin", "user"])}),
        HashTrieMap({"id": 2, "name": "Bob", "roles": HashTrieSet(["user"])}),
        HashTrieMap({"id": 3, "name": "Charlie", "roles": HashTrieSet(["moderator", "user"])})
    ]),
    "metadata": HashTrieMap({
        "version": "1.0",
        "created": "2023-01-01",
        "permissions": HashTrieSet(["read", "write", "admin"])
    })
})

print("Nested Structure Navigation:")
users_list = list(user_data["users"])
print(f"Total users: {len(users_list)}")
print(f"First user: {dict(users_list[0])}")
print(f"System version: {user_data['metadata']['version']}")

# Complex update operations
print("\nComplex Updates:")
# Add new user to the list
new_user = HashTrieMap({"id": 4, "name": "Diana", "roles": HashTrieSet(["editor", "user"])})
current_users = user_data["users"]
updated_users = current_users.push_front(new_user)
updated_data = user_data.insert("users", updated_users)
print(f"Added new user, total users now: {len(list(updated_data['users']))}")

# Update specific user's roles (get first user from list)
users_py_list = list(updated_data["users"])
first_user = users_py_list[0]
updated_roles = first_user["roles"].insert("superadmin")
updated_first_user = first_user.insert("roles", updated_roles)
print(f"Updated first user roles: {set(updated_first_user['roles'])}")

# Performance comparison
print("\nPerformance Analysis:")
# Create large data structures for timing
large_dict_data = {f"key_{i}": f"value_{i}" for i in range(1000)}

# Time Python dict operations
start_time = time.time()
python_dict = large_dict_data.copy()
python_dict["new_key"] = "new_value"
python_time = time.time() - start_time

# Time RPDS HashTrieMap operations
start_time = time.time()
rpds_map = HashTrieMap(large_dict_data)
new_rpds_map = rpds_map.insert("new_key", "new_value")
rpds_time = time.time() - start_time

print(f"Python dict copy + insert: {python_time:.6f} seconds")
print(f"RPDS map insert: {rpds_time:.6f} seconds")

# Memory efficiency demonstration
print(f"\nMemory Efficiency:")
base_map = HashTrieMap(large_dict_data)
variations = []

# Create multiple variations (sharing structure)
for i in range(5):
    variation = base_map.insert(f"extra_key_{i}", f"extra_value_{i}")
    variations.append(variation)

print(f"Base map size: {len(base_map)}")
print(f"Each variation size: {len(variations[0])}")
print("✅ All variations share most of their structure efficiently")

assert len(list(user_data["users"])) == 3  # the original nested structure is unchanged
assert len(list(updated_data["users"])) == 4
assert set(updated_first_user["roles"]) == {"editor", "user", "superadmin"}
assert set(first_user["roles"]) == {"editor", "user"}
assert len(base_map) == 1000 and all(len(v) == 1001 for v in variations)
assert new_rpds_map["new_key"] == "new_value" and "new_key" not in rpds_map
print("✅ Advanced data structure operations completed")
```

## Example 5: Production Use Cases and Functional Programming Patterns

```python
# Example 5: Production Use Cases and Functional Programming Patterns
from rpds import HashTrieMap, List, HashTrieSet
from functools import reduce

print("Production Use Cases:")

# Immutable Configuration Management
class ConfigManager:
    def __init__(self, initial_config=None):
        self.config = HashTrieMap(initial_config or {})
        self.history = List([self.config])
    
    def update_config(self, **kwargs):
        """Update configuration and maintain history"""
        new_config = self.config
        for key, value in kwargs.items():
            new_config = new_config.insert(key, value)
        
        # Create new manager with updated config and history
        new_manager = ConfigManager()
        new_manager.config = new_config
        new_manager.history = self.history.push_front(new_config)
        return new_manager
    
    def get_config_at_version(self, version):
        """Get configuration at specific version (0 is most recent)"""
        history_list = list(self.history)
        if 0 <= version < len(history_list):
            return history_list[version]
        return None
    
    def rollback(self, steps=1):
        """Rollback configuration by specified steps"""
        history_list = list(self.history)
        target_version = min(steps, len(history_list) - 1)
        target_config = history_list[target_version]
        
        new_manager = ConfigManager()
        new_manager.config = target_config
        new_manager.history = self.history.push_front(target_config)
        return new_manager

# Example usage
print("Configuration Management System:")
initial_config = {"db_host": "localhost", "db_port": 5432, "debug": False}
config_mgr = ConfigManager(initial_config)
print(f"Initial config: {dict(config_mgr.config)}")

# Update configuration
config_mgr_v2 = config_mgr.update_config(debug=True, api_key="secret123")
print(f"V2 config: {dict(config_mgr_v2.config)}")

config_mgr_v3 = config_mgr_v2.update_config(db_host="production-db", cache_size=1000)
print(f"V3 config: {dict(config_mgr_v3.config)}")

# Rollback
config_mgr_rollback = config_mgr_v3.rollback(1)
print(f"Rollback config: {dict(config_mgr_rollback.config)}")

print(f"Configuration history length: {len(list(config_mgr_v3.history))}")

# Functional Programming with Immutable Structures
print("\nFunctional Programming Patterns:")

# Event sourcing simulation
events = [
    {"type": "user_created", "user_id": 1, "name": "Alice"},
    {"type": "user_updated", "user_id": 1, "name": "Alice Smith"},
    {"type": "user_created", "user_id": 2, "name": "Bob"},
    {"type": "role_assigned", "user_id": 1, "role": "admin"},
    {"type": "role_assigned", "user_id": 2, "role": "user"}
]

def apply_event(state, event):
    """Apply single event to state"""
    if event["type"] == "user_created":
        user = HashTrieMap({
            "id": event["user_id"],
            "name": event["name"],
            "roles": HashTrieSet()
        })
        return state.insert(event["user_id"], user)
    
    elif event["type"] == "user_updated":
        if event["user_id"] in state:
            user = state[event["user_id"]].insert("name", event["name"])
            return state.insert(event["user_id"], user)
    
    elif event["type"] == "role_assigned":
        if event["user_id"] in state:
            user = state[event["user_id"]]
            updated_roles = user["roles"].insert(event["role"])
            updated_user = user.insert("roles", updated_roles)
            return state.insert(event["user_id"], updated_user)
    
    return state

# Event sourcing - rebuild state from events
initial_state = HashTrieMap()
final_state = reduce(apply_event, events, initial_state)

print("Event Sourcing Results:")
for user_id, user in final_state.items():
    print(f"User {user_id}: {user['name']}, roles: {set(user['roles'])}")

# Immutable state machine
print("\nImmutable State Machine:")

class ImmutableStateMachine:
    def __init__(self, state="idle", data=None):
        self.state = state
        self.data = HashTrieMap(data or {})
    
    def transition(self, action, **kwargs):
        """Create new state machine with updated state"""
        new_machine = ImmutableStateMachine(self.state, dict(self.data))
        
        if self.state == "idle" and action == "start":
            new_machine.state = "running"
            new_machine.data = self.data.insert("start_time", kwargs.get("time", 0))
        elif self.state == "running" and action == "pause":
            new_machine.state = "paused"
            new_machine.data = self.data.insert("pause_time", kwargs.get("time", 0))
        elif self.state == "paused" and action == "resume":
            new_machine.state = "running"
        elif action == "stop":
            new_machine.state = "stopped"
            new_machine.data = self.data.insert("stop_time", kwargs.get("time", 0))
        
        return new_machine

# State machine usage
sm = ImmutableStateMachine()
print(f"Initial state: {sm.state}")

sm = sm.transition("start", time=100)
print(f"After start: {sm.state}, data: {dict(sm.data)}")

sm = sm.transition("pause", time=200)
print(f"After pause: {sm.state}, data: {dict(sm.data)}")

sm = sm.transition("resume")
print(f"After resume: {sm.state}")

sm = sm.transition("stop", time=300)
print(f"After stop: {sm.state}, data: {dict(sm.data)}")

assert dict(config_mgr.config) == initial_config  # earlier versions are unchanged
assert dict(config_mgr_rollback.config) == dict(config_mgr_v2.config)
assert len(list(config_mgr_v3.history)) == 3
assert final_state[1]["name"] == "Alice Smith" and set(final_state[1]["roles"]) == {"admin"}
assert final_state[2]["name"] == "Bob" and set(final_state[2]["roles"]) == {"user"}
assert sm.state == "stopped" and dict(sm.data) == {"start_time": 100, "pause_time": 200, "stop_time": 300}
print("✅ Production use cases and functional programming patterns completed")
```
