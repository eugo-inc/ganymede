# rsa: 5 Basic Usage Examples

```python
# Example 1: Generate RSA key pair
import rsa
(pubkey, privkey) = rsa.newkeys(512)
print(pubkey, privkey)
# The public key is (n, e); the private key holds the factors p and q of n
assert pubkey.n == privkey.n and pubkey.e == privkey.e == 65537
assert privkey.p * privkey.q == privkey.n
assert pubkey.n.bit_length() == 512, pubkey.n.bit_length()
```

```python
# Example 2: Encrypt and decrypt a message
import rsa
(pubkey, privkey) = rsa.newkeys(512)
message = b'hello'
cipher = rsa.encrypt(message, pubkey)
plain = rsa.decrypt(cipher, privkey)
print(plain)
assert plain == message, plain
assert cipher != message and len(cipher) == 64  # 512-bit key -> 64-byte ciphertext
```

```python
# Example 3: Sign and verify a message
import rsa
(pubkey, privkey) = rsa.newkeys(512)
message = b'hello'
signature = rsa.sign(message, privkey, 'SHA-256')
# verify() returns the name of the hash used for the signature, and raises if the message changed
assert rsa.verify(message, signature, pubkey) == 'SHA-256'
try:
    rsa.verify(b'hellO', signature, pubkey)
except rsa.VerificationError:
    pass
else:
    raise AssertionError('a modified message must not verify')
print('Signature verified!')
```

```python
# Example 4: Save and load keys
import rsa
(pubkey, privkey) = rsa.newkeys(512)
with open('pub.pem', 'wb') as f:
    f.write(pubkey.save_pkcs1('PEM'))
with open('priv.pem', 'wb') as f:
    f.write(privkey.save_pkcs1('PEM'))
with open('pub.pem', 'rb') as f:
    pubkey2 = rsa.PublicKey.load_pkcs1(f.read())
with open('priv.pem', 'rb') as f:
    privkey2 = rsa.PrivateKey.load_pkcs1(f.read())
print(pubkey2, privkey2)
assert pubkey2 == pubkey and privkey2 == privkey
assert rsa.decrypt(rsa.encrypt(b'round trip', pubkey2), privkey2) == b'round trip'
```

```python
# Example 5: Clean up generated key files
import os
os.remove('pub.pem')
os.remove('priv.pem')
print('Removed pub.pem and priv.pem')
```
