## Example 1: ruamel.yaml functionality test 1

```python
from ruamel.yaml import YAML
import io

# Example 1: Basic YAML loading and dumping with round-trip preservation
print("Example 1: Basic YAML operations")

try:
    yaml = YAML()
    
    # YAML string with comments and formatting
    yaml_str = """
    # Configuration file
    server:
      host: localhost  # Default host
      port: 8080
      debug: true
    
    database:
      url: "postgresql://user:pass@localhost/db"
      pool_size: 10
    """
    
    # Load YAML while preserving comments and formatting
    data = yaml.load(yaml_str)
    print(f"Loaded data type: {type(data)}")
    print(f"Server host: {data['server']['host']}")
    print(f"Database pool size: {data['database']['pool_size']}")
    
    # Dump back to YAML string
    output = io.StringIO()
    yaml.dump(data, output)
    print("\nRound-trip YAML output:")
    print(output.getvalue())
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"ruamel.yaml example (cell 1) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 2: ruamel.yaml functionality test 2

```python
from ruamel.yaml import YAML
from ruamel.yaml.comments import CommentedMap, CommentedSeq
import io

# Example 2: Advanced YAML with comments and custom data structures
print("Example 2: Comments and data structure manipulation")

try:
    yaml = YAML()
    yaml.preserve_quotes = True
    
    # Create commented data structure programmatically
    data = CommentedMap()
    data['application'] = CommentedMap()
    data['application']['name'] = 'MyApp'
    data['application']['version'] = '1.0.0'
    
    # Add comments to the data
    data.yaml_add_eol_comment('Application configuration', 'application')
    data['application'].yaml_add_eol_comment('App name', 'name')
    data['application'].yaml_add_eol_comment('Current version', 'version')
    
    # Create a list with comments
    features = CommentedSeq(['auth', 'logging', 'metrics'])
    features.yaml_add_eol_comment('Authentication module', 0)
    features.yaml_add_eol_comment('Logging system', 1)
    data['features'] = features
    
    # Output the structured YAML with comments
    output = io.StringIO()
    yaml.dump(data, output)
    result = output.getvalue()
    print("Generated YAML with comments:")
    print(result)
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"ruamel.yaml example (cell 3) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 3: ruamel.yaml functionality test 3

```python
from ruamel.yaml import YAML
from datetime import datetime, date
import io

# Example 3: Custom data types and YAML tags
print("Example 3: Custom data types and tags")

yaml = YAML()
yaml.preserve_quotes = True

# Data with various Python types (excluding Decimal)
data = {
    'string_value': '"quoted string"',
    'number': 42,
    'float_val': 3.14159,
    'boolean': True,
    'null_value': None,
    'date_val': date(2023, 12, 25),
    'list_values': [1, 2, 3, 'mixed', True],
    'nested': {'level2': {'deep_value': 'nested data'}}
}

# Convert to YAML and back
output = io.StringIO()
yaml.dump(data, output)
print(f"✓ YAML dumped: {len(output.getvalue())} chars")

# Load back
reloaded = yaml.load(output.getvalue())
print(f"✓ Date preserved: {type(reloaded['date_val'])}")
print(f"✓ Nested access: {reloaded['nested']['level2']['deep_value']}")
```

## Example 4: ruamel.yaml functionality test 4

```python
from ruamel.yaml import YAML
import io
import tempfile

# Example 4: File operations and configuration management
print("Example 4: File I/O and configuration updates")

try:
    yaml = YAML()
    yaml.indent(mapping=2, sequence=4, offset=2)
    yaml.preserve_quotes = True
    
    # Create a configuration structure
    config = {
        'environment': 'development',
        'services': {
            'web': {
                'enabled': True,
                'port': 8080,
                'workers': 4
            },
            'database': {
                'enabled': True,
                'host': 'localhost',
                'port': 5432
            }
        },
        'logging': {
            'level': 'INFO',
            'format': '%(asctime)s - %(name)s - %(levelname)s - %(message)s'
        }
    }
    
    # Write to temporary file
    with tempfile.NamedTemporaryFile(mode='w', suffix='.yaml', delete=False) as f:
        yaml.dump(config, f)
        temp_file = f.name
    
    print(f"Configuration written to: {temp_file}")
    
    # Read back and modify
    with open(temp_file, 'r') as f:
        loaded_config = yaml.load(f)
    
    # Update configuration
    loaded_config['environment'] = 'production'
    loaded_config['services']['web']['workers'] = 8
    
    # Write updated configuration
    output = io.StringIO()
    yaml.dump(loaded_config, output)
    print("Updated configuration:")
    print(output.getvalue())
    
except Exception as e:
    print(f"Error: {e}")
    raise AssertionError(
        f"ruamel.yaml example (cell 7) failed: {type(e).__name__}: {e}"
    ) from e
```

## Example 5: ruamel.yaml functionality test 5

```python
from ruamel.yaml import YAML
import io

# Example 5: Multi-document YAML
print("Example 5: Multi-document YAML processing")

yaml = YAML()

# Multi-document YAML content with proper indentation
multi_doc_yaml = """---
document: 1
title: First Document
data:
  - item1
  - item2
---
document: 2
title: Second Document
data:
  - item3
  - item4
"""

# Load all documents
documents = list(yaml.load_all(multi_doc_yaml))
print(f"✓ Loaded {len(documents)} documents")

# Process each document
for i, doc in enumerate(documents, 1):
    print(f"✓ Document {i}: {doc.get('title')}, {len(doc.get('data', []))} items")

# Create multi-document output
output = io.StringIO()
new_docs = [{'doc_id': 1, 'status': 'active'}, {'doc_id': 2, 'status': 'pending'}]
yaml.dump_all(new_docs, output)
print(f"✓ Generated multi-doc: {len(output.getvalue())} chars")
```
