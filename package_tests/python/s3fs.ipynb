## Example 1: s3fs functionality test 1

```python
# Example 1: Basic S3FileSystem Creation and Configuration
import s3fs

print("=== Basic S3 FileSystem Setup ===")

# Create S3FileSystem with anonymous access
fs_basic = s3fs.S3FileSystem(anon=True)
print(f"Created: {type(fs_basic).__name__}")
print(f"Anonymous: {fs_basic.anon}")
print(f"Protocol: {fs_basic.protocol}")

# S3FileSystem with custom configuration
fs_configured = s3fs.S3FileSystem(
    anon=True,
    cache_regions=True,
    use_ssl=True,
    client_kwargs={'region_name': 'us-east-1'}
)

print(f"\nConfigured filesystem:")
print(f"  SSL: {fs_configured.use_ssl}")
print(f"  Cache regions: {fs_configured.cache_regions}")

# Test basic interface methods availability
methods = ['ls', 'exists', 'open', 'info', 'glob']
print(f"\nAvailable methods: {', '.join(m for m in methods if hasattr(fs_basic, m))}")
print("Basic S3FileSystem setup completed!")
```

## Example 2: s3fs functionality test 2

```python
# Example 2: S3 Path Operations and URL Handling
import s3fs
from urllib.parse import urlparse

print("=== S3 Path Operations ===")

fs = s3fs.S3FileSystem(anon=True)

# Demonstrate S3 path parsing
paths = ["s3://bucket/data/file.txt", "bucket/folder/doc.pdf", "s3://dataset/logs/app.log"]

for path in paths:
    if path.startswith('s3://'):
        parsed = urlparse(path)
        bucket, key = parsed.netloc, parsed.path.lstrip('/')
    else:
        parts = path.split('/', 1)
        bucket, key = parts[0], parts[1] if len(parts) > 1 else ''
    
    print(f"Path: {path}")
    print(f"  Bucket: '{bucket}', Key: '{key}'")
    if key:
        print(f"  Filename: {key.split('/')[-1]}")

print("Path operations completed!")
```

## Example 3: s3fs functionality test 3

```python
# Example 3: S3 File Interface and Methods
import s3fs

print("=== S3 File Interface ===")

fs = s3fs.S3FileSystem(anon=True)

# Check available filesystem methods
core_methods = ['ls', 'exists', 'open', 'put', 'get', 'copy', 'rm', 'info', 'glob', 'find']
available = [m for m in core_methods if hasattr(fs, m)]

print(f"Core methods available ({len(available)}/{len(core_methods)}):")
for method in available[:5]:
    print(f"  - {method}")

# Demonstrate path manipulation
test_path = "s3://my-bucket/data/2024/report.csv"
print(f"\nPath manipulation for: {test_path}")
print(f"  Split: {fs.split_path(test_path) if hasattr(fs, 'split_path') else 'N/A'}")
print(f"  Protocol: {fs.protocol}")

# Check filesystem attributes
print(f"\nFilesystem attributes:")
print(f"  Anonymous: {fs.anon}")
print(f"  Session: {type(fs.session).__name__ if hasattr(fs, 'session') else 'N/A'}")

print("File interface check completed!")
```

## Example 4: s3fs functionality test 4

```python
# Example 4: S3 Configuration Options
import s3fs

print("=== S3 Configuration Options ===")

# Configuration with caching
fs_cached = s3fs.S3FileSystem(anon=True, cache_regions=True, use_ssl=True)
print("Cached filesystem:")
print(f"  SSL: {fs_cached.use_ssl}, Cache regions: {fs_cached.cache_regions}")

# Configuration with custom client settings
fs_custom = s3fs.S3FileSystem(
    anon=True,
    client_kwargs={'region_name': 'us-west-2'}
)
print(f"\nCustom client: configured with region settings")

# Configuration with retry settings
fs_retry = s3fs.S3FileSystem(
    anon=True,
    config_kwargs={'retries': {'max_attempts': 3}, 'read_timeout': 60}
)
print(f"Retry configuration: max 3 attempts, 60s timeout")

# Compare configurations
configs = [fs_cached, fs_custom, fs_retry]
print(f"\nCreated {len(configs)} filesystem instances with different configs")
print("Configuration options completed!")
```
