## Example 3: s3transfer functionality test 3

```python
# Example 3: Progress Callback System
import s3transfer
import time

print("=== Progress Callback System ===")

# Create a progress callback
class ProgressCallback:
    def __init__(self, name):
        self.name = name
        self.bytes_transferred = 0
        self.start_time = time.time()
    
    def __call__(self, bytes_amount):
        self.bytes_transferred += bytes_amount
        elapsed = time.time() - self.start_time
        rate = self.bytes_transferred / elapsed if elapsed > 0 else 0
        print(f"  {self.name}: {self.bytes_transferred:,} bytes, Rate: {rate:.0f} B/s")

# Test callback
callback = ProgressCallback("Test Transfer")
print("Simulating progress updates:")
for i in range(5):
    callback(1024)
    time.sleep(0.01)

print(f"\nFinal: {callback.bytes_transferred:,} bytes transferred")
print("Progress callback system completed!")
```

## Example 4: s3transfer functionality test 4

```python
# Example 4: Transfer Configuration Optimization
import s3transfer

print("=== Transfer Configuration Optimization ===")

def optimize_config(file_size_mb, network_speed_mbps=10):
    """Optimize transfer config based on file size and network"""
    file_size = file_size_mb * 1024 * 1024
    
    # Determine thresholds
    if file_size > 100 * 1024 * 1024:  # >100MB
        threshold = 25 * 1024 * 1024
        concurrency = 10
    elif file_size > 10 * 1024 * 1024:  # >10MB
        threshold = 5 * 1024 * 1024
        concurrency = 5
    else:
        threshold = 1 * 1024 * 1024
        concurrency = 3
    
    return s3transfer.TransferConfig(
        multipart_threshold=threshold,
        max_concurrency=concurrency
    )

# Test optimization
test_sizes = [5, 50, 200]  # MB
for size in test_sizes:
    config = optimize_config(size)
    print(f"\nFile size: {size}MB - Threshold: {config.multipart_threshold/(1024*1024):.0f}MB, Concurrency: {config.max_concurrency}")

print("Configuration optimization completed!")
```
