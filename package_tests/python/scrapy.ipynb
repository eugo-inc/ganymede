```python
# Example 1: Define a basic Spider subclass and check its attributes.
import scrapy

class QuotesSpider(scrapy.Spider):
    name = 'quotes'
    start_urls = ['https://example.com/']

    def parse(self, response):
        yield {'title': response.css('title::text').get()}

assert QuotesSpider.name == 'quotes'
assert 'https://example.com/' in QuotesSpider.start_urls
assert callable(QuotesSpider.parse)
print(f"Spider class: {QuotesSpider.__name__}, name={QuotesSpider.name}")
```

```python
# Example 2: Define a scrapy.Item with Field declarations.
import scrapy
from scrapy import Field, Item

class Product(Item):
    name = Field()
    price = Field()
    tags = Field()

p = Product()
p['name'] = 'widget'
p['price'] = 9.99
p['tags'] = ['a', 'b']

assert p['name'] == 'widget'
assert p['price'] == 9.99
assert set(p.fields.keys()) == {'name', 'price', 'tags'}
print(f"Item populated: {dict(p)}")
```

```python
# Example 3: Parse HTML with scrapy.http.HtmlResponse + selectors.
from scrapy.http import HtmlResponse

body = (
    b'<html><body>'
    b'<h1 class="title">Hello</h1>'
    b'<ul><li>one</li><li>two</li><li>three</li></ul>'
    b'<a href="/next">next</a>'
    b'</body></html>'
)
response = HtmlResponse(url='https://example.com/', body=body, encoding='utf-8')

title = response.css('h1.title::text').get()
items = response.css('li::text').getall()
link = response.xpath('//a/@href').get()

assert title == 'Hello'
assert items == ['one', 'two', 'three']
assert link == '/next'
print(f"title={title!r} items={items} link={link!r}")
```
