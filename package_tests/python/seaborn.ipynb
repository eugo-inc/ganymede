## Example 1: seaborn functionality test 1

```python
# Example 1: Basic statistical visualization with built-in datasets
import seaborn as sns
import matplotlib
matplotlib.use('Agg')  # Use non-GUI backend for Jupyter
import matplotlib.pyplot as plt
import pandas as pd

# Set seaborn style and color palette
sns.set_style("whitegrid")
sns.set_palette("husl")

# Load built-in dataset
tips = sns.load_dataset('tips')
print(f"Tips dataset shape: {tips.shape}")
print(f"Columns: {list(tips.columns)}")
print(f"First few rows:")
print(tips.head())

# Create a simple scatter plot with regression line
plt.figure(figsize=(10, 6))

# Subplot 1: Scatter plot with regression line
plt.subplot(1, 2, 1)
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time')
sns.regplot(data=tips, x='total_bill', y='tip', scatter=False, color='red', ax=plt.gca())
plt.title('Tips vs Total Bill with Regression Line')

# Subplot 2: Box plot by day
plt.subplot(1, 2, 2)
sns.boxplot(data=tips, x='day', y='total_bill')
plt.title('Total Bill Distribution by Day')
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig('/tmp/seaborn_basic.png', dpi=80, bbox_inches='tight')
plt.close()

# Basic statistics
print(f"\nDataset statistics:")
print(f"Average tip: ${tips['tip'].mean():.2f}")
print(f"Average total bill: ${tips['total_bill'].mean():.2f}")
print(f"Tip percentage: {(tips['tip'] / tips['total_bill'] * 100).mean():.1f}%")

# Correlation analysis
correlation = tips[['total_bill', 'tip', 'size']].corr()
print(f"\nCorrelation matrix:")
print(correlation.round(3))

print("Example 1 completed successfully - basic statistical plots created")
```

## Example 2: seaborn functionality test 2

```python
# Example 2: Advanced distribution and categorical plots
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Load the flights dataset for time series visualization
flights = sns.load_dataset('flights')
iris = sns.load_dataset('iris')

print(f"Flights dataset: {flights.shape}")
print(f"Iris dataset: {iris.shape}")

# Create a complex figure with multiple plot types
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# 1. Heatmap with correlation matrix
plt.sca(axes[0, 0])
iris_corr = iris.select_dtypes(include=[np.number]).corr()
sns.heatmap(iris_corr, annot=True, cmap='coolwarm', center=0, square=True)
plt.title('Iris Features Correlation Heatmap')

# 2. Distribution plot with multiple categories
plt.sca(axes[0, 1])
sns.violinplot(data=iris, x='species', y='petal_length', inner='box')
plt.title('Petal Length Distribution by Species')
plt.xticks(rotation=45)

# 3. Time series heatmap
plt.sca(axes[1, 0])
flights_pivot = flights.pivot_table(values='passengers', index='month', columns='year')
sns.heatmap(flights_pivot, cmap='YlOrRd', linewidths=0.5)
plt.title('Flight Passengers Heatmap (Year vs Month)')

# 4. Pair plot style visualization
plt.sca(axes[1, 1])
# Create a subset for faster plotting
iris_subset = iris.sample(n=50, random_state=42)
sns.scatterplot(data=iris_subset, x='sepal_length', y='sepal_width', 
                hue='species', style='species', s=100)
plt.title('Sepal Dimensions by Species')

plt.tight_layout()
plt.savefig('/tmp/seaborn_advanced.png', dpi=80, bbox_inches='tight')
plt.close()

# Statistical analysis
print(f"\nFlights dataset analysis:")
print(f"Time range: {flights['year'].min()} to {flights['year'].max()}")
print(f"Average passengers per month: {flights['passengers'].mean():.0f}")
print(f"Peak month: {flights.loc[flights['passengers'].idxmax(), 'month']}")

print(f"\nIris dataset analysis:")
species_stats = iris.groupby('species').agg({
    'petal_length': ['mean', 'std'],
    'sepal_length': ['mean', 'std']
}).round(2)
print("Species statistics:")
print(species_stats)

# Create categorical count plot
plt.figure(figsize=(8, 4))
sns.countplot(data=iris, x='species', palette='Set2')
plt.title('Species Distribution in Iris Dataset')
plt.ylabel('Count')
plt.savefig('/tmp/seaborn_counts.png', dpi=80, bbox_inches='tight')
plt.close()

print("Example 2 completed successfully - advanced visualizations created")
```

## Example 3: seaborn functionality test 3

```python
# Example 3: Statistical modeling and regression visualization
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from scipy import stats

# Create synthetic data for regression analysis
np.random.seed(42)
n_samples = 200

# Generate correlated data with noise
x = np.random.normal(0, 1, n_samples)
y = 2.5 * x + 1.5 + np.random.normal(0, 0.8, n_samples)
category = np.random.choice(['A', 'B', 'C'], n_samples, p=[0.4, 0.35, 0.25])
group = np.random.choice(['Group1', 'Group2'], n_samples)

# Create DataFrame
df = pd.DataFrame({
    'x_var': x,
    'y_var': y,
    'category': category,
    'group': group,
    'x_squared': x**2,
    'interaction': x * (category == 'A').astype(int)
})

print(f"Generated dataset shape: {df.shape}")
print(f"Category distribution: {df['category'].value_counts().to_dict()}")

# Create comprehensive regression analysis
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Basic regression plot with confidence intervals
plt.sca(axes[0, 0])
sns.regplot(data=df, x='x_var', y='y_var', scatter_kws={'alpha': 0.6})
plt.title('Linear Regression with Confidence Interval')
plt.xlabel('X Variable')
plt.ylabel('Y Variable')

# 2. Regression by category with different slopes
plt.sca(axes[0, 1])
sns.lmplot(data=df, x='x_var', y='y_var', hue='category', 
          height=4, aspect=1, scatter_kws={'alpha': 0.6})
plt.gca().set_title('Regression Lines by Category')
plt.savefig('/tmp/seaborn_regression_by_category.png', dpi=80, bbox_inches='tight')
plt.close()

# Recreate the subplot since lmplot created a new figure
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Redo subplot 1
plt.sca(axes[0, 0])
sns.regplot(data=df, x='x_var', y='y_var', scatter_kws={'alpha': 0.6})
plt.title('Linear Regression with Confidence Interval')

# 2. Residuals plot for model diagnostics
plt.sca(axes[0, 1])
sns.residplot(data=df, x='x_var', y='y_var', scatter_kws={'alpha': 0.6})
plt.title('Residuals Plot')
plt.axhline(y=0, color='red', linestyle='--', alpha=0.7)

# 3. Joint plot style correlation
plt.sca(axes[1, 0])
sns.scatterplot(data=df, x='x_var', y='y_var', hue='group', alpha=0.7)
plt.title('Scatter Plot by Group')

# 4. Distribution comparison
plt.sca(axes[1, 1])
sns.boxplot(data=df, x='category', y='y_var')
sns.stripplot(data=df, x='category', y='y_var', color='black', alpha=0.3, size=3)
plt.title('Y Variable Distribution by Category')

plt.tight_layout()
plt.savefig('/tmp/seaborn_statistical.png', dpi=80, bbox_inches='tight')
plt.close()

# Statistical analysis and model fitting
correlation = stats.pearsonr(df['x_var'], df['y_var'])
print(f"\nStatistical Analysis:")
print(f"Pearson correlation: r={correlation[0]:.4f}, p-value={correlation[1]:.4f}")

# Group-wise analysis
group_stats = df.groupby('category')['y_var'].agg(['mean', 'std', 'count'])
print(f"\nGroup statistics:")
print(group_stats.round(3))

# Linear regression coefficients (simple approximation)
slope, intercept, r_value, p_value, std_err = stats.linregress(df['x_var'], df['y_var'])
print(f"\nLinear regression results:")
print(f"Slope: {slope:.4f} ± {std_err:.4f}")
print(f"Intercept: {intercept:.4f}")
print(f"R-squared: {r_value**2:.4f}")
print(f"P-value: {p_value:.4e}")

# Create distribution plots
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
sns.histplot(data=df, x='y_var', hue='category', multiple='stack', alpha=0.7)
plt.title('Y Variable Distribution by Category')

plt.subplot(1, 2, 2)
sns.kdeplot(data=df, x='y_var', hue='category', multiple='layer', alpha=0.7)
plt.title('Y Variable Density by Category')

plt.tight_layout()
plt.savefig('/tmp/seaborn_distributions.png', dpi=80, bbox_inches='tight')
plt.close()

print("Example 3 completed successfully - statistical modeling visualizations created")
```

## Example 4: seaborn functionality test 4

```python
# Example 4: Multi-panel grid plots and advanced styling
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Load multiple datasets for comprehensive analysis
titanic = sns.load_dataset('titanic')
mpg = sns.load_dataset('mpg')

print(f"Titanic dataset: {titanic.shape}")
print(f"MPG dataset: {mpg.shape}")
print(f"Titanic columns: {list(titanic.columns)}")

# Create FacetGrid for multi-panel visualization
sns.set_style("darkgrid")
sns.set_context("notebook", font_scale=0.9)

# 1. FacetGrid with survival analysis
g1 = sns.FacetGrid(titanic, col='sex', row='class', hue='survived', 
                   height=3, aspect=1, margin_titles=True)
g1.map_dataframe(sns.scatterplot, x='age', y='fare', alpha=0.7, s=50)
g1.add_legend(title='Survived', bbox_to_anchor=(1.05, 1), loc='upper left')
g1.fig.suptitle('Titanic: Age vs Fare by Class and Gender', y=1.02, fontsize=14)
plt.savefig('/tmp/seaborn_facetgrid.png', dpi=80, bbox_inches='tight')
plt.close()

# 2. PairGrid for correlation analysis
# Use a subset of MPG data for faster processing
mpg_numeric = mpg[['mpg', 'displacement', 'horsepower', 'weight']].dropna()
mpg_subset = mpg_numeric.sample(n=200, random_state=42)

g2 = sns.PairGrid(mpg_subset, height=2.5)
g2.map_upper(sns.scatterplot, alpha=0.6)
g2.map_lower(sns.kdeplot, fill=True, alpha=0.7)
g2.map_diag(sns.histplot, kde=True, alpha=0.7)
plt.savefig('/tmp/seaborn_pairplot.png', dpi=80, bbox_inches='tight')
plt.close()

# 3. Custom color palette and styling
custom_palette = sns.color_palette("husl", n_colors=len(titanic['class'].unique()))
sns.set_palette(custom_palette)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Age distribution by class
plt.sca(axes[0, 0])
sns.histplot(data=titanic, x='age', hue='class', multiple='layer', alpha=0.7)
plt.title('Age Distribution by Passenger Class')

# Survival rate by different factors
plt.sca(axes[0, 1])
survival_by_class = titanic.groupby(['class', 'sex'])['survived'].mean().reset_index()
sns.barplot(data=survival_by_class, x='class', y='survived', hue='sex')
plt.title('Survival Rate by Class and Gender')
plt.ylabel('Survival Rate')

# Fare distribution with violin plots
plt.sca(axes[1, 0])
sns.violinplot(data=titanic.dropna(subset=['age']), x='class', y='age', 
               hue='survived', split=True, inner='quart')
plt.title('Age Distribution by Class and Survival')

# Correlation heatmap for MPG data
plt.sca(axes[1, 1])
mpg_corr = mpg_numeric.corr()
mask = np.triu(np.ones_like(mpg_corr, dtype=bool))
sns.heatmap(mpg_corr, mask=mask, annot=True, cmap='RdBu_r', center=0,
            square=True, fmt='.2f', cbar_kws={'shrink': 0.8})
plt.title('MPG Dataset Correlation Matrix')

plt.tight_layout()
plt.savefig('/tmp/seaborn_styled.png', dpi=80, bbox_inches='tight')
plt.close()

# Statistical summaries
print(f"\nTitanic Dataset Analysis:")
print(f"Overall survival rate: {titanic['survived'].mean():.2%}")
survival_by_class = titanic.groupby('class')['survived'].agg(['mean', 'count'])
print("Survival by class:")
print(survival_by_class.round(3))

print(f"\nMPG Dataset Analysis:")
print(f"Average MPG: {mpg['mpg'].mean():.1f}")
mpg_by_origin = mpg.groupby('origin').agg({
    'mpg': ['mean', 'std'], 
    'horsepower': ['mean', 'std']
}).round(2)
print("MPG and Horsepower by origin:")
print(mpg_by_origin)

# Advanced plot: Ridge plot style with multiple distributions
plt.figure(figsize=(12, 6))

# Create a grid of subplots for ridge plot effect
classes = titanic['class'].unique()
fig, axes = plt.subplots(len(classes), 1, figsize=(10, 6), sharex=True)

for i, cls in enumerate(classes):
    data = titanic[titanic['class'] == cls]['age'].dropna()
    sns.kdeplot(data, ax=axes[i], fill=True, alpha=0.7, label=f'{cls} class')
    axes[i].set_ylabel(f'{cls}')
    axes[i].legend()

plt.xlabel('Age')
plt.suptitle('Age Distribution by Passenger Class (Ridge Plot Style)', fontsize=14)
plt.tight_layout()
plt.savefig('/tmp/seaborn_ridge.png', dpi=80, bbox_inches='tight')
plt.close()

print("Example 4 completed successfully - multi-panel grids and advanced styling created")
```

## Example 5: seaborn functionality test 5

```python
# Example 5: Time series visualization and custom themes
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Create time series data for visualization
np.random.seed(42)
dates = pd.date_range('2020-01-01', '2023-12-31', freq='D')
n_days = len(dates)

# Generate synthetic time series with trends and seasonality
trend = np.linspace(100, 150, n_days)
seasonal = 10 * np.sin(2 * np.pi * np.arange(n_days) / 365.25)
noise = np.random.normal(0, 5, n_days)
values = trend + seasonal + noise

# Create additional synthetic metrics
metric_a = values + np.random.normal(0, 8, n_days)
metric_b = values * 0.8 + np.random.normal(20, 6, n_days)
category = np.random.choice(['Product_A', 'Product_B', 'Product_C'], n_days, p=[0.4, 0.3, 0.3])

time_df = pd.DataFrame({
    'date': dates,
    'primary_metric': values,
    'metric_a': metric_a,
    'metric_b': metric_b,
    'category': category,
    'year': dates.year,
    'month': dates.month,
    'day_of_year': dates.day_of_year
})

print(f"Time series dataset: {time_df.shape}")
print(f"Date range: {time_df['date'].min()} to {time_df['date'].max()}")
print(f"Categories: {time_df['category'].value_counts().to_dict()}")

# Custom theme and styling
plt.style.use('seaborn-v0_8')
custom_colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4', '#FFEAA7']
sns.set_palette(custom_colors)

# 1. Time series line plot with confidence intervals
fig, axes = plt.subplots(3, 2, figsize=(16, 12))

# Main time series
plt.sca(axes[0, 0])
monthly_data = time_df.groupby([time_df['date'].dt.to_period('M'), 'category'])['primary_metric'].mean().reset_index()
monthly_data['date'] = monthly_data['date'].dt.to_timestamp()

for cat in time_df['category'].unique():
    cat_data = monthly_data[monthly_data['category'] == cat]
    plt.plot(cat_data['date'], cat_data['primary_metric'], label=cat, linewidth=2, alpha=0.8)

plt.title('Primary Metric Time Series by Category')
plt.xlabel('Date')
plt.ylabel('Primary Metric')
plt.legend()
plt.xticks(rotation=45)

# Seasonal decomposition visualization
plt.sca(axes[0, 1])
quarterly_data = time_df.groupby([time_df['date'].dt.quarter, 'year'])['primary_metric'].mean().reset_index()
sns.boxplot(data=time_df, x=time_df['date'].dt.month, y='primary_metric')
plt.title('Monthly Seasonality Pattern')
plt.xlabel('Month')
plt.ylabel('Primary Metric')

# Year-over-year comparison
plt.sca(axes[1, 0])
yearly_pivot = time_df.pivot_table(
    values='primary_metric', 
    index=time_df['date'].dt.day_of_year, 
    columns=time_df['date'].dt.year,
    aggfunc='mean'
)
for year in yearly_pivot.columns:
    plt.plot(yearly_pivot.index, yearly_pivot[year], label=f'{year}', alpha=0.8, linewidth=2)
plt.title('Year-over-Year Comparison')
plt.xlabel('Day of Year')
plt.ylabel('Primary Metric')
plt.legend()

# Correlation heatmap of metrics
plt.sca(axes[1, 1])
metrics_corr = time_df[['primary_metric', 'metric_a', 'metric_b', 'day_of_year']].corr()
sns.heatmap(metrics_corr, annot=True, cmap='RdBu_r', center=0, square=True, 
            fmt='.3f', cbar_kws={'shrink': 0.8})
plt.title('Metrics Correlation Matrix')

# Distribution analysis
plt.sca(axes[2, 0])
time_df['quarter'] = time_df['date'].dt.quarter
sns.violinplot(data=time_df, x='quarter', y='primary_metric', inner='box', alpha=0.7)
plt.title('Quarterly Distribution of Primary Metric')
plt.xlabel('Quarter')

# Anomaly detection visualization
plt.sca(axes[2, 1])
rolling_mean = time_df['primary_metric'].rolling(window=30).mean()
rolling_std = time_df['primary_metric'].rolling(window=30).std()
upper_bound = rolling_mean + 2 * rolling_std
lower_bound = rolling_mean - 2 * rolling_std

# Plot subset for clarity
subset_idx = slice(0, 500)  # First 500 days
plt.plot(time_df['date'][subset_idx], time_df['primary_metric'][subset_idx], 
         label='Original', alpha=0.7, linewidth=1)
plt.plot(time_df['date'][subset_idx], rolling_mean[subset_idx], 
         label='30-day MA', linewidth=2, color='red')
plt.fill_between(time_df['date'][subset_idx], 
                 lower_bound[subset_idx], upper_bound[subset_idx], 
                 alpha=0.2, color='gray', label='±2σ bounds')

# Mark potential anomalies
anomalies = (time_df['primary_metric'] > upper_bound) | (time_df['primary_metric'] < lower_bound)
anomaly_points = time_df[anomalies][subset_idx]
if len(anomaly_points) > 0:
    plt.scatter(anomaly_points['date'], anomaly_points['primary_metric'], 
                color='orange', s=50, label='Potential Anomalies', zorder=5)

plt.title('Anomaly Detection with Rolling Statistics')
plt.xlabel('Date')
plt.ylabel('Primary Metric')
plt.legend()
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig('/tmp/seaborn_timeseries.png', dpi=80, bbox_inches='tight')
plt.close()

# Statistical summary
print(f"\nTime Series Analysis:")
print(f"Mean primary metric: {time_df['primary_metric'].mean():.2f}")
print(f"Std primary metric: {time_df['primary_metric'].std():.2f}")

# Seasonal analysis
seasonal_stats = time_df.groupby(time_df['date'].dt.quarter)['primary_metric'].agg(['mean', 'std'])
print(f"\nQuarterly statistics:")
print(seasonal_stats.round(2))

# Trend analysis (simple linear trend)
from scipy.stats import linregress
x_numeric = np.arange(len(time_df))
slope, intercept, r_value, p_value, std_err = linregress(x_numeric, time_df['primary_metric'])
print(f"\nTrend analysis:")
print(f"Daily trend: {slope:.4f} units/day")
print(f"Annual trend: {slope * 365.25:.2f} units/year")
print(f"R-squared: {r_value**2:.4f}")

# Category performance comparison
category_performance = time_df.groupby('category')['primary_metric'].agg(['mean', 'std', 'min', 'max'])
print(f"\nCategory performance:")
print(category_performance.round(2))

print("Example 5 completed successfully - time series analysis and custom themes created")
```
