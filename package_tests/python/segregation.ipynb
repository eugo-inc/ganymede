## Example 2: segregation functionality test 2

```python
# Example 2: Dissimilarity Index Calculation
import pandas as pd
import numpy as np

print("=== Dissimilarity Index ===")

# Create demographic data
np.random.seed(42)
df = pd.DataFrame({
    'group_1': np.random.randint(100, 800, 20),
    'group_2': np.random.randint(50, 600, 20)
})

# Manual dissimilarity calculation
total_g1 = df['group_1'].sum()
total_g2 = df['group_2'].sum()
dissim = 0.5 * sum(abs(df['group_1']/total_g1 - df['group_2']/total_g2))

print(f"Group 1 total: {total_g1:,}")
print(f"Group 2 total: {total_g2:,}")
print(f"Dissimilarity Index: {dissim:.4f}")
print(f"Interpretation: {dissim*100:.1f}% would need to move for even distribution")

print("Example 2 completed!")
```

## Example 3: segregation functionality test 3

```python
# Example 3: Exposure and Isolation Indices
import pandas as pd
import numpy as np

print("=== Exposure and Isolation ===")

# Create demographic data
np.random.seed(42)
df = pd.DataFrame({
    'group_a': np.random.randint(100, 600, 15),
    'group_b': np.random.randint(80, 500, 15)
})
df['total'] = df['group_a'] + df['group_b']

# Calculate exposure index (group A to group B)
total_a = df['group_a'].sum()
exposure = sum((df['group_a'] / total_a) * (df['group_b'] / df['total']))

# Calculate isolation index (group A to itself)
isolation = sum((df['group_a'] / total_a) * (df['group_a'] / df['total']))

print(f"Exposure (A to B): {exposure:.4f}")
print(f"Isolation (A): {isolation:.4f}")
print(f"Sum: {exposure + isolation:.4f}")

print("Example 3 completed!")
```

## Example 4: segregation functionality test 4

```python
# Example 4: Entropy and Diversity Measures
import pandas as pd
import numpy as np

print("=== Entropy and Diversity ===")

# Create multi-group data
np.random.seed(42)
df = pd.DataFrame({
    'white': np.random.randint(100, 500, 12),
    'black': np.random.randint(50, 400, 12),
    'hispanic': np.random.randint(30, 300, 12)
})
df['total'] = df[['white', 'black', 'hispanic']].sum(axis=1)

# Calculate entropy for each area
def calc_entropy(row):
    props = [row['white']/row['total'], row['black']/row['total'], row['hispanic']/row['total']]
    return -sum(p * np.log(p) if p > 0 else 0 for p in props)

df['entropy'] = df.apply(calc_entropy, axis=1)

print(f"Average entropy: {df['entropy'].mean():.4f}")
print(f"Max entropy (most diverse): {df['entropy'].max():.4f}")
print(f"Min entropy (least diverse): {df['entropy'].min():.4f}")

print("Example 4 completed!")
```

## Example 5: segregation functionality test 5

```python
# Example 5: Comparative Segregation Analysis
import pandas as pd
import numpy as np

print("=== Comparative Analysis ===")

# Create data for two cities
np.random.seed(42)
city_a = pd.DataFrame({'g1': np.random.randint(200, 800, 10), 'g2': np.random.randint(50, 300, 10)})
city_b = pd.DataFrame({'g1': np.random.randint(150, 400, 10), 'g2': np.random.randint(150, 400, 10)})

# Calculate dissimilarity for both
def dissim(df):
    t1, t2 = df['g1'].sum(), df['g2'].sum()
    return 0.5 * sum(abs(df['g1']/t1 - df['g2']/t2))

d_a = dissim(city_a)
d_b = dissim(city_b)

print(f"City A dissimilarity: {d_a:.4f}")
print(f"City B dissimilarity: {d_b:.4f}")
print(f"Difference: {abs(d_a - d_b):.4f}")
print(f"More segregated: {'City A' if d_a > d_b else 'City B'}")

print("Example 5 completed!")
```
