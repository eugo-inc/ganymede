## Example 1: Version creation

```python
# Comprehensive version creation and parsing examples
import semantic_version
from typing import List, Optional

# Basic version creation
v1 = semantic_version.Version('1.2.3')
print(f"Basic version: {v1}")

# Version with prerelease and build metadata
v2 = semantic_version.Version('2.0.0-alpha.1+build.123')
print(f"Complex version: {v2}")

# Access individual components
print(f"\nVersion components for {v1}:")
print(f"  Major: {v1.major}")
print(f"  Minor: {v1.minor}")
print(f"  Patch: {v1.patch}")
print(f"  Prerelease: {v1.prerelease}")
print(f"  Build: {v1.build}")

print(f"\nVersion components for {v2}:")
print(f"  Major: {v2.major}")
print(f"  Minor: {v2.minor}")
print(f"  Patch: {v2.patch}")
print(f"  Prerelease: {v2.prerelease}")
print(f"  Build: {v2.build}")

# Create versions from individual components
v3 = semantic_version.Version(major=3, minor=1, patch=4, prerelease=['rc', '2'])
print(f"\nVersion from components: {v3}")

# Parse and validate different version formats
version_strings = [
    '1.0.0',
    '2.1.3-beta',
    '1.0.0-alpha.beta',
    '10.2.3-DEV.SNAPSHOT+build.1',
    '1.2.3-alpha.10.beta.0+build.unicorn.rainbow'
]

print(f"\nParsing various version formats:")
for version_str in version_strings:
    try:
        version = semantic_version.Version(version_str)
        print(f"  ✓ {version_str} -> {version}")
    except ValueError as e:
        print(f"  ✗ {version_str} -> Invalid: {e}")

# Demonstrate coercion of non-semantic versions
print(f"\nCoercing non-semantic versions:")
non_semantic_versions = ['1.2', '1', '1.2.3.4', '1.2-SNAPSHOT']
for version_str in non_semantic_versions:
    try:
        # Try to coerce non-standard versions
        coerced = semantic_version.Version.coerce(version_str)
        print(f"  {version_str} -> {coerced}")
    except ValueError as e:
        print(f"  {version_str} -> Cannot coerce: {e}")
```

## Example 2: Version comparison

```python
# Comprehensive version comparison and sorting
import semantic_version
from typing import List

# Create various versions for comparison
versions_data = [
    '1.0.0',
    '1.0.1',
    '1.1.0',
    '2.0.0',
    '2.0.0-alpha',
    '2.0.0-alpha.1',
    '2.0.0-alpha.beta',
    '2.0.0-beta',
    '2.0.0-beta.2',
    '2.0.0-beta.11',
    '2.0.0-rc.1',
    '2.0.0+build.1',
    '2.0.0+build.2'
]

versions = [semantic_version.Version(v) for v in versions_data]

print("Version comparison examples:")
v1 = semantic_version.Version('1.2.3')
v2 = semantic_version.Version('2.0.0')
v3 = semantic_version.Version('1.2.3-alpha')
v4 = semantic_version.Version('1.2.3+build.1')

print(f"  {v1} < {v2}: {v1 < v2}")
print(f"  {v1} > {v3}: {v1 > v3}")  # Release > prerelease
print(f"  {v1} == {v4}: {v1 == v4}")  # Build metadata ignored in comparison
print(f"  {v1} != {v4}: {v1 != v4}")

# Demonstrate precedence rules
print(f"\nPrecedence rules demonstration:")
precedence_examples = [
    ('1.0.0-alpha', '1.0.0-alpha.1'),
    ('1.0.0-alpha.1', '1.0.0-alpha.beta'),
    ('1.0.0-alpha.beta', '1.0.0-beta'),
    ('1.0.0-beta', '1.0.0-beta.2'),
    ('1.0.0-beta.2', '1.0.0-beta.11'),
    ('1.0.0-beta.11', '1.0.0-rc.1'),
    ('1.0.0-rc.1', '1.0.0'),
]

for v_lower, v_higher in precedence_examples:
    lower = semantic_version.Version(v_lower)
    higher = semantic_version.Version(v_higher)
    print(f"  {lower} < {higher}: {lower < higher}")

# Sort versions
print(f"\nSorting versions:")
unsorted_versions = [
    '2.0.0',
    '1.0.0',
    '2.0.0-alpha',
    '1.2.0',
    '2.0.0-beta',
    '1.0.1',
    '2.0.0-rc.1'
]

sorted_versions = sorted([semantic_version.Version(v) for v in unsorted_versions])
print("Unsorted:", unsorted_versions)
print("Sorted:  ", [str(v) for v in sorted_versions])

# Find latest version
print(f"\nFinding latest version:")
latest = max(versions)
print(f"Latest version: {latest}")

# Filter versions by major version
print(f"\nFiltering by major version:")
major_2_versions = [v for v in versions if v.major == 2]
print(f"Major version 2: {[str(v) for v in major_2_versions]}")

# Find compatible versions (same major.minor)
target_version = semantic_version.Version('2.0.5')
compatible = [v for v in versions if v.major == target_version.major and v.minor == target_version.minor]
print(f"Compatible with {target_version} (same major.minor): {[str(v) for v in compatible]}")
```

## Example 3: Version specs

```python
# Advanced version specifications and matching
import semantic_version
from typing import List

# Create test versions
test_versions = [
    '0.9.0',
    '1.0.0',
    '1.0.1',
    '1.1.0',
    '1.2.3',
    '2.0.0-alpha',
    '2.0.0',
    '2.1.0',
    '3.0.0'
]

versions = [semantic_version.Version(v) for v in test_versions]

print("Version specification examples:")

# Basic specifications
specs_examples = [
    '>=1.0.0',           # Greater than or equal
    '<2.0.0',            # Less than
    '~=1.2.0',           # Compatible release (>= 1.2.0, < 1.3.0)
    '==1.2.3',           # Exact match
    '!=2.0.0',           # Not equal
    '>1.0.0,<2.0.0',     # Range specification
    '>=1.0.0,!=1.2.3',   # Multiple conditions
]

for spec_str in specs_examples:
    spec = semantic_version.Spec(spec_str)
    matching = [str(v) for v in versions if v in spec]
    print(f"  {spec_str:20} matches: {matching}")

# Complex specifications
print(f"\nComplex specification examples:")
complex_specs = [
    '>=1.0.0,<2.0.0,!=1.2.3',  # Range with exclusion
    '~=1.2',                    # Compatible release operator
    '^1.2.3',                   # Caret range (npm-style, if supported)
    '>=2.0.0-alpha',            # Including prereleases
]

for spec_str in complex_specs:
    try:
        spec = semantic_version.Spec(spec_str)
        matching = [str(v) for v in versions if v in spec]
        print(f"  {spec_str:25} matches: {matching}")
    except ValueError as e:
        print(f"  {spec_str:25} invalid: {e}")

# Demonstrate specification parsing and validation
print(f"\nSpecification validation:")
spec_strings = [
    '>=1.0.0',
    '>=1.0.0,<2.0.0',
    '~=1.2.0',
    '==1.2.*',  # May not be supported
    'invalid_spec'
]

for spec_str in spec_strings:
    try:
        spec = semantic_version.Spec(spec_str)
        print(f"  ✓ '{spec_str}' -> {spec}")
    except ValueError as e:
        print(f"  ✗ '{spec_str}' -> Invalid: {e}")

# Practical use case: dependency resolution
print(f"\nPractical example - Package dependency resolution:")
package_versions = {
    'requests': ['2.25.1', '2.26.0', '2.27.0', '2.28.0'],
    'urllib3': ['1.26.0', '1.26.5', '1.26.8', '1.26.12'],
    'certifi': ['2021.5.30', '2021.10.8', '2022.5.18', '2022.9.24']
}

# Define dependency requirements
requirements = {
    'requests': '>=2.26.0,<3.0.0',
    'urllib3': '~=1.26.0',
    'certifi': '>=2021.1.1'
}

print("Available versions and requirements:")
for package, requirement in requirements.items():
    spec = semantic_version.Spec(requirement)
    available = package_versions[package]
    compatible = []
    
    for version_str in available:
        try:
            version = semantic_version.Version(version_str)
            if version in spec:
                compatible.append(version_str)
        except ValueError:
            # Skip invalid versions
            continue
    
    print(f"  {package}: {requirement}")
    print(f"    Available: {available}")
    print(f"    Compatible: {compatible}")
    print(f"    Recommended: {compatible[-1] if compatible else 'None'}")
```

## Example 4: Version parsing

```python
# Version manipulation and increment operations
import semantic_version
from typing import Optional

# Create a base version for manipulation
base_version = semantic_version.Version('1.2.3')
print(f"Base version: {base_version}")

# Demonstrate version increment operations
print(f"\nVersion increment operations:")
print(f"  Original:     {base_version}")
print(f"  Next patch:   {base_version.next_patch()}")
print(f"  Next minor:   {base_version.next_minor()}")
print(f"  Next major:   {base_version.next_major()}")

# Working with prerelease versions
prerelease_version = semantic_version.Version('2.0.0-alpha.1')
print(f"\nPrerelease version operations:")
print(f"  Original:     {prerelease_version}")
print(f"  Next patch:   {prerelease_version.next_patch()}")
print(f"  Next minor:   {prerelease_version.next_minor()}")
print(f"  Next major:   {prerelease_version.next_major()}")

# Create versions with different prerelease identifiers
print(f"\nCreating prerelease versions:")
release_version = semantic_version.Version('1.5.0')
alpha = semantic_version.Version('1.5.0-alpha')
beta = semantic_version.Version('1.5.0-beta')
rc = semantic_version.Version('1.5.0-rc.1')

print(f"  Release:      {release_version}")
print(f"  Alpha:        {alpha}")
print(f"  Beta:         {beta}")
print(f"  Release Candidate: {rc}")

# Version range operations
print(f"\nVersion range operations:")
versions_list = [
    '1.0.0', '1.0.1', '1.0.2',
    '1.1.0', '1.1.1',
    '1.2.0', '1.2.1', '1.2.3',
    '2.0.0', '2.0.1'
]

versions = [semantic_version.Version(v) for v in versions_list]
print(f"Available versions: {[str(v) for v in versions]}")

# Find versions in different ranges
target = semantic_version.Version('1.2.0')
print(f"\nVersions relative to {target}:")

# Same major version
same_major = [v for v in versions if v.major == target.major]
print(f"  Same major ({target.major}.x.x): {[str(v) for v in same_major]}")

# Same minor version
same_minor = [v for v in versions if v.major == target.major and v.minor == target.minor]
print(f"  Same minor ({target.major}.{target.minor}.x): {[str(v) for v in same_minor]}")

# Patch updates only
patch_updates = [v for v in versions if v.major == target.major and v.minor == target.minor and v > target]
print(f"  Patch updates: {[str(v) for v in patch_updates]}")

# Minor updates
minor_updates = [v for v in versions if v.major == target.major and v.minor > target.minor]
print(f"  Minor updates: {[str(v) for v in minor_updates]}")

# Major updates
major_updates = [v for v in versions if v.major > target.major]
print(f"  Major updates: {[str(v) for v in major_updates]}")

# Practical version selection function
def select_compatible_version(available_versions: List[str], requirement: str) -> Optional[str]:
    """Select the highest compatible version from available versions."""
    try:
        spec = semantic_version.Spec(requirement)
        compatible_versions = []
        
        for version_str in available_versions:
            try:
                version = semantic_version.Version(version_str)
                if version in spec:
                    compatible_versions.append(version)
            except ValueError:
                continue
        
        if compatible_versions:
            return str(max(compatible_versions))
        return None
    except ValueError:
        return None

# Test the selection function
print(f"\nVersion selection examples:")
available = ['1.0.0', '1.2.0', '1.2.5', '1.3.0', '2.0.0', '2.1.0']
requirements = ['>=1.0.0,<2.0.0', '~=1.2.0', '>=1.3.0', '==1.2.5']

for req in requirements:
    selected = select_compatible_version(available, req)
    print(f"  Requirement: {req:15} -> Selected: {selected or 'None'}")
```

## Example 5: Module info

```python
# Real-world use cases and best practices
import semantic_version
from typing import Dict, List, Optional, Tuple
import json

print(f"semantic_version module: {semantic_version.__name__}")

# Use case 1: API versioning system
class APIVersionManager:
    """Manage API versions using semantic versioning."""
    
    def __init__(self):
        self.supported_versions = [
            '1.0.0', '1.1.0', '1.2.0',
            '2.0.0', '2.1.0', '2.2.0',
            '3.0.0-beta.1', '3.0.0-rc.1'
        ]
        self.versions = [semantic_version.Version(v) for v in self.supported_versions]
    
    def is_supported(self, version_str: str) -> bool:
        """Check if a version is supported."""
        try:
            version = semantic_version.Version(version_str)
            return version in self.versions
        except ValueError:
            return False
    
    def get_compatible_version(self, requested: str) -> Optional[str]:
        """Find the highest compatible version."""
        try:
            requested_version = semantic_version.Version(requested)
            compatible = [
                v for v in self.versions 
                if (v.major == requested_version.major and 
                    v >= requested_version and
                    not v.prerelease)  # Exclude prereleases
            ]
            return str(max(compatible)) if compatible else None
        except ValueError:
            return None
    
    def get_breaking_changes(self, from_version: str, to_version: str) -> List[str]:
        """Identify potential breaking changes between versions."""
        try:
            from_v = semantic_version.Version(from_version)
            to_v = semantic_version.Version(to_version)
            
            changes = []
            if to_v.major > from_v.major:
                changes.append(f"Major version change: {from_v.major} -> {to_v.major} (breaking changes expected)")
            if to_v.minor > from_v.minor:
                changes.append(f"Minor version change: {from_v.minor} -> {to_v.minor} (new features)")
            if to_v.patch > from_v.patch:
                changes.append(f"Patch version change: {from_v.patch} -> {to_v.patch} (bug fixes)")
            
            return changes
        except ValueError:
            return ["Invalid version format"]

# Test the API version manager
api_manager = APIVersionManager()
print(f"\nAPI Version Manager Example:")
print(f"Supported versions: {api_manager.supported_versions}")

test_requests = ['1.0.0', '1.1.5', '2.0.0', '2.5.0', '3.0.0']
for request in test_requests:
    is_supported = api_manager.is_supported(request)
    compatible = api_manager.get_compatible_version(request)
    print(f"  Request: {request} -> Supported: {is_supported}, Compatible: {compatible}")

# Use case 2: Dependency conflict resolution
print(f"\nDependency Conflict Resolution:")
dependencies = {
    'package-a': {
        'requires': {
            'shared-lib': '>=1.0.0,<2.0.0',
            'util-lib': '>=2.1.0'
        }
    },
    'package-b': {
        'requires': {
            'shared-lib': '>=1.5.0,<3.0.0',
            'util-lib': '>=2.0.0,<2.2.0'
        }
    }
}

available_versions = {
    'shared-lib': ['1.0.0', '1.2.0', '1.5.0', '1.8.0', '2.0.0', '2.1.0'],
    'util-lib': ['2.0.0', '2.0.5', '2.1.0', '2.1.2', '2.2.0', '2.3.0']
}

def find_compatible_versions(deps: Dict, available: Dict) -> Dict[str, Optional[str]]:
    """Find versions that satisfy all dependency requirements."""
    result = {}
    
    for lib_name in available.keys():
        # Collect all requirements for this library
        all_specs = []
        for package, package_deps in deps.items():
            if lib_name in package_deps['requires']:
                all_specs.append(package_deps['requires'][lib_name])
        
        if not all_specs:
            continue
            
        # Combine all specifications
        combined_spec_str = ','.join(all_specs)
        try:
            combined_spec = semantic_version.Spec(combined_spec_str)
            
            # Find compatible versions
            compatible = []
            for version_str in available[lib_name]:
                try:
                    version = semantic_version.Version(version_str)
                    if version in combined_spec:
                        compatible.append(version)
                except ValueError:
                    continue
            
            result[lib_name] = str(max(compatible)) if compatible else None
        except ValueError:
            result[lib_name] = None
    
    return result

resolution = find_compatible_versions(dependencies, available_versions)
print("Dependency resolution:")
for lib, version in resolution.items():
    print(f"  {lib}: {version or 'No compatible version found'}")

# Use case 3: Release planning
print(f"\nRelease Planning Example:")
current_version = semantic_version.Version('2.1.5')
planned_changes = [
    {'type': 'feature', 'description': 'Add new API endpoint'},
    {'type': 'fix', 'description': 'Fix memory leak'},
    {'type': 'breaking', 'description': 'Remove deprecated methods'},
    {'type': 'feature', 'description': 'Improve performance'},
    {'type': 'fix', 'description': 'Fix validation bug'}
]

def plan_next_release(current: semantic_version.Version, changes: List[Dict]) -> Tuple[semantic_version.Version, str]:
    """Plan the next release version based on changes."""
    has_breaking = any(change['type'] == 'breaking' for change in changes)
    has_features = any(change['type'] == 'feature' for change in changes)
    has_fixes = any(change['type'] == 'fix' for change in changes)
    
    if has_breaking:
        next_version = current.next_major()
        release_type = "Major"
    elif has_features:
        next_version = current.next_minor()
        release_type = "Minor"
    elif has_fixes:
        next_version = current.next_patch()
        release_type = "Patch"
    else:
        next_version = current
        release_type = "No changes"
    
    return next_version, release_type

next_version, release_type = plan_next_release(current_version, planned_changes)
print(f"Current version: {current_version}")
print(f"Planned changes: {len(planned_changes)} changes")
for change in planned_changes:
    print(f"  - {change['type']}: {change['description']}")
print(f"Next version: {next_version} ({release_type} release)")

# Summary of semantic_version capabilities
print(f"\nSemantic Version Package Summary:")
print("✓ Complete semantic versioning support (SemVer 2.0.0)")
print("✓ Version parsing, validation, and comparison")
print("✓ Flexible version specifications and matching")
print("✓ Support for prerelease and build metadata")
print("✓ Version increment operations")
print("✓ Dependency resolution capabilities")
print("✓ API versioning and compatibility checking")
print("✓ Release planning and change impact analysis")
```
