```python
# Example 1: Construct a Version from parts and inspect its components.
from semver import Version

v = Version(major=1, minor=2, patch=3, prerelease="alpha.1", build="build.42")
print(f"version  = {v}")
print(f"major    = {v.major}")
print(f"minor    = {v.minor}")
print(f"patch    = {v.patch}")
print(f"prerelease = {v.prerelease}")
print(f"build    = {v.build}")

assert v.major == 1
assert v.minor == 2
assert v.patch == 3
assert v.prerelease == "alpha.1"
assert v.build == "build.42"
assert str(v) == "1.2.3-alpha.1+build.42"
```

```python
# Example 2: Parse version strings via Version.parse and validate detection.
from semver import Version

parsed = Version.parse("2.10.5-rc.1+sha.abc123")
print(f"parsed = {parsed!r}")
print(f"  major={parsed.major} minor={parsed.minor} patch={parsed.patch}")
print(f"  prerelease={parsed.prerelease} build={parsed.build}")

assert parsed.major == 2 and parsed.minor == 10 and parsed.patch == 5

assert Version.is_valid("1.0.0")
assert Version.is_valid("0.0.1-alpha+meta")
assert not Version.is_valid("not.a.version")
assert not Version.is_valid("1.2")
print("is_valid checks pass")
```

```python
# Example 3: Bump operations produce a new Version with the expected component changed.
from semver import Version

base = Version.parse("1.2.3")
print(f"base         = {base}")
print(f"bump_major() = {base.bump_major()}")
print(f"bump_minor() = {base.bump_minor()}")
print(f"bump_patch() = {base.bump_patch()}")
print(f"bump_prerelease() = {base.bump_prerelease()}")

assert base.bump_major() == Version.parse("2.0.0")
assert base.bump_minor() == Version.parse("1.3.0")
assert base.bump_patch() == Version.parse("1.2.4")
# bump_prerelease on a release version creates a prerelease on the next patch
assert base.bump_prerelease().prerelease is not None
```

```python
# Example 4: Compare and sort versions according to semver precedence rules.
from semver import Version, compare

versions = [
    Version.parse("1.0.0"),
    Version.parse("1.0.0-alpha"),
    Version.parse("1.0.0-alpha.1"),
    Version.parse("1.0.0-beta"),
    Version.parse("2.0.0"),
    Version.parse("1.10.0"),
]
ordered = sorted(versions)
for v in ordered:
    print(v)

assert ordered[0] == Version.parse("1.0.0-alpha")
assert ordered[-1] == Version.parse("2.0.0")

# compare() returns -1 / 0 / 1
assert compare("1.0.0", "2.0.0") == -1
assert compare("1.2.3", "1.2.3") == 0
assert compare("2.0.0", "1.9.9") == 1
print("compare() and sort precedence pass")
```

```python
# Example 5: match() and replace() — range matching + immutable component swap.
from semver import Version, match

# match(version, expression) -> bool
assert match("1.2.3", ">=1.0.0")
assert match("1.2.3", "<2.0.0")
assert not match("1.2.3", ">2.0.0")
assert match("1.2.3", "==1.2.3")
assert match("1.2.3", "!=1.0.0")

v = Version.parse("1.2.3-alpha+meta")
replaced = v.replace(minor=99, prerelease="beta")
print(f"original = {v}")
print(f"replaced = {replaced}")

assert replaced.minor == 99
assert replaced.prerelease == "beta"
# original is unchanged (Version is immutable)
assert v.minor == 2 and v.prerelease == "alpha"
print("match() and replace() pass")
```
