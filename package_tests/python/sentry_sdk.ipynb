```python
# Example 1: init the SDK with an in-memory transport that captures events locally.
import sentry_sdk
from sentry_sdk.transport import Transport

captured_events = []

class InMemoryTransport(Transport):
    def __init__(self, options=None):
        super().__init__(options)
    def capture_envelope(self, envelope):
        captured_events.append(envelope)
    def capture_event(self, event):
        captured_events.append(event)
    def flush(self, timeout, callback=None):
        pass
    def kill(self):
        pass

client = sentry_sdk.init(
    dsn='https://public@example.invalid/1',
    transport=InMemoryTransport,
    default_integrations=False,
    auto_enabling_integrations=False,
)
print(f"init returned: {type(client).__name__}")
print(f"client active: {sentry_sdk.Hub.current.client is not None}")
assert sentry_sdk.Hub.current.client is not None
```

```python
# Example 2: Hub / current client / integrations introspection.
import sentry_sdk
from sentry_sdk import Hub

hub = Hub.current
client = hub.client
print(f"hub: {type(hub).__name__}")
print(f"client: {type(client).__name__ if client else None}")
print(f"dsn (parsed): {client.dsn if client else None}")

integrations = client.integrations if client else {}
print(f"active integrations ({len(integrations)}): {sorted(integrations.keys())[:10]}")

# set_tag / set_user at module level routes through the current hub/scope.
sentry_sdk.set_tag('module-level-tag', 'ok')
sentry_sdk.set_user({'id': 'module-user'})
assert client is not None
```
