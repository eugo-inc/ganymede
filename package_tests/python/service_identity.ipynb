## Example 1: service_identity functionality test 1

```python
# Example 1: Basic service_identity Module
import service_identity
from cryptography import x509
from cryptography.x509.oid import NameOID
from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.primitives.asymmetric import rsa
from datetime import datetime, timedelta

print("=== service_identity Module ===")
print(f"Version: {getattr(service_identity, '__version__', 'Unknown')}")

# Create simple test certificate
key = rsa.generate_private_key(public_exponent=65537, key_size=2048)
subject = issuer = x509.Name([x509.NameAttribute(NameOID.COMMON_NAME, "example.com")])
cert = x509.CertificateBuilder().subject_name(subject).issuer_name(issuer).public_key(
    key.public_key()
).serial_number(x509.random_serial_number()).not_valid_before(
    datetime.utcnow()
).not_valid_after(
    datetime.utcnow() + timedelta(days=365)
).sign(key, hashes.SHA256())

print(f"Certificate subject: {cert.subject.rfc4514_string()}")
print("Example 1 completed!")
```

## Example 2: service_identity functionality test 2

```python
# Example 2: Certificate with Subject Alternative Names
from cryptography import x509
from cryptography.x509.oid import NameOID
from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.primitives.asymmetric import rsa
from datetime import datetime, timedelta
import ipaddress

print("=== Certificate with SANs ===")

key = rsa.generate_private_key(public_exponent=65537, key_size=2048)
subject = x509.Name([x509.NameAttribute(NameOID.COMMON_NAME, "example.com")])
san_list = [x509.DNSName("example.com"), x509.DNSName("www.example.com"), 
            x509.IPAddress(ipaddress.ip_address("127.0.0.1"))]

cert = x509.CertificateBuilder().subject_name(subject).issuer_name(subject).public_key(
    key.public_key()
).serial_number(x509.random_serial_number()).not_valid_before(
    datetime.utcnow()
).not_valid_after(
    datetime.utcnow() + timedelta(days=365)
).add_extension(x509.SubjectAlternativeName(san_list), critical=False).sign(key, hashes.SHA256())

san_ext = cert.extensions.get_extension_for_oid(x509.oid.ExtensionOID.SUBJECT_ALTERNATIVE_NAME)
print(f"SANs: {[str(name.value) for name in san_ext.value]}")
print("Example 2 completed!")
```

## Example 3: service_identity functionality test 3

```python
# Example 3: Certificate Time Validity Check
from cryptography import x509
from cryptography.x509.oid import NameOID
from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.primitives.asymmetric import rsa
from datetime import datetime, timedelta

print("=== Certificate Time Validity ===")

key = rsa.generate_private_key(public_exponent=65537, key_size=2048)
subject = x509.Name([x509.NameAttribute(NameOID.COMMON_NAME, "test.com")])

# Valid certificate
valid_cert = x509.CertificateBuilder().subject_name(subject).issuer_name(subject).public_key(
    key.public_key()
).serial_number(x509.random_serial_number()).not_valid_before(
    datetime.utcnow()
).not_valid_after(
    datetime.utcnow() + timedelta(days=365)
).sign(key, hashes.SHA256())

now = datetime.utcnow()
is_valid = valid_cert.not_valid_before <= now <= valid_cert.not_valid_after
print(f"Certificate time valid: {is_valid}")
print("Example 3 completed!")
```

## Example 4: service_identity functionality test 4

```python
# Example 4: Certificate Extensions
from cryptography import x509
from cryptography.x509.oid import NameOID
from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.primitives.asymmetric import rsa
from datetime import datetime, timedelta

print("=== Certificate Extensions ===")

key = rsa.generate_private_key(public_exponent=65537, key_size=2048)
subject = x509.Name([x509.NameAttribute(NameOID.COMMON_NAME, "server.com")])

cert = x509.CertificateBuilder().subject_name(subject).issuer_name(subject).public_key(
    key.public_key()
).serial_number(x509.random_serial_number()).not_valid_before(
    datetime.utcnow()
).not_valid_after(
    datetime.utcnow() + timedelta(days=365)
).add_extension(x509.KeyUsage(digital_signature=True, key_encipherment=True, 
    content_commitment=False, data_encipherment=False, key_agreement=False,
    key_cert_sign=False, crl_sign=False, encipher_only=False, decipher_only=False), 
    critical=True).sign(key, hashes.SHA256())

ku = cert.extensions.get_extension_for_oid(x509.oid.ExtensionOID.KEY_USAGE).value
print(f"Digital signature: {ku.digital_signature}, Key encipherment: {ku.key_encipherment}")
print("Example 4 completed!")
```

## Example 5: service_identity functionality test 5

```python
# Example 5: Certificate Fingerprint
from cryptography import x509
from cryptography.x509.oid import NameOID
from cryptography.hazmat.primitives import hashes, serialization
from cryptography.hazmat.primitives.asymmetric import rsa
from datetime import datetime, timedelta

print("=== Certificate Fingerprint ===")

key = rsa.generate_private_key(public_exponent=65537, key_size=2048)
subject = x509.Name([x509.NameAttribute(NameOID.COMMON_NAME, "fingerprint.com")])

cert = x509.CertificateBuilder().subject_name(subject).issuer_name(subject).public_key(
    key.public_key()
).serial_number(x509.random_serial_number()).not_valid_before(
    datetime.utcnow()
).not_valid_after(
    datetime.utcnow() + timedelta(days=365)
).sign(key, hashes.SHA256())

# Calculate fingerprint
cert_der = cert.public_bytes(serialization.Encoding.DER)
fingerprint = hashes.Hash(hashes.SHA256())
fingerprint.update(cert_der)
print(f"SHA256 fingerprint: {fingerprint.finalize().hex()[:32]}...")
print("Example 5 completed!")
```
