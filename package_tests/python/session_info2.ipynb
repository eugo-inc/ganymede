## Example 1: the packages a session imported

```python
# Example 1: session_info() lists the distributions behind the imported modules, with versions
import importlib.metadata
import numpy
import session_info2

info = session_info2.session_info()
assert "numpy" in info.imported_dists, list(info.imported_dists)
text = info.format("markdown")
assert importlib.metadata.version("numpy") in text, text[:400]
print("Example 1:", sorted(info.imported_dists)[:6])
```

## Example 2: Markdown and plain-text formats

```python
# Example 2: format("markdown") gives tables; the object's text repr carries the same rows
import session_info2

info = session_info2.session_info()
md = info.format("markdown")
assert md.lstrip().startswith("| Package") and "| ---" in md, md[:200]
assert "numpy" in md or "session-info2" in md or "| Package" in md
print("Example 2:", md.splitlines()[0])
```

## Example 3: the OS and CPU section

```python
# Example 3: os=True, cpu=True add the platform and the CPU count to the report
import os
import platform
import session_info2

md = session_info2.session_info(os=True, cpu=True).format("markdown")
assert platform.system() in md or platform.machine() in md, md[-500:]
assert str(os.cpu_count()) in md, md[-500:]
print("Example 3:", [line for line in md.splitlines() if "CPU" in line or "OS" in line][:3])
```

## Example 4: a package imported later appears

```python
# Example 4: the report reads sys.modules when it is built
import session_info2

before = set(session_info2.session_info().imported_dists)
import scipy  # noqa: E402
after = set(session_info2.session_info().imported_dists)
assert "scipy" in after, sorted(after)
print("Example 4: scipy in the report after import:", "scipy" in after, "(before:", "scipy" in before, ")")
```

## Example 5: dependencies on request

```python
# Example 5: dependencies=True adds the distributions the imported ones depend on
import pandas
import session_info2

info = session_info2.session_info(dependencies=True)
deps = set(info.deps_dists)
assert deps, "no dependency distributions were listed"
assert "numpy" in deps or "numpy" in info.imported_dists, sorted(deps)[:20]
print("Example 5:", len(deps), "dependency distributions, e.g.", sorted(deps)[:5])
```
