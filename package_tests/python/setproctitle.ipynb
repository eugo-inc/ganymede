```python
# Example 1: Basic Process Title Management
import setproctitle
import os
import time

# Get current process title
original_title = setproctitle.getproctitle()
print(f"Original process title: {original_title}")
print(f"Process ID: {os.getpid()}")

# Set a new process title
new_title = "MyPythonApplication"
setproctitle.setproctitle(new_title)
current_title = setproctitle.getproctitle()
print(f"New process title: {current_title}")

# Restore original title
setproctitle.setproctitle(original_title)
restored_title = setproctitle.getproctitle()
print(f"Restored title: {restored_title}")
print("Basic process title management complete")
```

```python
# Example 2: Dynamic Process Status Updates
import setproctitle
import time
from datetime import datetime

# Simulate a worker process with status updates
def simulate_worker_process():
    base_title = "DataProcessor"
    
    # Initialization phase
    setproctitle.setproctitle(f"{base_title}: Initializing")
    print(f"Status: {setproctitle.getproctitle()}")
    time.sleep(0.1)
    
    # Processing phase
    for i in range(3):
        status = f"{base_title}: Processing batch {i+1}/3"
        setproctitle.setproctitle(status)
        print(f"Status: {setproctitle.getproctitle()}")
        time.sleep(0.1)
    
    # Completion phase
    completion_time = datetime.now().strftime("%H:%M:%S")
    setproctitle.setproctitle(f"{base_title}: Completed at {completion_time}")
    print(f"Final status: {setproctitle.getproctitle()}")

simulate_worker_process()
print("Dynamic status updates demonstration complete")
```

```python
# Example 3: Server Application Process Identification
import setproctitle
import threading
import time

class MockServer:
    def __init__(self, server_name, port):
        self.server_name = server_name
        self.port = port
        self.connections = 0
        self.running = False
    
    def start(self):
        self.running = True
        # Set descriptive process title
        setproctitle.setproctitle(f"{self.server_name}[{self.port}]: Starting")
        print(f"Server starting: {setproctitle.getproctitle()}")
        
        # Simulate server ready state
        time.sleep(0.1)
        setproctitle.setproctitle(f"{self.server_name}[{self.port}]: Ready - 0 connections")
        print(f"Server ready: {setproctitle.getproctitle()}")
        
        # Simulate connections
        for i in range(3):
            self.connections += 1
            title = f"{self.server_name}[{self.port}]: Active - {self.connections} connections"
            setproctitle.setproctitle(title)
            print(f"Connection {i+1}: {setproctitle.getproctitle()}")
            time.sleep(0.1)
    
    def stop(self):
        self.running = False
        setproctitle.setproctitle(f"{self.server_name}[{self.port}]: Shutting down")
        print(f"Server stopping: {setproctitle.getproctitle()}")

# Demonstrate server process identification
server = MockServer("WebServer", 8080)
server.start()
server.stop()
print("Server application process identification complete")
```

```python
# Example 4: Process Title with Resource Monitoring
import setproctitle
import psutil
import time
import os

def monitor_with_title():
    """Demonstrate process title with resource monitoring information"""
    process = psutil.Process(os.getpid())
    base_title = "ResourceMonitor"
    
    for i in range(3):
        # Get current resource usage
        try:
            memory_mb = process.memory_info().rss / 1024 / 1024
            cpu_percent = process.cpu_percent()
            
            # Update process title with resource info
            title = f"{base_title}: Mem:{memory_mb:.1f}MB CPU:{cpu_percent:.1f}%"
            setproctitle.setproctitle(title)
            print(f"Monitor {i+1}: {setproctitle.getproctitle()}")
            
        except (psutil.NoSuchProcess, psutil.AccessDenied):
            # Fallback if psutil access is limited
            title = f"{base_title}: Monitoring cycle {i+1}"
            setproctitle.setproctitle(title)
            print(f"Monitor {i+1}: {setproctitle.getproctitle()}")
        
        time.sleep(0.1)
    
    # Reset to simple title
    setproctitle.setproctitle(f"{base_title}: Idle")
    print(f"Final: {setproctitle.getproctitle()}")

monitor_with_title()
print("Resource monitoring with process title complete")
```

```python
# Example 5: Production Process Management and Best Practices
import setproctitle
import time
import logging
from contextlib import contextmanager

# Setup logging for production patterns
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(message)s')
logger = logging.getLogger(__name__)

class ProcessTitleManager:
    """Production-ready process title management"""
    
    def __init__(self, application_name):
        self.application_name = application_name
        self.original_title = setproctitle.getproctitle()
        self.current_operation = None
    
    @contextmanager
    def operation(self, operation_name):
        """Context manager for operation-specific titles"""
        previous_operation = self.current_operation
        self.current_operation = operation_name
        
        try:
            title = f"{self.application_name}: {operation_name}"
            setproctitle.setproctitle(title)
            logger.info(f"Started operation: {operation_name}")
            yield
        except Exception as e:
            error_title = f"{self.application_name}: ERROR in {operation_name}"
            setproctitle.setproctitle(error_title)
            logger.error(f"Error in {operation_name}: {e}")
            raise
        finally:
            self.current_operation = previous_operation
            if previous_operation:
                setproctitle.setproctitle(f"{self.application_name}: {previous_operation}")
            else:
                setproctitle.setproctitle(f"{self.application_name}: Idle")
            logger.info(f"Completed operation: {operation_name}")
    
    def restore_title(self):
        """Restore original process title"""
        setproctitle.setproctitle(self.original_title)
        logger.info("Process title restored to original")

# Demonstrate production usage
manager = ProcessTitleManager("DataAnalyzer")

try:
    with manager.operation("Loading data"):
        print(f"Title: {setproctitle.getproctitle()}")
        time.sleep(0.1)
    
    with manager.operation("Processing analysis"):
        print(f"Title: {setproctitle.getproctitle()}")
        time.sleep(0.1)
    
    with manager.operation("Generating report"):
        print(f"Title: {setproctitle.getproctitle()}")
        time.sleep(0.1)
    
    print(f"Final title: {setproctitle.getproctitle()}")
    
finally:
    manager.restore_title()
    print(f"Restored title: {setproctitle.getproctitle()}")

print("Production process management demonstration complete")
```
