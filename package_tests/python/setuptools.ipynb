# setuptools: 5 Basic Usage Examples

```python
# Example 1: Check setuptools version
import setuptools
print(setuptools.__version__)
```

```python
# Example 2: Distribution metadata via importlib.metadata
# (pkg_resources was removed from modern setuptools; importlib.metadata is its
# documented replacement for get_distribution()).
import setuptools
from importlib.metadata import distribution

dist = distribution('setuptools')
print(dist.metadata['Name'], dist.version)
assert dist.metadata['Name'] == 'setuptools'
assert dist.version == setuptools.__version__
```

```python
# Example 3: Package discovery — find_packages / find_namespace_packages
# (replaces the removed pkg_resources.working_set listing with a real
# setuptools API exercise).
import os
import tempfile
from setuptools import find_packages, find_namespace_packages

with tempfile.TemporaryDirectory() as root:
    os.makedirs(os.path.join(root, 'mypkg', 'sub'))
    open(os.path.join(root, 'mypkg', '__init__.py'), 'w').close()
    open(os.path.join(root, 'mypkg', 'sub', '__init__.py'), 'w').close()
    os.makedirs(os.path.join(root, 'ns', 'plain'))
    open(os.path.join(root, 'ns', 'plain', '__init__.py'), 'w').close()
    pkgs = find_packages(where=root)
    ns_pkgs = find_namespace_packages(where=root)

print(f"find_packages: {sorted(pkgs)}")
print(f"find_namespace_packages: {sorted(ns_pkgs)}")
assert sorted(pkgs) == ['mypkg', 'mypkg.sub']
assert 'ns.plain' in ns_pkgs
```

```python
# Example 4: Requirements + metadata via setuptools' Distribution
# (replaces the removed pkg_resources.Requirement.parse).
from setuptools.dist import Distribution

dist = Distribution({
    'name': 'demo-project',
    'version': '1.2.3',
    'install_requires': ['requests>=2.0.0', 'packaging'],
})
print(dist.metadata.name, dist.metadata.version)
print('install_requires:', dist.install_requires)
assert dist.metadata.name == 'demo-project'
assert dist.metadata.version == '1.2.3'
assert 'requests>=2.0.0' in dist.install_requires
```

```python
# Example 5: Locate package resource files via importlib.resources
# (modern replacement for pkg_resources.resource_filename).
from importlib.resources import files

init_path = files('setuptools') / '__init__.py'
print(init_path)
assert init_path.is_file()
print(f"head: {init_path.read_text()[:50]!r}")
```
