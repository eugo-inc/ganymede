## Example 2: Setup.py Configuration with Rust Extensions

```python
# Example 2: RustBin and Strip Configuration
from setuptools_rust import RustBin, Strip, RustExtension, Binding

print("Example 2: RustBin and Strip Configuration")
print("=" * 60)

# Demonstrate RustBin for binary targets
print("RustBin - Building Rust Binary Tools:")
print("-" * 40)

# Create different RustBin configurations
rust_bins = [
    RustBin("cli-tool", path="Cargo.toml"),
    RustBin("data-processor", path="tools/Cargo.toml", features=["fast"]),
    RustBin({"name": "custom_name", "output_name": "mytool"}, features=["full"])
]

for i, rust_bin in enumerate(rust_bins, 1):
    print(f"\nRustBin {i}:")
    print(f"  Target: {rust_bin.target}")
    print(f"  Path: {rust_bin.path}")
    print(f"  Features: {rust_bin.features}")
    print(f"  Debug: {rust_bin.debug}")
    print(f"  Strip: {rust_bin.strip}")

# Demonstrate Strip enum for binary optimization
print("\n" + "=" * 60)
print("Strip Options for Binary Size Optimization:")
print("-" * 40)

strip_configs = [
    (Strip.No, "No stripping - keep all symbols"),
    (Strip.Debug, "Strip debug symbols only"),
    (Strip.All, "Strip all symbols - maximum size reduction")
]

for strip_option, description in strip_configs:
    print(f"\n{strip_option.name} ({strip_option.value}):")
    print(f"  Description: {description}")
    
    # Create RustBin with each strip option
    bin_with_strip = RustBin(
        "optimized-tool",
        strip=strip_option,
        debug=False if strip_option != Strip.No else None
    )
    print(f"  ✅ Created binary with strip={bin_with_strip.strip.name}")

# Advanced RustBin configurations
print("\n" + "=" * 60)
print("Advanced RustBin Configurations:")
print("-" * 40)

advanced_configs = {
    "Production Binary": RustBin(
        "prod-tool",
        path="Cargo.toml",
        features=["production", "optimized"],
        strip=Strip.All,
        debug=False,
        quiet=True
    ),
    "Debug Binary": RustBin(
        "debug-tool", 
        path="Cargo.toml",
        features=["dev"],
        strip=Strip.No,
        debug=True,
        quiet=False
    ),
    "Custom Environment": RustBin(
        "env-tool",
        path="Cargo.toml",
        env={"RUSTFLAGS": "-C target-cpu=native"},
        args=["--verbose"]
    )
}

for config_name, rust_bin in advanced_configs.items():
    print(f"\n{config_name}:")
    print(f"  Target: {rust_bin.target}")
    print(f"  Strip: {rust_bin.strip.name}")
    print(f"  Debug: {rust_bin.debug}")
    print(f"  Features: {rust_bin.features}")
    print(f"  Quiet: {rust_bin.quiet}")
    if rust_bin.env:
        print(f"  Environment: {rust_bin.env}")
    if rust_bin.args:
        print(f"  Args: {rust_bin.args}")

# Compare RustExtension vs RustBin
print("\n" + "=" * 60)
print("RustExtension vs RustBin Comparison:")
print("-" * 40)

rust_ext = RustExtension("mypackage.module", binding=Binding.PyO3)
rust_bin = RustBin("mytool", strip=Strip.All)

print(f"\nRustExtension (for Python modules):")
print(f"  Name: {rust_ext.name}")
print(f"  Binding: {rust_ext.binding.name}")
print(f"  Purpose: Python-importable extension module")

print(f"\nRustBin (for standalone binaries):")
print(f"  Target: {rust_bin.target}")
print(f"  Strip: {rust_bin.strip.name}")
print(f"  Purpose: Standalone executable tool")

print("\n✅ RustBin and Strip configuration demonstrated")
```

## Example 4: Build Process and Development Workflow

```python
# Example 4: build_rust Command and Build Options
from setuptools_rust import build_rust, RustExtension, Binding
from setuptools.dist import Distribution
import inspect

print("Example 4: build_rust Command and Build Options")
print("=" * 60)

# Inspect the build_rust class
print("build_rust Command Class:")
print("-" * 40)

print(f"Class: {build_rust}")
print(f"Module: {build_rust.__module__}")
print(f"Base classes: {[base.__name__ for base in build_rust.__bases__]}")

# Get build_rust user options
print("\nbuild_rust User Options:")
print("-" * 40)

if hasattr(build_rust, 'user_options'):
    for option in build_rust.user_options:
        if len(option) >= 3:
            long_name, short_name, description = option[0], option[1], option[2]
            print(f"  --{long_name.rstrip('=')}")
            if short_name:
                print(f"    Short: -{short_name}")
            print(f"    Description: {description}")

# Get boolean options
print("\nbuild_rust Boolean Options:")
print("-" * 40)

if hasattr(build_rust, 'boolean_options'):
    for bool_opt in build_rust.boolean_options:
        print(f"  --{bool_opt}")

# Create a Distribution with RustExtension
print("\n" + "=" * 60)
print("Creating Distribution with build_rust:")
print("-" * 40)

# Create sample RustExtensions
rust_extensions = [
    RustExtension(
        "mypackage.core",
        path="Cargo.toml",
        binding=Binding.PyO3,
        features=["default"]
    ),
    RustExtension(
        "mypackage.utils",
        path="utils/Cargo.toml",
        binding=Binding.PyO3,
        debug=False
    )
]

# Create a Distribution object
dist = Distribution({
    'name': 'test-rust-package',
    'version': '0.1.0',
    'rust_extensions': rust_extensions
})

print(f"Created Distribution: {dist.get_name()} v{dist.get_version()}")
print(f"Rust extensions count: {len(rust_extensions)}")

for i, ext in enumerate(rust_extensions, 1):
    print(f"\n  Extension {i}:")
    print(f"    Name: {ext.name}")
    print(f"    Path: {ext.path}")
    print(f"    Binding: {ext.binding.name}")
    print(f"    Features: {ext.features}")
    print(f"    Debug: {ext.debug}")

# Create build_rust command instance
print("\n" + "=" * 60)
print("build_rust Command Instance:")
print("-" * 40)

try:
    build_cmd = build_rust(dist)

    print(f"✅ build_rust command created")
    print(f"   Distribution: {build_cmd.distribution.get_name()}")

    # Inspect command attributes
    cmd_attrs = {
        'inplace': 'Build extensions in-place',
        'release': 'Build in release mode',
        'debug': 'Build in debug mode',
        'qbuild': 'Quiet cargo build',
        'build_temp': 'Temporary build directory',
        'build_lib': 'Build library directory',
    }

    print(f"\n   Command Attributes:")
    for attr, description in cmd_attrs.items():
        if hasattr(build_cmd, attr):
            value = getattr(build_cmd, attr)
            print(f"     {attr}: {value} ({description})")

except Exception as e:
    print(f"⚠️ Could not create build_rust command: {e}")
    raise

# Demonstrate build configurations
print("\n" + "=" * 60)
print("Build Configuration Options:")
print("-" * 40)

build_configs = {
    "Debug Build": {
        "description": "Development build with debug symbols",
        "options": {
            "debug": True,
            "release": False,
            "inplace": True,
            "qbuild": False
        }
    },
    "Release Build": {
        "description": "Optimized production build",
        "options": {
            "debug": False,
            "release": True,
            "inplace": False,
            "qbuild": False
        }
    },
    "Quiet Release": {
        "description": "Quiet optimized build for CI/CD",
        "options": {
            "debug": False,
            "release": True,
            "inplace": False,
            "qbuild": True
        }
    }
}

for config_name, config_data in build_configs.items():
    print(f"\n{config_name}:")
    print(f"  {config_data['description']}")
    print(f"  Options:")
    for opt, value in config_data['options'].items():
        print(f"    --{opt}: {value}")

# Show typical build workflow
print("\n" + "=" * 60)
print("Typical Build Workflow:")
print("-" * 40)

workflow = [
    ("Initialize", "Create Distribution with RustExtensions"),
    ("Configure", "Set build options (debug/release, inplace, etc.)"),
    ("Build", "Execute build_rust command to compile Rust code"),
    ("Install", "Copy built extensions to appropriate locations"),
    ("Test", "Run tests against built extensions")
]

for i, (step, description) in enumerate(workflow, 1):
    print(f"{i}. {step}: {description}")

# Inspect build_rust methods
print("\n" + "=" * 60)
print("build_rust Command Methods:")
print("-" * 40)

methods_to_show = ['initialize_options', 'finalize_options', 'run', 'get_dylib_ext_path']
for method_name in methods_to_show:
    if hasattr(build_rust, method_name):
        method = getattr(build_rust, method_name)
        print(f"  {method_name}()")
        if method.__doc__:
            doc_first_line = method.__doc__.strip().split('\n')[0]
            print(f"    {doc_first_line}")
```

## Example 5: Production Package Creation and CI/CD Integration

```python
# Example 5: Complete Production Package Configuration
from setuptools_rust import RustExtension, RustBin, Binding, Strip
from setuptools.dist import Distribution
import json

print("Example 5: Complete Production Package Configuration")
print("=" * 60)

# Create production-ready RustExtensions
print("Production RustExtension Configurations:")
print("-" * 40)

production_extensions = [
    RustExtension(
        "mypackage.core",
        path="rust/core/Cargo.toml",
        binding=Binding.PyO3,
        features=["production", "simd"],
        debug=False
    ),
    RustExtension(
        "mypackage.algorithms",
        path="rust/algorithms/Cargo.toml",
        binding=Binding.PyO3,
        features=["parallel", "optimized"],
        debug=False,
        args=["--jobs", "4"]
    ),
    RustExtension(
        "mypackage.io",
        path="rust/io/Cargo.toml",
        binding=Binding.PyO3,
        features=["async", "compression"],
        debug=False,
        quiet=True
    )
]

for i, ext in enumerate(production_extensions, 1):
    print(f"\nExtension {i}: {ext.name}")
    print(f"  Path: {ext.path}")
    print(f"  Binding: {ext.binding.name}")
    print(f"  Features: {ext.features}")
    print(f"  Debug: {ext.debug}")
    if hasattr(ext, 'optional'):
        print(f"  Optional: {ext.optional}")
    if hasattr(ext, 'args') and ext.args:
        print(f"  Build args: {ext.args}")
    if hasattr(ext, 'quiet') and ext.quiet:
        print(f"  Quiet mode: {ext.quiet}")

# Create production-ready RustBin configurations
print("\n" + "=" * 60)
print("Production RustBin Configurations:")
print("-" * 40)

production_bins = [
    RustBin(
        "mypackage-cli",
        path="rust/cli/Cargo.toml",
        features=["cli", "colors"],
        strip=Strip.All,
        debug=False
    ),
    RustBin(
        {"name": "data-processor", "output_name": "mypackage-process"},
        path="rust/processors/Cargo.toml",
        features=["parallel", "simd"],
        strip=Strip.All,
        debug=False,
        quiet=True
    ),
    RustBin(
        "benchmark-tool",
        path="rust/benchmarks/Cargo.toml",
        features=["bench"],
        strip=Strip.Debug,
        debug=False,
        args=["--release"]
    )
]

for i, bin_ext in enumerate(production_bins, 1):
    print(f"\nBinary {i}: {bin_ext.target}")
    print(f"  Path: {bin_ext.path}")
    print(f"  Features: {bin_ext.features}")
    print(f"  Strip: {bin_ext.strip.name}")
    print(f"  Debug: {bin_ext.debug}")
    if hasattr(bin_ext, 'quiet') and bin_ext.quiet:
        print(f"  Quiet mode: {bin_ext.quiet}")
    if hasattr(bin_ext, 'args') and bin_ext.args:
        print(f"  Build args: {bin_ext.args}")

# Create complete Distribution
print("\n" + "=" * 60)
print("Complete Production Distribution:")
print("-" * 40)

production_dist = Distribution({
    'name': 'production-rust-package',
    'version': '1.0.0',
    'author': 'Development Team',
    'author_email': 'dev@company.com',
    'description': 'Production Python package with Rust extensions',
    'url': 'https://github.com/company/production-rust-package',
    'rust_extensions': production_extensions,
})

print(f"Package name: {production_dist.get_name()}")
print(f"Version: {production_dist.get_version()}")
print(f"Author: {production_dist.get_author()}")
print(f"Author email: {production_dist.get_author_email()}")
print(f"Description: {production_dist.get_description()}")
print(f"URL: {production_dist.get_url()}")

# Analyze package composition
print("\n" + "=" * 60)
print("Package Composition Analysis:")
print("-" * 40)

total_extensions = len(production_extensions)
total_binaries = len(production_bins)
total_features = set()
bindings_used = set()
strip_options = set()

for ext in production_extensions:
    total_features.update(ext.features)
    bindings_used.add(ext.binding.name)

for bin_ext in production_bins:
    total_features.update(bin_ext.features)
    strip_options.add(bin_ext.strip.name)

print(f"Total Rust extensions: {total_extensions}")
print(f"Total Rust binaries: {total_binaries}")
print(f"Unique features used: {len(total_features)}")
print(f"  Features: {', '.join(sorted(total_features))}")
print(f"Bindings used: {', '.join(bindings_used)}")
print(f"Strip options: {', '.join(strip_options)}")

# Platform-specific configurations
print("\n" + "=" * 60)
print("Platform-Specific Build Configurations:")
print("-" * 40)

platform_configs = {
    "Linux": {
        "extensions": [
            RustExtension(
                "mypackage.linux_specific",
                path="rust/linux/Cargo.toml",
                binding=Binding.PyO3,
                features=["linux-optimizations"]
            )
        ],
        "platform": "linux"
    },
    "macOS": {
        "extensions": [
            RustExtension(
                "mypackage.macos_specific",
                path="rust/macos/Cargo.toml",
                binding=Binding.PyO3,
                features=["macos-frameworks"]
            )
        ],
        "platform": "darwin"
    },
    "Windows": {
        "extensions": [
            RustExtension(
                "mypackage.windows_specific",
                path="rust/windows/Cargo.toml",
                binding=Binding.PyO3,
                features=["windows-api"]
            )
        ],
        "platform": "win32"
    }
}

for platform_name, config in platform_configs.items():
    print(f"\n{platform_name}:")
    print(f"  Platform identifier: {config['platform']}")
    for ext in config['extensions']:
        print(f"  Extension: {ext.name}")
        print(f"    Features: {ext.features}")

# Multi-binding example
print("\n" + "=" * 60)
print("Multi-Binding Support Example:")
print("-" * 40)

multi_binding_extensions = [
    RustExtension(
        "compat.pyo3",
        path="Cargo.toml",
        binding=Binding.PyO3,
        features=["pyo3-features"]
    ),
    RustExtension(
        "compat.rustcpython",
        path="Cargo.toml",
        binding=Binding.RustCPython,
        features=["cpython-features"]
    ),
    RustExtension(
        "standalone.tool",
        path="tools/Cargo.toml",
        binding=Binding.NoBinding
    )
]

for ext in multi_binding_extensions:
    print(f"\n{ext.name}:")
    print(f"  Binding: {ext.binding.name}")
    print(f"  Features: {ext.features}")
    binding_desc = {
        "PyO3": "Modern Rust-Python bindings with async support",
        "RustCPython": "Legacy CPython API bindings",
        "NoBinding": "Standalone Rust code without Python bindings"
    }
    print(f"  Description: {binding_desc.get(ext.binding.name, 'Unknown')}")

# Summary
print("\n" + "=" * 60)
print("Production Package Summary:")
print("-" * 40)

summary_stats = {
    "Total Rust Extensions": len(production_extensions),
    "Total Rust Binaries": len(production_bins),
    "Binding Types Used": len(bindings_used),
    "Unique Features": len(total_features),
    "Platform-Specific Extensions": len(platform_configs),
    "Multi-Binding Support": len(multi_binding_extensions)
}

for stat_name, value in summary_stats.items():
    print(f"{stat_name}: {value}")

print("\n✅ Complete production package configuration demonstrated")
print("Ready for: Development, Testing, CI/CD, and Distribution!")
```
