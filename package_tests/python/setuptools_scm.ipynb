# setuptools_scm Package Testing

This notebook tests the setuptools_scm package for version management from SCM.

## Example 1: Basic import

```python
# setuptools_scm: Automatic version detection from SCM (Git, Mercurial, etc.)
import setuptools_scm
import tempfile
import os
import subprocess
from pathlib import Path

# Import key components
from setuptools_scm import get_version, Configuration

print(f"setuptools_scm module: {setuptools_scm.__name__}")

# Show available functions and classes
important_attrs = [attr for attr in dir(setuptools_scm) 
                  if not attr.startswith('_') and (attr[0].isupper() or attr in ['get_version'])]
print(f"Key components: {', '.join(important_attrs)}")

# Demonstrate version detection with fallback
print("\nVersion detection with fallback:")
try:
    # This will likely fail in notebook environment, but shows proper usage
    version = setuptools_scm.get_version(fallback_version="0.1.0.dev0")
    print(f"Detected version: {version}")
except Exception as e:
    print(f"Expected error (no SCM): {type(e).__name__}")
    print("Using fallback version: 0.1.0.dev0")
    raise
    
# Show how to check if we're in a git repository
current_dir = Path.cwd()
git_dir = current_dir / '.git'
print(f"\nCurrent directory: {current_dir}")
print(f"Git repository detected: {git_dir.exists()}")
```

## Example 2: Version detection

```python
# Demonstrate Configuration class and setup.py integration
import setuptools_scm
from setuptools_scm import Configuration
import tempfile
import textwrap

# Create a Configuration object with various options
config = Configuration(
    root='.',
    relative_to=None,
    local_scheme="no-local-version",  # Don't add local version identifiers
    version_scheme="guess-next-dev",   # Version scheme for development versions
    fallback_version="1.0.0.dev0",   # Fallback when no SCM is detected
    write_to=None,                    # Don't write version to file
    tag_regex=r'^v?(?P<version>[^\+]+)(?P<local>\+.*)?$',  # Tag pattern
)

print("Configuration object created:")
print(f"  Root: {config.root}")
print(f"  Local scheme: {config.local_scheme}")
print(f"  Version scheme: {config.version_scheme}")
print(f"  Fallback version: {config.fallback_version}")
print(f"  Tag regex: {config.tag_regex}")

# Show how to create a setup.py with setuptools_scm
setup_py_content = textwrap.dedent('''
    from setuptools import setup, find_packages
    
    setup(
        name="example-package",
        use_scm_version={
            "local_scheme": "no-local-version",
            "version_scheme": "release-branch-semver",
            "fallback_version": "0.1.0.dev0"
        },
        setup_requires=["setuptools_scm"],
        packages=find_packages(),
        author="Example Author",
        author_email="author@example.com",
        description="Example package using setuptools_scm",
        python_requires=">=3.8",
    )
''').strip()

print(f"\nExample setup.py integration:")
print(setup_py_content)

# Show pyproject.toml configuration alternative
pyproject_content = textwrap.dedent('''
    [build-system]
    requires = ["setuptools>=45", "setuptools_scm[toml]>=6.2"]
    build-backend = "setuptools.build_meta"
    
    [project]
    name = "example-package"
    dynamic = ["version"]
    
    [tool.setuptools_scm]
    version_scheme = "guess-next-dev"
    local_scheme = "no-local-version"
    fallback_version = "0.1.0.dev0"
''').strip()

print(f"\nAlternative pyproject.toml configuration:")
print(pyproject_content)
```

## Example 3: Configuration

```python
# Demonstrate version schemes and customization
import setuptools_scm
from setuptools_scm import Configuration
import re

# Show different version schemes available
version_schemes = [
    "guess-next-dev",
    "post-release", 
    "python-simplified-semver",
    "release-branch-semver",
    "no-guess-dev"
]

print("Available version schemes:")
for scheme in version_schemes:
    print(f"  - {scheme}")

# Show different local schemes
local_schemes = [
    "node-and-date",
    "node-and-timestamp", 
    "dirty-tag",
    "no-local-version"
]

print(f"\nAvailable local schemes:")
for scheme in local_schemes:
    print(f"  - {scheme}")

# Demonstrate custom version formatting
def custom_version_scheme(version):
    """Custom version scheme example."""
    if version.exact:
        return version.format_with("{tag}")
    else:
        return version.format_with("{tag}.dev{distance}+{node}")

def custom_local_scheme(version):
    """Custom local scheme example."""
    if version.dirty:
        return version.format_with("+dirty.{node}")
    else:
        return ""

print(f"\nCustom scheme functions created:")
print(f"  custom_version_scheme: {custom_version_scheme.__doc__.strip()}")
print(f"  custom_local_scheme: {custom_local_scheme.__doc__.strip()}")

# Create configuration with custom schemes
try:
    custom_config = Configuration(
        version_scheme=custom_version_scheme,
        local_scheme=custom_local_scheme,
        fallback_version="2.0.0.dev0"
    )
    print(f"\nCustom configuration created successfully")
    print(f"  Fallback version: {custom_config.fallback_version}")
except Exception as e:
    print(f"Custom config error: {e}")
    raise

# Show tag regex patterns for different tagging conventions
tag_patterns = {
    "simple": r'^(?P<version>[^\+]+)(?P<local>\+.*)?$',
    "v-prefix": r'^v(?P<version>[^\+]+)(?P<local>\+.*)?$',
    "release-prefix": r'^release-(?P<version>[^\+]+)(?P<local>\+.*)?$',
    "complex": r'^(?P<prefix>v?)(?P<version>[^\+]+)(?P<local>\+.*)?$'
}

print(f"\nTag regex patterns for different conventions:")
for name, pattern in tag_patterns.items():
    print(f"  {name}: {pattern}")
```

## Example 4: Build requirements

```python
# Create a simulated git repository for version detection
import setuptools_scm
import tempfile
import subprocess
import os
import re
from pathlib import Path

# Create a temporary directory and simulate a git repository
with tempfile.TemporaryDirectory() as temp_dir:
    temp_path = Path(temp_dir)
    
    # Initialize git repository
    try:
        subprocess.run(['git', 'init'], cwd=temp_dir, capture_output=True, check=True)
        subprocess.run(['git', 'config', 'user.name', 'Test User'], cwd=temp_dir, capture_output=True, check=True)
        subprocess.run(['git', 'config', 'user.email', 'test@example.com'], cwd=temp_dir, capture_output=True, check=True)
        
        # Create initial commit
        readme_file = temp_path / 'README.md'
        readme_file.write_text('# Test Project\nA test project for setuptools_scm')
        
        subprocess.run(['git', 'add', 'README.md'], cwd=temp_dir, capture_output=True, check=True)
        subprocess.run(['git', 'commit', '-m', 'Initial commit'], cwd=temp_dir, capture_output=True, check=True)
        
        # Create a tag
        subprocess.run(['git', 'tag', 'v1.0.0'], cwd=temp_dir, capture_output=True, check=True)
        
        # Make another commit after tag
        setup_file = temp_path / 'setup.py'
        setup_file.write_text('''
from setuptools import setup
setup(
    name="test-package",
    use_scm_version=True,
    setup_requires=["setuptools_scm"],
)
''')
        subprocess.run(['git', 'add', 'setup.py'], cwd=temp_dir, capture_output=True, check=True)
        subprocess.run(['git', 'commit', '-m', 'Add setup.py'], cwd=temp_dir, capture_output=True, check=True)
        
        print("Simulated git repository created:")
        print(f"  Location: {temp_dir}")
        print(f"  Files: {[f.name for f in temp_path.iterdir()]}")
        
        # Now try to get version from this repository
        try:
            version = setuptools_scm.get_version(root=temp_dir)
            print(f"\nVersion detected from git: {version}")
            
            # Show version with different schemes. `get_version` takes the scheme as a keyword; its first
            # positional parameter is `root`, so a Configuration object passed there is refused.
            version_dev = setuptools_scm.get_version(root=temp_dir, version_scheme="guess-next-dev")
            print(f"Version with guess-next-dev: {version_dev}")
            
            version_post = setuptools_scm.get_version(root=temp_dir, version_scheme="post-release")
            print(f"Version with post-release: {version_post}")
            
            # One commit after tag v1.0.0 on a clean tree: guess-next-dev (the default) bumps to the next patch
            # as a dev release, post-release marks it post1; both carry the "+g<commit>" local part.
            assert re.fullmatch(r"1\.0\.1\.dev1\+g[0-9a-f]+", version), version
            assert version_dev == version, (version_dev, version)
            assert re.fullmatch(r"1\.0\.0\.post1\+g[0-9a-f]+", version_post), version_post
            
        except Exception as e:
            print(f"Version detection error: {e}")
            raise
            
        # Show git log for context
        try:
            log_result = subprocess.run(['git', 'log', '--oneline', '--decorate'], 
                                      cwd=temp_dir, capture_output=True, text=True, check=True)
            print(f"\nGit history:")
            for line in log_result.stdout.strip().split('\n'):
                print(f"  {line}")
        except Exception as e:
            print(f"Git log error: {e}")
            raise
            
    except subprocess.CalledProcessError as e:
        print(f"Git command failed: {e}")
        print("Git may not be available in this environment")
    except FileNotFoundError:
        print("Git executable not found - demonstrating without actual git operations")
        
        # Show what would happen with version detection
        print("Simulated version detection (without git):")
        fallback_version = setuptools_scm.get_version(fallback_version="1.0.0.dev0+no.git")
        print(f"  Fallback version used: {fallback_version}")

print(f"\nsetuptools_scm provides automatic version management from SCM tags and commits")
```

## Example 5: Version info

```python
# Demonstrate programmatic version retrieval and advanced features
import setuptools_scm
from setuptools_scm import Configuration
import os
import textwrap

# Show environment variable overrides
print("Environment variable overrides:")
env_vars = {
    'SETUPTOOLS_SCM_PRETEND_VERSION': 'Override version detection',
    'SETUPTOOLS_SCM_DEBUG': 'Enable debug logging',
    'SETUPTOOLS_SCM_SUBPROCESS_TIMEOUT': 'Set subprocess timeout (default: 40s)',
    'SOURCE_DATE_EPOCH': 'Set reproducible build timestamp'
}

for var, description in env_vars.items():
    current_value = os.environ.get(var, 'Not set')
    print(f"  {var}: {description}")
    print(f"    Current value: {current_value}")

# Demonstrate version pretending for testing
print(f"\nDemonstrating version pretending:")
pretend_version = "2.5.0"
os.environ['SETUPTOOLS_SCM_PRETEND_VERSION'] = pretend_version

try:
    version = setuptools_scm.get_version()
    print(f"Pretended version: {version}")
except Exception as e:
    print(f"Pretend version result: {pretend_version} (would be used)")
    raise
finally:
    # Clean up environment
    if 'SETUPTOOLS_SCM_PRETEND_VERSION' in os.environ:
        del os.environ['SETUPTOOLS_SCM_PRETEND_VERSION']

# Show version file writing capability
print(f"\nVersion file writing configuration:")
version_file_template = textwrap.dedent('''
    # coding: utf-8
    # file generated by setuptools_scm
    # don't change, don't track in version control
    __version__ = version = '{version}'
    __version_tuple__ = version_tuple = {version_tuple!r}
''').strip()

print("Template for version file:")
print(version_file_template)

# Create configuration that writes version to file
config_with_file = Configuration(
    write_to="version.py",
    write_to_template=version_file_template,
    fallback_version="3.0.0.dev0"
)

print(f"\nConfiguration for version file writing:")
print(f"  Write to: {config_with_file.write_to}")
print(f"  Has template: {config_with_file.write_to_template is not None}")

# Show integration with different build systems
build_systems = {
    "setuptools": {
        "setup.py": "use_scm_version=True",
        "pyproject.toml": "[tool.setuptools_scm]"
    },
    "flit": {
        "pyproject.toml": "[build-system]\nrequires = ['flit_core >=3.2,<4', 'setuptools_scm']"
    },
    "poetry": {
        "pyproject.toml": "[tool.poetry.dependencies]\nsetuptools-scm = '*'"
    },
    "hatch": {
        "pyproject.toml": "[tool.hatch.version]\nsource = 'vcs'"
    }
}

print(f"\nIntegration with build systems:")
for system, configs in build_systems.items():
    print(f"  {system}:")
    for file, config in configs.items():
        print(f"    {file}: {config}")

print(f"\nsetuptools_scm key benefits:")
print("✓ Automatic version detection from VCS tags")
print("✓ Support for Git, Mercurial, and other SCM systems")
print("✓ Configurable version schemes and local schemes")
print("✓ Integration with modern build systems")
print("✓ Reproducible builds with SOURCE_DATE_EPOCH")
print("✓ Development version generation")
print("✓ Version file writing for runtime access")
```
