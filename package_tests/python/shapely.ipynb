## Example 1: shapely functionality test 1

```python
# Example 1: Basic Shapely Geometric Objects
from shapely.geometry import Point, LineString, Polygon

print("=== Shapely Geometric Objects ===")

# Create basic geometries
point = Point(5, 10)
line = LineString([(0, 0), (1, 1), (2, 3), (3, 5)])
polygon = Polygon([(0, 0), (0, 5), (5, 5), (5, 0)])

print(f"Point: x={point.x}, y={point.y}")
print(f"Point bounds: {point.bounds}")

print(f"\nLineString length: {line.length:.2f}")
print(f"Coordinates: {list(line.coords)}")

print(f"\nPolygon area: {polygon.area}")
print(f"Perimeter: {polygon.length:.2f}")
print(f"Is valid: {polygon.is_valid}")

# Polygon with hole
exterior = [(0, 0), (0, 10), (10, 10), (10, 0)]
hole = [(2, 2), (2, 8), (8, 8), (8, 2)]
poly_with_hole = Polygon(exterior, [hole])
print(f"\nPolygon with hole area: {poly_with_hole.area}")

print("Example 1 completed!")
```

## Example 2: shapely functionality test 2

```python
# Example 2: Spatial Relationships
from shapely.geometry import Point, Polygon, LineString

print("=== Spatial Relationships ===")

# Create test geometries
point1 = Point(2, 2)
point2 = Point(10, 10)
polygon1 = Polygon([(1, 1), (1, 4), (4, 4), (4, 1)])
polygon2 = Polygon([(3, 3), (3, 6), (6, 6), (6, 3)])

print(f"Polygon1 contains Point1: {polygon1.contains(point1)}")
print(f"Point1 within Polygon1: {point1.within(polygon1)}")
print(f"Polygon1 intersects Polygon2: {polygon1.intersects(polygon2)}")
print(f"Polygon1 disjoint from Point2: {polygon1.disjoint(point2)}")

# Distance calculations
print(f"\nDistance Point1 to Point2: {point1.distance(point2):.2f}")
print(f"Distance Polygon1 to Polygon2: {polygon1.distance(polygon2):.2f}")

# Equality test
point1_copy = Point(2, 2)
print(f"\nPoint1 equals copy: {point1.equals(point1_copy)}")

# Overlaps
print(f"Polygon1 overlaps Polygon2: {polygon1.overlaps(polygon2)}")

print("Example 2 completed!")
```

## Example 3: shapely functionality test 3

```python
# Example 3: Geometric Operations
from shapely.geometry import Polygon, LineString
from shapely.ops import unary_union

print("=== Geometric Operations ===")

# Create polygons
poly1 = Polygon([(0, 0), (0, 4), (4, 4), (4, 0)])
poly2 = Polygon([(2, 2), (2, 6), (6, 6), (6, 2)])

# Set operations
union = poly1.union(poly2)
intersection = poly1.intersection(poly2)
difference = poly1.difference(poly2)

print(f"Union area: {union.area}")
print(f"Intersection area: {intersection.area}")
print(f"Difference area: {difference.area}")

# Buffer operations
line = LineString([(0, 0), (5, 0), (5, 5)])
buffered = line.buffer(1.0)
print(f"\nLine length: {line.length:.2f}")
print(f"Buffered area: {buffered.area:.2f}")

# Simplification
print(f"\nCentroid: {poly1.centroid}")
print(f"Convex hull area: {poly1.convex_hull.area}")

print("Example 3 completed!")
```

## Example 4: shapely functionality test 4

```python
# Example 4: Affine Transformations
from shapely.geometry import Polygon
from shapely import affinity

print("=== Affine Transformations ===")

# Create test polygon
poly = Polygon([(0, 0), (1, 0), (1, 1), (0, 1)])
print(f"Original area: {poly.area}")
print(f"Original centroid: {poly.centroid}")

# Translation
translated = affinity.translate(poly, xoff=5, yoff=3)
print(f"\nTranslated centroid: {translated.centroid}")

# Scaling
scaled = affinity.scale(poly, xfact=2, yfact=3, origin='center')
print(f"Scaled area: {scaled.area} (6x original)")

# Rotation
rotated = affinity.rotate(poly, 45, origin='center')
print(f"Rotated bounds: {rotated.bounds}")

# Skew
skewed = affinity.skew(poly, xs=15, ys=0, origin='center')
print(f"Skewed area: {skewed.area:.3f}")

print("Example 4 completed!")
```

## Example 5: shapely functionality test 5

```python
# Example 5: Spatial Indexing and WKT
from shapely.geometry import Point, Polygon
from shapely.strtree import STRtree
from shapely import wkt

print("=== Spatial Indexing ===")

# Create polygons for indexing
polygons = []
for i in range(10):
    x, y = i * 2, i * 2
    poly = Polygon([(x, y), (x+1, y), (x+1, y+1), (x, y+1)])
    polygons.append(poly)

print(f"Created {len(polygons)} polygons")

# Build spatial index
tree = STRtree(polygons)
query_poly = Polygon([(4.5, 4.5), (6.5, 4.5), (6.5, 6.5), (4.5, 6.5)])
matches = list(tree.query(query_poly))
print(f"Query found {len(matches)} candidates")

# WKT serialization
test_poly = Polygon([(0, 0), (0, 3), (3, 3), (3, 0)])
wkt_str = test_poly.wkt
print(f"\nWKT: {wkt_str[:50]}...")
reconstructed = wkt.loads(wkt_str)
print(f"Round-trip successful: {test_poly.equals(reconstructed)}")

print("Example 5 completed!")
```
