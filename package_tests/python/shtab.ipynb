## Example 1: shtab functionality test 1

```python
# Example 1: Basic shtab Shell Completion
import shtab
import argparse

print("=== shtab Shell Completion ===")

# Create simple argument parser
parser = argparse.ArgumentParser(prog="myapp", description="Sample CLI")
parser.add_argument("command", choices=["build", "test", "deploy"], help="Command")
parser.add_argument("--verbose", "-v", action="store_true", help="Verbose mode")
parser.add_argument("--config", type=str, help="Config file")

print(f"Program: {parser.prog}")

# Generate bash completion
bash_completion = shtab.complete(parser, shell="bash")
print(f"\nBash completion: {len(bash_completion)} characters")
print(f"Preview: {bash_completion[:80]}...")

# Generate zsh completion
zsh_completion = shtab.complete(parser, shell="zsh")
print(f"\nZsh completion: {len(zsh_completion)} characters")
print(f"Preview: {zsh_completion[:80]}...")

print(f"\nShtab version: {getattr(shtab, '__version__', 'Unknown')}")
print("Example 1 completed!")
```

## Example 2: shtab functionality test 2

```python
# Example 2: Subcommands and Choices
import shtab
import argparse

print("=== Subcommands and Choices ===")

# Create parser with subcommands
parser = argparse.ArgumentParser(prog="tool")
parser.add_argument("--env", choices=["dev", "prod"], default="dev")

subparsers = parser.add_subparsers(dest="command")

# Build subcommand
build = subparsers.add_parser("build")
build.add_argument("--target", choices=["debug", "release"], default="debug")

# Test subcommand
test = subparsers.add_parser("test")
test.add_argument("--coverage", action="store_true")

print(f"Parser: {parser.prog}")
print(f"Subcommands: build, test")

# Generate completion
completion = shtab.complete(parser, shell="bash")
print(f"\nCompletion size: {len(completion)} characters")
print(f"Has subcommands: {'build' in completion and 'test' in completion}")

print("Example 2 completed!")
```

## Example 3: shtab functionality test 3

```python
# Example 3: Multiple Shell Formats
import shtab
import argparse

print("=== Multiple Shell Formats ===")

# Create parser
parser = argparse.ArgumentParser(prog="multicli")
parser.add_argument("action", choices=["start", "stop", "restart"])
parser.add_argument("--port", type=int, default=8080)
parser.add_argument("--host", type=str, default="localhost")

print(f"Parser: {parser.prog}")

# Generate for different shells
shells = ["bash", "zsh", "tcsh"]
for shell in shells:
    try:
        completion = shtab.complete(parser, shell=shell)
        print(f"\n{shell.upper()}:")
        print(f"  Size: {len(completion)} characters")
        print(f"  Lines: {completion.count(chr(10)) + 1}")
        print(f"  Has choices: {'start' in completion}")
    except Exception as e:
        print(f"\n{shell.upper()}: Error - {e}")
        raise

print("\nExample 3 completed!")
```

## Example 4: shtab functionality test 4

```python
# Example 4: Completion with Preamble
import shtab
import argparse

print("=== Completion with Preamble ===")

# Create parser
parser = argparse.ArgumentParser(prog="app", description="Sample application")
parser.add_argument("--log-level", choices=["DEBUG", "INFO", "ERROR"], default="INFO")
parser.add_argument("--output", type=str, help="Output file")

# Generate with custom preamble
preamble = "# Custom completion script\n# Generated by shtab"
completion = shtab.complete(parser, shell="bash", preamble=preamble)

print(f"Parser: {parser.prog}")
print(f"Completion size: {len(completion)} characters")
print(f"Has preamble: {'Custom completion' in completion}")
print(f"Has log levels: {'DEBUG' in completion}")

# Check for argument completion
print(f"\nFeatures:")
print(f"  --log-level: {'--log-level' in completion}")
print(f"  --output: {'--output' in completion}")

print("Example 4 completed!")
```

## Example 5: shtab functionality test 5

```python
# Example 5: Nested Subcommands
import shtab
import argparse

print("=== Nested Subcommands ===")

# Create parser with nested structure
parser = argparse.ArgumentParser(prog="cli")
subparsers = parser.add_subparsers(dest="cmd")

# Database commands
db = subparsers.add_parser("db", help="Database commands")
db_sub = db.add_subparsers(dest="db_cmd")
db_sub.add_parser("migrate", help="Run migrations")
db_sub.add_parser("backup", help="Backup database")

# Server commands
server = subparsers.add_parser("server", help="Server commands")
server.add_argument("--port", type=int, default=8000)

print(f"Parser: {parser.prog}")

# Generate completion
completion = shtab.complete(parser, shell="bash")
print(f"\nCompletion size: {len(completion)} characters")
print(f"Has db commands: {'migrate' in completion and 'backup' in completion}")
print(f"Has server: {'server' in completion}")

print("Example 5 completed!")
```
