```python
# Example 1: Sidecar widget class is exposed at the top level.
from sidecar import Sidecar

assert Sidecar is not None
print(f"Sidecar class: {Sidecar}")
print(f"Sidecar module: {Sidecar.__module__}")
print(f"Sidecar MRO: {[c.__name__ for c in Sidecar.__mro__]}")
```

```python
# Example 2: Sidecar is an ipywidgets DOMWidget — verify inheritance.
import ipywidgets
from sidecar import Sidecar

assert issubclass(Sidecar, ipywidgets.DOMWidget), "Sidecar must extend ipywidgets.DOMWidget"
print(f"ipywidgets version: {ipywidgets.__version__}")
print(f"Sidecar is a DOMWidget subclass: True")
```

```python
# Example 3: Instantiate Sidecar with a title — verify traitlets work.
from sidecar import Sidecar

sc = Sidecar(title="My Sidecar")
assert sc.title == "My Sidecar"
print(f"title: {sc.title!r}")
print(f"model_id: {sc.model_id}")
sc.close()
```

```python
# Example 4: Verify the `anchor` trait exists and accepts a valid value.
from sidecar import Sidecar

sc = Sidecar(title="Anchored", anchor="split-right")
assert sc.anchor == "split-right"
valid_anchors = sc.trait_metadata("anchor", "values") or sc.traits()["anchor"].values
print(f"anchor: {sc.anchor!r}")
print(f"valid anchor values: {valid_anchors}")
sc.close()
```

```python
# Example 5: Sidecar works as a context manager (Output-like behavior).
from sidecar import Sidecar

sc = Sidecar(title="Context Test")
with sc:
    print("hello from inside sidecar")

# Traits we expect on every Sidecar instance
expected_traits = {"title", "anchor"}
actual_traits = set(sc.traits().keys())
missing = expected_traits - actual_traits
assert not missing, f"missing traits: {missing}"
print(f"all expected traits present: {sorted(expected_traits)}")
sc.close()
```
