## Example 1: simpervisor functionality test 1

```python
# Example 1: Basic Process Supervision
import simpervisor
import asyncio
import sys

print("=== Basic Process Supervision ===")

# Create supervised process
if sys.platform.startswith('win'):
    cmd = ['ping', '-n', '3', '127.0.0.1']
else:
    cmd = ['sleep', '2']

proc = simpervisor.SupervisedProcess('test-proc', *cmd)
print(f"Created: {proc.name}")

async def test_basic():
    print("Starting process...")
    await proc.start()
    print(f"PID: {proc.pid}")
    
    await asyncio.sleep(1)
    print("Terminating...")
    await proc.terminate()
    print("Terminated!")

await test_basic()
print("Example 1 completed!")
```

## Example 2: simpervisor functionality test 2

```python
# Example 2: Process Monitoring
import simpervisor
import asyncio
import sys

print("=== Process Monitoring ===")

cmd = ['sleep', '3'] if not sys.platform.startswith('win') else ['ping', '-n', '3', '127.0.0.1']
proc = simpervisor.SupervisedProcess('monitor-test', *cmd)

async def test_monitoring():
    print(f"Starting {proc.name}...")
    await proc.start()
    print(f"Started with PID: {proc.pid}")
    
    print("Monitoring process for 2 seconds...")
    for i in range(4):
        await asyncio.sleep(0.5)
        if proc.proc:
            running = proc.proc.returncode is None
            print(f"  Check {i+1}: {'Running' if running else 'Stopped'}")
    
    await proc.terminate()
    print("Process terminated")

await test_monitoring()
print("Example 2 completed!")
```

## Example 3: simpervisor functionality test 3

```python
# Example 3: Multiple Processes
import simpervisor
import asyncio
import sys

print("=== Multiple Processes ===")

cmd = ['sleep', '3'] if not sys.platform.startswith('win') else ['ping', '-n', '3', '127.0.0.1']

processes = [
    simpervisor.SupervisedProcess(f'proc-{i}', *cmd)
    for i in range(3)
]

async def test_multiple():
    print(f"Starting {len(processes)} processes...")
    
    for proc in processes:
        await proc.start()
        print(f"  Started {proc.name} (PID: {proc.pid})")
    
    await asyncio.sleep(2)
    
    print("Stopping all processes...")
    for proc in processes:
        await proc.terminate()
        print(f"  Stopped {proc.name}")

await test_multiple()
print("Example 3 completed!")
```

## Example 4: simpervisor functionality test 4

```python
# Example 4: Process Kill
import simpervisor
import asyncio
import sys

print("=== Process Kill ===")

cmd = ['sleep', '10'] if not sys.platform.startswith('win') else ['ping', '-t', '127.0.0.1']
proc = simpervisor.SupervisedProcess('kill-test', *cmd)

async def test_kill():
    print("Starting long-running process...")
    await proc.start()
    print(f"Started PID: {proc.pid}")
    
    await asyncio.sleep(1)
    
    print("Attempting terminate...")
    await proc.terminate()
    await asyncio.sleep(0.5)
    
    if proc.proc and proc.proc.returncode is None:
        print("Still running, using kill...")
        await proc.kill()
    
    print("Process stopped")

await test_kill()
print("Example 4 completed!")
```

## Example 5: simpervisor functionality test 5

```python
# Example 5: Process Lifecycle
import simpervisor
import asyncio
import sys

print("=== Process Lifecycle ===")

cmd = ['sleep', '5'] if not sys.platform.startswith('win') else ['ping', '-n', '5', '127.0.0.1']
proc = simpervisor.SupervisedProcess('lifecycle', *cmd)

async def test_lifecycle():
    print(f"Initial state - Name: {proc.name}")
    
    print("\n1. Starting process...")
    await proc.start()
    print(f"   PID after start: {proc.pid}")
    
    print("\n2. Checking process status...")
    if proc.proc:
        print(f"   Running: {proc.proc.returncode is None}")
    
    await asyncio.sleep(2)
    
    print("\n3. Terminating process...")
    await proc.terminate()
    await asyncio.sleep(0.5)
    
    print(f"   Final return code: {proc.proc.returncode if proc.proc else 'N/A'}")

await test_lifecycle()
print("\nExample 5 completed!")
```
