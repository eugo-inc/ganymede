## Example 1: Basic Command-line Argument Parsing

```python
# Example 1: Basic Argument Parsing
from dataclasses import dataclass
import simple_parsing

print("=== Basic Argument Parsing ===")

@dataclass
class Config:
    name: str
    count: int = 10
    verbose: bool = False

parser = simple_parsing.ArgumentParser()
parser.add_arguments(Config, dest="config")

# Test parsing
args = parser.parse_args(["--name", "test", "--count", "5", "--verbose"])
config = args.config

print(f"Name: {config.name}")
print(f"Count: {config.count}")
print(f"Verbose: {config.verbose}")
print(f"Type check - count is int: {isinstance(config.count, int)}")

print("Example 1 completed!")
```

## Example 2: Nested Configuration and Inheritance

```python
# Example 2: Lists and Optional Values
from dataclasses import dataclass, field
from typing import List, Optional
import simple_parsing

print("=== Lists and Optional Values ===")

@dataclass
class Config:
    tags: List[str] = field(default_factory=list)
    values: List[float] = field(default_factory=list)
    description: Optional[str] = None

parser = simple_parsing.ArgumentParser()
parser.add_arguments(Config, dest="config")

args = parser.parse_args(["--tags", "ml", "test", "--values", "1.5", "2.3", "--description", "demo"])
config = args.config

print(f"Tags: {config.tags}")
print(f"Values: {config.values}")
print(f"Description: {config.description}")
print(f"All values are float: {all(isinstance(v, float) for v in config.values)}")

print("Example 2 completed!")
```

## Example 3: Subcommands and Validation

```python
# Example 3: Nested Configuration
from dataclasses import dataclass, field
import simple_parsing

print("=== Nested Configuration ===")

@dataclass
class DatabaseConfig:
    host: str = "localhost"
    port: int = 5432

@dataclass
class AppConfig:
    name: str = "app"
    debug: bool = False
    database: DatabaseConfig = field(default_factory=DatabaseConfig)

parser = simple_parsing.ArgumentParser()
parser.add_arguments(AppConfig, dest="config")

args = parser.parse_args(["--name", "myapp", "--host", "db.example.com", "--port", "3306"])
config = args.config

print(f"App: {config.name}, Debug: {config.debug}")
print(f"DB Host: {config.database.host}")
print(f"DB Port: {config.database.port}")

print("Example 3 completed!")
```

## Example 4: Configuration Files and Serialization

```python
# Example 4: Type Validation
from dataclasses import dataclass
from typing import Literal
import simple_parsing

print("=== Type Validation ===")

@dataclass
class Config:
    model: str = "resnet"
    optimizer: Literal["adam", "sgd", "rmsprop"] = "adam"
    learning_rate: float = 0.001
    epochs: int = 10

parser = simple_parsing.ArgumentParser()
parser.add_arguments(Config, dest="config")

args = parser.parse_args(["--model", "vgg", "--optimizer", "sgd", "--learning_rate", "0.01", "--epochs", "20"])
config = args.config

print(f"Model: {config.model}")
print(f"Optimizer: {config.optimizer}")
print(f"LR: {config.learning_rate} (type: {type(config.learning_rate).__name__})")
print(f"Epochs: {config.epochs} (type: {type(config.epochs).__name__})")

print("Example 4 completed!")
```

## Example 5: Production Patterns and Best Practices

```python
# Example 5: Default Values
from dataclasses import dataclass, field
from typing import List
import simple_parsing

print("=== Default Values ===")

@dataclass
class Config:
    name: str = "default_app"
    port: int = 8000
    workers: int = 4
    features: List[str] = field(default_factory=lambda: ["api", "web"])

parser = simple_parsing.ArgumentParser()
parser.add_arguments(Config, dest="config")

# Parse with minimal args (using defaults)
args1 = parser.parse_args([])
print(f"Defaults: name={args1.config.name}, port={args1.config.port}, features={args1.config.features}")

# Parse with overrides
args2 = parser.parse_args(["--name", "custom", "--port", "9000", "--features", "admin", "metrics"])
print(f"Custom: name={args2.config.name}, port={args2.config.port}, features={args2.config.features}")

print("Example 5 completed!")
```
