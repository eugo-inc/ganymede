## Example 1: simplejson functionality test 1

```python
# Example 1: Basic JSON Operations
import simplejson

print("=== Basic JSON Operations ===")

# Test data
data = {
    'name': 'John Doe',
    'age': 30,
    'languages': ['Python', 'JavaScript'],
    'is_dev': True
}

# Encode to JSON
json_str = simplejson.dumps(data)
print(f"Encoded: {json_str}")

# Decode from JSON
decoded = simplejson.loads(json_str)
print(f"Decoded: {decoded}")
print(f"Data preserved: {decoded == data}")

print(f"simplejson version: {getattr(simplejson, '__version__', 'Unknown')}")
print("Example 1 completed!")
```

## Example 2: simplejson functionality test 2

```python
# Example 2: Formatting Options
import simplejson

print("=== Formatting Options ===")

data = {'z': 1, 'a': 2, 'm': 3, 'nested': {'x': 10, 'y': 20}}

# Compact (default)
compact = simplejson.dumps(data)
print(f"Compact: {compact}")

# Pretty printing
pretty = simplejson.dumps(data, indent=2)
print(f"Pretty:\n{pretty}")

# Sorted keys
sorted_json = simplejson.dumps(data, sort_keys=True)
print(f"Sorted: {sorted_json}")

# Unicode handling
unicode_data = {'emoji': '🐍', 'chinese': '你好'}
ascii_encoded = simplejson.dumps(unicode_data, ensure_ascii=True)
unicode_allowed = simplejson.dumps(unicode_data, ensure_ascii=False)
print(f"ASCII: {ascii_encoded}")
print(f"Unicode: {unicode_allowed}")

print("Example 2 completed!")
```

## Example 3: simplejson functionality test 3

```python
# Example 3: Error Handling
import simplejson

print("=== Error Handling ===")

# Test invalid JSON
invalid_cases = [
    ('{"key": }', 'Missing value'),
    ('[1, 2, 3', 'Incomplete array'),
    ('{"a": 1 "b": 2}', 'Missing comma'),
]

for invalid_json, description in invalid_cases:
    print(f"\n{description}: {invalid_json}")
    try:
        result = simplejson.loads(invalid_json)
    except simplejson.JSONDecodeError as e:
        print(f"  ✓ Rejected at line {e.lineno}, col {e.colno}: {e.msg}")
    else:
        raise AssertionError(f"simplejson.loads accepted invalid JSON {invalid_json!r} -> {result!r}")

# Valid edge cases
valid_cases = ['42', '"string"', 'true', 'null', '[]', '{}']
print(f"\nValid edge cases:")
for valid_json in valid_cases:
    result = simplejson.loads(valid_json)
    print(f"  {valid_json} -> {result} ({type(result).__name__})")

print("Example 3 completed!")
```

## Example 4: simplejson functionality test 4

```python
# Example 4: Custom Serialization
import simplejson
from decimal import Decimal
from datetime import datetime

print("=== Custom Serialization ===")

def custom_serializer(obj):
    if isinstance(obj, datetime):
        return obj.isoformat()
    elif isinstance(obj, Decimal):
        return float(obj)
    return str(obj)

# Test data with custom types
data = {
    'timestamp': datetime(2024, 1, 15, 10, 30),
    'price': Decimal('19.99'),
    'custom': set([1, 2, 3])  # Sets not JSON serializable
}

json_str = simplejson.dumps(data, default=custom_serializer)
print(f"Custom serialized: {json_str}")

# Decimal handling on decode
decimal_json = '{"value": 123.456789}'
decoded = simplejson.loads(decimal_json, use_decimal=True)
print(f"Decoded with Decimal: {decoded}")
print(f"Type: {type(decoded['value'])}")

print("Example 4 completed!")
```

## Example 5: simplejson functionality test 5

```python
# Example 5: File I/O
import simplejson
import io

print("=== File I/O ===")

# Data to save
data = {
    'users': [{'id': i, 'name': f'User{i}'} for i in range(5)],
    'metadata': {'count': 5, 'version': '1.0'}
}

# Write to string buffer
output = io.StringIO()
simplejson.dump(data, output, indent=2)
json_str = output.getvalue()
print(f"JSON written to buffer ({len(json_str)} chars)")

# Read from string buffer
input_buffer = io.StringIO(json_str)
loaded = simplejson.load(input_buffer)
print(f"Loaded {len(loaded['users'])} users")
print(f"Metadata: {loaded['metadata']}")

# Special float values
special = {'inf': float('inf'), 'nan': float('nan')}
with_nan = simplejson.dumps(special, allow_nan=True)
print(f"\nWith NaN allowed: {with_nan}")

print("Example 5 completed!")
```
