# six: 5 Basic Usage Examples

```python
# Example 1: Using six for Python 2/3 compatibility
import six
print(six.PY2, six.PY3)
```

```python
# Example 2: Using six.moves for cross-version imports
import six
print(six.moves.range(5))
```

```python
# Example 3: Using six.iteritems for dict iteration
import six
d = {'a': 1, 'b': 2}
for k, v in six.iteritems(d):
    print(k, v)
```

```python
# Example 4: Using six.with_metaclass
import six
class Meta(type): pass
class MyClass(six.with_metaclass(Meta, object)): pass
print(type(MyClass))
```

```python
# Example 5: Using six.string_types
import six
print(isinstance('hello', six.string_types))
```
