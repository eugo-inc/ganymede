## Example 1: Basic Async Library Detection

```python
# Example 1: Basic Library Detection
import sniffio

print("=== Basic Library Detection ===")

print(f"sniffio version: {getattr(sniffio, '__version__', 'Unknown')}")

# Available functions
attrs = [attr for attr in dir(sniffio) if not attr.startswith('_')]
print(f"Available: {', '.join(attrs)}")

# Test detection in sync context
try:
    lib = sniffio.current_async_library()
    print(f"Detected library: {lib}")
except sniffio.AsyncLibraryNotFoundError:
    print("No async library detected (expected in sync context)")

# Test exception
try:
    sniffio.current_async_library()
except sniffio.AsyncLibraryNotFoundError as e:
    print(f"Exception type: {type(e).__name__}")
    print(f"Exception message: {str(e)[:50]}")

print("Example 1 completed!")
```

## Example 2: Cross-platform Compatibility Patterns

```python
# Example 2: Library Support Check
import sniffio

print("=== Library Support Check ===")

class LibraryChecker:
    def __init__(self):
        self.supported = ['asyncio', 'trio', 'curio']
    
    def check(self):
        try:
            lib = sniffio.current_async_library()
            is_supported = lib in self.supported
            return lib, is_supported
        except sniffio.AsyncLibraryNotFoundError:
            return None, False

checker = LibraryChecker()
lib, supported = checker.check()

if lib:
    print(f"Library: {lib}, Supported: {supported}")
else:
    print("No async library detected")

print(f"Supported libraries: {', '.join(checker.supported)}")
print("Example 2 completed!")
```

## Example 3: Advanced Integration Scenarios

```python
# Example 3: Error Handling
import sniffio

print("=== Error Handling ===")

def safe_detect():
    try:
        return sniffio.current_async_library()
    except sniffio.AsyncLibraryNotFoundError:
        return None
    except Exception as e:
        return f"Error: {e}"

# Test multiple calls
results = []
for i in range(3):
    result = safe_detect()
    results.append(result)
    if result:
        print(f"Detection {i+1}: {result}")
    else:
        print(f"Detection {i+1}: No library")

# Summary
unique_results = set(r for r in results if r)
print(f"\nUnique results: {unique_results if unique_results else 'None'}")
print(f"Total detections: {len(results)}")

print("Example 3 completed!")
```

## Example 4: Testing and Development Utilities

```python
# Example 4: Context Information
import sniffio

print("=== Context Information ===")

def get_context_info():
    info = {'has_library': False, 'library': None}
    
    try:
        lib = sniffio.current_async_library()
        info['has_library'] = True
        info['library'] = lib
    except sniffio.AsyncLibraryNotFoundError:
        pass
    
    return info

# Get context
context = get_context_info()

print(f"Has async library: {context['has_library']}")
print(f"Library name: {context['library'] or 'None'}")

# Additional info
print(f"\nModule attributes:")
for attr in ['AsyncLibraryNotFoundError', 'current_async_library']:
    has_it = hasattr(sniffio, attr)
    print(f"  {attr}: {has_it}")

print("Example 4 completed!")
```

## Example 5: Production Patterns and Best Practices

```python
# Example 5: Detection Wrapper
import sniffio

print("=== Detection Wrapper ===")

class AsyncDetector:
    def __init__(self):
        self.attempts = 0

    def detect(self):
        self.attempts += 1
        try:
            return sniffio.current_async_library()
        except sniffio.AsyncLibraryNotFoundError:
            return None

detector = AsyncDetector()

# Multiple detections
for i in range(3):
    lib = detector.detect()
    print(f"Attempt {i+1}: {lib or 'None'}")

print(f"Total attempts: {detector.attempts}")
print("Example 5 completed!")
```
