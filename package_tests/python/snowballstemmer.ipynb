## Example 1: snowballstemmer functionality test 1

```python
# Example 1: Basic Stemming
import snowballstemmer

print("=== Basic Stemming ===")

# Create English stemmer
stemmer = snowballstemmer.stemmer('english')

# Test words
words = ['running', 'runs', 'ran', 'runner', 'easily', 
         'computing', 'computer', 'computation']

print("English stemming:")
for word in words:
    stemmed = stemmer.stemWord(word)
    print(f"  {word:12} -> {stemmed}")

# Try French stemmer
fr_stemmer = snowballstemmer.stemmer('french')
fr_words = ['courir', 'courant', 'couru', 'coureur']

print("\nFrench stemming:")
for word in fr_words:
    stemmed = fr_stemmer.stemWord(word)
    print(f"  {word:12} -> {stemmed}")

print("Example 1 completed!")
```

## Example 2: snowballstemmer functionality test 2

```python
# Example 2: Text Processing
import snowballstemmer
from collections import Counter

print("=== Text Processing ===")

text = "The runners were running fast through the beautiful gardens"

stemmer = snowballstemmer.stemmer('english')
words = text.lower().split()

print(f"Original: {text}")
stemmed_words = [stemmer.stemWord(word) for word in words]
print(f"Stemmed: {' '.join(stemmed_words)}")

# Process multiple documents
docs = [
    "Computing systems require powerful computers",
    "Natural language processing involves stemming algorithms"
]

all_stems = []
for doc in docs:
    stems = [stemmer.stemWord(w) for w in doc.lower().split()]
    all_stems.extend(stems)

counts = Counter(all_stems)
print(f"\nTop stems: {counts.most_common(5)}")
print("Example 2 completed!")
```

## Example 3: snowballstemmer functionality test 3

```python
# Example 3: Search Indexing
import snowballstemmer

print("=== Search Indexing ===")

corpus = {
    "doc1": "Python programming language supports object-oriented programming",
    "doc2": "Machine learning algorithms process large datasets"
}

stemmer = snowballstemmer.stemmer('english')

# Build index
index = {}
for doc_id, text in corpus.items():
    words = text.lower().split()
    for word in words:
        stem = stemmer.stemWord(word)
        if stem not in index:
            index[stem] = []
        index[stem].append(doc_id)

print(f"Indexed {len(corpus)} documents, {len(index)} unique stems")

# Search
query = "programming languages"
query_stems = [stemmer.stemWord(w) for w in query.lower().split()]
matches = set()
for stem in query_stems:
    if stem in index:
        matches.update(index[stem])

print(f"\nQuery: '{query}'")
print(f"Matches: {list(matches)}")
print("Example 3 completed!")
```

## Example 4: snowballstemmer functionality test 4

```python
# Example 4: Stemming Quality Analysis
import snowballstemmer

print("=== Stemming Quality Analysis ===")

stemmer = snowballstemmer.stemmer('english')

# Test word groups
test_groups = {
    "verbs": ["run", "runs", "running", "ran", "runner"],
    "adjectives": ["happy", "happiness", "happier", "happiest"],
    "tech": ["compute", "computing", "computer", "computation"]
}

for category, words in test_groups.items():
    print(f"\n{category.title()}:")
    stems = []
    for word in words:
        stem = stemmer.stemWord(word)
        stems.append(stem)
        print(f"  {word:12} -> {stem}")
    unique = len(set(stems))
    print(f"  Unique stems: {unique}")

print("\nExample 4 completed!")
```

## Example 5: snowballstemmer functionality test 5

```python
# Example 5: Text Processing Pipeline
import snowballstemmer

print("=== Text Processing Pipeline ===")

class TextProcessor:
    def __init__(self):
        self.stemmer = snowballstemmer.stemmer('english')
        self.stop_words = {'the', 'a', 'an', 'and', 'or', 'in', 'on', 'at', 'to'}
    
    def process(self, text):
        words = text.lower().split()
        filtered = [w for w in words if w not in self.stop_words]
        stemmed = [self.stemmer.stemWord(w) for w in filtered]
        return stemmed

processor = TextProcessor()

text = "The algorithms are processing the datasets efficiently"
result = processor.process(text)

print(f"Original: {text}")
print(f"Processed: {' '.join(result)}")
print(f"Reduction: {len(text.split())} -> {len(result)} words")

# Batch processing
texts = ["Machine learning models", "Natural language processing"]
for t in texts:
    print(f"'{t}' -> {' '.join(processor.process(t))}")

print("Example 5 completed!")
```
