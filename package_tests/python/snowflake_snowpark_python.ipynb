## Example 1: Basic Import and Module Exploration

```python
# Example 1: Basic Import and Module Exploration
from snowflake.snowpark import Session
import snowflake.snowpark

print("=== Basic Import and Module Exploration ===")

# Check available attributes
attrs = [a for a in dir(snowflake.snowpark) if not a.startswith('_')]
print(f"Available attributes: {len(attrs)}")
print(f"Key classes: {[a for a in attrs if a[0].isupper()][:5]}")

# Session builder available
print(f"\nSession class available: {hasattr(Session, 'builder')}")

# Connection parameters structure
params = {
    'account': 'myaccount',
    'user': 'myuser',
    'password': 'mypass',
    'warehouse': 'COMPUTE_WH',
    'database': 'MYDB',
    'schema': 'PUBLIC'
}

print(f"\nConnection parameters:")
for key, val in params.items():
    masked = '*' * 8 if 'pass' in key else val
    print(f"  {key}: {masked}")

print("Example 1 completed!")
```

## Example 2: DataFrame Types and Functions

```python
# Example 2: DataFrame Types and Functions
from snowflake.snowpark import types, functions
import snowflake.snowpark

print("=== DataFrame Types and Functions ===")

# Check types module
type_attrs = [a for a in dir(types) if not a.startswith('_')]
print(f"Types module attributes: {len(type_attrs)}")
print(f"Data types: {[a for a in type_attrs if 'Type' in a][:5]}")

# Check functions module
func_attrs = [a for a in dir(functions) if not a.startswith('_')]
print(f"\nFunctions module attributes: {len(func_attrs)}")
print(f"Sample functions: {func_attrs[:8]}")

# Common data types
common_types = ['IntegerType', 'StringType', 'FloatType', 'BooleanType']
for dt in common_types:
    available = hasattr(types, dt)
    print(f"  {dt}: {available}")

print("\nExample 2 completed!")
```

## Example 3: Row and Schema Classes

```python
# Example 3: Row and Schema Classes
from snowflake.snowpark import Row
from snowflake.snowpark.types import StructType, StructField, IntegerType, StringType

print("=== Row and Schema Classes ===")

# Create Row objects
row1 = Row(name="Alice", age=30)
row2 = Row(name="Bob", age=25)

print(f"Row 1: {row1}")
print(f"Row 2: {row2}")
print(f"Access field: {row1.name}, {row1.age}")

# Define schema
schema = StructType([
    StructField("id", IntegerType()),
    StructField("name", StringType()),
    StructField("value", IntegerType())
])

print(f"\nSchema fields: {len(schema.fields)}")
for field in schema.fields:
    print(f"  {field.name}: {field.datatype}")

print("\nExample 3 completed!")
```

## Example 4: Functions Module Exploration

```python
# Example 4: Functions Module Exploration
from snowflake.snowpark import functions as F

print("=== Functions Module Exploration ===")

# Aggregate functions
agg_funcs = ['avg', 'count', 'max', 'min', 'sum']
print("Aggregate functions:")
for func_name in agg_funcs:
    available = hasattr(F, func_name)
    print(f"  {func_name}: {available}")

# String functions
str_funcs = ['upper', 'lower', 'trim', 'concat', 'length']
print("\nString functions:")
for func_name in str_funcs:
    available = hasattr(F, func_name)
    print(f"  {func_name}: {available}")

# Column operations
col_funcs = ['col', 'lit', 'when']
print("\nColumn functions:")
for func_name in col_funcs:
    available = hasattr(F, func_name)
    print(f"  {func_name}: {available}")

print("\nExample 4 completed!")
```

## Example 5: Session Builder Pattern

```python
# Example 5: Session Builder Pattern
from snowflake.snowpark import Session

print("=== Session Builder Pattern ===")

# Mock session configuration
class MockSessionBuilder:
    def __init__(self):
        self.config = {}
    
    def configs(self, params):
        self.config.update(params)
        print(f"  Configured {len(params)} parameters")
        return self
    
    def create(self):
        print(f"  Session created with config")
        return "MockSession"

# Demonstrate builder pattern
builder = MockSessionBuilder()

params = {
    'account': 'myaccount',
    'user': 'myuser',
    'warehouse': 'COMPUTE_WH'
}

print("Building session:")
session = builder.configs(params).create()
print(f"Result: {session}")

print(f"\nActual Session.builder available: {hasattr(Session, 'builder')}")
print("Example 5 completed!")
```
