## Example 1: Basic Import and URL Builder

```python
# Example 1: Basic Import and URL Builder
from sqlalchemy import create_engine
from snowflake.sqlalchemy import URL

print("=== Basic Import and URL Builder ===")

# Create URL string
url = URL(
    account='myaccount',
    user='myuser',
    password='mypass',
    database='mydb',
    schema='public',
    warehouse='compute_wh'
)

print(f"URL created: {type(url).__name__}")
print(f"URL string: {str(url)[:50]}...")

# Alternative string format
url_str = "snowflake://user:pass@account/db/schema?warehouse=wh"
print(f"\nAlternative format: {url_str[:50]}...")

# Check module attributes
print(f"\nURL function available: {callable(URL)}")

print("Example 1 completed!")
```

## Example 2: URL Configuration Options

```python
# Example 2: URL Configuration Options
from snowflake.sqlalchemy import URL

print("=== URL Configuration Options ===")

# Different URL configurations
url1 = URL(account='acc1', user='user1', password='pass1')
url2 = URL(account='acc2', user='user2', password='pass2', database='db')
url3 = URL(
    account='acc3',
    user='user3',
    password='pass3',
    database='mydb',
    schema='public',
    warehouse='wh',
    role='analyst'
)

configs = [("basic", url1), ("with_db", url2), ("full", url3)]

for name, url in configs:
    print(f"\n{name}: {str(url)[:40]}...")

print("\nAll URLs created successfully")
print("Example 2 completed!")
```

## Example 3: SQLAlchemy Integration

```python
# Example 3: SQLAlchemy Integration
from sqlalchemy import create_engine, MetaData, Table, Column, Integer, String
from snowflake.sqlalchemy import URL
import snowflake.sqlalchemy

print("=== SQLAlchemy Integration ===")

# Check available dialect
attrs = [a for a in dir(snowflake.sqlalchemy) if not a.startswith('_')]
print(f"Module attributes: {len(attrs)}")
print(f"Key items: {attrs[:5]}")

# Table definition example
metadata = MetaData()
users_table = Table('users', metadata,
    Column('id', Integer, primary_key=True),
    Column('name', String(50)),
    Column('email', String(100))
)

print(f"\nTable defined: {users_table.name}")
print(f"Columns: {[c.name for c in users_table.columns]}")
print(f"Primary key: {[c.name for c in users_table.primary_key]}")

print("\nExample 3 completed!")
```

## Example 4: Snowflake Data Types

```python
# Example 4: Snowflake Data Types
from snowflake.sqlalchemy import VARIANT, ARRAY, OBJECT, TIMESTAMP_LTZ
from sqlalchemy import Integer, String, Float

print("=== Snowflake Data Types ===")

# Snowflake-specific types
snowflake_types = [
    ('VARIANT', VARIANT),
    ('ARRAY', ARRAY),
    ('OBJECT', OBJECT),
    ('TIMESTAMP_LTZ', TIMESTAMP_LTZ)
]

print("Snowflake-specific types:")
for name, type_class in snowflake_types:
    instance = type_class()
    print(f"  {name}: {type(instance).__name__}")

# Standard SQL types
standard_types = [Integer(), String(100), Float()]
print(f"\nStandard SQL types: {[type(t).__name__ for t in standard_types]}")

print("\nExample 4 completed!")
```

## Example 5: Engine Creation Pattern

```python
# Example 5: Engine Creation Pattern
from sqlalchemy import create_engine
from snowflake.sqlalchemy import URL
import urllib.parse

print("=== Engine Creation Pattern ===")

# URL with special characters in password
password_with_special = "p@ss!word#123"
encoded_password = urllib.parse.quote(password_with_special)

print(f"Original password: {password_with_special}")
print(f"Encoded password: {encoded_password}")

# Create URL with encoded password
url = URL(
    account='myaccount',
    user='myuser',
    password=encoded_password,
    database='mydb'
)

print(f"\nURL created: {type(url).__name__}")
print(f"URL string: {str(url)[:50]}...")

print("\nExample 5 completed!")
```
