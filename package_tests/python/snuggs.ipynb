## Example 1: snuggs functionality test 1

```python
# Example 1: Basic S-Expression Evaluation
import snuggs
import numpy as np

print("=== Basic S-Expression Evaluation ===")

# Simple arithmetic with asarray
result1 = snuggs.eval("(+ (asarray 1 2 3) (asarray 4 5 6))")
print(f"Addition: (+ [1,2,3] [4,5,6]) = {result1}")

result2 = snuggs.eval("(* 2 (asarray 1 2 3))")
print(f"Multiplication: (* 2 [1,2,3]) = {result2}")

result3 = snuggs.eval("(- (asarray 10 20 30) (asarray 1 2 3))")
print(f"Subtraction: (- [10,20,30] [1,2,3]) = {result3}")

result4 = snuggs.eval("(/ (asarray 10 20 30) 2)")
print(f"Division: (/ [10,20,30] 2) = {result4}")

# Nested expressions
result5 = snuggs.eval("(+ (* 2 (asarray 1 2)) (asarray 3 4))")
print(f"Nested: (+ (* 2 [1,2]) [3,4]) = {result5}")

print("Example 1 completed!")
```

## Example 2: snuggs functionality test 2

```python
# Example 2: Using Context Variables
import snuggs
import numpy as np

print("=== Using Context Variables ===")

# Define arrays as context variables
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

result1 = snuggs.eval("(+ a b)", a=a, b=b)
print(f"(+ a b) with a={a}, b={b}")
print(f"Result: {result1}")

result2 = snuggs.eval("(* a 5)", a=a)
print(f"\n(* a 5) with a={a}")
print(f"Result: {result2}")

# Complex expression with multiple variables
c = np.array([2, 2, 2])
result3 = snuggs.eval("(+ (* a b) c)", a=a, b=b, c=c)
print(f"\n(+ (* a b) c) with a={a}, b={b}, c={c}")
print(f"Result: {result3}")

print("\nExample 2 completed!")
```

## Example 3: snuggs functionality test 3

```python
# Example 3: Comparison and Logical Operations
import snuggs
import numpy as np

print("=== Comparison and Logical Operations ===")

# Comparison operations
arr1 = np.array([1, 5, 10])
arr2 = np.array([2, 4, 8])

result1 = snuggs.eval("(> a b)", a=arr1, b=arr2)
print(f"(> {arr1} {arr2}) = {result1}")

result2 = snuggs.eval("(< a b)", a=arr1, b=arr2)
print(f"(< {arr1} {arr2}) = {result2}")

result3 = snuggs.eval("(== a (asarray 1 5 10))", a=arr1)
print(f"(== {arr1} [1,5,10]) = {result3}")

# Logical operations
result4 = snuggs.eval("(& (> a 0) (< a 10))", a=arr1)
print(f"\n(& (> {arr1} 0) (< {arr1} 10)) = {result4}")

print("\nExample 3 completed!")
```

## Example 4: snuggs functionality test 4

```python
# Example 4: Higher-Order Functions
import snuggs
import numpy as np

print("=== Higher-Order Functions ===")

# Using map with partial
result1 = snuggs.eval("(asarray (map (partial * 2) (asarray 1 2 3 4)))")
print(f"Map (partial * 2) over [1,2,3,4]: {result1}")

result2 = snuggs.eval("(asarray (map (partial + 10) (asarray 5 10 15)))")
print(f"Map (partial + 10) over [5,10,15]: {result2}")

# Combining map with other operations
arr = np.array([1, 2, 3])
result3 = snuggs.eval("(+ (asarray (map (partial * 2) a)) (asarray 1 1 1))", a=arr)
print(f"\nMap (* 2) over {arr} then add [1,1,1]: {result3}")

# Nested map operations
result4 = snuggs.eval("(asarray (map (partial / 2) (asarray 10 20 30)))")
print(f"Map (partial / 2) over [10,20,30]: {result4}")

print("\nExample 4 completed!")
```

## Example 5: snuggs functionality test 5

```python
# Example 5: Complex Mathematical Expressions
import snuggs
import numpy as np

print("=== Complex Mathematical Expressions ===")

# Polynomial-like expressions
x = np.array([1, 2, 3, 4])
result1 = snuggs.eval("(+ (* x x) (* 2 x) 1)", x=x)
print(f"x^2 + 2x + 1 where x={x}")
print(f"Result: {result1}")

# Weighted average calculation
a = np.array([10, 20, 30])
b = np.array([5, 10, 15])
result2 = snuggs.eval("(/ (+ (* a 0.7) (* b 0.3)) 1)", a=a, b=b)
print(f"\nWeighted avg: 0.7*{a} + 0.3*{b}")
print(f"Result: {result2}")

# Temperature conversion (C to F)
celsius = np.array([0, 10, 20, 30])
fahrenheit = snuggs.eval("(+ (* c (/ 9 5)) 32)", c=celsius)
print(f"\nCelsius {celsius} to Fahrenheit: {fahrenheit}")

print("\nExample 5 completed!")
```
