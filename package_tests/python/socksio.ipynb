## Example 1: socksio functionality test 1

```python
# Example 1: SOCKS4 Connection and Request
import socksio

print("=== SOCKS4 Connection and Request ===")

# Create SOCKS4 connection
conn = socksio.SOCKS4Connection(user_id=b"testuser")
print(f"Connection created: {type(conn).__name__}")

# Create a CONNECT request
request = socksio.SOCKS4Request.from_address(
    socksio.SOCKS4Command.CONNECT,
    ("93.184.216.34", 80)  # example.com IP
)

print(f"\nRequest type: {type(request).__name__}")
print(f"Command: {request.command}")
print(f"Port: {request.port}")
print(f"Address: {request.addr}")

# Send request and get data
conn.send(request)
data = conn.data_to_send()
print(f"\nData to send: {len(data)} bytes")
print(f"First 10 bytes: {data[:10]}")

print("Example 1 completed!")
```

## Example 2: socksio functionality test 2

```python
# Example 2: SOCKS4 with Multiple Requests
import socksio

print("=== SOCKS4 with Multiple Requests ===")

# Create connection
conn = socksio.SOCKS4Connection(user_id=b"myapp")

# Create multiple requests with different IPs
requests = [
    ("93.184.216.34", 80),   # example.com
    ("192.0.2.1", 443),      # test IP
    ("198.51.100.1", 8080)   # another test IP
]

for addr, port in requests:
    request = socksio.SOCKS4Request.from_address(
        socksio.SOCKS4Command.CONNECT,
        (addr, port)
    )
    print(f"\nRequest to {addr}:{port}")
    print(f"  Command: {request.command}")
    print(f"  Address: {request.addr}")
    print(f"  Port: {request.port}")

print("\nExample 2 completed!")
```

## Example 3: socksio functionality test 3

```python
# Example 3: SOCKS4 Reply Handling
import socksio

print("=== SOCKS4 Reply Handling ===")

# Create connection
conn = socksio.SOCKS4Connection(user_id=b"testapp")

# Create request
request = socksio.SOCKS4Request.from_address(
    socksio.SOCKS4Command.CONNECT,
    ("192.0.2.1", 80)
)

conn.send(request)
data = conn.data_to_send()
print(f"Sent request: {len(data)} bytes")
print(f"Request data: {data}")

# Simulate successful reply (code 90 = request granted)
reply_data = b'\x00\x5a\x00\x50\xc0\x00\x02\x01'
conn.receive_data(reply_data)

print(f"\nReceived reply: {len(reply_data)} bytes")
print(f"Reply data: {reply_data}")

print("\nExample 3 completed!")
```

## Example 4: socksio functionality test 4

```python
# Example 4: SOCKS4 Commands
import socksio

print("=== SOCKS4 Commands ===")

# Available SOCKS4 commands
print("SOCKS4 Commands:")
print(f"  CONNECT: {socksio.SOCKS4Command.CONNECT}")
print(f"  BIND: {socksio.SOCKS4Command.BIND}")

# Create BIND request (different from CONNECT)
conn = socksio.SOCKS4Connection(user_id=b"bindtest")

bind_request = socksio.SOCKS4Request.from_address(
    socksio.SOCKS4Command.BIND,
    ("192.0.2.1", 9000)
)

print(f"\nBIND request created:")
print(f"  Command: {bind_request.command}")
print(f"  Port: {bind_request.port}")

conn.send(bind_request)
data = conn.data_to_send()
print(f"\nData to send: {len(data)} bytes")

print("\nExample 4 completed!")
```

## Example 5: socksio functionality test 5

```python
# Example 5: Connection State and Data Flow
import socksio

print("=== Connection State and Data Flow ===")

# Create connection
conn = socksio.SOCKS4Connection(user_id=b"statetest")
print(f"Connection created: {type(conn).__name__}")

# Create and send request
request = socksio.SOCKS4Request.from_address(
    socksio.SOCKS4Command.CONNECT,
    ("198.51.100.1", 443)
)

print(f"\nSending request to 198.51.100.1:443")
conn.send(request)
outgoing_data = conn.data_to_send()

print(f"Data to send: {len(outgoing_data)} bytes")

# Simulate reply
reply_data = b'\x00\x5a\x01\xbb\xc6\x33\x64\x01'
conn.receive_data(reply_data)

print(f"\nReceived reply: {len(reply_data)} bytes")
print(f"Connection complete")

print("\nExample 5 completed!")
```
