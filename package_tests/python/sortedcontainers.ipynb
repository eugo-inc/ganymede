## Example 1: sortedcontainers functionality test 1

```python
from sortedcontainers import SortedList, SortedSet, SortedDict
import random

# Example 1: SortedList - Maintaining sorted order with fast operations
print("=== Example 1: SortedList Operations ===")

# Create and populate SortedList
sorted_list = SortedList()
print(f"Created empty SortedList: {sorted_list}")

# Add elements - they automatically stay sorted
elements = [15, 7, 23, 1, 45, 12, 3, 67, 9]
for element in elements:
    sorted_list.add(element)

print(f"After adding {elements}: {sorted_list}")
print(f"List length: {len(sorted_list)}")

# Demonstrate various operations
print(f"\nSortedList operations:")
print(f"  First element: {sorted_list[0]}")
print(f"  Last element: {sorted_list[-1]}")
print(f"  Element at index 3: {sorted_list[3]}")

# Binary search operations
print(f"  Index of 12: {sorted_list.index(12)}")
print(f"  Count of 7: {sorted_list.count(7)}")

# Range operations
print(f"  Elements < 15: {sorted_list[:sorted_list.bisect_left(15)]}")
print(f"  Elements >= 15: {sorted_list[sorted_list.bisect_left(15):]}")

# Remove operations
sorted_list.remove(23)
print(f"After removing 23: {sorted_list}")

# Bulk operations
more_elements = [2, 8, 14, 20]
sorted_list.update(more_elements)
print(f"After bulk update with {more_elements}: {sorted_list}")

# Performance comparison with regular list
print(f"\nPerformance comparison with regular list:")
regular_list = list(sorted_list)
print(f"  SortedList: O(log n) insertions, automatic sorting")
print(f"  Regular list: O(n) to maintain sort, O(n log n) to sort after")
print(f"  SortedList is ideal for frequent insertions with need for sorted order")

print("SortedList operations completed successfully")
```

## Example 2: sortedcontainers functionality test 2

```python
from sortedcontainers import SortedSet
from collections import Counter

# Example 2: SortedSet - Unique elements in sorted order
print("=== Example 2: SortedSet for Unique Sorted Collections ===")

# Create SortedSet with duplicate handling
data = [5, 2, 8, 2, 1, 9, 5, 3, 7, 1, 8]
sorted_set = SortedSet(data)

print(f"Original data: {data}")
print(f"SortedSet (unique, sorted): {sorted_set}")
print(f"Length: {len(sorted_set)} (vs {len(data)} original)")

# Set operations with automatic sorting
print(f"\nSet operations:")
set_a = SortedSet([1, 2, 3, 4, 5])
set_b = SortedSet([4, 5, 6, 7, 8])

print(f"Set A: {set_a}")
print(f"Set B: {set_b}")
print(f"Union A | B: {set_a | set_b}")
print(f"Intersection A & B: {set_a & set_b}")
print(f"Difference A - B: {set_a - set_b}")
print(f"Symmetric diff A ^ B: {set_a ^ set_b}")

# Range queries on SortedSet
print(f"\nRange queries:")
numbers = SortedSet(range(0, 100, 7))  # 0, 7, 14, 21, ...
print(f"Multiples of 7 < 100: {list(numbers)[:10]}...")

# Find elements in range [20, 50]
start_idx = numbers.bisect_left(20)
end_idx = numbers.bisect_right(50)
range_elements = numbers[start_idx:end_idx]
print(f"Elements in range [20, 50]: {list(range_elements)}")

# Use case: tracking unique visitors with timestamps
print(f"\nUse case: Unique visitor tracking")
visitor_timestamps = SortedSet()

# Simulate visitor arrivals (timestamps)
import time
base_time = int(time.time())
visitor_events = [
    base_time + i * 60 + random.randint(0, 30)  # visitors every ~1 minute
    for i in range(20)
]

# Add some duplicate timestamps (same visitor refresh)
visitor_events.extend([base_time + 120, base_time + 360])  # duplicates

for timestamp in visitor_events:
    visitor_timestamps.add(timestamp)

print(f"Total visit events: {len(visitor_events)}")
print(f"Unique visitors: {len(visitor_timestamps)}")
print(f"First 5 unique visit times: {list(visitor_timestamps)[:5]}")

# Find visitors in specific time window (5 minutes)
window_start = base_time + 300  # 5 minutes after start
window_end = window_start + 300  # 5-minute window
window_visitors = visitor_timestamps[
    visitor_timestamps.bisect_left(window_start):
    visitor_timestamps.bisect_right(window_end)
]
print(f"Visitors in 5-min window: {len(window_visitors)}")

print("SortedSet operations completed successfully")
```

## Example 3: sortedcontainers functionality test 3

```python
from sortedcontainers import SortedDict
from datetime import datetime, timedelta

# Example 3: SortedDict - Key-value mapping with sorted keys
print("=== Example 3: SortedDict for Ordered Key-Value Storage ===")

# Create SortedDict for time-series data
time_series = SortedDict()

# Add data points (timestamp -> value)
base_date = datetime(2024, 1, 1)
for i in range(10):
    timestamp = base_date + timedelta(hours=i * 6)
    value = 100 + random.randint(-20, 20)
    time_series[timestamp] = value

print(f"Time series data:")
for timestamp, value in list(time_series.items())[:5]:
    print(f"  {timestamp}: {value}")
print(f"  ... ({len(time_series)} total entries)")

# Range queries by key
print(f"\nRange queries:")
start_time = base_date + timedelta(hours=12)
end_time = base_date + timedelta(hours=36)

# Get all data in time range
start_idx = time_series.bisect_left(start_time)
end_idx = time_series.bisect_right(end_time)
range_data = time_series.items()[start_idx:end_idx]

print(f"Data between {start_time} and {end_time}:")
for timestamp, value in range_data:
    print(f"  {timestamp}: {value}")

# Use case: Configuration management with ordered keys
print(f"\nUse case: Configuration management")
config = SortedDict()

# Add configuration items (sorted by key)
config.update({
    'api.timeout': 30,
    'api.retries': 3,
    'api.base_url': 'https://api.example.com',
    'database.host': 'localhost',
    'database.port': 5432,
    'database.name': 'myapp',
    'cache.ttl': 3600,
    'cache.max_size': 1000,
    'logging.level': 'INFO',
    'logging.file': '/var/log/myapp.log'
})

print(f"Configuration (sorted by key):")
for key, value in config.items():
    print(f"  {key}: {value}")

# Get configuration sections
print(f"\nConfiguration sections:")
api_keys = list(config.irange('api.', 'api.z'))  # keys starting with 'api.'
db_keys = list(config.irange('database.', 'database.z'))

print(f"API settings: {[config[k] for k in api_keys]}")
print(f"Database settings: {[config[k] for k in db_keys]}")

# Performance characteristics
print(f"\nSortedDict characteristics:")
print(f"  Key access: O(log n)")  
print(f"  Key iteration: O(n) in sorted order")
print(f"  Range queries: O(log n + k) where k is result size")
print(f"  Insertion/deletion: O(log n)")

# Comparison with regular dict
regular_dict = dict(config)
sorted_keys_regular = sorted(regular_dict.keys())

print(f"\nComparison:")
print(f"  SortedDict: {list(config.keys())[:3]}... (always sorted)")
print(f"  Regular dict sorted: {sorted_keys_regular[:3]}... (requires sorting)")
print(f"  SortedDict maintains order automatically")

print("SortedDict operations completed successfully")
```

## Example 4: sortedcontainers functionality test 4

```python
from sortedcontainers import SortedList, SortedSet, SortedDict
import time

# Example 4: Performance benchmarking and comparison
print("=== Example 4: Performance Benchmarking ===")

def benchmark_sorted_containers():
    """Benchmark sorted containers vs standard library equivalents"""
    
    # Test data
    n = 1000
    data = [random.randint(1, 1000) for _ in range(n)]
    
    results = {}
    
    # Benchmark SortedList vs list + sort
    print("Benchmarking SortedList vs regular list:")
    
    # SortedList insertion
    start = time.perf_counter()
    sorted_list = SortedList()
    for item in data:
        sorted_list.add(item)
    sorted_list_time = time.perf_counter() - start
    results['SortedList insertion'] = sorted_list_time
    
    # Regular list + sort
    start = time.perf_counter()
    regular_list = []
    for item in data:
        regular_list.append(item)
        regular_list.sort()  # Inefficient: sorting after each insert
    regular_list_time = time.perf_counter() - start
    results['Regular list + sort'] = regular_list_time
    
    print(f"  SortedList insertion: {sorted_list_time:.4f}s")
    print(f"  Regular list + sort: {regular_list_time:.4f}s")
    print(f"  SortedList is {regular_list_time/sorted_list_time:.1f}x faster")
    
    # Benchmark SortedSet vs set + sorted
    print(f"\nBenchmarking SortedSet vs set + sorted:")
    
    start = time.perf_counter()
    sorted_set = SortedSet(data)
    sorted_set_list = list(sorted_set)  # Already sorted
    sorted_set_time = time.perf_counter() - start
    
    start = time.perf_counter()
    regular_set = set(data)
    regular_set_sorted = sorted(regular_set)
    regular_set_time = time.perf_counter() - start
    
    print(f"  SortedSet creation + list: {sorted_set_time:.4f}s")
    print(f"  Regular set + sorted: {regular_set_time:.4f}s")
    print(f"  Speedup: {regular_set_time/sorted_set_time:.1f}x")
    
    return results

# Run benchmarks
benchmark_results = benchmark_sorted_containers()

# Use case: Event processing system
print(f"\nUse case: Real-time event processing system")

class EventProcessor:
    """Process events in chronological order using SortedDict"""
    
    def __init__(self):
        self.events = SortedDict()  # timestamp -> event_data
        self.processed = SortedSet()  # processed event IDs
        
    def add_event(self, timestamp, event_id, event_data):
        """Add event to processing queue"""
        self.events[timestamp] = {
            'id': event_id,
            'data': event_data,
            'added_at': time.time()
        }
        
    def process_events_until(self, cutoff_time):
        """Process all events up to cutoff_time"""
        processed_count = 0
        
        # Process events in chronological order
        cutoff_idx = self.events.bisect_right(cutoff_time)
        events_to_process = list(self.events.items())[:cutoff_idx]
        
        for timestamp, event in events_to_process:
            event_id = event['id']
            if event_id not in self.processed:
                # Simulate processing
                self.processed.add(event_id)
                processed_count += 1
                
        # Remove processed events
        for timestamp, _ in events_to_process:
            del self.events[timestamp]
            
        return processed_count
    
    def get_stats(self):
        return {
            'pending_events': len(self.events),
            'processed_events': len(self.processed),
            'oldest_pending': min(self.events.keys()) if self.events else None,
            'newest_pending': max(self.events.keys()) if self.events else None
        }

# Demonstrate event processing
processor = EventProcessor()

# Add events with random timestamps
base_time = time.time()
for i in range(50):
    event_time = base_time - random.randint(0, 3600)  # Events from last hour
    processor.add_event(event_time, f"event_{i}", {"type": "user_action", "value": i})

print(f"Added 50 events")
stats = processor.get_stats()
print(f"Stats: {stats['pending_events']} pending, {stats['processed_events']} processed")

# Process events older than 30 minutes
cutoff = base_time - 1800  # 30 minutes ago
processed = processor.process_events_until(cutoff)
print(f"Processed {processed} events older than 30 minutes")

# Final stats
final_stats = processor.get_stats()
print(f"Final stats: {final_stats['pending_events']} pending, {final_stats['processed_events']} processed")

print("Performance benchmarking completed successfully")
```

## Example 5: sortedcontainers functionality test 5

```python
from sortedcontainers import SortedList, SortedSet, SortedDict
import json

# Example 5: Production data structures and advanced patterns
print("=== Example 5: Production Data Structures & Advanced Patterns ===")

class PriorityTaskQueue:
    """Production-ready task queue with priority and ordering"""
    
    def __init__(self):
        # Use SortedList with custom key function for priority + timestamp
        self.tasks = SortedList(key=lambda task: (task['priority'], task['created_at']))
        self.task_counter = 0
        
    def add_task(self, name, priority=0, data=None):
        """Add task with priority (lower number = higher priority)"""
        import time
        task = {
            'id': self.task_counter,
            'name': name,
            'priority': priority,
            'created_at': time.time(),
            'data': data or {}
        }
        self.tasks.add(task)
        self.task_counter += 1
        return task['id']
    
    def get_next_task(self):
        """Get highest priority task (and remove from queue)"""
        if self.tasks:
            return self.tasks.pop(0)
        return None
    
    def peek_next_task(self):
        """Look at next task without removing it"""
        return self.tasks[0] if self.tasks else None
    
    def get_tasks_by_priority(self, max_priority):
        """Get all tasks up to specified priority level"""
        # Use binary search to find tasks efficiently
        dummy_task = {'priority': max_priority + 1, 'created_at': float('inf')}
        cutoff_idx = self.tasks.bisect_left(dummy_task)
        return list(self.tasks[:cutoff_idx])
    
    def get_stats(self):
        if not self.tasks:
            return {'total': 0, 'by_priority': {}}
        
        priority_counts = {}
        for task in self.tasks:
            p = task['priority']
            priority_counts[p] = priority_counts.get(p, 0) + 1
            
        return {
            'total': len(self.tasks),
            'by_priority': priority_counts,
            'next_task': self.peek_next_task()['name'] if self.tasks else None
        }

# Demonstrate priority queue
print("Priority Task Queue demonstration:")
queue = PriorityTaskQueue()

# Add tasks with different priorities
tasks_to_add = [
    ("Send email", 2, {"recipient": "user@example.com"}),
    ("Process payment", 0, {"amount": 100.50}),  # High priority
    ("Generate report", 3, {"report_type": "monthly"}),
    ("Backup database", 1, {"database": "production"}),
    ("Send notification", 2, {"message": "Task completed"}),
    ("Update cache", 1, {"cache_key": "user_sessions"}),
]

for name, priority, data in tasks_to_add:
    task_id = queue.add_task(name, priority, data)
    print(f"Added task {task_id}: '{name}' (priority {priority})")

print(f"\nQueue stats: {queue.get_stats()}")

# Process high-priority tasks first
print(f"\nProcessing tasks in priority order:")
processed_count = 0
while queue.tasks and processed_count < 4:
    task = queue.get_next_task()
    print(f"Processing: '{task['name']}' (priority {task['priority']})")
    processed_count += 1

print(f"Remaining tasks: {queue.get_stats()['total']}")

# Advanced pattern: Multi-index data structure
print(f"\nAdvanced pattern: Multi-index customer database")

class CustomerDatabase:
    """Customer database with multiple sorted indices"""
    
    def __init__(self):
        self.customers_by_id = SortedDict()  # id -> customer
        self.customers_by_name = SortedDict()  # name -> id
        self.customers_by_signup = SortedDict()  # signup_date -> id
        self.customers_by_score = SortedList(key=lambda x: -x[1])  # (id, score) sorted by score desc
    
    def add_customer(self, customer_id, name, signup_date, credit_score):
        """Add customer with multiple indices"""
        customer = {
            'id': customer_id,
            'name': name,
            'signup_date': signup_date,
            'credit_score': credit_score
        }
        
        # Update all indices
        self.customers_by_id[customer_id] = customer
        self.customers_by_name[name] = customer_id
        self.customers_by_signup[signup_date] = customer_id
        self.customers_by_score.add((customer_id, credit_score))
        
    def get_customers_by_name_range(self, start_name, end_name):
        """Get customers with names in alphabetical range"""
        start_idx = self.customers_by_name.bisect_left(start_name)
        end_idx = self.customers_by_name.bisect_right(end_name)
        name_range = list(self.customers_by_name.items())[start_idx:end_idx]
        
        return [self.customers_by_id[customer_id] for name, customer_id in name_range]
    
    def get_top_customers_by_score(self, n=10):
        """Get top N customers by credit score"""
        top_entries = self.customers_by_score[:n]
        return [self.customers_by_id[customer_id] for customer_id, score in top_entries]
    
    def get_customers_since_date(self, since_date):
        """Get customers who signed up since given date"""
        since_idx = self.customers_by_signup.bisect_left(since_date)
        recent_signups = list(self.customers_by_signup.items())[since_idx:]
        return [self.customers_by_id[customer_id] for signup_date, customer_id in recent_signups]

# Demonstrate customer database
from datetime import datetime, timedelta
import random

db = CustomerDatabase()

# Add sample customers
base_date = datetime(2024, 1, 1)
names = ["Alice Johnson", "Bob Smith", "Carol Davis", "David Wilson", "Eva Brown", 
         "Frank Miller", "Grace Lee", "Henry Taylor", "Ivy Chen", "Jack Anderson"]

for i, name in enumerate(names):
    signup_date = base_date + timedelta(days=i * 10)
    credit_score = random.randint(600, 850)
    db.add_customer(f"cust_{i:03d}", name, signup_date, credit_score)

print(f"Added {len(names)} customers to database")

# Query examples
print(f"\nQuery examples:")
print(f"Customers with names A-E:")
a_to_e = db.get_customers_by_name_range("A", "E")
for customer in a_to_e:
    print(f"  {customer['name']} (score: {customer['credit_score']})")

print(f"\nTop 3 customers by credit score:")
top_customers = db.get_top_customers_by_score(3)
for customer in top_customers:
    print(f"  {customer['name']}: {customer['credit_score']}")

since_date = base_date + timedelta(days=50)
print(f"\nCustomers who signed up since {since_date.date()}:")
recent_customers = db.get_customers_since_date(since_date)
for customer in recent_customers:
    print(f"  {customer['name']} on {customer['signup_date'].date()}")

print("Production data structures and advanced patterns completed successfully")
```
