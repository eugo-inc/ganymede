## Example 1: soupsieve functionality test 1

```python
import soupsieve as sv
from bs4 import BeautifulSoup

# Example 1: Basic CSS selector usage with HTML parsing
print("=== Example 1: Basic CSS Selectors ===")

# Create sample HTML document
html_content = """
<!DOCTYPE html>
<html>
<head>
    <title>Sample Document</title>
</head>
<body>
    <div class="container">
        <h1 id="main-title">Welcome</h1>
        <p class="intro">This is an introduction paragraph.</p>
        <div class="content">
            <h2>Section 1</h2>
            <p class="highlight">Important information here.</p>
            <ul class="list">
                <li>Item 1</li>
                <li class="special">Item 2 (special)</li>
                <li>Item 3</li>
            </ul>
        </div>
        <div class="sidebar">
            <h3>Sidebar</h3>
            <p>Additional content</p>
        </div>
    </div>
</body>
</html>
"""

# Parse HTML with BeautifulSoup
soup = BeautifulSoup(html_content, 'html.parser')
print("HTML document parsed successfully")

# Basic CSS selectors
print(f"\nBasic selector examples:")

# Select by tag
titles = sv.select('h1, h2, h3', soup)
print(f"All headings (h1, h2, h3): {len(titles)} found")
for title in titles:
    print(f"  - {title.name}: {title.get_text()}")

# Select by class
highlighted = sv.select('.highlight', soup)
print(f"\nElements with class 'highlight': {len(highlighted)} found")
for elem in highlighted:
    print(f"  - {elem.get_text()}")

# Select by ID
main_title = sv.select('#main-title', soup)
print(f"\nElement with ID 'main-title': {main_title[0].get_text() if main_title else 'None'}")

# Select by attribute
special_items = sv.select('li.special', soup)
print(f"\nLi elements with class 'special': {len(special_items)} found")
for item in special_items:
    print(f"  - {item.get_text()}")

# Descendant selectors
container_paragraphs = sv.select('div.container p', soup)
print(f"\nParagraphs inside .container: {len(container_paragraphs)} found")
for p in container_paragraphs:
    print(f"  - {p.get_text()[:30]}...")

print("Basic CSS selector operations completed successfully")
```

## Example 3: soupsieve functionality test 3

```python
import soupsieve as sv
from bs4 import BeautifulSoup

# Example 3: Web scraping with CSS selectors
print("=== Example 3: Web Scraping with CSS Selectors ===")

# Sample e-commerce product page HTML
ecommerce_html = """
<html>
<body>
    <div class="product-page">
        <div class="breadcrumb">
            <a href="/">Home</a> > <a href="/electronics">Electronics</a> > <span>Laptop</span>
        </div>
        
        <div class="product-main">
            <div class="product-images">
                <img src="laptop1.jpg" alt="Laptop main image" class="main-image">
                <div class="thumbnails">
                    <img src="thumb1.jpg" alt="Thumbnail 1">
                    <img src="thumb2.jpg" alt="Thumbnail 2">
                </div>
            </div>
            
            <div class="product-info">
                <h1 class="product-title">Professional Laptop XYZ</h1>
                <div class="pricing">
                    <span class="price current">$1,299.99</span>
                    <span class="price original">$1,599.99</span>
                    <span class="discount">19% off</span>
                </div>
                
                <div class="specifications">
                    <h3>Specifications</h3>
                    <ul class="spec-list">
                        <li><span class="spec-name">Processor:</span> Intel i7</li>
                        <li><span class="spec-name">RAM:</span> 16GB</li>
                        <li><span class="spec-name">Storage:</span> 512GB SSD</li>
                        <li><span class="spec-name">Display:</span> 15.6" FHD</li>
                    </ul>
                </div>
                
                <div class="rating">
                    <div class="stars" data-rating="4.5">
                        <span class="star filled">★</span>
                        <span class="star filled">★</span>
                        <span class="star filled">★</span>
                        <span class="star filled">★</span>
                        <span class="star half">☆</span>
                    </div>
                    <span class="rating-text">(234 reviews)</span>
                </div>
            </div>
        </div>
        
        <div class="reviews-section">
            <h2>Customer Reviews</h2>
            <div class="review" data-helpful="15">
                <div class="review-header">
                    <span class="reviewer">John D.</span>
                    <span class="review-rating">5 stars</span>
                </div>
                <p class="review-text">Excellent laptop for professional work!</p>
            </div>
            <div class="review" data-helpful="8">
                <div class="review-header">
                    <span class="reviewer">Sarah M.</span>
                    <span class="review-rating">4 stars</span>
                </div>
                <p class="review-text">Good performance, slightly heavy.</p>
            </div>
        </div>
    </div>
</body>
</html>
"""

soup_ecom = BeautifulSoup(ecommerce_html, 'html.parser')

class ProductScraper:
    """Scrape product information using CSS selectors"""
    
    def __init__(self, soup):
        self.soup = soup
        
    def extract_product_info(self):
        """Extract main product information"""
        info = {}
        
        # Product title
        title = sv.select_one('.product-title', self.soup)
        info['title'] = title.get_text().strip() if title else None
        
        # Current price
        current_price = sv.select_one('.pricing .price.current', self.soup)
        info['current_price'] = current_price.get_text().strip() if current_price else None
        
        # Original price
        original_price = sv.select_one('.pricing .price.original', self.soup)
        info['original_price'] = original_price.get_text().strip() if original_price else None
        
        # Discount
        discount = sv.select_one('.pricing .discount', self.soup)
        info['discount'] = discount.get_text().strip() if discount else None
        
        # Rating
        rating_elem = sv.select_one('.rating .stars', self.soup)
        info['rating'] = rating_elem.get('data-rating') if rating_elem else None
        
        rating_text = sv.select_one('.rating .rating-text', self.soup)
        info['review_count'] = rating_text.get_text().strip() if rating_text else None
        
        return info
    
    def extract_specifications(self):
        """Extract product specifications"""
        specs = {}
        
        spec_items = sv.select('.specifications .spec-list li', self.soup)
        for item in spec_items:
            name_elem = sv.select_one('.spec-name', item)
            if name_elem:
                name = name_elem.get_text().strip().rstrip(':')
                value = item.get_text().replace(name_elem.get_text(), '').strip()
                specs[name] = value
                
        return specs
    
    def extract_reviews(self):
        """Extract customer reviews"""
        reviews = []
        
        review_elements = sv.select('.reviews-section .review', self.soup)
        for review_elem in review_elements:
            review = {}
            
            # Reviewer name
            reviewer = sv.select_one('.reviewer', review_elem)
            review['reviewer'] = reviewer.get_text().strip() if reviewer else None
            
            # Rating
            rating = sv.select_one('.review-rating', review_elem)
            review['rating'] = rating.get_text().strip() if rating else None
            
            # Review text
            text = sv.select_one('.review-text', review_elem)
            review['text'] = text.get_text().strip() if text else None
            
            # Helpful votes
            review['helpful_votes'] = review_elem.get('data-helpful')
            
            reviews.append(review)
            
        return reviews
    
    def extract_breadcrumb(self):
        """Extract navigation breadcrumb"""
        breadcrumb = []
        
        # Get all links and text in breadcrumb
        breadcrumb_links = sv.select('.breadcrumb a', self.soup)
        breadcrumb_text = sv.select('.breadcrumb span', self.soup)
        
        for link in breadcrumb_links:
            breadcrumb.append({
                'text': link.get_text().strip(),
                'url': link.get('href')
            })
            
        for text in breadcrumb_text:
            breadcrumb.append({
                'text': text.get_text().strip(),
                'url': None
            })
            
        return breadcrumb

# Demonstrate scraping
scraper = ProductScraper(soup_ecom)

print("Scraping product information:")
product_info = scraper.extract_product_info()
for key, value in product_info.items():
    print(f"  {key}: {value}")

print(f"\nProduct specifications:")
specs = scraper.extract_specifications()
for spec_name, spec_value in specs.items():
    print(f"  {spec_name}: {spec_value}")

print(f"\nCustomer reviews ({len(scraper.extract_reviews())} total):")
reviews = scraper.extract_reviews()
for i, review in enumerate(reviews, 1):
    print(f"  Review {i}: {review['reviewer']} - {review['rating']}")
    print(f"    \"{review['text']}\" ({review['helpful_votes']} helpful)")

print(f"\nBreadcrumb navigation:")
breadcrumb = scraper.extract_breadcrumb()
breadcrumb_text = " > ".join([item['text'] for item in breadcrumb])
print(f"  {breadcrumb_text}")

print("Web scraping with CSS selectors completed successfully")
```

```python
import soupsieve as sv
from bs4 import BeautifulSoup
import re

# Example 4: Custom selectors and filtering
print("=== Example 4: Custom Selectors and Filtering ===")

# Sample forum/discussion HTML
forum_html = """
<div class="forum">
    <div class="thread" data-priority="high" data-views="1500">
        <div class="post" data-user="admin" data-score="25">
            <div class="post-header">
                <span class="username">admin</span>
                <span class="timestamp">2024-01-15 10:30</span>
                <span class="score">Score: 25</span>
            </div>
            <div class="post-content">
                <p>Welcome to the new forum! This is an important announcement.</p>
                <p>Please read the rules carefully.</p>
            </div>
            <div class="tags">
                <span class="tag important">important</span>
                <span class="tag announcement">announcement</span>
            </div>
        </div>
    </div>
    
    <div class="thread" data-priority="medium" data-views="350">
        <div class="post" data-user="user123" data-score="8">
            <div class="post-header">
                <span class="username">user123</span>
                <span class="timestamp">2024-01-14 15:22</span>
                <span class="score">Score: 8</span>
            </div>
            <div class="post-content">
                <p>Has anyone tried the new API endpoints? I'm getting errors.</p>
                <p>Error code: 500</p>
            </div>
            <div class="tags">
                <span class="tag help">help</span>
                <span class="tag api">api</span>
            </div>
        </div>
        <div class="post reply" data-user="developer" data-score="15">
            <div class="post-header">
                <span class="username">developer</span>
                <span class="timestamp">2024-01-14 16:45</span>
                <span class="score">Score: 15</span>
            </div>
            <div class="post-content">
                <p>The API endpoints are working. Check your authentication headers.</p>
            </div>
        </div>
    </div>
    
    <div class="thread" data-priority="low" data-views="50">
        <div class="post" data-user="newbie" data-score="2">
            <div class="post-header">
                <span class="username">newbie</span>
                <span class="timestamp">2024-01-13 09:15</span>
                <span class="score">Score: 2</span>
            </div>
            <div class="post-content">
                <p>Hello everyone! I'm new here.</p>
            </div>
            <div class="tags">
                <span class="tag introduction">introduction</span>
            </div>
        </div>
    </div>
</div>
"""

soup_forum = BeautifulSoup(forum_html, 'html.parser')

# Custom filtering functions
class ForumAnalyzer:
    """Analyze forum data using advanced CSS selectors"""
    
    def __init__(self, soup):
        self.soup = soup
        
    def get_high_score_posts(self, min_score=10):
        """Get posts with score above threshold"""
        posts = sv.select('.post', self.soup)
        high_score_posts = []
        
        for post in posts:
            score_attr = post.get('data-score')
            if score_attr and int(score_attr) >= min_score:
                high_score_posts.append(post)
                
        return high_score_posts
    
    def get_posts_by_user_type(self, user_types):
        """Get posts by specific user types"""
        selector = ', '.join([f'[data-user="{user_type}"]' for user_type in user_types])
        return sv.select(selector, self.soup)
    
    def get_threads_by_popularity(self, min_views=100):
        """Get threads with minimum view count"""
        threads = sv.select('.thread', self.soup)
        popular_threads = []
        
        for thread in threads:
            views = thread.get('data-views')
            if views and int(views) >= min_views:
                popular_threads.append(thread)
                
        return popular_threads
    
    def extract_posts_with_keywords(self, keywords):
        """Find posts containing specific keywords"""
        matching_posts = []
        posts = sv.select('.post', self.soup)
        
        for post in posts:
            content_elem = sv.select_one('.post-content', post)
            if content_elem:
                content_text = content_elem.get_text().lower()
                for keyword in keywords:
                    if keyword.lower() in content_text:
                        matching_posts.append({
                            'post': post,
                            'keyword': keyword,
                            'user': post.get('data-user'),
                            'content': content_text[:100] + '...'
                        })
                        break
        
        return matching_posts
    
    def get_tag_statistics(self):
        """Analyze tag usage across forum"""
        tag_stats = {}
        
        tags = sv.select('.tags .tag', self.soup)
        for tag in tags:
            tag_text = tag.get_text().strip()
            if tag_text:
                tag_stats[tag_text] = tag_stats.get(tag_text, 0) + 1
        
        return tag_stats

# Demonstrate custom filtering
analyzer = ForumAnalyzer(soup_forum)

print("Forum analysis with custom selectors:")

# High score posts
high_score_posts = analyzer.get_high_score_posts(min_score=15)
print(f"\nHigh score posts (≥15): {len(high_score_posts)} found")
for post in high_score_posts:
    user = post.get('data-user')
    score = post.get('data-score')
    username = sv.select_one('.username', post)
    print(f"  - {username.get_text() if username else user}: {score} points")

# Posts by user types
admin_posts = analyzer.get_posts_by_user_type(['admin', 'developer'])
print(f"\nAdmin/Developer posts: {len(admin_posts)} found")
for post in admin_posts:
    user = post.get('data-user')
    username = sv.select_one('.username', post)
    print(f"  - {username.get_text() if username else user} ({user})")

# Popular threads
popular_threads = analyzer.get_threads_by_popularity(min_views=200)
print(f"\nPopular threads (≥200 views): {len(popular_threads)} found")
for thread in popular_threads:
    views = thread.get('data-views')
    priority = thread.get('data-priority')
    print(f"  - Thread: {views} views, {priority} priority")

# Keyword search
keyword_posts = analyzer.extract_posts_with_keywords(['api', 'error', 'new'])
print(f"\nPosts with keywords: {len(keyword_posts)} found")
for match in keyword_posts:
    print(f"  - User {match['user']} mentioned '{match['keyword']}'")
    print(f"    Content: {match['content']}")

# Tag statistics
tag_stats = analyzer.get_tag_statistics()
print(f"\nTag usage statistics:")
sorted_tags = sorted(tag_stats.items(), key=lambda x: x[1], reverse=True)
for tag, count in sorted_tags:
    print(f"  - {tag}: {count} uses")

# Advanced selector combinations
print(f"\nAdvanced selector combinations:")

# Posts in high-priority threads
high_priority_posts = sv.select('.thread[data-priority="high"] .post', soup_forum)
print(f"Posts in high-priority threads: {len(high_priority_posts)}")

# Reply posts (not first post in thread)
reply_posts = sv.select('.post.reply', soup_forum)
print(f"Reply posts: {len(reply_posts)}")

# Posts with both important and announcement tags
important_announcements = sv.select('.post:has(.tag.important):has(.tag.announcement)', soup_forum)
print(f"Important announcements: {len(important_announcements)}")

print("Custom selectors and filtering completed successfully")
```

## Example 5: soupsieve functionality test 5

```python
import soupsieve as sv
from bs4 import BeautifulSoup
import time

# Example 5: Production HTML processing and performance optimization
print("=== Example 5: Production HTML Processing & Performance ===")

class HTMLProcessor:
    """Production-ready HTML processing with CSS selectors"""
    
    def __init__(self):
        self.compiled_selectors = {}
        
    def compile_selector(self, selector_name, selector):
        """Pre-compile CSS selectors for better performance"""
        self.compiled_selectors[selector_name] = sv.compile(selector)
        
    def bulk_extract_data(self, html_documents, selectors_config):
        """Process multiple HTML documents efficiently"""
        results = []
        
        # Compile all selectors once
        compiled = {}
        for name, selector in selectors_config.items():
            compiled[name] = sv.compile(selector)
            
        for i, html in enumerate(html_documents):
            soup = BeautifulSoup(html, 'html.parser')
            doc_data = {'doc_id': i}
            
            for selector_name, compiled_selector in compiled.items():
                elements = compiled_selector.select(soup)
                doc_data[selector_name] = [elem.get_text().strip() for elem in elements]
                
            results.append(doc_data)
            
        return results
    
    def validate_html_structure(self, soup, required_elements):
        """Validate that HTML has required structure"""
        validation_results = {}
        
        for element_name, selector in required_elements.items():
            elements = sv.select(selector, soup)
            validation_results[element_name] = {
                'found': len(elements),
                'valid': len(elements) > 0,
                'elements': elements[:3]  # First 3 matches
            }
            
        return validation_results
    
    def extract_structured_data(self, soup, schema):
        """Extract structured data based on schema definition"""
        data = {}
        
        for field_name, field_config in schema.items():
            selector = field_config['selector']
            extract_type = field_config.get('type', 'text')
            is_list = field_config.get('list', False)
            
            elements = sv.select(selector, soup)
            
            if is_list:
                if extract_type == 'text':
                    data[field_name] = [elem.get_text().strip() for elem in elements]
                elif extract_type == 'attr':
                    attr_name = field_config['attr']
                    data[field_name] = [elem.get(attr_name) for elem in elements]
                elif extract_type == 'html':
                    data[field_name] = [str(elem) for elem in elements]
            else:
                element = elements[0] if elements else None
                if element:
                    if extract_type == 'text':
                        data[field_name] = element.get_text().strip()
                    elif extract_type == 'attr':
                        attr_name = field_config['attr']
                        data[field_name] = element.get(attr_name)
                    elif extract_type == 'html':
                        data[field_name] = str(element)
                else:
                    data[field_name] = None
                    
        return data

# Sample news article HTML for processing
news_articles = [
    """
    <article class="news-article">
        <h1 class="headline">Tech Industry Sees Major Growth</h1>
        <div class="byline">
            <span class="author">John Reporter</span>
            <time class="publish-date" datetime="2024-01-15">January 15, 2024</time>
        </div>
        <div class="article-body">
            <p class="lead">Technology companies report record profits this quarter.</p>
            <p>The growth is attributed to increased digital adoption...</p>
        </div>
        <div class="tags">
            <span class="tag">technology</span>
            <span class="tag">business</span>
            <span class="tag">earnings</span>
        </div>
    </article>
    """,
    """
    <article class="news-article">
        <h1 class="headline">Climate Change Summit Begins</h1>
        <div class="byline">
            <span class="author">Sarah Environmental</span>
            <time class="publish-date" datetime="2024-01-16">January 16, 2024</time>
        </div>
        <div class="article-body">
            <p class="lead">World leaders gather to discuss environmental policies.</p>
            <p>The summit focuses on renewable energy initiatives...</p>
        </div>
        <div class="tags">
            <span class="tag">environment</span>
            <span class="tag">politics</span>
            <span class="tag">climate</span>
        </div>
    </article>
    """,
    """
    <article class="news-article">
        <h1 class="headline">Sports Championship Finals</h1>
        <div class="byline">
            <span class="author">Mike Sports</span>
            <time class="publish-date" datetime="2024-01-17">January 17, 2024</time>
        </div>
        <div class="article-body">
            <p class="lead">Exciting championship game draws record viewership.</p>
            <p>The final score was 28-21 in a thrilling overtime...</p>
        </div>
        <div class="tags">
            <span class="tag">sports</span>
            <span class="tag">championship</span>
            <span class="tag">entertainment</span>
        </div>
    </article>
    """
]

# Initialize processor
processor = HTMLProcessor()

# Define extraction configuration
news_selectors = {
    'headlines': '.headline',
    'authors': '.author',
    'dates': '.publish-date',
    'lead_paragraphs': '.lead',
    'tags': '.tag'
}

print("Processing news articles with bulk extraction:")
start_time = time.perf_counter()

# Bulk process all articles
results = processor.bulk_extract_data(news_articles, news_selectors)
processing_time = time.perf_counter() - start_time

print(f"Processed {len(news_articles)} articles in {processing_time:.4f} seconds")
print(f"Average time per article: {processing_time/len(news_articles):.4f} seconds")

for i, result in enumerate(results, 1):
    print(f"\nArticle {i}:")
    print(f"  Headlines: {result.get('headlines', [])}")
    print(f"  Author: {result.get('authors', [])}")
    print(f"  Tags: {result.get('tags', [])}")

# Schema-based extraction
print(f"\nStructured data extraction using schema:")

news_schema = {
    'title': {'selector': '.headline', 'type': 'text'},
    'author': {'selector': '.author', 'type': 'text'},
    'publish_date': {'selector': '.publish-date', 'type': 'attr', 'attr': 'datetime'},
    'lead_paragraph': {'selector': '.lead', 'type': 'text'},
    'all_tags': {'selector': '.tag', 'type': 'text', 'list': True},
    'article_html': {'selector': '.article-body', 'type': 'html'}
}

# Extract structured data from first article
soup_sample = BeautifulSoup(news_articles[0], 'html.parser')
structured_data = processor.extract_structured_data(soup_sample, news_schema)

print("Structured data from first article:")
for field, value in structured_data.items():
    if field == 'article_html':
        print(f"  {field}: {str(value)[:60]}..." if value else f"  {field}: None")
    else:
        print(f"  {field}: {value}")

# HTML validation
print(f"\nHTML structure validation:")

required_elements = {
    'headline': '.headline',
    'author_info': '.byline .author',
    'publication_date': '.publish-date',
    'article_content': '.article-body p',
    'category_tags': '.tags .tag'
}

validation_result = processor.validate_html_structure(soup_sample, required_elements)
print("Validation results:")
for element, result in validation_result.items():
    status = "✓" if result['valid'] else "✗"
    print(f"  {status} {element}: {result['found']} found")

# Performance comparison
print(f"\nPerformance comparison:")

# Test with compiled vs non-compiled selectors
test_selector = '.news-article .headline'
test_soup = BeautifulSoup(news_articles[0], 'html.parser')

# Non-compiled performance
start = time.perf_counter()
for _ in range(1000):
    sv.select(test_selector, test_soup)
non_compiled_time = time.perf_counter() - start

# Compiled performance
compiled_selector = sv.compile(test_selector)
start = time.perf_counter()
for _ in range(1000):
    compiled_selector.select(test_soup)
compiled_time = time.perf_counter() - start

print(f"Non-compiled selector (1000x): {non_compiled_time:.4f}s")
print(f"Compiled selector (1000x): {compiled_time:.4f}s")
print(f"Performance improvement: {non_compiled_time/compiled_time:.1f}x faster")

print("Production HTML processing completed successfully")
```
