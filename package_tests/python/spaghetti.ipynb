## Example 1: spaghetti functionality test 1

```python
# Example 1: Create Simple Network
import spaghetti
import geopandas as gpd
from shapely.geometry import LineString

print("=== Create Simple Network ===")

# Create GeoDataFrame with street network
streets = gpd.GeoDataFrame({
    'geometry': [
        LineString([(0, 0), (1, 0)]),
        LineString([(1, 0), (1, 1)]),
        LineString([(1, 1), (0, 1)]),
        LineString([(0, 1), (0, 0)])
    ]
})

# Build network from GeoDataFrame
ntw = spaghetti.Network(in_data=streets)

print(f"Network created: {type(ntw).__name__}")
print(f"Number of vertices: {len(ntw.vertex_coords)}")
print(f"Number of edges: {len(ntw.edges)}")

print("\nExample 1 completed!")
```

## Example 2: spaghetti functionality test 2

```python
# Example 2: Point Pattern on Network
import spaghetti
import geopandas as gpd
from shapely.geometry import LineString, Point

print("=== Point Pattern on Network ===")

# Create network
streets = gpd.GeoDataFrame({
    'geometry': [
        LineString([(0, 0), (2, 0)]),
        LineString([(2, 0), (2, 2)]),
        LineString([(2, 2), (0, 2)])
    ]
})
ntw = spaghetti.Network(in_data=streets)

# Create point pattern GeoDataFrame
points_gdf = gpd.GeoDataFrame({
    'geometry': [Point(0.5, 0.3), Point(1.8, 0.2), Point(1.9, 1.5)]
})

# Snap points to network
ntw.snapobservations(points_gdf, "pp", attribute=True)

print(f"Original points: {len(points_gdf)}")
print(f"Snapped observations: 'pp' added to network")

print("\nExample 2 completed!")
```

## Example 3: spaghetti functionality test 3

```python
# Example 3: Network Analysis
import spaghetti
import geopandas as gpd
from shapely.geometry import LineString, Point

print("=== Network Analysis ===")

# Create network
streets = gpd.GeoDataFrame({
    'geometry': [
        LineString([(0, 0), (3, 0)]),
        LineString([(3, 0), (3, 3)]),
        LineString([(3, 3), (0, 3)]),
        LineString([(0, 3), (0, 0)])
    ]
})
ntw = spaghetti.Network(in_data=streets)

# Add observation points
points = gpd.GeoDataFrame({'geometry': [Point(1, 0.1), Point(2.9, 1.5)]})
ntw.snapobservations(points, "obs", attribute=True)

print(f"Network edges: {len(ntw.edges)}")
print(f"Observations snapped: 'obs'")

print("\nExample 3 completed!")
```

## Example 4: spaghetti functionality test 4

```python
# Example 4: Network Properties
import spaghetti
import geopandas as gpd
from shapely.geometry import LineString

print("=== Network Properties ===")

# Create L-shaped network
streets = gpd.GeoDataFrame({
    'geometry': [
        LineString([(0, 0), (5, 0)]),
        LineString([(5, 0), (5, 5)])
    ]
})
ntw = spaghetti.Network(in_data=streets)

# Network structure
print(f"Network edges: {len(ntw.edges)}")
print(f"Network vertices: {len(ntw.vertex_coords)}")

# Show vertex coordinates
print(f"\nVertex coordinates (first 2):")
for vertex_id in list(ntw.vertex_coords.keys())[:2]:
    coords = ntw.vertex_coords[vertex_id]
    print(f"  Vertex {vertex_id}: {coords}")

print("\nExample 4 completed!")
```

## Example 5: spaghetti functionality test 5

```python
# Example 5: Network Topology
import spaghetti
import geopandas as gpd
from shapely.geometry import LineString

print("=== Network Topology ===")

# Create branching network
streets = gpd.GeoDataFrame({
    'geometry': [
        LineString([(0, 0), (2, 0)]),
        LineString([(1, 0), (1, 1)]),
        LineString([(1, 0), (1, -1)]),
        LineString([(2, 0), (3, 1)])
    ]
})
ntw = spaghetti.Network(in_data=streets)

print(f"Topology:")
print(f"  Vertices: {len(ntw.vertex_coords)}")
print(f"  Edges: {len(ntw.edges)}")

# Adjacency
print(f"\nAdjacency (first 2 vertices):")
for vertex_id in list(ntw.adjacencylist.keys())[:2]:
    neighbors = ntw.adjacencylist[vertex_id]
    print(f"  Vertex {vertex_id}: {len(neighbors)} neighbors")

print("\nExample 5 completed!")
```
