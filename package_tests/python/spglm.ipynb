## Example 1: spglm functionality test 1

```python
# Example 1: Gaussian GLM
from spglm.glm import GLM
from spglm.family import Gaussian
import numpy as np

print("=== Gaussian GLM ===")

# Generate sample data
np.random.seed(42)
n = 100
X = np.random.randn(n, 3)
X = np.column_stack([np.ones(n), X])  # Shape: (100, 4)
beta_true = np.array([1.0, 2.0, -1.5, 0.8])
y = np.dot(X, beta_true) + np.random.randn(n) * 0.5
y = y.reshape(-1, 1)  # Reshape to column vector (100, 1)

# Fit Gaussian GLM
model = GLM(y, X, family=Gaussian())
results = model.fit()

print(f"Model fitted: {type(model).__name__}")
print(f"Coefficients: {results.params.flatten()}")
print(f"True betas: {beta_true}")

print("\nExample 1 completed!")
```

## Example 2: spglm functionality test 2

```python
# Example 2: Poisson GLM
from spglm.glm import GLM
from spglm.family import Poisson
import numpy as np

print("=== Poisson GLM ===")

# Generate count data
np.random.seed(123)
n = 80
X = np.random.randn(n, 2)
X = np.column_stack([np.ones(n), X])
beta = np.array([0.5, 1.2, -0.8])
mu = np.exp(np.dot(X, beta))
y = np.random.poisson(mu).reshape(-1, 1)

# Fit Poisson GLM
model = GLM(y, X, family=Poisson())
results = model.fit()

print(f"Family: {model.family.__class__.__name__}")
print(f"Observations: {len(y)}")
print(f"Estimated coefficients: {results.params.flatten()}")

print("\nExample 2 completed!")
```

## Example 3: spglm functionality test 3

```python
# Example 3: Binomial GLM (Logistic Regression)
from spglm.glm import GLM
from spglm.family import Binomial
import numpy as np

print("=== Binomial GLM ===")

# Generate binary data
np.random.seed(456)
n = 100
X = np.random.randn(n, 2)
X = np.column_stack([np.ones(n), X])
beta = np.array([0.0, 1.5, -1.0])
prob = 1 / (1 + np.exp(-np.dot(X, beta)))
y = np.random.binomial(1, prob).reshape(-1, 1)

# Fit Binomial GLM
model = GLM(y, X, family=Binomial())
results = model.fit()

print(f"Family: Binomial (Logistic)")
print(f"Binary outcomes: {np.sum(y)} ones, {n - int(np.sum(y))} zeros")
print(f"Coefficients: {results.params.flatten()}")

print("\nExample 3 completed!")
```

## Example 4: spglm functionality test 4

```python
# Example 4: QuasiPoisson GLM
from spglm.glm import GLM
from spglm.family import QuasiPoisson
import numpy as np

print("=== QuasiPoisson GLM ===")

# Generate overdispersed count data
np.random.seed(789)
n = 90
X = np.random.randn(n, 2)
X = np.column_stack([np.ones(n), X])
beta = np.array([1.0, 0.8, -0.6])
mu = np.exp(np.dot(X, beta))
y = np.random.negative_binomial(5, 5 / (5 + mu)).reshape(-1, 1)

# Fit QuasiPoisson GLM
model = GLM(y, X, family=QuasiPoisson())
results = model.fit()

print(f"Family: QuasiPoisson")
print(f"Coefficients: {results.params.flatten()}")

print("\nExample 4 completed!")
```

## Example 5: spglm functionality test 5

```python
# Example 5: Sparse Matrix Support
from spglm.glm import GLM
from spglm.family import Gaussian
import numpy as np
from scipy.sparse import csr_matrix

print("=== Sparse Matrix Support ===")

# Generate sparse design matrix
np.random.seed(101)
n, p = 100, 50
X_dense = np.random.randn(n, p)
X_dense[X_dense < 1.0] = 0  # Make sparse
X_sparse = csr_matrix(X_dense)
beta = np.random.randn(p)
y = (X_sparse.dot(beta) + np.random.randn(n) * 0.5).reshape(-1, 1)

# Fit with sparse matrix
model = GLM(y, X_sparse, family=Gaussian())
results = model.fit()

print(f"Sparse matrix shape: {X_sparse.shape}")
print(f"Sparsity: {1 - X_sparse.nnz / (n * p):.2%}")
print(f"Coefficients (first 5): {results.params.flatten()[:5]}")

print("\nExample 5 completed!")
```
