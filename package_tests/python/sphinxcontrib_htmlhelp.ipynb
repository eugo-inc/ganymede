# sphinxcontrib_htmlhelp

sphinxcontrib-htmlhelp 2.1.0 is Sphinx's `htmlhelp` builder: HTML pages plus the project (.hhp), contents (.hhc) and
index (.hhk) files Microsoft's HTML Help compiler reads. The compiler itself is Windows-only; these cells check the
files it would be given, their encoding setting, and the registration.

```python
# Example 1: an HTML Help build writes the pages and the three help-compiler files.
import pathlib, tempfile
from sphinx.cmd.build import build_main

src = pathlib.Path(tempfile.mkdtemp())
(src / "conf.py").write_text("project = 'Demo'\n")
(src / "index.rst").write_text("Demo\n====\n\n.. toctree::\n\n   page\n\nIntro with *emphasis*.\n")
(src / "page.rst").write_text("Page\n====\n\nA page about :index:`widgets`.\n")
rc = build_main(["-q", "-W", "-b", 'htmlhelp', str(src), str(src / "_out")])
out = src / "_out"
print("rc", rc, sorted(p.name for p in out.iterdir()))
assert rc == 0
assert all((out / f'demodoc.{ext}').exists() for ext in ('hhp', 'hhc', 'hhk'))
```

```python
# Example 2: the project file lists the start page and the contents/index files.
import pathlib, tempfile
from sphinx.cmd.build import build_main

src = pathlib.Path(tempfile.mkdtemp())
(src / "conf.py").write_text("project = 'Demo'\n")
(src / "index.rst").write_text("Demo\n====\n\n.. toctree::\n\n   page\n\nIntro with *emphasis*.\n")
(src / "page.rst").write_text("Page\n====\n\nA page about :index:`widgets`.\n")
rc = build_main(["-q", "-W", "-b", 'htmlhelp', str(src), str(src / "_out")])
out = src / "_out"
print("rc", rc, sorted(p.name for p in out.iterdir()))
assert rc == 0
hhp = (out / 'demodoc.hhp').read_text(encoding='cp1252')
print(hhp[:400])
assert '[OPTIONS]' in hhp and 'Default topic=index.html' in hhp
assert 'Contents file=demodoc.hhc' in hhp and 'Index file=demodoc.hhk' in hhp
```

```python
# Example 3: the contents file is the toctree; the index file carries the index entries.
import pathlib, tempfile
from sphinx.cmd.build import build_main

src = pathlib.Path(tempfile.mkdtemp())
(src / "conf.py").write_text("project = 'Demo'\n")
(src / "index.rst").write_text("Demo\n====\n\n.. toctree::\n\n   page\n\nIntro with *emphasis*.\n")
(src / "page.rst").write_text("Page\n====\n\nA page about :index:`widgets`.\n")
rc = build_main(["-q", "-W", "-b", 'htmlhelp', str(src), str(src / "_out")])
out = src / "_out"
print("rc", rc, sorted(p.name for p in out.iterdir()))
assert rc == 0
hhc = (out / 'demodoc.hhc').read_text(encoding='cp1252')
hhk = (out / 'demodoc.hhk').read_text(encoding='cp1252')
assert 'page.html' in hhc and 'widgets' in hhk
```

```python
# Example 4: the builder's registration.
import sphinxcontrib.htmlhelp as htmlhelp
from sphinxcontrib.htmlhelp import HTMLHelpBuilder

print(htmlhelp.__version__, HTMLHelpBuilder.name)
assert htmlhelp.__version__ == "2.1.0"
assert HTMLHelpBuilder.name == "htmlhelp" and HTMLHelpBuilder.embedded is True
```
