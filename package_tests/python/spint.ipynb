## Example 1: spint functionality test 1

```python
# Example 1: Gravity Model
from spint import Gravity
import numpy as np

print("=== Gravity Model ===")

# Generate flow data
np.random.seed(42)
n = 20
flows = np.random.poisson(50, (n, n)).reshape(-1, 1)
flows[flows == 0] = 1
origins = np.repeat(np.arange(n), n)
dests = np.tile(np.arange(n), n)
distances = np.random.uniform(1, 100, n * n).reshape(-1, 1)

# Create origin and destination attributes (population)
origin_pop = np.random.uniform(100, 1000, n)
dest_pop = np.random.uniform(100, 1000, n)

# Fit gravity model
model = Gravity(flows, origin_pop[origins], dest_pop[dests], distances, cost_func='exp')
results = model.fit()

print(f"Model type: {type(model).__name__}")
print(f"Parameters: {results.params.flatten()[:3]}")

print("\nExample 1 completed!")
```

## Example 2: spint functionality test 2

```python
# Example 2: Production-Constrained Model
from spint import Production
import numpy as np

print("=== Production-Constrained Model ===")

# Generate flow data
np.random.seed(123)
n = 15
flows = np.random.poisson(30, (n, n)).reshape(-1, 1)
flows[flows == 0] = 1  # Replace zeros
origins = np.repeat(np.arange(n), n)
dests = np.tile(np.arange(n), n)
distances = np.random.uniform(1, 50, n * n).reshape(-1, 1)

# Create destination attributes (population)
dest_pop = np.random.uniform(100, 1000, n)

# Fit production-constrained model with power cost function
model = Production(flows, origins, dest_pop[dests], distances, cost_func='pow')
results = model.fit()

print(f"Model type: {type(model).__name__}")
print(f"Parameters shape: {results.params.shape}")

print("\nExample 2 completed!")
```

## Example 3: spint functionality test 3

```python
# Example 3: Attraction-Constrained Model
from spint import Attraction
import numpy as np

print("=== Attraction-Constrained Model ===")

# Generate flow data
np.random.seed(456)
n = 15
flows = np.random.poisson(40, (n, n)).reshape(-1, 1)
flows[flows == 0] = 1  # Replace zeros
origins = np.repeat(np.arange(n), n)
dests = np.tile(np.arange(n), n)
distances = np.random.uniform(1, 60, n * n).reshape(-1, 1)

# Create origin attributes (population) - Attraction needs origin variables
origin_pop = np.random.uniform(100, 1000, n)

# Fit attraction-constrained model with exponential cost function
model = Attraction(flows, dests, origin_pop[origins], distances, cost_func='exp')
results = model.fit()

print(f"Model type: {type(model).__name__}")
print(f"Parameters shape: {results.params.shape}")

print("\nExample 3 completed!")
```

## Example 4: spint functionality test 4

```python
# Example 4: Doubly-Constrained Model
from spint import Doubly
import numpy as np

print("=== Doubly-Constrained Model ===")

# Generate flow data
np.random.seed(789)
n = 15
flows = np.random.poisson(35, (n, n)).reshape(-1, 1)
flows[flows == 0] = 1  # Replace zeros
origins = np.repeat(np.arange(n), n)
dests = np.tile(np.arange(n), n)
distances = np.random.uniform(1, 70, n * n).reshape(-1, 1)

# Fit doubly-constrained model (no additional attributes needed)
model = Doubly(flows, origins, dests, distances, cost_func='pow')
results = model.fit()

print(f"Model type: {type(model).__name__}")
print(f"Parameters shape: {results.params.shape}")

print("\nExample 4 completed!")
```

## Example 5: spint functionality test 5

```python
# Example 5: Gravity Model with Power Cost Function
from spint import Gravity
import numpy as np

print("=== Gravity Model with Power Cost ===")

# Generate flow data
np.random.seed(101)
n = 20
flows = np.random.poisson(45, (n, n)).reshape(-1, 1)
flows[flows == 0] = 1  # Replace zeros
origins = np.repeat(np.arange(n), n)
dests = np.tile(np.arange(n), n)
distances = np.random.uniform(1, 80, n * n).reshape(-1, 1)

# Create origin and destination attributes (population)
origin_pop = np.random.uniform(100, 1000, n)
dest_pop = np.random.uniform(100, 1000, n)

# Fit gravity model with power cost function
model = Gravity(flows, origin_pop[origins], dest_pop[dests], distances, cost_func='pow')
results = model.fit()

print(f"Model type: {type(model).__name__}")
print(f"Cost function: pow")
print(f"Parameters: {results.params.flatten()[:3]}")

print("\nExample 5 completed!")
```
