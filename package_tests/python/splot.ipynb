## Example 1: splot functionality test 1

```python
# Example 1: Moran Scatterplot
from splot.esda import moran_scatterplot
from esda import Moran
import geopandas as gpd
import numpy as np
from libpysal.weights import lat2W

print("=== Moran Scatterplot ===")

# Create synthetic spatial data
np.random.seed(42)
n = 25
coords = [(i % 5, i // 5) for i in range(n)]
values = np.random.randn(n) * 10 + 50

# Create spatial weights
w = lat2W(5, 5)

# Compute Moran's I
moran = Moran(values, w)

# Create Moran scatterplot
fig, ax = moran_scatterplot(moran)

print(f"Moran scatterplot created")
print(f"Moran's I: {moran.I:.4f}")
print(f"Figure type: {type(fig).__name__}")

print("\nExample 1 completed!")
```

## Example 2: splot functionality test 2

```python
# Example 2: LISA Cluster Map
from splot.esda import lisa_cluster
import geopandas as gpd
import numpy as np
from esda import Moran_Local
from libpysal.weights import Queen

print("=== LISA Cluster Map ===")

# Create grid of polygons
from shapely.geometry import box
n = 5
polygons = [box(i % n, i // n, (i % n) + 1, (i // n) + 1) for i in range(n * n)]

# Create synthetic data with spatial pattern
np.random.seed(123)
values = np.random.randn(n * n) * 10 + 50
gdf = gpd.GeoDataFrame({'value': values}, geometry=polygons)

# Create spatial weights and compute local Moran
w = Queen.from_dataframe(gdf)
moran_loc = Moran_Local(gdf['value'], w)

# Create LISA cluster map
fig, ax = lisa_cluster(moran_loc, gdf)

print(f"LISA cluster map created")
print(f"Figure type: {type(fig).__name__}")
print("\nExample 2 completed!")
```

## Example 3: splot functionality test 3

```python
# Example 3: Choropleth Map
from splot.mapping import vba_choropleth
import geopandas as gpd
import numpy as np

print("=== Choropleth Map ===")

# Create grid of polygons
from shapely.geometry import box
n = 4
polygons = [box(i % n, i // n, (i % n) + 1, (i // n) + 1) for i in range(n * n)]

# Create data with values
np.random.seed(456)
values = np.random.uniform(10, 100, n * n)
gdf = gpd.GeoDataFrame({'value': values}, geometry=polygons)

# Create value-by-alpha choropleth
fig, ax = vba_choropleth(gdf['value'], gdf['value'], gdf)

print(f"Choropleth map created")
print(f"Data range: {values.min():.2f} to {values.max():.2f}")
print(f"Figure type: {type(fig).__name__}")

print("\nExample 3 completed!")
```

## Example 4: splot functionality test 4

```python
# Example 4: Plot Spatial Weights
from splot.libpysal import plot_spatial_weights
import geopandas as gpd
from libpysal.weights import Queen
from shapely.geometry import box

print("=== Plot Spatial Weights ===")

# Create grid of polygons
n = 4
polygons = [box(i % n, i // n, (i % n) + 1, (i // n) + 1) for i in range(n * n)]
gdf = gpd.GeoDataFrame(geometry=polygons)

# Create Queen contiguity weights
w = Queen.from_dataframe(gdf)

# Plot spatial weights structure
fig, ax = plot_spatial_weights(w, gdf)

print(f"Spatial weights plot created")
print(f"Number of neighbors: {w.n}")
print(f"Figure type: {type(fig).__name__}")

print("\nExample 4 completed!")
```

## Example 5: splot functionality test 5

```python
# Example 5: Plot Local Moran Statistics
from splot.esda import plot_local_autocorrelation
from esda import Moran_Local
import geopandas as gpd
import numpy as np
from libpysal.weights import lat2W
from shapely.geometry import Point

print("=== Local Moran Statistics ===")

# Create synthetic spatial data
np.random.seed(101)
n = 25
coords = [(i % 5, i // 5) for i in range(n)]
values = np.random.randn(n) * 10 + 50
gdf = gpd.GeoDataFrame({'value': values}, geometry=[Point(x, y) for x, y in coords])

# Create spatial weights and compute local Moran
w = lat2W(5, 5)
moran_loc = Moran_Local(values, w)

# Plot local autocorrelation (creates multiple subplots)
fig = plot_local_autocorrelation(moran_loc, gdf, 'value')

print(f"Local autocorrelation plot created")
print(f"Figure type: {type(fig).__name__}")

print("\nExample 5 completed!")
```
