## Example 1: spopt functionality test 1

```python
# Example 1: P-Median Problem
from spopt.locate import PMedian
import numpy as np
import pulp

print("=== P-Median Problem ===")

# Create synthetic distance matrix
np.random.seed(42)
n_demand = 10
n_facilities = 5
distances = np.random.uniform(1, 100, (n_demand, n_facilities))
demand = np.random.uniform(10, 50, n_demand)

# Solve p-median problem (locate 2 facilities)
pmedian = PMedian.from_cost_matrix(distances, demand, p_facilities=2)
pmedian = pmedian.solve(pulp.PULP_CBC_CMD(msg=0))

print(f"Optimal facilities: {pmedian.fac2cli}")
print(f"Total cost: {pmedian.problem.objective.value():.2f}")
print(f"Status: {pulp.LpStatus[pmedian.problem.status]}")

print("\nExample 1 completed!")
```

## Example 2: spopt functionality test 2

```python
# Example 2: P-Center Problem
from spopt.locate import PCenter
import numpy as np
import pulp

print("=== P-Center Problem ===")

# Create synthetic distance matrix
np.random.seed(123)
n_demand = 12
n_facilities = 6
distances = np.random.uniform(5, 80, (n_demand, n_facilities))

# Solve p-center problem (minimize maximum distance)
pcenter = PCenter.from_cost_matrix(distances, p_facilities=3)
pcenter = pcenter.solve(pulp.PULP_CBC_CMD(msg=0))

print(f"Selected facilities: {pcenter.fac2cli}")
print(f"Maximum distance: {pcenter.problem.objective.value():.2f}")
print(f"Status: {pulp.LpStatus[pcenter.problem.status]}")

print("\nExample 2 completed!")
```

## Example 3: spopt functionality test 3

```python
# Example 3: LSCP (Location Set Covering Problem)
from spopt.locate import LSCP
import numpy as np
import pulp

print("=== Location Set Covering Problem ===")

# Create synthetic distance matrix
np.random.seed(456)
n_demand = 8
n_facilities = 8
distances = np.random.uniform(1, 50, (n_demand, n_facilities))
service_radius = 30

# Solve LSCP (minimum facilities to cover all demand)
lscp = LSCP.from_cost_matrix(distances, service_radius)
lscp = lscp.solve(pulp.PULP_CBC_CMD(msg=0))

print(f"Facilities to clients: {lscp.fac2cli}")
print(f"Clients to facilities: {lscp.cli2fac}")
print(f"Status: {pulp.LpStatus[lscp.problem.status]}")

print("\nExample 3 completed!")
```

## Example 4: spopt functionality test 4

```python
# Example 4: MCLP (Maximal Covering Location Problem)
from spopt.locate import MCLP
import numpy as np
import pulp

print("=== Maximal Covering Location Problem ===")

# Create synthetic data
np.random.seed(789)
n_demand = 10
n_facilities = 8
distances = np.random.uniform(1, 60, (n_demand, n_facilities))
demand = np.random.uniform(5, 30, n_demand)
service_radius = 40

# Solve MCLP (maximize coverage with p facilities)
mclp = MCLP.from_cost_matrix(distances, demand, service_radius, p_facilities=3)
mclp = mclp.solve(pulp.PULP_CBC_CMD(msg=0))

print(f"Facilities to clients: {mclp.fac2cli}")
print(f"Covered demand: {mclp.problem.objective.value():.2f}")
print(f"Status: {pulp.LpStatus[mclp.problem.status]}")

print("\nExample 4 completed!")
```

## Example 5: spopt functionality test 5

```python
# Example 5: Region K Means
from spopt.region import RegionKMeansHeuristic
import numpy as np
from libpysal.weights import lat2W

print("=== Region K-Means (Spatially Constrained Clustering) ===")

# Create synthetic spatial data
np.random.seed(101)
n = 25
X = np.random.randn(n, 3)

# Create spatial weights
w = lat2W(5, 5)

# Perform spatially constrained k-means
model = RegionKMeansHeuristic(X, n_clusters=4, w=w)
model.solve()

print(f"Number of clusters: {len(np.unique(model.labels_))}")
print(f"Labels: {model.labels_[:10]}...")
print(f"Each region is spatially contiguous")

print("\nExample 5 completed!")
```
