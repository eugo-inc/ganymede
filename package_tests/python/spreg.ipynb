## Example 1: spreg functionality test 1

```python
# Example 1: OLS Regression
from spreg import OLS
import numpy as np

print("=== OLS Regression ===")

# Generate synthetic data
np.random.seed(42)
n = 100
X = np.random.randn(n, 2)
beta = np.array([1.5, -0.8])
y = X @ beta + np.random.randn(n) * 0.5

# Fit OLS model
model = OLS(y, X)

print(f"Model: OLS")
print(f"Coefficients: {model.betas.flatten()}")
print(f"R-squared: {model.r2:.4f}")
print(f"Observations: {model.n}")

print("\nExample 1 completed!")
```

## Example 2: spreg functionality test 2

```python
# Example 2: Spatial Lag Model
from spreg import ML_Lag
import numpy as np
from libpysal.weights import lat2W

print("=== Spatial Lag Model ===")

# Generate spatial data
np.random.seed(123)
n = 25
X = np.random.randn(n, 2)
w = lat2W(5, 5)
y = np.random.randn(n) * 10 + 50

# Fit spatial lag model (Maximum Likelihood)
model = ML_Lag(y, X, w)

print(f"Model: Spatial Lag")
print(f"Rho (spatial lag): {model.rho:.4f}")
print(f"Coefficients: {model.betas.flatten()[:3]}")
print(f"Log-likelihood: {model.logll:.2f}")

print("\nExample 2 completed!")
```

## Example 3: spreg functionality test 3

```python
# Example 3: Spatial Error Model
from spreg import ML_Error
import numpy as np
from libpysal.weights import lat2W

print("=== Spatial Error Model ===")

# Generate spatial data with error autocorrelation
np.random.seed(456)
n = 25
X = np.random.randn(n, 2)
w = lat2W(5, 5)
y = np.random.randn(n) * 10 + 50

# Fit spatial error model (Maximum Likelihood)
model = ML_Error(y, X, w)

print(f"Model: Spatial Error")
print(f"Lambda (spatial error): {model.lam:.4f}")
print(f"Coefficients: {model.betas.flatten()}")
print(f"Log-likelihood: {model.logll:.2f}")

print("\nExample 3 completed!")
```

## Example 4: spreg functionality test 4

```python
# Example 4: Two-Stage Least Squares
from spreg import TSLS
import numpy as np

print("=== Two-Stage Least Squares ===")

# Generate data with endogenous variable
np.random.seed(789)
n = 100
X = np.random.randn(n, 2)
Z = np.random.randn(n, 2)  # Instruments
yend = np.random.randn(n, 1)  # Endogenous variable
y = X @ np.array([1.5, -0.8]) + yend.flatten() * 0.5 + np.random.randn(n) * 0.3

# Fit 2SLS model
model = TSLS(y, X, yend, Z)

print(f"Model: 2SLS")
print(f"Coefficients: {model.betas.flatten()}")
print(f"Observations: {model.n}")

print("\nExample 4 completed!")
```

## Example 5: spreg functionality test 5

```python
# Example 5: GM_Lag (Generalized Method of Moments Spatial Lag)
from spreg import GM_Lag
import numpy as np
from libpysal.weights import lat2W

print("=== GMM Spatial Lag Model ===")

# Generate spatial data
np.random.seed(101)
n = 25
X = np.random.randn(n, 2)
w = lat2W(5, 5)
y = np.random.randn(n) * 10 + 50

# Fit GMM spatial lag model
model = GM_Lag(y, X, w=w)

print(f"Model: GM_Lag")
print(f"Coefficients: {model.betas.flatten()}")
print(f"Observations: {model.n}")
print(f"Variables: {model.k}")

print("\nExample 5 completed!")
```
