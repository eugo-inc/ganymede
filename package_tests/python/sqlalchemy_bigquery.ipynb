## Example 1: BigQuery Dialect Registration

```python
# Example 1: BigQuery Dialect Registration
from sqlalchemy import create_engine
from sqlalchemy_bigquery import BigQueryDialect

print("=== BigQuery Dialect ===")

# Check dialect is registered
dialect = BigQueryDialect()

print(f"Dialect name: {dialect.name}")
print(f"Driver: {dialect.driver}")
print(f"Supports transactions: {dialect.supports_statement_cache}")

# BigQuery connection URL format
url_format = "bigquery://project-id/dataset-name"
print(f"\nConnection URL format: {url_format}")

print("\nExample 1 completed!")
```

## Example 2: BigQuery Types

```python
# Example 2: BigQuery Types
from sqlalchemy_bigquery import STRUCT, ARRAY
from sqlalchemy import Integer, String

print("=== BigQuery-Specific Types ===")

# BigQuery supports nested and repeated fields
struct_type = STRUCT()
array_type = ARRAY(Integer)

print(f"STRUCT type: {type(struct_type).__name__}")
print(f"ARRAY type: {type(array_type).__name__}")
print(f"Array item type: Integer")

print("\nBigQuery supports complex nested data structures")
print("STRUCT: Record/nested object type")
print("ARRAY: Repeated field type")

print("\nExample 2 completed!")
```

## Example 3: Table Definition with BigQuery Types

```python
# Example 3: Table Definition with BigQuery Types
from sqlalchemy import Table, Column, MetaData, String, Integer
from sqlalchemy_bigquery import ARRAY, STRUCT

print("=== BigQuery Table Definition ===")

metadata = MetaData()

# Define table with BigQuery-specific types
users_table = Table(
    'users',
    metadata,
    Column('id', Integer, primary_key=True),
    Column('name', String),
    Column('tags', ARRAY(String)),  # Array of strings
)

print(f"Table name: {users_table.name}")
print(f"Columns: {[c.name for c in users_table.columns]}")
print(f"Array column: tags (ARRAY<STRING>)")

print("\nExample 3 completed!")
```

## Example 4: BigQuery Compiler

```python
# Example 4: BigQuery Compiler
from sqlalchemy_bigquery import BigQueryDialect
from sqlalchemy import select, table, column

print("=== BigQuery SQL Compiler ===")

# Create dialect and compiler
dialect = BigQueryDialect()

# Simple table reference
users = table('users', column('id'), column('name'))

# Create a SELECT statement
stmt = select(users.c.id, users.c.name).limit(10)

# Compile to BigQuery SQL
compiled = stmt.compile(dialect=dialect)

print(f"SQLAlchemy statement compiled to BigQuery SQL")
print(f"Statement type: {type(stmt).__name__}")
print(f"Dialect: {dialect.name}")

print("\nExample 4 completed!")
```
