## Example 1: sqlalchemy_utils functionality test 1

```python
# Example 1: URLType for URL Fields
from sqlalchemy_utils import URLType
from sqlalchemy import Column, Integer, create_engine
from sqlalchemy.orm import declarative_base

print("=== URLType ===")

Base = declarative_base()

class Website(Base):
    __tablename__ = 'websites'
    id = Column(Integer, primary_key=True)
    url = Column(URLType)

# URLType automatically validates and parses URLs
print(f"Column type: {type(Website.url.type).__name__}")
print(f"Python type: URL objects with validation")
print(f"Use case: Store and validate web URLs")

print("\nExample 1 completed!")
```

## Example 2: sqlalchemy_utils functionality test 2

```python
# Example 2: ChoiceType for Enumerations
from sqlalchemy_utils import ChoiceType
from sqlalchemy import Column, Integer
from sqlalchemy.orm import declarative_base

print("=== ChoiceType ===")

Base = declarative_base()

class User(Base):
    __tablename__ = 'users'
    
    TYPES = [
        ('admin', 'Administrator'),
        ('user', 'Regular User'),
        ('guest', 'Guest')
    ]
    
    id = Column(Integer, primary_key=True)
    user_type = Column(ChoiceType(TYPES))

# ChoiceType stores value but provides label access
print(f"Column type: {type(User.user_type.type).__name__}")
print(f"Choices: {User.TYPES}")
print(f"Use case: Enum-like fields with human-readable labels")

print("\nExample 2 completed!")
```

## Example 3: sqlalchemy_utils functionality test 3

```python
# Example 3: JSONType for JSON Data
from sqlalchemy_utils import JSONType
from sqlalchemy import Column, Integer
from sqlalchemy.orm import declarative_base

print("=== JSONType ===")

Base = declarative_base()

class Document(Base):
    __tablename__ = 'documents'
    id = Column(Integer, primary_key=True)
    properties = Column(JSONType)

# JSONType automatically serializes/deserializes JSON
print(f"Column type: {type(Document.properties.type).__name__}")
print(f"Python type: dict/list (automatically serialized)")
print(f"Use case: Store structured data as JSON")
print(f"Features: Automatic serialization, query support")

print("\nExample 3 completed!")
```

## Example 5: sqlalchemy_utils functionality test 5

```python
# Example 5: UUIDType and Other Special Types
from sqlalchemy_utils import UUIDType, EmailType, IPAddressType
from sqlalchemy import Column
from sqlalchemy.orm import declarative_base

print("=== Special Column Types ===")

Base = declarative_base()

class Resource(Base):
    __tablename__ = 'resources'
    id = Column(UUIDType, primary_key=True)
    email = Column(EmailType)
    ip_address = Column(IPAddressType)

print(f"UUIDType: {type(Resource.id.type).__name__}")
print(f"EmailType: {type(Resource.email.type).__name__}")
print(f"IPAddressType: {type(Resource.ip_address.type).__name__}")

print("\nOther types: ColorType, PhoneNumberType, TimezoneType")
print("Use case: Type-safe fields with validation")

print("\nExample 5 completed!")
```
