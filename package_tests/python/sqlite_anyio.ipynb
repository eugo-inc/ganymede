## Example 1: sqlite_anyio functionality test 1

```python
# Example 1: Async Connection
import sqlite_anyio

print("=== Async SQLite Connection ===")

# Connect to in-memory database
async def connect_example():
    conn = await sqlite_anyio.connect(":memory:")
    print(f"Connection type: {type(conn).__name__}")
    print(f"Database: in-memory SQLite")
    await conn.close()
    return True

# Run async function (Jupyter supports top-level await)
result = await connect_example()

print(f"Connected successfully: {result}")
print(f"Use case: Async SQLite operations with AnyIO")

print("\nExample 1 completed!")
```

## Example 2: sqlite_anyio functionality test 2

```python
# Example 2: Async Execute and Fetch
import sqlite_anyio

print("=== Async Execute and Fetch ===")

async def execute_example():
    conn = await sqlite_anyio.connect(":memory:")
    
    # Create table
    cursor = await conn.cursor()
    await cursor.execute("CREATE TABLE users (id INTEGER, name TEXT)")
    
    # Insert data
    await cursor.execute("INSERT INTO users VALUES (1, 'Alice')")
    await cursor.execute("INSERT INTO users VALUES (2, 'Bob')")
    await conn.commit()
    
    # Fetch data
    await cursor.execute("SELECT * FROM users")
    rows = await cursor.fetchall()
    await conn.close()
    return rows

rows = await execute_example()
print(f"Fetched {len(rows)} rows: {rows}")

print("\nExample 2 completed!")
```

## Example 3: sqlite_anyio functionality test 3

```python
# Example 3: Async Fetch Multiple Rows
import sqlite_anyio

print("=== Async Fetch Multiple Rows ===")

async def fetch_example():
    conn = await sqlite_anyio.connect(":memory:")
    cursor = await conn.cursor()
    await cursor.execute("CREATE TABLE items (id INTEGER, value TEXT)")
    await cursor.executemany(
        "INSERT INTO items VALUES (?, ?)",
        [(1, 'A'), (2, 'B'), (3, 'C')]
    )
    
    # Fetch all results
    await cursor.execute("SELECT * FROM items")
    results = await cursor.fetchall()
    await conn.close()
    return results

items = await fetch_example()
print(f"Items: {items}")

print("\nExample 3 completed!")
```

## Example 4: sqlite_anyio functionality test 4

```python
# Example 4: Async Transactions
import sqlite_anyio

print("=== Async Transactions ===")

async def transaction_example():
    conn = await sqlite_anyio.connect(":memory:")
    cursor = await conn.cursor()
    await cursor.execute("CREATE TABLE accounts (id INTEGER, balance INTEGER)")
    await cursor.execute("INSERT INTO accounts VALUES (1, 100)")
    await conn.commit()
    
    # Transaction: transfer money
    await cursor.execute("UPDATE accounts SET balance = balance - 50 WHERE id = 1")
    await conn.commit()
    
    await cursor.execute("SELECT balance FROM accounts WHERE id = 1")
    row = await cursor.fetchone()
    await conn.close()
    return row[0]

balance = await transaction_example()
print(f"Final balance: {balance}")

print("\nExample 4 completed!")
```

## Example 5: sqlite_anyio functionality test 5

```python
# Example 5: Async Execute Many
import sqlite_anyio

print("=== Async Execute Many ===")

async def executemany_example():
    conn = await sqlite_anyio.connect(":memory:")
    cursor = await conn.cursor()
    await cursor.execute("CREATE TABLE products (id INTEGER, name TEXT, price REAL)")
    
    # Bulk insert
    products = [
        (1, 'Apple', 0.5),
        (2, 'Banana', 0.3),
        (3, 'Orange', 0.6)
    ]
    await cursor.executemany("INSERT INTO products VALUES (?, ?, ?)", products)
    await conn.commit()
    
    await cursor.execute("SELECT COUNT(*) FROM products")
    count = await cursor.fetchone()
    await conn.close()
    return count[0]

count = await executemany_example()
print(f"Inserted {count} products")

print("\nExample 5 completed!")
```
