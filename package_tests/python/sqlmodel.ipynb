## Example 1: Define SQLModel Class

```python
# Example 1: Define SQLModel Class
from sqlmodel import SQLModel, Field
from typing import Optional

print("=== SQLModel Class Definition ===")

class Hero(SQLModel, table=True):
    id: Optional[int] = Field(default=None, primary_key=True)
    name: str
    secret_name: str
    age: Optional[int] = None

# SQLModel combines Pydantic and SQLAlchemy
hero = Hero(name="Deadpond", secret_name="Dive Wilson", age=None)

print(f"Hero: {hero.name}")
print(f"Secret name: {hero.secret_name}")
print(f"Type: {type(hero).__name__}")
print(f"Is Pydantic model: {isinstance(hero, SQLModel)}")

print("\nExample 1 completed!")
```

## Example 2: Create Engine and Tables

```python
# Example 2: Create Engine and Tables
from sqlmodel import SQLModel, Field, create_engine
from typing import Optional

print("=== Create Database Tables ===")

class User(SQLModel, table=True):
    id: Optional[int] = Field(default=None, primary_key=True)
    username: str
    email: str

# Create in-memory SQLite database
engine = create_engine("sqlite:///:memory:")

# Create all tables
SQLModel.metadata.create_all(engine)

print(f"Engine: {engine.url}")
print(f"Tables created: {list(SQLModel.metadata.tables.keys())}")
print(f"User table: {User.__tablename__}")

print("\nExample 2 completed!")
```

## Example 3: Session and CRUD Operations

```python
# Example 3: Session and CRUD Operations
from sqlmodel import SQLModel, Field, create_engine, Session
from typing import Optional

print("=== Session and CRUD ===")

class Product(SQLModel, table=True):
    id: Optional[int] = Field(default=None, primary_key=True)
    name: str
    price: float

engine = create_engine("sqlite:///:memory:")
SQLModel.metadata.create_all(engine)

# Create and add products
with Session(engine) as session:
    product = Product(name="Laptop", price=999.99)
    session.add(product)
    session.commit()
    
    print(f"Added product: {product.name}")
    print(f"Product ID: {product.id}")

print("\nExample 3 completed!")
```

## Example 4: Query with Select

```python
# Example 4: Query with Select
from sqlmodel import SQLModel, Field, create_engine, Session, select
from typing import Optional

print("=== Query with Select ===")

class Item(SQLModel, table=True):
    id: Optional[int] = Field(default=None, primary_key=True)
    name: str
    quantity: int

engine = create_engine("sqlite:///:memory:")
SQLModel.metadata.create_all(engine)

with Session(engine) as session:
    session.add(Item(name="Apple", quantity=10))
    session.add(Item(name="Banana", quantity=5))
    session.commit()
    
    # Query using select
    statement = select(Item).where(Item.quantity > 5)
    results = session.exec(statement).all()
    
    print(f"Items with quantity > 5: {[item.name for item in results]}")

print("\nExample 4 completed!")
```

## Example 5: Field Validation with Pydantic

```python
# Example 5: Field Validation with Pydantic
from sqlmodel import SQLModel, Field
from typing import Optional

print("=== Field Validation ===")

class Account(SQLModel, table=True):
    id: Optional[int] = Field(default=None, primary_key=True)
    username: str = Field(min_length=3, max_length=20)
    balance: float = Field(ge=0.0)  # Greater than or equal to 0

# Valid account
account = Account(username="john", balance=100.0)
print(f"Valid account: {account.username}, balance: {account.balance}")

# SQLModel provides Pydantic validation
print(f"Username length: {len(account.username)}")
print(f"Balance is valid: {account.balance >= 0}")

print("\nExample 5 completed!")
```
