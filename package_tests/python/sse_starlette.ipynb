## Example 1: EventSourceResponse Basic Usage

```python
# Example 1: EventSourceResponse Basic Usage
from sse_starlette.sse import EventSourceResponse

print("=== EventSourceResponse ===")

# EventSourceResponse for Server-Sent Events
async def event_generator():
    for i in range(3):
        yield {"data": f"Message {i}"}

# Create SSE response
response_class = EventSourceResponse

print(f"Response class: {response_class.__name__}")
print(f"Use case: Server-Sent Events for real-time updates")
print(f"Protocol: SSE (Server-Sent Events)")
print(f"Content-Type: text/event-stream")

print("\nExample 1 completed!")
```

## Example 2: Event with ID and Event Type

```python
# Example 2: Event with ID and Event Type
from sse_starlette.sse import ServerSentEvent

print("=== ServerSentEvent ===")

# Create SSE event with custom fields
event = ServerSentEvent(
    data="User logged in",
    event="user_event",
    id="123",
    retry=5000
)

print(f"Event data: {event.data}")
print(f"Event type: {event.event}")
print(f"Event ID: {event.id}")
print(f"Retry timeout: {event.retry}ms")
print(f"\nUse case: Custom event types with IDs for client tracking")

print("\nExample 2 completed!")
```

## Example 3: Async Event Generator

```python
# Example 3: Async Event Generator
from sse_starlette.sse import EventSourceResponse
import asyncio

print("=== Async Event Generator ===")

async def number_generator():
    """Generate events asynchronously"""
    for i in range(5):
        await asyncio.sleep(0.1)  # Simulate delay
        yield {"data": str(i), "event": "number", "id": str(i)}

# EventSourceResponse accepts async generators
print(f"Generator: async number_generator")
print(f"Yields: 5 events with 0.1s delay")
print(f"Event fields: data, event, id")
print(f"\nUse case: Real-time data streaming to clients")

print("\nExample 3 completed!")
```

## Example 4: Event with JSON Data

```python
# Example 4: Event with JSON Data
from sse_starlette.sse import ServerSentEvent
import json

print("=== Event with JSON Data ===")

# SSE events can send JSON data
data = {"user": "alice", "action": "login", "timestamp": 1234567890}
event = ServerSentEvent(
    data=json.dumps(data),
    event="user_activity"
)

print(f"Event type: {event.event}")
print(f"Data (JSON): {event.data}")
print(f"Parsed: {json.loads(event.data)}")
print(f"\nUse case: Sending structured data via SSE")

print("\nExample 4 completed!")
```

## Example 5: Event Comment Fields

```python
# Example 5: Event Comment Fields
from sse_starlette.sse import ServerSentEvent

print("=== Event with Comments ===")

# SSE events can include comments (for keep-alive)
event = ServerSentEvent(
    data="Processing data",
    event="progress",
    comment="Keep-alive ping"
)

print(f"Event type: {event.event}")
print(f"Event data: {event.data}")
print(f"Comment: {event.comment}")
print(f"\nUse case: Keep-alive messages and debugging")
print(f"Comments are ignored by clients but useful for monitoring")

print("\nExample 5 completed!")
```
