## Example 1: SSEClient Basic Usage

```python
# Example 1: SSEClient Basic Usage
import sseclient
import io

print("=== SSEClient Basic Usage ===")

# Create a mock SSE response (bytes)
mock_response = io.BytesIO(
    b"data: Hello World\n\n"
    b"data: Second message\n\n"
)

# Parse SSE events
client = sseclient.SSEClient(mock_response)
events = list(client.events())

print(f"Client type: {type(client).__name__}")
print(f"Events received: {len(events)}")
print(f"First event data: {events[0].data}")
print(f"Use case: Consume Server-Sent Events from HTTP streams")

print("\nExample 1 completed!")
```

## Example 2: Event Fields (ID, Type, Retry)

```python
# Example 2: Event Fields (ID, Type, Retry)
import sseclient
import io

print("=== Event Fields ===")

# SSE events with all fields
mock_sse = io.BytesIO(
    b"id: 1\n"
    b"event: user_login\n"
    b"retry: 5000\n"
    b"data: User alice logged in\n\n"
    b"id: 2\n"
    b"event: user_logout\n"
    b"data: User alice logged out\n\n"
)

client = sseclient.SSEClient(mock_sse)
for event in client.events():
    print(f"Event ID: {event.id}, Type: {event.event}, Data: {event.data}")

print(f"\nUse case: Track events with IDs and custom types")

print("\nExample 2 completed!")
```

## Example 3: Multi-line Data Events

```python
# Example 3: Multi-line Data Events
import sseclient
import io

print("=== Multi-line Data ===")

# SSE supports multi-line data fields
mock_sse = io.BytesIO(
    b"data: Line 1\n"
    b"data: Line 2\n"
    b"data: Line 3\n\n"
    b"data: Single line\n\n"
)

client = sseclient.SSEClient(mock_sse)
events = list(client.events())

print(f"Event 1 (multi-line): {repr(events[0].data)}")
print(f"Event 2 (single-line): {repr(events[1].data)}")
print(f"\nUse case: Handle multi-line event data")

print("\nExample 3 completed!")
```

## Example 4: Event Iterator

```python
# Example 4: Event Iterator
import sseclient
import io

print("=== Event Iterator ===")

# Create SSE stream
mock_sse = io.BytesIO(
    b"data: Event A\n\n"
    b"data: Event B\n\n"
    b"data: Event C\n\n"
)

# Iterate over events as they arrive
client = sseclient.SSEClient(mock_sse)
count = 0
for event in client.events():
    count += 1
    print(f"Received event {count}: {event.data}")

print(f"\nTotal events: {count}")
print(f"Use case: Stream processing of SSE events")

print("\nExample 4 completed!")
```

## Example 5: Comments and Empty Events

```python
# Example 5: Comments and Empty Events
import sseclient
import io

print("=== Comments and Empty Events ===")

# SSE supports comment lines (ignored) and empty events
mock_sse = io.BytesIO(
    b": This is a comment\n"
    b"data: First event\n\n"
    b": Another comment\n"
    b"data:\n\n"  # Empty data
    b"data: Third event\n\n"
)

client = sseclient.SSEClient(mock_sse)
events = list(client.events())

for i, event in enumerate(events, 1):
    print(f"Event {i}: {repr(event.data)}")

print(f"\nUse case: Handle comments and empty events")

print("\nExample 5 completed!")
```
