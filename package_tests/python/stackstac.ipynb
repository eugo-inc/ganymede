## Example 1: stackstac functionality test 1

```python
# Example 1: Module Information
import stackstac

print("=== stackstac Module Information ===")

# Show module version and main components
print(f"Version: {stackstac.__version__}")
print(f"Main function: stackstac.stack")

# Show available functions
funcs = [f for f in dir(stackstac) if not f.startswith('_') and callable(getattr(stackstac, f))]
print(f"\nAvailable functions ({len(funcs)}):")
for func in funcs[:8]:
    print(f"  - {func}")

print(f"\nUse case: Convert STAC items to xarray DataArrays")

print("\nExample 1 completed!")
```

## Example 4: stackstac functionality test 4

```python
# Example 4: Show Function
import stackstac

print("=== Show Function ===")

# Access show function for visualization
print(f"stackstac.show function: {type(stackstac.show)}")
print(f"Purpose: Visualize stacked imagery")

# Show function details
import inspect
sig = inspect.signature(stackstac.show)
params = list(sig.parameters.keys())

print(f"\nShow parameters:")
for param in params[:6]:
    print(f"  - {param}")

print(f"\nUse case: Display satellite imagery arrays")

print("\nExample 4 completed!")
```
