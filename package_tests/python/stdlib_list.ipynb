```python
# Example 1: Basic Standard Library Module Detection
import stdlib_list
import sys

# Get standard library modules for current Python version
current_version = f"{sys.version_info.major}.{sys.version_info.minor}"
stdlib_modules = stdlib_list.stdlib_list(current_version)

print(f"Python version: {current_version}")
print(f"Total standard library modules: {len(stdlib_modules)}")
print(f"First 10 stdlib modules: {sorted(stdlib_modules)[:10]}")

# Test specific module detection
test_modules = ['os', 'sys', 'json', 'requests', 'numpy']
for module in test_modules:
    is_stdlib = module in stdlib_modules
    print(f"{module}: {'✓ Standard Library' if is_stdlib else '✗ Third-party'}")

print("Basic standard library module detection complete")
```

```python
# Example 2: Cross-Version Standard Library Comparison
import stdlib_list

# Compare stdlib modules across different Python versions
versions = ['3.8', '3.9', '3.10', '3.11']
version_modules = {}

for version in versions:
    try:
        modules = stdlib_list.stdlib_list(version)
        version_modules[version] = set(modules)
        print(f"Python {version}: {len(modules)} modules")
    except ValueError:
        print(f"Python {version}: Version not supported")

# Find modules added in newer versions
if len(version_modules) >= 2:
    versions_list = sorted(version_modules.keys())
    oldest = versions_list[0]
    newest = versions_list[-1]
    
    added_modules = version_modules[newest] - version_modules[oldest]
    removed_modules = version_modules[oldest] - version_modules[newest]
    
    print(f"\nModules added from {oldest} to {newest}: {len(added_modules)}")
    if added_modules:
        print(f"Examples: {sorted(list(added_modules))[:5]}")
    
    print(f"Modules removed from {oldest} to {newest}: {len(removed_modules)}")
    if removed_modules:
        print(f"Examples: {sorted(list(removed_modules))[:5]}")

print("Cross-version standard library comparison complete")
```

```python
# Example 3: Import Analysis and Dependency Validation
import stdlib_list
import sys
import ast
import textwrap

# Sample code for analysis
sample_code = textwrap.dedent("""
import os
import sys
import json
import requests
import numpy as np
from collections import defaultdict
from urllib.parse import urlparse
from datetime import datetime
""")

def analyze_imports(code_string):
    """Extract and categorize imports from Python code"""
    tree = ast.parse(code_string)
    imports = set()
    
    for node in ast.walk(tree):
        if isinstance(node, ast.Import):
            for alias in node.names:
                imports.add(alias.name.split('.')[0])
        elif isinstance(node, ast.ImportFrom):
            if node.module:
                imports.add(node.module.split('.')[0])
    
    return imports

# Analyze the sample code
current_version = f"{sys.version_info.major}.{sys.version_info.minor}"
stdlib_modules = set(stdlib_list.stdlib_list(current_version))
found_imports = analyze_imports(sample_code)

stdlib_imports = found_imports & stdlib_modules
third_party_imports = found_imports - stdlib_modules

print("Code Analysis Results:")
print(f"Total imports found: {len(found_imports)}")
print(f"Standard library imports: {sorted(stdlib_imports)}")
print(f"Third-party imports: {sorted(third_party_imports)}")

# Dependency report
print(f"\nDependency Summary:")
print(f"✓ {len(stdlib_imports)} standard library dependencies")
print(f"⚠ {len(third_party_imports)} external dependencies requiring installation")

print("Import analysis and dependency validation complete")
```

```python
# Example 4: Module Category Analysis and Statistics
import stdlib_list
import sys
from collections import Counter

def categorize_stdlib_modules(modules):
    """Categorize standard library modules by common patterns"""
    categories = {
        'File/IO': ['os', 'io', 'pathlib', 'tempfile', 'shutil', 'glob', 'filecmp'],
        'Network': ['urllib', 'http', 'socket', 'ssl', 'email', 'smtplib', 'ftplib'],
        'Data': ['json', 'csv', 'pickle', 'xml', 'sqlite3', 'dbm'],
        'Text': ['re', 'string', 'textwrap', 'unicodedata', 'codecs'],
        'Math/Science': ['math', 'statistics', 'random', 'decimal', 'fractions'],
        'Time/Date': ['time', 'datetime', 'calendar'],
        'Concurrency': ['threading', 'multiprocessing', 'concurrent', 'asyncio'],
        'System': ['sys', 'platform', 'ctypes', 'signal', 'subprocess'],
        'Development': ['unittest', 'doctest', 'pdb', 'profile', 'trace']
    }
    
    categorized = {cat: [] for cat in categories}
    uncategorized = []
    
    for module in modules:
        found_category = False
        for category, patterns in categories.items():
            if any(pattern in module for pattern in patterns):
                categorized[category].append(module)
                found_category = True
                break
        if not found_category:
            uncategorized.append(module)
    
    return categorized, uncategorized

# Analyze current Python version
current_version = f"{sys.version_info.major}.{sys.version_info.minor}"
stdlib_modules = stdlib_list.stdlib_list(current_version)
categorized, uncategorized = categorize_stdlib_modules(stdlib_modules)

print(f"Standard Library Analysis for Python {current_version}:")
print(f"Total modules: {len(stdlib_modules)}")

for category, modules in categorized.items():
    if modules:
        print(f"\n{category} ({len(modules)} modules):")
        print(f"  Examples: {sorted(modules)[:5]}")

print(f"\nUncategorized modules: {len(uncategorized)}")
if uncategorized:
    print(f"Examples: {sorted(uncategorized)[:10]}")

# Module name statistics
name_lengths = [len(name) for name in stdlib_modules]
print(f"\nModule Name Statistics:")
print(f"Average name length: {sum(name_lengths) / len(name_lengths):.1f} characters")
print(f"Shortest names: {sorted(stdlib_modules, key=len)[:5]}")
print(f"Longest names: {sorted(stdlib_modules, key=len, reverse=True)[:5]}")

print("Module category analysis and statistics complete")
```

```python
# Example 5: Production Environment Auditing and Validation
import stdlib_list
import sys
import importlib
from collections import defaultdict

class EnvironmentAuditor:
    """Production-ready environment auditing using stdlib-list"""
    
    def __init__(self):
        self.python_version = f"{sys.version_info.major}.{sys.version_info.minor}"
        try:
            self.stdlib_modules = set(stdlib_list.stdlib_list(self.python_version))
        except ValueError:
            # Fallback to a known version if current version not supported
            self.stdlib_modules = set(stdlib_list.stdlib_list('3.9'))
            print(f"Warning: Using Python 3.9 stdlib list as fallback")
    
    def audit_imports(self, module_names):
        """Audit a list of module names for stdlib vs third-party classification"""
        results = {
            'stdlib': [],
            'third_party': [],
            'unavailable': [],
            'importable': [],
            'not_importable': []
        }
        
        for module_name in module_names:
            # Check if it's stdlib
            if module_name in self.stdlib_modules:
                results['stdlib'].append(module_name)
            else:
                results['third_party'].append(module_name)
            
            # Check if it's importable
            try:
                importlib.import_module(module_name)
                results['importable'].append(module_name)
            except ImportError:
                results['not_importable'].append(module_name)
        
        return results
    
    def generate_report(self, audit_results):
        """Generate a comprehensive audit report"""
        total_modules = len(set(
            audit_results['stdlib'] + audit_results['third_party']
        ))
        
        report = []
        report.append(f"Environment Audit Report (Python {self.python_version})")
        report.append("=" * 50)
        report.append(f"Total modules analyzed: {total_modules}")
        report.append(f"Standard library modules: {len(audit_results['stdlib'])}")
        report.append(f"Third-party modules: {len(audit_results['third_party'])}")
        report.append(f"Importable modules: {len(audit_results['importable'])}")
        report.append(f"Missing modules: {len(audit_results['not_importable'])}")
        
        if audit_results['not_importable']:
            report.append("\nMissing Dependencies:")
            for module in sorted(audit_results['not_importable']):
                module_type = "stdlib" if module in self.stdlib_modules else "third-party"
                report.append(f"  - {module} ({module_type})")
        
        return "\n".join(report)

# Demonstrate production environment auditing
auditor = EnvironmentAuditor()

# Sample modules to audit (mix of stdlib and third-party)
test_modules = [
    'os', 'sys', 'json', 'csv', 'sqlite3',  # stdlib
    'requests', 'numpy', 'pandas', 'flask', 'django',  # third-party
    'fake_module', 'another_fake_module'  # non-existent
]

audit_results = auditor.audit_imports(test_modules)
report = auditor.generate_report(audit_results)

print(report)

# Summary statistics
stdlib_available = len([m for m in audit_results['stdlib'] if m in audit_results['importable']])
stdlib_missing = len([m for m in audit_results['stdlib'] if m in audit_results['not_importable']])

print(f"\nEnvironment Health Score:")
if audit_results['stdlib']:
    stdlib_health = (stdlib_available / len(audit_results['stdlib'])) * 100
    print(f"Standard library completeness: {stdlib_health:.1f}%")

total_health = (len(audit_results['importable']) / len(test_modules)) * 100
print(f"Overall module availability: {total_health:.1f}%")

print("Production environment auditing and validation complete")
```
