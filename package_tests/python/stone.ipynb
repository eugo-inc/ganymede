## Example 1: stone functionality test 1

```python
# Example 1: Basic Stone Types
from stone.ir import *

print("=== Basic Stone Types ===")

# Primitive types
string_type = String()
int_type = Int32()
bool_type = Boolean()

print(f"String type: {string_type}")
print(f"Int32 type: {int_type}")
print(f"Boolean type: {bool_type}")

# Type checking
print(f"\nType checks:")
print(f"is_string_type: {is_string_type(string_type)}")
print(f"is_integer_type: {is_integer_type(int_type)}")
print(f"is_boolean_type: {is_boolean_type(bool_type)}")

print(f"\nUse case: Define API data types")

print("\nExample 1 completed!")
```

## Example 2: stone functionality test 2

```python
# Example 2: Collection Types
from stone.ir import *

print("=== Collection Types ===")

# List and Map types
list_strings = List(String())
map_type = Map(String(), Int32())
nullable_int = Nullable(Int32())

print(f"List(String): {list_strings}")
print(f"Map(String, Int32): {map_type}")
print(f"Nullable(Int32): {nullable_int}")

# Type checks
print(f"\nType checks:")
print(f"is_list_type: {is_list_type(list_strings)}")
print(f"is_map_type: {is_map_type(map_type)}")
print(f"is_nullable_type: {is_nullable_type(nullable_int)}")

print(f"\nUse case: Complex data structures")

print("\nExample 2 completed!")
```

## Example 3: stone functionality test 3

```python
# Example 3: Struct Information
from stone.ir import *
import inspect

print("=== Struct Information ===")

# Show Struct class
print(f"Struct class: {Struct}")

# Show Struct signature
sig = inspect.signature(Struct.__init__)
print(f"\nStruct.__init__ signature: {sig}")

# Show StructField signature
field_sig = inspect.signature(StructField.__init__)
print(f"StructField.__init__ signature: {field_sig}")

# Basic types
print(f"\nBasic primitive types:")
print(f"  String: {String()}")
print(f"  Int32: {Int32()}")
print(f"  Boolean: {Boolean()}")

print(f"\nUse case: Understand Stone IR structure")

print("\nExample 3 completed!")
```

## Example 4: stone functionality test 4

```python
# Example 4: Type Aliases
from stone.ir import *

print("=== Type Aliases ===")

# Create type alias (name, data_type, doc)
user_id_alias = Alias('UserId', String(), 'Unique user identifier')

print(f"Alias name: {user_id_alias.name}")
print(f"Alias doc: {user_id_alias.doc}")
print(f"Underlying type: {type(user_id_alias.data_type).__name__}")

# Type checking
print(f"\nType checks:")
print(f"is_alias: {is_alias(user_id_alias)}")
print(f"is_string_type: {is_string_type(user_id_alias.data_type)}")

print(f"\nUse case: Create reusable type definitions")

print("\nExample 4 completed!")
```

## Example 5: stone functionality test 5

```python
# Example 5: Type Utilities
from stone.ir import *

print("=== Type Utilities ===")

# Composite types
list_type = List(String())
nullable_type = Nullable(Int32())

# Unwrap nullable
unwrapped = unwrap_nullable(nullable_type)
print(f"Nullable(Int32) unwrapped: {type(unwrapped).__name__}")

# Type checks
print(f"\nComposite type checks:")
print(f"is_composite_type(List): {is_composite_type(list_type)}")
print(f"is_composite_type(Nullable): {is_composite_type(nullable_type)}")
print(f"is_primitive_type(String): {is_primitive_type(String())}")

# Numeric types
print(f"\nNumeric type checks:")
print(f"is_numeric_type(Int32): {is_numeric_type(Int32())}")
print(f"is_numeric_type(Float64): {is_numeric_type(Float64())}")

print(f"\nUse case: Type introspection and validation")

print("\nExample 5 completed!")
```
