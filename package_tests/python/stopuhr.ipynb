## Example 1: Module Inspection

```python
# Example 1: Module Inspection
import stopuhr

print("=== Stopuhr Module ===")

# Show what's available
print(f"Module: {stopuhr.__name__}")

# List available attributes
attrs = [a for a in dir(stopuhr) if not a.startswith('_')]
print(f"\nAvailable items ({len(attrs)}):")
for attr in attrs[:10]:
    obj = getattr(stopuhr, attr)
    print(f"  {attr}: {type(obj).__name__}")

print(f"\nUse case: Timing and stopwatch utilities")

print("\nExample 1 completed!")
```

## Example 2: Stopwatch Class

```python
# Example 2: Stopwatch Class
import stopuhr

print("=== Stopwatch Class ===")

# Access stopwatch if available
if hasattr(stopuhr, 'stopwatch'):
    sw_class = stopuhr.stopwatch
    print(f"Stopwatch class: {sw_class}")
    print(f"Type: {type(sw_class).__name__}")

# Check for other timing utilities
for name in ['timer', 'clock', 'Stopwatch', 'Clock']:
    if hasattr(stopuhr, name):
        print(f"{name}: {type(getattr(stopuhr, name)).__name__}")

print(f"\nUse case: Access timing classes")

print("\nExample 2 completed!")
```

## Example 3: Module Functions

```python
# Example 3: Module Functions
import stopuhr
import inspect

print("=== Module Functions ===")

# Find callable functions
funcs = [name for name in dir(stopuhr) if callable(getattr(stopuhr, name)) and not name.startswith('_')]

print(f"Functions in stopuhr ({len(funcs)}):")
for func in funcs[:8]:
    sig = inspect.signature(getattr(stopuhr, func)) if inspect.isfunction(getattr(stopuhr, func)) else None
    if sig:
        print(f"  {func}{sig}")
    else:
        print(f"  {func}")

print(f"\nUse case: Explore timing functions")

print("\nExample 3 completed!")
```

## Example 4: Module Metadata

```python
# Example 4: Module Metadata
import stopuhr

print("=== Module Metadata ===")

# Module information
print(f"Module name: {stopuhr.__name__}")

if hasattr(stopuhr, '__version__'):
    print(f"Version: {stopuhr.__version__}")

if hasattr(stopuhr, '__doc__'):
    doc = stopuhr.__doc__
    if doc:
        print(f"Documentation: {doc[:100]}...")

if hasattr(stopuhr, '__file__'):
    print(f"File: {stopuhr.__file__.split('/')[-1]}")

print(f"\nUse case: Module information")

print("\nExample 4 completed!")
```

## Example 5: Package Structure

```python
# Example 5: Package Structure
import stopuhr

print("=== Package Structure ===")

# Check if it's a package
if hasattr(stopuhr, '__path__'):
    print(f"Package path: {stopuhr.__path__}")

# List all attributes
all_attrs = dir(stopuhr)
public_attrs = [a for a in all_attrs if not a.startswith('_')]

print(f"\nTotal attributes: {len(all_attrs)}")
print(f"Public attributes: {len(public_attrs)}")

print(f"\nPublic items:")
for attr in public_attrs[:10]:
    print(f"  - {attr}")

print(f"\nUse case: Understand package organization")

print("\nExample 5 completed!")
```
