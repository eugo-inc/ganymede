# sympy: 5 Basic Usage Examples

```python
# Example 1: Symbolic computation
import sympy as sp
x = sp.symbols('x')
expr = x**2 + 2*x + 1
print(sp.expand(expr))
```

```python
# Example 2: Solving equations
import sympy as sp
x = sp.symbols('x')
solution = sp.solve(x**2 - 4, x)
print(solution)
```

```python
# Example 3: Calculus (derivative and integral)
import sympy as sp
x = sp.symbols('x')
f = sp.sin(x)
print('Derivative:', sp.diff(f, x))
print('Integral:', sp.integrate(f, x))
```

```python
# Example 4: Matrix operations
import sympy as sp
M = sp.Matrix([[1, 2], [3, 4]])
print('Determinant:', M.det())
print('Inverse:', M.inv())
```

```python
# Example 5: Series expansion
import sympy as sp
x = sp.symbols('x')
series = sp.series(sp.exp(x), x, 0, 5)
print(series)
```
