## Example 1: Simple Moving Average

```python
# Example 1: Simple Moving Average
import talib
import numpy as np

print("=== Simple Moving Average ===")

# Generate sample price data
prices = np.array([10, 11, 12, 11, 10, 12, 13, 14, 13, 12], dtype=float)

# Calculate SMA
sma = talib.SMA(prices, timeperiod=3)

print(f"Prices: {prices}")
print(f"SMA(3): {sma}")
print(f"Non-NaN values: {sma[~np.isnan(sma)]}")

print(f"\nUse case: Calculate moving averages")

print("\nExample 1 completed!")
```

## Example 2: RSI (Relative Strength Index)

```python
# Example 2: RSI (Relative Strength Index)
import talib
import numpy as np

print("=== RSI Indicator ===")

# Generate sample price data with trend
prices = np.array([100, 102, 104, 103, 105, 107, 106, 108, 110, 109, 111, 113, 112, 114], dtype=float)

# Calculate RSI
rsi = talib.RSI(prices, timeperiod=14)

print(f"Prices: {prices}")
print(f"RSI(14): {rsi}")
print(f"Latest RSI: {rsi[-1]:.2f}")

# Interpret RSI
if rsi[-1] > 70:
    print("Signal: Overbought")
elif rsi[-1] < 30:
    print("Signal: Oversold")
else:
    print("Signal: Neutral")

print(f"\nUse case: Identify overbought/oversold conditions")

print("\nExample 2 completed!")
```

## Example 3: MACD (Moving Average Convergence Divergence)

```python
# Example 3: MACD (Moving Average Convergence Divergence)
import talib
import numpy as np

print("=== MACD Indicator ===")

# Generate sample price data
prices = np.array([100, 101, 102, 101, 103, 104, 105, 104, 106, 107, 108, 107, 109, 110, 111, 110, 112, 113, 114, 113, 115, 116, 117, 116, 118, 119, 120], dtype=float)

# Calculate MACD
macd, macd_signal, macd_hist = talib.MACD(prices, fastperiod=12, slowperiod=26, signalperiod=9)

print(f"Price data points: {len(prices)}")
print(f"Latest price: ${prices[-1]:.2f}")
print(f"MACD: {macd[-1]:.4f}")
print(f"Signal: {macd_signal[-1]:.4f}")
print(f"Histogram: {macd_hist[-1]:.4f}")

# Interpret MACD
if macd_hist[-1] > 0:
    print("Signal: Bullish (MACD above signal)")
else:
    print("Signal: Bearish (MACD below signal)")

print(f"\nUse case: Trend momentum analysis")

print("\nExample 3 completed!")
```

## Example 4: Bollinger Bands

```python
# Example 4: Bollinger Bands
import talib
import numpy as np

print("=== Bollinger Bands ===")

# Generate sample price data
prices = np.array([100, 102, 101, 103, 105, 104, 106, 108, 107, 109, 111, 110, 112, 114, 113, 115, 117, 116, 118, 120], dtype=float)

# Calculate Bollinger Bands
upper, middle, lower = talib.BBANDS(prices, timeperiod=5, nbdevup=2, nbdevdn=2, matype=0)

print(f"Latest price: ${prices[-1]:.2f}")
print(f"Upper band: ${upper[-1]:.2f}")
print(f"Middle band: ${middle[-1]:.2f}")
print(f"Lower band: ${lower[-1]:.2f}")

# Calculate bandwidth
bandwidth = upper[-1] - lower[-1]
print(f"Bandwidth: ${bandwidth:.2f}")

# Interpret position
if prices[-1] > upper[-1]:
    print("Signal: Price above upper band (overbought)")
elif prices[-1] < lower[-1]:
    print("Signal: Price below lower band (oversold)")
else:
    print("Signal: Price within bands")

print(f"\nUse case: Volatility and price range analysis")

print("\nExample 4 completed!")
```

## Example 5: ATR (Average True Range)

```python
# Example 5: ATR (Average True Range)
import talib
import numpy as np

print("=== ATR (Average True Range) ===")

# Generate sample OHLC data
high = np.array([102, 104, 106, 105, 107, 109, 108, 110, 112, 111, 113, 115, 114, 116], dtype=float)
low = np.array([98, 100, 102, 101, 103, 105, 104, 106, 108, 107, 109, 111, 110, 112], dtype=float)
close = np.array([100, 102, 104, 103, 105, 107, 106, 108, 110, 109, 111, 113, 112, 114], dtype=float)

# Calculate ATR
atr = talib.ATR(high, low, close, timeperiod=14)

print(f"Data points: {len(close)}")
print(f"Latest close: ${close[-1]:.2f}")
print(f"Latest high: ${high[-1]:.2f}")
print(f"Latest low: ${low[-1]:.2f}")
print(f"ATR(14): {atr[-1]:.2f}")

# Calculate volatility percentage
volatility_pct = (atr[-1] / close[-1]) * 100
print(f"Volatility: {volatility_pct:.2f}%")

print(f"\nUse case: Measure market volatility")

print("\nExample 5 completed!")
```

## Example 6: TA-Lib 0.8.1 Defaults, Streams and Math Transforms (asserted)

```python
import numpy as np
import talib

def same(a, b):
    return np.array_equal(a, b, equal_nan=True)

x = np.random.default_rng(0).normal(100, 5, 200)
default_bands, bands_20, bands_5 = talib.BBANDS(x), talib.BBANDS(x, timeperiod=20), talib.BBANDS(x, timeperiod=5)
assert all(map(same, default_bands, bands_20)), "BBANDS() no longer defaults to timeperiod=20"
assert not same(default_bands[1], bands_5[1]), "BBANDS() default is indistinguishable from timeperiod=5"
for name in ("PPO", "APO"):
    fn = getattr(talib, name)
    assert same(fn(x), fn(x, matype=talib.MA_Type.EMA)), f"{name}() no longer defaults to matype=EMA"
    assert not same(fn(x), fn(x, matype=talib.MA_Type.SMA)), f"{name}() default is indistinguishable from SMA"
handle, last = talib.stream.SMA(x, timeperiod=3), talib.SMA(x, timeperiod=3)[-1]
assert not isinstance(handle, (float, np.floating)), f"talib.stream.SMA returned a bare value, not a handle: {handle!r}"
assert np.isclose(handle.value, last, rtol=1e-12), f"stream SMA {handle.value!r} != function SMA {last!r}"
assert np.isclose(handle.update(101.0), (x[-2] + x[-1] + 101.0) / 3), "SMA stream update() drifted"
# The math transforms agree with numpy.
assert same(talib.SQRT(np.array([4.0, 9.0])), [2.0, 3.0]), talib.SQRT(np.array([4.0, 9.0]))
v = np.array([0.25, 0.5, 1.0])
for name, ref in (("LN", np.log), ("EXP", np.exp), ("COS", np.cos)):
    assert np.allclose(getattr(talib, name)(v), ref(v)), f"talib.{name} disagrees with numpy.{ref.__name__}"
print("Example 6: BBANDS/PPO/APO defaults, the SMA stream handle and the libm transforms all hold")
```
