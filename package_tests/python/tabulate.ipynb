## Example 1: tabulate functionality test 1

```python
import tabulate

# Example 1: Basic table formatting with different data types
data = [
    ["John", 25, "Engineer", 75000],
    ["Alice", 30, "Designer", 65000],
    ["Bob", 35, "Manager", 85000]
]
headers = ["Name", "Age", "Role", "Salary"]

print("Basic table formatting:")
print(tabulate.tabulate(data, headers=headers))

print("\nWith grid format:")
print(tabulate.tabulate(data, headers=headers, tablefmt="grid"))

print(f"Available table formats: {len(tabulate.tabulate_formats)}")
print(f"Sample formats: {list(tabulate.tabulate_formats)[:5]}")
```

## Example 2: tabulate functionality test 2

```python
import tabulate

# Example 2: Multiple table formats and styling options
data = [
    ["Product", "Price", "Stock", "Category"],
    ["Laptop", 999.99, 50, "Electronics"],
    ["Book", 29.99, 200, "Education"],
    ["Chair", 149.50, 75, "Furniture"]
]

print("GitHub markdown format:")
print(tabulate.tabulate(data[1:], headers=data[0], tablefmt="github"))

print("\nHTML format:")
print(tabulate.tabulate(data[1:], headers=data[0], tablefmt="html"))

print("\nLaTeX format:")
print(tabulate.tabulate(data[1:], headers=data[0], tablefmt="latex"))

print(f"Total available formats: {len(tabulate.tabulate_formats)}")
```

## Example 3: tabulate functionality test 3

```python
import tabulate

# Example 3: Advanced formatting with number formatting and alignment
data = [
    ["Metric", "Q1", "Q2", "Q3", "Q4"],
    ["Revenue ($M)", 10.5, 12.8, 15.2, 18.7],
    ["Users (K)", 125.6, 147.9, 189.3, 245.8],
    ["Conversion (%)", 3.45, 4.12, 4.89, 5.23]
]

print("Fancy grid with number formatting:")
print(tabulate.tabulate(data[1:], headers=data[0], tablefmt="fancy_grid", floatfmt=".2f"))

print("\nRight-aligned numbers:")
print(tabulate.tabulate(data[1:], headers=data[0], tablefmt="pretty", 
                       colalign=("left", "right", "right", "right", "right")))

print("\nPSQL format (database-like):")
print(tabulate.tabulate(data[1:], headers=data[0], tablefmt="psql"))
```

## Example 4: tabulate functionality test 4

```python
import tabulate

# Example 4: Working with dictionary data and missing values
# Dictionary data (common in data analysis)
dict_data = [
    {"Name": "Alice", "Score": 95, "Grade": "A", "Attended": True},
    {"Name": "Bob", "Score": 87, "Grade": "B", "Attended": True},
    {"Name": "Charlie", "Score": None, "Grade": "F", "Attended": False}
]

print("Dictionary data with missing values:")
print(tabulate.tabulate(dict_data, headers="keys", tablefmt="grid", 
                       missingval="N/A"))

# Mixed data types
mixed_data = [
    ["Item", "Count", "Active", "Last Update"],
    ["Server-1", 42, True, "2024-01-15"],
    ["Server-2", 0, False, None],
    ["Server-3", 156, True, "2024-01-20"]
]

print("\nMixed data types with custom missing value:")
print(tabulate.tabulate(mixed_data[1:], headers=mixed_data[0], 
                       tablefmt="outline", missingval="---"))
```

## Example 5: tabulate functionality test 5

```python
import tabulate

# Example 5: Production patterns and performance comparison
# Large dataset simulation for reporting
import time

performance_data = []
for i in range(100):
    performance_data.append([f"Task-{i:03d}", f"Worker-{i%5}", 
                           round(time.time() % 100, 2), "Completed"])

print("Production report format (first 5 rows):")
print(tabulate.tabulate(performance_data[:5], 
                       headers=["Task ID", "Worker", "Duration", "Status"],
                       tablefmt="rounded_grid"))

# Format comparison for different use cases
sample_data = [["API", "Requests", "Avg Response"], 
               ["GET /users", 1250, 45.2], 
               ["POST /data", 890, 123.8]]

formats = ["simple", "github", "html", "latex_booktabs"]
print(f"\nFormat comparison for documentation:")
for fmt in formats:
    print(f"\n{fmt.upper()} format:")
    print(tabulate.tabulate(sample_data[1:], headers=sample_data[0], 
                           tablefmt=fmt, floatfmt=".1f"))
```
