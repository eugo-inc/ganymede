## Example 1: Basic Unicode to ASCII Conversion

```python
import text_unidecode

# Example 1: Basic Unicode to ASCII transliteration
print("Basic Unicode to ASCII conversion:")

# Test various Unicode characters
test_strings = [
    "café",
    "naïve",
    "résumé",
    "Zürich",
    "München"
]

for original in test_strings:
    converted = text_unidecode.unidecode(original)
    print(f"'{original}' → '{converted}'")

# Show that it preserves ASCII characters
ascii_text = "Hello World 123!"
ascii_result = text_unidecode.unidecode(ascii_text)
print(f"\nASCII preserved: '{ascii_text}' → '{ascii_result}'")

print(f"Unidecode function available: {hasattr(text_unidecode, 'unidecode')}")
print("Converts Unicode characters to closest ASCII representation")
```

## Example 2: Multi-language Text Processing

```python
import text_unidecode

# Example 2: Multi-language text conversion
print("Multi-language Unicode conversion:")

# Test different language scripts
multilingual_texts = [
    ("Russian", "Привет мир!"),
    ("Japanese", "こんにちは世界"),
    ("Arabic", "مرحبا بالعالم"),
    ("Greek", "Γεια σας κόσμε"),
    ("Chinese", "你好世界"),
    ("Hebrew", "שלום עולם"),
    ("Korean", "안녕하세요 세계")
]

for language, original in multilingual_texts:
    converted = text_unidecode.unidecode(original)
    print(f"{language:8}: '{original}' → '{converted}'")

# Mixed script text
mixed_text = "Café de París 北京 москва"
mixed_result = text_unidecode.unidecode(mixed_text)
print(f"\nMixed scripts: '{mixed_text}' → '{mixed_result}'")

print(f"\nHandles multiple Unicode scripts and writing systems")
print("Useful for creating ASCII-safe filenames and URLs")
```

## Example 3: Special Characters and Symbols

```python
import text_unidecode

# Example 3: Special characters and symbols handling
print("Special characters and symbols conversion:")

# Mathematical and technical symbols
symbols = [
    "α β γ δ ε",  # Greek letters
    "∑ ∏ ∫ ∂ ∇",  # Mathematical symbols
    "← → ↑ ↓ ↔",  # Arrows
    "© ® ™ § ¶",   # Copyright and symbols
    "½ ¼ ¾ ⅓ ⅔",  # Fractions
    "€ £ ¥ $ ¢"   # Currency symbols
]

for symbol_group in symbols:
    converted = text_unidecode.unidecode(symbol_group)
    print(f"'{symbol_group}' → '{converted}'")

# Diacritical marks and accents
accented_chars = [
    "àáâãäåæçèéêë",
    "ìíîïñòóôõöø",
    "ùúûüýÿðþß"
]

print(f"\nDiacritical marks:")
for chars in accented_chars:
    converted = text_unidecode.unidecode(chars)
    print(f"'{chars}' → '{converted}'")

print(f"\nPreserves meaning while ensuring ASCII compatibility")
print("Handles symbols, accents, and special Unicode characters")
```

## Example 4: Practical Text Normalization Applications

```python
import text_unidecode
import unicodedata

# Example 4: Practical applications for text processing
print("Practical applications for text normalization:")

# URL-safe filename generation
filenames = [
    "Relatório Anual 2024.pdf",
    "Présentation François.pptx", 
    "データ分析レポート.xlsx",
    "Инструкция по установке.docx"
]

print("URL-safe filename generation:")
for original in filenames:
    safe_name = text_unidecode.unidecode(original).replace(' ', '_')
    print(f"'{original}' → '{safe_name}'")

# Search index normalization
search_terms = [
    "José María",
    "François",
    "Müller",
    "Sánchez",
    "Ñoño"
]

print(f"\nSearch term normalization:")
for term in search_terms:
    normalized = text_unidecode.unidecode(term).lower()
    print(f"'{term}' → '{normalized}' (searchable)")

# Compare with unicodedata normalization
test_word = "café"
unidecode_result = text_unidecode.unidecode(test_word)
nfd_result = unicodedata.normalize('NFD', test_word).encode('ascii', 'ignore').decode('ascii')

print(f"\nComparison with unicodedata:")
print(f"Original: '{test_word}'")
print(f"text_unidecode: '{unidecode_result}'")
print(f"unicodedata NFD: '{nfd_result}'")
print("text_unidecode provides more readable transliterations")
```

## Example 5: Production Patterns and Batch Processing

```python
import text_unidecode
import re

# Example 5: Production patterns and batch processing
class TextNormalizer:
    """Production text normalization service"""
    
    def __init__(self):
        self.stats = {"processed": 0, "characters_converted": 0}
    
    def normalize_text(self, text):
        """Normalize text to ASCII with statistics"""
        original_len = len(text)
        normalized = text_unidecode.unidecode(text)
        
        # Count non-ASCII characters converted
        non_ascii_count = sum(1 for char in text if ord(char) > 127)
        
        self.stats["processed"] += 1
        self.stats["characters_converted"] += non_ascii_count
        
        return normalized
    
    def normalize_batch(self, text_list):
        """Process multiple texts efficiently"""
        results = []
        for text in text_list:
            normalized = self.normalize_text(text)
            results.append({
                'original': text,
                'normalized': normalized,
                'length_change': len(text) - len(normalized)
            })
        return results
    
    def create_slug(self, text):
        """Create URL-safe slug from text"""
        # Normalize to ASCII
        normalized = text_unidecode.unidecode(text)
        
        # Convert to lowercase and replace non-alphanumeric with hyphens
        slug = re.sub(r'[^a-zA-Z0-9]+', '-', normalized.lower())
        
        # Remove leading/trailing hyphens
        return slug.strip('-')

print("Production text normalization patterns:")

# Create normalizer service
normalizer = TextNormalizer()

# Batch processing example
international_content = [
    "Café de la Paix",
    "北京烤鸭餐厅",
    "Москва - столица России",
    "François Müller & Sánchez Ltd.",
    "العالم العربي الجميل"
]

results = normalizer.normalize_batch(international_content)

print("Batch normalization results:")
for result in results:
    print(f"'{result['original'][:20]}...' → '{result['normalized'][:20]}...'")

# URL slug generation
print(f"\nURL slug generation:")
titles = ["My Great Article!", "François & José's Café", "北京旅行ガイド"]
for title in titles:
    slug = normalizer.create_slug(title)
    print(f"'{title}' → '{slug}'")

print(f"\nProcessing statistics:")
print(f"Total texts processed: {normalizer.stats['processed']}")
print(f"Characters converted: {normalizer.stats['characters_converted']}")

print(f"\nProduction use cases:")
print("- Database indexing and search")
print("- URL generation and routing")
print("- File naming and storage")
print("- API parameter normalization")
```
