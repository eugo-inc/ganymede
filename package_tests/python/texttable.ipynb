```python
# Example 1: Basic table construction with headers and rows.
from texttable import Texttable

t = Texttable()
t.add_rows([
    ["Name", "Age", "Role"],
    ["Alice", 30, "Engineer"],
    ["Bob", 25, "Designer"],
    ["Carol", 42, "Manager"],
])
rendered = t.draw()
print(rendered)

assert "Alice" in rendered
assert "Engineer" in rendered
assert "Name" in rendered
```

```python
# Example 2: Column alignment + data type control.
from texttable import Texttable

t = Texttable()
t.set_cols_align(["l", "r", "c"])
t.set_cols_dtype(["t", "i", "f"])  # text, int, float
t.add_rows([
    ["Item", "Count", "Price"],
    ["Apple", 10, 0.5],
    ["Banana", 25, 0.25],
    ["Cherry", 100, 0.05],
])
rendered = t.draw()
print(rendered)

assert "Apple" in rendered
assert "0.500" in rendered or "0.5" in rendered
```

```python
# Example 3: Custom column widths and max table width.
from texttable import Texttable

t = Texttable(max_width=80)
t.set_cols_width([10, 15, 20])
t.add_rows([
    ["Short", "Medium length", "This is a longer description"],
    ["foo", "bar baz", "the quick brown fox jumps"],
])
rendered = t.draw()
print(rendered)

# Every rendered line should respect max_width.
for line in rendered.split("\n"):
    assert len(line) <= 80, f"line exceeds 80 chars: {len(line)}"
assert "foo" in rendered
```

```python
# Example 4: Deco flags to toggle borders, headers, vlines, hlines.
from texttable import Texttable

t = Texttable()
t.set_deco(Texttable.HEADER | Texttable.VLINES)
t.add_rows([
    ["Col A", "Col B"],
    ["1", "2"],
    ["3", "4"],
])
rendered = t.draw()
print(rendered)

# With BORDER disabled, the rendered output should not start with a '+' corner.
assert not rendered.startswith("+--")
assert "Col A" in rendered
```

```python
# Example 5: Float precision via set_precision, and header() / add_row() incremental API.
from texttable import Texttable

t = Texttable()
t.set_precision(2)
t.set_cols_dtype(["t", "f"])
t.header(["Metric", "Value"])
t.add_row(["pi", 3.14159265])
t.add_row(["e", 2.71828182])
t.add_row(["sqrt2", 1.41421356])
rendered = t.draw()
print(rendered)

assert "3.14" in rendered
assert "2.72" in rendered or "2.71" in rendered
assert "Metric" in rendered
assert "Value" in rendered
print("\nAll texttable smoke checks passed.")
```
