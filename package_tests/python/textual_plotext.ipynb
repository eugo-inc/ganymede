## Example 1: The widget and its plotext handle

```python
from textual_plotext import Plot, PlotextPlot, themes

widget = PlotextPlot()
assert isinstance(widget.plt, Plot)
names = themes()  # the plotext theme names a PlotextPlot can use
assert "dark" in names and "default" in names, names
print(type(widget).__name__, "plt:", type(widget.plt).__name__, "| themes:", len(names))
```

## Example 2: Mount a line plot in a Textual app (headless)

```python
from textual.app import App, ComposeResult
from textual_plotext import PlotextPlot

class LineApp(App):
    def compose(self) -> ComposeResult:
        yield PlotextPlot()
    def on_mount(self) -> None:
        plt = self.query_one(PlotextPlot).plt
        plt.plot([1, 2, 3, 4], [1, 4, 9, 16])
        plt.title("squares")

app = LineApp()
async with app.run_test(size=(60, 20)) as pilot:
    await pilot.pause()
    widget = app.query_one(PlotextPlot)
    assert widget.size.width > 0 and widget.size.height > 0, widget.size
print("rendered", widget.size)
```

## Example 3: Bar chart and scatter in one figure

```python
from textual.app import App
from textual_plotext import PlotextPlot

class BarApp(App):
    def compose(self):
        yield PlotextPlot()

app = BarApp()
async with app.run_test(size=(60, 20)) as pilot:
    plt = app.query_one(PlotextPlot).plt
    plt.bar(["a", "b", "c"], [3, 1, 2])
    plt.scatter([1, 2, 3], [2, 1, 3])
    app.query_one(PlotextPlot).refresh()
    await pilot.pause()
print("bar + scatter ok")
```

## Example 4: A fixed plotext theme

```python
from textual.app import App
from textual_plotext import PlotextPlot

class ThemeApp(App):
    def compose(self):
        yield PlotextPlot()

app = ThemeApp()
async with app.run_test(size=(40, 15)) as pilot:
    widget = app.query_one(PlotextPlot)
    widget.theme = "dark"
    widget.plt.plot([0, 1], [1, 0])
    await pilot.pause()
    assert widget.theme == "dark"
print("theme", widget.theme)
```

## Example 5: Clear and redraw

```python
from textual.app import App
from textual_plotext import PlotextPlot

class RedrawApp(App):
    def compose(self):
        yield PlotextPlot()

app = RedrawApp()
async with app.run_test(size=(50, 15)) as pilot:
    widget = app.query_one(PlotextPlot)
    for n in range(3):
        widget.plt.clear_data()
        widget.plt.plot(list(range(5)), [n * x for x in range(5)])
        widget.refresh()
        await pilot.pause()
print("redrew 3 times")
```
