## Example 1: threadpoolctl functionality test 1

```python
import threadpoolctl
import os

# Example 1: Basic thread pool information and detection
print("Thread pool information and detection:")

# Get information about active thread pools
thread_info = threadpoolctl.threadpool_info()
print(f"Active thread pools detected: {len(thread_info)}")

if thread_info:
    for pool in thread_info:
        print(f"Pool: {pool.get('user_api', 'Unknown')} - {pool.get('internal_api', 'N/A')}")
        print(f"  Threads: {pool.get('num_threads', 'Unknown')}")
else:
    print("No thread pools currently detected")
    print("This is normal in environments without NumPy/SciPy with BLAS")

# Show available controllers
controllers = ['OpenBLASController', 'MKLController', 'OpenMPController']
print(f"Available controllers: {[c for c in controllers if hasattr(threadpoolctl, c)]}")
print(f"Main functions: threadpool_info, threadpool_limits")
```

## Example 2: threadpoolctl functionality test 2

```python
import threadpoolctl

# Example 2: Thread pool control and limiting
print("Thread pool control and limiting:")

# Create a context manager to limit threads
with threadpoolctl.threadpool_limits(limits=2, user_api='blas'):
    print("Inside context: BLAS threads limited to 2")
    
    # Check current thread pool status
    current_info = threadpoolctl.threadpool_info()
    if current_info:
        for pool in current_info:
            print(f"  {pool.get('user_api', 'Unknown')}: {pool.get('num_threads', 'N/A')} threads")
    else:
        print("  No thread pools affected (no BLAS libraries loaded)")

print("Outside context: Thread limits restored")

# Test with different APIs
api_types = ['blas', 'openmp']
for api in api_types:
    print(f"Can control {api} threads: {api in ['blas', 'openmp']}")

print("ThreadpoolController provides fine-grained thread management")
```

## Example 3: threadpoolctl functionality test 3

```python
import threadpoolctl

# Example 3: ThreadpoolController class usage
print("ThreadpoolController class usage:")

# Create a ThreadpoolController instance
controller = threadpoolctl.ThreadpoolController()
print(f"Controller created: {type(controller)}")

# Get information about thread pools
info = controller.info()
print(f"Thread pools found: {len(info)}")

if info:
    for pool in info:
        print(f"Library: {pool.get('filepath', 'Unknown')}")
        print(f"  API: {pool.get('user_api', 'N/A')}")
        print(f"  Version: {pool.get('version', 'N/A')}")
else:
    print("No thread pools detected by controller")

# Test context manager usage
try:
    with controller.limit(limits=1):
        print("Thread limits applied via controller")
        limited_info = controller.info()
        print(f"Pools under control: {len(limited_info)}")
except Exception as e:
    print(f"Controller limit operation: {e}")
    raise

print("ThreadpoolController provides object-oriented thread control")
```

## Example 4: threadpoolctl functionality test 4

```python
import threadpoolctl

# Example 4: Performance optimization and numerical library control
print("Performance optimization patterns:")

# Simulate numerical computation scenarios
def simulate_matrix_operations():
    """Simulate heavy numerical operations"""
    # In real use, this would involve NumPy/SciPy operations
    import time
    time.sleep(0.01)  # Simulate computation
    return "Matrix operation completed"

# Test different thread configurations
thread_configs = [1, 2, 4]

for num_threads in thread_configs:
    print(f"\nTesting with {num_threads} threads:")
    
    with threadpoolctl.threadpool_limits(limits=num_threads):
        # Get current configuration
        current_pools = threadpoolctl.threadpool_info()
        
        if current_pools:
            for pool in current_pools:
                actual_threads = pool.get('num_threads', 'Unknown')
                print(f"  {pool.get('user_api', 'API')}: {actual_threads} threads")
        else:
            print(f"  Requested: {num_threads} threads (no pools to control)")
        
        # Simulate computation
        result = simulate_matrix_operations()
        print(f"  Result: {result}")

print("\nOptimal thread count depends on:")
print("- CPU cores available")
print("- Memory bandwidth")
print("- Problem size and type")
print("- Other concurrent processes")
```

## Example 5: threadpoolctl functionality test 5

```python
import threadpoolctl
import os

# Example 5: Production patterns and environment management
class ComputeEnvironmentManager:
    """Production environment manager for numerical computations"""
    
    def __init__(self):
        self.original_info = threadpoolctl.threadpool_info()
        self.controlled_apis = ['blas', 'openmp']
    
    def get_system_info(self):
        """Get comprehensive system thread information"""
        info = {
            'cpu_count': os.cpu_count(),
            'thread_pools': len(self.original_info),
            'controlled_apis': self.controlled_apis
        }
        
        for pool in self.original_info:
            api = pool.get('user_api', 'unknown')
            info[f'{api}_threads'] = pool.get('num_threads', 0)
            
        return info
    
    def optimize_for_task(self, task_type='general'):
        """Optimize thread configuration for different task types"""
        cpu_count = os.cpu_count() or 4
        
        if task_type == 'memory_bound':
            return min(2, cpu_count)
        elif task_type == 'cpu_intensive':
            return cpu_count
        elif task_type == 'io_intensive':
            return 1
        else:
            return max(1, cpu_count // 2)
    
    def execute_with_limits(self, func, task_type='general'):
        """Execute function with optimized thread limits"""
        optimal_threads = self.optimize_for_task(task_type)
        
        with threadpoolctl.threadpool_limits(limits=optimal_threads):
            return func()

print("Production environment management:")

# Create environment manager
env_manager = ComputeEnvironmentManager()

# Get system information
sys_info = env_manager.get_system_info()
print(f"System info: {sys_info}")

# Test different optimization strategies
task_types = ['memory_bound', 'cpu_intensive', 'io_intensive', 'general']

for task_type in task_types:
    optimal = env_manager.optimize_for_task(task_type)
    print(f"{task_type:15}: {optimal} threads recommended")

# Simulate production task execution
def production_task():
    return "Production computation completed"

result = env_manager.execute_with_limits(production_task, 'cpu_intensive')
print(f"\nProduction result: {result}")

print("\nProduction benefits:")
print("- Prevents resource contention")
print("- Enables reproducible performance")
print("- Optimizes for specific workloads")
print("- Manages multi-tenant environments")
```
