## Example 1: Ultimate Test Case 1

```python
# Example 1: Write and Read TIFF
import tifffile
import numpy as np
import tempfile
import os

print("=== Write and Read TIFF ===")

# Create sample image
image = np.random.randint(0, 255, (100, 100, 3), dtype=np.uint8)

# Create temporary file
with tempfile.NamedTemporaryFile(suffix='.tif', delete=False) as tmp:
    tmp_path = tmp.name

# Write TIFF
tifffile.imwrite(tmp_path, image)
print(f"Wrote image: shape={image.shape}, dtype={image.dtype}")

# Read TIFF
loaded = tifffile.imread(tmp_path)
print(f"Read image: shape={loaded.shape}, dtype={loaded.dtype}")
print(f"Data matches: {np.array_equal(image, loaded)}")

# Cleanup
os.unlink(tmp_path)

print("\nExample 1 completed!")
```

## Example 2: Ultimate Test Case 2

```python
# Example 2: Multi-page TIFF
import tifffile
import numpy as np
import tempfile
import os

print("=== Multi-page TIFF ===")

# Create stack of images
stack = np.random.randint(0, 255, (5, 64, 64), dtype=np.uint8)

with tempfile.NamedTemporaryFile(suffix='.tif', delete=False) as tmp:
    tmp_path = tmp.name

# Write multi-page TIFF
tifffile.imwrite(tmp_path, stack)
print(f"Wrote stack: {stack.shape} ({stack.shape[0]} pages)")

# Read multi-page TIFF
loaded_stack = tifffile.imread(tmp_path)
print(f"Read stack: {loaded_stack.shape}")
print(f"Data matches: {np.array_equal(stack, loaded_stack)}")

# Cleanup
os.unlink(tmp_path)

print("\nExample 2 completed!")
```

## Example 3: Ultimate Test Case 3

```python
# Example 3: TiffFile Context Manager
import tifffile
import numpy as np
import tempfile
import os

print("=== TiffFile Context Manager ===")

# Create sample data
data = np.random.randint(0, 255, (3, 50, 50), dtype=np.uint8)

with tempfile.NamedTemporaryFile(suffix='.tif', delete=False) as tmp:
    tmp_path = tmp.name

tifffile.imwrite(tmp_path, data)

# Use TiffFile to read metadata
with tifffile.TiffFile(tmp_path) as tif:
    print(f"Number of pages: {len(tif.pages)}")
    print(f"Image shape: {tif.pages[0].shape}")
    print(f"Data type: {tif.pages[0].dtype}")
    print(f"Compression: {tif.pages[0].compression}")
    
    # Read data
    image = tif.asarray()
    print(f"Loaded shape: {image.shape}")

# Cleanup
os.unlink(tmp_path)

print("\nExample 3 completed!")
```

## Example 4: Ultimate Test Case 4

```python
# Example 4: Compressed TIFF
import tifffile
import numpy as np
import tempfile
import os

print("=== Compressed TIFF ===")

# Create large image
image = np.random.randint(0, 255, (200, 200), dtype=np.uint8)

with tempfile.NamedTemporaryFile(suffix='.tif', delete=False) as tmp:
    tmp_path = tmp.name

# Write with compression
tifffile.imwrite(tmp_path, image, compression='deflate')
print(f"Wrote compressed TIFF: {image.shape}")

# Check file size
file_size = os.path.getsize(tmp_path)
print(f"File size: {file_size} bytes")

# Read compressed TIFF
loaded = tifffile.imread(tmp_path)
print(f"Read image: {loaded.shape}")
print(f"Data matches: {np.array_equal(image, loaded)}")

# Cleanup
os.unlink(tmp_path)

print("\nExample 4 completed!")
```

## Example 5: Ultimate Test Case 5

```python
# Example 5: Metadata in TIFF
import tifffile
import numpy as np
import tempfile
import os

print("=== Metadata in TIFF ===")

# Create image with metadata
image = np.random.randint(0, 255, (80, 80), dtype=np.uint8)
metadata = {'Creator': 'tifffile test', 'Version': '1.0', 'Description': 'Test image'}

with tempfile.NamedTemporaryFile(suffix='.tif', delete=False) as tmp:
    tmp_path = tmp.name

# Write TIFF with metadata
tifffile.imwrite(tmp_path, image, metadata=metadata)
print(f"Wrote TIFF with metadata")

# Read and inspect metadata
with tifffile.TiffFile(tmp_path) as tif:
    print(f"Image shape: {tif.pages[0].shape}")
    if tif.pages[0].description:
        print(f"Description: {tif.pages[0].description}")

# Cleanup
os.unlink(tmp_path)

print("\nExample 5 completed!")
```
