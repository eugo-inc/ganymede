```python
# Example 1: top-level tldextract.extract() on a standard URL.
import tldextract

result = tldextract.extract('http://forums.news.cnn.com/')
print(f"subdomain={result.subdomain!r}")
print(f"domain={result.domain!r}")
print(f"suffix={result.suffix!r}")
print(f"registered_domain={result.registered_domain!r}")

assert result.subdomain == 'forums.news'
assert result.domain == 'cnn'
assert result.suffix == 'com'
assert result.registered_domain == 'cnn.com'
```

```python
# Example 2: multi-part public suffixes (e.g., .co.uk).
import tldextract

result = tldextract.extract('http://www.bbc.co.uk/news')
print(f"subdomain={result.subdomain!r}")
print(f"domain={result.domain!r}")
print(f"suffix={result.suffix!r}")

assert result.subdomain == 'www'
assert result.domain == 'bbc'
assert result.suffix == 'co.uk'
assert result.registered_domain == 'bbc.co.uk'

# fqdn convenience property
assert result.fqdn == 'www.bbc.co.uk'
```

```python
# Example 3: IP addresses get domain set, empty suffix.
import tldextract

ip_result = tldextract.extract('http://127.0.0.1:8080/deployed/')
print(f"ip: domain={ip_result.domain!r}, suffix={ip_result.suffix!r}, ipv4={ip_result.ipv4!r}")

assert ip_result.domain == '127.0.0.1'
assert ip_result.suffix == ''
assert ip_result.ipv4 == '127.0.0.1'

# Unknown TLD
unknown = tldextract.extract('http://example.bogusunknowntld')
print(f"unknown: domain={unknown.domain!r}, suffix={unknown.suffix!r}")
assert unknown.suffix == ''
```

```python
# Example 4: TLDExtract class with custom cache + suffix_list_urls=().
import tempfile

from tldextract import TLDExtract

with tempfile.TemporaryDirectory() as tmpdir:
    # suffix_list_urls=() forces use of the bundled snapshot (no network).
    extractor = TLDExtract(cache_dir=tmpdir, suffix_list_urls=())
    r = extractor('https://docs.python.org/3/library/index.html')
    print(f"subdomain={r.subdomain!r}, domain={r.domain!r}, suffix={r.suffix!r}")
    assert r.subdomain == 'docs'
    assert r.domain == 'python'
    assert r.suffix == 'org'

    # tuple-unpacking works (ExtractResult is a NamedTuple-style result)
    subdomain, domain, suffix = r.subdomain, r.domain, r.suffix
    assert (subdomain, domain, suffix) == ('docs', 'python', 'org')
```

```python
# Example 5: include_psl_private_domains for private suffixes like blogspot.com.
import tempfile

from tldextract import TLDExtract

with tempfile.TemporaryDirectory() as tmpdir:
    public_only = TLDExtract(cache_dir=tmpdir, suffix_list_urls=(), include_psl_private_domains=False)
    with_private = TLDExtract(cache_dir=tmpdir, suffix_list_urls=(), include_psl_private_domains=True)

    pub = public_only('http://waiterrant.blogspot.com')
    priv = with_private('http://waiterrant.blogspot.com')

    print(f"public_only: domain={pub.domain!r}, suffix={pub.suffix!r}")
    print(f"with_private: domain={priv.domain!r}, suffix={priv.suffix!r}")

    assert pub.domain == 'blogspot'
    assert pub.suffix == 'com'
    assert priv.domain == 'waiterrant'
    assert priv.suffix == 'blogspot.com'

print("\nAll tldextract smoke tests passed")
```
