## Example 1: tobler functionality test 1

```python
# Example 1: Area Weighted Interpolation
from tobler.area_weighted import area_interpolate
import geopandas as gpd
from shapely.geometry import Polygon

print("=== Area Weighted Interpolation ===")

# Create source polygons with data
source = gpd.GeoDataFrame({
    'population': [100, 200, 150],
    'geometry': [
        Polygon([(0, 0), (1, 0), (1, 1), (0, 1)]),
        Polygon([(1, 0), (2, 0), (2, 1), (1, 1)]),
        Polygon([(0, 1), (1, 1), (1, 2), (0, 2)])
    ]
}, crs='EPSG:4326')

# Create target polygons
target = gpd.GeoDataFrame({
    'geometry': [
        Polygon([(0.5, 0.5), (1.5, 0.5), (1.5, 1.5), (0.5, 1.5)])
    ]
}, crs='EPSG:4326')

# Perform area-weighted interpolation
result = area_interpolate(source, target, extensive_variables=['population'])

print(f"Source polygons: {len(source)}")
print(f"Target polygons: {len(target)}")
print(f"Interpolated population: {result['population'].values[0]:.2f}")

print("\nExample 1 completed!")
```

## Example 2: tobler functionality test 2

```python
# Example 2: Intensive Variables Interpolation
from tobler.area_weighted import area_interpolate
import geopandas as gpd
from shapely.geometry import Polygon

print("=== Intensive Variables Interpolation ===")

# Create source with intensive variables (ratios/rates)
source = gpd.GeoDataFrame({
    'density': [10.5, 15.2, 12.8],
    'geometry': [
        Polygon([(0, 0), (2, 0), (2, 2), (0, 2)]),
        Polygon([(2, 0), (4, 0), (4, 2), (2, 2)]),
        Polygon([(0, 2), (2, 2), (2, 4), (0, 4)])
    ]
}, crs='EPSG:4326')

# Target polygon
target = gpd.GeoDataFrame({
    'geometry': [Polygon([(1, 1), (3, 1), (3, 3), (1, 3)])]
}, crs='EPSG:4326')

# Interpolate intensive variable (area-weighted average)
result = area_interpolate(source, target, intensive_variables=['density'])

print(f"Interpolated density: {result['density'].values[0]:.2f}")

print("\nExample 2 completed!")
```

## Example 3: tobler functionality test 3

```python
# Example 3: Area Interpolate with Categorical Data
from tobler.area_weighted import area_interpolate
import geopandas as gpd
from shapely.geometry import Polygon

print("=== Categorical Area Interpolation ===")

# Create source polygons with categorical data
source = gpd.GeoDataFrame({
    'population': [1000, 2000, 1500],
    'households': [250, 500, 375],
    'geometry': [
        Polygon([(0, 0), (2, 0), (2, 2), (0, 2)]),
        Polygon([(2, 0), (4, 0), (4, 2), (2, 2)]),
        Polygon([(0, 2), (2, 2), (2, 4), (0, 4)])
    ]
}, crs='EPSG:4326')

# Target polygons (different boundaries)
target = gpd.GeoDataFrame({
    'geometry': [
        Polygon([(1, 1), (3, 1), (3, 3), (1, 3)]),
        Polygon([(0, 0), (1, 0), (1, 2), (0, 2)])
    ]
}, crs='EPSG:4326')

# Interpolate with categorical allocations
result = area_interpolate(source, target, extensive_variables=['population', 'households'])

print(f"Target polygon 1 - Population: {result['population'].iloc[0]:.0f}, Households: {result['households'].iloc[0]:.0f}")
print(f"Target polygon 2 - Population: {result['population'].iloc[1]:.0f}, Households: {result['households'].iloc[1]:.0f}")

print("\nExample 3 completed!")
```

## Example 4: tobler functionality test 4

```python
# Example 4: Area Interpolate with Different Geometries
from tobler.area_weighted import area_interpolate
import geopandas as gpd
from shapely.geometry import Polygon

print("=== Different Geometry Interpolation ===")

# Source polygons (larger zones)
source = gpd.GeoDataFrame({
    'jobs': [500, 300],
    'retail': [100, 50],
    'geometry': [
        Polygon([(0, 0), (3, 0), (3, 3), (0, 3)]),
        Polygon([(3, 0), (6, 0), (6, 3), (3, 3)])
    ]
}, crs='EPSG:4326')

# Target polygons (smaller zones)
target = gpd.GeoDataFrame({
    'geometry': [
        Polygon([(i, j), (i+1, j), (i+1, j+1), (i, j+1)])
        for i in range(0, 6) for j in range(0, 3)
    ]
}, crs='EPSG:4326')

# Interpolate employment data
result = area_interpolate(source, target, extensive_variables=['jobs', 'retail'])

print(f"Source zones: {len(source)}")
print(f"Target zones: {len(target)}")
print(f"Total jobs redistributed: {result['jobs'].sum():.0f}")
print(f"Total retail redistributed: {result['retail'].sum():.0f}")

print("\nExample 4 completed!")
```

## Example 5: tobler functionality test 5

```python
# Example 5: Multiple Variable Interpolation
from tobler.area_weighted import area_interpolate
import geopandas as gpd
from shapely.geometry import Polygon

print("=== Multiple Variable Interpolation ===")

# Source with multiple variables
source = gpd.GeoDataFrame({
    'population': [1000, 1500, 1200],
    'income': [50000, 60000, 55000],
    'area_km2': [10, 15, 12],
    'geometry': [
        Polygon([(0, 0), (3, 0), (3, 3), (0, 3)]),
        Polygon([(3, 0), (6, 0), (6, 3), (3, 3)]),
        Polygon([(0, 3), (3, 3), (3, 6), (0, 6)])
    ]
}, crs='EPSG:4326')

# Target polygons
target = gpd.GeoDataFrame({
    'geometry': [Polygon([(2, 2), (5, 2), (5, 5), (2, 5)])]
}, crs='EPSG:4326')

# Interpolate both extensive and intensive variables
result = area_interpolate(source, target, 
                          extensive_variables=['population'],
                          intensive_variables=['income'])

print(f"Interpolated population: {result['population'].values[0]:.0f}")
print(f"Interpolated income: ${result['income'].values[0]:.2f}")

print("\nExample 5 completed!")
```
