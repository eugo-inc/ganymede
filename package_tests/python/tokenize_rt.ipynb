## Example 1: tokenize_rt functionality test 1

```python
import tokenize_rt

# Example 1: Basic Python source code tokenization
print("Basic Python source code tokenization:")

# Sample Python code
python_code = """
def calculate_sum(a, b):
    '''Calculate sum of two numbers'''
    result = a + b
    return result

# Usage example
total = calculate_sum(10, 20)
print(f"Total: {total}")
"""

# Tokenize Python source code
tokens = tokenize_rt.src_to_tokens(python_code)
print(f"Tokenized {len(tokens)} tokens from Python source")

# Show first 10 tokens
print("First 10 tokens:")
for i, token in enumerate(tokens[:10]):
    print(f"  {i}: {token.name:15} '{token.src[:20]}{'...' if len(token.src) > 20 else ''}'")

# Convert back to source
reconstructed = tokenize_rt.tokens_to_src(tokens)
print(f"\nRound-trip preservation: {python_code == reconstructed}")
print(f"Original length: {len(python_code)} chars")
print(f"Reconstructed length: {len(reconstructed)} chars")

# Show token types
token_types = set(token.name for token in tokens)
print(f"Token types found: {sorted(token_types)}")
print("tokenize-rt provides lossless Python source tokenization")
```

## Example 2: tokenize_rt functionality test 2

```python
import tokenize_rt

# Example 2: Token manipulation and code transformation
print("Token manipulation and code transformation:")

# Original code with spaces and comments
original_code = """
# Data processing function
def process_data(items):
    # Filter and transform
    result = []
    for item in items:
        if item > 0:  # Only positive values
            result.append(item * 2)
    return result
"""

# Tokenize
tokens = tokenize_rt.src_to_tokens(original_code)
print(f"Original code has {len(tokens)} tokens")

# Find and modify specific tokens
modified_tokens = []
for token in tokens:
    if token.name == 'NAME' and token.src == 'process_data':
        # Rename function
        modified_token = token._replace(src='transform_data')
        modified_tokens.append(modified_token)
    elif token.name == 'NUMBER' and token.src == '2':
        # Change multiplier
        modified_token = token._replace(src='3')
        modified_tokens.append(modified_token)
    else:
        modified_tokens.append(token)

# Convert back to source
modified_code = tokenize_rt.tokens_to_src(modified_tokens)
print("Modified code:")
print(modified_code)

# Show token attributes
sample_token = next(t for t in tokens if t.name == 'NAME')
print(f"\nToken attributes:")
print(f"  name: {sample_token.name}")
print(f"  src: '{sample_token.src}'")
print(f"  line: {sample_token.line}")
print(f"  utf8_byte_offset: {sample_token.utf8_byte_offset}")

print("Tokens preserve exact source location and whitespace")
```

## Example 3: tokenize_rt functionality test 3

```python
# Example 3: Reversed Round-trip Tokenization
import tokenize_rt

print("=== Reversed Round-trip ===")

# Python code
code = """
x = 1 + 2
y = x * 3
"""

# Tokenize
tokens = tokenize_rt.src_to_tokens(code)
print(f"Original tokens: {len(tokens)}")

# Reverse the tokens
reversed_tokens = tokenize_rt.reversed_enumerate(tokens)
print("Reversed token enumeration:")
for offset, token in reversed_tokens:
    if token.name == 'NAME':
        print(f"  Offset {offset}: {token.src}")

# Use parse_string_literal
string_code = "'hello\\nworld'"
string_tokens = tokenize_rt.src_to_tokens(f"s = {string_code}")
for token in string_tokens:
    if token.name == 'STRING':
        parsed = tokenize_rt.parse_string_literal(token.src)
        print(f"\nString literal: {token.src}")
        print(f"Parsed value: {parsed}")

print("\nExample 3 completed!")
```

## Example 4: tokenize_rt functionality test 4

```python
# Example 4: Non-coding Tokens Filter
import tokenize_rt

print("=== Non-coding Tokens ===")

# Code with comments and whitespace
code = """
# This is a comment
def foo():  # inline comment
    x = 1  # another comment
    return x
"""

# Tokenize
tokens = tokenize_rt.src_to_tokens(code)
print(f"Total tokens: {len(tokens)}")

# Count non-coding tokens
non_coding = [t for t in tokens if t.name in tokenize_rt.NON_CODING_TOKENS]
print(f"Non-coding tokens: {len(non_coding)}")
print(f"NON_CODING_TOKENS set: {tokenize_rt.NON_CODING_TOKENS}")

# Show coding tokens only
coding_tokens = [t for t in tokens if t.name not in tokenize_rt.NON_CODING_TOKENS]
print(f"\nCoding tokens ({len(coding_tokens)}):")
for token in coding_tokens[:10]:
    print(f"  {token.name}: '{token.src}'")

print("\nExample 4 completed!")
```

## Example 5: tokenize_rt functionality test 5

```python
# Example 5: Offset Tracking and Token Finding
import tokenize_rt

print("=== Offset Tracking ===")

# Code with multiple lines
code = """x = 10
y = 20
z = x + y
"""

# Tokenize
tokens = tokenize_rt.src_to_tokens(code)

# Show token offsets
print("Token offsets:")
for i, token in enumerate(tokens):
    if token.name == 'NAME' or token.name == 'NUMBER':
        print(f"  Token {i}: '{token.src}' at line {token.line}, offset {token.utf8_byte_offset}")

# Use rfind_string_parts for string manipulation
string_code = 'msg = "Hello, " + "World!"'
string_tokens = tokenize_rt.src_to_tokens(string_code)

print("\nString tokens:")
for i, token in enumerate(string_tokens):
    if token.name == 'STRING':
        parts = tokenize_rt.rfind_string_parts(string_tokens, i)
        print(f"  String at {i}: {token.src}, parts info: {parts}")

print("\nExample 5 completed!")
```
