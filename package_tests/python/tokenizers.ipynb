## Example 1: tokenizers functionality test 1

```python
# Example 1: Basic tokenization with pre-trained tokenizer
from tokenizers import Tokenizer
from tokenizers.models import WordPiece
from tokenizers.pre_tokenizers import Whitespace
from tokenizers.trainers import WordPieceTrainer

print("Example 1: Basic Tokenization\n")

# Create a simple WordPiece tokenizer
print("Creating WordPiece tokenizer:")
tokenizer = Tokenizer(WordPiece(unk_token="[UNK]"))
tokenizer.pre_tokenizer = Whitespace()
print(f"  Tokenizer type: {type(tokenizer)}")
print(f"  Model: {type(tokenizer.model)}\n")

# Train on sample sentences
print("Training tokenizer on sample data:")
trainer = WordPieceTrainer(special_tokens=["[UNK]", "[PAD]", "[CLS]", "[SEP]", "[MASK]"])
sentences = [
    "Hello, how are you?",
    "I am doing great, thank you!",
    "This is a simple tokenization example.",
    "Tokenizers are very useful for NLP tasks.",
    "We can encode and decode text efficiently."
]
tokenizer.train_from_iterator(sentences, trainer=trainer)
print(f"  Vocabulary size: {tokenizer.get_vocab_size()}\n")

# Encode text
print("Encoding sample text:")
text = "Hello, this is a tokenization test."
encoded = tokenizer.encode(text)
print(f"  Input: {text}")
print(f"  Tokens: {encoded.tokens}")
print(f"  IDs: {encoded.ids}")

# Decode back
print("\nDecoding:")
decoded = tokenizer.decode(encoded.ids)
print(f"  Decoded: {decoded}")

print("\nTokenizers provides fast text tokenization")
```

## Example 2: tokenizers functionality test 2

```python
# Example 2: BPE tokenizer with encoding features
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

print("Example 2: BPE Tokenizer\n")

# Create BPE tokenizer
print("Creating BPE tokenizer:")
tokenizer = Tokenizer(BPE(unk_token="[UNK]"))
tokenizer.pre_tokenizer = Whitespace()

# Train on text corpus
print("Training BPE tokenizer:")
trainer = BpeTrainer(special_tokens=["[UNK]", "[PAD]", "[CLS]", "[SEP]"])
corpus = [
    "The quick brown fox jumps over the lazy dog.",
    "Machine learning is transforming the world.",
    "Natural language processing enables computers to understand text.",
    "Tokenization is the first step in text processing.",
    "Deep learning models require tokenized input."
]
tokenizer.train_from_iterator(corpus, trainer=trainer)
print(f"  Vocabulary size: {tokenizer.get_vocab_size()}\n")

# Encode with attention mask
print("Encoding text:")
text = "Tokenization transforms text into numbers."
encoding = tokenizer.encode(text)
print(f"  Text: {text}")
print(f"  Tokens: {encoding.tokens}")
print(f"  Token IDs: {encoding.ids}")
print(f"  Attention mask: {encoding.attention_mask}")

# Show token offsets
print("\nToken offsets (character positions):")
for i, (token, offset) in enumerate(zip(encoding.tokens, encoding.offsets)):
    print(f"  {i}: '{token}' -> {offset}")

# Batch encoding
print("\nBatch encoding:")
texts = ["First sentence.", "Second sentence here."]
encodings = tokenizer.encode_batch(texts)
for i, enc in enumerate(encodings):
    print(f"  Text {i+1}: {enc.tokens} ({len(enc.ids)} tokens)")

print("\nBPE tokenizer uses byte-pair encoding algorithm")
```

## Example 3: tokenizers functionality test 3

```python
# Example 3: Post-processing and special tokens
from tokenizers import Tokenizer
from tokenizers.models import WordPiece
from tokenizers.trainers import WordPieceTrainer
from tokenizers.pre_tokenizers import Whitespace
from tokenizers.processors import TemplateProcessing

print("Example 3: Post-Processing and Special Tokens\n")

# Create tokenizer
tokenizer = Tokenizer(WordPiece(unk_token="[UNK]"))
tokenizer.pre_tokenizer = Whitespace()

# Train tokenizer
print("Training tokenizer:")
trainer = WordPieceTrainer(
    special_tokens=["[UNK]", "[PAD]", "[CLS]", "[SEP]", "[MASK]"],
    vocab_size=100
)
training_data = [
    "This is the first training sentence.",
    "Here is another sentence for training.",
    "We need enough data to build vocabulary.",
    "Tokenizers learn subword units from text."
]
tokenizer.train_from_iterator(training_data, trainer=trainer)
print(f"  Vocabulary size: {tokenizer.get_vocab_size()}\n")

# Add post-processor for BERT-style format
print("Adding post-processor (BERT-style):")
tokenizer.post_processor = TemplateProcessing(
    single="[CLS] $A [SEP]",
    pair="[CLS] $A [SEP] $B:1 [SEP]:1",
    special_tokens=[
        ("[CLS]", tokenizer.token_to_id("[CLS]")),
        ("[SEP]", tokenizer.token_to_id("[SEP]")),
    ],
)
print("  Post-processor configured\n")

# Encode single sentence
print("Encoding single sentence:")
text = "This is a test."
encoding = tokenizer.encode(text)
print(f"  Text: {text}")
print(f"  Tokens (with special tokens): {encoding.tokens}")
print(f"  IDs: {encoding.ids}\n")

# Encode sentence pair
print("Encoding sentence pair:")
sent1 = "First sentence."
sent2 = "Second sentence."
encoding_pair = tokenizer.encode(sent1, sent2)
print(f"  Sentence 1: {sent1}")
print(f"  Sentence 2: {sent2}")
print(f"  Tokens: {encoding_pair.tokens}")
print(f"  Type IDs: {encoding_pair.type_ids}")

print("\nPost-processing adds special tokens and formatting")
```

## Example 4: tokenizers functionality test 4

```python
# Example 4: Padding and truncation
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

print("Example 4: Padding and Truncation\n")

# Create and train tokenizer
tokenizer = Tokenizer(BPE(unk_token="[UNK]"))
tokenizer.pre_tokenizer = Whitespace()

trainer = BpeTrainer(special_tokens=["[UNK]", "[PAD]", "[CLS]", "[SEP]"])
corpus = [
    "Short text.",
    "This is a medium length sentence for training.",
    "Here we have an even longer sentence that contains more words and tokens.",
    "Padding ensures all sequences have the same length."
]
tokenizer.train_from_iterator(corpus, trainer=trainer)
print(f"Trained tokenizer with vocab size: {tokenizer.get_vocab_size()}\n")

# Enable padding
print("Enabling padding:")
tokenizer.enable_padding(pad_id=tokenizer.token_to_id("[PAD]"), pad_token="[PAD]")
print("  Padding enabled\n")

# Encode texts of different lengths
print("Encoding texts with padding:")
texts = [
    "Short.",
    "This is longer.",
    "This is the longest sentence in the batch."
]
encodings = tokenizer.encode_batch(texts)
for i, (text, enc) in enumerate(zip(texts, encodings)):
    print(f"  Text {i+1}: '{text}'")
    print(f"    Tokens: {enc.tokens}")
    print(f"    Length: {len(enc.ids)}")
    print(f"    Attention mask: {enc.attention_mask}")

# Enable truncation
print("\nEnabling truncation (max_length=5):")
tokenizer.enable_truncation(max_length=5)
print("  Truncation enabled\n")

# Encode long text with truncation
print("Encoding with truncation:")
long_text = "This is a very long sentence that will be truncated to fit the maximum length."
enc_truncated = tokenizer.encode(long_text)
print(f"  Original: {long_text}")
print(f"  Tokens (truncated): {enc_truncated.tokens}")
print(f"  Length: {len(enc_truncated.ids)} (max: 5)")

print("\nPadding and truncation normalize sequence lengths")
```

## Example 5: tokenizers functionality test 5

```python
# Example 5: Saving and loading tokenizers
from tokenizers import Tokenizer
from tokenizers.models import WordPiece
from tokenizers.trainers import WordPieceTrainer
from tokenizers.pre_tokenizers import Whitespace
import tempfile
import os

print("Example 5: Saving and Loading Tokenizers\n")

# Create and train tokenizer
print("Creating and training tokenizer:")
tokenizer = Tokenizer(WordPiece(unk_token="[UNK]"))
tokenizer.pre_tokenizer = Whitespace()

trainer = WordPieceTrainer(
    special_tokens=["[UNK]", "[PAD]", "[CLS]", "[SEP]", "[MASK]"],
    vocab_size=200
)
training_corpus = [
    "Natural language processing is fascinating.",
    "Tokenizers convert text into numerical representations.",
    "Machine learning models process tokenized input.",
    "We can save and load trained tokenizers.",
    "This enables model deployment and reuse."
]
tokenizer.train_from_iterator(training_corpus, trainer=trainer)
print(f"  Vocabulary size: {tokenizer.get_vocab_size()}\n")

# Test original tokenizer
test_text = "Testing the tokenizer."
original_encoding = tokenizer.encode(test_text)
print(f"Original tokenizer encoding:")
print(f"  Text: {test_text}")
print(f"  Tokens: {original_encoding.tokens}")
print(f"  IDs: {original_encoding.ids}\n")

# Save tokenizer
print("Saving tokenizer:")
with tempfile.TemporaryDirectory() as tmpdir:
    save_path = os.path.join(tmpdir, "tokenizer.json")
    tokenizer.save(save_path)
    file_size = os.path.getsize(save_path)
    print(f"  Saved to: {save_path}")
    print(f"  File size: {file_size} bytes\n")
    
    # Load tokenizer
    print("Loading tokenizer:")
    loaded_tokenizer = Tokenizer.from_file(save_path)
    print(f"  Loaded successfully")
    print(f"  Vocabulary size: {loaded_tokenizer.get_vocab_size()}\n")
    
    # Test loaded tokenizer
    loaded_encoding = loaded_tokenizer.encode(test_text)
    print(f"Loaded tokenizer encoding:")
    print(f"  Text: {test_text}")
    print(f"  Tokens: {loaded_encoding.tokens}")
    print(f"  IDs: {loaded_encoding.ids}")
    
    # Verify they match
    print(f"\nVerification:")
    print(f"  Tokens match: {original_encoding.tokens == loaded_encoding.tokens}")
    print(f"  IDs match: {original_encoding.ids == loaded_encoding.ids}")

print("\nTokenizers can be saved and loaded for reuse")
```
