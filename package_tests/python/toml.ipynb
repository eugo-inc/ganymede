## Example 1: toml functionality test 1

```python
import toml

# Example 1: Basic TOML configuration parsing and generation
print("Basic TOML configuration handling:")

# Create a configuration dictionary
config = {
    "title": "My Application Config",
    "database": {
        "host": "localhost",
        "port": 5432,
        "name": "myapp_db",
        "user": "admin"
    },
    "servers": ["192.168.1.1", "192.168.1.2", "192.168.1.3"],
    "debug": True,
    "max_connections": 100
}

# Convert to TOML string
toml_string = toml.dumps(config)
print("Generated TOML:")
print(toml_string)

# Parse back to Python
parsed_config = toml.loads(toml_string)
print(f"Round-trip successful: {config == parsed_config}")
print(f"Database host: {parsed_config['database']['host']}")
print(f"Server count: {len(parsed_config['servers'])}")

# Show available functions
print(f"TOML functions: {[attr for attr in dir(toml) if not attr.startswith('_')]}")
```

## Example 2: toml functionality test 2

```python
import toml
import tempfile
import os

# Example 2: File operations and complex data structures
print("TOML file operations and complex structures:")

# Complex configuration with nested structures
complex_config = {
    "application": {
        "name": "WebApp",
        "version": "1.2.3",
        "features": ["auth", "api", "dashboard"]
    },
    "logging": {
        "level": "INFO",
        "handlers": {
            "file": {"path": "/var/log/app.log", "max_size": "10MB"},
            "console": {"enabled": True, "format": "%(levelname)s: %(message)s"}
        }
    },
    "cache": {
        "redis": {
            "host": "redis.example.com",
            "port": 6379,
            "timeout": 30.5
        }
    }
}

# Write to temporary file
with tempfile.NamedTemporaryFile(mode='w', suffix='.toml', delete=False) as f:
    toml.dump(complex_config, f)
    temp_file = f.name

print(f"Written to temporary file: {temp_file}")

# Read back from file
with open(temp_file, 'r') as f:
    loaded_config = toml.load(f)

print("Loaded configuration structure:")
print(f"App name: {loaded_config['application']['name']}")
print(f"Features: {loaded_config['application']['features']}")
print(f"Log level: {loaded_config['logging']['level']}")
print(f"Redis timeout: {loaded_config['cache']['redis']['timeout']}")

# Verify data types are preserved
print(f"Port is integer: {type(loaded_config['cache']['redis']['port']) == int}")
print(f"Timeout is float: {type(loaded_config['cache']['redis']['timeout']) == float}")

# Cleanup
os.unlink(temp_file)

print("File operations completed successfully")
```

## Example 3: toml functionality test 3

```python
import toml
from datetime import datetime, date

# Example 3: Data types and TOML specification compliance
print("TOML data types and specification compliance:")

# TOML supports various data types
typed_config = {
    # Strings
    "title": "Configuration Example",
    "description": "Multi-line string example",
    
    # Numbers
    "integer": 42,
    "float": 3.14159,
    "scientific": 1e6,
    
    # Booleans
    "enabled": True,
    "disabled": False,
    
    # Dates and times
    "created_date": date(2024, 1, 15),
    "created_time": datetime(2024, 1, 15, 14, 30, 0),
    
    # Arrays (must be homogeneous)
    "tags": ["config", "example", "toml"],
    "ports": [8080, 8081, 8082],
    "floats": [1.0, 2.5, 3.7],
    
    # Nested tables
    "database": {
        "primary": {"host": "db1.example.com", "port": 5432},
        "replica": {"host": "db2.example.com", "port": 5432}
    }
}

# Convert to TOML and parse back
toml_output = toml.dumps(typed_config)
parsed = toml.loads(toml_output)

print(f"Integer: {parsed['integer']} ({type(parsed['integer']).__name__})")
print(f"Float: {parsed['float']} ({type(parsed['float']).__name__})")
print(f"Date: {parsed['created_date']}")

print("\nExample 3 completed!")
```

## Example 4: toml functionality test 4

```python
import toml

# Example 4: Error handling and validation
print("TOML error handling and validation:")

# Valid TOML string
valid_toml = """
[server]
host = "localhost"
port = 8080
enabled = true

[[plugins]]
name = "auth"
version = "1.0"

[[plugins]]
name = "logging"
version = "2.1"
"""

# Parse valid TOML
try:
    valid_config = toml.loads(valid_toml)
    print("Valid TOML parsed successfully:")
    print(f"Server host: {valid_config['server']['host']}")
    print(f"Plugin count: {len(valid_config['plugins'])}")
    for plugin in valid_config['plugins']:
        print(f"  {plugin['name']} v{plugin['version']}")
except toml.TomlDecodeError as e:
    print(f"Parse error: {e}")

# Invalid TOML examples
invalid_toml_examples = [
    "key = ",  # Missing value
    "[section\nkey = value",  # Unclosed section
    "key = 'unclosed string",  # Unclosed string
    "duplicate = 1\nduplicate = 2"  # Duplicate keys
]

print(f"\nError handling tests:")
for i, invalid_toml in enumerate(invalid_toml_examples, 1):
    try:
        toml.loads(invalid_toml)
    except toml.TomlDecodeError as e:
        print(f"Test {i}: Correctly caught error - {str(e)[:50]}...")
    else:
        raise AssertionError(f"Test {i}: toml.loads accepted invalid TOML {invalid_toml!r}")

# Configuration validation helper
def validate_server_config(config_dict):
    """Validate server configuration structure"""
    required_fields = ['host', 'port']
    
    if 'server' not in config_dict:
        return False, "Missing 'server' section"
    
    server = config_dict['server']
    for field in required_fields:
        if field not in server:
            return False, f"Missing required field: {field}"
    
    if not isinstance(server['port'], int) or server['port'] <= 0:
        return False, "Port must be a positive integer"
    
    return True, "Configuration valid"

# Test validation
valid, message = validate_server_config(valid_config)
print(f"\nConfiguration validation: {valid} - {message}")
```

## Example 5: toml functionality test 5

```python
import toml
import os

# Example 5: Production patterns and configuration management
class ConfigManager:
    """Production configuration manager using TOML"""
    
    def __init__(self, config_file=None):
        self.config_file = config_file
        self.config = {}
        self.defaults = {
            "app": {
                "name": "Default App",
                "version": "1.0.0",
                "debug": False
            },
            "server": {
                "host": "0.0.0.0",
                "port": 8000,
                "workers": 4
            },
            "database": {
                "host": "localhost",
                "port": 5432,
                "timeout": 30
            }
        }
    
    def load_config(self, config_string=None):
        """Load configuration from string or file"""
        if config_string:
            try:
                self.config = toml.loads(config_string)
                return True, "Configuration loaded from string"
            except toml.TomlDecodeError as e:
                return False, f"TOML parse error: {e}"
        
        if self.config_file and os.path.exists(self.config_file):
            try:
                with open(self.config_file, 'r') as f:
                    self.config = toml.load(f)
                return True, f"Configuration loaded from {self.config_file}"
            except Exception as e:
                return False, f"File error: {e}"
        
        # Use defaults if no config provided
        self.config = self.defaults.copy()
        return True, "Using default configuration"
    
    def get(self, key_path, default=None):
        """Get configuration value using dot notation"""
        keys = key_path.split('.')
        value = self.config
        
        for key in keys:
            if isinstance(value, dict) and key in value:
                value = value[key]
            else:
                return default
        
        return value
    
    def merge_env_vars(self, prefix="APP_"):
        """Merge environment variables into configuration"""
        for env_key, env_value in os.environ.items():
            if env_key.startswith(prefix):
                # Convert APP_DATABASE_HOST to database.host
                config_key = env_key[len(prefix):].lower().replace('_', '.')
                
                # Try to convert to appropriate type
                if env_value.lower() in ('true', 'false'):
                    env_value = env_value.lower() == 'true'
                elif env_value.isdigit():
                    env_value = int(env_value)
                elif env_value.replace('.', '').isdigit():
                    env_value = float(env_value)
                
                # Set the value (simplified - real implementation would handle nested keys)
                print(f"Environment override: {config_key} = {env_value}")

print("Production configuration management:")

# Production configuration example
prod_config = """
[app]
name = "Production API"
version = "2.1.0"
debug = false
environment = "production"

[server]
host = "0.0.0.0"
port = 8080
workers = 8
timeout = 30

[database]
host = "prod-db.example.com"
port = 5432
name = "proddb"
pool_size = 20

[logging]
level = "INFO"
format = "%(asctime)s - %(name)s - %(levelname)s - %(message)s"

[[middleware]]
name = "cors"
enabled = true

[[middleware]]
name = "auth"
enabled = true
"""

# Create configuration manager
config_manager = ConfigManager()

# Load production configuration
success, message = config_manager.load_config(prod_config)
print(f"Config load: {success} - {message}")

# Access configuration values
print(f"\nConfiguration access:")
print(f"App name: {config_manager.get('app.name')}")
print(f"Server port: {config_manager.get('server.port')}")
print(f"Database host: {config_manager.get('database.host')}")
print(f"Debug mode: {config_manager.get('app.debug')}")
print(f"Missing key: {config_manager.get('nonexistent.key', 'default_value')}")

# Show environment variable merging capability
print(f"\nEnvironment variable integration:")
os.environ['APP_SERVER_PORT'] = '9000'
os.environ['APP_DEBUG'] = 'true'
config_manager.merge_env_vars()

print(f"\nProduction features:")
print("- TOML configuration parsing")
print("- Default value management")
print("- Environment variable overrides")
print("- Nested configuration access")
print("- Error handling and validation")
```
