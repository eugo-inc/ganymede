# tomli: 5 Basic Usage Examples

```python
# Example 1: Basic TOML parsing
import tomli

# Simple TOML content
toml_content = '''
title = "TOML Example"
version = "1.0.0"

[database]
server = "192.168.1.1"
ports = [8001, 8002, 8003]
enabled = true
'''

# Parse TOML content
data = tomli.loads(toml_content)

print(f"Parsed data type: {type(data)}")
print(f"Title: {data['title']}")
print(f"Version: {data['version']}")
print(f"Database server: {data['database']['server']}")
print(f"Database ports: {data['database']['ports']}")
print(f"Database enabled: {data['database']['enabled']}")
```

```python
# Example 2: TOML with arrays and tables
import tomli

complex_toml = '''
[package]
name = "my-package"
authors = ["Alice <alice@example.com>", "Bob <bob@example.com>"]

[package.dependencies]
requests = ">=2.25.0"
click = "^8.0.0"

[[tools]]
name = "pytest"
version = "6.0.0"
'''

parsed = tomli.loads(complex_toml)

print("Package info:")
print(f"  Name: {parsed['package']['name']}")
print(f"  Authors: {len(parsed['package']['authors'])} authors")
print(f"  Dependencies: {list(parsed['package']['dependencies'].keys())}")
print(f"  Tools: {[tool['name'] for tool in parsed['tools']]}")
```

```python
# Example 3: TOML file reading
import tomli
import tempfile
from pathlib import Path

# Create a temporary TOML file
temp_file = Path(tempfile.mkdtemp()) / "config.toml"

toml_content = '''
[server]
host = "localhost"
port = 8080

[features]
enable_api = true
max_connections = 100
'''

# Write and read TOML file
temp_file.write_text(toml_content)
with open(temp_file, "rb") as f:
    config = tomli.load(f)

print(f"Server: {config['server']['host']}:{config['server']['port']}")
print(f"API enabled: {config['features']['enable_api']}")
print(f"Max connections: {config['features']['max_connections']}")

temp_file.unlink()  # Clean up
```

```python
# Example 4: TOML error handling
import tomli

# Test invalid TOML strings
invalid_examples = [
    ('key = value without quotes', 'Unquoted string'),
    ('key = "value1"\nkey = "value2"', 'Duplicate key'),
    ('[invalid section]', 'Invalid section syntax')
]

print("Testing TOML error handling:")
for toml_str, description in invalid_examples:
    try:
        result = tomli.loads(toml_str)
    except tomli.TOMLDecodeError:
        print(f"✓ {description}: Caught TOMLDecodeError")
    else:
        raise AssertionError(f"{description}: tomli.loads accepted invalid TOML {toml_str!r}")

# Test valid edge case
valid_toml = 'empty_string = ""\nzero = 0\nbool_false = false'
result = tomli.loads(valid_toml)
print(f"\nValid edge cases: {result}")
```

```python
# Example 5: TOML data types
import tomli
from datetime import datetime, date, time

advanced_toml = '''
# Different data types
string = "Hello, World!"
integer = 42
float_val = 3.14
boolean = true
date_only = 1979-05-27
datetime_utc = 1979-05-27T00:32:00Z

# Arrays and inline tables
array = [1, 2, 3]
nested_array = [[1, 2], [3, 4]]
inline_table = { color = "red", size = "large" }
'''

data = tomli.loads(advanced_toml)

print("TOML data types:")
for key, value in data.items():
    print(f"  {key}: {value} ({type(value).__name__})")

print(f"\nTomli successfully parsed {len(data)} items")
```
