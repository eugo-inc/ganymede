## Example 1: tomlkit functionality test 1

```python
import tomlkit

# Example 1: TOMLKit document creation and formatting preservation
print("TOMLKit advanced TOML handling:")

# Create a new TOML document
doc = tomlkit.document()

# Add title and comments
doc.add("title", "Advanced Configuration")
doc.add(tomlkit.comment("Configuration file for production application"))
doc.add(tomlkit.nl())  # Add newline

# Add basic configuration
doc.add("version", "2.1.0")
doc.add("debug", False)
doc.add("features", ["auth", "logging", "metrics"])

# Add table with comments
doc.add(tomlkit.comment("Database configuration"))
database_table = tomlkit.table()
database_table.add("host", "prod-db.example.com")
database_table.add("port", 5432)
database_table.add("timeout", 30)
doc.add("database", database_table)

# Generate formatted TOML
formatted_toml = doc.as_string()
print("Generated TOML with formatting:")
print(formatted_toml)

# Show structure preservation
print(f"Document type: {type(doc)}")
print(f"Contains comments: {'#' in formatted_toml}")
print(f"Maintains structure: {tomlkit.nl() is not None}")

# Available TOMLKit functions
functions = [attr for attr in dir(tomlkit) if not attr.startswith('_')]
print(f"TOMLKit functions: {functions[:8]}...")  # Show first 8
```

## Example 2: tomlkit functionality test 2

```python
import tomlkit

# Example 2: Comment preservation and advanced document editing
print("Comment preservation and document editing:")

# Parse existing TOML with comments
existing_toml = """
# Application Configuration
title = "Production App"
version = "2.0.1"

# Server settings
[server]
host = "0.0.0.0"  # Listen on all interfaces
port = 8080       # HTTP port
workers = 4       # Number of worker processes

# Database configuration
[database]
# Primary database
host = "db.example.com"
port = 5432
name = "proddb"
"""

# Parse preserving comments and formatting
doc = tomlkit.parse(existing_toml)
print("Parsed document with preserved formatting:")
print(doc.as_string())

# Modify values while preserving structure
doc["server"]["workers"] = 8
doc["version"] = "2.0.2"

# Add new section with comment
doc.add(tomlkit.comment("Cache configuration"))
cache_table = tomlkit.table()
cache_table.add("enabled", True)
cache_table.add("ttl", 3600)
doc.add("cache", cache_table)

print("\nModified document (comments preserved):")
print(doc.as_string())

# Show item types
print(f"\nDocument item types:")
for key, value in doc.items():
    print(f"  {key}: {type(value).__name__}")

print("TOMLKit preserves formatting, comments, and structure")
```

## Example 3: tomlkit functionality test 3

```python
# Example 3: Array of Tables (AoT)
import tomlkit

print("=== Array of Tables ===")

# Create document with array of tables
doc = tomlkit.document()

# Create array of tables for plugins
plugins = tomlkit.aot()

# Add first plugin
plugin1 = tomlkit.table()
plugin1.add("name", "authentication")
plugin1.add("version", "1.0.0")
plugin1.add("enabled", True)
plugins.append(plugin1)

# Add second plugin
plugin2 = tomlkit.table()
plugin2.add("name", "logging")
plugin2.add("version", "2.1.0")
plugin2.add("enabled", True)
plugins.append(plugin2)

doc.add("plugins", plugins)

print(doc.as_string())
print(f"Array of tables type: {type(plugins).__name__}")
print(f"Number of plugins: {len(plugins)}")

print("\nExample 3 completed!")
```

## Example 4: tomlkit functionality test 4

```python
# Example 4: Inline Tables
import tomlkit

print("=== Inline Tables ===")

# Create document with inline tables
doc = tomlkit.document()

# Create inline table (compact format)
point = tomlkit.inline_table()
point.add("x", 10)
point.add("y", 20)
point.add("z", 30)

doc.add("coordinates", point)

# Create array of inline tables
colors = tomlkit.array()
colors.append(tomlkit.inline_table())
colors[0].add("name", "red")
colors[0].add("hex", "#FF0000")

colors.append(tomlkit.inline_table())
colors[1].add("name", "blue")
colors[1].add("hex", "#0000FF")

doc.add("colors", colors)

print(doc.as_string())
print(f"\nInline table type: {type(point).__name__}")
print(f"Coordinates: x={point['x']}, y={point['y']}, z={point['z']}")

print("\nExample 4 completed!")
```

## Example 5: tomlkit functionality test 5

```python
# Example 5: Item Types and Key Management
import tomlkit

print("=== Item Types and Key Management ===")

# Create document with typed items
doc = tomlkit.document()

# Create different item types
doc.add("string_value", tomlkit.string("Hello World"))
doc.add("integer_value", tomlkit.integer(42))
doc.add("float_value", tomlkit.float_(3.14))
doc.add("bool_value", tomlkit.boolean(True))

# Create array
numbers = tomlkit.array()
numbers.extend([1, 2, 3, 4, 5])
doc.add("numbers", numbers)

print(doc.as_string())

# Show item types
print("\nItem types in document:")
for key, value in doc.items():
    print(f"  {key}: {type(value).__name__} = {value}")

# Access values directly (automatic conversion)
print(f"\nDirect access (auto-converted):")
print(f"  string: {doc['string_value']} (type: {type(doc['string_value']).__name__})")
print(f"  integer: {doc['integer_value']} (type: {type(doc['integer_value']).__name__})")

print("\nExample 5 completed!")
```
