## Example 1: toolz functionality test 1

```python
import toolz

# Example 1: Basic functional programming with toolz
print("=== Basic toolz operations ===")

# Core sequence operations
numbers = list(range(10))
print(f"Original numbers: {numbers}")

# Take first N elements
first_five = list(toolz.take(5, numbers))
print(f"take(5): {first_five}")

# Drop first N elements  
dropped = list(toolz.drop(3, numbers))
print(f"drop(3): {dropped}")

# Get nth element
nth_element = toolz.nth(4, numbers)
print(f"nth(4): {nth_element}")

# Count elements in a sequence
count_result = toolz.count(numbers)
print(f"count(numbers): {count_result}")

# Interleave sequences
seq1 = [1, 3, 5]
seq2 = [2, 4, 6]
interleaved = list(toolz.interleave([seq1, seq2]))
print(f"interleave: {interleaved}")

print("\nExample 1 completed!")
```

## Example 2: toolz functionality test 2

```python
import toolz

# Example 2: Data transformation and grouping
print("=== Data transformation and grouping ===")

# Sample data
data = [
    {"name": "Alice", "age": 30, "department": "Engineering"},
    {"name": "Bob", "age": 25, "department": "Design"},
    {"name": "Charlie", "age": 35, "department": "Engineering"},
    {"name": "Diana", "age": 28, "department": "Marketing"},
    {"name": "Eve", "age": 32, "department": "Engineering"}
]

print(f"Original data: {len(data)} people")

# Group by department
grouped = toolz.groupby(lambda x: x["department"], data)
print(f"\nGrouped by department:")
for dept, people in grouped.items():
    print(f"  {dept}: {len(people)} people")

# Get specific field values
names = list(toolz.pluck("name", data))
print(f"\nNames: {names}")

# Filter and map
engineers = list(toolz.filter(lambda x: x["department"] == "Engineering", data))
engineer_names = list(toolz.pluck("name", engineers))
print(f"Engineers: {engineer_names}")

# Frequencies
ages = list(toolz.pluck("age", data))
print(f"Ages: {ages}")

print("\nExample 2 completed!")
```

## Example 3: toolz functionality test 3

```python
# Example 3: Function Composition
import toolz

print("=== Function Composition ===")

# Compose functions together
def add_ten(x):
    return x + 10

def multiply_two(x):
    return x * 2

def square(x):
    return x ** 2

# Compose functions (right to left execution)
composed = toolz.compose(add_ten, multiply_two, square)
result = composed(5)  # square(5) -> multiply_two(25) -> add_ten(50)
print(f"compose(add_ten, multiply_two, square)(5) = {result}")

# Pipe functions (left to right execution)
piped = toolz.pipe(5, square, multiply_two, add_ten)
print(f"pipe(5, square, multiply_two, add_ten) = {piped}")

# Curry a function
@toolz.curry
def power(base, exponent):
    return base ** exponent

square_func = power(exponent=2)
cube_func = power(exponent=3)

print(f"Curried power(2, 2) = {square_func(2)}")
print(f"Curried power(2, 3) = {cube_func(2)}")

print("\nExample 3 completed!")
```

## Example 4: toolz functionality test 4

```python
# Example 4: Dictionary and Sequence Utilities
import toolz

print("=== Dictionary and Sequence Utilities ===")

# Dictionary operations
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 3, "c": 4}

# Merge dictionaries
merged = toolz.merge(dict1, dict2)
print(f"merge: {merged}")

# Get nested values
nested = {"user": {"name": "Alice", "address": {"city": "NYC"}}}
city = toolz.get_in(["user", "address", "city"], nested)
print(f"get_in nested value: {city}")

# Partition sequence into chunks
numbers = list(range(10))
chunks = list(toolz.partition(3, numbers))
print(f"partition(3): {chunks}")

# Partition all (include remainder)
all_chunks = list(toolz.partition_all(3, numbers))
print(f"partition_all(3): {all_chunks}")

# Sliding window
windows = list(toolz.sliding_window(3, numbers))
print(f"sliding_window(3): {windows[:3]}...")

# Frequencies
items = ['a', 'b', 'a', 'c', 'b', 'a']
freq = toolz.frequencies(items)
print(f"frequencies: {freq}")

print("\nExample 4 completed!")
```

## Example 5: toolz functionality test 5

```python
# Example 5: Advanced Iterators
import toolz

print("=== Advanced Iterators ===")

# Concat - concatenate sequences
seq1 = [1, 2, 3]
seq2 = [4, 5, 6]
seq3 = [7, 8, 9]
concatenated = list(toolz.concat([seq1, seq2, seq3]))
print(f"concat: {concatenated}")

# Unique - preserve order while removing duplicates
items = [1, 2, 2, 3, 1, 4, 3, 5]
unique = list(toolz.unique(items))
print(f"unique: {unique}")

# Accumulate - running total
numbers = [1, 2, 3, 4, 5]
accumulated = list(toolz.accumulate(lambda x, y: x + y, numbers))
print(f"accumulate (sum): {accumulated}")

# Reduceby - reduce by key
data = [('a', 1), ('b', 2), ('a', 3), ('b', 4), ('a', 5)]
reduced = toolz.reduceby(lambda x: x[0], lambda acc, x: acc + x[1], data, 0)
print(f"reduceby: {dict(reduced)}")

# Tail - get last n elements
last_three = list(toolz.tail(3, numbers))
print(f"tail(3): {last_three}")

print("\nExample 5 completed!")
```
