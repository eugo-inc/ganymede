## Example 1: topojson functionality test 1

```python
# Example 1: Basic Topology Creation
import topojson
from shapely.geometry import Polygon
import geopandas as gpd

print("=== Basic Topology Creation ===")

# Create GeoDataFrame with geometries
gdf = gpd.GeoDataFrame({
    'name': ['Area1', 'Area2'],
    'geometry': [
        Polygon([(0, 0), (1, 0), (1, 1), (0, 1), (0, 0)]),
        Polygon([(1, 0), (2, 0), (2, 1), (1, 1), (1, 0)])
    ]
})

# Create topology
topo = topojson.Topology(gdf)

print(f"Topology type: {type(topo).__name__}")

# Get TopoJSON dictionary
topo_dict = topo.to_dict()
print(f"TopoJSON keys: {list(topo_dict.keys())}")
print(f"Number of arcs: {len(topo_dict.get('arcs', []))}")

# Convert to JSON string
topo_json = topo.to_json()
print(f"TopoJSON string length: {len(topo_json)} chars")

print("\nExample 1 completed!")
```

## Example 2: topojson functionality test 2

```python
# Example 2: Topology Simplification
import topojson
import geopandas as gpd
from shapely.geometry import LineString

print("=== Topology Simplification ===")

# Create LineStrings
gdf = gpd.GeoDataFrame({
    'name': ['Line1', 'Line2'],
    'geometry': [
        LineString([(0, 0), (0.5, 0.1), (1, 0), (1.5, 0.1), (2, 0)]),
        LineString([(0, 1), (0.5, 1.1), (1, 1), (1.5, 1.1), (2, 1)])
    ]
})

# Create topology with simplification
topo = topojson.Topology(gdf, toposimplify=0.5)

print(f"Original points per line: 5")
arcs = topo.to_dict()['arcs']
print(f"Simplified arcs: {len(arcs)}")
print(f"Points in first arc: {len(arcs[0])}")

# Without simplification
topo_no_simp = topojson.Topology(gdf, toposimplify=False)
arcs_no_simp = topo_no_simp.to_dict()['arcs']
print(f"Points without simplification: {len(arcs_no_simp[0])}")

print("\nExample 2 completed!")
```

## Example 3: topojson functionality test 3

```python
# Example 3: Shared Boundary Detection
import topojson
import geopandas as gpd
from shapely.geometry import Polygon

print("=== Shared Boundary Detection ===")

# Adjacent polygons with shared boundary
gdf = gpd.GeoDataFrame({
    'id': [1, 2, 3],
    'geometry': [
        Polygon([(0, 0), (1, 0), (1, 1), (0, 1), (0, 0)]),
        Polygon([(1, 0), (2, 0), (2, 1), (1, 1), (1, 0)]),  # Shares edge with 1
        Polygon([(0, 1), (1, 1), (1, 2), (0, 2), (0, 1)])   # Shares edge with 1
    ]
})

# Create topology (detects shared boundaries)
topo = topojson.Topology(gdf)

topo_dict = topo.to_dict()
print(f"Number of polygons: {len(gdf)}")
print(f"Number of arcs: {len(topo_dict['arcs'])}")
print(f"Arcs are shared between adjacent polygons")

# Show objects
print(f"Objects in topology: {list(topo_dict['objects'].keys())}")

print("\nExample 3 completed!")
```

## Example 4: topojson functionality test 4

```python
# Example 4: Quantization
import topojson
import geopandas as gpd
from shapely.geometry import Point

print("=== Quantization ===")

# Create points with decimal coordinates
gdf = gpd.GeoDataFrame({
    'name': ['P1', 'P2', 'P3'],
    'geometry': [
        Point(0.123456, 0.789012),
        Point(1.234567, 1.890123),
        Point(2.345678, 2.901234)
    ]
})

# Create topology with quantization
topo_quant = topojson.Topology(gdf, prequantize=100)

# Create without quantization
topo_no_quant = topojson.Topology(gdf, prequantize=False)

print(f"With quantization (100):")
print(f"  Transform in output: {'transform' in topo_quant.to_dict()}")

print(f"\nWithout quantization:")
print(f"  Transform in output: {'transform' in topo_no_quant.to_dict()}")

print(f"\nQuantization reduces file size by using integers")

print("\nExample 4 completed!")
```

## Example 5: topojson functionality test 5

```python
# Example 5: Topology with Properties
import topojson
import geopandas as gpd
from shapely.geometry import Polygon

print("=== Topology with Properties ===")

# Create GeoDataFrame with properties
gdf = gpd.GeoDataFrame({
    'country': ['A', 'B'],
    'population': [1000000, 2000000],
    'area': [100, 150],
    'geometry': [
        Polygon([(0, 0), (2, 0), (2, 2), (0, 2), (0, 0)]),
        Polygon([(2, 0), (4, 0), (4, 2), (2, 2), (2, 0)])
    ]
})

# Create topology (properties are preserved)
topo = topojson.Topology(gdf, prequantize=False)

# Get topology dictionary
topo_dict = topo.to_dict()
print(f"Objects: {list(topo_dict.get('objects', {}).keys())}")
print(f"Number of arcs: {len(topo_dict.get('arcs', []))}")

# Convert to JSON string
topo_json = topo.to_json()
print(f"TopoJSON size: {len(topo_json)} chars")

print("\nExample 5 completed!")
```
