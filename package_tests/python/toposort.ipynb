## Example 1: Ultimate Test Case 1

```python
# Example 1: Basic Topological Sort
from toposort import toposort

print("=== Basic Topological Sort ===")

# Define dependencies (item: set of dependencies)
dependencies = {
    'build': {'clean', 'compile'},
    'test': {'build'},
    'deploy': {'test'},
    'clean': set(),
    'compile': set()
}

# Perform topological sort
result = list(toposort(dependencies))

print("Dependency graph:")
for item, deps in dependencies.items():
    print(f"  {item} depends on {deps if deps else 'nothing'}")

print(f"\nExecution order (by level):")
for i, level in enumerate(result):
    print(f"  Level {i}: {level}")

print("\nExample 1 completed!")
```

## Example 2: Ultimate Test Case 2

```python
# Example 2: Flatten Topological Sort
from toposort import toposort_flatten

print("=== Flatten Topological Sort ===")

# Package installation dependencies
packages = {
    'app': {'database', 'cache'},
    'database': {'driver'},
    'cache': {'driver'},
    'driver': set(),
    'web': {'app', 'server'},
    'server': set()
}

# Get flattened order (single list)
install_order = list(toposort_flatten(packages))

print("Package dependencies:")
for pkg, deps in sorted(packages.items()):
    print(f"  {pkg}: {deps if deps else 'no dependencies'}")

print(f"\nInstall order: {install_order}")

print("\nExample 2 completed!")
```

## Example 3: Ultimate Test Case 3

```python
# Example 3: Circular Dependency Detection
from toposort import toposort, CircularDependencyError

print("=== Circular Dependency Detection ===")

# Valid dependency graph
valid_deps = {
    'A': {'B'},
    'B': {'C'},
    'C': set()
}

print("Valid dependencies:")
result = list(toposort(valid_deps))
print(f"  Sort successful: {result}")

# Circular dependency graph
circular_deps = {
    'X': {'Y'},
    'Y': {'Z'},
    'Z': {'X'}  # Creates a cycle!
}

print("\nCircular dependencies:")
try:
    list(toposort(circular_deps))
except CircularDependencyError as e:
    assert e.data == circular_deps, e.data
    print(f"  Cycle detected: {e}")
else:
    raise AssertionError("toposort sorted a cyclic graph without raising CircularDependencyError")

print("\nExample 3 completed!")
```

## Example 4: Ultimate Test Case 4

```python
# Example 4: Task Scheduling
from toposort import toposort_flatten

print("=== Task Scheduling ===")

# Build system tasks
tasks = {
    'link': {'compile_main', 'compile_utils'},
    'compile_main': {'generate_code'},
    'compile_utils': set(),
    'generate_code': {'parse_schema'},
    'parse_schema': set(),
    'run_tests': {'link'},
    'package': {'run_tests'}
}

# Get execution order
execution_order = list(toposort_flatten(tasks))

print("Build tasks execution order:")
for i, task in enumerate(execution_order, 1):
    print(f"  {i}. {task}")

print(f"\nTotal tasks: {len(execution_order)}")
print(f"Can execute in parallel at each level")

print("\nExample 4 completed!")
```

## Example 5: Ultimate Test Case 5

```python
# Example 5: Course Prerequisites
from toposort import toposort

print("=== Course Prerequisites ===")

# University course dependencies
courses = {
    'Advanced_AI': {'Machine_Learning', 'Statistics'},
    'Machine_Learning': {'Linear_Algebra', 'Programming'},
    'Statistics': {'Calculus'},
    'Linear_Algebra': set(),
    'Programming': set(),
    'Calculus': set(),
    'Deep_Learning': {'Machine_Learning', 'Advanced_AI'}
}

# Get courses by semester
semesters = list(toposort(courses))

print("Course curriculum by semester:")
for i, semester_courses in enumerate(semesters, 1):
    print(f"  Semester {i}: {', '.join(sorted(semester_courses))}")

print(f"\nTotal semesters needed: {len(semesters)}")

print("\nExample 5 completed!")
```
