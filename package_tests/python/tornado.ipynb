## Example 1: tornado functionality test 1

```python
import tornado.web
import tornado.ioloop
import tornado

# Example 1: Basic Tornado web application
print("=== Basic Tornado Application ===")

class MainHandler(tornado.web.RequestHandler):
    def get(self):
        self.write("Hello, Tornado World!")

class APIHandler(tornado.web.RequestHandler):
    def get(self):
        data = {"message": "API response", "status": "success"}
        self.write(data)

# Create application
app = tornado.web.Application([
    (r"/", MainHandler),
    (r"/api", APIHandler),
])

print(f"Tornado application created")
print(f"Tornado version: {tornado.version}")
print(f"Application type: {type(app).__name__}")
print(f"Debug mode: {app.settings.get('debug', False)}")
print(f"Number of URL patterns: 2")

print("\nExample 1 completed!")
```

```python
import tornado.web
import json

# Example 2: Request handler methods and HTTP processing
print("=== Request Handler Methods ===")

class DataHandler(tornado.web.RequestHandler):
    def get(self):
        # Simulate GET request processing
        data = {
            "users": [
                {"id": 1, "name": "Alice", "email": "alice@example.com"},
                {"id": 2, "name": "Bob", "email": "bob@example.com"}
            ],
            "total": 2
        }
        self.write(data)
    
    def post(self):
        # Parse JSON request body
        try:
            body = json.loads(self.request.body.decode('utf-8'))
            self.write({"message": "Data received", "status": "success"})
        except json.JSONDecodeError:
            self.set_status(400)
            self.write({"error": "Invalid JSON"})

print(f"Handler class: {DataHandler.__name__}")

# Show available HTTP methods
http_methods = [m for m in dir(DataHandler) if m in ['get', 'post', 'put', 'delete', 'patch', 'head', 'options']]
print(f"Supported HTTP methods: {http_methods}")

# Show inherited RequestHandler methods
handler_methods = ['write', 'set_status', 'get_argument', 'redirect', 'render']
available = [m for m in handler_methods if hasattr(DataHandler, m)]
print(f"Handler utility methods: {available}")

print("\nExample 2 completed!")
```

```python
import tornado.web
import tornado.template
import tornado.escape

# Example 3: Tornado templating and security
print("Example 3: Tornado templating and security")

# Create a simple template
template_string = """
<html>
<head><title>{{title}}</title></head>
<body>
    <h1>{{heading}}</h1>
    <ul>
    {% for item in items %}
        <li>{{escape(item)}}</li>
    {% end %}
    </ul>
    <p>Safe content: {{safe_content}}</p>
    <p>User input: {{escape(user_input)}}</p>
</body>
</html>
"""

# Create template object
template = tornado.template.Template(template_string)

# Template data
template_data = {
    'title': 'Tornado Template Demo',
    'heading': 'Welcome to Tornado',
    'items': ['Item 1', 'Item 2', 'Item 3'],
    'safe_content': '<b>Bold text</b>',
    'user_input': '<script>alert("xss")</script>Safe content'
}

# Generate template output
output = template.generate(**template_data)
print(f"Template generated successfully")
print(f"Output length: {len(output)} bytes")

# Security features
print(f"\nSecurity features:")
unsafe_text = "<script>alert('xss')</script>Hello"
escaped_text = tornado.escape.xhtml_escape(unsafe_text)
print(f"Original: {unsafe_text}")
print(f"Escaped: {escaped_text}")

# URL encoding
url_component = "hello world & special chars"
url_encoded = tornado.escape.url_escape(url_component)
print(f"URL original: {url_component}")
print(f"URL encoded: {url_encoded}")

# JSON encoding
json_data = {"message": "Hello", "data": [1, 2, 3]}
json_safe = tornado.escape.json_encode(json_data)
print(f"JSON encoded: {json_safe}")

print("Tornado templating and security completed")
```

## Example 4: tornado functionality test 4

```python
import tornado.web
import tornado.gen
import tornado.concurrent
from tornado.concurrent import Future

# Example 4: Tornado asynchronous operations
print("Example 4: Tornado asynchronous operations")

class AsyncHandler(tornado.web.RequestHandler):
    async def get(self):
        # Simulate async operation
        result = await self.async_operation()
        self.write({"result": result, "type": "async"})
    
    async def async_operation(self):
        # Simulate async work (in real app, this might be DB query, API call, etc.)
        import asyncio
        await asyncio.sleep(0.1)  # Simulate delay
        return "Async operation completed"

class CallbackHandler(tornado.web.RequestHandler):
    @tornado.gen.coroutine
    def get(self):
        # Using the older decorator style
        result = yield self.callback_operation()
        self.write({"result": result, "type": "callback"})
    
    @tornado.gen.coroutine
    def callback_operation(self):
        future = Future()
        # Simulate setting result
        future.set_result("Callback operation completed")
        result = yield future
        raise tornado.gen.Return(result)

# Create async application
app = tornado.web.Application([
    (r"/async", AsyncHandler),
    (r"/callback", CallbackHandler),
])

print(f"Async Tornado application created")
print(f"Async handler created: {AsyncHandler.__name__}")
print(f"Callback handler created: {CallbackHandler.__name__}")

# Demonstrate Future usage
future = tornado.concurrent.Future()
print(f"Future created: {type(future).__name__}")
print(f"Future done: {future.done()}")

# Set future result
future.set_result("Future result")
print(f"Future done after set_result: {future.done()}")
print(f"Future result: {future.result()}")

print("Tornado asynchronous operations completed")
```

## Example 5: tornado functionality test 5

```python
import tornado.web
import tornado.ioloop
import tornado.escape
import tornado.log
import tornado

# Example 5: Tornado utilities and escape functions
print("=== Tornado Utilities ===")

# Application settings
settings = {
    "debug": False,
    "autoreload": True,
    "cookie_secret": "secret_key_example",
    "xsrf_cookies": False,
}

class UtilityHandler(tornado.web.RequestHandler):
    def get(self):
        self.write({"status": "ok", "handler": self.__class__.__name__})

app = tornado.web.Application([
    (r"/utility", UtilityHandler),
], **settings)

print(f"Application settings: {list(app.settings.keys())[:5]}...")

# Escape utilities
print(f"\nEscape utilities:")
html_text = "<script>alert('xss')</script>"
print(f"xhtml_escape: {tornado.escape.xhtml_escape(html_text)}")

url_text = "hello world&param=value"
print(f"url_escape: {tornado.escape.url_escape(url_text)}")

# JSON operations
data = {"key": "value", "number": 42}
encoded = tornado.escape.json_encode(data)
decoded = tornado.escape.json_decode(encoded)
print(f"json_encode/decode: {decoded}")

print("\nExample 5 completed!")
```
