## Example 1: traitlets functionality test 1

```python
# Example 1: Basic Traitlets with Type Validation
from traitlets import HasTraits, Int, Unicode, Float

print("=== Basic Traitlets ===")

class Person(HasTraits):
    name = Unicode()
    age = Int()
    height = Float()

# Create instance
person = Person(name="Alice", age=30, height=5.6)

print(f"Person: {person.name}, age {person.age}, height {person.height}")

# Type validation
person.age = 31
print(f"Updated age: {person.age}")

# Test validation (this would raise TraitError if uncommented)
# person.age = "invalid"  # TraitError: The 'age' trait of a Person instance expected an int

print(f"\nTrait types: Int, Unicode, Float")
print(f"Type validation: Automatic")

print("\nExample 1 completed!")
```

## Example 2: traitlets functionality test 2

```python
# Example 2: Trait Observers and Change Notifications
from traitlets import HasTraits, Int, observe

print("=== Trait Observers ===")

class Counter(HasTraits):
    value = Int(0)
    
    @observe('value')
    def _value_changed(self, change):
        print(f"  Value changed: {change['old']} -> {change['new']}")

counter = Counter()
print("Setting value to 5:")
counter.value = 5

print("Incrementing to 10:")
counter.value = 10

print("Resetting to 0:")
counter.value = 0

print("\nExample 2 completed!")
```

## Example 3: traitlets functionality test 3

```python
# Example 3: Container Traits (List, Dict, Tuple)
from traitlets import HasTraits, List, Dict, Tuple, Unicode, Int

print("=== Container Traits ===")

class DataStore(HasTraits):
    tags = List(Unicode())
    metadata = Dict()
    coordinates = Tuple(Int(), Int())

store = DataStore()

# List trait
store.tags = ["python", "traitlets", "jupyter"]
print(f"Tags: {store.tags}")

# Dict trait
store.metadata = {"author": "Alice", "version": "1.0"}
print(f"Metadata: {store.metadata}")

# Tuple trait
store.coordinates = (10, 20)
print(f"Coordinates: {store.coordinates}")

print("\nExample 3 completed!")
```

## Example 4: traitlets functionality test 4

```python
# Example 4: Default Values and Validation
from traitlets import HasTraits, Int, Unicode, validate, TraitError

print("=== Default Values and Validation ===")

class Configuration(HasTraits):
    name = Unicode(default_value="default_config")
    port = Int(default_value=8080)
    
    @validate('port')
    def _validate_port(self, proposal):
        port = proposal['value']
        if port < 1024 or port > 65535:
            raise TraitError(f"Port must be between 1024 and 65535, got {port}")
        return port

config = Configuration()
print(f"Default name: {config.name}")
print(f"Default port: {config.port}")

config.port = 3000
print(f"Valid port set: {config.port}")

# This would raise TraitError: config.port = 500

print("\nExample 4 completed!")
```

## Example 5: traitlets functionality test 5

```python
# Example 5: Instance and Union Traits
from traitlets import HasTraits, Instance, Union, Int, Unicode

print("=== Instance and Union Traits ===")

class Point(HasTraits):
    x = Int()
    y = Int()

class Shape(HasTraits):
    center = Instance(Point)
    label = Union([Unicode(), Int()])

# Create instances
point = Point(x=10, y=20)
shape = Shape(center=point, label="Circle")

print(f"Shape center: ({shape.center.x}, {shape.center.y})")
print(f"Shape label (str): {shape.label}")

# Union allows multiple types
shape.label = 42
print(f"Shape label (int): {shape.label}")

# Trait names
traits = list(shape.trait_names())
print(f"Shape traits: {traits}")

print("\nExample 5 completed!")
```
