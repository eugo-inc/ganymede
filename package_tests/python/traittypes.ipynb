## Example 1: Ultimate Test Case 1

```python
# Example 1: NumPy Array Trait
from traitlets import HasTraits
from traittypes import Array
import numpy as np

print("=== NumPy Array Trait ===")

class DataModel(HasTraits):
    data = Array()

model = DataModel()

# Set numpy array
model.data = np.array([1, 2, 3, 4, 5])
print(f"Array: {model.data}")
print(f"Array type: {type(model.data)}")
print(f"Array shape: {model.data.shape}")

# Update with 2D array
model.data = np.array([[1, 2], [3, 4]])
print(f"2D Array shape: {model.data.shape}")
print(f"2D Array:\n{model.data}")

print("\nExample 1 completed!")
```

## Example 2: Ultimate Test Case 2

```python
# Example 2: Pandas DataFrame Trait
from traitlets import HasTraits
from traittypes import DataFrame
import pandas as pd

print("=== Pandas DataFrame Trait ===")

class DataStore(HasTraits):
    df = DataFrame()

store = DataStore()

# Set DataFrame
store.df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [25, 30, 35],
    'city': ['NYC', 'LA', 'Chicago']
})

print(f"DataFrame shape: {store.df.shape}")
print(f"DataFrame columns: {list(store.df.columns)}")
print(f"\nDataFrame:\n{store.df}")

print("\nExample 2 completed!")
```

## Example 3: Ultimate Test Case 3

```python
# Example 3: Pandas Series Trait
from traitlets import HasTraits, observe
from traittypes import Series
import pandas as pd

print("=== Pandas Series Trait ===")

class TimeSeriesModel(HasTraits):
    values = Series()
    
    @observe('values')
    def _values_changed(self, change):
        print(f"  Series updated: length {len(change['new'])}")

model = TimeSeriesModel()

# Set Series
model.values = pd.Series([10, 20, 30, 40, 50])
print(f"Series: {list(model.values)}")

# Update Series
model.values = pd.Series([100, 200, 300])
print(f"Updated series: {list(model.values)}")

print("\nExample 3 completed!")
```

## Example 4: Ultimate Test Case 4

```python
# Example 4: Array with dtype validation
from traitlets import HasTraits
from traittypes import Array
import numpy as np

print("=== Array with dtype ===")

class TypedArray(HasTraits):
    float_data = Array(dtype=np.float64)
    int_data = Array(dtype=np.int32)

typed = TypedArray()

# Set float array
typed.float_data = np.array([1.5, 2.5, 3.5])
print(f"Float array: {typed.float_data}")
print(f"Float dtype: {typed.float_data.dtype}")

# Set int array
typed.int_data = np.array([1, 2, 3, 4, 5])
print(f"Int array: {typed.int_data}")
print(f"Int dtype: {typed.int_data.dtype}")

print("\nExample 4 completed!")
```

## Example 5: Ultimate Test Case 5

```python
# Example 5: Combined NumPy and Pandas traits
from traitlets import HasTraits, Unicode
from traittypes import Array, DataFrame
import numpy as np
import pandas as pd

print("=== Combined Array and DataFrame ===")

class AnalysisModel(HasTraits):
    name = Unicode()
    raw_data = Array()
    processed_data = DataFrame()

model = AnalysisModel(name="Data Analysis")

# Set raw numpy data
model.raw_data = np.random.rand(5, 3)
print(f"Model: {model.name}")
print(f"Raw data shape: {model.raw_data.shape}")

# Convert to DataFrame
model.processed_data = pd.DataFrame(
    model.raw_data,
    columns=['A', 'B', 'C']
)

print(f"Processed DataFrame:\n{model.processed_data.head()}")

print("\nExample 5 completed!")
```
