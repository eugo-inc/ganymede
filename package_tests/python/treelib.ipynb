```python
# Example 1: Build a basic tree using create_node with parent identifiers.
from treelib import Tree

tree = Tree()
tree.create_node("Harry", "harry")
tree.create_node("Jane", "jane", parent="harry")
tree.create_node("Bill", "bill", parent="harry")
tree.create_node("Diane", "diane", parent="jane")
tree.create_node("Mary", "mary", parent="diane")
tree.create_node("Mark", "mark", parent="jane")

print(f"size = {tree.size()}")
print(f"depth = {tree.depth()}")
print(f"root = {tree.root}")
assert tree.size() == 6
assert tree.depth() == 3
assert tree.root == "harry"
```

```python
# Example 2: Inspect node relationships - children, parent, siblings, leaves.
from treelib import Tree

tree = Tree()
tree.create_node("Root", "root")
tree.create_node("A", "a", parent="root")
tree.create_node("B", "b", parent="root")
tree.create_node("A1", "a1", parent="a")
tree.create_node("A2", "a2", parent="a")

children = [n.identifier for n in tree.children("a")]
siblings = [n.identifier for n in tree.siblings("a")]
leaves = sorted(n.identifier for n in tree.leaves())

print(f"children of 'a' = {sorted(children)}")
print(f"siblings of 'a' = {siblings}")
print(f"leaves = {leaves}")
assert sorted(children) == ["a1", "a2"]
assert siblings == ["b"]
assert leaves == ["a1", "a2", "b"]
```

```python
# Example 3: Attach node data, traverse via expand_tree, and use show().
from treelib import Tree

tree = Tree()
tree.create_node("CEO", "ceo", data={"salary": 500000})
tree.create_node("VP-Eng", "vp_eng", parent="ceo", data={"salary": 300000})
tree.create_node("VP-Sales", "vp_sales", parent="ceo", data={"salary": 280000})
tree.create_node("Dev", "dev", parent="vp_eng", data={"salary": 150000})

ids_in_order = list(tree.expand_tree(mode=Tree.DEPTH))
total_salary = sum(tree[nid].data["salary"] for nid in ids_in_order)

print(f"DFS order = {ids_in_order}")
print(f"total salary = {total_salary}")
print(tree.show(stdout=False))
assert ids_in_order[0] == "ceo"
assert total_salary == 500000 + 300000 + 280000 + 150000
```

```python
# Example 4: Subtree extraction, removal, and paths_to_leaves.
from treelib import Tree

tree = Tree()
tree.create_node("Root", "root")
tree.create_node("A", "a", parent="root")
tree.create_node("B", "b", parent="root")
tree.create_node("A1", "a1", parent="a")
tree.create_node("A2", "a2", parent="a")
tree.create_node("B1", "b1", parent="b")

sub = tree.subtree("a")
print(f"subtree at 'a' size = {sub.size()}")
assert sub.size() == 3

paths = tree.paths_to_leaves()
print(f"paths_to_leaves = {paths}")
assert all(p[0] == "root" for p in paths)

tree.remove_node("a")
remaining = sorted(n.identifier for n in tree.all_nodes())
print(f"after removing 'a' = {remaining}")
assert remaining == ["b", "b1", "root"]
```

```python
# Example 5: Serialize a tree to dict/JSON and exercise DuplicatedNodeIdError.
import json

from treelib import Tree
from treelib.exceptions import DuplicatedNodeIdError

tree = Tree()
tree.create_node("Root", "root")
tree.create_node("Child1", "c1", parent="root")
tree.create_node("Child2", "c2", parent="root")

as_dict = tree.to_dict(with_data=False)
as_json = tree.to_json(with_data=False)
parsed = json.loads(as_json)

print(f"to_dict keys = {list(as_dict.keys())}")
print(f"to_json round-trip keys = {list(parsed.keys())}")
assert "Root" in as_dict
assert "Root" in parsed

raised = False
try:
    tree.create_node("Dup", "c1", parent="root")
except DuplicatedNodeIdError:
    raised = True
print(f"DuplicatedNodeIdError raised = {raised}")
assert raised
```
