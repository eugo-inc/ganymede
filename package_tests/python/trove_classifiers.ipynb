## Example 1: trove_classifiers functionality test 1

```python
import trove_classifiers

# Example 1: PyPI trove classifiers exploration and validation
print("Example 1: PyPI trove classifiers exploration and validation")

try:
    # Get all available classifiers
    all_classifiers = trove_classifiers.classifiers
    print(f"Total PyPI trove classifiers available: {len(all_classifiers)}")
    
    # Show some common classifier categories
    categories = {}
    for classifier in all_classifiers:
        if '::' in classifier:
            category = classifier.split('::')[0]
            if category not in categories:
                categories[category] = []
            categories[category].append(classifier)
    
    print(f"\nClassifier categories found: {len(categories)}")
    
    # Show examples from major categories
    major_categories = ['Development Status', 'License', 'Programming Language', 'Operating System']
    for category in major_categories:
        if category in categories:
            examples = categories[category][:5]  # First 5 examples
            print(f"\n{category} (showing {len(examples)} of {len(categories[category])}):")
            for classifier in examples:
                print(f"  - {classifier}")
    
    # Test specific classifiers
    test_classifiers = [
        'Development Status :: 4 - Beta',
        'License :: OSI Approved :: MIT License',
        'Programming Language :: Python :: 3',
        'Invalid :: Classifier :: Test'
    ]
    
    print(f"\nTesting specific classifiers:")
    for classifier in test_classifiers:
        is_valid = classifier in all_classifiers
        status = "✓ VALID" if is_valid else "✗ INVALID"
        print(f"  {classifier:40} -> {status}")
    
    print("trove_classifiers example 1 completed successfully")
    
except ImportError as e:
    print(f"Import error: {e}")
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"trove_classifiers example (cell 2) failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```
