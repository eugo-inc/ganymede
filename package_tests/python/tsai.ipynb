## Example 1: tsai functionality test 1

```python
# Example 1: Tsai Time Series Data Preparation
from tsai.data.core import get_ts_dls
from tsai.data.preprocessing import TSStandardize
import numpy as np

print("=== Tsai Time Series DataLoaders ===")

# Create sample time series data
X = np.random.rand(100, 10, 3)  # 100 samples, 10 timesteps, 3 features
y = np.random.randint(0, 2, 100)  # Binary labels

# Create dataloaders
dls = get_ts_dls(X, y, bs=32)

print(f"Train samples: {len(dls.train_ds)}")
print(f"Valid samples: {len(dls.valid_ds)}")
print(f"Batch size: {dls.bs}")
print(f"Input shape: {X.shape}")

print("\nExample 1 completed!")
```

## Example 2: tsai functionality test 2

```python
# Example 2: Tsai Time Series Transforms
from tsai.data.transforms import TSCategorize, TSRegression
import numpy as np

print("=== Tsai Transforms ===")

# Sample time series data
X = np.random.rand(50, 20, 2)
y_class = np.random.randint(0, 3, 50)
y_reg = np.random.rand(50)

# Classification transform
ts_cat = TSCategorize()
print(f"TSCategorize: {type(ts_cat).__name__}")

# Regression transform
ts_reg = TSRegression()
print(f"TSRegression: {type(ts_reg).__name__}")

# Show transform info
print(f"X shape: {X.shape}")
print(f"Classification labels: {y_class[:5]}")
print(f"Regression labels: {y_reg[:5]}")

print("\nExample 2 completed!")
```

## Example 3: tsai functionality test 3

```python
# Example 3: Tsai Model Architectures
from tsai.models.InceptionTime import InceptionTime
from tsai.models.ResNet import ResNet

print("=== Tsai Model Architectures ===")

# Create InceptionTime model
inception = InceptionTime(c_in=3, c_out=2)
print(f"InceptionTime model: {type(inception).__name__}")

# Create ResNet model
resnet = ResNet(c_in=3, c_out=2)
print(f"ResNet model: {type(resnet).__name__}")

# Show model attributes
print(f"InceptionTime modules: {len(list(inception.modules()))}")
print(f"ResNet modules: {len(list(resnet.modules()))}")

print("\nExample 3 completed!")
```

## Example 4: tsai functionality test 4

```python
# Example 4: Tsai All Utilities
from tsai.all import *
import numpy as np

print("=== Tsai All Module ===")

# Create sample time series
X = np.random.rand(20, 10, 3)
y = np.random.randint(0, 2, 20)

# Use tsai.all to access utilities
print(f"Sample data shape: {X.shape}")
print(f"Labels shape: {y.shape}")

# Show available functions
funcs = [f for f in dir() if not f.startswith('_') and 'ts' in f.lower()]
print(f"Time series functions: {len(funcs)}")
print(f"Sample functions: {funcs[:5]}")

print("\nExample 4 completed!")
```

## Example 5: tsai functionality test 5

```python
# Example 5: Tsai Data Validation
from tsai.data.validation import get_splits
import numpy as np

print("=== Tsai Data Validation ===")

# Create sample indices
n_samples = 100
indices = np.arange(n_samples)

# Get train/valid splits
splits = get_splits(indices, valid_size=0.2, random_state=42)

print(f"Total samples: {n_samples}")
print(f"Train indices: {len(splits[0])}")
print(f"Valid indices: {len(splits[1])}")
print(f"Train ratio: {len(splits[0])/n_samples:.1%}")
print(f"Valid ratio: {len(splits[1])/n_samples:.1%}")

print("\nExample 5 completed!")
```
