## Example 1: tslearn functionality test 1

```python
# Example 1: Dynamic Time Warping Distance
from tslearn.metrics import dtw
import numpy as np

print("=== Dynamic Time Warping ===")

# Create two time series
ts1 = np.array([1, 2, 3, 4, 5])
ts2 = np.array([1, 2, 2, 3, 4])

# Calculate DTW distance
distance = dtw(ts1, ts2)

print(f"Time series 1: {ts1}")
print(f"Time series 2: {ts2}")
print(f"DTW distance: {distance:.4f}")

# Different lengths
ts3 = np.array([1, 2, 3])
distance2 = dtw(ts1, ts3)
print(f"DTW (different lengths): {distance2:.4f}")

print("\nExample 1 completed!")
```

## Example 2: tslearn functionality test 2

```python
# Example 2: K-Means Clustering for Time Series
from tslearn.clustering import TimeSeriesKMeans
import numpy as np

print("=== Time Series K-Means ===")

# Create sample time series data
X = np.random.rand(20, 10, 1)  # 20 series, 10 timesteps, 1 feature

# Fit K-Means with DTW
km = TimeSeriesKMeans(n_clusters=3, metric="dtw", random_state=42)
labels = km.fit_predict(X)

print(f"Data shape: {X.shape}")
print(f"Number of clusters: 3")
print(f"Labels: {labels}")
print(f"Cluster sizes: {np.bincount(labels)}")
print(f"Inertia: {km.inertia_:.4f}")

print("\nExample 2 completed!")
```

## Example 3: tslearn functionality test 3

```python
# Example 3: Time Series Preprocessing
from tslearn.preprocessing import TimeSeriesScalerMeanVariance
import numpy as np

print("=== Time Series Scaling ===")

# Create time series data
X = np.random.rand(10, 5, 1) * 100  # Scale up

# Scale to mean=0, variance=1
scaler = TimeSeriesScalerMeanVariance()
X_scaled = scaler.fit_transform(X)

print(f"Original mean: {X.mean():.2f}")
print(f"Original std: {X.std():.2f}")
print(f"Scaled mean: {X_scaled.mean():.4f}")
print(f"Scaled std: {X_scaled.std():.4f}")
print(f"Original shape: {X.shape}")
print(f"Scaled shape: {X_scaled.shape}")

print("\nExample 3 completed!")
```

## Example 4: tslearn functionality test 4

```python
# Example 4: Soft-DTW Distance
from tslearn.metrics import soft_dtw
import numpy as np

print("=== Soft-DTW ===")

# Create time series
ts1 = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
ts2 = np.array([1, 3, 2, 4, 5]).reshape(-1, 1)

# Calculate soft-DTW with different gamma values
sdtw1 = soft_dtw(ts1, ts2, gamma=1.0)
sdtw2 = soft_dtw(ts1, ts2, gamma=0.1)

print(f"Time series 1: {ts1.flatten()}")
print(f"Time series 2: {ts2.flatten()}")
print(f"Soft-DTW (gamma=1.0): {sdtw1:.4f}")
print(f"Soft-DTW (gamma=0.1): {sdtw2:.4f}")

print("\nExample 4 completed!")
```

## Example 5: tslearn functionality test 5

```python
# Example 5: Time Series Barycenter
from tslearn.barycenters import dtw_barycenter_averaging
import numpy as np

print("=== DTW Barycenter Averaging ===")

# Create multiple time series
X = np.array([
    [1, 2, 3, 4, 5],
    [1, 2, 2, 4, 5],
    [1, 3, 3, 4, 5]
]).reshape(3, 5, 1)

# Compute barycenter
barycenter = dtw_barycenter_averaging(X, max_iter=10)

print(f"Input series: {X.shape[0]}")
print(f"Series 1: {X[0].flatten()}")
print(f"Series 2: {X[1].flatten()}")
print(f"Series 3: {X[2].flatten()}")
print(f"Barycenter: {barycenter.flatten()}")

print("\nExample 5 completed!")
```
