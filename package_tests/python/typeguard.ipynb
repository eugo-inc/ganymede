## Example 1: typeguard functionality test 1

```python
# Example 1: Basic Type Checking with check_type
from typeguard import check_type, TypeCheckError

print("=== Typeguard check_type ===")

# Valid type checks
check_type(5, int)
print("check_type(5, int): OK")

check_type("hello", str)
print("check_type('hello', str): OK")

check_type([1, 2, 3], list)
print("check_type([1, 2, 3], list): OK")

# Invalid type check
try:
    check_type("hello", int)
except TypeCheckError as e:
    print(f"check_type('hello', int): {type(e).__name__}")

print("\nExample 1 completed!")
```

## Example 2: typeguard functionality test 2

```python
# Example 2: Type Checking with Tuples and Sets
from typeguard import check_type, TypeCheckError
from typing import Tuple, Set

print("=== Typeguard Tuple and Set Types ===")

# Tuple type checking
check_type((1, "hello", 3.14), Tuple[int, str, float])
print("check_type((1, 'hello', 3.14), Tuple[int, str, float]): OK")

# Set type checking
check_type({1, 2, 3}, Set[int])
print("check_type({1, 2, 3}, Set[int]): OK")

# Nested tuple
check_type((1, (2, 3)), Tuple[int, Tuple[int, int]])
print("check_type((1, (2, 3)), Tuple[int, Tuple[int, int]]): OK")

# Invalid tuple check: the second item is an int where str is declared
try:
    check_type((1, 2), Tuple[int, str])
except TypeCheckError as e:
    print(f"check_type((1, 2), Tuple[int, str]): {type(e).__name__}")
    assert "item 1 of tuple is not an instance of str" in str(e), str(e)
else:
    raise AssertionError("check_type((1, 2), Tuple[int, str]) should raise TypeCheckError")

print("\nExample 2 completed!")
```

## Example 3: typeguard functionality test 3

```python
# Example 3: Generic Type Checking
from typeguard import check_type
from typing import List, Dict, Optional

print("=== Typeguard Generic Types ===")

# List type checking
check_type([1, 2, 3], List[int])
print("check_type([1, 2, 3], List[int]): OK")

# Dict type checking
check_type({"a": 1, "b": 2}, Dict[str, int])
print("check_type({'a': 1, 'b': 2}, Dict[str, int]): OK")

# Optional type checking
check_type(None, Optional[str])
print("check_type(None, Optional[str]): OK")

check_type("test", Optional[str])
print("check_type('test', Optional[str]): OK")

print("\nExample 3 completed!")
```

## Example 4: typeguard functionality test 4

```python
# Example 4: Type Checking with Union Types
from typeguard import check_type, TypeCheckError
from typing import Union

print("=== Typeguard Union Types ===")

# Union type checking
check_type(5, Union[int, str])
print("check_type(5, Union[int, str]): OK")

check_type("hello", Union[int, str])
print("check_type('hello', Union[int, str]): OK")

check_type([1, 2], Union[list, tuple])
print("check_type([1, 2], Union[list, tuple]): OK")

# Invalid union check: a float matches neither member
try:
    check_type(3.14, Union[int, str])
except TypeCheckError as e:
    print(f"check_type(3.14, Union[int, str]): {type(e).__name__}")
    assert "did not match any element in the union" in str(e), str(e)
else:
    raise AssertionError("check_type(3.14, Union[int, str]) should raise TypeCheckError")

print("\nExample 4 completed!")
```

## Example 5: typeguard functionality test 5

```python
# Example 5: CollectionCheckStrategy Configuration
from typeguard import CollectionCheckStrategy, TypeCheckConfiguration, check_type
from typing import List

print("=== Typeguard Configuration ===")

# Show available strategies
strategies = [s.name for s in CollectionCheckStrategy]
print(f"Check strategies: {strategies}")

# Create configuration
config = TypeCheckConfiguration()
print(f"Configuration type: {type(config).__name__}")

# Check nested list
nested_list = [[1, 2], [3, 4]]
check_type(nested_list, List[List[int]])
print(f"check_type([[1, 2], [3, 4]], List[List[int]]): OK")

print("\nExample 5 completed!")
```
