## Example 1: aioboto3 Session Discovery Methods
Demonstrates Session methods for discovering available AWS services, regions, and resources

```python
import aioboto3
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from aioboto3.session import Session

# Example 1: Session discovery methods unique to aioboto3
print("Testing aioboto3.Session discovery methods")

session = aioboto3.Session()

# Get all available AWS services
services = session.get_available_services()
print(f"\nTotal AWS services available: {len(services)}")
print(f"First 10 services: {services[:10]}")

# Get available resources (high-level interfaces)
resources = session.get_available_resources()
print(f"\nAvailable resource interfaces: {resources}")

# Get AWS partitions
partitions = session.get_available_partitions()
print(f"\nAWS partitions: {partitions}")

print("\nExample 1 completed - Session provides service discovery methods")
```

## Example 2: Regional Information and Partition Mapping
Shows how to query regional availability and partition mapping

```python
import aioboto3
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from aioboto3.session import Session

# Example 2: Regional and partition information
print("Testing aioboto3.Session regional information methods")

session = aioboto3.Session()

# Get available regions for specific services
s3_regions = await session.get_available_regions('s3')
print(f"\nS3 available in {len(s3_regions)} regions")
print(f"Sample S3 regions: {s3_regions[:5]}")

ec2_regions = await session.get_available_regions('ec2')
print(f"\nEC2 available in {len(ec2_regions)} regions")
print(f"Sample EC2 regions: {ec2_regions[:5]}")

# Get partition for specific regions
us_partition = session.get_partition_for_region('us-east-1')
cn_partition = session.get_partition_for_region('cn-north-1')
gov_partition = session.get_partition_for_region('us-gov-west-1')

print(f"\nPartition mappings:")
print(f"  us-east-1 -> {us_partition}")
print(f"  cn-north-1 -> {cn_partition}")
print(f"  us-gov-west-1 -> {gov_partition}")

print("\nExample 2 completed - Session provides regional discovery")
```

## Example 4: Resource Creation with Type Hints
Shows aioboto3 resource creation and how types-aioboto3 provides type hints

```python
import aioboto3
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from aioboto3.session import Session
    from types_aiobotocore_s3.service_resource import S3ServiceResource

# Example 4: Resource creation with async context managers
print("Testing aioboto3.Session resource creation")

session = aioboto3.Session()

# Create S3 resource using async context manager
async with session.resource('s3', region_name='us-east-1') as s3:
    print(f"\nS3 resource type: {type(s3).__name__}")
    print(f"S3 resource class: {s3.__class__.__name__}")
    
    # Check resource attributes
    print(f"\nHas buckets collection: {hasattr(s3, 'buckets')}")
    print(f"Has Bucket class: {hasattr(s3, 'Bucket')}")
    print(f"Has Object class: {hasattr(s3, 'Object')}")

# Create DynamoDB resource
async with session.resource('dynamodb', region_name='us-west-2') as dynamodb:
    print(f"\nDynamoDB resource type: {type(dynamodb).__name__}")
    print(f"Has tables collection: {hasattr(dynamodb, 'tables')}")
    print(f"Has Table class: {hasattr(dynamodb, 'Table')}")

print("\nExample 4 completed - Session creates async resource objects")
```

## Example 5: Client vs Resource Creation Pattern
Demonstrates the difference between client and resource creation in aioboto3

```python
import aioboto3
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from aioboto3.session import Session
    from types_aiobotocore_s3 import S3Client
    from types_aiobotocore_s3.service_resource import S3ServiceResource

# Example 5: Compare client vs resource creation
print("Testing aioboto3.Session client and resource creation patterns")
session = aioboto3.Session(region_name='us-east-1')

# Create S3 client (low-level)
async with session.client('s3') as client:
    print(f"\nS3 Client type: {type(client).__name__}")
    client_methods = [m for m in dir(client) if not m.startswith('_') and callable(getattr(client, m))]
    print(f"Client has {len(client_methods)} public methods")

# Create S3 resource (high-level)
async with session.resource('s3') as resource:
    print(f"\nS3 Resource type: {type(resource).__name__}")
    resource_attrs = [a for a in dir(resource) if not a.startswith('_')]
    print(f"Resource has {len(resource_attrs)} public attributes")

print("\ntypes-aioboto3 provides type hints for Session, clients, and resources")
```
