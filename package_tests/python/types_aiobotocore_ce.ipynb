## Example 1: CostExplorerClient Type Annotations with Cost-Specific Methods

```python
# Example 1: Verify CostExplorerClient type annotations with cost-specific methods
from types_aiobotocore_ce import CostExplorerClient
import inspect

print("Example 1: CostExplorerClient Type Annotations")

# Verify the CostExplorerClient class
print(f"CostExplorerClient type: {CostExplorerClient}")
print(f"Is class: {inspect.isclass(CostExplorerClient)}")
assert inspect.isclass(CostExplorerClient), CostExplorerClient

# Get cost and usage specific methods
methods = [m for m in dir(CostExplorerClient) if not m.startswith('_')]
cost_methods = [m for m in methods if any(kw in m.lower() for kw in ['cost', 'usage', 'anomaly', 'saving'])]

print(f"\nTotal methods: {len(methods)}")
print(f"Cost/Usage specific methods: {len(cost_methods)}")
print(f"\nKey Cost Explorer operations:")
for method in sorted(cost_methods)[:10]:
    print(f"  - {method}")

# Verify critical Cost Explorer methods
critical_ops = ['get_cost_and_usage', 'get_cost_forecast', 'get_anomaly_monitors']
available = [op for op in critical_ops if op in methods]
print(f"\nCritical operations available: {len(available)}/{len(critical_ops)}")
assert len(available) == len(critical_ops), f"CostExplorerClient is missing {sorted(set(critical_ops) - set(available))}"
assert cost_methods, "CostExplorerClient exposes no cost/usage/anomaly/savings methods"

print("\nExample 1 completed successfully")
```
