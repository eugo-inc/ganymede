## Example 1: CloudFrontClient Type Annotations with Distribution Methods

```python
# Example 1: Verify CloudFrontClient type annotations with distribution methods
from types_aiobotocore_cloudfront import CloudFrontClient
import inspect

print("Example 1: CloudFrontClient Distribution Methods")

# Verify the CloudFrontClient class
print(f"CloudFrontClient type: {CloudFrontClient}")
print(f"Is class: {inspect.isclass(CloudFrontClient)}")

# Get distribution-specific methods
methods = [m for m in dir(CloudFrontClient) if not m.startswith('_')]
dist_methods = [m for m in methods if 'distribution' in m.lower()]

print(f"\nTotal methods: {len(methods)}")
print(f"Distribution-specific methods: {len(dist_methods)}")
print(f"\nKey distribution operations:")
for method in sorted(dist_methods)[:12]:
    print(f"  - {method}")

# Check for critical CloudFront distribution operations
critical_ops = ['create_distribution', 'create_invalidation', 'list_distributions']
available = [op for op in critical_ops if op in methods]
print(f"\nCritical operations: {len(available)}/{len(critical_ops)}")

print("\nExample 1 completed successfully")
```

## Example 2: CloudFront Waiters for Deployment and Invalidation

```python
# Example 2: Verify CloudFront waiters for distribution deployment and cache invalidation
from types_aiobotocore_cloudfront.waiter import (
    DistributionDeployedWaiter,
    InvalidationCompletedWaiter,
    StreamingDistributionDeployedWaiter
)
import inspect

print("Example 2: CloudFront Waiters")

# CloudFront-specific waiters for async operations
waiters = [
    ('DistributionDeployedWaiter', DistributionDeployedWaiter),
    ('InvalidationCompletedWaiter', InvalidationCompletedWaiter),
    ('StreamingDistributionDeployedWaiter', StreamingDistributionDeployedWaiter)
]

for name, waiter_class in waiters:
    print(f"\n{name}:")
    print(f"  Type: {waiter_class}")
    print(f"  Is class: {inspect.isclass(waiter_class)}")
    
    methods = [m for m in dir(waiter_class) if not m.startswith('_')]
    print(f"  Methods: {methods}")
    
    if 'wait' in methods:
        print(f"  ✓ wait method available for async operations")

print("\nAll CloudFront waiters are properly typed")
print("Example 2 completed successfully")
```

## Example 3: CloudFront Paginators for Listing Resources

## Example 4: CloudFront Literal Types for Viewer Protocols and Cache Policies

```python
# Example 4: Explore CloudFront-specific literal types
from types_aiobotocore_cloudfront import literals

print("Example 4: CloudFront Literal Types")

# Get all CloudFront literals
literal_attrs = [attr for attr in dir(literals) if not attr.startswith('_')]
print(f"Total literal types: {len(literal_attrs)}")

# Check for key CloudFront-specific literals
cf_literals = [
    'ViewerProtocolPolicyType', 'HttpVersionType', 'PriceClassType',
    'OriginProtocolPolicyType', 'SSLSupportMethodType',
    'CachePolicyCookieBehaviorType', 'GeoRestrictionTypeType'
]

print("\nKey CloudFront literal types:")
for literal_name in cf_literals:
    if literal_name in literal_attrs:
        literal_type = getattr(literals, literal_name)
        print(f"  ✓ {literal_name}: {literal_type}")
    else:
        print(f"  ✗ {literal_name}: not found")

# Show cache policy related literals (unique to CloudFront)
cache_literals = [l for l in literal_attrs if 'Cache' in l or 'Origin' in l]
print(f"\nCache/Origin literals ({len(cache_literals)}):")
for lit in sorted(cache_literals)[:8]:
    print(f"  - {lit}")

print("\nExample 4 completed successfully")
```
