## Example 1: CodeBuildClient Type Annotations with Build and Project Methods

```python
# Example 1: Verify CodeBuildClient type annotations with build and project methods
from types_aiobotocore_codebuild import CodeBuildClient
import inspect

print("Example 1: CodeBuildClient Build and Project Methods")

# Verify the CodeBuildClient class
print(f"CodeBuildClient type: {CodeBuildClient}")
print(f"Is class: {inspect.isclass(CodeBuildClient)}")

# Get CodeBuild-specific methods
methods = [m for m in dir(CodeBuildClient) if not m.startswith('_')]
build_methods = [m for m in methods if 'build' in m.lower()]
project_methods = [m for m in methods if 'project' in m.lower()]

print(f"\nTotal methods: {len(methods)}")
print(f"Build-specific methods: {len(build_methods)}")
print(f"Project-specific methods: {len(project_methods)}")

print(f"\nKey build operations:")
for method in sorted(build_methods)[:8]:
    print(f"  - {method}")

print(f"\nKey project operations:")
for method in sorted(project_methods)[:6]:
    print(f"  - {method}")

print("\nExample 1 completed successfully")
```

## Example 2: CodeBuild Paginators for Builds, Projects, and Reports
