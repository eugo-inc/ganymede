## Example 1: CognitoIdentityProviderClient Type Annotations with User Pool Methods

```python
# Example 1: Verify CognitoIdentityProviderClient type annotations with user pool methods
from types_aiobotocore_cognito_idp import CognitoIdentityProviderClient
import inspect

print("Example 1: CognitoIdentityProviderClient User Pool Methods")

# Verify the CognitoIdentityProviderClient class
print(f"CognitoIdentityProviderClient type: {CognitoIdentityProviderClient}")
print(f"Is class: {inspect.isclass(CognitoIdentityProviderClient)}")

# Get Cognito IDP-specific methods
methods = [m for m in dir(CognitoIdentityProviderClient) if not m.startswith('_')]
user_methods = [m for m in methods if 'user' in m.lower() and 'admin' not in m.lower()]
admin_user_methods = [m for m in methods if 'user' in m.lower() and 'admin' in m.lower()]
pool_methods = [m for m in methods if 'pool' in m.lower()]

print(f"\nTotal methods: {len(methods)}")
print(f"User methods: {len(user_methods)}")
print(f"Admin user methods: {len(admin_user_methods)}")
print(f"User pool methods: {len(pool_methods)}")

print(f"\nKey user operations:")
for method in sorted(user_methods)[:8]:
    print(f"  - {method}")

print(f"\nKey admin user operations:")
for method in sorted(admin_user_methods)[:6]:
    print(f"  - {method}")

print("\nExample 1 completed successfully")
```

## Example 2: Cognito IDP Paginators for Users, Groups, and Pools

```python
# Example 2: Verify Cognito IDP paginators for users, groups, and user pools
from types_aiobotocore_cognito_idp.paginator import (
    ListUsersPaginator,
    ListGroupsPaginator,
    ListUserPoolsPaginator,
    ListUsersInGroupPaginator
)
import inspect

print("Example 2: Cognito IDP Paginators")

# Cognito IDP-specific paginators for user management
paginators = [
    ('ListUsersPaginator', ListUsersPaginator),
    ('ListGroupsPaginator', ListGroupsPaginator),
    ('ListUserPoolsPaginator', ListUserPoolsPaginator),
    ('ListUsersInGroupPaginator', ListUsersInGroupPaginator)
]

for name, paginator_class in paginators:
    print(f"\n{name}:")
    print(f"  Is class: {inspect.isclass(paginator_class)}")
    methods = [m for m in dir(paginator_class) if not m.startswith('_')]
    if 'paginate' in methods:
        print(f"  ✓ paginate method available")

print("\nCognito IDP paginators enable:")
print("  - Listing users in user pools")
print("  - Managing user groups")
print("  - Discovering user pool configurations")

print("\nExample 2 completed successfully")
```

## Example 3: Cognito IDP Literal Types for Authentication Flows

```python
# Example 3: Explore Cognito IDP-specific literal types for authentication flows
from types_aiobotocore_cognito_idp import literals

print("Example 3: Cognito IDP Literal Types")

# Get all Cognito IDP literals
literal_attrs = [attr for attr in dir(literals) if not attr.startswith('_') and not attr.startswith('s')]
print(f"Total literal types: {len(literal_attrs)}")

# Check for key Cognito IDP-specific literals
cognito_literals = [
    'AuthFlowTypeType',
    'ChallengeNameType',
    'UserStatusTypeType',
    'VerifiedAttributeTypeType',
    'MFAOptionTypeType'
]

print("\nKey Cognito IDP literal types:")
for literal_name in cognito_literals:
    if literal_name in literal_attrs:
        literal_type = getattr(literals, literal_name)
        print(f"  ✓ {literal_name}: {literal_type}")
    else:
        print(f"  ✗ {literal_name}: not found")

# Show auth flow related literals (unique to Cognito IDP)
auth_literals = [l for l in literal_attrs if 'Auth' in l or 'Challenge' in l or 'MFA' in l]
print(f"\nAuth/Challenge/MFA literals ({len(auth_literals)}):")
for lit in sorted(auth_literals)[:10]:
    print(f"  - {lit}")

print("\nExample 3 completed successfully")
```
