## Example 1: EC2Client Type Annotations with Describe Operations

```python
from types_aiobotocore_ec2 import EC2Client
from types_aiobotocore_ec2.type_defs import (
    DescribeInstancesResultTypeDef,
    FilterTypeDef,
    InstanceTypeDef
)
from aiobotocore.session import get_session
from botocore.exceptions import NoCredentialsError

# Example 1: Demonstrate type annotations for EC2Client
# This shows how types_aiobotocore_ec2 provides type hints for better IDE support

async def demonstrate_ec2_client_types():
    """Show type-annotated EC2 client usage"""
    session = get_session()

    # Create client with type annotation
    async with session.create_client("ec2", region_name="us-east-1") as client:
        client: EC2Client  # Type annotation enables autocomplete and type checking

        # Type-safe filter creation
        filters: list[FilterTypeDef] = [
            {"Name": "instance-state-name", "Values": ["running", "stopped"]}
        ]

        try:
            # describe_instances returns typed response
            response: DescribeInstancesResultTypeDef = await client.describe_instances(
                Filters=filters, MaxResults=5
            )

            print("EC2 Instances found:")
            for reservation in response.get("Reservations", []):
                for instance in reservation.get("Instances", []):
                    instance_typed: InstanceTypeDef = instance
                    print(f"  Instance ID: {instance_typed.get('InstanceId', 'N/A')}")
                    print(f"  State: {instance_typed.get('State', {}).get('Name', 'N/A')}")
        except NoCredentialsError:
            print("No AWS credentials (expected in test environment)")

        print("\nType annotations successfully demonstrated!")
        print("EC2Client provides 699 typed methods for EC2 operations")

await demonstrate_ec2_client_types()
```

## Example 2: Using Paginator Types for Listing EC2 Resources

```python
from types_aiobotocore_ec2 import EC2Client
from types_aiobotocore_ec2.paginator import (
    DescribeInstancesPaginator,
    DescribeVolumesPaginator,
    DescribeSnapshotsPaginator
)
from aiobotocore.session import get_session
from botocore.exceptions import NoCredentialsError

# Example 2: Demonstrate paginator type annotations
# Paginators help handle large result sets from AWS APIs

async def demonstrate_paginator_types():
    """Show type-annotated paginator usage for EC2"""
    session = get_session()

    async with session.create_client("ec2", region_name="us-west-2") as client:
        client: EC2Client

        # Create typed paginators for different EC2 resources
        instances_paginator: DescribeInstancesPaginator = client.get_paginator("describe_instances")
        volumes_paginator: DescribeVolumesPaginator = client.get_paginator("describe_volumes")
        snapshots_paginator: DescribeSnapshotsPaginator = client.get_paginator("describe_snapshots")

        print("Paginator types created successfully:")
        print(f"  DescribeInstancesPaginator: {type(instances_paginator).__name__}")
        print(f"  DescribeVolumesPaginator: {type(volumes_paginator).__name__}")
        print(f"  DescribeSnapshotsPaginator: {type(snapshots_paginator).__name__}")

        try:
            instance_count = 0
            async for page in instances_paginator.paginate(MaxResults=10):
                for reservation in page.get("Reservations", []):
                    instance_count += len(reservation.get("Instances", []))
            print(f"\nFound {instance_count} instances")
        except NoCredentialsError:
            print("\nNo AWS credentials (expected in test environment)")

        print("\nPaginator types successfully demonstrated!")
        print("Package provides 156 different paginator types")

await demonstrate_paginator_types()
```

## Example 4: Using TypeDefs for Request/Response Validation

```python
from types_aiobotocore_ec2 import EC2Client
from types_aiobotocore_ec2.type_defs import (
    TagTypeDef,
    DescribeSecurityGroupsResultTypeDef,
    SecurityGroupTypeDef,
    IpPermissionTypeDef
)
from aiobotocore.session import get_session
from botocore.exceptions import NoCredentialsError

# Example 4: Demonstrate TypeDefs for structured request/response data
# TypeDefs provide type safety for complex AWS API request parameters

async def demonstrate_typedef_usage():
    """Show how TypeDefs enable type-safe API requests"""
    session = get_session()

    async with session.create_client("ec2", region_name="ap-southeast-1") as client:
        client: EC2Client

        # Create type-safe tag definitions
        tags: list[TagTypeDef] = [
            {"Key": "Environment", "Value": "Development"},
            {"Key": "Application", "Value": "TestApp"},
            {"Key": "Owner", "Value": "DevTeam"}
        ]

        print("Type-safe Tag structures created:")
        for tag in tags:
            tag_typed: TagTypeDef = tag
            print(f"  {tag_typed['Key']}: {tag_typed['Value']}")

        try:
            # Type-safe security group query
            sg_response: DescribeSecurityGroupsResultTypeDef = await client.describe_security_groups(
                MaxResults=5
            )

            for sg in sg_response.get("SecurityGroups", []):
                sg_typed: SecurityGroupTypeDef = sg
                print(f"\nGroup: {sg_typed.get('GroupName', 'N/A')}")
                print(f"ID: {sg_typed.get('GroupId', 'N/A')}")
        except NoCredentialsError:
            print("\nNo AWS credentials (expected in test environment)")

        print("\nTypeDef usage successfully demonstrated!")
        print("Package provides 2905 TypeDef definitions for type-safe API usage")

await demonstrate_typedef_usage()
```

## Example 5: Using Literal Types for EC2 Enumerations

```python
from types_aiobotocore_ec2 import EC2Client
from types_aiobotocore_ec2.literals import (
    InstanceTypeType,
    ArchitectureValuesType,
    InstanceStateNameType,
    VolumeTypeType,
    ImageStateType
)
from aiobotocore.session import get_session

# Example 5: Demonstrate Literal types for type-safe enumerations
# Literals ensure only valid AWS EC2 enum values are used

async def demonstrate_literal_types():
    """Show how Literal types provide compile-time validation"""
    session = get_session()

    # Type-safe instance types
    instance_types: list[InstanceTypeType] = ["t2.micro", "t3.small", "m5.large"]
    
    # Type-safe architecture values
    architectures: list[ArchitectureValuesType] = ["x86_64", "arm64"]
    
    # Type-safe instance states
    valid_states: list[InstanceStateNameType] = ["pending", "running", "stopped", "terminated"]
    
    # Type-safe volume types
    volume_types: list[VolumeTypeType] = ["gp2", "gp3", "io1", "io2", "standard"]

    print("Literal Types Demonstration:")
    print(f"  Valid Instance Types: {instance_types}")
    print(f"  Valid Architectures: {architectures}")
    print(f"  Valid Instance States: {valid_states}")
    print(f"  Valid Volume Types: {volume_types}")

    async with session.create_client("ec2", region_name="us-east-1") as client:
        client: EC2Client

        # Use literals in API calls for type safety
        selected_instance_type: InstanceTypeType = "t2.micro"
        print(f"\nUsing type-safe instance type: {selected_instance_type}")

    print("\nLiteral types successfully demonstrated!")
    print("These types prevent typos and invalid values at development time")

await demonstrate_literal_types()
```
