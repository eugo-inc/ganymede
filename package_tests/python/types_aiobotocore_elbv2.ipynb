## Example 1: ELBv2Client Type Annotations for Load Balancer Operations

```python
from types_aiobotocore_elbv2 import ElasticLoadBalancingv2Client
from types_aiobotocore_elbv2.type_defs import (
    DescribeLoadBalancersOutputTypeDef,
    LoadBalancerTypeDef,
    DescribeTargetGroupsOutputTypeDef,
    TargetGroupTypeDef
)
from aiobotocore.session import get_session
from botocore.exceptions import NoCredentialsError

# Example 1: Demonstrate ELBv2 client type annotations for load balancer operations
# Shows how types_aiobotocore_elbv2 provides type hints for load balancer management

async def demonstrate_elbv2_client_types():
    """Show type-annotated ELBv2 client usage for load balancers"""
    session = get_session()

    async with session.create_client("elbv2", region_name="us-east-1") as client:
        client: ElasticLoadBalancingv2Client  # Type annotation enables autocomplete

        try:
            # Type-safe load balancer listing
            response: DescribeLoadBalancersOutputTypeDef = await client.describe_load_balancers()

            print("Application/Network Load Balancers found:")
            for lb in response.get("LoadBalancers", []):
                lb_typed: LoadBalancerTypeDef = lb
                print(f"  Name: {lb_typed.get('LoadBalancerName', 'N/A')}")
                print(f"  Type: {lb_typed.get('Type', 'N/A')}")
                print(f"  State: {lb_typed.get('State', {}).get('Code', 'N/A')}")
                print(f"  Scheme: {lb_typed.get('Scheme', 'N/A')}")
        except NoCredentialsError:
            print("No AWS credentials (expected in test environment)")

        print("\nELBv2 client type annotations successfully demonstrated!")
        print("ElasticLoadBalancingv2Client provides 56 typed methods for ALB/NLB operations")

await demonstrate_elbv2_client_types()
```

## Example 2: Using Paginator Types for ELBv2 Resources

```python
from types_aiobotocore_elbv2 import ElasticLoadBalancingv2Client
from types_aiobotocore_elbv2.paginator import (
    DescribeLoadBalancersPaginator,
    DescribeTargetGroupsPaginator,
    DescribeListenersPaginator,
    DescribeRulesPaginator
)
from aiobotocore.session import get_session
from botocore.exceptions import NoCredentialsError

# Example 2: Demonstrate ELBv2 paginator type annotations
# Paginators handle large result sets from ELBv2 list operations

async def demonstrate_elbv2_paginator_types():
    """Show type-annotated paginator usage for ELBv2 resources"""
    session = get_session()

    async with session.create_client("elbv2", region_name="us-west-2") as client:
        client: ElasticLoadBalancingv2Client

        # Create typed paginators for different ELBv2 resources
        lb_paginator: DescribeLoadBalancersPaginator = client.get_paginator("describe_load_balancers")
        tg_paginator: DescribeTargetGroupsPaginator = client.get_paginator("describe_target_groups")
        listener_paginator: DescribeListenersPaginator = client.get_paginator("describe_listeners")
        rules_paginator: DescribeRulesPaginator = client.get_paginator("describe_rules")

        print("ELBv2 Paginator types created successfully:")
        print(f"  DescribeLoadBalancersPaginator: {type(lb_paginator).__name__}")
        print(f"  DescribeTargetGroupsPaginator: {type(tg_paginator).__name__}")
        print(f"  DescribeListenersPaginator: {type(listener_paginator).__name__}")
        print(f"  DescribeRulesPaginator: {type(rules_paginator).__name__}")

        try:
            lb_count = 0
            async for page in lb_paginator.paginate():
                lb_count += len(page.get("LoadBalancers", []))
            print(f"\nFound {lb_count} load balancers")
        except NoCredentialsError:
            print("\nNo AWS credentials (expected in test environment)")

        print("\nELBv2 paginator types successfully demonstrated!")
        print("Package provides 10 different ELBv2 paginator types")

await demonstrate_elbv2_paginator_types()
```

## Example 4: Using TypeDefs for Target Groups and Listeners

```python
from types_aiobotocore_elbv2 import ElasticLoadBalancingv2Client
from types_aiobotocore_elbv2.type_defs import (
    TargetGroupTypeDef,
    TargetDescriptionTypeDef,
    ListenerTypeDef,
    ActionTypeDef,
    DescribeTargetGroupsOutputTypeDef,
    DescribeListenersOutputTypeDef
)
from aiobotocore.session import get_session
from botocore.exceptions import NoCredentialsError

# Example 4: Demonstrate TypeDefs for target groups and listeners
# TypeDefs provide type safety for complex ELBv2 target group and listener configurations

async def demonstrate_elbv2_typedef_usage():
    """Show how TypeDefs enable type-safe ELBv2 target group and listener operations"""
    session = get_session()

    async with session.create_client("elbv2", region_name="us-east-1") as client:
        client: ElasticLoadBalancingv2Client

        # Create type-safe target descriptions
        targets: list[TargetDescriptionTypeDef] = [
            {"Id": "i-1234567890abcdef0", "Port": 80},
            {"Id": "i-0987654321fedcba0", "Port": 80}
        ]

        # Create type-safe action configuration
        actions: list[ActionTypeDef] = [
            {
                "Type": "forward",
                "TargetGroupArn": "arn:aws:elasticloadbalancing:us-east-1:123456789012:targetgroup/my-targets/50dc6c495c0c9188",
                "Order": 1
            }
        ]

        print("Type-safe ELBv2 configurations created:")
        print(f"  Target Descriptions: {len(targets)}")
        for target in targets:
            target_typed: TargetDescriptionTypeDef = target
            print(f"    Instance: {target_typed['Id']}, Port: {target_typed.get('Port', 'N/A')}")
        print(f"  Listener Actions: {len(actions)}")

        try:
            # Type-safe target group query
            tg_response: DescribeTargetGroupsOutputTypeDef = await client.describe_target_groups()

            print(f"\nTarget Groups found: {len(tg_response.get('TargetGroups', []))}")
            for tg in tg_response.get("TargetGroups", [])[:3]:
                tg_typed: TargetGroupTypeDef = tg
                print(f"  Name: {tg_typed.get('TargetGroupName', 'N/A')}")
                print(f"  Protocol: {tg_typed.get('Protocol', 'N/A')}")
                print(f"  Port: {tg_typed.get('Port', 'N/A')}")
        except NoCredentialsError:
            print("\nNo AWS credentials (expected in test environment)")

        print("\nELBv2 TypeDef usage successfully demonstrated!")
        print("Package provides 215 TypeDef definitions for type-safe ELBv2 operations")

await demonstrate_elbv2_typedef_usage()
```
