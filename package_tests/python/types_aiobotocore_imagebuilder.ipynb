## Example 1: ImagebuilderClient Type Annotations for Image Management

```python
from types_aiobotocore_imagebuilder import ImagebuilderClient
from types_aiobotocore_imagebuilder.type_defs import (
    ListImagesResponseTypeDef,
    ImageSummaryTypeDef,
    ListImagePipelinesResponseTypeDef,
    ImagePipelineTypeDef,
    ImageStateTypeDef
)
from aiobotocore.session import get_session
from botocore.exceptions import NoCredentialsError

# Example 1: Demonstrate EC2 Image Builder client type annotations
# Shows how types_aiobotocore_imagebuilder provides type hints for image management

async def demonstrate_imagebuilder_client_types():
    """Show type-annotated Image Builder client usage"""
    session = get_session()

    async with session.create_client("imagebuilder", region_name="us-east-1") as client:
        client: ImagebuilderClient  # Type annotation enables autocomplete

        try:
            # Type-safe image listing
            images_response: ListImagesResponseTypeDef = await client.list_images(maxResults=10)

            print("EC2 Image Builder Images found:")
            for image in images_response.get("imageVersionList", []):
                image_typed: ImageSummaryTypeDef = image
                print(f"  Image ARN: {image_typed.get('arn', 'N/A')}")
                print(f"  Name: {image_typed.get('name', 'N/A')}")
                print(f"  Version: {image_typed.get('version', 'N/A')}")
                print(f"  Platform: {image_typed.get('platform', 'N/A')}")
                state: ImageStateTypeDef = image_typed.get('state', {})
                print(f"  State: {state.get('status', 'N/A')}")
                print()
                
            # Type-safe image pipeline listing
            pipelines_response: ListImagePipelinesResponseTypeDef = await client.list_image_pipelines(maxResults=10)
            
            print(f"Image Pipelines found: {len(pipelines_response.get('imagePipelineList', []))}")
            for pipeline in pipelines_response.get("imagePipelineList", [])[:3]:
                pipeline_typed: ImagePipelineTypeDef = pipeline
                print(f"  Pipeline ARN: {pipeline_typed.get('arn', 'N/A')}")
                print(f"  Name: {pipeline_typed.get('name', 'N/A')}")
                print(f"  Status: {pipeline_typed.get('status', 'N/A')}")
                
        except NoCredentialsError:
            print("No AWS credentials (expected in test environment)")

        print("\nImage Builder client type annotations successfully demonstrated!")
        print("ImagebuilderClient provides 90+ typed methods for EC2 Image Builder operations")

await demonstrate_imagebuilder_client_types()
```

## Example 2: Using Paginator Types for Image Builder Resources

```python
from types_aiobotocore_imagebuilder import ImagebuilderClient
from types_aiobotocore_imagebuilder.paginator import (
    ListComponentsPaginator,
    ListComponentBuildVersionsPaginator,
    ListContainerRecipesPaginator,
    ListImagesPaginator,
    ListImagePipelinesPaginator,
    ListImageRecipesPaginator,
    ListInfrastructureConfigurationsPaginator
)
from aiobotocore.session import get_session
from botocore.exceptions import NoCredentialsError

# Example 2: Demonstrate Image Builder paginator type annotations
# Paginators handle large result sets from Image Builder list operations

async def demonstrate_imagebuilder_paginator_types():
    """Show type-annotated paginator usage for Image Builder resources"""
    session = get_session()

    async with session.create_client("imagebuilder", region_name="us-east-1") as client:
        client: ImagebuilderClient

        # Create typed paginators for different Image Builder resources
        components_paginator: ListComponentsPaginator = client.get_paginator("list_components")
        component_versions_paginator: ListComponentBuildVersionsPaginator = client.get_paginator("list_component_build_versions")
        container_recipes_paginator: ListContainerRecipesPaginator = client.get_paginator("list_container_recipes")
        images_paginator: ListImagesPaginator = client.get_paginator("list_images")
        pipelines_paginator: ListImagePipelinesPaginator = client.get_paginator("list_image_pipelines")
        image_recipes_paginator: ListImageRecipesPaginator = client.get_paginator("list_image_recipes")
        infra_configs_paginator: ListInfrastructureConfigurationsPaginator = client.get_paginator("list_infrastructure_configurations")

        print("Image Builder Paginator types created successfully:")
        print(f"  ListComponentsPaginator: {type(components_paginator).__name__}")
        print(f"  ListComponentBuildVersionsPaginator: {type(component_versions_paginator).__name__}")
        print(f"  ListContainerRecipesPaginator: {type(container_recipes_paginator).__name__}")
        print(f"  ListImagesPaginator: {type(images_paginator).__name__}")
        print(f"  ListImagePipelinesPaginator: {type(pipelines_paginator).__name__}")
        print(f"  ListImageRecipesPaginator: {type(image_recipes_paginator).__name__}")
        print(f"  ListInfrastructureConfigurationsPaginator: {type(infra_configs_paginator).__name__}")

        try:
            component_count = 0
            async for page in components_paginator.paginate():
                component_count += len(page.get("componentVersionList", []))
            print(f"\nFound {component_count} components")
            
            image_count = 0
            async for page in images_paginator.paginate():
                image_count += len(page.get("imageVersionList", []))
            print(f"Found {image_count} images")
            
            pipeline_count = 0
            async for page in pipelines_paginator.paginate():
                pipeline_count += len(page.get("imagePipelineList", []))
            print(f"Found {pipeline_count} image pipelines")
            
        except NoCredentialsError:
            print("\nNo AWS credentials (expected in test environment)")

        print("\nImage Builder paginator types successfully demonstrated!")
        print("Package provides 19 different Image Builder paginator types")

await demonstrate_imagebuilder_paginator_types()
```

## Example 3: Using TypeDefs for Image Pipeline Configuration

```python
from types_aiobotocore_imagebuilder import ImagebuilderClient
from types_aiobotocore_imagebuilder.type_defs import (
    ImagePipelineTypeDef,
    ScheduleTypeDef,
    ImageTestsConfigurationTypeDef,
    CreateImagePipelineRequestTypeDef,
    GetImagePipelineResponseTypeDef
)
from aiobotocore.session import get_session
from botocore.exceptions import NoCredentialsError

# Example 3: Demonstrate TypeDefs for Image Pipeline configuration
# TypeDefs provide type safety for creating and managing Image Builder pipelines

async def demonstrate_image_pipeline_typedefs():
    """Show how TypeDefs enable type-safe Image Pipeline operations"""
    session = get_session()

    async with session.create_client("imagebuilder", region_name="us-east-1") as client:
        client: ImagebuilderClient

        # Type-safe schedule configuration
        schedule: ScheduleTypeDef = {
            "scheduleExpression": "cron(0 0 * * ? *)",  # Daily at midnight
            "pipelineExecutionStartCondition": "EXPRESSION_MATCH_AND_DEPENDENCY_UPDATES_AVAILABLE"
        }

        # Type-safe test configuration
        image_tests_config: ImageTestsConfigurationTypeDef = {
            "imageTestsEnabled": True,
            "timeoutMinutes": 90
        }

        print("Type-safe Image Pipeline Configuration:")
        print(f"  Schedule Expression: {schedule['scheduleExpression']}")
        print(f"  Start Condition: {schedule['pipelineExecutionStartCondition']}")
        print(f"  Tests Enabled: {image_tests_config['imageTestsEnabled']}")
        print(f"  Test Timeout: {image_tests_config['timeoutMinutes']} minutes")

        # Type-safe pipeline creation request (structure only, not executed)
        pipeline_request: CreateImagePipelineRequestTypeDef = {
            "name": "example-pipeline",
            "imageRecipeArn": "arn:aws:imagebuilder:us-east-1:123456789012:image-recipe/example/1.0.0",
            "infrastructureConfigurationArn": "arn:aws:imagebuilder:us-east-1:123456789012:infrastructure-configuration/example",
            "schedule": schedule,
            "imageTestsConfiguration": image_tests_config,
            "status": "ENABLED"
        }

        print("\nType-safe Pipeline Request Structure:")
        print(f"  Pipeline Name: {pipeline_request['name']}")
        print(f"  Image Recipe ARN: {pipeline_request['imageRecipeArn']}")
        print(f"  Infrastructure Config ARN: {pipeline_request['infrastructureConfigurationArn']}")
        print(f"  Status: {pipeline_request['status']}")

        try:
            # Type-safe pipeline retrieval
            pipeline_response: GetImagePipelineResponseTypeDef = await client.get_image_pipeline(
                imagePipelineArn="arn:aws:imagebuilder:us-east-1:123456789012:image-pipeline/example"
            )
            
            pipeline: ImagePipelineTypeDef = pipeline_response.get("imagePipeline", {})
            print(f"\nRetrieved Pipeline:")
            print(f"  ARN: {pipeline.get('arn', 'N/A')}")
            print(f"  Name: {pipeline.get('name', 'N/A')}")
            print(f"  Status: {pipeline.get('status', 'N/A')}")
        except NoCredentialsError:
            print("\nNo AWS credentials (expected in test environment)")
        except Exception as e:
            print(f"\nPipeline not found (expected in test environment): {type(e).__name__}")

        print("\nImage Pipeline TypeDef usage successfully demonstrated!")
        print("Package provides 200+ TypeDef definitions for type-safe Image Builder operations")

await demonstrate_image_pipeline_typedefs()
```

## Example 4: Using TypeDefs for Components and Image Recipes

```python
from types_aiobotocore_imagebuilder import ImagebuilderClient
from types_aiobotocore_imagebuilder.type_defs import (
    ComponentConfigurationTypeDef,
    ComponentParameterTypeDef,
    ImageRecipeTypeDef,
    ComponentSummaryTypeDef,
    InstanceBlockDeviceMappingTypeDef,
    EbsInstanceBlockDeviceSpecificationTypeDef,
    CreateImageRecipeRequestTypeDef
)
from aiobotocore.session import get_session
from botocore.exceptions import NoCredentialsError

# Example 4: Demonstrate TypeDefs for Components and Image Recipes
# TypeDefs provide type safety for building custom AMI recipes with components

async def demonstrate_component_recipe_typedefs():
    """Show how TypeDefs enable type-safe component and recipe operations"""
    session = get_session()

    async with session.create_client("imagebuilder", region_name="us-east-1") as client:
        client: ImagebuilderClient

        # Type-safe component parameter
        component_param: ComponentParameterTypeDef = {
            "name": "InstallPackages",
            "value": ["nginx", "python3", "git"]
        }

        # Type-safe component configuration
        component_config: ComponentConfigurationTypeDef = {
            "componentArn": "arn:aws:imagebuilder:us-east-1:123456789012:component/example/1.0.0",
            "parameters": [component_param]
        }

        print("Type-safe Component Configuration:")
        print(f"  Component ARN: {component_config['componentArn']}")
        print(f"  Parameter Name: {component_param['name']}")
        print(f"  Parameter Value: {component_param['value']}")

        # Type-safe EBS volume configuration
        ebs_config: EbsInstanceBlockDeviceSpecificationTypeDef = {
            "encrypted": True,
            "deleteOnTermination": True,
            "volumeSize": 30,
            "volumeType": "gp3"
        }

        # Type-safe block device mapping
        block_device: InstanceBlockDeviceMappingTypeDef = {
            "deviceName": "/dev/xvda",
            "ebs": ebs_config
        }

        print("\nType-safe Block Device Configuration:")
        print(f"  Device Name: {block_device['deviceName']}")
        print(f"  Volume Size: {ebs_config['volumeSize']} GB")
        print(f"  Volume Type: {ebs_config['volumeType']}")
        print(f"  Encrypted: {ebs_config['encrypted']}")

        # Type-safe image recipe creation request (structure only)
        recipe_request: CreateImageRecipeRequestTypeDef = {
            "name": "custom-ami-recipe",
            "semanticVersion": "1.0.0",
            "components": [component_config],
            "parentImage": "arn:aws:imagebuilder:us-east-1:aws:image/amazon-linux-2-x86/x.x.x",
            "blockDeviceMappings": [block_device]
        }

        print("\nType-safe Image Recipe Request:")
        print(f"  Recipe Name: {recipe_request['name']}")
        print(f"  Semantic Version: {recipe_request['semanticVersion']}")
        print(f"  Components: {len(recipe_request['components'])}")
        print(f"  Block Devices: {len(recipe_request['blockDeviceMappings'])}")

        try:
            # Type-safe component listing
            components_response = await client.list_components(owner="Self", maxResults=5)
            
            print(f"\nComponents found: {len(components_response.get('componentVersionList', []))}")
            for component in components_response.get("componentVersionList", [])[:3]:
                comp_typed: ComponentSummaryTypeDef = component
                print(f"  Component: {comp_typed.get('name', 'N/A')} v{comp_typed.get('version', 'N/A')}")
                print(f"    ARN: {comp_typed.get('arn', 'N/A')}")
                print(f"    Platform: {comp_typed.get('platform', 'N/A')}")
        except NoCredentialsError:
            print("\nNo AWS credentials (expected in test environment)")

        print("\nComponent and Recipe TypeDef usage successfully demonstrated!")
        print("TypeDefs ensure type safety for complex AMI building workflows")

await demonstrate_component_recipe_typedefs()
```

## Example 5: Using TypeDefs for Infrastructure and Distribution Configuration

```python
from types_aiobotocore_imagebuilder import ImagebuilderClient
from types_aiobotocore_imagebuilder.type_defs import (
    InfrastructureConfigurationTypeDef,
    S3LogsTypeDef,
    LoggingTypeDef,
    DistributionConfigurationTypeDef,
    DistributionTypeDef,
    AmiDistributionConfigurationTypeDef,
    LaunchPermissionConfigurationTypeDef,
    CreateInfrastructureConfigurationRequestTypeDef,
    CreateDistributionConfigurationRequestTypeDef
)
from aiobotocore.session import get_session
from botocore.exceptions import NoCredentialsError

# Example 5: Demonstrate TypeDefs for Infrastructure and Distribution Configuration
# TypeDefs provide type safety for EC2 instance and AMI distribution settings

async def demonstrate_infrastructure_distribution_typedefs():
    """Show how TypeDefs enable type-safe infrastructure and distribution configs"""
    session = get_session()

    async with session.create_client("imagebuilder", region_name="us-east-1") as client:
        client: ImagebuilderClient

        # Type-safe S3 logging configuration
        s3_logs: S3LogsTypeDef = {
            "s3BucketName": "my-imagebuilder-logs",
            "s3KeyPrefix": "build-logs/"
        }

        # Type-safe logging configuration
        logging: LoggingTypeDef = {
            "s3Logs": s3_logs
        }

        print("Type-safe Logging Configuration:")
        print(f"  S3 Bucket: {s3_logs['s3BucketName']}")
        print(f"  S3 Prefix: {s3_logs['s3KeyPrefix']}")

        # Type-safe infrastructure configuration request
        infra_request: CreateInfrastructureConfigurationRequestTypeDef = {
            "name": "example-infrastructure",
            "instanceTypes": ["t3.medium", "t3.large"],
            "instanceProfileName": "EC2ImageBuilderProfile",
            "logging": logging,
            "terminateInstanceOnFailure": True,
            "snsTopicArn": "arn:aws:sns:us-east-1:123456789012:imagebuilder-notifications"
        }

        print("\nType-safe Infrastructure Configuration:")
        print(f"  Name: {infra_request['name']}")
        print(f"  Instance Types: {infra_request['instanceTypes']}")
        print(f"  Instance Profile: {infra_request['instanceProfileName']}")
        print(f"  Terminate on Failure: {infra_request['terminateInstanceOnFailure']}")

        # Type-safe AMI launch permissions
        launch_perms: LaunchPermissionConfigurationTypeDef = {
            "userIds": ["123456789012"],
            "userGroups": ["all"]
        }

        # Type-safe AMI distribution configuration
        ami_distribution: AmiDistributionConfigurationTypeDef = {
            "name": "MyCustomAMI-{{ imagebuilder:buildDate }}",
            "description": "Custom AMI built with Image Builder",
            "launchPermission": launch_perms,
            "amiTags": {
                "Environment": "Production",
                "ManagedBy": "ImageBuilder"
            }
        }

        # Type-safe distribution configuration
        distribution: DistributionTypeDef = {
            "region": "us-east-1",
            "amiDistributionConfiguration": ami_distribution
        }

        print("\nType-safe Distribution Configuration:")
        print(f"  Region: {distribution['region']}")
        print(f"  AMI Name Pattern: {ami_distribution['name']}")
        print(f"  AMI Description: {ami_distribution['description']}")
        print(f"  Launch Permission User Groups: {launch_perms['userGroups']}")
        print(f"  AMI Tags: {ami_distribution['amiTags']}")

        # Type-safe distribution configuration request
        dist_request: CreateDistributionConfigurationRequestTypeDef = {
            "name": "example-distribution",
            "distributions": [distribution]
        }

        print(f"\nDistribution Configuration Request:")
        print(f"  Name: {dist_request['name']}")
        print(f"  Number of Distributions: {len(dist_request['distributions'])}")

        try:
            # Type-safe infrastructure configuration listing
            infra_response = await client.list_infrastructure_configurations(maxResults=5)
            
            print(f"\nInfrastructure Configurations found: {len(infra_response.get('infrastructureConfigurationSummaryList', []))}")
        except NoCredentialsError:
            print("\nNo AWS credentials (expected in test environment)")

        print("\nInfrastructure and Distribution TypeDef usage successfully demonstrated!")
        print("TypeDefs provide complete type safety for EC2 Image Builder deployment workflows")

await demonstrate_infrastructure_distribution_typedefs()
```
