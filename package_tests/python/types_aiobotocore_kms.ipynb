## Example 3: Using TypeDefs for KMS Key Creation

```python
from types_aiobotocore_kms.type_defs import CreateKeyRequestTypeDef, TagTypeDef, KeyMetadataTypeDef
from types_aiobotocore_kms import KMSClient
from types_aiobotocore_kms.type_defs import DescribeKeyResponseTypeDef, KeyListEntryTypeDef, ListKeysResponseTypeDef
from types_aiobotocore_kms.paginator import ListAliasesPaginator, ListGrantsPaginator, ListKeysPaginator

def demonstrate_key_creation_typedefs():
    tags: list[TagTypeDef] = [{"TagKey": "Environment", "TagValue": "Production"}, {"TagKey": "Application", "TagValue": "WebApp"}]
    create_key_request: CreateKeyRequestTypeDef = {
        "Description": "Encryption key for production data",
        "KeyUsage": "ENCRYPT_DECRYPT",
        "Origin": "AWS_KMS",
        "MultiRegion": False,
        "Tags": tags
    }
    print("Type-safe KMS Key Creation Request:")
    print(f"  Description: {create_key_request['Description']}")
    print(f"  Key Usage: {create_key_request['KeyUsage']}")
    print(f"  Origin: {create_key_request['Origin']}")
    print(f"  Multi-Region: {create_key_request['MultiRegion']}")
    print(f"  Tags: {len(create_key_request['Tags'])} tags")
    for tag in tags:
        print(f"    {tag['TagKey']}={tag['TagValue']}")
    print("TypeDefs ensure type safety for KMS key lifecycle management")

demonstrate_key_creation_typedefs()
```

## Example 4: Using TypeDefs for Encryption and Decryption

```python
from types_aiobotocore_kms.type_defs import EncryptRequestTypeDef, DecryptRequestTypeDef, EncryptResponseTypeDef, DecryptResponseTypeDef

def demonstrate_encryption_decryption_typedefs():
    plaintext_data = b"Sensitive production data"
    encrypt_request: EncryptRequestTypeDef = {
        "KeyId": "arn:aws:kms:us-east-1:123456789012:key/12345678-1234-1234-1234-123456789012",
        "Plaintext": plaintext_data,
        "EncryptionContext": {"Department": "Engineering", "Project": "DataPipeline"}
    }
    print("Type-safe KMS Encrypt Request:")
    print(f"  Key ID: {encrypt_request['KeyId'][-40:]}")
    print(f"  Plaintext Size: {len(encrypt_request['Plaintext'])} bytes")
    print(f"  Encryption Context: {encrypt_request['EncryptionContext']}")
    ciphertext = b"encrypted_data_blob"
    decrypt_request: DecryptRequestTypeDef = {
        "CiphertextBlob": ciphertext,
        "EncryptionContext": {"Department": "Engineering", "Project": "DataPipeline"}
    }
    print("\nType-safe KMS Decrypt Request:")
    print(f"  Ciphertext Size: {len(decrypt_request['CiphertextBlob'])} bytes")
    print(f"  Encryption Context: {decrypt_request['EncryptionContext']}")
    print("TypeDefs ensure type safety for KMS cryptographic operations")

demonstrate_encryption_decryption_typedefs()
```

## Example 5: Using TypeDefs for Grants and Key Policies

```python
from types_aiobotocore_kms.type_defs import CreateGrantRequestTypeDef, GrantConstraintsTypeDef, GrantListEntryTypeDef

def demonstrate_grant_policy_typedefs():
    constraints: GrantConstraintsTypeDef = {
        "EncryptionContextSubset": {"Department": "Engineering"},
        "EncryptionContextEquals": {"Project": "DataPipeline"}
    }
    grant_request: CreateGrantRequestTypeDef = {
        "KeyId": "arn:aws:kms:us-east-1:123456789012:key/12345678-1234-1234-1234-123456789012",
        "GranteePrincipal": "arn:aws:iam::123456789012:role/DataProcessorRole",
        "Operations": ["Encrypt", "Decrypt", "GenerateDataKey"],
        "Constraints": constraints
    }
    print("Type-safe KMS Grant Request:")
    print(f"  Key ID: {grant_request['KeyId'][-40:]}")
    print(f"  Grantee: {grant_request['GranteePrincipal'].split('/')[-1]}")
    print(f"  Operations: {', '.join(grant_request['Operations'])}")
    print(f"  Constraints:")
    print(f"    Encryption Context Subset: {constraints['EncryptionContextSubset']}")
    print(f"    Encryption Context Equals: {constraints['EncryptionContextEquals']}")
    grant_entry: GrantListEntryTypeDef = {"GrantId": "abc123", "GranteePrincipal": grant_request['GranteePrincipal'], "Operations": grant_request['Operations']}
    print(f"\nGrant Entry: ID={grant_entry['GrantId']}, Operations={len(grant_entry['Operations'])}")
    print("TypeDefs ensure type safety for KMS access control management")

demonstrate_grant_policy_typedefs()
```
