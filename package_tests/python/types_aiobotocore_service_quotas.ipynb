## Example 1: ServiceQuotasClient Type Annotations for Quota Management

```python
from types_aiobotocore_service_quotas import ServiceQuotasClient
from types_aiobotocore_service_quotas.type_defs import ListServicesResponseTypeDef, ServiceInfoTypeDef, ListServiceQuotasResponseTypeDef, ServiceQuotaTypeDef
from aiobotocore.session import get_session
from botocore.exceptions import NoCredentialsError

async def demonstrate_service_quotas_client_types():
    session = get_session()
    async with session.create_client("service-quotas", region_name="us-east-1") as client:
        client: ServiceQuotasClient  # Type annotation enables IDE autocomplete
        try:
            services_response: ListServicesResponseTypeDef = await client.list_services(MaxResults=5)
            print(f"AWS Services with Quotas: {len(services_response.get('Services', []))}")
            for service in services_response.get("Services", [])[:3]:
                service_typed: ServiceInfoTypeDef = service
                print(f"  Service: {service_typed.get('ServiceCode')} - {service_typed.get('ServiceName')}")
            if services_response.get('Services'):
                service_code = services_response['Services'][0]['ServiceCode']
                quotas_response: ListServiceQuotasResponseTypeDef = await client.list_service_quotas(ServiceCode=service_code, MaxResults=3)
                print(f"\nQuotas for {service_code}: {len(quotas_response.get('Quotas', []))}")
        except NoCredentialsError:
            print("No AWS credentials (expected)")
        except Exception as e:
            print(f"Service unavailable (expected): {type(e).__name__}")
        print("ServiceQuotasClient provides 20+ typed methods for quota management")

await demonstrate_service_quotas_client_types()
```

## Example 2: Using Paginator Types for Service Quotas Resources

```python
from types_aiobotocore_service_quotas import ServiceQuotasClient
from types_aiobotocore_service_quotas.paginator import ListServiceQuotasPaginator, ListAWSDefaultServiceQuotasPaginator, ListServicesPaginator
from aiobotocore.session import get_session
from botocore.exceptions import NoCredentialsError

async def demonstrate_service_quotas_paginator_types():
    session = get_session()
    async with session.create_client("service-quotas", region_name="us-east-1") as client:
        client: ServiceQuotasClient
        services_paginator: ListServicesPaginator = client.get_paginator("list_services")
        quotas_paginator: ListServiceQuotasPaginator = client.get_paginator("list_service_quotas")
        defaults_paginator: ListAWSDefaultServiceQuotasPaginator = client.get_paginator("list_aws_default_service_quotas")
        print("Service Quotas Paginator types:")
        print(f"  ListServicesPaginator: {type(services_paginator).__name__}")
        print(f"  ListServiceQuotasPaginator: {type(quotas_paginator).__name__}")
        print(f"  ListAWSDefaultServiceQuotasPaginator: {type(defaults_paginator).__name__}")
        try:
            service_count = 0
            async for page in services_paginator.paginate():
                service_count += len(page.get("Services", []))
            print(f"Found {service_count} services using paginator")
        except (NoCredentialsError, Exception):
            print("No credentials/service unavailable (expected)")
        print("Package provides 6+ Service Quotas paginator types")

await demonstrate_service_quotas_paginator_types()
```
