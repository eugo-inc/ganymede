## Example 1: STSClient Type Annotations for Security Token Operations

```python
from types_aiobotocore_sts import STSClient
from types_aiobotocore_sts.type_defs import GetCallerIdentityResponseTypeDef
from aiobotocore.session import get_session
from botocore.exceptions import ClientError, NoCredentialsError

async def demonstrate_sts_client_types():
    session = get_session()
    async with session.create_client("sts", region_name="us-east-1") as client:
        client: STSClient  # Type annotation enables IDE autocomplete
        try:
            identity_response: GetCallerIdentityResponseTypeDef = await client.get_caller_identity()
            print(f"Caller Identity:")
            print(f"  Account: {identity_response.get('Account')}")
            print(f"  User ID: {identity_response.get('UserId')}")
            print(f"  ARN: {identity_response.get('Arn')}")
        except (NoCredentialsError, ClientError) as e:
            print(f"No credentials or access denied (expected): {type(e).__name__}")
        print("STSClient provides 10+ typed methods for temporary credential management")

await demonstrate_sts_client_types()
```

## Example 2: Using TypeDefs for AssumeRole Operations

```python
from types_aiobotocore_sts.type_defs import AssumeRoleRequestTypeDef, TagTypeDef, PolicyDescriptorTypeTypeDef, AssumeRoleResponseTypeDef, AssumedRoleUserTypeDef, CredentialsTypeDef

def demonstrate_assume_role_typedefs():
    tags: list[TagTypeDef] = [{"Key": "Application", "Value": "DataPipeline"}, {"Key": "Environment", "Value": "Production"}]
    policy_arns: list[PolicyDescriptorTypeTypeDef] = [{"arn": "arn:aws:iam::aws:policy/ReadOnlyAccess"}]
    assume_role_request: AssumeRoleRequestTypeDef = {
        "RoleArn": "arn:aws:iam::123456789012:role/DataProcessorRole",
        "RoleSessionName": "data-pipeline-session",
        "DurationSeconds": 3600,
        "Tags": tags,
        "PolicyArns": policy_arns,
        "ExternalId": "unique-external-id-12345"
    }
    print("Type-safe AssumeRole Request:")
    print(f"  Role ARN: {assume_role_request['RoleArn'][-40:]}")
    print(f"  Session Name: {assume_role_request['RoleSessionName']}")
    print(f"  Duration: {assume_role_request['DurationSeconds']}s")
    print(f"  Tags: {len(tags)} tags, Policy ARNs: {len(policy_arns)}")
    assumed_role_user: AssumedRoleUserTypeDef = {"AssumedRoleId": "AROAEXAMPLEID:data-pipeline-session", "Arn": "arn:aws:sts::123456789012:assumed-role/DataProcessorRole/data-pipeline-session"}
    credentials: CredentialsTypeDef = {"AccessKeyId": "ASIAEXAMPLEKEY", "SecretAccessKey": "SECRET", "SessionToken": "TOKEN", "Expiration": "2024-01-15T11:00:00Z"}
    response: AssumeRoleResponseTypeDef = {"Credentials": credentials, "AssumedRoleUser": assumed_role_user}
    print(f"\nAssumed Role: {assumed_role_user['Arn'][-50:]}")
    print(f"Credentials expire: {credentials['Expiration']}")
    print("TypeDefs ensure type safety for STS AssumeRole operations")

demonstrate_assume_role_typedefs()
```

## Example 3: Using TypeDefs for Session Token Generation

```python
from types_aiobotocore_sts.type_defs import GetSessionTokenRequestTypeDef, GetSessionTokenResponseTypeDef, CredentialsTypeDef
from types_aiobotocore_sts import STSClient
from types_aiobotocore_sts.type_defs import GetCallerIdentityResponseTypeDef
from datetime import datetime, timedelta

def demonstrate_session_token_typedefs():
    session_token_request: GetSessionTokenRequestTypeDef = {
        "DurationSeconds": 43200,
        "SerialNumber": "arn:aws:iam::123456789012:mfa/user",
        "TokenCode": "123456"
    }
    print("Type-safe Session Token Request:")
    print(f"  Duration: {session_token_request['DurationSeconds']}s ({session_token_request['DurationSeconds'] // 3600} hours)")
    print(f"  MFA Device: {session_token_request['SerialNumber'][-30:]}")
    print(f"  MFA Token: {session_token_request['TokenCode']}")
    expiration_time = datetime.now() + timedelta(seconds=session_token_request['DurationSeconds'])
    credentials: CredentialsTypeDef = {
        "AccessKeyId": "ASIAEXAMPLEKEYID123",
        "SecretAccessKey": "wJalrXUtnFEMI/K7MDENG/bPxRfiCYEXAMPLEKEY",
        "SessionToken": "FwoGZXIvYXdzEBQaD...EXAMPLE_SESSION_TOKEN",
        "Expiration": expiration_time.isoformat()
    }
    response: GetSessionTokenResponseTypeDef = {"Credentials": credentials}
    print(f"\nTemporary Credentials Generated:")
    print(f"  Access Key: {credentials['AccessKeyId'][:20]}...")
    print(f"  Session Token Length: {len(credentials['SessionToken'])} chars")
    print(f"  Expires: {credentials['Expiration'][:19]}")
    print("TypeDefs ensure type safety for STS session token generation")

demonstrate_session_token_typedefs()
```

## Example 4: Using TypeDefs for Federated User Access

```python
from types_aiobotocore_sts.type_defs import GetFederationTokenRequestTypeDef, GetFederationTokenResponseTypeDef, FederatedUserTypeDef, CredentialsTypeDef
import json

def demonstrate_federated_user_typedefs():
    policy_document = {"Version": "2012-10-17", "Statement": [{"Effect": "Allow", "Action": ["s3:GetObject", "s3:PutObject"], "Resource": "arn:aws:s3:::my-bucket/*"}]}
    federation_request: GetFederationTokenRequestTypeDef = {
        "Name": "external-user-session",
        "Policy": json.dumps(policy_document),
        "DurationSeconds": 7200
    }
    print("Type-safe Federation Token Request:")
    print(f"  Federation Name: {federation_request['Name']}")
    print(f"  Duration: {federation_request['DurationSeconds']}s ({federation_request['DurationSeconds'] // 3600} hours)")
    print(f"  Policy Actions: {', '.join(policy_document['Statement'][0]['Action'])}")
    print(f"  Resource: {policy_document['Statement'][0]['Resource']}")
    federated_user: FederatedUserTypeDef = {
        "FederatedUserId": "123456789012:external-user-session",
        "Arn": "arn:aws:sts::123456789012:federated-user/external-user-session"
    }
    credentials: CredentialsTypeDef = {"AccessKeyId": "ASIAEXAMPLEFEDKEY", "SecretAccessKey": "SECRET", "SessionToken": "TOKEN", "Expiration": "2024-01-15T12:00:00Z"}
    response: GetFederationTokenResponseTypeDef = {"Credentials": credentials, "FederatedUser": federated_user}
    print(f"\nFederated User: {federated_user['Arn'][-50:]}")
    print(f"Credentials expire: {credentials['Expiration']}")
    print("TypeDefs ensure type safety for STS federated user access")

demonstrate_federated_user_typedefs()
```

## Example 5: Using TypeDefs for AssumeRoleWithWebIdentity (OIDC)
