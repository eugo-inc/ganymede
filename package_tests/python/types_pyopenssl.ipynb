## Example 1: Type-Safe Cryptographic Key Generation
Demonstrates PKey operations for RSA key generation with type annotations

```python
from OpenSSL import crypto
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from OpenSSL.crypto import PKey

def demonstrate_key_generation():
    """types-pyOpenSSL provides type annotations for cryptographic key operations"""
    
    # Generate RSA private key with type safety
    key: PKey = crypto.PKey()
    key.generate_key(crypto.TYPE_RSA, 2048)
    print(f"PKey generated: {type(key).__name__}")
    print(f"Key bits: {key.bits()}, Key check: {key.check()}")
    
    # Export key to PEM format
    private_pem = crypto.dump_privatekey(crypto.FILETYPE_PEM, key)
    public_pem = crypto.dump_publickey(crypto.FILETYPE_PEM, key)
    print(f"Private PEM: {len(private_pem)} bytes, Public PEM: {len(public_pem)} bytes")
    
    print("\ntypes-pyOpenSSL enables type-safe cryptographic key generation")

demonstrate_key_generation()
```

## Example 2: Type-Safe X.509 Certificate Creation
Demonstrates X509 and X509Name operations for certificate management

```python
from OpenSSL import crypto
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from OpenSSL.crypto import X509, X509Name

def demonstrate_certificate_creation():
    """types-pyOpenSSL provides type annotations for X.509 certificate operations"""
    
    # Create certificate with type safety
    cert: X509 = crypto.X509()
    cert.set_version(2)
    cert.set_serial_number(1000)
    cert.gmtime_adj_notBefore(0)
    cert.gmtime_adj_notAfter(365 * 24 * 60 * 60)
    
    # Set certificate subject
    subject: X509Name = cert.get_subject()
    subject.C = "US"
    subject.CN = "example.com"
    print(f"X509 certificate: {type(cert).__name__}, Subject: {subject.CN}")
    print(f"Serial: {cert.get_serial_number()}")
    
    print("\ntypes-pyOpenSSL enables type-safe X.509 certificate management")

demonstrate_certificate_creation()
```

## Example 3: Type-Safe Certificate Signing Operations
Demonstrates certificate signing with SHA256 algorithm

```python
from OpenSSL import crypto
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from OpenSSL.crypto import PKey, X509

def demonstrate_certificate_signing():
    """types-pyOpenSSL provides type annotations for certificate signing operations"""
    
    # Generate key and create certificate
    key: PKey = crypto.PKey()
    key.generate_key(crypto.TYPE_RSA, 2048)
    cert: X509 = crypto.X509()
    cert.set_serial_number(1000)
    cert.gmtime_adj_notBefore(0)
    cert.gmtime_adj_notAfter(365 * 24 * 60 * 60)
    cert.get_subject().CN = "example.com"
    cert.set_issuer(cert.get_subject())
    cert.set_pubkey(key)
    cert.sign(key, 'sha256')
    
    # Export signed certificate
    cert_pem = crypto.dump_certificate(crypto.FILETYPE_PEM, cert)
    print(f"Certificate signed with SHA256, PEM: {len(cert_pem)} bytes")
    print("\ntypes-pyOpenSSL ensures type-safe certificate signing")

demonstrate_certificate_signing()
```

## Example 4: Type-Safe SSL Context Configuration
Demonstrates SSL.Context for TLS configuration and cipher suite management

```python
from OpenSSL import SSL
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from OpenSSL.SSL import Context

def demonstrate_ssl_context():
    """types-pyOpenSSL provides type annotations for SSL context configuration"""
    
    # Create SSL context with type safety
    context: Context = SSL.Context(SSL.TLSv1_2_METHOD)
    print(f"SSL Context created: {type(context).__name__}, Method: TLSv1.2")
    
    # Configure context options
    context.set_options(SSL.OP_NO_SSLv2 | SSL.OP_NO_SSLv3)
    context.set_verify(SSL.VERIFY_PEER, lambda conn, cert, errno, depth, ok: ok)
    context.set_cipher_list(b"HIGH:!aNULL:!MD5")
    
    print(f"Configured: Disabled SSLv2/SSLv3, VERIFY_PEER, HIGH ciphers")
    print("\ntypes-pyOpenSSL enables type-safe SSL/TLS context configuration")

demonstrate_ssl_context()
```

## Example 5: Type-Safe Certificate Verification with X509Store

```python
from OpenSSL import crypto
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from OpenSSL.crypto import PKey, X509, X509Store, X509StoreContext

def demonstrate_store_verification():
    """types-pyOpenSSL provides type annotations for certificate-chain verification"""

    # Create a self-signed certificate
    key: PKey = crypto.PKey()
    key.generate_key(crypto.TYPE_RSA, 2048)
    cert: X509 = crypto.X509()
    cert.get_subject().CN = "example.com"
    cert.set_serial_number(1000)
    cert.gmtime_adj_notBefore(0)
    cert.gmtime_adj_notAfter(365 * 24 * 60 * 60)
    cert.set_issuer(cert.get_subject())
    cert.set_pubkey(key)
    cert.sign(key, 'sha256')

    # Verify it against a trust store containing itself (round-trip invariant)
    store: X509Store = crypto.X509Store()
    store.add_cert(cert)
    ctx: X509StoreContext = crypto.X509StoreContext(store, cert)
    ctx.verify_certificate()  # raises X509StoreContextError on failure
    print("Self-signed certificate verified against X509Store\n\ntypes-pyOpenSSL enables type-safe certificate verification")

demonstrate_store_verification()
```
