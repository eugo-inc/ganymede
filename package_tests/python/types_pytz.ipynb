## Example 1: Type-Safe Timezone Localization
Demonstrates timezone localization with type annotations

```python
import pytz
from datetime import datetime
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from pytz.tzinfo import BaseTzInfo

def demonstrate_timezone_localization():
    """types-pytz provides type annotations for timezone operations"""
    
    # Create timezone objects with type safety
    utc_tz: BaseTzInfo = pytz.UTC
    eastern_tz: BaseTzInfo = pytz.timezone('US/Eastern')
    tokyo_tz: BaseTzInfo = pytz.timezone('Asia/Tokyo')
    
    # Localize naive datetime
    naive_dt = datetime(2024, 1, 15, 12, 0, 0)
    utc_dt = utc_tz.localize(naive_dt)
    eastern_dt = eastern_tz.localize(naive_dt)
    
    print(f"UTC: {utc_dt}")
    print(f"Eastern: {eastern_dt}")
    print(f"Tokyo conversion: {utc_dt.astimezone(tokyo_tz)}")
    
    print("\ntypes-pytz enables type-safe timezone localization")

demonstrate_timezone_localization()
```

## Example 2: Type-Safe Timezone Discovery
Demonstrates timezone enumeration and discovery with type annotations

```python
import pytz
from typing import Set

def demonstrate_timezone_discovery():
    """types-pytz provides type annotations for timezone discovery"""
    
    # Access timezone collections with type safety
    all_tz: Set[str] = pytz.all_timezones_set
    common_tz: Set[str] = pytz.common_timezones_set
    
    print(f"Total timezones: {len(all_tz)}")
    print(f"Common timezones: {len(common_tz)}")
    print(f"Uncommon timezones: {len(all_tz - common_tz)}")
    
    # Get country timezones
    us_timezones = pytz.country_timezones.get('US', [])
    jp_timezones = pytz.country_timezones.get('JP', [])
    
    print(f"\nUS timezones: {len(us_timezones)}")
    print(f"Japan timezones: {len(jp_timezones)}")
    print(f"Sample US: {us_timezones[:3]}")
    
    print("\ntypes-pytz enables type-safe timezone discovery")

demonstrate_timezone_discovery()
```

## Example 3: Type-Safe DST (Daylight Saving Time) Handling
Demonstrates DST transitions with type annotations

```python
import pytz
from datetime import datetime
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from pytz.tzinfo import BaseTzInfo

def demonstrate_dst_handling():
    """types-pytz provides type annotations for DST transitions"""
    
    # Create timezone with DST
    eastern: BaseTzInfo = pytz.timezone('US/Eastern')
    
    # Test DST transition dates
    summer_dt = eastern.localize(datetime(2024, 7, 15, 12, 0, 0))
    winter_dt = eastern.localize(datetime(2024, 1, 15, 12, 0, 0))
    
    print(f"Summer (DST): {summer_dt.strftime('%Y-%m-%d %H:%M %Z')}")
    print(f"Winter (Standard): {winter_dt.strftime('%Y-%m-%d %H:%M %Z')}")
    print(f"Summer UTC offset: {summer_dt.utcoffset()}")
    print(f"Winter UTC offset: {winter_dt.utcoffset()}")
    print(f"DST active in summer: {bool(summer_dt.dst())}")
    print(f"DST active in winter: {bool(winter_dt.dst())}")
    
    print("\ntypes-pytz ensures type-safe DST handling")

demonstrate_dst_handling()
```

## Example 4: Type-Safe Timezone Normalization
Demonstrates timezone normalization for arithmetic operations

```python
import pytz
from datetime import datetime, timedelta
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from pytz.tzinfo import BaseTzInfo

def demonstrate_timezone_normalization():
    """types-pytz provides type annotations for timezone normalization"""
    
    # Create timezone and localized datetime
    eastern: BaseTzInfo = pytz.timezone('US/Eastern')
    dt = eastern.localize(datetime(2024, 3, 10, 1, 0, 0))
    
    print(f"Original: {dt.strftime('%Y-%m-%d %H:%M %Z')}")
    
    # Add time across DST boundary - requires normalization
    dt_plus_2h = dt + timedelta(hours=2)
    dt_normalized = eastern.normalize(dt_plus_2h)
    
    print(f"After +2h (not normalized): {dt_plus_2h.strftime('%Y-%m-%d %H:%M %Z')}")
    print(f"After normalization: {dt_normalized.strftime('%Y-%m-%d %H:%M %Z')}")
    print(f"Timezone changed: {dt.tzname() != dt_normalized.tzname()}")
    
    print("\ntypes-pytz enables type-safe timezone normalization")

demonstrate_timezone_normalization()
```

## Example 5: Type-Safe Multi-Timezone Conversion
Demonstrates converting times across multiple timezones

```python
import pytz
from datetime import datetime
from typing import TYPE_CHECKING, Dict

if TYPE_CHECKING:
    from pytz.tzinfo import BaseTzInfo

def demonstrate_multi_timezone_conversion():
    """types-pytz provides type annotations for multi-timezone conversions"""
    
    # Create multiple timezone objects
    timezones: Dict[str, BaseTzInfo] = {
        'UTC': pytz.UTC,
        'New York': pytz.timezone('America/New_York'),
        'London': pytz.timezone('Europe/London'),
        'Tokyo': pytz.timezone('Asia/Tokyo'),
        'Sydney': pytz.timezone('Australia/Sydney')
    }
    
    # Create UTC datetime
    utc_dt = pytz.UTC.localize(datetime(2024, 6, 15, 14, 30, 0))
    
    print("Same moment in different timezones:")
    for name, tz in timezones.items():
        local_dt = utc_dt.astimezone(tz)
        print(f"  {name}: {local_dt.strftime('%Y-%m-%d %H:%M %Z')}")
    
    print("\ntypes-pytz enables type-safe multi-timezone operations")

demonstrate_multi_timezone_conversion()
```
