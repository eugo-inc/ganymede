## Example 1: Type-Safe YAML Loading and Dumping
Demonstrates basic YAML load/dump operations with type annotations

```python
import yaml
from typing import Any, Dict

def demonstrate_yaml_load_dump():
    """types-pyyaml provides type annotations for YAML operations"""
    
    # YAML string to parse
    yaml_string = """
name: Configuration
version: 1.0
settings:
  debug: true
  timeout: 30
  servers:
    - host: server1.example.com
      port: 8080
    - host: server2.example.com
      port: 8081
"""
    
    # Load YAML with type safety
    data: Any = yaml.safe_load(yaml_string)
    print(f"Loaded data type: {type(data).__name__}")
    print(f"Name: {data['name']}, Version: {data['version']}")
    print(f"Debug: {data['settings']['debug']}, Servers: {len(data['settings']['servers'])}")
    
    # Dump back to YAML
    yaml_output = yaml.dump(data, default_flow_style=False)
    print(f"\nYAML output length: {len(yaml_output)} chars")
    print("\ntypes-pyyaml enables type-safe YAML serialization")

demonstrate_yaml_load_dump()
```

## Example 2: Type-Safe Custom YAML Tags
Demonstrates custom YAML tag handling with type annotations

```python
import yaml
from typing import Any

class Person:
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

def person_constructor(loader: yaml.SafeLoader, node: yaml.Node) -> Person:
    values = loader.construct_mapping(node)
    return Person(values['name'], values['age'])

def demonstrate_custom_tags():
    """types-pyyaml provides type annotations for custom tag handling"""
    
    yaml.add_constructor('!person', person_constructor, Loader=yaml.SafeLoader)
    
    yaml_data = "people:\n  - !person\n    name: Alice\n    age: 30\n  - !person\n    name: Bob\n    age: 25"
    
    data: Any = yaml.safe_load(yaml_data)
    print(f"Loaded {len(data['people'])} people")
    for person in data['people']:
        print(f"  {person.name}: {person.age} years old")
    print("\ntypes-pyyaml enables type-safe custom YAML tags")

demonstrate_custom_tags()
```

## Example 3: Type-Safe YAML Stream Processing
Demonstrates loading multiple YAML documents with type annotations

```python
import yaml
from typing import Any, Iterator

def demonstrate_yaml_stream():
    """types-pyyaml provides type annotations for multi-document YAML streams"""
    
    # Multiple YAML documents in one stream
    yaml_stream = """
---
document: 1
type: config
data:
  key1: value1
---
document: 2
type: log
entries:
  - timestamp: 2024-01-15
    message: Started
---
document: 3
type: metrics
count: 42
"""
    
    # Load all documents
    documents: Iterator[Any] = yaml.safe_load_all(yaml_stream)
    doc_list = list(documents)
    
    print(f"Loaded {len(doc_list)} YAML documents")
    for i, doc in enumerate(doc_list, 1):
        print(f"  Doc {i}: type={doc['type']}, keys={list(doc.keys())}")
    
    print("\ntypes-pyyaml enables type-safe YAML stream processing")

demonstrate_yaml_stream()
```

## Example 4: Type-Safe YAML Custom Dumper
Demonstrates custom YAML dumping with representers

```python
import yaml
from datetime import datetime

class Config:
    def __init__(self, name: str, created: datetime):
        self.name = name
        self.created = created

def config_representer(dumper: yaml.Dumper, data: Config) -> yaml.Node:
    return dumper.represent_mapping('!config', {'name': data.name, 'created': data.created.isoformat()})

def demonstrate_custom_dumper():
    """types-pyyaml provides type annotations for custom dumper operations"""
    
    yaml.add_representer(Config, config_representer)
    
    configs = [
        Config("production", datetime(2024, 1, 15, 10, 0, 0)),
        Config("staging", datetime(2024, 1, 16, 14, 30, 0))
    ]
    
    yaml_output = yaml.dump({'configs': configs}, default_flow_style=False)
    print("Custom objects dumped to YAML:")
    print(yaml_output)
    print("types-pyyaml enables type-safe custom YAML dumping")

demonstrate_custom_dumper()
```

## Example 5: Type-Safe YAML with Different Loaders
Demonstrates SafeLoader vs FullLoader with type annotations

```python
import yaml
from typing import Any

def demonstrate_yaml_loaders():
    """types-pyyaml provides type annotations for different YAML loaders"""
    
    # YAML with Python objects
    yaml_data = """
config:
  name: test
  values: !!python/tuple [1, 2, 3]
  mapping: {key1: value1, key2: value2}
"""
    
    # SafeLoader - only basic types
    safe_data: Any = yaml.safe_load(yaml_data.replace('!!python/tuple [1, 2, 3]', '[1, 2, 3]'))
    print(f"SafeLoader: {type(safe_data['config']['values']).__name__}")
    print(f"  Values: {safe_data['config']['values']}")
    
    # FullLoader - allows Python objects
    full_data: Any = yaml.full_load(yaml_data)
    print(f"\nFullLoader: {type(full_data['config']['values']).__name__}")
    print(f"  Values: {full_data['config']['values']}")
    
    # Compare loaders
    print(f"\nSafeLoader safer: True, FullLoader more flexible: True")
    print("types-pyyaml enables type-safe loader selection")

demonstrate_yaml_loaders()
```
