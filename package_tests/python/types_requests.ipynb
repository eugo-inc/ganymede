## Example 1: Type-Safe HTTP GET Requests
Demonstrates basic GET requests with type annotations

```python
# Builtin retry for flaky external endpoints — retries transient network errors with backoff.
import time as _t
def _retry(_fn, _tries=5, _delay=1.0):
    for _i in range(_tries):
        try:
            return _fn()
        except Exception:
            if _i == _tries - 1: raise
            _t.sleep(_delay * (_i + 1))
import requests
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from requests import Response

def demonstrate_get_request():
    """types-requests provides type annotations for HTTP GET operations"""
    
    # Make GET request with type safety
    response: Response = _retry(lambda: requests.get('https://httpbingo.org/get', params={'key': 'value'}))
    
    print(f"Status code: {response.status_code}")
    print(f"Content-Type: {response.headers.get('Content-Type')}")
    print(f"Response size: {len(response.content)} bytes")
    print(f"Encoding: {response.encoding}")
    print(f"OK: {response.ok}")
    
    # Parse JSON response
    data = response.json()
    print(f"URL accessed: {data.get('url')}")
    print(f"Args sent: {data.get('args')}")
    
    print("\ntypes-requests enables type-safe HTTP GET operations")

demonstrate_get_request()
```

## Example 2: Type-Safe HTTP POST with JSON
Demonstrates POST requests with JSON data and type annotations

```python
import requests
from typing import TYPE_CHECKING, Dict, Any

if TYPE_CHECKING:
    from requests import Response

def demonstrate_post_request():
    """types-requests provides type annotations for HTTP POST operations"""
    
    # Prepare JSON data
    payload: Dict[str, Any] = {
        'name': 'Test User',
        'email': 'test@example.com',
        'age': 30,
        'active': True
    }
    
    # Make POST request with JSON
    response: Response = _retry(lambda: requests.post('https://httpbingo.org/post', json=payload))
    
    print(f"Status: {response.status_code}")
    print(f"Request took: {response.elapsed.total_seconds():.3f}s")
    
    # Parse response
    data = response.json()
    print(f"Data sent: {data.get('json')}")
    
    print("\ntypes-requests enables type-safe HTTP POST with JSON")

demonstrate_post_request()
```

## Example 3: Type-Safe Session Management
Demonstrates persistent sessions with type annotations

```python
import requests
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from requests import Session, Response

def demonstrate_session():
    """types-requests provides type annotations for Session management"""
    
    # Create session with type safety
    session: Session = requests.Session()
    session.headers.update({'User-Agent': 'types-requests-demo/1.0'})
    
    # Make requests using session
    response1: Response = _retry(lambda: session.get('https://httpbingo.org/cookies/set?sessioncookie=123'))
    print(f"First request: {response1.status_code}")
    
    response2: Response = _retry(lambda: session.get('https://httpbingo.org/cookies'))
    cookies_data = response2.json()
    print(f"Session cookies: {cookies_data.get('cookies')}")
    
    # Check session state
    print(f"Session headers: {dict(session.headers)}")
    
    session.close()
    print("\ntypes-requests enables type-safe session management")

demonstrate_session()
```

## Example 4: Type-Safe Request Preparation
Demonstrates PreparedRequest for manual request building

```python
import requests
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from requests import Request, PreparedRequest, Session, Response

def demonstrate_prepared_request():
    """types-requests provides type annotations for PreparedRequest objects"""
    
    # Create request with type safety
    req: Request = requests.Request('POST', 'https://httpbingo.org/post', data={'key': 'value'})
    
    # Prepare request
    session: Session = requests.Session()
    prepared: PreparedRequest = session.prepare_request(req)
    
    print(f"Method: {prepared.method}")
    print(f"URL: {prepared.url}")
    print(f"Headers: {dict(prepared.headers)}")
    print(f"Body: {prepared.body}")
    
    # Send prepared request
    response: Response = _retry(lambda: session.send(prepared))
    print(f"\nResponse status: {response.status_code}")
    
    session.close()
    print("\ntypes-requests enables type-safe request preparation")

demonstrate_prepared_request()
```

## Example 5: Type-Safe Error Handling
Demonstrates exception handling with type annotations

```python
import requests
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from requests import Response

def demonstrate_error_handling():
    """types-requests provides type annotations for exception handling"""
    
    # Test timeout handling
    try:
        response: Response = requests.get('https://httpbingo.org/delay/10', timeout=2)
    except requests.Timeout:
        print("Timeout error caught (expected)")
    
    # Test HTTP error handling
    try:
        response: Response = requests.get('https://httpbingo.org/status/404')
        response.raise_for_status()
    except requests.HTTPError as e:
        print(f"HTTP error caught: {e.response.status_code}")
    
    # Test connection error
    try:
        requests.get('http://invalid-domain-12345.com', timeout=2)
    except requests.ConnectionError:
        print("Connection error caught (expected)")
    print("\ntypes-requests enables type-safe exception handling")

demonstrate_error_handling()
```
