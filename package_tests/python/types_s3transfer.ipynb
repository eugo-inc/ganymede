## Example 1: Type-Safe TransferConfig
Demonstrates TransferConfig for S3 transfer settings with type annotations

```python
from s3transfer.manager import TransferConfig
from typing import TYPE_CHECKING

if TYPE_CHECKING:
    from s3transfer.manager import TransferConfig as TC

def demonstrate_transfer_config():
    """types-s3transfer provides type annotations for TransferConfig"""
    
    # Create transfer config with type safety
    config: TransferConfig = TransferConfig(
        multipart_threshold=8 * 1024 * 1024,  # 8 MB
        max_request_concurrency=10,
        multipart_chunksize=8 * 1024 * 1024,  # 8 MB
        max_bandwidth=1024 * 1024 * 10  # 10 MB/s
    )
    
    print(f"TransferConfig created: {type(config).__name__}")
    print(f"Multipart threshold: {config.multipart_threshold / (1024*1024):.0f} MB")
    print(f"Max request concurrency: {config.max_request_concurrency}")
    print(f"Chunk size: {config.multipart_chunksize / (1024*1024):.0f} MB")
    print(f"Max bandwidth: {config.max_bandwidth / (1024*1024):.0f} MB/s")
    print(f"IO chunk size: {config.io_chunksize / 1024:.0f} KB")
    
    print("\ntypes-s3transfer enables type-safe transfer configuration")

demonstrate_transfer_config()
```

## Example 2: Type-Safe OSUtils
Demonstrates OSUtils for file system operations with type annotations

```python
from s3transfer.utils import OSUtils
import tempfile
import os

def demonstrate_os_utils():
    """types-s3transfer provides type annotations for OSUtils operations"""
    
    # Create OSUtils instance with type safety
    os_utils: OSUtils = OSUtils()
    
    # Create temporary file
    with tempfile.NamedTemporaryFile(mode='w', delete=False) as f:
        f.write("Test data for s3transfer operations")
        temp_file = f.name
    
    # Get file size
    file_size = os_utils.get_file_size(temp_file)
    print(f"File size: {file_size} bytes")
    
    # Open file with chunk reader
    with os_utils.open_file_chunk_reader(temp_file, 0, 20, []) as reader:
        chunk = reader.read(20)
        print(f"Read chunk: {chunk[:20]}")
    
    # Cleanup
    os.remove(temp_file)
    print("\ntypes-s3transfer enables type-safe file operations")

demonstrate_os_utils()
```

## Example 3: Type-Safe ReadFileChunk
Demonstrates ReadFileChunk for reading file chunks with type annotations

```python
from s3transfer.utils import ReadFileChunk
import tempfile
import os

def demonstrate_read_file_chunk():
    """types-s3transfer provides type annotations for ReadFileChunk"""
    
    # Create test file
    with tempfile.NamedTemporaryFile(mode='wb', delete=False) as f:
        f.write(b"0123456789" * 100)  # 1000 bytes
        temp_file = f.name
    
    # Read file chunk with type safety
    with open(temp_file, 'rb') as fileobj:
        chunk: ReadFileChunk = ReadFileChunk.from_filename(temp_file, start_byte=0, chunk_size=100)
        
        print(f"Chunk size: {len(chunk)} bytes")
        print(f"First 20 bytes: {chunk.read(20)}")
        print(f"Tell position: {chunk.tell()}")
        
        # Seek and read more
        chunk.seek(50)
        print(f"After seek(50): {chunk.read(10)}")
    
    os.remove(temp_file)
    print("\ntypes-s3transfer enables type-safe chunk reading")

demonstrate_read_file_chunk()
```

## Example 4: Type-Safe BandwidthLimiter
Demonstrates BandwidthLimiter for rate limiting with type annotations

```python
from s3transfer.bandwidth import BandwidthLimiter, LeakyBucket
import io

def demonstrate_bandwidth_limiter():
    """types-s3transfer provides type annotations for BandwidthLimiter"""
    
    # Create leaky bucket and bandwidth limiter with type safety (1 MB/s limit)
    max_rate = 1024 * 1024  # 1 MB/s
    leaky_bucket: LeakyBucket = LeakyBucket(max_rate=max_rate)
    limiter: BandwidthLimiter = BandwidthLimiter(leaky_bucket)
    
    print(f"BandwidthLimiter created: {type(limiter).__name__}")
    print(f"Configured max rate: {max_rate / (1024*1024):.1f} MB/s")
    
    # Create a test stream
    test_data = b"0" * (500 * 1024)  # 500 KB
    stream = io.BytesIO(test_data)
    
    # Get bandwidth-limited stream
    limited_stream = limiter.get_bandwith_limited_stream(stream, transfer_coordinator=None)
    
    # Read from limited stream
    chunk = limited_stream.read(100 * 1024)
    print(f"\nRead {len(chunk) / 1024:.0f} KB through bandwidth limiter")
    print("\ntypes-s3transfer enables type-safe bandwidth limiting")

demonstrate_bandwidth_limiter()
```

## Example 5: Type-Safe Transfer Constants
Demonstrates s3transfer constants and utilities with type annotations

```python
from s3transfer import MB
from s3transfer.constants import KB
from s3transfer.manager import TransferConfig
from s3transfer.utils import CallArgs

def demonstrate_transfer_constants():
    """types-s3transfer provides type annotations for constants and utilities"""
    
    # Use size constants with type safety
    print(f"KB constant: {KB} bytes")
    print(f"MB constant: {MB} bytes")
    
    # Create config using constants
    config: TransferConfig = TransferConfig(
        multipart_threshold=8 * MB,
        multipart_chunksize=8 * MB,
        max_request_concurrency=10
    )
    print(f"\nConfig threshold: {config.multipart_threshold / MB:.0f} MB")
    
    # Use CallArgs utility
    args: CallArgs = CallArgs(bucket='my-bucket', key='my-key', extra={'ACL': 'public-read'})
    print(f"\nCallArgs created:")
    print(f"  Bucket: {args.bucket}")
    print(f"  Key: {args.key}")
    
    print("\ntypes-s3transfer provides type-safe constants and utilities")

demonstrate_transfer_constants()
```
