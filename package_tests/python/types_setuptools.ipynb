## Example 2: find_packages() with type annotations

```python
# Demonstrate setuptools.find_packages() with type safety
from setuptools import find_packages
from typing import List
import tempfile
import os

# Create a temporary directory structure to demonstrate find_packages
with tempfile.TemporaryDirectory() as temp_dir:
    # Create a sample package structure
    package_dir = os.path.join(temp_dir, 'my_project')
    os.makedirs(package_dir)
    
    # Create __init__.py files to make directories into packages
    for subdir in ['', 'utils', 'core', 'tests']:
        dir_path = os.path.join(package_dir, subdir) if subdir else package_dir
        if subdir:
            os.makedirs(dir_path, exist_ok=True)
        with open(os.path.join(dir_path, '__init__.py'), 'w') as f:
            f.write('# Package init file\n')
    
    # Use find_packages with type annotation
    packages: List[str] = find_packages(where=package_dir)
    
    print("Found packages with type-safe find_packages():")
    for pkg in sorted(packages):
        print(f"  - {pkg}")
    
    print(f"\nTotal packages found: {len(packages)}")
```

## Example 3: Stub distribution metadata with importlib.metadata

```python
# pkg_resources was removed from setuptools 81+ (and its stubs from types-setuptools).
# Modern replacement: importlib.metadata — verify the setuptools stub files instead.
import importlib.metadata
from typing import List

# The stub distribution itself is queryable metadata.
stub_dist = importlib.metadata.distribution('types-setuptools')
print(f"types-setuptools version: {stub_dist.version}")
pyi: List[str] = [str(f) for f in (stub_dist.files or []) if str(f).endswith('.pyi')]
print(f"stub files installed: {len(pyi)}")
assert any('setuptools-stubs' in s for s in pyi), "setuptools-stubs .pyi files missing"

# Typed distribution queries for real installed packages.
for name in ('setuptools', 'types-setuptools'):
    version: str = importlib.metadata.version(name)
    print(f"  {name}: {version}")

total = sum(1 for _ in importlib.metadata.distributions())
print(f"Total installed distributions: {total}")
```

## Example 4: Extension and Command with type annotations

```python
# Demonstrate setuptools extension and command functionality
from setuptools import Extension, Command
from typing import List, Optional, Any
import tempfile
import os

# Create a type-safe Extension definition
def create_extension(name: str, sources: List[str], 
                    include_dirs: Optional[List[str]] = None) -> Extension:
    """Create a setuptools Extension with type safety."""
    return Extension(
        name=name,
        sources=sources,
        include_dirs=include_dirs or [],
        language='c'
    )

# Example extension configuration
ext_sources: List[str] = ['example.c', 'utils.c']
ext_includes: List[str] = ['/usr/local/include']

sample_extension = create_extension(
    name='example_module',
    sources=ext_sources,
    include_dirs=ext_includes
)

print("Created Extension with type safety:")
print(f"  Name: {sample_extension.name}")
print(f"  Sources: {sample_extension.sources}")
print(f"  Include dirs: {sample_extension.include_dirs}")
print(f"  Language: {sample_extension.language}")

# Demonstrate custom command class with type annotations
class CustomCommand(Command):
    """Custom setuptools command with type annotations."""
    
    description: str = 'Run custom build command'
    user_options: List[tuple] = [
        ('verbose', 'v', 'Enable verbose output'),
    ]
    
    def initialize_options(self) -> None:
        self.verbose: bool = False
    
    def finalize_options(self) -> None:
        pass
    
    def run(self) -> None:
        print(f"Running custom command (verbose={self.verbose})")

print("\nCustom command class created with type annotations")
print(f"Description: {CustomCommand.description}")
print(f"User options: {CustomCommand.user_options}")
```
