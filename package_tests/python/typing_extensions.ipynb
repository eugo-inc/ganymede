# typing_extensions: 5 Basic Usage Examples

```python
# Example 1: Using Annotated
from typing_extensions import Annotated
def f(x: Annotated[int, 'metadata']):
    print(x)
f(1)
```

```python
# Example 2: Using TypedDict
from typing_extensions import TypedDict
class Movie(TypedDict):
    name: str
    year: int
m = Movie(name='Inception', year=2010)
print(m)
```

```python
# Example 3: Using Literal
from typing_extensions import Literal
def accepts_only_foo(x: Literal['foo']):
    print(x)
accepts_only_foo('foo')
```

```python
# Example 4: Using Final
from typing_extensions import Final
PI: Final = 3.14159
print(PI)
```

```python
# Example 5: Using Protocol
from typing_extensions import Protocol
class Drawable(Protocol):
    def draw(self) -> None: ...
class Circle:
    def draw(self): print('Circle!')
c = Circle()
c.draw()
```
