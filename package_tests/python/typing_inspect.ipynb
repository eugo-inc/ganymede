## Example 1: Type Inspection for Generic Types
Demonstrates inspecting generic types like List, Dict, Optional

```python
import typing_inspect as ti
from typing import List, Dict, Optional

def demonstrate_generic_inspection():
    """typing_inspect provides runtime inspection of generic types"""
    
    # Inspect List[int]
    list_type = List[int]
    print(f"Type: {list_type}")
    print(f"Is generic: {ti.is_generic_type(list_type)}")
    print(f"Origin: {ti.get_origin(list_type)}")
    print(f"Args: {ti.get_args(list_type)}")
    
    # Inspect Dict[str, int]
    dict_type = Dict[str, int]
    print(f"\nType: {dict_type}")
    print(f"Origin: {ti.get_origin(dict_type)}")
    print(f"Args: {ti.get_args(dict_type)}")
    
    # Inspect Optional[str]
    optional_type = Optional[str]
    print(f"\nType: {optional_type}")
    print(f"Is optional: {ti.is_optional_type(optional_type)}")
    print(f"Args: {ti.get_args(optional_type)}")
    
    print("\ntyping_inspect enables runtime type introspection")

demonstrate_generic_inspection()
```

## Example 2: Callable Type Inspection
Demonstrates inspecting Callable types and their signatures

```python
import typing_inspect as ti
from typing import Callable

def demonstrate_callable_inspection():
    """typing_inspect provides inspection of Callable types"""
    
    # Inspect simple callable
    simple_callable = Callable[[int, str], bool]
    print(f"Type: {simple_callable}")
    print(f"Is callable: {ti.is_callable_type(simple_callable)}")
    print(f"Args: {ti.get_args(simple_callable)}")
    
    # Inspect callable with no args
    no_args_callable = Callable[[], None]
    print(f"\nType: {no_args_callable}")
    print(f"Is callable: {ti.is_callable_type(no_args_callable)}")
    print(f"Args: {ti.get_args(no_args_callable)}")
    
    # Inspect callable with multiple params
    multi_callable = Callable[[int, int, str], dict]
    print(f"\nType: {multi_callable}")
    args = ti.get_args(multi_callable)
    print(f"Param types: {args[0]}")
    print(f"Return type: {args[1]}")
    
    print("\ntyping_inspect enables callable type introspection")

demonstrate_callable_inspection()
```

## Example 3: TypeVar and Generic Base Inspection
Demonstrates inspecting TypeVars and generic base classes

```python
import typing_inspect as ti
from typing import TypeVar, Generic

T = TypeVar('T', bound=int)
U = TypeVar('U', int, str)

class Container(Generic[T]):
    pass

def demonstrate_typevar_inspection():
    """typing_inspect provides TypeVar and generic base inspection"""
    
    # Inspect TypeVar with bound
    print(f"TypeVar T: {T}, Is TypeVar: {ti.is_typevar(T)}")
    print(f"Bound: {ti.get_bound(T)}")
    
    # Inspect TypeVar with constraints
    print(f"\nTypeVar U: {U}, Is TypeVar: {ti.is_typevar(U)}")
    print(f"Constraints: {ti.get_constraints(U)}")
    
    # Inspect generic base classes
    container_int = Container[int]
    print(f"\nContainer[int]: {container_int}")
    print(f"Generic bases: {ti.get_generic_bases(container_int)}")
    print(f"Args: {ti.get_args(container_int)}")
    print("\ntyping_inspect enables TypeVar introspection")

demonstrate_typevar_inspection()
```

## Example 4: Union and Literal Type Inspection
Demonstrates inspecting Union and Literal types

```python
import typing_inspect as ti
from typing import Union, Literal

def demonstrate_union_literal_inspection():
    """typing_inspect provides Union and Literal type inspection"""
    
    # Inspect Union type
    union_type = Union[int, str, bool]
    print(f"Type: {union_type}")
    print(f"Is union: {ti.is_union_type(union_type)}")
    print(f"Args: {ti.get_args(union_type)}")
    
    # Inspect nested Union
    nested_union = Union[int, Union[str, float]]
    print(f"\nType: {nested_union}")
    print(f"Is union: {ti.is_union_type(nested_union)}")
    print(f"Args: {ti.get_args(nested_union)}")
    
    # Inspect Literal type
    literal_type = Literal['red', 'green', 'blue']
    print(f"\nType: {literal_type}")
    print(f"Is literal: {ti.is_literal_type(literal_type)}")
    print(f"Literal values: {ti.get_args(literal_type)}")
    
    print("\ntyping_inspect enables Union and Literal introspection")

demonstrate_union_literal_inspection()
```

## Example 5: Special Type Annotations Inspection
Demonstrates inspecting ClassVar, Final, and Tuple types

```python
import typing_inspect as ti
from typing import ClassVar, Final, Tuple

def demonstrate_special_annotations():
    """typing_inspect provides inspection of special type annotations"""
    
    # Inspect ClassVar
    classvar_type = ClassVar[int]
    print(f"ClassVar[int]: Is ClassVar: {ti.is_classvar(classvar_type)}")
    print(f"Args: {ti.get_args(classvar_type)}")
    
    # Inspect Final
    final_type = Final[str]
    print(f"\nFinal[str]: Is Final: {ti.is_final_type(final_type)}")
    print(f"Args: {ti.get_args(final_type)}")
    
    # Inspect Tuple
    tuple_type = Tuple[int, str, bool]
    print(f"\nTuple[int, str, bool]: Is Tuple: {ti.is_tuple_type(tuple_type)}")
    print(f"Args: {ti.get_args(tuple_type)}")
    
    # Inspect variable-length Tuple
    var_tuple = Tuple[int, ...]
    print(f"\nTuple[int, ...]: Args: {ti.get_args(var_tuple)}")
    
    print("\ntyping_inspect enables special annotation introspection")

demonstrate_special_annotations()
```
