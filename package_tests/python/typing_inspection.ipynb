## Example 1: Advanced Type Object Inspection
Demonstrates inspecting type annotations with typing_objects module

```python
from typing_inspection import typing_objects
from typing import ClassVar, Final, TypeVar, NewType, Literal, get_origin

def demonstrate_type_object_inspection():
    """typing_inspection.typing_objects provides advanced type inspection"""
    
    # Test ClassVar and Final (check origins)
    classvar_type = ClassVar[int]
    final_type = Final[str]
    print(f"ClassVar origin: {typing_objects.is_classvar(get_origin(classvar_type))}")
    print(f"Final origin: {typing_objects.is_final(get_origin(final_type))}")
    
    # Test TypeVar and NewType
    T = TypeVar('T')
    UserId = NewType('UserId', int)
    print(f"TypeVar T: {typing_objects.is_typevar(T)}")
    print(f"NewType UserId: {typing_objects.is_newtype(UserId)}")
    
    # Test Literal class
    print(f"Literal class: {typing_objects.is_literal(Literal)}")
    print("\ntyping_inspection enables advanced type origin detection")

demonstrate_type_object_inspection()
```

## Example 2: Literal Value Extraction
Demonstrates extracting literal values from Literal types

```python
from typing_inspection import introspection
from typing import Literal

def demonstrate_literal_extraction():
    """typing_inspection.introspection provides literal value extraction"""
    
    # Extract values from Literal types
    color_type = Literal['red', 'green', 'blue']
    color_values = list(introspection.get_literal_values(color_type))
    print(f"Color literals: {color_values}")
    
    # Extract from numeric literals
    num_type = Literal[1, 2, 3, 4, 5]
    num_values = list(introspection.get_literal_values(num_type))
    print(f"Number literals: {num_values}")
    
    # Extract from mixed literals
    mixed_type = Literal['a', 1, True, None]
    mixed_values = list(introspection.get_literal_values(mixed_type))
    print(f"Mixed literals: {mixed_values}, count: {len(color_values)}")
    print("\ntyping_inspection enables literal value extraction")

demonstrate_literal_extraction()
```

## Example 3: InspectedAnnotation for Detailed Analysis
Demonstrates analyzing annotations with InspectedAnnotation

```python
from typing_inspection import introspection
from typing import List, Optional

def demonstrate_inspected_annotation():
    """typing_inspection.introspection provides InspectedAnnotation for detailed analysis"""
    
    # Create InspectedAnnotation with proper parameters
    simple_ann = introspection.InspectedAnnotation(type=int, qualifiers=set(), metadata=[])
    print(f"Simple annotation type: {simple_ann.type}")
    print(f"Qualifiers: {simple_ann.qualifiers}")
    
    # Inspect List annotation
    list_ann = introspection.InspectedAnnotation(type=List[str], qualifiers=set(), metadata=[])
    print(f"\nList annotation type: {list_ann.type}")
    print(f"Origin: {introspection.get_origin(list_ann.type)}")
    print(f"Args: {introspection.get_args(list_ann.type)}")
    
    # Inspect Optional with metadata
    opt_ann = introspection.InspectedAnnotation(type=Optional[int], qualifiers=set(), metadata=["doc"])
    print(f"\nOptional annotation: {opt_ann.type}")
    print(f"Metadata: {opt_ann.metadata}")
    
    print("\ntyping_inspection enables detailed annotation inspection")

demonstrate_inspected_annotation()
```

## Example 4: Extended Type Checking Functions
Demonstrates additional type checking functions from typing_objects

```python
from typing_inspection import typing_objects
from typing import Any, ParamSpec, TypeVar, Concatenate, Union, get_origin

def demonstrate_extended_type_checks():
    """typing_inspection.typing_objects provides extended type checking"""
    
    # Test Any type
    print(f"Any is Any: {typing_objects.is_any(Any)}")
    print(f"int is Any: {typing_objects.is_any(int)}")
    
    # Test ParamSpec and TypeVar
    P = ParamSpec('P')
    print(f"\nParamSpec P: {typing_objects.is_paramspec(P)}")
    print(f"TypeVar T: {typing_objects.is_paramspec(TypeVar('T'))}")
    
    # Test Concatenate and Union (check origins)
    concat_type = Concatenate[int, P]
    union_type = Union[int, str]
    print(f"\nConcatenate: {typing_objects.is_concatenate(get_origin(concat_type))}")
    print(f"Union: {typing_objects.is_union(get_origin(union_type))}")
    print("\ntyping_inspection provides extended type checking")

demonstrate_extended_type_checks()
```

## Example 5: Annotation Source Detection
Demonstrates detecting annotation sources with AnnotationSource enum

```python
from typing_inspection import introspection
from typing import NamedTuple
from dataclasses import dataclass

def demonstrate_annotation_sources():
    """typing_inspection.introspection provides AnnotationSource for tracking origins"""
    
    # Show all annotation source types
    print("Available annotation sources:")
    for source in introspection.AnnotationSource:
        print(f"  {source.name}: {source.value}")
    
    # Create example structures
    @dataclass
    class Config:
        name: str
        value: int
    
    Point = NamedTuple('Point', [('x', int), ('y', int)])
    
    print(f"\nDataclass: {Config.__name__}, NamedTuple: {Point.__name__}")
    print("Annotation sources track where type hints originate")
    print("\ntyping_inspection enables annotation source tracking")

demonstrate_annotation_sources()
```
