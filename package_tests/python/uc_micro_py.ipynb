# uc_micro_py: 5 Basic Usage Examples

```python
# Example 1: Category regex sources
# uc_micro (the Python port of the JavaScript uc.micro package) exposes each Unicode
# category as a module whose REGEX attribute is a regular-expression source string.
import re
import uc_micro

for name in ('Any', 'Cc', 'Cf', 'P', 'Z'):
    source = getattr(uc_micro, name).REGEX
    print(f"uc_micro.{name}.REGEX: {len(source)} characters")
    assert isinstance(source, str) and source, f"uc_micro.{name}.REGEX is empty"
    re.compile(source)  # each source compiles as a Python regex
```

```python
# Example 2: Control and format characters (Cc, Cf)
import re
import uc_micro

assert re.fullmatch(uc_micro.Cc.REGEX, '\x07') is not None, "BEL (U+0007) is a control character"
assert re.fullmatch(uc_micro.Cc.REGEX, 'a') is None
assert re.fullmatch(uc_micro.Cf.REGEX, 'a') is None
assert re.fullmatch(uc_micro.Cf.REGEX, '\u200b') is not None, "ZERO WIDTH SPACE (U+200B) is a format character"
print("U+0007 is in Cc, U+200B is in Cf, 'a' is in neither")
```

```python
# Example 3: Membership is a regex match
import uc_micro

space_char = ord(' ')
print(f"Character code for space: {space_char}")

# Each category (Cc, Cf, P, Z) is a module whose REGEX attribute is a regex SOURCE string (the
# port of JavaScript's uc.micro) -- not a set, not a predicate. Membership is a regex match.
import re
print(f"Cc regex source: {uc_micro.Cc.REGEX!r}")
in_cc = re.fullmatch(uc_micro.Cc.REGEX, chr(space_char)) is not None
in_z = re.fullmatch(uc_micro.Z.REGEX, chr(space_char)) is not None
print(f"In Cc:", in_cc)
print(f"In Z:", in_z)
assert not in_cc and in_z, "space is a separator (Z), not a control character (Cc)"
assert re.fullmatch(uc_micro.Cc.REGEX, "\x07") is not None, "BEL (0x07) is a control character"
```

```python
# Example 4: Punctuation (P)
import re
import uc_micro

for ch in ('!', '?', '\u3002'):  # U+3002 is IDEOGRAPHIC FULL STOP
    assert re.fullmatch(uc_micro.P.REGEX, ch) is not None, f"{ch!r} should be punctuation"
assert re.fullmatch(uc_micro.P.REGEX, 'a') is None
print("'!', '?' and U+3002 are in P; 'a' is not")
```

```python
# Example 5: Separators (Z)
import re
import uc_micro

for ch in (' ', '\u00a0', '\u3000'):  # space, NO-BREAK SPACE, IDEOGRAPHIC SPACE
    assert re.fullmatch(uc_micro.Z.REGEX, ch) is not None, f"{ch!r} should be a separator"
assert re.fullmatch(uc_micro.Z.REGEX, 'a') is None
print("space, U+00A0 and U+3000 are in Z; 'a' is not")
```
