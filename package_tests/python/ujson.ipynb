## Example 1: High-performance JSON encoding and decoding

```python
import ujson
import json
import time

# Example 1: High-performance JSON encoding and decoding
print("High-performance JSON with ujson:")

# Sample data structure
data = {
    "user_id": 12345,
    "username": "alice_smith",
    "profile": {
        "email": "alice@example.com",
        "preferences": {
            "theme": "dark",
            "notifications": True,
            "language": "en"
        }
    },
    "activities": [
        {"type": "login", "timestamp": 1640995200},
        {"type": "view_page", "timestamp": 1640995260},
        {"type": "logout", "timestamp": 1640995800}
    ],
    "metadata": {
        "last_active": "2024-01-15T10:30:00Z",
        "session_count": 42,
        "total_time": 3600.5
    }
}

# Encode to JSON string
json_string = ujson.dumps(data)
print(f"Encoded JSON length: {len(json_string)} characters")

# Decode back to Python object
decoded_data = ujson.loads(json_string)
print(f"Round-trip successful: {data == decoded_data}")
assert decoded_data == data

# Show specific field access
print(f"Username: {decoded_data['username']}")
print(f"Theme: {decoded_data['profile']['preferences']['theme']}")
print(f"Activity count: {len(decoded_data['activities'])}")

# ujson features
print(f"ujson version: {getattr(ujson, '__version__', 'unknown')}")
print("ujson provides ultra-fast JSON processing for Python")
```

## Example 3: HTML-Safe JSON Encoding
Demonstrates ujson's encode_html_chars parameter for XSS protection

```python
import ujson

def demonstrate_html_safe_encoding():
    """ujson's encode_html_chars escapes HTML characters for XSS protection"""
    
    # Data with HTML/XSS risk characters
    risky_data = {
        "user_input": "<script>alert('XSS')</script>",
        "comment": "Hello & <goodbye>",
        "html_snippet": "<div class='test'>Content</div>"
    }
    
    # Without HTML encoding (unsafe for HTML embedding)
    unsafe_json = ujson.dumps(risky_data, encode_html_chars=False)
    print("Without encode_html_chars (unsafe):")
    print(unsafe_json[:80] + "...")
    
    # With HTML encoding (safe for HTML embedding)
    safe_json = ujson.dumps(risky_data, encode_html_chars=True)
    print("\nWith encode_html_chars=True (safe):")
    print(safe_json[:80] + "...")
    
    print("\nujson provides XSS-safe JSON encoding")

demonstrate_html_safe_encoding()
```

## Example 4: Unicode Without ASCII Escaping
Demonstrates ujson's ensure_ascii parameter for natural Unicode output

```python
import ujson

def demonstrate_unicode_handling():
    """ujson's ensure_ascii parameter controls Unicode escape sequences"""
    
    # Data with Unicode characters
    unicode_data = {
        "japanese": "こんにちは",
        "emoji": "🎉🚀✨",
        "russian": "Привет"
    }
    
    # With ensure_ascii=True (default) - uses escape sequences
    ascii_json = ujson.dumps(unicode_data, ensure_ascii=True)
    print("With ensure_ascii=True (escaped):")
    print(ascii_json[:70] + "...")
    
    # With ensure_ascii=False - natural Unicode output
    unicode_json = ujson.dumps(unicode_data, ensure_ascii=False)
    print("\nWith ensure_ascii=False (natural):")
    print(unicode_json)
    print(f"\nSize: {len(ascii_json)} vs {len(unicode_json)} chars")
    print("ujson supports natural Unicode output")

demonstrate_unicode_handling()
```

## Example 5: URL-Friendly JSON Encoding
Demonstrates ujson's escape_forward_slashes parameter for URL handling

```python
import ujson

def demonstrate_url_friendly_encoding():
    """ujson's escape_forward_slashes controls URL escaping"""
    
    # Data with URLs and paths
    url_data = {
        "api": "https://api.example.com/v1/users",
        "cdn": "https://cdn.example.com/images/photo.jpg",
        "path": "/home/user/file.txt"
    }
    
    # With escape_forward_slashes=True (default) - escapes /
    escaped_json = ujson.dumps(url_data, escape_forward_slashes=True)
    print("With escape_forward_slashes=True (escaped):")
    print(escaped_json[:70] + "...")
    
    # With escape_forward_slashes=False - keeps / as-is
    unescaped_json = ujson.dumps(url_data, escape_forward_slashes=False)
    print("\nWith escape_forward_slashes=False (readable):")
    print(unescaped_json)
    print("\nujson provides flexible URL encoding")

demonstrate_url_friendly_encoding()
```
