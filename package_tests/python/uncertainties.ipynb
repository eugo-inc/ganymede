## Example 1: Basic Error Propagation
Demonstrates automatic error propagation in calculations

```python
from uncertainties import ufloat

def demonstrate_error_propagation():
    """uncertainties automatically propagates errors through calculations"""
    
    # Create measurements with uncertainties
    length = ufloat(10.0, 0.1)  # 10.0 ± 0.1 cm
    width = ufloat(5.0, 0.2)    # 5.0 ± 0.2 cm
    
    print(f"Length: {length} cm")
    print(f"Width: {width} cm")
    
    # Automatic error propagation in operations
    area = length * width
    print(f"\nArea (L × W): {area} cm²")
    
    perimeter = 2 * (length + width)
    print(f"Perimeter: {perimeter} cm")
    
    ratio = length / width
    print(f"Aspect ratio: {ratio}")
    
    print("\nuncertainties handles error propagation automatically")

demonstrate_error_propagation()
```

## Example 2: Mathematical Functions with Uncertainties
Demonstrates error propagation through mathematical functions

```python
from uncertainties import ufloat
import uncertainties.umath as umath

def demonstrate_math_functions():
    """uncertainties.umath provides math functions with error propagation"""
    
    # Physical measurement with uncertainty
    angle = ufloat(0.785, 0.01)  # ~45° ± 0.01 rad
    distance = ufloat(100, 2)     # 100 ± 2 meters
    
    print(f"Angle: {angle} rad, Distance: {distance} m")
    
    # Trigonometric functions
    height = distance * umath.sin(angle)
    print(f"\nHeight (d × sin(θ)): {height} m")
    
    # Other math functions
    print(f"sqrt(distance): {umath.sqrt(distance)}")
    print(f"exp(angle): {umath.exp(angle)}")
    
    print("\nuncertainties.umath propagates errors through functions")

demonstrate_math_functions()
```

## Example 3: Correlated Variables
Demonstrates handling correlated uncertainties

```python
from uncertainties import correlated_values
import numpy as np

def demonstrate_correlated_variables():
    """uncertainties handles correlated measurements properly"""
    
    # Covariance matrix: [[var(x), cov(x,y)], [cov(y,x), var(y)]]
    covariance_matrix = [[1.0, 0.5],  # x variance=1, correlation=0.5
                         [0.5, 2.0]]  # y variance=2
    
    # Create correlated variables
    x, y = correlated_values([10, 20], covariance_matrix)
    
    print(f"Correlated variables:")
    print(f"x = {x}")
    print(f"y = {y}")
    
    # When correlated, x - x should have zero uncertainty
    diff_self = x - x
    print(f"\nx - x = {diff_self} (correctly zero uncertainty)")
    
    # Correlated sum
    sum_xy = x + y
    print(f"x + y = {sum_xy}")
    
    print("\nuncertainties properly handles correlated variables")

demonstrate_correlated_variables()
```

## Example 4: Accessing Error Components
Demonstrates extracting nominal values and uncertainties

```python
from uncertainties import ufloat

def demonstrate_component_access():
    """uncertainties allows accessing nominal values and standard deviations"""
    
    # Create measurement with uncertainty
    measurement = ufloat(9.81, 0.02)  # g = 9.81 ± 0.02 m/s²
    
    print(f"Full measurement: {measurement}")
    
    # Access components
    nominal = measurement.nominal_value
    error = measurement.std_dev
    
    print(f"\nNominal value: {nominal}")
    print(f"Standard deviation: {error}")
    print(f"Relative error: {error/nominal * 100:.2f}%")
    
    # Use in calculations needing only nominal value
    mass = 10  # kg
    force = mass * nominal
    print(f"\nForce (nominal only): {force} N")
    
    print("\nuncertainties provides easy component access")

demonstrate_component_access()
```

## Example 5: Custom Formatting Options
Demonstrates formatting uncertainties for different output styles

```python
from uncertainties import ufloat

def demonstrate_formatting():
    """uncertainties supports multiple formatting styles"""
    
    # Create measurement
    value = ufloat(123.456789, 0.123456)
    
    print("Formatting 123.456789 ± 0.123456:")
    print(f"Default: {value}")
    print(f"Short (.1u): {value:.1u}")
    print(f"LaTeX (:L): {value:L}")
    print(f"Pretty (:P): {value:P}")
    print(f"Scientific (.2e): {value:.2e}")
    print("\nuncertainties supports flexible formatting")

demonstrate_formatting()
```
