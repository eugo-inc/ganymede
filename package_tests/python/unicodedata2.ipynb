## Example 1: Basic Unicode character information and normalization

```python
import unicodedata2

# Example 1: Basic Unicode character information and normalization
print("Basic Unicode character operations:")

# Character name and properties lookup
test_chars = ['A', 'α', '中', '🦄', '€']
for char in test_chars:
    try:
        name = unicodedata2.name(char)
        category = unicodedata2.category(char)
        print(f"'{char}': {name} (Category: {category})")
    except ValueError:
        print(f"'{char}': No name available (Category: {unicodedata2.category(char)})")

# Unicode normalization forms
test_text = "café naïve résumé"
print(f"\nNormalization of '{test_text}':")

normalization_forms = ['NFC', 'NFD', 'NFKC', 'NFKD']
for form in normalization_forms:
    normalized = unicodedata2.normalize(form, test_text)
    print(f"{form:4}: {normalized} (length: {len(normalized)})")

# Show decomposition differences
print(f"\nDetailed NFD decomposition:")
nfd_text = unicodedata2.normalize('NFD', 'café')
for i, char in enumerate(nfd_text):
    print(f"  Position {i}: '{char}' (U+{ord(char):04X}) - {unicodedata2.name(char, 'NO NAME')}")

print(f"unicodedata2 provides Unicode {unicodedata2.unidata_version} data")
```

## Example 2: Advanced Unicode property analysis

```python
import unicodedata2

# Example 2: Advanced Unicode property analysis
print("Advanced Unicode property analysis:")

# Bidirectional properties for international text
bidi_text = "English العربية עברית"
print(f"Bidirectional text analysis: '{bidi_text}'")

for char in bidi_text:
    if char != ' ':
        bidi_type = unicodedata2.bidirectional(char)
        category = unicodedata2.category(char)
        print(f"  '{char}': bidirectional={bidi_type}, category={category}")

# Numeric value extraction
numeric_chars = ['1', '½', 'Ⅴ', '²', '₃']
print(f"\nNumeric value analysis:")
for char in numeric_chars:
    try:
        numeric_val = unicodedata2.numeric(char)
        decimal_val = unicodedata2.decimal(char, None)
        digit_val = unicodedata2.digit(char, None)
        print(f"'{char}': numeric={numeric_val}, decimal={decimal_val}, digit={digit_val}")
    except ValueError:
        print(f"'{char}': Not a numeric character")

# Case mapping properties
case_examples = ['A', 'a', 'ß', 'İ', 'Σ']
print(f"\nCase mapping analysis:")
for char in case_examples:
    mirrored = unicodedata2.mirrored(char)
    decomp = unicodedata2.decomposition(char)
    print(f"'{char}': mirrored={mirrored}, decomposition='{decomp}'")

print(f"Enhanced Unicode property access beyond standard library")
```

## Example 3: Text normalization and comparison utilities

```python
import unicodedata2

# Example 3: Text normalization and comparison utilities
print("Text normalization and comparison:")

# Normalization for text comparison
texts = [
    "café",      # Precomposed
    "cafe\u0301",  # Decomposed (e + combining acute)
    "CAFÉ",      # Different case
    "ℌₑˡˡₒ",    # Mathematical/special characters
]

print("Original texts and their normalizations:")
for text in texts:
    nfc = unicodedata2.normalize('NFC', text)
    nfd = unicodedata2.normalize('NFD', text)
    nfkc = unicodedata2.normalize('NFKC', text)
    nfkd = unicodedata2.normalize('NFKD', text)
    
    print(f"Original: '{text}' (len={len(text)})")
    print(f"  NFC:  '{nfc}' (len={len(nfc)})")
    print(f"  NFD:  '{nfd}' (len={len(nfd)})")
    print(f"  NFKC: '{nfkc}' (len={len(nfkc)})")
    print(f"  NFKD: '{nfkd}' (len={len(nfkd)})")

# Text similarity checking with normalization
def normalize_for_comparison(text):
    """Normalize text for case-insensitive comparison"""
    # Normalize to NFKD (compatibility decomposition)
    normalized = unicodedata2.normalize('NFKD', text)
    # Convert to lowercase using Unicode case mapping
    return normalized.lower()

comparison_pairs = [
    ("café", "cafe\u0301"),  # Composed vs decomposed
    ("Straße", "STRASSE"),   # German ß handling
    ("ﬁle", "file"),         # Ligature vs separate chars
]

print(f"\nText comparison after normalization:")
for text1, text2 in comparison_pairs:
    norm1 = normalize_for_comparison(text1)
    norm2 = normalize_for_comparison(text2)
    match = norm1 == norm2
    print(f"'{text1}' vs '{text2}': {match}")
    print(f"  Normalized: '{norm1}' vs '{norm2}'")
```

## Example 4: Character classification and script detection

```python
import unicodedata2
import re

# Example 4: Character classification and script detection
print("Character classification and script detection:")

# Test various character categories
test_chars = [
    ('A', 'Latin letter'),
    ('1', 'Digit'),
    (' ', 'Space'),
    ('+', 'Symbol'),
    ('α', 'Greek letter'),
    ('中', 'CJK character'),
    ('🦄', 'Emoji'),
    ('\u0300', 'Combining mark'),
    ('\u202A', 'Format character')
]

print("Character category analysis:")
for char, description in test_chars:
    category = unicodedata2.category(char)
    name = unicodedata2.name(char, 'NO NAME')
    east_asian_width = unicodedata2.east_asian_width(char)
    print(f"'{char}' ({description}):")
    print(f"  Category: {category}, Width: {east_asian_width}")
    print(f"  Name: {name}")

# Script and language detection patterns
def analyze_text_composition(text):
    """Analyze the composition of a text by character categories"""
    stats = {}
    for char in text:
        category = unicodedata2.category(char)
        stats[category] = stats.get(category, 0) + 1
    return stats

mixed_texts = [
    "Hello 世界! 🌍",
    "Café München François",
    "123ABC +-*/ αβγ",
    "E-mail: user@example.com"
]

print(f"\nText composition analysis:")
for text in mixed_texts:
    stats = analyze_text_composition(text)
    print(f"'{text}':")
    for category, count in sorted(stats.items()):
        print(f"  {category}: {count}")

# Combining character detection
def has_combining_marks(text):
    """Check if text contains combining marks"""
    return any(unicodedata2.category(char).startswith('M') for char in text)

combining_examples = ["café", "cafe\u0301", "a\u0300\u0301", "normal"]
print(f"\nCombining marks detection:")
for text in combining_examples:
    has_combining = has_combining_marks(text)
    print(f"'{text}': {'Has' if has_combining else 'No'} combining marks")
```

## Example 5: Production patterns for Unicode text processing

```python
import unicodedata2
import re

# Example 5: Production patterns for Unicode text processing
class UnicodeTextProcessor:
    """Production-ready Unicode text processing utility"""
    
    def __init__(self):
        self.stats = {"processed": 0, "normalized": 0, "issues_found": 0}
    
    def sanitize_text(self, text):
        """Sanitize text for storage and processing"""
        # Normalize to NFC (composed form)
        normalized = unicodedata2.normalize('NFC', text)
        
        # Remove control characters except common whitespace
        sanitized = ''.join(char for char in normalized 
                          if unicodedata2.category(char) not in ['Cc', 'Cf'] 
                          or char in ['\n', '\t', '\r'])
        
        self.stats["processed"] += 1
        if sanitized != text:
            self.stats["normalized"] += 1
        
        return sanitized
    
    def validate_identifier(self, identifier):
        """Validate Unicode identifier according to Python rules"""
        if not identifier:
            return False, "Empty identifier"
        
        # Check first character
        first_char = identifier[0]
        if not (unicodedata2.category(first_char) in ['Lu', 'Ll', 'Lt', 'Lm', 'Lo', 'Nl'] 
                or first_char == '_'):
            return False, f"Invalid start character: '{first_char}'"
        
        # Check remaining characters
        for i, char in enumerate(identifier[1:], 1):
            cat = unicodedata2.category(char)
            if not (cat in ['Lu', 'Ll', 'Lt', 'Lm', 'Lo', 'Nl', 'Mn', 'Mc', 'Nd', 'Pc'] 
                    or char == '_'):
                return False, f"Invalid character at position {i}: '{char}'"
        
        return True, "Valid identifier"
    
    def analyze_text_encoding(self, text):
        """Analyze text for encoding and normalization issues"""
        analysis = {
            "length": len(text),
            "normalized_length": len(unicodedata2.normalize('NFC', text)),
            "categories": {},
            "normalization_needed": False,
            "bidi_text": False
        }
        
        for char in text:
            category = unicodedata2.category(char)
            analysis["categories"][category] = analysis["categories"].get(category, 0) + 1
            
            # Check for bidirectional text
            if unicodedata2.bidirectional(char) in ['R', 'AL']:
                analysis["bidi_text"] = True
        
        # Check if normalization changes the text
        nfc_text = unicodedata2.normalize('NFC', text)
        analysis["normalization_needed"] = (text != nfc_text)
        
        return analysis

print("Production Unicode text processing:")

# Create text processor
processor = UnicodeTextProcessor()

# Test text sanitization
test_texts = [
    "Normal text",
    "Text with\u0000control\u001fcharacters",
    "Café with é vs e\u0301",
    "Mixed scripts: Hello 世界 مرحبا",
    "Emoji test: 👋🌍🚀"
]

print("Text sanitization:")
for text in test_texts:
    sanitized = processor.sanitize_text(text)
    changed = "✓" if text != sanitized else "○"
    print(f"{changed} '{text[:20]}...' → '{sanitized[:20]}...'")

# Test identifier validation
identifiers = [
    "valid_name",
    "ValidName",
    "name123",
    "変数名",  # Japanese variable name
    "123invalid",
    "my-name",
    "_private",
    "κόσμος"  # Greek word
]

print(f"\nIdentifier validation:")
for identifier in identifiers:
    valid, message = processor.validate_identifier(identifier)
    status = "✓" if valid else "✗"
    print(f"{status} '{identifier}': {message}")

# Text encoding analysis
analysis_texts = [
    "Simple ASCII text",
    "Mixed café naïve text",
    "Arabic: مرحبا بالعالم",
    "Emoji: 🦄🌈✨"
]

print(f"\nText encoding analysis:")
for text in analysis_texts:
    analysis = processor.analyze_text_encoding(text)
    print(f"'{text[:20]}...':")
    print(f"  Length: {analysis['length']} chars")
    print(f"  Needs normalization: {analysis['normalization_needed']}")
    print(f"  Contains bidirectional text: {analysis['bidi_text']}")
    
print(f"\nProcessing statistics:")
print(f"Texts processed: {processor.stats['processed']}")
print(f"Texts normalized: {processor.stats['normalized']}")

print(f"\nProduction features:")
print("- Text sanitization and normalization")
print("- Unicode identifier validation")
print("- Encoding analysis and detection")
print(f"- Unicode {unicodedata2.unidata_version} data")
```
