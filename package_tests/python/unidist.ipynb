## Example 1: Remote Function Execution
Demonstrates using @unidist.remote decorator for distributed execution

```python
import unidist

def demonstrate_remote_execution():
    """unidist.remote decorator enables distributed function execution"""
    unidist.init()
    
    # Define remote function
    @unidist.remote
    def compute_square(n):
        return n * n
    
    # Execute remotely and get reference
    result_ref = compute_square.remote(5)
    result = unidist.get(result_ref)
    print(f"Remote compute_square(5) = {result}")
    
    # Execute multiple tasks
    refs = [compute_square.remote(i) for i in range(5)]
    results = unidist.get(refs)
    print(f"Batch results: {results}")
    unidist.shutdown()
    print("\nunidist enables distributed function execution")

demonstrate_remote_execution()
```

## Example 2: Object Store with put/get
Demonstrates unidist's distributed object store

```python
import unidist

def demonstrate_object_store():
    """unidist.put and unidist.get provide distributed object storage"""
    
    unidist.init()
    
    # Put objects in distributed store
    data = {"name": "test", "values": [1, 2, 3, 4, 5]}
    data_ref = unidist.put(data)
    print(f"Stored object reference: {data_ref}")
    
    # Retrieve object
    retrieved = unidist.get(data_ref)
    print(f"Retrieved: {retrieved}")
    
    # Store multiple objects
    refs = [unidist.put(i * 10) for i in range(5)]
    values = unidist.get(refs)
    print(f"Multiple objects: {values}")
    
    unidist.shutdown()
    print("\nunidist provides distributed object storage")

demonstrate_object_store()
```

## Example 3: Cluster Resource Inspection
Demonstrates unidist.cluster_resources() for monitoring available resources

```python
import unidist

def demonstrate_cluster_resources():
    """unidist.cluster_resources() provides information about available resources"""
    unidist.init()
    
    # Get cluster resource information
    resources = unidist.cluster_resources()
    print(f"Cluster resources: {resources}")
    
    # Get number of CPUs available
    num_cpus = unidist.num_cpus()
    print(f"Available CPUs: {num_cpus}")
    
    # Check if unidist is initialized
    print(f"Is initialized: {unidist.is_initialized()}")
    
    # Get IP address of current node
    ip = unidist.get_ip()
    print(f"Node IP: {ip}")
    
    unidist.shutdown()
    print("\nunidist provides cluster resource monitoring")

demonstrate_cluster_resources()
```

## Example 4: Backend Configuration
Demonstrates checking and configuring unidist backends

```python
import unidist

def demonstrate_backend_config():
    """unidist supports multiple backends (Ray, Dask, MPI, Python)"""
    
    # Check current backend before init
    print(f"Default backend: {unidist.config.Backend.get()}")
    
    # Initialize with default backend
    unidist.init()
    
    print(f"Active backend: {unidist.config.Backend.get()}")
    
    # Test simple operation
    @unidist.remote
    def hello():
        return "Hello from unidist!"
    
    result = unidist.get(hello.remote())
    print(f"Result: {result}")
    
    unidist.shutdown()
    print("\nunidist provides backend-agnostic distributed computing")

demonstrate_backend_config()
```

## Example 5: Waiting for Specific Results
Demonstrates unidist.wait() for selective result retrieval

```python
import unidist
import time

def demonstrate_selective_waiting():
    """unidist.wait() enables selective result retrieval from multiple tasks"""
    unidist.init()
    
    # Define tasks with different execution times
    @unidist.remote
    def task(duration, value):
        time.sleep(duration)
        return value
    
    # Launch multiple tasks
    refs = [task.remote(0.1 * i, f"result_{i}") for i in range(1, 4)]
    print(f"Launched {len(refs)} tasks")
    
    # Wait for first task to complete
    ready, not_ready = unidist.wait(refs, num_returns=1)
    print(f"\nFirst task ready: {len(ready)}, Still running: {len(not_ready)}")
    print(f"First result: {unidist.get(list(ready)[0])}")
    unidist.shutdown()
    print("\nunidist.wait() enables selective result retrieval")

demonstrate_selective_waiting()
```
