## Example 1: Basic UPath operations with local filesystem

```python
from upath import UPath
import tempfile
import os

# Create a temporary directory for testing
with tempfile.TemporaryDirectory() as tmpdir:
    # Create UPath objects for local filesystem
    base_path = UPath(tmpdir)
    file_path = base_path / "test_file.txt"
    
    # Write content using UPath
    file_path.write_text("Hello from universal-pathlib!")
    
    # Read and verify
    content = file_path.read_text()
    print(f"Content: {content}")
    print(f"File exists: {file_path.exists()}")
    print(f"File name: {file_path.name}")
    print(f"File stem: {file_path.stem}")
    print(f"File suffix: {file_path.suffix}")
    print(f"Parent: {file_path.parent}")
    print(f"Is absolute: {file_path.is_absolute()}")
    
    # Create subdirectories
    subdir = base_path / "subdir" / "nested"
    subdir.mkdir(parents=True, exist_ok=True)
    print(f"\nCreated directory: {subdir}")
    print(f"Directory exists: {subdir.exists()}")
```

## Example 2: Memory filesystem operations using UPath

```python
from upath import UPath

# Use memory filesystem (fsspec's MemoryFileSystem)
mem_path = UPath("memory://data/example.json")

# Create parent directory
mem_path.parent.mkdir(parents=True, exist_ok=True)

# Write JSON-like content to memory
json_content = '{"name": "test", "value": 42}'
mem_path.write_text(json_content)

# Read back from memory
content = mem_path.read_text()
print(f"Content from memory: {content}")
print(f"Path protocol: {mem_path.protocol}")
print(f"Path exists: {mem_path.exists()}")

# List files in memory directory
mem_dir = UPath("memory://data")
print(f"\nFiles in memory directory:")
for item in mem_dir.iterdir():
    print(f"  - {item.name}")
```

## Example 3: Path manipulation and globbing with UPath

```python
from upath import UPath
import tempfile

# Create test structure
with tempfile.TemporaryDirectory() as tmpdir:
    base = UPath(tmpdir)
    
    # Create multiple files with different extensions
    (base / "file1.txt").write_text("Text file 1")
    (base / "file2.txt").write_text("Text file 2")
    (base / "data.json").write_text('{"key": "value"}')
    (base / "script.py").write_text("print('hello')")
    
    # Create subdirectory with files
    subdir = base / "subdir"
    subdir.mkdir()
    (subdir / "nested.txt").write_text("Nested file")
    
    # Glob patterns
    print("All .txt files:")
    for txt_file in base.glob("*.txt"):
        print(f"  - {txt_file.name}")
    
    print("\nAll files recursively:")
    for file in base.rglob("*.*"):
        print(f"  - {file.relative_to(base)}")
    
    # Path operations
    file_path = base / "file1.txt"
    print(f"\nPath parts: {file_path.parts}")
    print(f"With different suffix: {file_path.with_suffix('.md')}")
    print(f"With different name: {file_path.with_name('newfile.txt')}")
```

## Example 4: File metadata and stat operations with UPath

```python
from upath import UPath
import tempfile
import time

with tempfile.TemporaryDirectory() as tmpdir:
    base = UPath(tmpdir)
    
    # Create test files
    file1 = base / "test.txt"
    file1.write_text("Sample content for testing")
    
    # Get file statistics
    stat_info = file1.stat()
    print(f"File size: {stat_info.st_size} bytes")
    print(f"Modified time: {stat_info.st_mtime}")
    
    # Check file type
    print(f"\nIs file: {file1.is_file()}")
    print(f"Is directory: {file1.is_dir()}")
    print(f"Is directory (parent): {file1.parent.is_dir()}")
    
    # Read/write bytes
    binary_data = b"\x00\x01\x02\x03\x04"
    binary_file = base / "data.bin"
    binary_file.write_bytes(binary_data)
    
    read_data = binary_file.read_bytes()
    print(f"\nBinary data written and read: {read_data.hex()}")
    print(f"Data matches: {read_data == binary_data}")
    
    # Absolute path operations
    print(f"\nAbsolute path: {file1.absolute()}")
    print(f"Resolve path: {file1.resolve()}")
```

## Example 5: Advanced UPath features - joining paths and protocol detection

```python
from upath import UPath
import tempfile

# Demonstrate protocol detection and path joining
print("Protocol detection:")
local_path = UPath("/tmp/file.txt")
mem_path = UPath("memory://bucket/file.txt")
s3_path = UPath("s3://bucket/path/to/file.txt")

print(f"Local protocol: {local_path.protocol}")
print(f"Memory protocol: {mem_path.protocol}")
print(f"S3 protocol: {s3_path.protocol}")

# Path joining and manipulation
print("\nPath joining:")
base = UPath("memory://data")
full_path = base / "subfolder" / "file.txt"
print(f"Joined path: {full_path}")
print(f"As URI: {full_path.as_uri()}")

# Test actual operations with memory filesystem
with tempfile.TemporaryDirectory() as tmpdir:
    # Create a complex directory structure
    base = UPath(tmpdir)
    
    # Create nested structure
    for i in range(3):
        dir_path = base / f"dir{i}"
        dir_path.mkdir(exist_ok=True)
        for j in range(2):
            file_path = dir_path / f"file{j}.txt"
            file_path.write_text(f"Content {i}-{j}")
    
    # Walk through directory
    print("\nDirectory structure:")
    all_files = list(base.rglob("*.txt"))
    print(f"Total files found: {len(all_files)}")
    
    for file in sorted(all_files):
        rel_path = file.relative_to(base)
        content = file.read_text()
        print(f"  {rel_path}: {content}")
```
