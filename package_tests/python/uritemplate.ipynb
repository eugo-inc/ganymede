## Example 1: Basic URI template expansion

```python
import uritemplate

# Example 1: Basic URI template expansion
print("Basic URI template expansion:")

# Simple URI template
template = uritemplate.URITemplate('https://api.example.com/users/{user_id}')
uri = template.expand(user_id='12345')
print(f"Template: {template.uri}")
print(f"Expanded: {uri}")

# Multiple variables
api_template = uritemplate.URITemplate('https://api.example.com/{version}/users/{user_id}/posts{?limit,offset}')
api_uri = api_template.expand(
    version='v2',
    user_id='1001',
    limit=10,
    offset=20
)
print(f"\nMulti-variable template:")
print(f"Template: {api_template.uri}")
print(f"Expanded: {api_uri}")

# Path segments
path_template = uritemplate.URITemplate('/api/{category}{/subcategory*}')
path_uri = path_template.expand(
    category='products',
    subcategory=['electronics', 'phones', 'smartphones']
)
print(f"\nPath segments:")
print(f"Template: {path_template.uri}")
print(f"Expanded: {path_uri}")

# Show available variables in template
variables = api_template.variables
print(f"\nTemplate variables: {list(variables)}")
print("URI templates provide flexible URL generation")

# The expansions follow RFC 6570.
assert uri == 'https://api.example.com/users/12345', uri
assert api_uri == 'https://api.example.com/v2/users/1001/posts?limit=10&offset=20', api_uri
assert path_uri == '/api/products/electronics/phones/smartphones', path_uri
```

## Example 2: Query string expansion and form-style parameters

```python
import uritemplate

# Example 2: Query string and form-style parameters (unique to uritemplate)
print("Query string expansion with different operators:")

# Query string expansion with ?
query_template = uritemplate.URITemplate('https://api.github.com/search/repositories{?q,sort,order}')
query_uri = query_template.expand(
    q='python template',
    sort='stars',
    order='desc'
)
print(f"Query template: {query_template.uri}")
print(f"Expanded: {query_uri}")

# Query continuation with &
continuation_template = uritemplate.URITemplate('https://api.example.com/items?type=all{&page,per_page,filter}')
continuation_uri = continuation_template.expand(
    page=2,
    per_page=50,
    filter='active'
)
print(f"\nQuery continuation: {continuation_uri}")

# Form-style expansion with ;
form_template = uritemplate.URITemplate('/path{;x,y,z}')
form_uri = form_template.expand(x=10, y=20, z=30)
print(f"\nForm-style (;): {form_uri}")

# Fragment expansion with #
fragment_template = uritemplate.URITemplate('https://docs.example.com/page{#section}')
fragment_uri = fragment_template.expand(section='advanced-usage')
print(f"\nFragment expansion: {fragment_uri}")

# The expansions follow RFC 6570.
assert query_uri == 'https://api.github.com/search/repositories?q=python%20template&sort=stars&order=desc', query_uri
assert continuation_uri == 'https://api.example.com/items?type=all&page=2&per_page=50&filter=active', continuation_uri
assert form_uri == '/path;x=10;y=20;z=30', form_uri
assert fragment_uri == 'https://docs.example.com/page#advanced-usage', fragment_uri
```

## Example 3: List and dictionary expansion with explode modifier

```python
import uritemplate

# Example 3: List and dictionary expansion with explode modifier (*)
print("List expansion with explode modifier:")

# List expansion without explode
list_template = uritemplate.URITemplate('/search{?tags}')
list_uri = list_template.expand(tags=['python', 'api', 'rest'])
print(f"Without explode: {list_uri}")

# List expansion with explode modifier *
explode_template = uritemplate.URITemplate('/search{?tags*}')
explode_uri = explode_template.expand(tags=['python', 'api', 'rest'])
print(f"With explode (*): {explode_uri}")

# Dictionary/mapping expansion
print("\nDictionary expansion:")
dict_template = uritemplate.URITemplate('/filter{?params*}')
dict_uri = dict_template.expand(
    params={'category': 'books', 'price_min': 10, 'price_max': 50}
)
print(f"Dictionary explode: {dict_uri}")

# Path-style list expansion
path_list_template = uritemplate.URITemplate('/api/{path*}')
path_list_uri = path_list_template.expand(path=['v1', 'users', 'profile'])
print(f"\nPath segments: {path_list_uri}")

# Mixed expansion
mixed_template = uritemplate.URITemplate('https://api.example.com/{version}/search{?q,filters*}')
mixed_uri = mixed_template.expand(
    version='v2',
    q='python',
    filters={'lang': 'en', 'sort': 'date'}
)
print(f"\nMixed expansion: {mixed_uri}")

# The expansions follow RFC 6570.
assert list_uri == '/search?tags=python,api,rest', list_uri
assert explode_uri == '/search?tags=python&tags=api&tags=rest', explode_uri
assert path_list_uri == '/api/v1,users,profile', path_list_uri
```

## Example 4: Partial expansion and the expand() convenience function

```python
import uritemplate

# Example 4: Partial expansion and convenience functions
print("Partial expansion (unique feature):")

# Create a complex template
template = uritemplate.URITemplate('https://api.example.com/{version}/users/{user_id}/posts{?limit,offset,sort}')

# Partial expansion - expand only some variables
partial_result = template.partial(version='v2', limit=20)
print(f"Original template: {template.uri}")
print(f"After partial expansion: {partial_result.uri}")

# Continue expansion with remaining variables
final_uri = partial_result.expand(user_id='alice', offset=40, sort='date')
print(f"Final expanded URI: {final_uri}")

# Using the convenience expand() function
print("\nConvenience expand() function:")
simple_template = 'https://api.github.com/repos/{owner}/{repo}/issues{/issue_id}'
expanded_url = uritemplate.expand(simple_template, owner='python', repo='cpython', issue_id='12345')
print(f"Template: {simple_template}")
print(f"Expanded: {expanded_url}")

# Multiple partial expansions in sequence
print("\nChained partial expansions:")
base = uritemplate.URITemplate('/api/{version}/{resource}{/id}{?action,format}')
step1 = base.partial(version='v3')
step2 = step1.partial(resource='articles')
step3 = step2.partial(format='json')
final = step3.expand(id='42', action='view')
print(f"Final URI: {final}")

# The convenience function gives the same result as URITemplate(...).expand(...).
assert expanded_url == 'https://api.github.com/repos/python/cpython/issues/12345', expanded_url
```

## Example 5: Template variable inspection and reserved character expansion

```python
import uritemplate

# Example 5: Variable inspection and reserved character expansion
print("Template variable inspection:")

# Create a template
template = uritemplate.URITemplate('https://api.example.com/{version}/search{?query,limit,filters*}{#section}')

# Inspect variables in the template
variables = template.variables
print(f"Template: {template.uri}")
print(f"Variables found: {[str(v) for v in variables]}")
print(f"Number of variables: {len(variables)}")

# Reserved character expansion with +
print("\nReserved character expansion (+):")
reserved_template = uritemplate.URITemplate('/search{+path}')
reserved_uri = reserved_template.expand(path='/docs/guide/index.html')
print(f"With + operator: {reserved_uri}")

# Compare with regular expansion
regular_template = uritemplate.URITemplate('/search{path}')
regular_uri = regular_template.expand(path='/docs/guide/index.html')
print(f"Without + operator: {regular_uri}")

# Special characters in values
special_template = uritemplate.URITemplate('https://example.com/search{?q}')
special_uri = special_template.expand(q='hello world & special=chars')
print(f"\nURL encoding: {special_uri}")

# Reserved expansion with special characters
reserved_special = uritemplate.URITemplate('https://example.com{+path}')
reserved_special_uri = reserved_special.expand(path='/search?q=test&sort=date')
print(f"Reserved expansion: {reserved_special_uri}")

# Demonstrate variable_names property (if available)
print(f"\nTemplate string: {template.uri}")
print(f"Variable objects type: {type(variables)}")
for var in variables:
    print(f"  Variable: {var}")

# '+' keeps reserved characters; simple expansion percent-encodes them.
assert reserved_uri == '/search/docs/guide/index.html', reserved_uri
assert regular_uri == '/search%2Fdocs%2Fguide%2Findex.html', regular_uri
assert special_uri == 'https://example.com/search?q=hello%20world%20%26%20special%3Dchars', special_uri
```
