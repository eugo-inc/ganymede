```python
# Example 1: top-level url_normalize() on a plain URL adds a scheme + lowercases host.
from url_normalize import url_normalize

raw = "Example.com/Foo/Bar"
normalized = url_normalize(raw)
print(f"input:      {raw!r}")
print(f"normalized: {normalized!r}")

assert normalized.startswith("https://") or normalized.startswith("http://")
assert "example.com" in normalized.lower()
```

```python
# Example 2: default scheme handling + an https URL passes through cleanly.
from url_normalize import url_normalize

cases = [
    "https://example.com",
    "HTTP://Example.COM/Path",
    "example.org/some/path?b=2&a=1",
]
results = {raw: url_normalize(raw) for raw in cases}
for raw, out in results.items():
    print(f"{raw!r:50s} -> {out!r}")

assert results["https://example.com"].startswith("https://")
assert "example.com" in results["HTTP://Example.COM/Path"].lower()
```

```python
# Example 3: default_scheme keyword controls the implied scheme.
from url_normalize import url_normalize

raw = "example.com/path"
https_default = url_normalize(raw)
http_default = url_normalize(raw, default_scheme="http")

print(f"default scheme (https): {https_default!r}")
print(f"default_scheme='http':  {http_default!r}")

assert https_default.startswith("https://")
assert http_default.startswith("http://")
```

```python
# Example 4: IDN (internationalized domain name) gets punycoded via idna.
from url_normalize import url_normalize

raw = "https://\u4e2d\u56fd.icom.museum/"
normalized = url_normalize(raw)
print(f"input (IDN): {raw!r}")
print(f"normalized:  {normalized!r}")

# Punycode-encoded IDN labels begin with the 'xn--' ACE prefix.
assert "xn--" in normalized
```

```python
# Example 5: idempotence — normalizing twice yields the same result.
from url_normalize import url_normalize

samples = [
    "http://Example.com:80/A/./B/../C/",
    "https://example.com/?z=1&a=2",
    "example.net/path%7euser",
    "ftp://Example.com/file.txt",
]
for raw in samples:
    once = url_normalize(raw)
    twice = url_normalize(once)
    print(f"{raw!r:45s} -> {once!r}")
    assert once == twice, f"not idempotent: {once!r} != {twice!r}"

print("\nidempotence holds across all samples")
```
