```python
# Example 1: Basic uvloop Installation and Event Loop Creation
import uvloop
import time

# Install uvloop as the default event loop policy
uvloop.install()

print("uvloop Installation and Configuration:")
print(f"uvloop version: {uvloop.__version__}")

# Create a new uvloop event loop directly
loop = uvloop.new_event_loop()
print(f"\nEvent loop created:")
print(f"  Type: {type(loop).__name__}")
print(f"  Module: {loop.__class__.__module__}.{loop.__class__.__name__}")

# Demonstrate uvloop loop properties
print(f"\nuvloop Event Loop Properties:")
print(f"  Is running: {loop.is_running()}")
print(f"  Is closed: {loop.is_closed()}")
print(f"  Loop time: {loop.time():.6f}")
print(f"  Debug mode: {loop.get_debug()}")

# uvloop-specific features
print(f"\nuvloop Features:")
print(f"  Supports set_debug: {hasattr(loop, 'set_debug')}")
print(f"  Supports set_task_factory: {hasattr(loop, 'set_task_factory')}")
print(f"  Supports set_exception_handler: {hasattr(loop, 'set_exception_handler')}")

# Enable debug mode on uvloop
loop.set_debug(True)
print(f"  Debug mode enabled: {loop.get_debug()}")

# Disable debug mode
loop.set_debug(False)
print(f"  Debug mode disabled: {loop.get_debug()}")

# Close the loop
loop.close()
print(f"  Loop closed: {loop.is_closed()}")

# Create another loop to demonstrate
loop2 = uvloop.new_event_loop()
print(f"\nSecond uvloop created: {type(loop2).__name__}")
print(f"  Time: {loop2.time():.6f}")
loop2.close()

print("\n✅ Basic uvloop installation and event loop creation complete")
```

```python
# Example 2: uvloop Event Loop Policy Configuration
import uvloop
import sys

# Demonstrate event loop policy management
print("uvloop Event Loop Policy Configuration:")

# Create uvloop policy
policy = uvloop.EventLoopPolicy()
print(f"uvloop policy created: {type(policy).__name__}")
print(f"  Module: {policy.__class__.__module__}")

# Get a new event loop from the policy
loop = policy.new_event_loop()
print(f"\nEvent loop from policy:")
print(f"  Type: {type(loop).__name__}")
print(f"  Is uvloop: {isinstance(loop, uvloop.Loop)}")

# Demonstrate policy methods
print(f"\nPolicy features:")
print(f"  Can get event loop: {hasattr(policy, 'get_event_loop')}")
print(f"  Can set event loop: {hasattr(policy, 'set_event_loop')}")
print(f"  Can create new loop: {hasattr(policy, 'new_event_loop')}")

# Set the policy as current
policy.set_event_loop(loop)
retrieved_loop = policy.get_event_loop()
print(f"\nLoop identity check:")
print(f"  Set loop matches retrieved: {retrieved_loop is loop}")

# Create multiple loops from policy
loops = []
for i in range(3):
    new_loop = policy.new_event_loop()
    loops.append(new_loop)
    print(f"  Loop {i}: {type(new_loop).__name__} (time: {new_loop.time():.6f})")

# Clean up
loop.close()
for l in loops:
    l.close()

# Install uvloop globally
uvloop.install()
print(f"\nuvloop installed globally")

# Verify installation by checking default policy
import asyncio
default_policy = asyncio.get_event_loop_policy()
print(f"Default policy type: {type(default_policy).__name__}")
print(f"Is uvloop policy: {isinstance(default_policy, uvloop.EventLoopPolicy)}")

# Create a loop using the installed policy
test_loop = uvloop.new_event_loop()
print(f"\nNew loop after install:")
print(f"  Type: {type(test_loop).__name__}")
print(f"  Module: {test_loop.__class__.__module__}")
test_loop.close()

print("\n✅ uvloop event loop policy configuration complete")
```

```python
# Example 3: uvloop Loop Properties and Introspection
import uvloop
import time

# Create uvloop event loop
uvloop.install()

print("uvloop Loop Properties and Introspection:")

# Create multiple loops to demonstrate properties
loop1 = uvloop.new_event_loop()
loop2 = uvloop.new_event_loop()

print(f"\nLoop 1 properties:")
print(f"  Type: {type(loop1).__name__}")
print(f"  Module: {loop1.__class__.__module__}")
print(f"  Is running: {loop1.is_running()}")
print(f"  Is closed: {loop1.is_closed()}")
print(f"  Debug mode: {loop1.get_debug()}")
print(f"  Time: {loop1.time():.6f}")

print(f"\nLoop 2 properties:")
print(f"  Type: {type(loop2).__name__}")
print(f"  Is running: {loop2.is_running()}")
print(f"  Is closed: {loop2.is_closed()}")
print(f"  Time: {loop2.time():.6f}")

# Demonstrate time tracking
print(f"\nTime tracking comparison:")
time1_before = loop1.time()
time2_before = loop2.time()

# Wait a bit
time.sleep(0.05)

time1_after = loop1.time()
time2_after = loop2.time()

print(f"  Loop 1 elapsed: {time1_after - time1_before:.6f}s")
print(f"  Loop 2 elapsed: {time2_after - time2_before:.6f}s")

# Demonstrate debug mode toggling
print(f"\nDebug mode configuration:")
print(f"  Loop 1 debug: {loop1.get_debug()}")

loop1.set_debug(True)
print(f"  Loop 1 debug enabled: {loop1.get_debug()}")

loop1.set_debug(False)
print(f"  Loop 1 debug disabled: {loop1.get_debug()}")

# Check key methods (avoid problematic attributes)
print(f"\nuvloop key methods:")
key_methods = ['call_soon', 'call_later', 'call_at', 'create_task', 'create_future', 
               'set_debug', 'get_debug', 'set_exception_handler', 'set_task_factory',
               'is_running', 'is_closed', 'time', 'close']

for method in key_methods:
    has_method = hasattr(loop1, method)
    print(f"  {method}: {has_method}")

# Demonstrate loop state management
print(f"\nLoop state management:")
print(f"  Loop 1 closed: {loop1.is_closed()}")

loop1.close()
print(f"  Loop 1 closed after close(): {loop1.is_closed()}")

print(f"  Loop 2 closed: {loop2.is_closed()}")
loop2.close()
print(f"  Loop 2 closed after close(): {loop2.is_closed()}")

# Create new loop for additional introspection
loop3 = uvloop.new_event_loop()

print(f"\nLoop introspection:")
print(f"  Has call_soon: {hasattr(loop3, 'call_soon')}")
print(f"  Has call_later: {hasattr(loop3, 'call_later')}")
print(f"  Has call_at: {hasattr(loop3, 'call_at')}")
print(f"  Has create_task: {hasattr(loop3, 'create_task')}")
print(f"  Has create_future: {hasattr(loop3, 'create_future')}")
print(f"  Has set_exception_handler: {hasattr(loop3, 'set_exception_handler')}")
print(f"  Has set_task_factory: {hasattr(loop3, 'set_task_factory')}")

loop3.close()

print("\n✅ uvloop loop properties and introspection complete")
```

```python
# Example 4: uvloop Futures and Handles
import uvloop
import time

# Create uvloop event loop
uvloop.install()
loop = uvloop.new_event_loop()

print("uvloop Futures and Handles:")
print(f"Event loop: {type(loop).__name__}")

# Create futures using uvloop
print(f"\nCreating futures:")

future1 = loop.create_future()
future2 = loop.create_future()
future3 = loop.create_future()

print(f"Future 1: {future1}")
print(f"  Done: {future1.done()}")
print(f"  Cancelled: {future1.cancelled()}")

print(f"Future 2: {future2}")
print(f"  Done: {future2.done()}")

# Set result on futures
future1.set_result("Result 1")
print(f"\nAfter setting result on Future 1:")
print(f"  Done: {future1.done()}")
print(f"  Result: {future1.result()}")

# Set exception on future
future2.set_exception(ValueError("Test exception"))
print(f"\nAfter setting exception on Future 2:")
print(f"  Done: {future2.done()}")
try:
    future2.result()
except ValueError as e:
    print(f"  Exception: {e}")

# Cancel future
future3.cancel()
print(f"\nAfter cancelling Future 3:")
print(f"  Cancelled: {future3.cancelled()}")
print(f"  Done: {future3.done()}")

# Demonstrate handle creation
print(f"\nCreating callback handles:")

handle_log = []

def test_callback(callback_id):
    handle_log.append(callback_id)
    print(f"  Callback {callback_id} would execute")

# Create handles (but don't run them since we can't run_forever in Jupyter)
handle1 = loop.call_soon(test_callback, 1)
handle2 = loop.call_later(0.01, test_callback, 2)
handle3 = loop.call_at(loop.time() + 0.02, test_callback, 3)

print(f"\nHandle 1 (call_soon): {handle1}")
print(f"  Cancelled: {handle1.cancelled()}")

print(f"Handle 2 (call_later): {handle2}")
print(f"  Cancelled: {handle2.cancelled()}")

print(f"Handle 3 (call_at): {handle3}")
print(f"  Cancelled: {handle3.cancelled()}")

# Cancel handles
handle1.cancel()
handle2.cancel()

print(f"\nAfter cancelling handles:")
print(f"  Handle 1 cancelled: {handle1.cancelled()}")
print(f"  Handle 2 cancelled: {handle2.cancelled()}")
print(f"  Handle 3 cancelled: {handle3.cancelled()}")

# Cancel remaining handle
handle3.cancel()
print(f"  Handle 3 cancelled: {handle3.cancelled()}")

# Clean up
loop.close()

# Create new loop for more future demonstrations
loop2 = uvloop.new_event_loop()

print(f"\nFuture chaining demonstration:")

# Create multiple futures
futures = []
for i in range(5):
    future = loop2.create_future()
    futures.append(future)
    print(f"  Future {i}: Created")

# Set results
for i, future in enumerate(futures):
    future.set_result(f"Result_{i}")

print(f"\nFuture results:")
for i, future in enumerate(futures):
    print(f"  Future {i}: {future.result() if future.done() else 'Not done'}")

loop2.close()

print("\n✅ uvloop futures and handles complete")
```

```python
# Example 5: uvloop Exception Handlers and Configuration
import uvloop
import time

# Create uvloop with exception handling
uvloop.install()
loop = uvloop.new_event_loop()

print("uvloop Exception Handlers and Configuration:")
print(f"Event loop: {type(loop).__name__}")

# Track exceptions
exceptions_caught = []

def custom_exception_handler(loop_ref, context):
    """Custom exception handler for uvloop"""
    exception = context.get('exception')
    message = context.get('message', 'No message')
    
    exceptions_caught.append({
        'exception': exception,
        'message': message,
        'context_keys': list(context.keys())
    })
    
    print(f"  Exception caught: {message}")
    if exception:
        print(f"    Type: {type(exception).__name__}")

# Set custom exception handler
loop.set_exception_handler(custom_exception_handler)
print(f"Custom exception handler set: {hasattr(loop, 'set_exception_handler')}")

# Test exception handler manually
print(f"\nTesting exception handler:")
loop.call_exception_handler({
    'message': 'Test exception 1',
    'exception': ValueError('Test error 1')
})

loop.call_exception_handler({
    'message': 'Test exception 2',
    'exception': RuntimeError('Test error 2')
})

print(f"Exceptions caught: {len(exceptions_caught)}")

# Debug mode configuration
print(f"\nDebug mode configuration:")
print(f"  Current debug mode: {loop.get_debug()}")

loop.set_debug(True)
print(f"  Debug enabled: {loop.get_debug()}")

# Test exception in debug mode
loop.call_exception_handler({
    'message': 'Debug mode exception',
    'exception': TypeError('Debug test error')
})

loop.set_debug(False)
print(f"  Debug disabled: {loop.get_debug()}")

print(f"Total exceptions: {len(exceptions_caught)}")

# Remove exception handler
loop.set_exception_handler(None)
print(f"\nException handler removed")

# Clean up
loop.close()

# Demonstrate policy-level configuration
print(f"\nEvent loop policy configuration:")

policy = uvloop.EventLoopPolicy()
print(f"Policy created: {type(policy).__name__}")

# Create loop from policy
policy_loop = policy.new_event_loop()
print(f"Loop from policy: {type(policy_loop).__name__}")

# Set as current loop in policy
policy.set_event_loop(policy_loop)
retrieved = policy.get_event_loop()

print(f"Loop set and retrieved: {retrieved is policy_loop}")

policy_loop.close()

# Install uvloop globally and verify
uvloop.install()

import asyncio
current_policy = asyncio.get_event_loop_policy()
print(f"\nGlobal policy after install:")
print(f"  Type: {type(current_policy).__name__}")
print(f"  Is uvloop: {isinstance(current_policy, uvloop.EventLoopPolicy)}")

# Create loop using installed policy
final_loop = uvloop.new_event_loop()
print(f"\nLoop from installed policy:")
print(f"  Type: {type(final_loop).__name__}")
print(f"  Module: {final_loop.__class__.__module__}")
print(f"  Debug: {final_loop.get_debug()}")
print(f"  Closed: {final_loop.is_closed()}")

# Configure final loop
final_loop.set_debug(True)
print(f"  Debug enabled: {final_loop.get_debug()}")

# Set exception handler on final loop
final_exceptions = []

def final_handler(loop_ref, context):
    final_exceptions.append(context.get('message', 'Unknown'))

final_loop.set_exception_handler(final_handler)

# Test handler
final_loop.call_exception_handler({'message': 'Final test'})
print(f"  Exceptions: {len(final_exceptions)}")

final_loop.close()

print("\n✅ uvloop exception handlers and configuration complete")
```
