```python
# Example 1: varname() retrieves the name of the variable a function call is assigned to.
from varname import varname

def make_thing():
    name = varname()
    return name

my_widget = make_thing()
print(f"caller assigned result to variable named: {my_widget!r}")
assert my_widget == 'my_widget'
```

```python
# Example 2: nameof() returns the literal name of the passed identifier.
from varname import nameof

alpha = 1
beta = 2
gamma = 3

n_alpha = nameof(alpha)
n_multi = nameof(alpha, beta, gamma)

print(f"nameof(alpha) = {n_alpha!r}")
print(f"nameof(alpha, beta, gamma) = {n_multi!r}")

assert n_alpha == 'alpha'
assert n_multi == ('alpha', 'beta', 'gamma')
```

```python
# Example 3: argname() retrieves the source-expression names of function arguments.
# vars_only=False also returns the source of full expressions, not just variables.
from varname import argname

def show_arg(x):
    return argname('x', vars_only=False)

value = 42
name = show_arg(value)
print(f"argname of the parameter we passed: {name!r}")
assert name == 'value'

# Works with expressions too.
expr_name = show_arg(value + 1)
print(f"argname for an expression: {expr_name!r}")
assert expr_name == 'value + 1'
```

```python
# Example 4: varname() with multi-target assignment via multi_vars=True.
from varname import varname

def make_pair():
    names = varname(multi_vars=True)
    return (names, names)

left, right = make_pair()
print(f"varname captured both LHS names: left={left}, right={right}")
assert left == right == ('left', 'right')
```

```python
# Example 5: varname.helpers.Wrapper attaches the variable name to a value.
from varname.helpers import Wrapper

foo = Wrapper(123)
bar = Wrapper('hello')

print(f"foo: name={foo.name!r}, value={foo.value!r}")
print(f"bar: name={bar.name!r}, value={bar.value!r}")

assert foo.name == 'foo'
assert foo.value == 123
assert bar.name == 'bar'
assert bar.value == 'hello'
```
