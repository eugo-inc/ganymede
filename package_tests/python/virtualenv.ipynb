# virtualenv: 5 Basic Usage Examples

```python
# Example 1: Exploring virtualenv (safe approach)
import virtualenv

print('virtualenv imported successfully')
attrs = [attr for attr in dir(virtualenv) if not attr.startswith('_')]
print(f'Available attributes: {attrs[:10]}...')

# Show version info
if hasattr(virtualenv, '__version__'):
    print(f'\nVirtualenv version: {virtualenv.__version__}')

# Test if we can access basic functionality safely
try:
    if hasattr(virtualenv, 'session_via_cli'):
        print('\nCLI session functionality available')
    
    if hasattr(virtualenv, 'create'):
        print('Create functionality available')
        
    # Don't actually create a virtualenv, just show it's possible
    print('\nNote: Virtualenv creation would require proper arguments and filesystem access')
    
except Exception as e:
    print(f'Virtualenv functionality note: {e}')
    raise

print('Virtualenv exploration complete')
```

```python
# Example 2: Create a virtual environment programmatically
import virtualenv
virtualenv.cli_run(['venv_test'])
import os
print(os.path.exists('venv_test'))
```

```python
# Example 3: Show help for CLI
import virtualenv
print(virtualenv.__doc__)
```

```python
# Example 4: interpreter discovery with PythonInfo.current()
import virtualenv
from virtualenv.discovery import py_info

# Get current Python interpreter info
current_python = py_info.PythonInfo.current()
print(f"Python executable: {current_python.executable}")
print(f"Python version: {current_python.version}")
print(f"Platform: {current_python.platform}")
print(f"Architecture: {current_python.architecture}")
```

```python
# Example 5: Remove the created virtual environment
import shutil
shutil.rmtree('venv_test', ignore_errors=True)
print('Removed venv_test')
```
