## Example 1: Basic Environment Information

````python
import watermark
from datetime import datetime

# Example 1: Basic environment information and system details
print("Basic watermark functionality:")

# Get basic system and Python information
basic_info = watermark.watermark()
print("Basic watermark:")
print(basic_info)

# Get Python-specific information
python_info = watermark.watermark(python=True)
print("\nPython information:")
print(python_info)

# Get machine and system information
machine_info = watermark.watermark(machine=True)
print("\nMachine information:")  
print(machine_info)

# Combine multiple options
combined_info = watermark.watermark(
    python=True,
    machine=True,
    iso8601=True
)
print("\nCombined information:")
print(combined_info)

print(f"Watermark version: {watermark.__version__}")
print("Watermark provides comprehensive environment documentation")
````

## Example 2: Package Version Tracking

````python
import watermark

# Example 2: Package version tracking and dependency information
print("Package version tracking:")

# Track specific packages
single_package = watermark.watermark(
    packages="watermark",
    python=True
)
print("Single package tracking:")
print(single_package)

# Track multiple packages (common scientific stack)
science_packages = watermark.watermark(
    packages="numpy,pandas,scipy,matplotlib",
    python=True
)
print("\nScience packages:")
print(science_packages)

# Track development packages
dev_packages = watermark.watermark(
    packages="setuptools,pip,wheel",
    python=True
)
print("\nDevelopment packages:")
print(dev_packages)

# Try to track packages that might not exist
try:
    missing_packages = watermark.watermark(
        packages="nonexistent_package,another_missing",
        python=True
    )
    print("\nMissing packages test:")
    print(missing_packages)
except Exception as e:
    print(f"\nHandling missing packages: {type(e).__name__}")
    raise

# Track packages with machine info for complete environment snapshot
complete_snapshot = watermark.watermark(
    packages="watermark,setuptools",
    python=True,
    machine=True,
    iso8601=True
)
print("\nComplete environment snapshot:")
print(complete_snapshot)

print("Package tracking enables reproducible environment documentation")
````

## Example 3: Customized Output Formatting

````python
import watermark

# Example 3: Customized output formatting and options
print("Customized watermark formatting:")

# Minimal output - just Python version
minimal = watermark.watermark(python=True)
print("Minimal (Python only):")
print(minimal)

# Machine details only
machine_only = watermark.watermark(machine=True)
print("\nMachine details only:")
print(machine_only)

# ISO8601 timestamp format
iso_timestamp = watermark.watermark(iso8601=True)
print("\nISO8601 timestamp:")
print(iso_timestamp)

# Custom author and date information
try:
    # Some versions support custom author
    custom_info = watermark.watermark(
        python=True,
        machine=True,
        author="Data Science Team",
        iso8601=True
    )
    print("\nWith custom author:")
    print(custom_info)
except TypeError:
    # Fallback if author parameter not supported
    print("\nCustom author not supported in this version")

# Comprehensive documentation format
comprehensive = watermark.watermark(
    python=True,
    machine=True,
    packages="watermark,setuptools,pip",
    iso8601=True
)
print("\nComprehensive documentation:")
print(comprehensive)

# Test different package combinations
package_sets = [
    ("core", "setuptools,pip,wheel"),
    ("web", "requests,urllib3"),
    ("dev", "pytest,black,flake8")
]

print(f"\nDifferent package sets:")
for name, packages in package_sets:
    try:
        result = watermark.watermark(
            packages=packages,
            python=True
        )
        print(f"\n{name.upper()} packages:")
        # Extract just the package lines
        lines = result.strip().split('\n')
        package_lines = [line for line in lines if ':' in line and not line.startswith('Python')]
        for line in package_lines:
            print(f"  {line}")
    except Exception as e:
        print(f"{name} packages: Some packages not available")
        raise

print("\nWatermark supports flexible output formatting for different use cases")
````

## Example 4: IPython Magic Integration

````python
import watermark
import os
import platform

# Example 4: IPython magic command integration
print("IPython magic command integration:")

# Show how watermark works as IPython magic
print("Watermark as IPython magic command:")
print("Usage in Jupyter/IPython:")
print("  %watermark")
print("  %watermark -p numpy,pandas,matplotlib")
print("  %watermark -v -m -t")
print("  %watermark --iso8601 --python --machine")

# Demonstrate programmatic access to watermark components
print("\nProgrammatic access to watermark information:")

# Show available magic class and methods
if hasattr(watermark, 'WaterMark'):
    print("WaterMark magic class available")
    print("Magic methods for Jupyter integration")

# Environment variable integration
print("\nEnvironment-aware watermarking:")

# Simulate different environments
environments = [
    ("development", {"DEBUG": "true", "ENV": "dev"}),
    ("production", {"ENV": "prod", "LOGGING": "warning"}),
    ("testing", {"ENV": "test", "CI": "true"})
]

for env_name, env_vars in environments:
    print(f"\n{env_name.upper()} environment:")
    
    # Temporarily set environment variables
    original_env = {}
    for key, value in env_vars.items():
        original_env[key] = os.environ.get(key)
        os.environ[key] = value
    
    # Create environment-specific watermark
    env_watermark = watermark.watermark(
        python=True,
        packages="watermark"
    )
    
    # Add environment info manually
    print(f"Environment: {env_name}")
    for key, value in env_vars.items():
        print(f"{key}: {value}")
    print(env_watermark)
    
    # Restore environment
    for key, original_value in original_env.items():
        if original_value is None:
            os.environ.pop(key, None)
        else:
            os.environ[key] = original_value

# Cross-platform compatibility check
print(f"\nCross-platform compatibility:")
system_info = {
    "Platform": platform.platform(),
    "System": platform.system(),
    "Architecture": platform.architecture()[0],
    "Python": platform.python_version()
}

for key, value in system_info.items():
    print(f"{key}: {value}")

print("\nWatermark adapts to different environments and platforms")
print("Ideal for Jupyter notebooks and reproducible research")
````

## Example 5: Production Documentation Patterns

````python
import watermark
import json
from datetime import datetime

# Example 5: Production patterns for reproducible documentation
class EnvironmentDocumenter:
    """Production-ready environment documentation using watermark"""
    
    def __init__(self, project_name="DataScience Project"):
        self.project_name = project_name
        self.documentation_format = "markdown"
    
    def generate_environment_report(self, packages=None, format_type="text"):
        """Generate comprehensive environment report"""
        package_list = packages or "numpy,pandas,scipy,matplotlib,seaborn,scikit-learn"
        
        if format_type == "json":
            return self._generate_json_report(package_list)
        else:
            return self._generate_text_report(package_list)
    
    def _generate_text_report(self, package_list):
        """Generate text-based environment report"""
        report = f"# {self.project_name} - Environment Report\n\n"
        
        # Get comprehensive watermark
        try:
            watermark_info = watermark.watermark(
                python=True,
                machine=True,
                packages=package_list,
                iso8601=True
            )
            
            report += "## Environment Details\n\n"
            report += "```\n"
            report += watermark_info
            report += "\n```\n\n"
            
        except Exception as e:
            report += f"Error generating watermark: {e}\n\n"
        
        # Add reproducibility section
        report += "## Reproducibility Information\n\n"
        report += "This environment report was generated to ensure reproducible results.\n"
        report += "Install the exact package versions listed above to replicate this environment.\n\n"
        
        return report
    
    def _generate_json_report(self, package_list):
        """Generate JSON-formatted environment report"""
        try:
            # Get watermark info
            watermark_info = watermark.watermark(
                python=True,
                machine=True,
                packages=package_list,
                iso8601=True
            )
            
            # Parse watermark info into structured data
            lines = watermark_info.strip().split('\n')
            
            report = {
                "project": self.project_name,
                "generated_at": datetime.now().isoformat(),
                "environment": {},
                "packages": {},
                "system": {}
            }
            
            current_section = "environment"
            for line in lines:
                line = line.strip()
                if not line:
                    continue
                
                if ':' in line:
                    key, value = line.split(':', 1)
                    key = key.strip()
                    value = value.strip()
                    
                    if "version" in key.lower():
                        report["packages"][key] = value
                    elif any(sys_key in key.lower() for sys_key in ["os", "machine", "processor", "compiler"]):
                        report["system"][key] = value
                    else:
                        report["environment"][key] = value
            
            return json.dumps(report, indent=2)
            
        except Exception as e:
            return json.dumps({"error": str(e)}, indent=2)
    
    def save_environment_snapshot(self, filename=None, packages=None):
        """Save environment snapshot to file"""
        if filename is None:
            timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
            filename = f"environment_snapshot_{timestamp}.md"
        
        report = self.generate_environment_report(packages)
        
        # Simulate file writing (in real usage, would write to file)
        print(f"Environment snapshot saved to: {filename}")
        print(f"Report size: {len(report)} characters")
        
        return filename
    
    def compare_environments(self, env1_packages, env2_packages):
        """Compare two environment configurations"""
        print("Environment comparison:")
        
        # Generate reports for comparison
        env1_report = self.generate_environment_report(env1_packages)
        env2_report = self.generate_environment_report(env2_packages)
        
        # Simple comparison (in production, would be more sophisticated)
        env1_lines = set(env1_report.split('\n'))
        env2_lines = set(env2_report.split('\n'))
        
        differences = env1_lines.symmetric_difference(env2_lines)
        
        if differences:
            print(f"Found {len(differences)} differences between environments")
            for diff in sorted(differences)[:5]:  # Show first 5 differences
                if diff.strip():
                    print(f"  {diff}")
        else:
            print("Environments are identical")

print("Production environment documentation patterns:")

# Create documenter
documenter = EnvironmentDocumenter("ML Research Project")

# Generate different types of reports
print("1. Text-based environment report:")
text_report = documenter.generate_environment_report(
    packages="watermark,numpy,pandas",
    format_type="text"
)
print(text_report[:500] + "..." if len(text_report) > 500 else text_report)

print("\n2. JSON-formatted environment report:")
json_report = documenter.generate_environment_report(
    packages="watermark,setuptools",
    format_type="json"
)
print(json_report)

print("\n3. Environment snapshot management:")
snapshot_file = documenter.save_environment_snapshot(
    packages="watermark,pip,setuptools"
)

print("\n4. Environment comparison:")
documenter.compare_environments(
    env1_packages="watermark,numpy",
    env2_packages="watermark,pandas"
)

# Best practices demonstration
print("\n5. Best practices for reproducible research:")
best_practices = [
    "Always include watermark at the beginning of notebooks",
    "Track all critical package versions",
    "Include system information for hardware-dependent code",
    "Generate environment snapshots before major changes",
    "Use version control for environment files",
    "Document both development and production environments"
]

for i, practice in enumerate(best_practices, 1):
    print(f"  {i}. {practice}")

print("\nWatermark enables production-grade environment documentation")
print("Essential for reproducible research and deployment consistency")
````
