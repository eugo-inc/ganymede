## Example 1: Basic Character Width Calculation

```python
import wcwidth

# Example 1: Basic character width calculation
print("Basic character width calculation:")

# Test different character types
test_chars = [
    ('A', 'ASCII letter'),
    ('1', 'ASCII digit'),
    (' ', 'ASCII space'),
    ('中', 'Chinese character'),
    ('日', 'Japanese character'),
    ('한', 'Korean character'),
    ('🦄', 'Emoji unicorn'),
    ('🌈', 'Emoji rainbow'),
    ('\u200b', 'Zero-width space'),
    ('\u0301', 'Combining acute accent'),
]

print("Character width analysis:")
for char, description in test_chars:
    char_width = wcwidth.wcwidth(char)
    print(f"'{char}' ({description}): width = {char_width}")

# Test string width calculation
test_strings = [
    "Hello",
    "Hello World",
    "Hello 世界",
    "🦄🌈✨",
    "Café naïve",
    "مرحبا",  # Arabic
    "Привет"  # Russian
]

print(f"\nString width calculation:")
for text in test_strings:
    string_width = wcwidth.wcswidth(text)
    actual_length = len(text)
    print(f"'{text}': display width = {string_width}, char count = {actual_length}")

print(f"\nWcwidth helps calculate display width for terminal applications")
print("Essential for proper text alignment in console interfaces")
```

## Example 2: Terminal Text Alignment and Formatting

```python
import wcwidth

# Example 2: Text alignment and formatting for terminals
print("Terminal text alignment and formatting:")

class TerminalFormatter:
    """Text formatter using wcwidth for proper alignment"""
    
    def __init__(self, width=50):
        self.width = width
    
    def pad_text(self, text, align='left'):
        """Pad text to specific width with proper alignment"""
        display_width = wcwidth.wcswidth(text) or 0
        padding_needed = max(0, self.width - display_width)
        
        if align == 'left':
            return text + ' ' * padding_needed
        elif align == 'right':
            return ' ' * padding_needed + text
        elif align == 'center':
            left_pad = padding_needed // 2
            right_pad = padding_needed - left_pad
            return ' ' * left_pad + text + ' ' * right_pad
        
        return text
    
    def truncate_text(self, text, max_width=None):
        """Truncate text to fit within display width"""
        if max_width is None:
            max_width = self.width
        
        if wcwidth.wcswidth(text) <= max_width:
            return text
        
        # Find truncation point
        current_width = 0
        for i, char in enumerate(text):
            char_width = wcwidth.wcwidth(char) or 0
            if current_width + char_width > max_width - 3:  # Leave space for "..."
                return text[:i] + "..."
            current_width += char_width
        
        return text

# Create formatter
formatter = TerminalFormatter(width=30)

# Test different text types with alignment
test_texts = [
    "Simple ASCII text",
    "Mixed 中文 text",
    "Emoji 🦄🌈 content",
    "Arabic مرحبا world",
    "Long text that needs truncation example"
]

print("Text alignment examples (width=30):")
print("=" * 35)

for text in test_texts:
    left = formatter.pad_text(text, 'left')
    center = formatter.pad_text(text, 'center')
    right = formatter.pad_text(text, 'right')
    truncated = formatter.truncate_text(text)
    
    print(f"Original: '{text}'")
    print(f"Left:     |{left}|")
    print(f"Center:   |{center}|")
    print(f"Right:    |{right}|")
    print(f"Truncate: |{truncated}|")
    print("-" * 35)

# Demonstrate table formatting
print(f"\nTable formatting with mixed content:")
table_data = [
    ("Name", "Country", "Language"),
    ("Alice", "USA", "English"),
    ("田中", "Japan", "日本語"),
    ("Ahmed", "مصر", "العربية"),
    ("Владимир", "Россия", "Русский")
]

# Calculate column widths
col_widths = [0, 0, 0]
for row in table_data:
    for i, cell in enumerate(row):
        width = wcwidth.wcswidth(cell) or 0
        col_widths[i] = max(col_widths[i], width)

# Print formatted table
for row in table_data:
    formatted_row = []
    for i, cell in enumerate(row):
        cell_width = wcwidth.wcswidth(cell) or 0
        padding = col_widths[i] - cell_width
        formatted_cell = cell + ' ' * padding
        formatted_row.append(formatted_cell)
    print(f"| {' | '.join(formatted_row)} |")

print(f"\nWcwidth enables proper text formatting across different scripts and emojis")
```

## Example 3: Unicode Categories and Character Analysis

```python
import wcwidth

# Example 3: Unicode version support and character categories
print("Unicode version support and character categories:")

# Show available Unicode versions
unicode_versions = wcwidth.list_versions()
print(f"Available Unicode versions: {unicode_versions}")

# Test different character categories
character_categories = [
    ("Zero-width", ['\u200b', '\u200c', '\u200d', '\ufeff']),  # Zero-width chars
    ("Combining", ['\u0300', '\u0301', '\u0302', '\u0308']),   # Combining marks
    ("Wide CJK", ['中', '国', '日', '本', '한', '글']),         # CJK characters
    ("Emoji", ['😀', '🦄', '🌈', '⭐', '🔥', '💻']),           # Emoji
    ("ASCII", ['A', 'B', '1', '2', ' ', '!']),                # ASCII
    ("Extended", ['café', 'naïve', 'résumé']),                # Latin extended
]

print(f"\nCharacter category analysis:")
for category, chars in character_categories:
    print(f"\n{category} characters:")
    for char in chars:
        if isinstance(char, str) and len(char) == 1:
            width = wcwidth.wcwidth(char)
            unicode_point = ord(char)
            print(f"  '{char}' (U+{unicode_point:04X}): width = {width}")
        else:
            # Handle multi-character strings
            width = wcwidth.wcswidth(char)
            print(f"  '{char}': width = {width}")

# Test control characters and special cases
special_cases = [
    ('\t', 'Tab character'),
    ('\n', 'Newline'),
    ('\r', 'Carriage return'),
    ('\x00', 'Null character'),
    ('\x7f', 'DEL character'),
]

print(f"\nSpecial character handling:")
for char, description in special_cases:
    width = wcwidth.wcwidth(char)
    print(f"'{repr(char)}' ({description}): width = {width}")

# Test East Asian width property
east_asian_examples = [
    ('A', 'Narrow'),
    ('ア', 'Halfwidth Katakana'),
    ('ア', 'Fullwidth Katakana'),
    ('中', 'Wide'),
    ('︰', 'Fullwidth colon'),
]

print(f"\nEast Asian width examples:")
for char, description in east_asian_examples:
    width = wcwidth.wcwidth(char)
    print(f"'{char}' ({description}): width = {width}")

print(f"\nWcwidth handles complex Unicode character width rules")
print("Critical for internationalized terminal applications")
```

## Example 4: Progress Bars and Console UI Elements

```python
import wcwidth

# Example 4: Progress bars and console UI elements
print("Progress bars and console UI elements:")

class ConsoleProgressBar:
    """Progress bar with proper width handling"""
    
    def __init__(self, total, width=50):
        self.total = total
        self.width = width
        self.current = 0
    
    def update(self, progress, message=""):
        """Update progress bar with message"""
        self.current = progress
        percentage = min(100, (progress / self.total) * 100)
        
        # Calculate bar components
        filled_width = int((progress / self.total) * self.width)
        empty_width = self.width - filled_width
        
        # Create bar visual
        bar = '█' * filled_width + '░' * empty_width
        
        # Handle message width
        if message:
            msg_width = wcwidth.wcswidth(message) or 0
            # Truncate message if too long
            max_msg_width = 30
            if msg_width > max_msg_width:
                truncated = ""
                current_width = 0
                for char in message:
                    char_width = wcwidth.wcwidth(char) or 0
                    if current_width + char_width > max_msg_width - 3:
                        truncated += "..."
                        break
                    truncated += char
                    current_width += char_width
                message = truncated
        
        return f"[{bar}] {percentage:6.1f}% {message}"
    
    def demo_progress(self):
        """Demonstrate progress bar with different content"""
        import time
        
        tasks = [
            (10, "Starting..."),
            (25, "Loading 数据..."),
            (40, "Processing 🔄"),
            (60, "Analyzing results"),
            (80, "Finishing العمل..."),
            (100, "Complete! ✓")
        ]
        
        print("Progress bar demonstration:")
        for progress, message in tasks:
            bar_display = self.update(progress, message)
            print(f"\r{bar_display}", end="", flush=True)
            time.sleep(0.2)
        print()  # New line after completion

# Demo progress bar
progress_bar = ConsoleProgressBar(100, width=40)
progress_bar.demo_progress()

# Console menu with proper alignment
print(f"\nConsole menu with international text:")

class ConsoleMenu:
    """Console menu with proper text alignment"""
    
    def __init__(self, width=60):
        self.width = width
    
    def format_menu_item(self, number, text, shortcut=""):
        """Format menu item with proper alignment"""
        # Create the basic format
        item_text = f"{number}. {text}"
        
        if shortcut:
            # Calculate spaces needed for right alignment
            item_width = wcwidth.wcswidth(item_text) or 0
            shortcut_width = wcwidth.wcswidth(shortcut) or 0
            spacing_needed = self.width - item_width - shortcut_width - 2  # 2 for brackets
            
            if spacing_needed > 0:
                return f"{item_text}{' ' * spacing_needed}[{shortcut}]"
            else:
                return f"{item_text} [{shortcut}]"
        
        return item_text
    
    def display_menu(self, items):
        """Display formatted menu"""
        border = "=" * self.width
        print(border)
        print(self.center_text("International Application Menu"))
        print(border)
        
        for i, (text, shortcut) in enumerate(items, 1):
            formatted = self.format_menu_item(i, text, shortcut)
            print(formatted)
        
        print(border)
    
    def center_text(self, text):
        """Center text within width"""
        text_width = wcwidth.wcswidth(text) or 0
        padding = (self.width - text_width) // 2
        return " " * padding + text

# Demo menu
menu = ConsoleMenu(width=50)
menu_items = [
    ("View Profile", "Ctrl+P"),
    ("设置 Settings", "Ctrl+S"),
    ("تقارير Reports", "Ctrl+R"),
    ("Справка Help", "F1"),
    ("Exit 🚪", "Ctrl+Q")
]

menu.display_menu(menu_items)

print(f"\nWcwidth enables professional console interfaces")
print("Handles mixed scripts and emojis correctly in UI elements")
```

## Example 5: Production Terminal Applications

```python
import wcwidth
import re

# Example 5: Production patterns for terminal applications
class TerminalTextProcessor:
    """Production-ready text processing for terminal applications"""
    
    def __init__(self):
        # Get the latest Unicode version
        versions = wcwidth.list_versions()
        self.unicode_version = versions[-1] if versions else "unknown"
        self.stats = {"processed": 0, "truncated": 0, "aligned": 0}
    
    def safe_width(self, text):
        """Safely calculate display width with fallback"""
        try:
            width = wcwidth.wcswidth(text)
            return width if width is not None else len(text)
        except (TypeError, ValueError):
            return len(text)
    
    def normalize_line_endings(self, text):
        """Normalize line endings and remove problematic characters"""
        # Replace different line endings with \n
        text = re.sub(r'\r\n|\r', '\n', text)
        
        # Remove or replace control characters except tabs and newlines
        cleaned = ''
        for char in text:
            if char in ['\n', '\t']:
                cleaned += char
            elif wcwidth.wcwidth(char) == -1 and ord(char) < 32:
                cleaned += ' '  # Replace control chars with space
            else:
                cleaned += char
        
        return cleaned
    
    def smart_truncate(self, text, max_width, suffix="..."):
        """Intelligently truncate text at word boundaries when possible"""
        if self.safe_width(text) <= max_width:
            return text
        
        suffix_width = self.safe_width(suffix)
        target_width = max_width - suffix_width
        
        if target_width <= 0:
            return suffix[:max_width]
        
        # Try to break at word boundary
        words = text.split()
        result = ""
        current_width = 0
        
        for word in words:
            word_width = self.safe_width(word)
            space_width = 1 if result else 0
            
            if current_width + space_width + word_width <= target_width:
                if result:
                    result += " "
                    current_width += 1
                result += word
                current_width += word_width
            else:
                break
        
        # If we couldn't fit any complete words, truncate character by character
        if not result:
            for i, char in enumerate(text):
                char_width = wcwidth.wcwidth(char) or 0
                if current_width + char_width > target_width:
                    result = text[:i]
                    break
                current_width += char_width
        
        self.stats["truncated"] += 1
        return result + suffix
    
    def create_table(self, data, headers=None, max_col_width=20):
        """Create a properly formatted table with international content"""
        if not data:
            return ""
        
        # Determine number of columns
        if headers:
            num_cols = len(headers)
            table_data = [headers] + data
        else:
            num_cols = len(data[0]) if data else 0
            table_data = data
        
        # Calculate column widths
        col_widths = [0] * num_cols
        processed_data = []
        
        for row in table_data:
            processed_row = []
            for i, cell in enumerate(row):
                cell_str = str(cell)
                # Truncate if necessary
                if self.safe_width(cell_str) > max_col_width:
                    cell_str = self.smart_truncate(cell_str, max_col_width)
                
                processed_row.append(cell_str)
                col_widths[i] = max(col_widths[i], self.safe_width(cell_str))
            processed_data.append(processed_row)
        
        # Generate table
        lines = []
        separator = "+" + "+".join("-" * (w + 2) for w in col_widths) + "+"
        
        for i, row in enumerate(processed_data):
            if i == 0:
                lines.append(separator)
            
            formatted_cells = []
            for j, cell in enumerate(row):
                cell_width = self.safe_width(cell)
                padding = col_widths[j] - cell_width
                formatted_cells.append(f" {cell}{' ' * padding} ")
            
            lines.append("|" + "|".join(formatted_cells) + "|")
            
            if i == 0 and headers:  # Add separator after headers
                lines.append(separator)
        
        lines.append(separator)
        self.stats["aligned"] += len(processed_data)
        return "\n".join(lines)
    
    def format_log_entry(self, timestamp, level, source, message, width=100):
        """Format log entry with proper alignment"""
        # Create fixed-width components
        time_str = timestamp[:19]  # YYYY-MM-DD HH:MM:SS
        level_str = f"[{level:5s}]"
        source_str = f"{source:15s}"[:15]
        
        # Calculate remaining space for message
        prefix = f"{time_str} {level_str} {source_str} "
        prefix_width = self.safe_width(prefix)
        message_width = width - prefix_width
        
        # Truncate message if needed
        if self.safe_width(message) > message_width:
            message = self.smart_truncate(message, message_width)
        
        self.stats["processed"] += 1
        return prefix + message
    
    def get_stats(self):
        """Get processing statistics"""
        return self.stats.copy()

print("Production terminal text processing:")

# Create processor
processor = TerminalTextProcessor()

# Test international table data
print("1. International data table:")
table_data = [
    ["Name", "Country", "Population", "Language"],
    ["东京", "日本", "13,960,000", "日本語"],
    ["القاهرة", "مصر", "20,900,000", "العربية"],  
    ["Mumbai", "India", "20,400,000", "हिन्दी/English"],
    ["São Paulo", "Brasil", "22,400,000", "Português"],
    ["Москва", "Россия", "12,500,000", "Русский"]
]

formatted_table = processor.create_table(table_data[1:], headers=table_data[0])
print(formatted_table)

# Test log formatting
print(f"\n2. Log entry formatting:")
log_entries = [
    ("2024-01-15 10:30:15", "INFO", "auth", "User login successful for 田中太郎"),
    ("2024-01-15 10:30:16", "WARN", "api", "Rate limit approaching for client 🏢 Corp"),
    ("2024-01-15 10:30:17", "ERROR", "db", "Connection failed to database التطبيق_prod"),
    ("2024-01-15 10:30:18", "DEBUG", "cache", "Cache miss for key user:settings:Владимир123")
]

for entry in log_entries:
    formatted = processor.format_log_entry(*entry, width=80)
    print(formatted)

# Test text normalization and truncation
print(f"\n3. Text normalization and smart truncation:")
test_texts = [
    "This is a very long message that contains mixed content like 中文字符 and emojis",
    "Another example with different scripts: العربية русский 한글"
]

for text in test_texts:
    normalized = processor.normalize_line_endings(text)
    truncated = processor.smart_truncate(normalized, 50)
    print(f"Original : {text}")
    print(f"Truncated: {truncated}")
    print()

# Show processing statistics
stats = processor.get_stats()
print(f"Processing statistics:")
for key, value in stats.items():
    print(f"  {key}: {value}")

print(f"\nProduction benefits:")
print(f"  Unicode version: {processor.unicode_version}")
benefits = [
    "Accurate width calculation for all Unicode scripts",
    "Safe handling of control characters and edge cases",
    "Smart truncation at word boundaries",
    "Professional table formatting",
    "International log processing"
]

for benefit in benefits:
    print(f"  - {benefit}")

print(f"\nWcwidth enables robust international terminal applications")
```
