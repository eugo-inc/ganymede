## Example 1: Basic color name to hex conversion

```python
import webcolors

# Example 1: Basic color name to hex conversion
print("Example 1: Basic color name to hex conversion")

try:
    # Convert common color names to hex values
    colors = ['red', 'blue', 'green', 'yellow', 'purple', 'orange']
    
    print("Color name -> Hex conversion:")
    for color in colors:
        hex_value = webcolors.name_to_hex(color)
        print(f"  {color:8} -> {hex_value}")
    
    # Demonstrate case insensitivity
    upper_red = webcolors.name_to_hex('RED')
    print(f"\nCase insensitive: RED -> {upper_red}")
    
    print("webcolors example 1 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"webcolors Example 1 failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: Hex to color name and RGB conversion

```python
import webcolors

# Example 2: Hex to color name and RGB conversion
print("Example 2: Hex to color name and RGB conversion")

try:
    # Convert hex values back to color names
    hex_colors = ['#ff0000', '#0000ff', '#008000', '#ffff00', '#800080']
    
    print("Hex -> Color name and RGB:")
    for hex_color in hex_colors:
        try:
            name = webcolors.hex_to_name(hex_color)
            rgb = webcolors.hex_to_rgb(hex_color)
            print(f"  {hex_color} -> {name:8} -> RGB{rgb}")
        except ValueError:
            # Some hex values might not have standard names
            rgb = webcolors.hex_to_rgb(hex_color)
            print(f"  {hex_color} -> (no name) -> RGB{rgb}")
    
    print("webcolors example 2 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"webcolors Example 2 failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: RGB tuple to hex and percentage conversions

```python
import webcolors

# Example 3: RGB tuple to hex and percentage conversions
print("Example 3: RGB tuple to hex and percentage conversions")

try:
    # Define RGB color tuples
    rgb_colors = [
        (255, 0, 0),    # Red
        (0, 128, 0),    # Green
        (0, 0, 255),    # Blue
        (255, 255, 0),  # Yellow
        (128, 0, 128)   # Purple
    ]
    
    print("RGB -> Hex and Percentage:")
    for rgb in rgb_colors:
        hex_value = webcolors.rgb_to_hex(rgb)
        percent = webcolors.rgb_to_rgb_percent(rgb)
        
        try:
            name = webcolors.rgb_to_name(rgb)
            print(f"  RGB{rgb} -> {hex_value} -> {name} -> {percent}")
        except ValueError:
            print(f"  RGB{rgb} -> {hex_value} -> (no name) -> {percent}")
    
    print("webcolors example 3 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"webcolors Example 3 failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: HTML5 color parsing and validation

```python
import webcolors

# Example 4: HTML5 color parsing and validation
print("Example 4: HTML5 color parsing and validation")

try:
    # Test HTML5 color parsing capabilities
    test_colors = [
        'red',           # Named color
        '#ff0000',       # Hex color
        'rgb(255,0,0)',  # RGB functional notation
        'hsl(0,100%,50%)' # HSL (if supported)
    ]
    
    print("HTML5 Color Parsing:")
    for color in test_colors:
        try:
            # Try HTML5 simple color parsing
            parsed = webcolors.html5_parse_simple_color(color)
            serialized = webcolors.html5_serialize_simple_color(parsed)
            print(f"  '{color}' -> parsed: {parsed} -> serialized: '{serialized}'")
        except ValueError as ve:
            print(f"  '{color}' -> Parse error: {ve}")
        except AttributeError:
            print(f"  '{color}' -> Function not available")
    
    # List available color specifications
    print(f"\nSupported specifications: CSS2, CSS21, CSS3, HTML4")
    
    print("webcolors example 4 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"webcolors Example 4 failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: Color specification standards and available colors

```python
import webcolors

# Example 5: Color specification standards and available colors
print("Example 5: Color specification standards and available colors")

try:
    # Get all available color names for different specifications
    print("Available color specifications and sample colors:")
    
    # Get color names from different specs
    css3_colors = webcolors.names(webcolors.CSS3)
    html4_colors = webcolors.names(webcolors.HTML4)
    
    print(f"CSS3 colors available: {len(css3_colors)}")
    print(f"Sample CSS3 colors: {list(css3_colors)[:8]}")
    
    print(f"\nHTML4 colors available: {len(html4_colors)}")
    print(f"Sample HTML4 colors: {list(html4_colors)[:8]}")
    
    # Test specific color in different specs
    test_color = 'darkblue'
    print(f"\nTesting '{test_color}' in different specs:")
    
    for spec in [webcolors.CSS3, webcolors.HTML4]:
        try:
            hex_val = webcolors.name_to_hex(test_color, spec=spec)
            print(f"  {spec}: {test_color} -> {hex_val}")
        except ValueError:
            print(f"  {spec}: {test_color} -> Not available")
    
    print("webcolors example 5 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"webcolors Example 5 failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
