## Example 1: Basic xarray DataArray creation

```python
import xarray as xr
import numpy as np

# Example 1: Basic xarray DataArray creation
print("Example 1: Testing basic xarray DataArray")

try:
    # Create a simple DataArray
    data = np.random.rand(3, 4)
    coords = {'x': [1, 2, 3], 'y': ['a', 'b', 'c', 'd']}
    
    da = xr.DataArray(data, coords=coords, dims=['x', 'y'])
    print("Created xarray DataArray successfully")
    print(f"DataArray shape: {da.shape}")
    print(f"DataArray dimensions: {da.dims}")
    print(f"DataArray coordinates: {list(da.coords.keys())}")
    
    print("xarray example 1 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"xarray Example 1 failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 1 finished")
```

## Example 2: xarray Dataset creation

```python
import xarray as xr
import pandas as pd

# Example 2: xarray Dataset creation
print("Example 2: Testing xarray Dataset")

try:
    # Create a simple Dataset
    temps = xr.DataArray([15, 25, 35], dims=['time'], coords={'time': pd.date_range('2023-01-01', periods=3)})
    humidity = xr.DataArray([60, 70, 80], dims=['time'], coords={'time': pd.date_range('2023-01-01', periods=3)})
    
    ds = xr.Dataset({'temperature': temps, 'humidity': humidity})
    print("Created xarray Dataset successfully")
    print(f"Dataset variables: {list(ds.data_vars)}")
    print(f"Dataset coordinates: {list(ds.coords)}")
    print(f"Dataset dimensions: {dict(ds.dims)}")
    
    print("xarray example 2 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"xarray Example 2 failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 2 finished")
```

## Example 3: xarray arithmetic operations

```python
import xarray as xr
import numpy as np

# Example 3: xarray arithmetic operations
print("Example 3: Testing xarray arithmetic operations")

try:
    # Create two DataArrays for arithmetic
    data1 = np.array([[1, 2], [3, 4]])
    data2 = np.array([[5, 6], [7, 8]])
    
    da1 = xr.DataArray(data1, dims=['x', 'y'])
    da2 = xr.DataArray(data2, dims=['x', 'y'])
    
    # Perform arithmetic operations
    result_add = da1 + da2
    result_mult = da1 * da2
    
    print("Created DataArrays for arithmetic")
    print(f"da1 + da2 = {result_add.values}")
    print(f"da1 * da2 = {result_mult.values}")
    print("Arithmetic operations work with labeled dimensions")
    
    print("xarray example 3 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"xarray Example 3 failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 3 finished")
```

## Example 4: xarray indexing and selection

```python
import xarray as xr
import numpy as np

# Example 4: xarray indexing and selection
print("Example 4: Testing xarray indexing and selection")

try:
    # Create DataArray with named coordinates
    data = np.random.rand(4, 3)
    coords = {'time': [2020, 2021, 2022, 2023], 'location': ['A', 'B', 'C']}
    da = xr.DataArray(data, coords=coords, dims=['time', 'location'])
    
    # Test indexing operations
    print("Created DataArray with time and location coordinates")
    
    # Select by coordinate value
    selected_2021 = da.sel(time=2021)
    print(f"Selected data for 2021: shape {selected_2021.shape}")
    
    # Select location 'A'
    location_a = da.sel(location='A')
    print(f"Selected data for location A: shape {location_a.shape}")
    
    print("xarray example 4 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"xarray Example 4 failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 4 finished")
```

## Example 5: xarray aggregation operations

```python
import xarray as xr
import numpy as np

# Example 5: xarray aggregation operations
print("Example 5: Testing xarray aggregation operations")

try:
    # Create 3D DataArray for aggregation testing
    data = np.random.rand(2, 3, 4)
    coords = {'time': [2022, 2023], 'lat': [10, 20, 30], 'lon': [100, 110, 120, 130]}
    da = xr.DataArray(data, coords=coords, dims=['time', 'lat', 'lon'])
    
    print("Created 3D DataArray for aggregation")
    print(f"Original shape: {da.shape}")
    
    # Test aggregation operations
    mean_all = da.mean()
    mean_time = da.mean(dim='time')
    sum_lat_lon = da.sum(dim=['lat', 'lon'])
    
    print(f"Mean of all values: {mean_all.values:.3f}")
    print(f"Mean over time dimension shape: {mean_time.shape}")
    print(f"Sum over lat/lon dimensions shape: {sum_lat_lon.shape}")
    
    print("xarray example 5 completed successfully")
    
except Exception as e:
    print(f"Runtime error: {e}")
    raise AssertionError(
        f"xarray Example 5 failed: {type(e).__name__}: {e}"
    ) from e
    
print("Example 5 finished")
```
