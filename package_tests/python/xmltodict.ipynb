## Example 1: Basic XML to dictionary conversion

```python
import xmltodict

# Example 1: Basic XML to dictionary conversion
print("Basic XML to dictionary conversion:")

# Simple XML string
xml_data = """<book>
    <title>Python Programming</title>
    <author>John Smith</author>
    <price>29.99</price>
</book>"""

# Convert XML to dictionary
result = xmltodict.parse(xml_data)
print(f"XML parsed to: {type(result)}")
print(f"Root element: {list(result.keys())[0]}")

# Access data
book = result['book']
print(f"Title: {book['title']}")
print(f"Author: {book['author']}")
print(f"Price: ${book['price']}")

print("Basic conversion completed")
```

## Example 2: XML with attributes

```python
import xmltodict

# Example 2: XML with attributes
print("XML with attributes:")

# XML with attributes
xml_with_attrs = """<product id="123" category="electronics">
    <name>Laptop</name>
    <price currency="USD">999.99</price>
    <specs>
        <cpu cores="4">Intel i5</cpu>
        <memory>16GB</memory>
    </specs>
</product>"""

# Parse XML with attributes
result = xmltodict.parse(xml_with_attrs)
product = result['product']

print(f"Product ID: {product['@id']}")
print(f"Category: {product['@category']}")
print(f"Name: {product['name']}")
print(f"Price: {product['price']['#text']} {product['price']['@currency']}")

# Nested attributes
cpu = product['specs']['cpu']
print(f"CPU: {cpu['#text']} ({cpu['@cores']} cores)")
print(f"Memory: {product['specs']['memory']}")

print("Attribute parsing completed")
```

## Example 3: Dictionary to XML conversion

```python
import xmltodict

# Example 3: Dictionary to XML conversion  
print("Dictionary to XML conversion:")

# Create dictionary structure
data = {
    'order': {
        '@id': '12345',
        'customer': 'Jane Doe',
        'items': {
            'item': [
                {'name': 'Widget A', 'quantity': '2'},
                {'name': 'Widget B', 'quantity': '1'}
            ]
        },
        'total': '59.98'
    }
}

# Convert dictionary to XML
xml_output = xmltodict.unparse(data, pretty=True)
print("Generated XML:")
print(xml_output)

# Test round-trip conversion
parsed_back = xmltodict.parse(xml_output)
print(f"\nRound-trip test successful: {data.keys() == parsed_back.keys()}")

print("Dictionary to XML conversion completed")
```

## Example 4: Complex XML with lists

```python
import xmltodict

# Example 4: Complex XML with lists
print("Complex XML with lists:")

# XML with repeated elements (lists)
complex_xml = """<library>
    <books>
        <book id="1">
            <title>Python Guide</title>
            <tags>
                <tag>programming</tag>
                <tag>python</tag>
                <tag>tutorial</tag>
            </tags>
        </book>
        <book id="2">
            <title>Data Science</title>
            <tags>
                <tag>data</tag>
                <tag>analytics</tag>
            </tags>
        </book>
    </books>
</library>"""

# Parse complex XML
result = xmltodict.parse(complex_xml)
library = result['library']

# Access books (list)
books = library['books']['book']
print(f"Number of books: {len(books)}")

for i, book in enumerate(books):
    print(f"\nBook {i+1}:")
    print(f"  ID: {book['@id']}")
    print(f"  Title: {book['title']}")
    
    # Handle tags (can be single item or list)
    tags = book['tags']['tag']
    if isinstance(tags, list):
        print(f"  Tags: {', '.join(tags)}")
    else:
        print(f"  Tags: {tags}")

print("Complex XML parsing completed")
```

## Example 5: Custom parsing options

```python
import xmltodict
from collections import OrderedDict

# Example 5: Custom parsing options
print("Custom parsing options:")

# XML with namespaces
xml_with_ns = """<config xmlns:db="http://database.com">
    <db:connection>
        <db:host>localhost</db:host>
        <db:port>5432</db:port>
        <db:name>mydb</db:name>
    </db:connection>
    <settings>
        <debug>true</debug>
        <timeout>30</timeout>
    </settings>
</config>"""

# Parse with different options
print("1. Default parsing:")
default = xmltodict.parse(xml_with_ns)
print(f"   Root keys: {list(default.keys())}")

print("2. Namespace processing:")
ns_processed = xmltodict.parse(xml_with_ns, process_namespaces=True)
print(f"   Root keys: {list(ns_processed.keys())}")

print("3. Using OrderedDict:")
ordered = xmltodict.parse(xml_with_ns, dict_constructor=OrderedDict)
print(f"   Type: {type(ordered['config'])}")

# Custom postprocessor
def postprocessor(path, key, value):
    # Convert numeric strings to numbers
    if key in ['port', 'timeout'] and isinstance(value, str):
        return key, int(value)
    if key == 'debug' and value == 'true':
        return key, True
    return key, value

processed = xmltodict.parse(xml_with_ns, postprocessor=postprocessor)
config = processed['config']
print(f"4. With postprocessor:")
print(f"   Port type: {type(config['settings']['timeout'])}")
print(f"   Debug type: {type(config['settings']['debug'])}")

print("Custom parsing options completed")
```
