```python
# Example 1: Basic URL Creation and Component Access
from yarl import URL

# Create URL objects from strings
simple_url = URL('https://example.com/path')
complex_url = URL('https://user:pass@api.example.com:8080/v1/users?page=2&limit=10#section1')

print("Basic URL Creation and Component Access:")
print(f"Simple URL: {simple_url}")
print(f"Complex URL: {complex_url}")

# Access URL components
print(f"\nURL Components for: {complex_url}")
print(f"Scheme: {complex_url.scheme}")
print(f"Host: {complex_url.host}")
print(f"Port: {complex_url.port}")
print(f"Path: {complex_url.path}")
print(f"Query string: {complex_url.query_string}")
print(f"Fragment: {complex_url.fragment}")
print(f"User: {complex_url.user}")
print(f"Password: {complex_url.password}")

# Parse query parameters
print(f"\nQuery Parameters:")
for key, value in complex_url.query.items():
    print(f"  {key}: {value}")

# URL properties
print(f"\nURL Properties:")
print(f"Is absolute: {complex_url.is_absolute()}")
print(f"Authority: {complex_url.authority}")
print(f"Raw path: {complex_url.raw_path}")
print(f"Human repr: {complex_url.human_repr()}")

print("Basic URL creation and component access complete")
```

```python
# Example 2: URL Manipulation and Transformation
from yarl import URL

# Start with a base URL
base_url = URL('https://api.example.com/v1')
print(f"Base URL: {base_url}")

# URL manipulation methods (immutable - returns new URL objects)
print("\nURL Manipulation Examples:")

# Change scheme
secure_url = base_url.with_scheme('https')
print(f"With HTTPS: {secure_url}")

# Change host
dev_url = base_url.with_host('dev.example.com')
print(f"Dev environment: {dev_url}")

# Change port
port_url = base_url.with_port(8080)
print(f"With custom port: {port_url}")

# Modify path
users_url = base_url.with_path('/v1/users')
print(f"Users endpoint: {users_url}")

# Join paths
user_detail_url = users_url / '123' / 'profile'
print(f"User profile: {user_detail_url}")

# Add query parameters
filtered_url = users_url.with_query({
    'page': 2,
    'limit': 50,
    'status': 'active',
    'sort': 'created_at'
})
print(f"With query params: {filtered_url}")

# Update query parameters
updated_query = filtered_url.update_query(limit=25, order='desc')
print(f"Updated query: {updated_query}")

# Add fragment
fragment_url = user_detail_url.with_fragment('contact-info')
print(f"With fragment: {fragment_url}")

# Chain multiple operations
complex_transform = (base_url
                    .with_host('secure.api.example.com')
                    .with_port(443)
                    .with_path('/v2/search')
                    .with_query(q='python', type='library', limit=100)
                    .with_fragment('results'))
print(f"Chained operations: {complex_transform}")

print("URL manipulation and transformation complete")
```

```python
# Example 3: URL Joining and Resolution
from yarl import URL

# Base URL for relative resolution
base = URL('https://api.example.com/v1/users/')
print(f"Base URL: {base}")

# Relative URL resolution
print("\nRelative URL Resolution:")

# Join with relative paths
relative_paths = ['123', '../posts', './profile', '/admin/users', '?filter=active']

for path in relative_paths:
    resolved = base.join(URL(path))
    print(f"'{path}' -> {resolved}")

# Alternative joining syntax
print("\nAlternative Joining Methods:")
user_123 = base / '123'
user_posts = user_123 / 'posts'
recent_posts = user_posts.with_query(recent='true', limit=10)

print(f"User 123: {user_123}")
print(f"User posts: {user_posts}")
print(f"Recent posts: {recent_posts}")

# Working with different base URLs
print("\nDifferent Base URL Scenarios:")

bases = [
    URL('https://example.com/api/'),
    URL('https://example.com/api/v1'),
    URL('https://example.com/api/v1/'),
    URL('https://example.com/')
]

relative_url = 'users/123'

for base_url in bases:
    joined = base_url.join(URL(relative_url))
    print(f"{base_url} + '{relative_url}' = {joined}")

# Parent directory operations
print("\nParent Directory Operations:")
deep_url = URL('https://example.com/api/v1/users/123/posts/456/comments')
print(f"Original: {deep_url}")

current = deep_url
for i in range(4):
    current = current.parent
    print(f"Parent {i+1}: {current}")

# URL normalization
print("\nURL Normalization:")
messy_urls = [
    'https://example.com/api/../api/v1/./users/../users/123',
    'https://example.com//api//v1///users/123',
    'https://example.com/api/v1/users/123/..',
]

for messy in messy_urls:
    clean = URL(messy)
    print(f"Messy:  {messy}")
    print(f"Clean:  {clean}")
    print()

print("URL joining and resolution complete")
```

```python
# Example 4: Advanced Query Parameter Handling
from yarl import URL
from urllib.parse import quote, unquote

# Complex query parameter scenarios
base_url = URL('https://api.example.com/search')

print("Advanced Query Parameter Handling:")

# Multiple values for same parameter
multi_value_query = base_url.with_query([
    ('tag', 'python'),
    ('tag', 'web'),
    ('tag', 'api'),
    ('format', 'json'),
    ('limit', '50')
])
print(f"Multi-value params: {multi_value_query}")

# Working with special characters
special_chars_url = base_url.with_query({
    'q': 'hello world & friends',
    'filter': 'name="John Doe"',
    'path': '/home/user/file.txt',
    'symbols': '!@#$%^&*()'
})
print(f"Special characters: {special_chars_url}")
print(f"Raw query string: {special_chars_url.raw_query_string}")

# Query parameter manipulation
print("\nQuery Parameter Manipulation:")

# Start with existing query
existing_query = URL('https://example.com/api?page=1&limit=10&sort=name')
print(f"Original: {existing_query}")

# Update specific parameters
updated = existing_query.update_query(page=2, limit=25)
print(f"Updated page & limit: {updated}")

# Add new parameters
added = updated.update_query(filter='active', order='desc')
print(f"Added filter & order: {added}")

# Remove parameters - need to rebuild query without the parameter
query_dict = dict(added.query)
query_dict.pop('sort', None)  # Remove 'sort' if it exists
removed = added.with_query(query_dict)
print(f"Removed sort: {removed}")

# Query parameter access patterns
print("\nQuery Parameter Access:")

query_url = URL('https://api.example.com/users?page=2&limit=50&tags=python&tags=web&active=true')
query_dict = query_url.query

print(f"URL: {query_url}")
print(f"Query as dict: {dict(query_dict)}")
print(f"All 'tags' values: {query_dict.getall('tags')}")
print(f"Single 'page' value: {query_dict.get('page')}")
print(f"Missing param with default: {query_dict.get('missing', 'default_value')}")

# Building query strings programmatically
print("\nProgrammatic Query Building:")

def build_search_url(base, search_params):
    """Build search URL with dynamic parameters"""
    url = URL(base)
    
    # Filter out None values and empty strings, convert booleans to strings
    clean_params = {}
    for k, v in search_params.items():
        if v is not None and v != '':
            # Convert boolean to lowercase string
            if isinstance(v, bool):
                clean_params[k] = str(v).lower()
            else:
                clean_params[k] = v
    
    if clean_params:
        return url.with_query(clean_params)
    return url

# Example search parameters
search_configs = [
    {'q': 'python', 'category': 'programming', 'limit': 20},
    {'q': 'machine learning', 'tags': 'ai,ml', 'sort': 'relevance', 'page': None},
    {'category': 'web development', 'min_rating': 4.5, 'format': 'json'},
    {'limit': 100, 'include_inactive': True}  # Boolean value
]

for i, params in enumerate(search_configs, 1):
    search_url = build_search_url('https://api.example.com/search', params)
    print(f"Search {i}: {search_url}")

print("Advanced query parameter handling complete")
```

```python
# Example 5: Production URL Management and Validation
from yarl import URL
import re
from typing import Optional, List, Dict, Union

class URLManager:
    """Production-ready URL management and validation toolkit"""
    
    def __init__(self, base_url: str):
        self.base_url = URL(base_url)
        self.allowed_schemes = {'http', 'https'}
        self.allowed_hosts = set()
        self.url_patterns = {}
    
    def add_allowed_host(self, host: str):
        """Add allowed host for URL validation"""
        self.allowed_hosts.add(host)
    
    def register_pattern(self, name: str, pattern: str):
        """Register URL pattern for validation"""
        self.url_patterns[name] = re.compile(pattern)
    
    def validate_url(self, url: Union[str, URL]) -> Dict[str, Union[bool, str, List[str]]]:
        """Comprehensive URL validation"""
        if isinstance(url, str):
            try:
                url = URL(url)
            except Exception as e:
                return {'valid': False, 'errors': [f'Invalid URL format: {e}']}
        
        errors = []
        warnings = []
        
        # Scheme validation
        if url.scheme not in self.allowed_schemes:
            errors.append(f'Invalid scheme: {url.scheme}. Allowed: {self.allowed_schemes}')
        
        # Host validation
        if self.allowed_hosts and url.host not in self.allowed_hosts:
            errors.append(f'Host not allowed: {url.host}')
        
        # Basic security checks
        if url.user or url.password:
            warnings.append('URL contains credentials - security risk')
        
        # Path validation
        if '..' in url.path:
            warnings.append('Path contains ".." - potential security risk')
        
        if url.path and not url.path.startswith('/'):
            errors.append('Path should start with "/"')
        
        # Port validation
        if url.port and (url.port < 1 or url.port > 65535):
            errors.append(f'Invalid port: {url.port}')
        
        return {
            'valid': len(errors) == 0,
            'errors': errors,
            'warnings': warnings,
            'url': str(url)
        }
    
    def build_api_url(self, endpoint: str, version: str = 'v1', **params) -> URL:
        """Build standardized API URLs"""
        # Ensure endpoint starts with /
        if not endpoint.startswith('/'):
            endpoint = '/' + endpoint
        
        # Build path with version
        api_path = f'/api/{version}{endpoint}'
        
        # Create URL with base and path
        url = self.base_url.with_path(api_path)
        
        # Add query parameters if provided
        if params:
            url = url.with_query(params)
        
        return url
    
    def extract_api_info(self, url: Union[str, URL]) -> Dict[str, Optional[str]]:
        """Extract API information from URL"""
        if isinstance(url, str):
            url = URL(url)
        
        path_parts = url.path.strip('/').split('/')
        
        api_info = {
            'base_url': f"{url.scheme}://{url.host}{':' + str(url.port) if url.port else ''}",
            'api_prefix': None,
            'version': None,
            'endpoint': None,
            'resource_id': None
        }
        
        if len(path_parts) >= 1 and path_parts[0] == 'api':
            api_info['api_prefix'] = 'api'
            
            if len(path_parts) >= 2 and path_parts[1].startswith('v'):
                api_info['version'] = path_parts[1]
                
                if len(path_parts) >= 3:
                    api_info['endpoint'] = path_parts[2]
                    
                    if len(path_parts) >= 4:
                        api_info['resource_id'] = path_parts[3]
        
        return api_info
    
    def generate_sitemap_urls(self, base_patterns: List[str], 
                            dynamic_data: Dict[str, List[str]]) -> List[URL]:
        """Generate sitemap URLs from patterns and data"""
        urls = []
        
        for pattern in base_patterns:
            if '{' in pattern:  # Dynamic pattern
                # Extract placeholder names
                import string
                formatter = string.Formatter()
                placeholders = [field_name for _, field_name, _, _ 
                              in formatter.parse(pattern) if field_name]
                
                # Generate combinations
                if all(ph in dynamic_data for ph in placeholders):
                    if len(placeholders) == 1:
                        ph = placeholders[0]
                        for value in dynamic_data[ph]:
                            url_path = pattern.format(**{ph: value})
                            urls.append(self.base_url.with_path(url_path))
                    else:
                        # For multiple placeholders, use first few combinations
                        for i, value1 in enumerate(dynamic_data[placeholders[0]][:5]):
                            for j, value2 in enumerate(dynamic_data[placeholders[1]][:3]):
                                if len(placeholders) == 2:
                                    url_path = pattern.format(
                                        **{placeholders[0]: value1, placeholders[1]: value2}
                                    )
                                    urls.append(self.base_url.with_path(url_path))
            else:  # Static pattern
                urls.append(self.base_url.with_path(pattern))
        
        return urls

# Demonstrate production URL management
print("Production URL Management Demo:")

# Initialize URL manager
manager = URLManager('https://api.example.com')
manager.add_allowed_host('api.example.com')
manager.add_allowed_host('staging.example.com')

# URL validation examples
print("\nURL Validation:")
test_urls = [
    'https://api.example.com/users',
    'http://malicious.com/attack',
    'https://user:pass@api.example.com/data',
    'https://api.example.com/../admin',
    'ftp://api.example.com/file'
]

for test_url in test_urls:
    result = manager.validate_url(test_url)
    status = "✓" if result['valid'] else "✗"
    print(f"{status} {test_url}")
    if result['errors']:
        print(f"   Errors: {'; '.join(result['errors'])}")
    if result['warnings']:
        print(f"   Warnings: {'; '.join(result['warnings'])}")

# API URL building
print("\nAPI URL Building:")
api_urls = [
    manager.build_api_url('users', version='v1', page=1, limit=50),
    manager.build_api_url('posts', version='v2', author_id=123, status='published'),
    manager.build_api_url('search', q='python', category='programming')
]

for api_url in api_urls:
    print(f"  {api_url}")
    info = manager.extract_api_info(api_url)
    print(f"    API Info: {info}")

# Sitemap generation
print("\nSitemap URL Generation:")
patterns = [
    '/',
    '/about',
    '/users/{user_id}',
    '/categories/{category}',
    '/posts/{post_id}/comments'
]

dynamic_data = {
    'user_id': ['1', '2', '3', '4', '5'],
    'category': ['tech', 'science', 'art'],
    'post_id': ['101', '102', '103']
}

sitemap_urls = manager.generate_sitemap_urls(patterns, dynamic_data)
print(f"Generated {len(sitemap_urls)} sitemap URLs:")
for url in sitemap_urls[:10]:  # Show first 10
    print(f"  {url}")
if len(sitemap_urls) > 10:
    print(f"  ... and {len(sitemap_urls) - 10} more")

print("Production URL management and validation complete")
```
