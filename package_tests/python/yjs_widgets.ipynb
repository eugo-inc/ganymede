## Example 1: JupyterLab Extension Detection

```python
import yjs_widgets

# Example 1: Verify yjs-widgets JupyterLab extension installation
print("yjs-widgets JupyterLab Extension Check:")

# Check version
print(f"Package version: {yjs_widgets.__version__}")

# Get JupyterLab extension paths (unique to this package)
extension_paths = yjs_widgets._jupyter_labextension_paths()
print(f"\nExtension paths: {extension_paths}")

for path_info in extension_paths:
    print(f"  Source: {path_info['src']}")
    print(f"  Destination: {path_info['dest']}")

print("\nyjs-widgets provides collaborative editing for JupyterLab")
print("Extension enables real-time synchronization of notebook content")
```

## Example 2: Extension Configuration Check

```python
import yjs_widgets
import json

# Example 2: Examine extension configuration
print("Extension Configuration:")

# Get extension information using unique function
labextension_paths = yjs_widgets._jupyter_labextension_paths()
print(f"Number of extensions: {len(labextension_paths)}")

# Display extension metadata
for i, ext in enumerate(labextension_paths, 1):
    print(f"\nExtension {i}:")
    print(json.dumps(ext, indent=2))

print("\nThis extension integrates Yjs CRDT library with JupyterLab")
print("Enables conflict-free replicated data types for collaboration")
```

## Example 3: Package Metadata and Version Info

```python
import yjs_widgets
from yjs_widgets import _version

# Example 3: Access package version information
print("Package Metadata:")

# Version from main module
print(f"yjs_widgets version: {yjs_widgets.__version__}")

# Version from _version module
print(f"_version.__version__: {_version.__version__}")

# Check version info details if available
if hasattr(_version, 'version_info'):
    print(f"Version info: {_version.version_info}")

print("\nPackage provides JupyterLab integration for Yjs")
print("Used for real-time collaborative notebook editing")
```

## Example 4: Extension Path Resolution

```python
import yjs_widgets
import os

# Example 4: Resolve extension installation paths
print("Extension Path Resolution:")

# Get extension paths using package function
extensions = yjs_widgets._jupyter_labextension_paths()

for ext in extensions:
    src = ext['src']
    dest = ext['dest']
    
    print(f"\nExtension: {dest}")
    print(f"  Source directory: {src}")
    print(f"  Install destination: {dest}")
    
    # Check if package directory exists
    pkg_dir = os.path.dirname(yjs_widgets.__file__)
    print(f"  Package location: {pkg_dir}")

print("\nExtension provides collaborative editing widgets in JupyterLab")
```
