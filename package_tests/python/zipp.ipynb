# zipp: 5 Basic Usage Examples

```python
# Example 1: List files in a zip archive
import zipp
import zipfile
with zipfile.ZipFile('test.zip', 'w') as zf:
    zf.writestr('file.txt', 'hello')
zf = zipp.Path('test.zip')
print(list(zf.iterdir()))
```

```python
# Example 2: Read file contents from a zip archive
import zipp
zf = zipp.Path('test.zip')
print(zf.joinpath('file.txt').read_text())
```

```python
# Example 3: Check if a file exists in a zip archive
import zipp
zf = zipp.Path('test.zip')
print(zf.joinpath('file.txt').exists())
```

```python
# Example 4: Get the name of the zip archive
import zipp
zf = zipp.Path('test.zip')
print(zf.name)
```

```python
# Example 5: Remove the test zip file
import os
os.remove('test.zip')
print('Removed test.zip')
```
