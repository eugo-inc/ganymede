## Example 1: zstandard functionality test 1

```python
import zstandard as zstd

# Example 1: Basic compression and decompression
print("Example 1: Basic compression and decompression")

# Original data
data = b"Hello World! " * 100
print(f"Original size: {len(data)} bytes")

# Compress
compressor = zstd.ZstdCompressor()
compressed = compressor.compress(data)
print(f"Compressed size: {len(compressed)} bytes")
print(f"Compression ratio: {len(data)/len(compressed):.2f}x")

# Decompress
decompressor = zstd.ZstdDecompressor()
decompressed = decompressor.decompress(compressed)
print(f"Decompressed size: {len(decompressed)} bytes")
print(f"Data matches: {data == decompressed}")

print("Example 1 completed")
```

## Example 2: zstandard functionality test 2

```python
import zstandard as zstd

# Example 2: Compression levels and performance
print("Example 2: Compression levels")

data = b"The quick brown fox jumps over the lazy dog. " * 50

# Test different compression levels (1=fastest, 22=best compression)
levels = [1, 5, 10, 15, 22]
print(f"Original size: {len(data)} bytes\n")

for level in levels:
    compressor = zstd.ZstdCompressor(level=level)
    compressed = compressor.compress(data)
    ratio = len(data) / len(compressed)
    print(f"Level {level:2d}: {len(compressed):4d} bytes, ratio: {ratio:.2f}x")

# Default level
default_compressor = zstd.ZstdCompressor()
default_compressed = default_compressor.compress(data)
print(f"\nDefault: {len(default_compressed):4d} bytes")

print("Example 2 completed")
```

## Example 3: zstandard functionality test 3

```python
import zstandard as zstd
import io

# Example 3: Streaming compression with stream_writer
print("Example 3: Streaming compression")

# Create chunks of data
chunks = [b"Chunk %d: " % i + b"data " * 20 for i in range(5)]
print(f"Number of chunks: {len(chunks)}")
print(f"Total uncompressed: {sum(len(c) for c in chunks)} bytes")

# Stream compression using stream_writer
compressor = zstd.ZstdCompressor()
output = io.BytesIO()

with compressor.stream_writer(output, closefd=False) as writer:
    for chunk in chunks:
        writer.write(chunk)
        print(f"  Wrote chunk: {len(chunk)} bytes")

compressed = output.getvalue()
print(f"\nTotal compressed: {len(compressed)} bytes")
print(f"Compression ratio: {sum(len(c) for c in chunks)/len(compressed):.2f}x")

print("Example 3 completed")
```

## Example 4: zstandard functionality test 4

```python
import zstandard as zstd

# Example 4: Compression dictionaries
print("Example 4: Compression dictionaries")

# Training data with common patterns - need many diverse samples
samples = []
for i in range(20):
    samples.append(b"Hello World number " + str(i).encode() + b" " * 200)
    samples.append(b"Python programming " + str(i).encode() + b" " * 200)

print(f"Number of samples: {len(samples)}")
print(f"Total training data: {sum(len(s) for s in samples)} bytes")

# Train a dictionary - use dict_size parameter
dict_data = zstd.train_dictionary(dict_size=8192, samples=samples)
print(f"Dictionary created successfully")

# Test compression with and without dictionary
data = b"Hello World number 5 Python programming " * 20

compressed_no_dict = zstd.ZstdCompressor().compress(data)
compressed_with_dict = zstd.ZstdCompressor(dict_data=dict_data).compress(data)

print(f"Original: {len(data)} bytes")
print(f"Without dict: {len(compressed_no_dict)} bytes")
print(f"With dict: {len(compressed_with_dict)} bytes")

print("Example 4 completed")
```

## Example 5: zstandard functionality test 5

```python
import zstandard as zstd

# Example 5: Content size and frame parameters
print("Example 5: Frame parameters and content size")

data = b"Testing zstandard frame parameters! " * 30

# Compress with content size in frame
compressor_with_size = zstd.ZstdCompressor(write_content_size=True)
compressed_with_size = compressor_with_size.compress(data)

# Compress without content size
compressor_no_size = zstd.ZstdCompressor(write_content_size=False)
compressed_no_size = compressor_no_size.compress(data)

print(f"Original data: {len(data)} bytes")
print(f"With content size: {len(compressed_with_size)} bytes")
print(f"Without content size: {len(compressed_no_size)} bytes")

# Get frame parameters
params = zstd.get_frame_parameters(compressed_with_size)
print(f"\nFrame parameters:")
print(f"  Content size: {params.content_size}")
print(f"  Window size: {params.window_size}")
print(f"  Dict ID: {params.dict_id}")

print("Example 5 completed")
```
