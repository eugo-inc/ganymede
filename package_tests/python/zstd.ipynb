## Example 1: zstd functionality test 1

```python
import zstd

# Example 1: Basic ZSTD_compress and ZSTD_uncompress
print("Example 1: Basic compression and decompression")

# Original data
data = b"Hello World! " * 100
print(f"Original size: {len(data)} bytes")

# Compress using ZSTD_compress (unique to zstd package)
compressed = zstd.ZSTD_compress(data)
print(f"Compressed size: {len(compressed)} bytes")
print(f"Compression ratio: {len(data)/len(compressed):.2f}x")

# Decompress using ZSTD_uncompress (unique to zstd package)
decompressed = zstd.ZSTD_uncompress(compressed)
print(f"Decompressed size: {len(decompressed)} bytes")
print(f"Data matches: {data == decompressed}")

print("Example 1 completed")
```

## Example 2: zstd functionality test 2

```python
import zstd

# Example 2: Compression levels with ZSTD_compress
print("Example 2: Compression levels")

data = b"The quick brown fox jumps over the lazy dog. " * 50
print(f"Original size: {len(data)} bytes\n")

# Test different compression levels (1=fast, 22=max compression)
for level in [1, 5, 10, 15, 22]:
    compressed = zstd.ZSTD_compress(data, level)
    ratio = len(data) / len(compressed)
    print(f"Level {level:2d}: {len(compressed):4d} bytes, ratio: {ratio:.2f}x")

# Default compression level
default_level = zstd.ZSTD_default_compression_level()
print(f"\nDefault level: {default_level}")
compressed_default = zstd.ZSTD_compress(data, default_level)
print(f"Default compression: {len(compressed_default)} bytes")

print("Example 2 completed")
```

## Example 3: zstd functionality test 3

```python
import zstd

# Example 3: Using dumps and loads aliases
print("Example 3: Aliases dumps and loads")

# Original data - can be any Python bytes
data = b"Testing zstd dumps and loads functions! " * 30
print(f"Original size: {len(data)} bytes")

# dumps is an alias for ZSTD_compress (unique to zstd)
compressed = zstd.dumps(data)
print(f"Compressed with dumps(): {len(compressed)} bytes")

# loads is an alias for ZSTD_uncompress (unique to zstd)
decompressed = zstd.loads(compressed)
print(f"Decompressed with loads(): {len(decompressed)} bytes")
print(f"Data matches: {data == decompressed}")

# Compare with direct functions
compressed2 = zstd.ZSTD_compress(data)
print(f"\nUsing ZSTD_compress: {len(compressed2)} bytes")
print(f"Results identical: {compressed == compressed2}")

print("Example 3 completed")
```

## Example 4: zstd functionality test 4

```python
import zstd

# Example 4: Using compress and decompress aliases
print("Example 4: Aliases compress and decompress")

# Test data
data = b"Testing compress and decompress aliases! " * 25
print(f"Original size: {len(data)} bytes")

# compress is an alias for ZSTD_compress (unique to zstd)
compressed = zstd.compress(data)
print(f"Compressed with compress(): {len(compressed)} bytes")

# decompress is an alias for ZSTD_uncompress (unique to zstd)
decompressed = zstd.decompress(compressed)
print(f"Decompressed with decompress(): {len(decompressed)} bytes")
print(f"Data matches: {data == decompressed}")

# Test with different compression level
compressed_level_5 = zstd.compress(data, 5)
compressed_level_15 = zstd.compress(data, 15)
print(f"\nLevel 5: {len(compressed_level_5)} bytes")
print(f"Level 15: {len(compressed_level_15)} bytes")
print(f"Improvement: {len(compressed_level_5) - len(compressed_level_15)} bytes")

print("Example 4 completed")
```

## Example 5: zstd functionality test 5

```python
import zstd

# Example 5: Version information and high compression levels
print("Example 5: Version and compression levels")

# Get zstd library version string (unique to zstd)
version_str = zstd.ZSTD_version()
print(f"ZSTD library version: {version_str}")

# Get version as number (unique to zstd)
version_num = zstd.ZSTD_version_number()
print(f"Version number: {version_num}")

# Test various high compression levels
data = b"Testing various compression levels " * 20
print(f"\nOriginal size: {len(data)} bytes")

# Test levels from default to high compression
levels_to_test = [3, 10, 15, 20, 22]
for level in levels_to_test:
    compressed = zstd.ZSTD_compress(data, level)
    print(f"Level {level:2d}: {len(compressed)} bytes")

print("Example 5 completed")
```
